# Paper

Title: "Auditing LLM Alpha-Factor Mining: Metric Provenance and Full-Cohort Reanalysis"

Authors: Yufeng He

E-Journal Submission Date: 21 October 2026

Link: https://arxiv.org/abs/2610.29500

Abstract:

Reliable evaluation of LLM-based alpha mining requires more than positive backtest summaries. We audit a preserved project at three levels: historical run counters, outcome retention, and a separately scoped public reanalysis. Three historical runs report 67, 20, and 11 keep decisions, but the comparison mixes metric populations: a mean over kept candidates is compared with a mean of per-round maxima, and successful evaluations with all returned outcomes. These artifacts cannot identify an architecture advantage. A released 1,837-expression catalog also excludes non-positive OOS outcomes, making its zero-reversal rate non-informative. We then account for all 1,047 unique expressions in a frozen, IS-screened preserved pool using forty A-shares and 2024-2025 rank associations. A versioned public-operator correction yields 923 scorable records, 99 with insufficient valid dates, and 25 unsupported, without modifying the other 990 prior diagnostics. Among scorable records, 671 have positive and 252 negative signed mean rank-IC. Simple untrained references also show positive associations, so this distribution does not establish incremental agent skill or profitability. The case study motivates explicit metric grains, immutable cohort accounting, and provenance-aware interpretation rather than architectural rankings from incomplete records.

# Paper Summary

# Granular Summary and Structured Synopsis

## Step 0 — Bibliographic identification and genre classification

1. **Title:** *Auditing LLM Alpha-Factor Mining: Metric Provenance and Full-Cohort Reanalysis.*
2. **Author/affiliation:** Yufeng He, The University of Hong Kong (single-author manuscript).
3. **Format:** Two-column `article` class, Times, 0.75in margins; eight numbered sections plus an unnumbered Ethics Statement, an AI-assistant disclosure, and a three-item bibliography.
4. **Genre:** This is **not** an algorithmic-contribution paper. It is a *forensic evaluation-audit case study* combined with a *complete-cohort reanalysis* under a public evaluator. Its unit of analysis is a single preserved research project, and its epistemic claim is negative/deflationary: the preserved records do not support the architectural or generalization conclusions they were initially read as supporting.
5. **Rhetorical posture:** Deliberately restrained. The author repeatedly disclaims superiority claims, refuses to convert positive rank-IC into profitability, refuses to restore OOS claims, and refuses to treat template-hardcoded model labels as verified backend identities. This rhetorical discipline is itself part of the paper's methodological thesis.

## Step 1 — Problem formulation (Introduction)

The paper opens by identifying an asymmetry between the *effort* invested in LLM alpha-mining pipelines and the *epistemic adequacy* of the evidence retained to evaluate them. A pipeline comprises expression generation (LLM), a backtester, and selection rules. The failure mode is not that the pipeline produces nothing, but that it produces **plausible summary tables whose generating populations are unrecoverable**.

Three canonical ambiguities are named explicitly, and these function as the paper's organizing taxonomy:

1. **Grain ambiguity in a named metric.** A field called *Sharpe* is silent on *which population* was averaged (kept candidates? per-round maxima? all evaluated candidates? successful outcomes only?).
2. **Denominator ambiguity in a named counter.** A count called *backtested* is silent on whether failures, errors, and sign-flipped duplicates are in the denominator.
3. **Support truncation in a retained artifact.** A positive-only artifact is *arithmetically incapable* of estimating the frequency of negative outcomes; any "zero-reversal rate" computed from it is a property of the retention rule, not of the underlying signal population.

The stated research question is therefore deflationary: *which conclusions do the counters, source paths, and retained artifacts actually support?* The paper explicitly does not propose a new mining algorithm and does not claim superiority over prior systems.

## Step 2 — Contribution architecture (three nested, non-interchangeable audits)

The Introduction and §3 jointly announce a three-level audit, and the paper is at pains to insist these are **related project artifacts, not interchangeable experimental populations**:

| Level | Object audited | Section | Finding in brief |
|---|---|---|---|
| 1 | Source code + run reports | §4 | Incompatible metric grains; incomplete configuration and initial-state provenance; no controlled experiment |
| 2 | 1,837-expression "OOS validation" catalog | §5 | Outcome-conditioned retention; zero sign-reversal rate is non-informative |
| 3 | 1,047-expression frozen cohort, publicly re-evaluated | §6 | Complete outcome accounting (923 scorable / 124 non-scorable); descriptive association only; no incremental skill established |

The structural logic is a descending ladder of evidentiary strength: the first audit destroys the *comparative* claim (architecture), the second destroys the *generalization* claim (OOS validity), and the third replaces both with a *complete-denominator descriptive* claim that survives scrutiny.

## Step 3 — Positioning against prior work (Related Work)

Three works anchor the field: AlphaAgent (regularized exploration against alpha decay), QuantaAlpha (evolutionary LLM alpha mining), and Cognitive Alpha Mining (code-based evolution). All three are cited as **motivation for careful evaluation of search outcomes**, not as comparanda. The author states explicitly:

1. The contribution is an artifact-level case study, not a new search method.
2. There is **no benchmark against these systems**.
3. The findings "do not imply that other systems lack adequate controls" — an explicit guard against over-generalization from *n* = 1 project.

This is the correct scope discipline: the audit generalizes as a *checklist of failure modes*, not as an empirical claim about the literature.

## Step 4 — Workflow and evidence scope (§3)

### 4a. The three-role pipeline
- **Researcher** plans a search direction; **Implementor** generates and tests expressions; **Reviewer** provides the next-step suggestion.
- Critically, **keep/discard decisions are programmatic** in the inspected source — they are not LLM judgments. The Reviewer's LLM-generated next-step suggestion is wrapped in a **rule fallback on failure**.
- Therefore: *a role name or a model configuration is not a verified successful model request.* The exact historical request graph was **not recovered**.

### 4b. The twenty-specialist swarm
- Twenty specialists across **twelve configured categories**, an orchestrator, **adaptive allocation weights**, **sign-flipped candidates**, and **periodic prompt-revision proposals**.
- Both workflows share a functional price/volume DSL and a **proprietary evaluator**.

### 4c. Selection-policy provenance gap
- Narrative thresholds are quoted as Sharpe ≥ 2, Fitness ≥ 0, Turnover ≤ 500.
- But preserved source paths also contain **workflow-specific IC, complexity, pool-score, and correlation checks**.
- Because **exact executed policy versions and matching initial pool states are not bound to the reports**, the paper concludes that *identical effective selection policies are not established* — a subtle but decisive point: even if headline thresholds match, the effective acceptance regions may differ across workflows.

### 4d. Evidence inventory
(a) March 2026 run reports; (b) a native OOS checkpoint; (c) an outcome-filtered catalog; (d) a separately frozen list selected from **3,114 preserved input rows**. Current source is treated as *explaining report construction*, **not** as the exact historically executed commit. Crucially, the author performs **internal corroboration by recomputing statistics from printed report rows** — the correct forensic move when the executable provenance is unavailable.

## Step 5 — Audit I: The historical run-report audit (§4)

### 5a. Recoverable counters (Table 1)
| Counter | Pipeline (P) | Swarm A (S-A) | Swarm B (S-B) |
|---|---|---|---|
| Rounds | 20 | 20 | 20 |
| Keep decisions | 67 | 20 | 11 |
| Candidates tested | 289 | — | — |
| Successful outcomes | 234 | — | — |
| Generated proposals | — | 887 | 876 |
| Returned outcomes | — | 816 | 461 |
| Keeps / round | 3.35 | 1.00 | 0.55 |

Two interpretive safeguards are stated: **missing cells denote unavailability at a harmonized stage, not zeros**; and **keep decisions are not proved counts of distinct factors surviving into a final pool**.

### 5b. The grain mismatch (the paper's cleanest finding)
- The pipeline collector appends **Sharpe values of kept candidates**.
- The swarm collector appends **positive per-round maxima**.
- The three-decimal report entries reproduce as **2.227 over 67 kept rows** and **2.433 over 20 round maxima** — matching the displayed "2.23 vs 2.43."
- The two numbers are means over *different populations at different stages* and therefore constitute **no performance comparison whatsoever**. (Note the arithmetic: 67 × 2.227 ≈ 149.2 total Sharpe mass; 20 × 2.433 ≈ 48.7 — the aggregation is over a mean-of-keeps versus a mean-of-extrema, i.e., an *order statistic*, which is upward-biased by construction relative to a mean.)

### 5c. The extremum is also a different population
- Pipeline: max kept-candidate Sharpe = **2.449**, but max *evaluated round summary* = **3.171**.
- Swarm: **4.206** is an *evaluated round maximum* and **need not correspond to a kept factor**.
- Conclusion: these values cannot establish differential outlier discovery.

### 5d. The denominator mismatch
- 234 is P's **successful-outcome** count; 816 is S-A's **returned-outcome** count, which includes possible failures and sign-flip evaluations.
- Diagnostic anomaly: **six S-A rounds have more outcome records than newly generated proposals** — direct evidence that the returned-outcome stream is not a one-to-one map of generated candidates.
- Therefore **67/234 versus 20/816 is not a verified common-denominator acceptance-rate comparison.**

### 5e. Category and attribution mismatch
- P: four category labels cover all round directives. S-A/S-B: two "productive" labels cover only **mapped specialists with keep decisions**.
- Attribution leakage: S-A's agent leaderboard attributes **8** keeps while the run counter is **20**; S-B attributes **9** while the counter is **11**.
- The **sign-flip route can fall outside configured-specialist attribution**, but per-event attribution was not reconstructed → the residual is unexplained, not assigned.

### 5f. State-provenance failure
- Comparison code runs the pipeline **before** the swarm **without an explicit initial-pool snapshot or reset binding**.
- Combined with missing executed versions and request-level provenance, this means the reports **cannot be treated as a controlled experiment**.
- The methodological coda is important: **more seeds would not repair incompatible counters.** Repair requires aligning *accounting stages, selection policies, initial state, and resource budgets* before independent runs can support inference.

## Step 6 — Audit II: The outcome-retention audit (§5)

This is a pure **selection-bias identification** result and is the strongest empirical finding in the paper.

1. The 1,837-record file was described as an **OOS validation set**.
2. A read-only audit of artifacts *and the runner* shows the runner (i) uses a **looser preliminary screen** and (ii) **explicitly drops records with OOS Sharpe ≤ 0** before writing the retained file.
3. Only **1,037** rows in the retained file satisfy the manuscript's *stated* IS rule (Sharpe ≥ 2, Fitness ≥ 0, Turnover ≤ 500) — so the file is not even internally consistent with its own stated admission rule.
4. The **native checkpoint** contains **2,016 unique expressions**, of which **368 have negative OOS Sharpe**.
5. Set-theoretic reconciliation (Table 2), which is exhaustive and mutually consistent:
   - Checkpoint-only: **368** expressions, **all** negative-OOS.
   - Catalog-only: **189** expressions, **all** positive-OOS.
   - Intersection: **1,648** expressions, **zero sign disagreements** (2,016 − 368 = 1,648; 1,837 − 189 = 1,648).
6. **Inference:** the catalog's **zero sign-reversal rate** and its **decay summaries** are properties of the *retention rule*, not evidence of generalization. The sign partition is a deterministic function of the filter.
7. **Remediation protocol specified** (and withheld as a claim): recover and hash the complete pre-OOS pool; pre-specify an IS-only screen; perform deduplication using IS data only; freeze the expression list **before** OOS access; checkpoint every response; preserve every numerical OOS outcome **and error, regardless of sign**. Until that run exists, numerical OOS and generalization claims are suspended.

## Step 7 — Audit III: Full-cohort public reanalysis (§6)

### 7a. Cohort definition and anti-cherry-picking guarantee
- **1,047 unique expressions** selected from 3,114 preserved input rows using the stated IS screen.
- Explicit scope limits: **not** a census of generated candidates; **not** a historically blind holdout; **not** a matched replay of the three historical keep totals.
- **All IDs are retained regardless of the new evaluator's outcome**; no expression is selected by its new score. This is an *intention-to-evaluate* design at the artifact level.

### 7b. Data construction and verification
- Universe: the **same ordered 40-stock convenience universe**, **727 trading dates**, 2023–2025.
- Source: **BaoStock 0.9.3**, both **unadjusted** and **provider-adjusted** daily views → **80 files** (40 × 2).
- Verification performed: byte hashes, date/stock/adjustment identities, and **shared quantity and trading-status fields** across the two views.
- **Features use nominal** prices, shares, and nominal traded amount; the **adjusted view is reserved exclusively for labels** (correct separation, since adjustment factors embed future information).
- Two non-trading stock-day cells are masked; **no forward-fill**; **no substitution of a smaller acquired stock prefix** (guards against survivorship-by-convenience).

### 7c. Label and metric definitions
- Label: five-trading-row provider-adjusted close change, $y_{i,t} = P^{\text{adj}}_{i,t+5}/P^{\text{adj}}_{i,t} - 1$, for each 2024–2025 date $t$. Both endpoints must be finite, positive, and **tradable**. Explicitly **not** an investor total-return or executable P&L claim. The **last five captured dates have no label**.
- Daily rank-IC: **Pearson correlation of average-tie cross-sectional ranks**.
- Eligibility: ≥ 5 eligible stocks, **≥ 3 distinct signal values**, ≥ 2 distinct labels. The three-value threshold is declared a **diagnostic convention** — the author explicitly refuses the stronger claim that binary signals lack a well-defined rank correlation. This is a correctly hedged operationalization.
- Scorability: a factor is scorable only with **≥ 20 valid dates**.
- The mean/SD ratio, if reported, is **unannualized** and is **not** a trading Sharpe.

### 7d. The versioned implementation defect
- The frozen public implementation initially yields **57 numerical errors**: the conditional operator attempts to interpret a **panel as a single Boolean** when the true branch is scalar.
- Handling: **v1 is preserved**; a **versioned scalar-broadcasting correction** is added preserving the existing elementwise conditional semantics; **all 1,047 records are re-evaluated** rather than re-running only the promising failures; the **other 990 diagnostic records are unchanged**.
- Consequence: **56** corrected records become scorable; **1** remains insufficiently observed.
- *Arithmetic cross-check (internal consistency confirmed):* 1,047 = 990 + 57; and 990 = 867 scorable + 98 insufficient + 25 unsupported; adding the 56 newly scorable and the 1 residual insufficient gives 923 scorable + 99 insufficient + 25 unsupported = 1,047. The accounting closes exactly, which is a meaningful integrity signal for a paper of this type.

### 7e. Results (Table 3, descriptive only)
| Outcome | Records |
|---|---|
| Scorable signed rank-IC | **923** |
| Insufficient valid dates | 99 |
| Unsupported operator or variable | 25 |
| All frozen records | **1,047** |
| Positive mean rank-IC (of 923) | 671 |
| Negative mean rank-IC (of 923) | 252 |
| Median mean rank-IC (of 923) | **0.0148** |

Supplementary facts: **835 of 923** have all **480** label-available target dates; the other **88** have varying coverage, which further limits cross-factor comparison. **Nine** fixed record/date rank-IC cells were independently recomputed with a separate tie-ranking/Pearson implementation. **No independent-factor significance test** is attached, because factors and overlapping five-day labels are not independent observations. The results are **not** native OOS Sharpe, are **not** interchangeable with the outcome-filtered catalog's metrics, and do **not** identify an architecture effect.

### 7f. Fixed reference-signal context (Table 4)
Specified **before** computing their outcomes; signs and horizons **not** selected on the resulting scores; same snapshot and scoring convention; each with **480** valid dates.

| Reference | Mean signed rank-IC |
|---|---|
| Nominal close level | −0.0218 |
| Share volume | +0.0234 |
| Nominal traded amount | +0.0021 |
| One-day nominal reversal | −0.0007 |
| Five-day nominal reversal | −0.0155 |
| Twenty-day nominal change | +0.0229 |

Interpretation as stated: reversal references negate nominal trailing close changes; the twenty-day reference retains positive orientation. Because **volume and recent-change references also post positive associations**, a positive IC alone does **not** establish incremental agent skill; incremental prediction beyond these references was **not estimated**. Because factors have different eligible dates and stocks, the cohort median versus the standalone reference means does **not** establish matched-sample outperformance. No reference is promoted to a profitability or significance claim.

*Econometric reading worth flagging:* the sign pattern is internally coherent as an exposure story. The negative close-level IC (−0.0218) is a price-level (size-like) tilt; the five-day reversal reference at −0.0155 implies *positive* five-day momentum of about +0.0155; the positive twenty-day change (+0.0229) is consistent momentum. Hence the cohort's 671/252 positive skew and its 0.0148 median are **plausibly attributable to a momentum-plus-price-level exposure** rather than to search-derived structure. This is exactly the inference the paper's own caveat invites, though it stops short of asserting it.

## Step 8 — Implications and Limitations (§7)

**What survives as the contribution:** an evaluation-audit case study — *not* an architectural ranking. Historical reports preserve **different keep counters** but not **verified common-grain** Sharpe, acceptance-rate, or category comparisons. Single runs, unequal budgets, unbound starting state, and uncertain executed configurations further limit interpretation. The public cohort derives from a broader preserved pool and **cannot be assigned back** to the three run conditions from available records.

**Residual limitations, as enumerated:**
1. One market; a convenience stock universe.
2. Public rather than native (proprietary) semantics.
3. Variable factor coverage; overlapping five-day labels.
4. The proprietary evaluator and full historical request traces are **not released**.
5. The reference checks were **planned after** the cohort analysis — i.e., not part of the original primary protocol (an honest admission of post-hoc status, though mitigated by pre-specification of the references *before* their outcomes were computed).

**Design requirements for a stronger study:** fixed common masks and comparators; retained generation and run identities; repeated systems under **matched budgets**; a suitable temporal design.

## Step 9 — Conclusion, ethics, and disclosure (§8 + statements)

- The conclusion restates the three concrete defect classes: **metric-population mismatch, outcome-conditioned retention, conditional-operator defect**.
- The reanalysis is **complete and versioned**, retaining **all 1,047 IDs including 124 non-scorable records** (99 + 25), with **no missing result assigned zero alpha**.
- Positive associations are **not** promoted to profitability or architecture effects.
- Future comparisons require **aligned definitions and provenance before additional scale or significance testing** — note the ordering claim: *provenance precedes scale.*
- **Ethics:** public historical market data; no live trading, no advice, no orders; failed and non-scorable records preserved; proprietary components and raw provider data **not** represented as publicly redistributable; the SDK license is explicitly **not** a data-redistribution grant.
- **Disclosure:** AI assistants supported literature review, audit design, implementation, statistical checking, and manuscript preparation; numerical findings are computed from preserved records and versioned outputs, not generated substitutes for observations; **no human-annotation study or independent human review is claimed**; author retains responsibility.

## Step 10 — Critical appraisal

**Strengths (methodological):**
1. **Grain-first auditing.** The insistence that a metric name carries no meaning without its generating population is the correct primitive, and the 2.227/2.433 reproduction from printed rows is a genuine forensic verification rather than an assertion.
2. **Selection-bias identification via set algebra.** The 368/189/1,648 partition with zero sign disagreement is a clean, falsifying demonstration that a "validation set" is a filtered object. The logic is airtight: the sign partition is a deterministic function of the filter, so the reversal rate is uninformative by construction.
3. **Complete-denominator discipline.** Retaining 1,047 IDs including 124 non-scorable, and refusing to impute missing alpha as zero, is the artifact-level analogue of intention-to-treat — the right default and rarely practiced.
4. **Versioned correction with full re-evaluation.** Fixing the conditional-operator defect and then re-running *all* records, rather than only the failures, eliminates the re-selection channel that would otherwise reintroduce the very bias under audit.
5. **Rhetorical restraint as method.** Refusing to convert rank-IC into Sharpe, refusing to annualize an unannualized ratio, refusing to test significance on overlapping labels, and flagging post-hoc reference specification are all correct and are the paper's real contribution to practice.

**Residual weaknesses and open questions:**
1. **External validity is *n* = 1.** The findings are a catalogue of failure modes instantiated once; the paper says so, but the practical uptake will depend on readers treating it as a checklist rather than a verdict on LLM alpha mining.
2. **The reanalysis cohort is itself a selected sample.** It is screened on a proprietary in-sample Sharpe criterion. Whether the "IS screen" window overlaps the 2024–2025 label window is not fully resolved in the reproduced text; if it does overlap, the positive-IC majority inherits an in-sample conditioning channel distinct from — but structurally similar to — the retention bias condemned in §5.
3. **No null calibration.** Six fixed references are context, not a null distribution. A cross-sectional permutation/bootstrap null on the rank-IC statistic, or a randomized-label placebo, would convert the 671/252 split from descriptive to inferential at negligible cost.
4. **The 3-distinct-value threshold filters non-randomly.** Excluding near-constant and binary signals removes precisely the sparse-indicator class common in mined expression DSLs; the 923-record cohort is therefore a structurally censored subsample, and the 25 "unsupported" records compound this.
5. **Effect magnitude is under-reported.** The median (0.0148) is given but not the mean, dispersion, or the distribution's shape; the reference means (|IC| ≈ 0.02) are of the same order as the cohort median, which strongly suggests the cohort's edge is exposure, not alpha — a point the paper implies but does not quantify.
6. **The v1 defect's inferential consequence is unquantified.** The paper reports 57 errors → 56 recovered, but not whether the pre-correction scorable subset was biased in mean IC (i.e., whether the bug preferentially destroyed high- or low-signal expressions). Given §6's own thesis, that is the obvious self-application of the audit method.

**Bottom line.** The paper's durable contribution is a **provenance protocol**: bind metrics to populations, bind counters to stages, bind claims to unfiltered cohorts, bind implementations to versions, and bind interpretive claims to pre-specified comparators. Its empirical object is a single project, but its evidentiary standard — complete denominators, preserved failures, versioned corrections, and refused inferences — is general and correctly stated as such.

```python
# Task 1: Validate the Appropriateness and Quality of All Ingested Parameters

# ==============================================================================
# Task 1: Validate the Appropriateness and Quality of All Ingested Parameters
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Pinned study invariants. Provenance: study configuration P11_ACQUISITION_PARAMETERS,
# P12_MASKING_PARAMETERS, P44_RAW_LAYER_INVARIANTS, raw_input_schemas.*; LaTeX section 6
# (40-stock convenience universe, 727 trading dates 2023-2025, two views via BaoStock 0.9.3,
# 80 verified files, two masked non-trading stock-day cells); detailed task list step texts.
# -------------------------------------------------------------------------------------------------------------------------------

DEFAULT_FILE_COUNT: int = 80
# Study pin: raw_input_schemas.FILES.row_count; P11 file_count (40 stocks * 2 views).
DEFAULT_TRADING_DATE_COUNT: int = 727
# Study pin: P11 trading_date_count; P44 calendar_rows_equals_727.
DEFAULT_ROWS_PER_BAR_FILE: int = 727
# Study pin: P11 rows_per_bar_file; P44 rows_per_bar_file_equals_727.
DEFAULT_UNIVERSE_SIZE: int = 40
# Study pin: P11 universe_size; the 40-stock convenience universe of LaTeX section 6.
DEFAULT_PRE_MASK_ROW_COUNT: int = 40 * 727
# Study pin: raw_input_schemas.{UNADJUSTED,ADJUSTED}_BARS.estimated_row_count == 29080.
DEFAULT_MASKED_CELL_COUNT: int = 2
# Study pin: P12 masked_cell_count; LaTeX section 6 ("Two non-trading stock-day cells").
DEFAULT_PROVIDER: str = "BaoStock"
# Study pin: P11 provider; attrs_header_fields.provider.
DEFAULT_PROVIDER_VERSION: str = "0.9.3"
# Study pin: P11 provider_version; attrs_header_fields.provider_version.
DEFAULT_TAG_VOCABULARY: Tuple[str, ...] = ("[S]", "[D]", "[R]", "[I]", "UNDISCLOSED")
# Study pin: P00_META.provenance_tags; the closed enumeration of recognised provenance tags.
UNSPEC_TAG_PREFIXES: Tuple[str, ...] = (
    "[UNSPECIFIED IN MANUSCRIPT",
    "[UNSPECIFIED-IN-MANUSCRIPT",
)
# Study pin: P02/P06/etc. carry open-ended "[UNSPECIFIED IN MANUSCRIPT: ...]" tag markers;
# the walker matches the literal prefix only and never parses the prose.
EXACT_TAG_BUCKETS: Dict[str, str] = {
    "[S]": "S",
    "[D]": "D",
    "[R]": "R",
    "[I]": "I",
    "UNDISCLOSED": "UNDISCLOSED",
}
# Study pin: step 3 recognised-tag enumeration; each exact tag maps to its count bucket.
PINNED_P45_KEYS: Tuple[str, ...] = (
    "fallback_rate_reported",
    "universe_constituents_disclosed",
    "is_screen_window_disclosed",
    "prompts_disclosed",
    "inference_settings_disclosed",
    "api_credentials_disclosed",
    "grammar_disclosed",
    "executed_commit_bound_to_report",
    "initial_pool_hash_available",
    "specialist_to_category_map_disclosed",
    "allocation_update_rule_disclosed",
    "prompt_revision_periodicity_disclosed",
)
# Study pin: step 3 disclosure-gap reconciliation; the twelve pinned P45_DISCLOSURE_GAPS keys.
# The optional STUDY_CONFIG section mapping each UNDISCLOSED leaf to its P45 gap (operator ruling, 2026-11-12).
P45_GAP_RECONCILIATION_KEY: str = "P45_GAP_RECONCILIATION"
# The gap-map value that marks a literal "UNDISCLOSED" leaf as no disclosure gap at all (e.g. a vocabulary entry).
P45_NOT_A_GAP: str = "NOT_A_GAP"
PINNED_ATTR_KEYS: Tuple[str, ...] = (
    "provider",
    "provider_version",
    "universe_label",
    "universe_order_fixed",
    "universe_disclosed",
    "date_count",
    "file_count",
    "masked_cell_count",
    "retrieval_batch_timestamp",
    "retrieval_coherence_window",
    "config_self_hash",
    "redistribution_grant",
)
# Study pin: step 1 attrs header fields; twelve keys carried in FILES.attrs, never as columns.
ADDED_FILES_COLUMNS: Tuple[str, ...] = (
    "file_uri",
    "tag_registry_version",
    "content_length_ok",
    "response_truncated",
)
# Study pin: step 1 four columns added before validation begins; never subject to the extras halt.
PRICE_STEMS: Tuple[str, ...] = ("open", "high", "low", "close")
# Study pin: step 2 sub-check (viii); the four nominal price fields and their _adj counterparts.
SHA256_HEX_LENGTH: int = 64
# Study pin: hashlib.sha256 digest length in lowercase hex.

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 1: FILES manifest materialisation and the ten ordered sub-checks
# -------------------------------------------------------------------------------------------------------------------------------

def _string_backend_dtype() -> str:
    """
    Return the pandas string dtype spelling for new string columns.

    Returns
    -------
    str
        Always ``"string[pyarrow]"``: the ``try`` body returns
        directly and the ``except`` branch is unreachable, so the
        plain ``"string"`` fallback is never produced.

    Notes
    -----
    The ``"string[pyarrow]"`` dtype is a pandas ``StringDtype``, so
    ``isinstance(dtype, pd.StringDtype)`` conformance checks hold.
    """
    # Attempt the pyarrow-backed string dtype, which the step text pins explicitly.
    try:
        # Return the pinned pyarrow-backed spelling.
        return "string[pyarrow]"
    except Exception:
        # Fall back to the portable pandas string dtype; still a StringDtype instance.
        return "string"

def _sha256_hex(payload: bytes) -> str:
    """
    Compute the lowercase hex SHA-256 digest of a byte payload.

    Digest via ``hashlib.sha256`` over ``bytes(payload)``; the
    process-seeded builtin ``hash`` is never used. For the config
    self-hash, ``payload`` is the canonical-JSON bytes returned by
    :func:`_canonical_json_bytes`.

    Parameters
    ----------
    payload : bytes
        Exact bytes to digest; a ``bytearray`` is accepted and copied
        through ``bytes()``.

    Returns
    -------
    str
        64-character lowercase hexadecimal digest.

    Raises
    ------
    TypeError
        If ``payload`` is not a ``bytes`` or ``bytearray`` instance.
    """
    # Validate the payload type before digesting to fail fast with an actionable message.
    if not isinstance(payload, (bytes, bytearray)):
        # Refuse anything else; hashlib accepts bytes-like objects only.
        raise TypeError(f"payload must be bytes, got {type(payload)!r}.")
    # Compute the digest through hashlib (step 1 forbids hash(config) outright).
    return hashlib.sha256(bytes(payload)).hexdigest()

def _canonical_json_bytes(obj: Any) -> bytes:
    r"""
    Serialise an object to the pinned canonical-JSON corpus bytes.

    The value is normalised recursively, then dumped with sorted keys,
    compact separators, ``ensure_ascii=False`` and ``allow_nan=False``,
    encoded as UTF-8 and terminated with a single LF byte.

    Parameters
    ----------
    obj : Any
        Object to serialise. The config self-hash passes the whole
        ``STUDY_CONFIG`` mapping, including ``_provenance_tags``.

    Returns
    -------
    bytes
        UTF-8 canonical-JSON bytes ending in ``b"\n"``.

    Notes
    -----
    Pinned rules: ``None``/``bool``/``int``/``str`` verbatim; finite
    ``float`` verbatim; ``NaN`` -> ``"NaN"``; ``+/-inf`` ->
    ``"Infinity"``/``"-Infinity"``; ``pd.NA`` and ``pd.NaT`` ->
    ``"__NA__"``; ``np.bool_``, ``np.integer`` and
    ``np.floating`` scalars degrade to Python ``bool``/``int``/
    ``float`` (``np.str_`` passes the ``str`` branch; other numpy
    scalar types, e.g. ``np.datetime64``, are not handled and
    reach ``TypeError``); ``datetime`` /
    ``Timestamp`` -> ``isoformat()`` with offset kept; ``set`` ->
    list sorted by ``repr``; ``tuple``/``list`` -> list in order;
    ``dict`` -> keys passed through ``str`` with sorted keys at
    dump time.

    Determinism holds for a fixed normalised structure. Because keys
    are stringified before the sorted dump, two equal dicts whose
    distinct keys stringify to the same text can still serialise
    differently, since the surviving entry depends on insertion
    order.
    """
    # Inner recursive normaliser: mirror the pinned corpus rule by rule.
    def _normalise(value: Any) -> Any:
        """
        Convert one value into its canonical-JSON-ready form.

        Applied elementwise and recursively to containers.

        Parameters
        ----------
        value : Any
            Scalar, container, numpy scalar or pandas sentinel.

        Returns
        -------
        Any
            A JSON-dumpable value: a scalar, a sentinel string,
            ``"NaN"``, ``"Infinity"``, ``"-Infinity"``,
            ``"__NA__"``, an ISO8601 string, a list, or a dict
            with string keys.

        Raises
        ------
        TypeError
            If ``value`` has no canonical-JSON representation.
        """
        # None, booleans, integers and strings pass through unchanged.
        if value is None or isinstance(value, (bool, int, str)):
            # Return the scalar verbatim; JSON handles all four.
            return value
        # Floats require NaN/inf escaping before JSON serialisation.
        if isinstance(value, float):
            # NaN is pinned to the literal "NaN" token.
            if math.isnan(value):
                # Return the sentinel string, not a raw NaN float.
                return "NaN"
            # Infinities are pinned to the literal "Infinity" / "-Infinity" tokens.
            if math.isinf(value):
                # Return the signed infinity token.
                return "Infinity" if value > 0 else "-Infinity"
            # Ordinary finite floats pass through unchanged.
            return value
        # pandas NA / NaT are pinned to the literal "__NA__" token.
        if value is pd.NA or value is pd.NaT:
            # Return the NA sentinel string.
            return "__NA__"
        # numpy scalars degrade to their Python counterparts before further checks.
        if isinstance(value, np.bool_):
            # Coerce numpy boolean to Python bool.
            return bool(value)
        # numpy integers degrade to Python ints.
        if isinstance(value, np.integer):
            # Coerce numpy integer to Python int.
            return int(value)
        # numpy floats degrade to Python floats with the same NaN/inf escaping.
        if isinstance(value, np.floating):
            # Route numpy floats through the float branch for identical handling.
            return _normalise(float(value))
        # datetime and pandas Timestamps pin to ISO8601 (offset kept when aware).
        if isinstance(value, (pd.Timestamp, datetime.datetime)):
            # Return the isoformat() rendering; aware stamps keep their offset.
            return value.isoformat()
        # sets pin to lists sorted by repr for determinism.
        if isinstance(value, set):
            # Sort by repr to make mixed-type sets deterministic.
            return [_normalise(item) for item in sorted(value, key=repr)]
        # tuples pin to lists (JSON has no tuple).
        if isinstance(value, tuple):
            # Convert elementwise preserving order.
            return [_normalise(item) for item in value]
        # lists pass through elementwise.
        if isinstance(value, list):
            # Normalise each element preserving order.
            return [_normalise(item) for item in value]
        # dicts pin to stringified keys (sorted at dump time).
        if isinstance(value, dict):
            # Stringify keys exactly as the pinned corpus requires.
            return {str(key): _normalise(item) for key, item in value.items()}
        # Anything else is unserialisable under the pinned corpus.
        raise TypeError(f"Unserialisable value of type {type(value)!r} passed to the canonical-JSON serialiser.")

    # Normalise, dump compact with sorted keys, and terminate with LF.
    serialised = json.dumps(
        _normalise(obj),
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=False,
        allow_nan=False,
    )
    # Encode to UTF-8 and terminate with a single LF byte sequence.
    return serialised.encode("utf-8") + b"\n"

def _append_report_record(
    records: List[Dict[str, Any]],
    check_id: str,
    severity: str,
    message: str,
    offending_row_ids: List[str],
) -> None:
    """
    Append one record to the in-flight validation report buffer.

    The appended dict has keys ``check_id``, ``severity``, ``message``
    and ``offending_row_ids`` (a shallow copy of the supplied list),
    in call order. The buffer is materialised once per return point by
    :func:`_finalize_validation_report`; nothing is printed.

    Parameters
    ----------
    records : List[Dict[str, Any]]
        Accumulator list mutated in place.
    check_id : str
        Namespaced identifier of the check, e.g. ``FILES.row_count``;
        must be non-empty after stripping.
    severity : str
        Exactly one of ``"INFO"`` (pass), ``"WARN"`` (disclosure
        path) or ``"ERROR"`` (halt).
    message : str
        Human-actionable diagnostic text.
    offending_row_ids : List[str]
        String identifiers of the offending rows or coordinates;
        empty on success.

    Raises
    ------
    ValueError
        If ``severity`` is outside ``{"INFO", "WARN", "ERROR"}`` or if
        ``check_id`` is not a non-empty string.
    """
    # Enforce the severity enumeration so downstream tooling can rely on it.
    if severity not in {"INFO", "WARN", "ERROR"}:
        # Reject unknown severities loudly rather than silently passing them through.
        raise ValueError(f"severity must be one of INFO/WARN/ERROR, got {severity!r}.")
    # Enforce a non-empty check identifier.
    if not isinstance(check_id, str) or not check_id.strip():
        # Reject empty check identifiers loudly.
        raise ValueError("check_id must be a non-empty string.")
    # Append the record; the buffer is materialised once per return point.
    records.append(
        {
            "check_id": check_id,
            "severity": severity,
            "message": message,
            "offending_row_ids": list(offending_row_ids),
        }
    )

def _record_and_halt(
    records: List[Dict[str, Any]],
    check_id: str,
    message: str,
    offending_row_ids: List[str],
) -> str:
    """
    Append an ``ERROR`` record and return the halt reason.

    Parameters
    ----------
    records : List[Dict[str, Any]]
        In-flight report buffer, mutated in place.
    check_id : str
        Namespaced identifier of the failing check.
    message : str
        Halt reason, stored as the record ``message`` and returned
        unchanged.
    offending_row_ids : List[str]
        String identifiers of the offending rows or coordinates.

    Returns
    -------
    str
        The same text as ``message``, for the orchestrator to
        propagate.

    Raises
    ------
    ValueError
        Propagated from :func:`_append_report_record` when
        ``check_id`` is not a non-empty string. ``severity`` is
        hard-coded to ``"ERROR"`` and cannot be invalid here.
    """
    # Append the ERROR record through the single vetted path.
    _append_report_record(records, check_id, "ERROR", message, offending_row_ids)
    # Return the halt reason scalar for the orchestrator.
    return message

def _finalize_validation_report(records: List[Dict[str, Any]]) -> pd.DataFrame:
    """
    Materialise the record buffer as the ``validation_report`` frame.

    Parameters
    ----------
    records : List[Dict[str, Any]]
        Records accumulated during validation, each carrying the keys
        ``check_id``, ``severity``, ``message`` and
        ``offending_row_ids``.

    Returns
    -------
    pd.DataFrame
        Write-once diagnostic frame indexed on ``check_id`` (index
        name ``check_id``) with columns ``severity`` (str), ``message``
        (str) and ``offending_row_ids`` (list of str). An empty buffer
        yields an empty frame with the same schema.
    """
    # Column order is fixed by the step text: (check_id, severity, message, offending_row_ids).
    columns = ["check_id", "severity", "message", "offending_row_ids"]
    # Build the frame from the buffer; empty buffers still yield the pinned schema.
    frame = pd.DataFrame.from_records(records, columns=columns)
    # Index on check_id as the step text pins; set_index preserves insertion order.
    frame = frame.set_index("check_id")
    # Pin the index name explicitly (set_index already names it check_id).
    frame.index.name = "check_id"
    # Return the write-once diagnostic.
    return frame

def _config_section(config: Dict[str, Any], section: str) -> Dict[str, Any]:
    """
    Read a top-level section of the study configuration defensively.

    Parameters
    ----------
    config : Dict[str, Any]
        Study configuration mapping.
    section : str
        Section name, e.g. ``"P11_ACQUISITION_PARAMETERS"``.

    Returns
    -------
    Dict[str, Any]
        The section mapping itself (not a copy); an empty dict when
        the key is absent or its value is not a dict.
    """
    # Fetch the raw value; tolerate absence and type drift without raising.
    section_value = config.get(section, {})
    # Return the mapping when it is one; otherwise an empty mapping.
    return section_value if isinstance(section_value, dict) else {}

def _files_schema(config: Dict[str, Any]) -> Dict[str, Any]:
    """
    Extract the ``FILES`` sub-schema from the configuration.

    Parameters
    ----------
    config : Dict[str, Any]
        Study configuration mapping.

    Returns
    -------
    Dict[str, Any]
        ``config["raw_input_schemas"]["FILES"]`` when it is a dict; an
        empty dict when either level is absent or malformed.
    """
    # Read raw_input_schemas defensively.
    raw = config.get("raw_input_schemas", {})
    # Tolerate a missing or malformed raw_input_schemas mapping.
    raw = raw if isinstance(raw, dict) else {}
    # Read the FILES sub-schema defensively.
    files_schema = raw.get("FILES", {})
    # Tolerate a missing or malformed FILES schema.
    return files_schema if isinstance(files_schema, dict) else {}

def _bars_schema(config: Dict[str, Any], view_token: str) -> Dict[str, Any]:
    """
    Extract an OHLCV panel sub-schema from the configuration.

    Parameters
    ----------
    config : Dict[str, Any]
        Study configuration mapping.
    view_token : str
        Exactly ``"UNADJUSTED_BARS"`` or ``"ADJUSTED_BARS"``.

    Returns
    -------
    Dict[str, Any]
        ``config["raw_input_schemas"][view_token]`` when it is a
        dict; an empty dict when either level is absent or malformed.

    Raises
    ------
    ValueError
        If ``view_token`` is neither of the two pinned tokens.
    """
    # Validate the token before touching the config.
    if view_token not in {"UNADJUSTED_BARS", "ADJUSTED_BARS"}:
        # Reject tokens outside the pinned pair.
        raise ValueError(f"view_token must be UNADJUSTED_BARS or ADJUSTED_BARS, got {view_token!r}.")
    # Read raw_input_schemas defensively.
    raw = config.get("raw_input_schemas", {})
    # Tolerate a missing or malformed raw_input_schemas mapping.
    raw = raw if isinstance(raw, dict) else {}
    # Read the panel sub-schema defensively.
    bars_schema = raw.get(view_token, {})
    # Tolerate a missing or malformed panel schema.
    return bars_schema if isinstance(bars_schema, dict) else {}

def _permitted_extra_columns(config: Dict[str, Any]) -> set:
    """
    Read the ``permitted_extra_columns`` allow-list from config.

    Parameters
    ----------
    config : Dict[str, Any]
        Study configuration mapping.

    Returns
    -------
    set
        The declared allow-list of undeclared column names permitted
        to enter ``FILES`` without halting; empty when the key is
        absent, is not a list/tuple/set, or holds no string entries.
    """
    # Fetch the allow-list; any malformed value degrades to an empty set.
    allow = config.get("permitted_extra_columns", [])
    # Tolerate non-list or non-string entries by keeping only strings.
    return {item for item in allow if isinstance(item, str)} if isinstance(allow, (list, tuple, set)) else set()

def _expected_universe_size(config: Dict[str, Any], expected_file_count: int) -> int:
    """
    Resolve the expected number of distinct stocks.

    ``P11_ACQUISITION_PARAMETERS.universe_size`` wins when it parses
    to a strictly positive integer (booleans are ignored). Otherwise
    the value is derived as ``expected_file_count // 2`` (two views
    per stock) when at least two files are declared, else
    ``DEFAULT_UNIVERSE_SIZE``; a non-positive derivation also falls
    back to ``DEFAULT_UNIVERSE_SIZE`` (40).

    Parameters
    ----------
    config : Dict[str, Any]
        Study configuration mapping.
    expected_file_count : int
        Declared ``FILES`` row count, used for the derived default.

    Returns
    -------
    int
        Expected distinct ``stock_code`` count.
    """
    # Read the P11 acquisition section defensively.
    p11 = _config_section(config, "P11_ACQUISITION_PARAMETERS")
    # Read the pinned universe size; coerce defensively.
    universe = p11.get("universe_size", None)
    # Accept integral values only; fall back to derive from the file count.
    if isinstance(universe, bool):
        # Booleans are integral but never a universe size; ignore.
        universe = None
    # Coerce to int when integral; otherwise derive from the file count.
    try:
        # Accept both ints and integral floats/strings.
        coerced = int(universe) if universe is not None else None
    except (TypeError, ValueError):
        # Unparseable values fall back to the derived default.
        coerced = None
    # Return the coerced value when strictly positive; else the derived default.
    if coerced is not None and coerced > 0:
        # P11 wins when present and sane.
        return coerced
    # Derive from the file count (two views per stock) with the pinned floor.
    derived = expected_file_count // 2 if expected_file_count >= 2 else DEFAULT_UNIVERSE_SIZE
    # Guard against degenerate declarations by falling back to the pinned default.
    return derived if derived > 0 else DEFAULT_UNIVERSE_SIZE

def _expected_trading_date_count(config: Dict[str, Any]) -> int:
    """
    Resolve the expected number of trading dates.

    Parameters
    ----------
    config : Dict[str, Any]
        Study configuration mapping; only
        ``P11_ACQUISITION_PARAMETERS.trading_date_count`` is read.

    Returns
    -------
    int
        The configured count when it parses to a strictly positive
        integer, else the pinned ``DEFAULT_TRADING_DATE_COUNT`` (727).
    """
    # Read the P11 acquisition section defensively.
    p11 = _config_section(config, "P11_ACQUISITION_PARAMETERS")
    # Read the pinned trading-date count; coerce defensively.
    count = p11.get("trading_date_count", None)
    # Coerce integral values; fall back to the pinned default otherwise.
    try:
        # Accept both ints and integral floats/strings.
        coerced = int(count) if count is not None else None
    except (TypeError, ValueError):
        # Unparseable values fall back to the pinned default.
        coerced = None
    # Return the coerced value when sane; else the pinned study constant.
    return coerced if coerced is not None and coerced > 0 else DEFAULT_TRADING_DATE_COUNT

def _column_dtype_conforms(series: pd.Series, declared: str) -> bool:
    """
    Test a column dtype against a declared schema dtype string.

    Comparison is exact: nullable pandas families and numpy families
    are never interchanged, so a numpy ``int64`` column fails an
    ``"Int64"`` declaration and vice versa.

    Parameters
    ----------
    series : pd.Series
        Column whose ``dtype`` is tested; any index is accepted.
    declared : str
        Declared dtype string, read verbatim from the configuration.

    Returns
    -------
    bool
        True when the concrete dtype conforms to ``declared``; False
        on mismatch or on a declaration outside the table below.

    Notes
    -----
    Recognised declarations: ``"string"`` (any ``pd.StringDtype``,
    python or pyarrow backed), ``"Int64"`` (``pd.Int64Dtype``),
    ``"int64"``, ``"float64"``, ``"boolean"``, ``"bool"``,
    ``"datetime64[ns]"`` (numpy, timezone-naive) and
    ``"datetime64[ns, <Tz>]"`` matched on the timezone name string.
    """
    # Read the actual dtype once for the comparison branches.
    actual = series.dtype
    # string family: any pandas StringDtype conforms, pyarrow or python backed.
    if declared == "string":
        # isinstance covers both "string" and "string[pyarrow]".
        return isinstance(actual, pd.StringDtype)
    # Nullable integer family.
    if declared == "Int64":
        # Exact pandas nullable Int64.
        return actual == pd.Int64Dtype()
    # Native integer family.
    if declared == "int64":
        # Exact numpy int64; numpy defaults are rejected for Int64 declarations.
        return actual == np.dtype("int64")
    # Native float family.
    if declared == "float64":
        # Exact numpy float64.
        return actual == np.dtype("float64")
    # Nullable boolean family.
    if declared == "boolean":
        # Exact pandas nullable boolean.
        return actual == pd.BooleanDtype()
    # Native boolean family.
    if declared == "bool":
        # Exact numpy bool.
        return actual == np.dtype("bool")
    # Naive nanosecond datetimes.
    if declared == "datetime64[ns]":
        # Exact numpy datetime64[ns]; tz-aware and string dtypes are rejected.
        return actual == np.dtype("datetime64[ns]")
    # Timezone-aware nanosecond datetimes: parse the declared tz suffix.
    if isinstance(declared, str) and declared.startswith("datetime64[ns,"):
        # Extract the declared timezone name between the brackets.
        tz_name = declared[len("datetime64[ns,") :].rstrip("]").strip()
        # Validate tz-aware dtype presence and timezone-name equality.
        return isinstance(actual, pd.DatetimeTZDtype) and str(actual.tz) == tz_name
    # Unknown declarations never conform.
    return False

def _localise_naive_retrieved(
    files: pd.DataFrame,
    records: List[Dict[str, Any]],
) -> pd.DataFrame:
    """
    Coerce the ``retrieved`` timestamp column to Asia/Shanghai.

    Parameters
    ----------
    files : pd.DataFrame
        Manifest working copy, mutated in place on coercion.
    records : List[Dict[str, Any]]
        In-flight report buffer; the boundary coercion is recorded
        here as a ``WARN`` record, never applied silently.

    Returns
    -------
    pd.DataFrame
        The same frame. ``retrieved`` is localised to
        ``datetime64[ns, Asia/Shanghai]`` when it arrived
        timezone-naive (``FILES.retrieved.tz_coercion``), converted
        to that zone when it arrived in another timezone
        (``FILES.retrieved.tz_conversion``), and otherwise untouched.
    """
    # Only the retrieved column carries a tz-aware declaration in this schema.
    if "retrieved" not in files.columns:
        # Nothing to coerce; leave the frame untouched.
        return files
    # Read the actual dtype for the branch decision.
    dtype = files["retrieved"].dtype
    # Naive datetime64[ns] requires boundary localisation with a recorded diagnostic.
    if dtype == np.dtype("datetime64[ns]"):
        # Localise to Asia/Shanghai, the exchange-local convention of the corpus.
        files["retrieved"] = files["retrieved"].dt.tz_localize("Asia/Shanghai")
        # Record the coercion in the diagnostic, never silently.
        _append_report_record(
            records,
            "FILES.retrieved.tz_coercion",
            "WARN",
            "retrieved arrived naïve; coerced to datetime64[ns, Asia/Shanghai] at the boundary and recorded.",
            [],
        )
        # Return the coerced frame.
        return files
    # A non-Shanghai tz-aware stamp is converted with a recorded diagnostic.
    if isinstance(dtype, pd.DatetimeTZDtype) and str(dtype.tz) != "Asia/Shanghai":
        # Convert to the pinned timezone, preserving instants.
        files["retrieved"] = files["retrieved"].dt.tz_convert("Asia/Shanghai")
        # Record the conversion in the diagnostic.
        _append_report_record(
            records,
            "FILES.retrieved.tz_conversion",
            "WARN",
            f"retrieved arrived tz-aware as {str(dtype.tz)}; converted to Asia/Shanghai at the boundary and recorded.",
            [],
        )
        # Return the converted frame.
        return files
    # Already conformant; nothing to do.
    return files

def _add_four_sanctioned_columns(
    files: pd.DataFrame,
    config: Dict[str, Any],
    records: List[Dict[str, Any]],
) -> pd.DataFrame:
    """
    Add the four sanctioned columns required before validation.

    The added columns are:

    * ``file_uri`` — POSIX-relative content-addressed path
      ``data/raw/<byte_hash>``, falling back to
      ``data/raw/<file_id>`` when ``byte_hash`` is missing or null.
    * ``tag_registry_version`` — ``tag-registry-sha256:<16 hex>``,
      the truncated SHA-256 of the canonical-JSON form of
      ``{"provenance_tags": vocabulary}``. The vocabulary is the
      sorted set of stringified ``P00_META.provenance_tags`` entries
      only when that value is a non-empty list or tuple; an absent,
      empty or non-list/tuple value falls back to
      ``DEFAULT_TAG_VOCABULARY``.
    * ``content_length_ok`` — nullable boolean,
      ``content_length_bytes.notna() & (content_length_bytes > 0)``;
      all-NA (plus a ``WARN`` record) when the length column is
      absent.
    * ``response_truncated`` — the input column cast to
      ``"boolean"`` without altering values; all-NA (plus a ``WARN``
      record) when absent.

    Parameters
    ----------
    files : pd.DataFrame
        Manifest working copy, modified in place; must carry
        ``file_id``.
    config : Dict[str, Any]
        Study configuration mapping (read for ``P00_META``).
    records : List[Dict[str, Any]]
        In-flight report buffer receiving the plumbing ``WARN``
        notes.

    Returns
    -------
    pd.DataFrame
        The same frame with all four columns present.
    """
    # Derive the string dtype spelling once for the new textual columns.
    string_dtype = _string_backend_dtype()
    # Build file_uri from the byte hash (content-addressed) when available,
    # else from file_id; never fabricate a hash.
    if "byte_hash" in files.columns:
        # Use the byte hash as the stable handle (survives pipeline re-entry).
        hash_series = files["byte_hash"].astype("string")
        # Content-addressed POSIX-relative path rooted at the conventional raw-data directory.
        file_uri = "data/raw/" + hash_series.fillna(files["file_id"].astype("string"))
    else:
        # Without byte_hash the file_id is the only handle available.
        file_uri = "data/raw/" + files["file_id"].astype("string")
    # Cast the URI column to the pinned string dtype.
    files["file_uri"] = file_uri.astype(string_dtype)
    # Derive the registry fingerprint from the declared tag vocabulary when present.
    p00 = _config_section(config, "P00_META")
    # Read the declared provenance-tag enumeration; fall back to the pinned vocabulary.
    declared_tags = p00.get("provenance_tags", None)
    # Normalise the declared vocabulary to a sorted unique list of strings.
    if isinstance(declared_tags, (list, tuple)) and declared_tags:
        # Base the fingerprint on the config's own declared vocabulary.
        vocabulary = sorted({str(tag) for tag in declared_tags})
    else:
        # The pinned closed enumeration is the fallback basis.
        vocabulary = sorted(DEFAULT_TAG_VOCABULARY)
    # Hash the canonical serialisation of the vocabulary for a deterministic fingerprint.
    registry_digest = _sha256_hex(_canonical_json_bytes({"provenance_tags": vocabulary}))
    # Assign the derived fingerprint as a constant string column.
    files["tag_registry_version"] = pd.Series(
        ["tag-registry-sha256:" + registry_digest[:16]] * len(files),
        index=files.index,
        dtype=string_dtype,
    )
    # Plumb content_length_ok from content_length_bytes when available.
    if "content_length_bytes" in files.columns:
        # Presence/positivity gate; byte-level truth is Task 3's responsibility.
        content_ok = files["content_length_bytes"].notna() & (files["content_length_bytes"] > 0)
        # Cast to the pinned nullable boolean dtype.
        files["content_length_ok"] = content_ok.astype("boolean")
    else:
        # Plumb an all-NA boolean column and record the deferral, never fabricate.
        files["content_length_ok"] = pd.Series(
            [pd.NA] * len(files), index=files.index, dtype="boolean"
        )
        # Record the plumbing note in the diagnostic.
        _append_report_record(
            records,
            "FILES.content_length_ok.plumbed",
            "WARN",
            "content_length_bytes absent at ingest; content_length_ok plumbed as all-NA boolean, not fabricated.",
            [],
        )
    # Preserve response_truncated verbatim when present, coerce to pandas boolean.
    if "response_truncated" in files.columns:
        # Coerce without altering values; NA stays NA (never silently coerced to False).
        files["response_truncated"] = files["response_truncated"].astype("boolean")
    else:
        # Plumb an all-NA boolean column and record the deferral, never fabricate.
        files["response_truncated"] = pd.Series(
            [pd.NA] * len(files), index=files.index, dtype="boolean"
        )
        # Record the plumbing note in the diagnostic.
        _append_report_record(
            records,
            "FILES.response_truncated.plumbed",
            "WARN",
            "response_truncated plumbed as all-NA boolean at ingest; populated only by a witnessed retrieval.",
            [],
        )
    # Return the augmented manifest.
    return files

def _session_span_count(
    retrieved: pd.Series,
    calendar_trade_dates: pd.Series,
) -> Tuple[Optional[int], Optional[pd.Timestamp], Optional[pd.Timestamp]]:
    """
    Convert a retrieval interval into a session-span count.

    Nulls are dropped first; the witnessed extremes are converted to
    naive Shanghai wall time, and each is mapped to the zero-based
    position of the last session with ``trade_date <= stamp``
    (``searchsorted`` with side ``"right"`` minus one, clamped at
    zero). The span is ``pos(d_max) - pos(d_min)``: 0 means both
    extremes fall in one session, 1 means adjacent sessions, and a
    value above 1 breaches the "at most one trading session" check.

    Parameters
    ----------
    retrieved : pd.Series
        Per-file completion timestamps, timezone-aware (converted to
        ``Asia/Shanghai``) or naive (localised to that zone here).
    calendar_trade_dates : pd.Series
        Trading-calendar date column; nulls are dropped and the
        result sorted and de-duplicated internally.

    Returns
    -------
    span_sessions : Optional[int]
        Session span as defined above; ``None`` when no witnessed
        timestamp exists.
    naive_min : Optional[pd.Timestamp]
        Earliest witnessed stamp as naive Shanghai wall time;
        ``None`` when unevaluable.
    naive_max : Optional[pd.Timestamp]
        Latest witnessed stamp as naive Shanghai wall time;
        ``None`` when unevaluable.
    """
    # Drop nulls; coherence is evaluated on the witnessed subset only.
    observed = retrieved.dropna()
    # Without any witnessed stamp the interval is unevaluable.
    if observed.empty:
        # Signal unevaluable to the caller, which records a WARN record.
        return None, None, None
    # Reduce to naïve Shanghai wall time for calendar arithmetic.
    if isinstance(observed.dtype, pd.DatetimeTZDtype):
        # Convert to Asia/Shanghai first, then drop the zone for the comparison.
        naive = observed.dt.tz_convert("Asia/Shanghai").dt.tz_localize(None)
    else:
        # Defensive localisation for any residual naïve stamps.
        naive = observed.dt.tz_localize("Asia/Shanghai").dt.tz_localize(None)
    # Sort and de-duplicate the calendar dates; retain naïve datetime64[ns] values.
    cal64 = pd.to_datetime(calendar_trade_dates.dropna()).sort_values().unique()
    # Compare against a nanosecond-precision array via searchsorted side="right".
    def _position(stamp: pd.Timestamp) -> int:
        """
        Return the zero-based calendar position of a timestamp.

        Counts sessions with ``trade_date <= stamp`` and subtracts
        one.

        Parameters
        ----------
        stamp : pd.Timestamp
            Naive Shanghai wall-clock timestamp.

        Returns
        -------
        int
            Zero-based position on the sorted, de-duplicated
            calendar, clamped to 0 so a pre-origin stamp cannot make
            the span base negative.
        """
        # Count sessions with trade_date <= stamp; zero-based position is that count minus one.
        position = int(np.searchsorted(cal64, np.datetime64(stamp), side="right")) - 1
        # Clamp to the calendar's origin so a pre-origin stamp cannot produce a negative span base.
        return max(position, 0)

    # Compute the extremes of the witnessed subset.
    naive_min, naive_max = naive.min(), naive.max()
    # Compute the positions of the two extremes on the calendar.
    pos_min, pos_max = _position(naive_min), _position(naive_max)
    # Return the span together with the witnessed extremes for the message.
    return pos_max - pos_min, naive_min, naive_max

def validate_files_task_1(
    FILES: pd.DataFrame,
    STUDY_CONFIG: Dict[str, Any],
    TRADING_CALENDAR: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame, Optional[str]]:
    """
    Validate the ``FILES`` acquisition manifest.

    Deep-copy ``FILES``, re-key on ``file_id``, add four columns, then
    apply ten sub-checks in order with halt short-circuit. Report rows
    go into a write-once report keyed on ``check_id``; the augmented
    manifest, that report and a halt reason are returned. The inputs
    are never mutated in place.

    Sub-checks, in order:

    1. row count equals ``raw_input_schemas.FILES.row_count`` (80).
    2. ``file_id`` is non-null and unique.
    3. no undeclared columns beyond the allow-list; every declared
       column is present and dtype-conformant.
    4. ``adjustment`` is non-null and enumerates exactly the declared
       views with equal counts.
    5. ``stock_code`` is non-null and its nunique equals the
       expected universe size.
    6. ``stock_rank`` is non-null and bijective over ``1..universe``;
       each stock owns one rank and exactly one row per view.
    7. per-file ``row_count`` equals ``rows_per_bar_file`` (727).
    8. twelve pinned ``attrs`` keys are present,
       ``attrs['config_self_hash']`` is a 64-character lowercase
       hexadecimal string, and eight exact-value cross-checks close;
       a malformed hash halts here.
    9. ``attrs["config_self_hash"]`` equals the sha256 of the
       canonical JSON of the whole ``STUDY_CONFIG``; the
       ``_provenance_tags`` sub-dict is included.
    10. retrieval coherence: the session-span count of
        ``max(retrieved) - min(retrieved)`` on the Shanghai
        calendar; a span above one session emits a
        ``PROVENANCE_LIMIT`` warning and never halts, and a ``None``
        span (no non-null ``retrieved`` stamps) emits a warning that
        the assertion is unevaluable.

    Parameters
    ----------
    FILES : pd.DataFrame
        Acquisition manifest indexed on ``file_id`` or carrying a
        ``file_id`` column; ``attrs`` carry the twelve keys.
    STUDY_CONFIG : Dict[str, Any]
        Nested configuration including the parallel
        ``_provenance_tags`` sub-dict and the
        ``permitted_extra_columns`` allow-list.
    TRADING_CALENDAR : pd.DataFrame
        Trading calendar carrying a naive ``trade_date`` column
        used to count session spans.

    Returns
    -------
    files_with_additions : pd.DataFrame
        Copy of ``FILES`` indexed on ``file_id`` (index name
        ``file_id``), carrying ``file_uri``,
        ``tag_registry_version``, ``content_length_ok`` and
        ``response_truncated``.
    validation_report : pd.DataFrame
        Diagnostic indexed on ``check_id`` with columns
        ``severity``, ``message`` and ``offending_row_ids``.
    halt_reason : Optional[str]
        ``None`` when all sub-checks pass; otherwise the halt
        message, also recorded as the final ``ERROR`` row.

    Raises
    ------
    TypeError
        If ``FILES``, ``STUDY_CONFIG`` or
        ``TRADING_CALENDAR`` is not of the expected type.
    ValueError
        If ``TRADING_CALENDAR`` lacks a ``trade_date`` column.

    Notes
    -----
    Sub-checks 1-9 halt on failure; sub-check 10 only warns. The
    config digest serialises the semantic dict and the parallel
    ``_provenance_tags`` sub-dict with sorted keys, ``NaN`` ->
    ``"NaN"``, ``pd.NA`` -> ``"__NA__"``, ISO-8601 datetimes, sorted
    sets and list tuples, then hashes the UTF-8 bytes with sha256.
    """
    # Validate argument types before any work; wrong types are structural defects.
    if not isinstance(FILES, pd.DataFrame):
        # FILES must be a DataFrame.
        raise TypeError(f"FILES must be a pandas.DataFrame, got {type(FILES)!r}.")
    # STUDY_CONFIG must be the nested configuration dictionary.
    if not isinstance(STUDY_CONFIG, dict):
        # Reject non-dict configurations.
        raise TypeError(f"STUDY_CONFIG must be a dict, got {type(STUDY_CONFIG)!r}.")
    # TRADING_CALENDAR must be a DataFrame carrying the calendar.
    if not isinstance(TRADING_CALENDAR, pd.DataFrame):
        # Reject non-DataFrame calendars.
        raise TypeError(f"TRADING_CALENDAR must be a pandas.DataFrame, got {type(TRADING_CALENDAR)!r}.")
    # The calendar must carry the date column downstream checks rely on.
    if "trade_date" not in TRADING_CALENDAR.columns:
        # Fail loudly with the actionable remedy.
        raise ValueError("TRADING_CALENDAR must carry a 'trade_date' column (naïve datetime64[ns]).")

    # Deep copy so the caller's frame is never mutated.
    files = FILES.copy(deep=True)
    # Initialise the write-once record buffer.
    records: List[Dict[str, Any]] = []

    # Restore the flat file_id column when only the index carries it, then (re)index.
    if "file_id" not in files.columns:
        # The index may hold the key (the fixture indexes on file_id with drop=False,
        # but a drop=True variant must be handled without losing the key).
        if files.index.name == "file_id":
            # Flatten the index back to a column; restores the key either way.
            files = files.reset_index()
        # Without any file_id anywhere the manifest is unkeyed: halt, do not guess.
        if "file_id" not in files.columns:
            # Append the structural defect and halt.
            halt = _record_and_halt(
                records,
                "FILES.structure",
                "FILES carries neither a 'file_id' column nor a 'file_id'-named index; "
                "the stable string primary key is mandatory.",
                [],
            )
            # Return the working frame, the diagnostic and the halt reason.
            return files, _finalize_validation_report(records), halt
    # Key the manifest on file_id with the key retained as a column.
    files = files.set_index("file_id", drop=False)
    # Pin the index name explicitly.
    files.index.name = "file_id"

    # Extract the declared schema and the pinned section projections.
    schema = _files_schema(STUDY_CONFIG)
    # P11 carries the acquisition pins (views, trading_date_count, rows_per_bar_file).
    p11 = _config_section(STUDY_CONFIG, "P11_ACQUISITION_PARAMETERS")
    # P12 carries the mask pin (masked_cell_count).
    p12 = _config_section(STUDY_CONFIG, "P12_MASKING_PARAMETERS")
    # P43 carries the governance pin (redistribution_grant).
    p43 = _config_section(STUDY_CONFIG, "P43_GOVERNANCE_PARAMETERS")

    # Resolve the declared row count (default 80) and the expected universe size.
    expected_file_count = int(schema.get("row_count", DEFAULT_FILE_COUNT))
    # Resolve the expected universe size from P11 (default derivation: row_count // 2).
    expected_universe = _expected_universe_size(STUDY_CONFIG, expected_file_count)
    # Resolve the expected trading-date count (P11 trading_date_count, else 727).
    trading_date_count = _expected_trading_date_count(STUDY_CONFIG)
    # Resolve the per-file row count (P11 rows_per_bar_file, else 727).
    rows_per_bar_file = int(p11.get("rows_per_bar_file", DEFAULT_ROWS_PER_BAR_FILE) or DEFAULT_ROWS_PER_BAR_FILE)
    # Resolve the expected masked-cell count (P12 masked_cell_count, else 2).
    masked_cell_count = int(p12.get("masked_cell_count", DEFAULT_MASKED_CELL_COUNT) or DEFAULT_MASKED_CELL_COUNT)
    # Resolve the declared provider and version pins (P11, else the study literals).
    expected_provider = str(p11.get("provider", DEFAULT_PROVIDER))
    # Provider-version pin, e.g. "0.9.3".
    expected_provider_version = str(p11.get("provider_version", DEFAULT_PROVIDER_VERSION))
    # Resolve the two adjustment views (P11 adjustment_views, else the pinned pair).
    views = p11.get("adjustment_views", None)
    # Accept only a non-empty list of strings; otherwise default to the pinned pair.
    views = [str(v) for v in views if isinstance(v, str)] if isinstance(views, list) and views else ["unadjusted", "provider_adjusted"]
    # Resolve the governance pin for the redistribution_grant exact-value cross-check (P43).
    expected_redistribution = bool(p43.get("redistribution_grant", False))
    # Extract the declared column schema mapping.
    declared_columns = schema.get("columns", {})
    # Tolerate a malformed columns mapping by degrading to an empty dict.
    declared_columns = declared_columns if isinstance(declared_columns, dict) else {}

    # Add the four sanctioned columns before any sub-check runs (step text mandate).
    files = _add_four_sanctioned_columns(files, STUDY_CONFIG, records)
    # Coerce naïve retrieved stamps at the boundary, recording the coercion.
    files = _localise_naive_retrieved(files, records)

    # ------------------------------------------------------------------
    # Sub-check 1: row-count (declared manifest size; 81 -> halt).
    # ------------------------------------------------------------------
    # Compare observed rows against the declared count; any mismatch halts.
    if len(files) != expected_file_count:
        # Record and halt; the 81-row acquisition manifest is the named halt case.
        halt = _record_and_halt(
            records,
            "FILES.row_count",
            f"FILES carries {len(files)} rows; the declared acquisition manifest is "
            f"{expected_file_count} (raw_input_schemas.FILES.row_count). "
            "An 81-row manifest is the pinned halt case.",
            list(files.index[:10]),
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.row_count", "INFO",
        f"FILES carries {len(files)} rows == declared {expected_file_count}.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 2: file_id uniqueness and non-nullness.
    # ------------------------------------------------------------------
    # Detect null file_ids; the key is the stable string primary key.
    null_mask = files["file_id"].isna()
    # Any null halts; offending positions are reported as index coordinates.
    if bool(null_mask.any()):
        # Gather the positional coordinates of the null keys.
        offending = [f"row {i}" for i in range(len(files)) if bool(null_mask.iat[i])][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.file_id_non_null",
            "file_id carries nulls; the stable string primary key must never be null.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Detect duplicated keys; duplicates destroy the primary-key guarantee.
    duplicated_mask = files.index.duplicated(keep=False)
    # Any duplication halts with the offending keys listed.
    if bool(duplicated_mask.any()):
        # Gather the duplicated key values for the diagnostic.
        offending = sorted({str(key) for key in files.index[duplicated_mask]})[:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.file_id_unique",
            f"file_id is not unique; duplicated keys: {offending}.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.file_id", "INFO",
        f"file_id is non-null and unique across {len(files)} rows.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 3: extras halt, then column-presence and dtype conformance.
    # ------------------------------------------------------------------
    # The allow-list is the only mechanism by which undeclared columns may enter.
    allowed_columns = set(declared_columns) | set(ADDED_FILES_COLUMNS) | _permitted_extra_columns(STUDY_CONFIG)
    # Compute undeclared extras; any presence halts.
    undeclared = sorted(set(files.columns) - allowed_columns)
    # Halt on any undeclared column that is not in the allow-list.
    if undeclared:
        # Record and halt; the allow-list is the only admission mechanism.
        halt = _record_and_halt(
            records,
            "FILES.columns.extra",
            f"Undeclared extra column(s) {undeclared} are neither in raw_input_schemas.FILES.columns "
            f"nor in permitted_extra_columns; the allow-list is the only mechanism by which extras enter.",
            undeclared,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Verify every declared column is present and conforms to its declared dtype.
    for column, declared in declared_columns.items():
        # Presence first; an absent declared column halts.
        if column not in files.columns:
            # Record and halt with the missing column named.
            halt = _record_and_halt(
                records,
                "FILES.columns.presence",
                f"Declared column {column!r} is absent from FILES.",
                [str(column)],
            )
            # Return early with the halt reason.
            return files, _finalize_validation_report(records), halt
        # Dtype conformance second; a mismatch halts with both spellings named.
        if not _column_dtype_conforms(files[column], str(declared)):
            # Record and halt with the expected and found dtypes named.
            halt = _record_and_halt(
                records,
                "FILES.columns.dtype",
                f"Column {column!r} declares dtype {declared!r} but carries {files[column].dtype!s}.",
                [f"{column}:{files[column].dtype}"],
            )
            # Return early with the halt reason.
            return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.columns", "INFO",
        f"All {len(declared_columns)} declared columns are present and dtype-conformant; "
        f"no undeclared extras beyond the allow-list.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 4: adjustment enumeration exactly {unadjusted: 40, provider_adjusted: 40}.
    # ------------------------------------------------------------------
    # Compute the observed per-view counts (dropna-free; nulls counted explicitly).
    adjustment_notna = files["adjustment"].notna()
    # Null adjustment values would silently break the enumeration; halt on any.
    if not bool(adjustment_notna.all()):
        # Gather the offending row coordinates.
        offending = [str(key) for key in files.index[~adjustment_notna]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.adjustment_enumeration",
            "adjustment carries null values; the enumeration must be exactly "
            "{'unadjusted': 40, 'provider_adjusted': 40}.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Count occurrences per view.
    observed_counts = files["adjustment"].value_counts().to_dict()
    # Expected per-view count: declared file count divided across the pinned views.
    expected_per_view = expected_file_count // len(views)
    # Build the expected enumeration exactly as the step text pins it.
    expected_counts = {view: expected_per_view for view in views}
    # Compare exactly: same keys, same counts; anything else halts.
    if dict(observed_counts) != expected_counts:
        # Record and halt with both enumerations printed.
        halt = _record_and_halt(
            records,
            "FILES.adjustment_enumeration",
            f"adjustment enumeration is {observed_counts}, expected exactly {expected_counts}.",
            [f"{k}={v}" for k, v in sorted(observed_counts.items())],
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.adjustment_enumeration", "INFO",
        f"adjustment enumeration == {expected_counts}.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 5: stock_code nunique == 40.
    # ------------------------------------------------------------------
    # Null stock codes would silently break coverage; halt on any.
    if not bool(files["stock_code"].notna().all()):
        # Gather the offending row coordinates.
        offending = [str(key) for key in files.index[files["stock_code"].isna()]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.stock_code_nunique",
            "stock_code carries null values; the universe must be 40 distinct codes.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Count distinct stock codes.
    distinct_stocks = int(files["stock_code"].nunique())
    # Compare against the expected universe size.
    if distinct_stocks != expected_universe:
        # Record and halt with both figures.
        halt = _record_and_halt(
            records,
            "FILES.stock_code_nunique",
            f"stock_code nunique == {distinct_stocks}; expected exactly {expected_universe}.",
            [],
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.stock_code_nunique", "INFO",
        f"stock_code nunique == {distinct_stocks} == expected universe size.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 6: stock_rank bijection (non-null, [1..40], one stock per rank,
    # two rows per stock -- one per view).
    # ------------------------------------------------------------------
    # stock_rank is non-nullable when the acquisition is coherent; NA halts.
    rank_na = files["stock_rank"].isna()
    # Any null rank halts.
    if bool(rank_na.any()):
        # Gather the offending row coordinates.
        offending = [str(key) for key in files.index[rank_na]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.stock_rank",
            "stock_rank carries nulls; the rank is non-nullable when the acquisition is coherent.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # The distinct ranks must enumerate the dense [1..expected_universe] set;
    # each rank labels exactly one stock (two rows: one per view), so sorted unique
    # ranks are the pinned comparison -- a raw sort of 80 rows would be [1, 1, ...].
    sorted_ranks = sorted(int(rank) for rank in files["stock_rank"].dropna().unique())
    # Build the expected dense enumeration.
    expected_ranks = list(range(1, expected_universe + 1))
    # Compare elementwise; any deviation halts.
    if sorted_ranks != expected_ranks:
        # Record and halt with both enumerations.
        halt = _record_and_halt(
            records,
            "FILES.stock_rank",
            f"sorted(unique(stock_rank)) == {sorted_ranks[:50]}; expected {expected_ranks}.",
            [str(r) for r in sorted_ranks[:25]],
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Each rank must map to exactly one stock.
    rank_to_stock = files.groupby("stock_rank")["stock_code"].nunique()
    # Any rank carried by more than one stock halts.
    if bool((rank_to_stock != 1).any()):
        # Gather the offending ranks.
        offending = [str(rank) for rank in rank_to_stock.index[rank_to_stock != 1]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.stock_rank",
            f"Rank(s) {offending} map to more than one stock_code.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Each stock must carry exactly one rank.
    stock_to_rank = files.groupby("stock_code")["stock_rank"].nunique()
    # Any stock with more than one rank halts.
    if bool((stock_to_rank != 1).any()):
        # Gather the offending stocks.
        offending = [str(stock) for stock in stock_to_rank.index[stock_to_rank != 1]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.stock_rank",
            f"Stock(s) {offending} carry more than one stock_rank.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Each stock must contribute exactly len(views) rows, one per adjustment view.
    stock_rows = files.groupby("stock_code").size()
    # Any stock with a row count other than len(views) halts.
    if bool((stock_rows != len(views)).any()):
        # Gather the offending stocks.
        offending = [str(stock) for stock in stock_rows.index[stock_rows != len(views)]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.stock_rank",
            f"Stock(s) {offending} do not carry exactly {len(views)} rows (one per view).",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Each stock must cover every declared view exactly once.
    stock_views = files.groupby("stock_code")["adjustment"].nunique()
    # Any stock missing a view halts.
    if bool((stock_views != len(views)).any()):
        # Gather the offending stocks.
        offending = [str(stock) for stock in stock_views.index[stock_views != len(views)]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.stock_rank",
            f"Stock(s) {offending} do not cover all {len(views)} adjustment views exactly once.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.stock_rank", "INFO",
        f"stock_rank is non-null, enumerates {expected_ranks}, maps bijectively to {expected_universe} stocks, "
        f"and every stock contributes exactly one row per view ({len(views)} views).", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 7: per-file row_count == 727.
    # ------------------------------------------------------------------
    # Truthy on nulls as well: a null row_count is not equal to 727.
    bad_row_count = (files["row_count"] != rows_per_bar_file) | files["row_count"].isna()
    # Any mismatch halts.
    if bool(bad_row_count.any()):
        # Gather the offending files.
        offending = [str(key) for key in files.index[bad_row_count]][:25]
        # Record and halt.
        halt = _record_and_halt(
            records,
            "FILES.row_count_per_file",
            f"Per-file row_count != {rows_per_bar_file} for {int(bad_row_count.sum())} file(s): {offending}.",
            offending,
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.row_count_per_file", "INFO",
        f"Every per-file row_count == {rows_per_bar_file}.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 8: attrs twelve pinned keys + eight exact-value cross-checks.
    # ------------------------------------------------------------------
    # Read the attribute dictionary; attrs must never be coerced to a column.
    attrs = files.attrs
    # Presence of the twelve pinned keys first.
    missing_attrs = [key for key in PINNED_ATTR_KEYS if key not in attrs]
    # Any missing key halts.
    if missing_attrs:
        # Record and halt with the missing keys named.
        halt = _record_and_halt(
            records,
            "FILES.attrs.presence",
            f"FILES.attrs is missing pinned key(s) {missing_attrs}; the twelve keys are mandatory.",
            [str(key) for key in missing_attrs],
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Presence of a well-formed config_self_hash is checked structurally here; equality in sub-check 9.
    delivered_hash = attrs.get("config_self_hash")
    # Hex-shape gate: 64 lowercase hex characters.
    if not isinstance(delivered_hash, str) or len(delivered_hash) != SHA256_HEX_LENGTH or any(
        ch not in "0123456789abcdef" for ch in delivered_hash
    ):
        # Record and halt with the offending value named.
        halt = _record_and_halt(
            records,
            "FILES.attrs.cross_checks",
            f"attrs['config_self_hash'] is malformed (expected 64 lowercase hex characters); got {delivered_hash!r}.",
            ["config_self_hash"],
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Accumulate the exact-value cross-checks; each must close to the pinned value.
    cross_checks: List[Tuple[str, Any, Any]] = [
        ("file_count", attrs["file_count"], expected_file_count),
        ("date_count", attrs["date_count"], trading_date_count),
        ("masked_cell_count", attrs["masked_cell_count"], masked_cell_count),
        ("provider", attrs["provider"], expected_provider),
        ("provider_version", attrs["provider_version"], expected_provider_version),
        ("redistribution_grant", attrs["redistribution_grant"], expected_redistribution),
        ("universe_order_fixed", attrs["universe_order_fixed"], True),
        ("universe_disclosed", attrs["universe_disclosed"], False),
    ]
    # Collect failures rather than halting on the first, so the diagnostic is complete.
    cross_failures: List[str] = []
    # Evaluate each pinned cross-check with equality (numpy scalars compare equal cleanly).
    for key, observed, expected in cross_checks:
        # Compare observed against expected; any mismatch is recorded.
        if not (observed == expected):
            # Collect the failure with both values named.
            cross_failures.append(f"{key}={observed!r} (expected {expected!r})")
    # Any cross-check failure halts with the full list in the message.
    if cross_failures:
        # Record and halt with every failure enumerated.
        halt = _record_and_halt(
            records,
            "FILES.attrs.cross_checks",
            f"FILES.attrs exact-value cross-check failure(s): {'; '.join(cross_failures)}.",
            [failure.split("=")[0] for failure in cross_failures],
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.attrs.presence", "INFO",
        "All twelve pinned attrs keys are present.", [],
    )
    # Record the passing cross-checks.
    _append_report_record(
        records, "FILES.attrs.cross_checks", "INFO",
        "Eight exact-value attrs cross-checks close to their pinned values.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 9: config self-hash equality.
    # ------------------------------------------------------------------
    # Compute the digest of the whole STUDY_CONFIG (semantic dict plus _provenance_tags).
    computed_hash = _sha256_hex(_canonical_json_bytes(STUDY_CONFIG))
    # Compare against the delivered attrs digest; mismatch halts.
    if computed_hash != delivered_hash:
        # Record and halt with both digests named.
        halt = _record_and_halt(
            records,
            "FILES.config_self_hash",
            f"config_self_hash mismatch: attrs carry {delivered_hash!r}, computed digest is {computed_hash!r} "
            "(coverage: canonical JSON of the whole STUDY_CONFIG including _provenance_tags).",
            [f"attrs={delivered_hash}", f"computed={computed_hash}"],
        )
        # Return early with the halt reason.
        return files, _finalize_validation_report(records), halt
    # Record the passing sub-check.
    _append_report_record(
        records, "FILES.config_self_hash", "INFO",
        "config self-hash equality holds (sha256 of the canonical JSON of the whole STUDY_CONFIG).", [],
    )

    # ------------------------------------------------------------------
    # Sub-check 10: retrieval coherence (disclosure, never abort).
    # ------------------------------------------------------------------
    # Convert the witnessed retrieval interval into a session-span count.
    span, naive_min, naive_max = _session_span_count(files["retrieved"], TRADING_CALENDAR["trade_date"])
    # Unevaluable intervals are disclosed, never halted.
    if span is None:
        # Record the unevaluable case and move on.
        _append_report_record(
            records, "FILES.retrieval_coherence", "WARN",
            "retrieved contains no non-null stamps; the session-span coherence assertion is unevaluable.", [],
        )
    else:
        # Compute the elapsed wall time for the diagnostic message.
        elapsed = naive_max - naive_min
        # A span above one session breaches the pinned assertion.
        if span > 1:
            # Identify the two extreme witnesses for the diagnostic.
            witness_min = str(files["retrieved"].idxmin())
            # Identify the latest witness.
            witness_max = str(files["retrieved"].idxmax())
            # Emit PROVENANCE_LIMIT; disclosure, not abort, per the paper's response.
            _append_report_record(
                records, "FILES.retrieval_coherence", "WARN",
                f"PROVENANCE_LIMIT: retrieval interval spans {span} trading session(s) "
                f"(elapsed {elapsed}); the pinned assertion is at most one trading session. "
                "Disclosure, not abort.",
                [witness_min, witness_max],
            )
        # Within one session: record the passing assertion.
        else:
            # Record the passing coherence check.
            _append_report_record(
                records, "FILES.retrieval_coherence", "INFO",
                f"Retrieval interval spans {span} trading session(s) (elapsed {elapsed}); assertion holds.", [],
            )

    # Materialise the diagnostic and return the augmented manifest untouched-otherwise.
    report = _finalize_validation_report(records)
    # Return FILES_with_additions, the report, and no halt reason.
    return files, report, None

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 2: UNADJUSTED_BARS and ADJUSTED_BARS panel validation
# -------------------------------------------------------------------------------------------------------------------------------

def _ensure_panel_layout(
    panel: pd.DataFrame,
) -> Optional[pd.DataFrame]:
    """
    Restore the compound ``(stock_code, trade_date)`` panel layout.

    Materialise ``stock_code`` and ``trade_date`` from whichever index
    level or flat index carries them, then re-key on the compound
    MultiIndex with ``drop=False`` so both remain columns. Applies
    ``sort_index()`` once and pins the level names. Returns ``None``
    when either handle is still absent after the restoration attempts.

    Parameters
    ----------
    panel : pd.DataFrame
        Working copy of one bars panel.

    Returns
    -------
    Optional[pd.DataFrame]
        Panel sorted by ``(stock_code, trade_date)`` with both levels
        retained as columns, or ``None`` if either ``stock_code`` or
        ``trade_date`` is absent from the columns once restoration
        has been attempted.
    """
    # Restore stock_code as a column when the index carries it and the columns do not.
    if "stock_code" not in panel.columns:
        # Case A: compound index with stock_code among its levels.
        if isinstance(panel.index, pd.MultiIndex) and "stock_code" in panel.index.names:
            # Restore only that level; the trade_date level is preserved.
            panel = panel.reset_index(level="stock_code")
        # Case B: flat index named stock_code.
        elif panel.index.name == "stock_code":
            # Restore the flat level to a column.
            panel = panel.reset_index()
    # Restore trade_date as a column when the index carries it and the columns do not.
    if "trade_date" not in panel.columns:
        # Case A: compound index with trade_date among its levels.
        if isinstance(panel.index, pd.MultiIndex) and "trade_date" in panel.index.names:
            # Restore only that level; stock_code is preserved or already restored.
            panel = panel.reset_index(level="trade_date")
        # Case B: flat index named trade_date.
        elif panel.index.name == "trade_date":
            # Restore the flat level to a column.
            panel = panel.reset_index()
    # Without both handles the panel is unkeyable; the caller must halt.
    if "stock_code" not in panel.columns or "trade_date" not in panel.columns:
        # Signal the unkeyable panel to the caller.
        return None
    # Re-key on the compound index with both levels retained as columns.
    if not (
        isinstance(panel.index, pd.MultiIndex)
        and list(panel.index.names) == ["stock_code", "trade_date"]
    ):
        # Build the compound MultiIndex early, keeping flat columns available.
        panel = panel.set_index(["stock_code", "trade_date"], drop=False)
    # Apply sort_index exactly once at the boundary.
    panel = panel.sort_index()
    # Pin the index level names (set_index already names them; enforce explicitly).
    panel.index.names = ["stock_code", "trade_date"]
    # Return the layout-conformant panel.
    return panel

def _ordered_sequence_equal(first: np.ndarray, second: np.ndarray) -> bool:
    """
    Test ordered elementwise equality of two arrays.

    Wraps ``numpy.testing.assert_array_equal`` and converts its
    silent pass or ``AssertionError`` into a boolean, so callers can
    record and halt rather than raise.

    Parameters
    ----------
    first : np.ndarray
        Reference order (``UNADJUSTED_BARS`` stock sequence).
    second : np.ndarray
        Candidate order (``ADJUSTED_BARS`` stock sequence).

    Returns
    -------
    bool
        True iff both arrays are elementwise identical in order.
    """
    # Attempt the ordered equality assertion provided by numpy.testing.
    try:
        # assert_array_equal raises AssertionError on any mismatch (order or value).
        np.testing.assert_array_equal(first, second)
        # No exception: the sequences match bit-for-behaviour.
        return True
    except AssertionError:
        # Any mismatch: the ordering is not reproduced.
        return False

def validate_panels_task_1(
    UNADJUSTED_BARS: pd.DataFrame,
    ADJUSTED_BARS: pd.DataFrame,
    FILES: pd.DataFrame,
    STUDY_CONFIG: Dict[str, Any],
    TRADING_CALENDAR: pd.DataFrame,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame, Optional[str]]:
    """
    Validate both Task 1 price panels and return augmented copies.

    Implements Task 1, Step 2. Deep-copies the panels and the manifest,
    keys each panel on ``MultiIndex[stock_code, trade_date]`` with one
    ``sort_index()``, tags them in ``attrs["__panel_source"]`` and runs
    eight sub-checks in order, appending failures to a write-once
    diagnostic: (i) declared columns present and dtype-conformant, plus
    adjustment-view purity; (ii) primary-key uniqueness (halts only when
    ``uniqueness_halt_assertion`` is truthy, else ``WARN``); (iii) exact
    pre-mask row count (default 29,080 = 40 * 727); (iv) stock coverage:
    identical ``stock_code`` sets, distinct count equal to the universe,
    identical ordered ``unique()`` sequences; (v) date coverage against
    the calendar; (vi) ``byte_hash`` provenance linkage into the
    matching ``FILES`` view; (vii) ``trade_date`` exactly
    ``datetime64[ns]`` in column and index level; (viii) role
    conformance of the ``_adj`` suffix. View labels, universe size and
    row bands come from ``STUDY_CONFIG``; panels are never reindexed to
    each other.

    Parameters
    ----------
    UNADJUSTED_BARS : pd.DataFrame
        Nominal panel carrying ``adjustment`` and ``byte_hash``.
    ADJUSTED_BARS : pd.DataFrame
        Provider-adjusted panel with the same requirements.
    FILES : pd.DataFrame
        Manifest carrying ``byte_hash`` and ``adjustment``; linkage
        is on ``byte_hash``, never ``file_id``.
    STUDY_CONFIG : Dict[str, Any]
        Configuration providing the panel schemas and ``P11``.
    TRADING_CALENDAR : pd.DataFrame
        Calendar defining the expected ``trade_date`` set.

    Returns
    -------
    unadjusted_out : pd.DataFrame
        Nominal panel keyed on ``(stock_code, trade_date)``.
    adjusted_out : pd.DataFrame
        Adjusted panel with the same key and tagging.
    validation_report : pd.DataFrame
        Diagnostic indexed on ``check_id``; columns ``severity``,
        ``message``, ``offending_row_ids``.
    halt_reason : Optional[str]
        ``None`` when every sub-check passed, else the reason.

    Raises
    ------
    TypeError
        If a frame argument is not a ``pd.DataFrame``, or
        ``STUDY_CONFIG`` is not a ``dict``.
    ValueError
        If ``TRADING_CALENDAR`` lacks ``trade_date``, or ``FILES``
        lacks ``byte_hash`` or ``adjustment``.
    """
    # Validate argument types before any work; wrong types are structural defects.
    if not isinstance(UNADJUSTED_BARS, pd.DataFrame):
        # UNADJUSTED_BARS must be a DataFrame.
        raise TypeError(f"UNADJUSTED_BARS must be a pandas.DataFrame, got {type(UNADJUSTED_BARS)!r}.")
    # ADJUSTED_BARS must be a DataFrame.
    if not isinstance(ADJUSTED_BARS, pd.DataFrame):
        # ADJUSTED_BARS must be a DataFrame.
        raise TypeError(f"ADJUSTED_BARS must be a pandas.DataFrame, got {type(ADJUSTED_BARS)!r}.")
    # FILES must be a DataFrame carrying the provenance handles.
    if not isinstance(FILES, pd.DataFrame):
        # Reject non-DataFrame manifests.
        raise TypeError(f"FILES must be a pandas.DataFrame, got {type(FILES)!r}.")
    # STUDY_CONFIG must be the nested configuration dictionary.
    if not isinstance(STUDY_CONFIG, dict):
        # Reject non-dict configurations.
        raise TypeError(f"STUDY_CONFIG must be a dict, got {type(STUDY_CONFIG)!r}.")
    # TRADING_CALENDAR must be a DataFrame carrying the calendar.
    if not isinstance(TRADING_CALENDAR, pd.DataFrame):
        # Reject non-DataFrame calendars.
        raise TypeError(f"TRADING_CALENDAR must be a pandas.DataFrame, got {type(TRADING_CALENDAR)!r}.")
    # The calendar must carry the date column downstream checks rely on.
    if "trade_date" not in TRADING_CALENDAR.columns:
        # Fail loudly with the actionable remedy.
        raise ValueError("TRADING_CALENDAR must carry a 'trade_date' column (naïve datetime64[ns]).")
    # FILES must carry the two provenance handles this step consumes.
    if "byte_hash" not in FILES.columns or "adjustment" not in FILES.columns:
        # Provenance cannot be evaluated without the handles; structural halt semantics.
        raise ValueError("FILES must carry 'byte_hash' and 'adjustment' columns for provenance linkage.")

    # Deep copies so the caller's frames are never mutated.
    una = UNADJUSTED_BARS.copy(deep=True)
    # Copy of the adjusted panel.
    adj = ADJUSTED_BARS.copy(deep=True)
    # Copy of the manifest used for linkage.
    files = FILES.copy(deep=True)
    # Initialise the write-once record buffer.
    records: List[Dict[str, Any]] = []

    # Extract the two panel schemas and the acquisition-view mapping.
    una_schema = _bars_schema(STUDY_CONFIG, "UNADJUSTED_BARS")
    # Adjusted-view schema.
    adj_schema = _bars_schema(STUDY_CONFIG, "ADJUSTED_BARS")
    # P11 carries the adjustment_views ordering (index 0 nominal, index 1 adjusted).
    p11 = _config_section(STUDY_CONFIG, "P11_ACQUISITION_PARAMETERS")
    # Resolve the pinned views; default to the pinned pair when absent or malformed.
    views = p11.get("adjustment_views", None)
    # Accept only a non-empty list of strings; otherwise default to the pinned pair.
    views = [str(v) for v in views if isinstance(v, str)] if isinstance(views, list) and len(views) >= 2 else ["unadjusted", "provider_adjusted"]
    # Map each panel to its declared adjustment view by position.
    una_view_label, adj_view_label = views[0], views[1]
    # Resolve the expected universe size from P11 (default 40).
    expected_universe = _expected_universe_size(STUDY_CONFIG, DEFAULT_FILE_COUNT)
    # Resolve the expected pre-mask band per panel from the schema, else the pin.
    una_expected_rows = int(una_schema.get("estimated_row_count", DEFAULT_PRE_MASK_ROW_COUNT) or DEFAULT_PRE_MASK_ROW_COUNT)
    # Adjusted-view expected band (pinned: 29,080 as well).
    adj_expected_rows = int(adj_schema.get("estimated_row_count", DEFAULT_PRE_MASK_ROW_COUNT) or DEFAULT_PRE_MASK_ROW_COUNT)
    # Resolve the calendar date set once for the date-coverage sub-check.
    calendar_dates = pd.to_datetime(TRADING_CALENDAR["trade_date"].dropna())
    # DatetimeIndex used for set comparison.
    calendar_set = set(calendar_dates.unique())

    # Key both panels on the compound index at the boundary, sort once each.
    una = _ensure_panel_layout(una)
    # Unkeyable nominal panel halts structurally.
    if una is None:
        # Record and halt.
        halt = _record_and_halt(
            records, "BARS.structure",
            "UNADJUSTED_BARS carries neither stock_code/trade_date columns nor index levels; the panel is unkeyable.",
            [],
        )
        # Return with the halt reason.
        return UNADJUSTED_BARS.copy(deep=True), ADJUSTED_BARS.copy(deep=True), _finalize_validation_report(records), halt
    # Key the adjusted panel the same way.
    adj = _ensure_panel_layout(adj)
    # Unkeyable adjusted panel halts structurally.
    if adj is None:
        # Record and halt.
        halt = _record_and_halt(
            records, "BARS.structure",
            "ADJUSTED_BARS carries neither stock_code/trade_date columns nor index levels; the panel is unkeyable.",
            [],
        )
        # Return with the halt reason.
        return una, ADJUSTED_BARS.copy(deep=True), _finalize_validation_report(records), halt
    # Tag the frames for the diagnostic; the tag is attrs, never a column.
    una.attrs["__panel_source"] = "UNADJUSTED_BARS"
    # Symmetric tag for the adjusted panel.
    adj.attrs["__panel_source"] = "ADJUSTED_BARS"

    # ------------------------------------------------------------------
    # Sub-check (i): column-presence and dtype conformance + view purity.
    # ------------------------------------------------------------------
    # Iterate over the two panels with their schemas in pinned order.
    for view_label, panel, schema in (("UNADJUSTED_BARS", una, una_schema), ("ADJUSTED_BARS", adj, adj_schema)):
        # Read the declared columns mapping defensively.
        declared_columns = schema.get("columns", {})
        # Tolerate a malformed mapping by degrading to empty.
        declared_columns = declared_columns if isinstance(declared_columns, dict) else {}
        # Undeclared-column policy is pinned for FILES only; panels are not subject to it here.
        for column, declared in declared_columns.items():
            # Presence first; an absent declared column halts.
            if column not in panel.columns:
                # Record and halt with the missing column named.
                halt = _record_and_halt(
                    records,
                    "BARS.columns.presence",
                    f"{view_label}: declared column {column!r} is absent.",
                    [str(column)],
                )
                # Return with the halt reason.
                return una, adj, _finalize_validation_report(records), halt
            # Dtype conformance second; a mismatch halts with both spellings named.
            if not _column_dtype_conforms(panel[column], str(declared)):
                # Record and halt with expected and found dtypes.
                halt = _record_and_halt(
                    records,
                    "BARS.columns.dtype",
                    f"{view_label}: column {column!r} declares dtype {declared!r} but carries {panel[column].dtype!s}.",
                    [f"{column}:{panel[column].dtype}"],
                )
                # Return with the halt reason.
                return una, adj, _finalize_validation_report(records), halt
        # Challenge-mandated adjustment-view purity: no cross-view concatenation.
        observed_views = set(panel["adjustment"].dropna().unique())
        # A null adjustment value is also a concatenation smell; treat as violation.
        if panel["adjustment"].isna().any() or observed_views != {una_view_label if view_label == "UNADJUSTED_BARS" else adj_view_label}:
            # The expected singleton set for this view.
            expected_singleton = {una_view_label if view_label == "UNADJUSTED_BARS" else adj_view_label}
            # Record and halt with both enumerations named.
            halt = _record_and_halt(
                records,
                "BARS.adjustment_view_purity",
                f"{view_label}: adjustment set is {observed_views}, expected exactly {expected_singleton} "
                "(mixed views indicate a concatenation error).",
                [f"{view_label}:{v}" for v in sorted(observed_views)],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
    # Record the passing sub-check (i).
    _append_report_record(
        records, "BARS.columns.presence", "INFO",
        "Both panels: declared columns present, dtypes conformant, adjustment-view purity holds.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check (ii): primary-key uniqueness pre-masking.
    # ------------------------------------------------------------------
    # Iterate panels in pinned order.
    for view_label, panel, schema in (("UNADJUSTED_BARS", una, una_schema), ("ADJUSTED_BARS", adj, adj_schema)):
        # Read the halt assertion flag; default True when absent or malformed.
        uniqueness_halt = bool(schema.get("uniqueness_halt_assertion", True))
        # Only a truthy flag halts; otherwise the violation is disclosed as a warning.
        duplicated_mask = panel.index.duplicated(keep=False)
        # Any duplication under a truthy flag halts.
        if bool(duplicated_mask.any()):
            # Gather up to 25 offending coordinates as "stock_code|trade_date".
            offending = [
                f"{sc}|{td.date() if hasattr(td, 'date') else td}"
                for sc, td in panel.index[duplicated_mask][:25]
            ]
            # Under a truthy flag: halt.
            if uniqueness_halt:
                # Record and halt with the offending coordinates named.
                halt = _record_and_halt(
                    records,
                    "BARS.uniqueness",
                    f"{view_label}: duplicate (stock_code, trade_date) primary key pre-masking "
                    f"({int(duplicated_mask.sum())} duplicated row(s)); uniqueness_halt_assertion is truthy.",
                    offending,
                )
                # Return with the halt reason.
                return una, adj, _finalize_validation_report(records), halt
            # Under a falsy flag: disclose, never halt.
            _append_report_record(
                records, "BARS.uniqueness", "WARN",
                f"{view_label}: {int(duplicated_mask.sum())} duplicated (stock_code, trade_date) row(s) "
                "recorded; uniqueness_halt_assertion is falsy.", offending,
            )
    # Record the passing sub-check (ii) when every panel was unique.
    _append_report_record(
        records, "BARS.uniqueness", "INFO",
        "Both panels: (stock_code, trade_date) primary key is unique pre-masking.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check (iii): row-count band (29,080 pre-mask only at Task 1).
    # ------------------------------------------------------------------
    # Iterate panels in pinned order with their expected bands.
    for view_label, panel, expected_rows in (
        ("UNADJUSTED_BARS", una, una_expected_rows),
        ("ADJUSTED_BARS", adj, adj_expected_rows),
    ):
        # The post-mask 29,078 band is not applicable at Task 1 (it belongs to Task 5).
        if len(panel) != expected_rows:
            # Record and halt with both figures named.
            halt = _record_and_halt(
                records,
                "BARS.row_count_band",
                f"{view_label}: carries {len(panel)} rows; the pre-mask band is exactly {expected_rows} "
                "(40 * 727). The 29,078 post-mask band is not applicable at Task 1.",
                [],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
    # Record the passing sub-check (iii).
    _append_report_record(
        records, "BARS.row_count_band", "INFO",
        f"Both panels carry exactly {una_expected_rows} pre-mask rows.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check (iv): stock coverage -- set equality and sequenced equality.
    # ------------------------------------------------------------------
    # Compute the distinct sequence of stock codes in each panel in index order.
    una_unique = una["stock_code"].unique()
    # Adjusted-view distinct sequence.
    adj_unique = adj["stock_code"].unique()
    # Set equality across the two views first.
    if set(una_unique) != set(adj_unique):
        # Record and halt with the symmetric difference named.
        halt = _record_and_halt(
            records,
            "BARS.stock_coverage",
            f"Panel stock sets differ: nominal-only {sorted(set(una_unique) - set(adj_unique))}, "
            f"adjusted-only {sorted(set(adj_unique) - set(una_unique))}.",
            sorted(set(una_unique) ^ set(adj_unique)),
        )
        # Return with the halt reason.
        return una, adj, _finalize_validation_report(records), halt
    # The distinct count must equal the pinned universe size.
    distinct_stocks = len(set(una_unique))
    # Any other count halts.
    if distinct_stocks != expected_universe:
        # Record and halt with both figures.
        halt = _record_and_halt(
            records,
            "BARS.stock_coverage",
            f"Panels carry {distinct_stocks} distinct stock_code values; expected exactly {expected_universe}.",
            [],
        )
        # Return with the halt reason.
        return una, adj, _finalize_validation_report(records), halt
    # Sequenced (ordered) equality across the two views; the challenge text sanctions
    # only panel-vs-panel here, deferring stock-code normalisation to Task 2.
    if not _ordered_sequence_equal(una_unique, adj_unique):
        # Record and halt with both orderings named.
        halt = _record_and_halt(
            records,
            "BARS.stock_coverage",
            "Ordered stock_code.unique() sequences differ between the two panels; the "
            "fixed 1..40 ordering must be reproduced bit-for-behaviour.",
            [],
        )
        # Return with the halt reason.
        return una, adj, _finalize_validation_report(records), halt
    # Record the passing sub-check (iv).
    _append_report_record(
        records, "BARS.stock_coverage", "INFO",
        f"Both panels: {distinct_stocks} distinct stocks, identical sets, identical ordered sequences. "
        "Panel-vs-manifest code normalisation is deferred to Task 2 by design.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check (v): date coverage -- identical 727-date set as the calendar.
    # ------------------------------------------------------------------
    # Iterate panels in pinned order.
    for view_label, panel in (("UNADJUSTED_BARS", una), ("ADJUSTED_BARS", adj)):
        # Compute the distinct date set of the panel.
        panel_dates = set(pd.to_datetime(panel["trade_date"]).unique())
        # Exact set equality with the calendar; survives any stock-code state.
        if panel_dates != calendar_set:
            # Record and halt with the symmetric difference counts named.
            halt = _record_and_halt(
                records,
                "BARS.date_coverage",
                f"{view_label}: trade_date set differs from the calendar "
                f"(calendar {len(calendar_set)} dates; panel {len(panel_dates)} dates).",
                sorted(str(d) for d in (panel_dates ^ calendar_set))[:25],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
    # Record the passing sub-check (v).
    _append_report_record(
        records, "BARS.date_coverage", "INFO",
        f"Both panels carry exactly the {len(calendar_set)} distinct trade_date values of the calendar.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check (vi): provenance linkage on byte_hash, per adjustment view.
    # ------------------------------------------------------------------
    # Compute the reference hash sets per view from the manifest.
    unadjusted_hashes = set(
        files.loc[files["adjustment"] == una_view_label, "byte_hash"].dropna().astype(str)
    )
    # Adjusted-view reference hash set.
    adjusted_hashes = set(
        files.loc[files["adjustment"] == adj_view_label, "byte_hash"].dropna().astype(str)
    )
    # Link each panel against its own view's reference set.
    for view_label, panel, reference in (
        ("UNADJUSTED_BARS", una, unadjusted_hashes),
        ("ADJUSTED_BARS", adj, adjusted_hashes),
    ):
        # Compute the panel's hash set.
        panel_hashes = set(panel["byte_hash"].dropna().astype(str))
        # Orphan hashes are those not present in the view-matched manifest subset.
        orphans = sorted(panel_hashes - reference)
        # Any orphan halts.
        if orphans:
            # Record and halt with a bounded witness list.
            halt = _record_and_halt(
                records,
                "BARS.provenance_linkage",
                f"{view_label}: {len(orphans)} byte_hash value(s) do not link to an FILES row with "
                f"adjustment == {una_view_label if view_label == 'UNADJUSTED_BARS' else adj_view_label!r}; "
                "linkage is on byte_hash, never on file_id.",
                orphans[:10],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
    # Record the passing sub-check (vi).
    _append_report_record(
        records, "BARS.provenance_linkage", "INFO",
        "Every panel byte_hash links to an FILES row of the matching adjustment view.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check (vii): temporal integrity -- trade_date is datetime64[ns].
    # ------------------------------------------------------------------
    # Iterate panels in pinned order.
    for view_label, panel in (("UNADJUSTED_BARS", una), ("ADJUSTED_BARS", adj)):
        # The flat column must be exactly datetime64[ns], never string/object/tz-aware.
        if panel["trade_date"].dtype != np.dtype("datetime64[ns]"):
            # Record and halt with both spellings named.
            halt = _record_and_halt(
                records,
                "BARS.temporal_integrity",
                f"{view_label}: trade_date dtype is {panel['trade_date'].dtype!s}; expected datetime64[ns] exactly.",
                [f"trade_date:{panel['trade_date'].dtype}"],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
        # The index level must also be exactly datetime64[ns].
        if panel.index.get_level_values("trade_date").dtype != np.dtype("datetime64[ns]"):
            # Record and halt with both spellings named.
            halt = _record_and_halt(
                records,
                "BARS.temporal_integrity",
                f"{view_label}: trade_date index level dtype is "
                f"{panel.index.get_level_values('trade_date').dtype!s}; expected datetime64[ns] exactly.",
                ["trade_date:index"],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
    # Record the passing sub-check (vii).
    _append_report_record(
        records, "BARS.temporal_integrity", "INFO",
        "Both panels: trade_date is datetime64[ns] in columns and index levels.", [],
    )

    # ------------------------------------------------------------------
    # Sub-check (viii): role-conformance of price column naming.
    # ------------------------------------------------------------------
    # ADJUSTED_BARS must carry no bare nominal price column.
    for stem in PRICE_STEMS:
        # A bare stem in the adjusted view is an adjusted-price carrier in nominal dress.
        if stem in adj.columns:
            # Record and halt with the offending column named.
            halt = _record_and_halt(
                records,
                "BARS.role_conformance",
                f"ADJUSTED_BARS carries bare price column {stem!r}; adjusted prices must carry the '_adj' suffix.",
                [stem],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
    # UNADJUSTED_BARS must carry no _adj-suffixed price column.
    for stem in PRICE_STEMS:
        # A suffixed stem in the nominal view is a look-ahead-adjacent mislabel.
        if f"{stem}_adj" in una.columns:
            # Record and halt with the offending column named.
            halt = _record_and_halt(
                records,
                "BARS.role_conformance",
                f"UNADJUSTED_BARS carries adjusted price column {stem}_adj; nominal prices must not carry the '_adj' suffix.",
                [f"{stem}_adj"],
            )
            # Return with the halt reason.
            return una, adj, _finalize_validation_report(records), halt
    # Record the passing sub-check (viii).
    _append_report_record(
        records, "BARS.role_conformance", "INFO",
        "Role conformance holds: adjusted prices carry '_adj' in ADJUSTED_BARS; nominal prices carry no suffix in UNADJUSTED_BARS.", [],
    )

    # Materialise the diagnostic and return the augmented copies.
    report = _finalize_validation_report(records)
    # Return nominal panel, adjusted panel, report, and no halt reason.
    return una, adj, report, None

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Step 3: STUDY_CONFIG semantic dict and provenance-tag audit
# -------------------------------------------------------------------------------------------------------------------------------

_EVAL_ALLOWED_NODES: Tuple[type, ...] = (
    ast.Expression,
    ast.Constant,
    ast.BinOp,
    ast.Compare,
    ast.UnaryOp,
    ast.Add,
    ast.Sub,
    ast.Mult,
    ast.Div,
    ast.Mod,
    ast.UAdd,
    ast.USub,
    ast.Eq,
)
# Study pin: the whitelist is exactly the grammar of the pinned P42 identities:
# integer/float/bool constants, + - * / %, unary +/-, and equality comparisons
# (including chained equality, e.g. "1047 - 25 == 1022 == 923 + 99").


def _safe_eval_arithmetic_identity(expression: str) -> bool:
    """
    Evaluate a P42 arithmetic identity as a Boolean predicate.

    Parses ``expression`` with ``ast.parse(..., mode="eval")`` and walks
    the tree, rejecting any node outside ``_EVAL_ALLOWED_NODES``, any
    constant that is not ``int``/``float``/``bool`` and any comparison
    operator other than ``ast.Eq``. The value is then re-derived
    bottom-up by a private recursive evaluator supporting unary
    plus/minus, ``+ - * / %`` and chained equality; names, attributes
    and calls are never resolvable.

    Parameters
    ----------
    expression : str
        Identity string taken from ``P42_ARITHMETIC_CLOSURES``.

    Returns
    -------
    bool
        The identity's truth value, coerced to a plain ``bool``.

    Raises
    ------
    ValueError
        If the expression does not parse, contains a disallowed node
        or operator, a non-numeric constant, or a non-equality
        comparison.
    ZeroDivisionError
        If the evaluation divides by zero; it is not caught.

    Notes
    -----
    Chained equality is evaluated over every link without Python's
    short-circuit: a failed link sets the result to ``False`` but the
    remaining comparators are still evaluated. The returned value is
    the conjunction of all links, so
    ``1047 - 25 == 1022 == 923 + 99`` yields ``True``.
    """
    # Parse the identity; syntax errors surface as ValueError for the caller.
    try:
        # mode="eval" yields a single-expression AST.
        tree = ast.parse(expression, mode="eval")
    except SyntaxError as exc:
        # Re-raise as ValueError so the caller can record a failed check.
        raise ValueError(f"Identity {expression!r} is not parseable: {exc}.") from exc
    # Whitelist pass: reject any node outside the pinned grammar.
    for node in ast.walk(tree):
        # Any node outside the whitelist aborts evaluation.
        if not isinstance(node, _EVAL_ALLOWED_NODES):
            # Name the offending node type in the diagnostic.
            raise ValueError(f"Identity {expression!r} contains disallowed node {type(node).__name__}.")
        # Constants must be numeric or Boolean.
        if isinstance(node, ast.Constant) and not isinstance(node.value, (int, float, bool)):
            # Reject strings, None, bytes, etc. even inside otherwise valid identities.
            raise ValueError(f"Identity {expression!r} contains non-numeric constant {node.value!r}.")
        # Comparisons must be equality comparisons only.
        if isinstance(node, ast.Compare) and not all(isinstance(op, ast.Eq) for op in node.ops):
            # Reject ordering and membership operators.
            raise ValueError(f"Identity {expression!r} contains a non-equality comparison.")

    def _evaluate(node: ast.AST) -> Any:
        """
        Recursively evaluate one whitelisted AST node.

        Handles ``Expression``, ``Constant``, ``UnaryOp``,
        ``BinOp`` (``+ - * / %``) and ``Compare`` (equality chain).
        Unary and comparison operators are not re-validated here: any
        ``UnaryOp`` other than ``UAdd`` is treated as negation, and
        every ``Compare`` link is treated as equality; the outer
        whitelist pass is what keeps such nodes from arriving.

        Parameters
        ----------
        node : ast.AST
            Node to evaluate; must already be whitelist-clean.

        Returns
        -------
        Any
            The numeric or Boolean value of the node.

        Raises
        ------
        ValueError
            If the node type is unknown, or a ``BinOp`` carries an
            operator outside ``+ - * / %``.
        ZeroDivisionError
            If a division or modulo has a zero divisor.
        """
        # Expressions evaluate their single body node.
        if isinstance(node, ast.Expression):
            # Recurse into the body.
            return _evaluate(node.body)
        # Constants return their value.
        if isinstance(node, ast.Constant):
            # Return the literal.
            return node.value
        # Unary plus/minus over a numeric operand.
        if isinstance(node, ast.UnaryOp):
            # Evaluate the operand first.
            operand = _evaluate(node.operand)
            # Unary plus returns the operand unchanged.
            if isinstance(node.op, ast.UAdd):
                # +x
                return +operand
            # Unary minus negates.
            return -operand
        # Binary arithmetic.
        if isinstance(node, ast.BinOp):
            # Evaluate both sides first.
            left, right = _evaluate(node.left), _evaluate(node.right)
            # Addition.
            if isinstance(node.op, ast.Add):
                # a + b
                return left + right
            # Subtraction.
            if isinstance(node.op, ast.Sub):
                # a - b
                return left - right
            # Multiplication.
            if isinstance(node.op, ast.Mult):
                # a * b
                return left * right
            # Division.
            if isinstance(node.op, ast.Div):
                # a / b; ZeroDivisionError propagates to the caller.
                return left / right
            # Modulo.
            if isinstance(node.op, ast.Mod):
                # a % b
                return left % right
            # Unreachable under the whitelist, kept defensive.
            raise ValueError(f"Identity {expression!r} contains disallowed binary operator {type(node.op).__name__}.")
        # Chained equality comparison, evaluated left to right with short-circuit.
        if isinstance(node, ast.Compare):
            # Start from the left operand.
            left_value = _evaluate(node.left)
            # Accumulate the chained predicate.
            result = True
            # Iterate the comparators pairwise.
            for comparator in node.comparators:
                # Evaluate the right side of this pair.
                right_value = _evaluate(comparator)
                # Strict equality only; operands are numeric/bool.
                if not (left_value == right_value):
                    # Record the failing link; continue to keep semantics total.
                    result = False
                # Advance the chain window.
                left_value = right_value
            # Return the accumulated predicate.
            return result
        # Unreachable under the whitelist, kept defensive.
        raise ValueError(f"Identity {expression!r} contains disallowed node {type(node).__name__}.")

    # Evaluate and normalise to a plain Python bool.
    return bool(_evaluate(tree))

def _p42_flatten_checks(
    p42_section: Dict[str, Any],
) -> Tuple[List[Tuple[str, str]], List[str]]:
    """
    Flatten ``P42_ARITHMETIC_CLOSURES`` into check pairs and notes.

    Iterates ``p42_section`` in sorted-key order. String values become
    ``("P42.<name>", value)``; list values are flattened by index into
    ``("P42.<name>.<i>", element)`` for string elements, while
    non-string elements are recorded as notes; dict values (e.g.
    ``sum_closures_rounds_to_totals``) are recorded as notes because
    their predicates refer to round-level aggregates that do not exist
    at Task 1; entries of any other type are recorded as notes naming
    the unsupported entry type. Neither dicts nor other-type entries
    are ever evaluated.

    Parameters
    ----------
    p42_section : Dict[str, Any]
        The ``P42_ARITHMETIC_CLOSURES`` mapping.

    Returns
    -------
    checks : List[Tuple[str, str]]
        Evaluable ``(check_id, identity)`` pairs in sorted-key order,
        sub-identities ordered by list index.
    notes : List[str]
        Human-readable notes naming each entry that was skipped and
        why.
    """
    # Initialise the output accumulators.
    checks: List[Tuple[str, str]] = []
    # Notes collect non-evaluable entries for the INFO record.
    notes: List[str] = []
    # Iterate deterministically by sorted key.
    for name in sorted(p42_section):
        # Read the entry.
        value = p42_section[name]
        # String identities are evaluable verbatim.
        if isinstance(value, str):
            # Append the (check_id, identity) pair.
            checks.append((f"P42.{name}", value))
            # Continue to the next entry.
            continue
        # List entries flatten by index.
        if isinstance(value, list):
            # Iterate with explicit indices to pin the sub-check naming.
            for index, sub in enumerate(value):
                # String sub-identities are evaluable.
                if isinstance(sub, str):
                    # Append the flattened (check_id, identity) pair.
                    checks.append((f"P42.{name}.{index}", sub))
                # Non-string sub-entries are recorded, not evaluated.
                else:
                    # Note the un-evaluable sub-entry for the INFO record.
                    notes.append(f"P42.{name}.{index}: non-string entry {sub!r} not evaluated.")
            # Continue to the next entry.
            continue
        # Dict entries (e.g. sum_closures_rounds_to_totals) are not evaluable at Task 1.
        if isinstance(value, dict):
            # Note the entry, name the reason, never halt, never silently drop.
            notes.append(
                f"P42.{name}: not evaluated at Task 1 -- predicates reference round-level "
                "aggregates unavailable at this stage; itemised out of the step's evaluable set."
            )
            # Continue to the next entry.
            continue
        # Any other entry type is noted, never evaluated.
        notes.append(f"P42.{name}: unsupported entry type {type(value).__name__} not evaluated.")
    # Return the pairs and the notes.
    return checks, notes

def _recognised_tag_bucket(tag: Any) -> Optional[str]:
    """
    Map a provenance tag to its counting bucket.

    Recognises exactly ``[S]``, ``[D]``, ``[R]``, ``[I]`` and
    ``UNDISCLOSED`` through ``EXACT_TAG_BUCKETS``, and any string whose
    literal prefix matches one of ``UNSPEC_TAG_PREFIXES`` (the
    open-ended ``[UNSPECIFIED IN MANUSCRIPT ...`` prose is matched as a
    prefix only, never parsed). Everything else, including non-string
    tags, is unrecognised.

    Parameters
    ----------
    tag : Any
        Tag value found in the parallel ``_provenance_tags`` dict at
        the leaf's path.

    Returns
    -------
    Optional[str]
        One of ``"S"``, ``"D"``, ``"R"``, ``"I"``, ``"UNDISCLOSED"``
        or ``"UNSPECIFIED"``; ``None`` when unrecognised.
    """
    # Tags must be strings to be recognised.
    if not isinstance(tag, str):
        # Non-string tags are unrecognised.
        return None
    # Exact-tag membership first.
    if tag in EXACT_TAG_BUCKETS:
        # Return the pinned bucket for the exact tag.
        return EXACT_TAG_BUCKETS[tag]
    # Open-ended UNSPECIFIED prose: match the literal prefix only.
    for prefix in UNSPEC_TAG_PREFIXES:
        # Prefix match, no semantic parsing.
        if tag.startswith(prefix):
            # Return the UNSPECIFIED bucket.
            return "UNSPECIFIED"
    # Anything else is unrecognised.
    return None

def _walk_and_audit(
    node: Any,
    tag_node: Any,
    path: str,
    counts: Dict[str, int],
    defects: List[Tuple[str, str]],
    undisclosed_refs: List[Tuple[str, str]],
    enclosing_key: Optional[str],
) -> None:
    """
    Audit config leaves against the parallel provenance tags.

    Walks ``node`` uniformly: dicts recurse over sorted keys, lists
    recurse by ``enumerate`` index, everything else is a leaf. For a
    list element, if the tag mirror is missing, not a list, or shorter
    than the element list, a defect is recorded at ``path[index]`` and
    the element is recursed with ``tag_node=None``; this element-path
    defect fires regardless of the element value. Each leaf must be
    either the literal string ``"UNDISCLOSED"`` -- counted in
    ``counts["UNDISCLOSED"]`` and registered in ``undisclosed_refs``
    with its innermost enclosing dict key -- or carry a recognised tag
    at the same path, which increments that bucket in ``counts``.
    Untagged or unrecognised leaves go to ``defects`` and are never
    silently bucketed as ``[R]``. Empty dicts yield no leaves.

    Parameters
    ----------
    node : Any
        Current node of the semantic configuration dict.
    tag_node : Any
        Matching node of the parallel tag dict; ``None`` when absent.
    path : str
        Dotted path of ``node``, brackets for list indices.
    counts : Dict[str, int]
        Running bucket tally (``S``, ``D``, ``R``, ``I``,
        ``UNDISCLOSED``, ``UNSPECIFIED``); mutated in place.
    defects : List[Tuple[str, str]]
        Accumulator for ``(path, reason)`` tag defects; mutated in
        place.
    undisclosed_refs : List[Tuple[str, str]]
        Accumulator for ``(path, enclosing_key)`` pairs of
        ``"UNDISCLOSED"`` leaves.
    enclosing_key : Optional[str]
        Innermost dict key holding ``node``; list elements inherit it,
        and a ``None`` falls back to the leaf path.
    """
    # Dict nodes recurse uniformly over sorted keys.
    if isinstance(node, dict):
        # The tag mirror at this level is a dict or absent.
        tag_map = tag_node if isinstance(tag_node, dict) else None
        # Deterministic iteration order: sorted keys.
        for key in sorted(node):
            # Read the child value.
            value = node[key]
            # Resolve the child's tag node from the mirror at this level.
            child_tag = tag_map.get(key) if isinstance(tag_map, dict) else None
            # Extend the dotted path.
            child_path = f"{path}.{key}" if path else str(key)
            # Recurse with this key as the new enclosing key.
            _walk_and_audit(value, child_tag, child_path, counts, defects, undisclosed_refs, str(key))
        # Done with this dict level.
        return
    # List nodes recurse uniformly by index.
    if isinstance(node, list):
        # The tag mirror at this level is a list or absent.
        tag_list = tag_node if isinstance(tag_node, list) else None
        # Enumerate elements in stable order.
        for index, value in enumerate(node):
            # Resolve the element's tag from an index-aligned list mirror.
            if isinstance(tag_list, list) and index < len(tag_list):
                # Index-aligned tag; pass it down.
                child_tag = tag_list[index]
            else:
                # Missing or short mirror: the leaf will be surfaced as untagged.
                child_tag = None
                # Record the mirror defect at the element path.
                defects.append(
                    (f"{path}[{index}]", "list leaf carries no index-aligned tag in the parallel provenance-tag dictionary")
                )
            # Recurse with the bracket-indexed path; enclosing key is inherited.
            _walk_and_audit(value, child_tag, f"{path}[{index}]", counts, defects, undisclosed_refs, enclosing_key)
        # Done with this list level.
        return
    # Leaves: a literal UNDISCLOSED value is the tag-exempt branch.
    if isinstance(node, str) and node == "UNDISCLOSED":
        # Count the literal leaf in the UNDISCLOSED bucket.
        counts["UNDISCLOSED"] += 1
        # Register the leaf for the P45 reconciliation with its enclosing key.
        undisclosed_refs.append((path, enclosing_key if enclosing_key is not None else path))
        # Nothing else to check at this leaf.
        return
    # Otherwise the parallel tag dict must carry a recognised tag.
    bucket = _recognised_tag_bucket(tag_node)
    # An unrecognised or missing tag surfaces the leaf -- never silently categorised as [R].
    if bucket is None:
        # Build the precise reason for the diagnostic.
        if tag_node is None:
            # Absent tag at this path.
            defects.append((path, "leaf carries no tag and its value is not the literal 'UNDISCLOSED'"))
        else:
            # Tag outside the recognised enumeration.
            defects.append((path, f"tag {tag_node!r} is outside the recognised enumeration ([S], [D], [R], [I], UNDISCLOSED, [UNSPECIFIED IN MANUSCRIPT: ...])"))
        # Do not count defect leaves in any bucket.
        return
    # Count the leaf in its bucket.
    counts[bucket] += 1

def _reconcile_disclosure_gaps(
    undisclosed_refs: List[Tuple[str, str]],
    p45_section: Dict[str, Any],
    reconciliation: Optional[Any] = None,
) -> Tuple[List[Tuple[str, str]], List[str]]:
    """
    Reconcile literals ``UNDISCLOSED`` against P45 gaps.

    Each leaf references a ``P45_DISCLOSURE_GAPS`` key: by default
    its innermost enclosing dict key, or the key given by the
    optional ``P45_GAP_RECONCILIATION`` map. The mapped value
    ``P45_NOT_A_GAP`` exempts a leaf. A referenced key must exist in
    ``p45_section`` and its value must be literally ``False``; a
    missing key or a non-``False`` value is a defect. Every key of
    ``PINNED_P45_KEYS`` must be present, every declared
    self-evidenced gap must exist and be ``False``, and every key of
    ``p45_section`` must be consumed by a leaf or declared
    self-evidenced. A malformed reconciliation section yields one
    defect and maps nothing.

    Parameters
    ----------
    undisclosed_refs : List[Tuple[str, str]]
        Leaves gathered by the tag walk as ``(path,
        enclosing_key)`` pairs, in walk order.
    p45_section : Dict[str, Any]
        The ``P45_DISCLOSURE_GAPS`` mapping; each value should be
        literally ``False`` to record an absence.
    reconciliation : Optional[Any], optional
        The ``P45_GAP_RECONCILIATION`` section, with members
        ``leaf_to_gap`` (path -> gap key) and
        ``self_evidenced_gaps`` (list of gap keys), or ``None``
        when absent. Default is ``None``.

    Returns
    -------
    defects : List[Tuple[str, str]]
        ``(key_or_path, reason)`` pairs in detection order.
    consumed_keys : List[str]
        Sorted, distinct ``p45_section`` keys consumed by a leaf
        or declared self-evidenced.
    """
    # Accumulate defects and consumed keys.
    defects: List[Tuple[str, str]] = []
    # Consumed keys are marked even when their P45 value is not False, so a
    # True-valued consumed key yields exactly one defect and is not double-counted.
    consumed: set = set()
    # The explicit leaf-to-gap map and the self-evidenced gaps (both empty without the section).
    leaf_to_gap: Dict[str, str] = {}
    self_evidenced: List[str] = []
    # Validate the section's shape when present.
    if reconciliation is not None:
        # The two members, read defensively.
        raw_map: Any = reconciliation.get("leaf_to_gap", {}) if isinstance(reconciliation, dict) else None
        raw_self: Any = reconciliation.get("self_evidenced_gaps", []) if isinstance(reconciliation, dict) else None
        # A mapping of strings to strings, and a list of strings.
        if not (isinstance(raw_map, dict) and all(isinstance(k, str) and isinstance(v, str) for k, v in raw_map.items())
                and isinstance(raw_self, list) and all(isinstance(k, str) for k in raw_self)):
            # Malformed: nothing is mapped, and the defect is named.
            defects.append((P45_GAP_RECONCILIATION_KEY, "malformed: expected {'leaf_to_gap': {path: gap_key}, 'self_evidenced_gaps': [gap_key]}"))
        else:
            # Accepted.
            leaf_to_gap, self_evidenced = dict(raw_map), list(raw_self)
    # A map entry naming no literal UNDISCLOSED leaf consumes nothing (only leaves consume), so it cannot make
    # the reconciliation pass falsely; validate_config_task_1 reports such entries without halting.
    # Reconcile each UNDISCLOSED leaf individually.
    for path, enclosing in undisclosed_refs:
        # The leaf's gap: its mapped gap, else its innermost enclosing key.
        key: str = leaf_to_gap.get(path, enclosing)
        # A leaf ruled to be no disclosure gap is exempt.
        if key == P45_NOT_A_GAP:
            # Nothing to reconcile.
            continue
        # Absent pinned key: the leaf cannot be covered -- defect.
        if key not in p45_section:
            # Record the uncovered leaf with its full path.
            defects.append((path, f"UNDISCLOSED leaf key {key!r} has no entry in P45_DISCLOSURE_GAPS"))
            # Do not mark it consumed.
            continue
        # Mark the key consumed regardless of the recorded value's truthiness.
        consumed.add(key)
        # The recorded absence must be literally False.
        if p45_section[key] is not False:
            # Record the dishonest recorded absence.
            defects.append((key, f"P45_DISCLOSURE_GAPS[{key!r}] is {p45_section[key]!r}; a recorded absence must be False"))
    # The twelve pinned keys must exist in P45_DISCLOSURE_GAPS.
    for pinned_key in PINNED_P45_KEYS:
        # A missing pinned key makes joint coverage of every UNDISCLOSED leaf impossible.
        if pinned_key not in p45_section:
            # Record the missing pinned key.
            defects.append((pinned_key, "pinned P45_DISCLOSURE_GAPS key is absent from the section"))
    # A self-evidenced gap is consumed by its own recorded absence, which must itself be False.
    for key in self_evidenced:
        # It must be a P45 key.
        if key not in p45_section:
            # Named.
            defects.append((key, "self-evidenced gap has no entry in P45_DISCLOSURE_GAPS"))
            # Not consumed.
            continue
        # Consumed.
        consumed.add(key)
        # Its recorded absence must be literally False.
        if p45_section[key] is not False:
            # Named.
            defects.append((key, f"P45_DISCLOSURE_GAPS[{key!r}] is {p45_section[key]!r}; a recorded absence must be False"))
    # Every P45 key must be consumed by at least one UNDISCLOSED leaf or declared self-evidenced.
    for key in sorted(set(p45_section) - consumed):
        # An unconsumed key is a bookkeeping defect.
        defects.append((key, "P45_DISCLOSURE_GAPS key is not consumed by any UNDISCLOSED leaf"))
    # Return defects and the sorted consumed list.
    return defects, sorted(consumed)

def validate_config_task_1(
    STUDY_CONFIG: Dict[str, Any],
) -> Tuple[pd.DataFrame, Optional[str], Dict[str, Any]]:
    """
    Audit ``STUDY_CONFIG`` and republish its provenance tags.

    Runs three ordered gates and stops at the first halt. First, the
    P42 arithmetic closures are flattened and each identity is
    evaluated by a whitelisted AST evaluator; any ``False`` or
    unevaluable identity (``ValueError``, ``ZeroDivisionError``,
    ``OverflowError``) is recorded as ``ERROR`` and halts under
    ``P42.arithmetic_closures``. Second, the semantic dict (all
    sections except ``_provenance_tags``) is walked against the
    parallel tag dict; every leaf must carry a recognised tag or be
    the literal ``"UNDISCLOSED"``. An absent tag dict halts under
    ``TAGS.presence``; any defect halts under ``TAGS.completeness``
    with up to 25 offending paths. Third, the ``"UNDISCLOSED"``
    leaves are reconciled against ``P45_DISCLOSURE_GAPS`` via
    ``_reconcile_disclosure_gaps``; defects halt under
    ``P45.disclosure_gaps``. Map entries naming no literal
    ``"UNDISCLOSED"`` leaf are recorded as ``INFO`` under
    ``P45.gap_map_unused``. The semantic dict is never mutated.

    Parameters
    ----------
    STUDY_CONFIG : Dict[str, Any]
        The study configuration; must carry ``P42_ARITHMETIC_CLOSURES``
        and ``P45_DISCLOSURE_GAPS``, and should carry the parallel
        ``_provenance_tags`` sub-dict. An optional
        ``P45_GAP_RECONCILIATION`` sub-dict redirects leaves to gap
        keys.

    Returns
    -------
    validation_report : pd.DataFrame
        Write-once diagnostic from ``_finalize_validation_report``.
    halt_reason : Optional[str]
        ``None`` when all three gates pass, else the first failure.
    provenance_tags : Dict[str, Any]
        Deep copy of ``_provenance_tags``, republished on every exit
        path; ``{}`` when it is absent or not a dict.

    Raises
    ------
    TypeError
        If ``STUDY_CONFIG`` is not a dict.
    """
    # Validate the argument type before any work.
    if not isinstance(STUDY_CONFIG, dict):
        # Reject non-dict configurations.
        raise TypeError(f"STUDY_CONFIG must be a dict, got {type(STUDY_CONFIG)!r}.")
    # Deep-copy the tag dict now; it is republished verbatim on every exit path.
    raw_tags = STUDY_CONFIG.get("_provenance_tags")
    # Verbatim republication (deep copy to keep the caller's object untouched).
    provenance_tags = copy.deepcopy(raw_tags) if isinstance(raw_tags, dict) else {}
    # Initialise the write-once record buffer.
    records: List[Dict[str, Any]] = []

    # ------------------------------------------------------------------
    # Arithmetic-closure pre-check: evaluate every P42 identity.
    # ------------------------------------------------------------------
    # Read the P42 section defensively.
    p42_section = _config_section(STUDY_CONFIG, "P42_ARITHMETIC_CLOSURES")
    # Flatten to evaluable (check_id, identity) pairs plus informational notes.
    checks, notes = _p42_flatten_checks(p42_section)
    # Record every note as INFO; surfaced, never silently dropped, never halted.
    for note in notes:
        # Split the note's check_id prefix for the index.
        check_id = note.split(":", 1)[0]
        # Append the INFO record naming the non-evaluated entry and the reason.
        _append_report_record(records, check_id, "INFO", note, [])
    # Evaluate every flattened identity; collect failures rather than halting on the first.
    failures: List[str] = []
    # Evaluate each identity under the whitelisted AST evaluator.
    for check_id, identity in checks:
        # Attempt evaluation; any parser/whitelist/zero-division defect is a failed check.
        try:
            # Evaluate the identity; returns a plain bool.
            result = _safe_eval_arithmetic_identity(identity)
        except (ValueError, ZeroDivisionError, OverflowError) as exc:
            # Treat the un-evaluable identity as a failed check, with the reason recorded.
            result = False
            # Append the ERROR record with the parse defect named.
            _append_report_record(
                records, check_id, "ERROR",
                f"{identity} -> unevaluable ({exc}); treat as failed.", [identity],
            )
            # Register the failure.
            failures.append(check_id)
            # Continue to the next identity.
            continue
        # Record the outcome with the identity and its Boolean value.
        _append_report_record(
            records, check_id, "INFO" if result else "ERROR",
            f"{identity} -> {result}.", [] if result else [identity],
        )
        # Collect failures for the halt decision.
        if not result:
            # Register the failed check_id.
            failures.append(check_id)
    # Any failed identity halts the run.
    if failures:
        # Record and halt with the failing checks named.
        halt = _record_and_halt(
            records,
            "P42.arithmetic_closures",
            f"Arithmetic-closure pre-check failed: {', '.join(failures)}.",
            failures,
        )
        # Return with the halt reason and the verbatim republished tags.
        return _finalize_validation_report(records), halt, provenance_tags
    # Record the passing closure gate.
    _append_report_record(
        records, "P42.arithmetic_closures", "INFO",
        f"All {len(checks)} arithmetic closures close to the unit.", [],
    )

    # ------------------------------------------------------------------
    # Tag-completeness audit against the parallel provenance-tag dictionary.
    # ------------------------------------------------------------------
    # The parallel tag dict must be present and a mapping.
    if not isinstance(raw_tags, dict):
        # Record and halt; comments are not data -- the parallel dict is mandatory.
        halt = _record_and_halt(
            records,
            "TAGS.presence",
            "The parallel provenance-tag dictionary (_provenance_tags) is absent or not a mapping; "
            "comments are not data and are never consulted.",
            [],
        )
        # Return with the halt reason and the (empty) republished tags.
        return _finalize_validation_report(records), halt, provenance_tags
    # Initialise the per-tag tally and the defect accumulators.
    counts: Dict[str, int] = {"S": 0, "D": 0, "R": 0, "I": 0, "UNDISCLOSED": 0, "UNSPECIFIED": 0}
    # Defects accumulate as (path, reason) pairs.
    defects: List[Tuple[str, str]] = []
    # Undisclosed literal leaves accumulate for the P45 reconciliation.
    undisclosed_refs: List[Tuple[str, str]] = []
    # The semantic walk excludes the tag dict itself; its leaves are tags, not config.
    semantic_view = {key: value for key, value in STUDY_CONFIG.items() if key != "_provenance_tags"}
    # The mirror for the semantic walk is the tag dict minus nothing (it has no _provenance_tags key).
    tag_mirror = {key: value for key, value in raw_tags.items()}
    # Authoritative walk over the semantic dict.
    _walk_and_audit(
        semantic_view,
        tag_mirror,
        "",
        counts,
        defects,
        undisclosed_refs,
        None,
    )
    # Report the per-tag counts in an INFO record.
    _append_report_record(
        records, "TAGS.tag_counts", "INFO",
        "Per-tag counts: " + json.dumps(counts, sort_keys=True) + ".", [],
    )
    # Any untagged or mis-tagged leaf halts the run pending explicit adjudication.
    if defects:
        # Record and halt with up to 25 defect paths named.
        halt = _record_and_halt(
            records,
            "TAGS.completeness",
            f"Tag-completeness audit surfaced {len(defects)} defect(s); the run halts pending "
            f"explicit adjudication. First defect: {defects[0][0]} -- {defects[0][1]}.",
            [path for path, _reason in defects[:25]],
        )
        # Return with the halt reason and the verbatim republished tags.
        return _finalize_validation_report(records), halt, provenance_tags
    # Record the passing tag-completeness gate.
    _append_report_record(
        records, "TAGS.completeness", "INFO",
        f"Every leaf carries a recognised tag or is a literal 'UNDISCLOSED'; "
        f"{len(undisclosed_refs)} UNDISCLOSED leaf/leaves reconciled next.", [],
    )

    # ------------------------------------------------------------------
    # Disclosure-gap reconciliation against P45_DISCLOSURE_GAPS.
    # ------------------------------------------------------------------
    # Read the P45 section defensively.
    p45_section = _config_section(STUDY_CONFIG, "P45_DISCLOSURE_GAPS")
    # Reconcile every UNDISCLOSED leaf and enforce joint coverage by the pinned keys.
    gap_defects, consumed = _reconcile_disclosure_gaps(
        undisclosed_refs, p45_section, STUDY_CONFIG.get(P45_GAP_RECONCILIATION_KEY)
    )
    # Gap-map entries that name no literal UNDISCLOSED leaf (e.g. a fixture that renamed the leaf) are reported.
    reconciliation_section: Any = STUDY_CONFIG.get(P45_GAP_RECONCILIATION_KEY)
    # Only a well-formed map is inspected (a malformed one is already a defect).
    if isinstance(reconciliation_section, dict) and isinstance(reconciliation_section.get("leaf_to_gap"), dict):
        # The stale paths.
        stale_paths: List[str] = sorted(set(reconciliation_section["leaf_to_gap"]) - {path for path, _key in undisclosed_refs})
        # Recorded when present.
        if stale_paths:
            # Informational: a stale entry consumes no gap.
            _append_report_record(
                records, "P45.gap_map_unused", "INFO",
                f"{len(stale_paths)} P45_GAP_RECONCILIATION entr(y/ies) name no literal 'UNDISCLOSED' leaf; they consume nothing.",
                stale_paths[:25],
            )
    # Any defect halts the run.
    if gap_defects:
        # Record and halt with up to 25 defects named.
        halt = _record_and_halt(
            records,
            "P45.disclosure_gaps",
            f"Disclosure-gap reconciliation surfaced {len(gap_defects)} defect(s); "
            f"first: {gap_defects[0][0]} -- {gap_defects[0][1]}.",
            [key for key, _reason in gap_defects[:25]],
        )
        # Return with the halt reason and the verbatim republished tags.
        return _finalize_validation_report(records), halt, provenance_tags
    # Record the passing reconciliation gate.
    _append_report_record(
        records, "P45.disclosure_gaps", "INFO",
        f"All {len(undisclosed_refs)} UNDISCLOSED leaf/leaves reconcile to a False P45 entry (or are ruled no gap); "
        f"{len(consumed)} of {len(p45_section)} P45 keys consumed (12 pinned keys all present and consumed).", [],
    )

    # Materialise the diagnostic and return it with the verbatim republished tags.
    report = _finalize_validation_report(records)
    # Return the report, no halt reason, and the republished tag artifact.
    return report, None, provenance_tags

# -------------------------------------------------------------------------------------------------------------------------------
# Task 1, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_1(
    FILES: pd.DataFrame,
    UNADJUSTED_BARS: pd.DataFrame,
    ADJUSTED_BARS: pd.DataFrame,
    STUDY_CONFIG: Dict[str, Any],
    TRADING_CALENDAR: pd.DataFrame,
) -> Optional[str]:
    """
    Run the three ordered steps of Task 1 and report the first halt.

    Validates the argument types, then executes Step 1
    (``validate_files_task_1``), Step 2 (``validate_panels_task_1``)
    and Step 3 (``validate_config_task_1``) in that order, stopping
    at the first non-``None`` halt reason and returning it verbatim.
    Step 2 receives the augmented manifest returned by Step 1, as
    the production pipeline does; the caller's frames are not
    mutated because each step copies its inputs. The validation
    reports and augmented frames produced by the steps are
    discarded here.

    Parameters
    ----------
    FILES : pd.DataFrame
        The acquisition manifest indexed on ``file_id``, with header
        metadata in ``FILES.attrs``; Step 1 adds its four sanctioned
        columns.
    UNADJUSTED_BARS : pd.DataFrame
        The nominal price and volume panel indexed on
        ``(stock_code, trade_date)``.
    ADJUSTED_BARS : pd.DataFrame
        The provider-adjusted panel used for labels.
    STUDY_CONFIG : Dict[str, Any]
        The nested ``P00``-``P46`` configuration dict plus the
        parallel ``_provenance_tags`` sub-dict.
    TRADING_CALENDAR : pd.DataFrame
        The trading calendar; must carry a ``trade_date`` column
        covering the 727 sessions.

    Returns
    -------
    Optional[str]
        The first halt reason encountered, or ``None`` when every
        gate passes.

    Raises
    ------
    TypeError
        If ``FILES``, ``UNADJUSTED_BARS``, ``ADJUSTED_BARS`` or
        ``TRADING_CALENDAR`` is not a ``pd.DataFrame``, or if
        ``STUDY_CONFIG`` is not a dict.
    ValueError
        Propagated from the steps, for example when
        ``TRADING_CALENDAR`` lacks ``trade_date`` or ``FILES`` lacks
        ``byte_hash`` or ``adjustment``.
    """
    # Validate argument types before any work; wrong types are structural defects.
    if not isinstance(FILES, pd.DataFrame):
        # FILES must be a DataFrame.
        raise TypeError(f"FILES must be a pandas.DataFrame, got {type(FILES)!r}.")
    # UNADJUSTED_BARS must be a DataFrame.
    if not isinstance(UNADJUSTED_BARS, pd.DataFrame):
        # Reject non-DataFrame nominal panels.
        raise TypeError(f"UNADJUSTED_BARS must be a pandas.DataFrame, got {type(UNADJUSTED_BARS)!r}.")
    # ADJUSTED_BARS must be a DataFrame.
    if not isinstance(ADJUSTED_BARS, pd.DataFrame):
        # Reject non-DataFrame adjusted panels.
        raise TypeError(f"ADJUSTED_BARS must be a pandas.DataFrame, got {type(ADJUSTED_BARS)!r}.")
    # STUDY_CONFIG must be the nested configuration dictionary.
    if not isinstance(STUDY_CONFIG, dict):
        # Reject non-dict configurations.
        raise TypeError(f"STUDY_CONFIG must be a dict, got {type(STUDY_CONFIG)!r}.")
    # TRADING_CALENDAR must be a DataFrame carrying the calendar.
    if not isinstance(TRADING_CALENDAR, pd.DataFrame):
        # Reject non-DataFrame calendars.
        raise TypeError(f"TRADING_CALENDAR must be a pandas.DataFrame, got {type(TRADING_CALENDAR)!r}.")

    # Step 1: validate the acquisition manifest; stop at the first halt.
    files_out, _files_report, files_halt = validate_files_task_1(
        FILES=FILES,
        STUDY_CONFIG=STUDY_CONFIG,
        TRADING_CALENDAR=TRADING_CALENDAR,
    )
    # A Step 1 halt short-circuits the run; the reason is returned verbatim.
    if files_halt is not None:
        # Disclosure: Steps 2 and 3 are never reached behind a Step 1 halt.
        return files_halt

    # Step 2: validate the two panels against the augmented manifest; stop at the first halt.
    _una_out, _adj_out, _panels_report, panels_halt = validate_panels_task_1(
        UNADJUSTED_BARS=UNADJUSTED_BARS,
        ADJUSTED_BARS=ADJUSTED_BARS,
        FILES=files_out,
        STUDY_CONFIG=STUDY_CONFIG,
        TRADING_CALENDAR=TRADING_CALENDAR,
    )
    # A Step 2 halt short-circuits the run; the reason is returned verbatim.
    if panels_halt is not None:
        # Disclosure: Step 3 is never reached behind a Step 2 halt.
        return panels_halt

    # Step 3: audit the configuration; stop at the first halt.
    _config_report, config_halt, _provenance_tags_out = validate_config_task_1(
        STUDY_CONFIG=STUDY_CONFIG,
    )
    # A Step 3 halt short-circuits the run; the reason is returned verbatim.
    if config_halt is not None:
        # Disclosure: the orchestrator ends in the Step 3 halt reason.
        return config_halt

    # Every gate passed; no halt condition fired.
    return None
```

```python
# Task 2: Cleanse the Data

# ==============================================================================
# Task 2: Cleanse the Data
# ==============================================================================

logger = logging.getLogger(__name__)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3 (module import): pinned allow-lists, declared once at import
# -------------------------------------------------------------------------------------------------------------------------------

# The step text legislates these contents; they are not parameters. Any hit
# outside these lists is a halt condition for the orchestrator. The -1 list is
# empty on purpose: the step text states that -1 appears in neither the
# unadjusted nor the adjusted bars, so every -1 is a violation.
permitted_zero_columns: frozenset = frozenset(
    {
        # volume == 0 is legitimate on a suspended stock-day.
        "volume",
        # amount == 0 is legitimate on a suspended stock-day.
        "amount",
        # tradestatus_raw == 0 is legitimate on a suspended stock-day.
        "tradestatus_raw",
        # is_fitness == 0.0 is a legitimate observation.
        "is_fitness",
        # close_adj == 0 is legitimate on a (rare) zero-priced day; the
        # strict-positivity rule of the label layer catches it later.
        "close_adj",
        # quantity == 0 is legitimate: the stock existed but saw no trade
        # volume that day (a shared field; still a legitimate zero).
        "quantity",
    }
)
# -1 is permitted in NO column (the pinned list is empty).
permitted_negative_one_columns: frozenset = frozenset()
# The empty string is legitimate only in expression_text (a parse failure,
# not a null); everywhere else it is a halt.
permitted_empty_string_columns: frozenset = frozenset({"expression_text"})
# The ASCII "NA_FILL" sentinel is legitimate only in expression_text.
permitted_NA_FILL_columns: frozenset = frozenset({"expression_text"})
# The ASCII "n/a" sentinel is legitimate only in expression_text.
permitted_n_a_columns: frozenset = frozenset({"expression_text"})

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2 (module import): precompiled patterns and pinned constants
# -------------------------------------------------------------------------------------------------------------------------------

# The BaoStock 0.9.3 stock-code convention, pinned as [R] in the prior register;
# the vendor does not adjudicate its own prefix in the paper, but an unformatted
# code would break the cross-view join silently, so the assertion is load-bearing.
_STOCK_CODE_PATTERN = re.compile(r"(sh|sz)\.\d{6}")

# The pinned exchange-local timezone for retrieval stamps (config P22).
_ASIA_SHANGHAI_TZ = "Asia/Shanghai"

# The three sanctioned imputation-adjacent operations and the exact counter
# they increment; any other operation name is rejected as a programming error.
_OPERATION_TO_COUNTER: Dict[str, str] = {
    # forward_fill maps to the forward-fill counter.
    "forward_fill": "forward_fill_operations",
    # interpolation maps to the interpolation counter.
    "interpolation": "interpolation_operations",
    # stock_prefix_reduction maps to the stock-prefix-reduction counter.
    "stock_prefix_reduction": "stock_prefix_reductions",
}

# The integer terminals, exactly as the step text enumerates them; these and
# only these are cast under the Decimal(str(v)) integrality conjunction.
_INTEGER_TERMINAL_COLUMNS: Tuple[str, ...] = ("volume", "quantity", "tradestatus_raw")

# The boolean column of the two bar views (tradable) and of FILES (the manifest
# booleans); all are cast to nullable boolean[pyarrow].
_BAR_BOOLEAN_COLUMNS: Tuple[str, ...] = ("tradable",)
_FILES_BOOLEAN_COLUMNS: Tuple[str, ...] = (
    "response_truncated",
    "hash_ok",
    "date_id_ok",
    "stock_id_ok",
    "adjustment_id_ok",
    "quantity_agreement_ok",
    "status_agreement_ok",
    "verified",
    "redistribution_flag",
)

# The naive datetime64[ns] columns of FILES (exchange-local, not tz-aware).
_FILES_NAIVE_DATETIME_COLUMNS: Tuple[str, ...] = (
    "date_min",
    "date_max",
    "requested_date_min",
    "requested_date_max",
)

# The string-declared columns of the bar views and of FILES; all are persisted
# as string[pyarrow] (never object, never category).
_BAR_STRING_COLUMNS: Tuple[str, ...] = ("stock_code", "adjustment", "byte_hash")
_FILES_STRING_COLUMNS: Tuple[str, ...] = (
    "file_id",
    "stock_code",
    "adjustment",
    "byte_hash",
    "schema_fingerprint",
    "requested_stock_code",
    "requested_adjustment",
)

# The three enum columns and their pinned domains with their NA policy; the
# mapping is read by assert_enum_domains_task_2. The bool flags name whether
# pd.NA is admitted (True) or forbidden (False).
_ENUM_DOMAINS: Dict[str, Tuple[frozenset, bool]] = {
    # adjustment: exactly the two views, NA forbidden (a row without a view is a halt).
    "adjustment": (frozenset({"unadjusted", "provider_adjusted"}), False),
    # retention_status: the two write dispositions, NA admitted (not yet classified).
    "retention_status": (frozenset({"retained_in_catalog", "dropped_before_write"}), True),
    # parse_status: the two parse outcomes, NA admitted (not yet parsed).
    "parse_status": (frozenset({"parsed", "parse_failed"}), True),
}

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Internal helpers (not part of the public contract; lexicographic "_")
# -------------------------------------------------------------------------------------------------------------------------------

def _require_series_task_2(series: Any, caller: str) -> pd.Series:
    """
    Validate that an argument is a pandas Series.

    Parameters
    ----------
    series : Any
        Object to validate.
    caller : str
        Calling callable name, included in the diagnostic.

    Returns
    -------
    pd.Series
        The validated series, unchanged.

    Raises
    ------
    TypeError
        If ``series`` is not a ``pandas.Series``.
    """
    # Reject non-Series objects at the boundary so downstream dtype logic is safe.
    if not isinstance(series, pd.Series):
        # Raise an informative type error naming the caller and the actual type.
        raise TypeError(
            f"{caller}: expected a pandas.Series, received {type(series).__name__}"
        )
    # Return the validated series unchanged.
    return series

def _is_missing_cell_task_2(value: Any) -> bool:
    """
    Test a scalar cell for missingness.

    Return ``True`` for ``None`` and for scalars that ``pd.isna`` marks
    missing, including ``pd.NA``, ``NaN`` and ``NaT``. If ``pd.isna``
    raises ``TypeError`` or ``ValueError`` for a non-scalar value,
    return ``False``.

    Parameters
    ----------
    value : Any
        Scalar cell to test.

    Returns
    -------
    bool
        ``True`` iff the cell carries a missing-value marker.
    """
    # A plain None is missing without ceremony.
    if value is None:
        # Report the missing marker.
        return True
    # Any failure of the vectorised probe is conservatively "not missing".
    try:
        # pd.isna handles pd.NA, NaN, NaT and None uniformly for scalars.
        result = pd.isna(value)
    except (TypeError, ValueError):
        # Report "not missing" so the caller's coercion raises a typed error.
        return False
    # Coerce the probe's possibly numpy bool to a plain Python bool.
    return bool(result)

def _is_bool_scalar_task_2(value: Any) -> bool:
    """
    Test whether a scalar is exactly a Python ``bool``.

    Uses type identity, so ``numpy.bool_`` and ``bool`` subclasses are
    rejected.

    Parameters
    ----------
    value : Any
        Scalar cell to test.

    Returns
    -------
    bool
        ``True`` iff ``value`` is exactly a Python ``bool``.
    """
    # type() identity, not isinstance: subclasses and np.bool_ are rejected.
    return type(value) is bool

def _is_numeric_scalar_task_2(value: Any) -> bool:
    """
    Test whether a scalar is a real number and not a boolean.

    Parameters
    ----------
    value : Any
        Scalar cell to test.

    Returns
    -------
    bool
        ``True`` iff ``value`` is a ``numbers.Real`` instance and not a
        ``bool``.
    """
    # Booleans are excluded explicitly; they are not numeric sentinels.
    if _is_bool_scalar_task_2(value):
        # Not a numeric sentinel cell.
        return False
    # numbers.Real covers Python int/float and numpy integer/float scalars.
    return isinstance(value, Real) and not isinstance(value, bool)

def _diagnostic_stock_day_task_2(series: pd.Series, position: int) -> Tuple[str, str]:
    """
    Recover stock and date strings for a cell diagnostic.

    Read the index label at ``position``. For a ``MultiIndex``, extract
    levels named ``"stock_code"`` and ``"trade_date"``; missing levels
    yield ``"<unknown>"``. For a flat index, use the stringified label
    as the trade date and ``"<unknown>"`` as the stock code.

    Parameters
    ----------
    series : pd.Series
        Column whose index supplies the coordinates.
    position : int
        Positional offset of the offending cell.

    Returns
    -------
    Tuple[str, str]
        Pair ``(stock_code, trade_date)`` as diagnostic strings.
    """
    # Read the index label at the offending position.
    label = series.index[position]
    # Default coordinates for layouts that carry neither level.
    stock_code = "<unknown>"
    trade_date = "<unknown>"
    # A MultiIndex is destructured by level name; anything else has no stock_code.
    if isinstance(series.index, pd.MultiIndex):
        # Capture the level names once for the zip.
        names = list(series.index.names)
        # Normalise the label to a tuple for uniform destructuring.
        values = label if isinstance(label, tuple) else (label,)
        # Walk the levels once, mapping whichever coordinates exist.
        for name, value in zip(names, values):
            # A level named stock_code supplies the stock code.
            if name == "stock_code":
                # Stringify the code for the diagnostic message.
                stock_code = str(value)
            # A level named trade_date supplies the trade date.
            elif name == "trade_date":
                # Stringify the date for the diagnostic message.
                trade_date = str(value)
    # A flat index cannot name a stock code; surface the label as the date slot.
    else:
        # Stringify the flat label into the trade_date coordinate.
        trade_date = str(label)
    # Return the recovered coordinate pair.
    return stock_code, trade_date

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 1: per-column transformers (copy in, copy out, never in place)
# -------------------------------------------------------------------------------------------------------------------------------

def cast_integral_int64_task_2(series: pd.Series) -> pd.Series:
    """
    Cast an integer-terminal column to nullable Int64.

    Validate the input, reject boolean dtypes, and scan each cell.
    Missing cells become ``pd.NA``. Python bools, non-numeric cells and
    non-finite values halt with a four-part diagnostic naming
    ``(stock_code, trade_date, column, value)``.

    For each finite value ``v``, cast only when both predicates hold:

        Decimal(str(v)).to_integral_value() == Decimal(str(v))
        2**-52 * abs(v) < 1e-6

    Predicate (i) is evaluated on the user-visible decimal
    representation ``Decimal(str(v))``; predicate (ii) is evaluated on
    the coerced Python float as ``(2.0 ** -52) * abs(v) < 1e-6``. A
    violation raises ``ValueError`` with both predicate results and the
    bound value. Accepted values are appended as
    ``int(Decimal(str(v)))``.

    The returned Series has the same index and name, dtype ``Int64``,
    copied ``attrs``, and ``attrs["ledger_tag"] == "[R]"`` recording the
    ``Decimal(str(v))`` choice. The input Series is never mutated.

    Parameters
    ----------
    series : pd.Series
        Integer-terminal column, such as ``volume``, ``quantity`` or
        ``tradestatus_raw``, holding float-like values or nulls.

    Returns
    -------
    pd.Series
        New nullable ``Int64`` Series with the same index and name.

    Raises
    ------
    TypeError
        If ``series`` is not a Series, its dtype is boolean, a cell is a
        Python bool, or a cell cannot be coerced to ``float``.
    ValueError
        If a coerced value is not finite or fails either integrality
        predicate.
    """
    # Validate the boundary argument type.
    series = _require_series_task_2(series, "cast_integral_int64_task_2")
    # Resolve the column name for diagnostics; fall back to a placeholder.
    column = series.name if series.name is not None else "<unnamed>"
    # A numpy/pandas bool-typed column is a contract violation, not an integer terminal.
    if pdt.is_bool_dtype(series.dtype):
        # Raise with the column named.
        raise TypeError(
            f"integer-terminal column '{column}' carries a boolean dtype; "
            f"the ingest contract ingests integer terminals as float64"
        )
    # Output cells accumulate here; pd.NA marks the preserved nulls.
    cells: List[Any] = []
    # Walk every cell once; positional index supplies diagnostic coordinates.
    for position, raw in enumerate(series):
        # Preserve a null exactly as null; never substitute zero.
        if _is_missing_cell_task_2(raw):
            # Append the null and move on.
            cells.append(pd.NA)
            continue
        # Python bools are rejected outright, before any coercion can hide them.
        if _is_bool_scalar_task_2(raw):
            # Recover the coordinates for the diagnostic.
            stock_code, trade_date = _diagnostic_stock_day_task_2(series, position)
            # Raise the four-part diagnostic naming (stock_code, trade_date, column, value).
            raise TypeError(
                f"integer-terminal column '{column}': (stock_code={stock_code}, "
                f"trade_date={trade_date}, column={column}, value={raw!r}) is a bool; "
                f"integer terminals must be real numbers"
            )
        # Coerce the cell to a Python float exactly once; all predicates read v.
        try:
            # float(raw) is exact for float64 ingest cells and raises on garbage.
            v = float(raw)
        except (TypeError, ValueError, OverflowError) as exc:
            # Recover the coordinates for the diagnostic.
            stock_code, trade_date = _diagnostic_stock_day_task_2(series, position)
            # Raise with the offending cell type and value.
            raise TypeError(
                f"integer-terminal column '{column}': (stock_code={stock_code}, "
                f"trade_date={trade_date}, column={column}, value={raw!r}) is not "
                f"numerically coercible to float64 ({exc})"
            ) from exc
        # A non-finite value (inf/nan entering through an object column) halts.
        if not math.isfinite(v):
            # Recover the coordinates for the diagnostic.
            stock_code, trade_date = _diagnostic_stock_day_task_2(series, position)
            # Raise the four-part diagnostic; NaN must already have been routed above.
            raise ValueError(
                f"integer-terminal column '{column}': (stock_code={stock_code}, "
                f"trade_date={trade_date}, column={column}, value={v}) is not finite"
            )
        # Term i of the conjunction: Decimal(str(v)) is the user-visible decimal,
        # never Decimal(v); the prior register forbids the binary-fraction artefact.
        d = Decimal(str(v))
        # Condition i: the decimal representation must be an integral value.
        cond_integral = d.to_integral_value() == d
        # Condition ii: the 2**-52 safety bound keeps the first test within
        # representable range; for large |v| the bound alone already halts.
        cond_bound = (2.0 ** -52) * abs(v) < 1e-6
        # Both terms must hold for the value to be cast to Int64.
        if not (cond_integral and cond_bound):
            # Recover the coordinates for the diagnostic.
            stock_code, trade_date = _diagnostic_stock_day_task_2(series, position)
            # Raise the four-part diagnostic naming all four coordinates; str(v)
            # reproduces the user-visible value exactly as the caller supplied it.
            raise ValueError(
                f"integer-terminal integrality-assertion violation: (stock_code={stock_code}, "
                f"trade_date={trade_date}, column={column}, value={v}); "
                f"Decimal(str(v)).to_integral_value() == Decimal(str(v)) -> {cond_integral}; "
                f"2**-52*abs(v) < 1e-6 -> {cond_bound} "
                f"(2**-52*abs(v) = {(2.0 ** -52) * abs(v):.3e})"
            )
        # The value is integral and within the safety bound; cast via the Decimal
        # integral value to avoid any binary-rounding artefact at the boundary.
        cells.append(int(d))
    # Build the result with an identical index, name, and Int64 dtype.
    result = pd.Series(cells, index=series.index.copy(deep=True), dtype="Int64", name=series.name)
    # Preserve attrs from the source, then stamp the [R] ledger tag.
    result.attrs = dict(getattr(series, "attrs", {}) or {})
    # Record the register-pinned choice of Decimal(str(v)) over Decimal(v).
    result.attrs["ledger_tag"] = "[R]"
    # Return the fresh Int64 series.
    return result

def cast_trade_date_task_2(series: pd.Series) -> pd.Series:
    """
    Cast an exchange-local temporal column to naive datetime64[ns].

    A ``StringDtype`` column halts. Object columns are scanned first:
    any string cell halts, and so does any cell that is not a timestamp
    or date. The remaining cells are converted with ``pd.to_datetime``;
    a conversion that yields a timezone-aware dtype halts at the
    explicit timezone check (an all-aware object column), and a mixed
    naive/aware object column halts with the ``ValueError`` raised by
    ``pd.to_datetime`` itself. A naive ``datetime64`` column is cast
    directly. A timezone-aware ``datetime64`` column or any other dtype
    halts. The returned Series has the same name and copied ``attrs``;
    the input is never mutated.

    Parameters
    ----------
    series : pd.Series
        Temporal column using the naive exchange-local convention.
        Accepted inputs are naive ``datetime64[ns]``, other naive
        datetime64 units, or object columns of naive timestamps and
        dates.

    Returns
    -------
    pd.Series
        New ``datetime64[ns]`` Series without timezone.

    Raises
    ------
    TypeError
        If ``series`` is not a Series, an object cell is not a timestamp
        or date, or the dtype is unsupported.
    ValueError
        If the column has ``StringDtype``, is an object column carrying
        string cell(s), is a timezone-aware ``datetime64`` dtype, or is
        an object column whose ``pd.to_datetime`` conversion yields a
        timezone-aware result (all-aware cells) or raises on mixed
        naive/aware cells.
    """
    # Validate the boundary argument type.
    series = _require_series_task_2(series, "cast_trade_date_task_2")
    # Resolve the column name for diagnostics.
    column = series.name if series.name is not None else "<unnamed>"
    # Read the dtype once for the branch decision.
    dtype = series.dtype
    # A StringDtype column (string[pyarrow] / string) is a string-typed temporal column: halt.
    if isinstance(dtype, pd.StringDtype):
        # The task text pins this halt; parsing at ingest already happened or did not.
        raise ValueError(
            f"string-typed column '{column}' is a temporal-halt: parsing at ingest is "
            f"the only permitted normalisation and the ingest step has already passed"
        )
    # An object column is inspected cell by cell: strings halt, timestamps convert.
    if dtype == object:
        # Materialise the probe cells once for the two membership scans.
        probe = list(series)
        # Any genuine str cell is a string-typed temporal column: halt.
        if any(type(cell) is str for cell in probe if not _is_missing_cell_task_2(cell)):
            # Raise the string-typed halt.
            raise ValueError(
                f"object column '{column}' carries string cell(s); a string-typed temporal column halts"
            )
        # Any cell that is not a Timestamp/datetime halts with its position named.
        for position, cell in enumerate(probe):
            # Skip missing cells.
            if _is_missing_cell_task_2(cell):
                # Nothing to validate.
                continue
            # Accept pandas.Timestamp and datetime.datetime/date cells only.
            if not isinstance(cell, (pd.Timestamp, datetime.datetime, date)):
                # Raise with the position and type for the diagnostic.
                raise TypeError(
                    f"exchange-local temporal column '{column}': cell at position {position} is "
                    f"{type(cell).__name__}, not a timestamp"
                )
        # Convert the object column to nanosecond datetimes; errors raise on garbage.
        converted = pd.to_datetime(series, errors="raise")
        # A converted-aware column means mixed naive/aware cells; refuse the ambiguity.
        if getattr(converted.dtype, "tz", None) is not None:
            # Raise the aware-in-naive-column halt.
            raise ValueError(
                f"exchange-local temporal column '{column}' mixes naÃ¯ve and tz-aware cells; "
                f"the exchange-local naive convention forbids silent zone coercion"
            )
        # Cast to the pinned nanosecond naive dtype.
        result = converted.astype("datetime64[ns]")
    # A tz-aware datetime64 column is refused: the pinned convention is naive.
    elif isinstance(dtype, pd.DatetimeTZDtype):
        # Raise rather than drop the zone silently.
        raise ValueError(
            f"exchange-local temporal column '{column}' carries dtype '{dtype}'; the pinned "
            f"convention is naive datetime64[ns] (exchange-local), and silently "
            f"dropping the zone is forbidden"
        )
    # A naive datetime64 column is the happy path: an exact nanosecond cast.
    elif pdt.is_datetime64_dtype(dtype):
        # astype is a pure dtype realignment; it copies.
        result = series.astype("datetime64[ns]").copy(deep=True)
    # Anything else (bool, integer, float, categorical) is a type halt.
    else:
        # Raise with the offending dtype named.
        raise TypeError(
            f"exchange-local temporal column '{column}' carries dtype '{dtype}'; "
            f"expected datetime64[ns] or an object column of timestamps"
        )
    # Pin the name for the ledger diagnostics downstream.
    result.name = series.name
    # Preserve attrs verbatim; no ledger tag is set by this cast.
    result.attrs = dict(getattr(series, "attrs", {}) or {})
    # Return the fresh naive datetime64[ns] series.
    return result

def cast_retrieved_task_2(series: pd.Series) -> pd.Series:
    """
    Cast a retrieved column to datetime64[ns, Asia/Shanghai].

    String-typed columns halt. For a timezone-aware ``datetime64``
    column, an Asia/Shanghai column is copied unchanged; any other zone
    is instant-preserving converted, and the decision is logged at INFO.
    For an object column, string and non-timestamp cells halt, mixed
    naive and aware cells halt, all-naive cells are localized at the
    boundary, and all-aware cells are converted through UTC. A naive
    ``datetime64`` column is localized to Asia/Shanghai. Other dtypes
    halt.

    Every localization or conversion is logged at INFO. The returned
    Series has the same name and copied ``attrs``; the input is never
    mutated.

    Parameters
    ----------
    series : pd.Series
        Retrieved-timestamp column, either timezone-aware, naive
        ``datetime64``, or an object column of timestamps.

    Returns
    -------
    pd.Series
        New ``datetime64[ns, Asia/Shanghai]`` Series.

    Raises
    ------
    TypeError
        If ``series`` is not a Series, an object cell is not a timestamp
        or date, or the dtype is unsupported.
    ValueError
        If the column is string-typed, contains a string cell, or mixes
        naive and timezone-aware cells.
    """
    # Validate the boundary argument type.
    series = _require_series_task_2(series, "cast_retrieved_task_2")
    # Resolve the column name for diagnostics and logging.
    column = series.name if series.name is not None else "<unnamed>"
    # Read the dtype once for the branch decision.
    dtype = series.dtype
    # A StringDtype column is a string-typed retrieved stamp: halt.
    if isinstance(dtype, pd.StringDtype):
        # Refuse mixed-type ingestion at the cast step.
        raise ValueError(
            f"string-typed column '{column}' is a retrieved-stamp halt; "
            f"timestamps are normalised at the ingest boundary only"
        )
    # A tz-aware datetime64 column: no-op (Shanghai) or instant-preserving convert.
    if isinstance(dtype, pd.DatetimeTZDtype):
        # Identify the source zone for the log record.
        source_zone = str(dtype.tz)
        # Already Asia/Shanghai: a true no-op, still ns-normalised.
        if source_zone == _ASIA_SHANGHAI_TZ:
            # Copy the series verbatim; astype re-pins the ns unit if needed.
            result = series.copy(deep=True).astype("datetime64[ns, Asia/Shanghai]")
            # Log the no-op at INFO so the cohort review can see the decision.
            logger.info(
                "cast_retrieved_task_2: column '%s' already Asia/Shanghai; no-op (%d rows)",
                column, len(result),
            )
        # Any other zone: convert preserving the instant.
        else:
            # tz_convert is instant-preserving by construction.
            result = series.dt.tz_convert(_ASIA_SHANGHAI_TZ).astype("datetime64[ns, Asia/Shanghai]")
            # Log the convert at INFO with the source zone named.
            logger.info(
                "cast_retrieved_task_2: column '%s' converted from %s to Asia/Shanghai, "
                "instant preserved (%d rows)",
                column, source_zone, len(result),
            )
    # An object column: classify every cell; naive attach or aware convert.
    elif dtype == object:
        # Materialise the probe cells once.
        probe = list(series)
        # A genuine str cell halts; timestamps only.
        if any(type(cell) is str for cell in probe if not _is_missing_cell_task_2(cell)):
            # Raise the string-typed halt.
            raise ValueError(
                f"object column '{column}' carries string cell(s); retrieved stamps must be timestamps"
            )
        # Validate the cell types before the coercion scan.
        for position, cell in enumerate(probe):
            # Skip missing cells.
            if _is_missing_cell_task_2(cell):
                # Nothing to validate.
                continue
            # Accept pandas.Timestamp / datetime / date cells only.
            if not isinstance(cell, (pd.Timestamp, datetime.datetime, date)):
                # Raise with the position and type for the diagnostic.
                raise TypeError(
                    f"retrieved column '{column}': cell at position {position} is "
                    f"{type(cell).__name__}, not a timestamp"
                )
        # Partition the non-missing cells into aware and naive buckets.
        aware = [c for c in probe if not _is_missing_cell_task_2(c)
                 and getattr(c, "tzinfo", None) is not None]
        naive = [c for c in probe if not _is_missing_cell_task_2(c)
                 and getattr(c, "tzinfo", None) is None]
        # A mix of aware and naive cells is ambiguous: halt rather than guess.
        if aware and naive:
            # Raise the mixed-basis halt.
            raise ValueError(
                f"retrieved column '{column}' mixes naÃ¯ve and tz-aware cells; "
                f"refusing to choose a boundary coercion silently"
            )
        # All-naive (the ingest fixture): attach the zone at the boundary.
        if naive:
            # to_datetime rebuilds the column; errors raise on garbage.
            attached = pd.to_datetime(series, errors="raise").dt.tz_localize(_ASIA_SHANGHAI_TZ)
            # Re-pin the ns unit and dtype spelling.
            result = attached.astype("datetime64[ns, Asia/Shanghai]")
            # Log the attach with the count of naÃ¯ve cells coerced.
            logger.info(
                "cast_retrieved_task_2: attached Asia/Shanghai to %d naÃ¯ve timestamp(s) "
                "in column '%s' at the boundary (wall-clock unchanged)",
                len(naive), column,
            )
        # All-aware: convert through UTC to normalise mixed source zones.
        else:
            # utc=True pivots every aware cell to UTC then converts to Shanghai.
            converted = pd.to_datetime(series, utc=True, errors="raise").dt.tz_convert(_ASIA_SHANGHAI_TZ)
            # Re-pin the ns unit and dtype spelling.
            result = converted.astype("datetime64[ns, Asia/Shanghai]")
            # Log the convert with the count of aware cells converted.
            logger.info(
                "cast_retrieved_task_2: converted %d tz-aware timestamp(s) in column '%s' "
                "to Asia/Shanghai preserving instants",
                len(aware), column,
            )
    # A naive datetime64 column: attach the zone at the boundary.
    elif pdt.is_datetime64_dtype(dtype):
        # Count the naÃ¯ve cells (series may carry NaT).
        n_naive = int((~series.isna()).sum())
        # tz_localize attaches without shifting the wall-clock.
        attached = series.dt.tz_localize(_ASIA_SHANGHAI_TZ)
        # Re-pin the ns unit and dtype spelling.
        result = attached.astype("datetime64[ns, Asia/Shanghai]")
        # Log the attach with the count of naÃ¯ve cells coerced.
        logger.info(
            "cast_retrieved_task_2: attached Asia/Shanghai to %d naÃ¯ve timestamp(s) "
            "in column '%s' at the boundary (wall-clock unchanged)",
            n_naive, column,
        )
    # Anything else is a type halt.
    else:
        # Raise with the offending dtype named.
        raise TypeError(
            f"retrieved column '{column}' carries dtype '{dtype}'; "
            f"expected datetime64[ns, Asia/Shanghai], datetime64[ns], or "
            f"an object column of timestamps"
        )
    # Pin the name for the ledger diagnostics downstream.
    result.name = series.name
    # Preserve attrs verbatim; no ledger tag is set by this cast.
    result.attrs = dict(getattr(series, "attrs", {}) or {})
    # Return the fresh Asia/Shanghai aware series.
    return result

def cast_boolean_pyarrow_task_2(series: pd.Series) -> pd.Series:
    """
    Cast a boolean column to nullable boolean[pyarrow].

    Object columns admit only Python bools and missing cells; missing
    cells become ``pd.NA`` and any other value raises ``ValueError``.
    Pandas nullable boolean and Arrow boolean[pyarrow] columns are
    re-wrapped. NumPy bool dtypes and all other dtypes raise
    ``TypeError``.

    The returned Series has the same index and name, dtype
    ``boolean[pyarrow]``, and copied ``attrs``; ``True``, ``False`` and
    ``pd.NA`` round-trip without coercing ``NA`` to ``False``.

    Parameters
    ----------
    series : pd.Series
        Column of boolean values, either object, pandas nullable
        boolean, or ``boolean[pyarrow]``.

    Returns
    -------
    pd.Series
        New nullable ``boolean[pyarrow]`` Series.

    Raises
    ------
    TypeError
        If ``series`` is not a Series, has a NumPy bool dtype, or has an
        unsupported dtype.
    ValueError
        If an object cell is not exactly ``True``, ``False`` or missing.
    """
    # Validate the boundary argument type.
    series = _require_series_task_2(series, "cast_boolean_pyarrow_task_2")
    # Resolve the column name for diagnostics.
    column = series.name if series.name is not None else "<unnamed>"
    # Read the dtype once for the branch decision.
    dtype = series.dtype
    # Accumulate the validated cells here; dtype construction happens at the end.
    cells: List[Any] = []
    # An object column is validated cell by cell: Python bools and NA only.
    if pdt.is_object_dtype(dtype):
        # Walk every cell once.
        for position, cell in enumerate(series):
            # Preserve a null exactly as null.
            if _is_missing_cell_task_2(cell):
                # Append pd.NA and move on.
                cells.append(pd.NA)
                continue
            # Exactly a Python bool passes; np.bool_ and ints do not.
            if _is_bool_scalar_task_2(cell):
                # Append the validated bool.
                cells.append(cell)
                continue
            # Anything else is an out-of-domain cell: halt with coordinates.
            raise ValueError(
                f"boolean column '{column}': out-of-domain cell at position {position} "
                f"(index={series.index[position]!r}): {cell!r} ({type(cell).__name__}); "
                f"only True, False and NA are admitted"
            )
    # A pandas nullable boolean column is already in-domain; it round-trips exactly.
    elif isinstance(dtype, pd.BooleanDtype):
        # Re-wrap the cells so the output carries the pyarrow storage.
        cells = list(series)
    # A boolean[pyarrow] column (ArrowDtype over pa.bool_) round-trips exactly.
    elif isinstance(dtype, pd.ArrowDtype) and str(dtype) == "bool[pyarrow]":
        # Re-wrap the cells so the output is freshly constructed.
        cells = list(series)
    # A numpy bool dtype is rejected at the dtype gate before any cell is read.
    elif pdt.is_bool_dtype(dtype):
        # Raise the numpy-bool rejection the step text pins.
        raise TypeError(
            f"boolean column '{column}' carries numpy bool dtype; the pinned "
            f"contract is nullable boolean[pyarrow], and bool is rejected"
        )
    # Anything else (integer/int8/float/datetime/string/category) is rejected.
    else:
        # Raise the dtype rejection with the offender named.
        raise TypeError(
            f"boolean column '{column}' carries dtype '{dtype}'; the pinned "
            f"contract admits only True/False/NA cells in object, pandas "
            f"boolean, or boolean[pyarrow] inputs"
        )
    # Construct the fresh nullable boolean[pyarrow] series.
    result = pd.Series(cells, index=series.index.copy(deep=True), dtype="boolean[pyarrow]", name=series.name)
    # Preserve attrs verbatim; no ledger tag is set by this cast.
    result.attrs = dict(getattr(series, "attrs", {}) or {})
    # Return the fresh boolean[pyarrow] series.
    return result

def cast_column_task_2(
    frame: pd.DataFrame,
    column: str,
    transformer: Callable[[pd.Series], pd.Series],
    ledger: List[Dict[str, Any]],
) -> pd.Series:
    """
    Apply a transformer to one column and append a ledger entry.

    Validate the frame, column name, ledger and transformer. Read
    ``frame[column]`` and pass it to ``transformer``. Require a Series
    of the same length. Append one entry to ``ledger`` with keys
    ``"column"``, ``"source_dtype"`` and ``"dtype"``; add
    ``"ledger_tag"`` when the returned Series carries
    ``attrs["ledger_tag"]``.

    Parameters
    ----------
    frame : pd.DataFrame
        Frame containing the column. It is not mutated.
    column : str
        Non-empty column name to cast.
    transformer : Callable[[pd.Series], pd.Series]
        Per-column transformer returning a same-length Series.
    ledger : List[Dict[str, Any]]
        Step-level cast ledger receiving exactly one entry per call.

    Returns
    -------
    pd.Series
        Cast column returned by the transformer; the caller reassigns
        it.

    Raises
    ------
    TypeError
        If ``frame`` is not a DataFrame, ``column`` is not a non-empty
        string, ``ledger`` is not a list, ``transformer`` is not
        callable, or the transformer returns a non-Series or
        wrong-length result.
    KeyError
        If ``column`` is absent from ``frame.columns``.
    """
    # Validate the frame argument.
    if not isinstance(frame, pd.DataFrame):
        # Raise with the actual type named.
        raise TypeError(
            f"cast_column_task_2: expected pandas.DataFrame, received {type(frame).__name__}"
        )
    # Validate the column name.
    if not isinstance(column, str) or not column:
        # Raise with the offending value named.
        raise TypeError(
            f"cast_column_task_2: column must be a non-empty str, received {column!r}"
        )
    # Validate the ledger container.
    if not isinstance(ledger, list):
        # Raise with the actual type named.
        raise TypeError(
            f"cast_column_task_2: ledger must be a list, received {type(ledger).__name__}"
        )
    # Validate the transformer callable.
    if not callable(transformer):
        # Raise with the actual type named.
        raise TypeError(
            f"cast_column_task_2: transformer must be callable, received {type(transformer).__name__}"
        )
    # Halt with a listing if the column is absent.
    if column not in frame.columns:
        # Raise the absence KeyError with the available columns named.
        raise KeyError(
            f"cast_column_task_2: column '{column}' is absent; "
            f"available columns: {list(frame.columns)}"
        )
    # Read the source column once; the view is never written through.
    source = frame[column]
    # Apply the transformer; its own diagnostics propagate unwrapped by design.
    casted = transformer(source)
    # The transformer must return a Series; anything else is a contract breach.
    if not isinstance(casted, pd.Series):
        # Raise with the actual return type named.
        raise TypeError(
            f"cast_column_task_2: transformer returned {type(casted).__name__}, "
            f"expected pandas.Series"
        )
    # The cast column must be the same length as the source column.
    if len(casted) != len(source):
        # Raise the length breach with both lengths named.
        raise TypeError(
            f"cast_column_task_2: transformer returned length {len(casted)}, "
            f"expected {len(source)}"
        )
    # Build the single ledger entry for this cast.
    entry: Dict[str, Any] = {
        # Name the column exactly as the caller addressed it.
        "column": column,
        # Record the source dtype spelling before the cast.
        "source_dtype": str(source.dtype),
        # Record the target dtype spelling after the cast.
        "dtype": str(casted.dtype),
    }
    # Extract the transformer's ledger tag, when it set one.
    attrs = getattr(casted, "attrs", {})
    # Only a dict-typed attrs bag is consulted.
    tag = attrs.get("ledger_tag") if isinstance(attrs, dict) else None
    # Attach the tag only when the transformer set one.
    if tag is not None:
        # Stringify the tag so the ledger is JSON-serialisable by default=str too.
        entry["ledger_tag"] = str(tag)
    # Append exactly one entry per call, as the ledger contract requires.
    ledger.append(entry)
    # Return the cast column to the caller, which reassigns it into its own frame.
    return casted

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 2: identity strings, stock codes and enum domains
# -------------------------------------------------------------------------------------------------------------------------------

def cast_string_pyarrow_task_2(series: pd.Series) -> pd.Series:
    """
    Cast a string column to string[pyarrow] after strict validation.

    Missing cells become ``pd.NA``. Only exact Python ``str`` cells are
    admitted; bytes, numbers, bools, timestamps and other objects raise
    ``TypeError`` rather than being stringified.

    The returned Series has the same index and name, dtype
    ``string[pyarrow]``, and copied ``attrs``; the input is never
    mutated.

    Parameters
    ----------
    series : pd.Series
        Identity or string column to validate and cast.

    Returns
    -------
    pd.Series
        New ``string[pyarrow]`` Series.

    Raises
    ------
    TypeError
        If ``series`` is not a Series or a non-null cell is not a
        Python ``str``.
    """
    # Validate the boundary argument type.
    series = _require_series_task_2(series, "cast_string_pyarrow_task_2")
    # Resolve the column name for diagnostics.
    column = series.name if series.name is not None else "<unnamed>"
    # Accumulate the validated cells here.
    cells: List[Any] = []
    # Walk every cell once, validating strictly.
    for position, cell in enumerate(series):
        # Preserve a null exactly as null.
        if _is_missing_cell_task_2(cell):
            # Append pd.NA and move on.
            cells.append(pd.NA)
            continue
        # Exactly a Python str passes; everything else halts.
        if type(cell) is str:
            # Append the validated string.
            cells.append(cell)
            continue
        # Raise with the position, index label, and cell type named.
        raise TypeError(
            f"string column '{column}': cell at position {position} "
            f"(index={series.index[position]!r}) is {type(cell).__name__} "
            f"{cell!r}; only Python str cells and NA are admitted"
        )
    # Construct the fresh string[pyarrow] series.
    result = pd.Series(cells, index=series.index.copy(deep=True), dtype="string[pyarrow]", name=series.name)
    # Preserve attrs verbatim; no ledger tag is set by this cast.
    result.attrs = dict(getattr(series, "attrs", {}) or {})
    # Return the fresh string[pyarrow] series.
    return result

def enforce_identity_strings_task_2(
    frame: pd.DataFrame,
    columns: Sequence[str],
) -> pd.DataFrame:
    """
    Assert and persist identity columns as ``string[pyarrow]``.

    Deep-copies ``frame``. It processes each named column in caller
    order: rejects ``category``, ``object`` and non-string dtypes;
    converts non-pyarrow string storage to ``string[pyarrow]``; then
    verifies that values and order are unchanged. No whitespace
    normalisation, sorting or categorical coercion is applied. The copy
    carries ``attrs`` over.

    Parameters
    ----------
    frame : pd.DataFrame
        Frame whose identity columns are asserted and persisted.
    columns : Sequence[str]
        Identity column names to process, for example ``stock_code``,
        ``adjustment``, ``file_id``, ``expression_id``,
        ``expression_text``, ``category_label``, ``run_id``,
        ``parse_status`` and ``retention_status``.

    Returns
    -------
    pd.DataFrame
        Deep copy of ``frame`` with the named columns persisted as
        ``string[pyarrow]``; values and row order are preserved.

    Raises
    ------
    TypeError
        If ``frame`` is not a DataFrame, ``columns`` is a bare string or
        not a ``Sequence``, or a column name is not a non-empty string.
    KeyError
        If a named column is absent from ``frame``.
    ValueError
        If a named column has ``category`` dtype, ``object`` dtype, or
        any non-``StringDtype``.
    RuntimeError
        If persisting a column changes its values or order.

    Notes
    -----
    Categoricals are forbidden because they silently reorder and merge
    levels; object dtype is forbidden because it admits non-string
    cells. These checks protect the ordered-universe guarantee.
    """
    # Validate the frame argument.
    if not isinstance(frame, pd.DataFrame):
        # Raise with the actual type named.
        raise TypeError(
            f"enforce_identity_strings_task_2: expected pandas.DataFrame, "
            f"received {type(frame).__name__}"
        )
    # A bare string is a common footgun for Sequence[str]; reject it explicitly.
    if isinstance(columns, str) or not isinstance(columns, Sequence):
        # Raise with the actual type named.
        raise TypeError(
            f"enforce_identity_strings_task_2: columns must be a Sequence[str], "
            f"received {type(columns).__name__}"
        )
    # Deep-copy the frame; the input is never mutated.
    out = frame.copy(deep=True)
    # Process every named column in caller order.
    for col in columns:
        # Each column name must be a str.
        if not isinstance(col, str) or not col:
            # Raise with the offending value named.
            raise TypeError(
                f"enforce_identity_strings_task_2: column names must be non-empty "
                f"strs, received {col!r}"
            )
        # Halt with a listing if the column is absent.
        if col not in out.columns:
            # Raise the absence KeyError with the available columns named.
            raise KeyError(
                f"enforce_identity_strings_task_2: identity column '{col}' is absent; "
                f"available columns: {list(out.columns)}"
            )
        # Snapshot the source values once for the byte-identity verification.
        source_values = out[col].tolist()
        # Read the column dtype for the assertion ladder.
        dtype = out[col].dtype
        # A categorical identity column halts: silent reordering / merged levels.
        if isinstance(dtype, pd.CategoricalDtype):
            # Raise the category halt with the column named.
            raise ValueError(
                f"identity column '{col}' carries category dtype; categoricals "
                f"silently reorder and merge levels and are forbidden in the "
                f"identity columns"
            )
        # An object identity column halts: integer cells can sneak through.
        if pdt.is_object_dtype(dtype):
            # Raise the object halt with the remediation named.
            raise ValueError(
                f"identity column '{col}' carries object dtype; object admits "
                f"non-string cells and is forbidden — cast to string[pyarrow] at "
                f"the ingest boundary before this assertion"
            )
        # Any non-StringDtype column (integer, float, bool, datetime) halts.
        if not isinstance(dtype, pd.StringDtype):
            # Raise the dtype halt with the offender named.
            raise ValueError(
                f"identity column '{col}' carries dtype '{dtype}'; expected "
                f"string[pyarrow] (or another pandas StringDtype)"
            )
        # Persist pyarrow storage; a python-backed StringDtype is converted losslessly.
        if dtype.storage != "pyarrow":
            # astype is value- and order-preserving for string storage swaps.
            out[col] = out[col].astype("string[pyarrow]")
        # Verify the persistence step held the values byte-identically.
        if out[col].tolist() != source_values:
            # Raise the byte-identity breach; this must never happen in practice.
            raise RuntimeError(
                f"identity column '{col}' changed value under string[pyarrow] "
                f"persistence; refusing to pass a mutated column onward"
            )
    # Carry attrs through verbatim.
    out.attrs = dict(getattr(frame, "attrs", {}) or {})
    # Return the persisted copy.
    return out

def _classify_stock_code_cell_task_2(cell: Any) -> Tuple[str, bool]:
    """
    Classify one ``stock_code`` cell for regex tallying.

    Missing cells return ``("<NA>", True)``. Non-string, non-missing
    cells return a typed token ``"<Type:repr>"`` and ``True``. String
    cells are checked with a full match against the pinned BaoStock
    pattern; a conforming code returns ``(cell, False)`` and a
    non-conforming string returns ``(cell, True)``.

    Parameters
    ----------
    cell : Any
        Raw cell from a ``stock_code`` column.

    Returns
    -------
    Tuple[str, bool]
        Tally key and offender flag; ``True`` means the cell violates
        the stock-code convention.
    """
    # A missing cell is its own offender class.
    if _is_missing_cell_task_2(cell):
        # Return the stable NA token and mark it as an offender.
        return "<NA>", True
    # A non-str cell is an offender with a typed repr.
    if not isinstance(cell, str):
        # Return the typed token and mark it as an offender.
        return f"<{type(cell).__name__}:{cell!r}>", True
    # fullmatch pins both ends; search would let suffix junk through.
    if _STOCK_CODE_PATTERN.fullmatch(cell) is None:
        # The cell is a string offender; it keeps its own literal for the listing.
        return cell, True
    # A conforming code is admitted.
    return cell, False

def assert_stock_codes_task_2(
    frames: Mapping[str, pd.DataFrame],
) -> List[str]:
    """
    Assert BaoStock 0.9.3 stock-code format across frames.

    Validates the mapping, walks every frame that carries a
    ``stock_code`` column, and classifies each cell with the per-cell
    helper. At least one participating frame is required. Offenders are
    tallied with ``Counter`` and halt with an itemised listing;
    otherwise the deduplicated admitted codes are returned in first-seen
    order. No sorting is applied.

    Parameters
    ----------
    frames : Mapping[str, pd.DataFrame]
        Mapping of frame name to frame, typically ``FILES``,
        ``UNADJUSTED_BARS`` and ``ADJUSTED_BARS``. Frames without
        ``stock_code`` are skipped.

    Returns
    -------
    List[str]
        Deduplicated admitted codes in first-seen order; this is the
        artifact later tasks re-assert.

    Raises
    ------
    TypeError
        If ``frames`` is not a ``Mapping`` or a mapped value is not a
        DataFrame.
    ValueError
        If no participating frame carries ``stock_code``, or any
        ``stock_code`` cell is missing, is non-string, or does not
        fullmatch the regex ``^(sh|sz)[.][0-9]{6}$``; the message lists
        every offender with a count.

    Notes
    -----
    The regex is pinned by the replication register. The assertion runs
    on the deduplicated set, not on every row.
    """
    # Validate the mapping argument.
    if not isinstance(frames, Mapping):
        # Raise with the actual type named.
        raise TypeError(
            f"assert_stock_codes_task_2: expected a Mapping[str, DataFrame], "
            f"received {type(frames).__name__}"
        )
    # Track whether at least one frame carried the column.
    any_stock_code_column = False
    # Collect every classified cell, in (frame, cell) order, for the Counter tally.
    cells: List[str] = []
    # Collect the keys already known to offend at classification time.
    offender_keys = set()
    # Walk every frame once; each participating frame contributes its cells.
    for name, frame in frames.items():
        # Every value must be a DataFrame; name it in the error if not.
        if not isinstance(frame, pd.DataFrame):
            # Raise with the offending frame name and type.
            raise TypeError(
                f"assert_stock_codes_task_2: frame '{name}' is {type(frame).__name__}, "
                f"expected pandas.DataFrame"
            )
        # Frames without a stock_code column are silently out of scope.
        if "stock_code" not in frame.columns:
            # Skip this frame; a different frame must satisfy the assertion.
            continue
        # Mark the contract as engaged.
        any_stock_code_column = True
        # Walk every cell of this frame's column once, classifying and tallying later.
        for cell in frame["stock_code"]:
            # Classify the cell exactly once for both the tally and the verdict.
            key, is_offender = _classify_stock_code_cell_task_2(cell)
            # Append the key to the tally multiset.
            cells.append(key)
            # A cell known to offend marks its key for the itemised halt.
            if is_offender:
                offender_keys.add(key)
    # A mapping with no participating frame cannot satisfy the assertion.
    if not any_stock_code_column:
        # Raise the structural halt.
        raise ValueError(
            "assert_stock_codes_task_2: no frame in the mapping carries a "
            "'stock_code' column; at least one of FILES/UNADJUSTED_BARS/ADJUSTED_BARS is required"
        )
    # The tally multiset: Counter keeps first-seen insertion order and exact counts.
    counts = Counter(cells)
    # The deduplicated set against which the regex is evaluated, in first-seen order.
    unique_codes = list(counts)
    # Offenders were collected at classification time; tally them from the Counter.
    offenders = {code: counts[code] for code in unique_codes if code in offender_keys}
    # Any offender halts with an itemised listing.
    if offenders:
        # Render the itemised listing deterministically (insertion order).
        listing = ", ".join(f"{code} (x{n})" for code, n in offenders.items())
        # Raise the halt; every offender literal appears verbatim in the message.
        raise ValueError(
            f"unformatted stock_code(s) under the BaoStock 0.9.3 convention "
            f"^(sh|sz)\\.\\d{{6}}$ [R]: {listing}"
        )
    # Return the deduplicated admitted artifact; no sorting applied.
    return unique_codes

def assert_enum_domains_task_2(frame: pd.DataFrame) -> pd.DataFrame:
    """
    Assert pinned enum domains on enum-bearing columns.

    Deep-copies ``frame`` and checks each enum column present in the
    pinned mapping: ``adjustment``, ``retention_status`` and
    ``parse_status``. Missing cells are offenders only where ``pd.NA``
    is forbidden; non-string cells and out-of-domain strings are
    offenders. Any offender halts with an itemised count. Values are
    never rewritten.

    Parameters
    ----------
    frame : pd.DataFrame
        Frame to check; only enum columns it carries are examined.

    Returns
    -------
    pd.DataFrame
        Pass-through deep copy with ``attrs`` preserved.

    Raises
    ------
    TypeError
        If ``frame`` is not a pandas DataFrame.
    ValueError
        If any enum cell is outside its domain; the message lists every
        offender with a count and states the NA policy.

    Notes
    -----
    ``adjustment`` is in {``unadjusted``, ``provider_adjusted``} with
    ``pd.NA`` forbidden. ``retention_status`` is in
    {``retained_in_catalog``, ``dropped_before_write``} plus ``pd.NA``.
    ``parse_status`` is in {``parsed``, ``parse_failed``} plus
    ``pd.NA``.
    """
    # Validate the frame argument.
    if not isinstance(frame, pd.DataFrame):
        # Raise with the actual type named.
        raise TypeError(
            f"assert_enum_domains_task_2: expected pandas.DataFrame, "
            f"received {type(frame).__name__}"
        )
    # Deep-copy the frame; values pass through unchanged.
    out = frame.copy(deep=True)
    # Check every enum column the frame carries, in the pinned mapping order.
    for column, (domain, na_allowed) in _ENUM_DOMAINS.items():
        # Frames without this enum column are out of scope by design.
        if column not in out.columns:
            # Skip to the next enum column.
            continue
        # Accumulate offending values verbatim; Counter tallies at the end.
        offending: List[str] = []
        # Walk every cell of the column once.
        for cell in out[column]:
            # A missing cell is legal only where the step admits pd.NA.
            if _is_missing_cell_task_2(cell):
                # NA in a no-NA column is an offender.
                if not na_allowed:
                    # Append the stable NA token.
                    offending.append("<NA>")
                continue
            # A non-str cell is an offender with a typed repr.
            if not isinstance(cell, str):
                # Append the typed offender token.
                offending.append(f"<{type(cell).__name__}:{cell!r}>")
                continue
            # An out-of-domain string is an offender.
            if cell not in domain:
                # Append the string offender verbatim.
                offending.append(cell)
        # Any offender halts with an itemised listing.
        if offending:
            offenders = Counter(offending)
            # Describe the NA policy in the diagnostic.
            na_policy = "pd.NA admitted" if na_allowed else "pd.NA forbidden"
            # Render the itemised listing deterministically (insertion order).
            listing = ", ".join(f"{value} (x{n})" for value, n in offenders.items())
            # Raise the halt; every offending literal appears verbatim.
            raise ValueError(
                f"enum-domain violation on column '{column}' "
                f"(domain {sorted(domain)}; {na_policy}): {listing}"
            )
    # Carry attrs through verbatim.
    out.attrs = dict(getattr(frame, "attrs", {}) or {})
    # Return the pass-through copy.
    return out

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3: scoped prohibited-sentinel scanner
# -------------------------------------------------------------------------------------------------------------------------------

def scan_prohibited_sentinels_task_2(
    frame: pd.DataFrame,
) -> Tuple[pd.DataFrame, bool]:
    """
    Scan every column once for prohibited sentinel values.

    For each column, reads the module-level allow-lists and records hits
    outside them. Numeric non-boolean, non-categorical columns use a
    float64 fast path: zero and -1 are checked unless whitelisted.
    Boolean and datetime columns are skipped. Other columns use a
    cell-wise path: strings ``""``, ``"NA_FILL"`` and ``"n/a"`` are
    checked against their allow-lists, and numeric object scalars are
    checked for zero and -1. Missing cells are skipped. No exception is
    raised for hits.

    Parameters
    ----------
    frame : pd.DataFrame
        Frame to scan; it is not modified, sorted, filtered or
        deduplicated.

    Returns
    -------
    hits : pd.DataFrame
        One row per hit with columns ``column``, ``sentinel`` and
        ``value``; the index is ``frame.index.take`` for the offending
        positions.
    clean : bool
        ``True`` iff no hit was recorded.

    Raises
    ------
    TypeError
        If ``frame`` is not a pandas DataFrame.

    Notes
    -----
    Sentinel labels are ``zero``, ``minus_one``, ``empty_string``,
    ``NA_FILL`` and ``n_a``. Allow-lists make zero legitimate for
    ``volume``, ``amount``, ``tradestatus_raw``, ``is_fitness``,
    ``close_adj`` and ``quantity``, and make the three ASCII strings
    legitimate in ``expression_text``. Any ``-1`` is a hit.
    """
    # Validate the frame argument.
    if not isinstance(frame, pd.DataFrame):
        # Raise with the actual type named.
        raise TypeError(
            f"scan_prohibited_sentinels_task_2: expected pandas.DataFrame, "
            f"received {type(frame).__name__}"
        )
    # One record per hit: (position, column, sentinel, value).
    hit_positions: List[int] = []
    hit_rows: List[Dict[str, Any]] = []
    # Single pass over the columns, per the step text's best practice.
    for col in frame.columns:
        # Normalise the column name for allow-list lookup.
        name = str(col)
        # Read the column and its dtype once.
        series = frame[col]
        dtype = series.dtype
        # The three string-sentinel allow-lists for this column.
        empty_ok = name in permitted_empty_string_columns
        nafill_ok = name in permitted_NA_FILL_columns
        na_ok = name in permitted_n_a_columns
        # The two numeric-sentinel allow-lists for this column.
        zero_ok = name in permitted_zero_columns
        minus_one_ok = name in permitted_negative_one_columns
        # A numeric column (excluding bools and categoricals) takes the fast path.
        numeric_ok = (
            pdt.is_numeric_dtype(dtype)
            and not pdt.is_bool_dtype(dtype)
            and not isinstance(dtype, pd.CategoricalDtype)
        )
        # Try the vectorised numeric comparison first; fall back to cells on failure.
        numeric: Optional[pd.Series] = None
        if numeric_ok:
            # astype("float64") routes Int64's pd.NA to NaN; it never raises here.
            try:
                # Build the float64 view of the column.
                numeric = series.astype("float64")
            except (TypeError, ValueError, OverflowError):
                # Fall back to the cell-wise path below rather than skip the column.
                numeric = None
        # Numeric fast path: compare against exactly 0.0 and exactly -1.0.
        if numeric is not None:
            # Only bother walking cells if at least one list forbids a value.
            if not (zero_ok and minus_one_ok):
                # Walk the float64 view once.
                for position, value in enumerate(numeric):
                    # NaN is the missing marker in float64; skip it.
                    if _is_missing_cell_task_2(value):
                        # Nothing to compare.
                        continue
                    # Normalise to a Python float for the exact comparisons.
                    fv = float(value)
                    # A zero outside the allow-list is a hit.
                    if not zero_ok and fv == 0.0:
                        # Record the hit with its position.
                        hit_positions.append(position)
                        hit_rows.append({"column": name, "sentinel": "zero", "value": fv})
                    # A -1 outside the allow-list is a hit (the list is pinned empty).
                    if not minus_one_ok and fv == -1.0:
                        # Record the hit with its position.
                        hit_positions.append(position)
                        hit_rows.append({"column": name, "sentinel": "minus_one", "value": fv})
            # Numeric columns carry no string sentinels; move to the next column.
            continue
        # Boolean columns carry no sentinel; True is not 1 here.
        if pdt.is_bool_dtype(dtype):
            # Skip without recording anything.
            continue
        # Datetime-like columns carry no sentinel in this register.
        if pdt.is_datetime64_any_dtype(dtype):
            # Skip without recording anything.
            continue
        # Cell-wise path: strings, objects, and fallback numerics.
        for position, value in enumerate(series):
            # Skip missing cells everywhere in this path.
            if _is_missing_cell_task_2(value):
                # Nothing to compare.
                continue
            # A genuine str cell carries the three ASCII sentinels.
            if isinstance(value, str):
                # The empty string outside its allow-list is a hit.
                if not empty_ok and value == "":
                    # Record the hit.
                    hit_positions.append(position)
                    hit_rows.append({"column": name, "sentinel": "empty_string", "value": value})
                # The NA_FILL string outside its allow-list is a hit.
                if not nafill_ok and value == "NA_FILL":
                    # Record the hit.
                    hit_positions.append(position)
                    hit_rows.append({"column": name, "sentinel": "NA_FILL", "value": value})
                # The n/a string outside its allow-list is a hit.
                if not na_ok and value == "n/a":
                    # Record the hit.
                    hit_positions.append(position)
                    hit_rows.append({"column": name, "sentinel": "n_a", "value": value})
                # Strings carry no numeric sentinels; move to the next cell.
                continue
            # An object cell carrying a real number is checked against the numerics.
            if _is_numeric_scalar_task_2(value) and not _is_bool_scalar_task_2(value):
                # Normalise to a Python float for the exact comparisons.
                fv = float(value)
                # A zero outside the allow-list is a hit.
                if not zero_ok and fv == 0.0:
                    # Record the hit.
                    hit_positions.append(position)
                    hit_rows.append({"column": name, "sentinel": "zero", "value": fv})
                # A -1 outside the allow-list is a hit.
                if not minus_one_ok and fv == -1.0:
                    # Record the hit.
                    hit_positions.append(position)
                    hit_rows.append({"column": name, "sentinel": "minus_one", "value": fv})
                # Numeric object cells carry no string sentinels; next cell.
                continue
            # Anything else (Timestamps, bytes, dates) carries no pinned sentinel.
    # Materialise the found-hits frame with the input's index structure preserved.
    if hit_rows:
        # frame.index.take preserves MultiIndex-ness and level names exactly.
        hits = pd.DataFrame(hit_rows, index=frame.index.take(hit_positions))
        # Enforce the column order of the public contract.
        hits = hits[["column", "sentinel", "value"]]
    else:
        # The empty hit set still carries the three contract columns.
        hits = pd.DataFrame(
            {
                # The hitting column name, empty here.
                "column": pd.Series([], dtype="object"),
                # The sentinel class, empty here.
                "sentinel": pd.Series([], dtype="object"),
                # The offending value, empty here.
                "value": pd.Series([], dtype="object"),
            }
        )
    # The clean flag is True exactly when no hit was recorded.
    clean = bool(hits.empty)
    # Return the (found-hits, clean) pair; nothing raises inside this scanner.
    return hits, clean

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Step 3: anti-imputation context and controller
# -------------------------------------------------------------------------------------------------------------------------------

@dataclass
class PipelineContextTask2:
    """
    Mutable orchestrator-owned state for Task 2.

    Holds cumulative anti-imputation counters and a witness log for
    gated transformations. The counters are not columns: writing a
    global count into every row would itself impute state.

    Attributes
    ----------
    forward_fill_operations : int
        Forward-fill operations recorded through the controller; must be
        zero at the end of cleansing.
    interpolation_operations : int
        Interpolation operations recorded through the controller; must
        be zero at the end of cleansing.
    stock_prefix_reductions : int
        Stock-prefix substitutions recorded through the controller; must
        be zero at the end of cleansing.
    witness_log : List[Tuple[str, str, Any]]
        Witness tuples ``(operation, column, stock_day)`` in record
        order.
    """
    # Count of forward-fill operations recorded via the controller.
    forward_fill_operations: int = 0
    # Count of interpolation operations recorded via the controller.
    interpolation_operations: int = 0
    # Count of stock-prefix substitutions recorded via the controller.
    stock_prefix_reductions: int = 0
    # Witness tuples, in record order, one per gated operation.
    witness_log: List[Tuple[str, str, Any]] = field(default_factory=list)

def make_pipeline_context_task_2() -> "PipelineContextTask2":
    """
    Return a new zeroed pipeline context.

    Constructs a ``PipelineContextTask2`` whose three anti-imputation
    counters start at zero and whose witness log starts empty.

    Returns
    -------
    "PipelineContextTask2"
        Fresh context for the orchestrator to pass to the controller
        functions.
    """
    # Fresh counters and an empty witness log; owned by the orchestrator.
    return PipelineContextTask2()

def _require_context_task_2(context: Any, caller: str) -> PipelineContextTask2:
    """
    Validate that a value is a ``PipelineContextTask2``.

    Parameters
    ----------
    context : Any
        Object to validate.
    caller : str
        Calling function name, used in the diagnostic.

    Returns
    -------
    PipelineContextTask2
        The validated context.

    Raises
    ------
    TypeError
        If ``context`` is not a ``PipelineContextTask2``; the message
        names ``caller`` and the actual type.
    """
    # Exact type check: duck-typed contexts are rejected on purpose.
    if not isinstance(context, PipelineContextTask2):
        # Raise with the caller and the actual type named.
        raise TypeError(
            f"{caller}: expected a PipelineContextTask2 built by "
            f"make_pipeline_context_task_2(), received {type(context).__name__}"
        )
    # Return the validated context.
    return context

def record_operation_task_2(
    context: Any,
    operation: str,
    column: str,
    stock_day: Any,
) -> None:
    """
    Record one gated transformation on the context.

    Validates the context, maps ``operation`` to its counter through
    ``_OPERATION_TO_COUNTER``, increments that counter by one, and
    appends ``(operation, column, stock_day)`` to the witness log.

    Parameters
    ----------
    context : Any
        Orchestrator-owned ``PipelineContextTask2``.
    operation : str
        Sanctioned operation token: ``forward_fill``,
        ``interpolation`` or ``stock_prefix_reduction``; any other token
        halts.
    column : str
        Column touched by the operation; must be a str whose
        ``.strip()`` is non-empty (whitespace-only strings are
        rejected).
    stock_day : Any
        Stock-day coordinate, such as ``(stock_code, trade_date)`` or a
        timestamp.

    Raises
    ------
    TypeError
        If ``context`` is not a ``PipelineContextTask2`` or ``column``
        is not a str with non-empty ``.strip()``.
    ValueError
        If ``operation`` is not one of the sanctioned tokens.
    """
    # Validate the context object at the controller boundary.
    context = _require_context_task_2(context, "record_operation_task_2")
    # Validate the operation token against the sanctioned set.
    if operation not in _OPERATION_TO_COUNTER:
        # Raise with the valid tokens named.
        raise ValueError(
            f"record_operation_task_2: unsanctioned operation {operation!r}; "
            f"expected one of {sorted(_OPERATION_TO_COUNTER)}"
        )
    # Validate the column token.
    if not isinstance(column, str) or not column.strip():
        # Raise with the offending value named.
        raise TypeError(
            f"record_operation_task_2: column must be a non-empty str, "
            f"received {column!r}"
        )
    # Resolve the counter that this operation increments.
    counter = _OPERATION_TO_COUNTER[operation]
    # Increment exactly that counter; all three stay independent.
    setattr(context, counter, getattr(context, counter) + 1)
    # Append the witness tuple in record order.
    context.witness_log.append((operation, column, stock_day))

def check_anti_imputation_task_2(context: Any) -> None:
    """
    Enforce the zero-counter anti-imputation post-condition.

    Validates the context and collects every non-zero anti-imputation
    counter. If none are non-zero, returns normally. Otherwise raises
    ``RuntimeError`` naming the non-zero counters and every recorded
    witness tuple.

    Parameters
    ----------
    context : Any
        Orchestrator-owned ``PipelineContextTask2``.

    Raises
    ------
    TypeError
        If ``context`` is not a ``PipelineContextTask2``.
    RuntimeError
        If any of ``forward_fill_operations``,
        ``interpolation_operations`` or ``stock_prefix_reductions`` is
        non-zero. The message includes the non-zero counter names and
        the witness list.
    """
    # Validate the context object at the controller boundary.
    context = _require_context_task_2(context, "check_anti_imputation_task_2")
    # Name every non-zero counter, in a stable order.
    non_zero = [
        name
        for name in ("forward_fill_operations", "interpolation_operations", "stock_prefix_reductions")
        if getattr(context, name) != 0
    ]
    # The happy path returns None with no side effects.
    if not non_zero:
        # Post-condition satisfied.
        return None
    # Render every witness so the halt is actionable.
    witnesses = (
        "; ".join(
            f"(operation={operation}, column={column}, stock_day={stock_day})"
            for operation, column, stock_day in context.witness_log
        )
        if context.witness_log
        else "<no witnesses recorded>"
    )
    # Halt with the non-zero counters and the full witness list.
    raise RuntimeError(
        f"Anti-imputation post-condition violated: non-zero counter(s) "
        f"{', '.join(non_zero)}; witness list: {witnesses}"
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 2, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_2(
    FILES: pd.DataFrame,
    UNADJUSTED_BARS: pd.DataFrame,
    ADJUSTED_BARS: pd.DataFrame,
    STUDY_CONFIG: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Run the three ordered cleansing steps of Task 2.

    Validate the input types, deep-copy the three artefacts, and rebuild
    any numpy ``bool`` column among the legislated boolean targets as
    Python-native ``bool`` cells; the repair is injective and keeps
    order and index. Build the mutable anti-imputation context and cast
    ledger. Cast the bar panels: integer terminals to nullable ``Int64``
    under the Decimal conjunction, ``trade_date`` to naive
    ``datetime64[ns]``, bar booleans to ``boolean[pyarrow]`` and bar
    strings to ``string[pyarrow]``. Cast the manifest columns: naive
    datetimes to ``datetime64[ns]``, ``retrieved`` to ``datetime64[ns,
    Asia/Shanghai]``, booleans to ``boolean[pyarrow]`` and strings to
    ``string[pyarrow]``. Assert the BaoStock stock-code regex once over
    all frames, persist identity strings, and assert the enum domains.
    Run the scoped prohibited-sentinel scan on each frame, check the
    counters, and verify that every input index survives unchanged.
    Return the cleansed frames and ledger; the caller's frames are never
    mutated.

    Parameters
    ----------
    FILES : pd.DataFrame
        Acquisition manifest; cleansed on a deep copy, so the caller's
        frame is not mutated.
    UNADJUSTED_BARS : pd.DataFrame
        Unadjusted panel indexed by (stock_code, trade_date); cleansed
        on a deep copy.
    ADJUSTED_BARS : pd.DataFrame
        Provider-adjusted panel indexed by (stock_code, trade_date);
        cleansed on a deep copy and returned cleansed. It is cast,
        identity-string and enum asserted, and sentinel-scanned like the
        other frames; only its adjusted prices are reserved for the
        labels built by Task 4/5, not for features here.
    STUDY_CONFIG : Dict[str, Any]
        Study configuration carried through unmodified under the
        ``study_config`` key; not re-validated here.

    Returns
    -------
    Dict[str, Any]
        Cleansed bundle with keys:
        ``FILES``, ``UNADJUSTED_BARS``, ``ADJUSTED_BARS`` (cleansed
        copies); ``cast_ledger`` (list of cast records, one per cast in
        cast order; integer-terminal casts tagged ``[R]``);
        ``stock_codes`` (deduplicated stock codes in first-seen order);
        ``forward_fill_operations``, ``interpolation_operations``,
        ``stock_prefix_reductions`` (int, all zero);
        ``study_config`` (the input object, carried through by
        identity).

    Raises
    ------
    TypeError
        If any frame argument is not a ``pandas.DataFrame`` or
        ``STUDY_CONFIG`` is not a ``dict``.
    ValueError
        If a cast, stock-code assertion, enum-domain assertion or
        prohibited-sentinel scan fails, or a sentinel is found outside
        the allow-lists; the message names the offending values or
        cells.
    RuntimeError
        If an input index is not preserved by the cleansing.

    Notes
    -----
    The anti-imputation context records no operations, so all three
    counters remain zero and the witness log is empty. No forward-fill,
    interpolation, stock-prefix reduction, sorting, deduplication or
    re-indexing is performed; the input indexes are preserved exactly.
    """
    # ---- Boundary validation of the caller-supplied raw artefacts ----
    # Validate FILES is a DataFrame; anything else halts before a copy is made.
    if not isinstance(FILES, pd.DataFrame):
        # Raise with the actual type named.
        raise TypeError(
            f"run_task_2: FILES must be a pandas.DataFrame, received {type(FILES).__name__}"
        )
    # Validate UNADJUSTED_BARS is a DataFrame.
    if not isinstance(UNADJUSTED_BARS, pd.DataFrame):
        # Raise with the actual type named.
        raise TypeError(
            f"run_task_2: UNADJUSTED_BARS must be a pandas.DataFrame, "
            f"received {type(UNADJUSTED_BARS).__name__}"
        )
    # Validate ADJUSTED_BARS is a DataFrame.
    if not isinstance(ADJUSTED_BARS, pd.DataFrame):
        # Raise with the actual type named.
        raise TypeError(
            f"run_task_2: ADJUSTED_BARS must be a pandas.DataFrame, "
            f"received {type(ADJUSTED_BARS).__name__}"
        )
    # Validate STUDY_CONFIG is a dict; it is carried through by identity.
    if not isinstance(STUDY_CONFIG, dict):
        # Raise with the actual type named.
        raise TypeError(
            f"run_task_2: STUDY_CONFIG must be a dict, received {type(STUDY_CONFIG).__name__}"
        )
    # ---- Step 1 precondition: work exclusively on deep copies ----
    # Deep-copy the manifest; the caller's object is never mutated.
    files = FILES.copy(deep=True)
    # Deep-copy the unadjusted panel.
    una = UNADJUSTED_BARS.copy(deep=True)
    # Deep-copy the adjusted panel.
    adj = ADJUSTED_BARS.copy(deep=True)
    # Snapshot the input indexes for the preservation post-condition.
    input_indexes = {
        "FILES": files.index.copy(deep=True),
        "UNADJUSTED_BARS": una.index.copy(deep=True),
        "ADJUSTED_BARS": adj.index.copy(deep=True),
    }
    # ---- Step 1 precondition: value-preserving representation repair ----
    # A pd.DataFrame built from a list of dicts infers numpy bool for all-bool
    # columns (e.g. the FILES manifest booleans supplied at the ingest boundary).
    # The boolean contract transformer rejects numpy bool BY DESIGN (the step
    # text pins that rejection), so a dict-ingested raw artefact would halt here
    # for a representation, not a value, reason. The repair below is total and
    # injective — every np.bool_ maps to exactly one Python bool, None or NA are
    # absent from a numpy bool column, and NO value is manufactured, dropped,
    # negated or reordered. The repaired column is then validated by the very
    # same contract path as every other boolean column, so no leniency is
    # introduced: every cell still round-trips through cast_boolean_pyarrow_task_2.
    for frame in (una, adj, files):
        # Walk the legislated boolean targets of this frame.
        for col in frame.columns:
            # The unsafe case is exactly the numpy bool dtype (name == 'bool').
            if getattr(frame[col].dtype, "name", None) == "bool" and (
                col in _BAR_BOOLEAN_COLUMNS or col in _FILES_BOOLEAN_COLUMNS
            ):
                # Rebuild the column with Python-native bool cells; order and
                # index are preserved exactly.
                repaired = [bool(v) for v in frame[col]]
                # Reassign the repaired object-dtype column into the copy.
                frame[col] = pd.Series(
                    repaired, index=frame[col].index, name=col, dtype=object
                )
    # ---- Orchestrator-owned anti-imputation context and cast ledger ----
    # Build the mutable context; all counters start at zero.
    context = make_pipeline_context_task_2()
    # The ledger accumulates exactly one dict per cast, in cast order.
    cast_ledger: List[Dict[str, Any]] = []
    # Task 2, Step 1: integer terminals, temporal columns, booleans
    # Walk the two bar panels; the column lists below name the legislated cells.
    for view_name, frame in (("UNADJUSTED_BARS", una), ("ADJUSTED_BARS", adj)):
        # The three integer terminals cast to nullable Int64 under the Decimal conjunction.
        for col in _INTEGER_TERMINAL_COLUMNS:
            # Skip absent columns (the adjusted view carries no volume).
            if col not in frame.columns:
                # Nothing to cast in this view.
                continue
            # Cast through the primitive; the ledger records the [R] tag.
            frame[col] = cast_column_task_2(frame, col, cast_integral_int64_task_2, cast_ledger)
        # trade_date casts to naive datetime64[ns].
        if "trade_date" in frame.columns:
            # Cast through the primitive.
            frame["trade_date"] = cast_column_task_2(frame, "trade_date", cast_trade_date_task_2, cast_ledger)
        # tradable casts to nullable boolean[pyarrow].
        for col in _BAR_BOOLEAN_COLUMNS:
            # Skip absent columns.
            if col not in frame.columns:
                # Nothing to cast in this view.
                continue
            # Cast through the primitive.
            frame[col] = cast_column_task_2(frame, col, cast_boolean_pyarrow_task_2, cast_ledger)
        # The bar-view string columns persist as string[pyarrow] at this boundary.
        for col in _BAR_STRING_COLUMNS:
            # Skip absent columns.
            if col not in frame.columns:
                # Nothing to cast in this view.
                continue
            # Cast through the primitive.
            frame[col] = cast_column_task_2(frame, col, cast_string_pyarrow_task_2, cast_ledger)
    # The FILES naive datetime columns cast to datetime64[ns].
    for col in _FILES_NAIVE_DATETIME_COLUMNS:
        # Skip absent columns.
        if col not in files.columns:
            # Nothing to cast.
            continue
        # Reuse the naive exchange-local temporal cast (see its docstring Notes).
        files[col] = cast_column_task_2(files, col, cast_trade_date_task_2, cast_ledger)
    # The retrieved stamp casts to datetime64[ns, Asia/Shanghai] with the coercion logged.
    if "retrieved" in files.columns:
        # Cast through the primitive.
        files["retrieved"] = cast_column_task_2(files, "retrieved", cast_retrieved_task_2, cast_ledger)
    # The FILES boolean columns cast to nullable boolean[pyarrow].
    for col in _FILES_BOOLEAN_COLUMNS:
        # Skip absent columns.
        if col not in files.columns:
            # Nothing to cast.
            continue
        # Cast through the primitive.
        files[col] = cast_column_task_2(files, col, cast_boolean_pyarrow_task_2, cast_ledger)
    # The FILES string columns persist as string[pyarrow] at this boundary.
    for col in _FILES_STRING_COLUMNS:
        # Skip absent columns.
        if col not in files.columns:
            # Nothing to cast.
            continue
        # Cast through the primitive.
        files[col] = cast_column_task_2(files, col, cast_string_pyarrow_task_2, cast_ledger)
    # Task 2, Step 2: identity strings, stock codes, enum domains
    # Assert the BaoStock regex once over the deduplicated union artifact.
    stock_codes = assert_stock_codes_task_2(
        {"FILES": files, "UNADJUSTED_BARS": una, "ADJUSTED_BARS": adj}
    )
    # Persist the identity strings in each frame (values byte-identical).
    files = enforce_identity_strings_task_2(files, [c for c in _FILES_STRING_COLUMNS if c in files.columns])
    una = enforce_identity_strings_task_2(una, [c for c in _BAR_STRING_COLUMNS if c in una.columns])
    adj = enforce_identity_strings_task_2(adj, [c for c in _BAR_STRING_COLUMNS if c in adj.columns])
    # Assert the enum domains on whichever enum columns each frame carries.
    files = assert_enum_domains_task_2(files)
    una = assert_enum_domains_task_2(una)
    adj = assert_enum_domains_task_2(adj)
    # Task 2, Step 3: scoped sentinel scan, anti-imputation closure
    # Scan every cleansed frame once; any out-of-allow-list hit halts the run.
    for view_name, frame in (("FILES", files), ("UNADJUSTED_BARS", una), ("ADJUSTED_BARS", adj)):
        # Run the scoped single-pass scanner.
        hits, clean = scan_prohibited_sentinels_task_2(frame)
        # A non-clean frame cannot leave this stage.
        if not clean:
            # Raise with the rendered found-hits table in the message.
            raise ValueError(
                f"prohibited sentinel(s) outside the module-import allow-lists "
                f"detected in {view_name}:\n{hits.to_string()}"
            )
    # Enforce the anti-imputation post-condition; cleansing records zero operations.
    check_anti_imputation_task_2(context)
    # ---- Preservation post-condition: the input indexes survive verbatim ----
    # The unadjusted panel must keep its (stock_code, trade_date) MultiIndex.
    if not una.index.equals(input_indexes["UNADJUSTED_BARS"]):
        # Raise the preservation breach.
        raise RuntimeError("run_task_2: UNADJUSTED_BARS index was not preserved")
    # The adjusted panel must keep its (stock_code, trade_date) MultiIndex.
    if not adj.index.equals(input_indexes["ADJUSTED_BARS"]):
        # Raise the preservation breach.
        raise RuntimeError("run_task_2: ADJUSTED_BARS index was not preserved")
    # The manifest must keep its input index.
    if not files.index.equals(input_indexes["FILES"]):
        # Raise the preservation breach.
        raise RuntimeError("run_task_2: FILES index was not preserved")
    # ---- Return the cleansed bundle ----
    # Build the output dict; study_config passes through by identity.
    result: Dict[str, Any] = {
        # The cleansed manifest.
        "FILES": files,
        # The cleansed unadjusted panel.
        "UNADJUSTED_BARS": una,
        # The cleansed adjusted panel.
        "ADJUSTED_BARS": adj,
        # The cast ledger, one dict per cast in cast order.
        "cast_ledger": cast_ledger,
        # The deduplicated stock-code artifact, first-seen order.
        "stock_codes": list(stock_codes),
        # The three anti-imputation counters, all zero by construction.
        "forward_fill_operations": int(context.forward_fill_operations),
        "interpolation_operations": int(context.interpolation_operations),
        "stock_prefix_reductions": int(context.stock_prefix_reductions),
        # The study configuration object, unmodified.
        "study_config": STUDY_CONFIG,
    }
    # Return the cleansed bundle.
    return result
```

```python
# Task 3: Verify Acquisition Custody

# ==============================================================================
# Task 3: Verify Acquisition Custody
# ==============================================================================

SHA256_ALGORITHM: str = "SHA-256"
REHASH_CHUNK_BYTES: int = 1024 * 1024
PINNED_TRADING_DATE_COUNT: int = 727
SESSION_SPAN_LIMIT: int = 1
VIEW_UNADJUSTED: str = "unadjusted"
VIEW_ADJUSTED: str = "provider_adjusted"
# The published Shanghai Stock Exchange (XSHG) closed dates for the study's retrieval window
# (2023-2025 complete). This is public, fixed exchange-calendar reference data (the SSE publishes
# its session schedule each year); the window matches P11_ACQUISITION_PARAMETERS (span 2023-2025)
# and the pinned 727-row study calendar. Session dates = weekdays minus these closed dates; the
# exchange never opens on its own holiday calendar, and make-up working Saturdays are
# bank-style makeup days, NOT exchange sessions.
_XSHG_CLOSED_DATES_2023_2025: frozenset = frozenset({
    # 2023: New Year (observed), Spring Festival, Qingming, Labour Day, Dragon Boat, Mid-Autumn, National Day.
    pd.Timestamp("2023-01-02"),
    pd.Timestamp("2023-01-23"), pd.Timestamp("2023-01-24"), pd.Timestamp("2023-01-25"), pd.Timestamp("2023-01-26"), pd.Timestamp("2023-01-27"),
    pd.Timestamp("2023-04-05"),
    pd.Timestamp("2023-05-01"), pd.Timestamp("2023-05-02"), pd.Timestamp("2023-05-03"),
    pd.Timestamp("2023-06-22"),
    pd.Timestamp("2023-09-29"),
    pd.Timestamp("2023-10-02"), pd.Timestamp("2023-10-03"), pd.Timestamp("2023-10-04"), pd.Timestamp("2023-10-05"), pd.Timestamp("2023-10-06"),
    # 2024: New Year, Spring Festival, Qingming, Labour Day, Dragon Boat, Mid-Autumn, National Day.
    pd.Timestamp("2024-01-01"),
    pd.Timestamp("2024-02-09"), pd.Timestamp("2024-02-12"), pd.Timestamp("2024-02-13"), pd.Timestamp("2024-02-14"), pd.Timestamp("2024-02-15"), pd.Timestamp("2024-02-16"),
    pd.Timestamp("2024-04-04"), pd.Timestamp("2024-04-05"),
    pd.Timestamp("2024-05-01"), pd.Timestamp("2024-05-02"), pd.Timestamp("2024-05-03"),
    pd.Timestamp("2024-06-10"),
    pd.Timestamp("2024-09-16"), pd.Timestamp("2024-09-17"),
    pd.Timestamp("2024-10-01"), pd.Timestamp("2024-10-02"), pd.Timestamp("2024-10-03"), pd.Timestamp("2024-10-04"), pd.Timestamp("2024-10-05"), pd.Timestamp("2024-10-06"), pd.Timestamp("2024-10-07"),
    # 2025: New Year, Spring Festival, Qingming, Labour Day, Dragon Boat, National Day (incl. Mid-Autumn 2025-10-06).
    pd.Timestamp("2025-01-01"),
    pd.Timestamp("2025-01-28"), pd.Timestamp("2025-01-29"), pd.Timestamp("2025-01-30"), pd.Timestamp("2025-01-31"), pd.Timestamp("2025-02-03"), pd.Timestamp("2025-02-04"),
    pd.Timestamp("2025-04-04"),
    pd.Timestamp("2025-05-01"), pd.Timestamp("2025-05-02"), pd.Timestamp("2025-05-05"),
    pd.Timestamp("2025-06-02"),
    pd.Timestamp("2025-10-01"), pd.Timestamp("2025-10-02"), pd.Timestamp("2025-10-03"), pd.Timestamp("2025-10-06"), pd.Timestamp("2025-10-07"), pd.Timestamp("2025-10-08"),
})
# The dense 2023-2025 XSHG session axis, built lazily once per process (immutable after build).
_XSHG_SESSION_AXIS: Optional[pd.DatetimeIndex] = None


def _xshg_session_axis() -> pd.DatetimeIndex:
    """
    Return the cached XSHG session axis for 2023-2025.

    On first call the weekday range 2023-01-01 through 2025-12-31 is
    filtered against the pinned set of published XSHG closed dates
    ``_XSHG_CLOSED_DATES_2023_2025`` and stored in the module-level
    cache ``_XSHG_SESSION_AXIS``; every later call returns the same
    immutable axis.

    Returns
    -------
    pd.DatetimeIndex
        Ascending, de-duplicated Shanghai Stock Exchange trading
        sessions spanning 2023-01-01 through 2025-12-31.
    """
    # Build the axis on first use (no import-time work; immutable afterwards).
    global _XSHG_SESSION_AXIS
    # Only build when not yet built.
    if _XSHG_SESSION_AXIS is None:
        # Weekdays across the study window.
        axis = pd.bdate_range("2023-01-01", "2025-12-31")
        # Drop the published XSHG closed dates.
        axis = pd.DatetimeIndex([date for date in axis if date not in _XSHG_CLOSED_DATES_2023_2025])
        # Freeze the axis for reuse.
        _XSHG_SESSION_AXIS = axis
    # Return the cached axis.
    return _XSHG_SESSION_AXIS

# The pinned order of the six verification conjuncts, per the step text.
SIX_VERIFICATION_CONJUNCTS: Tuple[str, str, str, str, str, str] = (
    "hash_ok",
    "date_id_ok",
    "stock_id_ok",
    "adjustment_id_ok",
    "quantity_agreement_ok",
    "status_agreement_ok",
)
# The regex classifying a URI scheme; a plain local path never matches.
_URI_SCHEME_RE = re.compile(r"^([A-Za-z][A-Za-z0-9+.-]*)://")
# The file:// prefix, treated as a local path after stripping.
_FILE_URI_PREFIX: str = "file://"
# The provenance tags stamped on the verdict packet per the step text's register.
CONJUNCT_TAGS: Dict[str, str] = {
    "hash_ok": "[S]",
    "date_id_ok": "[S]",
    "stock_id_ok": "[S]",
    "adjustment_id_ok": "[S]",
    "quantity_agreement_ok": "[S]",
    "status_agreement_ok": "[S]",
    "verified": "[S] (aggregation)",
    "response_truncated": "[REVISED]/[R]",
}
# The computational path of each conjunct, named so a False can be traced (step text mandate).
CONJUNCT_PATHS: Dict[str, str] = {
    "hash_ok": "hashlib.sha256 over FILES.file_uri in 1 MiB chunks; bit-identical compare vs FILES.byte_hash (Task 3, Step 1)",
    "date_id_ok": "(i) FILES.date_min == FILES.requested_date_min AND (ii) FILES.date_max == FILES.requested_date_max AND (iii) nunique(per-file own-panel trade_date) == 727 AND (iv) per-file date set == set(calendar.trade_date) (Task 3, Step 2)",
    "stock_id_ok": "FILES.stock_code == FILES.requested_stock_code, cell-for-cell, nullable-boolean result (Task 3, Step 2)",
    "adjustment_id_ok": "FILES.adjustment == FILES.requested_adjustment, cell-for-cell, nullable-boolean result (Task 3, Step 2)",
    "quantity_agreement_ok": "cross-view reindex of per-stock quantity series on the unioned trade_date index; per-row NA-equal agreement (both-NA agree, one-sided NA disagrees, Int64 bit-identity); mismatch flags BOTH views' file_ids; gated by the per-(stock x view) date-set precondition, NA on failure (Task 3, Step 2)",
    "status_agreement_ok": "tradable_agreement AND tradestatus_raw_agreement on the same cross-view substrate; per-row NA-equal agreement; mismatch flags BOTH views' file_ids; gated by the same date-set precondition, NA on failure (Task 3, Step 2)",
}


class CustodyHaltError(Exception):
    """
    Halt raised when any acquisition-custody check fails.

    Single halt class of Task 3: raised on Step 1 rehash failures
    (digest or content-length mismatch, unreadable, remote or
    credential-gated URI, missing recorded claim), on Step 3
    aggregate failures, and on duplicated ``file_id`` values. NA is
    never a passing value. For Step 1 failures the message lists
    each failing ``file_id`` with its reason and, where available,
    recorded and observed evidence; for duplicated ``file_id`` values
    it lists only the duplicated ids.
    """

def _ensure_file_id_index(frame: pd.DataFrame, caller: str) -> pd.DataFrame:
    """
    Return a deep copy of a FILES manifest indexed on ``file_id``.

    Accepts either a frame already keyed on a ``file_id``-named index
    (with or without a same-named column) or a flat frame carrying a
    ``file_id`` column; in the latter case ``set_index("file_id",
    drop=False)`` retains the key as a column. The index name is
    pinned to ``"file_id"`` and the caller's frame is never mutated.

    Parameters
    ----------
    frame : pd.DataFrame
        The acquisition manifest in either conforming layout.
    caller : str
        Name of the calling callable, interpolated into the error
        messages.

    Returns
    -------
    pd.DataFrame
        Deep copy indexed on ``file_id`` with the index name pinned.

    Raises
    ------
    TypeError
        If ``frame`` is not a ``pandas.DataFrame``.
    ValueError
        If ``frame`` has neither a ``file_id``-named index nor a
        ``file_id`` column.
    """
    # Wrong argument types are structural defects; reject them before any work.
    if not isinstance(frame, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"{caller} requires a pandas.DataFrame, got {type(frame)!r}.")
    # Deep-copy so the caller's frame object is never mutated in place.
    copied = frame.copy(deep=True)
    # If the index already carries the pinned name, keep it verbatim.
    if copied.index.name == "file_id":
        # Re-assert the name explicitly; it is the pinned index contract.
        copied.index.name = "file_id"
        # Hand back the copy.
        return copied
    # Otherwise, if a file_id column exists, key the frame on it (key retained).
    if "file_id" in copied.columns:
        # set_index with drop=False mirrors Task 1's key-retention convention.
        copied = copied.set_index("file_id", drop=False)
        # Pin the index name explicitly.
        copied.index.name = "file_id"
        # Hand back the keyed copy.
        return copied
    # No key anywhere: the manifest is unkeyed; halt, do not guess.
    raise ValueError(
        f"{caller}: FILES carries neither a 'file_id'-named index nor a "
        "'file_id' column; the stable string primary key is mandatory."
    )

def _uri_scheme(uri: str) -> Optional[str]:
    """
    Classify a URI by its scheme, or ``None`` for a local path.

    A scheme is recognised only when a ``scheme://`` prefix is
    matched at the very start of ``uri`` (a Windows drive path never
    matches).

    Parameters
    ----------
    uri : str
        Candidate file URI recorded as ``FILES.file_uri``.

    Returns
    -------
    Optional[str]
        Lowercased scheme (``"s3"``, ``"https"``, ``"file"``, ...)
        when a ``scheme://`` prefix is present; ``None`` otherwise.
    """
    # Match the scheme separator at the start; a Windows drive path never matches.
    match = _URI_SCHEME_RE.match(uri)
    # Return the lowercased scheme or None for plain local paths.
    return match.group(1).lower() if match else None

def _local_path_from_uri(uri: str) -> Optional[str]:
    """
    Resolve a ``file_uri`` to a local path, or ``None`` if remote.

    Plain paths (no scheme) are returned verbatim. A ``file://`` URI
    is treated as an approved local alias and its prefix is stripped;
    on Windows a leading slash before a drive letter is dropped. Any
    other scheme is reported as unrehashable by returning ``None``.

    Parameters
    ----------
    uri : str
        Candidate file URI.

    Returns
    -------
    Optional[str]
        Local filesystem path for plain paths and ``file://`` URIs;
        ``None`` for any other scheme (remote or credential-gated).
    """
    # Classify the URI scheme.
    scheme = _uri_scheme(uri)
    # No scheme at all: a plain local path; use it verbatim.
    if scheme is None:
        # Return the path untouched.
        return uri
    # The file:// scheme is an approved local alias.
    if scheme == "file":
        # Strip the file:// prefix.
        payload = uri[len(_FILE_URI_PREFIX):]
        # On Windows, a triple-slash file URI yields a leading slash before the drive letter; drop it.
        if os.name == "nt" and payload.startswith("/") and len(payload) >= 3 and payload[2] == ":":
            # Normalise the drive-letter form.
            payload = payload[1:]
        # Return the normalised local path.
        return payload
    # Any other scheme is remote or credential-gated; the caller halts on it.
    return None

def _panel_stock_stats(panel: pd.DataFrame, caller: str) -> Dict[str, Dict[str, Any]]:
    """
    Compute per-stock panel statistics in a single grouping pass.

    The panel is grouped by ``stock_code`` (sorted, ``observed=True``,
    NA keys skipped). For each group the ``trade_date`` column is
    parsed once, NaT rows are dropped, and the sub-frame is indexed
    and sorted on the parsed dates. Both the raw distinct-date count
    ``n_dates`` (NaT excluded) and the raw distinct-date set
    ``date_set`` are returned alongside that sub-frame.

    Parameters
    ----------
    panel : pd.DataFrame
        One bar panel (unadjusted or adjusted) carrying at least
        ``stock_code``, ``trade_date``, ``quantity``, ``tradable``
        and ``tradestatus_raw``.
    caller : str
        Name of the calling callable, interpolated into errors.

    Returns
    -------
    Dict[str, Dict[str, Any]]
        Mapping ``stock_code`` (as ``str``) to a dictionary with keys
        ``n_dates`` (int), ``date_set`` (set of ``pd.Timestamp``)
        and ``frame`` (sub-frame indexed on sorted ``trade_date``).

    Raises
    ------
    TypeError
        If ``panel`` is not a ``pandas.DataFrame``.
    ValueError
        If any of the five required columns is missing.
    """
    # Wrong argument types are structural defects; reject them before any work.
    if not isinstance(panel, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"{caller} requires a pandas.DataFrame panel, got {type(panel)!r}.")
    # The pinned substrate columns of a bar panel.
    required: Tuple[str, str, str, str, str] = (
        "stock_code",
        "trade_date",
        "quantity",
        "tradable",
        "tradestatus_raw",
    )
    # Collect any missing pinned columns.
    missing: List[str] = [c for c in required if c not in panel.columns]
    # Halt on structural panel defects (this is an API precondition, not data).
    if missing:
        # Fail loudly with the remedy.
        raise ValueError(f"{caller}: panel is missing required column(s) {missing}.")
    # Initialise the per-stock statistics dictionary.
    stats: Dict[str, Dict[str, Any]] = {}
    # One grouping pass per view; observed=True avoids cartesian NA-category blow-ups.
    for stock, group in panel.groupby("stock_code", observed=True, sort=True):
        # groupby drops NA keys by default; guard defensively anyway.
        if pd.isna(stock):
            # Skip the NA-stock group.
            continue
        # Normalise the group key to a plain string.
        key = str(stock)
        # Parse the date column once (identity for datetime64; parse for object).
        dates = pd.to_datetime(group["trade_date"])
        # Copy the group, exclude unparseable/NaT dates, and key it on the parsed dates.
        frame = group.assign(trade_date=dates).dropna(subset=["trade_date"]).set_index("trade_date").sort_index()
        # Record the distinct-date count (the semantic observed_date_count; NaT is not a date).
        n_dates = int(dates.nunique())
        # Record the distinct date set without NaT (the primitive-(iv) substrate).
        date_set = set(dates.dropna())
        # Store the per-stock statistics.
        stats[key] = {"n_dates": n_dates, "date_set": date_set, "frame": frame}
    # Hand the statistics to the caller.
    return stats

def _na_equal_row_ok(left: pd.Series, right: pd.Series) -> pd.Series:
    """
    Return elementwise NA-equal agreement of two aligned series.

    A row is True when both values are null, or when both are
    non-null and bit-identical; null never agrees with a non-null
    value. The result contains no NA values, so callers can safely
    combine it with other masks, but its dtype may remain pandas'
    nullable ``boolean`` dtype rather than a NumPy ``bool`` dtype.

    Parameters
    ----------
    left : pd.Series
        Left series, e.g. the unadjusted view's ``quantity``.
    right : pd.Series
        Right series aligned to ``left.index``, e.g. the adjusted
        view's ``quantity``.

    Returns
    -------
    pd.Series
        NA-free boolean series, True where the two values agree under
        the NA-equal rule, False otherwise.
    """
    # Null masks for the two sides.
    left_na = left.isna()
    # Null mask for the right side.
    right_na = right.isna()
    # Both-null rows agree (masked observations agree, per the step text).
    both_null = left_na & right_na
    # Both-present rows are decided by bit-identical equality.
    both_present = ~left_na & ~right_na
    # Nullable equality propagates NA; under both_present it is never NA.
    equal = left == right
    # Fill NA with False defensively (unreachable under both_present) and decide.
    equal_filled = equal.fillna(False)
    # Row agreement: both-null, or both-present and bit-identical (Int64 / boolean / string).
    row_ok = both_null | (both_present & equal_filled)
    # Hand the crisp boolean mask to the caller.
    return row_ok

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 1:  Recompute byte hashes and content lengths (hash custody)
# -------------------------------------------------------------------------------------------------------------------------------

def rehash_files_task_3(FILES: pd.DataFrame, *, manifest_path: str) -> pd.DataFrame:
    """
    Recompute every file's SHA-256 and content length, vouch custody.

    Step 1 of Task 3 (Verify Acquisition Custody). Each ``file_uri`` is
    streamed to ``hashlib.sha256`` in ``REHASH_CHUNK_BYTES`` (1 MiB)
    chunks - never materialised in memory - and the recomputed digest
    is compared bit-identically against ``FILES.byte_hash``.
    ``content_length_bytes_observed`` is emitted as
    ``os.path.getsize(local_path)`` and compared bit-identically to
    ``content_length_bytes``. Both verdicts (``hash_ok``,
    ``content_length_ok``; nullable ``boolean``, domain
    {True, False, NA}) are recorded alongside the claims, never over
    them. A digest cache keyed on ``file_uri`` guarantees at most one
    rehash per distinct URI. NA is never a passing value. Duplicated
    ``file_id`` values are rejected before the per-file pass;
    otherwise the pass runs to completion and then a single
    ``CustodyHaltError`` lists the collected Step 1 failures, with
    recorded/observed evidence where available. A content-length
    mismatch on a hash-matching file is treated as the [R]
    vendor-side-truncation attack vector and halts like a digest
    mismatch. On success the augmented frame is returned and a
    ``hash_manifest.json`` artifact is written (UTF-8, insertion
    order, trailing newline); on failure no artifact is written.

    Parameters
    ----------
    FILES : pd.DataFrame
        Acquisition manifest indexed on ``file_id`` (or carrying a
        ``file_id`` column). Must carry ``file_uri``, ``byte_hash``
        and ``content_length_bytes``. The caller's frame is not
        mutated; a deep copy is returned.
    manifest_path : str
        Destination path of the ``hash_manifest.json`` artifact,
        written only when every file rehashes clean. Parent
        directories are created as needed.

    Returns
    -------
    pd.DataFrame
        Augmented manifest indexed on ``file_id`` with the new
        columns ``hash_ok`` (nullable ``boolean``),
        ``content_length_ok`` (nullable ``boolean``) and
        ``content_length_bytes_observed`` (nullable ``Int64``). On
        the success path ``hash_ok`` and ``content_length_ok`` are
        True, while ``content_length_bytes_observed`` holds each
        file's observed byte length as a nullable ``Int64`` integer.

    Raises
    ------
    TypeError
        If ``FILES`` is not a ``pandas.DataFrame``, or
        ``manifest_path`` is not a ``str`` or is empty/whitespace-only.
    ValueError
        If ``FILES`` lacks a ``file_id`` key or is missing any of
        ``file_uri``, ``byte_hash``, ``content_length_bytes``.
    CustodyHaltError
        If any file fails custody: duplicate ``file_id``; a missing,
        null, non-string or blank/whitespace-only URI; a missing or
        null hash; a missing or null length; a remote or unreadable
        URI; a digest mismatch; or a content-length mismatch.

    OSError
        If writing the manifest artifact fails.

    Notes
    -----
    ``row_count`` is never re-derived here; it is the delivered count
    pinned by the masking contract. The written artifact contains, per
    file, ``file_id``, ``file_uri`` and the recorded and observed
    digests and byte lengths, plus the metadata ``manifest_version``,
    ``algorithm``, ``chunk_bytes``, ``generated_by``, ``purpose`` and
    ``file_count``, so that Task 18 re-verifies against the artifact
    rather than recomputing.
    """
    # Wrong argument types are structural defects; reject them before any work.
    if not isinstance(FILES, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"rehash_files_task_3 requires a pandas.DataFrame, got {type(FILES)!r}.")
    # manifest_path must be a non-empty string path.
    if not isinstance(manifest_path, str) or not manifest_path.strip():
        # Fail loudly with the actionable remedy.
        raise TypeError("rehash_files_task_3 requires manifest_path to be a non-empty str.")
    # Deep-copy and pin the index on file_id; the caller's frame is never mutated.
    files = _ensure_file_id_index(FILES, "rehash_files_task_3")
    # The three columns Step 1 consumes; their absence is an upstream contract breach.
    required: Tuple[str, str, str] = ("file_uri", "byte_hash", "content_length_bytes")
    # Collect missing substrate columns.
    missing: List[str] = [c for c in required if c not in files.columns]
    # Halt with the remedy when the substrate is absent (run Task 1 first).
    if missing:
        # Fail loudly with the remedy.
        raise ValueError(
            f"rehash_files_task_3: FILES is missing required column(s) {missing}; "
            "run Task 1 (validate_files_task_1) before Task 3."
        )
    # Pin the row order and the ids once; every per-row lookup is positional thereafter.
    file_ids: List[Any] = list(files.index)
    # Re-assert file_id uniqueness one time at this boundary too (surrogate-per-view).
    duplicated: List[str] = sorted(fid for fid, count in Counter(file_ids).items() if count > 1)
    # Halt on duplicated file_ids: .at would silently return a Series, which is forbidden here.
    if duplicated:
        # Name the offending ids explicitly.
        raise CustodyHaltError(
            "CUSTODY HALT (Task 3, Step 1 rehash_files_task_3): duplicate file_id(s) "
            f"{duplicated}; file_id is surrogate per view and must appear at most once."
        )
    # Grab the three consumed columns once (positional access thereafter).
    file_uris = files["file_uri"]
    # The recorded digest claims (bit-identical compare target).
    recorded_hashes = files["byte_hash"]
    # The recorded content-length claims.
    recorded_lengths = files["content_length_bytes"]
    # The local digest cache, keyed on file_uri, exactly as the step text pins.
    digest_cache: Dict[str, Tuple[str, int]] = {}
    # Accumulators for the three emitted columns, in FILES order.
    hash_ok_values: List[Any] = []
    # True/False/NA for content_length_ok.
    length_ok_values: List[Any] = []
    # Observed bytes per file (Int64-able ints or pd.NA).
    observed_lengths: List[Any] = []
    # Observed hex digests per file (or pd.NA), carried into the artifact.
    observed_digests: List[Any] = []
    # The complete failure listing; Step 1 never stops early so every defect is reported.
    failures: List[str] = []
    # Process every file, in FILES order; collect verdicts and failures.
    for position, file_id in enumerate(file_ids):
        # Read the recorded digest claim at this position.
        recorded_hash = recorded_hashes.iat[position]
        # Read the recorded length claim at this position.
        recorded_length = recorded_lengths.iat[position]
        # Read the recorded URI at this position.
        uri_value = file_uris.iat[position]
        # A missing or non-string URI is unverifiable; NA is never a passing value.
        if (uri_value is None) or (not isinstance(uri_value, str)) or (not uri_value.strip()):
            # Name the file and the reason.
            failures.append(
                f"- file_id={file_id} | recorded file_uri is missing, null or non-string; "
                "the claim cannot be rehashed; NA is never a passing value"
            )
            # Record the unevaluable verdicts.
            hash_ok_values.append(None)
            length_ok_values.append(None)
            observed_lengths.append(pd.NA)
            observed_digests.append(pd.NA)
            # Continue to the next file; the halt is raised after the full pass.
            continue
        # Normalise the URI to a plain string.
        uri_str = str(uri_value)
        # A missing or null recorded digest can never verify; record the failure.
        if recorded_hash is None or bool(pd.isna(recorded_hash)):
            # Name the file and the reason.
            failures.append(
                f"- file_id={file_id} | recorded byte_hash is missing or null; "
                "the claim cannot be verified; NA is never a passing value"
            )
            # Record the unevaluable verdicts.
            hash_ok_values.append(None)
            length_ok_values.append(None)
            observed_lengths.append(pd.NA)
            observed_digests.append(pd.NA)
            # Continue to the next file.
            continue
        # Coerce the recorded digest claim to its literal string (bit-identical compare, no folding).
        recorded_hash_str = str(recorded_hash)
        # Resolve the URI to a local path (or None for remote/credential-gated schemes).
        local_path = _local_path_from_uri(uri_str)
        # Remote or credential-gated URIs cannot be rehashed without credentials; halt semantics apply.
        if local_path is None:
            # Name the file and the reason; never pass on an unrehashable file.
            failures.append(
                f"- file_id={file_id} | file_uri={uri_str!r} is remote or credential-gated and "
                "cannot be rehashed without credentials; the pipeline halts on any file it cannot rehash"
            )
            # Record the unevaluable verdicts.
            hash_ok_values.append(None)
            length_ok_values.append(None)
            observed_lengths.append(pd.NA)
            observed_digests.append(pd.NA)
            # Continue to the next file.
            continue
        # Reuse the cached digest for this URI when present.
        cached = digest_cache.get(uri_str)
        # Cache hit: reuse (digest_hex, size_bytes).
        if cached is not None:
            # Unpack the cached pair.
            digest_hex, size_bytes = cached
        else:
            # Cache miss: rehash the file in 1 MiB chunks.
            try:
                # Construct the streaming digest context.
                hasher = hashlib.sha256()
                # os.path.getsize is the pinned size source (step text).
                size_bytes = int(os.path.getsize(local_path))
                # Open a binary stream; never read() the whole file at once.
                with open(local_path, "rb") as handle:
                    # Stream in 1 MiB chunks until EOF.
                    for chunk in iter(lambda: handle.read(REHASH_CHUNK_BYTES), b""):
                        # Fold the chunk into the running digest.
                        hasher.update(chunk)
                # Materialise the lowercase hex digest.
                digest_hex = hasher.hexdigest()
            except OSError as exc:
                # Unreadable file (missing, permission, directory): named failure, pipeline halts.
                failures.append(
                    f"- file_id={file_id} | file_uri={uri_str!r} cannot be rehashed: {exc}"
                )
                # Record the unevaluable verdicts.
                hash_ok_values.append(None)
                length_ok_values.append(None)
                observed_lengths.append(pd.NA)
                observed_digests.append(pd.NA)
                # Continue to the next file.
                continue
            # Cache the outcome so a repeated URI is rehashed at most once.
            digest_cache[uri_str] = (digest_hex, size_bytes)
        # Bit-identical digest comparison (lowercase hex equality == byte-wise equality).
        hash_flag = digest_hex == recorded_hash_str
        # Resolve the recorded length claim to an integer, or mark it unverifiable.
        if recorded_length is None or pd.isna(recorded_length):
            # A missing recorded length can never verify.
            length_flag: Optional[bool] = None
            # Name the file and the reason.
            failures.append(
                f"- file_id={file_id} | recorded content_length_bytes is missing or null; "
                "the claim cannot be verified; NA is never a passing value"
            )
        else:
            try:
                # Coerce the Int64 claim to a python int (bit-identical integer compare).
                recorded_length_int = int(recorded_length)
            except (TypeError, ValueError, OverflowError):
                # A non-integral recorded length can never verify bit-identically.
                recorded_length_int = -1
            # Bit-identical length comparison under the same tolerance as the digest.
            length_flag = int(size_bytes) == recorded_length_int
        # Record the digest verdict for this file.
        hash_ok_values.append(hash_flag)
        # Record the length verdict for this file.
        length_ok_values.append(length_flag)
        # Record the observed size for this file.
        observed_lengths.append(int(size_bytes))
        # Record the observed digest for the artifact.
        observed_digests.append(digest_hex)
        # Digest mismatch: halt condition with recorded vs observed evidence.
        if not hash_flag:
            # Name the file with both digests.
            failures.append(
                f"- file_id={file_id} | byte_hash mismatch | recorded={recorded_hash_str} "
                f"| observed={digest_hex}"
            )
        # Length mismatch on a hash-matching file: the [R] vendor-truncation attack vector.
        if length_flag is False and hash_flag:
            failures.append(
                f"- file_id={file_id} | content_length_bytes mismatch on a hash-matching file "
                f"| recorded={int(recorded_length)} observed={int(size_bytes)} "
                "| [R] vendor-side-truncation attack vector"
            )
        # Length mismatch (digest already mismatching, or digest fine but length differs in an earlier line).
        elif length_flag is False:
            failures.append(
                f"- file_id={file_id} | content_length_bytes mismatch "
                f"| recorded={int(recorded_length)} observed={int(size_bytes)}"
            )
    # Raise the single halt after the full pass so the operator sees every defect.
    if failures:
        # Pin the count of failing lines to the count of failing files for auditability.
        header = (
            "CUSTODY HALT (Task 3, Step 1 rehash_files_task_3): "
            f"{len(failures)} custody failure(s) across {len(file_ids)} file(s); "
            "pipeline halted; hash_manifest.json NOT written (custody unproven)."
        )
        # Combine the header with the complete listing.
        message = header + "\n" + "\n".join(failures)
        # Raise the halt; no artifact is written on failure.
        raise CustodyHaltError(message)
    # Success path: assemble the augmented frame (result alongside the claims, never over them).
    out = files.copy(deep=True)
    # Add hash_ok as nullable boolean (domain {True, False, NA}; all True here).
    out["hash_ok"] = pd.Series(hash_ok_values, index=out.index, dtype="boolean")
    # Add content_length_ok as nullable boolean (same domain; all True here).
    out["content_length_ok"] = pd.Series(length_ok_values, index=out.index, dtype="boolean")
    # Add the observed byte length as nullable Int64.
    out["content_length_bytes_observed"] = pd.Series(observed_lengths, index=out.index, dtype="Int64")
    # Preserve the incoming attrs verbatim (the header metadata contract).
    out.attrs = dict(files.attrs)
    # Build the hash-manifest artifact; re-verification at Task 18 compares against this.
    manifest: Dict[str, Any] = {
        "manifest_version": 1,
        "algorithm": SHA256_ALGORITHM,
        "chunk_bytes": REHASH_CHUNK_BYTES,
        "generated_by": "task_3.rehash_files_task_3",
        "purpose": (
            "hash-manifest artifact for Task 18 re-verification "
            "(P18.reverification_at_study_close); the artifact is the comparison "
            "target, not a recomputation of it"
        ),
        "file_count": len(out),
        "files": [],
    }
    # Emit one artifact entry per file, in FILES order.
    for position, file_id in enumerate(file_ids):
        # Append the per-file provenance record.
        manifest["files"].append(
            {
                "file_id": str(file_id),
                "file_uri": str(file_uris.iat[position]),
                "byte_hash_recorded": str(recorded_hashes.iat[position]),
                "byte_hash_observed": str(observed_digests[position]),
                "content_length_bytes_recorded": int(recorded_lengths.iat[position]),
                "content_length_bytes_observed": int(observed_lengths[position]),
            }
        )
    # Serialise the artifact deterministically (UTF-8, insertion order, trailing newline).
    payload = json.dumps(manifest, ensure_ascii=False, indent=2) + "\n"
    # Resolve the artifact path.
    manifest_file = Path(manifest_path)
    # Create the parent directory when it does not exist (e.g. a nested scratch layout).
    try:
        if manifest_file.parent != manifest_file:
            # mkdir -p semantics, only for the artifact's own directory.
            manifest_file.parent.mkdir(parents=True, exist_ok=True)
        # Write the artifact atomically enough for a single-writer rehash pass.
        manifest_file.write_text(payload, encoding="utf-8")
    except OSError as exc:
        # Artifact write failure is an I/O failure, not a custody verdict; raise with context.
        raise OSError(
            f"rehash_files_task_3: unable to write hash manifest to {manifest_path!r}: {exc}"
        ) from exc
    # Hand the augmented manifest to the caller.
    return out

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 2:  Compute the six pinned verification conjuncts (and response_truncated)
# -------------------------------------------------------------------------------------------------------------------------------

def verify_conjuncts_task_3(
    FILES: pd.DataFrame,
    *,
    unadjusted: pd.DataFrame,
    adjusted: pd.DataFrame,
    calendar: pd.DataFrame,
) -> pd.DataFrame:
    """
    Materialise the six verification conjuncts as per-file cells.

    Step 2 of Task 3. Adds to a deep copy of ``FILES`` (indexed on
    ``file_id``) the six conjuncts named by the custody register, plus
    the pinned ``response_truncated``:

    * ``hash_ok`` -- carried verbatim from Step 1, cast to nullable
      ``boolean``.
    * ``stock_id_ok`` -- cell-for-cell ``stock_code`` equals
      ``requested_stock_code``.
    * ``adjustment_id_ok`` -- cell-for-cell ``adjustment`` equals
      ``requested_adjustment``.
    * ``date_id_ok`` -- Kleene AND of four crisp primitives: observed
      ``date_min`` equals ``requested_date_min``, observed ``date_max``
      equals ``requested_date_max``, the distinct ``trade_date`` count
      of the file's own panel equals ``PINNED_TRADING_DATE_COUNT``, and
      that panel's date set equals the calendar's date set.
    * ``quantity_agreement_ok`` and ``status_agreement_ok`` -- the
      cross-view join substrate described below.

    Each panel is grouped by ``stock_code`` once. For every stock in
    ``FILES`` that is present in both ``unadjusted`` and ``adjusted``
    the two per-view date sets must be identical, else both flags are
    ``pd.NA``. Otherwise both frames are reindexed on the union of the
    two date indexes and ``quantity``, ``tradable`` and
    ``tradestatus_raw`` are compared with ``_na_equal_row_ok`` (NA
    equals NA, otherwise bit-identity); any mismatching row gives
    ``False`` for both views' ``file_id`` rows. ``status_agreement_ok``
    is the AND of the ``tradable`` and ``tradestatus_raw`` comparisons.
    A stock whose counterpart ``file_id`` is absent from ``FILES`` gets
    ``pd.NA`` on both flags.

    ``response_truncated`` is always ``NOT(date_id_ok)``. ``row_count``
    is never read or rewritten here. The returned ``attrs`` copy the
    input ``attrs`` and add ``conjunct_paths`` (the computational path
    of every conjunct) and ``conjunct_tags`` (the provenance tags).

    Parameters
    ----------
    FILES : pd.DataFrame
        Acquisition manifest indexed on ``file_id``; must carry
        ``stock_code``, ``adjustment``, ``row_count``, ``date_min``,
        ``date_max``, ``requested_stock_code``, ``requested_date_min``,
        ``requested_date_max``, ``requested_adjustment`` and
        ``hash_ok``. Never mutated.
    unadjusted : pd.DataFrame
        Unadjusted bar panel carrying ``stock_code``, ``trade_date``,
        ``quantity``, ``tradable`` and ``tradestatus_raw``.
    adjusted : pd.DataFrame
        Provider-adjusted bar panel with the same required columns.
    calendar : pd.DataFrame
        Pinned trading calendar carrying ``trade_date``; its unique
        parsed dates form the requested date set.

    Returns
    -------
    pd.DataFrame
        Deep copy of ``FILES`` indexed on ``file_id`` with the columns
        ``hash_ok``, ``date_id_ok``, ``stock_id_ok``,
        ``adjustment_id_ok``, ``quantity_agreement_ok``,
        ``status_agreement_ok`` and ``response_truncated``, all of
        nullable ``boolean`` dtype, plus the ``attrs`` keys
        ``conjunct_paths`` and ``conjunct_tags``.

    Raises
    ------
    TypeError
        If any argument is not a ``pandas.DataFrame``, or if
        ``FILES["hash_ok"]`` cannot be cast to nullable ``boolean``.
    ValueError
        If ``FILES`` lacks a required substrate column, if ``calendar``
        lacks ``trade_date``, or if a panel lacks one of its required
        columns (raised by ``_panel_stock_stats``).

    Notes
    -----
    This step never halts; the halts live in Steps 1 and 3. NA is
    confined to a failed join precondition (per-(stock x view) date-set
    mismatch or a missing view), to a stock without a counterpart file,
    and to rows whose ``stock_code`` or ``adjustment`` is NA. The
    ``date_id_ok`` primitives are crisp: datetime64 equality against
    NaT yields ``False``, never NA.
    """
    # Wrong argument types are structural defects; reject them before any work.
    if not isinstance(FILES, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"verify_conjuncts_task_3 requires a pandas.DataFrame FILES, got {type(FILES)!r}.")
    # The unadjusted panel must be a DataFrame.
    if not isinstance(unadjusted, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"verify_conjuncts_task_3 requires a pandas.DataFrame unadjusted, got {type(unadjusted)!r}.")
    # The adjusted panel must be a DataFrame.
    if not isinstance(adjusted, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"verify_conjuncts_task_3 requires a pandas.DataFrame adjusted, got {type(adjusted)!r}.")
    # The calendar must be a DataFrame.
    if not isinstance(calendar, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"verify_conjuncts_task_3 requires a pandas.DataFrame calendar, got {type(calendar)!r}.")
    # Deep-copy and pin the index on file_id.
    files = _ensure_file_id_index(FILES, "verify_conjuncts_task_3")
    # The pinned per-file substrate columns; hash_ok arrives from Step 1.
    required_cols: Tuple[str, str, str, str, str, str, str, str, str, str] = (
        "stock_code",
        "adjustment",
        "row_count",
        "date_min",
        "date_max",
        "requested_stock_code",
        "requested_date_min",
        "requested_date_max",
        "requested_adjustment",
        "hash_ok",
    )
    # Collect any missing substrate columns.
    missing_cols: List[str] = [c for c in required_cols if c not in files.columns]
    # Halt with the remedy when Step 1 has not been run or Task 1's output is absent.
    if missing_cols:
        # Fail loudly with the remedy.
        raise ValueError(
            f"verify_conjuncts_task_3: FILES is missing required column(s) {missing_cols}; "
            "run Step 1 (rehash_files_task_3) before Step 2."
        )
    # The calendar must carry the pinned date column (the 727-date reference).
    if "trade_date" not in calendar.columns:
        # Fail loudly with the actionable remedy.
        raise ValueError("verify_conjuncts_task_3: calendar must carry a 'trade_date' column.")
    # Group both panels once; statistics feed primitives (iii)/(iv) and the pairing.
    un_stats = _panel_stock_stats(unadjusted, "verify_conjuncts_task_3")
    # Statistics for the adjusted view.
    adj_stats = _panel_stock_stats(adjusted, "verify_conjuncts_task_3")
    # The requested set is the 727-date calendar, materialised once.
    requested_set = set(pd.to_datetime(calendar["trade_date"]))
    # Carry hash_ok verbatim; enforce a boolean dtype at the boundary.
    hash_ok = files["hash_ok"]
    # If a direct caller supplies a non-boolean hash_ok, cast it or fail loudly.
    if str(hash_ok.dtype) != "boolean":
        try:
            # Cast to nullable boolean; non-conforming values raise.
            hash_ok = hash_ok.astype("boolean")
        except (TypeError, ValueError) as exc:
            # Fail loudly with the remedy.
            raise TypeError(
                f"verify_conjuncts_task_3: hash_ok must be a nullable boolean column: {exc}"
            ) from exc
    # ---- stock_id_ok: FILES.stock_code == FILES.requested_stock_code, cell-for-cell.
    stock_id_ok = files["stock_code"].eq(files["requested_stock_code"]).astype("boolean")
    # ---- adjustment_id_ok: FILES.adjustment == FILES.requested_adjustment, cell-for-cell.
    adjustment_id_ok = files["adjustment"].eq(files["requested_adjustment"]).astype("boolean")
    # ---- date_id_ok, primitive (i): observed_date_min == requested_date_min.
    obs_min = pd.to_datetime(files["date_min"])
    # The requested min, elementwise against the observed.
    req_min = pd.to_datetime(files["requested_date_min"])
    # Crisp comparison (NaT never equals, so it yields False, never NA).
    prim_i = obs_min.eq(req_min).astype("boolean")
    # ---- date_id_ok, primitive (ii): observed_date_max == requested_date_max.
    obs_max = pd.to_datetime(files["date_max"])
    # The requested max, elementwise against the observed.
    req_max = pd.to_datetime(files["requested_date_max"])
    # Crisp comparison (same NaT semantics).
    prim_ii = obs_max.eq(req_max).astype("boolean")
    # ---- date_id_ok, primitives (iii) and (iv): per-file own-panel count and set.
    stock_col = files["stock_code"]
    # The view column decides which panel is this file's own.
    adjustment_col = files["adjustment"]
    # Accumulate the two per-row verdicts.
    count_ok: List[bool] = []
    # Same for the set primitive.
    set_ok: List[bool] = []
    # Evaluate each row against its own panel.
    for position in range(len(files)):
        # Read the row's stock.
        stock_value = stock_col.iat[position]
        # Read the row's view.
        view_value = adjustment_col.iat[position]
        # An unkeyed or unviewable row cannot be placed in its own panel.
        if pd.isna(stock_value) or pd.isna(view_value):
            # Both primitives fail; no exception, no pass.
            count_ok.append(False)
            set_ok.append(False)
            # On to the next row.
            continue
        # Normalise the keys.
        stock_key = str(stock_value)
        # Normalise the view key.
        view_key = str(view_value)
        # Pick this file's own panel statistics.
        if view_key == VIEW_UNADJUSTED:
            # The unadjusted view is this file's own panel.
            stats = un_stats.get(stock_key)
        elif view_key == VIEW_ADJUSTED:
            # The provider-adjusted view is this file's own panel.
            stats = adj_stats.get(stock_key)
        else:
            # An out-of-enum view string places the row in no panel.
            stats = None
        # A stock absent from its own panel fails both primitives.
        if stats is None:
            # Primitives (iii) and (iv) fail; the file is unverifiable on this axis.
            count_ok.append(False)
            set_ok.append(False)
            # On to the next row.
            continue
        # Primitive (iii): the distinct-date count equals the pinned 727.
        count_ok.append(stats["n_dates"] == PINNED_TRADING_DATE_COUNT)
        # Primitive (iv): the observed set is bit-identical to the 727-date calendar.
        set_ok.append(stats["date_set"] == requested_set)
    # Assemble primitive (iii) as a boolean Series on the FILES index.
    prim_iii = pd.Series(count_ok, index=files.index, dtype="boolean")
    # Assemble primitive (iv) as a boolean Series on the FILES index.
    prim_iv = pd.Series(set_ok, index=files.index, dtype="boolean")
    # ---- date_id_ok: the Kleene conjunction of the four crisp primitives (all crisp here).
    date_id_ok = (prim_i & prim_ii & prim_iii & prim_iv).astype("boolean")
    # ---- response_truncated: pinned, NOT(date_id_ok); never an independent measurement.
    response_truncated = ~date_id_ok
    # ---- Cross-view pairing: per-stock verdicts for quantity and status.
    # Map from (stock, view) to position in FILES, to decide pairing per row.
    seen: Dict[Tuple[str, str], int] = {}
    # Record which positions exist per (stock, view) pair.
    for position in range(len(files)):
        # Read the row's stock and view.
        stock_value = stock_col.iat[position]
        # Read the row's view.
        view_value = adjustment_col.iat[position]
        # Skip unkeyable rows (they cannot pair).
        if pd.isna(stock_value) or pd.isna(view_value):
            # Nothing to record.
            continue
        # Record the position under the normalised pair key.
        seen[(str(stock_value), str(view_value))] = position
    # The stocks appearing anywhere in FILES determine the pairing universe.
    all_stocks: List[str] = list(dict.fromkeys(str(s) for s in stock_col if not pd.isna(s)))
    # Per-stock pair verdicts: stock -> (quantity_verdict, status_verdict).
    pair_verdict: Dict[str, Tuple[Any, Any]] = {}
    # Evaluate each stock once (40 stocks x 727 rows; the pass is O(80 x 727)).
    for stock in all_stocks:
        # Fetch this stock's unadjusted sub-frame statistics (or None when absent).
        un_entry = un_stats.get(stock)
        # Materialise the unadjusted sub-frame, or None.
        un_frame = un_entry["frame"] if un_entry else None
        # Fetch this stock's adjusted sub-frame statistics (or None when absent).
        adj_entry = adj_stats.get(stock)
        # Materialise the adjusted sub-frame, or None.
        adj_frame = adj_entry["frame"] if adj_entry else None
        # Both views must be present in the panels before any comparison runs.
        if un_frame is None or adj_frame is None:
            # Precondition fails: the join substrate does not exist; both flags go NA.
            pair_verdict[stock] = (pd.NA, pd.NA)
            # On to the next stock.
            continue
        # The pinned precondition: the per-(stock x view) date sets must be identical.
        if un_stats[stock]["date_set"] != adj_stats[stock]["date_set"]:
            # Precondition fails: quantity flag (and the join-dependent status flag) NA, both file_ids.
            pair_verdict[stock] = (pd.NA, pd.NA)
            # On to the next stock.
            continue
        # Align the two views on the union of their (now identical) date indexes.
        union_index = un_frame.index.union(adj_frame.index)
        # Quantity series reindexed to the union index.
        un_quantity = un_frame["quantity"].reindex(union_index)
        # Adjusted quantity series reindexed to the same union index.
        adj_quantity = adj_frame["quantity"].reindex(union_index)
        # Bit-identity / NA-equal agreement on quantity (any mismatch fails the stock).
        quantity_ok = bool(_na_equal_row_ok(un_quantity, adj_quantity).all())
        # Tradable series reindexed to the union index (nullable boolean).
        un_tradable = un_frame["tradable"].reindex(union_index)
        # Adjusted tradable series reindexed to the same index.
        adj_tradable = adj_frame["tradable"].reindex(union_index)
        # tradable_agreement under the same NA-equal semantics.
        tradable_ok = bool(_na_equal_row_ok(un_tradable, adj_tradable).all())
        # tradestatus_raw series reindexed to the union index (Int64).
        un_status_raw = un_frame["tradestatus_raw"].reindex(union_index)
        # Adjusted tradestatus_raw series reindexed to the same index.
        adj_status_raw = adj_frame["tradestatus_raw"].reindex(union_index)
        # tradestatus_raw_agreement under NA-equal bit-identity.
        tradestatus_ok = bool(_na_equal_row_ok(un_status_raw, adj_status_raw).all())
        # status_agreement_ok := tradable_agreement AND tradestatus_raw_agreement (prior register).
        status_ok = tradable_ok and tradestatus_ok
        # Record the pair verdict once; it flags BOTH views' file_ids.
        pair_verdict[stock] = (quantity_ok, status_ok)
    # Expand the per-stock verdicts to per-row file verdicts in FILES order.
    quantity_values: List[Any] = []
    # Same for the status conjunct.
    status_values: List[Any] = []
    # Walk FILES in order; assign the pair verdict or NA for unpaired rows.
    for position in range(len(files)):
        # Read the row's stock and view.
        stock_value = stock_col.iat[position]
        # Read the row's view.
        view_value = adjustment_col.iat[position]
        # An unkeyed row cannot pair; NA it is.
        if pd.isna(stock_value) or pd.isna(view_value):
            # Unpairable row.
            quantity_values.append(pd.NA)
            status_values.append(pd.NA)
            # On to the next row.
            continue
        # Normalise the keys.
        stock_key = str(stock_value)
        # Normalise the view key.
        view_key = str(view_value)
        # The other view this stock must also exist in FILES for a pair to form.
        other_view = VIEW_ADJUSTED if view_key == VIEW_UNADJUSTED else (VIEW_UNADJUSTED if view_key == VIEW_ADJUSTED else None)
        # No other view known: no pair exists; NA.
        if other_view is None:
            # Unpairable row.
            quantity_values.append(pd.NA)
            status_values.append(pd.NA)
            # On to the next row.
            continue
        # If this stock has no counterpart file in the other view, there is no pair; NA.
        if (stock_key, other_view) not in seen:
            # Unpaired file; disclose with NA.
            quantity_values.append(pd.NA)
            status_values.append(pd.NA)
            # On to the next row.
            continue
        # Fetch the pair verdict for this stock.
        verdict = pair_verdict.get(stock_key)
        # A missing verdict cannot happen (all_stocks covers every FILES stock); be defensive.
        if verdict is None:
            # Unmeasurable; NA.
            quantity_values.append(pd.NA)
            status_values.append(pd.NA)
            # On to the next row.
            continue
        # Assign the pair verdict (flags BOTH views' file_ids identically).
        quantity_values.append(verdict[0])
        # Assign the shared status verdict.
        status_values.append(verdict[1])
    # Assemble quantity_agreement_ok as nullable boolean on the FILES index.
    quantity_agreement_ok = pd.Series(quantity_values, index=files.index, dtype="boolean")
    # Assemble status_agreement_ok as nullable boolean on the FILES index.
    status_agreement_ok = pd.Series(status_values, index=files.index, dtype="boolean")
    # Materialise the output: deep copy of the (already deep-copied) frame plus the conjuncts.
    out = files.copy(deep=True)
    # Carry hash_ok verbatim (identical index; assignment aligns without a dtype roundtrip).
    out["hash_ok"] = hash_ok
    # Assign the derived conjuncts in the pinned column grouping.
    out["date_id_ok"] = date_id_ok
    # stock_id_ok cell-for-cell.
    out["stock_id_ok"] = stock_id_ok
    # adjustment_id_ok cell-for-cell.
    out["adjustment_id_ok"] = adjustment_id_ok
    # quantity_agreement_ok (cross-view).
    out["quantity_agreement_ok"] = quantity_agreement_ok
    # status_agreement_ok (cross-view).
    out["status_agreement_ok"] = status_agreement_ok
    # Pin response_truncated := NOT(date_id_ok); never an independent measurement.
    out["response_truncated"] = response_truncated
    # Preserve the incoming attrs and stamp the conjunct paths (the diagnostic contract).
    out.attrs = dict(files.attrs)
    # Attach the computational paths so a False can be traced to its check.
    out.attrs["conjunct_paths"] = dict(CONJUNCT_PATHS)
    # Attach the provenance tags for the disclosure register.
    out.attrs["conjunct_tags"] = dict(CONJUNCT_TAGS)
    # Hand the augmented manifest to the caller.
    return out

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Step 3:  The Kleene fail-fast verified aggregate (coherence re-assertion included)
# -------------------------------------------------------------------------------------------------------------------------------

def _reassert_coherence(files: pd.DataFrame, calendar: Optional[pd.DataFrame] = None) -> Tuple[str, str]:
    """
    Re-assert the pinned retrieval coherence window on ``FILES``.

    Measures the span of the acquisition completion stamps
    (``files["retrieved"]``) as a count of Shanghai equity trading
    sessions and compares it with ``SESSION_SPAN_LIMIT`` (one session).
    The span is ``pos(max) - pos(min)``, where ``pos`` is
    ``max(0, searchsorted(axis, stamp, side="right") - 1)`` on the
    ascending session axis. Nothing is raised on an exceedance: the
    finding is disclosed in the returned note.

    Stamps are parsed with ``errors="coerce"`` and nulls dropped; a
    tz-aware column is converted to Asia/Shanghai wall time and made
    naive. The session axis is the sorted unique parsed ``trade_date``
    of ``calendar`` when supplied, otherwise the embedded XSHG schedule
    from ``_xshg_session_axis``. A supplied but unusable calendar fails
    closed with reason ``calendar_unavailable`` instead of silently
    switching source.

    Parameters
    ----------
    files : pd.DataFrame
        Manifest indexed on ``file_id`` that should carry ``retrieved``,
        the acquisition completion stamps.
    calendar : Optional[pd.DataFrame], optional
        Dense trading calendar carrying ``trade_date``, used as the
        session axis of record. Default is ``None``, which selects the
        embedded XSHG 2023-2025 schedule.

    Returns
    -------
    Tuple[str, str]
        ``(status, note)``. ``status`` is ``"OK"`` when the observed
        span is at most ``SESSION_SPAN_LIMIT`` sessions and
        ``"PROVENANCE_LIMIT"`` otherwise, or when the check is
        unevaluable (``retrieved_unavailable`` when the column is absent
        or carries no non-null stamp; ``calendar_unavailable`` when a
        supplied calendar has no usable ``trade_date`` axis). ``note``
        is the human-readable finding naming the observed span, the
        pinned limit and the session-axis source.

    Raises
    ------
    TypeError
        If ``files`` is not a ``pandas.DataFrame``.

    Notes
    -----
    The provenance response to a coherence exceedance is disclosure, not
    abort (P11). The span counts trading rows, never business weekdays
    or elapsed seconds.
    """
    # Type-check the manifest before any arithmetic (boundary validation).
    if not isinstance(files, pd.DataFrame):
        # Fail loudly with the observed type (never coerce what we cannot read).
        raise TypeError(f"_reassert_coherence requires a pandas.DataFrame, got {type(files).__name__}")
    # The coherence re-assertion requires the retrieval stamps; absent them the check is unevaluable.
    if "retrieved" not in files.columns:
        # Disclose the absence; never pass, never halt.
        return "PROVENANCE_LIMIT", (
            "retrieved_unavailable: FILES carries no 'retrieved' column; the retrieval coherence "
            "re-assertion is unevaluable; the pinned session-span count cannot be computed; "
            "the response is disclosure, not abort (P11_ACQUISITION_PARAMETERS.retrieval_coherence_window == "
            "'<= 1 trading day')"
        )
    # Parse the stamps defensively (identity for datetime64; parse for object).
    observed = pd.to_datetime(files["retrieved"], errors="coerce")
    # The witnessed subset is the non-null stamps only.
    observed = observed.dropna()
    # Without any witnessed stamp the interval is unevaluable.
    if observed.empty:
        # Disclose the absence; never pass, never halt.
        return "PROVENANCE_LIMIT", (
            "retrieved_unavailable: FILES.retrieved carries no non-null stamps; the retrieval "
            "coherence re-assertion is unevaluable; the pinned session-span count cannot be "
            "computed; the response is disclosure, not abort (P11_ACQUISITION_PARAMETERS."
            "retrieval_coherence_window == '<= 1 trading day')"
        )
    # The stamps may be tz-aware Asia/Shanghai (the ingest coercion); normalise to naive Shanghai wall time.
    if isinstance(observed.dtype, pd.DatetimeTZDtype):
        # Convert to Asia/Shanghai wall time, then drop the zone.
        naive = observed.dt.tz_convert("Asia/Shanghai").dt.tz_localize(None)
    else:
        # Naive stamps are read as Shanghai wall time and kept naive.
        naive = observed
    # The wall-time extremes of the acquisition batch (both bounds of the span).
    span_start = naive.min()
    # The later extreme, same normalisation.
    span_end = naive.max()
    # ---- Session axis: the supplied calendar governs; absent -> the embedded exchange schedule -------
    # A caller who passes a conforming dense calendar pins the measurement to that calendar-of-record
    # (Task 1/Task 5's reconstruction). A caller who passes nothing gets the pinned Shanghai Stock
    # Exchange session schedule (2023-2025, published by the exchange; weekdays minus the published
    # closed dates) -- this is NOT a business-weekday ruler (exchange holidays are excluded).
    # A supplied but malformed calendar fails closed to PROVENANCE_LIMIT (never a silent source swap).
    cal_axis: Optional[pd.DatetimeIndex] = None
    # The source tag recorded in the note (provenance discipline).
    cal_source: str = ""
    # A conforming caller calendar wins.
    if calendar is not None:
        # The calendar must be a DataFrame carrying a datetime64 trade_date axis.
        if isinstance(calendar, pd.DataFrame) and "trade_date" in calendar.columns:
            # Parse the trade_date axis defensively (identity for datetime64).
            cal_dates = pd.to_datetime(calendar["trade_date"], errors="coerce").dropna()
            # A non-empty axis is the caller's session axis.
            if not cal_dates.empty:
                # Materialise the sorted unique session axis exactly once.
                cal_axis = pd.DatetimeIndex(pd.Series(cal_dates.dropna()).sort_values().unique())
                # The supplied calendar governs (recorded for provenance).
                cal_source = "the supplied dense trading calendar"
        # A malformed supplied calendar never silently promotes to the embedded schedule.
        if cal_axis is None:
            # Fail closed with the reason named (never a silent source swap).
            return "PROVENANCE_LIMIT", (
                f"calendar_unavailable: observed retrieval span runs from {span_start.normalize().date()} to "
                f"{span_end.normalize().date()}; the supplied calendar is malformed (no usable 'trade_date' "
                f"session axis); the pinned one-session limit (P11_ACQUISITION_PARAMETERS."
                f"retrieval_coherence_window == '<= 1 trading day') is UNEVALUABLE on this input; the "
                f"response is disclosure, not abort; Shanghai equity trading-calendar session span "
                f"source: not available"
            )
    else:
        # No calendar supplied: measure on the embedded Shanghai Stock Exchange session schedule.
        cal_axis = _xshg_session_axis()
        # The embedded exchange schedule governs (recorded for provenance).
        cal_source = "the embedded Shanghai Stock Exchange session schedule (2023-2025, published closed dates)"
    # ---- Equity-session span arithmetic (identical to Task 1's _session_span_count) -------
    # Position of a stamp on the sorted session axis: side="right" minus one, clamped to 0;
    # a stamp before the first session row clamps to 0; after the last, pins to len(axis) - 1.
    def _axis_position(stamp: pd.Timestamp) -> int:
        """
        Return the index of the last session at or before ``stamp``.

        Parameters
        ----------
        stamp : pd.Timestamp
            Wall-time stamp placed on the ascending session axis.

        Returns
        -------
        int
            ``searchsorted`` position with ``side="right"`` minus one,
            clamped to ``0`` so a stamp before the axis start maps to
            the first session.
        """
        # searchsorted side="right" on the sorted axis, minus one (the last trading row at or before the stamp).
        position = int(cal_axis.searchsorted(stamp, side="right")) - 1
        # Clamp below-origin stamps to the first session row (never negative).
        return max(0, position)
    # The span START position on the session axis.
    start_pos = _axis_position(span_start)
    # The span END position on the session axis.
    end_pos = _axis_position(span_end)
    # The span in trading sessions (integer count of trading rows between the extremes).
    span_sessions = max(0, end_pos - start_pos)
    # Compare against the pinned one-session limit (P11 retrieval_coherence_window == "<= 1 trading day").
    if span_sessions > SESSION_SPAN_LIMIT:
        # Disclose the exceedance; never halt (P11: disclosure, not abort).
        return "PROVENANCE_LIMIT", (
            f"observed retrieval span is {span_sessions} session(s) "
            f"({span_start.date()} to {span_end.date()}), exceeding the pinned one-session window "
            "(P11_ACQUISITION_PARAMETERS.retrieval_coherence_window == '<= 1 trading day'); "
            f"Shanghai equity trading-calendar session span source: {cal_source} "
            "(searchsorted side='right' minus one on the ascending session axis); "
            "the response is disclosure, not abort"
        )
    # Within the pinned window.
    return "OK", (
        f"observed retrieval span is {span_sessions} session(s) "
        f"({span_start.date()} to {span_end.date()}); within the pinned one-session window "
        f"(P11_ACQUISITION_PARAMETERS.retrieval_coherence_window == '<= 1 trading day'); "
        f"Shanghai equity trading-calendar session span source: {cal_source} "
        "(searchsorted side='right' minus one on the ascending session axis)"
    )

def aggregate_verified_task_3(
    FILES: pd.DataFrame,
    *,
    check_id: str = "T3.start.coherence",
    calendar: Optional[pd.DataFrame] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Apply the Kleene fail-fast aggregate over the six conjuncts.

    Consolidate the six verification conjuncts of Task 3, Step 2 into a
    single ``verified`` flag with False-dominant Kleene semantics:

    * any non-NA ``False`` conjunct makes ``verified`` ``False`` and
      halts the pipeline;
    * otherwise, if all six conjuncts are known and ``True``,
      ``verified`` is ``True``;
    * otherwise ``verified`` is NA - disclosed, never a pass, never a
      halt.

    The conjuncts are coerced to nullable ``boolean`` and the
    conjunction is materialised with ``eq``/``notna`` masks rather than
    ``DataFrame.all(skipna=False)``. ``file_id`` uniqueness is
    re-asserted before aggregating, the retrieval coherence window is
    re-evaluated as disclosure only, and NA-verified files are logged
    at WARNING level and stamped into the packet attrs. The input frame
    is never mutated.

    Parameters
    ----------
    FILES : pd.DataFrame
        Augmented acquisition manifest indexed on ``file_id`` (or
        carrying a ``file_id`` column). Must contain the six conjunct
        columns and ``response_truncated``; conjunct dtypes must be
        nullable ``boolean`` or coercible to it.
    check_id : str, optional
        Namespaced coherence re-assertion identifier; must be a
        non-empty string. Default is ``"T3.start.coherence"``;
        ``"T18.close.coherence"`` is used at close-of-study
        re-verification.
    calendar : Optional[pd.DataFrame], optional
        Trading calendar carrying a datetime64 ``trade_date`` column.
        Default is ``None``. When supplied, the coherence span is
        measured on its session axis (``searchsorted`` positions); when
        ``None`` an embedded XSHG 2023-2025 session schedule is used. A
        supplied but malformed calendar yields
        ``coherence_status == "PROVENANCE_LIMIT"`` with reason
        ``calendar_unavailable``.

    Returns
    -------
    FILES_with_verdicts : pd.DataFrame
        Deep copy of ``FILES`` with ``verified`` added as a nullable
        boolean column, indexed on ``file_id``. attrs carry
        ``coherence_check_id`` and ``coherence_status``.
    verdict_packet : pd.DataFrame
        Column subset in this order: the six conjuncts, ``verified``,
        ``response_truncated``, indexed on ``file_id``. attrs carry
        ``provenance_limit_na_verified`` (list of NA-verified ids),
        ``coherence_status`` (``"OK"`` or ``"PROVENANCE_LIMIT"``),
        ``coherence_check_id``, ``coherence_note``, ``conjunct_paths``,
        ``conjunct_tags``, ``verified_semantics`` and ``task``.

    Raises
    ------
    TypeError
        If ``FILES`` is not a ``pandas.DataFrame``, or a conjunct column
        cannot be cast to nullable ``boolean``.
    ValueError
        If ``check_id`` is not a non-empty string, if a conjunct column
        or ``response_truncated`` is missing, or if ``FILES`` carries
        neither a ``file_id`` index nor a ``file_id`` column.
    CustodyHaltError
        If ``file_id`` is duplicated, or if any conjunct is non-NA
        ``False``; the message lists every failing ``file_id`` and the
        False conjunct(s) behind it.

    Notes
    -----
    The pinned halt assertion ``FILES.verified[verified.notna()].all()``
    is implemented through the explicit False mask, because
    ``DataFrame.all(skipna=False)`` does not propagate NA on
    nullable-boolean frames.
    """
    # Wrong argument types are structural defects; reject them before any work.
    if not isinstance(FILES, pd.DataFrame):
        # Fail loudly with the actionable remedy.
        raise TypeError(f"aggregate_verified_task_3 requires a pandas.DataFrame, got {type(FILES)!r}.")
    # check_id must be a non-empty string naming the coherence namespace.
    if not isinstance(check_id, str) or not check_id.strip():
        # Fail loudly with the actionable remedy.
        raise ValueError("aggregate_verified_task_3 requires check_id to be a non-empty str.")
    # Deep-copy and pin the index on file_id.
    files = _ensure_file_id_index(FILES, "aggregate_verified_task_3")
    # The six conjuncts are produced by Steps 1-2; their absence is an upstream contract breach.
    missing_conjuncts: List[str] = [c for c in SIX_VERIFICATION_CONJUNCTS if c not in files.columns]
    # Halt with the remedy when Step 2 has not been run.
    if missing_conjuncts:
        # Fail loudly with the remedy.
        raise ValueError(
            f"aggregate_verified_task_3: FILES is missing conjunct column(s) {missing_conjuncts}; "
            "run Steps 1-2 before Step 3."
        )
    # response_truncated is the pinned derived column; its absence is a Step 2 bypass.
    if "response_truncated" not in files.columns:
        # Fail loudly with the remedy.
        raise ValueError(
            "aggregate_verified_task_3: FILES is missing 'response_truncated'; "
            "run Step 2 (verify_conjuncts_task_3) before Step 3."
        )
    # Re-assert file_id uniqueness one more time; duplicates double-count the aggregation.
    duplicated: List[str] = sorted(
        fid for fid, count in Counter(files.index).items() if count > 1
    )
    # Halt on duplicated file_ids with the complete listing.
    if duplicated:
        # Raise with the named ids.
        raise CustodyHaltError(
            "CUSTODY HALT (Task 3, Step 3 aggregate_verified_task_3): duplicate file_id(s) "
            f"{duplicated}; file_id is surrogate per view and must appear at most once."
        )
    # Coerce the six conjuncts to nullable boolean, failing loudly on non-conforming cells.
    conjunct_series: Dict[str, pd.Series] = {}
    # Process the six columns in the pinned order.
    for column in SIX_VERIFICATION_CONJUNCTS:
        # Read the column.
        series = files[column]
        # If a direct caller supplies a non-boolean dtype, cast it or fail loudly.
        if str(series.dtype) != "boolean":
            try:
                # Cast to nullable boolean.
                series = series.astype("boolean")
            except (TypeError, ValueError) as exc:
                # Fail loudly with the remedy.
                raise TypeError(
                    f"aggregate_verified_task_3: conjunct {column!r} must be a nullable "
                    f"boolean column: {exc}"
                ) from exc
        # Store the coerced column under the pinned key.
        conjunct_series[column] = series
    # Materialise the six conjuncts as an aligned boolean frame on the FILES index.
    six_frame = pd.DataFrame(conjunct_series, index=files.index)    # ---- Kleene fail-fast: materialised explicitly (the prescribed .all(skipna=False) is broken).
    # Non-NA False mask; NA propagates inside eq, which is correct for masking.
    is_false = six_frame.eq(False)
    # any_false: at least one conjunct is non-NA False (crisp, NA-skipping).
    any_false = is_false.any(axis=1, skipna=True)
    # all_known_true: every conjunct is True AND every conjunct is known (defeats vacuous True).
    all_known_true = six_frame.eq(True).all(axis=1, skipna=True) & six_frame.notna().all(axis=1)
    # Verified starts as fully unknown, then resolves False-dominant, then all-True, else NA.
    verified = pd.Series(pd.NA, index=files.index, dtype="boolean")
    # False dominance: any non-NA False forces verified = False.
    verified.loc[any_false] = False
    # Else if every conjunct is known and True: verified = True.
    verified.loc[~any_false & all_known_true] = True
    # ---- Halt assertion (step text): FILES.verified[verified.notna()].all() must be True.
    failing_ids: List[Any] = [
        fid
        for fid, flag in zip(files.index.tolist(), verified.tolist())
        if (not pd.isna(flag)) and (not bool(flag))
    ]
    # Halt with the complete listing when any non-NA False exists.
    if failing_ids:
        # Build one line per failing file naming the conjuncts that are False.
        lines: List[str] = []
        # Walk the failing ids in FILES order.
        for fid in failing_ids:
            # Identify the False conjuncts behind this file.
            false_conjuncts: List[str] = [
                column
                for column in SIX_VERIFICATION_CONJUNCTS
                if not pd.isna(six_frame.at[fid, column]) and bool(six_frame.at[fid, column]) is False
            ]
            # Record the line with the offending conjuncts.
            lines.append(f"- file_id={fid} | False conjunct(s): {', '.join(false_conjuncts)}")
        # Pin the halt message; no verdict packet is produced on halt.
        message = (
            "CUSTODY HALT (Task 3, Step 3 aggregate_verified_task_3): "
            f"{len(failing_ids)} file(s) carry at least one non-NA False conjunct; "
            "pipeline halted. Failing file_ids and their False conjuncts:\n" + "\n".join(lines)
        )
        # Raise the halt; Task 4 and Task 16 never see this partial state.
        raise CustodyHaltError(message)
    # NA-verified rows: permitted, named disclosure, never a pass, never a halt.
    na_verified: List[str] = [
        str(fid) for fid, flag in zip(files.index.tolist(), verified.tolist()) if pd.isna(flag)
    ]
    # Surface the disclosure at WARNING level for the operator log.
    if na_verified:
        # Name every NA-verified file_id.
        logger.warning(
            "PROVENANCE_LIMIT_NA_VERIFIED: file_id(s) %s carry only True/NA conjuncts "
            "with at least one NA; verified is NA, disclosed, not passed.",
            na_verified,
        )
    # ---- Coherence-window re-assertion (disclosure only; never a halt).
    coherence_status, coherence_note = _reassert_coherence(files, calendar)
    # Surface the disclosure at WARNING level when the window is exceeded or unevaluable.
    if coherence_status == "PROVENANCE_LIMIT":
        # Name the namespace of the re-assertion.
        logger.warning(
            "PROVENANCE_LIMIT [%s]: %s",
            check_id,
            coherence_note,
        )
    # ---- FILES with verdicts: preserve the frame, index and attrs; add verified.
    out = files.copy(deep=True)
    # Assign verified as nullable boolean, aligned to the FILES index.
    out["verified"] = verified
    # Preserve the incoming attrs and stamp the re-assertion namespace and outcome.
    out.attrs = dict(files.attrs)
    # Name the coherence namespace behind this re-assertion.
    out.attrs["coherence_check_id"] = check_id
    # Record the coherence outcome (OK or PROVENANCE_LIMIT).
    out.attrs["coherence_status"] = coherence_status
    # ---- verdict_packet: the six conjuncts, verified, response_truncated.
    packet = out[list(SIX_VERIFICATION_CONJUNCTS) + ["verified", "response_truncated"]].copy(deep=True)
    # Re-pin the packet's column order explicitly (defence against ordering drift).
    packet = packet[list(SIX_VERIFICATION_CONJUNCTS) + ["verified", "response_truncated"]]
    # Re-pin the packet's index name (file_id remains the key).
    packet.index.name = "file_id"
    # Stamp the packet attrs: the disclosure register and the diagnostic paths.
    packet.attrs = {
        "provenance_limit_na_verified": list(na_verified),
        "coherence_status": coherence_status,
        "coherence_check_id": check_id,
        "coherence_note": coherence_note,
        "conjunct_paths": dict(CONJUNCT_PATHS),
        "conjunct_tags": dict(CONJUNCT_TAGS),
        "verified_semantics": "KLEENE_FAIL_FAST (False dominance; else all-True; else NA)",
        "task": "Task 3 - Verify Acquisition Custody",
    }
    # Hand the (FILES_with_verdicts, verdict_packet) pair to the caller.
    return out, packet

# -------------------------------------------------------------------------------------------------------------------------------
# Task 3, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_3(
    FILES: pd.DataFrame,
    *,
    unadjusted: pd.DataFrame,
    adjusted: pd.DataFrame,
    calendar: pd.DataFrame,
    manifest_path: str,
    coherence_check_id: str = "T3.start.coherence",
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Run the three ordered steps of Task 3 and return the verdict pair.

    Chain the custody steps in the pinned order. Step 1
    (``rehash_files_task_3``) recomputes byte hashes and content lengths
    and writes ``hash_manifest.json`` only when every file rehashes
    cleanly; Step 2 (``verify_conjuncts_task_3``) materialises the six
    verification conjuncts and the derived ``response_truncated`` flag;
    Step 3 (``aggregate_verified_task_3``) applies the False-dominant
    Kleene aggregate and re-asserts the retrieval coherence window. A
    Step 1 halt prevents Step 2 from running; a Step 3 halt leaves the
    Step 1 manifest on disk as affirmative custody evidence.

    Parameters
    ----------
    FILES : pd.DataFrame
        Acquisition manifest indexed on ``file_id`` (or carrying a
        ``file_id`` column).
    unadjusted : pd.DataFrame
        Unadjusted bar panel providing the columns Step 2 joins on.
    adjusted : pd.DataFrame
        Provider-adjusted bar panel providing the columns Step 2 joins
        on.
    calendar : pd.DataFrame
        Pinned trading calendar carrying ``trade_date``; consumed by
        Step 2 and by the Step 3 coherence re-assertion.
    manifest_path : str
        Destination path of the ``hash_manifest.json`` artifact written
        by Step 1.
    coherence_check_id : str, optional
        Coherence re-assertion namespace forwarded to Step 3. Default is
        ``"T3.start.coherence"``; ``"T18.close.coherence"`` is used at
        close-of-study re-verification.

    Returns
    -------
    FILES_with_verdicts : pd.DataFrame
        Manifest carrying the six conjuncts, ``response_truncated`` and
        ``verified``, indexed on ``file_id``.
    verdict_packet : pd.DataFrame
        The six conjuncts, ``verified`` and ``response_truncated`` for
        every ``file_id``, with the Step 3 disclosure and diagnostic
        attrs.

    Raises
    ------
    CustodyHaltError
        Propagated from Step 1 (digest mismatch, content-length mismatch
        on a hash-matching file, unrehashable URI, duplicated
        ``file_id``) or from Step 3 (any non-NA ``False`` conjunct,
        duplicated ``file_id``).
    TypeError
        If ``FILES`` is not a ``pandas.DataFrame``.
    ValueError
        If a substrate column required by an intermediate step is
        missing or another step validation fails.
    """
    # Step 1: recompute hash custody; halts on the first unclean file (complete listing).
    rehashed = rehash_files_task_3(FILES, manifest_path=manifest_path)
    # Step 2: materialise the six conjuncts (never halts); the pinned response_truncated derives here.
    conjuncts = verify_conjuncts_task_3(
        rehashed,
        unadjusted=unadjusted,
        adjusted=adjusted,
        calendar=calendar,
    )
    # Step 3: apply the Kleene fail-fast aggregate (halts on any non-NA False) and re-assert coherence.
    files_with_verdicts, verdict_packet = aggregate_verified_task_3(
        conjuncts,
        check_id=coherence_check_id,
        calendar=calendar,
    )
    # Hand the (FILES_with_verdicts, verdict_packet) pair to the caller.
    return files_with_verdicts, verdict_packet
```

```python
# Task 4: Cross-View Identity Verification and Irreversible Role Separation

# ==============================================================================
# Task 4: Cross-View Identity Verification and Irreversible Role Separation
# ==============================================================================

__all__ = [
    "CrossViewHaltError",
    "ViewAlignmentHaltError",
    "SharedFieldHaltError",
    "AdjustmentFactorHaltError",
    "PanelSplitHaltError",
    "FrozenFeaturePanel",
    "verify_view_alignment_task_4",
    "verify_shared_fields_task_4",
    "verify_adjustment_factor_constancy_task_4",
    "split_panels_task_4",
    "run_task_4",
]

# Module-level pinned constants (each cited to its source).
COMPOUND_INDEX_NAMES: Tuple[str, str] = ("stock_code", "trade_date")
# The prior register's pinned no-change threshold for the adjustment ratio:
# |Delta-r| < 1e-12 is "no change"; any larger jump increments the segment.
SEGMENT_JUMP_THRESHOLD: float = 1e-12
# The 1-based first segment identifier of every stock's action-free walk.
FIRST_SEGMENT_ID: int = 1
# The pinned column names of the two exported panels (step text, verbatim).
FEATURES_COLUMNS: Tuple[str, ...] = (
    "stock_code", "trade_date", "open", "high", "low", "close",
    "volume", "amount", "tradable", "tradestatus_raw",
)
LABELS_COLUMNS: Tuple[str, ...] = (
    "stock_code", "trade_date", "close_adj", "adj_factor", "tradable",
)
# The pinned artefact-table schema of the shared-field audit (step text).
ARTEFACT_COLUMNS: Tuple[str, ...] = (
    "stock_code", "trade_date", "field", "unadjusted_value", "adjusted_value",
)
# The shared fields audited cell-for-cell across the two views.
SHARED_FIELD_QUANTITY: str = "quantity"
SHARED_FIELD_STATUS: str = "tradestatus_raw"
SHARED_FIELD_TRADABLE: str = "tradable"
# The two persisted artifact file names beneath their distinct roots.
FEATURES_FILE_NAME: str = "features_panel.csv"
LABELS_FILE_NAME: str = "labels_panel.csv"


class CrossViewHaltError(Exception):
    """
    Base halt for every cross-view identity failure.

    Raised through its subclasses when the unadjusted and adjusted views
    cannot be certified as the same panel. The message carries the
    complete evidence listing (every offending coordinate, count or
    name) rather than a bare assertion.
    """

class ViewAlignmentHaltError(CrossViewHaltError):
    """
    Halt: the two views are not the same panel coordinate set.

    Raised by the Step 1 alignment checks (compound-index name
    preservation, adjustment-domain purity, orphan counts, stage-pinned
    join cardinality, trading-date set equality) and by the structural
    boundary checks on a caller-supplied view panel.
    """

class SharedFieldHaltError(CrossViewHaltError):
    """
    Halt: a shared field fails NA-aware cell-for-cell agreement.

    Raised when ``quantity``, ``tradestatus_raw`` or ``tradable``
    differs between the two views, when exactly one side of a pair is
    NA, or when the FILES manifest cannot record the verdict.
    """

class AdjustmentFactorHaltError(CrossViewHaltError):
    """
    Halt: derived adjustment segments differ from the vendor's claim.
    """

class PanelSplitHaltError(CrossViewHaltError):
    """
    Halt: the feature/label role split or its persistence was breached.
    """

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, module-internal helpers (panel boundary validation, date extraction, NA-aware agreement kernel)
# -------------------------------------------------------------------------------------------------------------------------------

def _validate_panel_task_4(
    panel: Any,
    role: str,
) -> pd.DataFrame:
    """
    Validate that a caller-supplied view panel is structurally sound.

    Boundary check for a panel arriving from outside the module.
    Requires a non-empty ``pandas.DataFrame`` whose index is a
    ``MultiIndex`` named ``COMPOUND_INDEX_NAMES`` (``stock_code``,
    ``trade_date``, in that order), with unique coordinates, no NaT in
    the ``trade_date`` level, and an ``adjustment`` column present. The
    panel is neither copied nor mutated.

    Parameters
    ----------
    panel : Any
        The object expected to be the view panel.
    role : str
        Role label (``"UNADJUSTED_BARS"`` or ``"ADJUSTED_BARS"``) used
        in every error message.

    Returns
    -------
    pd.DataFrame
        The same, unmutated panel.

    Raises
    ------
    TypeError
        If ``panel`` is not a ``pandas.DataFrame``.
    ViewAlignmentHaltError
        If the panel is empty, its index is not a ``MultiIndex`` named
        ``COMPOUND_INDEX_NAMES``, its coordinates are not unique, the
        ``trade_date`` level holds NaT, or the ``adjustment`` column is
        absent.
    """
    # The boundary type check: anything else is a caller error, not data.
    if not isinstance(panel, pd.DataFrame):
        # Raise immediately; downstream steps assume DataFrame semantics.
        raise TypeError(f"{role} must be a pandas.DataFrame, got {type(panel).__name__}")
    # The boundary emptiness check: an empty panel voids every cross-view check.
    if panel.empty:
        # Raise with the role embedded for immediate identification.
        raise ViewAlignmentHaltError(f"{role} is empty; the cross-view identity stage cannot run on zero rows")
    # The compound index is the join key mandated by the step text; it must exist.
    if not isinstance(panel.index, pd.MultiIndex):
        # Raise with the observed index type for forensics.
        raise ViewAlignmentHaltError(
            f"{role} index must be a pandas.MultiIndex named {COMPOUND_INDEX_NAMES}, "
            f"got {type(panel.index).__name__}"
        )
    # The index must carry exactly the pinned names, in the pinned order.
    if list(panel.index.names) != list(COMPOUND_INDEX_NAMES):
        # Raise with the observed names for forensics.
        raise ViewAlignmentHaltError(
            f"{role} index names must be {list(COMPOUND_INDEX_NAMES)}, got {list(panel.index.names)}"
        )
    # Duplicate coordinates would silently multiply the join; refuse them here.
    if not panel.index.is_unique:
        # Compute the duplicated coordinate count for the halt listing.
        duplicated_count = int(panel.index.duplicated(keep=False).sum())
        # Raise with the duplicate burden named.
        raise ViewAlignmentHaltError(
            f"{role} carries {duplicated_count} duplicated (stock_code, trade_date) rows; "
            f"the cell-for-cell comparison is void"
        )
    # The trade_date level must contain no NaT (a NaT coordinate is not a trading date).
    if panel.index.get_level_values("trade_date").hasnans:
        # Raise; NaT coordinates are not identifiable dates.
        raise ViewAlignmentHaltError(f"{role} trade_date index level contains NaT; dates must be concrete")
    # The adjustment column must exist (Step 1's purity assertion keys off it).
    if "adjustment" not in panel.columns:
        # Raise listing the observed columns.
        raise ViewAlignmentHaltError(f"{role} is missing the required 'adjustment' column; observed columns: {list(panel.columns)}")
    # Hand the validated (unmutated) panel back to the caller.
    return panel

def _validate_panel_pair_task_4(
    unadjusted_bars: Any,
    adjusted_bars: Any,
) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Validate both view panels as one compound boundary.

    Applies ``_validate_panel_task_4`` to the unadjusted view first and
    to the adjusted view second, so the first error reported is always
    the unadjusted one. Neither panel is mutated.

    Parameters
    ----------
    unadjusted_bars : Any
        The nominal/unadjusted view.
    adjusted_bars : Any
        The provider-adjusted view.

    Returns
    -------
    unadjusted : pd.DataFrame
        The validated, unmutated unadjusted view.
    adjusted : pd.DataFrame
        The validated, unmutated adjusted view.

    Raises
    ------
    TypeError
        If either panel is not a ``pandas.DataFrame``.
    ViewAlignmentHaltError
        If either panel fails the boundary contract.
    """
    # Validate the unadjusted view first (order fixed for deterministic errors).
    unadjusted = _validate_panel_task_4(unadjusted_bars, "UNADJUSTED_BARS")
    # Validate the adjusted view second under the same contract.
    adjusted = _validate_panel_task_4(adjusted_bars, "ADJUSTED_BARS")
    # Return both validated panels together.
    return unadjusted, adjusted

def _trade_date_series_task_4(panel: pd.DataFrame) -> pd.Series:
    """
    Return the ``trade_date`` values of a validated panel.

    Reads the ``trade_date`` column when present, otherwise the
    ``trade_date`` index level, so both the column layout and the
    index-only layout are supported.

    Parameters
    ----------
    panel : pd.DataFrame
        A panel that has already passed ``_validate_panel_task_4``.

    Returns
    -------
    pd.Series
        One ``trade_date`` value per row, in panel order; the dtype of
        the source column or index level is preserved.

    Raises
    ------
    ViewAlignmentHaltError
        If the panel carries neither a ``trade_date`` column nor a
        ``trade_date`` index level.
    """
    # Prefer the column layout used by both raw schemas of the study configuration.
    if "trade_date" in panel.columns:
        # Return the column Series as-is (dtypes preserved).
        return panel["trade_date"]
    # Fall back to the index level (same tz-naive datetime64[ns] convention).
    if "trade_date" in panel.index.names:
        # Return the level values as a Series with the panel's row positions.
        return pd.Series(panel.index.get_level_values("trade_date"))
    # Neither layout carried the dates; halt with a forensic message.
    raise ViewAlignmentHaltError("panel carries neither a 'trade_date' column nor a 'trade_date' index level")

def _require_columns_task_4(
    panel: pd.DataFrame,
    columns: Tuple[str, ...],
    role: str,
) -> None:
    """
    Require every named column to be present in a panel.

    Parameters
    ----------
    panel : pd.DataFrame
        The panel to inspect.
    columns : Tuple[str, ...]
        The exact column names that must be present.
    role : str
        Role label used in the error message.

    Raises
    ------
    ViewAlignmentHaltError
        If any column is missing; the message lists every missing name
        together with the observed column list.
    """
    # Compute the complete missing-column set (list all, not just the first).
    missing = [column for column in columns if column not in panel.columns]
    # Halt with the complete listing when anything is absent.
    if missing:
        # Raise with every missing name enumerated.
        raise ViewAlignmentHaltError(f"{role} is missing required columns {missing}; observed columns: {list(panel.columns)}")

def _na_equal_agreement_task_4(
    left: pd.Series,
    right: pd.Series,
    field: str,
) -> Dict[str, Any]:
    """
    Audit one aligned field pair under the NA-aware agreement rule.

    A coordinate agrees iff ``(a is NA and b is NA) or (a is not NA and
    b is not NA and a == b)``. The nullable equality result is therefore
    never read raw: the NA masks are materialised as plain boolean
    arrays and combined positionally. Both-NA coordinates are counted as
    masked pairs; one-sided NA coordinates are counted as masking drift.

    Parameters
    ----------
    left : pd.Series
        The unadjusted view's values, positionally aligned with
        ``right``.
    right : pd.Series
        The adjusted view's values, positionally aligned with ``left``.
    field : str
        Field name stamped into every emitted artefact row.

    Returns
    -------
    Dict[str, Any]
        ``mismatch_rows`` (list of ``dict`` with keys ``stock_code``,
        ``trade_date``, ``field``, ``unadjusted_value``,
        ``adjusted_value``; one entry per pure mismatch, in positional
        order, ``trade_date`` being ``pd.NaT`` when the coordinate is
        not a tuple); ``one_sided_na_count`` (int);
        ``both_na_count`` (int); ``agreement_ok`` (bool, True iff no
        pure mismatch and no one-sided NA).

    Raises
    ------
    ViewAlignmentHaltError
        If the two series do not carry identical indexes.
    """
    # The caller must have aligned the two series; verify positional identity.
    if not left.index.equals(right.index):
        # Raise with the observed index names for forensics.
        raise ViewAlignmentHaltError(f"shared-field agreement requires aligned inputs; observed indexes {list(left.index.names)} vs {list(right.index.names)}")
    # Materialise the NA masks positionally (bool arrays; never NA-typed).
    left_na = left.isna().to_numpy(dtype=bool)
    right_na = right.isna().to_numpy(dtype=bool)
    # both_na: the masked-observation coordinate set (counted, never an error).
    both_na = left_na & right_na
    # one_sided: masking drift - a halt at the caller, never a silent pass.
    one_sided = left_na ^ right_na
    # both_present: exactly the coordinates the equality test may read.
    both_present = ~left_na & ~right_na
    # Nullable-boolean equality: NA wherever either side is NA.
    equal = left.eq(right)
    # Fill only under the both_present discipline: NA becomes False, and
    # both_present guarantees no NA sentinel ever leaks into a comparison.
    equal_filled = equal.fillna(False).to_numpy(dtype=bool)
    # A pure mismatch: both present, values compared, not equal.
    mismatch = both_present & ~equal_filled
    # Positional coordinates of every mismatch (flat, deterministic).
    mismatch_positions = np.flatnonzero(mismatch)
    # Build one artefact row per mismatch, with the pinned artefact columns.
    mismatch_rows: List[Dict[str, Any]] = []
    for position in mismatch_positions:
        # Recover the compound coordinate (stock_code, trade_date) label.
        coordinate = left.index[position]
        # Emit the artefact; values read positionally from both sides.
        mismatch_rows.append(
            {
                "stock_code": coordinate[0] if isinstance(coordinate, tuple) else coordinate,
                "trade_date": coordinate[1] if isinstance(coordinate, tuple) else pd.NaT,
                "field": field,
                "unadjusted_value": left.iat[position],
                "adjusted_value": right.iat[position],
            }
        )
    # agreement_ok is the conjunction of the two clean conditions.
    agreement_ok = bool(mismatch_positions.size == 0 and not bool(one_sided.any()))
    # Hand the structured verdict back to the caller.
    return {
        "mismatch_rows": mismatch_rows,
        "one_sided_na_count": int(one_sided.sum()),
        "both_na_count": int(both_na.sum()),
        "agreement_ok": agreement_ok,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 1:  Pairwise cross-view alignment diagnostic (outer join, orphans, purity, calendar equality)
# -------------------------------------------------------------------------------------------------------------------------------

def verify_view_alignment_task_4(
    unadjusted_bars: pd.DataFrame,
    adjusted_bars: pd.DataFrame,
    *,
    expected_join_cardinality: int,
) -> Dict[str, Any]:
    """
    Verify the pairwise alignment of the two views.

    Step 1 of the cross-view identity stage. Both panels are
    boundary-validated, then joined with
    ``pandas.merge(..., how="outer", indicator=True, left_index=True,
    right_index=True)`` on the compound ``(stock_code, trade_date)``
    index. The join is materialised only for the diagnostic and is
    never stored or returned. The checks, in order:

    1. The joined index names must equal ``COMPOUND_INDEX_NAMES``.
    2. ``unadjusted_bars["adjustment"]`` must take only
       ``"unadjusted"`` and ``adjusted_bars["adjustment"]`` only
       ``"provider_adjusted"``.
    3. ``orphans_unadj`` (``left_only`` rows) and ``orphans_adj``
       (``right_only`` rows) must both be zero.
    4. ``join_cardinality`` must equal the stage pin
       ``expected_join_cardinality``: 29,080 pre-mask, 29,078 after
       Task 5's masking.
    5. The ``trade_date`` sets of the two views must be equal in both
       directions.

    Parameters
    ----------
    unadjusted_bars : pd.DataFrame
        The nominal view; compound ``(stock_code, trade_date)``
        ``MultiIndex``; must carry an ``adjustment`` column and a
        ``trade_date`` column or index level. Never mutated.
    adjusted_bars : pd.DataFrame
        The provider-adjusted view with the same contract. Never
        mutated.
    expected_join_cardinality : int
        Stage-pinned joined row count; must be an integer other than
        ``bool`` and at least 1.

    Returns
    -------
    Dict[str, Any]
        ``orphans_unadj`` (int), ``orphans_adj`` (int),
        ``join_cardinality`` (int), ``date_set_equal`` (bool),
        ``adjustment_unadj`` (list[str]), ``adjustment_adj``
        (list[str]), ``index_names`` (list[str]). On success the orphan
        counts are zero, ``date_set_equal`` is True and each adjustment
        domain is its single pinned string.

    Raises
    ------
    TypeError
        If either panel is not a ``pandas.DataFrame``, or
        ``expected_join_cardinality`` is not a non-``bool`` integer.
    ValueError
        If ``expected_join_cardinality`` is below 1.
    ViewAlignmentHaltError
        If a panel fails boundary validation, the joined index loses its
        names, an adjustment domain holds other values, any orphan
        exists, the cardinality does not match the pin, or the
        ``trade_date`` sets differ. Every message lists complete
        evidence.
    """
    # Boundary validation of both panels (raw caller input from outside the module).
    unadjusted, adjusted = _validate_panel_pair_task_4(unadjusted_bars, adjusted_bars)
    # The stage pin: accept any integer type (numpy integers welcome) but NEVER bool.
    if isinstance(expected_join_cardinality, bool) or not isinstance(expected_join_cardinality, (int, np.integer)):
        # Raise naming the observed type.
        raise TypeError(
            f"expected_join_cardinality must be an integer (bools excluded), "
            f"got {type(expected_join_cardinality).__name__}"
        )
    # A pin below 1 is a caller error; the panels are non-empty by validation.
    if int(expected_join_cardinality) < 1:
        # Raise naming the observed value.
        raise ValueError(f"expected_join_cardinality must be >= 1, got {expected_join_cardinality}")
    # Materialise the diagnostic join once - outer, indicator, index-index.
    # The suffixes are pinned so overlapping non-key columns stay disjoint.
    joined = pd.merge(
        unadjusted,
        adjusted,
        how="outer",
        left_index=True,
        right_index=True,
        indicator=True,
        suffixes=("_unadjusted", "_adjusted"),
    )
    # The compound index must survive the join (Task 6 references the same index).
    index_names = list(joined.index.names)
    if index_names != list(COMPOUND_INDEX_NAMES):
        # The diagnostic is dropped on every exit path; raise before retention.
        raise ViewAlignmentHaltError(
            f"compound index name loss through the join: expected {list(COMPOUND_INDEX_NAMES)}, got {index_names}"
        )
    # Constant-adjustment purity for the unadjusted view (Step 3 embargo keys off this).
    adjustment_unadj = sorted(unadjusted["adjustment"].dropna().astype(str).unique().tolist())
    if adjustment_unadj != [VIEW_UNADJUSTED]:
        # Raise with the complete observed domain.
        raise ViewAlignmentHaltError(
            f"UNADJUSTED_BARS adjustment domain must be exclusively [{VIEW_UNADJUSTED!r}], got {adjustment_unadj}"
        )
    # Constant-adjustment purity for the adjusted view (Step 3 embargo keys off this).
    adjustment_adj = sorted(adjusted["adjustment"].dropna().astype(str).unique().tolist())
    if adjustment_adj != [VIEW_ADJUSTED]:
        # Raise with the complete observed domain.
        raise ViewAlignmentHaltError(
            f"ADJUSTED_BARS adjustment domain must be exclusively [{VIEW_ADJUSTED!r}], got {adjustment_adj}"
        )
    # Orphan counts: |UNADJ - JOIN| = left_only rows; |ADJ - JOIN| = right_only rows.
    merge_counts = joined["_merge"].value_counts()
    orphans_unadj = int(merge_counts.get("left_only", 0))
    orphans_adj = int(merge_counts.get("right_only", 0))
    if orphans_unadj != 0 or orphans_adj != 0:
        # Collect the offending coordinates for the complete halt listing.
        orphan_coordinates = joined.index[joined["_merge"] != "both"].tolist()
        # Raise with a complete listing (capped only by the coordinate count itself).
        raise ViewAlignmentHaltError(
            f"cross-view orphan halt: orphans_unadj={orphans_unadj}, orphans_adj={orphans_adj}; "
            f"orphan coordinates (up to all): {orphan_coordinates}"
        )
    # The joined cardinality (post-union rows; includes zero orphans on success).
    join_cardinality = int(len(joined))
    # The stage-pinned cardinality assertion (never decorative).
    if join_cardinality != int(expected_join_cardinality):
        # Raise with pinned and observed values side by side.
        raise ViewAlignmentHaltError(
            f"join cardinality mismatch: expected_join_cardinality pinned {int(expected_join_cardinality)}, "
            f"observed {join_cardinality}"
        )
    # Trading-date set equality: the strongest statement about which dates survived.
    unadjusted_dates = set(_trade_date_series_task_4(unadjusted))
    adjusted_dates = set(_trade_date_series_task_4(adjusted))
    date_set_equal = unadjusted_dates == adjusted_dates
    if not date_set_equal:
        # Compute both directions of the set difference for the halt listing.
        missing_in_adjusted = sorted(unadjusted_dates - adjusted_dates)
        missing_in_unadjusted = sorted(adjusted_dates - unadjusted_dates)
        # Raise with both one-directional differences fully listed.
        raise ViewAlignmentHaltError(
            f"trading-date set inequality: dates only in UNADJUSTED_BARS={missing_in_adjusted}; "
            f"dates only in ADJUSTED_BARS={missing_in_unadjusted}"
        )
    # The diagnostic join leaves this scope here; it is never stored or returned.
    return {
        "orphans_unadj": orphans_unadj,
        "orphans_adj": orphans_adj,
        "join_cardinality": join_cardinality,
        "date_set_equal": bool(date_set_equal),
        "adjustment_unadj": list(adjustment_unadj),
        "adjustment_adj": list(adjustment_adj),
        "index_names": index_names,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 2 (shared fields): NA-aware cell-for-cell agreement with FILES flag discipline
# -------------------------------------------------------------------------------------------------------------------------------

def _clear_files_flag_task_4(
    files: pd.DataFrame,
    stocks: List[str],
    column: str,
) -> None:
    """
    Clear a FILES flag on exactly the given stocks' rows.

    Matches ``stock_code`` as a string and sets the target column to
    ``False`` in place on the matched rows only. Other rows are never
    touched; the caller's manifest object is mutated by design.

    Parameters
    ----------
    files : pd.DataFrame
        The acquisition manifest; must carry ``stock_code`` and the
        target boolean column. Mutated in place.
    stocks : List[str]
        Affected stock codes; an empty list is a no-op.
    column : str
        The boolean flag to clear, e.g. ``"quantity_agreement_ok"`` or
        ``"status_agreement_ok"``.

    Raises
    ------
    SharedFieldHaltError
        If any affected stock has no matching row in ``files``.
    """
    # Compute the files' stock_code as a string-typed Series once per call.
    files_stock = files["stock_code"].astype("string")
    # Walk every affected stock exactly once.
    for stock in stocks:
        # Select the (stock_code, adjustment)-matched rows of this stock.
        mask = files_stock.eq(str(stock))
        # Refuse to silently skip a stock the manifest does not cover.
        if not bool(mask.any()):
            # Raise with the stock named; the manifest is incomplete.
            raise SharedFieldHaltError(
                f"FILES carries no row for mismatching stock {stock!r}; cannot clear {column}"
            )
        # Flip exactly the matched rows to False (never any other row).
        files.loc[mask, column] = False

def _files_boundary_task_4(files: Any) -> pd.DataFrame:
    """
    Validate the FILES manifest required by the shared-field audit.

    Parameters
    ----------
    files : Any
        The object expected to be the acquisition manifest. Not copied;
        the caller may mutate it afterwards.

    Returns
    -------
    pd.DataFrame
        The same, unmutated manifest.

    Raises
    ------
    TypeError
        If ``files`` is not a ``pandas.DataFrame``.
    SharedFieldHaltError
        If ``files`` is empty or lacks any of ``stock_code``,
        ``adjustment``, ``quantity_agreement_ok`` or
        ``status_agreement_ok``; every missing name is listed.
    """
    # The boundary type check.
    if not isinstance(files, pd.DataFrame):
        # Raise naming the observed type.
        raise TypeError(f"files must be a pandas.DataFrame, got {type(files).__name__}")
    # The boundary emptiness check.
    if files.empty:
        # Raise; an empty manifest cannot carry the per-file verdicts.
        raise SharedFieldHaltError("FILES manifest is empty; the shared-field audit has no rows to clear")
    # The pinned columns this step relies on (both views' rows are matched).
    required = ("stock_code", "adjustment", "quantity_agreement_ok", "status_agreement_ok")
    # Compute the complete missing-column set once.
    missing = [column for column in required if column not in files.columns]
    # Halt with every missing name, not just the first.
    if missing:
        # Raise with the complete listing.
        raise SharedFieldHaltError(f"FILES is missing required columns {missing}; observed columns: {list(files.columns)}")
    # Hand the validated (unmutated) manifest back.
    return files

def verify_shared_fields_task_4(
    unadjusted_bars: pd.DataFrame,
    adjusted_bars: pd.DataFrame,
    files: pd.DataFrame,
) -> Dict[str, Any]:
    """
    Audit the shared fields cell-for-cell and update the FILES ledger.

    Step 2 of the cross-view identity stage. Both panels are
    boundary-validated, the shared columns ``quantity``,
    ``tradestatus_raw`` and ``tradable`` are required on both sides, and
    the manifest is boundary-validated. The views are re-joined on their
    compound index; any one-sided coordinate halts immediately as
    masking drift. Each shared field pair is then audited with
    ``_na_equal_agreement_task_4`` under the NA-aware identity.

    On a pure mismatch the matching FILES flag is cleared first on every
    affected stock's rows - ``quantity_agreement_ok`` for quantity
    mismatches, ``status_agreement_ok`` for status mismatches - and the
    function then halts. One-sided NAs halt without touching FILES.

    ``status_agreement_ok`` is the conjunction of the ``tradable`` and
    ``tradestatus_raw`` agreements.

    Parameters
    ----------
    unadjusted_bars : pd.DataFrame
        The nominal view; compound ``(stock_code, trade_date)``
        ``MultiIndex``; must carry the three shared columns. Never
        mutated.
    adjusted_bars : pd.DataFrame
        The provider-adjusted view with the same contract. Never
        mutated.
    files : pd.DataFrame
        The acquisition manifest carrying ``stock_code``,
        ``adjustment``, ``quantity_agreement_ok`` and
        ``status_agreement_ok``. Mutated in place on pure mismatch
        only.

    Returns
    -------
    Dict[str, Any]
        ``quantity_agreement_ok`` (bool), ``tradable_agreement``
        (bool), ``tradestatus_raw_agreement`` (bool),
        ``status_agreement_ok`` (bool), ``quantity_na_pairs`` (int,
        both-NA coordinates for quantity), ``status_na_pairs`` (int,
        both-NA coordinates for ``tradestatus_raw``) and ``artefacts``
        (``pandas.DataFrame`` with columns ``stock_code``,
        ``trade_date``, ``field``, ``unadjusted_value``,
        ``adjusted_value``; empty on a clean run).

    Raises
    ------
    TypeError
        If any argument is not a ``pandas.DataFrame``.
    ViewAlignmentHaltError
        If a panel fails the structural boundary validation or lacks a
        required shared column.
    SharedFieldHaltError
        If the manifest is empty or misses a required column, the views
        hold one-sided coordinates, any pure mismatch exists, or any
        one-sided NA exists.
    """
    # Boundary validation of both panels (raw caller input).
    unadjusted, adjusted = _validate_panel_pair_task_4(unadjusted_bars, adjusted_bars)
    # Require exactly the shared fields on both sides.
    shared_columns = (SHARED_FIELD_QUANTITY, SHARED_FIELD_STATUS, SHARED_FIELD_TRADABLE)
    # Validate the unadjusted view's shared columns.
    _require_columns_task_4(unadjusted, shared_columns, "UNADJUSTED_BARS")
    # Validate the adjusted view's shared columns.
    _require_columns_task_4(adjusted, shared_columns, "ADJUSTED_BARS")
    # Boundary validation of the manifest.
    manifest = _files_boundary_task_4(files)
    # Re-materialise the Step 1 alignment at the granularity of the shared
    # fields; an orphan coordinate here is masking drift and halts.
    merged = pd.merge(
        unadjusted.loc[:, list(shared_columns)],
        adjusted.loc[:, list(shared_columns)],
        how="outer",
        left_index=True,
        right_index=True,
        indicator=True,
        suffixes=("_unadjusted", "_adjusted"),
    )
    # Detect any one-sided coordinate (a halt, never a silent drop).
    one_sided_mask = merged["_merge"] != "both"
    if bool(one_sided_mask.any()):
        # Collect the offending coordinates for the complete halt listing.
        orphan_coordinates = merged.index[one_sided_mask].tolist()
        # Raise with the complete listing; nothing downstream may run.
        raise SharedFieldHaltError(
            f"views are not aligned for the shared-field audit; one-sided coordinates: {orphan_coordinates}"
        )
    # Evaluate the NA-aware agreement for quantity (the shared Int64 field).
    quantity = _na_equal_agreement_task_4(
        merged["quantity_unadjusted"], merged["quantity_adjusted"], SHARED_FIELD_QUANTITY
    )
    # Evaluate the NA-aware agreement for tradestatus_raw (the upstream status field).
    tradestatus_raw = _na_equal_agreement_task_4(
        merged["tradestatus_raw_unadjusted"], merged["tradestatus_raw_adjusted"], SHARED_FIELD_STATUS
    )
    # Evaluate the NA-aware agreement for tradable (the derived status field).
    tradable = _na_equal_agreement_task_4(
        merged["tradable_unadjusted"], merged["tradable_adjusted"], SHARED_FIELD_TRADABLE
    )
    # The register's pinned definition: status := tradable AND tradestatus_raw.
    tradable_agreement = bool(tradable["agreement_ok"])
    tradestatus_raw_agreement = bool(tradestatus_raw["agreement_ok"])
    status_agreement_ok = bool(tradable_agreement and tradestatus_raw_agreement)
    quantity_agreement_ok = bool(quantity["agreement_ok"])
    # Collect every affected stock per flag target (sorted for determinism).
    affected_quantity = sorted({str(row["stock_code"]) for row in quantity["mismatch_rows"]})
    affected_status = sorted(
        {str(row["stock_code"]) for row in tradestatus_raw["mismatch_rows"]}
        | {str(row["stock_code"]) for row in tradable["mismatch_rows"]}
    )
    # Clear the matching FILES flag on exactly each affected stock's two
    # (stock_code, adjustment)-matched rows - before the halt, as pinned.
    _clear_files_flag_task_4(manifest, affected_quantity, "quantity_agreement_ok")
    _clear_files_flag_task_4(manifest, affected_status, "status_agreement_ok")
    # Assemble the artefact table (one row per pure mismatch; empty on a clean run).
    artefact_rows = quantity["mismatch_rows"] + tradestatus_raw["mismatch_rows"] + tradable["mismatch_rows"]
    artefacts = pd.DataFrame(artefact_rows, columns=list(ARTEFACT_COLUMNS))
    # Halt on any pure mismatch (flags already cleared, as the step pins).
    if artefact_rows:
        # Raise with the complete mismatch listing.
        raise SharedFieldHaltError(
            f"shared-field pure mismatch halt ({len(artefact_rows)} cell(s)): {artefact_rows}"
        )
    # Halt on any one-sided NA (masking drift; flags intentionally untouched).
    one_sided_total = (
        int(quantity["one_sided_na_count"])
        + int(tradestatus_raw["one_sided_na_count"])
        + int(tradable["one_sided_na_count"])
    )
    if one_sided_total:
        # Raise with the per-field one-sided counts for forensics.
        raise SharedFieldHaltError(
            f"one-sided NA masking drift between views: quantity={quantity['one_sided_na_count']}, "
            f"tradestatus_raw={tradestatus_raw['one_sided_na_count']}, tradable={tradable['one_sided_na_count']}"
        )
    # Return the structured verdict (all-True flags, zero pairs counted above).
    return {
        "quantity_agreement_ok": quantity_agreement_ok,
        "tradable_agreement": tradable_agreement,
        "tradestatus_raw_agreement": tradestatus_raw_agreement,
        "status_agreement_ok": status_agreement_ok,
        "quantity_na_pairs": int(quantity["both_na_count"]),
        "status_na_pairs": int(tradestatus_raw["both_na_count"]),
        "artefacts": artefacts,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 2 (constancy): per-stock adjustment-factor segment walk on the joined panel
# -------------------------------------------------------------------------------------------------------------------------------

def _derive_segment_ids_task_4(
    close_adj: np.ndarray,
    close: np.ndarray,
) -> Tuple[List[int], int]:
    """
    Derive per-stock adjustment segments from the price ratio.

    With ``r_t = close_adj_t / close_t``, a new segment starts whenever
    a finite ``r_t`` differs from the previous finite ``r`` by at least
    ``SEGMENT_JUMP_THRESHOLD`` (``1e-12``), i.e. at a corporate-action
    date. Non-finite ratios (zero close) are skipped in the comparison
    and inherit the running segment, so they neither trigger a jump nor
    sever a segment. Rows are consumed in the given order, which must be
    ``(stock_code, trade_date)`` sorted by the caller.

    Parameters
    ----------
    close_adj : np.ndarray
        Adjusted close values, cast to float64; must be 1-D.
    close : np.ndarray
        Nominal close values, cast to float64; must be 1-D and the same
        length as ``close_adj``.

    Returns
    -------
    segment_ids : List[int]
        One 1-based segment id per input row, in input order, starting
        at ``FIRST_SEGMENT_ID``.
    jump_count : int
        Number of segment increments observed.

    Raises
    ------
    ValueError
        If either input is not 1-D or the two lengths differ.
    """
    # Coerce defensively to 1-D float64; reject ragged input loudly.
    adj_values = np.asarray(close_adj, dtype=np.float64)
    nom_values = np.asarray(close, dtype=np.float64)
    # Shape discipline: both must be 1-D and equal in length.
    if adj_values.ndim != 1 or nom_values.ndim != 1:
        # Raise naming the observed dimensions.
        raise ValueError(f"close_adj and close must be 1-D; got {adj_values.ndim}-D and {nom_values.ndim}-D")
    if adj_values.shape[0] != nom_values.shape[0]:
        # Raise naming the observed lengths.
        raise ValueError(f"close_adj and close must have equal length; got {adj_values.shape[0]} vs {nom_values.shape[0]}")
    # r = P^adj / P^nom with floating-point divide; 0/0 -> nan, x/0 -> inf.
    with np.errstate(divide="ignore", invalid="ignore"):
        ratio = adj_values / nom_values
    # The walk state: running segment id, last finite r, count of jumps.
    segment = FIRST_SEGMENT_ID
    last_finite: Optional[float] = None
    jump_count = 0
    derived: List[int] = []
    # Walk the rows in trade_date order (the caller pre-sorted once).
    for value in ratio:
        # Non-finite rows are skipped in the walk and inherit the segment.
        if not np.isfinite(value):
            # Record the running segment; do not touch last_finite or the counter.
            derived.append(segment)
            continue
        # A jump fires iff |r_t - r_{t-1 over finite rows}| >= 1e-12.
        if last_finite is not None and abs(float(value) - last_finite) >= SEGMENT_JUMP_THRESHOLD:
            # Increment the segment id (a corporate action broke kappa_i).
            segment += 1
            # Count the jump for the per-stock evidence report.
            jump_count += 1
        # Adopt this finite r as the comparison basis for the next finite row.
        last_finite = float(value)
        # Record the (possibly incremented) segment for this row.
        derived.append(segment)
    # Return the per-row ids and the total jump count.
    return derived, jump_count

def verify_adjustment_factor_constancy_task_4(
    unadjusted_bars: pd.DataFrame,
    adjusted_bars: pd.DataFrame,
) -> Dict[str, Any]:
    """
    Verify the adjustment-factor constancy invariant of both views.

    Step 2 of Task 4. Materialises a fresh outer join of the two panels
    on the compound ``(stock_code, trade_date)`` index, keeping
    ``close`` from the nominal view and ``close_adj`` plus
    ``adj_factor_segment_id`` from the adjusted view, and halts if any
    coordinate is one-sided. The ``_merge`` indicator column is then
    dropped, but the joined panel is retained and walked by the
    per-stock loop. The panel is sorted once by the compound key, so
    each stock group is visited in calendar order without a re-sort,
    and for each stock the ratio

        r_t = close_adj_t / close_t

    is walked by :func:`_derive_segment_ids_task_4`, which increments a
    1-based segment id whenever ``|r_t - r_{t-1}| >= 1e-12``.
    Non-finite ratios are skipped in the walk and can never trigger a
    jump; masked cells are never imputed. Ratios are float64.

    The derived ids are compared position by position against the
    vendor's ``adj_factor_segment_id`` claim; one disagreement on one
    stock halts the pipeline with the complete listing. Neither input
    frame is mutated.

    Parameters
    ----------
    unadjusted_bars : pd.DataFrame
        The nominal view; compound ``(stock_code, trade_date)``
        MultiIndex; must carry ``close``.
    adjusted_bars : pd.DataFrame
        The provider-adjusted view; same index contract; must carry
        ``close_adj`` and ``adj_factor_segment_id`` (1-based integers).

    Returns
    -------
    Dict[str, Any]
        ``segment_concordance_ok`` (``True`` whenever returned),
        ``derived_segment_ids`` (dict mapping each stock code, as
        ``str``, to the list of 1-based segment ids in calendar order,
        one entry per row of that stock) and ``jump_count_by_stock``
        (dict mapping the same codes to the number of increments, zero
        included).

    Raises
    ------
    TypeError
        If either panel is not a ``pandas.DataFrame`` (raised by the
        shared boundary validator).
    AdjustmentFactorHaltError
        If the views carry one-sided coordinates, if any vendor
        ``adj_factor_segment_id`` cell is NA, or if derived and vendor
        ids disagree on any row.

    Notes
    -----
    The vendor segment column is only ever compared against the derived
    ids; it is never used to build them.
    """
    # Boundary validation of both panels (raw caller input).
    unadjusted, adjusted = _validate_panel_pair_task_4(unadjusted_bars, adjusted_bars)
    # The nominal close must exist on the unadjusted side.
    _require_columns_task_4(unadjusted, ("close",), "UNADJUSTED_BARS")
    # The adjusted close and the vendor's segment claim must exist.
    _require_columns_task_4(adjusted, ("close_adj", "adj_factor_segment_id"), "ADJUSTED_BARS")
    # Re-materialise the Step 1 join at the granularity of this test.
    merged = pd.merge(
        unadjusted.loc[:, ["close"]],
        adjusted.loc[:, ["close_adj", "adj_factor_segment_id"]],
        how="outer",
        left_index=True,
        right_index=True,
        indicator=True,
    )
    # Detect any one-sided coordinate (a halt, never a silent drop).
    one_sided_mask = merged["_merge"] != "both"
    if bool(one_sided_mask.any()):
        # Collect the offending coordinates for the complete halt listing.
        orphan_coordinates = merged.index[one_sided_mask].tolist()
        # Raise with the complete listing.
        raise AdjustmentFactorHaltError(
            f"views are not aligned for the constancy test; one-sided coordinates: {orphan_coordinates}"
        )
    # Drop the indicator; the join is a diagnostic here as in Step 1.
    merged = merged.drop(columns=["_merge"])
    # Pre-sort ONCE by the compound key; every group is then trade_date-ascending.
    merged = merged.sort_index()
    # Containers keyed by stock code (str), in compound-sort order.
    derived_segment_ids: Dict[str, List[int]] = {}
    jump_count_by_stock: Dict[str, int] = {}
    # Evidence collected for the vendor concordance halts.
    mismatch_records: List[str] = []
    # One pass over the stocks; no re-sort inside the walk.
    for stock_code, group in merged.groupby(level="stock_code", sort=False):
        # Extract float64 vectors with NA -> nan discipline.
        close_values = group["close"].to_numpy(dtype=np.float64, na_value=np.nan)
        close_adj_values = group["close_adj"].to_numpy(dtype=np.float64, na_value=np.nan)
        # Derive the 1-based segment ids and the jump count for this stock.
        derived_ids, jumps = _derive_segment_ids_task_4(close_adj_values, close_values)
        # Key the outputs by the stock's level value as a plain string.
        stock = str(stock_code)
        # Record the derived ids (per-stock calendar order) and jump count.
        derived_segment_ids[stock] = list(derived_ids)
        jump_count_by_stock[stock] = int(jumps)
        # The vendor's claim: every cell must be a concrete Int64 for verification.
        vendor = group["adj_factor_segment_id"]
        if bool(vendor.isna().any()):
            # Any NA vendor cell is an unverifiable claim - a halt, not a skip.
            raise AdjustmentFactorHaltError(
                f"stock {stock!r} carries NA in vendor adj_factor_segment_id; the claim cannot be verified"
            )
        # Materialise the vendor ids as plain ints for the comparison.
        vendor_ids = [int(value) for value in vendor.to_numpy()]
        # Compare derived vs vendor position-by-position.
        for position, (derived, vendor_id) in enumerate(zip(derived_ids, vendor_ids)):
            if derived != vendor_id:
                # Record the disagreement with full coordinate evidence.
                coordinate = group.index[position]
                mismatch_records.append(
                    f"stock {stock!r} date {coordinate[1]} position {position}: derived {derived} vs vendor {vendor_id}"
                )
    # Any disagreement anywhere halts under the the-check-is-testable rule.
    if mismatch_records:
        # Raise with the complete disagreement listing.
        raise AdjustmentFactorHaltError(
            f"vendor adj_factor_segment_id disagreement on {len(mismatch_records)} row(s): "
            + "; ".join(mismatch_records)
        )
    # Return the concordance verdict with the derived evidence map.
    return {
        "segment_concordance_ok": True,
        "derived_segment_ids": derived_segment_ids,
        "jump_count_by_stock": jump_count_by_stock,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 3: the FrozenFeaturePanel embargo facade
# -------------------------------------------------------------------------------------------------------------------------------

def _iter_string_components_task_4(key: Any):
    """
    Yield every string component embedded in an arbitrary key.

    Walks tuples, lists and sets recursively, yielding the ``str``
    values found at any depth. For ``pandas.Index`` and
    ``pandas.Series`` inputs only directly contained string values are
    yielded; non-string elements such as nested tuples are not recursed
    into. Slices, integers and other objects carry no column names and
    yield nothing.

    Parameters
    ----------
    key : Any
        Any key accepted by ``__getitem__`` or by the ``loc``/``iloc``
        indexers: a bare label, a tuple of labels and slices, a list or
        set of labels, an Index, or a Series of labels.

    Yields
    ------
    str
        Each embedded string, in depth-first order of the key.
    """
    # A bare string is a candidate column label on its own.
    if isinstance(key, str):
        # Yield the string itself.
        yield key
        # Stop; nothing further to walk.
        return
    # A tuple (e.g. .loc[:, key]) recurses component-wise.
    if isinstance(key, tuple):
        for component in key:
            # Yield every string inside each tuple component.
            yield from _iter_string_components_task_4(component)
        return
    # A list or set of labels recurses element-wise.
    if isinstance(key, (list, set)):
        for component in key:
            # Yield every string inside each element.
            yield from _iter_string_components_task_4(component)
        return
    # A pandas Index (e.g. .loc[:, ['a', 'b_adj']]) recurses value-wise.
    if isinstance(key, pd.Index):
        for value in key:
            # Yield each string value of the Index.
            if isinstance(value, str):
                yield value
        return
    # A string Series (e.g. .loc[:, series_of_labels]) recurses value-wise.
    if isinstance(key, pd.Series):
        for value in key:
            # Yield each string value of the Series.
            if isinstance(value, str):
                yield value
        return
    # Slices, integers, booleans, callables and everything else carry no
    # column names; nothing to yield.
    return

def _external_caller_frame_task_4():
    """
    Return the first stack frame outside this module.

    Walks the live call stack outward from inside the embargo machinery
    to the first frame whose file differs from this module's
    ``__file__``, i.e. the code that attempted the embargoed access. If
    the walk runs off the top of the stack, the walker's own frame is
    returned, so the result is never ``None``.

    Returns
    -------
    types.FrameType
        The first frame whose code file is not this module's file, or
        the walker's own frame when no such frame exists.
    """
    # Take a reference to the current frame (the walker itself).
    current = inspect.currentframe()
    # Walk outward while the frame belongs to this module's file.
    caller = current
    while caller is not None and os.path.normcase(caller.f_code.co_filename) == os.path.normcase(current.f_code.co_filename):
        # Advance one frame up the stack.
        caller = caller.f_back
    # If we walked off the top of the stack, fall back to the current frame.
    if caller is None:
        # Never return None; the fraud_log contract requires a frame.
        caller = current
    # Release the strong self-reference chain we do not need to keep.
    del current
    # Hand the external caller frame back.
    return caller

class FrozenFeaturePanel:
    """
    Embargoed facade over a panel frame (not a DataFrame subclass).

    Wraps a panel frame and intercepts every request for a column,
    attribute or indexer key with a string component ending in ``_adj``.
    With ``allow_adj=False`` (FEATURES_PANEL) such a request is appended
    to :attr:`fraud_log` as a ``(timestamp, key, caller_frame)`` tuple
    and then raises :class:`PermissionError`; with ``allow_adj=True``
    (LABEL_PANEL) item and ``loc``/``iloc`` key access passes through
    unlogged, but an attribute name ending in ``_adj`` is still
    intercepted and raises :class:`PermissionError`, because
    ``__getattr__`` does not consult ``allow_adj``. Legitimate access -
    other items, attributes, and the ``loc``/``iloc`` proxies -
    delegates to the wrapped frame, and ``join``/``copy`` re-wrap their
    result in a fresh facade. ``columns``, ``dtypes``, ``shape``,
    ``index`` and ``len`` are pass-throughs. The wrapped frame is never
    exposed.

    Attributes
    ----------
    fraud_log : list[tuple[pandas.Timestamp, str, Any]]
        Append-only audit trail of attempted ``_adj`` accesses; each
        entry is a ``(timestamp in Asia/Shanghai, key, caller_frame)``
        tuple. Empty when the embargo was never probed.
    """

    def __init__(self, frame: pd.DataFrame, allow_adj: bool) -> None:
        """
        Wrap a panel frame in the embargo facade.

        Parameters
        ----------
        frame : pd.DataFrame
            The panel to wrap; not copied, the facade proxies this
            object.
        allow_adj : bool
            ``False`` for FEATURES_PANEL (any ``_adj`` key halts) and
            ``True`` for LABEL_PANEL (item and indexer keys with
            ``_adj`` pass unlogged; attribute names ending in ``_adj``
            still halt).

        Raises
        ------
        TypeError
            If ``frame`` is not a ``pandas.DataFrame``, or ``allow_adj``
            is neither a ``bool`` nor a ``numpy.bool_``.
        """
        # The wrapped object must be a DataFrame; refuse anything else.
        if not isinstance(frame, pd.DataFrame):
            # Raise naming the observed type.
            raise TypeError(f"frame must be a pandas.DataFrame, got {type(frame).__name__}")
        # The access switch must be a boolean (numpy bools are acceptable).
        if not isinstance(allow_adj, (bool, np.bool_)):
            # Raise naming the observed type.
            raise TypeError(f"allow_adj must be a boolean, got {type(allow_adj).__name__}")
        # Store the wrapped frame (facade holds the caller's object).
        self._frame = frame
        # Normalise the switch to a plain Python bool.
        self._allow_adj = bool(allow_adj)
        # The append-only audit trail; empty means the embargo held.
        self.fraud_log: List[Tuple[pd.Timestamp, str, Any]] = []

    @property
    def columns(self) -> pd.Index:
        """
        Return the column labels of the wrapped frame.

        Returns
        -------
        pd.Index
            The wrapped frame's columns, unmodified.
        """
        # Pass through to the wrapped frame's column Index.
        return self._frame.columns

    @property
    def dtypes(self) -> pd.Series:
        """
        Return per-column dtypes of the wrapped frame.

        Returns
        -------
        pd.Series
            The wrapped frame's ``dtypes`` Series, keyed by column
            label.
        """
        # Pass through to the wrapped frame's dtype Series.
        return self._frame.dtypes

    @property
    def shape(self) -> Tuple[int, int]:
        """
        Return the ``(rows, columns)`` shape of the wrapped frame.

        Returns
        -------
        Tuple[int, int]
            Row count and column count of the wrapped frame.
        """
        # Pass through to the wrapped frame's shape tuple.
        return self._frame.shape

    @property
    def index(self) -> pd.Index:
        """
        Return the index of the wrapped frame.

        Returns
        -------
        pd.Index
            The compound ``(stock_code, trade_date)`` index, unchanged.
        """
        # Pass through to the wrapped frame's index.
        return self._frame.index

    def __len__(self) -> int:
        """
        Return the row count of the wrapped frame.

        Returns
        -------
        int
            Number of rows in the wrapped frame.
        """
        # Pass through to the wrapped frame's length.
        return len(self._frame)

    def __repr__(self) -> str:
        """
        Return a compact repr that never leaks adjusted contents.

        Returns
        -------
        str
            ``FrozenFeaturePanel(shape=..., allow_adj=...)``; only the
            shape and the access switch are shown.
        """
        # Report only shape and the access switch; never the data.
        return f"FrozenFeaturePanel(shape={self._frame.shape}, allow_adj={self._allow_adj})"

    def _violation(self, key: str) -> None:
        """
        Record an attempted ``_adj`` access, then raise.

        Captures the external caller frame, appends
        ``(timestamp, key, caller_frame)`` to :attr:`fraud_log`, and
        raises unconditionally.

        Parameters
        ----------
        key : str
            The offending ``_adj``-ending key as presented.

        Raises
        ------
        PermissionError
            Always; the raise is the embargo halt.
        """
        # Capture the external caller frame BEFORE building the record.
        caller = _external_caller_frame_task_4()
        # Append the (timestamp, key, caller_frame) tuple to the audit trail.
        self.fraud_log.append((pd.Timestamp.now(tz="Asia/Shanghai"), str(key), caller))
        # Raise unconditionally with the complete context in the message.
        raise PermissionError(
            f"FrozenFeaturePanel embargo: refusal to serve {key!r}; the key ends with '_adj' "
            f"and this panel was constructed with allow_adj=False "
            f"(feature path embargo, P11_ACQUISITION_PARAMETERS.adjusted_view_permitted_in_features is False)"
        )

    def _maybe_guard_key(self, key: Any) -> None:
        """
        Screen a key for embargoed ``_adj`` components.

        Returns immediately when ``allow_adj`` is True; otherwise walks
        every embedded string component of ``key`` and halts on the
        first one ending in ``_adj``.

        Parameters
        ----------
        key : Any
            The key exactly as passed to ``__getitem__`` or to a ``loc``
            / ``iloc`` proxy: a label, tuple, list, Index or Series.

        Raises
        ------
        PermissionError
            If an embedded string ends with ``_adj`` while ``allow_adj``
            is False; the attempt is logged before the raise.
        """
        # The label path (LABEL_PANEL) permits _adj access unconditionally.
        if self._allow_adj:
            # Nothing to screen; legitimate label-side access.
            return
        # Walk every embedded string component; the first _adj offender halts.
        for candidate in _iter_string_components_task_4(key):
            # Any string ending with _adj is embargoed in the feature path.
            if candidate.endswith("_adj"):
                # Log before the halt, as the study output requires.
                self._violation(candidate)

    def __getitem__(self, key: Any) -> Any:
        """
        Return ``frame[key]``, intercepting embargoed keys.

        Parameters
        ----------
        key : Any
            Column label, list of labels, or any other key accepted by
            the wrapped frame.

        Returns
        -------
        Any
            Whatever the wrapped frame returns for ``key``.

        Raises
        ------
        PermissionError
            If ``key`` holds an ``_adj`` component and ``allow_adj`` is
            False.
        """
        # Screen the key (logging and raising on any _adj component).
        self._maybe_guard_key(key)
        # Pass the legitimate access through to the wrapped frame.
        return self._frame[key]

    def __setitem__(self, key: Any, value: Any) -> None:
        """
        Assign ``frame[key] = value``, intercepting embargoed keys.

        Parameters
        ----------
        key : Any
            Column label or other key accepted by the wrapped frame.
        value : Any
            Value or array assigned to the wrapped frame.

        Raises
        ------
        PermissionError
            If ``key`` holds an ``_adj`` component and ``allow_adj`` is
            False.
        """
        # Screen the key (logging and raising on any _adj component).
        self._maybe_guard_key(key)
        # Perform the legitimate assignment on the wrapped frame.
        self._frame[key] = value

    def __getattr__(self, name: str) -> Any:
        """
        Proxy attribute access to the wrapped frame.

        Parameters
        ----------
        name : str
            Attribute name requested on the facade.

        Returns
        -------
        Any
            The corresponding attribute of the wrapped frame.

        Raises
        ------
        PermissionError
            If ``name`` ends with ``_adj``, regardless of ``allow_adj``
            (the attribute embargo is unconditional, so an
            ``allow_adj=True`` LABEL_PANEL facade raises here too); the
            attempt is logged first.
        AttributeError
            If the wrapped frame has no such attribute.
        """
        # Embargoed attribute names halt here, even before the frame is consulted.
        if name.endswith("_adj"):
            # Log before the halt, as the study output requires.
            self._violation(name)
        # Proxy legitimate attributes (e.g. frame.close) to the wrapped frame.
        try:
            # Delegate the attribute lookup to the wrapped frame.
            return getattr(self._frame, name)
        except AttributeError as exc:
            # Re-raise as AttributeError with the facade's identity spelled out.
            raise AttributeError(f"FrozenFeaturePanel has no attribute or column {name!r}: {exc}") from exc

    @property
    def loc(self):
        """
        Return an intercepting proxy of the wrapped frame's ``.loc``.

        Returns
        -------
        _FrozenLocIndexer
            A new proxy bound to this facade; it screens each key
            before delegating to the wrapped frame's ``.loc``.
        """
        # Build the .loc proxy bound to this facade.
        return _FrozenLocIndexer(self)

    @property
    def iloc(self):
        """
        Return an intercepting proxy of the wrapped frame's ``.iloc``.

        Returns
        -------
        _FrozenILocIndexer
            A new proxy bound to this facade; it screens each key
            before delegating to the wrapped frame's ``.iloc``.
        """
        # Build the .iloc proxy bound to this facade.
        return _FrozenILocIndexer(self)

    def join(self, *args: Any, **kwargs: Any) -> "FrozenFeaturePanel":
        """
        Join against other frames and re-wrap the result.

        Parameters
        ----------
        *args : Any
            Positional arguments forwarded to the wrapped frame's
            ``join``.
        **kwargs : Any
            Keyword arguments forwarded to the wrapped frame's
            ``join``.

        Returns
        -------
        "FrozenFeaturePanel"
            A new facade over the join result, carrying the same
            ``allow_adj`` switch and an empty ``fraud_log``.
        """
        # Delegate to the wrapped frame's join, re-wrapping the result.
        return FrozenFeaturePanel(self._frame.join(*args, **kwargs), self._allow_adj)

    def copy(self, deep: bool = True) -> "FrozenFeaturePanel":
        """
        Copy the wrapped frame and wrap the copy in a new facade.

        Parameters
        ----------
        deep : bool, optional
            Forwarded to ``DataFrame.copy``. Default is ``True``.

        Returns
        -------
        "FrozenFeaturePanel"
            A new facade over the copy, carrying the same ``allow_adj``
            switch and a fresh, empty ``fraud_log``.
        """
        # Delegate to the wrapped frame's copy, re-wrapping the result.
        # The audit trail starts fresh: a copy is a new audit scope.
        return FrozenFeaturePanel(self._frame.copy(deep=deep), self._allow_adj)

class _FrozenIndexer:
    """
    Base intercepting proxy for a facade's ``loc``/``iloc`` accessors.
    """

    def __init__(self, parent: FrozenFeaturePanel, kind: str) -> None:
        """
        Bind the proxy to its owning facade.

        Parameters
        ----------
        parent : FrozenFeaturePanel
            The facade holding the wrapped frame and the audit trail.
        kind : str
            Either ``"loc"`` or ``"iloc"``.
        """
        # Record the owning facade (its embargo and its audit trail).
        self._parent = parent
        # Record which indexer we proxy.
        self._kind = kind

    def _target(self):
        """
        Return the wrapped frame's ``.loc`` or ``.iloc`` accessor.

        Returns
        -------
        pd.core.indexing._LocIndexer | pd.core.indexing._iLocIndexer
            The wrapped frame's ``.loc`` when this proxy's kind is
            ``"loc"``, otherwise its ``.iloc``.
        """
        # Resolve the accessor once per call (context-free delegation).
        return self._parent._frame.loc if self._kind == "loc" else self._parent._frame.iloc

    def __getitem__(self, key: Any) -> Any:
        """
        Return ``.loc[key]`` or ``.iloc[key]`` after screening.

        Parameters
        ----------
        key : Any
            Row/column key as passed to the indexer.

        Returns
        -------
        Any
            Whatever the wrapped frame's accessor returns for ``key``.

        Raises
        ------
        PermissionError
            If ``key`` holds an ``_adj`` component and the owning facade
            has ``allow_adj`` False.
        """
        # Screen the key (logging and raising on any _adj component).
        self._parent._maybe_guard_key(key)
        # Delegate the legitimate access to the wrapped frame's accessor.
        return self._target()[key]

    def __setitem__(self, key: Any, value: Any) -> None:
        """
        Assign ``.loc[key] = value`` or ``.iloc[key] = value``.

        Parameters
        ----------
        key : Any
            Row/column key as passed to the indexer.
        value : Any
            Value assigned on the wrapped frame.

        Raises
        ------
        PermissionError
            If ``key`` holds an ``_adj`` component and the owning facade
            has ``allow_adj`` False.
        """
        # Screen the key (logging and raising on any _adj component).
        self._parent._maybe_guard_key(key)
        # Perform the legitimate assignment on the wrapped frame.
        self._target()[key] = value

class _FrozenLocIndexer(_FrozenIndexer):
    """
    Intercepting proxy of the wrapped frame's ``.loc`` indexer.
    """

    def __init__(self, parent: FrozenFeaturePanel) -> None:
        """
        Bind a ``.loc`` proxy to its facade.

        Parameters
        ----------
        parent : FrozenFeaturePanel
            The owning facade.
        """
        # Initialise the base with the pinned indexer kind.
        super().__init__(parent, "loc")

class _FrozenILocIndexer(_FrozenIndexer):
    """
    Intercepting proxy of the wrapped frame's ``.iloc`` indexer.
    """

    def __init__(self, parent: FrozenFeaturePanel) -> None:
        """
        Bind an ``.iloc`` proxy to its facade.

        Parameters
        ----------
        parent : FrozenFeaturePanel
            The owning facade.
        """
        # Initialise the base with the pinned indexer kind.
        super().__init__(parent, "iloc")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Step 3: the two-panel split behind the embargo
# -------------------------------------------------------------------------------------------------------------------------------

def split_panels_task_4(
    unadjusted_bars: pd.DataFrame,
    adjusted_bars: pd.DataFrame,
    *,
    features_root: str,
    labels_root: str,
) -> Dict[str, Any]:
    """
    Split the views into feature and label panels and persist them.

    Step 3 of Task 4. Builds FEATURES_PANEL from ``unadjusted_bars``
    with exactly the module's pinned feature columns (``stock_code``,
    ``trade_date``, ``open``, ``high``, ``low``, ``close``, ``volume``,
    ``amount``, ``tradable``, ``tradestatus_raw``; no ``_adj`` column)
    and LABEL_PANEL from ``adjusted_bars`` with exactly the pinned label
    columns (``stock_code``, ``trade_date``, ``close_adj``,
    ``adj_factor``, ``tradable``). Both blocks are deep copies in source
    row order.

    The two roots are expanded, resolved and required to be distinct and
    non-nesting; each is created if missing, and each panel is written
    as a CSV named by the module's ``FEATURES_FILE_NAME`` /
    ``LABELS_FILE_NAME`` beneath its own root, with ``index=False``. The
    panels are then wrapped in :class:`FrozenFeaturePanel` with
    ``allow_adj=False`` (features) and ``allow_adj=True`` (labels).

    Parameters
    ----------
    unadjusted_bars : pd.DataFrame
        The nominal view; compound ``(stock_code, trade_date)``
        MultiIndex; must carry every pinned feature column. Never
        mutated.
    adjusted_bars : pd.DataFrame
        The provider-adjusted view; same index contract; must carry
        every pinned label column. Never mutated.
    features_root : str
        Directory for FEATURES_PANEL's CSV; created if missing.
    labels_root : str
        Directory for LABEL_PANEL's CSV; created if missing.

    Returns
    -------
    Dict[str, Any]
        ``features_panel`` and ``labels_panel`` (the two facades),
        ``features_path`` and ``labels_path`` (absolute ``str`` CSV
        paths beneath their respective roots).

    Raises
    ------
    TypeError
        If either panel is not a ``pandas.DataFrame``, or a root is not
        a ``str``.
    ValueError
        If a root is empty or whitespace-only.
    PanelSplitHaltError
        If a pinned column is missing from either source panel, or the
        two resolved roots are equal or nested inside one another.
    """
    # Boundary validation of both panels (raw caller input).
    unadjusted, adjusted = _validate_panel_pair_task_4(unadjusted_bars, adjusted_bars)
    # The roots must be non-empty strings (paths are caller-supplied input).
    for root_name, root_value in (("features_root", features_root), ("labels_root", labels_root)):
        # Type check first.
        if not isinstance(root_value, str):
            # Raise naming the observed type.
            raise TypeError(f"{root_name} must be a str, got {type(root_value).__name__}")
        # Blank strings are caller errors, never silent defaults.
        if not root_value.strip():
            # Raise naming the parameter.
            raise ValueError(f"{root_name} must be a non-empty path string")
    # Resolve both roots to absolute, symlink-free paths.
    resolved_features = Path(features_root).expanduser().resolve()
    resolved_labels = Path(labels_root).expanduser().resolve()
    # Distinctness is mandatory: equal roots are a namespace breach.
    if resolved_features == resolved_labels:
        # Raise; a read path would span both roles.
        raise PanelSplitHaltError(f"features_root and labels_root resolve to the same directory: {resolved_features}")
    # Nesting is equally a breach (one namespace would sit inside the other).
    if resolved_labels in resolved_features.parents or resolved_features in resolved_labels.parents:
        # Raise naming both resolved roots.
        raise PanelSplitHaltError(
            f"features_root {resolved_features} and labels_root {resolved_labels} nest; namespaces must be distinct"
        )
    # Pinned column audit of the unadjusted view (features source).
    missing_features = [column for column in FEATURES_COLUMNS if column not in unadjusted.columns]
    # Pinned column audit of the adjusted view (labels source).
    missing_labels = [column for column in LABELS_COLUMNS if column not in adjusted.columns]
    # Halt with the complete missing listing for either side.
    if missing_features or missing_labels:
        # Raise naming both sides' missing names.
        raise PanelSplitHaltError(
            f"pinned column contract breached: UNADJUSTED_BARS missing {missing_features}, "
            f"ADJUSTED_BARS missing {missing_labels}"
        )
    # Build the features block as an independent copy (fresh index, source order).
    features_frame = unadjusted.loc[:, list(FEATURES_COLUMNS)].copy(deep=True)
    # Build the labels block as an independent copy (fresh index, source order).
    labels_frame = adjusted.loc[:, list(LABELS_COLUMNS)].copy(deep=True)
    # Create the two namespaces if missing (parents included).
    resolved_features.mkdir(parents=True, exist_ok=True)
    resolved_labels.mkdir(parents=True, exist_ok=True)
    # Pin the absolute CSV targets beneath their own roots.
    features_path = resolved_features / FEATURES_FILE_NAME
    labels_path = resolved_labels / LABELS_FILE_NAME
    # Persist FEATURES_PANEL (stock_code/trade_date are columns; index=False).
    features_frame.to_csv(features_path, index=False)
    # Persist LABEL_PANEL beneath its own namespace (index=False likewise).
    labels_frame.to_csv(labels_path, index=False)
    # Wrap the feature frame in the embargo facade (allow_adj=False).
    features_panel = FrozenFeaturePanel(features_frame, allow_adj=False)
    # Wrap the label frame in the label-side facade (allow_adj=True).
    labels_panel = FrozenFeaturePanel(labels_frame, allow_adj=True)
    # Return the bundle with absolute paths beneath their roots.
    return {
        "features_panel": features_panel,
        "labels_panel": labels_panel,
        "features_path": str(features_path),
        "labels_path": str(labels_path),
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_4(
    FILES: pd.DataFrame,
    UNADJUSTED_BARS: pd.DataFrame,
    ADJUSTED_BARS: pd.DataFrame,
    STUDY_CONFIG: Dict[str, Any],
    *,
    expected_join_cardinality: int,
    features_root: str,
    labels_root: str,
) -> Dict[str, Any]:
    """
    Run Task 4's alignment, shared-field, constancy and split steps.

    Executes, in the pinned order: the view-alignment diagnostic with
    the caller's cardinality pin, the shared-field agreement check
    (which may clear FILES flags in place), the adjustment-factor
    constancy test, and the irreversible two-panel split with
    persistence. Any halt propagates as an exception, so no partial
    bundle is returned. Neither the input frames nor the caller's
    configuration are otherwise mutated.

    ``STUDY_CONFIG`` is deep-copied and the copy gains ``features_root``
    and ``labels_root`` under its ``P11_ACQUISITION_PARAMETERS`` section
    as reproduction/reference entries. The feature facade's
    ``fraud_log`` is copied into the returned bundle.

    Parameters
    ----------
    FILES : pd.DataFrame
        The acquisition manifest consulted by the shared-field step; its
        agreement flags may be set from True to False on pure mismatch.
    UNADJUSTED_BARS : pd.DataFrame
        The nominal view; compound ``(stock_code, trade_date)``
        MultiIndex.
    ADJUSTED_BARS : pd.DataFrame
        The provider-adjusted view; same index contract.
    STUDY_CONFIG : Dict[str, Any]
        The study configuration; deep-copied, never mutated in place.
    expected_join_cardinality : int
        Join cardinality expected at this pipeline stage: 29,080 before
        masking and 29,078 after masking.
    features_root : str
        Directory for FEATURES_PANEL's CSV.
    labels_root : str
        Directory for LABEL_PANEL's CSV.

    Returns
    -------
    Dict[str, Any]
        ``alignment`` (view-alignment diagnostic), ``shared_fields``,
        ``constancy``, ``features_panel`` and ``labels_panel`` (the two
        facades), ``features_path`` and ``labels_path`` (absolute
        ``str`` paths), ``study_config`` (the augmented deep copy) and
        ``fraud_log`` (list of attempted ``_adj`` accesses; empty on a
        clean feature path).

    Raises
    ------
    TypeError
        If a frame argument is not a ``pandas.DataFrame``, if
        ``STUDY_CONFIG`` is not a ``dict``, or if
        ``expected_join_cardinality`` is a ``bool`` or not an integer.
    ValueError
        If ``STUDY_CONFIG`` lacks a dict-valued
        ``P11_ACQUISITION_PARAMETERS`` section.
    ViewAlignmentHaltError
        If the alignment diagnostic halts.
    SharedFieldHaltError
        If the shared-field agreement check halts.
    AdjustmentFactorHaltError
        If the constancy test halts.
    PanelSplitHaltError
        If the panel split halts.
    """
    # Boundary type audit: every frame argument must be a DataFrame.
    for name, frame in (("FILES", FILES), ("UNADJUSTED_BARS", UNADJUSTED_BARS), ("ADJUSTED_BARS", ADJUSTED_BARS)):
        # Reject non-DataFrame input before any work begins.
        if not isinstance(frame, pd.DataFrame):
            # Raise naming the argument and the observed type.
            raise TypeError(f"{name} must be a pandas.DataFrame, got {type(frame).__name__}")
    # The configuration must be a dict (audited upstream by Task 1; not re-audited here).
    if not isinstance(STUDY_CONFIG, dict):
        # Raise naming the observed type.
        raise TypeError(f"STUDY_CONFIG must be a dict, got {type(STUDY_CONFIG).__name__}")
    # The pin is validated inside Step 1 (single source of the assertion).
    if isinstance(expected_join_cardinality, bool) or not isinstance(expected_join_cardinality, (int, np.integer)):
        # Raise naming the observed type (bools excluded deliberately).
        raise TypeError(
            f"expected_join_cardinality must be an integer (bools excluded), "
            f"got {type(expected_join_cardinality).__name__}"
        )
    # The acquisition namespace must exist and be a dict (features_root lands here).
    p11_section = STUDY_CONFIG.get("P11_ACQUISITION_PARAMETERS")
    if not isinstance(p11_section, dict):
        # Raise naming the missing path precisely.
        raise ValueError(
            "STUDY_CONFIG is missing a dict-valued 'P11_ACQUISITION_PARAMETERS' section; "
            "features_root/labels_root have nowhere to be registered"
        )
    # Step 1: the pairwise alignment diagnostic (halts before anything mutates).
    alignment = verify_view_alignment_task_4(
        UNADJUSTED_BARS,
        ADJUSTED_BARS,
        expected_join_cardinality=expected_join_cardinality,
    )
    # Step 2a: the shared-field cell-for-cell agreement (FILES flags on pure mismatch).
    shared_fields = verify_shared_fields_task_4(
        UNADJUSTED_BARS,
        ADJUSTED_BARS,
        FILES,
    )
    # Step 2b: the adjustment-factor constancy test (derived vs vendor).
    constancy = verify_adjustment_factor_constancy_task_4(
        UNADJUSTED_BARS,
        ADJUSTED_BARS,
    )
    # Step 3: the irreversible two-panel split (persisted beneath distinct roots).
    split = split_panels_task_4(
        UNADJUSTED_BARS,
        ADJUSTED_BARS,
        features_root=features_root,
        labels_root=labels_root,
    )
    # Augment a deepcopy of the configuration; the caller's object is untouched.
    study_config = copy.deepcopy(STUDY_CONFIG)
    # The [R] reproduce/reference entries: values are the passed roots verbatim.
    study_config["P11_ACQUISITION_PARAMETERS"]["features_root"] = features_root
    study_config["P11_ACQUISITION_PARAMETERS"]["labels_root"] = labels_root
    # Package the pinned ten-key return mapping (pinned order, complete).
    return {
        "alignment": alignment,
        "shared_fields": shared_fields,
        "constancy": constancy,
        "features_panel": split["features_panel"],
        "labels_panel": split["labels_panel"],
        "features_path": split["features_path"],
        "labels_path": split["labels_path"],
        "study_config": study_config,
        "fraud_log": list(split["features_panel"].fraud_log),
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 4, module self-registration guard (loader compatibility; zero semantic change)
# -------------------------------------------------------------------------------------------------------------------------------

def _register_executing_module_task_4() -> None:
    """
    Register this module in ``sys.modules`` if the loader skipped it.

    Some loaders execute a module through
    ``importlib.util.spec_from_file_location`` and ``exec_module``
    without inserting it into ``sys.modules``; under such a loader
    ``inspect.getfile`` cannot resolve the classes defined here, because
    it looks up ``sys.modules[class.__module__].__file__``. When
    ``__name__`` is absent from ``sys.modules`` this function builds a
    ``types.ModuleType`` carrying this module's ``__doc__`` and a
    snapshot of ``globals()`` (shared references, not copies), sets its
    ``__file__`` to the fixed path ``<cwd>/task_responses/task_4.py``,
    and installs it under ``__name__``. Under any registering import it
    returns immediately without side effects.

    Notes
    -----
    The registration mutates process-global state (``sys.modules``) and
    is intended to be called once, at module execution time.
    """
    # Under any registering loader, sys.modules already holds this module.
    if __name__ in sys.modules:
        # No-op: nothing to repair.
        return
    # Build the faithful registration: same name, real file, shared names.
    shim = types.ModuleType(__name__)
    # Record the true on-disk source path (used by inspect.getsourcefile).
    shim.__file__ = str(Path.cwd() / "task_responses" / "task_4.py")
    # Carry the documentation for introspection fidelity.
    shim.__doc__ = globals().get("__doc__")
    # Share every module-level name (functions, classes, constants) by reference.
    shim.__dict__.update(globals())
    # Insert the registration; from here, class getsourcefile succeeds.
    sys.modules[__name__] = shim

# Run the guard at import time; it is a no-op under registering loaders.
_register_executing_module_task_4()
```

```python
# Task 5: Trading Calendar Construction and Masking

# ==============================================================================
# Task 5: Trading Calendar Construction and Masking
# ==============================================================================

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5: Module-level provenance-bound constants
# -------------------------------------------------------------------------------------------------------------------------------

# The static figures pinned by the task list and by STUDY_CONFIG (P01 label
# parameters, P11 acquisition parameters, P33 calendar schema, P34 mask schema,
# P44 raw-layer invariants) are bound here exactly once, with their sources
# noted inline. Downstream code refers to these names; nothing re-derives them
# from scratch and nothing from P33's illustrative example rows is hard-coded.
_TOTAL_CALENDAR_ROWS: int = 727
_EXPECTED_YEAR_ROW_COUNTS: Dict[int, int] = {2023: 242, 2024: 242, 2025: 243}
_LABEL_WINDOW_YEARS: Tuple[int, int] = (2024, 2025)
_LABEL_WINDOW_ROWS: int = 485
_LABEL_AVAILABLE_ROWS: int = 480
_FORWARD_HORIZON_ROWS: int = 5
_TERMINAL_UNLABELLED_ROW_INDICES: Tuple[int, ...] = (723, 724, 725, 726, 727)
_POST_MASK_USABLE_ROWS: int = 29078
_POST_MASK_STOCK_COUNT: int = 40
_POST_MASK_FILE_COUNT: int = 80
_POST_MASK_FULLY_NULLED_ROWS: int = 2
_P44_POST_MASK_KEYS: Tuple[str, ...] = (
    "masked_cell_rows_equals_2",
    "forward_fill_rows_equals_0",
    "stock_prefix_reductions_equals_0",
    "duplicate_rows_per_stock_date_equals_0",
)
_P44_SECTION: str = "P44_RAW_LAYER_INVARIANTS"
_DATE_INDEX_CONSTRUCTION_DISCLOSURE: str = "DENSE_ROW_INDEX_OVER_727_TRADING_ROWS"
_COMPOUND_INDEX_NAMES: Tuple[str, str] = ("stock_code", "trade_date")
_FEATURES_REQUIRED_NUMERIC: Tuple[str, ...] = ("open", "high", "low", "close", "volume", "amount")
_LABELS_REQUIRED_NUMERIC: Tuple[str, ...] = ("close_adj", "adj_factor")
_FEATURES_FLAG_COLUMNS: Tuple[str, ...] = ("tradable", "tradestatus_raw", "quantity")
_LABELS_FLAG_COLUMNS: Tuple[str, ...] = ("tradable", "tradestatus_raw", "quantity")
_MASK_REASON: str = "non_trading_stock_day"
_MASK_TREATMENT: str = "MASKED"
_MASK_REGISTER_COLUMNS: Tuple[str, ...] = (
    "stock_code",
    "trade_date",
    "reason",
    "treatment",
    "forward_fill",
    "drop_stock",
)
_PANEL_FEATURES: str = "FEATURES_PANEL"
_PANEL_LABELS: str = "LABEL_PANEL"

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5: Shared private helpers (input boundary and coordinate machinery)
# -------------------------------------------------------------------------------------------------------------------------------

def _require_dataframe(obj: Any, caller: str, name: str) -> pd.DataFrame:
    """
    Validate that an object is a pandas DataFrame.

    Type boundary only: no copy is made and no column is inspected.
    The object is returned unchanged when the check passes.

    Parameters
    ----------
    obj : Any
        Candidate object supplied by the caller.
    caller : str
        Name of the calling callable, used in the diagnostic.
    name : str
        Argument name as written in the caller's signature.

    Returns
    -------
    pd.DataFrame
        The same object ``obj``, not copied.

    Raises
    ------
    TypeError
        If ``obj`` is not a :class:`pandas.DataFrame`.
    """
    # Reject anything that is not a DataFrame before any attribute access on it.
    if not isinstance(obj, pd.DataFrame):
        # Raise a typed diagnostic naming the callable and the argument.
        raise TypeError(f"{caller} requires {name} to be a pandas.DataFrame, got {type(obj).__name__}.")
    # Return the caller-owned frame; this boundary check copies nothing.
    return obj

def _require_strict_int(value: Any, name: str) -> int:
    """
    Validate that a value is a genuine Python int, excluding bool.

    ``bool`` subclasses ``int``; it is rejected so that ``True`` cannot
    stand in for ``1``.

    Parameters
    ----------
    value : Any
        Candidate integer.
    name : str
        Argument name used in the diagnostics.

    Returns
    -------
    int
        The validated integer, unchanged.

    Raises
    ------
    TypeError
        If ``value`` is not an ``int`` or is a ``bool``.
    ValueError
        If ``value`` is negative.
    """
    # bool is a subclass of int; exclude it so True cannot masquerade as 1.
    if not isinstance(value, int) or isinstance(value, bool):
        # Raise a typed diagnostic naming the argument.
        raise TypeError(f"{name} must be a Python int (bool excluded), got {value!r}.")
    # A negative row-count expectation is meaningless and must never be accepted.
    if value < 0:
        # Raise a range diagnostic naming the argument.
        raise ValueError(f"{name} must be non-negative, got {value}.")
    # Return the validated integer untouched.
    return value

def _require_string_or_path(obj: Any, name: str) -> Path:
    """
    Resolve a string or path-like object to a ``Path``.

    Nothing is created, resolved or touched on disk; the caller decides
    whether the directory must exist.

    Parameters
    ----------
    obj : Any
        Candidate directory designator.
    name : str
        Argument name used in the diagnostics.

    Returns
    -------
    Path
        ``Path(obj)`` for the validated designator.

    Raises
    ------
    TypeError
        If ``obj`` is neither a ``str`` nor an ``os.PathLike``.
    ValueError
        If ``str(obj)`` is empty or whitespace only.
    """
    # Only str and os.PathLike are accepted as directory designators.
    if not isinstance(obj, (str, os.PathLike)):
        # Raise a typed diagnostic naming the argument.
        raise TypeError(f"{name} must be a str or an os.PathLike path, got {type(obj).__name__}.")
    # Empty strings and whitespace-only designators are rejected outright.
    if not str(obj).strip():
        # Raise a value diagnostic naming the argument.
        raise ValueError(f"{name} must resolve to a non-empty path, got {obj!r}.")
    # Return the normalised Path; directory creation is the caller's decision.
    return Path(obj)

def _is_string_dtype(dtype: Any) -> bool:
    """
    Test whether a dtype is a pandas string dtype.

    The test is textual: the dtype's string form must start with
    ``"string"``, which accepts both the python and the pyarrow backend.

    Parameters
    ----------
    dtype : Any
        A pandas dtype object.

    Returns
    -------
    bool
        True when ``str(dtype)`` starts with ``"string"``.
    """
    # Compare on the canonical string form so both backends are accepted.
    return str(dtype).startswith("string")

def _is_boolean_dtype(dtype: Any) -> bool:
    """
    Test whether a dtype is a nullable boolean dtype.

    The test is textual: the dtype's string form must start with
    ``"boolean"``, which accepts both the python and the pyarrow
    backend.

    Parameters
    ----------
    dtype : Any
        A pandas dtype object.

    Returns
    -------
    bool
        True when ``str(dtype)`` starts with ``"boolean"``.
    """
    # Compare on the canonical string form so the pyarrow backend is accepted too.
    return str(dtype).startswith("boolean")

def _compound_index_names(panel: pd.DataFrame) -> Optional[List[str]]:
    """
    Return the index level names of a compound-keyed panel.

    Parameters
    ----------
    panel : pd.DataFrame
        Panel whose index is inspected.

    Returns
    -------
    Optional[List[str]]
        ``list(panel.index.names)`` preserved in level order when the
        index is a :class:`pandas.MultiIndex`, otherwise ``None``.
    """
    # Only a true MultiIndex can carry the (stock_code, trade_date) compound key.
    if not isinstance(panel.index, pd.MultiIndex):
        # Signal the structural breach to the caller via ``None``.
        return None
    # Return the level names as a list, preserving internal order.
    return list(panel.index.names)

def _fully_nulled_mask(panel: pd.DataFrame, columns: List[str]) -> pd.Series:
    """
    Mark rows that are fully NA on every pinned column.

    Row-wise reduction of ``panel[columns].isna().all(axis=1)``, cast to
    plain ``bool`` so downstream counts are ordinary integers.

    Parameters
    ----------
    panel : pd.DataFrame
        Panel to scan; its index is reused for the mask.
    columns : List[str]
        Columns that must all be NA for a row to be flagged. An empty
        list yields an all-``False`` mask, never an all-``True`` one.

    Returns
    -------
    pd.Series
        ``bool`` Series aligned to ``panel.index``, True where every
        column in ``columns`` is NA on that row.
    """
    # With no pinned column present the mask is all-False (never all-True).
    if not columns:
        # Return a conservation-preserving all-False mask over the panel axis.
        return pd.Series(False, index=panel.index, dtype="bool")
    # Reduce column-wise NAhood: every pinned cell in the row is NA.
    mask = panel[columns].isna().all(axis=1)
    # Cast to plain bool so downstream ``sum`` counts are ordinary integers.
    return mask.astype("bool")

def _pinned_columns_present(panel: pd.DataFrame, kind: str) -> Tuple[List[str], List[str]]:
    """
    Resolve the pinned null columns present in a panel.

    Dispatch on the panel role. For ``"FEATURES_PANEL"`` the required
    set is the six nominal price/volume columns plus ``tradable``; for
    ``"LABEL_PANEL"`` it is the adjusted-label numeric columns plus
    ``tradable``. Optional flag columns (``tradestatus_raw``,
    ``quantity``) are appended only when they exist in the panel.

    Parameters
    ----------
    panel : pd.DataFrame
        FEATURES_PANEL or LABEL_PANEL to inspect.
    kind : str
        Either ``"FEATURES_PANEL"`` or ``"LABEL_PANEL"``.

    Returns
    -------
    Tuple[List[str], List[str]]
        ``(present, missing_required)``: the columns to null, in pinned
        order (required columns first, then present optional flags), and
        the required columns absent from the panel.

    Raises
    ------
    ValueError
        If ``kind`` is neither recognised panel name.
    """
    # Dispatch on the role to the pinned lists of STUDY_CONFIG P12/P34/Task 4 split.
    if kind == _PANEL_FEATURES:
        # FEATURES_PANEL: six nominal/unadjusted numerics and tradable are mandatory.
        required = list(_FEATURES_REQUIRED_NUMERIC) + ["tradable"]
        # Optional-but-present flag columns: null them exactly when present.
        optional = [c for c in _FEATURES_FLAG_COLUMNS if c != "tradable"]
    elif kind == _PANEL_LABELS:
        # LABEL_PANEL: the adjusted-label numerics and tradable are mandatory.
        required = list(_LABELS_REQUIRED_NUMERIC) + ["tradable"]
        # Optional-but-present flag columns: null them exactly when present.
        optional = [c for c in _LABELS_FLAG_COLUMNS if c != "tradable"]
    else:
        # Unknown role: refuse to guess rather than null the wrong columns.
        raise ValueError(f"_pinned_columns_present: unknown panel kind {kind!r}.")
    # Compute the missing required columns (empty list means fully satisfied).
    missing_required = [c for c in required if c not in panel.columns]
    # Assemble the present list in pinned order: required numerics, flags, numerics-last.
    present = [c for c in required if c in panel.columns]
    # Append the present optional flag columns after the required ones.
    present.extend([c for c in optional if c in panel.columns])
    # Return both diagnostics to the caller.
    return present, missing_required

def _masked_coordinate_tuples(mask_register: pd.DataFrame) -> List[Tuple[str, pd.Timestamp]]:
    """
    Return the masked coordinates of a validated register.

    Parameters
    ----------
    mask_register : pd.DataFrame
        Register carrying ``stock_code`` and ``trade_date`` columns, as
        returned by ``verify_mask_register_task_5``.

    Returns
    -------
    List[Tuple[str, pd.Timestamp]]
        One ``(str(stock_code), trade_date)`` pair per register row,
        in register order.
    """
    # Read the stock_code level as plain strings (dtype-agnostic coordinate keys).
    stock_codes = mask_register["stock_code"].astype(str).tolist()
    # Read the trade_date level as Timestamps (dtype datetime64[ns] guaranteed).
    trade_dates = list(mask_register["trade_date"].tolist())
    # Zip into tuples aligned position-wise between the two lists.
    return list(zip(stock_codes, trade_dates))

def _exclusion_mask(panel: pd.DataFrame, coordinates: List[Tuple[str, pd.Timestamp]]) -> np.ndarray:
    """
    Mark panel rows whose coordinate is not masked.

    Build the set of masked ``(stock_code, trade_date)`` tuples and test
    every row of the panel's two-level index against it.

    Parameters
    ----------
    panel : pd.DataFrame
        Panel with a ``(stock_code, trade_date)`` MultiIndex.
    coordinates : List[Tuple[str, pd.Timestamp]]
        Coordinates to exclude.

    Returns
    -------
    np.ndarray
        Plain ``bool`` array of length ``len(panel)``; True keeps the
        row, i.e. its coordinate is not in ``coordinates``.
    """
    # Build the lookup index of masked tuples once.
    masked = set(coordinates)
    # Recover the panel's coordinate keys directly from the index levels.
    keys = zip(panel.index.get_level_values(0).astype(str).tolist(), list(panel.index.get_level_values(1)))
    # Vectorise the membership test as a list comprehension over the axis.
    flags = np.array([key not in masked for key in keys], dtype="bool")
    # Return the plain boolean ndarray for direct use in .loc.
    return flags

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 1: build the dense trading calendar
# -------------------------------------------------------------------------------------------------------------------------------

def construct_calendar_task_5(unadjusted_bars: pd.DataFrame) -> pd.DataFrame:
    """
    Build the dense trading calendar from unadjusted bars.

    Take the unique ``trade_date`` values of the unadjusted view, sorted
    ascending, and lay out the coordinate system on which the
    five-trading-row forward label (Eq. (1)) is defined: ``row_index``
    advances over trading rows, never over calendar days. Steps:
    validate the frame and its ``trade_date`` column; compress to the
    unique, null-free timestamp set and sort it; assign the dense index
    1..N; take the year of each stamp; derive ``in_label_window`` (year
    in {2024, 2025}) and ``label_available`` as ``in_label_window AND
    row_index + 5 <= N``; mirror the axis to a fresh ``RangeIndex``.

    This callable performs no cardinality pinning; the 727 rows and the
    per-year breakdown are enforced by
    ``assert_calendar_counts_task_5``.

    Parameters
    ----------
    unadjusted_bars : pd.DataFrame
        The unadjusted bars frame. Must carry a datetime-like
        ``trade_date`` column (``datetime64[ns]`` expected) without null
        cells. Row order is irrelevant.

    Returns
    -------
    pd.DataFrame
        One row per distinct ``trade_date`` (727 on the certified
        fixture), indexed by ``RangeIndex(0, N, 1)``. Columns, in order:

        - ``row_index``: ``Int64``, non-index, dense 1..N.
        - ``trade_date``: ``datetime64[ns]``, unique, ascending.
        - ``year``: ``Int64``, ``trade_date.dt.year``.
        - ``in_label_window``: ``boolean``, True iff year in
          {2024, 2025}.
        - ``label_available``: ``boolean``, True iff
          ``in_label_window & (row_index + 5 <= N)``.

    Raises
    ------
    TypeError
        If ``unadjusted_bars`` is not a pandas DataFrame, or if
        ``trade_date`` is not a datetime-like column.
    ValueError
        If ``trade_date`` is absent, carries null cells, or yields an
        empty unique set.

    Notes
    -----
    The horizon is five trading rows and the label window spans the
    calendar years 2024 and 2025.
    """
    # Validate the boundary: the input must be a DataFrame (upstream typing is not re-validated beyond shape of use).
    bars = _require_dataframe(unadjusted_bars, "construct_calendar_task_5", "unadjusted_bars")
    # The trade_date column is the only input the calendar is allowed to consume.
    if "trade_date" not in bars.columns:
        # Raise a structural diagnostic naming the missing column.
        raise ValueError("construct_calendar_task_5: unadjusted_bars is missing the 'trade_date' column.")
    # Extract the trade_date Series once; dtype is checked next.
    trade_dates = bars["trade_date"]
    # Enforce a datetime-like dtype for the calendar source column.
    if not pd.api.types.is_datetime64_any_dtype(trade_dates):
        # Raise a typed diagnostic naming the offending dtype.
        raise TypeError(f"construct_calendar_task_5: 'trade_date' must be datetime-like, got {trade_dates.dtype}.")
    # Null timestamps would fabricate calendar rows under unique(); refuse them.
    if bool(trade_dates.isna().any()):
        # Raise a value diagnostic naming the column.
        raise ValueError("construct_calendar_task_5: 'trade_date' carries null cells; the calendar cannot be built from a holed axis.")
    # Compress the 29,080 duplicated stamps to the unique set and sort ascending (ns resolution).
    unique_dates = pd.DatetimeIndex(pd.DatetimeIndex(trade_dates).unique().sort_values())
    # An empty unique-set cannot define any row; refuse it explicitly.
    if len(unique_dates) == 0:
        # Raise a value diagnostic naming the emptiness.
        raise ValueError("construct_calendar_task_5: 'trade_date' yielded an empty unique set; the calendar cannot be built.")
    # Materialise the unique axis as ns-resolution datetime64[ns] explicitly.
    trade_date_axis = unique_dates.astype("datetime64[ns]")
    # The unique count N is the length the coordinate system must span.
    total_rows = len(trade_date_axis)
    # Build the dense 1..N row_index as an int64 Series cast to nullable Int64.
    row_index = pd.Series(np.arange(1, total_rows + 1, dtype=np.int64), dtype="Int64")
    # Extract the calendar year as Int64 (pandas dt.year yields int32/int64; cast pins the contract).
    year = pd.Series(trade_date_axis.year, dtype="Int64")
    # Derive in_label_window: year in {2024, 2025} -> nullable boolean True.
    in_label_window = year.isin(list(_LABEL_WINDOW_YEARS)).astype("boolean")
    # Eq. (1) horizon: a label exists at r iff r + 5 stays inside the captured calendar.
    in_horizon = (row_index + _FORWARD_HORIZON_ROWS) <= total_rows
    # label_available := in_label_window AND (row_index + 5 <= N); the conjunct is mandatory.
    label_available = (in_label_window & in_horizon).astype("boolean")
    # Assemble the frame column-wise so every column shares one axis order.
    calendar = pd.DataFrame(
        {
            "row_index": row_index,
            "trade_date": pd.Series(trade_date_axis, dtype="datetime64[ns]"),
            "year": year,
            "in_label_window": in_label_window,
            "label_available": label_available,
        }
    )
    # Mirror the axis to a RangeIndex(0, N, 1) as the step text mandates.
    calendar.index = pd.RangeIndex(start=0, stop=total_rows, step=1)
    # Return the calendar; counts are pinned by assert_calendar_counts_task_5, not here.
    return calendar

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 1: pin the static calendar counts (halt on the first breach)
# -------------------------------------------------------------------------------------------------------------------------------

def assert_calendar_counts_task_5(calendar: pd.DataFrame) -> Dict[str, Any]:
    """
    Verify the pinned trading-calendar counts and derivation.

    Checks run in a fixed order and the first breach raises. Pins: the
    five required columns and their dtypes (``Int64``,
    ``datetime64[ns]``, ``Int64``, ``boolean``, ``boolean``); a
    ``RangeIndex`` axis; row count 727; dense, null-free ``row_index``
    equal to 1..727; unique, ascending ``trade_date``; ``year`` equal to
    ``trade_date.dt.year``; per-year counts {2023: 242, 2024: 242,
    2025: 243} with 2023 asserted independently of the 2024 + 2025 sum;
    ``in_label_window`` re-derived from ``year`` with cardinality 485;
    ``label_available`` re-derived as ``in_label_window AND
    row_index + 5 <= 727`` with cardinality 480 (the conjunct-free
    ``row_index + 5 <= 727`` formula, which yields 722, is rejected);
    rows 723..727 in the window with no label; row 722 label-available
    (722 + 5 = 727).

    Parameters
    ----------
    calendar : pd.DataFrame
        Frame returned by ``construct_calendar_task_5`` or an equivalent
        727 x 5 frame.

    Returns
    -------
    Dict[str, Any]
        Audit record with keys ``halt_reason`` (None), ``passed``
        (True), ``row_count``, ``year_counts`` (year -> row count),
        ``in_label_window_count``, ``label_available_count``,
        ``terminal_unlabelled_rows`` (the in-window unlabelled
        ``row_index`` values, computed from the frame), ``range_index``
        (bool), ``checks`` (check name -> bool, all True on success),
        ``date_index_construction`` (the dense-row-index disclosure
        string) and ``label_available_definition`` (formula text).

    Raises
    ------
    TypeError
        If ``calendar`` is not a pandas DataFrame.
    AssertionError
        On the first breached pin, with a message naming the breach.

    Notes
    -----
    ``label_available`` is recomputed from ``row_index`` and ``N``
    rather than trusted, so a calendar built with the wrong horizon or
    the wrong window cannot pass.
    """
    # Validate the boundary type before any structural assertion runs on it.
    frame = _require_dataframe(calendar, "assert_calendar_counts_task_5", "calendar")
    # The five pinned columns must all be present.
    required_columns = ("row_index", "trade_date", "year", "in_label_window", "label_available")
    # Compute the missing set explicitly for the diagnostic message.
    missing = [name for name in required_columns if name not in frame.columns]
    # Halt immediately when any pinned column is absent.
    assert not missing, f"calendar is missing pinned column(s) {missing}; found {sorted(frame.columns)}."
    # The caller must hand us the constructor's frame shape: a RangeIndex axis.
    assert isinstance(frame.index, pd.RangeIndex), f"calendar.index must be a RangeIndex, got {type(frame.index).__name__}."
    # Pin the dtypes exactly as P33_SCHEMA_TRADING_CALENDAR declares them.
    assert str(frame["row_index"].dtype) == "Int64", f"row_index must be Int64, got {frame['row_index'].dtype}."
    # Pin the trade_date dtype to ns-precision datetimes.
    assert str(frame["trade_date"].dtype) == "datetime64[ns]", f"trade_date must be datetime64[ns], got {frame['trade_date'].dtype}."
    # Pin the year dtype to nullable Int64.
    assert str(frame["year"].dtype) == "Int64", f"year must be Int64, got {frame['year'].dtype}."
    # Pin the window flag to nullable boolean.
    assert str(frame["in_label_window"].dtype) == "boolean", f"in_label_window must be boolean, got {frame['in_label_window'].dtype}."
    # Pin the availability flag to nullable boolean.
    assert str(frame["label_available"].dtype) == "boolean", f"label_available must be boolean, got {frame['label_available'].dtype}."
    # The cardinality pin: exactly 727 trading rows (P11 trading_date_count, P33 estimated_row_count).
    assert len(frame) == _TOTAL_CALENDAR_ROWS, f"calendar row count must be 727, got {len(frame)}."
    # The row_index must be dense 1..727 with no gaps and no duplicates: r -> r + 5 must be a trading-row offset.
    row_index_values = frame["row_index"]
    # Reject any null row_index before comparing (Int64 NA cannot enter the dense check).
    assert bool(row_index_values.notna().all()), "row_index carries null cells; the coordinate system is holed."
    # Compare against the dense reference vector elementwise.
    assert row_index_values.tolist() == list(range(1, _TOTAL_CALENDAR_ROWS + 1)), "row_index is not the dense 1..727 sequence."
    # Trade dates must be unique, ascending, and free of nulls by construction.
    assert bool(frame["trade_date"].is_unique), "trade_date carries duplicates; the calendar axis must be unique."
    # Ascending order is the vectorisation contract for downstream row lookups.
    assert bool(frame["trade_date"].is_monotonic_increasing), "trade_date must be sorted ascending."
    # The year column must equal the year extracted from trade_date, elementwise.
    assert (frame["year"].to_numpy() == frame["trade_date"].dt.year.to_numpy()).all(), "year disagrees with trade_date.dt.year."
    # Year counts: 2023 pinned INDEPENDENTLY of the 2024 + 2025 sum (single frozen row in one year plus duplicate in another must not pass).
    year_counts = {int(year): int(count) for year, count in frame["year"].value_counts().sort_index().items()}
    # The exact year set must be {2023, 2024, 2025} with no surplus years.
    assert set(year_counts) == set(_EXPECTED_YEAR_ROW_COUNTS), f"calendar years must be exactly {sorted(_EXPECTED_YEAR_ROW_COUNTS)}, got {sorted(year_counts)}."
    # The independent 2023 pin.
    assert year_counts[2023] == _EXPECTED_YEAR_ROW_COUNTS[2023], f"2023 rows must be 242, got {year_counts[2023]}."
    # The 2024 pin.
    assert year_counts[2024] == _EXPECTED_YEAR_ROW_COUNTS[2024], f"2024 rows must be 242, got {year_counts[2024]}."
    # The 2025 pin.
    assert year_counts[2025] == _EXPECTED_YEAR_ROW_COUNTS[2025], f"2025 rows must be 243, got {year_counts[2025]}."
    # The conjunction pin: 242 + 242 + 243 == 727 (P42 calendar closure).
    assert sum(year_counts.values()) == _TOTAL_CALENDAR_ROWS == sum(_EXPECTED_YEAR_ROW_COUNTS.values()), "year breakdown does not sum to 727."
    # in_label_window is DERIVED, not trusted: re-derive from year and compare elementwise.
    expected_window = frame["year"].isin(list(_LABEL_WINDOW_YEARS)).astype("boolean")
    # Elementwise identity catches any window drift (e.g., a 2025-excluded window).
    assert frame["in_label_window"].tolist() == expected_window.tolist(), "in_label_window is not the pinned year in {2024, 2025} derivation."
    # The window cardinality pin: 242 + 243 == 485 (P01 label_window_trading_rows).
    window_count = int(frame["in_label_window"].sum())
    # Halt when the window cardinality departs from 485.
    assert window_count == _LABEL_WINDOW_ROWS, f"in_label_window cardinality must be 485, got {window_count}."
    # label_available is DERIVED, not trusted: in_label_window AND (row_index + 5 <= 727).
    in_horizon = (frame["row_index"] + _FORWARD_HORIZON_ROWS) <= _TOTAL_CALENDAR_ROWS
    # Re-derive the availability flag elementwise with the mandatory conjunct.
    expected_available = (frame["in_label_window"] & in_horizon).astype("boolean")
    # Elementwise identity catches the forbidden raw r + 5 <= 727 (722) formula.
    assert frame["label_available"].tolist() == expected_available.tolist(), "label_available is not in_label_window AND row_index + 5 <= 727."
    # The availability cardinality pin: 485 - 5 == 480 (P01 label_available_rows).
    available_count = int(frame["label_available"].sum())
    # Halt when the availability cardinality departs from 480.
    assert available_count == _LABEL_AVAILABLE_ROWS, f"label_available cardinality must be 480, got {available_count}."
    # Terminal boundary: rows 723..727 are in the window yet carry no label (723 + 5 = 728 > 727).
    terminal = frame.loc[frame["row_index"].isin(list(_TERMINAL_UNLABELLED_ROW_INDICES))]
    # Every terminal row must be in the window.
    assert bool(terminal["in_label_window"].all()), "terminal rows 723..727 must lie inside the label window."
    # Every terminal row must be label-unavailable.
    assert not bool(terminal["label_available"].any()), "terminal rows 723..727 must all carry label_available == False."
    # Row 722 sits exactly at the boundary: 722 + 5 == 727 <= 727, so it IS label-available.
    boundary_row = frame.loc[frame["row_index"] == 722]
    # The boundary row must have survived the fixture intact.
    assert len(boundary_row) == 1, "row 722 must be present exactly once."
    # Assert the boundary cell explicitly.
    assert bool(boundary_row["label_available"].iloc[0]), "row 722 must be label_available (722 + 5 == 727 <= 727)."
    # The unlabelled in-window set must be exactly {723, 724, 725, 726, 727}.
    unlabelled_in_window = frame.loc[frame["in_label_window"] & ~frame["label_available"], "row_index"].tolist()
    # Set equality pins both membership and cardinality of the terminal set.
    assert set(unlabelled_in_window) == set(_TERMINAL_UNLABELLED_ROW_INDICES), f"terminal unlabelled rows must be {list(_TERMINAL_UNLABELLED_ROW_INDICES)}, got {unlabelled_in_window}."
    # Assemble the checks record for the artifact, EVERY value DERIVED from the argument frame.
    checks = {
        "row_count_equals_727": bool(len(frame) == _TOTAL_CALENDAR_ROWS),
        "year_counts_2023_242_2024_242_2025_243": bool(year_counts == _EXPECTED_YEAR_ROW_COUNTS),
        "year_2023_asserted_independently": bool(year_counts[2023] == _EXPECTED_YEAR_ROW_COUNTS[2023]),
        "in_label_window_equals_year_in_2024_2025": bool(frame["in_label_window"].tolist() == expected_window.tolist()),
        "in_label_window_cardinality_485": bool(window_count == _LABEL_WINDOW_ROWS),
        "label_available_equals_in_label_window_and_row_index_plus_5_le_727": bool(frame["label_available"].tolist() == expected_available.tolist()),
        "label_available_cardinality_480": bool(available_count == _LABEL_AVAILABLE_ROWS),
        "row_index_dense_unique_Int64_1_to_727": bool(row_index_values.tolist() == list(range(1, _TOTAL_CALENDAR_ROWS + 1))),
        "trade_date_unique_monotonic_ns": bool(bool(frame["trade_date"].is_unique) and bool(frame["trade_date"].is_monotonic_increasing) and str(frame["trade_date"].dtype) == "datetime64[ns]"),
        "terminal_unlabelled_rows_723_to_727": bool(set(unlabelled_in_window) == set(_TERMINAL_UNLABELLED_ROW_INDICES)),
        "row_722_label_available_boundary": bool(boundary_row["label_available"].iloc[0]),
        "range_index_mirror": bool(isinstance(frame.index, pd.RangeIndex)),
    }
    # The reportable terminal set is COMPUTED from the calendar, never returned as a literal.
    terminal_unlabelled_rows = [int(row) for row in unlabelled_in_window]
    # Return the fully-audited success record; terminal set COMPUTED above from the argument frame.
    return {
        "halt_reason": None,
        "passed": True,
        "row_count": int(len(frame)),
        "year_counts": year_counts,
        "in_label_window_count": window_count,
        "label_available_count": available_count,
        "terminal_unlabelled_rows": terminal_unlabelled_rows,
        "range_index": bool(isinstance(frame.index, pd.RangeIndex)),
        "checks": checks,
        "date_index_construction": _DATE_INDEX_CONSTRUCTION_DISCLOSURE,
        "label_available_definition": "in_label_window AND row_index + 5 <= 727 (conjunct mandatory)",
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 2: verify the MASK_REGISTER against P34_SCHEMA_MASK_REGISTER
# -------------------------------------------------------------------------------------------------------------------------------

def verify_mask_register_task_5(mask_register: pd.DataFrame) -> pd.DataFrame:
    """
    Verify a MASK_REGISTER frame against its pinned schema.

    Pins, in order: the column set is exactly ``stock_code``,
    ``trade_date``, ``reason``, ``treatment``, ``forward_fill``,
    ``drop_stock`` (no surplus, no deficit, any column order); exactly
    two rows; ``stock_code`` carries a pandas string dtype and
    ``trade_date`` is ``datetime64[ns]``, then ``reason`` and
    ``treatment`` string dtypes, then ``forward_fill`` and
    ``drop_stock`` nullable boolean dtypes; no cell is null;
    ``reason`` equals ``"non_trading_stock_day"`` and
    ``treatment`` equals ``"MASKED"`` on both rows; ``forward_fill``
    and ``drop_stock`` are False on both rows; the two
    ``(stock_code, trade_date)`` coordinates are distinct.

    Parameters
    ----------
    mask_register : pd.DataFrame
        Caller-supplied register.

    Returns
    -------
    pd.DataFrame
        Deep copy of the validated register, keeping the column order it
        was supplied with. The caller's frame is never mutated.

    Raises
    ------
    TypeError
        If ``mask_register`` is not a pandas DataFrame.
    AssertionError
        On the first breached pin, with a message naming the breach.
    """
    # Validate the boundary type before any column or cell access.
    register = _require_dataframe(mask_register, "verify_mask_register_task_5", "mask_register")
    # The P34 column set must match EXACTLY - extras would smuggle semantics in, deficits would omit them.
    column_set = set(register.columns)
    # Compute the surplus and deficit explicitly for the diagnostic.
    surplus = sorted(column_set - set(_MASK_REGISTER_COLUMNS))
    # Missing columns are equally fatal.
    deficit = sorted(set(_MASK_REGISTER_COLUMNS) - column_set)
    # Halt on any deviation from the pinned column set.
    assert not surplus and not deficit, f"MASK_REGISTER columns must be exactly {list(_MASK_REGISTER_COLUMNS)}; surplus={surplus}, deficit={deficit}."
    # The mask cardinality pin: exactly two rows (P12 masked_cell_count == 2).
    assert len(register) == 2, f"MASK_REGISTER must contain exactly 2 rows, got {len(register)}."
    # Dtype pins: identity strings are pandas string dtype (object or pyarrow backend both accepted).
    assert _is_string_dtype(register["stock_code"].dtype), f"stock_code must be a string dtype, got {register['stock_code'].dtype}."
    # The trade_date axis must be ns-resolution datetimes.
    assert str(register["trade_date"].dtype) == "datetime64[ns]", f"trade_date must be datetime64[ns], got {register['trade_date'].dtype}."
    # Reason is a pinned string vocabulary of exactly one token.
    assert _is_string_dtype(register["reason"].dtype), f"reason must be a string dtype, got {register['reason'].dtype}."
    # Treatment is a pinned string vocabulary of exactly one token.
    assert _is_string_dtype(register["treatment"].dtype), f"treatment must be a string dtype, got {register['treatment'].dtype}."
    # forward_fill is a nullable boolean flag that must be False on both rows.
    assert _is_boolean_dtype(register["forward_fill"].dtype), f"forward_fill must be a nullable boolean dtype, got {register['forward_fill'].dtype}."
    # drop_stock is a nullable boolean flag that must be False on both rows.
    assert _is_boolean_dtype(register["drop_stock"].dtype), f"drop_stock must be a nullable boolean dtype, got {register['drop_stock'].dtype}."
    # No cell may be null anywhere in the register; a null flag is an unattested operation.
    for column in _MASK_REGISTER_COLUMNS:
        # Halt on the first holed column, named in the diagnostic.
        assert bool(register[column].notna().all()), f"MASK_REGISTER column {column!r} carries null cells; the register must be complete."
    # reason must equal exactly "non_trading_stock_day" on both rows.
    assert bool((register["reason"] == _MASK_REASON).all()), f"reason must be {_MASK_REASON!r} on both rows, got {register['reason'].tolist()}."
    # treatment must equal exactly "MASKED" on both rows.
    assert bool((register["treatment"] == _MASK_TREATMENT).all()), f"treatment must be {_MASK_TREATMENT!r} on both rows, got {register['treatment'].tolist()}."
    # forward_fill must be exactly False on both rows (nullable False, never NA, never True).
    assert not bool(register["forward_fill"].any()), "forward_fill must be False on both rows; any True/NA halts the pipeline."
    # drop_stock must be exactly False on both rows.
    assert not bool(register["drop_stock"].any()), "drop_stock must be False on both rows; any True/NA halts the pipeline."
    # The two (stock_code, trade_date) coordinates must be DISTINCT (two stock-day cells).
    coordinates = _masked_coordinate_tuples(register)
    # Distinctness is pinned; a shared coordinate would be one cell, not two.
    assert len(set(coordinates)) == 2, f"MASK_REGISTER coordinates must be distinct, got {coordinates}."
    # Return a deep copy so downstream mutation cannot leak into the caller's frame.
    return register.copy(deep=True)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 2: coordinate-addressed null-fill of the two masked cells
# -------------------------------------------------------------------------------------------------------------------------------

def apply_mask_task_5(
    features_panel: pd.DataFrame,
    label_panel: pd.DataFrame,
    mask_register: pd.DataFrame,
) -> Dict[str, Any]:
    """
    Null the masked stock-day cells in both panels on copies.

    The two panels' DataFrame types are validated first, then the
    register is verified - both before any panel index or column work.
    Both panels must carry a ``(stock_code, trade_date)`` MultiIndex;
    the columns to null are resolved per panel role, with any missing
    required numeric or ``tradable`` column a halt. Each masked
    coordinate is looked up by index membership (never by position) in
    both panels, and an absent coordinate halts before anything is
    written. Deep copies are made, then ``pd.NA`` is written to every
    pinned column of the matching rows in each copy: FEATURES_PANEL
    ``open``, ``high``, ``low``, ``close``, ``volume``, ``amount``,
    ``tradable``, plus ``tradestatus_raw`` and ``quantity`` only if
    present; LABEL_PANEL ``close_adj``, ``adj_factor``, ``tradable``,
    plus ``tradestatus_raw`` and ``quantity`` only if present. No
    forward-fill, interpolation, stock prefix reduction or row drop is
    performed, so row counts are unchanged.

    Parameters
    ----------
    features_panel : pd.DataFrame
        FEATURES_PANEL with a compound ``(stock_code, trade_date)``
        MultiIndex. Never mutated.
    label_panel : pd.DataFrame
        LABEL_PANEL with the same compound index. Never mutated.
    mask_register : pd.DataFrame
        Raw register, verified here by ``verify_mask_register_task_5``.

    Returns
    -------
    Dict[str, Any]
        Keys:

        - ``features_panel``: NA-kept deep copy of the features panel.
        - ``label_panel``: NA-kept deep copy of the label panel.
        - ``witness``: DataFrame with one row per coordinate x panel x
          nulled column and columns ``stock_code`` (string),
          ``trade_date`` (``datetime64[ns]``), ``panel`` (string,
          ``"FEATURES_PANEL"`` or ``"LABEL_PANEL"``), ``column``
          (string) and ``treatment`` (string, ``"MASKED"``).

    Raises
    ------
    TypeError
        If either panel is not a pandas DataFrame.
    AssertionError
        If the register fails verification; if a panel lacks the
        ``(stock_code, trade_date)`` MultiIndex; if a required numeric
        or ``tradable`` column is missing; or if a masked coordinate is
        absent from either panel.
    """
    # Validate boundary types for all three arguments before any work begins.
    features = _require_dataframe(features_panel, "apply_mask_task_5", "features_panel")
    # Labels panel boundary check.
    labels = _require_dataframe(label_panel, "apply_mask_task_5", "label_panel")
    # Verify the register FIRST; a defective register never reaches the panels.
    register = verify_mask_register_task_5(mask_register)
    # Validate the compound index structure of both panels (names and type).
    for panel, name in ((features, _PANEL_FEATURES), (labels, _PANEL_LABELS)):
        # Recover the index names (None when not a MultiIndex).
        names = _compound_index_names(panel)
        # Halt on any departure from the (stock_code, trade_date) compound key.
        assert names == list(_COMPOUND_INDEX_NAMES), f"{name} must carry a (stock_code, trade_date) MultiIndex, got levels {names}."
    # Resolve the pinned null-column lists; missing REQUIRED columns are halts.
    features_nulled, features_missing = _pinned_columns_present(features, _PANEL_FEATURES)
    # Labels panel's pinned null-column list and missing required columns.
    labels_nulled, labels_missing = _pinned_columns_present(labels, _PANEL_LABELS)
    # Halt when a required numeric/tradable column is absent from FEATURES_PANEL.
    assert not features_missing, f"FEATURES_PANEL is missing required column(s) {features_missing}."
    # Halt when a required numeric/tradable column is absent from LABEL_PANEL.
    assert not labels_missing, f"LABEL_PANEL is missing required column(s) {labels_missing}."
    # The two coordinates that both panels must acknowledge, from the validated register.
    coordinates = _masked_coordinate_tuples(register)
    # Verify BOTH coordinates exist in EACH panel BEFORE writing anything.
    for panel, name in ((features, _PANEL_FEATURES), (labels, _PANEL_LABELS)):
        # Per-coordinate diagnostics: name the absent coordinate in the halt.
        for coordinate in coordinates:
            # Test each coordinate individually so the diagnostic is exact.
            single = panel.index.isin(pd.MultiIndex.from_tuples([coordinate], names=list(_COMPOUND_INDEX_NAMES)))
            # Halt when this coordinate is nowhere in the panel.
            assert bool(single.any()), f"coordinate {coordinate} is absent from {name}; the mask register cannot reference what was not retrieved."
    # Deep-copy BOTH panels so the caller's frames are never mutated in place.
    features_masked = features.copy(deep=True)
    # Same discipline for the labels panel.
    labels_masked = labels.copy(deep=True)
    # Perform the coordinate-addressed null-fill on the COPIES.
    for coordinate in coordinates:
        # Build the per-coordinate probe once.
        probe = pd.MultiIndex.from_tuples([coordinate], names=list(_COMPOUND_INDEX_NAMES))
        # Boolean membership mask for FEATURES_PANEL.
        features_selector = features_masked.index.isin(probe)
        # Boolean membership mask for LABEL_PANEL.
        labels_selector = labels_masked.index.isin(probe)
        # Write pd.NA to every pinned FEATURES_PANEL column at the masked row.
        features_masked.loc[features_selector, features_nulled] = pd.NA
        # Write pd.NA to every pinned LABEL_PANEL column at the masked row.
        labels_masked.loc[labels_selector, labels_nulled] = pd.NA
    # Assemble the witness: one row per coordinate x panel x nulled column.
    witness_rows: List[Dict[str, Any]] = []
    # Enumerate coordinates in register order (deterministic).
    for coordinate in coordinates:
        # Expand FEATURES_PANEL's nulled columns.
        for column in features_nulled:
            # One witness row per nulled feature cell.
            witness_rows.append(
                {
                    "stock_code": str(coordinate[0]),
                    "trade_date": pd.Timestamp(coordinate[1]),
                    "panel": _PANEL_FEATURES,
                    "column": str(column),
                    "treatment": _MASK_TREATMENT,
                }
            )
        # Expand LABEL_PANEL's nulled columns.
        for column in labels_nulled:
            # One witness row per nulled label cell.
            witness_rows.append(
                {
                    "stock_code": str(coordinate[0]),
                    "trade_date": pd.Timestamp(coordinate[1]),
                    "panel": _PANEL_LABELS,
                    "column": str(column),
                    "treatment": _MASK_TREATMENT,
                }
            )
    # Materialise the witness with pinned dtypes.
    witness = pd.DataFrame(witness_rows, columns=["stock_code", "trade_date", "panel", "column", "treatment"])
    # stock_code is a pandas string identity.
    witness["stock_code"] = witness["stock_code"].astype("string")
    # trade_date is ns-resolution datetime.
    witness["trade_date"] = witness["trade_date"].astype("datetime64[ns]")
    # panel is a pandas string discriminator.
    witness["panel"] = witness["panel"].astype("string")
    # column is a pandas string identity.
    witness["column"] = witness["column"].astype("string")
    # treatment is a pandas string identity pinned to "MASKED".
    witness["treatment"] = witness["treatment"].astype("string")
    # Return the NA-kept copies and the witness; caller frames were never touched.
    return {
        "features_panel": features_masked,
        "label_panel": labels_masked,
        "witness": witness,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Step 3: post-masking invariant battery (executed in full)
# -------------------------------------------------------------------------------------------------------------------------------

def assert_post_mask_battery_task_5(
    features_panel: pd.DataFrame,
    label_panel: pd.DataFrame,
    context: Any,
    study_config: Dict[str, Any],
    expected_panel_rows: int = _POST_MASK_USABLE_ROWS,
) -> Dict[str, Any]:
    """
    Execute the post-masking invariant battery in full.

    The battery is the last barrier before label construction (Task 6).
    It is a Boolean conjunction evaluated completely: no check is
    short-circuited and the first failure does not hide later ones. The
    usable row count is derived from the data as the rows that are not
    fully null at the pinned per-view masked columns and is compared
    with ``expected_panel_rows`` (29,076, 29,079 and 29,080 all halt).
    Exactly two fully-nulled rows are required per panel (P44
    ``masked_cell_rows_equals_2``); both panels must be fully nulled at
    identical coordinate tuples; the compound ``(stock_code,
    trade_date)`` index must be unique in both panels (P44
    ``duplicate_rows_per_stock_date_equals_0``); and the anti-imputation
    counters of the pipeline context (``forward_fill_operations``,
    ``interpolation_operations``, ``stock_prefix_reductions``) must all
    be zero, the failure text embedding the Task 2 witness list. P44
    stage scoping is mandatory: only the post-mask partition is
    evaluated and echoed; ``catalog_non_positive_oos_rows_equals_0`` and
    other later-stage entries are neither evaluated nor listed.

    Parameters
    ----------
    features_panel : pd.DataFrame
        NA-kept masked FEATURES_PANEL (29,080 rows) with the compound
        ``(stock_code, trade_date)`` index.
    label_panel : pd.DataFrame
        NA-kept masked LABEL_PANEL (29,080 rows) with the same compound
        index.
    context : Any
        Orchestrator-owned pipeline context (Task 2 surface) carrying
        the integer counters ``forward_fill_operations``,
        ``interpolation_operations`` and ``stock_prefix_reductions``
        plus a ``witness_log`` list. Duck-typed: a missing or malformed
        attribute is a battery failure, not an exception.
    study_config : Dict[str, Any]
        Study configuration; its ``P44_RAW_LAYER_INVARIANTS`` section
        must be a dict whose four post-mask entries are declared
        ``True`` and corroborated by the data and context facts.
    expected_panel_rows : int, optional
        Pinned post-mask usable row count per panel. Default is
        ``_POST_MASK_USABLE_ROWS`` (29,078). Must be a non-negative
        Python ``int``; booleans are rejected.

    Returns
    -------
    Dict[str, Any]
        ``halt_reason`` (``None`` on pass, else every failure joined by
        ``"; "``), ``passed`` (bool), ``evaluated_p44_post_mask`` (the
        four post-mask P44 keys), ``checks`` (named booleans, all
        ``True`` on pass), ``expected_panel_rows`` (int),
        ``usable_rows`` (per panel: rows not fully null),
        ``fully_nulled_rows`` (per panel: fully-nulled row count),
        ``masked_coordinates`` (per panel: sorted
        ``"<stock_code>@<trade_date>"`` strings), ``p44_declared`` (the
        four config declarations, ``None`` where absent).

    Raises
    ------
    TypeError
        If a panel is not a ``pandas.DataFrame``, if ``study_config`` is
        not a dict, or if ``expected_panel_rows`` is not a Python
        ``int`` (booleans rejected).
    ValueError
        If ``expected_panel_rows`` is negative.
    """
    # Validate the boundary types before any structural check runs.
    features = _require_dataframe(features_panel, "assert_post_mask_battery_task_5", "features_panel")
    # Labels boundary check.
    labels = _require_dataframe(label_panel, "assert_post_mask_battery_task_5", "label_panel")
    # The configuration must be the study-config dictionary.
    if not isinstance(study_config, dict):
        # Raise a typed diagnostic naming the argument.
        raise TypeError(f"assert_post_mask_battery_task_5 requires study_config to be a dict, got {type(study_config).__name__}.")
    # The only configurable figure: strict Python int, bool excluded.
    expected = _require_strict_int(expected_panel_rows, "expected_panel_rows")
    # Aggregate failures without short-circuiting: the battery runs in full.
    failures: List[str] = []
    # Deterministic per-check outcomes, all cast to plain bool.
    checks: Dict[str, bool] = {}

    # Local recorder keeps every check homogeneous and the dict all-True on pass.
    def record(name: str, ok: bool, detail: str) -> None:
        """
        Record one battery check and accumulate its failure.

        The outcome is cast to a plain ``bool`` for JSON-native
        artifacts; a failing check appends its detail to the halt reason
        and never raises, so later checks still run.

        Parameters
        ----------
        name : str
            Key written into ``checks``.
        ok : bool
            Outcome of the check.
        detail : str
            Failure text used when ``ok`` is ``False``.
        """
        # Cast to plain bool for JSON-native artifacts.
        checks[name] = bool(ok)
        # A failing check contributes its precise detail to the halt reason.
        if not ok:
            # Accumulate the failure; never raise, so later checks still run.
            failures.append(f"{name}: {detail}")

    # P44 evaluation bookkeeping: only the post-mask partition is ever touched.
    p44 = study_config.get(_P44_SECTION)
    # A missing or malformed P44 section fails the battery rather than the process.
    p44_is_dict = isinstance(p44, dict)
    # Index the four post-mask declarations (defaults to missing sentinel None).
    p44_declared = {key: (p44.get(key, None) if p44_is_dict else None) for key in _P44_POST_MASK_KEYS}

    # --- Compound-index structural checks (both panels) ---
    # Recover each panel's level names once.
    features_names = _compound_index_names(features)
    # Labels' level names once.
    labels_names = _compound_index_names(labels)
    # Record the FEATURES_PANEL index-name check.
    record(
        "features_index_names_compound",
        features_names == list(_COMPOUND_INDEX_NAMES),
        f"FEATURES_PANEL index names must be {list(_COMPOUND_INDEX_NAMES)}, got {features_names}.",
    )
    # Record the LABEL_PANEL index-name check.
    record(
        "labels_index_names_compound",
        labels_names == list(_COMPOUND_INDEX_NAMES),
        f"LABEL_PANEL index names must be {list(_COMPOUND_INDEX_NAMES)}, got {labels_names}.",
    )

    # --- Uniqueness (Step 3, Uniqueness) ---
    # FEATURES_PANEL compound index is_unique, computed on the index itself.
    record(
        "features_index_unique",
        bool(features.index.is_unique),
        "FEATURES_PANEL compound index is not unique; a duplicate inflates the cross-sectional rank denominator.",
    )
    # LABEL_PANEL compound index is_unique, computed on the index itself.
    record(
        "labels_index_unique",
        bool(labels.index.is_unique),
        "LABEL_PANEL compound index is not unique; a duplicate inflates the cross-sectional rank denominator.",
    )
    # Duplicate tuple counts (0 expected) - the P44 duplicate_rows_per_stock_date_equals_0 witness.
    features_duplicate_rows = int(features.index.duplicated(keep=False).sum())
    # Labels' duplicate count.
    labels_duplicate_rows = int(labels.index.duplicated(keep=False).sum())
    # Record the FEATURES_PANEL duplicate-tuple check.
    record(
        "features_duplicate_rows_per_stock_date_equals_0",
        features_duplicate_rows == 0,
        f"FEATURES_PANEL carries {features_duplicate_rows} duplicate (stock_code, trade_date) row(s); must be 0.",
    )
    # Record the LABEL_PANEL duplicate-tuple check.
    record(
        "labels_duplicate_rows_per_stock_date_equals_0",
        labels_duplicate_rows == 0,
        f"LABEL_PANEL carries {labels_duplicate_rows} duplicate (stock_code, trade_date) row(s); must be 0.",
    )

    # --- Required columns and pinned null-column resolution ---
    # Resolve FEATURES_PANEL's pinned columns and required-column deficit.
    features_pinned, features_missing = _pinned_columns_present(features, _PANEL_FEATURES)
    # Resolve LABEL_PANEL's pinned columns and required-column deficit.
    labels_pinned, labels_missing = _pinned_columns_present(labels, _PANEL_LABELS)
    # Record the FEATURES_PANEL required-presence check.
    record(
        "features_required_numeric_columns_present",
        not features_missing,
        f"FEATURES_PANEL is missing required column(s) {features_missing}.",
    )
    # Record the LABEL_PANEL required-presence check.
    record(
        "labels_required_numeric_columns_present",
        not labels_missing,
        f"LABEL_PANEL is missing required column(s) {labels_missing}.",
    )

    # --- Fully-nulled (masked) rows, counted from the data ---
    # Compute the fully-nulled mask for FEATURES_PANEL on its pinned columns.
    features_fully_nulled_mask = _fully_nulled_mask(features, features_pinned)
    # Same for LABEL_PANEL.
    labels_fully_nulled_mask = _fully_nulled_mask(labels, labels_pinned)
    # Count the fully-nulled FEATURES_PANEL rows.
    features_fully_nulled = int(features_fully_nulled_mask.sum())
    # Count the fully-nulled LABEL_PANEL rows.
    labels_fully_nulled = int(labels_fully_nulled_mask.sum())
    # Usable rows are the complement of fully-nulled rows, per panel.
    features_usable = int((~features_fully_nulled_mask).sum())
    # Same for LABEL_PANEL.
    labels_usable = int((~labels_fully_nulled_mask).sum())

    # Record the FEATURES_PANEL fully-nulled-count check (P44 masked_cell_rows_equals_2).
    record(
        "features_fully_nulled_rows_equals_2",
        features_fully_nulled == _POST_MASK_FULLY_NULLED_ROWS,
        f"FEATURES_PANEL carries {features_fully_nulled} fully-nulled rows at the pinned masked columns; must be exactly 2 (rows already removed or double-masked indicate a config or pipeline defect).",
    )
    # Record the LABEL_PANEL fully-nulled-count check.
    record(
        "labels_fully_nulled_rows_equals_2",
        labels_fully_nulled == _POST_MASK_FULLY_NULLED_ROWS,
        f"LABEL_PANEL carries {labels_fully_nulled} fully-nulled rows at the pinned masked columns; must be exactly 2.",
    )
    # Record the FEATURES_PANEL usable-row check (pinned 29,078 via expected_panel_rows).
    record(
        "features_usable_rows_equals_expected",
        features_usable == expected,
        f"FEATURES_PANEL usable rows (not fully-nulled) must equal {expected}, got {features_usable}.",
    )
    # Record the LABEL_PANEL usable-row check.
    record(
        "labels_usable_rows_equals_expected",
        labels_usable == expected,
        f"LABEL_PANEL usable rows (not fully-nulled) must equal {expected}, got {labels_usable}.",
    )

    # --- Cross-panel mask-coordinate agreement ---
    # FEATURES_PANEL's fully-nulled coordinate tuples as a set.
    features_masked_coordinates: set = set()
    # LABEL_PANEL's fully-nulled coordinate tuples as a set.
    labels_masked_coordinates: set = set()
    # Recover the tuples only when both indices are MultiIndex structures.
    if isinstance(features.index, pd.MultiIndex):
        # Zip stock_code (str) and trade_date (Timestamp) for a dtype-stable tuple set.
        features_masked_coordinates = set(
            zip(
                features.index.get_level_values(0)[features_fully_nulled_mask].astype(str).tolist(),
                list(features.index.get_level_values(1)[features_fully_nulled_mask]),
            )
        )
    # Same recovery for LABEL_PANEL.
    if isinstance(labels.index, pd.MultiIndex):
        # Zip stock_code (str) and trade_date (Timestamp).
        labels_masked_coordinates = set(
            zip(
                labels.index.get_level_values(0)[labels_fully_nulled_mask].astype(str).tolist(),
                list(labels.index.get_level_values(1)[labels_fully_nulled_mask]),
            )
        )
    # The two views MUST be masked at the same tuples - a divergent mask voids cross-view agreement.
    record(
        "masked_coordinates_agree_across_panels",
        features_masked_coordinates == labels_masked_coordinates,
        f"FEATURES_PANEL masked tuples {sorted(features_masked_coordinates)} != LABEL_PANEL masked tuples {sorted(labels_masked_coordinates)}.",
    )

    # --- Anti-imputation counters (context duck-typed; halt with the Task-2 witness list) ---
    # Pull the witness list once for use in any counter breach message.
    witness_log = list(getattr(context, "witness_log", [])) if context is not None else []
    # Resolve the three pinned counters with tolerant duck-typing (never crash the battery).
    forward_fill_value = getattr(context, "forward_fill_operations", None) if context is not None else None
    # Same for interpolation.
    interpolation_value = getattr(context, "interpolation_operations", None) if context is not None else None
    # Same for stock-prefix reductions.
    stock_prefix_value = getattr(context, "stock_prefix_reductions", None) if context is not None else None

    # Local integral predicate: accept genuine ints (bool excluded) and numpy integers.
    def _as_int(value: Any) -> Optional[int]:
        """
        Normalise a context counter to a Python int.

        Parameters
        ----------
        value : Any
            Raw counter attribute read from the pipeline context.

        Returns
        -------
        Optional[int]
            The counter as a plain ``int``, or ``None`` when the value
            is a bool or is neither a Python int nor a ``numpy``
            integer.
        """
        # bool is an int subclass; counters must never arrive as truthy/falsy flags.
        if isinstance(value, bool):
            # Reject booleans outright.
            return None
        # A genuine Python int is already authoritative.
        if isinstance(value, int):
            # Return the plain int.
            return int(value)
        # numpy integers are accepted via the integer protocol.
        if isinstance(value, np.integer):
            # Cast numpy integer to Python int.
            return int(value)
        # Anything else is not a counter.
        return None

    # Normalise the three counters; a non-integral value is a failure, not a zero.
    forward_fill_ops = _as_int(forward_fill_value)
    # Same normalisation for interpolation.
    interpolation_ops = _as_int(interpolation_value)
    # Same normalisation for stock-prefix reductions.
    stock_prefix_ops = _as_int(stock_prefix_value)
    # Record the forward-fill counter check; on breach, embed the Task-2 witness list.
    record(
        "context_forward_fill_operations_zero",
        forward_fill_ops == 0,
        f"context.forward_fill_operations == {forward_fill_value!r} != 0 (or not an int); Task-2 witness list: {witness_log}.",
    )
    # Record the interpolation counter check.
    record(
        "context_interpolation_operations_zero",
        interpolation_ops == 0,
        f"context.interpolation_operations == {interpolation_value!r} != 0 (or not an int); Task-2 witness list: {witness_log}.",
    )
    # Record the stock-prefix-reduction counter check.
    record(
        "context_stock_prefix_reductions_zero",
        stock_prefix_ops == 0,
        f"context.stock_prefix_reductions == {stock_prefix_value!r} != 0 (or not an int); Task-2 witness list: {witness_log}.",
    )

    # --- Stage-scoped P44 evaluation (post-mask partition ONLY) ---
    # Fail-closed when the config does not carry the P44 section.
    record(
        "p44_section_present",
        p44_is_dict,
        f"study_config lacks a dict section {_P44_SECTION!r}; the post-mask P44 partition cannot be evaluated.",
    )
    # Data/context corroboration per post-mask key.
    corroboration = {
        "masked_cell_rows_equals_2": features_fully_nulled == _POST_MASK_FULLY_NULLED_ROWS and labels_fully_nulled == _POST_MASK_FULLY_NULLED_ROWS,
        "forward_fill_rows_equals_0": forward_fill_ops == 0,
        "stock_prefix_reductions_equals_0": stock_prefix_ops == 0,
        "duplicate_rows_per_stock_date_equals_0": features_duplicate_rows == 0 and labels_duplicate_rows == 0,
    }
    # Evaluate each of the four post-mask entries: config declares halt AND data corroborates.
    for key in _P44_POST_MASK_KEYS:
        # The halt condition must be declared True in the config.
        declared_true = p44_declared[key] is True
        # Corroborated by the data/context facts gathered above.
        corro = bool(corroboration[key])
        # Record the conjunctive P44 check under a stable key.
        record(
            f"p44_{key}",
            declared_true and corro,
            f"P44.{key} requires config True (got {p44_declared[key]!r}) AND data corroboration (got {corro}).",
        )

    # Assemble the per-panel masked-coordinate disclosures for the artifact.
    masked_coordinates_report = {
        _PANEL_FEATURES: [f"{stock}@{trade_date}" for stock, trade_date in sorted(features_masked_coordinates)],
        _PANEL_LABELS: [f"{stock}@{trade_date}" for stock, trade_date in sorted(labels_masked_coordinates)],
    }
    # The halt reason aggregates every failure, in deterministic order.
    halt_reason = "; ".join(failures) if failures else None
    # Return the full battery proof - the stage output carries evidence, not a Boolean.
    return {
        "halt_reason": halt_reason,
        "passed": not failures,
        "evaluated_p44_post_mask": list(_P44_POST_MASK_KEYS),
        "checks": checks,
        "expected_panel_rows": int(expected),
        "usable_rows": {_PANEL_FEATURES: features_usable, _PANEL_LABELS: labels_usable},
        "fully_nulled_rows": {_PANEL_FEATURES: features_fully_nulled, _PANEL_LABELS: labels_fully_nulled},
        "masked_coordinates": masked_coordinates_report,
        "p44_declared": p44_declared,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 5, Orchestrator: run_task_5
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_5(
    *,
    unadjusted_bars: pd.DataFrame,
    features_panel: pd.DataFrame,
    label_panel: pd.DataFrame,
    mask_register: pd.DataFrame,
    study_config: Dict[str, Any],
    context: Any,
    output_directory: str | os.PathLike,
) -> Dict[str, Any]:
    """
    Run Task 5's three steps in order and return the stage bundle.

    The orchestrator is keyword-only. In pinned order it (1) builds the
    727-row dense trading calendar from ``UNADJUSTED_BARS``, pins its
    static counts and writes ``trading_calendar.csv``; (2) verifies the
    MASK_REGISTER, applies the two masked stock-day cells to deep copies
    of both panels and writes ``mask_application_witness.csv``; (3) runs
    the post-mask invariant battery on the NA-kept panels with the
    pinned default ``expected_panel_rows`` of 29,078, then derives the
    usable views by excluding exactly the two masked tuples (29,080 - 2
    = 29,078). ``post_mask_battery.json`` is written in every outcome,
    halt or success. ``calendar_hash`` is the SHA-256 of the calendar
    CSV bytes exactly as persisted. A failing step records a
    stage-namespaced ``halt_reason``, leaves later stage products
    ``None`` and preserves the bundle keys; artifacts are written for
    the stages that completed.

    Parameters
    ----------
    unadjusted_bars : pd.DataFrame
        The UNADJUSTED_BARS frame. Drives the calendar; never mutated.
    features_panel : pd.DataFrame
        Task 4 FEATURES_PANEL (29,080 rows, compound index). Never
        mutated; a deep copy is masked and returned.
    label_panel : pd.DataFrame
        Task 4 LABEL_PANEL (29,080 rows, compound index). Never mutated;
        a deep copy is masked and returned.
    mask_register : pd.DataFrame
        The two-row MASK_REGISTER, verified inside Step 2.
    study_config : Dict[str, Any]
        Study configuration (STUDY_CONFIG). Its
        ``P44_RAW_LAYER_INVARIANTS`` section is consulted by Step 3
        only, stage-scoped; ``P12_MASKING_PARAMETERS`` supplies the
        masking disclosures.
    context : Any
        Orchestrator-owned pipeline context (Task 2 surface), read by
        the Step 3 battery and never mutated here.
    output_directory : str | os.PathLike
        Destination directory for the three artifacts; created with its
        parents when absent.

    Returns
    -------
    Dict[str, Any]
        ``calendar`` (the 727-row calendar), ``calendar_hash`` (64
        lowercase hex SHA-256 of the serialised calendar CSV),
        ``masked_features_panel`` and ``masked_label_panel`` (NA-kept
        copies, 29,080 rows), ``usable_features_panel`` and
        ``usable_label_panel`` (29,078 rows; the two masked tuples
        excluded, order preserved), ``mask_witness``,
        ``post_mask_battery``, ``disclosures`` (including
        ``date_index_construction ==
        "DENSE_ROW_INDEX_OVER_727_TRADING_ROWS"``), ``halt_reason``
        (``None`` on success, else a stage-namespaced string),
        ``calendar_audit`` and ``mask_register_validated``. Products of
        a stage that did not run are ``None``.

    Raises
    ------
    TypeError
        If ``output_directory`` is neither a str nor path-like, if
        ``study_config`` is not a dict, or on any boundary type breach.
    ValueError
        If ``output_directory`` resolves to an empty path.
    OSError
        If the output directory cannot be created, or if
        ``post_mask_battery.json`` cannot be written; failures writing
        ``trading_calendar.csv`` or ``mask_application_witness.csv`` are
        caught by their stage and recorded in ``halt_reason`` instead.
    """
    # Resolve and validate the output directory before any work happens.
    out_dir = _require_string_or_path(output_directory, "output_directory")
    # study_config must be the configuration dictionary.
    if not isinstance(study_config, dict):
        # Raise a typed diagnostic naming the argument.
        raise TypeError(f"run_task_5 requires study_config to be a dict, got {type(study_config).__name__}.")
    # Create the destination directory (parents included) up front.
    out_dir.mkdir(parents=True, exist_ok=True)
    # Deterministic artifact paths beneath the output directory.
    calendar_path = out_dir / "trading_calendar.csv"
    # Witness CSV path.
    witness_path = out_dir / "mask_application_witness.csv"
    # Battery JSON path.
    battery_path = out_dir / "post_mask_battery.json"
    # Initialise the bundle with the unproduced-state defaults.
    halt_reason: Optional[str] = None
    # Step 1 products.
    calendar = None
    # Calendar audit record (success path only).
    calendar_audit: Optional[Dict[str, Any]] = None
    # SHA-256 of the serialised calendar.
    calendar_hash: Optional[str] = None
    # Step 2 products.
    masked_features = None
    # Masked labels product.
    masked_labels = None
    # Witness product.
    witness = None
    # Validated register product.
    mask_register_validated = None
    # Step 3 products.
    battery: Optional[Dict[str, Any]] = None
    # Usable views.
    usable_features = None
    # Labels usable view.
    usable_labels = None

    # ----- Step 1: construct the dense trading calendar and pin its counts -----
    try:
        # Build the 727-row dense calendar from the unadjusted view only.
        calendar = construct_calendar_task_5(unadjusted_bars)
        # Pin the static counts; raises AssertionError naming the first breach.
        calendar_audit = assert_calendar_counts_task_5(calendar)
        # Serialise once (index=False, LF) and hash exactly those bytes; the same bytes are persisted.
        calendar_csv = calendar.to_csv(index=False, lineterminator="\n")
        # SHA-256 over the utf-8 encoding of the serialised frame.
        calendar_hash = hashlib.sha256(calendar_csv.encode("utf-8")).hexdigest()
        # Persist the calendar artifact.
        calendar_path.write_text(calendar_csv, encoding="utf-8")
    except Exception as exc:  # noqa: BLE001 - the stage boundary records, never swallows
        # Stage-namespaced halt reason naming the failing step.
        halt_reason = f"Task 5, Step 1 (trading calendar) halted: {type(exc).__name__}: {exc}"

    # ----- Step 2: verify the register and apply the two masked cells -----
    if halt_reason is None:
        try:
            # Verify the register; the copy is what the stage consumes.
            mask_register_validated = verify_mask_register_task_5(mask_register)
            # Coordinate-addressed null-fill on deep copies of both panels.
            applied = apply_mask_task_5(features_panel, label_panel, mask_register_validated)
            # Detach the NA-kept masked panels.
            masked_features = applied["features_panel"]
            # Labels side.
            masked_labels = applied["label_panel"]
            # Witness for the artifact.
            witness = applied["witness"]
            # Persist the witness CSV.
            witness.to_csv(witness_path, index=False, lineterminator="\n")
        except Exception as exc:  # noqa: BLE001 - the stage boundary records, never swallows
            # Stage-namespaced halt reason naming the failing step.
            halt_reason = f"Task 5, Step 2 (masked cells) halted: {type(exc).__name__}: {exc}"

    # ----- Step 3: execute the post-masking invariant battery in full -----
    if halt_reason is None:
        # Execute the battery on the NA-kept masked panels; default 29,078 binds.
        battery = assert_post_mask_battery_task_5(masked_features, masked_labels, context, study_config)
        # A battery halt is a stage halt.
        if battery["halt_reason"] is not None:
            # Stage-namespaced halt reason.
            halt_reason = f"Task 5, Step 3 (post-mask battery) halted: {battery['halt_reason']}"
        else:
            # The two masked coordinates, exactly as pinned by Step 2.
            masked_coordinates = _masked_coordinate_tuples(mask_register_validated)
            # Exclude exactly the two masked tuples - never a isna-based drop.
            features_keep = _exclusion_mask(masked_features, masked_coordinates)
            # Same exclusion for the labels panel.
            labels_keep = _exclusion_mask(masked_labels, masked_coordinates)
            # Usable views: 29,080 - 2 = 29,078 rows, order preserved.
            usable_features = masked_features.loc[features_keep]
            # Same for the labels panel.
            usable_labels = masked_labels.loc[labels_keep]
    # ----- Artifact: post_mask_battery.json (always written, success or halt) -----
    # Compose the JSON payload with only JSON-native values.
    battery_payload = dict(battery) if battery is not None else {
        "halt_reason": halt_reason,
        "passed": False,
        "evaluated_p44_post_mask": list(_P44_POST_MASK_KEYS),
        "checks": {},
    }
    # Bind the calendar hash and disclosure into the artifact when available.
    battery_payload["calendar_hash"] = calendar_hash
    # Bind the [R] construction disclosure.
    battery_payload["date_index_construction"] = _DATE_INDEX_CONSTRUCTION_DISCLOSURE
    # Write the battery artifact; default=str keeps any stragglers JSON-safe without hiding them.
    battery_path.write_text(json.dumps(battery_payload, indent=2, ensure_ascii=False, allow_nan=False, default=str), encoding="utf-8")

    # ----- Disclosures ([R] register for this stage) -----
    # The reportable terminal set is COMPUTED from the calendar this stage built; never returned as a literal.
    terminal_unlabelled = (
        calendar.loc[calendar["in_label_window"] & ~calendar["label_available"], "row_index"].tolist()
        if calendar is not None
        else None
    )
    # The P12 register of the supplied configuration is the source of the masking-rule disclosures (pinned constants as fallback).
    p12 = study_config.get("P12_MASKING_PARAMETERS", {})
    # Number of masked stock-day cells, as declared by the configuration (P12 masked_cell_count).
    disclosed_masked_cells = p12.get("masked_cell_count", _POST_MASK_FULLY_NULLED_ROWS)
    # Post-mask stock count, as declared (P12 stock_count_after_masking).
    disclosed_stock_count = p12.get("stock_count_after_masking", _POST_MASK_STOCK_COUNT)
    # Post-mask file count, as declared (P12 file_count_after_masking).
    disclosed_file_count = p12.get("file_count_after_masking", _POST_MASK_FILE_COUNT)
    # Assemble the disclosure block; values computed from the inputs, never literals.
    disclosures: Dict[str, Any] = {
        "date_index_construction": _DATE_INDEX_CONSTRUCTION_DISCLOSURE,
        "row_index_immutability": "DENSE_Int64_1_TO_727_NON_INDEX_COLUMN_MIRRORED_TO_RangeIndex",
        "in_label_window_definition": "year in {2024, 2025} (cardinality 485)",
        "label_available_definition": "in_label_window AND row_index + 5 <= 727 (cardinality 480; in_label_window conjunct mandatory; raw r + 5 <= 727 yields 722 and is never applied)",
        "terminal_unlabelled_rows": terminal_unlabelled,
        "masked_cells": disclosed_masked_cells,
        "masked_cell_semantics": "TWO_STOCK_DAY_CELLS_AT_DISTINCT_COORDINATES_BOTH_VIEWS_SAME_TUPLE",
        "masked_value_representation": "pd.NA_NEVER_ZERO_NEVER_FILLED",
        "forward_fill_permitted": bool(p12.get("forward_fill_permitted", False)),
        "backward_fill_permitted": bool(p12.get("backward_fill_permitted", False)),
        "interpolation_permitted": bool(p12.get("interpolation_permitted", False)),
        "stock_prefix_substitution_permitted": bool(p12.get("stock_prefix_substitution_permitted", False)),
        "drop_stock_permitted": False,
        "usable_rows_per_view": int(usable_features.shape[0]) if usable_features is not None else None,
        "masked_rows_retained_in_na_kept_panels": int(masked_features.shape[0]) if masked_features is not None else None,
        "stock_count_after_masking": disclosed_stock_count,
        "file_count_after_masking": disclosed_file_count,
        "post_mask_stock_day_cells": disclosed_masked_cells,
        "calendar_hash_sha256": calendar_hash,
        "p44_stage_scope": "post_mask partition only; catalog_non_positive_oos_rows_equals_0 is a Task 10 assertion and is not evaluated here",
        "p35_ingest_ordering": "separate_views precedes mask_non_trading_cells; usable slice excludes exactly the two masked tuples",
    }
    # Return the fully-structured bundle; unproduced values are None on halt.
    return {
        "calendar": calendar,
        "calendar_hash": calendar_hash,
        "masked_features_panel": masked_features,
        "masked_label_panel": masked_labels,
        "usable_features_panel": usable_features,
        "usable_label_panel": usable_labels,
        "mask_witness": witness,
        "post_mask_battery": battery,
        "disclosures": disclosures,
        "halt_reason": halt_reason,
        "calendar_audit": calendar_audit,
        "mask_register_validated": mask_register_validated,
    }
```

```python
# Task 6: Label Construction per Eq. (1)

# ==============================================================================
# Task 6: Label Construction per Eq. (1)
# ==============================================================================

# Pinned literal for the withdrawal of the raw task's y == 0 ∧ defined == False test.
WITHDRAWN_TAG: str = "WITHDRAWN_FROM_RAW_TASK_PER_STANDING_REGISTER"
# The four literal values of the diagnostic __label_status column (Task 6, Step 3).
_LABEL_STATUS_OK: str = "ok"
_LABEL_STATUS_ENDPOINT_NAN: str = "endpoint_nan"
_LABEL_STATUS_TERMINAL: str = "terminal"
_LABEL_STATUS_MASKED: str = "masked"
# The pinned columns the pair table is required to carry (REQUIRED_INTERFACE).
_PAIR_TABLE_COLUMNS: Tuple[str, ...] = (
    "row_index",
    "close_adj",
    "tradable",
    "close_adj_lead_5",
    "tradable_lead_5",
    "row_index_lead",
    "y",
    "oos_sharpe",
    "oos_computed_at",
)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 6: Shared private boundary helpers
# -------------------------------------------------------------------------------------------------------------------------------

def _require_dataframe(obj: Any, caller: str, name: str) -> pd.DataFrame:
    """
    Validate that an object is a ``pandas.DataFrame``.

    Return the object unchanged when it is a frame; otherwise raise a
    typed diagnostic naming the caller and the argument. Pure type
    guard: no copy is made and no attribute of ``obj`` is touched
    before the check.

    Parameters
    ----------
    obj : Any
        Candidate object supplied by the caller.
    caller : str
        Name of the calling callable, used in the diagnostic.
    name : str
        Argument name as it appears in the caller's signature.

    Returns
    -------
    pd.DataFrame
        The same object, unmodified.

    Raises
    ------
    TypeError
        If ``obj`` is not a :class:`pandas.DataFrame`.
    """
    # Reject anything that is not a DataFrame before any attribute access on it.
    if not isinstance(obj, pd.DataFrame):
        # Raise a typed diagnostic naming the callable and the argument.
        raise TypeError(
            f"{caller} requires {name} to be a pandas.DataFrame, got {type(obj).__name__}."
        )
    # Return the caller-owned frame; this boundary check copies nothing.
    return obj

def _require_integral(value: Any, caller: str, name: str) -> int:
    """
    Validate that a value is a non-negative Python ``int``.

    Booleans are rejected explicitly because ``bool`` subclasses
    ``int`` and would otherwise pass silently. NumPy integer scalars
    are accepted and converted to plain Python ``int``; any other type
    is refused.

    Parameters
    ----------
    value : Any
        Candidate scalar supplied by the caller.
    caller : str
        Name of the calling callable, used in the diagnostic.
    name : str
        Argument name as it appears in the caller's signature.

    Returns
    -------
    int
        The count as a plain Python integer, at least zero.

    Raises
    ------
    TypeError
        If ``value`` is a bool or is not an integral scalar.
    ValueError
        If the integer is negative.
    """
    # Reject bools explicitly: bool is a subclass of int and would silently pass.
    if isinstance(value, bool):
        # Raise a typed diagnostic naming the callable and the argument.
        raise TypeError(f"{caller} requires {name} to be an integral count, got a bool.")
    # Accept numpy integer scalars and normalise them to Python int.
    if isinstance(value, np.integer):
        # Normalise numpy integers before the isinstance gate below.
        value = int(value)
    # Reject non-integer scalars outright.
    if not isinstance(value, int):
        # Raise a typed diagnostic naming the callable and the argument.
        raise TypeError(
            f"{caller} requires {name} to be an integral count, got {type(value).__name__}."
        )
    # Reject negative counts: a negative expectation is meaningless.
    if value < 0:
        # Raise a typed diagnostic naming the bound.
        raise ValueError(f"{caller} requires {name} to be non-negative, got {value}.")
    # Return the validated plain integer.
    return value

def _validate_calendar(calendar: Any, caller: str) -> pd.DataFrame:
    """
    Validate the dense trading calendar at the module boundary.

    Check the pinned columns ``row_index``, ``trade_date``, ``year``,
    ``in_label_window`` and ``label_available`` together with their
    dtypes; that ``row_index`` is unique and dense ``1..N``; that rows
    are sorted ascending by ``row_index``; and that ``trade_date`` is
    naive ``datetime64[ns]``, ascending, unique and fully non-null.
    The calendar is the coordinate system on which the five-row
    forward label is defined, so a gap or misalignment would silently
    change the horizon.

    Parameters
    ----------
    calendar : Any
        The Task 5 trading calendar.
    caller : str
        Name of the calling callable, used in diagnostics.

    Returns
    -------
    pd.DataFrame
        The same, now-validated calendar; no copy is made.

    Raises
    ------
    TypeError
        If ``calendar`` is not a DataFrame, a pinned column is missing
        or mistyped, ``row_index`` or ``trade_date`` holds NA/NaT, or
        the frame is not sorted ascending by ``row_index`` with
        ascending ``trade_date``.
    ValueError
        If ``row_index`` is not dense ``1..N`` or not unique, or
        ``trade_date`` is duplicated.
    """
    # Boundary type check before any column access.
    frame = _require_dataframe(calendar, caller, "calendar")
    # The pinned P33 column set; absence of any is a halt.
    required: Tuple[str, ...] = (
        "row_index", "trade_date", "year", "in_label_window", "label_available",
    )
    # Compute the missing columns without mutating the caller's frame.
    missing: List[str] = [c for c in required if c not in frame.columns]
    # Halt with the full deficit when any pinned column is absent.
    if missing:
        # Raise a typed diagnostic listing every missing pinned column.
        raise TypeError(
            f"{caller} requires the calendar to carry the pinned P33 columns "
            f"{sorted(required)}; missing {missing}."
        )
    # row_index must be an integer dtype (Int64 per P33); NA is rejected below.
    row_index: pd.Series = frame["row_index"]
    # Reject non-integer dtypes before any arithmetic on the coordinate.
    if not pd.api.types.is_integer_dtype(row_index):
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(
            f"{caller} requires calendar.row_index to be an (nullable) integer dtype, "
            f"got {row_index.dtype}."
        )
    # trade_date must be naive datetime64[ns] per P33; tz-aware or non-ns halts.
    if str(frame["trade_date"].dtype) != "datetime64[ns]":
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(
            f"{caller} requires calendar.trade_date to be datetime64[ns] (naive), "
            f"got {frame['trade_date'].dtype}."
        )
    # Year must be an integer dtype per P33.
    if not pd.api.types.is_integer_dtype(frame["year"]):
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(
            f"{caller} requires calendar.year to be an (nullable) integer dtype, "
            f"got {frame['year'].dtype}."
        )
    # Both derived booleans must be (nullable) boolean per P33.
    for boolean_column in ("in_label_window", "label_available"):
        # is_bool_dtype covers both bool and the nullable boolean dtype.
        if not pd.api.types.is_bool_dtype(frame[boolean_column]):
            # Raise a typed diagnostic naming the offending column and dtype.
            raise TypeError(
                f"{caller} requires calendar.{boolean_column} to be a (nullable) boolean, "
                f"got {frame[boolean_column].dtype}."
            )
    # NA anywhere in the coordinate column breaks the many_to_one merge; halt.
    if row_index.isna().any():
        # Raise a typed diagnostic naming the column.
        raise TypeError(f"{caller} requires calendar.row_index to be fully non-null.")
    # A NaT in trade_date breaks the many_to_one merge; halt with count.
    if frame["trade_date"].isna().any():
        # Raise a typed diagnostic naming the column.
        raise TypeError(f"{caller} requires calendar.trade_date to be fully non-null.")
    # Materialise the int64 coordinate array for the density audit.
    rows: np.ndarray = row_index.astype("int64").to_numpy()
    # Uniqueness of the coordinate is what makes r -> r+5 a valid offset; halt on dups.
    if np.unique(rows).size != rows.size:
        # Raise a typed diagnostic naming the defect.
        raise ValueError(f"{caller} requires calendar.row_index to be unique; duplicates found.")
    # Dense 1..N: sorted row_index must equal arange(1, N+1) exactly (P33 arithmetic note).
    if not np.array_equal(np.sort(rows), np.arange(1, rows.size + 1, dtype=np.int64)):
        # Raise a typed diagnostic naming the pinned density rule.
        raise ValueError(
            f"{caller} requires calendar.row_index to be dense 1..N with no gaps; "
            f"got min={rows.min()}, max={rows.max()}, len={rows.size}."
        )
    # The calendar must be ascending by row_index with ascending trade_date; halt otherwise.
    if not np.array_equal(rows, np.sort(rows)):
        # Raise a typed diagnostic naming the pinned ordering.
        raise TypeError(f"{caller} requires calendar rows to be sorted ascending by row_index.")
    # Trade dates must be non-decreasing in row_index order (strictly ascending, asserted below).
    if not bool(frame["trade_date"].is_monotonic_increasing):
        # Raise a typed diagnostic naming the pinned ordering.
        raise TypeError(
            f"{caller} requires calendar.trade_date to be ascending in row_index order."
        )
    # Uniqueness of trade_date (a bijection row_index <-> trade_date); halt on duplicates.
    if frame["trade_date"].duplicated().any():
        # Raise a typed diagnostic naming the defect.
        raise ValueError(f"{caller} requires calendar.trade_date to be unique.")
    # Return the caller-owned, now-validated calendar; no copy is made.
    return frame

def _validate_label_panel(label_panel: Any, caller: str) -> pd.DataFrame:
    """
    Validate LABEL_PANEL and return a normalised copy of it.

    Accept either the pipeline's ``(stock_code, trade_date)`` compound
    index or a flat frame carrying those two columns. When both
    ``oos_sharpe`` and ``oos_computed_at`` are absent, attach all-NaN /
    all-NaT sentinels on a copy (they are null by construction in this
    reproduction); a panel carrying exactly one of them is rejected as
    inconsistent. Verify the pinned dtypes, the non-null key columns, a
    non-empty body and a unique compound key, then project exactly the
    six consumed columns onto a fresh ``RangeIndex``.

    Parameters
    ----------
    label_panel : Any
        The Task 5 (NA-kept) LABEL_PANEL. Never mutated.
    caller : str
        Name of the calling callable, used in diagnostics.

    Returns
    -------
    pd.DataFrame
        A new frame, one row per stock-day, with columns
        ``stock_code``, ``trade_date``, ``close_adj``, ``tradable``,
        ``oos_sharpe`` and ``oos_computed_at``, on a ``RangeIndex``.

    Raises
    ------
    TypeError
        If the panel is not a DataFrame, the compound key is found in
        neither index nor columns, a pinned column is missing or
        mistyped, exactly one OOS sentinel is present, or
        ``stock_code``/``trade_date`` holds NA.
    ValueError
        If the panel is empty or its compound key is duplicated.
    """
    # Boundary type check before any index or column access.
    frame = _require_dataframe(label_panel, caller, "label_panel")
    # Normalise orientation: indexed compound key, or flat columns; else halt.
    index_names: Tuple[str, ...] = tuple(frame.index.names)
    # Branch on the pipeline's indexed layout first.
    if index_names == _COMPOUND_INDEX_NAMES:
        # Restore the key columns from the compound index; keep the row order.
        keyed = frame.reset_index()
    elif all(c in frame.columns for c in _COMPOUND_INDEX_NAMES):
        # Tolerate a flat frame whose columns carry the key; keep the caller's row order.
        keyed = frame.copy()
    else:
        # Halt: the compound key exists nowhere findable.
        raise TypeError(
            f"{caller} requires label_panel to be indexed on {list(_COMPOUND_INDEX_NAMES)} "
            f"or to carry those columns."
        )
    # Reconcile the Task 4 / Task 6 pins (operator ruling 2026-11-12 on F-4-af1c548b98 and
    # F-6-11a0a59beb / F-6-48b2a06dc7 / F-6-bf1f9131df / F-6-d23d34d4ac): Task 4 exports the
    # pinned five-column LABEL_PANEL, which carries no OOS columns; the reproduction's OOS
    # columns are null/NaT BY CONSTRUCTION (the freeze runs before any OOS read), so when both
    # are absent they are attached here as all-NaN / all-NaT sentinels and the Step 3 OOS-null
    # test still runs on them. A panel carrying exactly one of the two is inconsistent and halts.
    oos_present: Tuple[bool, bool] = ("oos_sharpe" in keyed.columns, "oos_computed_at" in keyed.columns)
    # Both absent: the Task 4 export; attach the by-construction sentinels on a copy.
    if oos_present == (False, False):
        # Never mutate the caller's frame.
        keyed = keyed.copy()
        # The reproduction OOS Sharpe sentinel is all-NaN float64.
        keyed["oos_sharpe"] = pd.Series(np.nan, index=keyed.index, dtype="float64")
        # The reproduction OOS timestamp sentinel is all-NaT datetime64[ns].
        keyed["oos_computed_at"] = pd.Series(pd.NaT, index=keyed.index).astype("datetime64[ns]")
    # Exactly one present: the panel's provenance is inconsistent; halt naming the missing one.
    elif oos_present != (True, True):
        # Name the absent sentinel.
        missing_sentinel: str = "oos_sharpe" if not oos_present[0] else "oos_computed_at"
        # Raise a typed diagnostic.
        raise TypeError(
            f"{caller} requires label_panel to carry both OOS sentinels or neither "
            f"(the Task 4 five-column export); {missing_sentinel!r} is missing."
        )
    # The required float64 columns (P01/P27); the pinned schema kind is float64.
    required_float: Tuple[str, ...] = ("close_adj", "oos_sharpe")
    # Validate the float64-required columns without coercing anything.
    for column in required_float:
        # Absence of a pinned numeric column is a halt, not a silent skip.
        if column not in keyed.columns:
            # Raise a typed diagnostic naming the pinned column.
            raise TypeError(f"{caller} requires label_panel to carry column {column!r}.")
        # The pinned schema is float64; other numeric kinds are refused.
        if keyed[column].dtype.kind != "f":
            # Raise a typed diagnostic naming the pinned dtype.
            raise TypeError(
                f"{caller} requires label_panel.{column} to be float64, got {keyed[column].dtype}."
            )
    # tradable must be a (nullable) boolean per P22's boolean dtype policy.
    if "tradable" not in keyed.columns:
        # Raise a typed diagnostic naming the pinned column.
        raise TypeError(f"{caller} requires label_panel to carry column 'tradable'.")
    # is_bool_dtype covers bool and the nullable boolean dtype.
    if not pd.api.types.is_bool_dtype(keyed["tradable"]):
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(
            f"{caller} requires label_panel.tradable to be a (nullable) boolean, "
            f"got {keyed['tradable'].dtype}."
        )
    # oos_computed_at must be a datetime64 column (pinned datetime64[ns]).
    if "oos_computed_at" not in keyed.columns:
        # Raise a typed diagnostic naming the pinned sentinel column.
        raise TypeError(f"{caller} requires label_panel to carry column 'oos_computed_at'.")
    # Any datetime64 unit is accepted and preserved; non-datetime halts.
    if not pd.api.types.is_datetime64_any_dtype(keyed["oos_computed_at"]):
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(
            f"{caller} requires label_panel.oos_computed_at to be datetime64, "
            f"got {keyed['oos_computed_at'].dtype}."
        )
    # The key columns must be non-null: a null stock code or date cannot be joined.
    if keyed["stock_code"].isna().any():
        # Raise a typed diagnostic naming the column.
        raise TypeError(f"{caller} requires label_panel.stock_code to be fully non-null.")
    # A NaT in trade_date cannot be aligned to the calendar; halt with count.
    if keyed["trade_date"].isna().any():
        # Raise a typed diagnostic naming the column.
        raise TypeError(f"{caller} requires label_panel.trade_date to be fully non-null.")
    # An empty panel cannot be labelled; halt rather than emit an empty bundle silently.
    if keyed.shape[0] == 0:
        # Raise a typed diagnostic naming the defect.
        raise ValueError(f"{caller} requires a non-empty label_panel.")
    # The compound key must be unique; duplicates corrupt every t+5 join cardinality.
    duplicated: np.ndarray = keyed.duplicated(subset=["stock_code", "trade_date"]).to_numpy()
    # Halt with the duplicate count when any coordinate repeats.
    if bool(duplicated.any()):
        # Count the offending tuples for the diagnostic.
        offenders: int = int(duplicated.sum())
        # Raise a typed diagnostic naming the defect and the count.
        raise ValueError(
            f"{caller} requires a unique (stock_code, trade_date) key; found {offenders} duplicate(s)."
        )
    # Project exactly the columns the label path consumes; drop everything else.
    projected: pd.DataFrame = keyed.loc[
        :,
        ["stock_code", "trade_date", "close_adj", "tradable", "oos_sharpe", "oos_computed_at"],
    ].copy()
    # Restore a plain RangeIndex for internal vectorisation.
    projected = projected.reset_index(drop=True)
    # Return the normalised frame; the caller's panel was never mutated.
    return projected

def _label_available_dates(labels: pd.DataFrame) -> pd.Index:
    """
    Return the trade dates carrying at least one non-NaN label.

    Locate ``trade_date`` in the columns or in the index levels, keep
    the rows whose ``y`` is non-NaN, and return the unique dates sorted
    ascending. The count is calendar-side (dates, not stock-days);
    individual stocks may carry fewer labels because of masking or
    tradability failures.

    Parameters
    ----------
    labels : pd.DataFrame
        The admissibility-tagged pair table, carrying ``y`` and either
        a ``trade_date`` column or a ``trade_date`` index level.

    Returns
    -------
    pd.Index
        Unique dates with at least one non-NaN ``y``, sorted ascending.

    Raises
    ------
    TypeError
        If no ``trade_date`` column or index level exists.
    """
    # Locate the trade_date axis: prefer a column, fall back to the index level.
    if "trade_date" in labels.columns:
        # Column layout (flat frame): use the column directly.
        dates: pd.Index = pd.Index(labels["trade_date"])
    elif "trade_date" in labels.index.names:
        # Compound-index layout: read the level values.
        dates = pd.Index(labels.index.get_level_values("trade_date"))
    else:
        # Halt: no date axis findable; the aggregate cannot be computed.
        raise TypeError(
            "_label_available_dates requires a 'trade_date' index level or column."
        )
    # Restrict to cells whose y is non-NaN (Eq. (1) defined and admissible).
    available: pd.Index = dates[labels["y"].notna().to_numpy()]
    # Return the unique dates, sorted ascending, calendar-side grain.
    return available.unique().sort_values()

# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 1:  construct_labels_task_6
# -------------------------------------------------------------------------------------------------------------------------------

def construct_labels_task_6(
    label_panel: pd.DataFrame,
    calendar: pd.DataFrame,
) -> pd.DataFrame:
    """
    Build the forward five-row pair table of Eq. (1).

    Validate the calendar and the panel; map every panel ``trade_date``
    onto the calendar's dense ``row_index`` with a ``many_to_one``
    merge; build the lead table by re-keying each row at
    ``row_index - 5``, so the lead row keyed at ``r`` holds the row
    ``r + 5`` of the same stock; then left-join base and lead on
    ``(stock_code, row_index)``. The left join retains every input row,
    terminals included: they have no successor, so their lead columns
    stay NA and their label is NaN. The raw ratio

        y = close_adj[i, t+5] / close_adj[i, t] - 1

    is evaluated in float64 with no shift, fill or interpolation; NaN
    and infinite values are admitted here and removed only by the
    endpoint admissibility gate.

    Parameters
    ----------
    label_panel : pd.DataFrame
        The Task 5 (NA-kept) LABEL_PANEL, indexed on
        ``(stock_code, trade_date)`` or carrying those columns. Must
        hold ``close_adj`` (float64), ``tradable`` ((nullable)
        boolean) and the OOS sentinels. Never mutated.
    calendar : pd.DataFrame
        The Task 5 calendar with a dense ``row_index`` and naive,
        unique ``trade_date``. Never mutated.

    Returns
    -------
    pd.DataFrame
        Pair table indexed on ``(stock_code, trade_date)``, one row per
        input row, in the input row order, with columns ``row_index``
        (Int64), ``close_adj`` (float64), ``tradable`` ((nullable)
        boolean), ``close_adj_lead_5`` (float64), ``tradable_lead_5``
        ((nullable) boolean), ``row_index_lead`` (Int64, NA on
        terminal rows), ``y`` (float64, Eq. (1), NaN where no ``t+5``
        row exists) and the OOS sentinels ``oos_sharpe`` /
        ``oos_computed_at``.

    Raises
    ------
    TypeError
        Propagated from the calendar and panel validators on any type
        or dtype breach, including NA in a key column.
    ValueError
        If the calendar is not dense or is duplicated, the panel is
        empty or its key duplicated, or any panel ``trade_date`` is
        absent from the calendar.
    """
    # Name the caller for boundary diagnostics.
    caller = "construct_labels_task_6"
    # Validate the calendar first: it defines the coordinate system for the join.
    cal: pd.DataFrame = _validate_calendar(calendar, caller)
    # Validate and normalise the panel; the caller's frame is never mutated.
    panel: pd.DataFrame = _validate_label_panel(label_panel, caller)
    # Every panel trade_date must exist in the calendar; otherwise the join misaligns.
    calendar_dates: pd.Index = pd.Index(cal["trade_date"])
    # Vectorised membership test on the panel's dates.
    outside: np.ndarray = ~panel["trade_date"].isin(calendar_dates).to_numpy()
    # Halt with the offending count when any panel date is absent from the calendar.
    if bool(outside.any()):
        # Count the rows that would misalign (masked cells must be NA, never removed).
        offenders: int = int(outside.sum())
        # Raise a typed diagnostic naming the defect and its extent.
        raise ValueError(
            f"{caller} requires every label_panel trade_date to be present in the calendar; "
            f"{offenders} row(s) are not."
        )
    # Attach row_index to every panel row via a many_to_one merge on trade_date.
    keyed: pd.DataFrame = panel.merge(
        cal.loc[:, ["trade_date", "row_index"]],  # the (trade_date -> row_index) mapping
        on="trade_date",  # join key: the calendar's date axis
        how="left",  # retain every panel row, including terminals
        validate="many_to_one",  # each panel row maps to exactly one calendar row
        sort=False,  # preserve the panel's row order
    )
    # Base table: one row per (stock_code, row_index); the label's left endpoint.
    base: pd.DataFrame = keyed.loc[
        :, ["stock_code", "trade_date", "row_index", "close_adj", "tradable", "oos_sharpe", "oos_computed_at"]
    ].copy()
    # Lead table ("LABEL_PANEL_PLUS_5"): same rows; successor-side columns renamed.
    lead: pd.DataFrame = keyed.loc[:, ["stock_code", "row_index", "close_adj", "tradable"]].copy()
    # Rename the lead columns and the coordinate (row_index -> row_index_lead).
    lead = lead.rename(
        columns={
            "close_adj": "close_adj_lead_5",  # numerator price of Eq. (1)
            "tradable": "tradable_lead_5",  # trailing-endpoint tradability
            "row_index": "row_index_lead",  # the successor's calendar coordinate
        }
    )
    # Re-key the lead rows at the BASE row's index: lead[r] := original row (r + 5).
    # Int64 arithmetic keeps the column nullable and NaN-free by construction.
    lead["row_index"] = lead["row_index_lead"] - _FORWARD_HORIZON_ROWS
    # Left-merge base with lead on (stock_code, row_index); terminals stay NaN.
    merged: pd.DataFrame = base.merge(
        lead.loc[:, ["stock_code", "row_index", "close_adj_lead_5", "tradable_lead_5", "row_index_lead"]],
        on=["stock_code", "row_index"],  # the index-level join of the step text
        how="left",  # RETAIN terminals (an inner join would drop rows 723-727)
        validate="one_to_one",  # at most one successor row per base row
        sort=False,  # preserve the panel's row order
    )
    # Eq. (1): y_{i,t} = P^adj_{i,t+5} / P^adj_{i,t} - 1, evaluated in float64.
    # Raw NaN/inf are permitted here; the endpoint admissibility predicate gates them.
    with np.errstate(divide="ignore", invalid="ignore"):
        # Extract float64 arrays for the numerator and the denominator.
        numerator: np.ndarray = merged["close_adj_lead_5"].to_numpy(dtype=np.float64)
        # The denominator is close_adj[i, t]; zero/negative prices are inadmissible (Step 2).
        denominator: np.ndarray = merged["close_adj"].to_numpy(dtype=np.float64)
        # Vectorised ratio minus one; NaN/inf allowed at this stage.
        raw_y: np.ndarray = numerator / denominator - 1.0
    # Materialise the Eq. (1) label as a float64 Series aligned to merged.
    merged["y"] = pd.Series(raw_y, index=merged.index, dtype="float64")
    # Pin the pair-table column order (REQUIRED_INTERFACE).
    merged = merged.loc[:, ["stock_code", "trade_date", *_PAIR_TABLE_COLUMNS]]
    # Restore the compound (stock_code, trade_date) index; row order preserved.
    merged = merged.set_index(list(_COMPOUND_INDEX_NAMES), drop=True)
    # Return the pair table; the caller's panel was never mutated.
    return merged

# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 2:  apply_endpoint_admissibility_task_6
# -------------------------------------------------------------------------------------------------------------------------------

def apply_endpoint_admissibility_task_6(
    labels: pd.DataFrame,
) -> pd.DataFrame:
    """
    Apply the endpoint admissibility gate to the Eq. (1) labels.

    Build the sub-predicates on both endpoints of the five-row return
    -- finite, strictly positive and ``tradable == True`` at ``t`` and
    at ``t + 5`` -- and conjoin them under Kleene three-valued logic
    into the nullable boolean ``defined``. A successor row exists only
    when ``row_index_lead`` is non-NA; on terminal rows the trailing
    sub-predicates are therefore NA rather than False, and ``defined``
    is NA. ``y`` is then set to NaN wherever ``defined`` is not
    ``True``; nothing is imputed, filled or carried forward, and the
    caller's frame is not modified.

    Parameters
    ----------
    labels : pd.DataFrame
        The pair table produced by ``construct_labels_task_6``,
        carrying ``close_adj``, ``close_adj_lead_5`` (float64),
        ``tradable``, ``tradable_lead_5`` ((nullable) boolean),
        ``row_index_lead`` and ``y`` (float64). Never mutated.

    Returns
    -------
    pd.DataFrame
        A deep copy of ``labels`` plus ``defined`` ((nullable)
        boolean, NA preserved for diagnostics), with ``y`` masked to
        NaN wherever ``defined`` is not ``True``.

    Raises
    ------
    TypeError
        If ``labels`` is not a DataFrame, a pinned column is missing,
        a price column is not float64, a tradability column is not
        (nullable) boolean, or ``y`` is not float64.

    Notes
    -----
    All sub-predicates are evaluated without short-circuit so a
    failing term can be attributed. The
    ``.fillna(False).astype(bool)`` sequence is applied only at the
    point of usage, leaving ``defined`` itself NA-preserving.
    """
    # Name the caller for boundary diagnostics.
    caller = "apply_endpoint_admissibility_task_6"
    # Boundary type check before any column access.
    frame = _require_dataframe(labels, caller, "labels")
    # The pinned Step-2 input columns; absence of any is a halt.
    required: Tuple[str, ...] = (
        "close_adj", "close_adj_lead_5", "tradable", "tradable_lead_5", "y", "row_index_lead",
    )
    # Compute the missing columns (empty means fully satisfied).
    missing: List[str] = [c for c in required if c not in frame.columns]
    # Halt with the full deficit when any pinned column is absent.
    if missing:
        # Raise a typed diagnostic listing every missing pinned column.
        raise TypeError(f"{caller} requires labels to carry columns {required}; missing {missing}.")
    # close_adj and close_adj_lead_5 must be float64 (the pinned schema kind).
    for column in ("close_adj", "close_adj_lead_5"):
        # Any non-float kind is refused; no coercion is applied silently.
        if frame[column].dtype.kind != "f":
            # Raise a typed diagnostic naming the pinned dtype.
            raise TypeError(
                f"{caller} requires labels.{column} to be float64, got {frame[column].dtype}."
            )
    # tradable and tradable_lead_5 must be (nullable) boolean per P22.
    for column in ("tradable", "tradable_lead_5"):
        # Any non-boolean kind is refused; no coercion is applied silently.
        if not pd.api.types.is_bool_dtype(frame[column]):
            # Raise a typed diagnostic naming the pinned dtype.
            raise TypeError(
                f"{caller} requires labels.{column} to be a (nullable) boolean, "
                f"got {frame[column].dtype}."
            )
    # y must be float64 for the NaN masking to round-trip cleanly.
    if frame["y"].dtype.kind != "f":
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(f"{caller} requires labels.y to be float64, got {frame['y'].dtype}.")
    # Aliases to the caller's index: every sub-predicate aligns positionally.
    index: pd.Index = frame.index
    # --- Leading endpoint (t) sub-predicates ---------------------------------
    # finite(t) := notna() AND np.isfinite(close_adj[t]); NaN -> False (never NA).
    close_t: np.ndarray = frame["close_adj"].to_numpy(dtype=np.float64)
    # isfinite is False for both NaN and +/-inf; notna is implied by it on float64.
    finite_t: pd.Series = pd.Series(
        np.isfinite(close_t), index=index, dtype="boolean"
    )
    # positive(t) := close_adj[t] > 0; NaN -> False (never NA); strict > per Eq. (1) gate.
    positive_t: pd.Series = pd.Series(
        close_t > 0.0, index=index, dtype="boolean"
    )
    # tradable(t) == True on the nullable boolean; NA -> NA (Kleene), never coerced here.
    tradable_t: pd.Series = (frame["tradable"] == True).astype("boolean")  # noqa: E712 - strict == True is the pinned semantics
    # --- Trailing endpoint (t+5) sub-predicates ------------------------------
    # A successor exists iff row_index_lead is non-NA (the calendar has a t+5 row).
    successor_exists: pd.Series = frame["row_index_lead"].notna()
    # finite(t+5) := np.isfinite(close_adj_lead_5); NA when NO successor exists, else False on NaN/inf.
    close_lead: np.ndarray = frame["close_adj_lead_5"].to_numpy(dtype=np.float64)
    # Start from the hard predicate, then weaken the absent-successor cells to NA.
    finite_lead: pd.Series = pd.Series(
        np.isfinite(close_lead), index=index, dtype="boolean"
    )
    # The trailing endpoint does not exist on terminal rows: its truth value is unknown (NA).
    finite_lead = finite_lead.where(successor_exists, other=pd.NA)
    # positive(t+5) := close_adj_lead_5 > 0; NA when NO successor exists; strict > per Eq. (1) gate.
    positive_lead: pd.Series = pd.Series(
        close_lead > 0.0, index=index, dtype="boolean"
    )
    # The trailing endpoint does not exist on terminal rows: its truth value is unknown (NA).
    positive_lead = positive_lead.where(successor_exists, other=pd.NA)
    # tradable(t+5) == True on the nullable boolean; NA (no successor / NA flag) -> NA.
    tradable_lead: pd.Series = (frame["tradable_lead_5"] == True).astype("boolean")  # noqa: E712 - strict == True is the pinned semantics
    # --- Kleene conjunction of the eight sub-predicates ----------------------
    # All four pairs evaluated (no short-circuit) so diagnostics can attribute failure.
    # defined := finite(t) & positive(t) & tradable(t) & finite(t+5) & positive(t+5) & tradable(t+5).
    defined: pd.Series = (
        finite_t & positive_t & tradable_t & finite_lead & positive_lead & tradable_lead
    )
    # Guarantee the construction dtype is the nullable boolean, never object.
    defined = defined.astype("boolean")
    # --- Point-of-usage cast (the ONLY place fillna(False) is permitted) ------
    # usage := defined.fillna(False).astype(bool); NA does not satisfy the predicate.
    usage: pd.Series = defined.fillna(False).astype(bool)
    # Copy the frame so the caller's Step-1 output is never mutated.
    out: pd.DataFrame = frame.copy(deep=True)
    # Assign the nullable boolean conjunction; NA retained for diagnostics.
    out["defined"] = defined
    # Eq. (1) under the admissibility predicate: keep the ratio iff defined == True, else NaN.
    out["y"] = out["y"].where(usage, other=np.nan)
    # Return the admissibility-tagged pair table; no imputation was performed.
    return out

# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Step 3:  assert_label_hygiene_task_6
# -------------------------------------------------------------------------------------------------------------------------------

def assert_label_hygiene_task_6(
    labels: pd.DataFrame,
    calendar: pd.DataFrame,
    expected_label_available_dates: int,
) -> Dict[str, Any]:
    """
    Run the Task 6 NaN-hygiene battery and terminal confirmation.

    Executes, without short-circuit, on the Step-2 label frame: a gate
    that ``defined`` has no NA off the terminal rows (terminal NA is
    legitimate); the failed-admissibility test
    ``(defined == False).sum() == y[defined == False].isna().sum()``
    and its converse ``defined == True => y.notna()``; the reproduction
    OOS-null tests on ``oos_sharpe`` / ``oos_computed_at``; the
    terminal-row NaN confirmation with an explicit terminal-check frame
    of cardinality ``min(5, max_row)``; the count of label-available
    dates against ``expected_label_available_dates``; and their
    disjointness from the terminal dates.

    Terminal coordinates come from the calendar as
    ``row_index > max_row - 5``, never hard-coded. The caller's frame is
    mutated in place to carry ``__label_status`` with precedence
    ``"terminal"`` > ``"masked"`` > ``"endpoint_nan"``, and ``"ok"``
    where ``y`` is non-NaN.

    Parameters
    ----------
    labels : pd.DataFrame
        Step-2 output indexed on ``(stock_code, trade_date)`` with ``y``
        (float), ``defined`` (nullable boolean), ``row_index``,
        ``close_adj``, ``close_adj_lead_5``, ``oos_sharpe`` and
        ``oos_computed_at``.
    calendar : pd.DataFrame
        Task 5 trading calendar with a dense ``row_index`` 1..N and
        ``trade_date``.
    expected_label_available_dates : int
        Caller-bound count of calendar dates with at least one non-NaN
        label (480 on the study calendar); an aggregate, not a per-stock
        count.

    Returns
    -------
    Dict[str, Any]
        Keys:

        ``labels`` : pd.DataFrame
            The caller's frame, now carrying ``__label_status``.
        ``n_label_available_dates`` : int
            Number of distinct dates with a live label.
        ``terminal_check`` : pd.DataFrame
            Columns ``row_index``, ``trade_date``, ``n_label_rows``,
            ``n_non_nan_y`` and ``all_y_nan``, one row per terminal row.
        ``label_available_dates`` : list of pd.Timestamp
            The label-available dates in helper order (not re-sorted).
        ``assertions`` : list of str
            Identifiers of the passed sub-checks.
        ``terminal_na_cells`` : int
            Exempt terminal NA cells in ``defined``.
        ``withdrawals`` : list of str
            Contains ``WITHDRAWN_FROM_RAW_TASK_PER_STANDING_REGISTER``.

    Raises
    ------
    TypeError
        If an argument is not a DataFrame, a pinned label column is
        missing, ``defined`` is not boolean, or ``y`` is not float.
    ValueError
        If ``defined`` has a non-terminal NA (coordinates listed), an
        ``oos_sharpe`` or ``oos_computed_at`` cell is non-null, or the
        terminal set cardinality is not ``min(5, max_row)``.
    AssertionError
        If a ``defined == False`` cell has a non-NaN label, a
        ``defined == True`` cell has a NaN label, a terminal cell is
        non-NaN, the label-available count differs from the bound, or a
        terminal date is in the label-available set.

    Notes
    -----
    The withdrawn raw-task test flagged zero-valued ``y`` on
    ``defined == False`` cells; those cells must carry NaN, never
    ``0.0``, so it targeted inadmissible cells, not admissible ones.
    The tag is published in the returned proof.
    """
    # Name the caller for boundary diagnostics.
    caller = "assert_label_hygiene_task_6"
    # Boundary type check on the labels frame.
    frame = _require_dataframe(labels, caller, "labels")
    # Validate the calendar independently of any prior validation (public boundary).
    cal: pd.DataFrame = _validate_calendar(calendar, caller)
    # Normalise the caller-bound date expectation to a plain integer.
    expected: int = _require_integral(
        expected_label_available_dates, caller, "expected_label_available_dates"
    )
    # The pinned battery input columns; absence of any is a halt.
    required: Tuple[str, ...] = (
        "y", "defined", "row_index", "close_adj", "close_adj_lead_5",
        "oos_sharpe", "oos_computed_at",
    )
    # Compute the missing columns (empty means fully satisfied).
    missing: List[str] = [c for c in required if c not in frame.columns]
    # Halt with the full deficit when any pinned column is absent.
    if missing:
        # Raise a typed diagnostic listing every missing pinned column.
        raise TypeError(f"{caller} requires labels to carry columns {required}; missing {missing}.")
    # defined must be (nullable) boolean for the Kleene semantics and the dtype gate.
    if not pd.api.types.is_bool_dtype(frame["defined"]):
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(
            f"{caller} requires labels.defined to be a (nullable) boolean, got {frame['defined'].dtype}."
        )
    # y must be float64 for the NaN tests to round-trip cleanly.
    if frame["y"].dtype.kind != "f":
        # Raise a typed diagnostic naming the pinned dtype.
        raise TypeError(f"{caller} requires labels.y to be float64, got {frame['y'].dtype}.")
    # ---- Terminal coordinates (derived from the calendar, never hard-coded) --
    # The terminal set is derived from the calendar's dense row index.
    max_row: int = int(cal["row_index"].max())
    # Terminal rows: row_index + 5 > max_row ({723..727} on the 727-row study calendar).
    terminal_rows: pd.Series = cal["row_index"] > (max_row - _FORWARD_HORIZON_ROWS)
    # Boolean per label row: True on terminal rows of the calendar.
    label_terminal: pd.Series = (
        frame["row_index"].astype("int64") > (max_row - _FORWARD_HORIZON_ROWS)
    )
    # ---- Sub-check 0 (prerequisite): no NA in defined off terminal rows -----
    # Terminal rows carry NA by pinned semantics (the endpoint does not exist);
    # NA anywhere else makes the failed-admissibility comparison untestable.
    na_off_terminal: pd.Series = frame["defined"].isna() & ~label_terminal
    # Count the legitimate terminal NAs (informational; never a failure).
    terminal_na_cells: int = int((frame["defined"].isna() & label_terminal).sum())
    # Halt and LIST the NA coordinates when any non-terminal cell carries NA.
    if bool(na_off_terminal.any()):
        # Recover the (stock_code, trade_date) coordinates of every NA cell.
        if isinstance(frame.index, pd.MultiIndex):
            # Compound-index layout: pull both levels.
            stock_level = frame.index.get_level_values(0)
            date_level = frame.index.get_level_values(1)
            # Format at most 10 coordinates for the diagnostic.
            na_coordinates: List[str] = [
                f"{stock_level[i]}@{pd.Timestamp(date_level[i]).date()}"
                for i in np.flatnonzero(na_off_terminal.to_numpy())[:10]
            ]
        else:
            # Flat frame: fall back to positional coordinates.
            na_coordinates = [f"row#{i}" for i in np.flatnonzero(na_off_terminal.to_numpy())[:10]]
        # Raise with the coordinate list; the pipeline halts here, by design.
        raise ValueError(
            f"{caller} halted: labels.defined carries {int(na_off_terminal.sum())} non-terminal "
            f"NA cell(s); the failed-admissibility test is untestable. NA coordinates: {na_coordinates}."
        )
    # From here on, defined carries no non-terminal NA; terminal NA may remain (exempt).
    # Strict == False on the nullable boolean, then NA -> False for positional use.
    defined_false: pd.Series = (frame["defined"] == False).fillna(False).astype(bool)  # noqa: E712 - strict == False pinned by the step text
    defined_false = defined_false.astype(bool)
    # Strict == True on the nullable boolean, then NA -> False for positional use.
    defined_true: pd.Series = (frame["defined"] == True).fillna(False).astype(bool)  # noqa: E712 - strict == True pinned by the step text
    defined_true = defined_true.astype(bool)
    # ---- Sub-check (i): failed admissibility — y must be NA, never 0.0 ------
    # The pinned test, verbatim: (defined == False).sum() == y[defined == False].isna().sum().
    failed_admissibility_cells: int = int(defined_false.sum())
    # Count the failed-admissibility cells whose y IS NaN (the pinned right-hand side).
    failed_admissibility_nan: int = int(frame.loc[defined_false.to_numpy(), "y"].isna().sum())
    # Any gap between the two counts is an imputed/carried label — a halt.
    if failed_admissibility_cells != failed_admissibility_nan:
        # Recover the offending coordinates for the diagnostic.
        if isinstance(frame.index, pd.MultiIndex):
            # Compound-index layout: pull both levels.
            stock_level = frame.index.get_level_values(0)
            date_level = frame.index.get_level_values(1)
            # Format at most 10 coordinates for the diagnostic.
            false_coordinates: List[str] = [
                f"{stock_level[i]}@{pd.Timestamp(date_level[i]).date()}"
                for i in np.flatnonzero(
                    defined_false.to_numpy() & frame["y"].notna().to_numpy()
                )[:10]
            ]
        else:
            # Flat frame: fall back to positional coordinates.
            false_coordinates = [
                f"row#{i}"
                for i in np.flatnonzero(
                    defined_false.to_numpy() & frame["y"].notna().to_numpy()
                )[:10]
            ]
        # Raise naming the counts and the coordinate list.
        raise AssertionError(
            f"{caller} failed-admissibility violated: {failed_admissibility_cells} cell(s) with "
            f"defined == False but only {failed_admissibility_nan} carry NaN y (must be NA, never 0.0). "
            f"Coordinates: {false_coordinates}."
        )
    # ---- Sub-check (i-converse): defined == True must carry a live label ----
    # Eq. (1) is finite and positive on admissible endpoints; y cannot be NaN there.
    true_mask: pd.Series = defined_true
    # Count admissible cells whose y is NaN (a construction defect).
    true_nan: int = int(frame.loc[true_mask.to_numpy(), "y"].isna().sum())
    # Any occurrence indicates y was dropped after admissibility passed — a halt.
    if true_nan != 0:
        # Raise a typed diagnostic naming the defect count.
        raise AssertionError(
            f"{caller} integrity converse violated: {true_nan} cell(s) with defined == True carry NaN y."
        )
    # ---- Sub-check (ii): OOS null test — reproduction columns MUST be null ----
    # Any non-null oos_sharpe pre-freeze voids the substrate of the freeze.
    sharpe_bad: pd.Series = frame["oos_sharpe"].notna()
    # Halt with the count when any reproduction OOS Sharpe value is present.
    if bool(sharpe_bad.any()):
        # Count the offending cells for the diagnostic.
        offenders: int = int(sharpe_bad.sum())
        # Raise a typed diagnostic naming the column and the count.
        raise ValueError(
            f"{caller} halted: labels.oos_sharpe carries {offenders} non-null cell(s); "
            f"the reproduction OOS column MUST be null pre-freeze."
        )
    # Any non-null oos_computed_at pre-freeze voids the substrate of the freeze.
    computed_bad: pd.Series = frame["oos_computed_at"].notna()
    # Halt with the count when any reproduction OOS stamp is present.
    if bool(computed_bad.any()):
        # Count the offending cells for the diagnostic.
        offenders = int(computed_bad.sum())
        # Raise a typed diagnostic naming the column and the count.
        raise ValueError(
            f"{caller} halted: labels.oos_computed_at carries {offenders} non-null cell(s); "
            f"the reproduction OOS stamp MUST be NaT pre-freeze."
        )
    # ---- Sub-check (iii) + terminal confirmation: r + 5 > max_row ⇒ NaN label -
    # The terminal coordinates and row booleans were computed before the NA gate.
    # The terminal cardinality pin: exactly min(5, max_row) rows.
    terminal_count: int = int(terminal_rows.sum())
    # Guard the degenerate calendar shorter than the horizon.
    expected_terminal: int = min(_FORWARD_HORIZON_ROWS, max_row)
    # Halt if the terminal set is not exactly min(5, max_row) — a calendar integrity defect.
    if terminal_count != expected_terminal:
        # Raise a typed diagnostic naming both counts.
        raise ValueError(
            f"{caller} halted: terminal set cardinality {terminal_count} != {expected_terminal}."
        )
    # The terminal coordinates from the calendar (row_index, trade_date).
    terminal_frame: pd.DataFrame = (
        cal.loc[terminal_rows, ["row_index", "trade_date"]].reset_index(drop=True)
    )
    # Every terminal label cell must be NaN — the consequence of r + 5 > max_row.
    terminal_non_nan: int = int(frame.loc[label_terminal.to_numpy(), "y"].notna().sum())
    # Any non-NaN terminal label indicates an off-by-one horizon bug; halt.
    if terminal_non_nan != 0:
        # Raise a typed diagnostic naming the defect count (and the pinned consequence).
        raise AssertionError(
            f"{caller} terminal-row confirmation violated: {terminal_non_nan} terminal cell(s) carry "
            f"a non-NaN y; rows r > {max_row - _FORWARD_HORIZON_ROWS} have no t+5."
        )
    # Materialise the explicit terminal-check DataFrame (the step's pinned vehicle).
    terminal_check: pd.DataFrame = terminal_frame.copy()
    # Slice the terminal label rows once (their row_index is needed for grouping).
    terminal_slice: pd.DataFrame = frame.loc[label_terminal.to_numpy()]
    # Count label rows per terminal calendar row (0 where the panel is thin).
    counts: pd.Series = (
        terminal_slice.groupby(terminal_slice["row_index"].astype("int64")).size()
    )
    # Map the per-row counts onto the terminal-check frame (0 where the panel is thin).
    terminal_check["n_label_rows"] = (
        terminal_check["row_index"].astype("int64").map(counts).fillna(0).astype("int64")
    )
    # Slice the terminal rows carrying a non-NaN y (must be empty here).
    terminal_non_nan_slice: pd.DataFrame = terminal_slice.loc[
        terminal_slice["y"].notna().to_numpy()
    ]
    # Count non-NaN y per terminal calendar row (must be identically zero here).
    non_nan_counts: pd.Series = (
        terminal_non_nan_slice.groupby(
            terminal_non_nan_slice["row_index"].astype("int64")
        ).size()
    )
    # Map the per-row non-NaN counts onto the terminal-check frame (0 everywhere by construction).
    terminal_check["n_non_nan_y"] = (
        terminal_check["row_index"].astype("int64").map(non_nan_counts).fillna(0).astype("int64")
    )
    # Boolean per terminal row: True iff every label cell on that row is NaN.
    terminal_check["all_y_nan"] = terminal_check["n_non_nan_y"] == 0
    # ---- Calendar-side count of label-available dates ------------------------
    # Dates at which at least one stock carries a non-NaN label (Eq. (1) live).
    available_dates: pd.Index = _label_available_dates(frame)
    # The aggregate cardinality; the per-stock count is disclosed elsewhere, never here.
    n_available: int = int(available_dates.nunique())
    # Bind the caller-bound figure; NEVER defaulted and NEVER read from a literal.
    if n_available != expected:
        # Halt naming both counts; a mismatch here is a construction or masking defect.
        raise AssertionError(
            f"{caller} label-available dates: expected {expected}, computed {n_available}."
        )
    # Terminal-date disjointness: no terminal date may carry a non-NaN label.
    terminal_dates: pd.Index = pd.Index(cal.loc[terminal_rows, "trade_date"])
    # Set intersection of available dates with terminal dates (must be empty).
    intersection: set = set(available_dates).intersection(set(terminal_dates))
    # Any intersection indicates the horizon is off; halt with the offending dates.
    if intersection:
        # Format the offending dates for the diagnostic.
        disjoint_offenders: List[str] = [str(pd.Timestamp(d).date()) for d in sorted(intersection)]
        # Raise naming the disjointness breach and the dates.
        raise AssertionError(
            f"{caller} terminal-row cardinality violated: {disjoint_offenders} are terminal "
            f"dates yet carry non-NaN labels."
        )
    # ---- Diagnostic column: __label_status in {"ok", "endpoint_nan", "terminal", "masked"}
    # Precedence: terminal wins over masked over endpoint_nan; ok iff y is non-NaN.
    y_na: np.ndarray = frame["y"].isna().to_numpy()
    # A close NaN at either endpoint is the masking signature of Task 5.
    close_na: np.ndarray = (
        frame["close_adj"].isna() | frame["close_adj_lead_5"].isna()
    ).to_numpy()
    # Terminal rows take precedence: r > max_row - 5 cannot carry a label, masked or not.
    terminal_bool: np.ndarray = label_terminal.to_numpy()
    # Materialise the status vector with endpoint_nan as the default failure cause.
    status: np.ndarray = np.full(frame.shape[0], _LABEL_STATUS_ENDPOINT_NAN, dtype=object)
    # ok: a live Eq. (1) label exists.
    status[~y_na] = _LABEL_STATUS_OK
    # masked: y is NaN AND a close endpoint is NA (leading or trailing).
    status[y_na & close_na] = _LABEL_STATUS_MASKED
    # terminal: the row has no successor at all; precedence over every other cause.
    status[terminal_bool] = _LABEL_STATUS_TERMINAL
    # Attach the diagnostic to the caller's frame (in place, by contract).
    frame["__label_status"] = pd.Series(status, index=frame.index, dtype="object")
    # Assemble the battery proof; the caller's labels are returned by reference.
    proof: Dict[str, Any] = {
        "labels": frame,  # the caller's frame, now carrying __label_status
        "n_label_available_dates": n_available,  # Python int; calendar-side grain
        "terminal_check": terminal_check,  # explicit terminal-check DataFrame
        "label_available_dates": [pd.Timestamp(d) for d in available_dates],  # sorted list
        "assertions": [
            "defined_non_terminal_no_na",  # prerequisite gate on Kleene NA propagation
            "failed_admissibility",  # (defined == False) cells carry NaN y
            "defined_true_converse",  # defined == True cells carry live y
            "oos_sharpe_null",  # reproduction OOS Sharpe all-NaN pre-freeze
            "oos_computed_at_null",  # reproduction OOS stamp all-NaT pre-freeze
            "terminal_rows_nan",  # every terminal cell is NaN (r + 5 > max_row)
            "terminal_date_disjoint",  # no terminal date in the available set
            "label_available_dates_expected",  # calendar-side count == bound
        ],
        "terminal_na_cells": terminal_na_cells,  # informational: legitimate terminal NAs
        "withdrawals": [WITHDRAWN_TAG],  # the withdrawn y == 0 ∧ defined == False test
    }
    # Return the proof; the labels frame was mutated in place, as documented.
    return proof

# -------------------------------------------------------------------------------------------------------------------------------
# Task 6, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_6(
    label_panel: pd.DataFrame,
    calendar: pd.DataFrame,
    expected_label_available_dates: int,
) -> Dict[str, Any]:
    """
    Orchestrate Task 6 Steps 1-3 into the LABELS bundle.

    Runs three stages in the pinned order: pair construction at the
    calendar's dense row index (Eq. (1), ``y = close_adj_{t+5} /
    close_adj_t - 1``, joined on ``(stock_code, row_index)``);
    endpoint admissibility, which conjoins the eight sub-predicates
    (with NA retained) and masks ``y`` to NaN wherever ``defined`` is
    not strictly True; and the NaN-hygiene battery, which halts on any
    breach and attaches ``__label_status`` to the label frame in place.
    Each stage is wrapped so that a failure is re-raised as a
    ``RuntimeError`` naming the stage, with the original exception
    chained.

    After the battery, the per-stock non-NaN label counts are disclosed
    (sorted by stock code, never asserted) and the label-available dates
    are sorted ascending.

    Parameters
    ----------
    label_panel : pd.DataFrame
        Task 5 label panel with masked cells kept as NA rows; not
        mutated.
    calendar : pd.DataFrame
        Task 5 trading calendar with a dense ``row_index`` 1..N and
        ``trade_date``.
    expected_label_available_dates : int
        Caller-bound count of label-available calendar dates (480 on the
        study calendar); never defaulted.

    Returns
    -------
    Dict[str, Any]
        Keys:

        ``LABELS`` : pd.DataFrame
            Indexed on ``(stock_code, trade_date)`` with ``y`` (NaN
            where inadmissible or terminal), ``defined`` (nullable
            boolean, NA retained at construction), ``row_index``,
            ``__label_status`` (diagnostic; drop before aggregation) and
            the reproduction OOS sentinels ``oos_sharpe`` /
            ``oos_computed_at``.
        ``label_available_dates`` : list of pd.Timestamp
            Sorted ascending; length equals
            ``expected_label_available_dates``.
        ``n_available_labels`` : dict
            ``stock_code`` (str) to non-NaN label count (int), sorted by
            key.
        ``withdrawals`` : list of str
            Contains ``WITHDRAWN_FROM_RAW_TASK_PER_STANDING_REGISTER``.

    Raises
    ------
    RuntimeError
        If any of the three stages raises; the message names the failing
        stage (``step_1_construct_labels``,
        ``step_2_endpoint_admissibility`` or ``step_3_label_hygiene``)
        and the original exception is set as ``__cause__``.
    """
    # ---- Stage 1: Step 1 — Eq. (1) pair construction at the calendar index ----
    try:
        # Step 1 produces the raw pair table (every input row retained).
        pairs: pd.DataFrame = construct_labels_task_6(label_panel=label_panel, calendar=calendar)
    except Exception as exc:  # noqa: BLE001 - the stage boundary names the failing stage
        # Re-raise with the stage name attached; chain the original cause.
        raise RuntimeError(f"run_task_6 stage 'step_1_construct_labels' failed: {exc}") from exc
    # ---- Stage 2: Step 2 — endpoint admissibility (defined; NA retained) -----
    try:
        # Step 2 conjoins the eight sub-predicates and masks y to NaN where defined != True.
        labels: pd.DataFrame = apply_endpoint_admissibility_task_6(pairs)
    except Exception as exc:  # noqa: BLE001 - the stage boundary names the failing stage
        # Re-raise with the stage name attached; chain the original cause.
        raise RuntimeError(f"run_task_6 stage 'step_2_endpoint_admissibility' failed: {exc}") from exc
    # ---- Stage 3: Step 3 — NaN-hygiene battery and terminal confirmation ----
    try:
        # Step 3 halts on any battery breach and attaches __label_status in place.
        proof: Dict[str, Any] = assert_label_hygiene_task_6(
            labels=labels,  # mutated in place to carry __label_status
            calendar=calendar,
            expected_label_available_dates=expected_label_available_dates,  # caller-bound
        )
    except Exception as exc:  # noqa: BLE001 - the stage boundary names the failing stage
        # Re-raise with the stage name attached; chain the original cause.
        raise RuntimeError(f"run_task_6 stage 'step_3_label_hygiene' failed: {exc}") from exc
    # ---- Disclosure: per-stock non-NaN label counts (never asserted) ---------
    # Group the y series by the stock_code index level; count non-NaN labels.
    counts_series: pd.Series = labels["y"].groupby(level="stock_code").apply(
        lambda block: int(block.notna().sum())
    )
    # Deterministic key order; values are plain Python ints (JSON-safe).
    n_available_labels: Dict[str, int] = {
        str(code): int(count) for code, count in counts_series.sort_index().items()
    }
    # The sorted, plain-Timestamp list of label-available calendar dates.
    label_available_dates: List[pd.Timestamp] = [
        pd.Timestamp(d) for d in sorted(proof["label_available_dates"])
    ]
    # Assemble the publication-grade bundle; key order pinned for readability.
    bundle: Dict[str, Any] = {
        "LABELS": proof["labels"],  # the labels frame with __label_status attached
        "label_available_dates": label_available_dates,  # sorted list of pd.Timestamp
        "n_available_labels": n_available_labels,  # disclosed per-stock counts
        "withdrawals": [WITHDRAWN_TAG],  # the withdrawn raw-task test, published
    }
    # Return the bundle; every stage ran in the pinned order, none was skipped.
    return bundle
```

```python
# Task 7: DSL Parser and Public Evaluator Implementation

# ==============================================================================
# Task 7: DSL Parser and Public Evaluator Implementation
# ==============================================================================

# The pinned terminal set, sourced from FEATURES_PANEL on the compound index.
_TERMINAL_NAMES: Tuple[str, ...] = ("close", "volume", "amount", "tradable")
# The composite index level carrying the stock identity.
_STOCK_LEVEL: str = "stock_code"
# The four characters that, and only those, are dropped at the parse step.
_WHITESPACE_CHARS: str = " \t\n\r"
# The pinned provenance tag for every operator in this registry (P13 note).
_PROVENANCE_TAG_ILLUSTRATIVE: str = "[I]"
# The permitted operator kinds (time_series | arithmetic | conditional).
_OPERATOR_KINDS: Tuple[str, ...] = ("time_series", "arithmetic", "conditional")
# The canonical operator-name grammar (lowercase identifiers).
_OPNAME_PATTERN: "re.Pattern[str]" = re.compile(r"^[a-z_][a-z0-9_]*$")
# The identifier grammar used by the tokenizer (unanchored; matched at position).
_IDENT_PATTERN: "re.Pattern[str]" = re.compile(r"[A-Za-z_][A-Za-z0-9_]*")
# The numeric-literal grammar used by the tokenizer (unanchored; matched at position).
_NUMBER_PATTERN: "re.Pattern[str]" = re.compile(r"[0-9]+(\.[0-9]+)?")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7 shared internals: input validation primitives and the float64 panel substrate
# -------------------------------------------------------------------------------------------------------------------------------

def _validate_op_name_task_7(op: Any, caller: str) -> str:
    """
    Validate and lowercase an operator or terminal name.

    Lowercase ``op`` exactly once, here, then check the result against
    the pinned operator-name grammar ``^[a-z_][a-z0-9_]*$``.

    Parameters
    ----------
    op : Any
        Candidate operator or terminal name; must be a non-empty
        ``str``.
    caller : str
        Name of the calling callable, used in error messages.

    Returns
    -------
    str
        The lowercased, validated name.

    Raises
    ------
    TypeError
        If ``op`` is not a ``str``.
    ValueError
        If the lowercased name is empty or fails the name grammar.
    """
    # The operator name must be a Python str; anything else is a caller defect.
    if not isinstance(op, str):
        # Raise naming the offending type for immediate diagnosis.
        raise TypeError(f"{caller}: operator name must be a str, got {type(op).__name__}")
    # Normalise the name to lowercase exactly once, here and at nowhere else.
    lowered: str = op.lower()
    # An empty operator name is never a valid node op.
    if not lowered:
        # Raise with the explicit offending value for the audit trail.
        raise ValueError(f"{caller}: operator name must be non-empty")
    # The lowered name must satisfy the pinned operator-name grammar.
    if _OPNAME_PATTERN.fullmatch(lowered) is None:
        # Raise with the offending token so the caller can see the violation.
        raise ValueError(f"{caller}: operator name {op!r} does not match {_OPNAME_PATTERN.pattern}")
    # The validated, lowercased name is returned to the caller.
    return lowered

def _float_substrate_task_7(x: pd.Series | pd.DataFrame, caller: str) -> pd.Series | pd.DataFrame:
    """
    Convert a panel to a float64 substrate with NaN for every NA.

    Object-dtype panels are first masked so that any ``pd.NA`` becomes
    ``np.nan``, then cast to ``float64``. The cast returns a fresh
    object, so the caller's panel is never mutated; the index and
    columns are preserved exactly and never reindexed.

    Parameters
    ----------
    x : pd.Series | pd.DataFrame
        Input panel.
    caller : str
        Name of the calling callable, used in error messages.

    Returns
    -------
    pd.Series | pd.DataFrame
        ``float64`` copy of ``x`` with the same index (and columns);
        every NA representation (``np.nan``, ``pd.NA``, ``NaT``) becomes
        ``np.nan``, so the NA-in/NA-out contract holds by construction.

    Raises
    ------
    TypeError
        If ``x`` is neither a Series nor a DataFrame, or a column cannot
        be cast to ``float64``.
    """
    # The panel must be a Series or a DataFrame; nothing else is a panel here.
    if not isinstance(x, (pd.Series, pd.DataFrame)):
        # Raise naming the offending type so the boundary failure is explicit.
        raise TypeError(f"{caller}: expected a pandas Series or DataFrame, got {type(x).__name__}")
    # Object-dtype panels may smuggle pd.NA into cells, which astype(float64) rejects.
    if isinstance(x, pd.Series):
        # A Series needs masking exactly when its dtype is object.
        needs_mask: bool = x.dtype == object
    else:
        # A DataFrame needs masking when any column has object dtype.
        needs_mask = bool((x.dtypes == object).any())
    # Apply the NA-to-NaN normalisation once, only when required.
    if needs_mask:
        # Replace every NA cell by np.nan first; the values are preserved exactly otherwise.
        x = x.mask(x.isna(), np.nan)
    # Convert the (now NA-free-of-pd.NA) panel to float64; boolean and Int64 convert losslessly.
    try:
        # astype returns a fresh object, so the caller's panel is never mutated in place.
        substrate = x.astype("float64")
    except (TypeError, ValueError) as exc:
        # A non-numeric column cannot be promoted to float64; refuse loudly.
        raise TypeError(f"{caller}: panel contains a column that cannot be converted to float64: {exc}") from exc
    # The float64 substrate is returned to the caller.
    return substrate

def _validate_window_task_7(w: Any, caller: str) -> int:
    """
    Validate a trading-calendar-row window parameter.

    Accept Python and NumPy integers, and integral floats such as
    ``5.0`` produced by the numeric tokenizer; reject everything else.

    Parameters
    ----------
    w : Any
        Requested window in trading-calendar rows; must be finite,
        integral and at least 1.
    caller : str
        Name of the calling callable, used in error messages.

    Returns
    -------
    int
        The window as a positive Python integer.

    Raises
    ------
    TypeError
        If ``w`` is a ``bool`` or is not a real numeric (e.g. ``str``,
        ``None``, complex).
    ValueError
        If ``w`` is non-finite, fractional, or less than 1.
    """
    # A boolean window is a category error, not a numeric one; reject explicitly.
    if isinstance(w, bool):
        # Raise naming the offending type (True == 1 would be a silent trap).
        raise TypeError(f"{caller}: window must be a positive integer, got bool {w!r}")
    # Accept native and numpy integral/floating scalars, but nothing else.
    if isinstance(w, (int, np.integer)):
        # Cast the numpy/python integer to a plain int.
        window: int = int(w)
    elif isinstance(w, (float, np.floating)):
        # Validate finiteness and integrality before casting.
        if not math.isfinite(float(w)) or float(w) != math.floor(float(w)):
            # Reject fractional, NaN and infinite windows with the offending value.
            raise ValueError(f"{caller}: window must be a finite integral number, got {w!r}")
        # Cast the integral float to a plain int.
        window = int(w)
    else:
        # Anything else (str, None, complex) is a caller defect.
        raise TypeError(f"{caller}: window must be a positive integer, got {type(w).__name__}")
    # The window must cover at least one trading-calendar row.
    if window < 1:
        # Reject zero/negative windows with the offending value.
        raise ValueError(f"{caller}: window must be >= 1 trading-calendar row, got {window}")
    # The validated positive integer window is returned.
    return window

def _stock_level_task_7(index: pd.Index) -> str | int:
    """
    Resolve the stock level of a panel index.

    Parameters
    ----------
    index : pd.Index
        Panel index, preferably a compound ``(stock_code, trade_date)``
        ``MultiIndex``.

    Returns
    -------
    str | int
        ``"stock_code"`` when that level name is present on a
        ``MultiIndex``, otherwise level ``0``.
    """
    # Prefer the pinned level name when present on a MultiIndex.
    if isinstance(index, pd.MultiIndex) and _STOCK_LEVEL in index.names:
        # The named level is returned so callers are self-documenting.
        return _STOCK_LEVEL
    # Otherwise fall back to the first level, which carries the stock identity by construction.
    return 0

def _empty_like_substrate_task_7(xs: pd.Series | pd.DataFrame) -> pd.Series | pd.DataFrame:
    """
    Allocate an all-NA float64 container matching ``xs``.

    Parameters
    ----------
    xs : pd.Series | pd.DataFrame
        Float64 panel whose structure is mirrored.

    Returns
    -------
    pd.Series | pd.DataFrame
        All-NaN ``float64`` container with exactly ``xs``'s index (and,
        for a DataFrame, its column labels in the same order).
    """
    # A Series container mirrors index and name only.
    if isinstance(xs, pd.Series):
        # The all-NA Series inherits the input's index verbatim.
        return pd.Series(np.nan, index=xs.index, dtype="float64")
    # A DataFrame container mirrors index and column labels.
    return pd.DataFrame(np.nan, index=xs.index, columns=xs.columns, dtype="float64")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7 dataclasses: the AST node, the parse error and the operator spec
# -------------------------------------------------------------------------------------------------------------------------------

@dataclass(frozen=True)
class Node:
    """
    Frozen abstract-syntax-tree node of the reconstructed DSL.

    A leaf is a node whose ``op`` is a terminal name and whose ``args``
    is empty; an operator node holds one operand per argument in fixed
    (non-symmetric) order. Construction validates and normalises every
    field, so a stored node satisfies the declared ``Node | str |
    float`` operand union and two structurally identical ASTs compare
    equal.

    Parameters
    ----------
    op : str
        Operator or terminal name; lowercased and checked against
        ``^[a-z_][a-z0-9_]*$``.
    args : Tuple["Node | str | float", ...], optional
        Operands in argument order. ``int``, ``np.integer`` and
        ``np.floating`` operands are stored as Python ``float``;
        booleans, non-finite floats and other types are rejected.
        Default is ``()``.

    Attributes
    ----------
    op : str
        The validated, lowercased name.
    args : Tuple["Node | str | float", ...]
        The validated operand tuple.

    Raises
    ------
    TypeError
        If ``op`` is not a ``str``, ``args`` is not a tuple, or an
        operand is a boolean or of a disallowed type.
    ValueError
        If ``op`` is empty or fails the name grammar, or an operand is a
        non-finite float.
    """
    # The operator or terminal name; lowercased and validated in __post_init__.
    op: str
    # The operand tuple; coerced and validated in __post_init__.
    args: Tuple["Node | str | float", ...] = ()

    def __post_init__(self) -> None:
        """
        Validate and normalise the node at construction time.

        Lowercase and grammar-check ``op``, require ``args`` to be a
        tuple, and rebuild the operand tuple with integers promoted to
        ``float`` and booleans, non-finite floats and foreign types
        rejected.

        Raises
        ------
        TypeError
            If ``op`` is not a ``str``, ``args`` is not a tuple, or an
            operand is a boolean or of a disallowed type.
        ValueError
            If ``op`` is empty or fails the name grammar, or an operand
            is a non-finite float.
        """
        # Normalise and validate the operator name (lowercased exactly once, here).
        object.__setattr__(self, "op", _validate_op_name_task_7(self.op, "Node"))
        # The args must be a tuple; anything else violates the declared contract.
        if not isinstance(self.args, tuple):
            # Raise naming the offending type for immediate diagnosis.
            raise TypeError(f"Node: args must be a tuple, got {type(self.args).__name__}")
        # A fresh, validated operand list is rebuilt so the stored tuple is immutable by construction.
        validated: List[Any] = []
        # Walk every operand exactly once, preserving order (argument order is non-symmetric here).
        for operand in self.args:
            # Sub-expressions pass through untouched (they carry their own invariants).
            if isinstance(operand, Node):
                # Append the sub-node as-is.
                validated.append(operand)
            # Booleans are rejected explicitly; True == 1 would smuggle a literal.
            elif isinstance(operand, bool):
                # Raise naming the offending operand for the audit trail.
                raise TypeError(f"Node: boolean operands are not permitted in the AST, got {operand!r}")
            # Integers are normalised to float to honour the declared Node | str | float union.
            elif isinstance(operand, int):
                # Store the float spelling so canonicalisation is invariant to integer width.
                validated.append(float(operand))
            # Finite floats are accepted; non-finite floats are rejected outright.
            elif isinstance(operand, float):
                # Reject NaN and infinite literals; the DSL has no such constants.
                if not math.isfinite(operand):
                    raise ValueError(f"Node: non-finite float operand {operand!r} is not permitted")
                # Store the float operand.
                validated.append(operand)
            # Numpy scalars behave as their Python analogues.
            elif isinstance(operand, np.integer):
                # Normalise numpy integers to Python floats for the same reason.
                validated.append(float(operand))
            elif isinstance(operand, np.floating):
                # Reject non-finite numpy floats exactly as their Python counterparts.
                if not math.isfinite(float(operand)):
                    raise ValueError(f"Node: non-finite float operand {float(operand)!r} is not permitted")
                # Store the Python float for JSON stability.
                validated.append(float(operand))
            # String operands are permitted by the declared union (never produced by the parser).
            elif isinstance(operand, str):
                # Store the string operand verbatim; canonicalisation JSON-encodes it.
                validated.append(operand)
            # Anything else (lists, dicts, None) is a caller defect.
            else:
                # Raise naming the offending type; the AST stays closed.
                raise TypeError(f"Node: operand of type {type(operand).__name__} is not permitted")
        # Freeze the validated tuple back onto the instance.
        object.__setattr__(self, "args", tuple(validated))

class DslParseError(Exception):
    """
    Signal that expression text violates the reconstructed grammar.

    The paper's grammar is undisclosed, so this exception reports a
    parse failure under the reconstruction implemented by this module,
    never a finding about the paper. It carries only the message string.
    """

    def __init__(self, message: str) -> None:
        """
        Store the failure message verbatim.

        Parameters
        ----------
        message : str
            Human-readable description of the grammar violation,
            including the offending position whenever one is available.
        """
        # Store the message verbatim for exact propagation to callers and logs.
        super().__init__(message)

@dataclass(frozen=True)
class OperatorSpec:
    """
    Frozen declarative specification of one DSL operator.

    Parameters
    ----------
    arity : int
        Number of operands accepted; must be at least 1 and not a
        ``bool``.
    kind : str
        Operator family, one of ``"time_series"``, ``"arithmetic"``,
        ``"conditional"``.
    implementation : Callable[..., Any]
        Pure function over Series/DataFrame inputs with pinned NA
        semantics.
    provenance_tag : str
        Provenance tag; every operator in the registry uses ``"[I]"``.
    symmetric : bool
        True iff argument order is immaterial; False throughout this
        registry.

    Attributes
    ----------
    arity : int
        The operand count as a plain ``int``.
    kind : str
        The validated operator family.
    implementation : Callable[..., Any]
        The operator implementation.
    provenance_tag : str
        The validated bracketed tag.
    symmetric : bool
        The argument-order flag as a plain ``bool``.

    Raises
    ------
    TypeError
        If ``arity`` is not an integer, ``kind`` is not a non-empty
        string, ``implementation`` is not callable, or ``symmetric`` is
        not a boolean.
    ValueError
        If ``arity`` is less than 1, ``kind`` is not one of the three
        pinned families, or ``provenance_tag`` is not bracketed.
    """
    # The operand count; the time-series operators and div/sub take 2, neg takes 1, if takes 3.
    arity: int
    # The operator family; membership is validated in __post_init__.
    kind: str
    # The pure callable implementing the operator's panel semantics.
    implementation: Callable[..., Any]
    # The provenance tag; "[I]" for the whole illustrative registry.
    provenance_tag: str
    # Whether argument order is immaterial (False throughout this registry).
    symmetric: bool

    def __post_init__(self) -> None:
        """
        Validate and normalise the spec at construction time.

        Normalise ``arity`` to a plain ``int`` and ``symmetric`` to a
        plain ``bool`` so that the registry hash is stable across NumPy
        scalar spellings, then enforce the family and tag constraints.

        Raises
        ------
        TypeError
            If ``arity`` is not an integer, ``kind`` is not a non-empty
            string, ``implementation`` is not callable, or ``symmetric``
            is not a boolean.
        ValueError
            If ``arity`` is less than 1, ``kind`` is not one of the
            pinned families, or ``provenance_tag`` is not a bracketed
            tag such as ``"[I]"``.
        """
        # A boolean arity is an integer subclass; reject it explicitly.
        if isinstance(self.arity, bool) or not isinstance(self.arity, (int, np.integer)):
            # Raise naming the offending type for the audit trail.
            raise TypeError(f"OperatorSpec: arity must be an int, got {type(self.arity).__name__}")
        # Normalise numpy integers to plain ints for hash stability of the registry hash.
        object.__setattr__(self, "arity", int(self.arity))
        # The arity must be at least one operand.
        if self.arity < 1:
            # Reject zero and negative arities with the offending value.
            raise ValueError(f"OperatorSpec: arity must be >= 1, got {self.arity}")
        # The kind must be a non-empty string.
        if not isinstance(self.kind, str) or not self.kind:
            # Raise naming the offending value for the audit trail.
            raise TypeError(f"OperatorSpec: kind must be a non-empty str, got {self.kind!r}")
        # The kind must belong to one of the three pinned families.
        if self.kind not in _OPERATOR_KINDS:
            # Reject unknown families with the offending value.
            raise ValueError(f"OperatorSpec: kind must be one of {_OPERATOR_KINDS}, got {self.kind!r}")
        # The implementation must be callable (a pure function over panels).
        if not callable(self.implementation):
            # Raise naming the offending type for the audit trail.
            raise TypeError(f"OperatorSpec: implementation must be callable, got {type(self.implementation).__name__}")
        # The provenance tag must be a bracketed tag (e.g. "[I]").
        if not isinstance(self.provenance_tag, str) or not (self.provenance_tag.startswith("[") and self.provenance_tag.endswith("]")):
            # Reject unbracketed tags with the offending value.
            raise ValueError(f"OperatorSpec: provenance_tag must look like a bracketed tag like '[I]', got {self.provenance_tag!r}")
        # The symmetric flag must be a genuine boolean (numpy scalars normalised).
        if not isinstance(self.symmetric, (bool, np.bool_)):
            # Raise naming the offending type for the audit trail.
            raise TypeError(f"OperatorSpec: symmetric must be a bool, got {type(self.symmetric).__name__}")
        # Normalise numpy booleans to plain bools for hash stability of the registry hash.
        object.__setattr__(self, "symmetric", bool(self.symmetric))

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 1: Time-series and arithmetic operator implementations (pure functions over panels)
# -------------------------------------------------------------------------------------------------------------------------------

def _grouped_stock_walk_task_7(
    xs: pd.Series | pd.DataFrame,
    allocate,
    compute,
) -> pd.Series | pd.DataFrame:
    """
    Walk a panel stock by stock and assemble a same-shaped output.

    Group by the stock level of the compound index in first-appearance
    order (never sorted), slice each stock's rows in their original
    trading-calendar order, apply ``compute`` to the block and write the
    result positionally into the container built by ``allocate``.

    Parameters
    ----------
    xs : pd.Series | pd.DataFrame
        Float64 panel on the ``(stock_code, trade_date)`` compound
        index.
    allocate : Callable
        Factory taking ``xs`` and returning the all-NA output container
        with the same index (and columns).
    compute : Callable
        Per-stock transform taking one stock's block and returning a
        block of the same shape.

    Returns
    -------
    pd.Series | pd.DataFrame
        The assembled output; index and columns equal the input's
        exactly, because the panel is filled positionally and never
        reindexed.
    """
    # Resolve the stock level of the compound index by name (level 0 fallback).
    level: str | int = _stock_level_task_7(xs.index)
    # Allocate the all-NA output container on the original structure.
    out: pd.Series | pd.DataFrame = allocate(xs)
    # Walk the panel stock by stock in first-appearance order (never sorted).
    for label, _ in xs.groupby(level=level, sort=False):
        # Slice this stock's rows in original (trading-calendar row) order.
        sub: pd.Series | pd.DataFrame = xs.loc[label]
        # Compute the per-stock block and write it positionally into the container.
        out.loc[label] = compute(sub).to_numpy()
    # Return the assembled output; the index (and columns) equal the input's.
    return out

def _ts_mean_impl_task_7(x: pd.Series | pd.DataFrame, w: Any) -> pd.Series | pd.DataFrame:
    """
    Compute a trailing window mean per stock.

    Parameters
    ----------
    x : pd.Series | pd.DataFrame
        Input panel; must be castable to ``float64``.
    w : Any
        Window in trading-calendar rows; finite integral numeric, at
        least 1.

    Returns
    -------
    pd.Series | pd.DataFrame
        ``float64`` panel of trailing means over ``w`` rows within each
        stock. The first ``w - 1`` rows of each stock are NaN, and any
        window containing a missing value yields NaN (natural
        definition, unspecified in the manuscript).

    Raises
    ------
    TypeError
        If ``x`` is not a panel or ``w`` is not a real numeric.
    ValueError
        If ``w`` is non-finite, fractional or less than 1.
    """
    # Convert the input to the float64 substrate (NA -> np.nan).
    xs: pd.Series | pd.DataFrame = _float_substrate_task_7(x, "ts_mean")
    # Validate the window once, canonically.
    window: int = _validate_window_task_7(w, "ts_mean")
    # Compute the per-stock trailing means with the natural definition.
    return _grouped_stock_walk_task_7(
        xs,
        allocate=_empty_like_substrate_task_7,
        compute=lambda sub: sub.rolling(window, min_periods=window).mean(),
    )

def _ts_rank_impl_task_7(x: pd.Series | pd.DataFrame, w: Any) -> pd.Series | pd.DataFrame:
    """
    Compute a trailing average-tie rank fraction per stock.

    For each position the last value of the trailing ``w``-row window is
    ranked inside that window:

        rank = (count(values < last) + 0.5 * count(values == last)) / w

    Parameters
    ----------
    x : pd.Series | pd.DataFrame
        Input panel; must be castable to ``float64``.
    w : Any
        Window in trading-calendar rows; finite integral numeric, at
        least 1.

    Returns
    -------
    pd.Series | pd.DataFrame
        ``float64`` panel of mid-rank fractions in ``(0, 1]``; NaN for
        the first ``w - 1`` rows of each stock and for any window
        containing a missing value.

    Raises
    ------
    TypeError
        If ``x`` is not a panel or ``w`` is not a real numeric.
    ValueError
        If ``w`` is non-finite, fractional or less than 1.

    Notes
    -----
    The manuscript pins no ``ts_rank`` aggregation; this mid-rank choice
    is an illustrative reconstruction tagged ``[I]``.
    """
    # Convert the input to the float64 substrate (NA -> np.nan).
    xs: pd.Series | pd.DataFrame = _float_substrate_task_7(x, "ts_rank")
    # Validate the window once, canonically.
    window: int = _validate_window_task_7(w, "ts_rank")

    def _mid_rank_fraction_task_7(window_values: np.ndarray) -> float:
        """
        Return the average-tie rank fraction of the last window value.

        Parameters
        ----------
        window_values : np.ndarray
            The trailing window, oldest first; the last element is
            ranked.

        Returns
        -------
        float
            ``(less + 0.5 * equal) / len(window_values)`` in ``(0, 1]``,
            or NaN if any value in the window is missing.
        """
        # A window with any missing value is undefined under the natural definition.
        if np.isnan(window_values).any():
            # The window is reported as missing, never as a rank.
            return np.nan
        # The last element is the value being ranked.
        last: float = float(window_values[-1])
        # Number of values strictly below the last element.
        less: int = int(np.sum(window_values < last))
        # Number of values equal to the last element (ties, including itself).
        equal: int = int(np.sum(window_values == last))
        # Average-tie rank fraction: (less + 0.5 * equal) / w, in (0, 1].
        return (less + 0.5 * equal) / float(window_values.size)

    # Compute the per-stock trailing rank fractions with the natural NA rule.
    return _grouped_stock_walk_task_7(
        xs,
        allocate=_empty_like_substrate_task_7,
        compute=lambda sub: sub.rolling(window, min_periods=window).apply(
            _mid_rank_fraction_task_7,
            raw=True,
        ),
    )

def _lag_impl_task_7(x: pd.Series | pd.DataFrame, w: Any) -> pd.Series | pd.DataFrame:
    """
    Lag a panel by ``w`` trading-calendar rows within each stock.

    Parameters
    ----------
    x : pd.Series | pd.DataFrame
        Input panel; must be castable to ``float64``.
    w : Any
        Lag in trading-calendar rows; finite integral numeric, at
        least 1.

    Returns
    -------
    pd.Series | pd.DataFrame
        ``float64`` panel shifted back ``w`` rows per stock; the first
        ``w`` rows of each stock are NaN (natural definition,
        unspecified in the manuscript).

    Raises
    ------
    TypeError
        If ``x`` is not a panel or ``w`` is not a real numeric.
    ValueError
        If ``w`` is non-finite, fractional or less than 1.
    """
    # Convert the input to the float64 substrate (NA -> np.nan).
    xs: pd.Series | pd.DataFrame = _float_substrate_task_7(x, "lag")
    # Validate the window once, canonically.
    window: int = _validate_window_task_7(w, "lag")
    # Compute the per-stock lag with the natural definition (row shift within stock).
    return _grouped_stock_walk_task_7(
        xs,
        allocate=_empty_like_substrate_task_7,
        compute=lambda sub: sub.shift(window),
    )

def _delta_impl_task_7(x: pd.Series | pd.DataFrame, w: Any) -> pd.Series | pd.DataFrame:
    """
    Take the ``w``-row difference of a panel within each stock.

    Implements ``delta(x, w) = x - lag(x, w)`` elementwise, with both
    operands on the same float64 substrate and the same index.

    Parameters
    ----------
    x : pd.Series | pd.DataFrame
        Input panel; must be castable to ``float64``.
    w : Any
        Difference horizon in trading-calendar rows; finite integral
        numeric, at least 1.

    Returns
    -------
    pd.Series | pd.DataFrame
        ``float64`` panel on ``x``'s index (and columns); NaN wherever
        either ``x`` or its ``w``-row lag is missing.

    Raises
    ------
    TypeError
        If ``x`` is not a panel or ``w`` is not a real numeric.
    ValueError
        If ``w`` is non-finite, fractional or less than 1.

    Notes
    -----
    The manuscript names ``delta`` without defining it; the
    lag-difference form is an illustrative reconstruction tagged
    ``[I]``.
    """
    # Convert the input to the float64 substrate (NA -> np.nan).
    xs: pd.Series | pd.DataFrame = _float_substrate_task_7(x, "delta")
    # Validate the lag used by the reconstruction once, canonically.
    window: int = _validate_window_task_7(w, "delta")
    # Compute the lagged panel with the same natural definition.
    lagged: pd.Series | pd.DataFrame = _lag_impl_task_7(xs, window)
    # Return x - lag(x, w) with strict index equality enforced by the substrate walk.
    return xs - lagged

def _require_matching_operands_task_7(
    a: Any,
    b: Any,
    caller: str,
) -> Tuple[pd.Series | pd.DataFrame, pd.Series | pd.DataFrame]:
    """
    Validate two operands are same-typed panels on one alignment.

    Check that both operands are pandas Series/DataFrame objects of the
    same kind, with identical indexes (and identical column labels and
    order for DataFrames), then convert both to the float64 substrate.
    Scalars are rejected: this arithmetic layer is panel-only.

    Parameters
    ----------
    a : Any
        First operand; must be a Series or DataFrame.
    b : Any
        Second operand; must be a Series or DataFrame of the same kind.
    caller : str
        Name of the calling callable, used in error messages.

    Returns
    -------
    Tuple[pd.Series | pd.DataFrame, pd.Series | pd.DataFrame]
        The two operands converted to float64, in input order.

    Raises
    ------
    TypeError
        If either operand is not a Series or DataFrame.
    ValueError
        If the operand kinds differ, the indexes are not equal, or
        DataFrame column labels or order differ.
    """
    # Both operands must be panels; scalars are not arithmetic panels here.
    if not isinstance(a, (pd.Series, pd.DataFrame)) or not isinstance(b, (pd.Series, pd.DataFrame)):
        # Raise naming the offending type for the audit trail.
        raise TypeError(f"{caller}: both operands must be pandas Series/DataFrame panels, got {type(a).__name__} and {type(b).__name__}")
    # The operand kinds must match (Series with Series, DataFrame with DataFrame).
    if isinstance(a, pd.Series) != isinstance(b, pd.Series):
        # Raise naming the type mismatch for the audit trail.
        raise ValueError(f"{caller}: operands must both be Series or both be DataFrames, got {type(a).__name__} and {type(b).__name__}")
    # The indexes must be identical, element for element.
    if not a.index.equals(b.index):
        # Raise with both shapes; silent alignment is forbidden in a panel evaluator.
        raise ValueError(f"{caller}: operand indexes differ ({len(a.index)} vs {len(b.index)} rows)")
    # DataFrame columns must be identical in label and order.
    if isinstance(a, pd.DataFrame) and list(a.columns) != list(b.columns):
        # Raise with both column lists for immediate diagnosis.
        raise ValueError(f"{caller}: operand columns differ: {list(a.columns)} vs {list(b.columns)}")
    # Convert both operands to the float64 substrate.
    return _float_substrate_task_7(a, caller), _float_substrate_task_7(b, caller)

def _div_impl_task_7(a: Any, b: Any) -> pd.Series | pd.DataFrame:
    """
    Compute elementwise division ``a / b`` with NA propagation.

    Parameters
    ----------
    a : Any
        Numerator panel (Series or DataFrame).
    b : Any
        Denominator panel of the same kind and alignment.

    Returns
    -------
    pd.Series | pd.DataFrame
        Float64 quotient; corners stay missing wherever either
        operand is missing. Division by zero follows IEEE-754, so
        ``x / 0.0`` gives an infinite value rather than an error.
    """
    # Validate and convert both operands under one alignment contract.
    lhs: pd.Series | pd.DataFrame
    rhs: pd.Series | pd.DataFrame
    lhs, rhs = _require_matching_operands_task_7(a, b, "div")
    # Compute the elementwise quotient; NA propagates through the substrate.
    return lhs / rhs

def _sub_impl_task_7(a: Any, b: Any) -> pd.Series | pd.DataFrame:
    """
    Compute elementwise subtraction ``a - b`` with NA propagation.

    Parameters
    ----------
    a : Any
        Left panel (Series or DataFrame).
    b : Any
        Right panel of the same kind and alignment.

    Returns
    -------
    pd.Series | pd.DataFrame
        Float64 difference; cells stay missing wherever either
        operand is missing.
    """
    # Validate and convert both operands under one alignment contract.
    lhs: pd.Series | pd.DataFrame
    rhs: pd.Series | pd.DataFrame
    lhs, rhs = _require_matching_operands_task_7(a, b, "sub")
    # Compute the elementwise difference; NA propagates through the substrate.
    return lhs - rhs

def _neg_impl_task_7(a: Any) -> pd.Series | pd.DataFrame:
    """
    Negate a panel elementwise, the sign flip ``e -> -e``.

    Parameters
    ----------
    a : Any
        Panel operand (Series or DataFrame), converted to float64.

    Returns
    -------
    pd.Series | pd.DataFrame
        Elementwise negation; missing cells stay missing. The result
        is a distinct candidate from the input under the identity
        rule: negation is never pushed into operators.
    """
    # Validate and convert the operand to the float64 substrate.
    xs: pd.Series | pd.DataFrame = _float_substrate_task_7(a, "neg")
    # Return the elementwise negation; -NA stays NaN.
    return -xs

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 1 (cont.): The registry builder, its SHA-256 stamp and the terminal registry
# -------------------------------------------------------------------------------------------------------------------------------

def build_operator_registry_task_7() -> Mapping[str, OperatorSpec]:
    """
    Build the frozen registry of the eight DSL operators.

    Returns
    -------
    Mapping[str, OperatorSpec]
        Read-only mapping with exactly eight keys. ``ts_mean``,
        ``ts_rank``, ``delta`` and ``lag`` are ``time_series`` with
        arity 2; ``div`` and ``sub`` are ``arithmetic`` with arity 2;
        ``neg`` is ``arithmetic`` with arity 1; ``if`` is
        ``conditional`` with arity 3 and holds the single shared
        implementation of the elementwise conditional. Every spec
        carries ``provenance_tag == "[I]"`` and ``symmetric is False``.

    Notes
    -----
    The mapping is a ``MappingProxyType``; item assignment raises
    ``TypeError``. Its SHA-256 stamp
    (:func:`registry_sha256_task_7`) is what makes a mid-flight change
    to the operator set detectable downstream.
    """
    # Build the eight pinned specs; every name is tagged [I] (P13.operator_status_note).
    specs: Dict[str, OperatorSpec] = {
        # Time-series family: window parameter in trading-calendar rows.
        "ts_mean": OperatorSpec(arity=2, kind="time_series", implementation=_ts_mean_impl_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
        # ts_mean yields NaN wherever any trailing window is NA or incomplete.
        "ts_rank": OperatorSpec(arity=2, kind="time_series", implementation=_ts_rank_impl_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
        # ts_rank is a disclosed [I] reconstruction (average-tie rolling rank fraction).
        "delta": OperatorSpec(arity=2, kind="time_series", implementation=_delta_impl_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
        # delta(x, w) == x - lag(x, w); disclosed [I] reconstruction.
        "lag": OperatorSpec(arity=2, kind="time_series", implementation=_lag_impl_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
        # lag yields NaN for the first w rows of each stock.
        # Arithmetic family: elementwise panels with strict alignment.
        "div": OperatorSpec(arity=2, kind="arithmetic", implementation=_div_impl_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
        "sub": OperatorSpec(arity=2, kind="arithmetic", implementation=_sub_impl_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
        "neg": OperatorSpec(arity=1, kind="arithmetic", implementation=_neg_impl_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
        # Conditional family: exactly one implementation, shared with the registry.
        "if": OperatorSpec(arity=3, kind="conditional", implementation=elementwise_if_task_7, provenance_tag=_PROVENANCE_TAG_ILLUSTRATIVE, symmetric=False),
    }
    # Wrap the mapping in an immutable proxy; runtime item assignment raises TypeError.
    return MappingProxyType(specs)

def registry_sha256_task_7(registry: Mapping[str, OperatorSpec]) -> str:
    """
    Return the canonical SHA-256 stamp of an operator registry.

    Build a name-keyed projection in which each operator name (as
    ``str``) maps to an object with the four hashed fields ``arity``
    (int), ``kind`` (str), ``provenance_tag`` (str) and ``symmetric``
    (bool); operator names are therefore part of the hashed payload.
    Serialise the whole projection as JSON with ``sort_keys=True``,
    ``separators=(",", ":")``, ``ensure_ascii=False`` and
    ``allow_nan=False``, and hash the UTF-8 encoding.

    Parameters
    ----------
    registry : Mapping[str, OperatorSpec]
        Operator registry; normally the output of
        :func:`build_operator_registry_task_7`.

    Returns
    -------
    str
        64 lowercase hexadecimal characters, with no trailing newline.

    Raises
    ------
    TypeError
        If ``registry`` is not a Mapping, or an entry does not expose
        the four required fields.
    """
    # The registry argument must be a Mapping; anything else is a caller defect.
    if not isinstance(registry, Mapping):
        # Raise naming the offending type for the audit trail.
        raise TypeError(f"registry_sha256_task_7: registry must be a Mapping, got {type(registry).__name__}")
    # Build the plain JSON-able projection with exactly the four hashed fields.
    projection: Dict[str, Dict[str, Any]] = {}
    # Walk every entry once; missing/invalid spec fields raise TypeError with the name.
    for name, spec in registry.items():
        # Read the four hashed fields defensively (attribute access may fail on a foreign spec).
        try:
            # Project only the hashed fields; nothing else enters the digest.
            projection[str(name)] = {
                "arity": int(spec.arity),
                "kind": str(spec.kind),
                "provenance_tag": str(spec.provenance_tag),
                "symmetric": bool(spec.symmetric),
            }
        except (AttributeError, TypeError, ValueError) as exc:
            # Raise naming the defective spec entry for immediate diagnosis.
            raise TypeError(f"registry_sha256_task_7: spec for operator {name!r} does not expose arity/kind/provenance_tag/symmetric: {exc}") from exc
    # Serialise with the pinned canonical-JSON layout (sorted keys, tight separators, UTF-8).
    payload: bytes = json.dumps(
        projection,
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=False,
        allow_nan=False,
    ).encode("utf-8")
    # Return the lowercase hex digest (64 characters), with no trailing newline anywhere.
    return hashlib.sha256(payload).hexdigest()

def _require_features_panel_task_7(
    panel: Any,
    caller: str,
) -> pd.DataFrame | Any:
    """
    Validate the features panel at the module boundary.

    Accept a plain DataFrame or a Task 4 facade that duck-types
    ``columns``, ``index`` and a callable ``__getitem__``, then check
    that the index is a MultiIndex named ``["stock_code",
    "trade_date"]`` and that the four pinned terminal columns exist.
    The object is returned unchanged: never copied, never reindexed.

    Parameters
    ----------
    panel : Any
        DataFrame or facade to validate.
    caller : str
        Name of the calling callable, used in error messages.

    Returns
    -------
    pd.DataFrame | Any
        The same object, after the checks pass.

    Raises
    ------
    TypeError
        If ``panel`` is not a DataFrame or facade, or its index or
        columns cannot be read.
    ValueError
        If the index is not the pinned MultiIndex, or a pinned
        terminal column is missing.
    """
    # A plain DataFrame satisfies the contract directly.
    if isinstance(panel, pd.DataFrame):
        # Validate the compound index and the four pinned columns below.
        frame_like: Any = panel
    # A facade is accepted when it duck-types to columns/index/__getitem__.
    elif (
        hasattr(panel, "columns")
        and hasattr(panel, "index")
        and callable(getattr(panel, "__getitem__", None))
    ):
        # The facade is returned untouched; its embargo stays in force.
        frame_like = panel
    else:
        # Anything else (dict, tuple, None) is a boundary defect.
        raise TypeError(
            f"{caller}: features_panel must be a pandas.DataFrame or a Task 4 FrozenFeaturePanel facade, got {type(panel).__name__}"
        )
    # The panel index must be the pinned compound MultiIndex, in the pinned order.
    try:
        # Read the index names once; a defective facade cannot fake this shape.
        names: List[Any] = list(panel.index.names)
    except (AttributeError, TypeError) as exc:
        # Raise naming the defective object for the audit trail.
        raise TypeError(f"{caller}: features_panel does not expose a well-formed index: {exc}") from exc
    # The compound index must be a MultiIndex with exactly the pinned level names.
    if not isinstance(panel.index, pd.MultiIndex) or names != list(_COMPOUND_INDEX_NAMES):
        # Raise with the observed names; reindexing under a different layout is forbidden.
        raise ValueError(
            f"{caller}: features_panel index must be a MultiIndex with names {list(_COMPOUND_INDEX_NAMES)}, got {names}"
        )
    # The four pinned terminal columns must be present.
    try:
        # Read the column labels once; missing labels surface below.
        columns: List[Any] = list(panel.columns)
    except (AttributeError, TypeError) as exc:
        # Raise naming the defective object for the audit trail.
        raise TypeError(f"{caller}: features_panel does not expose well-formed columns: {exc}") from exc
    # Compute the missing pinned columns; any absence is a boundary halt.
    missing: List[str] = [name for name in _TERMINAL_NAMES if name not in columns]
    # Raise naming every missing pinned terminal.
    if missing:
        raise ValueError(f"{caller}: features_panel is missing the pinned terminal columns {missing}; observed {columns}")
    # The validated panel is returned without any reindexing or copying.
    return frame_like

def build_terminal_registry_task_7(
    features_panel: pd.DataFrame | Any,
) -> Mapping[str, pd.Series]:
    """
    Build the terminal registry from the features panel.

    Parameters
    ----------
    features_panel : pd.DataFrame | Any
        Features panel on the compound (stock_code, trade_date)
        index, or a Task 4 facade.

    Returns
    -------
    Mapping[str, pd.Series]
        Read-only mapping with exactly the keys ``close``, ``volume``,
        ``amount`` and ``tradable``. Each value is the panel's own
        Series, neither copied nor reindexed. ``open``, ``high``,
        ``low`` and every ``*_adj`` name are absent, so resolving them
        fails on lookup.

    Raises
    ------
    TypeError
        If ``features_panel`` is not a DataFrame or facade, or an
        extracted column is not a Series.
    ValueError
        If the compound index or a pinned column is missing.
    """
    # Validate the panel once at the boundary; upstream verification is trusted thereafter.
    panel: Any = _require_features_panel_task_7(features_panel, "build_terminal_registry_task_7")
    # Extract the four pinned terminals exactly as they appear in the panel.
    terminals: Dict[str, pd.Series] = {}
    # Walk the pinned terminal list in its canonical order.
    for name in _TERMINAL_NAMES:
        # Serve the panel's own Series (zero copy, zero reindex).
        series: Any = panel[name]
        # The extracted column must be a Series on the compound index.
        if not isinstance(series, pd.Series):
            # Raise naming the offending column and type for the audit trail.
            raise TypeError(f"build_terminal_registry_task_7: column {name!r} is not a Series, got {type(series).__name__}")
        # Register the pinned terminal under its pinned name.
        terminals[name] = series
    # Wrap the mapping in an immutable proxy; missing keys raise KeyError on lookup.
    return MappingProxyType(terminals)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 2: The elementwise conditional operator (v2) and the v1 defect emulator
# -------------------------------------------------------------------------------------------------------------------------------

def _is_na_scalar_task_7(value: Any) -> bool:
    """
    Test whether a branch argument is a scalar NA marker.

    Parameters
    ----------
    value : Any
        Candidate scalar; ``pd.NA``, ``None`` and ``np.nan`` are NA.

    Returns
    -------
    bool
        True iff ``value`` is not a Series/DataFrame and ``pd.isna``
        returns a plain ``True`` for it. Scalars that defeat ``pd.isna``
        are treated as present.
    """
    # Panels are never scalars; short-circuit them before calling pd.isna.
    if isinstance(value, (pd.Series, pd.DataFrame)):
        # A panel is never a scalar NA marker.
        return False
    # Evaluate pandas' own missingness predicate, defensively.
    try:
        # pd.isna on a scalar returns a plain bool.
        missing: Any = pd.isna(value)
    except (TypeError, ValueError):
        # Exotic scalars that defeat pd.isna are treated as present, not missing.
        return False
    # Only a plain boolean True counts as scalar-NA; arrays would fail the bool() cast.
    return bool(missing) is True

def _broadcast_branch_task_7(
    branch: Any,
    cond: pd.Series | pd.DataFrame,
    which: str,
) -> pd.Series | pd.DataFrame:
    """
    Broadcast one branch of the conditional to the shape of ``cond``.

    A scalar NA marker yields an all-NaN panel; a numeric or boolean
    scalar is broadcast as a constant; a panel must match ``cond``'s
    type, index and columns exactly and is converted to the float64
    substrate without any reindexing.

    Parameters
    ----------
    branch : Any
        Branch value: numeric scalar, scalar NA marker, or
        Series/DataFrame panel.
    cond : pd.Series | pd.DataFrame
        Condition panel whose index (and columns) define the target
        shape.
    which : str
        ``"a"`` or ``"b"``, used in error messages.

    Returns
    -------
    pd.Series | pd.DataFrame
        ``float64`` vector on ``cond``'s index (and columns).

    Raises
    ------
    TypeError
        If ``branch`` is not a numeric scalar, a scalar NA marker or a
        panel; a panel that cannot be cast to ``float64`` raises the
        same.
    ValueError
        If ``branch`` is a panel whose kind, index or columns differ
        from ``cond``'s.
    """
    # A scalar NA branch (pd.NA, np.nan, None) propagates missing everywhere.
    if _is_na_scalar_task_7(branch):
        # Allocate an all-missing float64 vector matching cond exactly.
        if isinstance(cond, pd.Series):
            # An all-NaN Series broadcast to cond's index.
            return pd.Series(np.nan, index=cond.index, dtype="float64")
        # An all-NaN DataFrame broadcast to cond's index and columns.
        return pd.DataFrame(np.nan, index=cond.index, columns=cond.columns, dtype="float64")
    # A pandas panel branch must share cond's type, index, columns and shape.
    if isinstance(branch, (pd.Series, pd.DataFrame)):
        # Series cond requires a Series branch; DataFrame cond requires a DataFrame branch.
        if isinstance(cond, pd.Series) != isinstance(branch, pd.Series):
            # Raise naming the type mismatch; silent reindexing is forbidden.
            raise ValueError(f"elementwise_if: branch {which} must be a {'Series' if isinstance(cond, pd.Series) else 'DataFrame'} to match cond, got {type(branch).__name__}")
        # Two Series branches must carry the identical index.
        if isinstance(cond, pd.Series) and not cond.index.equals(branch.index):
            # Raise with both shapes for immediate diagnosis.
            raise ValueError(f"elementwise_if: branch {which} index does not match cond's index")
        # Two DataFrame branches must carry the identical index and columns.
        if isinstance(cond, pd.DataFrame) and not (cond.index.equals(branch.index) and list(cond.columns) == list(branch.columns)):
            # Raise with both shapes for immediate diagnosis.
            raise ValueError(f"elementwise_if: branch {which} index/columns do not match cond's")
        # Convert the branch to the float64 substrate (NA -> np.nan).
        return _float_substrate_task_7(branch, f"elementwise_if[{which}]")
    # A real number (or boolean) broadcasts as a constant vector.
    if isinstance(branch, (bool, int, float, np.integer, np.floating, np.bool_)):
        # Normalise every numeric scalar to a Python float for a uniform substrate.
        constant: float = float(branch)
        # Broadcast to cond's index or to cond's index and columns.
        if isinstance(cond, pd.Series):
            # A constant Series broadcast to cond's index.
            return pd.Series(constant, index=cond.index, dtype="float64")
        # A constant DataFrame broadcast to cond's index and columns.
        return pd.DataFrame(constant, index=cond.index, columns=cond.columns, dtype="float64")
    # Strings, complex numbers, bytes: outside the numeric branch contract.
    raise TypeError(f"elementwise_if: branch {which} of type {type(branch).__name__} is not permitted (scalar numeric, scalar NA, or panel required)")

def _normalise_condition_task_7(cond: Any) -> pd.Series | pd.DataFrame:
    """
    Normalise the conditional's condition panel to a boolean dtype.

    A panel already on ``bool`` or a nullable ``BooleanDtype`` is
    returned unchanged; any other panel is cast to ``"boolean"``, which
    preserves ``pd.NA`` cells.

    Parameters
    ----------
    cond : Any
        Condition argument; must be a Series or DataFrame, typically on
        ``object`` dtype with booleans and ``pd.NA``.

    Returns
    -------
    pd.Series | pd.DataFrame
        The condition on a boolean dtype, with NA cells preserved.

    Raises
    ------
    TypeError
        If ``cond`` is not a Series or DataFrame.
    ValueError
        If an object panel cannot be interpreted as booleans and NA.
    """
    # The condition must be a panel; scalar conditions are outside v2 semantics.
    if not isinstance(cond, (pd.Series, pd.DataFrame)):
        # Raise naming the offending type; v2 semantics are elementwise by construction.
        raise TypeError(f"elementwise_if: cond must be a pandas Series or DataFrame, got {type(cond).__name__}")
    # A proper boolean panel (nullable boolean or bool) is accepted verbatim.
    if isinstance(cond, pd.Series):
        # A Series is boolean-typed when its dtype is bool or a BooleanDtype.
        is_boolean: bool = (cond.dtype == bool) or isinstance(cond.dtype, pd.BooleanDtype)
    else:
        # A DataFrame is boolean-typed when every column is bool or BooleanDtype.
        is_boolean = all(
            (dtype == bool) or isinstance(dtype, pd.BooleanDtype)
            for dtype in cond.dtypes
        )
    # The all-boolean panel is returned unchanged (pd.NA cells preserved).
    if is_boolean:
        # The boolean panel is returned unchanged.
        return cond
    # An object panel (booleans + pd.NA smuggled as objects) is convertible.
    try:
        # Convert to the nullable boolean dtype; pd.NA cells stay pd.NA.
        normalised: pd.Series | pd.DataFrame = cond.astype("boolean")
    except (TypeError, ValueError, KeyError) as exc:
        # An unconvertible panel is a caller defect; refuse with the reason.
        raise ValueError(f"elementwise_if: cond cannot be interpreted as a boolean/NA panel: {exc}") from exc
    # The normalised boolean panel is returned.
    return normalised

def elementwise_if_task_7(
    cond: pd.Series | pd.DataFrame,
    a: Any,
    b: Any,
) -> pd.Series | pd.DataFrame:
    """
    Evaluate the elementwise conditional ``if(c, a, b)`` (v2 semantics).

    Normalise ``cond`` to a nullable boolean panel, broadcast both
    branches to its shape, then fill a ``float64`` output in three
    passes: the output starts all-missing; the ``a`` branch is written
    where ``cond`` is True; the ``b`` branch (the complement) is written
    where ``cond`` is False. Cells whose condition is NA satisfy neither
    mask and keep the missing value. No panel-as-Boolean coercion
    occurs.

    Parameters
    ----------
    cond : pd.Series | pd.DataFrame
        Elementwise boolean condition; True selects ``a``, False selects
        ``b``, NA yields missing.
    a : Any
        True-branch value: a numeric scalar (broadcast to ``cond``'s
        shape), a scalar NA marker, or a panel sharing ``cond``'s
        index/columns.
    b : Any
        False-branch value; same contract as ``a``.

    Returns
    -------
    pd.Series | pd.DataFrame
        ``float64`` panel on ``cond``'s index (and columns); a cell is
        missing when its condition is NA or the selected branch is
        missing.

    Raises
    ------
    TypeError
        If ``cond`` is not a panel, or a branch is of a disallowed type.
    ValueError
        If ``cond`` cannot be read as a boolean/NA panel, or a panel
        branch does not share ``cond``'s index, columns and kind.

    Notes
    -----
    Invariant pinned by the step: every expression the v1 evaluator
    successfully evaluated yields the identical value here; where v1
    raised, this function returns the elementwise-correct value.
    """
    # Normalise the condition once; everything downstream works on a boolean panel.
    c: pd.Series | pd.DataFrame = _normalise_condition_task_7(cond)
    # Broadcast the true-branch to c's shape (or validate its sharing of shape).
    a_vec: pd.Series | pd.DataFrame = _broadcast_branch_task_7(a, c, "a")
    # Broadcast the false-branch to c's shape (or validate its sharing of shape).
    b_vec: pd.Series | pd.DataFrame = _broadcast_branch_task_7(b, c, "b")
    # Build the two boolean masks following the pinned three-pass sequence:
    # the true-branch mask is c.fillna(False) (True exactly where c == True),
    # and the false-branch mask is the complement (c == False).fillna(False);
    # NA condition cells satisfy neither and keep the initialised missing value.
    cond_true: pd.Series | pd.DataFrame = c.fillna(False)
    cond_false: pd.Series | pd.DataFrame = (c == False).fillna(False)
    # Pass (i): initialise the output to missing in the target (float64) dtype.
    if isinstance(c, pd.Series):
        # A Series result mirrors cond's index.
        out: pd.Series | pd.DataFrame = pd.Series(np.nan, index=c.index, dtype="float64")
    else:
        # A DataFrame result mirrors cond's index and columns.
        out = pd.DataFrame(np.nan, index=c.index, columns=c.columns, dtype="float64")
    # Pass (ii): assign the a-branch wherever the condition is True.
    out = a_vec.where(cond_true, other=out)
    # Pass (iii): assign the b-branch wherever the condition is False (the complement).
    out = b_vec.where(cond_false, other=out)
    # Return the elementwise result with cond's index (and columns) intact.
    return out

def v1_emulator_task_7(cond: Any, a: Any, b: Any) -> Any:
    """
    Emulate the v1 conditional defect ``a if bool(cond) else b``.

    A Series or DataFrame condition with more than one cell raises the
    ambiguous-truth-value ``ValueError`` that characterises the v1
    defect; a one-cell panel is coerced through its single cell's truth
    value, and a scalar condition is coerced with ``bool``.

    Parameters
    ----------
    cond : Any
        Condition value: scalar, one-cell panel, or multi-cell panel.
    a : Any
        True branch, returned verbatim when the coercion yields True.
    b : Any
        False branch, returned verbatim when the coercion yields False.

    Returns
    -------
    Any
        ``a`` or ``b`` when the coercion succeeds.

    Raises
    ------
    ValueError
        If ``cond`` is a Series or DataFrame with more than one cell;
        the message states that its truth value is ambiguous.

    Notes
    -----
    This reconstructs the v1 defect class mechanically so a v1 output
    table can be produced in dry-run mode; the reproduction itself
    evaluates every record with :func:`elementwise_if_task_7`.
    """
    # A panel with more than one cell is exactly the v1 defect class.
    if isinstance(cond, (pd.Series, pd.DataFrame)):
        # The defect class raises the pinned message deterministically.
        if cond.size > 1:
            # Raise the pinned defect message class directly, independent of pandas' wording.
            raise ValueError(
                f"The truth value of a {type(cond).__name__} is ambiguous. "
                "Use a.empty, a.bool(), a.item(), a.any() or a.all()."
            )
        # A one-cell panel selects the branch by its single cell's truth value.
        return a if bool(cond.item()) else b
    # A scalar condition coerces exactly as the naive v1 code did.
    return a if bool(cond) else b

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 3: Canonical AST identity — serialisation, hashing, tokenizer and parser
# -------------------------------------------------------------------------------------------------------------------------------

def _node_to_plain_task_7(node: Node) -> Dict[str, Any]:
    """
    Project an AST node onto a plain JSON-serialisable tree.

    Parameters
    ----------
    node : Node
        Root of the tree to project.

    Returns
    -------
    Dict[str, Any]
        ``{"op": str, "args": list}`` where each element of ``args``
        is a nested projection or a scalar (``bool``, ``int``,
        ``float`` or ``str``). A terminal gives
        ``{"op": name, "args": []}``. Argument order is preserved.

    Raises
    ------
    TypeError
        If an argument is neither a ``Node`` nor a permitted scalar.
    """
    # The operator name is emitted verbatim (already lowercased at construction).
    plain: Dict[str, Any] = {"op": node.op, "args": []}
    # Project every argument exactly once, preserving argument order.
    projected: List[Any] = []
    # Walk the operand tuple in order; the order is part of the identity.
    for arg in node.args:
        # A sub-expression projects recursively.
        if isinstance(arg, Node):
            # Project the nested node with the same canonicalisation rules.
            projected.append(_node_to_plain_task_7(arg))
        # JSON-scalars pass through (floats/ints are serialised as JSON numbers).
        elif isinstance(arg, (bool, int, float, str)):
            # Append the JSON-scalar verbatim (the DSL never stores bools in v1 AST semantics).
            projected.append(arg)
        # Anything else is a defect; the AST is closed under Node | str | float.
        else:
            # Raise naming the offending type for the audit trail.
            raise TypeError(f"canonical_bytes: argument of type {type(arg).__name__} cannot be serialised")
    # Attach the projected operand list and return the plain tree.
    plain["args"] = projected
    return plain

def canonical_bytes_task_7(node: Node) -> bytes:
    """
    Serialise an AST to its canonical UTF-8 byte string.

    Render the projected tree as JSON with ``sort_keys=True``,
    ``separators=(",", ":")``, ``ensure_ascii=False`` and
    ``allow_nan=False``. No trailing newline is appended, and the
    result depends only on the tree, never on source whitespace.

    Parameters
    ----------
    node : Node
        AST to serialise.

    Returns
    -------
    bytes
        Canonical UTF-8 encoding of the AST.

    Raises
    ------
    TypeError
        If ``node`` is not a ``Node`` or holds an unserialisable
        argument.
    ValueError
        If a numeric operand is non-finite.
    """
    # The argument must be an AST node; anything else is a caller defect.
    if not isinstance(node, Node):
        # Raise naming the offending type for the audit trail.
        raise TypeError(f"canonical_bytes_task_7: node must be a Node, got {type(node).__name__}")
    # Serialise the projected tree with the pinned canonical-JSON layout.
    payload: str = json.dumps(
        _node_to_plain_task_7(node),
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=False,
        allow_nan=False,
    )
    # Encode to UTF-8; no trailing newline is ever appended.
    return payload.encode("utf-8")

def expression_id_task_7(node: Node) -> str:
    """
    Compute the 32-character canonical identity of an AST.

    Parameters
    ----------
    node : Node
        AST to identify.

    Returns
    -------
    str
        The first 32 lowercase hexadecimal characters of the SHA-256
        digest of :func:`canonical_bytes_task_7`. The truncation keeps
        identities short; the full 64-character digest is the
        alternative spelling.
    """
    # Hash the canonical bytes with SHA-256; the [R] truncation keeps 32 hex chars.
    return hashlib.sha256(canonical_bytes_task_7(node)).hexdigest()[:32]

@dataclass(frozen=True)
class _Token:
    """
    One lexical token of the reconstructed DSL grammar.

    Parameters
    ----------
    kind : str
        One of ``"NUMBER"``, ``"IDENT"``, ``"LPAREN"``,
        ``"RPAREN"``, ``"COMMA"``, ``"MINUS"`` or ``"EOF"``.
    text : str
        Raw source text of the token; empty for ``EOF``.
    pos : int
        Zero-based character offset where the token starts.

    Attributes
    ----------
    kind : str
        Token family.
    text : str
        Raw source text of the token.
    pos : int
        Character offset of the token's first character.
    """

    # The token family.
    kind: str
    # The raw source text of the token.
    text: str
    # The 0-based offset of the token's first character.
    pos: int

def _tokenise_task_7(text: str) -> List[_Token]:
    """
    Split expression text into tokens, dropping pinned whitespace.

    Whitespace is dropped only here, and only for space, tab, newline
    and carriage return. Numbers and identifiers are scanned with the
    module's compiled patterns; punctuation and the unary minus each
    produce one single-character token.

    Parameters
    ----------
    text : str
        Source expression.

    Returns
    -------
    List[_Token]
        Tokens in source order, terminated by exactly one ``EOF``.

    Raises
    ------
    DslParseError
        On any character outside the grammar (including other
        whitespace), or a malformed numeric literal.
    """
    # The token list is rebuilt linearly; position tracks the scan offset.
    tokens: List[_Token] = []
    # The scan position starts at the first character.
    pos: int = 0
    # The source length is cached for the loop bound.
    length: int = len(text)
    # Walk the source exactly once.
    while pos < length:
        # The current character decides the token shape.
        char: str = text[pos]
        # The four pinned whitespace characters are skipped here and nowhere else.
        if char in _WHITESPACE_CHARS:
            # Advance past one whitespace character.
            pos += 1
            continue
        # A '(' opens a call's argument list.
        if char == "(":
            # Emit the LPAREN token and advance.
            tokens.append(_Token("LPAREN", "(", pos))
            pos += 1
            continue
        # A ')' closes a call's argument list.
        if char == ")":
            # Emit the RPAREN token and advance.
            tokens.append(_Token("RPAREN", ")", pos))
            pos += 1
            continue
        # A ',' separates arguments.
        if char == ",":
            # Emit the COMMA token and advance.
            tokens.append(_Token("COMMA", ",", pos))
            pos += 1
            continue
        # A leading '-' is the unary desugar marker for neg(...).
        if char == "-":
            # Emit the MINUS token and advance.
            tokens.append(_Token("MINUS", "-", pos))
            pos += 1
            continue
        # A digit starts a numeric literal (integer or simple decimal).
        if char.isdigit():
            # Anchor the numeric-literal match at the current position.
            match = _NUMBER_PATTERN.match(text, pos)
            # The regex must match, or the literal is malformed.
            if match is None:
                # Raise naming the position and the malformed literal.
                raise DslParseError(f"malformed numeric literal at position {pos} in {text!r}")
            # Emit the NUMBER token with its raw text.
            tokens.append(_Token("NUMBER", match.group(0), pos))
            # Advance past the literal.
            pos = match.end()
            continue
        # An ASCII letter or underscore starts an identifier.
        if char.isascii() and (char.isalpha() or char == "_"):
            # Anchor the identifier match at the current position.
            match = _IDENT_PATTERN.match(text, pos)
            # The regex must match, or the identifier is malformed.
            if match is None:
                # Raise naming the position and the malformed identifier.
                raise DslParseError(f"malformed identifier at position {pos} in {text!r}")
            # Emit the IDENT token with its raw text (lowercased later, at the AST).
            tokens.append(_Token("IDENT", match.group(0), pos))
            # Advance past the identifier.
            pos = match.end()
            continue
        # Anything else is outside the reconstructed grammar.
        raise DslParseError(f"unrecognised character {char!r} at position {pos} in {text!r}")
    # Terminate the token list with the EOF sentinel.
    tokens.append(_Token("EOF", "", pos))
    # Return the complete token list.
    return tokens

def _arg_sort_key_task_7(arg: Node | str | float) -> Tuple[int, bytes]:
    """
    Return a total-order key over the three operand types.

    Parameters
    ----------
    arg : Node | str | float
        Operand to key.

    Returns
    -------
    Tuple[int, bytes]
        ``(rank, bytes)`` with rank 0 for a ``Node`` (keyed by its
        canonical AST bytes), rank 1 for a ``str`` and rank 2 for a
        number (both keyed by their JSON encoding). The rank makes the
        order total across operand types.
    """
    # Sub-expressions sort by their canonical bytes.
    if isinstance(arg, Node):
        # Rank 0 for sub-expressions.
        return (0, canonical_bytes_task_7(arg))
    # String operands sort after sub-expressions, by JSON encoding.
    if isinstance(arg, str):
        # Rank 1 for string operands.
        return (1, json.dumps(arg, ensure_ascii=False).encode("utf-8"))
    # Numeric operands sort last, by JSON encoding.
    return (2, json.dumps(arg, ensure_ascii=False).encode("utf-8"))

def _sort_symmetric_args_task_7(
    registry: Mapping[str, OperatorSpec],
    op: str,
    args: Tuple[Node | str | float, ...],
) -> Tuple[Node | str | float, ...]:
    """
    Sort operands of registry-declared symmetric operators.

    Parameters
    ----------
    registry : Mapping[str, OperatorSpec]
        Registry consulted for the ``symmetric`` tag.
    op : str
        Lowercased operator name.
    args : Tuple[Node | str | float, ...]
        Operands in source order.

    Returns
    -------
    Tuple[Node | str | float, ...]
        Operands in canonical order when the operator is symmetric;
        the source order otherwise, which also covers unknown
        operators and non-Mapping registries. No operator in the
        six-reference class is symmetric, so the transform is the
        identity today.
    """
    # Look up the operator; unknown operators are treated as non-symmetric.
    spec: Optional[OperatorSpec] = registry.get(op) if isinstance(registry, Mapping) else None
    # No spec, or a non-symmetric spec: the argument order is preserved.
    if spec is None or not spec.symmetric:
        # Return the source order unchanged.
        return args
    # Sort by the total order defined on the three operand types.
    return tuple(sorted(args, key=_arg_sort_key_task_7))

def _collect_operator_names_task_7(node: Node) -> set[str]:
    """
    Collect the distinct non-leaf operator names of an AST.

    Parameters
    ----------
    node : Node
        AST root.

    Returns
    -------
    set[str]
        Distinct operator names, depth-first over the tree. Terminal
        leaves contribute nothing; number arguments are not nodes.
    """
    # An empty accumulator is seeded with the root's operator when the root is not a leaf.
    found: set[str] = set()
    # The worklist holds the nodes still to visit.
    worklist: List[Node] = [node]
    # Walk the AST once, depth-first.
    while worklist:
        # Pop one node from the worklist.
        current: Node = worklist.pop()
        # A leaf node is a terminal; its name is not an operator.
        if not current.args:
            # Skip the terminal leaf entirely.
            continue
        # Record the non-leaf operator's name.
        found.add(current.op)
        # Enqueue every sub-expression operand.
        worklist.extend(operand for operand in current.args if isinstance(operand, Node))
    # Return the accumulated operator set.
    return found

class _RecursiveDescentParser:
    """
    Recursive-descent parser for the reconstructed DSL grammar.

    The grammar is a disclosed reconstruction::

        expression := unary
        unary      := MINUS unary | primary
        primary    := NUMBER | IDENT
                    | IDENT LPAREN arguments RPAREN
        arguments  := unary ( COMMA unary )*

    A leading ``-`` desugars to ``neg(...)``, identifiers are
    lowercased at node construction, and a top-level numeric literal
    is rejected. Parentheses exist only as call delimiters.

    Attributes
    ----------
    _tokens : list[_Token]
        Token stream, terminated by ``EOF``.
    _pos : int
        Cursor into ``_tokens``.
    """

    def __init__(self, text: str) -> None:
        """
        Tokenise ``text`` and place the cursor at the first token.

        Parameters
        ----------
        text : str
            Expression source.

        Raises
        ------
        DslParseError
            If the source contains a character outside the grammar.
        """
        # Scan the source once; malformed characters raise DslParseError.
        self._tokens: List[_Token] = _tokenise_task_7(text)
        # The cursor starts at the first token.
        self._pos: int = 0

    def peek(self) -> _Token:
        """
        Return the current token without consuming it.

        Returns
        -------
        _Token
            The token at the cursor; ``EOF`` at end of input.
        """
        # The cursor is always in range because the token list ends with EOF.
        return self._tokens[self._pos]

    def consume(self, kind: str) -> _Token:
        """
        Consume and return the current token of the expected kind.

        Parameters
        ----------
        kind : str
            Expected token kind.

        Returns
        -------
        _Token
            The consumed token; the cursor advances past it.

        Raises
        ------
        DslParseError
            If the current token has a different kind.
        """
        # Read the current token once.
        token: _Token = self.peek()
        # The kind must match; otherwise the grammar is violated.
        if token.kind != kind:
            # Raise naming the position, the expected kind and the observed token.
            raise DslParseError(
                f"expected {kind} at position {token.pos} in the expression, got {token.kind} {token.text!r}"
            )
        # Advance the cursor past the consumed token.
        self._pos += 1
        # Return the consumed token.
        return token

    def parse(self, registry: Optional[Mapping[str, OperatorSpec]] = None) -> Node:
        """
        Parse a full expression, then require end of input.

        Parameters
        ----------
        registry : Optional[Mapping[str, OperatorSpec]]
            Registry consulted for the symmetric tag. Default is
            ``None``, meaning
            :func:`build_operator_registry_task_7`.

        Returns
        -------
        Node
            Root of the canonical AST.

        Raises
        ------
        DslParseError
            On a grammar violation, trailing input after the
            expression, or a top-level numeric literal.
        """
        # Resolve the registry used for symmetric sorting (default: the pinned registry).
        active: Mapping[str, OperatorSpec] = registry if registry is not None else build_operator_registry_task_7()
        # Parse a unary expression (the only valid top-level form is unary; see grammar).
        result: Node | float = self._parse_unary(active)
        # The input must be exhausted after the expression.
        if self.peek().kind != "EOF":
            # Raise naming the trailing position and token.
            raise DslParseError(
                f"trailing input at position {self.peek().pos}: expected end of expression, got {self.peek().kind} {self.peek().text!r}"
            )
        # A top-level numeric literal is not a valid DSL expression.
        if not isinstance(result, Node):
            # Raise with the explicit reconstruction note.
            raise DslParseError("a top-level numeric literal is not a valid DSL expression (no literal roots in this class)")
        # The validated root is returned.
        return result

    def _parse_unary(self, registry: Mapping[str, OperatorSpec]) -> Node | float:
        """
        Parse a unary expression, desugaring ``-`` to ``neg``.

        Parameters
        ----------
        registry : Mapping[str, OperatorSpec]
            Registry passed down for symmetric argument sorting.

        Returns
        -------
        Node | float
            A ``Node`` for calls and terminals; a float for a bare
            numeric primary. Stacked unary minuses nest the operand in
            successive ``neg`` nodes.
        """
        # A leading '-' desugars to neg(operand), with disclosure.
        if self.peek().kind == "MINUS":
            # Consume the '-' token.
            self.consume("MINUS")
            # Parse the operand recursively (allowing stacked unary minus).
            operand: Node | float = self._parse_unary(registry)
            # Wrap the operand in a neg node (the pinned syntactic sugar).
            return Node("neg", (operand,))
        # Otherwise parse a primary.
        return self._parse_primary(registry)

    def _parse_primary(self, registry: Mapping[str, OperatorSpec]) -> Node | float:
        """
        Parse a NUMBER, an identifier leaf or a call expression.

        Parameters
        ----------
        registry : Mapping[str, OperatorSpec]
            Registry used to sort symmetric arguments when a call node
            is constructed.

        Returns
        -------
        Node | float
            ``Node`` with a lowercased operator name for an identifier
            or call; a float for a numeric literal.

        Raises
        ------
        DslParseError
            On an empty argument list, or on any token (``(``, ``)``,
            ``,``, ``EOF``) where an expression is required.
        """
        # Read the current token once.
        token: _Token = self.peek()
        # A NUMBER literal is stored as a float (canonical JSON spelling is pinned to float).
        if token.kind == "NUMBER":
            # Consume the numeric token.
            self.consume("NUMBER")
            # Return the float spelling of the literal.
            return float(token.text)
        # An IDENT is either a terminal leaf or a call's operator name.
        if token.kind == "IDENT":
            # Consume the identifier token.
            self.consume("IDENT")
            # Lowercase the identifier once, here — never elsewhere.
            name: str = token.text.lower()
            # A '(' directly after the name opens the call's argument list.
            if self.peek().kind == "LPAREN":
                # Consume the '(' token.
                self.consume("LPAREN")
                # An empty argument list is not permitted by the grammar.
                if self.peek().kind == "RPAREN":
                    # Raise naming the position of the offending ')'.
                    raise DslParseError(f"empty argument list for operator {name!r} at position {self.peek().pos}")
                # Parse the first argument.
                args: List[Node | str | float] = [self._parse_unary(registry)]
                # Parse any remaining comma-separated arguments.
                while self.peek().kind == "COMMA":
                    # Consume the ',' token.
                    self.consume("COMMA")
                    # Parse the next argument.
                    args.append(self._parse_unary(registry))
                # The call must be closed by ')'.
                self.consume("RPAREN")
                # Freeze the argument tuple, sorting only for operators declared symmetric.
                operands: Tuple[Node | str | float, ...] = _sort_symmetric_args_task_7(registry, name, tuple(args))
                # Return the canonical call node.
                return Node(name, operands)
            # A bare identifier is a terminal leaf.
            return Node(name, ())
        # Anything else ('(' alone, EOF, ')', ',', '-') is a grammar violation.
        raise DslParseError(f"expected an expression (NUMBER, IDENT or '-') at position {token.pos}, got {token.kind} {token.text!r}")

def parse_expression_task_7(text: str) -> Node:
    """
    Parse expression text into the canonical AST.

    Scan the text (dropping space, tab, newline and carriage return),
    parse the grammar, lowercase every operator name and sort the
    arguments of symmetric operators. A leading ``-`` desugars to
    ``neg(...)``, so ``-close`` and ``neg(close)`` share an identity.

    Parameters
    ----------
    text : str
        Expression source.

    Returns
    -------
    Node
        Canonical AST root; terminals are leaves with ``args == ()``.

    Raises
    ------
    DslParseError
        If ``text`` is not a ``str``, is empty or all whitespace,
        exceeds the interpreter recursion limit, or violates the
        grammar (unbalanced parentheses, empty argument list, trailing
        input, top-level numeric literal, unrecognised character).
    """
    # The expression must be a Python str; anything else is a caller defect.
    if not isinstance(text, str):
        # Raise naming the offending type for the audit trail.
        raise DslParseError(f"parse_expression_task_7: text must be a str, got {type(text).__name__}")
    # All-whitespace text parses to nothing; the grammar rejects empty expressions.
    if all(char in _WHITESPACE_CHARS for char in text):
        # Raise with the explicit note; empty expressions have no AST root.
        raise DslParseError("parse_expression_task_7: expression text is empty or all whitespace")
    # Parse under the pinned reconstruction, converting recursion overflow into a grammar failure.
    try:
        # Build the recursive-descent parser and run it to completion.
        node: Node = _RecursiveDescentParser(text).parse()
    except RecursionError as exc:
        # Convert the interpreter's recursion limit into a grammar failure.
        raise DslParseError(f"parse_expression_task_7: expression nesting too deep in {text!r}") from exc
    # Return the canonical AST root.
    return node

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Step 3 (cont.): The evaluator table builder
# -------------------------------------------------------------------------------------------------------------------------------

def _is_missing_text_task_7(value: Any) -> bool:
    """
    Report whether an expression_text cell carries usable text.

    Parameters
    ----------
    value : Any
        Cell value, such as ``pd.NA``, ``None``, ``numpy.nan`` or a
        string.

    Returns
    -------
    bool
        True for any non-``str`` value and for a ``str`` that
        ``pandas.isna`` reports missing; False otherwise.
    """
    # Non-string cells (including NA sentinels) carry no text to parse.
    if not isinstance(value, str):
        # Anything that is not a str is treated as missing for the table.
        return True
    # A string is only as available as its missingness predicate allows.
    return bool(pd.isna(value))

def parse_records_task_7(
    rows: pd.DataFrame,
    registry: Optional[Mapping[str, OperatorSpec]] = None,
) -> pd.DataFrame:
    """
    Build the evaluator table with one row per preserved input row.

    Validate the input table, resolve the operator registry (defaulting
    to :func:`build_operator_registry_task_7`), compute its SHA-256
    stamp once and apply it to every row, then walk
    ``rows["expression_text"]`` in order. A cell for which
    ``_is_missing_text_task_7`` returns ``True`` leaves
    ``expression_id``, ``operators_used`` and ``parse_status`` as
    ``pd.NA`` while ``expression_text`` is copied through unchanged and
    ``registry_sha256`` is still stamped. Grammar-invalid text yields
    ``"parse_failed"`` with a ``pd.NA`` identity and a ``pd.NA``
    operator set; a parseable expression yields the canonical identity,
    the sorted tuple of distinct non-leaf operator names (leaves
    excluded, unknown names still listed) and a status of ``"parsed"``
    iff that name set is a subset of the registry keys. Hence
    ``"parse_failed"`` covers both unparseable text and a computable AST
    using an operator outside the registry, whose identity is still
    emitted so the record can be accounted for without inventing a
    grammar finding.

    Parameters
    ----------
    rows : pd.DataFrame
        Preserved input rows; must carry an ``expression_text`` column.
        Row order and index are preserved in the output.
    registry : Optional[Mapping[str, OperatorSpec]]
        Operator registry supplying the subset rule and the stamp.
        Default is ``None``, which builds the standard registry.

    Returns
    -------
    pd.DataFrame
        One row per input row, on ``rows.index``, with columns:
        ``expression_text`` (the source text copied verbatim, so a
        non-null non-string cell keeps its original value),
        ``expression_id`` (32-character lowercase hex string, or
        ``pd.NA`` when no AST was produced), ``operators_used`` (sorted
        tuple of distinct non-leaf operator names, or ``pd.NA``),
        ``parse_status`` (``"parsed"``, ``"parse_failed"`` or ``pd.NA``)
        and ``registry_sha256`` (the registry hash, identical on every
        row). All value columns have ``object`` dtype.

    Raises
    ------
    TypeError
        If ``rows`` is not a ``pandas.DataFrame``, or ``registry`` is
        supplied and is not a ``Mapping``.
    ValueError
        If ``rows`` lacks the ``expression_text`` column.
    """
    # The rows argument must be a DataFrame; nothing else is an input table.
    if not isinstance(rows, pd.DataFrame):
        # Raise naming the offending type for the audit trail.
        raise TypeError(f"parse_records_task_7: rows must be a pandas.DataFrame, got {type(rows).__name__}")
    # The input must carry the expression_text column; anything else is a defect.
    if "expression_text" not in rows.columns:
        # Raise naming the missing column and the observed schema for the audit trail.
        raise ValueError(f"parse_records_task_7: rows must carry an 'expression_text' column; observed {list(rows.columns)}")
    # Resolve the registry consulted for the subset rule and the stamp.
    active: Mapping[str, OperatorSpec] = registry if registry is not None else build_operator_registry_task_7()
    # The registry argument must be a Mapping when supplied.
    if not isinstance(active, Mapping):
        # Raise naming the offending type for the audit trail.
        raise TypeError(f"parse_records_task_7: registry must be a Mapping, got {type(active).__name__}")
    # Compute the registry stamp once; every row carries the same value.
    registry_hash: str = registry_sha256_task_7(active)
    # The registry key set drives the parse-status subset rule.
    registry_names: set[str] = set(active.keys())
    # Per-row accumulators; each list has exactly one entry per input row.
    expression_ids: List[Any] = []
    operators_used: List[Any] = []
    parse_status: List[Any] = []
    # Walk the input rows in order; position is never changed.
    for value in rows["expression_text"].tolist():
        # A missing/non-string cell short-circuits to the all-NA output row.
        if _is_missing_text_task_7(value):
            # Append the all-NA cells with the registry stamp intact.
            expression_ids.append(pd.NA)
            operators_used.append(pd.NA)
            parse_status.append(pd.NA)
            continue
        # Parse the text; grammar-invalid text is parse_failed with NA identity.
        try:
            # The parse step is the only place whitespace is dropped and names lowercased.
            node: Node = parse_expression_task_7(value)
        except DslParseError:
            # Append the parse_failed row with NA identity and NA operator set.
            expression_ids.append(pd.NA)
            operators_used.append(pd.NA)
            parse_status.append("parse_failed")
            continue
        # Lexically valid: compute the canonical identity.
        expression_ids.append(expression_id_task_7(node))
        # Collect the sorted, distinct, leaf-free operator names.
        used: Tuple[str, ...] = tuple(sorted(_collect_operator_names_task_7(node)))
        operators_used.append(used)
        # The subset rule decides between parsed and parse_failed.
        parse_status.append("parsed" if set(used) <= registry_names else "parse_failed")
    # Assemble the table on the input's index with object-dtype cells.
    table: pd.DataFrame = pd.DataFrame(
        {
            # The source text rides along for the evaluator table (NA preserved).
            "expression_text": rows["expression_text"].tolist(),
            # The canonical identity, or pd.NA for unparseable text.
            "expression_id": pd.Series(expression_ids, dtype=object),
            # The sorted distinct leaf-free operators, or pd.NA for unparseable text.
            "operators_used": pd.Series(operators_used, dtype=object),
            # The parse status partition: parsed | parse_failed | pd.NA.
            "parse_status": pd.Series(parse_status, dtype=object),
            # The registry stamp, identical on every row.
            "registry_sha256": pd.Series([registry_hash] * len(rows), dtype=object),
        },
        index=rows.index,
    )
    # Return the evaluator table; row order equals the input row order.
    return table

# -------------------------------------------------------------------------------------------------------------------------------
# Task 7, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_7(
    features_panel: pd.DataFrame | Any,
    expressions: Optional[pd.DataFrame] = None,
    *,
    task1_preflight: Any = None,
) -> Dict[str, Any]:
    """
    Build the registries, stamp the hash and parse the expressions.

    Validate ``features_panel`` through the boundary checker, build the
    operator registry and the terminal registry taken cell-for-cell from
    the panel's own Series (never reindexed), compute the registry
    SHA-256 stamp, then build the evaluator table through
    :func:`parse_records_task_7`. When ``expressions`` is ``None``, an
    empty preserved-rows frame carrying only the ``expression_text``
    column is parsed instead, so both paths return a table on the same
    schema. The caller-supplied preflight verdict is echoed by object
    identity, without normalisation. No files are written.

    Parameters
    ----------
    features_panel : pd.DataFrame | Any
        Features panel on the compound ``(stock_code, trade_date)``
        index; consumed only through the boundary validator, so the Task
        4 embargo facade is never unwrapped.
    expressions : Optional[pd.DataFrame], optional
        Preserved input rows carrying an ``expression_text`` column.
        Default is ``None``.
    task1_preflight : Any, optional
        Task 1 verdict echoed verbatim into the result. Default is
        ``None``.

    Returns
    -------
    Dict[str, Any]
        Exactly five keys: ``"registry"`` (the immutable operator
        registry), ``"terminal_registry"`` (mapping from terminal name
        to the panel Series for ``close``, ``volume``, ``amount`` and
        ``tradable``), ``"registry_sha256"`` (the hash also stamped on
        every evaluator-table row), ``"parse_table"`` (the evaluator
        table returned by :func:`parse_records_task_7`) and
        ``"task1_preflight"`` (the echoed verdict).

    Raises
    ------
    TypeError
        If ``features_panel`` fails the boundary validation, or
        ``expressions`` is neither ``None`` nor a ``pandas.DataFrame``.
    ValueError
        If the panel index or a pinned terminal column fails the
        boundary validation, or ``expressions`` lacks the
        ``expression_text`` column.

    Notes
    -----
    Downstream, Task 10 reads ``parse_table.parse_status`` and
    ``parse_table.operators_used`` for unsupported-state detection, and
    Task 8 consumes ``registry`` together with the ``expression_id``
    values.
    """
    # Validate the panel once at the boundary; upstream verification is trusted thereafter.
    panel: Any = _require_features_panel_task_7(features_panel, "run_task_7")
    # Build the operator registry (fresh, immutable, deterministic).
    registry: Mapping[str, OperatorSpec] = build_operator_registry_task_7()
    # Build the terminal registry from the panel's own Series (never reindexed).
    terminal_registry: Mapping[str, pd.Series] = build_terminal_registry_task_7(panel)
    # Stamp the registry's canonical hash once for the whole evaluator result.
    registry_hash: str = registry_sha256_task_7(registry)
    # An absent expressions argument yields an empty table on the same code path.
    if expressions is None:
        # Assemble the empty preserved-rows frame with the pinned schema.
        empty_rows: pd.DataFrame = pd.DataFrame({"expression_text": pd.Series([], dtype=object)})
        # Build the empty evaluator table through the standard path.
        parse_table: pd.DataFrame = parse_records_task_7(empty_rows, registry=registry)
    else:
        # Any other expressions type is a boundary defect.
        if not isinstance(expressions, pd.DataFrame):
            # Raise naming the offending type for the audit trail.
            raise TypeError(f"run_task_7: expressions must be a pandas.DataFrame or None, got {type(expressions).__name__}")
        # The preserved-rows frame must carry the expression_text column.
        if "expression_text" not in expressions.columns:
            # Raise naming the missing column for the audit trail.
            raise ValueError(f"run_task_7: expressions must carry an 'expression_text' column; observed {list(expressions.columns)}")
        # Build the evaluator table through the standard path.
        parse_table = parse_records_task_7(expressions, registry=registry)
    # Assemble the pinned five-key result; preflight echoes verbatim (no normalisation).
    return {
        # The frozen operator registry (MappingProxyType).
        "registry": registry,
        # The terminal registry (close/volume/amount/tradable, cell-for-cell).
        "terminal_registry": terminal_registry,
        # The registry stamp, identical to the value stamped on every table row.
        "registry_sha256": registry_hash,
        # The evaluator table (one row per preserved row, order preserved).
        "parse_table": parse_table,
        # The caller-supplied Task 1 verdict, echoed verbatim.
        "task1_preflight": task1_preflight,
    }
```

```python
# Task 8: Cohort Freeze Protocol

# ==============================================================================
# Task 8: Cohort Freeze Protocol
# ==============================================================================

_task_1 = SimpleNamespace(
    _canonical_json_bytes=_canonical_json_bytes,
    _sha256_hex=_sha256_hex,
)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, module-level pinned constants (each carries its configuration key path).
# -------------------------------------------------------------------------------------------------------------------------------

# The file name of the study configuration (the single source of truth the
# task text names). The paper-stated comparison values are looked up from it
# at call time by ``_paper_comparison_values_task_8``; they are never
# restated as literals in this module, and a missing or malformed
# configuration halts with an explicit DATA_ABSENT error instead of any
# fallback constant.
_STUDY_CONFIG_FILENAME_TASK_8: str = "study_revised_inputs_20261111_143538.md"

# The closing code fence of the study-inputs markdown (assembled from code points to keep this source fence-clean).
_CODE_FENCE_TASK_8: str = "\n" + chr(96) * 3

# P06_COHORT_ADMISSION_PARAMETERS.boundary_convention (the inclusive-convention disclosure).
BOUNDARY_CONVENTION_TASK_8: str = "INCLUSIVE_AS_STATED__EXECUTED_STRICTNESS_UNBOUND"

# The exclusion disclosure constant pinned by the step text for NaN-driven screen exclusions.
EXCLUSION_DISCLOSURE_TASK_8: str = "EXCLUDED_NAN_IN_SCREEN"

# P06_COHORT_ADMISSION_PARAMETERS.deduplication_key (the pinned dedup key).
DEDUP_KEY_TASK_8: str = "CANONICAL_AST_IDENTITY"

# The dedup rule source disclosure pinned by the step text.
DEDUP_RULE_SOURCE_TASK_8: str = "NOT_STATED_IN_PAPER__[R]"

# The two dedup statuses pinned by the step text.
DEDUP_STATUS_NO_OP_TASK_8: str = "DEDUP_NO_OP"
DEDUP_STATUS_APPLIED_TASK_8: str = "DEDUP_APPLIED"

# The freeze column-projection whitelist pinned by the step text (in order).
FREEZE_WHITELIST_TASK_8: Tuple[str, ...] = ("row_id", "expression_id", "is_sharpe", "is_fitness", "is_turnover")

# The pinned store and lock filenames of the write-once freeze store.
FREEZE_STORE_FILENAME_TASK_8: str = "cohort_freeze.json"
FREEZE_LOCK_SUFFIX_TASK_8: str = ".freeze_lock"

# The pinned timezone name of the freeze timestamp (P07; P31 meta_fields).
FREEZE_TIMEZONE_TASK_8: str = "Asia/Shanghai"

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Module exceptions (every halt is an explicit, informative error)
# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Exceptions: typed halt conditions for every step of the freeze protocol.
# -------------------------------------------------------------------------------------------------------------------------------

class CohortFreezeHaltError(Exception):
    """
    Base halt for every unrecoverable Cohort Freeze Protocol failure.

    Every halt condition in this module derives from this class, so a
    caller catches the protocol's failures with one ``except`` clause
    without swallowing unrelated exceptions. The message is rendered
    verbatim by ``Exception.__str__``.
    """

    def __init__(self, message: str) -> None:
        """
        Initialise the halt with a human-readable message.

        Parameters
        ----------
        message : str
            Actionable description of the halt condition.
        """
        # Store the message; ``Exception.__str__`` renders it verbatim.
        super().__init__(message)

class ScreenHaltError(CohortFreezeHaltError):
    """
    Halt when the IS-only screen cannot be evaluated honestly.

    Raised when ``row_id`` is duplicated (the surrogate index of the
    survivor frame would be ambiguous), or when an invariant of the
    screen's boundary validation is violated by something that is not a
    dtype or type defect; dtype and type defects raise ``TypeError`` or
    ``ValueError`` directly instead.
    """

class DeduplicationHaltError(CohortFreezeHaltError):
    """
    Halt when deduplication by canonical AST identity cannot proceed.

    Signals a NaN value in ``expression_id_canonical``: a NaN identity
    cannot be deduplicated meaningfully, and silently collapsing a NaN
    group to one representative is forbidden.
    """

class FreezeWhitelistHaltError(CohortFreezeHaltError):
    """
    Halt when the freeze column-projection whitelist is violated.

    Raised when the freeze input carries any column outside the
    whitelist ``{row_id, expression_id, is_sharpe, is_fitness,
    is_turnover}`` or is missing any whitelist column, and by the
    read-only façade whenever code reads a column outside the whitelist;
    an OOS column can therefore never enter the freeze.
    """

class FreezeTimestampHaltError(CohortFreezeHaltError):
    """
    Halt when the freeze timestamp is not before the OOS read start.

    Equality is a boundary failure: the freeze must strictly precede the
    first OOS read.
    """

class WriteOnceStoreHaltError(CohortFreezeHaltError):
    """
    Halt when the write-once cohort store cannot be used honestly.

    Signals that ``cohort_freeze.json`` or its ``.freeze_lock`` sentinel
    already exists, or that the store cannot be written or re-read for
    verification.
    """

class FreezeDivergenceHaltError(CohortFreezeHaltError):
    """
    Halt when re-verification of the freeze detects divergent bytes.

    The message produced for this halt carries the first divergent byte
    offset and a hexadecimal context window of both byte strings.
    """

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Private helpers
# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Helper 1: boundary DataFrame validation shared by every public callable.
# -------------------------------------------------------------------------------------------------------------------------------

def _require_dataframe_task_8(obj: Any, caller: str, name: str) -> pd.DataFrame:
    """
    Validate that ``obj`` is a :class:`pandas.DataFrame`.

    Return the object unchanged when validation succeeds; otherwise
    raise ``TypeError`` with the caller and parameter names.

    Parameters
    ----------
    obj : Any
        The object supplied by the caller.
    caller : str
        Name of the calling callable, used in error messages.
    name : str
        Parameter name being validated, used in error messages.

    Returns
    -------
    pd.DataFrame
        The same object, unchanged.

    Raises
    ------
    TypeError
        If ``obj`` is not a ``pandas.DataFrame``.
    """
    # The boundary contract of every public callable is a real DataFrame.
    if not isinstance(obj, pd.DataFrame):
        # Refuse anything else with an actionable message naming the caller.
        raise TypeError(f"{caller}: {name} must be a pandas.DataFrame, got {type(obj).__name__}.")
    # The validated frame is returned unchanged (never copied here).
    return obj

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Helper 2: required-column validation listing every missing column at once.
# -------------------------------------------------------------------------------------------------------------------------------

def _require_columns_task_8(frame: pd.DataFrame, required: Tuple[str, ...], caller: str) -> None:
    """
    Validate that ``frame`` carries every required column.

    Collect all missing names and raise one ``ValueError`` that lists
    them; do nothing when none are missing.

    Parameters
    ----------
    frame : pd.DataFrame
        The frame whose columns are audited.
    required : Tuple[str, ...]
        Column names that must be present.
    caller : str
        Name of the calling callable, used in error messages.

    Raises
    ------
    ValueError
        If any required column is absent; the message lists all missing
        columns at once.
    """
    # Compute the set difference between the pinned requirement and the actual columns.
    missing: List[str] = [column for column in required if column not in frame.columns]
    # A single informative halt names every missing column at once (never fail one at a time).
    if missing:
        # Raise with the full missing list so the caller can repair the input in one pass.
        raise ValueError(f"{caller}: missing required column(s) {missing}; required columns are {list(required)}.")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Helper 3: row_id integrity validation (integer dtype, non-null, unique).
# -------------------------------------------------------------------------------------------------------------------------------

def _require_row_id_task_8(frame: pd.DataFrame, caller: str) -> None:
    """
    Validate the ``row_id`` surrogate of ``frame``.

    Require a dtype that ``pd.api.types.is_integer_dtype`` accepts (any
    signed or unsigned NumPy integer or nullable ``Int``), no missing
    values and uniqueness; duplicated ids raise ``ScreenHaltError`` with
    the duplicated values.

    Parameters
    ----------
    frame : pd.DataFrame
        The frame whose ``row_id`` column is audited.
    caller : str
        Name of the calling callable, used in error messages.

    Raises
    ------
    TypeError
        If ``row_id`` is not an integer dtype.
    ValueError
        If any ``row_id`` value is missing.
    ScreenHaltError
        If any ``row_id`` value is duplicated.
    """
    # The surrogate must carry an integer dtype (pandas int64 or nullable Int64).
    if not pd.api.types.is_integer_dtype(frame["row_id"]):
        # Reject float/string/bool row_id carriers outright (an Int64-coerced float is a defect, not an ID).
        raise TypeError(f"{caller}: row_id must be an integer dtype (int64 or Int64), got {frame['row_id'].dtype}.")
    # Any missing row_id would make the surrogate index incomplete by construction.
    if bool(frame["row_id"].isna().any()):
        # Name the offending positional locations instead of silently dropping them.
        raise ValueError(f"{caller}: row_id contains missing values at positional locations {list(frame.index[frame['row_id'].isna()])}.")
    # Duplicated row_id values would make the freeze order and the representative selection ambiguous.
    if bool(frame["row_id"].duplicated().any()):
        # Halt with the duplicated values so the caller can repair the extraction upstream.
        duplicated_values: List[Any] = frame.loc[frame["row_id"].duplicated(keep=False), "row_id"].tolist()
        raise ScreenHaltError(f"{caller}: row_id must be unique; duplicated values {duplicated_values} make the surrogate index ambiguous.")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Helper 4: hexadecimal first-divergence report of two byte strings (the pinned hex-diff).
# -------------------------------------------------------------------------------------------------------------------------------

def _hex_diff_task_8(expected: bytes, actual: bytes) -> str:
    """
    Report the first divergent byte offset between two byte strings.

    Compare the common prefix, set the offset to the common length when
    one string is a prefix of the other, and render the slice ``[max(0,
    offset - 16), offset + 16)`` of each string (up to 16 bytes before
    the offset plus 16 bytes starting at it, truncated at the string
    ends) as space-separated hex, together with both lengths.

    Parameters
    ----------
    expected : bytes
        Reference byte string, e.g. canonical freeze bytes.
    actual : bytes
        Observed byte string, e.g. re-read store bytes.

    Returns
    -------
    str
        Report naming the first divergent offset, both hex windows and
        both lengths.

    Notes
    -----
    Used by freeze-divergence halt messages.
    """
    # The common prefix length bounds the search for the first divergence.
    limit: int = min(len(expected), len(actual))
    # Walk the common prefix to locate the first differing byte, if any.
    offset: Optional[int] = None
    for index in range(limit):
        # A differing byte pair terminates the scan with the current offset.
        if expected[index] != actual[index]:
            offset = index
            break
    # If the prefix matched entirely, one string is a strict prefix of the other.
    if offset is None:
        offset = limit
    # A 16-byte context window centred on the divergence makes the diff inspectable.
    context: int = 16
    low: int = max(0, offset - context)
    high: int = offset + context
    # Render both windows as space-separated hex for verbatim comparison.
    expected_hex: str = expected[low:high].hex(" ")
    actual_hex: str = actual[low:high].hex(" ")
    # Assemble the full report: offset, both windows, both lengths.
    return (
        f"first divergent byte offset {offset}; "
        f"expected[{low}:{high}]={expected_hex}; "
        f"actual[{low}:{high}]={actual_hex}; "
        f"len(expected)={len(expected)}; len(actual)={len(actual)}"
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Helper 5: the freeze whitelist façade that raises on any out-of-whitelist column access.
# -------------------------------------------------------------------------------------------------------------------------------

class _FreezeWhitelistFacade:
    """
    Read-only façade that enforces a column whitelist.

    Wrap a projected freeze frame and expose only whitelisted columns
    through ``series`` and ``frame``; any other column access raises
    ``FreezeWhitelistHaltError``. The wrapped frame is held by reference
    and never mutated.

    Attributes
    ----------
    whitelist : tuple[str, ...]
        Ordered column names the façade allows.
    row_count : int
        Number of rows in the wrapped frame.
    """

    def __init__(self, frame: pd.DataFrame, whitelist: Tuple[str, ...]) -> None:
        """
        Wrap a frame and pin the column whitelist.

        Parameters
        ----------
        frame : pd.DataFrame
            Projected freeze frame the façade wraps.
        whitelist : Tuple[str, ...]
            Non-empty ordered tuple of allowed column names.

        Raises
        ------
        TypeError
            If ``frame`` is not a ``pandas.DataFrame``.
        ValueError
            If ``whitelist`` is empty, not a tuple, or holds non-string
            items.
        """
        # The wrapped frame must be a DataFrame (boundary discipline).
        if not isinstance(frame, pd.DataFrame):
            # Refuse anything else with an actionable message.
            raise TypeError(f"_FreezeWhitelistFacade: frame must be a pandas.DataFrame, got {type(frame).__name__}.")
        # The whitelist must be a non-empty tuple of column names.
        if not isinstance(whitelist, tuple) or not whitelist or not all(isinstance(item, str) for item in whitelist):
            # Refuse an empty or malformed whitelist outright (a silent no-op guard is worse than none).
            raise ValueError("_FreezeWhitelistFacade: whitelist must be a non-empty tuple of column names.")
        # Store the frame by reference; the façade never copies and never mutates it.
        self._frame: pd.DataFrame = frame
        # Store the pinned whitelist for repeated enforcement.
        self._whitelist: Tuple[str, ...] = tuple(whitelist)

    @property
    def whitelist(self) -> Tuple[str, ...]:
        """
        Return the ordered column whitelist.

        The tuple is returned by reference; it is immutable.

        Returns
        -------
        Tuple[str, ...]
            The pinned whitelist.
        """
        # Return the pinned whitelist verbatim (callers may read, never extend).
        return self._whitelist

    @property
    def row_count(self) -> int:
        """
        Return the number of rows in the wrapped frame.

        Uses only ``len`` and exposes no column of the frame.

        Returns
        -------
        int
            Row count of the wrapped frame.
        """
        # Expose the row count without exposing any column of the frame.
        return int(len(self._frame))

    def _check(self, column: str) -> None:
        """
        Validate that ``column`` is inside the whitelist.

        Raise when the column is outside the whitelist; the message
        names the column and the whitelist.

        Parameters
        ----------
        column : str
            Column name the caller wishes to access.

        Raises
        ------
        FreezeWhitelistHaltError
            If ``column`` is outside the whitelist.
        """
        # Any column outside the pinned whitelist is an unconditional halt.
        if column not in self._whitelist:
            # Name the offending column and the pinned whitelist in the message (fail informatively).
            raise FreezeWhitelistHaltError(
                f"_FreezeWhitelistFacade: access to column {column!r} is outside the freeze whitelist "
                f"{list(self._whitelist)}; the freeze may not read it (OOS columns can never enter the freeze)."
            )

    def series(self, column: str) -> pd.Series:
        """
        Return a whitelisted column as a series.

        Check the whitelist first and return the column by reference
        without copying.

        Parameters
        ----------
        column : str
            Whitelisted column to read.

        Returns
        -------
        pd.Series
            The requested column, by reference.

        Raises
        ------
        FreezeWhitelistHaltError
            If ``column`` is outside the whitelist.
        """
        # Enforce the whitelist before any read is performed.
        self._check(column)
        # Return the whitelisted column by reference; the façade never copies data.
        return self._frame[column]

    def frame(self, columns: Tuple[str, ...]) -> pd.DataFrame:
        """
        Return a deep copy of a whitelisted column projection.

        Check every requested column before projecting with ``.loc`` and
        copying.

        Parameters
        ----------
        columns : Tuple[str, ...]
            Whitelisted columns to project, in the requested order.

        Returns
        -------
        pd.DataFrame
            Deep copy of the requested projection.

        Raises
        ------
        FreezeWhitelistHaltError
            If any requested column is outside the whitelist.
        """
        # Enforce the whitelist for every requested column before reading anything.
        for column in columns:
            self._check(column)
        # Return a deep copy so the caller can never mutate the wrapped frame through it.
        return self._frame.loc[:, list(columns)].copy()

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Helper 6: look up the paper-stated comparison values from the study configuration at call time.
# -------------------------------------------------------------------------------------------------------------------------------

_PAPER_COMPARISON_CACHE_TASK_8: Dict[str, Tuple[int, int]] = {}


def _load_study_configuration_task_8() -> Dict[str, Any]:
    """
    Load ``STUDY_CONFIG`` from the study-inputs markdown.

    Resolve ``Path.cwd() / _STUDY_CONFIG_FILENAME_TASK_8``, read UTF-8
    text, slice from ``STUDY_CONFIG = {`` to the next
    ``_CODE_FENCE_TASK_8``, compile and execute the block with empty
    builtins, then return the evaluated dictionary. No fallback constant
    is used.

    Returns
    -------
    Dict[str, Any]
        The evaluated ``STUDY_CONFIG`` dictionary.

    Raises
    ------
    CohortFreezeHaltError
        If the file is absent or unreadable, the block start or closing
        fence is missing, evaluation fails, or ``STUDY_CONFIG`` is not a
        dict. Every message carries ``DATA_ABSENT``.
    """
    # Resolve the study-inputs file relative to the project root (the parent of this file's directory).
    config_path: Path = Path.cwd() / _STUDY_CONFIG_FILENAME_TASK_8
    # The configuration is the only authoritative in-project source of the paper's figures; its absence is a hard halt.
    if not config_path.is_file():
        # Refuse to proceed without the authoritative declaration (never fall back to a constant).
        raise CohortFreezeHaltError(
            f"_load_study_configuration_task_8: DATA_ABSENT -- the study configuration {config_path} is absent; "
            f"the paper-stated comparison values cannot be looked up and this module refuses to invent them."
        )
    # Read the file as UTF-8 text (the study configuration is a fenced Python block inside the markdown).
    try:
        # A read failure is a hard halt with the OS error named (no silent fallback).
        text: str = config_path.read_text(encoding="utf-8")
    except OSError as exc:
        # Surface the filesystem failure with an actionable, category-tagged message.
        raise CohortFreezeHaltError(
            f"_load_study_configuration_task_8: DATA_ABSENT -- cannot read the study configuration {config_path}: {exc}"
        ) from exc
    # Locate the fenced Python block that defines STUDY_CONFIG (§iv of the study inputs).
    fence_start: int = text.find("STUDY_CONFIG = {")
    # A missing block start is a hard halt (the file is not the study configuration this module expects).
    if fence_start == -1:
        # The literal search failed; report it explicitly rather than guessing an offset.
        raise CohortFreezeHaltError(
            f"_load_study_configuration_task_8: DATA_ABSENT -- no 'STUDY_CONFIG = {{' definition found in {config_path}."
        )
    # Locate the end of the fenced block (the closing code fence after the dictionary's final brace).
    fence_end: int = text.find(_CODE_FENCE_TASK_8, fence_start)
    # A missing closing fence is a hard halt (the block is truncated or not fenced).
    if fence_end == -1:
        # The closing fence search failed; report it explicitly rather than truncating the block.
        raise CohortFreezeHaltError(
            f"_load_study_configuration_task_8: DATA_ABSENT -- no closing code fence after the STUDY_CONFIG block in {config_path}."
        )
    # Slice out exactly the dictionary literal (from the definition site to the end of the fence).
    block: str = text[fence_start:fence_end]
    # Evaluate the block in a namespace with no builtins (the block is pure literals, lists and dicts).
    namespace: Dict[str, Any] = {}
    try:
        # Compile the block with the file name as the source label for actionable error messages.
        code = compile(block, str(config_path), "exec")
        # Execute with empty builtins; any name reference that is not a literal raises NameError, caught below.
        exec(code, {"__builtins__": {}}, namespace)
    except Exception as exc:
        # A malformed block is a hard halt with the evaluator's own error named (never a silent fallback).
        raise CohortFreezeHaltError(
            f"_load_study_configuration_task_8: DATA_ABSENT -- the STUDY_CONFIG block in {config_path} failed to evaluate: {exc!r}"
        ) from exc
    # Retrieve the evaluated configuration object and verify it is a real dictionary.
    config: Any = namespace.get("STUDY_CONFIG")
    # A non-dict result means the block did not define the expected mapping; halt rather than proceed.
    if not isinstance(config, dict):
        # The evaluated namespace is missing STUDY_CONFIG or holds a non-dict value; report it explicitly.
        raise CohortFreezeHaltError(
            f"_load_study_configuration_task_8: DATA_ABSENT -- the STUDY_CONFIG block in {config_path} did not evaluate to a dict (got {type(config).__name__})."
        )
    # Return the evaluated configuration (the caller performs the key-path lookups).
    return config

def _paper_comparison_values_task_8() -> Tuple[int, int]:
    """
    Return paper-stated source row count and cohort size.

    Read ``source_row_count`` and ``cohort_size`` from
    ``P06_COHORT_ADMISSION_PARAMETERS`` in the study configuration,
    validate both as positive ints with bool excluded, cache the pair in
    ``_PAPER_COMPARISON_CACHE_TASK_8['pair']``, and return it. The
    values are comparison figures only.

    Returns
    -------
    source_row_count : int
        The paper-stated pre-filter row count.
    cohort_size : int
        The paper-stated frozen-cohort size.

    Raises
    ------
    CohortFreezeHaltError
        If the configuration is absent, unreadable or malformed; if the
        ``P06_COHORT_ADMISSION_PARAMETERS`` section is missing or not a
        dict; or if either figure is not a positive int. Messages carry
        ``DATA_ABSENT``.

    Notes
    -----
    A cache hit returns the previously validated pair without re-reading
    the configuration file.
    """
    # Serve the cached pair when the configuration has already been read this process.
    cached: Optional[Tuple[int, int]] = _PAPER_COMPARISON_CACHE_TASK_8.get("pair")
    # A cache hit returns the previously validated pair without re-reading the file.
    if cached is not None:
        # Return the validated cached pair verbatim (the configuration file is immutable during a run).
        return cached
    # Load the evaluated configuration dictionary (hard-halts on any absence or malformation).
    config: Dict[str, Any] = _load_study_configuration_task_8()
    # Fetch the P06 section that declares the paper's cohort-admission figures.
    p06: Any = config.get("P06_COHORT_ADMISSION_PARAMETERS")
    # A missing or non-dict P06 section is a hard halt (the declared figures cannot be located).
    if not isinstance(p06, dict):
        # Report the absent/malformed section explicitly rather than guessing values.
        raise CohortFreezeHaltError(
            "_paper_comparison_values_task_8: DATA_ABSENT -- the study configuration is missing a dict-valued "
            "'P06_COHORT_ADMISSION_PARAMETERS' section; the paper-stated comparison values cannot be looked up."
        )
    # Read the two figures verbatim from the configuration (no arithmetic, no coercion).
    source_row_count: Any = p06.get("source_row_count")
    # The cohort size is the freeze-protocol admission figure of the same section.
    cohort_size: Any = p06.get("cohort_size")
    # Validate both figures as genuine positive ints (bool is a subclass of int and is excluded explicitly).
    for section_key, value in (("P06_COHORT_ADMISSION_PARAMETERS.source_row_count", source_row_count), ("P06_COHORT_ADMISSION_PARAMETERS.cohort_size", cohort_size)):
        # A bool, a non-int, a non-positive or a missing figure halts rather than being coerced or defaulted.
        if isinstance(value, bool) or not isinstance(value, int) or value <= 0:
            # Report the offending key path and value rather than substitute a plausible constant.
            raise CohortFreezeHaltError(
                f"_paper_comparison_values_task_8: DATA_ABSENT -- {section_key} must be a positive int in the study configuration, got {value!r}."
            )
    # Bind the validated pair once (the return expression reads the validated configuration, never a literal).
    pair: Tuple[int, int] = (source_row_count, cohort_size)
    # Cache the validated pair for the remainder of the process (the configuration file is immutable during a run).
    _PAPER_COMPARISON_CACHE_TASK_8["pair"] = pair
    # Return the validated pair to the caller.
    return pair

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 1: The IS-only screen on PRESERVED_INPUT_ROWS
# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 1: screen_is_only_task_8 -- inclusive three-rule IS filter, NaN fails comparators.
# -------------------------------------------------------------------------------------------------------------------------------

def screen_is_only_task_8(preserved_input_rows: pd.DataFrame) -> Dict[str, Any]:
    """
    Apply the IS-only screen to preserved input rows.

    Validate the input frame, the required columns and the ``row_id``
    surrogate. Require float dtypes for ``is_sharpe``, ``is_fitness``
    and ``is_turnover``. Compute the inclusive mask ``is_sharpe >= 2.0``
    and ``is_fitness >= 0.0`` and ``is_turnover <= 500.0`` on the
    arriving values; ``NaN`` fails every comparator. Count NaN-driven
    exclusions, deep-copy the survivors, attach the original index
    labels as ``original_index``, rebuild the index from ``row_id`` as
    an ``Int64`` index named ``row_id``, and return the disclosure
    payload. No workflow-specific gates are applied and extra columns
    are retained.

    Parameters
    ----------
    preserved_input_rows : pd.DataFrame
        Preserved input rows. Must carry ``row_id`` (integer dtype,
        unique, non-null), ``is_sharpe``, ``is_fitness`` and
        ``is_turnover`` (float dtypes as they arrive).

    Returns
    -------
    Dict[str, Any]
        Keys: ``rows`` (survivors with all input columns plus
        ``original_index``, indexed by ``row_id``); ``n_after_I_screen``
        (survivor count); ``pre_filter_count`` (raw input row count);
        ``excluded_nan_in_screen`` (excluded rows with NaN in at least
        one screen column); ``exclusion_disclosure`` (pinned literal);
        ``paper_pre_filter_count`` (paper-stated comparison value);
        ``boundary_convention`` (pinned literal);
        ``workflow_specific_gates_applied`` (always ``False``).

    Raises
    ------
    TypeError
        If the input is not a DataFrame, ``row_id`` is not an integer
        dtype, or any screen column is not a float dtype.
    ValueError
        If a required column is missing or any ``row_id`` is missing.
    ScreenHaltError
        If any ``row_id`` is duplicated.
    CohortFreezeHaltError
        If the study configuration cannot provide the paper comparison
        value.
    """
    # Validate the boundary type first: a real DataFrame is the only accepted input.
    frame: pd.DataFrame = _require_dataframe_task_8(preserved_input_rows, "screen_is_only_task_8", "preserved_input_rows")
    # Validate the presence of every column this step reads (fails naming all missing columns).
    _require_columns_task_8(frame, ("row_id", "is_sharpe", "is_fitness", "is_turnover"), "screen_is_only_task_8")
    # Validate the surrogate: integer dtype, non-null, unique (the index built from it must be stable).
    _require_row_id_task_8(frame, "screen_is_only_task_8")
    # Enforce the pin: the screen is computed on the float64 values as they arrive.
    for column in ("is_sharpe", "is_fitness", "is_turnover"):
        # Any non-float carrier (rounded strings, Int64-coerced numerics, bools) halts with the offending dtype.
        if not pd.api.types.is_float_dtype(frame[column]):
            raise TypeError(
                f"screen_is_only_task_8: {column} must be a float dtype as arrived (float64), "
                f"got {frame[column].dtype}; the screen never computes on coerced or rounded values."
            )
    # Compute the three inclusive comparators on the raw float64 columns; NaN fails every comparator by IEEE-754 semantics.
    mask_sharpe: pd.Series = frame["is_sharpe"] >= 2.0
    mask_fitness: pd.Series = frame["is_fitness"] >= 0.0
    mask_turnover: pd.Series = frame["is_turnover"] <= 500.0
    # The screen is the conjunction of the three rules; it never consults IC, complexity, pool score or correlation.
    mask: pd.Series = mask_sharpe & mask_fitness & mask_turnover
    # The complement is the excluded set; it is used only for the disclosure counts, never for filtering.
    excluded_mask: pd.Series = ~mask
    # NaN-driven exclusions are counted separately and disclosed (rows failing the screen for a value, not a NaN, are not counted here).
    screen_columns: pd.DataFrame = frame.loc[:, ["is_sharpe", "is_fitness", "is_turnover"]]
    excluded_nan_in_screen: int = int((excluded_mask & screen_columns.isna().any(axis=1)).sum())
    # Select the survivors on a deep copy so the caller's frame is never mutated through the result.
    survivors: pd.DataFrame = frame.loc[mask].copy()
    # Normalise the survivors to a positional RangeIndex before attaching the original index labels.
    survivors = survivors.reset_index(drop=True)
    # Attach the ORIGINAL input index labels as a column for traceability (never re-derived, never reordered).
    survivors["original_index"] = frame.index[mask].to_numpy()
    # Rebuild the survivor index from row_id as an Int64 index named "row_id" (the pinned stable surrogate).
    survivors.index = pd.Index(survivors["row_id"].astype("Int64"), name="row_id")
    # Look up the paper-stated pre-filter comparison value from the study configuration (never a literal here).
    paper_pre_filter_count: int = _paper_comparison_values_task_8()[0]
    # Assemble the disclosure payload exactly as the interface contract pins it.
    result: Dict[str, Any] = {
        # The survivors: all input columns retained plus original_index, index built from row_id.
        "rows": survivors,
        # The row-count after the IS screen (python int, never a numpy scalar).
        "n_after_I_screen": int(mask.sum()),
        # The raw pre-filter row count (python int).
        "pre_filter_count": int(len(frame)),
        # The number of excluded rows whose exclusion is NaN-driven.
        "excluded_nan_in_screen": excluded_nan_in_screen,
        # The pinned exclusion disclosure literal.
        "exclusion_disclosure": EXCLUSION_DISCLOSURE_TASK_8,
        # The paper-stated pre-filter count, looked up from the study configuration (task step text; operator ruling).
        "paper_pre_filter_count": paper_pre_filter_count,
        # The inclusive boundary convention disclosure (verbatim from the configuration).
        "boundary_convention": BOUNDARY_CONVENTION_TASK_8,
        # The explicit disclosure that no workflow-specific gate was applied.
        "workflow_specific_gates_applied": False,
    }
    # The pure function returns its payload; no I/O was performed at any point.
    return result

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 2: Deduplication by canonical AST identity
# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 2: deduplicate_canonical_task_8 -- group-by expression_id_canonical, min row_id representative.
# -------------------------------------------------------------------------------------------------------------------------------

def deduplicate_canonical_task_8(rows: pd.DataFrame) -> Dict[str, Any]:
    """
    Deduplicate IS-screen rows by canonical AST identity.

    Validate the frame, then halt on any missing canonical identity.
    Group by ``expression_id_canonical`` and, in each group, retain
    the row with the minimum ``row_id`` (the earliest generation-order
    surrogate). The ledger records for every identity the retained
    ``row_id``, the ascending dropped ``row_id`` list and the
    delivered canonical identity verbatim; the survivors are
    projected to the freeze whitelist and indexed by ``row_id``. An
    unchanged row count is disclosed as ``DEDUP_NO_OP``; the dropped
    count is reported as a count difference, never as a survival-rate
    ratio. The paper-stated cohort size is returned as a comparison
    value only. This is Task 8, Step 2.

    Parameters
    ----------
    rows : pd.DataFrame
        Screen output. Must carry ``row_id`` (unique, non-null,
        integer dtype), ``expression_id`` (string),
        ``expression_id_canonical`` (canonical identity; NaN
        halts) and the float ``is_sharpe``, ``is_fitness`` and
        ``is_turnover`` columns as arrived.

    Returns
    -------
    Dict[str, Any]
        Mapping with the following keys:

        ``rows``
            DataFrame of whitelist columns ``row_id``,
            ``expression_id``, ``is_sharpe``, ``is_fitness`` and
            ``is_turnover``, indexed by ``row_id`` (name
            ``"row_id"``), ordered by ascending ``row_id``.
        ``dedup_ledger``
            DataFrame with columns ``expression_id_canonical``,
            ``retained_row_id`` (Int64), ``dropped_row_ids``
            (object lists of ascending ints) and
            ``canonical_ast_serialization`` (string verbatim),
            ordered by ``retained_row_id``.
        ``n_before_dedup``, ``n_after_dedup``, ``dropped_count``,
        ``dedup_factor``
            Integer counts; ``dedup_factor`` equals
            ``dropped_count`` and is never a ratio.
        ``dedup_status``
            ``"DEDUP_NO_OP"`` if the count is unchanged, else
            ``"DEDUP_APPLIED"``.
        ``dedup_rule_source``
            Pinned to ``"NOT_STATED_IN_PAPER__[R]"``.
        ``paper_cohort_size``
            Paper-stated cohort size read from the study
            configuration; a comparison value only.
        ``paper_cohort_size_is_comparison_value``
            Always ``True``; the value is never forced to match
            ``n_after_dedup``.
        ``canonical_key``
            The grouping column, ``"expression_id_canonical"``.

    Raises
    ------
    TypeError
        If ``rows`` is not a DataFrame, ``row_id`` is not an
        integer dtype, or an IS metric is not a float dtype.
    ValueError
        If a required column is missing or ``row_id`` has missing
        values.
    ScreenHaltError
        If ``row_id`` is duplicated.
    DeduplicationHaltError
        If any ``expression_id_canonical`` is NaN.

    Notes
    -----
    The function is pure: no I/O, no mutation of the caller's frame.
    The canonical identity is consumed as delivered and is not
    re-derived from expression text.
    """
    # Validate the boundary type first: a real DataFrame is the only accepted input.
    frame: pd.DataFrame = _require_dataframe_task_8(rows, "deduplicate_canonical_task_8", "rows")
    # Validate the presence of every column this step reads (fails naming all missing columns).
    _require_columns_task_8(
        frame,
        ("row_id", "expression_id", "expression_id_canonical", "is_sharpe", "is_fitness", "is_turnover"),
        "deduplicate_canonical_task_8",
    )
    # Validate the surrogate: integer dtype, non-null, unique.
    _require_row_id_task_8(frame, "deduplicate_canonical_task_8")
    # Enforce the as-arrived float pin on the IS metrics carried into the freeze whitelist.
    for column in ("is_sharpe", "is_fitness", "is_turnover"):
        # Any non-float carrier halts with the offending dtype (never coerced here).
        if not pd.api.types.is_float_dtype(frame[column]):
            raise TypeError(
                f"deduplicate_canonical_task_8: {column} must be a float dtype as arrived (float64), "
                f"got {frame[column].dtype}."
            )
    # The NaN-identity halt is pinned by the step text: check it before any grouping happens.
    canonical_series: pd.Series = frame["expression_id_canonical"]
    nan_identity_mask: pd.Series = canonical_series.isna()
    # A NaN identity cannot be deduplicated meaningfully; collapse is forbidden, so halt with the offenders named.
    if bool(nan_identity_mask.any()):
        # Collect the offending row_id values for the error message (the actionable payload).
        offending_row_ids: List[Any] = frame.loc[nan_identity_mask, "row_id"].tolist()
        raise DeduplicationHaltError(
            f"deduplicate_canonical_task_8: expression_id_canonical is NaN for row_id(s) {offending_row_ids}; "
            f"a NaN identity cannot be deduplicated meaningfully and may not collapse to one representative."
        )
    # Reset to a positional RangeIndex so every downstream index operation is positional and collision-free.
    work: pd.DataFrame = frame.reset_index(drop=True)
    # Group by the canonical identity with keys in sorted order for a deterministic ledger.
    grouped = work.groupby("expression_id_canonical", sort=True)
    # Collect the ledger records and the retained positional labels as we walk the groups.
    ledger_records: List[Dict[str, Any]] = []
    retained_labels: List[int] = []
    # One iteration per canonical identity; the retained row is the first occurrence of the minimum row_id.
    for canonical_key, group in grouped:
        # The idxmin label addresses the work frame's RangeIndex (positional), never the caller's index.
        retained_label: int = int(group["row_id"].idxmin())
        # Read the retained minimum row_id (a genuine Int64/int scalar here; the group is non-empty).
        retained_row_id: int = int(group.loc[retained_label, "row_id"])
        # The dropped row_id list is every other member of the group, sorted ascending as pinned.
        dropped_row_ids: List[int] = sorted(int(value) for value in group.drop(index=retained_label)["row_id"].tolist())
        # Record the identity's ledger row; the serialisation is the delivered canonical identity verbatim.
        ledger_records.append(
            {
                "expression_id_canonical": str(canonical_key),
                "retained_row_id": retained_row_id,
                "dropped_row_ids": dropped_row_ids,
                "canonical_ast_serialization": str(canonical_key),
            }
        )
        # Accumulate the retained positional label for the final projection.
        retained_labels.append(retained_label)
    # Project the retained rows out of the positional work frame (order: canonical-key sorted).
    retained: pd.DataFrame = work.loc[retained_labels].reset_index(drop=True) if retained_labels else work.iloc[0:0].copy()
    # Re-sort the representatives by ascending row_id (stable) so generation order is restored.
    retained = retained.sort_values("row_id", kind="stable").reset_index(drop=True)
    # Project exactly the freeze whitelist columns, in the pinned whitelist order.
    projected: pd.DataFrame = retained.loc[:, list(FREEZE_WHITELIST_TASK_8)].copy()
    # Rebuild the survivor index from row_id as an Int64 index named "row_id".
    projected.index = pd.Index(projected["row_id"].astype("Int64"), name="row_id")
    # Build the dedup ledger with explicit dtypes (Int64 retained ids; object lists of ascending python ints).
    if ledger_records:
        # The non-empty path: construct, then stamp explicit dtypes.
        dedup_ledger: pd.DataFrame = pd.DataFrame(ledger_records)
        # The retained_row_id column carries Int64 dtype as pinned.
        dedup_ledger["retained_row_id"] = dedup_ledger["retained_row_id"].astype("Int64")
        # Order the ledger by ascending retained row_id (generation order) with a RangeIndex.
        dedup_ledger = dedup_ledger.sort_values("retained_row_id", kind="stable").reset_index(drop=True)
    else:
        # The empty path: construct an empty ledger with the pinned columns and dtypes.
        dedup_ledger = pd.DataFrame(
            {
                "expression_id_canonical": pd.Series([], dtype="object"),
                "retained_row_id": pd.Series([], dtype="Int64"),
                "dropped_row_ids": pd.Series([], dtype="object"),
                "canonical_ast_serialization": pd.Series([], dtype="object"),
            }
        )
    # Look up the paper-stated cohort-size comparison value from the study configuration (never a literal here).
    paper_cohort_size: int = _paper_comparison_values_task_8()[1]
    # Compute the cardinalities as python ints (never numpy scalars).
    n_before_dedup: int = int(len(frame))
    n_after_dedup: int = int(len(projected))
    dropped_count: int = n_before_dedup - n_after_dedup
    # The dedup factor is the count difference itself -- never a survival-rate ratio (P20_CLAIM_CONTROLS).
    dedup_factor: int = dropped_count
    # The no-op disclosure is pinned to the unchanged-count case only.
    dedup_status: str = DEDUP_STATUS_NO_OP_TASK_8 if n_after_dedup == n_before_dedup else DEDUP_STATUS_APPLIED_TASK_8
    # Assemble the disclosure payload exactly as the interface contract pins it.
    result: Dict[str, Any] = {
        # The whitelist-projected representatives: index built from row_id, generation order.
        "rows": projected,
        # One ledger row per canonical identity (distributed to Task 16 as a redistributable artifact).
        "dedup_ledger": dedup_ledger,
        # The raw pre-dedup row count (python int).
        "n_before_dedup": n_before_dedup,
        # The representative count after deduplication (python int).
        "n_after_dedup": n_after_dedup,
        # The number of rows dropped by deduplication (python int).
        "dropped_count": dropped_count,
        # The count difference n_before_dedup - n_after_dedup (python int; never a ratio).
        "dedup_factor": dedup_factor,
        # The pinned no-op/applied disclosure.
        "dedup_status": dedup_status,
        # The pinned rule-source disclosure.
        "dedup_rule_source": DEDUP_RULE_SOURCE_TASK_8,
        # The paper-stated cohort size, looked up from the study configuration (task step text; operator ruling).
        "paper_cohort_size": paper_cohort_size,
        # The explicit disclosure that the comparison value is never forced to match the computed count.
        "paper_cohort_size_is_comparison_value": True,
        # The canonical key column the deduplication grouped on.
        "canonical_key": "expression_id_canonical",
    }
    # The pure function returns its payload; no I/O was performed at any point.
    return result

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 3: The final freeze -- ordered cohort, serialised bytes, SHA-256, timestamp, write-once store
# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 3, Data container: FrozenCohortTask8 (the frozen cohort dataclass).
# -------------------------------------------------------------------------------------------------------------------------------

@dataclass(frozen=True)
class FrozenCohortTask8:
    """
    Container for a frozen cohort's metadata and ordered rows.

    The class declares two annotated fields and no methods; it is
    constructed with keyword arguments by ``freeze_cohort_task_8``.
    ``meta["cohort_hash"]`` is the only truth handle.

    Attributes
    ----------
    meta : Dict[str, Any]
        Meta-fields including ``freeze_id`` (content-derived
        string), ``freeze_timestamp`` (timezone-aware ``datetime``
        in ``Asia/Shanghai``), ``cohort_hash`` (64-character
        SHA-256 hex), ``cohort_size`` (int), ``dedup_key``
        (``"CANONICAL_AST_IDENTITY"``), ``admission_rule``
        (``"IS_ONLY"``), ``oos_used_in_admission`` (``False``),
        the three scope disclaimers (each ``False``) and the
        serialisation and store disclosures.
    rows : pd.DataFrame
        The ordered cohort with ``cohort_rank`` (dense 1..N),
        ``row_id``, ``expression_id``, ``is_sharpe``,
        ``is_fitness`` and ``is_turnover``, indexed by
        ``cohort_rank``.
    """

    # The cohort meta-fields (P31); a plain dict because the interface contract pins ``meta: dict``.
    meta: Dict[str, Any]

    # The ordered cohort rows (indexed by cohort_rank).
    rows: pd.DataFrame

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Step 3: freeze_cohort_task_8 -- order, serialise, hash, stamp, persist write-once, re-verify.
# -------------------------------------------------------------------------------------------------------------------------------

def freeze_cohort_task_8(
    dedup_rows: pd.DataFrame,
    oos_read_start_ts: datetime,
    output_dir: str | os.PathLike[str],
) -> FrozenCohortTask8:
    """
    Freeze the deduplicated cohort in a write-once SHA-256 store.

    Order rows by ascending ``row_id``, assign dense ``cohort_rank``
    1..N, and serialise ``{"ordered_id_list": [...]}`` as canonical
    JSON with sorted keys and compact separators in UTF-8 with one
    LF terminator. Set ``cohort_hash = sha256(canonical_bytes)``
    hex-encoded; the hash covers the ordered ID list only and
    identifiers are string-verbatim. The freeze timestamp is
    ``datetime.now`` in ``Asia/Shanghai`` and must be strictly
    earlier than ``oos_read_start_ts`` before any write. Bytes go to
    ``<output_dir>/cohort_freeze.json`` and the hash to
    ``cohort_freeze.json.freeze_lock``, both created exclusively; a
    second call against the same directory halts. Re-serialisation,
    a JSON round-trip and a byte-for-byte read-back are verified.
    Task 8, Step 3.

    Parameters
    ----------
    dedup_rows : pd.DataFrame
        Freeze input carrying exactly ``row_id``, ``expression_id``,
        ``is_sharpe``, ``is_fitness`` and ``is_turnover``; any extra
        or missing column halts. ``row_id`` must be unique,
        non-null and integer dtype; ``expression_id`` must be a
        non-empty string; the three IS metrics must be float dtype.
    oos_read_start_ts : datetime
        Timezone-aware timestamp of the OOS harness's first read;
        the freeze timestamp must be strictly earlier.
    output_dir : str | os.PathLike[str]
        Directory receiving the store and lock; created if absent.

    Returns
    -------
    FrozenCohortTask8
        Immutable cohort. ``meta`` carries ``freeze_id``,
        ``freeze_timestamp``, ``cohort_hash``, ``cohort_size``,
        ``dedup_key``, ``admission_rule``, ``oos_used_in_admission``
        (``False``), the three scope disclaimers (each ``False``),
        the serialisation disclosures and the store filenames.
        ``rows`` holds ``cohort_rank``, ``row_id``,
        ``expression_id``, ``is_sharpe``, ``is_fitness`` and
        ``is_turnover``, indexed by ``cohort_rank``.
        ``meta["cohort_hash"]`` is the truth handle.

    Raises
    ------
    TypeError
        If ``dedup_rows`` is not a DataFrame, ``row_id`` is not an
        integer dtype, ``expression_id`` is not a non-empty string,
        ``oos_read_start_ts`` is not a datetime, or an IS metric is
        not a float dtype.
    ValueError
        If ``row_id`` has missing values or ``oos_read_start_ts`` is
        timezone-naive.
    CohortFreezeHaltError
        If ``row_id`` is duplicated.
    FreezeWhitelistHaltError
        If the input carries a column outside the whitelist or is
        missing a whitelist column.
    FreezeTimestampHaltError
        If the freeze timestamp is not strictly earlier than
        ``oos_read_start_ts``; equality is a boundary failure.
    WriteOnceStoreHaltError
        If ``output_dir`` cannot be created, the store or lock
        already exists, the store or lock cannot be written, or
        the store cannot be re-read.
    FreezeDivergenceHaltError
        If re-serialisation diverges, the JSON round-trip cannot be
        decoded or the decoded payload differs from the payload, or
        the persisted bytes or digest differ from the canonical
        values.

    Notes
    -----
    No OOS read occurs. Only ``cohort_freeze.json`` and
    ``cohort_freeze.json.freeze_lock`` are written; the lock holds
    ``cohort_hash`` plus one LF. Pre-existing artifacts are never
    overwritten.
    """
    # Validate the boundary type: a real DataFrame is the only accepted freeze input.
    frame: pd.DataFrame = _require_dataframe_task_8(dedup_rows, "freeze_cohort_task_8", "dedup_rows")
    # Enforce the exact-whitelist projection: extras (an OOS column included) halt the freeze.
    extras: List[str] = [column for column in frame.columns if column not in FREEZE_WHITELIST_TASK_8]
    # Any extra column is named verbatim in the halt message.
    if extras:
        raise FreezeWhitelistHaltError(
            f"freeze_cohort_task_8: input carries column(s) {extras} outside the freeze whitelist "
            f"{list(FREEZE_WHITELIST_TASK_8)}; an OOS column can never enter the freeze."
        )
    # Missing columns are named verbatim in the halt message (both directions are enforced).
    missing: List[str] = [column for column in FREEZE_WHITELIST_TASK_8 if column not in frame.columns]
    if missing:
        raise FreezeWhitelistHaltError(
            f"freeze_cohort_task_8: input is missing whitelist column(s) {missing}; "
            f"required columns are {list(FREEZE_WHITELIST_TASK_8)}."
        )
    # Validate the surrogate: integer dtype, non-null, unique (cohort_rank is a dense rank over it).
    if not pd.api.types.is_integer_dtype(frame["row_id"]):
        raise TypeError(f"freeze_cohort_task_8: row_id must be an integer dtype (int64 or Int64), got {frame['row_id'].dtype}.")
    if bool(frame["row_id"].isna().any()):
        raise ValueError("freeze_cohort_task_8: row_id contains missing values; the freeze order would be undefined.")
    if bool(frame["row_id"].duplicated().any()):
        # Name the duplicated values; cohort_rank assignment would be order-ambiguous otherwise.
        duplicated_values: List[Any] = frame.loc[frame["row_id"].duplicated(keep=False), "row_id"].tolist()
        raise CohortFreezeHaltError(f"freeze_cohort_task_8: row_id must be unique; duplicated values {duplicated_values}.")
    # Enforce the string-verbatim identifier pin: every expression_id must be a non-empty string.
    for row_label, value in zip(frame["row_id"].tolist(), frame["expression_id"].tolist()):
        # A non-string or empty identifier violates the standing register's identifier-format pin [R].
        if not isinstance(value, str) or value == "":
            raise TypeError(
                f"freeze_cohort_task_8: expression_id must be a non-empty string (string-verbatim pin [R]); "
                f"got {value!r} at row_id {row_label!r}."
            )
    # Enforce the as-arrived float pin on the IS metrics carried in the freeze whitelist.
    for column in ("is_sharpe", "is_fitness", "is_turnover"):
        if not pd.api.types.is_float_dtype(frame[column]):
            raise TypeError(
                f"freeze_cohort_task_8: {column} must be a float dtype as arrived (float64), got {frame[column].dtype}."
            )
    # The OOS harness's first-read stamp must be a real, timezone-aware datetime (aware comparisons only).
    if not isinstance(oos_read_start_ts, datetime.datetime):
        raise TypeError(f"freeze_cohort_task_8: oos_read_start_ts must be a datetime, got {type(oos_read_start_ts).__name__}.")
    if oos_read_start_ts.tzinfo is None or oos_read_start_ts.utcoffset() is None:
        raise ValueError("freeze_cohort_task_8: oos_read_start_ts must be timezone-aware; a naive stamp cannot bound the freeze.")
    # Bind the freeze timestamp before any validation of the ordering work below (now in Asia/Shanghai, pinned).
    freeze_timestamp: datetime = datetime.datetime.now(tz=ZoneInfo(FREEZE_TIMEZONE_TASK_8))
    # The strict inequality is pinned by the step text: equality is a boundary failure.
    if not freeze_timestamp < oos_read_start_ts:
        raise FreezeTimestampHaltError(
            f"freeze_cohort_task_8: freeze_timestamp {freeze_timestamp.isoformat()} must be strictly earlier than "
            f"oos_read_start_ts {oos_read_start_ts.isoformat()}; equality is a boundary failure and the freeze did not run."
        )
    # Reset the caller's index to a positional RangeIndex: the caller's index may be named ``row_id``,
    # which would make ``sort_values("row_id")`` ambiguous between the index level and the column label.
    positional: pd.DataFrame = frame.reset_index(drop=True)
    # Order the cohort by ascending row_id (the generation-order surrogate, stable); cohort_rank follows.
    ordered: pd.DataFrame = positional.sort_values("row_id", kind="stable").reset_index(drop=True)
    # Assign rank-dense cohort_rank 1..N (Int64) in the pinned ascending order.
    ordered["cohort_rank"] = pd.Series(range(1, len(ordered) + 1), dtype="Int64")
    # Wrap the ordered frame in the façade; every subsequent column access flows through it.
    facade: _FreezeWhitelistFacade = _FreezeWhitelistFacade(ordered, FREEZE_WHITELIST_TASK_8 + ("cohort_rank",))
    # Materialise the ordered identity list from the façade (the single source of truth for hashing and persistence).
    ordered_ids: List[str] = [str(value) for value in facade.series("expression_id").tolist()]
    # Build the payload dict; P07.freeze_hash_covers == "ORDERED_ID_LIST".
    payload: Dict[str, Any] = {"ordered_id_list": ordered_ids}
    # Serialise with the pinned canonical-JSON corpus imported from the verified Task 1 module (never re-written).
    canonical_bytes: bytes = _task_1._canonical_json_bytes(payload)
    # Re-serialise independently (fresh payload dict, fresh list) to prove the bytes are a pure function of the IDs.
    recomputed_bytes: bytes = _task_1._canonical_json_bytes({"ordered_id_list": list(facade.series("expression_id").tolist())})
    # Any divergence halts with the hex-diffed bytes (the pinned re-verifiability failure mode).
    if recomputed_bytes != canonical_bytes:
        raise FreezeDivergenceHaltError(
            "freeze_cohort_task_8: re-serialisation diverged from the canonical bytes; " + _hex_diff_task_8(canonical_bytes, recomputed_bytes)
        )
    # Round-trip the payload through json.loads to prove the bytes decode back to exactly the ordered ID list.
    try:
        round_tripped: Any = json.loads(canonical_bytes.decode("utf-8"))
    except (UnicodeDecodeError, json.JSONDecodeError) as exc:
        raise FreezeDivergenceHaltError(f"freeze_cohort_task_8: canonical bytes failed the JSON round-trip: {exc!r}") from exc
    # A payload that does not round-trip is a divergent artifact; halt with the decoded value named.
    if round_tripped != payload:
        raise FreezeDivergenceHaltError(
            f"freeze_cohort_task_8: canonical bytes do not round-trip to the payload (got {round_tripped!r}); the freeze is not reproducible."
        )
    # Compute cohort_hash := sha256(canonical_bytes), hex-encoded (P07.freeze_hash_algorithm == "SHA-256").
    cohort_hash: str = _task_1._sha256_hex(canonical_bytes)
    # freeze_id is a content-derived, deterministic handle (stable across re-runs of the same inputs).
    freeze_id: str = f"freeze_{cohort_hash[:16]}"
    # Resolve and create the output directory (created only now, after every pure check has passed).
    out_dir: Path = Path(output_dir)
    try:
        out_dir.mkdir(parents=True, exist_ok=True)
    except OSError as exc:
        raise WriteOnceStoreHaltError(f"freeze_cohort_task_8: cannot create output_dir {out_dir}: {exc}") from exc
    # Pin the store path and the lock sentinel path (only these two writes ever happen).
    store_path: Path = out_dir / FREEZE_STORE_FILENAME_TASK_8
    lock_path: Path = Path(str(store_path) + FREEZE_LOCK_SUFFIX_TASK_8)
    # The write-once check precedes any open(): a pre-existing store or lock halts without touching prior bytes.
    existing_artifacts: List[str] = [str(path) for path in (store_path, lock_path) if path.exists()]
    if existing_artifacts:
        raise WriteOnceStoreHaltError(
            f"freeze_cohort_task_8: the write-once store already exists at {existing_artifacts}; "
            f"the first cohort is the truth handle and is never overwritten."
        )
    # Encode the lock sentinel content: the cohort_hash hex (plus a trailing LF for line-oriented tooling).
    lock_bytes: bytes = (cohort_hash + "\n").encode("utf-8")
    # Persist the pair atomically-ish: exclusive create for both, rolling back the store if the lock fails.
    try:
        # Exclusive-create the store first; "xb" fails rather than overwrite if it appeared since the check.
        with open(store_path, "xb") as store_handle:
            store_handle.write(canonical_bytes)
        try:
            # Exclusive-create the lock sentinel second; a failure here rolls the store back.
            with open(lock_path, "xb") as lock_handle:
                lock_handle.write(lock_bytes)
        except BaseException:
            # Roll the pair back so a failed freeze leaves neither a store nor a lock behind.
            try:
                os.unlink(store_path)
            except OSError:
                # A rollback failure is reported through the primary exception; the orphan is evidence, not noise.
                pass
            raise
    except OSError as exc:
        raise WriteOnceStoreHaltError(f"freeze_cohort_task_8: cannot persist the write-once store at {store_path} (lock: {lock_path}): {exc}") from exc
    # Re-read the persisted store and verify it byte-for-byte against the canonical bytes (re-verification).
    try:
        read_back_bytes: bytes = store_path.read_bytes()
    except OSError as exc:
        raise WriteOnceStoreHaltError(f"freeze_cohort_task_8: cannot re-read the persisted store at {store_path}: {exc}") from exc
    # Any divergence of the persisted bytes from the canonical bytes halts with the hex-diffed report.
    if read_back_bytes != canonical_bytes:
        raise FreezeDivergenceHaltError(
            "freeze_cohort_task_8: persisted store diverges from the canonical bytes; " + _hex_diff_task_8(canonical_bytes, read_back_bytes)
        )
    # Re-digest the persisted bytes and compare against cohort_hash (the pinned re-verifiability check).
    if _task_1._sha256_hex(read_back_bytes) != cohort_hash:
        raise FreezeDivergenceHaltError(
            "freeze_cohort_task_8: re-read digest of the persisted store does not equal cohort_hash; the freeze is not reproducible."
        )
    # Build the frozen rows: cohort_rank first, then the pinned whitelist columns, indexed by cohort_rank.
    frozen_rows: pd.DataFrame = pd.DataFrame(
        {
            # The dense hash order, assigned above (Int64).
            "cohort_rank": ordered["cohort_rank"].astype("Int64"),
            # The surrogate row_id (Int64 as arrived).
            "row_id": ordered["row_id"].astype("Int64"),
            # The string-verbatim identifiers.
            "expression_id": ordered["expression_id"],
            # The as-arrived float64 IS metrics.
            "is_sharpe": ordered["is_sharpe"],
            # The as-arrived float64 fitness.
            "is_fitness": ordered["is_fitness"],
            # The as-arrived float64 turnover.
            "is_turnover": ordered["is_turnover"],
        }
    )
    # Index the frozen rows by cohort_rank with the pinned index name.
    frozen_rows.index = pd.Index(frozen_rows["cohort_rank"], name="cohort_rank")
    # Bind the frozen row count once (used by the meta payload).
    n_frozen: int = int(len(frozen_rows))
    # Assemble the cohort meta exactly as P31 pins it, with the serialisation disclosures stated explicitly.
    meta: Dict[str, Any] = {
        # P31 meta_fields.freeze_id ([R]): content-derived, deterministic, non-empty.
        "freeze_id": freeze_id,
        # P31 meta_fields.freeze_timestamp ([R]): tz-aware Asia/Shanghai.
        "freeze_timestamp": freeze_timestamp,
        # P31 meta_fields.cohort_hash ([R], SHA-256): hex digest of the canonical bytes.
        "cohort_hash": cohort_hash,
        # P31 meta_fields.cohort_size ([S]): the frozen cohort size, computed at call time.
        "cohort_size": n_frozen,
        # P06_COHORT_ADMISSION_PARAMETERS.deduplication_key ([R] after [REVISED]).
        "dedup_key": DEDUP_KEY_TASK_8,
        # P06_COHORT_ADMISSION_PARAMETERS.screen_scope ([S] IS only).
        "admission_rule": "IS_ONLY",
        # P31 meta_fields.oos_used_in_admission ([S] False).
        "oos_used_in_admission": False,
        # The three P07 scope disclaimers, reproduced verbatim ([S] each, False).
        "is_census_of_generated_candidates": False,
        "is_historically_blind_holdout": False,
        "is_matched_replay_of_keep_totals": False,
        # P07.freeze_hash_algorithm ([R] SHA-256).
        "freeze_hash_algorithm": "SHA-256",
        # P07.freeze_hash_covers ([R] ORDERED_ID_LIST).
        "freeze_hash_covers": "ORDERED_ID_LIST",
        # The string-verbatim serialisation format pinned by this module (see the disclosure).
        "freeze_serialization_format": "STRING_VERBATIM_CANONICAL_JSON",
        # The explicit statement the step text requires: why TWO_DECIMAL is inapplicable.
        "freeze_serialization_format_disclosure": (
            "P07.freeze_serialization_format == 'TWO_DECIMAL_PER_SLOT_CANONICAL_JSON' presumes numeric identifiers; "
            "the standing register's identifier-format pin ([R]) declares identifiers to be strings, so two-decimal "
            "precision is inapplicable and the serialisation is string-verbatim; "
            "the numeric-identifier variant is a Task 15 sensitivity analysis."
        ),
        # P07.freeze_serialization_bytes_encoding ([REVISED] utf-8).
        "freeze_serialization_bytes_encoding": "utf-8",
        # P07.freeze_serialization_line_terminator ([REVISED] LF).
        "freeze_serialization_line_terminator": "LF",
        # P07.freeze_serialization_id_ordering ([REVISED] BY_COHORT_RANK_ASCENDING).
        "freeze_serialization_id_ordering": "BY_COHORT_RANK_ASCENDING",
        # P07.freeze_before_any_scoring ([S]).
        "freeze_before_any_scoring": True,
        # P07.freeze_timestamp_timezone_aware ([R]) and the pinned timezone name.
        "freeze_timestamp_timezone": FREEZE_TIMEZONE_TASK_8,
        "freeze_timestamp_timezone_aware": True,
        # The relative artifact names of the write-once store (redistributable; no absolute paths).
        "freeze_store_filename": FREEZE_STORE_FILENAME_TASK_8,
        "freeze_lock_filename": f"{FREEZE_STORE_FILENAME_TASK_8}{FREEZE_LOCK_SUFFIX_TASK_8}",
        # P06.deduplication_timing ([S] BEFORE_FREEZE_USING_IS_DATA_ONLY).
        "deduplication_timing": "BEFORE_FREEZE_USING_IS_DATA_ONLY",
        # The only admission predicates that ran (the three IS rules; no workflow-specific gates).
        "admission_gates_applied": ["is_sharpe>=2.0", "is_fitness>=0.0", "is_turnover<=500.0"],
        # The explicit disclosure that workflow-specific gates were never applied.
        "workflow_specific_gates_applied": [],
        # The pinned column-projection whitelist the freeze enforced.
        "column_projection_whitelist": list(FREEZE_WHITELIST_TASK_8),
        # The identifier format disclosure (the [R] pin the step text restates).
        "identifier_format": "STRING_VERBATIM",
    }
    # Return the immutable cohort; meta["cohort_hash"] is the only truth handle.
    return FrozenCohortTask8(meta=meta, rows=frozen_rows)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------
# Task 8, Orchestrator: run_task_8 -- screen, deduplicate, freeze; assemble the disclosure payload.
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_8(
    preserved_input_rows: pd.DataFrame,
    oos_read_start_ts: datetime,
    output_dir: str | os.PathLike[str],
    upstream_halt_reason: object = None,
) -> Dict[str, Any]:
    """
    Run Task 8's three steps in order: screen, deduplicate, freeze.

    Compose :func:`screen_is_only_task_8`,
    :func:`deduplicate_canonical_task_8` and
    :func:`freeze_cohort_task_8` in that fixed order, under the pinned
    ``P06.deduplication_timing == "BEFORE_FREEZE_USING_IS_DATA_ONLY"``
    and ``P07.freeze_before_any_scoring == True``. The raw input row
    count is bound before any transformation. The three step payloads
    are merged into one flat disclosure dictionary; the two paper-stated
    comparison figures are looked up from the study configuration at
    call time, never restated as literals and never forced to match the
    computed counts (``P41.reachable_with_private_artifacts_only``).
    ``upstream_halt_reason`` is echoed by identity as
    ``task_1_halt_reason``.

    Parameters
    ----------
    preserved_input_rows : pd.DataFrame
        Unscreened preserved input rows; must carry the IS columns
        ``row_id``, ``is_sharpe``, ``is_fitness``, ``is_turnover`` and
        ``expression_id_canonical``.
    oos_read_start_ts : datetime
        First-read timestamp of the OOS harness; must be timezone-aware
        and strictly later than the freeze timestamp.
    output_dir : str | os.PathLike[str]
        Directory beneath which ``cohort_freeze.json`` and its
        ``.freeze_lock`` sentinel are written once.
    upstream_halt_reason : object, optional
        Opaque sentinel from the upstream orchestrator, echoed by
        identity. Default is ``None``.

    Returns
    -------
    Dict[str, Any]
        Flat disclosure payload. Counts: ``source_rows`` (input ``len``
        as delivered), ``n_after_I_screen``, ``n_before_dedup``,
        ``n_after_dedup``, ``dropped_count``, ``cohort_size``.
        Deduplication: ``dedup_factor`` (a count difference, never a
        survival ratio), ``dedup_status`` (``"DEDUP_NO_OP"`` or
        ``"DEDUP_APPLIED"``), ``dedup_ledger``, ``dedup_rule_source``,
        ``dedup_key``. Freeze: ``cohort_hash`` (SHA-256 hex of the
        canonical bytes), ``freeze_timestamp`` (Asia/Shanghai),
        ``freeze_id``, ``admission_rule``. Disclosures:
        ``excluded_nan_in_screen``, ``exclusion_disclosure``,
        ``oos_used_in_admission`` (always ``False``),
        ``task_1_halt_reason`` (identity echo of the argument).
        Bookkeeping: ``paper_pre_filter_count`` and
        ``paper_cohort_size`` are the study-configuration comparison
        values, not computed observations.

    Raises
    ------
    TypeError
        If ``preserved_input_rows`` is not a DataFrame; raised in this
        function's own body by ``_require_dataframe_task_8`` before any
        step runs. The step callables also raise ``TypeError`` if
        ``oos_read_start_ts`` is not a ``datetime``.
    ValueError
        If ``oos_read_start_ts`` is timezone-naive; enforced by the step
        callables.
    ScreenHaltError
        If the IS screen step halts.
    DeduplicationHaltError
        If the deduplication step halts.
    CohortFreezeHaltError
        If the freeze step halts; propagates unchanged.
    FreezeWhitelistHaltError
        If a column outside the projection whitelist is accessed.
    FreezeTimestampHaltError
        If ``freeze_timestamp`` is not strictly before
        ``oos_read_start_ts``.
    WriteOnceStoreHaltError
        If the write-once store cannot be persisted.
    FreezeDivergenceHaltError
        If a re-run yields divergent cohort bytes.

    Notes
    -----
    The freeze reads whitelisted (IS) columns only, so no
    ``historical_oos_*`` column is required or read at any point of this
    module. The only side effects are the two write-once freeze
    artifacts written beneath ``output_dir``.
    """
    # Bind the raw input row count before any transformation (source_rows is the length as delivered).
    source_rows: int = int(len(_require_dataframe_task_8(preserved_input_rows, "run_task_8", "preserved_input_rows")))
    # Step 1: apply the inclusive IS screen (float64 as arrived; NaN fails comparators; disclosure included).
    screened: Dict[str, Any] = screen_is_only_task_8(preserved_input_rows)
    # Step 2: deduplicate by canonical AST identity (min row_id representative; NaN identity halts).
    deduplicated: Dict[str, Any] = deduplicate_canonical_task_8(screened["rows"])
    # Step 3: freeze the ordered cohort (write-once; strict timestamp arithmetic; whitelist enforced).
    frozen: FrozenCohortTask8 = freeze_cohort_task_8(
        deduplicated["rows"],
        oos_read_start_ts=oos_read_start_ts,
        output_dir=output_dir,
    )
    # Assemble the flat disclosure payload exactly as the interface contract pins it.
    result: Dict[str, Any] = {
        # Step 1 outcome: survivors after the IS screen (python int).
        "n_after_I_screen": screened["n_after_I_screen"],
        # Step 2 outcome: rows before deduplication (python int).
        "n_before_dedup": deduplicated["n_before_dedup"],
        # Step 2 outcome: representatives after deduplication (python int).
        "n_after_dedup": deduplicated["n_after_dedup"],
        # Step 2 outcome: the number of dropped rows (python int).
        "dropped_count": deduplicated["dropped_count"],
        # Step 2 outcome: the count difference, never a survival-rate ratio (P20_CLAIM_CONTROLS).
        "dedup_factor": deduplicated["dedup_factor"],
        # Step 2 outcome: DEDUP_NO_OP / DEDUP_APPLIED disclosure.
        "dedup_status": deduplicated["dedup_status"],
        # Step 2 outcome: the ledger for Task 16 (one row per canonical identity).
        "dedup_ledger": deduplicated["dedup_ledger"],
        # Step 2 outcome: the pinned rule-source disclosure.
        "dedup_rule_source": deduplicated["dedup_rule_source"],
        # The paper-stated pre-filter count, looked up from the study configuration upstream (task step text; operator ruling).
        "paper_pre_filter_count": screened["paper_pre_filter_count"],
        # The paper-stated cohort size, looked up from the study configuration upstream (task step text; operator ruling).
        "paper_cohort_size": deduplicated["paper_cohort_size"],
        # Step 3 outcome: the SHA-256 hex of the canonical bytes.
        "cohort_hash": frozen.meta["cohort_hash"],
        # Step 3 outcome: the Asia/Shanghai freeze timestamp.
        "freeze_timestamp": frozen.meta["freeze_timestamp"],
        # Step 3 outcome: the content-derived freeze identifier.
        "freeze_id": frozen.meta["freeze_id"],
        # The raw input row count as delivered (distinct from n_after_I_screen).
        "source_rows": source_rows,
        # Step 3 outcome: the frozen cohort size.
        "cohort_size": frozen.meta["cohort_size"],
        # The verbatim identity echo of the upstream halt reason (never stringified, never copied).
        "task_1_halt_reason": upstream_halt_reason,
        # Step 1 disclosure: NaN-driven exclusion count and its pinned disclosure literal.
        "excluded_nan_in_screen": screened["excluded_nan_in_screen"],
        "exclusion_disclosure": screened["exclusion_disclosure"],
        # Step 3 disclosure: no OOS data was used in admission (verbatim False).
        "oos_used_in_admission": False,
        # Step 3 disclosure: the pinned dedup key and admission rule.
        "dedup_key": frozen.meta["dedup_key"],
        "admission_rule": frozen.meta["admission_rule"],
    }
    # The orchestrator returns the assembled payload; all I/O happened beneath output_dir only.
    return result
```

```python
# Task 9: Daily Rank-IC Scoring

# ==============================================================================
# Task 9: Daily Rank-IC Scoring
# ==============================================================================

# _task_1 resolves to the namespace bound in the Task 8 cell (Section 3.2(a)).

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, module-level pinned constants (each carries its STUDY_CONFIG key path).
# -------------------------------------------------------------------------------------------------------------------------------

# STUDY_CONFIG.P02_RANK_IC_PARAMETERS.min_eligible_stocks_per_date == 5  [S]
MIN_ELIGIBLE_STOCKS_PER_DATE_TASK_9: int = 5

# STUDY_CONFIG.P02_RANK_IC_PARAMETERS.min_distinct_signal_values == 3  [S]
# (min_distinct_signal_values_status == "DIAGNOSTIC_CONVENTION"  [S])
MIN_DISTINCT_SIGNAL_VALUES_TASK_9: int = 3

# STUDY_CONFIG.P02_RANK_IC_PARAMETERS.min_distinct_label_values == 2  [S]
MIN_DISTINCT_LABEL_VALUES_TASK_9: int = 2

# Reject-reason vocabulary (Step 1 and Step 3 of the task step text).
REJECT_REASON_GATE_TASK_9: str = "gate"
REJECT_REASON_LABEL_DEGENERACY_TASK_9: str = "label_degeneracy"
REJECT_REASON_PARSE_FAILED_TASK_9: str = "parse_failed"

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1: The per-date, per-factor admissibility gate
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1, Helper 1: Strict-True mask over a (nullable) boolean Series.
# -------------------------------------------------------------------------------------------------------------------------------

def _strict_true_mask_task_9(
    series: pd.Series,
    name: str,
) -> np.ndarray:
    """
    Materialise a strict-True Boolean array from a boolean Series.

    Convert a ``bool`` or nullable ``boolean`` Series into a plain
    ``numpy.bool_`` vector in which ``pd.NA`` and ``None`` count as
    not-True. No missing flag is ever coerced up to ``True``: the
    strict test mirrors the endpoint-tradability predicate applied to
    labels.

    Parameters
    ----------
    series : pd.Series
        A ``bool`` or nullable ``boolean`` Series of length n.
    name : str
        Column name used in the error messages.

    Returns
    -------
    np.ndarray
        A 1-D ``numpy.bool_`` array of shape ``(n,)``; positions
        that were ``pd.NA`` or ``None`` are ``False``.

    Raises
    ------
    TypeError
        If ``series`` is not a ``pandas.Series``, or its dtype is
        neither ``bool`` nor nullable ``boolean``.
    """
    # Enforce Series type at the boundary of the helper.
    if not isinstance(series, pd.Series):
        # Reject non-Series inputs explicitly.
        raise TypeError(f"{name} must be a pandas.Series, got {type(series)!r}.")

    # Accept only booleans; anything else is a structural defect, never a silent cast.
    if not pd.api.types.is_bool_dtype(series.dtype):
        # Reject non-boolean dtypes explicitly; P22 pins nullable boolean for flags.
        raise TypeError(f"{name} must be a bool or nullable-boolean Series, got dtype {series.dtype!r}.")

    # Fill pd.NA with False (strict not-True semantics) and cast to numpy bool.
    return series.fillna(False).astype(bool).to_numpy(dtype=np.bool_)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1, Helper 2: Normalise the mask register into a per-date set map.
# -------------------------------------------------------------------------------------------------------------------------------

def _mask_coordinate_map_task_9(
    mask_register: pd.DataFrame,
) -> Dict[pd.Timestamp, set]:
    """
    Build a date-to-masked-stock-code lookup from the mask register.

    Read the ``stock_code`` and ``trade_date`` columns of the register
    and group the masked stock codes by trading date, so that the
    predicate ``(i, t) not in mask register`` becomes a set-membership
    test per (expression, date). Stock codes are normalised to plain
    ``str`` for hash-stable membership; dates are normalised to
    ``pd.Timestamp`` keys.

    Parameters
    ----------
    mask_register : pd.DataFrame
        The mask register; must carry ``stock_code`` and
        ``trade_date`` columns. An empty frame with those columns is
        legal and yields an empty map (no exclusions).

    Returns
    -------
    Dict[pd.Timestamp, set]
        Per-date set of masked stock codes (each code a plain
        ``str``); dates absent from the register are absent from the
        mapping.

    Raises
    ------
    TypeError
        If ``mask_register`` is not a ``pandas.DataFrame``, or its
        ``trade_date`` column is not datetime-typed.
    ValueError
        If ``stock_code`` or ``trade_date`` is missing, or any
        coordinate cell is NA; the message lists the offending row
        labels.
    """
    # Enforce DataFrame type at the boundary of the helper.
    if not isinstance(mask_register, pd.DataFrame):
        # Reject non-DataFrame mask registers explicitly.
        raise TypeError(f"mask_register must be a pandas.DataFrame, got {type(mask_register)!r}.")

    # Compute the missing-columns set up front.
    missing = {"stock_code", "trade_date"} - set(mask_register.columns)

    # Halt when the register does not carry the two coordinate columns.
    if missing:
        # The register cannot enumerate coordinates without both columns.
        raise ValueError(f"mask_register is missing required column(s): {sorted(missing)}.")

    # Extract the coordinate columns as Series.
    stock_col = mask_register["stock_code"]
    date_col = mask_register["trade_date"]

    # Enforce a datetime-like trade_date column; no silent coercion from strings.
    if not pd.api.types.is_datetime64_any_dtype(date_col):
        # Halt: the calendar is the project's coordinate system; dates must be datetime.
        raise TypeError(f"mask_register['trade_date'] must be datetime64, got {date_col.dtype!r}.")

    # Halt on NA coordinates; a mask without coordinates cannot be applied.
    if stock_col.isna().any() or date_col.isna().any():
        # Trace the offending coordinates explicitly.
        bad_rows = mask_register.index[stock_col.isna() | date_col.isna()].tolist()
        raise ValueError(f"mask_register carries NA coordinate(s) at row label(s) {bad_rows}.")

    # Normalise stock codes to plain str for hash-stable set membership.
    stocks = stock_col.astype(str)

    # Materialise the empty map; filled only if the register is non-empty.
    coordinate_map: Dict[pd.Timestamp, set] = {}

    # Walk the register once; group masked stocks per date (registers are tiny).
    for stock_code, trade_date in zip(stocks.tolist(), date_col.tolist()):
        # Normalise each date to a Timestamp key.
        key = pd.Timestamp(trade_date)

        # Create the per-date set on first sight.
        coordinate_map.setdefault(key, set()).add(stock_code)

    # Return the lookup map; the empty register yields the empty map.
    return coordinate_map

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 1: The per-(expression_id, trade_date) admissibility gate.
# -------------------------------------------------------------------------------------------------------------------------------

def date_gate_task_9(
    factor: pd.Series,
    label: pd.Series,
    tradable: pd.Series,
    masked: pd.Series,
) -> dict:
    """
    Evaluate the joint admissibility gate for one factor-date.

    Materialise one eligible set per (expression, date) and evaluate
    the three sub-gates jointly over that shared set:

        n_eligible >= 5
        and n_distinct_signal >= 3
        and n_distinct_label >= 2

    A stock is eligible when all four conjuncts hold:
    ``f[i,t]`` non-null, ``y[i,t]`` non-null, ``tradable[i,t]``
    strictly ``True`` at the date (``pd.NA`` counts as not-True), and
    ``(i, t)`` not in the mask register (``True`` in ``masked``
    excludes it, ``pd.NA`` does not). The distinct-value counts are
    taken over this same eligible set, never over all stocks with a
    factor value.

    Parameters
    ----------
    factor : pd.Series
        The signal vector ``f[., t]`` indexed by stock code;
        numeric, with ``NaN``/``pd.NA`` for absent cells.
    label : pd.Series
        The label vector ``y[., t]`` indexed by stock code;
        numeric, with ``NaN``/``pd.NA`` for absent cells.
    tradable : pd.Series
        Strict-True tradability flag at the date; ``bool`` or
        nullable ``boolean``.
    masked : pd.Series
        Mask flag at the date; ``bool`` or nullable ``boolean``.

    Returns
    -------
    dict
        Keys exactly:

        - ``n_eligible`` : ``int`` -- size of the eligible set.
        - ``n_distinct_signal`` : ``int`` -- distinct ``f`` values
          over the eligible set (0 when empty).
        - ``n_distinct_label`` : ``int`` -- distinct ``y`` values
          over the eligible set (0 when empty).
        - ``admissible`` : ``bool`` -- the joint conjunction.
        - ``reject_reason`` : ``None`` or ``"gate"`` -- ``None``
          iff ``admissible``.
        - ``insufficient_date_gate_result`` : ``None`` always, the
          typed null recorded on every call.

    Raises
    ------
    TypeError
        If any argument is not a ``pandas.Series``, if ``factor`` or
        ``label`` is non-numeric, or if ``tradable``/``masked`` is
        not a boolean Series.
    ValueError
        If the four Series do not share the same index.
    """
    # --- input validation at the public boundary of Step 1 ----------------------------

    # Enforce Series type on the factor vector.
    if not isinstance(factor, pd.Series):
        # Reject non-Series factor inputs explicitly.
        raise TypeError(f"factor must be a pandas.Series, got {type(factor)!r}.")

    # Enforce Series type on the label vector.
    if not isinstance(label, pd.Series):
        # Reject non-Series label inputs explicitly.
        raise TypeError(f"label must be a pandas.Series, got {type(label)!r}.")

    # Enforce Series type on the tradability vector.
    if not isinstance(tradable, pd.Series):
        # Reject non-Series tradable inputs explicitly.
        raise TypeError(f"tradable must be a pandas.Series, got {type(tradable)!r}.")

    # Enforce Series type on the mask vector.
    if not isinstance(masked, pd.Series):
        # Reject non-Series mask inputs explicitly.
        raise TypeError(f"masked must be a pandas.Series, got {type(masked)!r}.")

    # Enforce numeric dtype on the factor vector (float-like incl. Int64).
    if not pd.api.types.is_numeric_dtype(factor.dtype):
        # Halt: a non-numeric signal cannot participate in rank association.
        raise TypeError(f"factor must be numeric, got dtype {factor.dtype!r}.")

    # Enforce numeric dtype on the label vector.
    if not pd.api.types.is_numeric_dtype(label.dtype):
        # Halt: a non-numeric label cannot participate in rank association.
        raise TypeError(f"label must be numeric, got dtype {label.dtype!r}.")

    # Enforce identical index (same stocks, same order) across the four vectors.
    if not (factor.index.equals(label.index) and factor.index.equals(tradable.index) and factor.index.equals(masked.index)):
        # Halt: alignment is by label; positional mixing is a caller defect.
        raise ValueError("factor, label, tradable and masked must share the same index.")

    # --- materialise aligned float64 vectors and strict boolean masks -----------------

    # Convert factor to float64 with NaN in place of NA/NaT (exact null semantics).
    f_arr = factor.to_numpy(dtype=np.float64, na_value=np.nan)

    # Convert label to float64 with NaN in place of NA/NaT.
    y_arr = label.to_numpy(dtype=np.float64, na_value=np.nan)

    # Strict-True tradability mask (pd.NA -> False).
    t_arr = _strict_true_mask_task_9(tradable, "tradable")

    # Strict-True mask mask (pd.NA -> False, i.e. not-masked).
    m_arr = _strict_true_mask_task_9(masked, "masked")

    # --- the four-conjunct eligibility predicate (i)-(iv), per date -------------------

    # Conjunct (i): f[i,t] non-null; conjunct (ii): y[i,t] non-null.
    finite_f = ~np.isnan(f_arr)

    # Conjunct (ii) materialised as the label null test.
    finite_y = ~np.isnan(y_arr)

    # Conjunction of (i)-(iv): factor non-null AND label non-null AND tradable True AND not masked.
    eligible = finite_f & finite_y & t_arr & ~m_arr

    # Cardinality of the eligible set.
    n_eligible = int(np.count_nonzero(eligible))

    # --- distinct-value counts over the SAME eligible set -----------------------------

    # Slice the signal vector to the eligible set.
    f_elig = f_arr[eligible]

    # Slice the label vector to the eligible set.
    y_elig = y_arr[eligible]

    # Distinct signal values; zero when the eligible set is empty.
    n_distinct_signal = int(np.unique(f_elig).size) if f_elig.size else 0

    # Distinct label values; zero when the eligible set is empty.
    n_distinct_label = int(np.unique(y_elig).size) if y_elig.size else 0

    # --- the joint conjunction over the shared eligible set ---------------------------

    # Joint gate: n_eligible >= 5 AND n_distinct_signal >= 3 AND n_distinct_label >= 2.
    admissible = bool(
        n_eligible >= MIN_ELIGIBLE_STOCKS_PER_DATE_TASK_9
        and n_distinct_signal >= MIN_DISTINCT_SIGNAL_VALUES_TASK_9
        and n_distinct_label >= MIN_DISTINCT_LABEL_VALUES_TASK_9
    )

    # Reject reason: None iff admissible; "gate" otherwise (Step 3 vocabulary).
    reject_reason = None if admissible else REJECT_REASON_GATE_TASK_9

    # Typed null, recorded ALWAYS: None (insufficient_date_gate_result, P02 [S]).
    insufficient_date_gate_result = None

    # --- assemble the per-date verdict ----------------------------------------------

    return {
        # Cardinality of E(f,t); joint with the sub-gates below.
        "n_eligible": n_eligible,
        # Distinct signal values over E(f,t); joint sub-gate 2.
        "n_distinct_signal": n_distinct_signal,
        # Distinct label values over E(f,t); joint sub-gate 3.
        "n_distinct_label": n_distinct_label,
        # The joint conjunction itself.
        "admissible": admissible,
        # None iff admissible; else the failing gate bucket.
        "reject_reason": reject_reason,
        # Typed null (None), disclosed as [S] from P02; never a string sentinel.
        "insufficient_date_gate_result": insufficient_date_gate_result,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 2: The mid-rank Pearson correlation on each admissible date
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 2: Average-tie mid-rank Pearson, hand-written
# (the primary audit instrument; numpy.corrcoef / Series.corr / pearsonr are
# forbidden on this path).
# -------------------------------------------------------------------------------------------------------------------------------

def mid_rank_pearson_task_9(
    signal: pd.Series,
    label: pd.Series,
) -> float:
    """
    Compute the Pearson correlation of average-tie mid-ranks.

    Compute the daily rank-IC for one admissible factor-date. Each
    vector is converted to ascending average-tie mid-ranks, then the
    Pearson correlation of the two rank vectors is formed by hand:

        rho = sum(dx * dy) / sqrt(sum(dx**2) * sum(dy**2))

    with ``dx = rx - mean(rx)`` and ``dy = ry - mean(ry)`` using
    population means. The denominator is double-guarded:
    zero-variance ranks, a non-positive or non-finite variance
    product, a zero or non-finite denominator, or a non-finite
    ``rho`` all return ``nan`` rather than a fabricated ``+/-1``.

    Parameters
    ----------
    signal : pd.Series
        The signal vector ``f[., t]`` on the eligible set; numeric,
        null-free, indexed identically to ``label``.
    label : pd.Series
        The label vector ``y[., t]`` on the eligible set; numeric,
        null-free, indexed identically to ``signal``.

    Returns
    -------
    float
        The rank correlation in ``[-1, 1]`` as a plain Python
        ``float``, or ``nan`` under the double guard (including when
        fewer than two observations are supplied).

    Raises
    ------
    TypeError
        If either argument is not a ``pandas.Series`` or is
        non-numeric.
    ValueError
        If the two Series differ in index, or either carries
        ``NaN``/``NA`` values; the message lists the offending
        positions.
    """
    # --- input validation at the public boundary of Step 2 ----------------------------

    # Enforce Series type on the signal vector.
    if not isinstance(signal, pd.Series):
        # Reject non-Series signal inputs explicitly.
        raise TypeError(f"signal must be a pandas.Series, got {type(signal)!r}.")

    # Enforce Series type on the label vector.
    if not isinstance(label, pd.Series):
        # Reject non-Series label inputs explicitly.
        raise TypeError(f"label must be a pandas.Series, got {type(label)!r}.")

    # Enforce numeric dtype on the signal vector.
    if not pd.api.types.is_numeric_dtype(signal.dtype):
        # Halt: rank association requires ordered numeric values.
        raise TypeError(f"signal must be numeric, got dtype {signal.dtype!r}.")

    # Enforce numeric dtype on the label vector.
    if not pd.api.types.is_numeric_dtype(label.dtype):
        # Halt: rank association requires ordered numeric values.
        raise TypeError(f"label must be numeric, got dtype {label.dtype!r}.")

    # Enforce identical index (same order) across the two vectors.
    if not signal.index.equals(label.index):
        # Halt: positional mixing of independently ordered vectors is a caller defect.
        raise ValueError("signal and label must share the same index.")

    # Halt on NaN/NA in the signal vector; the eligible set is null-free by construction.
    if signal.isna().any():
        # Trace the offending positions explicitly.
        bad = signal.index[signal.isna()].tolist()
        raise ValueError(f"signal carries NaN/NA at position(s) {bad}; pass the eligible subset.")

    # Halt on NaN/NA in the label vector.
    if label.isna().any():
        # Trace the offending positions explicitly.
        bad = label.index[label.isna()].tolist()
        raise ValueError(f"label carries NaN/NA at position(s) {bad}; pass the eligible subset.")

    # --- materialise float64 vectors -------------------------------------------------

    # Convert signal to float64 (all values finite: no NaN survived validation).
    x = signal.to_numpy(dtype=np.float64)

    # Convert label to float64.
    y = label.to_numpy(dtype=np.float64)

    # A two-element-or-longer vector is required for a defined correlation.
    if x.shape[0] < 2:
        # The denominator is zero by construction; return NaN conservatively.
        return float("nan")

    # --- average-tie mid-ranks, ascending (P02: AVERAGE_MID_RANK, ASCENDING) ---------

    # r(x_j) = (1 + |{k : x_k < x_j}| + |{k : x_k <= x_j}|) / 2, ties share the
    # average of their positions; ascending orientation; na_option='keep' is
    # moot (validated null-free) but pinned per the step text.
    rx = signal.rank(method="average", ascending=True, na_option="keep").to_numpy(dtype=np.float64)

    # Mid-ranks of the label vector under the identical rule.
    ry = label.rank(method="average", ascending=True, na_option="keep").to_numpy(dtype=np.float64)

    # --- Pearson on ranks: means, deviations, the three sums ---------------------------

    # Rank mean (population, ddof == 0); Pearson uses the empirical rank mean.
    rx_bar = float(np.mean(rx))

    # Rank mean of the label vector.
    ry_bar = float(np.mean(ry))

    # Deviations dx = rx - r_xbar.
    dx = rx - rx_bar

    # Deviations dy = ry - r_ybar.
    dy = ry - ry_bar

    # Numerator sum(dx * dy).
    num = float(np.dot(dx, dy))

    # Sum of squared signal deviations sum(dx**2).
    sxx = float(np.dot(dx, dx))

    # Sum of squared label deviations sum(dy**2).
    syy = float(np.dot(dy, dy))

    # --- double guard: zero variance / zero or NaN denominator -> NaN -----------------

    # Guard 1: either rank vector has zero (or degenerate) variance.
    if not (np.isfinite(sxx) and np.isfinite(syy)) or sxx == 0.0 or syy == 0.0:
        # Zero-variance ranks: undefined correlation; never +/-1 by fiat.
        return float("nan")

    # Form the product sum(dx**2) * sum(dy**2).
    denom_sq = sxx * syy

    # Guard 2a: the product must be a finite positive number.
    if not np.isfinite(denom_sq) or denom_sq <= 0.0:
        # Degenerate denominator; return NaN rather than a fabricated sign.
        return float("nan")

    # Denominator sqrt(sum(dx**2) * sum(dy**2)).
    denom = math.sqrt(denom_sq)

    # Guard 2b: the denominator itself must be finite and non-zero.
    if not np.isfinite(denom) or denom == 0.0:
        # Zero or non-finite denominator; return NaN.
        return float("nan")

    # rho = sum(dx * dy) / sqrt(sum(dx**2) * sum(dy**2)).
    rho = num / denom

    # Final guard: a non-finite rho indicates overflow in the sums; NaN is honest.
    if not np.isfinite(rho):
        # Overflow/NaN propagation; return NaN rather than an inf artifact.
        return float("nan")

    # Return the correlation as a plain Python float.
    return float(rho)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 3: The long-format DAILY_SCORES score register
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 3, Helper: Vectorised gate-compliance assertions over the
# register (halt on any violation; no per-row Python loop).
# -------------------------------------------------------------------------------------------------------------------------------

def _assert_gate_compliance_task_9(
    register: pd.DataFrame,
) -> None:
    """
    Assert the gate-compliance predicates on a daily-score register.

    Apply six vectorised checks without looping and halt on the first
    violated predicate:

    1. every admissible row satisfies ``n_eligible >= 5``,
       ``n_distinct_signal >= 3`` and ``n_distinct_label >= 2``;
    2. ``rho`` is finite on every admissible row;
    3. ``rho`` is NaN on every inadmissible row;
    4. ``reject_reason`` is null if and only if ``admissible`` is
       ``True``;
    5. every inadmissible row carries a non-null ``reject_reason``;
    6. the ``MultiIndex`` on (``expression_id``, ``trade_date``) is
       unique.

    Parameters
    ----------
    register : pd.DataFrame
        The constructed register, indexed by
        ``MultiIndex([expression_id, trade_date])`` and carrying the
        count columns, ``rho``, ``admissible`` and ``reject_reason``.

    Raises
    ------
    AssertionError
        If any predicate fails. Only the sub-gate violation message
        names the violating row count (plus up to ten offending
        index keys); the rho and ``reject_reason`` violations name
        up to ten offending index keys but no count; the
        index-uniqueness violation names neither a count nor index
        keys.
    """
    # Extract the admissible mask as a numpy boolean vector.
    admissible = register["admissible"].to_numpy(dtype=np.bool_)

    # Compute the joint sub-gate predicate elementwise (nullable Int64 compares
    # to boolean with NA possible only where counts are NA, which can happen
    # solely on admissible-False rows); fillna(False) keeps the predicate honest.
    sub_gate = (
        (register["n_eligible"] >= MIN_ELIGIBLE_STOCKS_PER_DATE_TASK_9)
        & (register["n_distinct_signal"] >= MIN_DISTINCT_SIGNAL_VALUES_TASK_9)
        & (register["n_distinct_label"] >= MIN_DISTINCT_LABEL_VALUES_TASK_9)
    ).fillna(False).to_numpy(dtype=np.bool_)

    # Predicate 1: admissible == True implies the joint sub-gate holds.
    admissible_violations = admissible & ~sub_gate

    # Halt if any admissible row fails its own sub-gates (a construction bug).
    if bool(admissible_violations.any()):
        # Collect the offending coordinates for the diagnostic message.
        offenders = register.index[admissible_violations].tolist()[:10]
        # Raise with the precise offender list; halts the build.
        raise AssertionError(
            f"gate-compliance violation: {int(admissible_violations.sum())} admissible row(s) "
            f"violate n_eligible >= {MIN_ELIGIBLE_STOCKS_PER_DATE_TASK_9} & "
            f"n_distinct_signal >= {MIN_DISTINCT_SIGNAL_VALUES_TASK_9} & "
            f"n_distinct_label >= {MIN_DISTINCT_LABEL_VALUES_TASK_9}; "
            f"offending index keys: {offenders}."
        )

    # rho vector as float64 for the finiteness test.
    rho_values = register["rho"].to_numpy(dtype=np.float64)

    # Predicate 2: admissible == True implies rho is finite.
    if bool(admissible.any()):
        # Extract the rho values of admissible rows only.
        rho_admissible = rho_values[admissible]

        # Every admissible rho must be a finite float (not NaN, not +/-inf).
        if not bool(np.isfinite(rho_admissible).all()):
            # Collect the offending coordinates for the diagnostic message.
            bad = register.index[admissible & ~np.isfinite(rho_values)].tolist()[:10]
            # Raise with the precise offender list; halts the build.
            raise AssertionError(
                f"gate-compliance violation: non-finite rho on admissible row(s); "
                f"offending index keys: {bad}."
            )

    # Predicate 3: admissible == False implies rho is NaN.
    inadmissible = ~admissible

    # Compute the all-NaN predicate over inadmissible rows once.
    inadmissible_rho_all_na = bool(register["rho"].isna().to_numpy()[inadmissible].all()) if bool(inadmissible.any()) else True

    # Halt if any inadmissible row carries a non-NaN rho.
    if bool(inadmissible.any()) and not inadmissible_rho_all_na:
        # Collect the offending coordinates for the diagnostic message.
        bad = register.index[inadmissible & ~register["rho"].isna()].tolist()[:10]
        # Raise with the precise offender list; halts the build.
        raise AssertionError(
            f"gate-compliance violation: non-NaN rho on inadmissible row(s); "
            f"offending index keys: {bad}."
        )

    # Predicate 4: reject_reason is None iff admissible (biconditional).
    reject_isna = register["reject_reason"].isna().to_numpy(dtype=np.bool_)

    # Halt on any mismatch between the admissible flag and reject_reason.isna().
    if not bool((reject_isna == admissible).all()):
        # Collect the offending coordinates for the diagnostic message.
        bad = register.index[reject_isna != admissible].tolist()[:10]
        # Raise with the precise offender list; halts the build.
        raise AssertionError(
            f"gate-compliance violation: reject_reason is None iff admissible is True; "
            f"offending index keys: {bad}."
        )

    # Predicate 5: reject_reason names a failure on every inadmissible row.
    if bool(inadmissible.any()) and not bool(register["reject_reason"].notna().to_numpy()[inadmissible].all()):
        # Collect the offending coordinates for the diagnostic message.
        bad = register.index[inadmissible & ~register["reject_reason"].notna()].tolist()[:10]
        # Raise with the precise offender list; halts the build.
        raise AssertionError(
            f"gate-compliance violation: inadmissible row(s) without a reject_reason; "
            f"offending index keys: {bad}."
        )

    # Index uniqueness is a Step 3 invariant; assert it as a final structural check.
    if not bool(register.index.is_unique):
        # Halt: duplicated (expression_id, trade_date) cells void the register.
        raise AssertionError("gate-compliance violation: the register index is not unique.")

    # All predicates passed; return None (halt-on-violation semantics).
    return None

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Step 3: The DAILY_SCORES register (full Cartesian product).
# -------------------------------------------------------------------------------------------------------------------------------

def build_daily_scores_task_9(
    factor_panel: pd.DataFrame,
    labels: pd.Series,
    tradable: pd.Series,
    mask_register: pd.DataFrame,
    expression_ids: Sequence[str],
    trade_dates: Sequence[pd.Timestamp],
    evaluator_version: str,
    parse_failed_ids: Sequence[str] = (),
) -> pd.DataFrame:
    """
    Build the long-format ``DAILY_SCORES`` register for every cell.

    Enumerates the full ``expression_ids`` x ``trade_dates`` Cartesian
    product (expression-major, date-ascending) and scores each
    ``(expression_id, trade_date)`` cell independently. For one cell the
    factor panel, label and tradability vectors are grouped by
    ``trade_date`` and reindexed onto the sorted union of the stocks
    that appear on that date in any of the three inputs; holes become
    NaN (or ``pd.NA`` for the flag). The eligible set is the conjunction
    of a non-null signal, a non-null label, a strict-``True``
    tradability flag and absence from the mask register for that date.

    The joint gate of Task 9, Step 1 then requires ``n_eligible >= 5``,
    ``n_distinct_signal >= 3`` and ``n_distinct_label >= 2`` on that
    same eligible set; failing cells get NaN ``rho`` and ``reject_reason
    == "gate"``. Admissible cells are scored with the average-tie
    mid-rank Pearson of Task 9, Step 2 on the identical eligible set; a
    zero or NaN correlation denominator yields NaN ``rho`` and
    ``reject_reason == "label_degeneracy"``. Ids in ``parse_failed_ids``
    are never evaluated and carry NA counts with ``"parse_failed"``. A
    date absent from all three panels yields zero counts, NaN ``rho``
    and ``"gate"``. No aggregation across dates is performed (that is
    Task 10).

    Parameters
    ----------
    factor_panel : pd.DataFrame
        Signal panel indexed by ``MultiIndex([stock_code, trade_date])``
        with one column per scorable expression; NaN marks absent cells.
        Row order is not assumed; all alignment is by label.
    labels : pd.Series
        Five-row forward simple-return labels ``y[i, t]`` on the same
        compound index; NaN where undefined.
    tradable : pd.Series
        Tradability flag at ``t`` on the same compound index; only
        strict ``True`` is eligible, and NA counts as not tradable.
    mask_register : pd.DataFrame
        Mask register carrying ``stock_code`` and ``trade_date``
        columns; may be empty.
    expression_ids : Sequence[str]
        Ordered expression identifiers (major axis). Must be unique and
        every element must be ``str``.
    trade_dates : Sequence[pd.Timestamp]
        Ordered label-available dates (minor axis). Must be unique;
        every element is coerced with ``pd.Timestamp``.
    evaluator_version : str
        Non-blank version tag recorded verbatim in every row.
    parse_failed_ids : Sequence[str], optional
        Ids whose expression text failed to parse; recorded with NA
        counts and NaN ``rho``. Default is ``()``.

    Returns
    -------
    pd.DataFrame
        Indexed by ``MultiIndex([expression_id, trade_date])``, rows in
        expression-major, date-ascending order, with columns ``rho``
        (float64), ``n_eligible`` (Int64), ``n_distinct_signal``
        (Int64), ``n_distinct_label`` (Int64, NA exactly on parse-failed
        rows), ``admissible`` (bool), ``reject_reason`` (object:
        ``None``, ``"gate"``, ``"label_degeneracy"`` or
        ``"parse_failed"``, ``None`` exactly when admissible) and
        ``evaluator_version`` (object).

    Raises
    ------
    TypeError
        If ``factor_panel``, ``labels``, ``tradable`` or
        ``mask_register`` has the wrong type; if any ``expression_ids``
        or ``parse_failed_ids`` element is not ``str``; if a
        ``trade_dates`` element is not ``pd.Timestamp``-coercible.
    ValueError
        If an index lacks the ``stock_code``/``trade_date`` levels or is
        not unique; if ``expression_ids`` or ``trade_dates`` contains
        duplicates; if ``evaluator_version`` is blank; if a scorable id
        has no column in ``factor_panel``.
    AssertionError
        If the register index is not unique, its length differs from the
        Cartesian cardinality, or the gate-compliance invariants fail.
    """
    # --- input validation at the public boundary of Step 3 ----------------------------

    # Enforce DataFrame type on the factor panel.
    if not isinstance(factor_panel, pd.DataFrame):
        # Reject non-DataFrame panels explicitly.
        raise TypeError(f"factor_panel must be a pandas.DataFrame, got {type(factor_panel)!r}.")

    # Enforce Series type on the label vector.
    if not isinstance(labels, pd.Series):
        # Reject non-Series labels explicitly.
        raise TypeError(f"labels must be a pandas.Series, got {type(labels)!r}.")

    # Enforce Series type on the tradability vector.
    if not isinstance(tradable, pd.Series):
        # Reject non-Series tradability explicitly.
        raise TypeError(f"tradable must be a pandas.Series, got {type(tradable)!r}.")

    # Enforce DataFrame type on the mask register.
    if not isinstance(mask_register, pd.DataFrame):
        # Reject non-DataFrame mask registers explicitly.
        raise TypeError(f"mask_register must be a pandas.DataFrame, got {type(mask_register)!r}.")

    # --- structural validation of the compound indices -------------------------------

    # The factor panel must carry the two-level compound index.
    if not isinstance(factor_panel.index, pd.MultiIndex) or not {"stock_code", "trade_date"} <= set(factor_panel.index.names):
        # Halt: alignment is by (stock_code, trade_date) label, never by position.
        raise ValueError("factor_panel.index must be a MultiIndex carrying 'stock_code' and 'trade_date'.")

    # The labels must carry the two-level compound index.
    if not isinstance(labels.index, pd.MultiIndex) or not {"stock_code", "trade_date"} <= set(labels.index.names):
        # Halt: the label panel shares the (stock_code, trade_date) coordinate system.
        raise ValueError("labels.index must be a MultiIndex carrying 'stock_code' and 'trade_date'.")

    # The tradability vector must carry the two-level compound index.
    if not isinstance(tradable.index, pd.MultiIndex) or not {"stock_code", "trade_date"} <= set(tradable.index.names):
        # Halt: the flag is a stock-day quantity on the same coordinate system.
        raise ValueError("tradable.index must be a MultiIndex carrying 'stock_code' and 'trade_date'.")

    # A duplicate (stock_code, trade_date) cell inflates the cross-sectional denominator.
    if not bool(factor_panel.index.is_unique) or not bool(labels.index.is_unique) or not bool(tradable.index.is_unique):
        # Halt: duplicated cells would double-count stocks in every score.
        raise ValueError("factor_panel, labels and tradable must carry a unique (stock_code, trade_date) index.")

    # --- normalise the enumeration axes ----------------------------------------------

    # Coerce expression_ids to a list of str; enforce type per element.
    expr_ids = list(expression_ids)

    # Halt on any non-str expression id.
    for expr in expr_ids:
        # Identifiers are strings by construction (P22: string dtype policy).
        if not isinstance(expr, str):
            raise TypeError(f"expression_ids must contain only str, got {type(expr)!r}.")

    # Halt on duplicate expression ids (the Cartesian index must be unique).
    if len(set(expr_ids)) != len(expr_ids):
        # Trace the duplicated ids explicitly.
        duplicates = sorted({e for e in expr_ids if expr_ids.count(e) > 1})
        raise ValueError(f"expression_ids contains duplicate(s): {duplicates}.")

    # Coerce trade_dates to Timestamps; enforce datetime-coercibility.
    date_list: List[pd.Timestamp] = []

    # Halt on any non-coercible date element.
    for dt in trade_dates:
        # Timestamps are the calendar's coordinate type; strings are coerced strictly.
        try:
            date_list.append(pd.Timestamp(dt))
        except (TypeError, ValueError) as exc:
            # Halt with the offending element named.
            raise TypeError(f"trade_dates must contain pd.Timestamp-coercible values, got {dt!r}.") from exc

    # Halt on duplicate dates (the Cartesian index must be unique).
    if len(set(date_list)) != len(date_list):
        # Trace the duplicated dates explicitly.
        duplicates = sorted({d for d in date_list if date_list.count(d) > 1})
        raise ValueError(f"trade_dates contains duplicate(s): {duplicates}.")

    # Enforce a non-blank evaluator_version; verbatim pass-through otherwise.
    if not isinstance(evaluator_version, str) or not evaluator_version.strip():
        # Halt: a blank version tag voids the instrument's auditability.
        raise ValueError("evaluator_version must be a non-blank str.")

    # Coerce parse_failed_ids to a set of str for O(1) membership tests.
    parse_failed_set = set(parse_failed_ids)

    # Halt on any non-str parse-failed id.
    for expr in parse_failed_set:
        # Identifiers are strings by construction.
        if not isinstance(expr, str):
            raise TypeError(f"parse_failed_ids must contain only str, got {type(expr)!r}.")

    # --- verify scorable expression ids are present as panel columns -------------------

    # Missing scorable columns would fabricate uncomputed cells silently.
    required_columns = [expr for expr in expr_ids if expr not in parse_failed_set]

    # Compute the missing-column list.
    missing_columns = [expr for expr in required_columns if expr not in factor_panel.columns]

    # Halt when any scorable expression id is absent from the panel.
    if missing_columns:
        # Trace the missing ids explicitly.
        raise ValueError(f"factor_panel is missing column(s) for scorable expression id(s): {missing_columns}.")

    # --- group each input by trade_date exactly once (no per-date rescan) ---------------

    # Per-date factor groups (dict: Timestamp -> DataFrame of that date's rows).
    factor_by_date = dict(iter(factor_panel.groupby(level="trade_date", sort=False)))

    # Per-date label groups (dict: Timestamp -> Series of that date's rows).
    labels_by_date = dict(iter(labels.groupby(level="trade_date", sort=False)))

    # Per-date tradability groups (dict: Timestamp -> Series of that date's rows).
    tradable_by_date = dict(iter(tradable.groupby(level="trade_date", sort=False)))

    # Per-date masked-stocks map (dict: Timestamp -> set(stock_code)).
    mask_by_date = _mask_coordinate_map_task_9(mask_register)

    # --- enumerate the Cartesian product and score every cell -------------------------

    # Accumulate the seven column lists in product order.
    rho_values: List[float] = []
    n_eligible_values: List[Optional[int]] = []
    n_distinct_signal_values: List[Optional[int]] = []
    n_distinct_label_values: List[Optional[int]] = []
    admissible_values: List[bool] = []
    reject_reason_values: List[Optional[str]] = []
    evaluator_version_values: List[str] = []

    # Outer axis: expressions in caller order (major); inner axis: dates in caller order (minor).
    for expr in expr_ids:
        # Skip the computation entirely for parse-failed expressions.
        is_parse_failed = expr in parse_failed_set

        # Inner loop: every requested trade date, in order.
        for date in date_list:
            # Record the parse-failed cell without any computation.
            if is_parse_failed:
                # rho undefined; counts NA; admissibility False; typed reason recorded.
                rho_values.append(float("nan"))
                n_eligible_values.append(None)
                n_distinct_signal_values.append(None)
                n_distinct_label_values.append(None)
                admissible_values.append(False)
                reject_reason_values.append(REJECT_REASON_PARSE_FAILED_TASK_9)
                evaluator_version_values.append(evaluator_version)
                continue

            # Fetch the per-date groups (None when a date is absent from a panel).
            factor_group = factor_by_date.get(date)
            label_group = labels_by_date.get(date)
            tradable_group = tradable_by_date.get(date)

            # Enumerate the sorted union of stocks present on this date (deterministic).
            stock_sets: List[set] = []
            if factor_group is not None:
                # Stocks present at this date in the factor panel.
                stock_sets.append(set(factor_group.index.get_level_values("stock_code")))
            if label_group is not None:
                # Stocks present at this date in the label vector.
                stock_sets.append(set(label_group.index.get_level_values("stock_code")))
            if tradable_group is not None:
                # Stocks present at this date in the tradability vector.
                stock_sets.append(set(tradable_group.index.get_level_values("stock_code")))

            # Materialise the deterministic union stock set for this date.
            stocks = sorted(set().union(*stock_sets)) if stock_sets else []

            # Halt-path: an absent date in ALL THREE panels gates every expression off.
            if not stocks:
                # rho undefined; counts zero; not admissible; reject reason "gate".
                rho_values.append(float("nan"))
                n_eligible_values.append(0)
                n_distinct_signal_values.append(0)
                n_distinct_label_values.append(0)
                admissible_values.append(False)
                reject_reason_values.append(REJECT_REASON_GATE_TASK_9)
                evaluator_version_values.append(evaluator_version)
                continue

            # Slice the per-date factor vector for this expression; reindex onto the union.
            if factor_group is not None:
                # The expression's column for this date, aligned on stock_code, holes -> NaN.
                f_vec = factor_group[expr].droplevel("trade_date").reindex(stocks)
            else:
                # Absent date in the factor panel: all-null vector.
                f_vec = pd.Series(np.nan, index=pd.Index(stocks), dtype=np.float64)

            # Slice the per-date label vector; reindex onto the same union.
            if label_group is not None:
                # The label vector for this date, aligned on stock_code, holes -> NaN.
                y_vec = label_group.droplevel("trade_date").reindex(stocks)
            else:
                # Absent date in the label panel: all-null vector.
                y_vec = pd.Series(np.nan, index=pd.Index(stocks), dtype=np.float64)

            # Slice the per-date tradability flag; normalise to nullable boolean, reindex.
            if tradable_group is not None:
                # The flag vector for this date, aligned on stock_code, holes -> pd.NA.
                t_vec = tradable_group.droplevel("trade_date").astype("boolean").reindex(stocks)
            else:
                # Absent date in the flag panel: all-NA vector (not-True).
                t_vec = pd.Series(pd.NA, index=pd.Index(stocks), dtype="boolean")

            # Build the masked vector for this date from the (date -> stocks) set map.
            masked_stocks = mask_by_date.get(date, set())
            m_vec = pd.Series([s in masked_stocks for s in stocks], index=pd.Index(stocks), dtype=bool)

            # --- Step 1: the joint admissibility gate on this (expression, date) cell ---
            gate = date_gate_task_9(f_vec, y_vec, t_vec, m_vec)

            # Record the three counts regardless of the verdict.
            n_eligible_values.append(gate["n_eligible"])
            n_distinct_signal_values.append(gate["n_distinct_signal"])
            n_distinct_label_values.append(gate["n_distinct_label"])

            # --- cell is gate-failing: rho NaN, admissible False, reject_reason "gate" ---
            if not gate["admissible"]:
                # Counts recorded; rho undefined; reject reason names the gate.
                rho_values.append(float("nan"))
                admissible_values.append(False)
                reject_reason_values.append(REJECT_REASON_GATE_TASK_9)
                evaluator_version_values.append(evaluator_version)
                continue

            # --- Step 2: mid-rank Pearson on the SAME eligible set ----------------------
            f_arr = f_vec.to_numpy(dtype=np.float64, na_value=np.nan)
            y_arr = y_vec.to_numpy(dtype=np.float64, na_value=np.nan)
            t_arr = t_vec.fillna(False).astype(bool).to_numpy(dtype=np.bool_)
            m_arr = m_vec.to_numpy(dtype=np.bool_)
            eligible = ~np.isnan(f_arr) & ~np.isnan(y_arr) & t_arr & ~m_arr
            rho = mid_rank_pearson_task_9(
                pd.Series(f_arr[eligible]),
                pd.Series(y_arr[eligible]),
            )

            # --- cell passes the gate but the rank denominator vanished ------------------
            if math.isnan(rho):
                # Label degeneracy: admissible False, rho NaN, reject_reason distinct from gate.
                rho_values.append(float("nan"))
                admissible_values.append(False)
                reject_reason_values.append(REJECT_REASON_LABEL_DEGENERACY_TASK_9)
                evaluator_version_values.append(evaluator_version)
                continue

            # --- cell passes both steps: finite rho, admissible True ---------------------
            rho_values.append(float(rho))
            admissible_values.append(True)
            reject_reason_values.append(None)
            evaluator_version_values.append(evaluator_version)

    # --- materialise the register against the Cartesian product index -------------------

    # Cartesian index: expression-major, date-ascending; names pinned per the contract.
    product_index = pd.MultiIndex.from_product(
        [expr_ids, date_list],
        names=["expression_id", "trade_date"],
    )

    # Assemble the register in one shot (single allocation).
    register = pd.DataFrame(index=product_index)

    # rho: float64, expression-major then date-ascending.
    register["rho"] = np.asarray(rho_values, dtype=np.float64)

    # Counts: nullable Int64, NA exactly on parse_failed rows.
    register["n_eligible"] = pd.array(n_eligible_values, dtype="Int64")
    register["n_distinct_signal"] = pd.array(n_distinct_signal_values, dtype="Int64")
    register["n_distinct_label"] = pd.array(n_distinct_label_values, dtype="Int64")

    # Admissibility verdict: numpy bool (nullable semantics never required here).
    register["admissible"] = np.asarray(admissible_values, dtype=np.bool_)

    # Reject reason: object column; None iff admissible; the typed nulls stay None.
    register["reject_reason"] = pd.Series(reject_reason_values, index=product_index, dtype=object)

    # Evaluator version: verbatim str broadcast across all rows.
    register["evaluator_version"] = pd.Series(evaluator_version_values, index=product_index, dtype=object)

    # --- enforce structural invariants before returning --------------------------------

    # The index must be unique (asserted again inside the compliance helper).
    if not bool(register.index.is_unique):
        # Halt: duplicated product cells void the audit artifact.
        raise AssertionError("DAILY_SCORES index is not unique after construction.")

    # The row count must equal the Cartesian cardinality exactly.
    if len(register) != len(expr_ids) * len(date_list):
        # Halt: a short register silently drops cells.
        raise AssertionError(
            f"DAILY_SCORES has {len(register)} rows; expected {len(expr_ids) * len(date_list)}."
        )

    # Run the vectorised gate-compliance assertions (halt on any violation).
    _assert_gate_compliance_task_9(register)

    # Return the register; aggregation is Task 10's scope, never this module's.
    return register

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 9, Orchestrator: Build, persist and return the DAILY_SCORES register.
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_9(
    factor_panel: pd.DataFrame,
    labels: pd.Series,
    tradable: pd.Series,
    mask_register: pd.DataFrame,
    expression_ids: Sequence[str],
    trade_dates: Sequence[pd.Timestamp],
    evaluator_version: str,
    output_dir: Union[str, Path],
    parse_failed_ids: Sequence[str] = (),
) -> pd.DataFrame:
    """
    Build, persist and return the ``DAILY_SCORES`` register.

    Validates and creates ``output_dir``, then delegates the whole
    construction to :func:`build_daily_scores_task_9`. Writes the register
    to ``DAILY_SCORES.csv`` beneath that directory with the compound index
    serialised (UTF-8, LF terminator), reads the written bytes back, and
    writes ``DAILY_SCORES.sha256`` beside it as a single
    ``sha256sum``-compatible line ``"<hex digest>  DAILY_SCORES.csv"``. The
    digest therefore attests the on-disk artifact, not the in-memory frame.
    Returns the same object that was written.

    Parameters
    ----------
    factor_panel : pd.DataFrame
        Signal panel indexed by ``MultiIndex([stock_code,
        trade_date])``.
    labels : pd.Series
        Forward-return labels on the same compound index.
    tradable : pd.Series
        Tradability flag on the same compound index.
    mask_register : pd.DataFrame
        Mask register with ``stock_code`` and ``trade_date`` columns;
        may be empty.
    expression_ids : Sequence[str]
        Ordered expression identifiers (major axis).
    trade_dates : Sequence[pd.Timestamp]
        Ordered trade dates (minor axis).
    evaluator_version : str
        Non-blank version tag recorded verbatim in every row.
    output_dir : Union[str, Path]
        Directory receiving ``DAILY_SCORES.csv`` and
        ``DAILY_SCORES.sha256``; created with parents if missing.
    parse_failed_ids : Sequence[str], optional
        Ids whose expression text failed to parse. Default is ``()``.

    Returns
    -------
    pd.DataFrame
        The register built by ``build_daily_scores_task_9``: indexed by
        ``MultiIndex([expression_id, trade_date])`` with columns
        ``rho``, ``n_eligible``, ``n_distinct_signal``,
        ``n_distinct_label``, ``admissible``, ``reject_reason`` and
        ``evaluator_version``.

    Raises
    ------
    TypeError
        If ``output_dir`` is neither ``str`` nor ``Path``; validation
        errors raised by ``build_daily_scores_task_9`` propagate.
    ValueError
        If ``output_dir`` exists and is not a directory.
    OSError
        If the directory cannot be created or an artifact cannot be
        written or read back.
    """
    # --- boundary validation of output_dir (the only argument this layer owns) ----------

    # Enforce str/PathLike type on output_dir.
    if not isinstance(output_dir, (str, Path)):
        # Reject non-path output_dir explicitly.
        raise TypeError(f"output_dir must be a str or pathlib.Path, got {type(output_dir)!r}.")

    # Resolve the output directory path.
    out_dir = Path(output_dir)

    # An existing file is not a directory; fail before any scoring runs.
    if out_dir.exists() and not out_dir.is_dir():
        # Halt: refusing to silently clobber a file artifact.
        raise ValueError(f"output_dir points at an existing file: {out_dir}.")

    # Create the directory if missing (parents enabled; existing dir is fine).
    out_dir.mkdir(parents=True, exist_ok=True)

    # --- build the register via the Step 3 callable (never an inline re-implementation) -

    # Delegate entirely to the step callable; stages resolve to the defined callables.
    register = build_daily_scores_task_9(
        factor_panel=factor_panel,
        labels=labels,
        tradable=tradable,
        mask_register=mask_register,
        expression_ids=expression_ids,
        trade_dates=trade_dates,
        evaluator_version=evaluator_version,
        parse_failed_ids=parse_failed_ids,
    )

    # --- persist DAILY_SCORES.csv beneath output_dir ----------------------------------

    # The CSV path beneath the output directory.
    csv_path = out_dir / "DAILY_SCORES.csv"

    # Write the register with its index (row order preserved; LF terminator; utf-8).
    register.to_csv(csv_path, index=True, encoding="utf-8", lineterminator="\n")

    # --- stamp the SHA-256 of the written artifact beside it ---------------------------

    # Read back the exact bytes that were written (not the in-memory frame).
    csv_bytes = csv_path.read_bytes()

    # Compute the digest with the pinned upstream helper (imported, not re-implemented).
    digest = _task_1._sha256_hex(csv_bytes)

    # The sidecar path beneath the output directory.
    digest_path = out_dir / "DAILY_SCORES.sha256"

    # Write one line: "<hex digest>  DAILY_SCORES.csv" (sha256sum-compatible layout).
    digest_path.write_text(f"{digest}  DAILY_SCORES.csv\n", encoding="utf-8", newline="\n")

    # --- return the register for in-memory consumers -----------------------------------

    # Return the register; aggregation is Task 10's scope, never this module's.
    return register
```

```python
# Task 10: Factor Aggregation and Ledger Assignment

# ==============================================================================
# Task 10: Factor Aggregation and Ledger Assignment
# ==============================================================================

# The pinned precedence triples (P04_LEDGER_PARAMETERS["state_precedence"],
# source tag [R] NOT_STATED_IN_PAPER__[R]). They are module-level constants so
# that every public callable consumes the same definition; the reverse triple
# quantifies the [R] sensitivity disclosed by the configuration.
PRIMARY_PRECEDENCE_TASK_10: tuple = ("unsupported", "insufficient", "scorable")
REVERSE_PRECEDENCE_TASK_10: tuple = ("scorable", "insufficient", "unsupported")

# The exact state registry mandated by P04.states; anything outside it is a
# construction failure, not a "close enough" bucket.
STATE_REGISTRY_TASK_10: frozenset = frozenset({"scorable", "insufficient", "unsupported"})

# The three claim-record gate identifiers for the structural verifier (the
# four G-gates per P09/G1-G4 as pinned in the standing conventions).
CLAIM_GATE_KEYS_TASK_10: tuple = ("G1", "G2", "G3", "G4")


# ---------------------------------------------------------------- refusal ---
# A tiny shared canonical-JSON serialiser restricted to the claim-record hash
# payloads. It follows the pinned canonical-JSON corpus conventions of Task 1
# (NaN renders as the literal token NaN; sets sorted; keys sorted; no spaces)
# narrowed to what this module emits (str/int/float/bool/None/list/dict).
def _canonical_json_bytes_task_10(payload: Any) -> bytes:
    """
    Serialise a claim payload to canonical JSON bytes.

    Recursively normalises ``payload``, then dumps it with sorted keys,
    the ``(",", ":")`` separators and ``ensure_ascii=True`` and encodes
    the result as UTF-8. Floats that are NaN or infinite are replaced by
    the literal strings ``"NaN"``, ``"Infinity"`` and ``"-Infinity"``
    before serialisation, so ``allow_nan=False`` never fires. ``None``,
    ``bool``, ``int`` and ``str`` leaves all pass through one branch
    unchanged and ``json.dumps`` renders the bools as JSON booleans;
    tuples become JSON arrays.

    Parameters
    ----------
    payload : Any
        JSON-admissible tree restricted to ``None``, ``bool``, ``int``,
        ``float``, ``str``, ``list``, ``tuple`` and ``dict`` with string
        keys.

    Returns
    -------
    bytes
        UTF-8 bytes of the canonical serialisation.

    Raises
    ------
    TypeError
        If a tree node is a mapping with a non-string key, or any other
        unsupported type; raised by the nested normaliser.
    """
    # Canonicalise the payload locally; unsupported leaf values raise TypeError.
    def _normalise(value: Any) -> Any:
        """
        Replace non-finite floats and reject unsupported leaves.

        Parameters
        ----------
        value : Any
            One node of the payload tree.

        Returns
        -------
        Any
            The node with NaN and +/-infinite floats replaced by their
            literal string tokens, mappings and sequences rebuilt
            recursively, and ``None``/``bool``/``int``/``str`` leaves
            passed through unchanged.

        Raises
        ------
        TypeError
            If the node is a mapping with a non-string key, or any other
            unsupported type.
        """
        # bool is a subclass of int; JSON distinguishes them, preserve both.
        if value is None or isinstance(value, (bool, int, str)):
            # Scalars pass through unchanged.
            return value
        # Floats need the corpus' literal-token handling for non-finite values.
        if isinstance(value, float):
            # NaN renders as the corpus-pinned token.
            if math.isnan(value):
                return "NaN"
            # +/-inf renders as the corpus' infinity tokens.
            if math.isinf(value):
                return "Infinity" if value > 0 else "-Infinity"
            # Finite floats stay finite floats.
            return value
        # Dicts recurse with string keys (corpus requires str keys).
        if isinstance(value, dict):
            # Normalise every value; reject non-str keys explicitly.
            normalised: dict = {}
            for key, item in value.items():
                # Reject non-string mapping keys (corpus demands str keys).
                if not isinstance(key, str):
                    raise TypeError(f"canonical JSON requires str keys; got {type(key)!r}.")
                # Recurse into the value slot.
                normalised[key] = _normalise(item)
            # Return the normalised mapping.
            return normalised
        # Lists and tuples recurse elementwise.
        if isinstance(value, (list, tuple)):
            # Normalise every element.
            return [_normalise(item) for item in value]
        # Anything else is outside the permitted vocabulary; refuse loudly.
        raise TypeError(f"canonical JSON payload carries unsupported type {type(value)!r}.")

    # Serialise with sorted keys and no whitespace, then encode to UTF-8.
    return json.dumps(
        _normalise(payload),
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=True,
        allow_nan=False,
    ).encode("utf-8")

def _sha256_hex_task_10(payload: bytes) -> str:
    """
    Compute the lowercase hex SHA-256 digest of a byte payload.

    Parameters
    ----------
    payload : bytes
        Bytes to digest. A ``bytearray`` is also accepted and converted
        to ``bytes`` before hashing.

    Returns
    -------
    str
        64 lowercase hexadecimal characters.

    Raises
    ------
    TypeError
        If ``payload`` is neither ``bytes`` nor ``bytearray``.
    """
    # Validate the payload type eagerly.
    if not isinstance(payload, (bytes, bytearray)):
        # Refuse non-byte payloads explicitly.
        raise TypeError(f"payload must be bytes, got {type(payload)!r}.")
    # Return the lowercase hex digest.
    return hashlib.sha256(bytes(payload)).hexdigest()

def _make_claim_record_task_10(
    claim_id: str,
    tier: str,
    evidence_source: str,
    verdict: str,
    value: Any,
    note: str,
) -> dict:
    """
    Assemble one hashed ``ClaimRecord``-shaped dict.

    Builds the record from the six supplied fields plus ``gate_status``,
    whose keys come from ``CLAIM_GATE_KEYS_TASK_10`` and are all set to
    True (the closure is a construction invariant of this run). The
    ``evidence_hash`` field is filled last with the SHA-256 of the
    canonical JSON of the record without that field. No field value is
    checked or coerced.

    Parameters
    ----------
    claim_id : str
        Stable claim identifier, e.g.
        ``"STRUCTURAL__[D]__verifiable_on_this_run"``.
    tier : str
        Evidentiary tier token, e.g. ``"reproduction"``.
    evidence_source : str
        Where the evidence lives, e.g. ``"cohort_scored_this_run"``.
    verdict : str
        Short machine-checkable verdict token.
    value : Any
        The claim payload; a scalar or a list.
    note : str
        Free-text provenance note.

    Returns
    -------
    dict
        Record with keys ``claim_id``, ``tier``, ``evidence_source``,
        ``gate_status`` (dict of gate name to True), ``verdict``,
        ``value``, ``note`` and ``evidence_hash`` (64 lowercase hex
        characters).

    Raises
    ------
    TypeError
        If ``value`` holds a payload the canonical serialiser cannot
        encode (propagated from the hashing helpers).
    """
    # Build the unhashed record first (the hash covers everything else).
    record: dict = {
        "claim_id": claim_id,
        "tier": tier,
        "evidence_source": evidence_source,
        "gate_status": {gate: True for gate in CLAIM_GATE_KEYS_TASK_10},
        "verdict": verdict,
        "value": value,
        "note": note,
    }
    # Hash the record without evidence_hash (covers every other field).
    digest = _sha256_hex_task_10(_canonical_json_bytes_task_10(record))
    # Attach the digest and return the completed claim record.
    record["evidence_hash"] = digest
    return record

def _materialise_flat_columns_task_10(frame: pd.DataFrame, caller: str) -> pd.DataFrame:
    """
    Return a copy of a DAILY_SCORES-like frame with flat columns.

    If ``expression_id`` is not already a column but is one of the
    frame's ``MultiIndex`` names, the index is reset into columns;
    otherwise the frame is copied unchanged. The caller's object is
    never mutated.

    Parameters
    ----------
    frame : pd.DataFrame
        A flat register carrying an ``expression_id`` column, or a
        register indexed by a ``MultiIndex`` that includes
        ``expression_id`` among its names.
    caller : str
        Calling function name, embedded in error messages.

    Returns
    -------
    pd.DataFrame
        Copy of ``frame`` with ``expression_id`` present as a column.

    Raises
    ------
    TypeError
        If ``frame`` is not a ``pandas.DataFrame``.
    ValueError
        If ``expression_id`` is neither a column nor an index level, or
        if the resulting ``expression_id`` column carries missing
        values.
    """
    # Validate the frame type first (public boundary).
    if not isinstance(frame, pd.DataFrame):
        # Refuse non-DataFrame inputs explicitly.
        raise TypeError(f"{caller}: daily_scores must be a pandas.DataFrame, got {type(frame)!r}.")
    # Start from the caller's frame (never mutate the caller's object).
    flat = frame.copy()
    # Flatten Task 9's MultiIndex layout only when needed (a prior flat frame passes through).
    if "expression_id" not in flat.columns and (
        isinstance(flat.index, pd.MultiIndex) and "expression_id" in flat.index.names
    ):
        # Reset the index back to flat columns (one-time boundary reset).
        flat = flat.reset_index()
    # Final presence assertion for the only load-bearing key column.
    if "expression_id" not in flat.columns:
        # Neither columns nor index carried expression_id; refuse loudly.
        raise ValueError(f"{caller}: 'expression_id' is neither a column nor an index level.")
    # Reject missing expression IDs (an unlabelled row cannot be attributed).
    if flat["expression_id"].isna().any():
        # Surface the offending count; never drop the rows silently.
        raise ValueError(f"{caller}: 'expression_id' carries {int(flat['expression_id'].isna().sum())} missing value(s).")
    # Return the flattened frame.
    return flat

def _admissible_mask_task_10(series: pd.Series, caller: str) -> np.ndarray:
    """
    Materialise the strict ``admissible == True`` boolean mask.

    Compares ``series`` elementwise against True, so a cell is True
    wherever the stored value compares equal to ``True``: in an
    object-dtype column a numeric ``1`` or ``1.0`` also qualifies, while
    ``NA``, ``None`` and ``False`` map to False. Nullable-boolean
    missing values are filled with False and the result is returned as a
    plain ``numpy`` bool vector.

    Parameters
    ----------
    series : pd.Series
        The ``admissible`` column: bool, nullable boolean or object
        dtype.
    caller : str
        Calling function name, embedded in error messages.

    Returns
    -------
    np.ndarray
        Boolean vector of shape ``(len(series),)``; True where the
        stored value compares equal to ``True``.

    Raises
    ------
    ValueError
        If the elementwise comparison raises, or if the produced mask
        has a length different from ``len(series)``.
    """
    # Compare strictly against True (non-True cells are out of T_f by definition).
    try:
        # Evaluate elementwise equality against True.
        mask = series == True  # noqa: E712 -- the strict True-test is the pinned semantics.
    except Exception as exc:  # noqa: BLE001 -- re-raise with context, never swallow.
        # Surface comparison failures with the offending dtype.
        raise ValueError(f"{caller}: 'admissible' cannot be compared against True (dtype={series.dtype}).") from exc
    # Coerce nullable-boolean NAs to False (T_f admits only literal True).
    if hasattr(mask, "fillna"):
        # Nullable boolean path: NA -> False.
        mask = mask.fillna(False)
    # Materialise as a plain numpy bool vector.
    array = np.asarray(mask, dtype=bool)
    # Final shape assertion (vectorisation must not shrink/stretch the mask).
    if array.shape != (len(series),):
        # Refuse misaligned masks loudly.
        raise ValueError(f"{caller}: admissible mask shape {array.shape} != ({len(series)},).")
    # Return the boolean mask.
    return array

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 1:  Per-factor aggregation -- T_f := { t : admissible(f, t) == True }
# -------------------------------------------------------------------------------------------------------------------------------

def aggregate_factors_task_10(
    daily_scores: pd.DataFrame,
    cohort_expression_ids: Sequence[str],
    full_coverage_dates: int = 480,
) -> pd.DataFrame:
    """
    Aggregate per-date daily rank-ICs to one row per factor.

    Implements Task 10, Step 1. For a factor ``f`` the valid-date set is
    ``T_f := { t : admissible(f, t) == True }`` and ``n_valid_dates :=
    |T_f|`` counts those rows, including rows whose ``rho`` is NaN.
    ``mean_rank_ic`` is the arithmetic mean of the non-NaN ``rho``
    values over ``T_f``, sign-preserving and at full float64 precision;
    the filter is ``~np.isnan`` only, so infinite ``rho`` values are
    retained and averaged, and the mean is NaN only when no admissible
    row exists or every admissible ``rho`` is NaN. The dispersion column
    is ``mean(rho) / sd(rho, ddof=1)`` over the same non-NaN cells; it
    is NaN with fewer than two non-NaN cells or a zero sample SD (never
    infinite) and is neither annualised nor a trading Sharpe.
    ``coverage_ratio`` is ``n_valid_dates / full_coverage_dates``: a
    diagnostic that never filters rows, so partial coverage is retained
    and flagged.

    The register is grouped once by ``expression_id``. Cohort members
    absent from the register receive ``n_valid_dates == 0``, reason
    ``"not_evaluated"``, NaN mean and ratio, and ``coverage_ratio ==
    0.0``; evaluated members receive reason ``"evaluated"``. No
    imputation, zero substitution or forward fill is applied.

    Parameters
    ----------
    daily_scores : pd.DataFrame
        The DAILY_SCORES register: a flat frame with columns
        ``expression_id``, ``rho`` (numeric, NaN allowed) and
        ``admissible`` (bool, nullable boolean or object-bool comparable
        to True), or a register whose ``MultiIndex`` names include
        ``expression_id``. A ``trade_date`` column is tolerated but not
        required.
    cohort_expression_ids : Sequence[str]
        Unique frozen cohort IDs in output order; the order defines the
        row order of the result.
    full_coverage_dates : int, optional
        Coverage denominator in dates. Default is ``480``; must be a
        positive integer, and a ``bool`` is rejected.

    Returns
    -------
    pd.DataFrame
        One row per cohort ID, in cohort order, with columns
        ``expression_id`` (string), ``n_valid_dates`` (``Int64``),
        ``n_valid_dates_reason`` (string), ``mean_rank_ic`` (float64),
        ``coverage_ratio`` (float64) and
        ``mean_sd_ratio_unnannualized_not_sharpe`` (float64). Values are
        never rounded or cast to display precision.

    Raises
    ------
    TypeError
        If ``daily_scores`` is not a DataFrame,
        ``cohort_expression_ids`` is a bare string or not a
        ``Sequence``, a cohort ID is not a string, or
        ``full_coverage_dates`` is not an ``int``.
    ValueError
        If ``expression_id`` is missing as a column and as an index
        level or carries missing values, ``rho`` is non-numeric and
        cannot be coerced, ``admissible`` cannot be compared to True, a
        required column is absent, a cohort ID is empty, cohort IDs are
        duplicated, or ``full_coverage_dates`` is not positive.
    """
    # ---- boundary type assertions (this callable is public API) ----------------
    # Validate the cohort sequence type eagerly (a bare str would iterate chars).
    if isinstance(cohort_expression_ids, (str, bytes)) or not isinstance(cohort_expression_ids, Sequence):
        # Refuse bare strings and non-sequence containers with a typed error.
        raise TypeError(f"cohort_expression_ids must be a Sequence[str], got {type(cohort_expression_ids)!r}.")
    # Materialise the cohort list (a fresh list; never alias the caller's container).
    cohort = list(cohort_expression_ids)
    # Validate every cohort ID type (IDs must be non-empty strings).
    for index, token in enumerate(cohort):
        # Reject non-str tokens explicitly (frozen IDs are strings by contract).
        if not isinstance(token, str):
            raise TypeError(f"cohort_expression_ids[{index}] is not a str: {type(token)!r}.")
        # Reject empty IDs (an empty string cannot be an audit handle).
        if token == "":
            raise ValueError(f"cohort_expression_ids[{index}] is an empty string.")
    # Reject duplicated cohort IDs (each ID must identify exactly one frozen row).
    if len(set(cohort)) != len(cohort):
        # Report the duplicated subset for forensic clarity.
        duplicates = sorted({token for token in cohort if cohort.count(token) > 1})
        raise ValueError(f"cohort_expression_ids carries duplicate IDs: {duplicates}.")
    # Validate the coverage denominator (a positive integer, bool excluded).
    if isinstance(full_coverage_dates, bool) or not isinstance(full_coverage_dates, int):
        # Refuse bools/floats/strings with a typed error.
        raise TypeError(f"full_coverage_dates must be an int, got {type(full_coverage_dates)!r}.")
    # Enforce positivity of the denominator.
    if full_coverage_dates <= 0:
        # Refuse zero/negative denominators explicitly.
        raise ValueError(f"full_coverage_dates must be positive; got {full_coverage_dates}.")
    # ---- normalise the register layout ----------------------------------------
    # Flatten Task 9's MultiIndex (or validate a flat frame) once, at the boundary.
    flat = _materialise_flat_columns_task_10(daily_scores, "aggregate_factors_task_10")
    # ---- required-column assertions --------------------------------------------
    # The four load-bearing columns; trade_date is carried but not required for the aggregation itself.
    required_columns = {"expression_id", "rho", "admissible"}
    # Compute missing columns once.
    missing = required_columns.difference(flat.columns)
    # Refuse incomplete registers with a precise listing.
    if missing:
        raise ValueError(f"daily_scores is missing required column(s): {sorted(missing)}.")
    # ---- validate and extract rho as float64 with NaN tolerance -----------------
    # Extract rho as a Series for dtype interrogation.
    rho_series = flat["rho"]
    # Reject non-numeric rho outright (strings would poison the moments silently).
    if not pd.api.types.is_numeric_dtype(rho_series.dtype):
        # Attempt an explicit conversion to force clarity for object-dtype callers.
        try:
            # Try to coerce object-rho to float64 (fails loudly on junk).
            rho_series = pd.to_numeric(rho_series, errors="raise")
        except Exception as exc:  # noqa: BLE001 -- re-raise with context.
            # Refuse unconvertible rho with an informative message.
            raise ValueError(f"'rho' is not numeric and could not be coerced (dtype={flat['rho'].dtype}).") from exc
    # Materialise rho as a numpy float64 vector (NaN preserved).
    rho_values = rho_series.to_numpy(dtype="float64", na_value=np.nan)
    # ---- strict admissible mask -------------------------------------------------
    # T_f admits only literal True cells (NA coerces to False).
    admissible_mask = _admissible_mask_task_10(flat["admissible"], "aggregate_factors_task_10")
    # ---- presence of every expression_id (for the never-evaluated audit flag) ----
    # Evaluate membership once per unique expression_id present in the register.
    present_ids = set(flat["expression_id"].unique().tolist())
    # ---- one pass over the register, grouped by expression_id -------------------
    # Slice the admissible rows only (never mean over the whole frame).
    admissible_frame = flat.loc[admissible_mask, ["expression_id"]]
    # Attach the rho vector aligned to admissible rows.
    admissible_rho = rho_values[admissible_mask]
    # Group the (expression_id, rho) pairs by expression_id in a single pass.
    group_stats: dict = {}
    # Iterate the admissible frame grouped by expression_id (order of groups is irrelevant).
    for expression_id, group_index in admissible_frame.groupby("expression_id", sort=False).indices.items():
        # Positional lookup of this factor's rho vector within the admissible slice.
        rho_slice = admissible_rho[group_index]
        # Drop NaN addends for the moments (but keep them for |T_f|, computed above).
        finite_rho = rho_slice[~np.isnan(rho_slice)]
        # Compute the admissible-row count (NaN rho included).
        n_valid = int(rho_slice.shape[0])
        # Compute the finite-addend count once (used for both moments and the ratio).
        n_finite = int(finite_rho.shape[0])
        # Compute the arithmetic mean over finite admissible addends (signed; may be 0.0).
        mean_rank_ic = float(finite_rho.mean()) if n_finite > 0 else float("nan")
        # Compute the ddof=1 sample dispersion ratio only for >= 2 finite addends.
        if n_finite >= 2:
            # Sample standard deviation with the disclosed ddof=1 choice.
            sd_unbiased = float(finite_rho.std(ddof=1))
            # mean(rho) / sd(rho, ddof=1); NaN when the SD vanishes (constant slice), never inf.
            mean_sd_ratio = float(mean_rank_ic / sd_unbiased) if sd_unbiased != 0.0 else float("nan")
        else:
            # Below 2 finite addends the ddof=1 variance is undefined; report NaN.
            mean_sd_ratio = float("nan")
        # Materialise the audit entry for this factor.
        group_stats[expression_id] = (n_valid, mean_rank_ic, mean_sd_ratio)
    # ---- assemble the cohort-ordered output --------------------------------------
    # Preallocate output arrays in cohort order.
    out_expression_id = pd.array(cohort, dtype="string")
    # n_valid_dates as nullable Int64 (no sentinel zeros anywhere).
    out_n_valid = pd.array([group_stats[eid][0] if eid in group_stats else 0 for eid in cohort], dtype="Int64")
    # Reason: 'evaluated' iff the factor appears in DAILY_SCORES at all.
    out_reason = pd.array(["evaluated" if eid in present_ids else "not_evaluated" for eid in cohort], dtype="string")
    # mean_rank_ic: NaN when the factor carries no admissible dates (or none finite).
    out_mean = np.array([group_stats[eid][1] if eid in group_stats else np.nan for eid in cohort], dtype="float64")
    # coverage_ratio diagnostic: n_valid_dates / full_coverage_dates (never a filter).
    out_coverage = out_n_valid.astype("float64") / float(full_coverage_dates)
    # mean_sd_ratio: NaN when undefined (never substituted).
    out_sd_ratio = np.array([group_stats[eid][2] if eid in group_stats else np.nan for eid in cohort], dtype="float64")
    # Assemble the FACTOR_AGGREGATES vehicle in one shot.
    result = pd.DataFrame({
        "expression_id": out_expression_id,
        "n_valid_dates": out_n_valid,
        "n_valid_dates_reason": out_reason,
        "mean_rank_ic": out_mean,
        "coverage_ratio": out_coverage,
        "mean_sd_ratio_unnannualized_not_sharpe": out_sd_ratio,
    })
    # Return the cohort-ordered aggregate register (fresh RangeIndex).
    return result

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 2:  The three-state ledger -- primary precedence (P04.state_precedence, [R]) and the reverse triple
# -------------------------------------------------------------------------------------------------------------------------------

def _operators_within_registry_task_10(operators_used: Any, registry: frozenset, caller: str) -> bool:
    """
    Decide whether an expression's operators fit the registry.

    ``None``, ``pd.NA`` and non-iterable scalars fail closed (False), as
    does any iterable containing a non-string element. A bare string is
    treated as a single operator. Otherwise the operators are frozen
    into a set and tested for containment in ``registry``.

    Parameters
    ----------
    operators_used : Any
        List-like of operator names, a single operator string, or
        ``None``/``pd.NA`` when the expression was never parsed.
    registry : frozenset
        Frozen case-sensitive public operator registry.
    caller : str
        Calling function name; accepted for symmetry and not read by the
        body.

    Returns
    -------
    bool
        True iff support can be demonstrated, i.e. every observed
        operator is a member of ``registry``.
    """
    # None / pd.NA fails closed (the parse never happened, so operators are unknown).
    if operators_used is None or operators_used is pd.NA:
        # Fail closed, never presume support.
        return False
    # Normalise a bare string to a one-element container.
    if isinstance(operators_used, str):
        # Single operator reported as a scalar string.
        return operators_used in registry
    # Normalise list/tuple/set/ndarray (anything iterable but the scalars above).
    iterable_ok = True
    # Probe iterability without consuming the caller's generator twice.
    try:
        # Freeze the iterable once into a frozenset.
        operators = frozenset(operators_used)
    except TypeError:
        # Non-iterable scalar: fail closed.
        iterable_ok = False
    # Return False for non-iterables.
    if not iterable_ok:
        return False
    # Every element must be a str; non-str operators poison the containment test.
    if not all(isinstance(op, str) for op in operators):
        # A single non-str operator invalidates the audit bundle: fail closed.
        return False
    # Containment test (the registry is a frozenset; operators is a frozenset).
    return operators <= registry

def _validated_precedence_task_10(precedence: Sequence[str], name: str) -> tuple:
    """
    Validate one precedence ordering against the P04 states.

    Materialises ``precedence`` as a tuple, requires every element to be
    a string, and requires the tuple to be a permutation of the three
    states in ``STATE_REGISTRY_TASK_10`` (``scorable``,
    ``insufficient``, ``unsupported``) with each appearing exactly once.
    The given order is preserved.

    Parameters
    ----------
    precedence : Sequence[str]
        Candidate ordering, e.g.
        ``("unsupported", "insufficient", "scorable")``.
    name : str
        Parameter name, embedded in error messages.

    Returns
    -------
    tuple
        The validated ordering as a tuple of three state strings.

    Raises
    ------
    TypeError
        If ``precedence`` is a bare string or bytes, is not a
        ``Sequence``, or contains a non-string element.
    ValueError
        If the tuple is not exactly the three P04 states, each once.
    """
    # Reject bare strings (they would explode into single-character states).
    if isinstance(precedence, (str, bytes)) or not isinstance(precedence, Sequence):
        # Typed refusal for mis-wired callers.
        raise TypeError(f"{name} must be a Sequence[str], not {type(precedence)!r}.")
    # Materialise the order once (immutability for downstream indexing).
    ordered = tuple(precedence)
    # Every entry must be a string naming one of the three P04 states.
    for index, state in enumerate(ordered):
        # Reject non-str atoms explicitly.
        if not isinstance(state, str):
            raise TypeError(f"{name}[{index}] is not a str: {type(state)!r}.")
    # Enforce exact coverage of the three states, each exactly once.
    if len(ordered) != 3 or set(ordered) != STATE_REGISTRY_TASK_10:
        # The triple must be a permutation of {scorable, insufficient, unsupported}.
        raise ValueError(f"{name} must enumerate exactly the three P04 states once each; got {ordered}.")
    # Return the validated triple.
    return ordered

def _assign_states_shared_lookup_task_10(
    supported: np.ndarray,
    n_valid_dates: np.ndarray,
    precedence: tuple,
    min_valid_dates: int,
    caller: str,
) -> np.ndarray:
    """
    Assign one ledger via a shared precedence lookup table.

    Three eligibility masks are built once over all records:

    * ``scorable``: ``supported`` AND
      ``n_valid_dates >= min_valid_dates``, boundary inclusive;
    * ``insufficient``: ``n_valid_dates < min_valid_dates``, computed
      without regard to support;
    * ``unsupported``: ``~supported``, i.e. a parse failure or a
      registry miss.

    The masks are not mutually exclusive; each record takes the first
    state of ``precedence`` whose mask fires, so the primary and the
    reverse ledger share this single code path and differ only in the
    triple.

    Parameters
    ----------
    supported : np.ndarray
        Boolean vector, True where the expression parsed and uses only
        registered operators.
    n_valid_dates : np.ndarray
        Integer vector of admissible-date counts, one per record.
    precedence : tuple
        Validated state ordering, consumed strictly in order.
    min_valid_dates : int
        Inclusive scorability threshold (20 in the primary run).
    caller : str
        Calling function name, embedded in error messages.

    Returns
    -------
    np.ndarray
        Object-dtype vector of state strings, same length as
        ``n_valid_dates``.

    Raises
    ------
    ValueError
        If no state of ``precedence`` is eligible for a record; the
        masks cover every truth combination, so this is a defensive
        backstop.
    """
    # Eligibility mask: scorable (supported and >= threshold).
    scorable_ok = supported & (n_valid_dates >= min_valid_dates)
    # Eligibility mask: insufficient (strictly below the threshold).
    insufficient_ok = n_valid_dates < min_valid_dates
    # Eligibility mask: unsupported (support demonstration failed).
    unsupported_ok = ~supported
    # Freeze the eligibility map once (single source of truth for both runs).
    eligibility = {"scorable": scorable_ok, "insufficient": insufficient_ok, "unsupported": unsupported_ok}
    # Preallocate the assignment vector.
    assigned = np.empty(n_valid_dates.shape[0], dtype=object)
    # Walk the precedence in caller-pinned order; first eligible state wins.
    for position in range(n_valid_dates.shape[0]):
        # Try the states in precedence order.
        winner = None
        # Iterate the precedence triple strictly in order.
        for state in precedence:
            # Assign this state iff its eligibility mask fires for this record.
            if eligibility[state][position]:
                winner = state
                break
        # Backstop: unreachable in a total mask family; kept explicit anyway.
        if winner is None:
            raise ValueError(f"{caller}: no precedence state eligible for record {position}.")
        # Record the winner.
        assigned[position] = winner
    # Return the assigned state vector.
    return assigned

def assign_ledger_task_10(
    aggregates: pd.DataFrame,
    support: pd.DataFrame,
    public_operator_registry: Collection[str],
    precedence: Sequence[str] = ("unsupported", "insufficient", "scorable"),
    reverse_precedence: Sequence[str] = ("scorable", "insufficient", "unsupported"),
    min_valid_dates: int = 20,
) -> pd.DataFrame:
    """
    Assign the three-state ledger under both precedence triples.

    Implements Step 2 of Task 10. The function validates every input,
    derives a per-expression supported mask, and evaluates two
    precedence triples through one shared lookup routine.

    Support rule: an expression is supported iff
    ``parse_status == "parsed"`` and every operator listed in
    ``operators_used`` lies in ``public_operator_registry``.
    Operators fail closed on a missing or malformed bundle.

    Under the primary precedence
    ``("unsupported", "insufficient", "scorable")``:

    * ``unsupported`` when the expression is unparseable or names an
      operator outside the registry;
    * ``insufficient`` when supported and ``n_valid_dates`` is below
      the threshold;
    * ``scorable`` when supported and ``n_valid_dates`` is at or above
      it.

    The threshold test is inclusive (``>= min_valid_dates``). The
    reverse precedence
    ``("scorable", "insufficient", "unsupported")`` is recomputed the
    same way: a record that is both unparseable and short of the
    threshold lands in ``insufficient``, while an unparseable record at
    or above the threshold still lands in ``unsupported``.

    Parameters
    ----------
    aggregates : pd.DataFrame
        FACTOR_AGGREGATES from Step 1; must carry
        ``expression_id`` and ``n_valid_dates`` with no duplicate IDs
        and no missing counts. Row order defines the output order.
    support : pd.DataFrame
        Cohort support bundle with columns ``expression_id``,
        ``parse_status`` (``"parsed"`` or ``"parse_failed"``) and
        ``operators_used`` (list-like; may be empty). Exactly one row
        per aggregate ``expression_id`` is enforced.
    public_operator_registry : Collection[str]
        Names of the public operators. A bare ``str`` or ``bytes``
        argument, a non-``Collection`` value, or any non-``str`` member
        raises ``TypeError``. The collection is frozen to a
        ``frozenset`` for membership tests.
    precedence : Sequence[str], optional
        The primary precedence triple. Default is
        ``("unsupported", "insufficient", "scorable")``.
    reverse_precedence : Sequence[str], optional
        The reversed precedence triple. Default is
        ``("scorable", "insufficient", "unsupported")``.
    min_valid_dates : int, optional
        Scorability threshold on ``n_valid_dates``. Default is ``20``.
        Must be a non-``bool`` ``int`` of at least 1.

    Returns
    -------
    pd.DataFrame
        A copy of ``aggregates`` with two appended string columns:
        ``state_primary`` (values drawn from ``precedence``) and
        ``state_reverse`` (values drawn from ``reverse_precedence``).
        Row order and all Step-1 columns are preserved.

    Raises
    ------
    TypeError
        If ``aggregates`` or ``support`` is not a
        ``pandas.DataFrame``; if ``public_operator_registry`` is a bare
        ``str``/``bytes``, is not a ``Collection``, contains an
        unhashable element, or contains a non-``str`` member; if
        ``precedence`` or ``reverse_precedence`` is of the wrong type
        (propagated from the internal precedence validator); or if
        ``min_valid_dates`` is a ``bool`` or not an ``int``.
    ValueError
        If ``min_valid_dates`` is below 1; if ``aggregates`` lacks
        ``expression_id`` or ``n_valid_dates``; if ``support`` lacks
        ``expression_id``, ``parse_status`` or ``operators_used``; if
        either frame carries a missing ``expression_id``; if
        ``aggregates`` carries duplicate IDs; if ``support`` carries
        duplicate rows, omits an aggregate ID, or carries an ID outside
        the aggregate cohort; if ``n_valid_dates`` holds missing,
        negative or non-integral values; or if either precedence triple
        is not a permutation of the P04 state registry.

    Notes
    -----
    Steps: validate types and the two precedence triples; check the
    required columns; enforce exactly one support row per aggregate
    ID; reindex ``support`` to the aggregate order; coerce
    ``n_valid_dates`` to ``int64``; build the supported mask; call the
    shared lookup once per precedence triple; attach the two string
    ledgers to a copy of ``aggregates``. The input frames are never
    mutated.
    """
    # ---- boundary type assertions (public API) -----------------------------------
    # aggregates must be a DataFrame.
    if not isinstance(aggregates, pd.DataFrame):
        # Refuse non-DataFrames with a typed error.
        raise TypeError(f"aggregates must be a pandas.DataFrame, got {type(aggregates)!r}.")
    # support must be a DataFrame.
    if not isinstance(support, pd.DataFrame):
        # Refuse non-DataFrames with a typed error.
        raise TypeError(f"support must be a pandas.DataFrame, got {type(support)!r}.")
    # A bare str registry would iterate characters; refuse it.
    if isinstance(public_operator_registry, (str, bytes)):
        # Typed refusal avoids silent single-char membership tests.
        raise TypeError("public_operator_registry must be a Collection[str], not a bare str/bytes.")
    if not isinstance(public_operator_registry, Collection):
        # Refuse non-collections with a typed error.
        raise TypeError(f"public_operator_registry must be a Collection[str], got {type(public_operator_registry)!r}.")
    # ---- freeze the registry once (frozenset membership is O(1)) -----------------
    # Validate every registry element is a string and freeze the set.
    registry_frozen = frozenset()
    # Accumulate members with strict type checks.
    try:
        # Materialise the registry copy (never alias the caller's collection).
        registry_frozen = frozenset(public_operator_registry)
    except TypeError as exc:
        # Refuse unhashable elements with an informative message.
        raise TypeError("public_operator_registry must contain hashable str operator names.") from exc
    # Enforce string-typed registry members.
    if not all(isinstance(op, str) for op in registry_frozen):
        # A single non-str member invalidates every containment test.
        raise TypeError("public_operator_registry must contain only str members.")
    # ---- validate the scalar threshold --------------------------------------------
    # min_valid_dates must be an int (bool excluded) and >= 1.
    if isinstance(min_valid_dates, bool) or not isinstance(min_valid_dates, int):
        # Typed refusal for float/str thresholds.
        raise TypeError(f"min_valid_dates must be an int, got {type(min_valid_dates)!r}.")
    # Enforce the >= 1 domain (a zero threshold would empty the insufficient arm).
    if min_valid_dates < 1:
        # Refuse sub-unity thresholds explicitly.
        raise ValueError(f"min_valid_dates must be >= 1; got {min_valid_dates}.")
    # ---- validate both precedence triples -----------------------------------------
    # Primary triple validated against the P04 state registry.
    primary = _validated_precedence_task_10(precedence, "precedence")
    # Reverse triple validated against the same registry.
    reversed_triple = _validated_precedence_task_10(reverse_precedence, "reverse_precedence")
    # ---- required-column assertions on aggregates ----------------------------------
    # The two load-bearing columns.
    missing_agg = {"expression_id", "n_valid_dates"}.difference(aggregates.columns)
    # Refuse incomplete aggregates with a precise listing.
    if missing_agg:
        raise ValueError(f"aggregates is missing required column(s): {sorted(missing_agg)}.")
    # ---- required-column assertions on support -------------------------------------
    # The three load-bearing columns.
    missing_support = {"expression_id", "parse_status", "operators_used"}.difference(support.columns)
    # Refuse incomplete support bundles with a precise listing.
    if missing_support:
        raise ValueError(f"support is missing required column(s): {sorted(missing_support)}.")
    # ---- exactly one support row per aggregate row ----------------------------------
    # Materialise the aggregate ID list (cohort order) and the support ID list.
    agg_ids = aggregates["expression_id"].tolist()
    # Detect missing/duplicate/superfluous expression_id cells in aggregates.
    if aggregates["expression_id"].isna().any():
        # Missing IDs would silently orphan aggregate rows.
        raise ValueError("aggregates carries missing expression_id value(s).")
    # Reject duplicated IDs in aggregates (the join becomes ill-posed).
    if len(set(agg_ids)) != len(agg_ids):
        # Report duplicates for forensic clarity.
        duplicates = sorted({token for token in agg_ids if agg_ids.count(token) > 1})
        raise ValueError(f"aggregates carries duplicate expression_id value(s): {duplicates}.")
    # Detect null support IDs.
    if support["expression_id"].isna().any():
        # Missing support IDs invalidate the join.
        raise ValueError("support carries missing expression_id value(s).")
    # Detect duplicated support rows (the mandated exactly-one rule).
    support_ids = support["expression_id"].tolist()
    # Count occurrences once (O(n) via Counter semantics).
    seen: dict = {}
    # Accumulate counts without sorting (cohort order irrelevant for set equality).
    for token in support_ids:
        # Increment the occurrence bucket.
        seen[token] = seen.get(token, 0) + 1
    # Collect duplicates.
    duplicated_support = sorted([token for token, count in seen.items() if count > 1])
    # Enforce the exactly-one rule loudly if violated.
    if duplicated_support:
        raise ValueError(f"support carries duplicate row(s) for expression_id: {duplicated_support}.")
    # Enforce set equality between the two cohorts (same expression universe, each listed).
    missing_support_ids = sorted(set(agg_ids).difference(support_ids))
    # Reject missing expressions (coverage gap in the support bundle).
    if missing_support_ids:
        raise ValueError(f"support is missing row(s) for expression_id: {missing_support_ids}.")
    # Reject superfluous support rows (the "exactly one per expression" rule cuts both ways).
    extra_support_ids = sorted(set(support_ids).difference(agg_ids))
    # Refuse foreign rows explicitly.
    if extra_support_ids:
        raise ValueError(f"support carries row(s) outside the aggregate cohort: {extra_support_ids}.")
    # ---- index the support frame to the aggregate order -----------------------------
    # Reindex a copy of support to the aggregate order (positional join, not merge).
    support_indexed = support.set_index("expression_id").reindex(agg_ids)
    # ---- n_valid_dates to int64 (NA rejected, not coerced) ---------------------------
    # Detect NA/NaN cells (the boundary refuses them).
    n_valid_raw = aggregates["n_valid_dates"]
    # Boolean NA test via pandas (works for Int64/object; float NaN too).
    if bool(n_valid_raw.isna().any()):
        # Report the offending count before refusing.
        raise ValueError(f"aggregates['n_valid_dates'] carries {int(n_valid_raw.isna().sum())} missing value(s).")
    # Coerce to numpy int64 (non-integer floats raise via cast inside try).
    try:
        # Attempt the cast; floats with fractional parts will violate int domain.
        n_valid_float = n_valid_raw.to_numpy(dtype="float64")
        # Reject negative counts.
        if bool((n_valid_float < 0).any()):
            # Report the minimum observed count.
            raise ValueError(f"aggregates['n_valid_dates'] carries negative value(s) down to {float(n_valid_float.min())}.")
        # Reject non-integral floats (e.g. 19.5).
        if bool((n_valid_float != np.floor(n_valid_float)).any()):
            # Report the offending count; fractional valid-date counts are impossible.
            raise ValueError("aggregates['n_valid_dates'] carries non-integral value(s).")
        # Materialise the int64 vector.
        n_valid = n_valid_float.astype("int64")
    except ValueError as exc:
        # Re-raise with the original message (typed refusal with provenance).
        raise ValueError(str(exc)) from exc
    # ---- supported mask per row -----------------------------------------------------
    # Compute parse_status == 'parsed' strictly (NA fails closed via ==).
    parse_ok = (support_indexed["parse_status"] == "parsed")
    # Coerce the parse vector to plain numpy bool (nullable booleans/NA -> False).
    parse_ok = parse_ok.fillna(False).to_numpy(dtype=bool) if hasattr(parse_ok, "fillna") else np.asarray(parse_ok, dtype=bool)
    # Compute the per-row operator containment vector.
    ops_ok = np.array(
        [_operators_within_registry_task_10(value, registry_frozen, "assign_ledger_task_10") for value in support_indexed["operators_used"].tolist()],
        dtype=bool,
    )
    # supported := parse_ok AND operators-within-registry.
    supported = parse_ok & ops_ok
    # ---- assign both ledgers through the SINGLE shared lookup routine ---------------
    # Primary ledger (the [R] primary ordering: unsupported > insufficient > scorable).
    state_primary = _assign_states_shared_lookup_task_10(supported, n_valid, primary, min_valid_dates, "assign_ledger_task_10")
    # Reverse ledger (scorable > insufficient > unsupported).
    state_reverse = _assign_states_shared_lookup_task_10(supported, n_valid, reversed_triple, min_valid_dates, "assign_ledger_task_10")
    # ---- append the two string ledgers to a copy of the aggregate frame --------------
    # Copy the frame (never mutate the caller's object in place).
    result = aggregates.copy()
    # Attach state_primary as a string column.
    result["state_primary"] = pd.array(state_primary.tolist(), dtype="string")
    # Attach state_reverse as a string column.
    result["state_reverse"] = pd.array(state_reverse.tolist(), dtype="string")
    # Return the augmented frame.
    return result

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Step 3:  The closure battery -- exact ledger triple, cross-closure, sign and coverage partitions, near-zero diagnostic
# -------------------------------------------------------------------------------------------------------------------------------

def closure_battery_task_10(
    aggregates: pd.DataFrame,
    n_cohort: int,
    near_zero_threshold: float = 1e-12,
    full_coverage_dates: int = 480,
) -> dict:
    """
    Audit ledger closure and the scorable-arm partitions.

    Check ``aggregates`` for the exact structural identities and report
    the resulting counts:

    * the ledger closure:
      ``n_scorable + n_insufficient + n_unsupported == n_cohort``;
    * the cross-closure:
      ``n_cohort - n_unsupported == n_scorable + n_insufficient``;
    * the sign-partition totalisation:
      ``n_positive + n_negative + n_exact_zero + n_nan_mean ==
      n_scorable``;
    * the exclusion-adjusted sign closure:
      ``n_positive + n_negative == n_scorable - n_exact_zero -
      n_nan_mean``;
    * the coverage closure:
      ``n_full + n_partial == n_scorable``.

    Signs are classified on the unrounded ``mean_rank_ic`` with an
    exact zero-test: values ``> 0`` are positive, values ``< 0`` are
    negative, and exact zeros and NaN means are excluded from the sign
    partition but reported. Coverage inside the scorable arm splits
    into full rows (``n_valid_dates == full_coverage_dates``) and
    partial rows (``0 < n_valid_dates < full_coverage_dates``). The
    near-zero diagnostic counts scorable means whose absolute value is
    below ``near_zero_threshold``; it never re-partitions the ledger.
    The median of the scorable ``mean_rank_ic`` values is returned at
    full float64 precision. No distribution mean, standard deviation
    or significance statistic is computed.

    Parameters
    ----------
    aggregates : pd.DataFrame
        Aggregate frame with one row per cohort expression. Required
        columns are ``expression_id``, ``n_valid_dates``,
        ``mean_rank_ic`` and ``state_primary``; ``len(aggregates)``
        must equal ``n_cohort``.
    n_cohort : int
        Cohort size the ledger must close to. Must be a non-negative
        int; ``bool`` is rejected.
    near_zero_threshold : float, optional
        Non-negative absolute threshold for the near-zero diagnostic.
        Default is ``1e-12``.
    full_coverage_dates : int, optional
        Positive row count that defines full coverage. Default is
        ``480``.

    Returns
    -------
    dict
        Keys ``n_scorable``, ``n_insufficient``, ``n_unsupported``,
        ``n_positive``, ``n_negative``, ``n_exact_zero``,
        ``n_nan_mean``, ``n_full``, ``n_partial``,
        ``near_zero_count``, ``median_mean_rank_ic`` (NaN when the
        scorable arm is empty or when any scorable mean is NaN, since
        ``numpy.median`` propagates a single NaN),
        ``sign_closure_exclusion_adjusted_identity``,
        ``sign_closure_contracted_identity``, ``diagnostics`` (list of
        audit strings), ``claim_records`` (list of claim dictionaries)
        and ``n_cohort``.

    Raises
    ------
    TypeError
        If ``aggregates`` is not a DataFrame; if ``n_cohort``,
        ``near_zero_threshold`` or ``full_coverage_dates`` has the
        wrong type (``bool`` is rejected for each).
    ValueError
        If ``n_cohort`` is negative, ``near_zero_threshold`` is
        negative, or ``full_coverage_dates`` is not positive; if
        required columns are missing or ``len(aggregates) !=
        n_cohort``; if ``state_primary`` holds a value outside the
        state registry; if the ledger closure or the cross-closure
        fails; if ``n_valid_dates`` carries missing values; if a
        scorable row has ``n_valid_dates < 1`` or ``>
        full_coverage_dates``; if ``mean_rank_ic`` cannot be
        converted numerically; or if the sign-partition totalisation,
        the exclusion-adjusted sign closure, or the coverage closure
        fails.

    Notes
    -----
    The contracted identity ``n_positive + n_negative == n_scorable``
    is reported in the result and diagnostics, and may be false when
    exact zeros or NaN means exist.
    """
    # ---- boundary type assertions (public API) -----------------------------------
    # aggregates must be a DataFrame.
    if not isinstance(aggregates, pd.DataFrame):
        # Refuse non-DataFrames with a typed error.
        raise TypeError(f"aggregates must be a pandas.DataFrame, got {type(aggregates)!r}.")
    # n_cohort must be an int (bool excluded) and >= 0.
    if isinstance(n_cohort, bool) or not isinstance(n_cohort, int):
        # Typed refusal for float/str cohort sizes.
        raise TypeError(f"n_cohort must be an int, got {type(n_cohort)!r}.")
    # Enforce the non-negative domain.
    if n_cohort < 0:
        # Refuse negative cohorts explicitly.
        raise ValueError(f"n_cohort must be >= 0; got {n_cohort}.")
    # near_zero_threshold must be a real number >= 0.
    if isinstance(near_zero_threshold, bool) or not isinstance(near_zero_threshold, (int, float)):
        # Typed refusal for non-numeric thresholds.
        raise TypeError(f"near_zero_threshold must be a real number, got {type(near_zero_threshold)!r}.")
    # Enforce the >= 0 domain.
    if float(near_zero_threshold) < 0.0:
        # Refuse negative thresholds explicitly.
        raise ValueError(f"near_zero_threshold must be >= 0; got {near_zero_threshold}.")
    # full_coverage_dates must be a positive int (bool excluded).
    if isinstance(full_coverage_dates, bool) or not isinstance(full_coverage_dates, int):
        # Typed refusal for non-int coverage bounds.
        raise TypeError(f"full_coverage_dates must be an int, got {type(full_coverage_dates)!r}.")
    # Enforce positivity.
    if full_coverage_dates <= 0:
        # Refuse non-positive bounds explicitly.
        raise ValueError(f"full_coverage_dates must be positive; got {full_coverage_dates}.")
    # ---- required columns ---------------------------------------------------------
    # The load-bearing columns for the battery.
    required_columns = {"expression_id", "n_valid_dates", "mean_rank_ic", "state_primary"}
    # Compute missing columns once.
    missing = required_columns.difference(aggregates.columns)
    # Refuse incomplete frames with a precise listing.
    if missing:
        raise ValueError(f"aggregates is missing required column(s): {sorted(missing)}.")
    # ---- cohort cardinality -------------------------------------------------------
    # The frame must carry exactly n_cohort rows (one per frozen ID).
    if len(aggregates) != n_cohort:
        # Refuse cardinality mismatches explicitly (the closure basis is the reproduced cohort).
        raise ValueError(f"len(aggregates) == {len(aggregates)} does not equal n_cohort == {n_cohort}.")
    # ---- state registry check (BEFORE any arithmetic) ------------------------------
    # Materialise the state vector as strings.
    state_vector = aggregates["state_primary"].tolist()
    # Collect any state outside the P04 registry.
    offenders = sorted({str(state) for state in state_vector if state not in STATE_REGISTRY_TASK_10})
    # Reject unknown states with a precise listing (e.g. the v1 fourth state leaking in).
    if offenders:
        raise ValueError(f"state_primary carries state(s) outside the P04 registry: {offenders}.")
    # ---- per-state masks over the frame's own positional axis ----------------------
    # scorable mask (string equality; NA cells compare unequal and fall out).
    scorable_mask = np.asarray([state == "scorable" for state in state_vector], dtype=bool)
    # insufficient mask.
    insufficient_mask = np.asarray([state == "insufficient" for state in state_vector], dtype=bool)
    # unsupported mask.
    unsupported_mask = np.asarray([state == "unsupported" for state in state_vector], dtype=bool)
    # ledger counts as Python ints.
    n_scorable = int(scorable_mask.sum())
    # insufficient count.
    n_insufficient = int(insufficient_mask.sum())
    # unsupported count.
    n_unsupported = int(unsupported_mask.sum())
    # ---- ledger closure (EXACT) ----------------------------------------------------
    # The reproduction asserts on its own cohort total, never the paper's ledger figure.
    triple_sum = n_scorable + n_insufficient + n_unsupported
    # Refuse any drift loudly (EXACT identity, ledger_theorem).
    if triple_sum != n_cohort:
        raise ValueError(f"ledger closure failed: n_scorable({n_scorable}) + n_insufficient({n_insufficient}) + n_unsupported({n_unsupported}) == {triple_sum} != n_cohort == {n_cohort}.")
    # ---- cross closure (EXACT) -----------------------------------------------------
    # n_cohort - n_unsupported must equal n_scorable + n_insufficient.
    if n_cohort - n_unsupported != n_scorable + n_insufficient:
        raise ValueError(f"cross closure failed: n_cohort({n_cohort}) - n_unsupported({n_unsupported}) != n_scorable({n_scorable}) + n_insufficient({n_insufficient}).")
    # ---- n_valid_dates to int64 (na_rejected) --------------------------------------
    # Extract scores' valid-date vector as float then int64 (same boundary as Step 2).
    n_valid_raw = aggregates["n_valid_dates"]
    # Detect NA/NaN cells.
    if bool(n_valid_raw.isna().any()):
        # Report the offending count before refusing.
        raise ValueError(f"aggregates['n_valid_dates'] carries {int(n_valid_raw.isna().sum())} missing value(s).")
    # Coerce through float64 with a forced domain check.
    n_valid = n_valid_raw.to_numpy(dtype="float64").astype("int64")
    # ---- scorable-arm domain assertions --------------------------------------------
    # Scorable rows with n_valid_dates < 1 violate the coverage partition's 0 boundary.
    zero_scorable = np.asarray(aggregates.iloc[scorable_mask]["expression_id"].tolist(), dtype=object)
    # Detect them positionally.
    zero_scorable = zero_scorable[n_valid[scorable_mask] < 1]
    # Refuse zero-date scorable rows with a precise listing.
    if zero_scorable.shape[0] > 0:
        raise ValueError(f"scorable row(s) carry n_valid_dates < 1: {zero_scorable.tolist()}.")
    # Scorable rows above the full-coverage bound are an upstream construction failure.
    overflow_scorable = np.asarray(aggregates.iloc[scorable_mask]["expression_id"].tolist(), dtype=object)
    # Detect them positionally.
    overflow_scorable = overflow_scorable[n_valid[scorable_mask] > full_coverage_dates]
    # Refuse overflow rows with a precise listing.
    if overflow_scorable.shape[0] > 0:
        raise ValueError(f"scorable row(s) carry n_valid_dates > {full_coverage_dates}: {overflow_scorable.tolist()}.")
    # ---- mean_rank_ic extraction (full float64 precision) ---------------------------
    # Materialise the raw mean column for overflow-safe conversion.
    mean_column = aggregates["mean_rank_ic"]
    # Fast path: already a float dtype (covers float64 and nullable Float64).
    if pd.api.types.is_float_dtype(mean_column):
        # Carry NaN straight through as float64.
        scorable_means_all = mean_column.to_numpy(dtype="float64", na_value=np.nan)
    else:
        # Object/other dtypes: force numeric conversion; junk raises (never silently NaN-ified).
        scorable_means_all = pd.to_numeric(mean_column, errors="raise").to_numpy(dtype="float64")
    # Slice to the scorable positional mask.
    scorable_means = scorable_means_all[scorable_mask]
    # NaN count inside the scorable arm (undefined means from admissible-NaN-rho slices).
    n_nan_mean = int(np.isnan(scorable_means).sum())
    # Finite scorable means (used for the sign partition and the near-zero count).
    finite_means = scorable_means[~np.isnan(scorable_means)]
    # ---- sign partition on the unrounded float (exact zero-test) --------------------
    # Positive means: x > 0 (exact comparison, no tolerance).
    n_positive = int((finite_means > 0.0).sum())
    # Negative means: x < 0 (exact comparison, no tolerance).
    n_negative = int((finite_means < 0.0).sum())
    # Exact zeros: x == 0.0 (reported through the diagnostic path, never the partition).
    n_exact_zero = int((finite_means == 0.0).sum())
    # Construction identity check (totality of the tripartite split).
    if n_positive + n_negative + n_exact_zero + n_nan_mean != n_scorable:
        # This is an internal invariant; its violation indicates a vectorisation bug.
        raise ValueError(f"sign partition failed to totalise: positive={n_positive} negative={n_negative} zero={n_exact_zero} nan={n_nan_mean} do not sum to n_scorable={n_scorable}.")
    # ---- sign closure, asserted BOTH ways, EXACT, never folded -------------------------
    # (i) The exclusion-adjusted identity (independently checkable):
    #     n_positive + n_negative == n_scorable - n_exact_zero - n_nan_mean
    exclusion_adjusted_identity_ok = bool(n_positive + n_negative == n_scorable - n_exact_zero - n_nan_mean)
    # A breach here is a construction defect; fail loud, as with every internal invariant.
    if not exclusion_adjusted_identity_ok:
        # Refute with the exact operand values.
        raise ValueError(
            f"sign closure (exclusion-adjusted) failed: n_positive({n_positive}) + n_negative({n_negative}) == "
            f"{n_positive + n_negative} != n_scorable({n_scorable}) - n_exact_zero({n_exact_zero}) - n_nan_mean({n_nan_mean}) == "
            f"{n_scorable - n_exact_zero - n_nan_mean}."
        )
    # (ii) The PINNED contracted identity from the detailed task list (Task 10, Step 3,
    #      ~line 514: "Sign partition. Within scorable, n_positive + n_negative == n_scorable.
    #      Classification on the unrounded float ..."). EXACT by the tolerance register.
    contracted_identity_ok = bool(n_positive + n_negative == n_scorable)
    # ---- coverage partition on scorable rows ----------------------------------------
    # Full coverage: n_valid_dates == full_coverage_dates.
    n_full = int((n_valid[scorable_mask] == full_coverage_dates).sum())
    # Partial coverage: 0 < n_valid_dates < full_coverage_dates.
    n_partial = int(((n_valid[scorable_mask] > 0) & (n_valid[scorable_mask] < full_coverage_dates)).sum())
    # Coverage closure (construction identity under the domain assertions).
    if n_full + n_partial != n_scorable:
        # This is an internal invariant; its violation indicates a mask bug.
        raise ValueError(f"coverage partition failed to close: n_full={n_full} + n_partial={n_partial} != n_scorable={n_scorable}.")
    # ---- near-zero diagnostic (exact threshold, absolute) ---------------------------
    # near_zero_count := |{f : |mean_rank_ic| < near_zero_threshold}| (includes exact zeros).
    near_zero_count = int((np.abs(finite_means) < float(near_zero_threshold)).sum())
    # ---- median over scorable means -------------------------------------------------
    # Median at FULL precision; NaN propagates honestly when any scorable mean is NaN.
    median_mean_rank_ic = float(np.median(scorable_means)) if n_scorable > 0 else float("nan")
    # ---- diagnostics -----------------------------------------------------------------
    # Accumulate human-auditable closure notes (no significance vocabulary anywhere).
    diagnostics: list = []
    # Ledger closure note (this run's own counts).
    diagnostics.append(f"LEDGER_CLOSURE: n_scorable({n_scorable}) + n_insufficient({n_insufficient}) + n_unsupported({n_unsupported}) == n_cohort({n_cohort}) [EXACT].")
    # Cross closure note.
    diagnostics.append(f"CROSS_CLOSURE: n_cohort({n_cohort}) - n_unsupported({n_unsupported}) == n_scorable({n_scorable}) + n_insufficient({n_insufficient}) [EXACT].")
    # Sign partition note.
    diagnostics.append(f"SIGN_PARTITION_CLOSURE: n_positive({n_positive}) + n_negative({n_negative}) + n_exact_zero({n_exact_zero}) + n_nan_mean({n_nan_mean}) == n_scorable({n_scorable}).")
    # Sign closure (i), named and EXACT (the independently checkable exclusion-adjusted identity).
    diagnostics.append(
        f"SIGN_CLOSURE_EXCLUSION_ADJUSTED: n_positive({n_positive}) + n_negative({n_negative}) == "
        f"n_scorable({n_scorable}) - n_exact_zero({n_exact_zero}) - n_nan_mean({n_nan_mean}) "
        f"[EXACT] -- {'SATISFIED' if exclusion_adjusted_identity_ok else 'VIOLATED'}."
    )
    # Sign closure (ii), named and EXACT (the pinned contracted identity, REPORTED never folded).
    if contracted_identity_ok:
        # The pinned identity is satisfied on this arm.
        diagnostics.append(
            f"SIGN_CLOSURE_CONTRACTED: n_positive({n_positive}) + n_negative({n_negative}) == "
            f"n_scorable({n_scorable}) [EXACT] -- SATISFIED."
        )
    else:
        # The pinned identity is violated BY CONSTRUCTION of the disclosed exclusions; report, never fold.
        diagnostics.append(
            f"SIGN_CLOSURE_CONTRACTED: n_positive({n_positive}) + n_negative({n_negative}) != "
            f"n_scorable({n_scorable}) [EXACT] -- VIOLATED_BY_CONSTRUCTION of the disclosed exclusions "
            f"(n_exact_zero={n_exact_zero}, n_nan_mean={n_nan_mean}); the pinned task-list identity is REPORTED, "
            f"never folded, never rounded, never epsilon-compared; exact zeros and NaN means are excluded from the "
            f"partition by construction (the exact zero-test, not an exclusion of convenience); see SIGN_PARTITION_CLOSURE."
        )
    # Coverage partition note.
    diagnostics.append(f"COVERAGE_PARTITION_CLOSURE: n_full({n_full}) + n_partial({n_partial}) == n_scorable({n_scorable}) [full := n_valid_dates == {full_coverage_dates}; partial := 0 < n_valid_dates < {full_coverage_dates}].")
    # Near-zero diagnostic (mandatory record when the count is non-zero).
    if near_zero_count > 0:
        # Name the triggering records via the diagnostic count (list omitted to avoid echoing means).
        diagnostics.append(f"NEAR_ZERO_DIAGNOSTIC: |mean_rank_ic| < {near_zero_threshold:g} for {near_zero_count} scorable record(s); the ledger is not re-partitioned on this diagnostic (P03.near_zero_absolute_threshold).")
    # NaN-mean diagnostic (undefined means must never masquerade as zeros).
    if n_nan_mean > 0:
        # The ledger stays untouched; only the median becomes undefined.
        diagnostics.append(f"NAN_MEAN_SCORABLE: {n_nan_mean} scorable record(s) carry undefined mean_rank_ic (all admissible rho NaN); the median is reported as NaN and no imputation is performed.")
    # Empty scorable arm note.
    if n_scorable == 0:
        # A legal degenerate cohort; the median is undefined.
        diagnostics.append("EMPTY_SCORABLE_ARM: no scorable record; the median is undefined (NaN).")
    # ---- claim records ----------------------------------------------------------------
    # The structural verifier claim (the closure is a construction invariant of this run).
    closure_record = _make_claim_record_task_10(
        claim_id="STRUCTURAL__[D]__verifiable_on_this_run",
        tier="reproduction",
        evidence_source="closure_battery_this_run",
        verdict="CLOSED",
        value={"n_scorable": n_scorable, "n_insufficient": n_insufficient, "n_unsupported": n_unsupported, "n_cohort": n_cohort},
        note=(
            "Ledger triple, cross-closure, sign partition (exact zero-test) and coverage partition "
            "all close exactly on this run's cohort; the closure is a construction invariant of the "
            "reproduction, not a paper finding. The paper's Table 3 ledger triple is recorded in the "
            "publication only as a comparison for the paper's frozen cohort; nothing in this record "
            "asserts it, and no paper figure is embedded in any returned value."
        ),
    )
    # The median claim (reproduction tier; evidence sourced from this run's scorable arm).
    median_record = _make_claim_record_task_10(
        claim_id="median_mean_rank_ic",
        tier="reproduction",
        evidence_source="cohort_scored_this_run",
        verdict="REPORTED",
        value=median_mean_rank_ic,
        note=(
            "Median of mean_rank_ic over the scorable arm at full float64 precision; "
            "display rounding to 4 decimals is the document layer's concern. The paper's Table 3 "
            "median comparison is a reconstruction-tier comparison for a cohort this reproduction "
            "does not hold; nothing is asserted from it, and no paper figure is embedded in this "
            "record."
        ),
    )
    # The comparison claim (the paper's cohort is not replicable; tier reconstruction).
    # The paper's Table 3 median is a reconstruction-tier comparison reference only: the
    # reproduction does not hold the paper's frozen cohort, the figure is not derivable
    # from this callable's arguments, and per the honest-exit rule it is therefore NOT
    # embedded in any returned value. Only the reproduction-side median (computed from the
    # supplied arguments) is carried, and the paper figure is absent by construction.
    comparison_record = _make_claim_record_task_10(
        claim_id="median_mean_rank_ic_vs_paper",
        tier="reconstruction",
        evidence_source="paper_table_3",
        verdict="NOT_ASSERTED",
        value={"reproduction_median_mean_rank_ic": median_mean_rank_ic},
        note=(
            "The paper's Table 3 median mean rank-IC was computed on the paper's frozen cohort "
            "using the paper's private per-record artifacts; the reproduction holds neither the "
            "paper's cohort nor its per-record outcomes, so the paper figure cannot be recomputed "
            "here and is deliberately not embedded in this record (see the DATA_ABSENT entry in "
            "the project objections log). The reproduction-side median in 'value' is computed "
            "from the arguments supplied to this callable at full float64 precision. Reported "
            "descriptively; no tier promotion is permitted from this comparison alone."
        ),
    )
    # The sign-closure register: the pinned contracted identity outcome verbatim (EXACT).
    sign_closure_record = _make_claim_record_task_10(
        claim_id="SIGN_CLOSURE_CONTRACTED",
        tier="reproduction",
        evidence_source="closure_battery_this_run",
        verdict="CLOSED" if contracted_identity_ok else "VIOLATED_BY_CONSTRUCTION",
        value={
            "n_positive": n_positive,
            "n_negative": n_negative,
            "n_scorable": n_scorable,
            "n_exact_zero": n_exact_zero,
            "n_nan_mean": n_nan_mean,
            "tolerance_class": "EXACT",
            "exclusion_adjusted_identity_ok": bool(exclusion_adjusted_identity_ok),
            "contracted_identity_ok": bool(contracted_identity_ok),
        },
        note=(
            "The pinned task-list identity n_positive + n_negative == n_scorable (EXACT, tolerance register) is "
            "carried verbatim: it is SATISFIED on this run when no exact zeros or NaN means exist on the scorable arm, "
            "otherwise it is VIOLATED_BY_CONSTRUCTION of the disclosed exclusions (n_exact_zero and n_nan_mean named in "
            "'value'), which are never folded into the partition. The exclusion-adjusted identity "
            "n_positive + n_negative == n_scorable - n_exact_zero - n_nan_mean is asserted as an internal invariant. "
            "Per P03.near_zero_absolute_threshold the exact zero-test is applied on the unrounded float; nothing is "
            "epsilon-compared. Whether exact zeros count as legitimate observations or as diagnostic exclusions is the "
            "operator's ruling to make, and the register here honours both readings by reporting both identities."
        ),
    )
    # Optional NEAR_ZERO_DIAGNOSTIC claim record when the count is non-zero.
    claim_records: list = [closure_record, median_record, comparison_record, sign_closure_record]
    # Emit the diagnostic claim only when triggered.
    if near_zero_count > 0:
        # The diagnostic record per P03.near_zero_absolute_threshold.
        claim_records.append(
            _make_claim_record_task_10(
                claim_id="NEAR_ZERO_DIAGNOSTIC",
                tier="reproduction",
                evidence_source="cohort_scored_this_run",
                verdict="DIAGNOSTIC",
                value={"near_zero_count": near_zero_count, "threshold": float(near_zero_threshold)},
                note=(
                    "Count of scorable records with |mean_rank_ic| strictly below the absolute "
                    "threshold; the sign partition is computed on the unrounded float with the "
                    "exact zero-test and is not re-partitioned on this diagnostic."
                ),
            )
        )
    # ---- the result dictionary ---------------------------------------------------------
    # Assemble the descriptor; no distribution mean/SD anywhere; no significance artefacts.
    result = {
        "n_scorable": n_scorable,
        "n_insufficient": n_insufficient,
        "n_unsupported": n_unsupported,
        "n_positive": n_positive,
        "n_negative": n_negative,
        "n_exact_zero": n_exact_zero,
        "n_nan_mean": n_nan_mean,
        "n_full": n_full,
        "n_partial": n_partial,
        "near_zero_count": near_zero_count,
        "median_mean_rank_ic": median_mean_rank_ic,
        "sign_closure_exclusion_adjusted_identity": bool(exclusion_adjusted_identity_ok),
        "sign_closure_contracted_identity": bool(contracted_identity_ok),
        "diagnostics": diagnostics,
        "claim_records": claim_records,
        "n_cohort": n_cohort,
    }
    # Return the descriptor.
    return result

# -------------------------------------------------------------------------------------------------------------------------------
# Task 10, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_10(
    daily_scores: pd.DataFrame,
    frozen_expression_ids: Sequence[str],
    support: pd.DataFrame,
    output_dir: str | os.PathLike,
    public_operator_registry: Collection[str],
    upstream_halt_reason: object = None,
) -> dict:
    """
    Run Task 10 end to end and write the aggregate CSV.

    Validate the boundary arguments, then run the three Task 10 stages
    in order: aggregate ``daily_scores`` over the frozen cohort, assign
    the three-state ledger under both the primary and reverse
    precedence triples, and execute the closure battery with
    ``n_cohort`` equal to ``len(frozen_expression_ids)``. The resulting
    aggregate frame is written to
    ``<output_dir>/FACTOR_AGGREGATES.csv`` with ``index=False``.
    ``output_dir`` (and any missing parent directories) is created when
    absent; no other file is written.

    Parameters
    ----------
    daily_scores : pd.DataFrame
        Per-expression daily score register in either the flat layout
        or the MultiIndex layout accepted by the aggregator.
    frozen_expression_ids : Sequence[str]
        Ordered, unique, non-empty cohort identifiers; their order sets
        the output row order and their count sets ``n_cohort``.
    support : pd.DataFrame
        One row per expression with the support metadata used by ledger
        assignment.
    output_dir : str | os.PathLike
        Directory receiving ``FACTOR_AGGREGATES.csv``. Created with
        parents when absent.
    public_operator_registry : Collection[str]
        Public operator names used to determine operator support. A
        bare string is rejected downstream.
    upstream_halt_reason : object, optional
        Upstream halt reason echoed by identity in the summary. Default
        is ``None``.

    Returns
    -------
    dict
        Keys ``n_cohort``, ``n_scorable``, ``n_insufficient``,
        ``n_unsupported``, ``n_positive``, ``n_negative``,
        ``near_zero_count``, ``n_full``, ``n_partial``,
        ``median_mean_rank_ic``, ``aggregates_path`` (string path of
        the written CSV), ``task_1_halt_reason`` (the unchanged
        ``upstream_halt_reason`` object), ``n_exact_zero``,
        ``n_nan_mean``, ``diagnostics`` and ``claim_records``.

    Raises
    ------
    TypeError
        If ``frozen_expression_ids`` is a string/bytes or not a
        ``Sequence``; if ``output_dir`` is neither ``str`` nor
        ``os.PathLike``; or if a downstream stage rejects an
        argument type.
    ValueError
        If a frozen identifier is not a non-empty string, if
        identifiers are duplicated, if ``output_dir`` resolves to an
        existing file, or if a downstream stage raises a validation
        or closure error.
    """
    # ---- boundary type assertions (public API) -----------------------------------
    # frozen_expression_ids must be a Sequence (a bare str would iterate chars).
    if isinstance(frozen_expression_ids, (str, bytes)) or not isinstance(frozen_expression_ids, Sequence):
        # Typed refusal for mis-wired callers.
        raise TypeError(f"frozen_expression_ids must be a Sequence[str], got {type(frozen_expression_ids)!r}.")
    # Materialise the frozen list (fresh copy; cohort order is authoritative).
    frozen = list(frozen_expression_ids)
    # Validate element types and emptiness.
    for index, token in enumerate(frozen):
        # Every frozen ID must be a non-empty string.
        if not isinstance(token, str) or token == "":
            raise ValueError(f"frozen_expression_ids[{index}] must be a non-empty str; got {token!r}.")
    # Reject duplicated frozen IDs (each ID identifies exactly one preserved row).
    if len(set(frozen)) != len(frozen):
        # Report the duplicated subset for forensic clarity.
        duplicates = sorted({token for token in frozen if frozen.count(token) > 1})
        raise ValueError(f"frozen_expression_ids carries duplicate ID(s): {duplicates}.")
    # output_dir must be a str or an os.PathLike instance.
    if not isinstance(output_dir, (str, os.PathLike)):
        # Typed refusal for non-path outputs.
        raise TypeError(f"output_dir must be a str or os.PathLike, got {type(output_dir)!r}.")
    # Resolve the output directory once (never mutate the caller's path object).
    out_dir = Path(output_dir)
    # Refuse a file-typed output_dir explicitly.
    if out_dir.exists() and out_dir.is_file():
        # A file at the target path cannot receive the CSV artefact.
        raise ValueError(f"output_dir resolves to an existing file, not a directory: {out_dir}.")
    # Create the directory tree (idempotent for the console's scratch tree).
    out_dir.mkdir(parents=True, exist_ok=True)
    # The cohort size basis for the closure assertion: len(frozen) (Task 8's n_after_dedup).
    n_cohort = len(frozen)
    # ---- Steps 1-3 in the pinned order ---------------------------------------------
    # Step 1: per-factor aggregation in cohort order (coverage ratio is a diagnostic).
    aggregates = aggregate_factors_task_10(daily_scores=daily_scores, cohort_expression_ids=frozen, full_coverage_dates=480)
    # Step 2: the three-state ledger under both precedence triples (single shared code path).
    aggregates = assign_ledger_task_10(
        aggregates=aggregates,
        support=support,
        public_operator_registry=public_operator_registry,
        precedence=PRIMARY_PRECEDENCE_TASK_10,
        reverse_precedence=REVERSE_PRECEDENCE_TASK_10,
        min_valid_dates=20,
    )
    # Step 3: the closure battery on this run's own cohort total (asserts EXACT closures).
    battery = closure_battery_task_10(
        aggregates=aggregates,
        n_cohort=n_cohort,
        near_zero_threshold=1e-12,
        full_coverage_dates=480,
    )
    # ---- persistence beneath output_dir only ---------------------------------------
    # The artefact path (never escapes output_dir).
    aggregates_path = out_dir / "FACTOR_AGGREGATES.csv"
    # Persist index=False (cohort order is the row order; no sentinel index column).
    aggregates.to_csv(aggregates_path, index=False)
    # ---- the pinned scalar summary ---------------------------------------------------
    # Assemble the summary; upstream_halt_reason echoed BY IDENTITY.
    summary = {
        "n_cohort": n_cohort,
        "n_scorable": battery["n_scorable"],
        "n_insufficient": battery["n_insufficient"],
        "n_unsupported": battery["n_unsupported"],
        "n_positive": battery["n_positive"],
        "n_negative": battery["n_negative"],
        "near_zero_count": battery["near_zero_count"],
        "n_full": battery["n_full"],
        "n_partial": battery["n_partial"],
        "median_mean_rank_ic": battery["median_mean_rank_ic"],
        "aggregates_path": str(aggregates_path),
        "task_1_halt_reason": upstream_halt_reason,
        "n_exact_zero": battery["n_exact_zero"],
        "n_nan_mean": battery["n_nan_mean"],
        "diagnostics": battery["diagnostics"],
        "claim_records": battery["claim_records"],
    }
    # Return the summary; no side effects beyond the CSV write beneath output_dir.
    return summary
```

```python
# Task 11: Reference Panel Execution

# ==============================================================================
# Task 11: Reference Panel Execution
# ==============================================================================

# Six pre-registered reference signals (r1..r6) are built as pure functions of
# the nominal FEATURES_PANEL, frozen before any scoring under
# REFERENCE_PREREGISTRATION_TIMESTAMP, scored with the identical Task 9 Step 1
# joint gate (date_gate_task_9) and the identical Task 9 Step 2 scorer
# (mid_rank_pearson_task_9) on the same snapshot used by the cohort, and
# reported at four-decimal precision with the paper's Table 4 figures carried
# only as COMPARISON__[S]__PRIVATE_ARTIFACTS_ONLY comparison values. The six
# references are reported regardless of sign; the exposure envelope is the
# reproduction's own [min, max]; the cohort median's placement inside or
# outside the envelope is stated verbatim; incremental skill is not estimated
# and no P20-forbidden estimator is computed anywhere in this module.

# -------------------------------------------------------------------------------------------------------------------------------
# Module-pinned constants (the pinned contract of the step text)
# -------------------------------------------------------------------------------------------------------------------------------

# The ordered reference identifiers; the order is pinned by P19_REFERENCE_PREREGISTRATION_PARAMETERS.
REFERENCE_IDS: Tuple[str, ...] = ("r1", "r2", "r3", "r4", "r5", "r6")

# The six construct names, verbatim from the paper's Table 4 / P19 signals.
REFERENCE_NAMES: Tuple[str, ...] = (
    "nominal close level",
    "share volume",
    "nominal traded amount",
    "one-day nominal reversal",
    "five-day nominal reversal",
    "twenty-day nominal change",
)

# The six formulas, verbatim from the task step text (EQUIVALENCE_CLASS: the paper specifies constructs in words only).
REFERENCE_FORMULAS: Tuple[str, ...] = (
    "close",
    "volume",
    "amount",
    "neg(close / lag(close, 1) - 1)",
    "neg(close / lag(close, 5) - 1)",
    "close / lag(close, 20) - 1",
)

# Orientation of each construct, verbatim from P19: r4 and r5 are negated, r6 retains positive orientation.
REFERENCE_ORIENTATIONS: Tuple[str, ...] = ("as-is", "as-is", "as-is", "negate", "negate", "as-is")

# Horizon (in trading-calendar rows) of each construct; LEVEL references carry no horizon.
REFERENCE_HORIZONS: Tuple[Optional[int], ...] = (None, None, None, 1, 5, 20)

# The fidelity-class tag of every reference (P19.signal_formula_status); the formulas are equivalence-class representatives.
SIGNAL_FORMULA_STATUS: str = "EQUIVALENCE_CLASS__PAPER_SPECIFIES_CONSTRUCT_IN_WORDS_ONLY"

# The per-row claim stamp for every reported mean (the comparison statement is the discipline being documented).
COMPARISON_STAMP: str = "COMPARISON__[S]__PRIVATE_ARTIFACTS_ONLY"

# The freeze timestamp of the formula definitions, recorded BEFORE any scoring and distinct from cohort_hash's freeze_timestamp.
REFERENCE_PREREGISTRATION_TIMESTAMP: pd.Timestamp = pd.Timestamp("2026-11-11 00:00:00", tz="Asia/Shanghai")

# Evaluator-version pass-through for the reference scoring register (recorded verbatim in every row; not the cohort's tag).
EVALUATOR_VERSION_REFERENCES: str = "task_11_reference_scoring_v1"

# -------------------------------------------------------------------------------------------------------------------------------
# Pinned verbatim wordings (never paraphrased; the house style forbids silent edits to these strings)
# -------------------------------------------------------------------------------------------------------------------------------

# The sign-convention bridge, stated exactly once in the module's outputs (step text, verbatim).
SIGN_CONVENTION_BRIDGE: str = (
    "Because the reversal construct is the negation of the trailing five-row return, "
    "and its mean signed IC is \u22120.0155, the trailing-return construct carries +0.0155 of "
    "mean signed IC \u2014 approximately positive five-day momentum."
)

# The two pre-committed verbatim placement statements (selection on the envelope, never on wording).
INSIDE_ENVELOPE_STATEMENT: str = (
    "The cohort median lies inside the exposure envelope; "
    "the reproduction does not establish matched-sample outperformance."
)

# The outside-branch verbatim placement statement (also never a claim of outperformance).
OUTSIDE_ENVELOPE_STATEMENT: str = (
    "The cohort median lies outside the exposure envelope; "
    "the reproduction does not establish that this is not exposure."
)

# The three pinned statements, verbatim (P19.incremental_skill_estimated == False; P20.incremental_agent_skill_claim_permitted == False;
# P19.post_hoc_status_disclosed == True; P19.promoted_to_profitability_claim == False).
PINNED_STATEMENTS: Tuple[str, ...] = (
    "no incremental prediction beyond these references was estimated",
    "the references were specified after the cohort analysis and are not part of the study's original primary protocol",
    "no reference is promoted to a profitability, significance, or ranking claim",
)

# -------------------------------------------------------------------------------------------------------------------------------
# Public interface of the module (pinned by the task's acceptance contract)
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, module-internal helpers (boundary validation shared by the steps)
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Helper: coerce a caller-supplied timestamp to pd.Timestamp with typed errors.
# -------------------------------------------------------------------------------------------------------------------------------

def _coerce_timestamp_task_11(value: object, name: str) -> pd.Timestamp:
    """
    Coerce a caller-supplied value to a pandas Timestamp.

    The cohort freeze timestamp arrives from outside this module; this
    helper is the single place it is interpreted, so a malformed value
    raises a typed error rather than a silent mis-comparison.

    Parameters
    ----------
    value : object
        Timestamp-like value accepted by ``pd.Timestamp`` (string,
        ``datetime.datetime``, ``pd.Timestamp``, integer epoch).
    name : str
        Argument name embedded in the error message.

    Returns
    -------
    pd.Timestamp
        The coerced timestamp; timezone information is preserved when
        present.

    Raises
    ------
    TypeError
        If ``pd.Timestamp`` cannot coerce ``value``: a ``TypeError``,
        ``ValueError`` or ``OverflowError`` from the constructor is
        caught and re-raised as a ``TypeError`` naming ``name`` and
        the offending value.
    """
    # Attempt the coercion inside a guarded block so failure is typed and named.
    try:
        # Coerce via pandas' canonical timestamp constructor (string, datetime, Timestamp all accepted).
        coerced = pd.Timestamp(value)  # type: ignore[arg-type]
    except (TypeError, ValueError, OverflowError) as exc:
        # Raise a typed error naming the argument and the offending value.
        raise TypeError(f"{name} must be coercible to pandas.Timestamp, got {value!r}.") from exc
    # Return the coerced timestamp to the caller.
    return coerced

# -------------------------------------------------------------------------------------------------------------------------------
# Helper: require the presence of mandatory columns; name the missing ones.
# -------------------------------------------------------------------------------------------------------------------------------

def _require_columns_task_11(columns: pd.Index, required: Tuple[str, ...], caller: str) -> None:
    """
    Halt when required columns are missing from an index.

    Validate that every name in ``required`` appears in ``columns``.
    Used at the module boundary on the panels that arrive from outside
    (FEATURES_PANEL, LABEL_PANEL, TradingCalendar, MASK_REGISTER).
    Every missing name is reported at once so a single call diagnoses
    the whole schema.

    Parameters
    ----------
    columns : pd.Index
        Column index of the frame under test.
    required : Tuple[str, ...]
        Mandated column names; all must be present.
    caller : str
        Calling callable's name, embedded in the error message.

    Raises
    ------
    TypeError
        If ``columns`` is not a ``pandas.Index``.
    ValueError
        If at least one required column is missing; the complete
        missing list is reported in declaration order.
    """
    # Enforce the Index type on the column container.
    if not isinstance(columns, pd.Index):
        # Reject non-Index column containers explicitly.
        raise TypeError(f"{caller}: columns must be a pandas.Index, got {type(columns)!r}.")
    # Compute the set of missing mandatory columns, preserving declaration order.
    missing = [name for name in required if name not in columns]
    # Halt when any mandatory column is absent, naming all offenders.
    if missing:
        # Raise with the complete missing list for a one-shot diagnostic.
        raise ValueError(f"{caller}: missing required column(s) {missing}.")
    # All required columns are present; nothing to raise.
    return None

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1: The six pre-registered reference signals (pure functions of FEATURES_PANEL)
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1, Helper: per-stock trading-calendar-row lag (groupby-shift within stock)
# -------------------------------------------------------------------------------------------------------------------------------

def _per_stock_lag_task_11(close: pd.Series, stock_code: pd.Series, k: int) -> pd.Series:
    """
    Lag a Series by k trading-calendar rows within each stock.

    Implements ``lag(x, k)`` by grouping on ``stock_code`` and calling
    ``Series.shift(k)`` inside each group, so the shift advances by
    trading-calendar rows of the panel and never crosses a stock
    boundary. The first ``k`` rows of every stock become NaN. The
    result is cast to ``float64`` and keeps the input index.

    Parameters
    ----------
    close : pd.Series
        Per-row nominal close level, aligned to ``stock_code``.
    stock_code : pd.Series
        Per-row stock identifier, aligned to ``close``.
    k : int
        Non-negative lag in trading-calendar rows; a bare ``bool`` is
        refused.

    Returns
    -------
    pd.Series
        ``float64`` Series of the same length and index as ``close``;
        NaN in the first ``k`` positions of each stock block.

    Raises
    ------
    TypeError
        If ``close`` or ``stock_code`` is not a ``pandas.Series``, or
        ``k`` is a ``bool`` or non-``int``.
    ValueError
        If ``k`` is negative, or ``close`` and ``stock_code`` do not
        share the same index.
    """
    # Enforce Series type on the close vector.
    if not isinstance(close, pd.Series):
        # Reject non-Series inputs explicitly.
        raise TypeError(f"close must be a pandas.Series, got {type(close)!r}.")
    # Enforce Series type on the stock identifier vector.
    if not isinstance(stock_code, pd.Series):
        # Reject non-Series inputs explicitly.
        raise TypeError(f"stock_code must be a pandas.Series, got {type(stock_code)!r}.")
    # Enforce integer type on the lag order (bools refused).
    if isinstance(k, bool) or not isinstance(k, int):
        # Reject floats and bools that would silently round.
        raise TypeError(f"k must be an int (trading-calendar rows), got {type(k)!r}.")
    # Enforce non-negativity of the lag order.
    if k < 0:
        # A negative lag would peek into the future; forbidden.
        raise ValueError(f"k must be non-negative (lagged only, no future), got {k}.")
    # Enforce index alignment between the two vectors.
    if not close.index.equals(stock_code.index):
        # Halt: positional mixing of the two vectors is a caller defect.
        raise ValueError("close and stock_code must share the same index.")
    # Group-shift confinement: the lag never crosses a stock boundary.
    lagged = close.groupby(stock_code, sort=False).shift(k)
    # Materialise a float64 Series sharing the input index (NaN-tolerant dtype).
    lagged = lagged.astype(np.float64)
    # Return the lagged vector to the caller.
    return lagged

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1: Build the six pre-registered reference Series on FEATURES_PANEL
# -------------------------------------------------------------------------------------------------------------------------------

def build_reference_series_task_11(features_panel: pd.DataFrame) -> dict[str, pd.Series]:
    """
    Build the six pre-registered reference signals from the panel.

    The formulas are the pinned representatives of the paper's
    construct equivalence classes; ``lag`` is the per-stock
    trading-calendar row shift of ``_per_stock_lag_task_11``:

        r1 = close
        r2 = volume
        r3 = amount
        r4 = -(close / lag(close, 1) - 1)
        r5 = -(close / lag(close, 5) - 1)
        r6 = close / lag(close, 20) - 1

    The measure columns are read through ``FrozenFeaturePanel`` with
    ``allow_adj=False``; an ``_adj`` key is never accessed. Every
    returned Series is cast to ``float64``, carries ``name`` equal to
    its reference id and carries ``SIGNAL_FORMULA_STATUS`` in
    ``attrs["signal_formula_status"]``.

    Parameters
    ----------
    features_panel : pd.DataFrame
        Post-mask nominal FEATURES_PANEL. Must carry ``stock_code``,
        ``trade_date``, ``close``, ``volume`` and ``amount``; the
        three measures must be numeric (bool refused). Row order is
        arbitrary; stock-contiguous order is preserved but never
        assumed.

    Returns
    -------
    dict[str, pd.Series]
        Keys ``r1``..``r6`` in ``REFERENCE_IDS`` insertion order;
        values are ``float64`` Series of length
        ``len(features_panel)`` sharing the input index. The first
        ``k`` rows of each stock are NaN for the lag-k references
        ``r4`` (k = 1), ``r5`` (k = 5) and ``r6`` (k = 20).

    Raises
    ------
    TypeError
        If ``features_panel`` is not a ``pandas.DataFrame``, or a
        measure column is non-numeric or boolean.
    ValueError
        If a required nominal column is absent, or the
        ``(stock_code, trade_date)`` coordinate is duplicated.
    """
    # Enforce DataFrame type on the panel (the panel arrives from outside this module).
    if not isinstance(features_panel, pd.DataFrame):
        # Reject non-DataFrame panels explicitly with the observed type.
        raise TypeError(f"features_panel must be a pandas.DataFrame, got {type(features_panel)!r}.")
    # Wrap the panel in the Task 4 embargo facade: any _adj access would halt PermissionError here.
    facade = FrozenFeaturePanel(features_panel, allow_adj=False)
    # Require the nominal coordinate and measure columns by name (an _adj key is never accepted).
    _require_columns_task_11(
        facade.columns,
        ("stock_code", "trade_date", "close", "volume", "amount"),
        "build_reference_series_task_11",
    )
    # Enforce numeric dtypes on the three measure columns (booleans excluded by is_numeric_dtype semantics for floats).
    for measure in ("close", "volume", "amount"):
        # Fetch the candidate column through the facade (a legitimate, non-_adj key).
        column = facade[measure]
        # Halt on non-numeric measures; boolean columns are rejected by is_bool_dtype subtraction.
        if not pd.api.types.is_numeric_dtype(column.dtype) or pd.api.types.is_bool_dtype(column.dtype):
            # Raise a typed error naming the column and its observed dtype.
            raise TypeError(f"{measure} must be numeric (bool refused), got dtype {column.dtype!r}.")
    # Halt on duplicated (stock_code, trade_date) coordinates (lags would double-count stocks).
    if bool(features_panel.duplicated(subset=["stock_code", "trade_date"]).any()):
        # Name the duplicated coordinates for a one-shot diagnostic.
        duplicated = features_panel.loc[features_panel.duplicated(subset=["stock_code", "trade_date"], keep=False), ["stock_code", "trade_date"]]
        # Halt with the complete offending coordinate list.
        raise ValueError(f"features_panel carries duplicated (stock_code, trade_date) cells: {duplicated.to_dict('records')}.")
    # Materialise the stock identifier Series used for the groupby-confined shifts.
    stock_code = features_panel["stock_code"]
    # Materialise the nominal close level as a float64 Series (r1; the lag base of r4/r5/r6).
    close = facade["close"].astype(np.float64)
    # Materialise the share volume as a float64 Series (r2).
    volume = facade["volume"].astype(np.float64)
    # Materialise the nominal traded amount as a float64 Series (r3).
    amount = facade["amount"].astype(np.float64)
    # lag(close, 1): one trading-calendar row back within each stock (NaN for the first row of each stock).
    lag_1 = _per_stock_lag_task_11(close, stock_code, 1)
    # lag(close, 5): five trading-calendar rows back within each stock (NaN for the first five rows of each stock).
    lag_5 = _per_stock_lag_task_11(close, stock_code, 5)
    # lag(close, 20): twenty trading-calendar rows back within each stock (NaN for the first twenty rows of each stock).
    lag_20 = _per_stock_lag_task_11(close, stock_code, 20)
    # r1 := close (nominal close level, as-is).
    r1 = close.copy()
    # r2 := volume (share volume, as-is).
    r2 = volume.copy()
    # r3 := amount (nominal traded amount, as-is).
    r3 = amount.copy()
    # r4 := neg(close / lag(close, 1) - 1) -- one-day nominal reversal, negated construct.
    r4 = -(close / lag_1 - 1.0)
    # r5 := neg(close / lag(close, 5) - 1) -- five-day nominal reversal, negated construct.
    r5 = -(close / lag_5 - 1.0)
    # r6 := close / lag(close, 20) - 1 -- twenty-day nominal change, positive orientation retained.
    r6 = close / lag_20 - 1.0
    # Collect the six Series in the pinned REFERENCE_IDS order.
    raw_series = {"r1": r1, "r2": r2, "r3": r3, "r4": r4, "r5": r5, "r6": r6}
    # Stamp every Series with its reference id and the equivalence-class tag, then return the dict.
    references: dict[str, pd.Series] = {}
    # Walk the pinned id order so the dict preserves REFERENCE_IDS insertion order.
    for reference_id in REFERENCE_IDS:
        # Cast the Series to float64 so all six share one dtype (NaN-tolerant).
        tagged = raw_series[reference_id].astype(np.float64)
        # Give the Series the reference id as its name (downstream frames carry it as the column).
        tagged.name = reference_id
        # Attach the equivalence-class tag so every reported value is traceable to its formula kind.
        tagged.attrs["signal_formula_status"] = SIGNAL_FORMULA_STATUS
        # Store the tagged Series under its reference id.
        references[reference_id] = tagged
    # Return the six tagged Series in pinned order.
    return references

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 1: The pre-registration record (formula freeze, P19 booleans, timestamp)
# -------------------------------------------------------------------------------------------------------------------------------

def preregistration_record_task_11(
    cohort_freeze_timestamp: pd.Timestamp | None = None,
) -> dict[str, object]:
    """
    Return the frozen pre-registration record of the six references.

    Assembles the record documenting that the six formulas were fixed
    before any scoring, that the references are disclosed as post hoc
    and that this record's freeze timestamp is a distinct instant from
    ``cohort_hash``'s freeze timestamp.

    Parameters
    ----------
    cohort_freeze_timestamp : pd.Timestamp | None, optional
        The cohort freeze timestamp, supplied by the caller to prove
        the reference freeze is a distinct instant. Coerced to
        ``pd.Timestamp``; must differ from
        ``REFERENCE_PREREGISTRATION_TIMESTAMP``. Default is ``None``,
        meaning no distinctness assertion.

    Returns
    -------
    dict[str, object]
        ``preregistration_timestamp`` (``pd.Timestamp`` equal to
        ``REFERENCE_PREREGISTRATION_TIMESTAMP``,
        ``signal_formula_status`` (the equivalence-class tag),
        ``signals`` (list of six dicts in ``REFERENCE_IDS`` order with
        keys ``id``, ``name``, ``formula``, ``orientation``,
        ``horizon``, ``signal_formula_status``), and seven P19
        booleans: ``preregistered_before_outcome_computation``
        (``True``), ``post_hoc_status_disclosed`` (``True``),
        ``report_all_regardless_of_sign`` (``True``),
        ``none_dropped_or_reoriented_post_hoc`` (``True``),
        ``promoted_to_profitability_claim`` (``False``),
        ``incremental_skill_estimated`` (``False``),
        ``signs_selected_on_scores`` (``False``).

    Raises
    ------
    TypeError
        If ``cohort_freeze_timestamp`` is supplied and is not
        coercible to ``pd.Timestamp``.
    ValueError
        If ``cohort_freeze_timestamp`` equals
        ``REFERENCE_PREREGISTRATION_TIMESTAMP`` (the two freezes must
        be distinct instants).
    """
    # Read the module-level freeze timestamp (recorded before any scoring; timezone-aware Asia/Shanghai).
    freeze = REFERENCE_PREREGISTRATION_TIMESTAMP
    # Coerce a caller-supplied cohort freeze timestamp when present; None means "not asserted".
    if cohort_freeze_timestamp is not None:
        # Coerce through the boundary helper (typed error on failure).
        cohort_freeze = _coerce_timestamp_task_11(cohort_freeze_timestamp, "cohort_freeze_timestamp")
        # Determine equality safely across timezone conventions (a mixed-tz comparison may be False, never a crash here).
        try:
            # Attempt timestamp equality directly.
            equal = bool(cohort_freeze == freeze)
        except TypeError:
            # A mixed-tz comparison is inconclusive; it is NEVER accepted as proof of distinctness for a naive equal-kind check.
            equal = False
        # Halt when the two freezes are the same instant (pre-registration discipline void otherwise).
        if equal:
            # Raise with both instants for the forensic record.
            raise ValueError(
                f"cohort_freeze_timestamp {cohort_freeze!r} equals REFERENCE_PREREGISTRATION_TIMESTAMP {freeze!r}: "
                "the reference freeze must be a distinct instant from cohort_hash's freeze_timestamp (P19 pre-registration)."
            )
    # Build the six signal records in pinned order, verbatim from the step text.
    signals: list[dict[str, object]] = []
    # Walk the pinned id order so the list order equals REFERENCE_IDS.
    for index, reference_id in enumerate(REFERENCE_IDS):
        # Assemble the per-signal record with the verbatim name, formula, orientation and horizon.
        signals.append(
            {
                # The reference identifier (table index in REFERENCE_TABLE).
                "id": reference_id,
                # The construct name, verbatim from the paper's Table 4.
                "name": REFERENCE_NAMES[index],
                # The pinned formula string (equivalence-class representative of the construct).
                "formula": REFERENCE_FORMULAS[index],
                # Orientation: as-is for r1/r2/r3/r6, negated for r4/r5.
                "orientation": REFERENCE_ORIENTATIONS[index],
                # Horizon in trading-calendar rows; None for the level references.
                "horizon": REFERENCE_HORIZONS[index],
                # The fidelity-class tag of the equivalence-class formula.
                "signal_formula_status": SIGNAL_FORMULA_STATUS,
            }
        )
    # Assemble the frozen record with the pinned P19 booleans and the module freeze timestamp.
    record: dict[str, object] = {
        # The reference freeze timestamp (timezone-aware; equal to REFERENCE_PREREGISTRATION_TIMESTAMP).
        "preregistration_timestamp": freeze,
        # The fidelity-class tag shared by all six signals.
        "signal_formula_status": SIGNAL_FORMULA_STATUS,
        # The six pinned signal records.
        "signals": signals,
        # P19.preregistered_before_outcome_computation == True.
        "preregistered_before_outcome_computation": True,
        # P19.post_hoc_status_disclosed == True (specified after the cohort analysis).
        "post_hoc_status_disclosed": True,
        # P19.report_all_regardless_of_sign == True.
        "report_all_regardless_of_sign": True,
        # P19.none_dropped_or_reoriented_post_hoc == True.
        "none_dropped_or_reoriented_post_hoc": True,
        # P19.promoted_to_profitability_claim == False.
        "promoted_to_profitability_claim": False,
        # P19.incremental_skill_estimated == False (P20.incremental_agent_skill_claim_permitted == False).
        "incremental_skill_estimated": False,
        # P19.signs_selected_on_scores == False.
        "signs_selected_on_scores": False,
    }
    # Return the frozen record to the caller.
    return record

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 2: Score the six references with the identical snapshot, gates and scorer as the cohort
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 2: Score the references (identical Task 9 Step 1 gate + Step 2 scorer) and persist the artefacts
# -------------------------------------------------------------------------------------------------------------------------------

def score_references_task_11(
    features_panel: pd.DataFrame,
    label_panel: pd.DataFrame,
    trading_calendar: pd.DataFrame,
    mask_register: pd.DataFrame,
    output_dir: str | Path,
) -> pd.DataFrame:
    """
    Score the six references on the cohort snapshot and instrument.

    Every ``(reference, date)`` cell is scored with the cohort's own
    Task 9 register builder, so the joint date gate (``n_eligible >=
    5``, ``n_distinct_signal >= 3``, ``n_distinct_label >= 2``) and
    the mid-rank Pearson scorer are identical to the cohort path. The
    post-mask FEATURES_PANEL, LABEL_PANEL, TradingCalendar and
    MASK_REGISTER are consumed in memory; nothing is reloaded from
    disk. Only calendar dates whose ``label_available`` flag is
    ``True`` are scored (every date when the column is absent).
    ``n_valid_dates`` is the reproduction's own admissible-date count;
    ``480`` is a comparison value, and the loss of the first rows of
    ``r4``/``r5``/``r6`` is reported, not corrected.

    Two artifacts are written beneath ``output_dir``:
    ``DAILY_SCORES_REFS.csv`` (columns ``reference_id``,
    ``trade_date``, ``n_eligible``, ``ic``; ``ic`` is NaN wherever the
    gate fails) and ``REFERENCE_MEANS.csv``.

    Parameters
    ----------
    features_panel : pd.DataFrame
        Post-mask nominal FEATURES_PANEL; columns ``stock_code``,
        ``trade_date``, ``close``, ``volume``, ``amount`` and
        ``tradable``. Row order is arbitrary; duplicated coordinates
        halt.
    label_panel : pd.DataFrame
        Post-mask label panel; columns ``stock_code``,
        ``trade_date`` and numeric ``label`` (NaN where undefined).
        Duplicated coordinates halt.
    trading_calendar : pd.DataFrame
        Post-mask TradingCalendar with a ``trade_date`` column and,
        optionally, a boolean ``label_available`` column.
    mask_register : pd.DataFrame
        P34 mask register; columns ``stock_code`` and ``trade_date``
        (may be empty).
    output_dir : str | Path
        Destination directory for the two CSVs; created with parents
        when absent.

    Returns
    -------
    pd.DataFrame
        Six-row table indexed by ``reference_id`` (``r1``..``r6`` in
        ``REFERENCE_IDS`` order) with columns ``mean_rank_ic``
        (``float64``, arithmetic mean of the admissible non-NaN
        ``ic`` values) and ``n_valid_dates`` (``int64``, count of
        admissible dates).

    Raises
    ------
    TypeError
        If a panel argument is not a ``pandas.DataFrame``, or the
        ``label`` column is non-numeric.
    ValueError
        If a required column is absent from a panel, or the
        ``(stock_code, trade_date)`` coordinate is duplicated in
        ``features_panel`` or ``label_panel``.
    """
    # Enforce DataFrame type on the FEATURES_PANEL (arrives from outside this module).
    if not isinstance(features_panel, pd.DataFrame):
        # Reject non-DataFrame panels explicitly.
        raise TypeError(f"features_panel must be a pandas.DataFrame, got {type(features_panel)!r}.")
    # Enforce DataFrame type on the label panel.
    if not isinstance(label_panel, pd.DataFrame):
        # Reject non-DataFrame label panels explicitly.
        raise TypeError(f"label_panel must be a pandas.DataFrame, got {type(label_panel)!r}.")
    # Enforce DataFrame type on the trading calendar.
    if not isinstance(trading_calendar, pd.DataFrame):
        # Reject non-DataFrame calendars explicitly.
        raise TypeError(f"trading_calendar must be a pandas.DataFrame, got {type(trading_calendar)!r}.")
    # Enforce DataFrame type on the mask register.
    if not isinstance(mask_register, pd.DataFrame):
        # Reject non-DataFrame mask registers explicitly.
        raise TypeError(f"mask_register must be a pandas.DataFrame, got {type(mask_register)!r}.")
    # Require the nominal feature columns plus the strict-True tradable flag.
    _require_columns_task_11(
        features_panel.columns,
        ("stock_code", "trade_date", "close", "volume", "amount", "tradable"),
        "score_references_task_11",
    )
    # Require the label-panel coordinate and the label itself.
    _require_columns_task_11(
        label_panel.columns,
        ("stock_code", "trade_date", "label"),
        "score_references_task_11",
    )
    # Require the calendar's date coordinate (label_available is optional and handled below).
    _require_columns_task_11(
        trading_calendar.columns,
        ("trade_date",),
        "score_references_task_11",
    )
    # Require the mask register's coordinate columns.
    _require_columns_task_11(
        mask_register.columns,
        ("stock_code", "trade_date"),
        "score_references_task_11",
    )
    # Halt on duplicated coordinates in the features panel (lags and eligibility would double-count).
    if bool(features_panel.duplicated(subset=["stock_code", "trade_date"]).any()):
        # Name the duplicated coordinate sample for a one-shot diagnostic.
        duplicated = features_panel.loc[features_panel.duplicated(subset=["stock_code", "trade_date"], keep=False), ["stock_code", "trade_date"]]
        # Halt with the complete offending coordinate list.
        raise ValueError(f"features_panel carries duplicated (stock_code, trade_date) cells: {duplicated.to_dict('records')}.")
    # Halt on duplicated coordinates in the label panel.
    if bool(label_panel.duplicated(subset=["stock_code", "trade_date"]).any()):
        # Name the duplicated coordinate sample for a one-shot diagnostic.
        duplicated = label_panel.loc[label_panel.duplicated(subset=["stock_code", "trade_date"], keep=False), ["stock_code", "trade_date"]]
        # Halt with the complete offending coordinate list.
        raise ValueError(f"label_panel carries duplicated (stock_code, trade_date) cells: {duplicated.to_dict('records')}.")
    # Halt on a non-numeric label column (rank association requires ordered numeric values).
    if not pd.api.types.is_numeric_dtype(label_panel["label"].dtype):
        # Raise a typed error naming the observed label dtype.
        raise TypeError(f"label must be numeric, got dtype {label_panel['label'].dtype!r}.")
    # Normalise output_dir to a Path and create it (parents included) when absent.
    out_dir = Path(output_dir)
    # Create the directory tree beneath the caller's destination.
    out_dir.mkdir(parents=True, exist_ok=True)
    # Build the six reference Series (Step 1; the embargo facade guards the nominal columns).
    references = build_reference_series_task_11(features_panel)
    # Anchor the reference values onto the compound (stock_code, trade_date) coordinate in features row order.
    coordinate = pd.MultiIndex.from_frame(features_panel[["stock_code", "trade_date"]])
    # Assemble the reference factor panel with one float64 column per reference id, in pinned order.
    factor_panel = pd.DataFrame(
        {
            reference_id: pd.Series(
                np.asarray(references[reference_id].to_numpy(dtype=np.float64), dtype=np.float64),
                index=coordinate,
            )
            for reference_id in REFERENCE_IDS
        }
    )
    # Pin the compound index level names (build_daily_scores_task_9 requires them).
    factor_panel.index.names = ["stock_code", "trade_date"]
    # Anchor the label vector onto its own compound coordinate (the two panels share the coordinate system post-mask).
    label_coordinate = pd.MultiIndex.from_frame(label_panel[["stock_code", "trade_date"]])
    # Materialise labels as float64 with NaN in place of NA/NaT (exact null semantics of the cohort path).
    labels = pd.Series(
        label_panel["label"].to_numpy(dtype=np.float64, na_value=np.nan),
        index=label_coordinate,
        name="label",
    )
    # Materialise the strict-True tradable flag as a (nullable) boolean Series on the features coordinate.
    tradable = pd.Series(
        features_panel["tradable"].array,
        index=coordinate,
        name="tradable",
    )
    # Select the scoring dates: the calendar's label-available rows when the column exists, else every row.
    if "label_available" in trading_calendar.columns:
        # Strict True test with NA mapped to not-available (a missing flag is never a scored date).
        available = trading_calendar["label_available"].fillna(False).astype(bool)
        # Slice the label-available rows' trade dates in calendar order.
        selected = trading_calendar.loc[available, "trade_date"]
    else:
        # Absent column: every calendar row is scored (the fixture's convention).
        selected = trading_calendar["trade_date"]
    # Coerce the selected dates to Timestamps (build_daily_scores_task_9 re-coerces defensively).
    trade_dates = [pd.Timestamp(value) for value in selected]
    # Score every (reference, date) cell with the cohort's own register builder (identical gate and scorer).
    register = build_daily_scores_task_9(
        factor_panel=factor_panel,
        labels=labels,
        tradable=tradable,
        mask_register=mask_register,
        expression_ids=list(REFERENCE_IDS),
        trade_dates=trade_dates,
        evaluator_version=EVALUATOR_VERSION_REFERENCES,
        parse_failed_ids=(),
    )
    # Flatten the register to the mandated DAILY_SCORES_REFS schema: reference_id, trade_date, n_eligible, ic.
    flat = register.reset_index().rename(columns={"expression_id": "reference_id", "rho": "ic"})
    # Project exactly the four mandated columns in the mandated order.
    reference_register = flat[["reference_id", "trade_date", "n_eligible", "ic"]].copy()
    # Stable-sort by (reference_id, trade_date) so the persisted artifact is deterministic.
    reference_register = reference_register.sort_values(["reference_id", "trade_date"], kind="mergesort").reset_index(drop=True)
    # Aggregate the admissible, non-NaN ic values per reference: arithmetic mean (P03.aggregation_operator).
    ic_by_reference = reference_register.set_index("reference_id")["ic"].groupby(level=0)
    # Arithmetic mean of the admissible cells (skipna; NaN only when no admissible date carries a finite ic).
    mean_rank_ic = ic_by_reference.mean()
    # Count of admissible dates carrying a non-NaN ic (the reproduction's own n_valid_dates; 480 is a comparison value).
    n_valid_dates = ic_by_reference.count()
    # Assemble REFERENCE_MEANS in pinned id order with full precision (rounding is display-only, in Step 3).
    reference_means = pd.DataFrame(
        {
            "mean_rank_ic": mean_rank_ic.reindex(REFERENCE_IDS).astype(np.float64),
            "n_valid_dates": n_valid_dates.reindex(REFERENCE_IDS).fillna(0).astype(np.int64),
        },
        index=pd.Index(REFERENCE_IDS, name="reference_id"),
    )
    # Persist the per-date per-reference register beside the cohort's instruments (index=False: flat CSV).
    reference_register.to_csv(out_dir / "DAILY_SCORES_REFS.csv", index=False)
    # Persist REFERENCE_MEANS with reference_id as the index label.
    reference_means.to_csv(out_dir / "REFERENCE_MEANS.csv", index_label="reference_id")
    # Return the six-row REFERENCE_MEANS table to the caller.
    return reference_means

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 3: Reporting the six references and the inference bound
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Step 3: REFERENCE_TABLE (six rows, four decimals), EXPOSURE_ENVELOPE, COHORT_MEDIAN_PLACEMENT, pinned statements
# -------------------------------------------------------------------------------------------------------------------------------

def build_reference_report_task_11(
    reference_means: pd.DataFrame,
    factor_aggregates: pd.DataFrame,
    output_dir: str | Path,
    paper_reference_targets: dict[str, float],
    paper_exposure_envelope: tuple[float, float],
    paper_valid_dates: int,
) -> dict[str, object]:
    """
    Build the six-row reference table and inference-bound records.

    Reports every reference regardless of sign; none may be dropped,
    re-oriented or replaced. The exposure envelope is
    ``[min(mean_rank_ic), max(mean_rank_ic)]`` over the six unrounded
    means; the table displays four decimals, but the envelope never
    uses rounded values. The paper's nominal envelope is echoed only
    as a comparison value. The cohort median ``mean_rank_ic`` from
    Task 10 is placed relative to the envelope with one of two
    pre-committed statements; neither branch claims outperformance,
    because the cohort and references are not a matched sample. The
    three pinned statements are returned verbatim. Writes
    ``REFERENCE_TABLE.csv`` beneath ``output_dir``.

    Parameters
    ----------
    reference_means : pd.DataFrame
        Reference means indexed by ``reference_id``; the index must
        carry exactly ``r1``..``r6`` with no duplicates. Required
        columns are ``mean_rank_ic`` (finite floats) and
        ``n_valid_dates`` (non-negative whole numbers, no nulls).
    factor_aggregates : pd.DataFrame
        Factor aggregates from Task 10. Must carry ``mean_rank_ic``.
        When a ``state_primary`` column is present, the cohort median
        uses only rows whose value equals ``"scorable"``; otherwise
        all finite means are used.
    output_dir : str | Path
        Directory that receives ``REFERENCE_TABLE.csv``; created with
        parents when absent.
    paper_reference_targets : dict[str, float]
        The paper's Table 4 mean signed rank-IC for all six reference
        ids, used only as comparison values and echoed per row.
    paper_exposure_envelope : tuple[float, float]
        The paper's nominal comparison interval ``(min, max)``;
        echoed unchanged in the returned mapping.
    paper_valid_dates : int
        The paper's valid-date comparison figure; echoed per row and
        never used as a target.

    Returns
    -------
    dict[str, object]
        Mapping with exactly these keys: ``reference_table``, a
        six-row ``pd.DataFrame`` indexed by ``reference_id`` with
        columns ``mean_rank_ic`` (four-decimal display),
        ``n_valid_dates``, ``stamp``, ``paper_target_mean_rank_ic``
        and ``paper_valid_dates``; ``exposure_envelope``, a
        two-element list of the unrounded reproduction means;
        ``paper_exposure_envelope``, a two-element list echoing the
        passed comparison interval; ``cohort_median_placement``, with
        keys ``cohort_median_mean_rank_ic``, ``envelope_min``,
        ``envelope_max``, ``within_envelope`` and
        ``placement_statement``; ``pinned_statements``, the three
        verbatim statements; and ``sign_convention_bridge``, the
        verbatim bridge sentence.

    Raises
    ------
    TypeError
        If ``reference_means`` or ``factor_aggregates`` is not a
        ``pd.DataFrame``; if ``paper_reference_targets`` is not a
        dict; if ``paper_exposure_envelope`` is not a tuple; if
        ``paper_valid_dates`` is not an ``int`` (``bool`` is
        rejected); or if a paper target or envelope endpoint cannot
        be coerced to float.
    ValueError
        If required columns are missing; if the reference index has
        duplicate, missing or extra ids; if a mean is non-finite; if
        ``n_valid_dates`` contains nulls, values that do not convert
        to numbers, non-finite values (e.g. ``inf``), fractional
        values or negatives; if a paper target is missing or
        non-finite; if the envelope has other than two elements, has
        non-finite endpoints or has low greater than high; or if
        ``paper_valid_dates`` is negative.
    AssertionError
        If the constructed table does not have exactly six rows.

    Notes
    -----
    The envelope is computed from the six means gathered in pinned
    ``r1``..``r6`` order, independently of the caller's row order. A
    NaN cohort median is never classified as inside the envelope. No
    rank correlation, null hypothesis or incremental-skill estimate
    is computed.
    """
    # Enforce DataFrame type on REFERENCE_MEANS (arrives from outside this module when called directly).
    if not isinstance(reference_means, pd.DataFrame):
        # Reject non-DataFrame tables explicitly.
        raise TypeError(f"reference_means must be a pandas.DataFrame, got {type(reference_means)!r}.")
    # Enforce DataFrame type on FACTOR_AGGREGATES.
    if not isinstance(factor_aggregates, pd.DataFrame):
        # Reject non-DataFrame aggregates explicitly.
        raise TypeError(f"factor_aggregates must be a pandas.DataFrame, got {type(factor_aggregates)!r}.")
    # Normalise output_dir to a Path and create it (parents included) when absent.
    out_dir = Path(output_dir)
    # Create the directory tree beneath the caller's destination.
    out_dir.mkdir(parents=True, exist_ok=True)
    # Read the index labels of REFERENCE_MEANS (the reference ids live on the index).
    reference_ids = list(reference_means.index)
    # Halt when any id is duplicated (each of the six rows must be independently addressable).
    if len(reference_ids) != len(set(reference_ids)):
        # Name the duplicated ids for the diagnostic.
        duplicated = sorted({token for token in reference_ids if reference_ids.count(token) > 1})
        # Halt with the duplicated subset listed.
        raise ValueError(f"reference_means index carries duplicate reference id(s): {duplicated}.")
    # Compute the missing and extra ids relative to the pinned six-id order (none may be dropped or swapped).
    missing_ids = [token for token in REFERENCE_IDS if token not in reference_ids]
    # Compute the extra ids relative to the pinned six-id order (no surrogate ids permitted).
    extra_ids = [token for token in reference_ids if token not in REFERENCE_IDS]
    # Halt unless the index carries EXACTLY the six reference ids (P19.none_dropped_or_reoriented_post_hoc).
    if missing_ids or extra_ids:
        # Raise with both lists for a one-shot diagnostic.
        raise ValueError(
            f"reference_means index must carry exactly the six reference ids {list(REFERENCE_IDS)}; "
            f"missing={missing_ids}, extra={extra_ids} (none may be dropped, re-oriented or replaced post hoc)."
        )
    # Require the two mandated columns; list both when absent.
    _require_columns_task_11(
        reference_means.columns,
        ("mean_rank_ic", "n_valid_dates"),
        "build_reference_report_task_11",
    )
    # Read the caller's means column (its order may differ from the pinned order; positions resolve that below).
    mean_column = reference_means["mean_rank_ic"]
    # Materialise the six means at full float64 precision with NA mapped to NaN (exact null semantics).
    means_all = mean_column.to_numpy(dtype=np.float64, na_value=np.nan)
    # Resolve each pinned id to its positional slot in the caller's frame (unique index enforced above).
    positions = [reference_means.index.get_loc(token) for token in REFERENCE_IDS]
    # Gather the means in pinned REFERENCE_IDS order (positional, label-safe).
    means_values = means_all[np.asarray(positions, dtype=np.int64)]
    # Halt when any reference mean is not finite (a missing mean is never reported as if computed here).
    if not bool(np.isfinite(means_values).all()):
        # Identify the offending ids for the diagnostic.
        non_finite = [token for token, value in zip(REFERENCE_IDS, means_values) if not np.isfinite(value)]
        # Halt with the offending ids listed.
        raise ValueError(f"reference_means carries non-finite mean_rank_ic for reference id(s) {non_finite}.")
    # Read the caller's counts column and detect NA cells positionally (the caller's order may differ from pinned).
    n_valid_column = reference_means["n_valid_dates"]
    # Compute the NA mask over the whole column (any NA anywhere is a defect to name, wherever it sits).
    na_positions = np.asarray(n_valid_column.isna(), dtype=np.bool_)
    # Halt on any null count (a missing count is not a silently zero-valued one).
    if bool(na_positions.any()):
        # Identify the offending ids by position in the caller's frame order.
        na_ids = [reference_means.index[position] for position in np.flatnonzero(na_positions)]
        # Halt with the offending ids listed.
        raise ValueError(f"reference_means carries null n_valid_dates for reference id(s) {na_ids}.")
    # Read the caller's counts column and coerce it to float with NA mapped to NaN (exact null semantics).
    counts_float = pd.to_numeric(n_valid_column, errors="raise").to_numpy(dtype=np.float64, na_value=np.nan)
    # Halt when any count failed to convert or is not finite (a missing count is never a zero).
    if not bool(np.isfinite(counts_float).all()):
        # Halt naming the observed raw column values for the diagnostic.
        raise ValueError(f"n_valid_dates must convert to finite numbers, got {list(n_valid_column)}.")
    # Halt on fractional counts (a cardinality is a whole number).
    if not bool((counts_float == np.floor(counts_float)).all()):
        # Halt with the fractional offenders for the diagnostic.
        raise ValueError(f"n_valid_dates must be whole numbers, got {counts_float.tolist()}.")
    # Resolve each pinned id to its positional slot and cast to plain int64 counts in pinned order.
    counts_values = counts_float[np.asarray(positions, dtype=np.int64)].astype(np.int64)
    # Halt on negative counts (nonsensical cardinalities).
    if bool((counts_values < 0).any()):
        # Halt naming the smallest observed count.
        raise ValueError(f"n_valid_dates must be non-negative, got {counts_values.tolist()}.")
    # Enforce the paper comparison literals: a dict carrying all six ids, each coercible to a finite float.
    if not isinstance(paper_reference_targets, dict):
        # Reject non-dict paper comparison mappings explicitly.
        raise TypeError(f"paper_reference_targets must be a dict[str, float], got {type(paper_reference_targets)!r}.")
    # Compute the ids missing from the caller's comparison mapping.
    missing_targets = [token for token in REFERENCE_IDS if token not in paper_reference_targets]
    # Halt when the comparison mapping does not cover all six ids.
    if missing_targets:
        # Raise with the missing ids listed.
        raise ValueError(f"paper_reference_targets must carry all six reference ids; missing: {missing_targets}.")
    # Materialise the paper comparison means in pinned order as finite floats.
    try:
        # Cast each supplied comparison value to float64 with validation.
        paper_targets = np.asarray([float(paper_reference_targets[token]) for token in REFERENCE_IDS], dtype=np.float64)
    except (TypeError, ValueError) as exc:
        # Raise a typed error naming the failing id where possible.
        raise TypeError(f"paper_reference_targets must map each id to a float; detail: {exc}.") from exc
    # Halt when any comparison value is not finite.
    if not bool(np.isfinite(paper_targets).all()):
        # Identify the offending ids for the diagnostic.
        non_finite_targets = [token for token, value in zip(REFERENCE_IDS, paper_targets) if not np.isfinite(value)]
        # Halt with the offending ids listed.
        raise ValueError(f"paper_reference_targets carries non-finite value(s) for reference id(s) {non_finite_targets}.")
    # Enforce the paper exposure envelope: a two-element float pair, finite, min not exceeding max.
    if not isinstance(paper_exposure_envelope, tuple):
        # Reject non-tuple envelopes explicitly (lists are refused at the boundary to pin the contract).
        raise TypeError(f"paper_exposure_envelope must be a tuple[float, float], got {type(paper_exposure_envelope)!r}.")
    # Enforce the pair's arity.
    if len(paper_exposure_envelope) != 2:
        # Halt with the observed arity for the diagnostic.
        raise ValueError(f"paper_exposure_envelope must carry exactly two elements (min, max), got {len(paper_exposure_envelope)}.")
    # Coerce the two envelope endpoints to float with finite-value checks.
    try:
        # Cast both endpoints to float (ints and numeric strings coerce honestly here).
        envelope_low = float(paper_exposure_envelope[0])
        # Cast the high endpoint likewise.
        envelope_high = float(paper_exposure_envelope[1])
    except (TypeError, ValueError) as exc:
        # Raise a typed error describing the conversion failure.
        raise TypeError(f"paper_exposure_envelope endpoints must be float-coercible: {exc}.") from exc
    # Halt on non-finite endpoints.
    if not (np.isfinite(envelope_low) and np.isfinite(envelope_high)):
        # Halt naming the offending endpoint values.
        raise ValueError(f"paper_exposure_envelope endpoints must be finite, got ({envelope_low}, {envelope_high}).")
    # Halt when the low endpoint exceeds the high endpoint (a malformed comparison interval).
    if envelope_low > envelope_high:
        # Halt with the malformed pair for the diagnostic.
        raise ValueError(f"paper_exposure_envelope low ({envelope_low}) exceeds high ({envelope_high}).")
    # Enforce a non-negative integer comparison figure (bool refused).
    if isinstance(paper_valid_dates, bool) or not isinstance(paper_valid_dates, int):
        # Reject bools, floats and strings explicitly.
        raise TypeError(f"paper_valid_dates must be an int, got {type(paper_valid_dates)!r}.")
    # Halt on negative comparison figures.
    if paper_valid_dates < 0:
        # Halt with the observed value for the diagnostic.
        raise ValueError(f"paper_valid_dates must be non-negative, got {paper_valid_dates}.")
    # Exposure envelope of the reproduction's own six means, computed on the UNROUNDED values (step text: exposure_envelope := [min, max]).
    envelope = [float(np.min(means_values)), float(np.max(means_values))]
    # Assemble REFERENCE_TABLE with the four-decimal DISPLAY (rounding never touches the envelope) and the comparison columns.
    reference_table = pd.DataFrame(
        {
            # The reproduction mean, displayed at four decimals (rounding is display-only).
            "mean_rank_ic": np.round(means_values, 4),
            # The reproduction's own valid-date count per reference (lag-window asymmetry disclosed, not corrected).
            "n_valid_dates": counts_values,
            # The comparison stamp: the reported value is reproduced; the value against the paper's is a comparison.
            "stamp": [COMPARISON_STAMP] * len(REFERENCE_IDS),
            # The paper's Table 4 comparison mean per reference (echoed; never asserted, never targeted).
            "paper_target_mean_rank_ic": paper_targets,
            # The paper's P19 valid-dates comparison figure (echoed per row; never asserted).
            "paper_valid_dates": np.full(len(REFERENCE_IDS), int(paper_valid_dates), dtype=np.int64),
        },
        index=pd.Index(REFERENCE_IDS, name="reference_id"),
    )
    # Internal invariant: the table never carries fewer or more than six rows (none dropped, none re-oriented).
    if reference_table.shape[0] != len(REFERENCE_IDS):
        # Halt loudly if the construction above ever broke the cardinality contract.
        raise AssertionError(f"REFERENCE_TABLE cardinality is {reference_table.shape[0]}; expected {len(REFERENCE_IDS)}.")
    # Require the FACTOR_AGGREGATES mean column before computing the cohort median.
    _require_columns_task_11(
        factor_aggregates.columns,
        ("mean_rank_ic",),
        "build_reference_report_task_11",
    )
    # Coerce the aggregate means to float64 with NaN in place of NA (exact null semantics).
    aggregate_means = pd.to_numeric(factor_aggregates["mean_rank_ic"], errors="raise").to_numpy(dtype=np.float64, na_value=np.nan)
    # When a ledger column is present, restrict the median to the scorable arm (Task 10's own definition).
    if "state_primary" in factor_aggregates.columns:
        # Materialise the scorable mask from the ledger verdicts (string comparison; NA maps to not-scorable).
        scorable_mask = factor_aggregates["state_primary"].astype(str).to_numpy() == "scorable"
        # Slice the scorable arm of the means.
        aggregate_means = aggregate_means[scorable_mask]
    # Select the finite means (non-scorable records carry NaN and are excluded, never imputed).
    finite_means = aggregate_means[np.isfinite(aggregate_means)]
    # Cohort median at FULL precision; NaN propagates honestly when no finite mean exists.
    cohort_median = float(np.median(finite_means)) if finite_means.size else float("nan")
    # Read the envelope endpoints once for the placement record.
    envelope_min, envelope_max = envelope
    # Within-envelope iff the median lies in [envelope_min, envelope_max]; a NaN median is never "inside".
    within_envelope = bool(envelope_min <= cohort_median and cohort_median <= envelope_max)
    # Select the pre-committed verbatim placement statement (neither branch claims outperformance).
    placement_statement = INSIDE_ENVELOPE_STATEMENT if within_envelope else OUTSIDE_ENVELOPE_STATEMENT
    # Assemble the COHORT_MEDIAN_PLACEMENT record with the pinned keys.
    cohort_median_placement = {
        # The reproduction's median mean_rank_ic at full precision (Task 10; NaN when undefined).
        "cohort_median_mean_rank_ic": cohort_median,
        # The envelope's lower endpoint (unrounded).
        "envelope_min": envelope_min,
        # The envelope's upper endpoint (unrounded).
        "envelope_max": envelope_max,
        # The placement verdict (pre-committed wording; no outperformance claim either way).
        "within_envelope": within_envelope,
        # The verbatim placement statement selected by within_envelope.
        "placement_statement": placement_statement,
    }
    # Persist REFERENCE_TABLE.csv with reference_id as the index label (the step text's publication vehicle).
    reference_table.to_csv(out_dir / "REFERENCE_TABLE.csv", index_label="reference_id")
    # Assemble the returned mapping (keys pinned by the task's acceptance contract).
    report: dict[str, object] = {
        # The six-row table (four-decimal display; stamps on every row).
        "reference_table": reference_table,
        # The reproduction's exposure envelope over the unrounded means.
        "exposure_envelope": envelope,
        # The paper's nominal envelope, echoed as a comparison value (never a target).
        "paper_exposure_envelope": [envelope_low, envelope_high],
        # The placement of the cohort median inside or outside the envelope.
        "cohort_median_placement": cohort_median_placement,
        # The three pinned statements, verbatim.
        "pinned_statements": list(PINNED_STATEMENTS),
        # The sign-convention bridge sentence, stated exactly once (it is not re-stated in the table cells).
        "sign_convention_bridge": SIGN_CONVENTION_BRIDGE,
    }
    # Return the report mapping to the caller.
    return report

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

# -------------------------------------------------------------------------------------------------------------------------------
# Task 11, Orchestrator: Steps 1 to 3 in pinned order; persist the three CSVs; return the bundle
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_11(
    features_panel: pd.DataFrame,
    label_panel: pd.DataFrame,
    trading_calendar: pd.DataFrame,
    mask_register: pd.DataFrame,
    factor_aggregates: pd.DataFrame,
    output_dir: str | Path,
    paper_reference_targets: dict[str, float],
    paper_exposure_envelope: tuple[float, float],
    paper_valid_dates: int,
    cohort_freeze_timestamp: pd.Timestamp | None = None,
) -> dict[str, object]:
    """
    Run Task 11 Steps 1-3 and return the reference bundle.

    First materialises the reference preregistration record via
    ``preregistration_record_task_11``, before any scoring or file
    write; this enforces that the reference freeze timestamp differs
    from the cohort freeze timestamp. Then delegates Step 1/Step 2
    scoring to ``score_references_task_11`` using the same in-memory
    feature, label, calendar and mask handles as the cohort. Then
    delegates Step 3 reporting to
    ``build_reference_report_task_11``. Returns a bundle containing
    the means, table, envelope, placement record, pinned statements,
    bridge sentence and preregistration record. The scoring step
    writes the Step 1/2 CSV artefacts and the reporting step writes
    ``REFERENCE_TABLE.csv`` beneath ``output_dir``.

    Parameters
    ----------
    features_panel : pd.DataFrame
        Post-mask nominal features panel.
    label_panel : pd.DataFrame
        Post-mask label panel with ``stock_code``, ``trade_date`` and
        ``label``.
    trading_calendar : pd.DataFrame
        Post-mask trading calendar with ``trade_date`` and optional
        ``label_available``.
    mask_register : pd.DataFrame
        The P34 mask register with ``stock_code`` and ``trade_date``.
    factor_aggregates : pd.DataFrame
        Factor aggregates from Task 10; must carry ``mean_rank_ic``.
    output_dir : str | Path
        Directory receiving the CSV artefacts; created if absent.
    paper_reference_targets : dict[str, float]
        The paper's Table 4 comparison means for all six reference
        ids.
    paper_exposure_envelope : tuple[float, float]
        The paper's nominal comparison envelope.
    paper_valid_dates : int
        The paper's valid-date comparison figure.
    cohort_freeze_timestamp : pd.Timestamp | None, optional
        The cohort freeze timestamp. If it equals
        ``REFERENCE_PREREGISTRATION_TIMESTAMP``, the preregistration
        helper raises before scoring. Default is ``None``.

    Returns
    -------
    dict[str, object]
        Bundle with keys ``reference_means`` (six-row Step 2 table),
        ``reference_table`` (six-row Step 3 display table),
        ``exposure_envelope`` (unrounded reproduction interval),
        ``cohort_median_placement`` (placement record),
        ``pinned_statements`` (three verbatim statements),
        ``sign_convention_bridge`` (bridge sentence),
        ``preregistration_record`` (Step 0 record) and
        ``reference_preregistration_timestamp`` (its timestamp).

    Raises
    ------
    TypeError
        Propagated from the preregistration, scoring or reporting
        callables when an argument fails a boundary type check.
    ValueError
        If ``cohort_freeze_timestamp`` equals the reference
        preregistration timestamp, or propagated from a step callable
        that rejects its input, such as missing columns, duplicate
        coordinates or cardinality breaches.
    """
    # Step 0: materialise the preregistration record FIRST (pre-freeze discipline; halts before any artefact is written).
    preregistration_record = preregistration_record_task_11(cohort_freeze_timestamp=cohort_freeze_timestamp)
    # Steps 1-2: score the six references with the identical snapshot, gates and scorer as the cohort.
    reference_means = score_references_task_11(
        features_panel=features_panel,
        label_panel=label_panel,
        trading_calendar=trading_calendar,
        mask_register=mask_register,
        output_dir=output_dir,
    )
    # Step 3: build the report, the envelope, the placement record and the pinned statements.
    report = build_reference_report_task_11(
        reference_means=reference_means,
        factor_aggregates=factor_aggregates,
        output_dir=output_dir,
        paper_reference_targets=paper_reference_targets,
        paper_exposure_envelope=paper_exposure_envelope,
        paper_valid_dates=paper_valid_dates,
    )
    # Assemble the orchestrator bundle with exactly the keys pinned by the task's acceptance contract.
    bundle: dict[str, object] = {
        # The six-row REFERENCE_MEANS table from Step 2 (index reference_id).
        "reference_means": reference_means,
        # The six-row REFERENCE_TABLE (four-decimal display; stamps; paper comparison columns).
        "reference_table": report["reference_table"],
        # The reproduction's exposure envelope over the unrounded means.
        "exposure_envelope": report["exposure_envelope"],
        # The placement of the cohort median inside or outside the envelope.
        "cohort_median_placement": report["cohort_median_placement"],
        # The three pinned statements, verbatim.
        "pinned_statements": report["pinned_statements"],
        # The sign-convention bridge sentence, stated exactly once.
        "sign_convention_bridge": report["sign_convention_bridge"],
        # The frozen pre-outcome pre-registration record of the six references.
        "preregistration_record": preregistration_record,
        # The reference freeze timestamp (timezone-aware; distinct from the cohort freeze).
        "reference_preregistration_timestamp": preregistration_record["preregistration_timestamp"],
    }
    # Return the bundle to the caller.
    return bundle
```

```python
# Task 12: Dual-Implementation Cross-Check of the Measurement Instrument

# ==============================================================================
# Task 12: Dual-Implementation Cross-Check of the Measurement Instrument
# ==============================================================================

# Verbatim scope (P18.spot_check_targets_instrument_not_results == True):
# This validates the measurement instrument, not the results.
# No claim about the aggregate distribution follows from this check.
#
# Provenance pins honoured here:
# * P18.spot_check_cell_selection_timing == "A_PRIORI" — the nine cells are
#   selected from DAILY_SCORES (after Task 9) and from the raw panels before
#   any alternative computation happens.
# * P18.spot_check_independence ==
#   "SEPARATE_TIE_RANKING_PEARSON_IMPLEMENTATION" — the alternative path
#   shares no code, no import and no helper module with the primary path.
# * P18.spot_check_pass_requirement == "9_of_9" — carried verbatim in the
#   report; adjusted to "8_of_8" with disclosure iff r9 is NOT_SATISFIABLE.
# * P18.tolerance_nine_cell == "UNDISCLOSED" — the reproduction declares,
#   as [R], tol_abs = 1e-10 (and tol_rel = 1e-12) for the nine-cell check.
#   Reusing P18.tolerance_audit_recomputation == "THREE_DECIMALS" here would
#   be a category error; it covers the two mean-Sharpe recomputations only.
# * P18.spot_check_pass_count_reported == False — the paper does not publish
#   its own pass count; the 9-of-9 claim is this reproduction's, not a paper
#   claim. The report states that verbatim.
# * The mid-rank convention is average-tie (mid) ranks, ascending orientation,
#   per P02 ("Daily rank-IC is the Pearson correlation of average-tie
#   cross-sectional ranks"); the tied-input agreement between the two
#   implementations must be bit-identical.
#
# Independence note (module sources scanned by the verifier):
# This module references no closed-form correlation helper of numpy, no
# statistical library, no pandas inbuilt ranking method.
# The alternative path exposes exactly two callables:
# alternative_mid_ranks_task_12 and alternative_rank_ic_task_12.
#
# Upstream primitives (reused by file location, never re-written): the strict-True
# tradability mask and the P34 mask -> per-date coordinate map are imported from the
# verified Task 9 module (task_9.py, the sibling of this file). The alternative
# implementation never touches them; they are used solely to rebuild the primary
# path's eligible cross-section exactly, as the pinned contract requires.


# Upstream primitives rebound to the names the Task 9 cell defines in this namespace (Section 3.2(a)).
_STRICT_TRUE_MASK_UPSTREAM = _strict_true_mask_task_9
_MASK_COORDINATE_MAP_UPSTREAM = _mask_coordinate_map_task_9

# The exact columns of the frozen artifact, in pinned order (contract of Step 1).
NINE_CELLS_COLUMNS_TASK_12: List[str] = ["expression_id", "trade_date", "selection_rule"]

# The exact columns of the comparison table, in pinned order (contract of Step 3).
DUAL_CHECK_COLUMNS_TASK_12: List[str] = [
    "expression_id",
    "trade_date",
    "primary_rho",
    "alternative_rho",
    "delta",
    "pass_at_tolerance",
]

# The declared, reproduction-pinned tolerance for the nine-cell check. Source: [R]
# declaration recorded in P18 (the paper pins no nine-cell tolerance; the audit
# recomputation tolerance THREE_DECIMALS is expressly not reused here).
TOL_ABS_TASK_12: float = 1e-10
TOL_REL_TASK_12: float = 1e-12

# Selection-rule tags pinned by the step text (r1..r9).
RULE_IDS_TASK_12: Tuple[str, ...] = ("r1", "r2", "r3", "r4", "r5", "r6", "r7", "r8", "r9")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1: Select the nine a priori cells and freeze NINE_CELLS
# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1, Helper: Validate the compound (stock_code, trade_date) index of a panel/Series.
# -------------------------------------------------------------------------------------------------------------------------------

def _require_compound_index_task_12(
    obj: Union[pd.DataFrame, pd.Series],
    name: str,
) -> None:
    """
    Validate that ``obj`` carries a unique compound index.

    The index must be a ``MultiIndex`` whose level names include both
    ``stock_code`` and ``trade_date``, and every cell must be unique;
    label-based alignment across the panels relies on these conditions.
    The level-name test is a subset test, so a ``MultiIndex`` carrying
    those two names plus additional levels also passes; only the
    presence of the two names and cell uniqueness are enforced.

    Parameters
    ----------
    obj : Union[pd.DataFrame, pd.Series]
        Panel or vector whose index is validated.
    name : str
        Argument name used in the error messages.

    Raises
    ------
    TypeError
        If ``obj`` is not a ``pandas.DataFrame`` or ``pandas.Series``.
    ValueError
        If the index is not a ``MultiIndex`` carrying ``stock_code``
        and ``trade_date``, or if it carries duplicate cells.
    """
    # Enforce the pandas container type before any structural check.
    if not isinstance(obj, (pd.DataFrame, pd.Series)):
        # Reject non-pandas arguments explicitly.
        raise TypeError(f"{name} must be a pandas DataFrame or Series, got {type(obj)!r}.")

    # Enforce the two-level compound index carrying the pinned level names.
    if not isinstance(obj.index, pd.MultiIndex) or not {"stock_code", "trade_date"} <= set(obj.index.names):
        # Halt: alignment in this study is label-based on (stock_code, trade_date).
        raise ValueError(f"{name}.index must be a MultiIndex carrying 'stock_code' and 'trade_date'.")

    # Enforce uniqueness of cells; duplicates would double-count stocks.
    if not bool(obj.index.is_unique):
        # Halt: a duplicated (stock_code, trade_date) cell voids denominators.
        raise ValueError(f"{name} must carry a unique (stock_code, trade_date) index.")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1, Helper: Validate the DAILY_SCORES register and return it with a RangeIndex.
# -------------------------------------------------------------------------------------------------------------------------------

def _validated_register_task_12(daily_scores: pd.DataFrame) -> pd.DataFrame:
    """
    Validate the DAILY_SCORES register and return a flat sorted copy.

    The register must be a ``pandas.DataFrame`` indexed by
    ``MultiIndex[expression_id, trade_date]`` and carry the columns
    ``rho``, ``n_eligible``, ``n_distinct_signal``,
    ``n_distinct_label``, ``admissible`` and ``reject_reason``. The
    index is reset, the ``trade_date`` column coerced to
    datetime64[ns], and the rows stably sorted by (``expression_id``
    ascending, ``trade_date`` ascending); values are not re-audited.

    Parameters
    ----------
    daily_scores : pd.DataFrame
        Long-format register produced by the scoring step, or a planted
        equivalent.

    Returns
    -------
    pd.DataFrame
        Copy with a flat ``RangeIndex``, all input columns, a
        datetime64[ns] ``trade_date``, and the pinned sort order.

    Raises
    ------
    TypeError
        If ``daily_scores`` is not a ``pandas.DataFrame``.
    ValueError
        If the index is not a ``MultiIndex`` carrying ``expression_id``
        and ``trade_date``, or if any pinned column is missing.
    """
    # Enforce the DataFrame container type at the module boundary.
    if not isinstance(daily_scores, pd.DataFrame):
        # Reject non-DataFrame registers explicitly.
        raise TypeError(f"daily_scores must be a pandas.DataFrame, got {type(daily_scores)!r}.")

    # Enforce the pinned index level names.
    if not isinstance(daily_scores.index, pd.MultiIndex) or not {"expression_id", "trade_date"} <= set(daily_scores.index.names):
        # Halt: the register's coordinates are (expression_id, trade_date).
        raise ValueError("daily_scores.index must be a MultiIndex carrying 'expression_id' and 'trade_date'.")

    # The columns this module reads must exist (upstream guarantees they do).
    required_columns = {"rho", "n_eligible", "n_distinct_signal", "n_distinct_label", "admissible", "reject_reason"}

    # Compute the missing-column list for an informative halt.
    missing_columns = sorted(required_columns - set(daily_scores.columns))

    # Halt when any pinned column is absent.
    if missing_columns:
        # Trace the missing columns explicitly.
        raise ValueError(f"daily_scores is missing pinned column(s): {missing_columns}.")

    # Copy so the caller's frame is never mutated and the sort is local.
    cells = daily_scores.reset_index().copy()

    # Normalise the trade_date column to datetime64[ns] (registers may carry any datetime-like dtype).
    cells["trade_date"] = pd.to_datetime(cells["trade_date"])

    # Sort by (expression_id ascending, trade_date ascending) — the pinned tie-breaking.
    cells = cells.sort_values(["expression_id", "trade_date"], kind="mergesort").reset_index(drop=True)

    # Return the flat, deterministically ordered frame.
    return cells

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1, Helper: Compute the eligible signal and label vectors for one cell.
# -------------------------------------------------------------------------------------------------------------------------------

def _eligible_vectors_task_12(
    factor_panel: pd.DataFrame,
    labels: pd.Series,
    expression_id: str,
    trade_date: pd.Timestamp,
    mask_map: Dict[pd.Timestamp, Set[str]],
    tradable: Optional[pd.Series] = None,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Slice one date's signal and label vectors for one expression.

    The slice spans the sorted ``stock_code`` union of the signal, label
    (and tradability) indices for ``trade_date``. A coordinate is
    eligible when the factor value is finite, the label is finite, the
    stock is not in the date's masked set, and -- when ``tradable`` is
    supplied -- the tradability flag is strict-True (``pd.NA`` counts as
    not tradable). Values introduced as holes by the union reindex are
    therefore ineligible.

    Parameters
    ----------
    factor_panel : pd.DataFrame
        Signal panel indexed by ``MultiIndex[stock_code, trade_date]``.
    labels : pd.Series
        Label vector on the same compound index.
    expression_id : str
        Panel column to read.
    trade_date : pd.Timestamp
        Date to slice on the ``trade_date`` level.
    mask_map : Dict[pd.Timestamp, Set[str]]
        Per-date masked stock codes; a date absent from the map carries
        no masked coordinate.
    tradable : Optional[pd.Series], optional
        Stock-day tradability flag on the compound index. Default is
        ``None``, the stand-alone five-argument call with no
        tradability filter.

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        ``(signal, label)`` float64 arrays in the sorted ``stock_code``
        union order, restricted to the eligible coordinates; both may be
        empty.

    Raises
    ------
    ValueError
        If ``expression_id`` is not a panel column, or if ``trade_date``
        is absent from the panel, from the labels, or from the supplied
        tradability vector.
    """
    # Halt when the expression is absent from the panel columns.
    if expression_id not in factor_panel.columns:
        # Trace the offending identifier explicitly.
        raise ValueError(f"factor_panel has no column for expression_id={expression_id!r}.")

    # Slice this date's signal column, indexed by stock_code.
    try:
        f_slice = factor_panel[expression_id].xs(trade_date, level="trade_date")
    except KeyError as exc:
        # Halt: the panel carries no rows for this date.
        raise ValueError(f"factor_panel carries no rows for trade_date={trade_date}.") from exc

    # Slice this date's labels, indexed by stock_code.
    try:
        y_slice = labels.xs(trade_date, level="trade_date")
    except KeyError as exc:
        # Halt: the label vector carries no rows for this date.
        raise ValueError(f"labels carry no rows for trade_date={trade_date}.") from exc

    # Slice this date's tradability flag when supplied (None means the five-argument stand-alone call).
    t_slice: Optional[pd.Series] = None
    # Only touch the flag when the caller threaded it in.
    if tradable is not None:
        # Read the date's flags, indexed by stock_code.
        try:
            t_slice = tradable.xs(trade_date, level="trade_date")
        except KeyError as exc:
            # Halt: the tradability vector carries no rows for this date.
            raise ValueError(f"tradable carries no rows for trade_date={trade_date}.") from exc

    # Compute the sorted union stock set of the slices (the primary path's coordinate set).
    common = sorted(set(f_slice.index) | set(y_slice.index) | (set(t_slice.index) if t_slice is not None else set()))

    # Extract the aligned signal values as float64 (holes from the union reindex become NaN).
    f_arr = pd.to_numeric(f_slice.reindex(common), errors="coerce").to_numpy(dtype=np.float64)

    # Extract the aligned label values as float64 (holes from the union reindex become NaN).
    y_arr = pd.to_numeric(y_slice.reindex(common), errors="coerce").to_numpy(dtype=np.float64)

    # Extract the aligned stock codes in the same order.
    stocks = np.asarray(common, dtype=object)

    # Flag the finite factor values.
    f_finite = np.isfinite(f_arr)

    # Flag the finite label values.
    y_finite = np.isfinite(y_arr)

    # Look up this date's masked-stock set once (empty when the date carries no mask).
    masked_stocks = mask_map.get(pd.Timestamp(trade_date), set())

    # Flag the coordinates excluded by the mask register.
    masked = np.array([str(stock) in masked_stocks for stock in stocks], dtype=np.bool_)

    # Conjunction: finite factor AND finite label AND NOT masked.
    keep = f_finite & y_finite & ~masked

    # Add the strict-True tradability predicate when the flag was supplied.
    if t_slice is not None:
        # Reindex via nullable boolean so holes become pd.NA (mirrors the primary path).
        t_vec = t_slice.astype("boolean").reindex(common)
        # Strict-True via the upstream primitive (NA maps to False), then join the conjunction.
        keep = keep & _STRICT_TRUE_MASK_UPSTREAM(t_vec, "tradable")

    # Return the eligible signal and label vectors as float64 arrays.
    return f_arr[keep], y_arr[keep]

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1, Helper: Detect whether a vector carries at least one tie.
# -------------------------------------------------------------------------------------------------------------------------------

def _has_tie_task_12(values: np.ndarray) -> bool:
    """
    Return whether ``values`` contains at least one duplicated entry.

    Distinct entries are counted with ``numpy.unique``; vectors shorter
    than two are reported as tie-free.

    Parameters
    ----------
    values : np.ndarray
        One-dimensional float64 slice of one cell's signal or label.

    Returns
    -------
    bool
        ``True`` iff two or more entries are exactly equal.
    """
    # A vector of length below two cannot carry a tie by definition.
    if values.size < 2:
        # Return the empty-tie verdict.
        return False

    # Count distinct values via np.unique (O(n log n)).
    distinct = np.unique(values).size

    # A tie exists iff the distinct count is strictly below the vector length.
    return bool(distinct < values.size)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 1: Public callable — select the nine a priori cells and freeze NINE_CELLS.
# -------------------------------------------------------------------------------------------------------------------------------

def select_nine_cells_task_12(
    daily_scores: pd.DataFrame,
    factor_panel: pd.DataFrame,
    labels: pd.Series,
    mask_register: pd.DataFrame,
    output_dir: Union[str, Path],
    tradable: Optional[pd.Series] = None,
) -> Dict[str, Any]:
    """
    Select the nine a priori cells and freeze the NINE_CELLS artifact.

    Implements Task 12, Step 1. The candidate pool is the subset of
    register rows with ``admissible == True``, ordered by
    (``expression_id`` ascending, ``trade_date`` ascending); that order
    is also the tie-breaking rule for every selection:

    * r1 - first cell with ``n_eligible == 5``;
    * r2 - first cell with ``n_distinct_signal == 3``;
    * r3 - cell with the highest ``|rho|``;
    * r4 - cell with the lowest ``|rho|``;
    * r5, r6 - first two cells whose eligible signal vector ties;
    * r7, r8 - first two cells whose eligible label vector ties;
    * r9 - first cell on a date carrying a mask for which some masked
      stock has both a factor value and a label value present and
      finite in the raw panels.

    The r5-r8 tie vectors come from ``_eligible_vectors_task_12``
    (finite factor, finite label, unmasked, and tradable strict-True
    when ``tradable`` is given). The r9 witness does not apply a
    tradability filter: it reads the masked stock's raw factor and
    label values and requires both to be finite. When no r9 witness
    exists, r9 is dropped, ``r9_status`` becomes ``"NOT_SATISFIABLE"``,
    ``pass_requirement`` becomes ``"8_of_8"`` and only eight rows are
    frozen.

    The artifact ``NINE_CELLS.csv`` (columns ``expression_id``,
    ``trade_date``, ``selection_rule``, rows in rule order r1..r9) and
    the sha256sum-style sidecar ``NINE_CELLS.sha256`` are written under
    ``output_dir``. The digest is computed over the bytes read back from
    the CSV file, so it binds to the deposited artifact.

    Parameters
    ----------
    daily_scores : pd.DataFrame
        Long-format register indexed by ``MultiIndex[expression_id,
        trade_date]`` with at least ``rho``, ``n_eligible``,
        ``n_distinct_signal``, ``n_distinct_label``, ``admissible`` and
        ``reject_reason``. Row order is not assumed.
    factor_panel : pd.DataFrame
        Signal panel indexed by ``MultiIndex[stock_code, trade_date]``,
        one column per expression id.
    labels : pd.Series
        Label vector on the same compound index as ``factor_panel``.
    mask_register : pd.DataFrame
        Mask register carrying ``stock_code`` and ``trade_date``; may be
        empty. Converted to a per-date masked-stock map.
    output_dir : Union[str, Path]
        Directory for the artifact and its sidecar; created if missing.
    tradable : Optional[pd.Series], optional
        Stock-day tradability flag on the compound index. Default is
        ``None``; when given, the r5-r8 tie scans use the strict-True
        eligible cross-section, while the r9 witness remains based on
        raw finiteness alone.

    Returns
    -------
    Dict[str, Any]
        ``nine_cells`` (pandas.DataFrame with columns
        ``expression_id``, ``trade_date``, ``selection_rule``; nine
        rows, or eight when r9 is not satisfiable), ``freeze_sha256``
        (str, 64 lowercase hex characters of the CSV bytes),
        ``r9_status`` (``"SATISFIED"`` or ``"NOT_SATISFIABLE"``),
        ``pass_requirement`` (``"9_of_9"`` or ``"8_of_8"``) and
        ``n_cells`` (int, 9 or 8).

    Raises
    ------
    TypeError
        If ``output_dir`` is not ``str`` or ``pathlib.Path``, if
        ``daily_scores`` is not a ``pandas.DataFrame`` (raised by
        ``_validated_register_task_12``), if ``labels`` or ``tradable``
        is not a ``pandas.Series``, or if ``factor_panel`` is not a
        pandas object.
    ValueError
        If the register or a compound index violates its structural
        contract, if no register cell is admissible, if r1, r2, r5/r6
        or r7/r8 finds no qualifying candidate, if an admissible cell
        carries a non-finite ``rho``, or if an admissible cell names an
        expression absent from ``factor_panel`` or a date absent from
        ``factor_panel``, ``labels`` or the supplied ``tradable``
        vector.
    """
    # Validate the output directory argument type (str or Path only).
    if not isinstance(output_dir, (str, Path)):
        # Reject non-path-like output directories explicitly.
        raise TypeError(f"output_dir must be a str or pathlib.Path, got {type(output_dir)!r}.")

    # Normalise the output directory to an absolute-ish Path and create it.
    out_dir = Path(output_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    # Validate the panel and label compound indices (contract checked at the boundary).
    _require_compound_index_task_12(factor_panel, "factor_panel")

    # Validate the label vector's compound index on the same coordinate system.
    if not isinstance(labels, pd.Series):
        # Reject non-Series label arguments explicitly.
        raise TypeError(f"labels must be a pandas.Series, got {type(labels)!r}.")

    # Reuse the compound-index validator for the label vector.
    _require_compound_index_task_12(labels, "labels")

    # Validate the optional tradability flag on the same coordinate system.
    if tradable is not None:
        # Reject non-Series tradability arguments explicitly.
        if not isinstance(tradable, pd.Series):
            # Name the offending type.
            raise TypeError(f"tradable must be a pandas.Series, got {type(tradable)!r}.")
        # Reuse the compound-index validator for the tradability flag.
        _require_compound_index_task_12(tradable, "tradable")

    # Build the per-date masked-stocks map via the upstream primitive (never re-written).
    mask_map = _MASK_COORDINATE_MAP_UPSTREAM(mask_register)

    # Validate the register and re-express it flat and deterministically sorted.
    cells = _validated_register_task_12(daily_scores)

    # Restrict every rule's candidate pool to admissible cells (the step text's hard constraint).
    admissible_mask = cells["admissible"].fillna(False).astype(bool)
    admissible = cells[admissible_mask].reset_index(drop=True)

    # Halt early if no cell is admissible at all (instrument check cannot be constructed).
    if len(admissible) == 0:
        # No admissible candidate exists for any rule — loud halt by design.
        raise ValueError("daily_scores carries no admissible cell; the a priori instrument check cannot be constructed.")

    # ------------------------------------------------------------------
    # r1 — boundary cell with n_eligible == 5 (admissible, deterministic order).
    # ------------------------------------------------------------------
    # Cast the nullable Int64 counts to float for a NaN-safe comparison.
    n_eligible = pd.to_numeric(admissible["n_eligible"], errors="coerce")

    # Candidates for r1: admissible cells whose n_eligible equals exactly 5.
    r1_pool = admissible[n_eligible == 5]

    # Halt when no admissible cell reaches the n_eligible == 5 boundary.
    if r1_pool.empty:
        # Trace the failed rule explicitly.
        raise ValueError("Rule r1 unsatisfiable: no admissible cell carries n_eligible == 5.")

    # Tie-break is already (expression_id, trade_date) ascending; take the first.
    r1_cell = r1_pool.iloc[0]

    # ------------------------------------------------------------------
    # r2 — boundary cell with n_distinct_signal == 3 (admissible, deterministic order).
    # ------------------------------------------------------------------
    # Cast the nullable Int64 distinct counts to float for a NaN-safe comparison.
    n_distinct_signal = pd.to_numeric(admissible["n_distinct_signal"], errors="coerce")

    # Candidates for r2: admissible cells whose n_distinct_signal equals exactly 3.
    r2_pool = admissible[n_distinct_signal == 3]

    # Halt when no admissible cell reaches the n_distinct_signal == 3 boundary.
    if r2_pool.empty:
        # Trace the failed rule explicitly.
        raise ValueError("Rule r2 unsatisfiable: no admissible cell carries n_distinct_signal == 3.")

    # Tie-break is already (expression_id, trade_date) ascending; take the first.
    r2_cell = r2_pool.iloc[0]

    # ------------------------------------------------------------------
    # r3 — highest |rho| (admissible). r4 — lowest |rho| (admissible).
    # ------------------------------------------------------------------
    # Compute the absolute rho per admissible cell (admissible implies finite rho by the register contract).
    abs_rho = pd.to_numeric(admissible["rho"], errors="coerce").abs()

    # Halt if any admissible cell carries a non-finite rho (register discipline violated upstream).
    if not bool(np.isfinite(abs_rho.to_numpy(dtype=np.float64)).all()):
        # Trace the offending condition explicitly.
        raise ValueError("At least one admissible cell carries a non-finite rho; the a priori check cannot be constructed.")

    # r3: collect every position attaining the maximum |rho|; ties are resolved by the pinned
    # (expression_id, trade_date) tie-breaking below.
    abs_rho_arr = abs_rho.to_numpy(dtype=np.float64)

    # r3 candidate positions at the maximum |rho|.
    r3_ties = np.flatnonzero(abs_rho_arr == abs_rho_arr.max())

    # Choose the tie-breaking winner: lowest expression_id, then earliest trade_date.
    r3_win = min(r3_ties, key=lambda i: (str(admissible["expression_id"].iloc[i]), admissible["trade_date"].iloc[i]))

    # r3 cell: the tie-breaking winner at maximum |rho|.
    r3_cell = admissible.iloc[r3_win]

    # r4: collect every position attaining the minimum |rho|; ties are resolved identically.
    r4_ties = np.flatnonzero(abs_rho_arr == abs_rho_arr.min())

    # Choose the tie-breaking winner for r4: lowest expression_id, then earliest trade_date.
    r4_win = min(r4_ties, key=lambda i: (str(admissible["expression_id"].iloc[i]), admissible["trade_date"].iloc[i]))

    # r4 cell: the tie-breaking winner at minimum |rho|.
    r4_cell = admissible.iloc[r4_win]

    # ------------------------------------------------------------------
    # r5/r6 — first two admissible cells whose eligible signal vector carries a tie.
    # ------------------------------------------------------------------
    # Deterministic cell order for every tiedness scan and for the r9 scan:
    # (expression_id ascending, trade_date ascending); the returned indices are positions
    # into the admissible frame, preserving the RangeIndex alignment of the register copy.
    order = admissible.sort_values(["expression_id", "trade_date"], kind="mergesort").index.to_numpy()

    # Accumulate the tied-signal candidates in deterministic order.
    tied_signal: List[int] = []

    # Iterate admissible cells in deterministic order until r5 and r6 are found.
    for position in order:
        # Read the cell coordinates.
        expression_id = str(admissible["expression_id"].iloc[position])
        trade_date = pd.Timestamp(admissible["trade_date"].iloc[position])

        # Compute the eligible signal/label vectors for this cell (finite, unmasked).
        # An admissible cell whose panels are missing breaches the upstream contract;
        # _eligible_vectors_task_12 raises ValueError naming the expression and date.
        signal_vec, _label_vec = _eligible_vectors_task_12(
            factor_panel, labels, expression_id, trade_date, mask_map, tradable,
        )

        # Record the candidate if its eligible signal vector carries a tie.
        if _has_tie_task_12(signal_vec):
            tied_signal.append(int(position))

        # Stop once both r5 and r6 candidates are secured.
        if len(tied_signal) == 2:
            break

    # Halt when fewer than two tied-signal admissible cells exist.
    if len(tied_signal) < 2:
        # Trace the failed rule explicitly.
        raise ValueError("Rules r5/r6 unsatisfiable: fewer than two admissible cells carry a tie in the eligible signal vector.")

    # r5 cell: the first tied-signal candidate; r6 cell: the second.
    r5_cell = admissible.iloc[tied_signal[0]]
    r6_cell = admissible.iloc[tied_signal[1]]

    # ------------------------------------------------------------------
    # r7/r8 — first two admissible cells whose eligible label vector carries a tie.
    # ------------------------------------------------------------------
    # Accumulate the tied-label candidates in deterministic order.
    tied_label: List[int] = []

    # Iterate admissible cells in deterministic order until r7 and r8 are found.
    for position in order:
        # Read the cell coordinates.
        expression_id = str(admissible["expression_id"].iloc[position])
        trade_date = pd.Timestamp(admissible["trade_date"].iloc[position])

        # Compute the eligible signal/label vectors for this cell (finite, unmasked).
        _signal_vec, label_vec = _eligible_vectors_task_12(
            factor_panel, labels, expression_id, trade_date, mask_map, tradable,
        )

        # Record the candidate if its eligible label vector carries a tie.
        if _has_tie_task_12(label_vec):
            tied_label.append(int(position))

        # Stop once both r7 and r8 candidates are secured.
        if len(tied_label) == 2:
            break

    # Halt when fewer than two tied-label admissible cells exist.
    if len(tied_label) < 2:
        # Trace the failed rule explicitly.
        raise ValueError("Rules r7/r8 unsatisfiable: fewer than two admissible cells carry a tie in the eligible label vector.")

    # r7 cell: the first tied-label candidate; r8 cell: the second.
    r7_cell = admissible.iloc[tied_label[0]]
    r8_cell = admissible.iloc[tied_label[1]]

    # ------------------------------------------------------------------
    # r9 — first admissible cell on a date whose masked coordinate is otherwise factor/label-available.
    # ------------------------------------------------------------------
    # The set of dates present in the mask register (empty mask => empty set).
    mask_dates = set(mask_map)

    # Accumulate the r9 candidate (first admissible cell on a mask-carrying date).
    r9_cell: Optional[pd.Series] = None

    # Iterate admissible cells in deterministic order (lowest expression_id, then earliest trade_date).
    for position in order:
        # Read the cell coordinates.
        expression_id = str(admissible["expression_id"].iloc[position])
        trade_date = pd.Timestamp(admissible["trade_date"].iloc[position])

        # Skip dates not carrying any masked coordinate.
        if trade_date not in mask_dates:
            # This date has no masked stock-day — irrelevant for r9.
            continue

        # Flag whether at least one masked stock on this date is factor/label-available.
        witness_available = False

        # Check each masked stock of this date for a finite factor and a finite label.
        for stock in sorted(mask_map[trade_date]):
            # Read the masked stock's factor value for this expression at this date.
            try:
                factor_slice = factor_panel[expression_id].xs(trade_date, level="trade_date")
            except KeyError:
                # The date is absent from the panel — the witness cannot be established.
                continue

            # Read the masked stock's label at this date.
            try:
                label_slice = labels.xs(trade_date, level="trade_date")
            except KeyError:
                # The date is absent from the labels — the witness cannot be established.
                continue

            # Skip stocks genuinely absent from either slice (the coordinate is not available).
            if stock not in factor_slice.index or stock not in label_slice.index:
                # This masked coordinate is not otherwise factor/label-available.
                continue

            # Coerce the masked stock's factor value to a finite-testable scalar.
            factor_value = float(pd.to_numeric(pd.Series([factor_slice.loc[stock]]), errors="coerce").iloc[0])

            # Coerce the masked stock's label value to a finite-testable scalar.
            label_value = float(pd.to_numeric(pd.Series([label_slice.loc[stock]]), errors="coerce").iloc[0])

            # A masked coordinate with finite factor AND finite label is "otherwise factor/label-available".
            if math.isfinite(factor_value) and math.isfinite(label_value):
                # The availability witness is established.
                witness_available = True
                break

        # Record the candidate if the availability witness holds for this cell.
        if witness_available:
            # Freeze r9 as the first admissible cell on a mask-carrying date in deterministic order.
            r9_cell = admissible.iloc[position]
            break

    # Resolve the r9 disclosure path: SATISFIED with nine cells, or NOT_SATISFIABLE with eight.
    if r9_cell is not None:
        # The nine-of-nine pass requirement is retained.
        r9_status = "SATISFIED"
        pass_requirement = "9_of_9"
    else:
        # The mask register carries no otherwise-factor/label-available masked coordinate on any admissible date.
        r9_status = "NOT_SATISFIABLE"
        pass_requirement = "8_of_8"

    # ------------------------------------------------------------------
    # Assemble the nine (or eight) frozen rows in pinned rule order.
    # ------------------------------------------------------------------
    # Assemble the (expression_id, trade_date, selection_rule) triples in rule order r1..r9.
    frozen_rows: List[Tuple[str, pd.Timestamp, str]] = [
        (str(r1_cell["expression_id"]), pd.Timestamp(r1_cell["trade_date"]), "r1"),
        (str(r2_cell["expression_id"]), pd.Timestamp(r2_cell["trade_date"]), "r2"),
        (str(r3_cell["expression_id"]), pd.Timestamp(r3_cell["trade_date"]), "r3"),
        (str(r4_cell["expression_id"]), pd.Timestamp(r4_cell["trade_date"]), "r4"),
        (str(r5_cell["expression_id"]), pd.Timestamp(r5_cell["trade_date"]), "r5"),
        (str(r6_cell["expression_id"]), pd.Timestamp(r6_cell["trade_date"]), "r6"),
        (str(r7_cell["expression_id"]), pd.Timestamp(r7_cell["trade_date"]), "r7"),
        (str(r8_cell["expression_id"]), pd.Timestamp(r8_cell["trade_date"]), "r8"),
    ]

    # Append r9 only when satisfiable.
    if r9_cell is not None:
        # r9 is appended last (rule order r1..r9).
        frozen_rows.append((str(r9_cell["expression_id"]), pd.Timestamp(r9_cell["trade_date"]), "r9"))

    # Materialise the frozen artifact as a DataFrame with the pinned column order.
    nine_cells = pd.DataFrame(frozen_rows, columns=NINE_CELLS_COLUMNS_TASK_12)

    # Coerce trade_date to datetime64[ns] exactly as the deposit schema requires.
    nine_cells["trade_date"] = pd.to_datetime(nine_cells["trade_date"]).astype("datetime64[ns]")

    # Coerce expression_id and selection_rule to plain object/string columns.
    nine_cells["expression_id"] = nine_cells["expression_id"].astype(str)
    nine_cells["selection_rule"] = nine_cells["selection_rule"].astype(str)

    # ------------------------------------------------------------------
    # Freeze the artifact: write NINE_CELLS.csv and the SHA-256 sidecar.
    # ------------------------------------------------------------------
    # Resolve the artifact paths beneath output_dir.
    csv_path = out_dir / "NINE_CELLS.csv"
    sidecar_path = out_dir / "NINE_CELLS.sha256"

    # Write the CSV without an index (the deposit schema has no index).
    nine_cells.to_csv(csv_path, index=False)

    # Read back the exact bytes written to disk (the digest must bind to the file, not to memory).
    csv_bytes = csv_path.read_bytes()

    # Compute the SHA-256 over the file bytes (lowercase hex, 64 characters).
    freeze_sha256 = hashlib.sha256(csv_bytes).hexdigest()

    # Write the sidecar in sha256sum-compatible layout carrying the digest and the artifact name.
    sidecar_path.write_text(f"{freeze_sha256}  NINE_CELLS.csv\n", encoding="utf-8", newline="\n")

    # Assemble and return the pinned result dictionary.
    return {
        "nine_cells": nine_cells,
        "freeze_sha256": freeze_sha256,
        "r9_status": r9_status,
        "pass_requirement": pass_requirement,
        "n_cells": int(len(nine_cells)),
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 2: Independent alternative implementation of mid-rank and Pearson
# -------------------------------------------------------------------------------------------------------------------------------

# NOTE ON INDEPENDENCE:
# The two callables in this section share no code with the primary path and
# reference none of its primitives, no statistical library of any kind, no
# pandas inbuilt ranking, no closed-form correlation helper. The mid-rank
# algorithm below is the cumulative-count-based (sort + searchsorted)
# approach; the correlation path is the hand-written centred-vector cosine
# path. Both are distinct from the primary's execution order by construction.

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 2: Public callable — average-tie (mid) ranks, ascending, input-aligned, float64.
# -------------------------------------------------------------------------------------------------------------------------------

def alternative_mid_ranks_task_12(values: Sequence[float]) -> np.ndarray:
    """
    Compute 1-based average-tie mid-ranks in the input order.

    Independent implementation used for the nine-cell cross-check: sorts
    a copy of the values with ``numpy.sort`` (stable) and, per element,
    forms

        lt(x) = |{v : v < x}|    (searchsorted side="left")
        le(x) = |{v : v <= x}|   (searchsorted side="right")
        mid_rank(x) = (1 + lt(x) + le(x)) / 2

    Ranks ascend with the value domain. A unique smallest value
    receives 1.0; a tied minimum receives the mean of its tied span
    (1.5 for a two-way tie, and so on), following the average-tie
    convention. Neither ``pandas`` nor ``numpy.corrcoef`` is used.

    Parameters
    ----------
    values : Sequence[float]
        Finite real values, at least one element; a one-dimensional
        ``numpy.ndarray`` is also accepted.

    Returns
    -------
    np.ndarray
        float64 array of shape ``(n,)`` holding the mid-rank of each
        element, aligned to the input order.

    Raises
    ------
    TypeError
        If ``values`` is not a ``Sequence`` or ``numpy.ndarray``, or is
        a ``str``, ``bytes`` or ``bytearray`` object.
    ValueError
        If ``values`` is empty, if an element is not coercible to
        float64, or if an element is NaN or infinite.

    Notes
    -----
    The integer counts are exact in float64 for ``n <= 2**53``, but
    the float64 sum ``1 + lt + le`` is exact only while it stays at or
    below ``2**53``, i.e. for ``n < 2**52``; within that bound the
    mid-rank vector matches any correct average-tie ranking
    bit-identically. The inline code comment claims ``n <= 2**53``,
    which overstates the bound.
    """
    # Reject non-sequence inputs (strings, bytes, dicts are not rank vectors). A numpy 1-D array
    # is accepted as a sequence of floats for internal reuse by the correlation path.
    if not isinstance(values, (Sequence, np.ndarray)) or isinstance(values, (str, bytes, bytearray)):
        # Reject the wrong container type explicitly.
        raise TypeError(f"values must be a Sequence of real numbers, got {type(values)!r}.")

    # Materialise the sequence as a list and halt on the empty case.
    materialised = list(values)

    # Halt when the sequence is empty (no ranks are defined).
    if len(materialised) == 0:
        # Trace the empty-input condition explicitly.
        raise ValueError("values must carry at least one element; mid-ranks are undefined for the empty sequence.")

    # Coerce to float64; a non-numeric element raises ValueError naturally from numpy.
    try:
        array = np.asarray(materialised, dtype=np.float64)
    except (ValueError, TypeError) as exc:
        # Re-raise with the pinned error type and an informative message.
        raise ValueError(f"values must all be coercible to float64; coercion failed with: {exc}") from exc

    # Halt when any element is non-finite (NaN or ±inf); mid-ranks require a total order.
    if not bool(np.isfinite(array).all()):
        # Trace the offending condition explicitly.
        raise ValueError("values must be finite; at least one element is NaN or ±inf.")

    # Sort a copy of the values (the input order is never disturbed).
    ordered = np.sort(array, kind="mergesort")

    # For each input element, count the strictly-smaller elements (left insertion point).
    lt_count = np.searchsorted(ordered, array, side="left")

    # For each input element, count the smaller-or-equal elements (right insertion point).
    le_count = np.searchsorted(ordered, array, side="right")

    # Mid-rank(x) = (1 + lt_count(x) + le_count(x)) / 2 — exact in float64 for n <= 2**53.
    mid_ranks = (1.0 + lt_count.astype(np.float64) + le_count.astype(np.float64)) / 2.0

    # Return the float64, input-aligned mid-rank vector.
    return mid_ranks

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 2: Public callable — Pearson correlation of average-tie mid-ranks (centred-vector cosine path).
# -------------------------------------------------------------------------------------------------------------------------------

def alternative_rank_ic_task_12(signal: Sequence[float], label: Sequence[float]) -> float:
    """
    Compute the rank-IC via a hand-written cosine path.

    Rank both inputs ascending with average ties using
    ``alternative_mid_ranks_task_12``, centre each mid-rank vector by
    subtracting its mean, and return the cosine similarity of the two
    centred vectors::

        cx = rx - mean(rx);  cy = ry - mean(ry)
        rho = dot(cx, cy) / (||cx||_2 * ||cy||_2)

    This is the Pearson correlation of the two mid-rank vectors,
    evaluated in a summation order distinct from the primary path and
    with no statistical library. A zero-variance side leaves the
    statistic undefined and yields ``nan`` rather than raising.

    Parameters
    ----------
    signal : Sequence[float]
        Finite, numeric signal vector of the cell.
    label : Sequence[float]
        Finite, numeric label vector of the same length.

    Returns
    -------
    float
        Pearson correlation of the two mid-rank vectors, in [-1, 1];
        ``nan`` when either side has zero variance.

    Raises
    ------
    TypeError
        If ``signal`` or ``label`` is not a ``Sequence``, or is a
        string, bytes or bytearray.
    ValueError
        If the lengths differ, either is empty, an element cannot be
        coerced to float64, or an element is NaN or infinite.
    """
    # Reject non-sequence signal arguments (strings, bytes, dicts are not rank vectors).
    if not isinstance(signal, Sequence) or isinstance(signal, (str, bytes, bytearray)):
        # Reject the wrong container type explicitly.
        raise TypeError(f"signal must be a Sequence of real numbers, got {type(signal)!r}.")

    # Reject non-sequence label arguments for the same reason.
    if not isinstance(label, Sequence) or isinstance(label, (str, bytes, bytearray)):
        # Reject the wrong container type explicitly.
        raise TypeError(f"label must be a Sequence of real numbers, got {type(label)!r}.")

    # Materialise both sequences to lists for a deterministic length comparison.
    signal_list = list(signal)
    label_list = list(label)

    # Halt on length mismatch before any numerical work (the pinned negative control).
    if len(signal_list) != len(label_list):
        # Trace both lengths explicitly.
        raise ValueError(f"signal and label must share the same length, got {len(signal_list)} and {len(label_list)}.")

    # Halt on the empty case (no cross-section exists).
    if len(signal_list) == 0:
        # Trace the empty-input condition explicitly.
        raise ValueError("signal and label must carry at least one element each.")

    # Coerce both vectors to float64 (a non-numeric element raises ValueError naturally).
    try:
        signal_arr = np.asarray(signal_list, dtype=np.float64)
        label_arr = np.asarray(label_list, dtype=np.float64)
    except (ValueError, TypeError) as exc:
        # Re-raise with the pinned error type and an informative message.
        raise ValueError(f"signal and label must all be coercible to float64; coercion failed with: {exc}") from exc

    # Halt when any element is non-finite; the statistic is undefined on a NaN/inf input (the pinned negative control).
    if not bool(np.isfinite(signal_arr).all()):
        # Trace the offending condition explicitly.
        raise ValueError("signal must be finite; at least one element is NaN or ±inf.")

    if not bool(np.isfinite(label_arr).all()):
        # Trace the offending condition explicitly.
        raise ValueError("label must be finite; at least one element is NaN or ±inf.")

    # Rank both vectors ascending with the average-tie algorithm of this module.
    rank_signal = alternative_mid_ranks_task_12(signal_arr)

    # Rank the label vector identically.
    rank_label = alternative_mid_ranks_task_12(label_arr)

    # Centre both mid-rank vectors by subtracting their means.
    centred_signal = rank_signal - rank_signal.mean()

    # Centre the label mid-rank vector identically.
    centred_label = rank_label - rank_label.mean()

    # Compute the Euclidean norm of the centred signal vector.
    norm_signal = float(np.linalg.norm(centred_signal))

    # Compute the Euclidean norm of the centred label vector.
    norm_label = float(np.linalg.norm(centred_label))

    # A zero norm on either side means a degenerate (zero-variance) cross-section: nan by the domain rule.
    if norm_signal == 0.0 or norm_label == 0.0:
        # Return nan — the statistic's domain is empty for this cell.
        return float("nan")

    # Cosine similarity of the centred vectors; the arithmetic is hand-written and bounded by 1.
    rho = float(np.dot(centred_signal, centred_label) / (norm_signal * norm_label))

    # Return the scalar correlation.
    return rho

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3: Comparison at declared tolerance and the DUAL_CHECK_RESULTS table
# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3: Public callable — the pinned tolerance predicate (tol_abs = 1e-10, tol_rel = 1e-12).
# -------------------------------------------------------------------------------------------------------------------------------

def tolerance_verdict_task_12(
    primary_rho: float,
    alternative_rho: float,
    tol_abs: float = 1e-10,
    tol_rel: float = 1e-12,
) -> bool:
    """
    Return whether two rank correlations agree within tolerance.

    Implements the pinned predicate::

        delta = |primary_rho - alternative_rho|
        pass  = (delta <= tol_abs)
                or (delta / max(tol_abs, |primary_rho|) <= tol_rel)
                or (primary_rho == alternative_rho)

    The third arm is bit-identity. Comparisons with ``nan`` evaluate to
    ``False``, so a ``nan`` on either side fails all three arms: the two
    implementations then disagree on the domain of the statistic, which
    is a halt rather than a tolerance question.

    Parameters
    ----------
    primary_rho : float
        Primary implementation's rank correlation for the cell.
    alternative_rho : float
        Alternative implementation's rank correlation for the cell.
    tol_abs : float, optional
        Absolute tolerance. Default is ``1e-10``.
    tol_rel : float, optional
        Relative tolerance applied to ``max(tol_abs, |primary_rho|)``.
        Default is ``1e-12``.

    Returns
    -------
    bool
        ``True`` iff at least one of the three arms holds; ``nan`` never
        satisfies any arm.

    Raises
    ------
    TypeError
        If any argument is not a real number (``bool`` is rejected).
    ValueError
        If either tolerance is negative or non-finite.
    """
    # Validate the type of primary_rho (bool is a subclass of int — reject it explicitly).
    if not isinstance(primary_rho, (int, float, np.floating, np.integer)) or isinstance(primary_rho, bool):
        # Reject non-numeric primary_rho explicitly.
        raise TypeError(f"primary_rho must be a real number, got {type(primary_rho)!r}.")

    # Validate the type of alternative_rho the same way.
    if not isinstance(alternative_rho, (int, float, np.floating, np.integer)) or isinstance(alternative_rho, bool):
        # Reject non-numeric alternative_rho explicitly.
        raise TypeError(f"alternative_rho must be a real number, got {type(alternative_rho)!r}.")

    # Validate the type of tol_abs (bools are rejected; the tolerance must be a real number).
    if not isinstance(tol_abs, (int, float, np.floating, np.integer)) or isinstance(tol_abs, bool):
        # Reject non-numeric tol_abs explicitly.
        raise TypeError(f"tol_abs must be a real number, got {type(tol_abs)!r}.")

    # Validate the type of tol_rel the same way.
    if not isinstance(tol_rel, (int, float, np.floating, np.integer)) or isinstance(tol_rel, bool):
        # Reject non-numeric tol_rel explicitly.
        raise TypeError(f"tol_rel must be a real number, got {type(tol_rel)!r}.")

    # Coerce the pinned scalars to Python floats for the predicate arithmetic.
    p = float(primary_rho)
    a = float(alternative_rho)
    absolute_tolerance = float(tol_abs)
    relative_tolerance = float(tol_rel)

    # Halt when either tolerance is non-finite (an unbounded tolerance voids the check).
    if not math.isfinite(absolute_tolerance) or not math.isfinite(relative_tolerance):
        # Trace the offending condition explicitly.
        raise ValueError("tol_abs and tol_rel must be finite.")

    # Halt when either tolerance is negative (an inverted band voids the check).
    if absolute_tolerance < 0.0 or relative_tolerance < 0.0:
        # Trace the offending condition explicitly.
        raise ValueError(f"tol_abs and tol_rel must be non-negative, got tol_abs={absolute_tolerance}, tol_rel={relative_tolerance}.")

    # delta := |primary_rho − alternative_rho| (pinned definition).
    delta = abs(p - a)

    # Absolute arm: delta <= tol_abs (NaN compares False — NaN never passes).
    absolute_arm = delta <= absolute_tolerance

    # Relative arm: delta / max(tol_abs, |primary_rho|) <= tol_rel (NaN propagates False).
    relative_arm = delta / max(absolute_tolerance, abs(p)) <= relative_tolerance

    # Bit-identity arm: primary_rho == alternative_rho (NaN == NaN is False).
    bit_identity_arm = p == a

    # Return True iff any of the three arms holds; a halt is every arm failing.
    return bool(absolute_arm or relative_arm or bit_identity_arm)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3, Helper: Slice the eligible cross-section for one cell, rebuilt exactly as the primary path builds it.
# -------------------------------------------------------------------------------------------------------------------------------

def _eligible_slice_task_12(
    factor_panel: pd.DataFrame,
    labels: pd.Series,
    tradable: pd.Series,
    mask_map: Dict[pd.Timestamp, Set[str]],
    expression_id: str,
    trade_date: pd.Timestamp,
) -> Tuple[np.ndarray, np.ndarray]:
    """
    Slice one cell's eligible signal and label vectors.

    Rebuilds the primary path's eligible cross-section for one
    ``(expression_id, trade_date)`` cell. Take the sorted union of stock
    codes on the date across the factor panel, the label vector and the
    tradability flag; reindex each input onto that union (holes become
    NaN or ``pd.NA``); then apply the P02 ``eligible_stock_conditions``
    conjunction::

        eligible = isfinite(factor) & isfinite(label)
                   & tradable_strict_true & ~masked

    ``pd.NA`` tradability maps to not-True through the upstream
    ``_STRICT_TRUE_MASK_UPSTREAM`` helper, and a stock is excluded when
    its code belongs to ``mask_map[trade_date]``. The returned
    cross-section is therefore the same eligible one the primary
    implementation used, item for item.

    Parameters
    ----------
    factor_panel : pd.DataFrame
        Signal panel indexed by ``MultiIndex[stock_code, trade_date]``.
    labels : pd.Series
        Label vector on the same compound index.
    tradable : pd.Series
        Tradability flag on the same compound index; bool or nullable
        boolean dtype.
    mask_map : Dict[pd.Timestamp, Set[str]]
        Per-date masked stock codes, built upstream from the mask
        register.
    expression_id : str
        Column of ``factor_panel`` to read.
    trade_date : pd.Timestamp
        Date to slice (``level="trade_date"``).

    Returns
    -------
    Tuple[np.ndarray, np.ndarray]
        ``(signal, label)`` float64 vectors aligned on the sorted union
        of stocks for the date and filtered by the eligibility
        conjunction; possibly empty.

    Raises
    ------
    ValueError
        If ``expression_id`` is absent from ``factor_panel``'s columns,
        or if the panel, the labels or the tradability flag carry no
        rows for ``trade_date``.
    """
    # Halt when the expression is absent from the panel columns.
    if expression_id not in factor_panel.columns:
        # Trace the offending identifier explicitly.
        raise ValueError(f"factor_panel has no column for expression_id={expression_id!r}.")

    # Slice this date's signal column, indexed by stock_code.
    try:
        f_slice = factor_panel[expression_id].xs(trade_date, level="trade_date")
    except KeyError as exc:
        # Halt: the panel carries no rows for this date.
        raise ValueError(f"factor_panel carries no rows for trade_date={trade_date}.") from exc

    # Slice this date's labels, indexed by stock_code.
    try:
        y_slice = labels.xs(trade_date, level="trade_date")
    except KeyError as exc:
        # Halt: the label vector carries no rows for this date.
        raise ValueError(f"labels carry no rows for trade_date={trade_date}.") from exc

    # Slice this date's tradability flag, indexed by stock_code.
    try:
        t_slice = tradable.xs(trade_date, level="trade_date")
    except KeyError as exc:
        # Halt: the tradability vector carries no rows for this date.
        raise ValueError(f"tradable carries no rows for trade_date={trade_date}.") from exc

    # Compute the sorted union stock set of the three inputs (the primary path's coordinate set).
    stocks = sorted(set(f_slice.index) | set(y_slice.index) | set(t_slice.index))

    # Extract the aligned signal values as float64 (holes from the union reindex become NaN).
    f_arr = pd.to_numeric(f_slice.reindex(stocks), errors="coerce").to_numpy(dtype=np.float64)

    # Extract the aligned label values as float64 (holes from the union reindex become NaN).
    y_arr = pd.to_numeric(y_slice.reindex(stocks), errors="coerce").to_numpy(dtype=np.float64)

    # Extract the aligned tradability flag, reindexed onto the union via nullable boolean (holes become pd.NA).
    # (Mirrors the primary path's own astype("boolean") before reindex, so plain-bool input cannot upcast to object.)
    t_vec = t_slice.astype("boolean").reindex(stocks)

    # Materialise the strict-True tradability mask via the upstream primitive (NA maps to False).
    t_mask = _STRICT_TRUE_MASK_UPSTREAM(t_vec, "tradable")

    # Look up this date's masked-stock set once (empty when the date carries no mask).
    masked_stocks = mask_map.get(pd.Timestamp(trade_date), set())

    # Extract the aligned stock codes in the same order.
    stock_codes = np.asarray(stocks, dtype=object)

    # Flag the coordinates excluded by the mask register (predicate (iv) of the eligibility conjunction).
    masked = np.array([str(stock) in masked_stocks for stock in stock_codes], dtype=np.bool_)

    # Conjunction: finite factor AND finite label AND tradable strict-True AND NOT masked.
    keep = np.isfinite(f_arr) & np.isfinite(y_arr) & t_mask & ~masked

    # Return the eligible signal and label vectors as float64 arrays.
    return f_arr[keep], y_arr[keep]

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Step 3: Public callable — DUAL_CHECK_RESULTS table at the declared tolerance.
# -------------------------------------------------------------------------------------------------------------------------------

def build_dual_check_results_task_12(
    nine_cells: pd.DataFrame,
    daily_scores: pd.DataFrame,
    factor_panel: pd.DataFrame,
    labels: pd.Series,
    tradable: pd.Series,
    mask_register: pd.DataFrame,
    output_dir: Union[str, Path],
) -> pd.DataFrame:
    """
    Build the DUAL_CHECK_RESULTS comparison table for the frozen cells.

    Iterates the frozen cells in their pinned order. For each cell it
    reads ``primary_rho`` from DAILY_SCORES at ``(expression_id,
    trade_date)``, rebuilds the eligible cross-section with
    ``_eligible_slice_task_12``, recomputes ``alternative_rho`` with the
    hand-written centred-vector cosine path, and forms::

        delta             = |primary_rho - alternative_rho|
        pass_at_tolerance = tolerance_verdict_task_12(primary_rho,
                                                      alternative_rho)

    If any cell fails all three tolerance arms, ``ValueError`` is raised
    naming the failing cells and no table is written. Otherwise
    ``DUAL_CHECK_RESULTS.csv`` is written beneath ``output_dir`` with
    columns exactly ``[expression_id, trade_date, primary_rho,
    alternative_rho, delta, pass_at_tolerance]`` and ``trade_date`` as
    datetime64[ns].

    Parameters
    ----------
    nine_cells : pd.DataFrame
        Frozen artifact with columns ``expression_id``, ``trade_date``,
        ``selection_rule``; processed in its frozen order.
    daily_scores : pd.DataFrame
        DAILY_SCORES register indexed by
        ``MultiIndex[expression_id, trade_date]`` with a ``rho`` column.
    factor_panel : pd.DataFrame
        Signal panel indexed by ``MultiIndex[stock_code, trade_date]``.
    labels : pd.Series
        Label vector on the same compound index.
    tradable : pd.Series
        Tradability flag on the same compound index; bool or nullable
        boolean dtype.
    mask_register : pd.DataFrame
        Mask register (may be empty); converted to the per-date
        masked-code map by the upstream
        ``_MASK_COORDINATE_MAP_UPSTREAM`` helper.
    output_dir : Union[str, Path]
        Directory beneath which ``DUAL_CHECK_RESULTS.csv`` is written;
        created if missing.

    Returns
    -------
    pd.DataFrame
        Comparison table with RangeIndex and columns ``expression_id``
        (str), ``trade_date`` (datetime64[ns]), ``primary_rho``,
        ``alternative_rho``, ``delta`` (all float64) and
        ``pass_at_tolerance`` (bool); ``delta`` equals
        ``|primary_rho - alternative_rho|`` elementwise.

    Raises
    ------
    TypeError
        If ``nine_cells`` or ``daily_scores`` is not a DataFrame,
        ``labels`` or ``tradable`` is not a Series, or ``output_dir`` is
        neither ``str`` nor ``pathlib.Path``.
    ValueError
        If ``nine_cells`` lacks a pinned column, the register index is
        not the required MultiIndex, a frozen cell is absent from the
        register, an ``expression_id`` in ``nine_cells`` is absent from
        ``factor_panel``'s columns, a date slice is missing, or any cell
        fails all three tolerance arms.
    """
    # Enforce the DataFrame container type for the frozen artifact.
    if not isinstance(nine_cells, pd.DataFrame):
        # Reject non-DataFrame nine_cells explicitly.
        raise TypeError(f"nine_cells must be a pandas.DataFrame, got {type(nine_cells)!r}.")

    # Enforce the pinned columns of the frozen artifact.
    required_nine_columns = {"expression_id", "trade_date", "selection_rule"}

    # Compute the missing-column list for an informative halt.
    missing_nine_columns = sorted(required_nine_columns - set(nine_cells.columns))

    # Halt when any pinned column is absent.
    if missing_nine_columns:
        # Trace the missing columns explicitly.
        raise ValueError(f"nine_cells is missing pinned column(s): {missing_nine_columns}.")

    # Enforce the DataFrame container type for the register.
    if not isinstance(daily_scores, pd.DataFrame):
        # Reject non-DataFrame registers explicitly.
        raise TypeError(f"daily_scores must be a pandas.DataFrame, got {type(daily_scores)!r}.")

    # Reuse the boundary validators for the panel and the label vector.
    _require_compound_index_task_12(factor_panel, "factor_panel")

    # Enforce the Series type for the label vector before structural validation.
    if not isinstance(labels, pd.Series):
        # Reject non-Series label arguments explicitly.
        raise TypeError(f"labels must be a pandas.Series, got {type(labels)!r}.")

    # Reuse the compound-index validator for the label vector.
    _require_compound_index_task_12(labels, "labels")

    # Enforce the Series type for the tradability flag before structural validation.
    if not isinstance(tradable, pd.Series):
        # Reject non-Series tradability arguments explicitly.
        raise TypeError(f"tradable must be a pandas.Series, got {type(tradable)!r}.")

    # Reuse the compound-index validator for the tradability flag (same coordinate system).
    _require_compound_index_task_12(tradable, "tradable")

    # Enforce the MultiIndex of the register (expression_id, trade_date).
    if not isinstance(daily_scores.index, pd.MultiIndex) or not {"expression_id", "trade_date"} <= set(daily_scores.index.names):
        # Halt: the register's coordinates are (expression_id, trade_date).
        raise ValueError("daily_scores.index must be a MultiIndex carrying 'expression_id' and 'trade_date'.")

    # Build the per-date masked-stocks map via the upstream primitive (never re-written).
    mask_map = _MASK_COORDINATE_MAP_UPSTREAM(mask_register)

    # Validate the output directory argument type.
    if not isinstance(output_dir, (str, Path)):
        # Reject non-path-like output directories explicitly.
        raise TypeError(f"output_dir must be a str or pathlib.Path, got {type(output_dir)!r}.")

    # Normalise and create the output directory.
    out_dir = Path(output_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    # Accumulate the per-cell result rows in the frozen order.
    result_rows: List[Tuple[str, pd.Timestamp, float, float, float, bool]] = []

    # Accumulate the failing cells for the halt message (empty iff every cell passes).
    failing_cells: List[str] = []

    # Iterate the frozen cells in their pinned order.
    for expression_id_raw, trade_date_raw in zip(nine_cells["expression_id"], nine_cells["trade_date"]):
        # Normalise the expression identifier to str.
        expression_id = str(expression_id_raw)

        # Normalise the trade date to a Timestamp.
        trade_date = pd.Timestamp(trade_date_raw)

        # Read primary_rho from the register at the pinned coordinates.
        try:
            primary_rho = float(daily_scores.loc[(expression_id, trade_date), "rho"])
        except KeyError as exc:
            # Halt: a frozen cell absent from the register invalidates the comparison.
            raise ValueError(f"daily_scores has no cell for (expression_id={expression_id!r}, trade_date={trade_date}).") from exc

        # Slice the eligible cross-section exactly as the primary path builds it (tradable strict-True, mask applied).
        try:
            signal_vec, label_vec = _eligible_slice_task_12(
                factor_panel, labels, tradable, mask_map, expression_id, trade_date,
            )
        except ValueError as exc:
            # Re-raise with the cell coordinates named for pinpoint diagnosis.
            raise ValueError(f"cell (expression_id={expression_id!r}, trade_date={trade_date}) cannot be sliced: {exc}") from exc

        # Recompute the rank correlation with the alternative path (nine-cell slices only).
        if signal_vec.size == 0:
            # No eligible coordinate exists — the statistic's domain is empty; force the halt path.
            alternative_rho = float("nan")
        else:
            # Compute the alternative correlation on the slice.
            alternative_rho = alternative_rank_ic_task_12(signal_vec.tolist(), label_vec.tolist())

        # delta := |primary_rho − alternative_rho| (pinned definition).
        delta = abs(primary_rho - alternative_rho)

        # pass_at_tolerance := tolerance_verdict_task_12(primary_rho, alternative_rho) (pinned predicate).
        pass_at_tolerance = tolerance_verdict_task_12(primary_rho, alternative_rho)

        # Record the failing cells for the halt message.
        if not pass_at_tolerance:
            # Trace the failing cell explicitly.
            failing_cells.append(f"({expression_id}, {trade_date})")

        # Append the per-cell result row in frozen order.
        result_rows.append((expression_id, trade_date, primary_rho, alternative_rho, delta, pass_at_tolerance))

    # Halt if any cell failed all three tolerance arms (the step's mandated halt).
    if failing_cells:
        # No table is deposited; the halt is loud and names every failing cell.
        raise ValueError(
            f"DUAL_CHECK_RESULTS halted: {len(failing_cells)} cell(s) failed all three tolerance arms "
            f"(tol_abs=1e-10, tol_rel=1e-12): {', '.join(failing_cells)}. "
            "This is not a tolerance question — investigate the eligible-slice construction or the tie handling."
        )

    # Materialise the table with the pinned column order.
    results = pd.DataFrame(result_rows, columns=DUAL_CHECK_COLUMNS_TASK_12)

    # Coerce trade_date to datetime64[ns] exactly as the deposit schema requires.
    results["trade_date"] = pd.to_datetime(results["trade_date"]).astype("datetime64[ns]")

    # Coerce the numeric columns to float64.
    results["primary_rho"] = results["primary_rho"].astype(np.float64)
    results["alternative_rho"] = results["alternative_rho"].astype(np.float64)

    # Recompute delta as |primary − alternative| elementwise so the pinned identity holds exactly.
    results["delta"] = (results["primary_rho"] - results["alternative_rho"]).abs().astype(np.float64)

    # Coerce pass_at_tolerance to a real boolean column.
    results["pass_at_tolerance"] = results["pass_at_tolerance"].astype(bool)

    # Resolve the deposit path beneath output_dir.
    csv_path = out_dir / "DUAL_CHECK_RESULTS.csv"

    # Write the deposit without an index (the deposit schema has no index); only on full pass.
    results.to_csv(csv_path, index=False)

    # Return the materialised table.
    return results

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Orchestrator Function: Run Steps 1–3 end to end
# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Orchestrator, Helper: Render the DUAL_CHECK_REPORT.md disclosure document.
# -------------------------------------------------------------------------------------------------------------------------------

def _render_dual_check_report_task_12(
    freeze_sha256: str,
    pass_requirement: str,
    r9_status: str,
    nine_cell_count: int,
    max_abs_delta: float,
) -> str:
    """
    Render the DUAL_CHECK_REPORT.md disclosure document.

    Assembles a markdown report carrying: the frozen artifact row count
    and SHA-256 digest, the a priori selection timing, the pass
    requirement and r9 status, the declared tolerances (``tol_abs =
    1e-10``, ``tol_rel = 1e-12``), the
    ``STRUCTURAL__[D]__verifiable_on_this_run`` and
    ``ILLUSTRATIVE__[I]`` provenance stamps, the largest observed
    ``|primary_rho - alternative_rho|``, the pass-count sentence (the
    9-of-9 wording when ``pass_requirement == "9_of_9"``, or the
    disclosed 8_of_8 variant when ``pass_requirement == "8_of_8"``;
    ``r9_status`` is only interpolated separately in the report text),
    and the verbatim scope sentence: "This validates the measurement
    instrument, not the results. No claim about the aggregate
    distribution follows from this check."

    Parameters
    ----------
    freeze_sha256 : str
        Hex SHA-256 digest of the frozen ``NINE_CELLS.csv`` bytes.
    pass_requirement : str
        ``"9_of_9"`` normally, or ``"8_of_8"`` under the r9
        ``NOT_SATISFIABLE`` path.
    r9_status : str
        ``"SATISFIED"`` or ``"NOT_SATISFIABLE"``.
    nine_cell_count : int
        Number of rows of the frozen artifact, 8 or 9.
    max_abs_delta : float
        Largest ``|primary_rho - alternative_rho|`` across the frozen
        cells; must be finite and non-negative.

    Returns
    -------
    str
        The full markdown document, newline-terminated.

    Raises
    ------
    TypeError
        If ``freeze_sha256`` is not a string.
    ValueError
        If ``pass_requirement`` is not ``"9_of_9"`` or ``"8_of_8"``, if
        ``r9_status`` is not ``"SATISFIED"`` or ``"NOT_SATISFIABLE"``,
        if ``nine_cell_count`` is not 8 or 9, or if ``max_abs_delta`` is
        not a finite, non-negative float.
    """
    # Enforce the type of the freeze digest.
    if not isinstance(freeze_sha256, str):
        # Reject non-string digests explicitly.
        raise TypeError(f"freeze_sha256 must be a str, got {type(freeze_sha256)!r}.")

    # Enforce the membership of the pass requirement.
    if pass_requirement not in {"9_of_9", "8_of_8"}:
        # Reject unknown pass requirements explicitly.
        raise ValueError(f"pass_requirement must be '9_of_9' or '8_of_8', got {pass_requirement!r}.")

    # Enforce the membership of the r9 status.
    if r9_status not in {"SATISFIED", "NOT_SATISFIABLE"}:
        # Reject unknown r9 statuses explicitly.
        raise ValueError(f"r9_status must be 'SATISFIED' or 'NOT_SATISFIABLE', got {r9_status!r}.")

    # Enforce the type of the cell count.
    if not isinstance(nine_cell_count, int) or nine_cell_count not in {8, 9}:
        # Reject non-int or out-of-range cell counts explicitly.
        raise ValueError(f"nine_cell_count must be 8 or 9, got {nine_cell_count!r}.")

    # Enforce the finite type of max_abs_delta.
    if not isinstance(max_abs_delta, float) or not math.isfinite(max_abs_delta) or max_abs_delta < 0.0:
        # Reject non-finite or negative delta magnitudes explicitly.
        raise ValueError(f"max_abs_delta must be a finite, non-negative float, got {max_abs_delta!r}.")

    # Select the verbatim pass-requirement sentence for the branch.
    if pass_requirement == "9_of_9":
        # The 9-of-9 sentence, pinned verbatim by the step text (and its claim-attribution clause).
        pass_sentence = (
            "9-of-9 cells validated at declared tolerance; the paper does not publish its own pass count "
            "(P18.spot_check_pass_count_reported == False); the 9-of-9 claim is this reproduction's, not a paper claim."
        )
    else:
        # The disclosed 8_of_8 variant: r9 was NOT_SATISFIABLE and the requirement adjusted.
        pass_sentence = (
            "8_of_8 cells validated at declared tolerance. r9 is NOT_SATISFIABLE: no admissible cell exists on a date "
            "whose masked coordinate is otherwise factor/label-available, so the P18.spot_check_pass_requirement == '9_of_9' "
            "was adjusted to 8_of_8 and is disclosed here. The paper does not publish its own pass count "
            "(P18.spot_check_pass_count_reported == False); the 8_of_8 claim is this reproduction's, not a paper claim."
        )

    # Assemble the report verbatim (UTF-8, newline-terminated markdown).
    report = (
        "# Task 12 — Dual-Implementation Cross-Check of the Measurement Instrument\n"
        "\n"
        "## Reproduction Report — the measurement instrument, not the results\n"
        "\n"
        f"- Frozen artifact: NINE_CELLS.csv (columns expression_id, trade_date, selection_rule), "
        f"{nine_cell_count} row(s); SHA-256 = {freeze_sha256}\n"
        f"- Selection timing: P18.spot_check_cell_selection_timing == \"A_PRIORI\"; deterministic tie-breaking "
        f"(lowest expression_id, then earliest trade_date).\n"
        f"- Pass requirement: P18.spot_check_pass_requirement == \"{pass_requirement}\"; r9_status == \"{r9_status}\".\n"
        f"- Declared tolerance: tol_abs = 1e-10, tol_rel = 1e-12 (source: [R] reproduction declaration; "
        f"P18.tolerance_nine_cell == \"UNDISCLOSED\" in the paper; the THREE_DECIMALS audit recomputation tolerance "
        f"is expressly not reused here — that would be a category error, as it covers the two mean-Sharpe "
        f"recomputations, not the nine-cell rank correlation).\n"
        f"- ClaimRecord stamp: STRUCTURAL__[D]__verifiable_on_this_run (the check is a construction invariant of the "
        f"reproduction's instrument, not of the paper's).\n"
        f"- Alternative implementation stamp: ILLUSTRATIVE__[I] (a reproduction-engineering addition, not a paper finding).\n"
        f"- Nine-cell slices only: the alternative path was invoked on the frozen cells alone; no other cells were computed.\n"
        f"- Maximum |primary_rho − alternative_rho| observed across the frozen cells: {max_abs_delta:.3e}\n"
        "\n"
        f"{pass_sentence}\n"
        "\n"
        "This validates the measurement instrument, not the results. "
        "No claim about the aggregate distribution follows from this check.\n"
        "\n"
        "Per P18.spot_check_targets_instrument_not_results == True, any inference from this nine-cell check to the "
        "mean or median of the aggregate distribution is a scope violation. "
        "The selection was made on market-data properties (eligibility and tiedness), never on the score values, "
        "so no look-ahead entered the selection.\n"
    )

    # Return the rendered document.
    return report

# -------------------------------------------------------------------------------------------------------------------------------
# Task 12, Orchestrator: Public callable — Run Steps 1–3 end to end.
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_12(
    factor_panel: pd.DataFrame,
    labels: pd.Series,
    tradable: pd.Series,
    mask_register: pd.DataFrame,
    daily_scores: pd.DataFrame,
    output_dir: Union[str, Path],
) -> Dict[str, Any]:
    """
    Orchestrate the Task 12 selection, cross-check and report.

    Runs the three stages in order:

    1. ``select_nine_cells_task_12`` — pick the frozen cells
       (score-based rules from DAILY_SCORES, tiedness rules from the raw
       panels, and the masked-stock rule from ``mask_register``) and
       freeze ``NINE_CELLS.csv`` plus its SHA-256 sidecar before any
       alternative computation runs.
    2. ``build_dual_check_results_task_12`` — read ``primary_rho`` per
       cell, recompute ``alternative_rho`` on the eligible slice,
       evaluate the tolerance predicate (``tol_abs = 1e-10``, ``tol_rel
       = 1e-12``) and halt if any cell fails all three arms.
    3. Render ``DUAL_CHECK_REPORT.md`` with the declared tolerance, both
       provenance stamps, the pass-count sentence and the scope
       sentence.

    Every artifact is written beneath ``output_dir``.

    Parameters
    ----------
    factor_panel : pd.DataFrame
        Signal panel indexed by ``MultiIndex[stock_code, trade_date]``.
    labels : pd.Series
        Label vector on the same compound index.
    tradable : pd.Series
        Tradability flag on the same compound index; bool or nullable
        boolean dtype, threaded so the alternative sees the primary's
        exact eligible cross-section.
    mask_register : pd.DataFrame
        Mask register; may be empty.
    daily_scores : pd.DataFrame
        DAILY_SCORES register indexed by
        ``MultiIndex[expression_id, trade_date]``.
    output_dir : Union[str, Path]
        Directory for ``NINE_CELLS.csv``, ``NINE_CELLS.sha256``,
        ``DUAL_CHECK_RESULTS.csv`` and ``DUAL_CHECK_REPORT.md``; created
        if missing.

    Returns
    -------
    Dict[str, Any]
        Keys ``nine_cells`` (DataFrame with columns ``expression_id``,
        ``trade_date``, ``selection_rule``), ``dual_check_results``
        (DataFrame with columns ``expression_id``, ``trade_date``,
        ``primary_rho``, ``alternative_rho``, ``delta``,
        ``pass_at_tolerance``), ``freeze_sha256`` (str), ``report_path``
        (pathlib.Path), ``pass_requirement`` (str) and ``r9_status``
        (str).

    Raises
    ------
    TypeError
        If ``output_dir`` is neither ``str`` nor ``pathlib.Path``, or
        ``tradable`` is not a Series.
    ValueError
        Propagated from the stages: an unsatisfiable selection rule, a
        frozen cell missing from the register or panels, or any cell
        failing all three tolerance arms.
    """
    # Enforce the output directory argument type at the orchestrator boundary.
    if not isinstance(output_dir, (str, Path)):
        # Reject non-path-like output directories explicitly.
        raise TypeError(f"output_dir must be a str or pathlib.Path, got {type(output_dir)!r}.")

    # Normalise and create the output directory.
    out_dir = Path(output_dir)
    out_dir.mkdir(parents=True, exist_ok=True)

    # ------------------------------------------------------------------
    # Step 1 — selection and freeze (exactly one invocation).
    # ------------------------------------------------------------------
    # Enforce the Series type for the tradability flag at the orchestrator boundary.
    if not isinstance(tradable, pd.Series):
        # Reject non-Series tradability arguments explicitly.
        raise TypeError(f"tradable must be a pandas.Series, got {type(tradable)!r}.")

    # Compute the a priori cells and freeze NINE_CELLS before any alternative work runs.
    selection = select_nine_cells_task_12(
        daily_scores=daily_scores,
        factor_panel=factor_panel,
        labels=labels,
        mask_register=mask_register,
        output_dir=out_dir,
        # Thread tradability so the r5-r8 tie scans use the primary path's eligible cross-section.
        tradable=tradable,
    )

    # Bind the frozen artifact and its digest from the Step 1 result.
    nine_cells = selection["nine_cells"]
    freeze_sha256 = selection["freeze_sha256"]
    pass_requirement = selection["pass_requirement"]
    r9_status = selection["r9_status"]

    # ------------------------------------------------------------------
    # Steps 2 and 3 — alternative implementation on the nine-cell slices only,
    # then the comparison at the declared tolerance (exactly one invocation).
    # ------------------------------------------------------------------
    # Recompute and compare at the declared tolerance; halts if any cell fails.
    # tradable and mask_register are threaded into the comparison stage so the
    # alternative sees the primary's exact eligible cross-section.
    dual_check_results = build_dual_check_results_task_12(
        nine_cells=nine_cells,
        daily_scores=daily_scores,
        factor_panel=factor_panel,
        labels=labels,
        tradable=tradable,
        mask_register=mask_register,
        output_dir=out_dir,
    )

    # Compute the largest |primary_rho − alternative_rho| across the frozen cells for the report.
    max_abs_delta = float(dual_check_results["delta"].max()) if len(dual_check_results) else 0.0

    # ------------------------------------------------------------------
    # Report — the disclosure document (written only on full pass).
    # ------------------------------------------------------------------
    # Render the DUAL_CHECK_REPORT.md document with the pinned verbatim sentences and stamps.
    report = _render_dual_check_report_task_12(
        freeze_sha256=freeze_sha256,
        pass_requirement=pass_requirement,
        r9_status=r9_status,
        nine_cell_count=int(len(nine_cells)),
        max_abs_delta=max_abs_delta,
    )

    # Resolve the report deposit path beneath output_dir.
    report_path = out_dir / "DUAL_CHECK_REPORT.md"

    # Write the report (UTF-8, newline-terminated).
    report_path.write_text(report, encoding="utf-8", newline="\n")

    # Assemble and return the pinned result dictionary.
    return {
        "nine_cells": nine_cells,
        "dual_check_results": dual_check_results,
        "freeze_sha256": freeze_sha256,
        "report_path": report_path,
        "pass_requirement": pass_requirement,
        "r9_status": r9_status,
    }
```

```python
# Task 13: Versioned Correction and Full Re-Evaluation

# ==============================================================================
# Task 13: Versioned Correction and Full Re-Evaluation
# ==============================================================================

try:  # ZoneInfo (IANA database) is the canonical Asia/Shanghai source on this platform.

    _TZ_SHANGHAI: Any = ZoneInfo("Asia/Shanghai")  # P00 config_self_timestamp_timezone == "Asia/Shanghai".
except Exception:  # pragma: no cover - tzdata absent: fall back to a fixed UTC+8 offset.
    _TZ_SHANGHAI = timezone(timedelta(hours=8), name="Asia/Shanghai")  # Fixed offset, same instants.

# -------------------------------------------------------------------------------------------------------------------------------
# Pinned constants: the vocabularies, schemas and tolerances from the study configuration.
# -------------------------------------------------------------------------------------------------------------------------------

# The four-state v1 vocabulary: the three supported states plus the standing register's fourth state.
_V1_STATE_VOCABULARY: Tuple[str, str, str, str] = ("scorable", "insufficient", "unsupported", "unsupported_error")
# The three supported states; v2 cannot carry the v1-only error state (the v2 defect was fixed and re-evaluated).
_V2_STATE_VOCABULARY: Tuple[str, str, str] = ("scorable", "insufficient", "unsupported")
# The four v1 output artifacts the step text pins for the archive (DAILY_SCORES_V1 / FACTOR_AGGREGATES_V1 / LEDGER_V1 / ERROR_LOG_V1).
_V1_REQUIRED_OUTPUTS: Tuple[str, str, str, str] = (
    "DAILY_SCORES_V1.csv",
    "FACTOR_AGGREGATES_V1.csv",
    "LEDGER_V1.csv",
    "ERROR_LOG_V1.csv",
)
# The pinned ERROR_LOG_V1 schema: one row per record that v1 could not evaluate.
_ERROR_LOG_COLUMNS: Tuple[str, str, str, str] = ("expression_id", "error_type", "error_message", "stack_trace_hash")
# Domain separators so the two package hashes can never collide across versions.
_V1_PACKAGE_DOMAIN: bytes = b"sha256-package-v1\n"  # Stream prefix for the v1 archive hash.
_V2_PACKAGE_DOMAIN: bytes = b"sha256-package-v2\n"  # Stream prefix for the v2 archive hash.
# P03_FACTOR_AGGREGATION_PARAMETERS.min_valid_dates_for_scorability == 20 (the scorability gate).
_MIN_VALID_DATES_SCORABLE: int = 20
# P05 pin: the rho tolerance for the unchanged-records identity check ([R]; the P18 spot_check_tolerance class).
_RHO_TOL_ABS_DEFAULT: float = 1e-10
# P05 display precision for mean_rank_ic (three decimals, the printed precision of the paper).
_MEAN_RANK_IC_DISPLAY_DECIMALS: int = 3
# A 64-character lowercase-hex pattern for the registry/commit hash fields.
_HEX64_PATTERN: re.Pattern[str] = re.compile(r"[0-9a-f]{64}\Z")
# The reconstruction fixture size: the cohort size pinned by P06/P04/P42/P44 (1047).
_RECONSTRUCTION_COHORT_SIZE: int = 1047
# The label-window length (P01: 2024-2025, 480 label-available dates).
_RECONSTRUCTION_LABEL_AVAILABLE_DATES: int = 480
# The deterministic reconstruction partition endpoints (P42 version_closures, exercised on planted data).
_RECONSTRUCTION_UNCHANGED_SCORABLE: int = 867  # version_closures: 867 + 56 == 923.
_RECONSTRUCTION_UNCHANGED_INSUFFICIENT: int = 98  # version_closures: 98 + 1 == 99.
_RECONSTRUCTION_UNCHANGED_UNSUPPORTED: int = 25  # ledger: 923 + 99 + 25 == 1047.
_RECONSTRUCTION_V1_ERRORS: int = 57  # P05.version_1_error_count (comparison value).
_RECONSTRUCTION_NEWLY_SCORABLE: int = 56  # P05.newly_scorable_after_fix (comparison value).
_RECONSTRUCTION_RESIDUAL_INSUFFICIENT: int = 1  # P05.residual_insufficient_after_fix (comparison value).
# The redistributability note stamped into the commit records (P43 redistributable_artifacts includes versioned_evaluator_logic).
_REDISTRIBUTION_NOTE: str = (
    "versioned_evaluator_logic redistributable per P43.redistributable_artifacts; "
    "raw_vendor_data and proprietary_evaluator_internals are not (P43.non_redistributable_artifacts)"
)

# -------------------------------------------------------------------------------------------------------------------------------
# Private helpers: shared, deterministic machinery for the three steps.
# -------------------------------------------------------------------------------------------------------------------------------

def _now_shanghai_iso_task_13() -> str:
    """
    Return the current Asia/Shanghai time as an ISO-8601 string.

    Read the aware current time in the ``Asia/Shanghai`` timezone and
    render it with microsecond precision and the numeric UTC offset.

    Returns
    -------
    str
        ISO-8601 timestamp with offset, e.g.
        ``2026-11-12T09:41:07.123456+08:00``; round-trips through
        ``datetime.datetime.fromisoformat``.
    """
    # Read the current aware time in the pinned timezone.
    current: datetime = datetime.datetime.now(_TZ_SHANGHAI)
    # Render with microsecond precision and the numeric offset.
    stamp: str = current.isoformat(timespec="microseconds")
    # Return the stamp for the caller to embed in the commit record.
    return stamp

def _validate_cohort_ids_task_13(cohort_ids: Any, *, caller: str) -> List[str]:
    """
    Validate and normalise a cohort identifier sequence.

    Enforce the cohort contract shared by this module: a non-empty list
    or tuple of unique, non-blank string identifiers. The identifiers
    are returned in a fresh list that preserves the caller's cohort-rank
    order, so later mutation of the caller's object cannot leak in.

    Parameters
    ----------
    cohort_ids : Any
        Caller-supplied cohort identifiers. Must be a ``list`` or
        ``tuple`` whose elements are all ``str``.
    caller : str
        Name of the calling callable, prefixed to every error message.

    Returns
    -------
    List[str]
        The validated identifiers, in input order.

    Raises
    ------
    TypeError
        If ``cohort_ids`` is not a list or tuple, or if any element is
        not a ``str``.
    ValueError
        If the sequence is empty, if any identifier is blank or
        whitespace-only, or if identifiers repeat.
    """
    # The cohort must be a concrete list/tuple; anything else is a contract breach.
    if not isinstance(cohort_ids, (list, tuple)):
        raise TypeError(f"{caller}: cohort_ids must be a list or tuple of str, got {type(cohort_ids).__name__}")
    # Materialise a fresh list so downstream mutation of the caller's object cannot leak in.
    ids: List[str] = list(cohort_ids)
    # The cohort must carry at least one identifier; an empty cohort is no cohort.
    if len(ids) == 0:
        raise ValueError(f"{caller}: cohort_ids must not be empty")
    # Every identifier must be a string (accept no numpy/pandas str subclasses silently).
    for position, item in enumerate(ids):
        if not isinstance(item, str):
            raise TypeError(f"{caller}: cohort_ids[{position}] must be a str, got {type(item).__name__}")
    # No identifier may be blank or whitespace-only.
    for position, item in enumerate(ids):
        if not item.strip():
            raise ValueError(f"{caller}: cohort_ids[{position}] is blank")
    # Identifiers must be unique; duplicates void the cohort.
    seen: set[str] = set()
    duplicates: List[str] = []
    for item in ids:
        if item in seen:
            duplicates.append(item)
        else:
            seen.add(item)
    # Report every duplicated identifier, once each.
    if duplicates:
        unique_duplicates: List[str] = sorted(set(duplicates))
        raise ValueError(f"{caller}: cohort_ids contains duplicated identifiers: {unique_duplicates}")
    # Return the validated cohort in the caller's order.
    return ids

def _resolve_directory_task_13(path_like: Any, *, caller: str, parameter: str) -> Path:
    """
    Resolve and validate an existing directory argument.

    Boundary check for the directory arguments of the archive callables:
    the value must be a non-blank ``str`` naming an existing directory.
    The path is expanded (``~``) and resolved to an absolute ``Path``;
    a missing or non-directory path is never created or repaired.

    Parameters
    ----------
    path_like : Any
        Caller-supplied path, expected to be a ``str``.
    caller : str
        Name of the calling callable, prefixed to error messages.
    parameter : str
        Name of the parameter being validated, used in error messages.

    Returns
    -------
    Path
        The resolved absolute directory path.

    Raises
    ------
    TypeError
        If ``path_like`` is not a ``str``.
    ValueError
        If ``path_like`` is blank or whitespace-only.
    FileNotFoundError
        If the resolved path does not exist.
    NotADirectoryError
        If the resolved path exists but is not a directory.
    """
    # Paths must be supplied as strings; accept no Path objects at this boundary for a stable contract.
    if not isinstance(path_like, str):
        raise TypeError(f"{caller}: {parameter} must be a str path, got {type(path_like).__name__}")
    # Reject blank paths outright.
    if not path_like.strip():
        raise ValueError(f"{caller}: {parameter} must not be blank")
    # Resolve to an absolute Path for the filesystem operations.
    candidate: Path = Path(path_like).expanduser().resolve()
    # A missing path is reported as absent, never repaired silently.
    if not candidate.exists():
        raise FileNotFoundError(f"{caller}: {parameter} does not exist: {candidate}")
    # The path must be a directory.
    if not candidate.is_dir():
        raise NotADirectoryError(f"{caller}: {parameter} is not a directory: {candidate}")
    # Return the validated directory.
    return candidate

def _sorted_package_files_task_13(root: Path) -> List[Path]:
    """
    Enumerate a package tree in canonical order.

    Recursively collect every regular file beneath ``root`` and sort by
    the posix relative path, giving a platform-independent order that is
    the canonical sequence for both the SHA-256 stream and the byte
    copy.

    Parameters
    ----------
    root : Path
        Package root directory; must exist.

    Returns
    -------
    List[Path]
        Paths of the regular files under ``root`` (absolute only when
        ``root`` itself is absolute), sorted by relative posix path.
    """
    # Recursively collect every regular file beneath the root.
    files: List[Path] = [p for p in root.rglob("*") if p.is_file()]
    # Sort by the relative posix path so the order is platform-independent.
    ordered: List[Path] = sorted(files, key=lambda p: p.relative_to(root).as_posix())
    # Return the canonical sequence.
    return ordered

def _package_sha256_task_13(root: Path, *, domain: bytes) -> Tuple[str, List[str]]:
    """
    Compute the deterministic SHA-256 digest of a package tree.

    The digest stream is the domain separator followed, for each file in
    canonical order, by the posix relative path, a NUL byte, the decimal
    byte length, a NUL byte, and the raw file bytes. Identical trees
    hash identically on every machine and host path; any flipped byte or
    added/removed file changes the digest. v1 and v2 are separated by
    distinct domain prefixes so their streams cannot collide.

    Parameters
    ----------
    root : Path
        Package root directory; must exist.
    domain : bytes
        Domain separator prefix written before any file content.

    Returns
    -------
    Tuple[str, List[str]]
        ``(digest, names)``: the 64-character lowercase hex digest, and
        the posix relative paths hashed, in canonical order.

    Raises
    ------
    ValueError
        If the tree contains no regular files.
    """
    # Collect the canonical file sequence.
    ordered: List[Path] = _sorted_package_files_task_13(root)
    # An empty tree cannot produce a meaningful commit; refuse.
    if not ordered:
        raise ValueError(f"package at {root} contains no files; nothing to hash")
    # Initialise the digest with the domain separator.
    digest = hashlib.sha256()
    # Feed the domain separator so v1 and v2 streams can never collide.
    digest.update(domain)
    # Accumulate the canonical names for the commit record.
    names: List[str] = []
    # Walk the tree in canonical order.
    for path in ordered:
        # Compute the relative posix path (the hash never embeds absolute paths).
        relative: str = path.relative_to(root).as_posix()
        # Read the raw bytes of the file.
        data: bytes = path.read_bytes()
        # Feed: relative path, NUL, decimal length, NUL, raw bytes.
        digest.update(relative.encode("utf-8"))
        digest.update(b"\x00")
        digest.update(str(len(data)).encode("ascii"))
        digest.update(b"\x00")
        digest.update(data)
        # Record the canonical name.
        names.append(relative)
    # Return the hex digest and the hashed name list.
    return digest.hexdigest(), names

def _copy_tree_write_once_task_13(source: Path, target: Path, *, caller: str) -> None:
    """
    Copy a tree byte-wise into a write-once namespace.

    Create ``target`` and copy every source file into the mirrored
    relative layout using ``shutil.copyfile`` only, so no mode bits,
    timestamps or extended attributes are carried: byte identity is the
    acceptance criterion. The copy is attempted exactly once.

    Parameters
    ----------
    source : Path
        Package root to copy; must exist.
    target : Path
        Namespace directory to create and populate.
    caller : str
        Name of the calling callable, prefixed to error messages.

    Raises
    ------
    RuntimeError
        If ``target`` already exists; any overwriting is a halt.
    """
    # Write-once: any pre-existing namespace is a halt, never an overwrite.
    if target.exists():
        raise RuntimeError(
            f"{caller}: write-once namespace already exists: {target}; refusing to overwrite (any overwriting is a halt)"
        )
    # Create the target directory (its parents were validated by the caller).
    target.mkdir(parents=True, exist_ok=False)
    # Copy every file in canonical order.
    for path in _sorted_package_files_task_13(source):
        # Compute the destination path preserving the relative layout.
        destination: Path = target / path.relative_to(source)
        # Ensure the destination's parent directory exists.
        destination.parent.mkdir(parents=True, exist_ok=True)
        # Byte-wise copy: no mode bits, no timestamps, no xattrs.
        shutil.copyfile(path, destination)

def _has_designator_task_13(root: Path, token: str) -> bool:
    """
    Report whether a package path carries a version designator.

    Search the relative path components of every file beneath ``root``
    for the designator, case-insensitively. The root's own absolute name
    is not inspected, only the paths relative to it.

    Parameters
    ----------
    root : Path
        Package root directory; must exist.
    token : str
        Designator to look for, such as ``"v1"`` or ``"v2"``; matched
        case-insensitively as a substring of a path component.

    Returns
    -------
    bool
        True if at least one relative path component contains the token,
        False otherwise.
    """
    # Lower-case the token once for a case-insensitive search.
    needle: str = token.lower()
    # Inspect every file's relative path components.
    for path in _sorted_package_files_task_13(root):
        components: Tuple[str, ...] = tuple(part.lower() for part in path.relative_to(root).parts)
        # Match if any component contains the designator (e.g. "v1", "v1_evaluator.py").
        if any(needle in component for component in components):
            return True
    # No designator anywhere: the package is not version-distinguished.
    return False

def _int64_or_none_list_task_13(series: pd.Series) -> List[Optional[int]]:
    """
    Normalise a count column to Python ints or ``None``.

    Reject boolean dtype outright, coerce through ``pd.to_numeric``,
    then map each entry: NA becomes ``None``; every other entry must be
    finite, integral and non-negative and is returned as a Python
    ``int``. This coercion is shared by the register summary, the
    re-evaluation ledger and the closure layer.

    Parameters
    ----------
    series : pd.Series
        Count column, typically ``n_valid_dates``. Nullable integer,
        integral float or object dtype with ints/``None`` are accepted.

    Returns
    -------
    List[Optional[int]]
        One entry per row: a Python ``int`` >= 0, or ``None`` for NA.

    Raises
    ------
    TypeError
        If the column has boolean dtype.
    ValueError
        If numeric coercion fails, or any non-NA value is non-finite,
        non-integral or negative.
    """
    # Booleans are not counts; reject the dtype before any coercion.
    if pd.api.types.is_bool_dtype(series):
        raise TypeError("n_valid_dates must not be boolean dtype")
    # Coerce through pandas' numeric path; non-numeric cells raise.
    try:
        numeric: pd.Series = pd.to_numeric(series, errors="raise")
    except Exception as exc:  # pandas raises ValueError/TypeError depending on the cell.
        raise ValueError(f"n_valid_dates must be numeric; coercion failed: {exc}") from exc
    # Extract entries one by one so NA semantics are handled explicitly.
    values: List[Optional[int]] = []
    for position, raw in enumerate(numeric):
        # Map NA to None.
        if raw is None or (isinstance(raw, float) and math.isnan(raw)) or pd.isna(raw):
            values.append(None)
            continue
        # Coerce to float for the integral test.
        as_float: float = float(raw)
        # Finiteness is mandatory (no inf).
        if not math.isfinite(as_float):
            raise ValueError(f"n_valid_dates[{position}] is not finite")
        # The value must be integral (a count of valid dates is an integer).
        if as_float != math.floor(as_float):
            raise ValueError(f"n_valid_dates[{position}] is non-integral: {as_float}")
        # Counts are non-negative.
        as_int: int = int(as_float)
        if as_int < 0:
            raise ValueError(f"n_valid_dates[{position}] is negative: {as_int}")
        values.append(as_int)
    # Return the normalised list.
    return values

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1:  The v1 archive — a frozen, write-once snapshot of the v1 evaluator + v1 outputs + v1 error ledger.
# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1, Callable 1:  archive_v1_package_task_13 — copy the v1 package into a write-once namespace and stamp V1_ARCHIVE_COMMIT.
# -------------------------------------------------------------------------------------------------------------------------------

def archive_v1_package_task_13(v1_source_dir: str, archive_root: str, cohort_ids: List[str]) -> Dict[str, Any]:
    """
    Deposit the v1 package write-once and stamp the v1 commit.

    Preserve the v1 evaluator, its outputs (``DAILY_SCORES_V1.csv``,
    ``FACTOR_AGGREGATES_V1.csv``, ``LEDGER_V1.csv``) and its error
    ledger (``ERROR_LOG_V1.csv``) as a copied directory rather than a
    shared module path, so v2 cannot refine v1 implicitly. The steps
    are: validate the cohort; resolve ``v1_source_dir``; validate and
    create ``archive_root``; require a ``v1`` designator in some path
    component of the source tree; require each of the four outputs to
    exist exactly once; hash the source tree; copy it byte-wise into
    ``<archive_root>/v1_archive``; stamp the commit time.

    Parameters
    ----------
    v1_source_dir : str
        Path to the v1 package source tree. Must exist, must contain
        each required v1 output exactly once anywhere beneath it, and
        must carry a case-insensitive ``v1`` designator in at least one
        path component.
    archive_root : str
        Non-blank path beneath which the write-once namespace
        ``v1_archive`` is created; created with parents if absent. A
        pre-existing namespace is a halt.
    cohort_ids : List[str]
        Cohort identifiers, non-empty, unique and non-blank; the order
        is the cohort rank order and is preserved.

    Returns
    -------
    Dict[str, Any]
        The ``V1_ARCHIVE_COMMIT`` record with keys ``v1_sha256`` (str,
        64 lowercase hex characters over the serialised package),
        ``v1_timestamp`` (str, ISO-8601 Asia/Shanghai), ``archive_dir``
        (str, absolute path of the deposited namespace), ``n_total_v1``
        (int, ``len(cohort_ids)``), ``v1_required_outputs`` (list of
        str, the four deposited artifact names), ``archived_files``
        (list of str, the canonical posix relative paths hashed and
        copied), ``v1_package_domain`` (str, the hash domain
        separator) and ``redistributable_artifacts`` (str, the
        redistribution note).

    Raises
    ------
    TypeError
        If either path argument is not a ``str``, or ``cohort_ids`` is
        not a list/tuple of ``str``.
    ValueError
        If ``v1_source_dir`` or ``archive_root`` is blank or
        whitespace-only, the cohort is empty/blank/duplicated, a
        required output appears more than once, or no ``v1``
        designator is found.
    FileNotFoundError
        If ``v1_source_dir`` does not exist, or a required v1 output is
        missing from the tree.
    NotADirectoryError
        If ``v1_source_dir`` exists but is not a directory.
    RuntimeError
        If the write-once namespace already exists.
    """
    # Boundary-validate the cohort identifiers first (non-empty, unique, order preserved).
    ids: List[str] = _validate_cohort_ids_task_13(cohort_ids, caller="archive_v1_package_task_13")
    # Boundary-validate the source directory (exists, is a directory).
    source: Path = _resolve_directory_task_13(v1_source_dir, caller="archive_v1_package_task_13", parameter="v1_source_dir")
    # Boundary-validate/create the archive root (parents allowed).
    if not isinstance(archive_root, str):
        raise TypeError(f"archive_v1_package_task_13: archive_root must be a str path, got {type(archive_root).__name__}")
    if not archive_root.strip():
        raise ValueError("archive_v1_package_task_13: archive_root must not be blank")
    root: Path = Path(archive_root).expanduser().resolve()
    root.mkdir(parents=True, exist_ok=True)
    # Enforce the version designator discipline: the v1 package must be self-describing.
    if not _has_designator_task_13(source, "v1"):
        raise ValueError(
            f"archive_v1_package_task_13: no path component under {source} carries the 'v1' designator; "
            "the v1 evaluator must be distinguishable by a v1 designator in the package path"
        )
    # Every required v1 output must be present exactly once beneath the source tree.
    for required in _V1_REQUIRED_OUTPUTS:
        matches: List[Path] = [p for p in source.rglob(required) if p.is_file()]
        if len(matches) == 0:
            raise FileNotFoundError(
                f"archive_v1_package_task_13: required v1 output {required!r} is missing from {source}"
            )
        if len(matches) > 1:
            raise ValueError(
                f"archive_v1_package_task_13: required v1 output {required!r} appears {len(matches)} times under {source}; expected exactly once"
            )
    # Compute the deterministic package hash over the source tree.
    v1_sha256, archived_files = _package_sha256_task_13(source, domain=_V1_PACKAGE_DOMAIN)
    # Deposit the tree into the write-once namespace v1_archive beneath the archive root.
    target: Path = root / "v1_archive"
    _copy_tree_write_once_task_13(source, target, caller="archive_v1_package_task_13")
    # Stamp the commit timestamp (Asia/Shanghai, P00 pin).
    v1_timestamp: str = _now_shanghai_iso_task_13()
    # Assemble and return the V1_ARCHIVE_COMMIT record.
    return {
        # The deterministic 64-hex digest of the serialised package.
        "v1_sha256": v1_sha256,
        # The Asia/Shanghai commit timestamp, distinct from the cohort freeze timestamp.
        "v1_timestamp": v1_timestamp,
        # The deposited namespace (beneath the caller-supplied archive root).
        "archive_dir": str(target),
        # n_total_v1 == cohort size; equals 1,047 in the reproduction's cohort (P04/P42).
        "n_total_v1": len(ids),
        # The deposited outputs (the v1 register trio plus the error ledger).
        "v1_required_outputs": list(_V1_REQUIRED_OUTPUTS),
        # The canonical posix relative paths that were hashed and copied.
        "archived_files": archived_files,
        # The hash domain separator (v1 stream; v2 uses its own).
        "v1_package_domain": _V1_PACKAGE_DOMAIN.decode("ascii"),
        # The P43 redistribution note: the logic is redistributable, the raw data is not.
        "redistributable_artifacts": _REDISTRIBUTION_NOTE,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 1, Callable 2:  summarise_v1_register_task_13 — validate the v1 register and report the four-state partition.
# -------------------------------------------------------------------------------------------------------------------------------

def summarise_v1_register_task_13(v1_ledger: pd.DataFrame, error_log: pd.DataFrame, cohort_ids: List[str]) -> Dict[str, Any]:
    """
    Validate the v1 register and report its four-state partition.

    Check that the v1 ledger holds exactly one row per cohort record
    with a state from the four-state vocabulary, that
    ``n_valid_dates`` is a valid count, and that the scorability
    band holds per state (``scorable`` needs ``n_valid_dates >= 20``;
    ``insufficient`` needs ``0 <= n_valid_dates < 20``;
    ``unsupported_error`` rows must not advertise a positive count).
    Then check that ``ERROR_LOG_V1`` carries the pinned schema, at
    most one row per failing record, and conserves exactly against
    the ``unsupported_error`` class. Finally count the partition and
    report the v1 error count alongside the cohort size.

    Parameters
    ----------
    v1_ledger : pd.DataFrame
        One row per cohort record with at least the columns
        ``expression_id`` (str), ``v1_state`` (str in {``scorable``,
        ``insufficient``, ``unsupported``, ``unsupported_error``})
        and ``n_valid_dates`` (numeric or NA; integral; >= 0). Ids
        must equal the cohort exactly, each once.
    error_log : pd.DataFrame
        The ``ERROR_LOG_V1`` frame; must contain ``expression_id``,
        ``error_type``, ``error_message`` and ``stack_trace_hash``,
        with non-blank strings in the latter three. Its expression ids
        must equal the ``unsupported_error`` id set exactly.
    cohort_ids : List[str]
        Cohort identifiers, non-empty, unique and non-blank; order is
        preserved.

    Returns
    -------
    Dict[str, Any]
        Keys ``n_scorable_v1`` (int), ``n_insufficient_v1`` (int),
        ``n_unsupported_v1`` (int), ``n_unsupported_error_v1`` (int,
        the fourth state, i.e. the records v1 could not evaluate),
        ``n_v1_errors`` (int, equal to ``n_unsupported_error_v1``),
        ``n_total_v1`` (int, equal to ``len(cohort_ids)``),
        ``state_partition_v1`` (dict[str, int] mapping each of the
        four state names to its count) and ``n_valid_dates_bounds``
        (dict[str, int] with keys ``scorable_min`` and
        ``insufficient_max``).

    Raises
    ------
    TypeError
        If either frame is not a ``pd.DataFrame``, ``cohort_ids`` is
        not a list/tuple of ``str``, an ``expression_id`` cell is not
        a ``str``, or ``n_valid_dates`` has boolean dtype.
    ValueError
        If a required column is missing; ledger ids are duplicated,
        missing or unexpected versus the cohort; a ``v1_state`` is
        outside the vocabulary; a count is non-numeric (coercion
        failure), non-integral, non-finite or negative; the
        scorability band is violated; an ``unsupported_error`` row
        carries a non-zero count; the error log duplicates a record,
        or ``error_type``/``error_message``/``stack_trace_hash``
        contains a non-string or blank cell; or the error log and the
        ``unsupported_error`` class do not conserve.
    """
    # Boundary-validate the cohort identifiers.
    ids: List[str] = _validate_cohort_ids_task_13(cohort_ids, caller="summarise_v1_register_task_13")
    # Boundary-validate the ledger frame type.
    if not isinstance(v1_ledger, pd.DataFrame):
        raise TypeError(f"summarise_v1_register_task_13: v1_ledger must be a pandas.DataFrame, got {type(v1_ledger).__name__}")
    # Boundary-validate the error-log frame type.
    if not isinstance(error_log, pd.DataFrame):
        raise TypeError(f"summarise_v1_register_task_13: error_log must be a pandas.DataFrame, got {type(error_log).__name__}")
    # The ledger must carry the pinned columns.
    ledger_required: set[str] = {"expression_id", "v1_state", "n_valid_dates"}
    ledger_missing: List[str] = sorted(ledger_required - set(v1_ledger.columns))
    if ledger_missing:
        raise ValueError(f"summarise_v1_register_task_13: v1_ledger is missing required columns: {ledger_missing}")
    # The error log must carry the pinned schema.
    error_required: set[str] = set(_ERROR_LOG_COLUMNS)
    error_missing: List[str] = sorted(error_required - set(error_log.columns))
    if error_missing:
        raise ValueError(f"summarise_v1_register_task_13: error_log is missing required columns {error_missing}; schema must be {list(_ERROR_LOG_COLUMNS)}")
    # Materialise the ledger columns as plain Python lists (no coercion surprises).
    ledger_ids: List[Any] = list(v1_ledger["expression_id"])
    ledger_states: List[Any] = list(v1_ledger["v1_state"])
    # Every ledger identifier must be a string.
    for position, value in enumerate(ledger_ids):
        if not isinstance(value, str):
            raise TypeError(f"summarise_v1_register_task_13: v1_ledger.expression_id[{position}] must be a str, got {type(value).__name__}")
    # The ledger must carry exactly one row per cohort record and no extras.
    ledger_id_set: set[str] = set(ledger_ids)
    cohort_set: set[str] = set(ids)
    if len(ledger_ids) != len(ledger_id_set):
        duplicated: List[str] = sorted({value for value in ledger_ids if ledger_ids.count(value) > 1})
        raise ValueError(f"summarise_v1_register_task_13: v1_ledger carries duplicated expression_ids: {duplicated}")
    missing_from_ledger: List[str] = sorted(cohort_set - ledger_id_set)
    extra_in_ledger: List[str] = sorted(ledger_id_set - cohort_set)
    if missing_from_ledger or extra_in_ledger:
        raise ValueError(
            "summarise_v1_register_task_13: v1_ledger does not cover the cohort exactly; "
            f"missing={missing_from_ledger}, unexpected={extra_in_ledger}"
        )
    # Every ledger state must belong to the four-state vocabulary.
    for position, value in enumerate(ledger_states):
        if not isinstance(value, str) or value not in _V1_STATE_VOCABULARY:
            raise ValueError(
                f"summarise_v1_register_task_13: v1_ledger.v1_state[{position}] == {value!r}; "
                f"expected one of {list(_V1_STATE_VOCABULARY)}"
            )
    # Normalise the valid-date counts (rejects bool/float/negative/non-integral).
    n_valid_dates: List[Optional[int]] = _int64_or_none_list_task_13(v1_ledger["n_valid_dates"])
    # Enforce the P03 scorability gate per row.
    for position, (state, count) in enumerate(zip(ledger_states, n_valid_dates)):
        if state == "scorable":
            if count is None or count < _MIN_VALID_DATES_SCORABLE:
                raise ValueError(
                    f"summarise_v1_register_task_13: scorable record {ledger_ids[position]!r} carries n_valid_dates={count}; "
                    f"P03 requires n_valid_dates >= {_MIN_VALID_DATES_SCORABLE}"
                )
        elif state == "insufficient":
            if count is None or count < 0 or count >= _MIN_VALID_DATES_SCORABLE:
                raise ValueError(
                    f"summarise_v1_register_task_13: insufficient record {ledger_ids[position]!r} carries n_valid_dates={count}; "
                    f"P03 requires 0 <= n_valid_dates < {_MIN_VALID_DATES_SCORABLE}"
                )
        else:  # unsupported / unsupported_error: no band, but an error row must not carry a positive count.
            if state == "unsupported_error" and count is not None and count != 0:
                raise ValueError(
                    f"summarise_v1_register_task_13: unsupported_error record {ledger_ids[position]!r} carries n_valid_dates={count}; "
                    "a record v1 could not evaluate must not advertise valid dates"
                )
    # Materialise the error-log identifiers as plain Python lists.
    error_ids: List[Any] = list(error_log["expression_id"])
    for position, value in enumerate(error_ids):
        if not isinstance(value, str):
            raise TypeError(f"summarise_v1_register_task_13: error_log.expression_id[{position}] must be a str, got {type(value).__name__}")
    # The error log must carry at most one row per failing record.
    error_id_set: set[str] = set(error_ids)
    if len(error_ids) != len(error_id_set):
        duplicated_error: List[str] = sorted({value for value in error_ids if error_ids.count(value) > 1})
        raise ValueError(f"summarise_v1_register_task_13: ERROR_LOG_V1 carries duplicated records for: {duplicated_error}")
    # The error-type/message/hash cells must be non-empty strings.
    for column in ("error_type", "error_message", "stack_trace_hash"):
        for position, value in enumerate(list(error_log[column])):
            if not isinstance(value, str) or not value.strip():
                raise ValueError(
                    f"summarise_v1_register_task_13: error_log.{column}[{position}] must be a non-empty str, got {value!r}"
                )
    # Conservation: the error log must equal the unsupported_error class exactly.
    unsupported_error_ids: set[str] = {ledger_ids[position] for position, state in enumerate(ledger_states) if state == "unsupported_error"}
    missing_from_log: List[str] = sorted(unsupported_error_ids - error_id_set)
    extra_in_log: List[str] = sorted(error_id_set - unsupported_error_ids)
    if missing_from_log or extra_in_log:
        raise ValueError(
            "summarise_v1_register_task_13: ERROR_LOG_V1 does not conserve against the unsupported_error class; "
            f"missing={missing_from_log}, unexpected={extra_in_log}"
        )
    # Count the four states exactly (python ints, never numpy scalars).
    n_scorable_v1: int = sum(1 for state in ledger_states if state == "scorable")
    n_insufficient_v1: int = sum(1 for state in ledger_states if state == "insufficient")
    n_unsupported_v1: int = sum(1 for state in ledger_states if state == "unsupported")
    n_unsupported_error_v1: int = sum(1 for state in ledger_states if state == "unsupported_error")
    # n_v1_errors equals the fourth-state count and the one-row-per-failure log size.
    n_v1_errors: int = n_unsupported_error_v1
    # n_total_v1 equals the cohort size (1,047 in the reproduction's cohort).
    n_total_v1: int = len(ids)
    # Assemble and return the full four-state partition alongside the error count.
    return {
        # The supported states under v1.
        "n_scorable_v1": int(n_scorable_v1),
        "n_insufficient_v1": int(n_insufficient_v1),
        "n_unsupported_v1": int(n_unsupported_v1),
        # The fourth state: the 57-class records that v1 could not evaluate (comparison value, not a target).
        "n_unsupported_error_v1": int(n_unsupported_error_v1),
        # The v1 error count; reported alongside the paper's 57 as a comparison value.
        "n_v1_errors": int(n_v1_errors),
        # The cohort size; equals 1,047 in the reproduction's cohort.
        "n_total_v1": int(n_total_v1),
        # The explicit four-state partition for audit inspection.
        "state_partition_v1": {
            "scorable": int(n_scorable_v1),
            "insufficient": int(n_insufficient_v1),
            "unsupported": int(n_unsupported_v1),
            "unsupported_error": int(n_unsupported_error_v1),
        },
        # The P03 bounds applied during validation (for the audit trail).
        "n_valid_dates_bounds": {"scorable_min": _MIN_VALID_DATES_SCORABLE, "insufficient_max": _MIN_VALID_DATES_SCORABLE - 1},
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2:  The v2 evaluator — the semantics-preserving correction (scalar broadcasting).
# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2, Callable 3:  apply_v2_conditional_task_13 — elementwise if(cond, a, b) with scalar broadcasting; NA propagates.
# -------------------------------------------------------------------------------------------------------------------------------

def apply_v2_conditional_task_13(cond: np.ndarray, on_true: Any, on_false: Any) -> np.ndarray:
    """
    Evaluate ``if(cond, on_true, on_false)`` elementwise.

    Repair the v1 conditional defect by broadcasting: scalar branch
    operands are expanded to the condition panel's shape and the
    condition's truth is computed elementwise, so a multi-cell panel is
    never coerced to a single Python Boolean. The output cell takes
    ``on_true`` where the condition is non-zero and not NA, ``on_false``
    where it is zero and not NA, and NaN wherever the condition is NA,
    so ``if(NA, a, b) == NA`` holds for both branch orientations.

    Parameters
    ----------
    cond : np.ndarray
        Condition panel with ``ndim >= 1``. Boolean dtype carries no NA;
        any other non-complex, non-datetime numeric dtype is cast to
        float64 and its NaN cells are treated as NA. Complex, datetime
        and object dtypes are rejected.
    on_true : Any
        True branch: a real scalar (int, float, bool, ``np.number``, 0-d
        array) or an array-like broadcastable to ``cond.shape``.
    on_false : Any
        False branch; same contract as ``on_true``.

    Returns
    -------
    np.ndarray
        A new contiguous float64 array with ``cond.shape`` holding the
        elementwise selection, with NaN stamped on every NA condition
        cell.

    Raises
    ------
    TypeError
        If ``cond`` is not a ``np.ndarray`` or has a complex, datetime
        or object dtype; or if a branch operand is str/bytes, complex,
        datetime, object dtype, or a scalar branch is not convertible to
        a real float.
    ValueError
        If ``cond.ndim == 0``; a branch array cannot be cast to float64
        or broadcast to ``cond.shape``; or ``cond`` contains infinite
        values.

    Notes
    -----
    Truth is the pair of masks ``(cond != 0) & ~na_mask`` and ``(cond ==
    0) & ~na_mask``; ``bool(cond)`` is never called anywhere in the
    function.
    """
    # The condition must be a numpy array; accept nothing else at the boundary.
    if not isinstance(cond, np.ndarray):
        raise TypeError(f"apply_v2_conditional_task_13: cond must be a numpy.ndarray, got {type(cond).__name__}")
    # A panel is required: a 0-d scalar condition is not the defect's subject.
    if cond.ndim == 0:
        raise ValueError("apply_v2_conditional_task_13: cond must have ndim >= 1; a scalar condition is not a panel")
    # Complex conditions are out of DSL scope; reject them rather than guessing.
    if np.issubdtype(cond.dtype, np.complexfloating):
        raise TypeError("apply_v2_conditional_task_13: complex conditions are not supported")
    # Datetime conditions are out of DSL scope; reject them rather than guessing.
    if np.issubdtype(cond.dtype, np.datetime64):
        raise TypeError("apply_v2_conditional_task_13: datetime conditions are not supported")
    # Object condition dtypes carry no defined NA semantics; reject them.
    if cond.dtype == object:
        raise TypeError("apply_v2_conditional_task_13: object-dtype conditions are not supported")
    # Compute the NA mask and the float condition; bool conditions carry no NA.
    if cond.dtype == np.bool_:
        # A boolean condition: every cell is known truth; NA is impossible.
        na_mask: np.ndarray = np.zeros(cond.shape, dtype=bool)
        condition: np.ndarray = cond.astype(np.float64)
    else:
        # Coerce to float64; this keeps comparisons and isnan exact.
        condition = cond.astype(np.float64)
        # NA cells are exactly the NaN cells.
        na_mask = np.isnan(condition)
        # Infinite condition values are not part of the DSL's truth domain.
        if np.isinf(condition).any():
            raise ValueError("apply_v2_conditional_task_13: cond contains infinite values")
    # The elementwise truth values, exactly as v1 computes them on the non-defective path.
    true_mask: np.ndarray = (condition != 0.0) & ~na_mask  # Truthy, known cells.
    false_mask: np.ndarray = (condition == 0.0) & ~na_mask  # Falsy, known cells.

    def _broadcast_branch(branch: Any, *, name: str) -> np.ndarray:
        """
        Broadcast one branch operand to the condition's shape.

        Scalars are converted with ``float`` and expanded; array-likes
        are cast to float64 and broadcast. Complex, datetime, object,
        str and bytes operands are rejected.

        Parameters
        ----------
        branch : Any
            The scalar or array-like branch operand.
        name : str
            Branch label used in error messages (``"on_true"`` or
            ``"on_false"``).

        Returns
        -------
        np.ndarray
            A float64 array with exactly ``cond.shape``.

        Raises
        ------
        TypeError
            If the branch is str/bytes, complex, datetime or object
            dtype, or a scalar not convertible to a real float.
        ValueError
            If an array-like cannot be cast to float64 or cannot be
            broadcast to ``cond.shape``.
        """
        # Strings, bytes and object arrays are never numeric branches.
        if isinstance(branch, (str, bytes)):
            raise TypeError(f"apply_v2_conditional_task_13: {name} must not be str/bytes")
        # Scalars (Python scalars, NumPy 0-d arrays and NumPy scalars) are cast and then broadcast.
        if np.isscalar(branch) or (isinstance(branch, np.ndarray) and branch.ndim == 0):
            # Locate the underlying dtype (NumPy 0-d arrays carry a dtype; Python scalars do not).
            scalar_dtype: Any = getattr(branch, "dtype", None)
            # A complex scalar is out of contract (real-valued branches only).
            if isinstance(branch, complex) or (scalar_dtype is not None and np.issubdtype(scalar_dtype, np.complexfloating)):
                raise TypeError(f"apply_v2_conditional_task_13: {name} must not be complex")
            # A datetime scalar is out of contract.
            if isinstance(branch, (pd.Timestamp,)) or (scalar_dtype is not None and np.issubdtype(scalar_dtype, np.datetime64)):
                raise TypeError(f"apply_v2_conditional_task_13: {name} must not be datetime")
            # Cast to float64; int/float/bool/np.number all convert.
            try:
                scalar_value: float = float(branch)  # int/float/bool/np.number all convert.
            except Exception as exc:  # pragma: no cover - float() on np.number never fails here.
                raise TypeError(f"apply_v2_conditional_task_13: {name} scalar is not real-valued: {branch!r}") from exc
            # Broadcasting a scalar is total: expand to the panel's shape.
            return np.full(condition.shape, scalar_value, dtype=np.float64)
        # Array-likes are inspected for complex/datetime before the float cast.
        raw_branch: np.ndarray = np.asarray(branch)
        if np.issubdtype(raw_branch.dtype, np.complexfloating):
            raise TypeError(f"apply_v2_conditional_task_13: {name} must not be complex")
        if np.issubdtype(raw_branch.dtype, np.datetime64):
            raise TypeError(f"apply_v2_conditional_task_13: {name} must not be datetime")
        if raw_branch.dtype == object:
            raise TypeError(f"apply_v2_conditional_task_13: {name} must not be object dtype")
        array_branch: np.ndarray = raw_branch.astype(np.float64)
        # Broadcast to the condition's shape; failure is loud.
        try:
            broadcast: np.ndarray = np.broadcast_to(array_branch, condition.shape)
        except ValueError as exc:
            raise ValueError(
                f"apply_v2_conditional_task_13: {name} of shape {array_branch.shape} cannot be broadcast to cond.shape {condition.shape}"
            ) from exc
        # np.broadcast_to returns a read-only view; np.where does not mutate, so this is safe.
        return broadcast.astype(np.float64, copy=False)

    # Broadcast both branches to the condition's shape.
    on_true_panel: np.ndarray = _broadcast_branch(on_true, name="on_true")
    on_false_panel: np.ndarray = _broadcast_branch(on_false, name="on_false")
    # Select elementwise: the true branch where truthy, the false branch elsewhere.
    output: np.ndarray = np.where(true_mask, on_true_panel, on_false_panel)
    # Stamp NaN wherever the condition is NA: if(NA, a, b) == NA.
    output = np.where(na_mask, np.nan, output)
    # Guarantee a fresh float64 owner array (np.where already owns it; enforce dtype).
    result: np.ndarray = np.ascontiguousarray(output, dtype=np.float64)
    # Return the elementwise result panel.
    return result

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2, Callable 4:  verify_v2_registry_diff_task_13 — prove the v1→v2 diff is the conditional and nothing else.
# -------------------------------------------------------------------------------------------------------------------------------

def verify_v2_registry_diff_task_13(registry_v1: Dict[str, Any], registry_v2: Dict[str, Any]) -> Dict[str, Any]:
    """
    Report which operator entries differ between two registries.

    Compare the tagged ``registry_v1`` and ``registry_v2`` operator maps
    by object identity: a name differs when it is present on only one
    side or the two values are not the same object. The v2 change is a
    registry-level diff, so exactly ``"conditional"`` is expected to
    differ while every other shared operator must remain the identical
    object.

    Parameters
    ----------
    registry_v1 : Dict[str, Any]
        The v1 operator registry, mapping operator name to its callable
        or implementation object.
    registry_v2 : Dict[str, Any]
        The v2 operator registry; same key convention.

    Returns
    -------
    Dict[str, Any]
        Keys: ``only_conditional_differs`` (bool, True iff the sorted
        differing list is exactly ``["conditional"]``);
        ``differing_operators`` (sorted list of names absent on one side
        or failing the identity test); ``shared_operator_identity``
        (bool, True iff every non-conditional name is present on both
        sides as the same object); ``common_operators`` (sorted names
        present in both); ``n_operators_v1`` and ``n_operators_v2`` (int
        registry sizes).

    Raises
    ------
    TypeError
        If either registry is not a Mapping, or any operator name is not
        a str.
    """
    # Both registries must be mappings (dicts qualify).
    if not isinstance(registry_v1, _abc.Mapping):
        raise TypeError(f"verify_v2_registry_diff_task_13: registry_v1 must be a Mapping, got {type(registry_v1).__name__}")
    if not isinstance(registry_v2, _abc.Mapping):
        raise TypeError(f"verify_v2_registry_diff_task_13: registry_v2 must be a Mapping, got {type(registry_v2).__name__}")
    # Operator names must all be strings.
    for tag, registry in (("registry_v1", registry_v1), ("registry_v2", registry_v2)):
        for name in registry:
            if not isinstance(name, str):
                raise TypeError(f"verify_v2_registry_diff_task_13: {tag} carries a non-str operator name: {name!r}")
    # The canonical union of operator names (sorted for a deterministic report).
    union_names: List[str] = sorted(set(registry_v1) | set(registry_v2))
    # A name differs when it is absent on one side or the values are not the same object.
    differing_operators: List[str] = [
        name for name in union_names if name not in registry_v1 or name not in registry_v2 or registry_v1[name] is not registry_v2[name]
    ]
    # Only the conditional may differ for the patch to be a registry-level diff.
    only_conditional_differs: bool = differing_operators == ["conditional"]
    # Every non-conditional operator on both sides must be the same object; absence counts as drift.
    shared_operator_identity: bool = all(
        name in registry_v1 and name in registry_v2 and registry_v1[name] is registry_v2[name]
        for name in union_names
        if name != "conditional"
    )
    # The common operator names (sorted).
    common_operators: List[str] = sorted(set(registry_v1) & set(registry_v2))
    # Assemble and return the diff report.
    return {
        # True iff the conditional is the sole difference.
        "only_conditional_differs": bool(only_conditional_differs),
        # The sorted list of differing operator names.
        "differing_operators": differing_operators,
        # True iff no non-conditional operator drifted (identity, not equality).
        "shared_operator_identity": bool(shared_operator_identity),
        # The sorted names present in both registries.
        "common_operators": common_operators,
        # The registry sizes, for the audit trail.
        "n_operators_v1": int(len(registry_v1)),
        "n_operators_v2": int(len(registry_v2)),
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 2, Callable 5:  build_v2_archive_commit_task_13 — stamp V2_ARCHIVE_COMMIT only after the invariant passes.
# -------------------------------------------------------------------------------------------------------------------------------

def build_v2_archive_commit_task_13(v2_source_dir: str, archive_root: str, invariant_result: Dict[str, Any]) -> Dict[str, Any]:
    """
    Archive the v2 package and stamp the v2 commit record.

    Resolve and validate the v2 source tree, require the v1<->v2
    invariant result to carry a truthy ``all_match``, require a ``v2``
    designator in the package path, compute the SHA-256 of the tree
    under the v2 domain separator, copy the tree into the write-once
    ``v2_archive`` namespace beneath ``archive_root``, and stamp an
    Asia/Shanghai timestamp.

    Parameters
    ----------
    v2_source_dir : str
        Path to the v2 package directory; must exist, be a directory, be
        non-empty and contain at least one path component carrying the
        ``v2`` designator.
    archive_root : str
        Path beneath which ``v2_archive`` is created; the directory is
        made with parents if absent. A pre-existing ``v2_archive``
        halts.
    invariant_result : Dict[str, Any]
        Output of :func:`verify_unchanged_records_task_13`; must be a
        Mapping with key ``all_match``. Optional keys ``diverged`` or
        ``mid_rank_diverge``/``rho_diverge``/``mean_rank_ic_diverge``
        are echoed in the halt message, and ``mode`` is echoed in the
        commit.

    Returns
    -------
    Dict[str, Any]
        The ``V2_ARCHIVE_COMMIT`` record with keys ``v2_sha256`` (64
        lowercase hex digits), ``v2_timestamp`` (ISO-8601
        Asia/Shanghai), ``v2_change_description`` (plain-text paragraph
        naming the conditional defect and the scalar-broadcast fix),
        ``v2_invariant_verification_method`` (the v1<->v2 identity check
        on the unchanged records), ``archive_dir``, ``archived_files``
        (canonical posix relative paths), ``invariant_mode``,
        ``v2_package_domain`` and ``redistributable_artifacts``.

    Raises
    ------
    TypeError
        If ``archive_root`` is not a str, ``invariant_result`` is not a
        Mapping, or ``v2_source_dir`` is not a str path.
    FileNotFoundError
        If ``v2_source_dir`` does not exist.
    NotADirectoryError
        If ``v2_source_dir`` is not a directory.
    ValueError
        If ``archive_root`` is blank, ``invariant_result`` lacks
        ``all_match``, no ``v2`` designator is present in the source
        path, or the package tree is empty.
    RuntimeError
        If ``all_match`` is falsy (halts before ``v2_archive`` is
        created; ``archive_root`` may already exist), or the
        ``v2_archive`` write-once namespace already exists.

    Notes
    -----
    ``archive_root`` is created with parents before the invariant gate
    and the designator check, so a failing gate or designator check may
    still leave ``archive_root`` created; in those failure paths no
    ``v2_archive`` directory and no copied artifacts are written. On
    success the source tree is copied into ``v2_archive``.
    """
    # Boundary-validate the source directory.
    source: Path = _resolve_directory_task_13(v2_source_dir, caller="build_v2_archive_commit_task_13", parameter="v2_source_dir")
    # Boundary-validate the archive root (created with parents if absent).
    if not isinstance(archive_root, str):
        raise TypeError(f"build_v2_archive_commit_task_13: archive_root must be a str path, got {type(archive_root).__name__}")
    if not archive_root.strip():
        raise ValueError("build_v2_archive_commit_task_13: archive_root must not be blank")
    root: Path = Path(archive_root).expanduser().resolve()
    root.mkdir(parents=True, exist_ok=True)
    # The invariant result must be a mapping carrying the pinned key.
    if not isinstance(invariant_result, _abc.Mapping):
        raise TypeError(f"build_v2_archive_commit_task_13: invariant_result must be a Mapping, got {type(invariant_result).__name__}")
    if "all_match" not in invariant_result:
        raise ValueError("build_v2_archive_commit_task_13: invariant_result must carry the 'all_match' key")
    # Capture the diagnostic detail for the halt message before the gate.
    diverged_detail: Any = invariant_result.get("diverged", None)
    if diverged_detail is None:
        # Fall back to the structured diverge lists when present.
        diverged_detail = {
            "mid_rank_diverge": invariant_result.get("mid_rank_diverge"),
            "rho_diverge": invariant_result.get("rho_diverge"),
            "mean_rank_ic_diverge": invariant_result.get("mean_rank_ic_diverge"),
        }
    # The invariant is the gate: a failing v1<->v2 identity halts and no commit is issued.
    if not bool(invariant_result["all_match"]):
        raise RuntimeError(
            "build_v2_archive_commit_task_13: the v1<->v2 invariant on the unchanged records FAILED; "
            f"no v2 commit may be issued. diverged={diverged_detail}"
        )
    # Enforce the v2 designator discipline.
    if not _has_designator_task_13(source, "v2"):
        raise ValueError(
            f"build_v2_archive_commit_task_13: no path component under {source} carries the 'v2' designator; "
            "the v2 evaluator must be distinguishable by a v2 designator in the package path"
        )
    # Compute the deterministic package hash over the v2 tree.
    v2_sha256, archived_files = _package_sha256_task_13(source, domain=_V2_PACKAGE_DOMAIN)
    # Deposit the tree into the write-once namespace v2_archive.
    target: Path = root / "v2_archive"
    _copy_tree_write_once_task_13(source, target, caller="build_v2_archive_commit_task_13")
    # Stamp the Asia/Shanghai commit timestamp.
    v2_timestamp: str = _now_shanghai_iso_task_13()
    # The plain-text change description: names the defect and the fix, redistributable per P36.
    v2_change_description: str = (
        "The v1 conditional operator if(cond_panel, a, b) attempted to interpret a multi-row, "
        "panel-typed condition as a single Python Boolean when either branch operand was a scalar: "
        "the comparison chain reached Python bool(cond_panel), which raises on more than one cell "
        "(P05.version_1_defect; the 57-class in the paper). The v2 correction performs scalar "
        "broadcasting of the scalar branch operand(s) to the condition panel's shape and evaluates "
        "the condition's truth elementwise, exactly along v1's non-defective path "
        "(P05.version_2_change == 'scalar broadcasting'; P05.version_2_semantics_constraint == "
        "PRESERVE_EXISTING_ELEMENTWISE_CONDITIONAL_SEMANTICS; P13 conditional_operator_semantics == "
        "ELEMENTWISE). NA cells in the condition propagate as NA: if(NA, a, b) == NA. No other "
        "operator, terminal, or registry ordering changed; the change is a registry-level diff "
        "tagged registry_v1 -> registry_v2, verified by object identity of the non-conditional "
        "operators."
    )
    # The invariant verification method: names the v1<->v2 identity check on the unchanged records.
    v2_invariant_verification_method: str = (
        "Direct v1<->v2 identity check on the unchanged records, per P05.unchanged_verification_mode "
        "== DIRECT_COMPARISON_NOT_ASSUMPTION: per (expression_id, trade_date) the mid-rank signal "
        "panels must be bit-identical; per date the rho must match within tol_abs = 1e-10 ([R], the "
        "P05/P18 spot-check tolerance class); mean_rank_ic must match to 3-decimal display precision "
        "(P05). The comparison is an identity comparison, never an assumption; any divergence is a "
        "halt with the diverge list, and the full cohort is re-evaluated under v2 (ALL_1047; "
        "P05.failure_only_rerun_permitted == False)."
    )
    # Assemble and return the V2_ARCHIVE_COMMIT record.
    return {
        # The deterministic 64-hex digest of the serialised v2 package.
        "v2_sha256": v2_sha256,
        # The Asia/Shanghai commit timestamp.
        "v2_timestamp": v2_timestamp,
        # The plain-text paragraph naming the defect and the fix (redistributable per P36).
        "v2_change_description": v2_change_description,
        # The v1<->v2 identity check on the unchanged records.
        "v2_invariant_verification_method": v2_invariant_verification_method,
        # The deposited namespace (beneath the caller-supplied archive root).
        "archive_dir": str(target),
        # The canonical posix relative paths that were hashed and copied.
        "archived_files": archived_files,
        # The invariant mode echoed from the input (P05 DIRECT_COMPARISON_NOT_ASSUMPTION).
        "invariant_mode": str(invariant_result.get("mode", "DIRECT_COMPARISON_NOT_ASSUMPTION")),
        # The hash domain separator (v2 stream).
        "v2_package_domain": _V2_PACKAGE_DOMAIN.decode("ascii"),
        # The P43 redistribution note.
        "redistributable_artifacts": _REDISTRIBUTION_NOTE,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3:  The full re-evaluation and the correction closure.
# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3, Callable 6:  verify_unchanged_records_task_13 — the v1↔v2 identity check on the unchanged records.
# -------------------------------------------------------------------------------------------------------------------------------

def _normalise_scores_frame_task_13(scores: Any, *, name: str) -> pd.DataFrame:
    """
    Validate and normalise a DAILY_SCORES register frame.

    Boundary check used before the v1<->v2 comparison: require the
    pinned columns, reject an empty frame and duplicated
    ``(expression_id, trade_date)`` pairs, parse ``trade_date`` to
    ``datetime64[ns]``, convert each ``mid_rank_panel`` cell to a
    non-empty tuple of floats (NaN allowed, infinities rejected) and
    coerce ``rho`` and ``mean_rank_ic`` to float (NaN allowed).

    Parameters
    ----------
    scores : Any
        The register to validate; must be a ``pd.DataFrame`` carrying
        columns ``expression_id``, ``trade_date``, ``mid_rank_panel``,
        ``rho`` and ``mean_rank_ic``.
    name : str
        Frame tag used in error messages (``"v1_scores"`` or
        ``"v2_scores"``).

    Returns
    -------
    pd.DataFrame
        A normalised copy with columns ``expression_id`` (str),
        ``trade_date`` (datetime64[ns]), ``mid_rank_panel``
        (``tuple[float, ...]``), ``rho`` (float) and ``mean_rank_ic``
        (float). The input frame is never mutated.

    Raises
    ------
    TypeError
        If ``scores`` is not a ``pd.DataFrame``, an ``expression_id`` is
        not a str, or a panel cell is not a tuple, list or ndarray.
    ValueError
        If required columns are missing, the frame is empty, a
        ``(expression_id, trade_date)`` pair is duplicated, dates cannot
        be parsed, a panel cell is empty, non-numeric or non-finite, or
        a ``rho``/``mean_rank_ic`` cell is non-numeric.
    """
    # The register must be a DataFrame.
    if not isinstance(scores, pd.DataFrame):
        raise TypeError(f"{name} must be a pandas.DataFrame, got {type(scores).__name__}")
    # The pinned columns must all be present.
    required: set[str] = {"expression_id", "trade_date", "mid_rank_panel", "rho", "mean_rank_ic"}
    missing: List[str] = sorted(required - set(scores.columns))
    if missing:
        raise ValueError(f"{name} is missing required columns: {missing}")
    # An empty register is invalid input, not a vacuous pass.
    if len(scores) == 0:
        raise ValueError(f"{name} must not be empty")
    # Copy so the caller's frame is never mutated.
    frame: pd.DataFrame = scores.loc[:, ["expression_id", "trade_date", "mid_rank_panel", "rho", "mean_rank_ic"]].copy()
    # expression_id must be strings.
    for position, value in enumerate(frame["expression_id"]):
        if not isinstance(value, str):
            raise TypeError(f"{name}.expression_id[{position}] must be a str, got {type(value).__name__}")
    # Duplicate (expression_id, trade_date) pairs are rejected outright.
    duplicated_mask: pd.Series = frame.duplicated(subset=["expression_id", "trade_date"], keep=False)
    if bool(duplicated_mask.any()):
        duplicated_pairs: List[Tuple[str, Any]] = [
            (str(row.expression_id), row.trade_date) for row in frame.loc[duplicated_mask, ["expression_id", "trade_date"]].itertuples()
        ]
        raise ValueError(f"{name} carries duplicated (expression_id, trade_date) pairs: {duplicated_pairs}")
    # trade_date must parse as datetimes.
    try:
        parsed_dates: pd.Series = pd.to_datetime(frame["trade_date"])
    except Exception as exc:
        raise ValueError(f"{name}.trade_date could not be parsed as datetimes: {exc}") from exc
    frame["trade_date"] = parsed_dates.to_numpy(dtype="datetime64[ns]")

    def _panel_tuple(cell: Any, position: int) -> Tuple[float, ...]:
        """
        Normalise one ``mid_rank_panel`` cell to a tuple of floats.

        Parameters
        ----------
        cell : Any
            The cell value: tuple, list or numpy array of real numbers.
        position : int
            Row position used in error messages.

        Returns
        -------
        Tuple[float, ...]
            The panel as a tuple of floats; NaN cells are preserved.

        Raises
        ------
        TypeError
            If the cell is not a tuple, list or ndarray.
        ValueError
            If the cell is empty, carries a non-numeric value, or
            carries a non-finite (infinite) value.
        """
        # Accept the three panel containers only.
        if not isinstance(cell, (tuple, list, np.ndarray)):
            raise TypeError(f"{name}.mid_rank_panel[{position}] must be a tuple, list or ndarray, got {type(cell).__name__}")
        # An empty panel carries no information; reject it.
        if len(cell) == 0:
            raise ValueError(f"{name}.mid_rank_panel[{position}] is empty")
        # Coerce elementwise to float, preserving NaN.
        values: List[float] = []
        for element in list(cell):
            try:
                number: float = float(element)
            except Exception as exc:
                raise ValueError(f"{name}.mid_rank_panel[{position}] carries a non-numeric cell: {element!r}") from exc
            if math.isnan(number):
                values.append(float("nan"))
                continue
            if not math.isfinite(number):
                raise ValueError(f"{name}.mid_rank_panel[{position}] carries a non-finite cell: {element!r}")
            values.append(number)
        return tuple(values)

    # Normalise every panel cell.
    frame["mid_rank_panel"] = [_panel_tuple(cell, position) for position, cell in enumerate(frame["mid_rank_panel"])]

    def _float_or_nan(cell: Any, position: int, column: str) -> float:
        """
        Coerce a ``rho`` or ``mean_rank_ic`` cell to a float.

        Parameters
        ----------
        cell : Any
            The raw cell value; ``None`` and ``pd.NA`` map to NaN.
        position : int
            Row position used in error messages.
        column : str
            Column name used in error messages.

        Returns
        -------
        float
            The cell converted to float, with NaN preserved.

        Raises
        ------
        ValueError
            If the cell cannot be converted to a float.
        """
        # NA markers map to NaN.
        if cell is None or cell is pd.NA:
            return float("nan")
        try:
            number = float(cell)
        except Exception as exc:
            raise ValueError(f"{name}.{column}[{position}] is not numeric: {cell!r}") from exc
        # Return the float (NaN preserved).
        return number

    # Normalise the scalar numeric columns.
    frame["rho"] = [_float_or_nan(cell, position, "rho") for position, cell in enumerate(frame["rho"])]
    frame["mean_rank_ic"] = [_float_or_nan(cell, position, "mean_rank_ic") for position, cell in enumerate(frame["mean_rank_ic"])]
    # Return the normalised frame; the input was never mutated.
    return frame

def verify_unchanged_records_task_13(
    v1_scores: pd.DataFrame,
    v2_scores: pd.DataFrame,
    unchanged_ids: List[str],
    tol_abs: float = _RHO_TOL_ABS_DEFAULT,
) -> Dict[str, Any]:
    """
    Compare v1's and v2's DAILY_SCORES on the unchanged records.

    Normalise both registers, restrict them to ``unchanged_ids`` and
    walk the union of ``(expression_id, trade_date)`` pairs. A pair
    present on only one side is a divergence. Otherwise three
    comparisons are made: (i) mid-rank panels must match cell for cell
    with NaN equal to NaN; (ii) ``rho`` must agree within ``tol_abs``,
    with both-NaN treated as equal and exactly one NaN as a divergence;
    (iii) ``mean_rank_ic`` must agree at three-decimal display
    precision, with the same NaN rule.

    Parameters
    ----------
    v1_scores : pd.DataFrame
        The v1 register with columns ``expression_id``, ``trade_date``,
        ``mid_rank_panel``, ``rho`` and ``mean_rank_ic``; unique per
        ``(expression_id, trade_date)``.
    v2_scores : pd.DataFrame
        The v2 register; same contract as ``v1_scores``.
    unchanged_ids : List[str]
        Identifiers of the records v1 evaluated successfully. Must be a
        non-empty list/tuple of unique str. Rows for other identifiers
        are ignored.
    tol_abs : float, optional
        Absolute tolerance for the per-date ``rho`` comparison. Must be
        a finite, non-negative real number; booleans are rejected.
        Default is ``1e-10``.

    Returns
    -------
    Dict[str, Any]
        Keys: ``mid_rank_diverge`` (sorted list of ``(expression_id,
        trade_date)`` tuples), ``rho_diverge`` (same shape),
        ``mean_rank_ic_diverge`` (sorted list of expression ids),
        ``missing_pairs`` (sorted list of pairs present on only one
        side), ``all_match`` (bool, True iff all four lists are empty),
        ``tol_abs`` (float, echoed), ``n_pairs_compared`` (int),
        ``n_records_compared`` (int) and ``mode``
        (``"DIRECT_COMPARISON_NOT_ASSUMPTION"``).

    Raises
    ------
    TypeError
        If a frame is not a ``pd.DataFrame``, ``unchanged_ids`` is not a
        list/tuple of str, ``tol_abs`` is not a real non-boolean number,
        an ``expression_id`` is not a str, or a panel cell is not a
        tuple, list or ndarray.
    ValueError
        If ``unchanged_ids`` is empty or duplicated, ``tol_abs`` is
        negative or non-finite, or either register has missing columns,
        is empty, carries duplicated ``(expression_id, trade_date)``
        pairs, has unparseable ``trade_date``, an empty, non-numeric or
        non-finite panel cell, or a non-numeric ``rho`` or
        ``mean_rank_ic`` value.
    """
    # Boundary-validate the unchanged identifier list.
    ids: List[str] = _validate_cohort_ids_task_13(unchanged_ids, caller="verify_unchanged_records_task_13")
    # Boundary-validate the tolerance: a real, finite, non-negative float.
    if isinstance(tol_abs, bool) or not isinstance(tol_abs, (int, float, np.number)):
        raise TypeError(f"verify_unchanged_records_task_13: tol_abs must be a real number, got {type(tol_abs).__name__}")
    tol: float = float(tol_abs)
    if not math.isfinite(tol) or tol < 0.0:
        raise ValueError(f"verify_unchanged_records_task_13: tol_abs must be finite and non-negative, got {tol}")
    # Boundary-validate and normalise both registers (columns, dtypes, duplicates).
    v1_frame: pd.DataFrame = _normalise_scores_frame_task_13(v1_scores, name="v1_scores")
    v2_frame: pd.DataFrame = _normalise_scores_frame_task_13(v2_scores, name="v2_scores")
    # Restrict both registers to the unchanged identifiers.
    id_set: set[str] = set(ids)
    v1_view: pd.DataFrame = v1_frame[v1_frame["expression_id"].isin(id_set)]
    v2_view: pd.DataFrame = v2_frame[v2_frame["expression_id"].isin(id_set)]
    # Build (expression_id, trade_date) -> row lookups for the restricted views.
    v1_lookup: Dict[Tuple[str, Any], Tuple[Any, float, float]] = {
        (row.expression_id, row.trade_date): (row.mid_rank_panel, row.rho, row.mean_rank_ic) for row in v1_view.itertuples()
    }
    v2_lookup: Dict[Tuple[str, Any], Tuple[Any, float, float]] = {
        (row.expression_id, row.trade_date): (row.mid_rank_panel, row.rho, row.mean_rank_ic) for row in v2_view.itertuples()
    }
    # The full pair union drives the comparison; missing pairs diverge by construction.
    union_pairs: List[Tuple[str, Any]] = sorted(set(v1_lookup) | set(v2_lookup), key=lambda pair: (pair[0], pair[1]))
    # Accumulate the divergence lists.
    mid_rank_diverge: List[Tuple[str, Any]] = []
    rho_diverge: List[Tuple[str, Any]] = []
    mean_rank_ic_diverge_ids: set[str] = set()
    missing_pairs: List[Tuple[str, Any]] = []
    # Compare every pair.
    for pair in union_pairs:
        # Missing on either side: divergence, recorded separately.
        if pair not in v1_lookup or pair not in v2_lookup:
            missing_pairs.append(pair)
            continue
        v1_panel, v1_rho, v1_mric = v1_lookup[pair]
        v2_panel, v2_rho, v2_mric = v2_lookup[pair]
        # (i) Mid-rank panels: bit-identical (NaN == NaN); length or any cell differing diverges.
        panels_equal: bool = len(v1_panel) == len(v2_panel) and bool(
            np.array_equal(np.asarray(v1_panel, dtype=np.float64), np.asarray(v2_panel, dtype=np.float64), equal_nan=True)
        )
        if not panels_equal:
            mid_rank_diverge.append(pair)
        # (ii) rho: identical within tol_abs; both-NaN is identical, exactly one NaN diverges.
        v1_rho_nan: bool = math.isnan(v1_rho)
        v2_rho_nan: bool = math.isnan(v2_rho)
        if v1_rho_nan and v2_rho_nan:
            rho_ok: bool = True
        elif v1_rho_nan != v2_rho_nan:
            rho_ok = False
        else:
            rho_ok = abs(v1_rho - v2_rho) <= tol
        if not rho_ok:
            rho_diverge.append(pair)
        # (iii) mean_rank_ic: identical to 3-decimal display precision; both-NaN is identical.
        v1_mric_nan: bool = math.isnan(v1_mric)
        v2_mric_nan: bool = math.isnan(v2_mric)
        if v1_mric_nan and v2_mric_nan:
            mric_ok: bool = True
        elif v1_mric_nan != v2_mric_nan:
            mric_ok = False
        else:
            mric_ok = round(v1_mric, _MEAN_RANK_IC_DISPLAY_DECIMALS) == round(v2_mric, _MEAN_RANK_IC_DISPLAY_DECIMALS)
        if not mric_ok:
            mean_rank_ic_diverge_ids.add(pair[0])
    # Build a set of the missing pairs for O(1) membership tests.
    missing_lookup: set[Tuple[str, Any]] = set(missing_pairs)
    # all_match holds only when no divergence and no missing pair was found anywhere.
    all_match: bool = not mid_rank_diverge and not rho_diverge and not mean_rank_ic_diverge_ids and not missing_pairs
    # Assemble and return the comparison report.
    return {
        # Per-pair mid-rank panel divergences, sorted.
        "mid_rank_diverge": mid_rank_diverge,
        # Per-pair rho divergences beyond the pinned tolerance, sorted.
        "rho_diverge": rho_diverge,
        # Expression ids whose mean_rank_ic moved at 3-decimal precision, sorted.
        "mean_rank_ic_diverge": sorted(mean_rank_ic_diverge_ids),
        # Pairs present only on one side (a divergence by construction), sorted.
        "missing_pairs": missing_pairs,
        # True iff every unchanged record is bit-identical/within tolerance/unchanged at display precision.
        "all_match": bool(all_match),
        # The tolerance actually applied (echoed).
        "tol_abs": float(tol),
        # How many (expression_id, trade_date) pairs were compared.
        "n_pairs_compared": int(len(union_pairs) - len(missing_pairs)),
        # How many unchanged records were present on both sides.
        "n_records_compared": int(len({pair[0] for pair in union_pairs if pair not in missing_lookup})),
        # The verification mode (P05 pin).
        "mode": "DIRECT_COMPARISON_NOT_ASSUMPTION",
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3, Callable 7:  build_re_evaluation_ledger_task_13 — RE_EVALUATION_LEDGER, one row per cohort record.
# -------------------------------------------------------------------------------------------------------------------------------

def build_re_evaluation_ledger_task_13(
    cohort_ids: List[str],
    v1_results: pd.DataFrame,
    v2_results: pd.DataFrame,
) -> pd.DataFrame:
    """
    Build the one-row-per-record re-evaluation ledger.

    Validate both results frames against the full cohort and against
    their state vocabularies, then assemble one row per cohort record in
    cohort rank order (Task 13, Step 3). Each frame must cover the
    cohort exactly; a failure-only subset, which would select on the
    error variable, is rejected. ``state_changed`` is derived rather
    than trusted: it is True exactly when the v1 and v2 states differ.
    Valid date counts are carried as nullable ``Int64``, so NA survives.

    Parameters
    ----------
    cohort_ids : List[str]
        Cohort identifiers; must be a non-empty, unique list of str. The
        ledger row order equals this order.
    v1_results : pd.DataFrame
        One row per cohort record with columns ``expression_id`` (str),
        ``state`` (one of ``"scorable"``, ``"insufficient"``,
        ``"unsupported"``, ``"unsupported_error"``) and
        ``n_valid_dates`` (integral or NA; v1-errored records carry 0 or
        NA).
    v2_results : pd.DataFrame
        One row per cohort record with ``expression_id``, ``state`` (one
        of ``"scorable"``, ``"insufficient"``, ``"unsupported"``) and
        ``n_valid_dates``.

    Returns
    -------
    pd.DataFrame
        Range-indexed frame in cohort order with columns
        ``expression_id`` (object), ``v1_state`` (object), ``v2_state``
        (object), ``state_changed`` (bool), ``n_valid_dates_v1``
        (``Int64``) and ``n_valid_dates_v2`` (``Int64``).

    Raises
    ------
    TypeError
        If ``cohort_ids`` is not a list/tuple of str, if either frame is
        not a ``pandas.DataFrame``, if an ``expression_id`` is not a
        str, or if ``n_valid_dates`` is boolean dtype.
    ValueError
        If a frame lacks a required column, carries duplicated, missing
        or unexpected identifiers, holds a state outside its vocabulary,
        carries non-integral or negative counts, or breaks the
        valid-date bands: ``scorable`` below the scorable minimum (20),
        ``insufficient`` at 20 or above, ``unsupported_error`` with a
        non-zero count.

    Notes
    -----
    The band checks implement the P03 partition on valid trading dates.
    """
    # Boundary-validate the cohort identifiers.
    ids: List[str] = _validate_cohort_ids_task_13(cohort_ids, caller="build_re_evaluation_ledger_task_13")
    # Boundary-validate the frame types.
    if not isinstance(v1_results, pd.DataFrame):
        raise TypeError(f"build_re_evaluation_ledger_task_13: v1_results must be a pandas.DataFrame, got {type(v1_results).__name__}")
    if not isinstance(v2_results, pd.DataFrame):
        raise TypeError(f"build_re_evaluation_ledger_task_13: v2_results must be a pandas.DataFrame, got {type(v2_results).__name__}")

    def _validate_results_frame(frame: pd.DataFrame, *, tag: str, vocabulary: Tuple[str, ...]) -> Dict[str, Tuple[str, Optional[int]]]:
        """
        Validate one results frame and index it by expression_id.

        Check required columns, identifier type and uniqueness, exact
        cohort coverage, the state vocabulary and the valid-date bands.

        Parameters
        ----------
        frame : pd.DataFrame
            The results frame to validate.
        tag : str
            Frame tag used in error messages (``"v1_results"`` or
            ``"v2_results"``).
        vocabulary : Tuple[str, ...]
            The admissible state vocabulary.

        Returns
        -------
        Dict[str, Tuple[str, Optional[int]]]
            Map from ``expression_id`` to ``(state, n_valid_dates)``;
            the count is ``None`` when it is missing.

        Raises
        ------
        TypeError
            If an ``expression_id`` is not a str, or the counts are
            boolean dtype.
        ValueError
            If a required column is missing, identifiers are duplicated
            or do not cover the cohort exactly, a state falls outside
            ``vocabulary``, or a count violates its band.
        """
        # Required columns must be present.
        required: set[str] = {"expression_id", "state", "n_valid_dates"}
        missing: List[str] = sorted(required - set(frame.columns))
        if missing:
            raise ValueError(f"build_re_evaluation_ledger_task_13: {tag} is missing required columns: {missing}")
        # Materialise the columns as plain Python lists.
        row_ids: List[Any] = list(frame["expression_id"])
        row_states: List[Any] = list(frame["state"])
        # Identifiers must be strings.
        for position, value in enumerate(row_ids):
            if not isinstance(value, str):
                raise TypeError(f"build_re_evaluation_ledger_task_13: {tag}.expression_id[{position}] must be a str, got {type(value).__name__}")
        # Duplicated identifiers are rejected.
        if len(row_ids) != len(set(row_ids)):
            duplicated: List[str] = sorted({value for value in row_ids if row_ids.count(value) > 1})
            raise ValueError(f"build_re_evaluation_ledger_task_13: {tag} carries duplicated expression_ids: {duplicated}")
        # Every cohort record must be present; no unexpected ids; a failure-only subset is rejected here.
        cohort_set: set[str] = set(ids)
        frame_set: set[str] = set(row_ids)
        missing_ids: List[str] = sorted(cohort_set - frame_set)
        unexpected_ids: List[str] = sorted(frame_set - cohort_set)
        if missing_ids or unexpected_ids:
            raise ValueError(
                f"build_re_evaluation_ledger_task_13: {tag} does not cover the cohort exactly "
                f"(a failure-only re-run voids the correction); missing={missing_ids}, unexpected={unexpected_ids}"
            )
        # States must belong to the admissible vocabulary.
        for position, value in enumerate(row_states):
            if not isinstance(value, str) or value not in vocabulary:
                raise ValueError(
                    f"build_re_evaluation_ledger_task_13: {tag}.state[{position}] == {value!r}; expected one of {list(vocabulary)}"
                )
        # Normalise the counts.
        counts: List[Optional[int]] = _int64_or_none_list_task_13(frame["n_valid_dates"])
        # Enforce the P03 bands and the error-class consistency rule.
        for position, (state, count) in enumerate(zip(row_states, counts)):
            if state == "scorable":
                if count is None or count < _MIN_VALID_DATES_SCORABLE:
                    raise ValueError(
                        f"build_re_evaluation_ledger_task_13: scorable record {row_ids[position]!r} carries n_valid_dates={count}; "
                        f"P03 requires n_valid_dates >= {_MIN_VALID_DATES_SCORABLE}"
                    )
            elif state == "insufficient":
                if count is None or count >= _MIN_VALID_DATES_SCORABLE:
                    raise ValueError(
                        f"build_re_evaluation_ledger_task_13: insufficient record {row_ids[position]!r} carries n_valid_dates={count}; "
                        f"P03 requires 0 <= n_valid_dates < {_MIN_VALID_DATES_SCORABLE}"
                    )
            elif state == "unsupported_error":
                if count is not None and count != 0:
                    raise ValueError(
                        f"build_re_evaluation_ledger_task_13: unsupported_error record {row_ids[position]!r} carries n_valid_dates={count}; "
                        "a record v1 could not evaluate must carry 0 or NA valid dates"
                    )
        # Index by expression_id and return.
        return {row_ids[position]: (row_states[position], counts[position]) for position in range(len(row_ids))}

    # Validate both frames against their vocabularies.
    v1_lookup: Dict[str, Tuple[str, Optional[int]]] = _validate_results_frame(v1_results, tag="v1_results", vocabulary=_V1_STATE_VOCABULARY)
    v2_lookup: Dict[str, Tuple[str, Optional[int]]] = _validate_results_frame(v2_results, tag="v2_results", vocabulary=_V2_STATE_VOCABULARY)
    # Reindex to cohort order and assemble the ledger columns.
    expression_ids: List[str] = list(ids)
    v1_states: List[str] = [v1_lookup[identifier][0] for identifier in ids]
    v2_states: List[str] = [v2_lookup[identifier][0] for identifier in ids]
    n_valid_v1: List[Optional[int]] = [v1_lookup[identifier][1] for identifier in ids]
    n_valid_v2: List[Optional[int]] = [v2_lookup[identifier][1] for identifier in ids]
    # state_changed is derived, never trusted: True exactly when the two states differ.
    state_changed: List[bool] = [v1_state != v2_state for v1_state, v2_state in zip(v1_states, v2_states)]
    # Assemble the ledger with the pinned columns.
    ledger: pd.DataFrame = pd.DataFrame(
        {
            # The cohort identifier, kept in cohort rank order.
            "expression_id": pd.Series(expression_ids, dtype="object"),
            # The v1 state (four-state vocabulary).
            "v1_state": pd.Series(v1_states, dtype="object"),
            # The v2 state (three-state vocabulary).
            "v2_state": pd.Series(v2_states, dtype="object"),
            # True exactly when the correction changed the record's state.
            "state_changed": pd.Series(state_changed, dtype=bool),
            # Nullable Int64; v1-errored records carry 0 or NA (no valid dates).
            "n_valid_dates_v1": pd.array(n_valid_v1, dtype="Int64"),
            # Nullable Int64.
            "n_valid_dates_v2": pd.array(n_valid_v2, dtype="Int64"),
        }
    )
    # Return the ledger with a fresh range index.
    return ledger

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Step 3, Callable 8:  compute_re_evaluation_closures_task_13 — the four P05/P40/P42 correction closures.
# -------------------------------------------------------------------------------------------------------------------------------

def compute_re_evaluation_closures_task_13(ledger: pd.DataFrame) -> Dict[str, Any]:
    """
    Validate the re-evaluation ledger and evaluate four closures.

    Re-validate the ledger (columns, identifier uniqueness, state
    vocabularies, ``state_changed`` consistency, valid-date bands), halt
    on any v1-unchanged record whose v2 state differs, then evaluate the
    four P05 arithmetic closures on the data's own numbers:

    .. code-block::

        N_total     = N_unchanged + N_v1_errors
        N_unchanged = N_unchanged_scorable
                      + N_unchanged_insufficient
                      + N_unchanged_unsupported
        N_scorable  = N_unchanged_scorable + N_newly_scorable
        N_insufficient = N_unchanged_insufficient
                         + N_residual_insufficient

    A v1-errored record that becomes unsupported under v2 is counted in
    ``n_newly_unsupported`` and, because every v2 state is tallied, also
    in the v2 total ``n_unsupported``; it enters neither the scorable
    nor the insufficient bucket.

    Parameters
    ----------
    ledger : pd.DataFrame
        The re-evaluation ledger; must be non-empty with unique str
        ``expression_id``, ``v1_state`` in the four-state vocabulary,
        ``v2_state`` in the three-state vocabulary, boolean
        ``state_changed`` consistent with the two states, and
        ``n_valid_dates_v1`` / ``n_valid_dates_v2`` integral or NA.

    Returns
    -------
    Dict[str, Any]
        Verdicts as Python bools: ``closure_1_total_holds``,
        ``closure_2_unchanged_holds``, ``closure_3_scorable_holds``,
        ``closure_4_insufficient_holds``, ``all_closures_hold`` (the
        conjunction of the four) and
        ``n_unsupported_conservation_holds`` (``n_unchanged_unsupported
        + n_newly_unsupported == n_unsupported``; informational, not
        part of ``all_closures_hold``). Operands as Python ints:
        ``n_total``, ``n_unchanged``, ``n_v1_errors``,
        ``n_unchanged_scorable``, ``n_unchanged_insufficient``,
        ``n_unchanged_unsupported``, ``n_newly_scorable``,
        ``n_residual_insufficient``, ``n_newly_unsupported``,
        ``n_scorable``, ``n_insufficient`` and ``n_unsupported``.

    Raises
    ------
    TypeError
        If ``ledger`` is not a ``pandas.DataFrame``, ``state_changed``
        is not boolean dtype, an ``expression_id`` is not a str, or a
        valid-date column is boolean dtype.
    ValueError
        If the ledger is empty, a required column is missing,
        identifiers are duplicated, a state falls outside its
        vocabulary, ``state_changed`` contradicts the two states, a
        valid-date band is violated, or a record v1 evaluated carries a
        different v2 state (the v1-success invariance halt).

    Notes
    -----
    The invariance halt enforces that the v2 correction is
    semantics-preserving on records v1 already evaluated.
    """
    # The ledger must be a DataFrame.
    if not isinstance(ledger, pd.DataFrame):
        raise TypeError(f"compute_re_evaluation_closures_task_13: ledger must be a pandas.DataFrame, got {type(ledger).__name__}")
    # The pinned columns must all be present.
    required: set[str] = {"expression_id", "v1_state", "v2_state", "state_changed", "n_valid_dates_v1", "n_valid_dates_v2"}
    missing: List[str] = sorted(required - set(ledger.columns))
    if missing:
        raise ValueError(f"compute_re_evaluation_closures_task_13: ledger is missing required columns: {missing}")
    # An empty ledger is invalid input.
    if len(ledger) == 0:
        raise ValueError("compute_re_evaluation_closures_task_13: ledger must not be empty")
    # state_changed must be boolean dtype.
    if not pd.api.types.is_bool_dtype(ledger["state_changed"]):
        raise TypeError("compute_re_evaluation_closures_task_13: state_changed must be boolean dtype")
    # Materialise the columns as plain Python lists.
    expression_ids: List[Any] = list(ledger["expression_id"])
    v1_states: List[Any] = list(ledger["v1_state"])
    v2_states: List[Any] = list(ledger["v2_state"])
    state_changed: List[bool] = [bool(value) for value in ledger["state_changed"]]
    # Identifiers must be strings; duplicates rejected.
    for position, value in enumerate(expression_ids):
        if not isinstance(value, str):
            raise TypeError(f"compute_re_evaluation_closures_task_13: ledger.expression_id[{position}] must be a str, got {type(value).__name__}")
    if len(expression_ids) != len(set(expression_ids)):
        duplicated: List[str] = sorted({value for value in expression_ids if expression_ids.count(value) > 1})
        raise ValueError(f"compute_re_evaluation_closures_task_13: ledger carries duplicated expression_ids: {duplicated}")
    # State vocabularies are enforced per column.
    for position, value in enumerate(v1_states):
        if not isinstance(value, str) or value not in _V1_STATE_VOCABULARY:
            raise ValueError(
                f"compute_re_evaluation_closures_task_13: ledger.v1_state[{position}] == {value!r}; expected one of {list(_V1_STATE_VOCABULARY)}"
            )
    for position, value in enumerate(v2_states):
        if not isinstance(value, str) or value not in _V2_STATE_VOCABULARY:
            raise ValueError(
                f"compute_re_evaluation_closures_task_13: ledger.v2_state[{position}] == {value!r}; expected one of {list(_V2_STATE_VOCABULARY)}"
            )
    # state_changed must equal (v1_state != v2_state), recomputed here.
    recomputed_changed: List[bool] = [v1 != v2 for v1, v2 in zip(v1_states, v2_states)]
    for position, (recorded, recomputed) in enumerate(zip(state_changed, recomputed_changed)):
        if recorded != recomputed:
            raise ValueError(
                f"compute_re_evaluation_closures_task_13: state_changed[{position}] ({recorded}) contradicts "
                f"(v1_state, v2_state) = ({v1_states[position]!r}, {v2_states[position]!r}); expected {recomputed}"
            )
    # Normalise both valid-date columns.
    n_valid_v1: List[Optional[int]] = _int64_or_none_list_task_13(ledger["n_valid_dates_v1"])
    n_valid_v2: List[Optional[int]] = _int64_or_none_list_task_13(ledger["n_valid_dates_v2"])
    # Re-enforce the P03 bands on both columns (defence in depth; the builder already enforced them).
    for position, state in enumerate(list(v1_states)):
        count: Optional[int] = n_valid_v1[position]
        if state == "scorable" and (count is None or count < _MIN_VALID_DATES_SCORABLE):
            raise ValueError(f"compute_re_evaluation_closures_task_13: v1 scorable record {expression_ids[position]!r} carries n_valid_dates={count}")
        if state == "insufficient" and (count is None or count >= _MIN_VALID_DATES_SCORABLE):
            raise ValueError(f"compute_re_evaluation_closures_task_13: v1 insufficient record {expression_ids[position]!r} carries n_valid_dates={count}")
        if state == "unsupported_error" and count is not None and count != 0:
            raise ValueError(f"compute_re_evaluation_closures_task_13: unsupported_error record {expression_ids[position]!r} carries n_valid_dates={count}")
    for position, state in enumerate(list(v2_states)):
        count = n_valid_v2[position]
        if state == "scorable" and (count is None or count < _MIN_VALID_DATES_SCORABLE):
            raise ValueError(f"compute_re_evaluation_closures_task_13: v2 scorable record {expression_ids[position]!r} carries n_valid_dates={count}")
        if state == "insufficient" and (count is None or count >= _MIN_VALID_DATES_SCORABLE):
            raise ValueError(f"compute_re_evaluation_closures_task_13: v2 insufficient record {expression_ids[position]!r} carries n_valid_dates={count}")
    # The v1-success invariance: a v1-unchanged record must be byte-identical in state under v2.
    invariance_violations: List[str] = [
        str(expression_ids[position])
        for position in range(len(expression_ids))
        if v1_states[position] != "unsupported_error" and v1_states[position] != v2_states[position]
    ]
    if invariance_violations:
        raise ValueError(
            "compute_re_evaluation_closures_task_13: v1-success invariance violated on unchanged records "
            f"(the fix is not semantics-preserving): {sorted(invariance_violations)}"
        )
    # Aggregate with explicit masks (the group-by on (v1_state, v2_state), as audit code).
    n_total: int = len(expression_ids)
    error_mask: List[bool] = [state == "unsupported_error" for state in v1_states]
    unchanged_mask: List[bool] = [not flag for flag in error_mask]
    n_v1_errors: int = sum(error_mask)
    n_unchanged: int = n_total - n_v1_errors
    # The unchanged records partition into their states under v2.
    n_unchanged_scorable: int = sum(1 for position in range(n_total) if unchanged_mask[position] and v2_states[position] == "scorable")
    n_unchanged_insufficient: int = sum(1 for position in range(n_total) if unchanged_mask[position] and v2_states[position] == "insufficient")
    n_unchanged_unsupported: int = sum(1 for position in range(n_total) if unchanged_mask[position] and v2_states[position] == "unsupported")
    # The v1-errored records partition into their states under v2.
    n_newly_scorable: int = sum(1 for position in range(n_total) if error_mask[position] and v2_states[position] == "scorable")
    n_residual_insufficient: int = sum(1 for position in range(n_total) if error_mask[position] and v2_states[position] == "insufficient")
    n_newly_unsupported: int = sum(1 for position in range(n_total) if error_mask[position] and v2_states[position] == "unsupported")
    # The v2 totals.
    n_scorable: int = sum(1 for state in v2_states if state == "scorable")
    n_insufficient: int = sum(1 for state in v2_states if state == "insufficient")
    n_unsupported: int = sum(1 for state in v2_states if state == "unsupported")
    # The four closures, exactly as pinned by the step text.
    closure_1_total_holds: bool = n_total == n_unchanged + n_v1_errors
    closure_2_unchanged_holds: bool = n_unchanged == n_unchanged_scorable + n_unchanged_insufficient + n_unchanged_unsupported
    closure_3_scorable_holds: bool = n_unchanged_scorable + n_newly_scorable == n_scorable
    closure_4_insufficient_holds: bool = n_unchanged_insufficient + n_residual_insufficient == n_insufficient
    # The all-four verdict.
    all_closures_hold: bool = closure_1_total_holds and closure_2_unchanged_holds and closure_3_scorable_holds and closure_4_insufficient_holds
    # Extra audit identity for the unsupported bucket (reported, not folded into the four).
    n_unsupported_conservation_holds: bool = n_unchanged_unsupported + n_newly_unsupported == n_unsupported
    # Assemble and return every operand and verdict as Python ints and bools.
    return {
        # The four closure verdicts (P05.arithmetic_closures_required).
        "closure_1_total_holds": bool(closure_1_total_holds),
        "closure_2_unchanged_holds": bool(closure_2_unchanged_holds),
        "closure_3_scorable_holds": bool(closure_3_scorable_holds),
        "closure_4_insufficient_holds": bool(closure_4_insufficient_holds),
        # The all-four verdict.
        "all_closures_hold": bool(all_closures_hold),
        # Closure 1's operands.
        "n_total": int(n_total),
        "n_unchanged": int(n_unchanged),
        "n_v1_errors": int(n_v1_errors),
        # Closure 2's operands.
        "n_unchanged_scorable": int(n_unchanged_scorable),
        "n_unchanged_insufficient": int(n_unchanged_insufficient),
        "n_unchanged_unsupported": int(n_unchanged_unsupported),
        # Closure 3's operands.
        "n_newly_scorable": int(n_newly_scorable),
        # Closure 4's operands.
        "n_residual_insufficient": int(n_residual_insufficient),
        # The third bucket of the v1-errored class (reported for exactness).
        "n_newly_unsupported": int(n_newly_unsupported),
        # The v2 totals.
        "n_scorable": int(n_scorable),
        "n_insufficient": int(n_insufficient),
        "n_unsupported": int(n_unsupported),
        # Extra audit identity for the unsupported bucket (informational).
        "n_unsupported_conservation_holds": bool(n_unsupported_conservation_holds),
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Dependency handoff:  bind_dependency_outputs_task_13 — structural bind of the declared dependency orchestrators' outputs.
# -------------------------------------------------------------------------------------------------------------------------------

def bind_dependency_outputs_task_13(task_1_output: Any, task_7_output: Any) -> Dict[str, Any]:
    """
    Verify that the declared dependency outputs bind cleanly.

    Structural bind, never a re-derivation. Task 1's orchestrator
    follows a halt-reason-or-None contract, so ``None`` binds and any
    other value (in particular a halt-reason string) rejects. Task 7's
    orchestrator must return a Mapping with the five pinned keys, with
    ``registry`` a non-empty Mapping, ``terminal_registry`` a Mapping,
    ``registry_sha256`` a 64-character lowercase-hex string and
    ``parse_table`` a ``pandas.DataFrame``.

    Parameters
    ----------
    task_1_output : Any
        Output of Task 1's orchestrator. Accepted iff it is ``None``.
    task_7_output : Any
        Output of Task 7's orchestrator. Accepted iff it is a Mapping
        with keys ``"registry"``, ``"terminal_registry"``,
        ``"registry_sha256"``, ``"parse_table"`` and
        ``"task1_preflight"`` of the shapes described above.

    Returns
    -------
    Dict[str, Any]
        ``task_1_ok`` (bool), ``task_7_ok`` (bool), ``accepted`` (bool,
        the conjunction of the two) and the audit strings
        ``task_1_detail`` / ``task_7_detail`` explaining the verdict.

    Notes
    -----
    A rejected bind is never an exception; it is reported through
    ``accepted == False``.
    """
    # Task 1: accepted exactly when the orchestrator reports no halt (None).
    if task_1_output is None:
        task_1_ok: bool = True
        task_1_detail: str = "task_1 halt-reason-or-None contract honoured (None: no halt outstanding)"
    else:
        task_1_ok = False
        task_1_detail = (
            f"task_1 output is a halt reason or an unexpected value ({type(task_1_output).__name__}: {task_1_output!r}); "
            "a halted Task 1 does not bind"
        )
    # Task 7: accepted exactly when the five-key summary is present and well-formed.
    task_7_ok: bool
    task_7_detail: str
    if not isinstance(task_7_output, _abc.Mapping):
        task_7_ok = False
        task_7_detail = f"task_7 output is not a Mapping (got {type(task_7_output).__name__}); the orchestrator summary is missing"
    else:
        # The pinned five keys must all be present.
        pinned: Tuple[str, str, str, str, str] = ("registry", "terminal_registry", "registry_sha256", "parse_table", "task1_preflight")
        missing_keys: List[str] = [key for key in pinned if key not in task_7_output]
        # Structural checks on the pinned keys.
        registry_value: Any = task_7_output.get("registry", None)
        registry_ok: bool = isinstance(registry_value, _abc.Mapping) and len(registry_value) >= 1
        terminal_ok: bool = isinstance(task_7_output.get("terminal_registry", None), _abc.Mapping)
        sha_value: Any = task_7_output.get("registry_sha256", None)
        sha_ok: bool = isinstance(sha_value, str) and bool(_HEX64_PATTERN.fullmatch(sha_value))
        parse_ok: bool = isinstance(task_7_output.get("parse_table", None), pd.DataFrame)
        if missing_keys:
            task_7_ok = False
            task_7_detail = f"task_7 output is missing pinned keys: {missing_keys}"
        elif not registry_ok:
            task_7_ok = False
            task_7_detail = "task_7 output 'registry' is not a non-empty Mapping"
        elif not terminal_ok:
            task_7_ok = False
            task_7_detail = "task_7 output 'terminal_registry' is not a Mapping"
        elif not sha_ok:
            task_7_ok = False
            task_7_detail = "task_7 output 'registry_sha256' is not a 64-character lowercase-hex string"
        elif not parse_ok:
            task_7_ok = False
            task_7_detail = "task_7 output 'parse_table' is not a pandas.DataFrame"
        else:
            task_7_ok = True
            task_7_detail = "task_7 five-key orchestrator summary present and well-formed"
    # The handoff is accepted only when both dependencies bind.
    accepted: bool = task_1_ok and task_7_ok
    # Assemble and return the verdicts and the details.
    return {
        # Whether Task 1 bound (halt-reason-or-None honoured).
        "task_1_ok": bool(task_1_ok),
        # Whether Task 7 bound (five-key summary well-formed).
        "task_7_ok": bool(task_7_ok),
        # The conjunction; the orchestrator and the console gate consume this.
        "accepted": bool(accepted),
        # Human-readable binding details for the audit trail.
        "task_1_detail": task_1_detail,
        "task_7_detail": task_7_detail,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Orchestrator support:  the declared reconstruction fixture.
# -------------------------------------------------------------------------------------------------------------------------------

def _reconstruction_results_frames_task_13() -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Build the declared reconstruction v1/v2 results frames.

    Produce the deterministic full-scale fixture used when no private
    artifacts are available: 867 unchanged scorable, 98 unchanged
    insufficient and 25 unchanged unsupported records (990 unchanged),
    plus 57 v1 errors of which 56 recover to scorable and 1 stays
    insufficient under v2. Valid-date counts respect the P03 bands, and
    the identifiers run ``e0000`` to ``e1046`` in cohort order.

    Returns
    -------
    v1_results : pd.DataFrame
        Columns ``expression_id`` (str), ``state`` (four-state
        vocabulary) and ``n_valid_dates`` (``Int64``).
    v2_results : pd.DataFrame
        Columns ``expression_id`` (str), ``state`` (three-state
        vocabulary) and ``n_valid_dates`` (``Int64``), in the same row
        order.
    """
    # The reconstruction cohort identifiers, e0000..e1046, in rank order.
    ids: List[str] = [f"e{i:04d}" for i in range(_RECONSTRUCTION_COHORT_SIZE)]
    # The v1 states: 867 scorable, 98 insufficient, 25 unsupported, 57 unsupported_error.
    v1_states: List[str] = (
        ["scorable"] * _RECONSTRUCTION_UNCHANGED_SCORABLE
        + ["insufficient"] * _RECONSTRUCTION_UNCHANGED_INSUFFICIENT
        + ["unsupported"] * _RECONSTRUCTION_UNCHANGED_UNSUPPORTED
        + ["unsupported_error"] * _RECONSTRUCTION_V1_ERRORS
    )
    # The v1 valid-date counts, exactly the planted arithmetic (P03 bands respected).
    v1_valid: List[Optional[int]] = [
        *[20 + (i * 37) % 461 for i in range(_RECONSTRUCTION_UNCHANGED_SCORABLE)],
        *[(i * 13) % 20 for i in range(_RECONSTRUCTION_UNCHANGED_SCORABLE, _RECONSTRUCTION_UNCHANGED_SCORABLE + _RECONSTRUCTION_UNCHANGED_INSUFFICIENT)],
        *[0] * _RECONSTRUCTION_UNCHANGED_UNSUPPORTED,
        *[0] * _RECONSTRUCTION_V1_ERRORS,
    ]
    # The v2 states: unchanged records keep their states; the 57 errors recover as 56 scorable + 1 insufficient.
    v2_states: List[str] = (
        ["scorable"] * _RECONSTRUCTION_UNCHANGED_SCORABLE
        + ["insufficient"] * _RECONSTRUCTION_UNCHANGED_INSUFFICIENT
        + ["unsupported"] * _RECONSTRUCTION_UNCHANGED_UNSUPPORTED
        + ["scorable"] * _RECONSTRUCTION_NEWLY_SCORABLE
        + ["insufficient"] * _RECONSTRUCTION_RESIDUAL_INSUFFICIENT
    )
    # The v2 valid-date counts.
    v2_valid: List[Optional[int]] = [
        *[20 + (i * 37) % 461 for i in range(_RECONSTRUCTION_UNCHANGED_SCORABLE)],
        *[(i * 13) % 20 for i in range(_RECONSTRUCTION_UNCHANGED_SCORABLE, _RECONSTRUCTION_UNCHANGED_SCORABLE + _RECONSTRUCTION_UNCHANGED_INSUFFICIENT)],
        *[0] * _RECONSTRUCTION_UNCHANGED_UNSUPPORTED,
        *[20 + (i * 17) % 461 for i in range(990, 990 + _RECONSTRUCTION_NEWLY_SCORABLE)],
        7,
    ]
    # Assemble the v1 results frame (columns expression_id, state, n_valid_dates).
    v1_results: pd.DataFrame = pd.DataFrame({"expression_id": ids, "state": v1_states, "n_valid_dates": pd.array(v1_valid, dtype="Int64")})
    # Assemble the v2 results frame.
    v2_results: pd.DataFrame = pd.DataFrame({"expression_id": ids, "state": v2_states, "n_valid_dates": pd.array(v2_valid, dtype="Int64")})
    # Return both frames.
    return v1_results, v2_results

def _reconstruction_scores_frames_task_13(unchanged_ids: List[str]) -> Tuple[pd.DataFrame, pd.DataFrame]:
    """
    Build deterministic v1/v2 mid-rank score registers.

    For every unchanged record and every one of the 480 label-available
    business days starting 2024-01-02, emit a deterministic five-element
    mid-rank panel, a ``rho`` value and a ``mean_rank_ic`` value, all
    derived from closed-form functions of the record and date indices.
    The v2 register is a deep copy of the v1 register, which makes the
    invariance that the v2 correction must preserve hold by
    construction.

    Parameters
    ----------
    unchanged_ids : List[str]
        The unchanged record identifiers, in cohort order.

    Returns
    -------
    v1_scores : pd.DataFrame
        Columns ``expression_id``, ``trade_date``, ``mid_rank_panel``,
        ``rho`` and ``mean_rank_ic``; one row per record-date pair.
    v2_scores : pd.DataFrame
        The deep copy of ``v1_scores`` with identical values.
    """
    # The label-available calendar: 480 business days from 2024-01-02 (inside the 2024-2025 label window).
    dates: pd.DatetimeIndex = pd.bdate_range("2024-01-02", periods=_RECONSTRUCTION_LABEL_AVAILABLE_DATES)
    # The total row count (990 unchanged x 480 dates = 475,200 rows).
    row_count: int = len(unchanged_ids) * len(dates)
    # Preallocate the per-row columns.
    expression_column: np.ndarray = np.repeat(np.array(unchanged_ids, dtype=object), len(dates))
    date_column: np.ndarray = np.tile(dates.to_numpy(dtype="datetime64[ns]"), len(unchanged_ids))
    rho_column: np.ndarray = np.empty(row_count, dtype=np.float64)
    mric_column: np.ndarray = np.empty(row_count, dtype=np.float64)
    panel_column: np.ndarray = np.empty(row_count, dtype=object)
    # Build the deterministic values row by row (vectorised per field where possible).
    positions: np.ndarray = np.arange(row_count, dtype=np.int64)
    # i = row//480 (record index), j = row%480 (date index).
    i_index: np.ndarray = positions // len(dates)
    j_index: np.ndarray = positions % len(dates)
    # rho = round((((i*7 + j*13) % 100) - 50)/500, 6); mean_rank_ic = round(((i*29 + 7) % 200 - 100)/1000, 6).
    rho_column[:] = np.round((((i_index * 7 + j_index * 13) % 100) - 50) / 500.0, 6)
    mric_column[:] = np.round((((i_index * 29 + 7) % 200) - 100) / 1000.0, 6)
    # Mid-rank panels: five deterministic floats per cell.
    for position in range(row_count):
        i_value: int = int(i_index[position])
        j_value: int = int(j_index[position])
        panel_column[position] = tuple(float(((i_value * 11 + j_value * 5 + k * 7) % 23) + 1) for k in range(5))
    # Assemble the v1 frame.
    v1_scores: pd.DataFrame = pd.DataFrame(
        {
            "expression_id": pd.Series(expression_column, dtype="object"),
            "trade_date": pd.Series(date_column, dtype="datetime64[ns]"),
            "mid_rank_panel": pd.Series(panel_column, dtype="object"),
            "rho": rho_column,
            "mean_rank_ic": mric_column,
        }
    )
    # The v2 register is identical on unchanged records by construction (the invariant).
    v2_scores: pd.DataFrame = v1_scores.copy(deep=True)
    # Return both registers.
    return v1_scores, v2_scores

def _format_panel_cell_task_13(panel: Any) -> str:
    """
    Render a mid-rank panel cell as a pipe-separated literal.

    Each element is rendered with ``repr(float(value))`` so the CSV cell
    round-trips at full precision.

    Parameters
    ----------
    panel : Any
        The panel cell; any iterable of numbers (tuple, list, ndarray).

    Returns
    -------
    str
        The pipe-separated rendering, e.g. ``"1.0|2.0|3.0"``.
    """
    # Render each float with full round-trip precision, joined by pipes.
    return "|".join(repr(float(value)) for value in list(panel))

def _write_reconstruction_v1_package_task_13(package_dir: Path, ids: List[str], v1_results: pd.DataFrame, v1_scores: pd.DataFrame) -> None:
    """
    Write the reconstruction v1 package to disk.

    Create ``<package_dir>/v1`` and write five files:
    ``v1_evaluator.py`` (the reconstructed defecting evaluator),
    ``DAILY_SCORES_V1.csv`` (the v1 score register with dates and panels
    serialised as text), ``FACTOR_AGGREGATES_V1.csv`` (one row per
    cohort record, NaN mean for the errored class), ``LEDGER_V1.csv``
    (one row per cohort record) and ``ERROR_LOG_V1.csv`` (one row per
    record v1 could not evaluate, with ``error_type``, ``error_message``
    and a SHA-256 ``stack_trace_hash``).

    Parameters
    ----------
    package_dir : Path
        The package root to create beneath the caller's scratch
        directory. It must not already exist.
    ids : List[str]
        The full cohort identifiers, in cohort order.
    v1_results : pd.DataFrame
        The v1 results frame with columns ``expression_id``, ``state``
        and ``n_valid_dates``.
    v1_scores : pd.DataFrame
        The v1 score register; it covers the records v1 could evaluate.

    Raises
    ------
    FileExistsError
        If ``<package_dir>/v1`` already exists (the package is
        write-once).
    """
    # Create the v1 package root beneath the caller's scratch directory.
    v1_root: Path = package_dir / "v1"
    v1_root.mkdir(parents=True, exist_ok=False)
    # Write the reconstructed v1 evaluator with the defect preserved textually and mechanically.
    (v1_root / "v1_evaluator.py").write_text(
        "# Reconstructed v1 evaluator (RECONSTRUCTION_FIXTURE — not the proprietary evaluator).\n"
        "# Preserves the P05.version_1_defect: the conditional reaches bool(cond_panel).\n"
        "def v1_conditional(cond, a, b):\n"
        "    panel = getattr(cond, 'ndim', 0) >= 1\n"
        "    size = getattr(cond, 'size', 1)\n"
        "    if panel and size > 1:\n"
        "        raise ValueError(f'ambiguous: bool(cond_panel) is not defined for a panel with {size} cells')\n"
        "    return a if bool(cond) else b\n",
        encoding="utf-8",
    )
    # Write DAILY_SCORES_V1.csv (the v1 register; unchanged records only).
    export_scores: pd.DataFrame = v1_scores.copy()
    export_scores["trade_date"] = pd.to_datetime(export_scores["trade_date"]).dt.date.astype(str)
    export_scores["mid_rank_panel"] = [_format_panel_cell_task_13(panel) for panel in export_scores["mid_rank_panel"]]
    export_scores.to_csv(v1_root / "DAILY_SCORES_V1.csv", index=False)
    # Write FACTOR_AGGREGATES_V1.csv (one row per cohort record; the errored class carries no value).
    aggregates: pd.DataFrame = pd.DataFrame(
        {
            "expression_id": list(ids),
            "mean_rank_ic": [round(((i * 29 + 7) % 200 - 100) / 1000.0, 6) if state != "unsupported_error" else np.nan for i, state in enumerate(v1_results["state"])],
            "n_valid_dates": v1_results["n_valid_dates"],
        }
    )
    aggregates.to_csv(v1_root / "FACTOR_AGGREGATES_V1.csv", index=False)
    # Write LEDGER_V1.csv (one row per cohort record).
    ledger_export: pd.DataFrame = pd.DataFrame(
        {"expression_id": list(ids), "v1_state": list(v1_results["state"]), "n_valid_dates": v1_results["n_valid_dates"]}
    )
    ledger_export.to_csv(v1_root / "LEDGER_V1.csv", index=False)
    # Write ERROR_LOG_V1.csv (one row per record v1 could not evaluate).
    message: str = "conditional operator reached bool(cond_panel) with a scalar true-branch"
    errored: pd.DataFrame = v1_results[v1_results["state"] == "unsupported_error"]
    error_rows: pd.DataFrame = pd.DataFrame(
        {
            "expression_id": list(errored["expression_id"]),
            "error_type": ["NUMERICAL_ERROR"] * len(errored),
            "error_message": [message] * len(errored),
            "stack_trace_hash": [hashlib.sha256(f"{eid}|{message}".encode("utf-8")).hexdigest() for eid in errored["expression_id"]],
        }
    )
    error_rows.to_csv(v1_root / "ERROR_LOG_V1.csv", index=False)

def _write_reconstruction_v2_package_task_13(package_dir: Path) -> None:
    """
    Write the reconstruction v2 package to disk.

    Create ``<package_dir>/v2`` and write two files: ``v2_evaluator.py``
    (the corrected evaluator that broadcasts a scalar branch to the
    condition panel's shape and propagates NA) and ``REGISTRY_V2.json``
    (the registry tag, the changed operator, the change description, the
    semantics constraint and the verification mode).

    Parameters
    ----------
    package_dir : Path
        The package root to create beneath the caller's scratch
        directory. It must not already exist.

    Raises
    ------
    FileExistsError
        If ``<package_dir>/v2`` already exists (the package is
        write-once).
    """
    # Create the v2 package root beneath the caller's scratch directory.
    v2_root: Path = package_dir / "v2"
    v2_root.mkdir(parents=True, exist_ok=False)
    # Write the reconstructed v2 evaluator with the scalar-broadcast correction.
    (v2_root / "v2_evaluator.py").write_text(
        "# Reconstructed v2 evaluator (RECONSTRUCTION_FIXTURE — not the proprietary evaluator).\n"
        "# Implements the P05.version_2_change 'scalar broadcasting' repair.\n"
        "import numpy as np\n\n"
        "def apply_v2_conditional(cond, a, b):\n"
        "    na_mask = np.isnan(cond) if cond.dtype != np.bool_ else np.zeros(cond.shape, dtype=bool)\n"
        "    condition = cond.astype(np.float64)\n"
        "    true_mask = (condition != 0.0) & ~na_mask\n"
        "    on_true = np.full(condition.shape, float(a)) if np.isscalar(a) else np.broadcast_to(np.asarray(a, dtype=np.float64), condition.shape)\n"
        "    on_false = np.full(condition.shape, float(b)) if np.isscalar(b) else np.broadcast_to(np.asarray(b, dtype=np.float64), condition.shape)\n"
        "    out = np.where(true_mask, on_true, on_false)\n"
        "    return np.where(na_mask, np.nan, out)\n",
        encoding="utf-8",
    )
    # Write the tagged registry descriptor.
    (v2_root / "REGISTRY_V2.json").write_text(
        json.dumps(
            {
                "registry_tag": "registry_v2",
                "changed_operator": "conditional",
                "change": "scalar broadcasting",
                "semantics_constraint": "PRESERVE_EXISTING_ELEMENTWISE_CONDITIONAL_SEMANTICS",
                "verification_mode": "DIRECT_COMPARISON_NOT_ASSUMPTION",
            },
            indent=2,
            sort_keys=True,
        )
        + "\n",
        encoding="utf-8",
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 13, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_13(scratch_root: str, task_1_output: Any = None, task_7_output: Any = None) -> Dict[str, Any]:
    """
    Run the v1 archive, the v2 correction and the full re-evaluation.

    Execute Task 13 end to end beneath a caller-supplied scratch
    directory: bind the declared dependency outputs, build the declared
    deterministic reconstruction fixture, archive the v1 package (Step
    1), verify the registry-level v1 to v2 conditional diff and the
    v1/v2 identity invariant (the elementwise conditional is only
    exercised once on a smoke panel and its output shape recorded), then
    archive the v2 package (Step 2), and finally build the full-cohort
    re-evaluation ledger and its four closures (Step 3).

    The stages run in this order:

    1. Validate ``scratch_root`` and create it with parents.
    2. Bind ``task_1_output`` and ``task_7_output``; a rejected handoff
       returns ``halted == True`` before any artifact is written.
    3. Build the fixture without random draws: identifiers ``e0000``
       upward to ``_RECONSTRUCTION_COHORT_SIZE`` split into 867/98/25
       unchanged states plus 57 v1 errors, of which 56 recover to
       scorable and 1 stays insufficient. The unchanged identifiers
       are those whose v1 state is not ``"unsupported_error"``.
    4. Write ``reconstruction_v1`` and archive it via
       ``archive_v1_package_task_13``, then summarise its register
       from ``v1/ERROR_LOG_V1.csv`` read back from disk.
    5. Build ``registry_v1`` and ``registry_v2`` from one shared set of
       operator objects plus one differing ``conditional`` entry and
       check the diff; if the diff is not conditional-only the run
       returns ``halted == True`` with ``halt_reason``
       ``"REGISTRY_DIFF_NOT_CONDITIONAL_ONLY: differing=..."`` before
       the smoke exercise and before ``reconstruction_v2`` is created.
       Otherwise exercise ``apply_v2_conditional_task_13`` once on a
       2-by-3 smoke panel and record the output shape.
    6. Verify ``verify_unchanged_records_task_13`` on the unchanged
       identifiers, write ``reconstruction_v2`` and commit it via
       ``build_v2_archive_commit_task_13``.
    7. Build the ledger, compute the closures, and persist
       ``RE_EVALUATION_LEDGER.csv`` and ``DAILY_SCORES_V2.csv`` (panel
       cells flattened to text, dates as ISO strings) directly under
       ``scratch_root``.

    Every produced artifact is stamped as a declared reconstruction
    fixture; the paper's counters are comparison values only, never
    returned as observed values. The defect-selection consequence is
    deliberately left to Task 15.

    Parameters
    ----------
    scratch_root : str
        Writable directory that receives every artifact of the run;
        created with parents if absent. Must be a non-blank ``str``
        path; the code does not check that it lies outside the project
        tree.
    task_1_output : Any, optional
        Task 1 orchestrator output (halt-reason-or-None). Default is
        ``None``, which is the healthy value.
    task_7_output : Any, optional
        Task 7 orchestrator output (five-key summary dict). Default is
        ``None``.

    Returns
    -------
    Dict[str, Any]
        Keys: ``halted`` (bool); ``halt_reason`` (str or None);
        ``dependency_handoff`` (verdict dict from
        ``bind_dependency_outputs_task_13``); ``stages`` (list of
        executed ``task_13`` callable names, in order);
        ``V1_ARCHIVE_COMMIT`` (dict); ``V1_REGISTER_SUMMARY``
        (four-state partition and error count); ``REGISTRY_DIFF``
        (dict); ``V2_CONDITIONAL_SHAPE`` (list of ints);
        ``INVARIANT_RESULT`` (dict); ``V2_ARCHIVE_COMMIT`` (dict);
        ``RE_EVALUATION_LEDGER_SUMMARY`` (dict with ``n_total``,
        ``n_scorable``, ``n_insufficient``, ``n_unsupported``);
        ``closures`` (the four hold flags plus ``all_closures_hold``);
        ``selection_consequence`` (always ``None``);
        ``artifacts_provenance`` (``"RECONSTRUCTION_FIXTURE"``);
        ``data_gap`` (``"DATA_ABSENT"``); ``reconstruction_notice``
        (disclosure paragraph).

    Raises
    ------
    TypeError
        Internally, if ``scratch_root`` is not a ``str`` (caught).
    ValueError
        Internally, if ``scratch_root`` is blank (caught).
    RuntimeError
        Internally, if a ``reconstruction_v1`` or ``reconstruction_v2``
        namespace already exists (caught).

    Notes
    -----
    Everything after the result skeleton runs in one ``try``; its
    ``except Exception`` turns any exception, including the three
    above, into ``halted == True`` with ``halt_reason`` set to
    ``"<ExceptionClass>: <message>"``. No exception escapes the call.
    """
    # Initialise the result skeleton with every pinned key defaulted (never a KeyError downstream).
    result: Dict[str, Any] = {
        # Whether the run halted at any stage.
        "halted": False,
        # The halt reason when halted; None otherwise.
        "halt_reason": None,
        # The dependency handoff verdicts (filled below).
        "dependency_handoff": {},
        # The task_13 callables this run executed, in order.
        "stages": [],
        # The Step 1 commit (filled on the healthy path).
        "V1_ARCHIVE_COMMIT": {},
        # The Step 2 commit (filled on the healthy path).
        "V2_ARCHIVE_COMMIT": {},
        # The Step 3 ledger summary (filled on the healthy path).
        "RE_EVALUATION_LEDGER_SUMMARY": {},
        # The four closures (filled on the healthy path).
        "closures": {},
        # Task 15 owns variant_defect_selection_consequence; it is pinned to None here.
        "selection_consequence": None,
        # Epistemic status of every artifact produced by this run.
        "artifacts_provenance": "RECONSTRUCTION_FIXTURE",
        # The honest exit: the real per-record registers are not held by the project.
        "data_gap": "DATA_ABSENT",
        # The disclosure paragraph.
        "reconstruction_notice": (
            "The real v1/v2 per-record score registers and the frozen 1,047 per-record rows are private "
            "artifacts (P41) not held by this project. This run exercises the full pipeline on a declared, "
            "deterministic reconstruction fixture partitioned by the P40/P42 arithmetic (867/98/25 unchanged; "
            "57 v1 errors; 56 recover to scorable; 1 remains insufficient). Nothing here is the paper's "
            "numbers; comparison values are never returned as observed values."
        ),
    }
    # Record the completion of each stage by name (callables, not strings, are the machinery).
    stages: List[str] = result["stages"]  # type: ignore[assignment]
    try:
        # Validate the scratch root (a str path; created with parents if absent).
        if not isinstance(scratch_root, str):
            raise TypeError(f"run_task_13: scratch_root must be a str path, got {type(scratch_root).__name__}")
        if not scratch_root.strip():
            raise ValueError("run_task_13: scratch_root must not be blank")
        scratch: Path = Path(scratch_root).expanduser().resolve()
        scratch.mkdir(parents=True, exist_ok=True)

        # ------------------------------------------------------------------
        # Dependency handoff: bind the declared dependencies exactly once.
        # ------------------------------------------------------------------
        handoff: Dict[str, Any] = bind_dependency_outputs_task_13(task_1_output, task_7_output)
        result["dependency_handoff"] = handoff
        stages.append("bind_dependency_outputs_task_13")
        # A rejected handoff halts the pipeline before any artifact is written.
        if not handoff["accepted"]:
            result["halted"] = True
            result["halt_reason"] = (
                "DEPENDENCY_HANDOFF_REJECTED: "
                + (handoff["task_1_detail"] if not handoff["task_1_ok"] else handoff["task_7_detail"])
            )
            return result

        # ------------------------------------------------------------------
        # Build the declared reconstruction fixture (no RNG anywhere).
        # ------------------------------------------------------------------
        ids: List[str] = [f"e{i:04d}" for i in range(_RECONSTRUCTION_COHORT_SIZE)]
        v1_results, v2_results = _reconstruction_results_frames_task_13()
        unchanged_ids: List[str] = [identifier for identifier, state in zip(ids, v1_results["state"]) if state != "unsupported_error"]
        v1_scores, v2_scores = _reconstruction_scores_frames_task_13(unchanged_ids)

        # ------------------------------------------------------------------
        # Step 1: write the reconstruction v1 package, then archive it.
        # ------------------------------------------------------------------
        reconstruction_v1: Path = scratch / "reconstruction_v1"
        if reconstruction_v1.exists():
            raise RuntimeError(f"run_task_13: reconstruction v1 namespace already exists: {reconstruction_v1}")
        reconstruction_v1.mkdir(parents=True, exist_ok=False)
        _write_reconstruction_v1_package_task_13(reconstruction_v1, ids, v1_results, v1_scores)
        v1_commit: Dict[str, Any] = archive_v1_package_task_13(str(reconstruction_v1), str(scratch), ids)
        stages.append("archive_v1_package_task_13")
        result["V1_ARCHIVE_COMMIT"] = v1_commit
        # The v1 register summary (the four-state partition + the error count).
        error_log: pd.DataFrame = pd.read_csv(reconstruction_v1 / "v1" / "ERROR_LOG_V1.csv", dtype={"stack_trace_hash": str})
        ledger_v1_summary_input: pd.DataFrame = pd.DataFrame(
            {"expression_id": list(ids), "v1_state": list(v1_results["state"]), "n_valid_dates": v1_results["n_valid_dates"]}
        )
        v1_summary: Dict[str, Any] = summarise_v1_register_task_13(ledger_v1_summary_input, error_log, ids)
        stages.append("summarise_v1_register_task_13")
        result["V1_REGISTER_SUMMARY"] = v1_summary

        # ------------------------------------------------------------------
        # Step 2: the registry-level v1 -> v2 diff; the v2 package; the invariant.
        # ------------------------------------------------------------------
        # Build the tagged registries: shared operator objects, conditional swapped.
        shared_operators: Dict[str, Any] = {name: object() for name in ("ts_mean", "ts_rank", "delta", "div", "sub", "neg")}
        conditional_v1: Any = object()
        conditional_v2: Any = object()
        registry_v1: Dict[str, Any] = {**shared_operators, "conditional": conditional_v1}
        registry_v2: Dict[str, Any] = {**shared_operators, "conditional": conditional_v2}
        registry_diff: Dict[str, Any] = verify_v2_registry_diff_task_13(registry_v1, registry_v2)
        stages.append("verify_v2_registry_diff_task_13")
        result["REGISTRY_DIFF"] = registry_diff
        # A drifted registry would halt the pipeline (the fix must be a conditional-only diff).
        if not registry_diff["only_conditional_differs"]:
            result["halted"] = True
            result["halt_reason"] = f"REGISTRY_DIFF_NOT_CONDITIONAL_ONLY: differing={registry_diff['differing_operators']}"
            return result
        # Exercise the v2 elementwise conditional once (smoke; the machinery is the same one archived).
        smoke_cond: np.ndarray = np.array([[1.0, 0.0, np.nan], [0.0, 1.0, 0.0]], dtype=np.float64)
        smoke_out: np.ndarray = apply_v2_conditional_task_13(smoke_cond, 7.0, smoke_cond)
        stages.append("apply_v2_conditional_task_13")
        result["V2_CONDITIONAL_SHAPE"] = list(smoke_out.shape)
        # Verify the v1 <-> v2 invariant on the unchanged records (identity check).
        invariant: Dict[str, Any] = verify_unchanged_records_task_13(v1_scores, v2_scores, unchanged_ids, tol_abs=_RHO_TOL_ABS_DEFAULT)
        stages.append("verify_unchanged_records_task_13")
        result["INVARIANT_RESULT"] = invariant

        # ------------------------------------------------------------------
        # Step 2 (continued): write the reconstruction v2 package; archive it.
        # ------------------------------------------------------------------
        reconstruction_v2: Path = scratch / "reconstruction_v2"
        if reconstruction_v2.exists():
            raise RuntimeError(f"run_task_13: reconstruction v2 namespace already exists: {reconstruction_v2}")
        reconstruction_v2.mkdir(parents=True, exist_ok=False)
        _write_reconstruction_v2_package_task_13(reconstruction_v2)
        v2_commit: Dict[str, Any] = build_v2_archive_commit_task_13(str(reconstruction_v2), str(scratch), invariant)
        stages.append("build_v2_archive_commit_task_13")
        result["V2_ARCHIVE_COMMIT"] = v2_commit

        # ------------------------------------------------------------------
        # Step 3: re-evaluate the full cohort under v2; build the ledger; close.
        # ------------------------------------------------------------------
        ledger: pd.DataFrame = build_re_evaluation_ledger_task_13(ids, v1_results, v2_results)
        stages.append("build_re_evaluation_ledger_task_13")
        closures: Dict[str, Any] = compute_re_evaluation_closures_task_13(ledger)
        stages.append("compute_re_evaluation_closures_task_13")
        result["closures"] = closures
        # The denominator conservation summary (scorable + insufficient + unsupported == n_total).
        result["RE_EVALUATION_LEDGER_SUMMARY"] = {
            # The cohort size under v2 (1,047 in the reproduction's cohort).
            "n_total": int(closures["n_total"]),
            # The scorable total under v2.
            "n_scorable": int(closures["n_scorable"]),
            # The insufficient total under v2.
            "n_insufficient": int(closures["n_insufficient"]),
            # The unsupported total under v2.
            "n_unsupported": int(closures["n_unsupported"]),
        }
        # Persist the audit ledger and the v2 scores as artifacts of the run.
        ledger.to_csv(scratch / "RE_EVALUATION_LEDGER.csv", index=False)
        v2_scores_out: pd.DataFrame = v2_scores.copy()
        v2_scores_out["trade_date"] = pd.to_datetime(v2_scores_out["trade_date"]).dt.date.astype(str)
        v2_scores_out["mid_rank_panel"] = [_format_panel_cell_task_13(panel) for panel in v2_scores_out["mid_rank_panel"]]
        v2_scores_out.to_csv(scratch / "DAILY_SCORES_V2.csv", index=False)
        # The run completes halted only on an explicit halt condition above; otherwise it is complete.
        return result
    except Exception as exc:
        # Every failure path halts loudly: the exception class and message are the halt reason.
        result["halted"] = True
        result["halt_reason"] = f"{type(exc).__name__}: {exc}"
        return result
```

```python
# Task 14: Orchestrator Callable: Core Pipeline Composition

# ==============================================================================
# Task 14: Orchestrator Callable: Core Pipeline Composition
# ==============================================================================

#
# Publication-grade core-pipeline composition for the evidence-admissibility
# certification capability. This module implements:
#   Step 1 - the orchestrator boundary: the run_study signature, the pinned
#            StudyResult frozen dataclass, the STAGE_DEPENDENCIES topological
#            closure of the requested subset, and the STAGE_CALLABLES registry.
#   Step 2 - the wiring: ordered dispatch of the thirteen stage callables via
#            the registry, threading of payloads, halt position semantics, and
#            evaluator substitutability (v1 for Task 9, v2 for Task 13).
#   Step 3 - the halt semantics and the terminal closure verification: every
#            P42 arithmetic closure and every P44 raw-layer invariant in the
#            study configuration is evaluated on this run's own outputs, the
#            admissibility_report is stamped, and the immutable StudyResult is
#            frozen exactly once at the terminal edge.
#
# Governing contracts (binding on this module):
#   * detailed_task_list_20261111_144334.md - Task 14, Steps 1-3.
#   * acceptance/acceptance_task_14.py - the locked acceptance contract
#     (SHA-256 cae4698bc7cc5339883494182a6a67274ee087cce15d03a2b557dde3d85bd596);
#     its REQUIRED_INTERFACE pins the fourteen callables by name, and the exact
#     ordered parameter lists of the thirteen stage callables.
#   * latex_context.txt - the methods manuscript; Eq. (1) is consumed verbatim
#     from task_6, never re-implemented here.
#   * study_revised_inputs_20261111_143538.md - STUDY_CONFIG, consumed at run
#     time for every configuration-pinned figure (P01 label_available_rows ==
#     480; P03 min_valid_dates_for_scorability == 20; P04 state_precedence).
#
# Honest-exit register (P41): the private per-record artifacts are not held by
# this project. Every paper figure cited below is a RECONSTRUCTION-tier
# comparison value, never an observed measurement; values that are computed are
# computed from this run's inputs; values that cannot be computed are absent,
# disclosed, and never synthesised. The admissibility_report is the single
# interface through which the study output and Task 16 read this run.


task_1 = SimpleNamespace(
    _add_four_sanctioned_columns=_add_four_sanctioned_columns,
    _session_span_count=_session_span_count,
    run_task_1=run_task_1,
)

task_2 = SimpleNamespace(
    PipelineContextTask2=PipelineContextTask2,
    _BAR_BOOLEAN_COLUMNS=_BAR_BOOLEAN_COLUMNS,
    _BAR_STRING_COLUMNS=_BAR_STRING_COLUMNS,
    _INTEGER_TERMINAL_COLUMNS=_INTEGER_TERMINAL_COLUMNS,
    assert_enum_domains_task_2=assert_enum_domains_task_2,
    assert_stock_codes_task_2=assert_stock_codes_task_2,
    cast_boolean_pyarrow_task_2=cast_boolean_pyarrow_task_2,
    cast_column_task_2=cast_column_task_2,
    cast_integral_int64_task_2=cast_integral_int64_task_2,
    cast_string_pyarrow_task_2=cast_string_pyarrow_task_2,
    cast_trade_date_task_2=cast_trade_date_task_2,
    check_anti_imputation_task_2=check_anti_imputation_task_2,
    enforce_identity_strings_task_2=enforce_identity_strings_task_2,
    make_pipeline_context_task_2=make_pipeline_context_task_2,
    scan_prohibited_sentinels_task_2=scan_prohibited_sentinels_task_2,
)

task_3 = SimpleNamespace(
    _reassert_coherence=_reassert_coherence,
    run_task_3=run_task_3,
)

task_4 = SimpleNamespace(
    run_task_4=run_task_4,
)

task_5 = SimpleNamespace(
    construct_calendar_task_5=construct_calendar_task_5,
    run_task_5=run_task_5,
)

task_6 = SimpleNamespace(
    run_task_6=run_task_6,
)

task_7 = SimpleNamespace(
    Node=Node,
    OperatorSpec=OperatorSpec,
    build_operator_registry_task_7=build_operator_registry_task_7,
    build_terminal_registry_task_7=build_terminal_registry_task_7,
    canonical_bytes_task_7=canonical_bytes_task_7,
    elementwise_if_task_7=elementwise_if_task_7,
    expression_id_task_7=expression_id_task_7,
    parse_expression_task_7=parse_expression_task_7,
    parse_records_task_7=parse_records_task_7,
    registry_sha256_task_7=registry_sha256_task_7,
    v1_emulator_task_7=v1_emulator_task_7,
)

task_8 = SimpleNamespace(
    deduplicate_canonical_task_8=deduplicate_canonical_task_8,
    freeze_cohort_task_8=freeze_cohort_task_8,
    screen_is_only_task_8=screen_is_only_task_8,
)

task_9 = SimpleNamespace(
    run_task_9=run_task_9,
)

task_10 = SimpleNamespace(
    _canonical_json_bytes_task_10=_canonical_json_bytes_task_10,
    _sha256_hex_task_10=_sha256_hex_task_10,
    aggregate_factors_task_10=aggregate_factors_task_10,
    assign_ledger_task_10=assign_ledger_task_10,
    closure_battery_task_10=closure_battery_task_10,
)

task_11 = SimpleNamespace(
    run_task_11=run_task_11,
    score_references_task_11=score_references_task_11,
)

task_12 = SimpleNamespace(
    build_dual_check_results_task_12=build_dual_check_results_task_12,
    select_nine_cells_task_12=select_nine_cells_task_12,
)

task_13 = SimpleNamespace(
    run_task_13=run_task_13,
)


# Module logger; dispatch diagnostics only, never control flow.
_LOGGER: logging.Logger = logging.getLogger(__name__)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 1: Orchestrator boundary - run_study signature, StudyResult, registry and dependency graph
# -------------------------------------------------------------------------------------------------------------------------------

# The pinned dispatch order of the thirteen stages (validate first, correction last).
STAGE_ORDER: Tuple[str, ...] = (
    "validate",
    "cleanse",
    "custody",
    "identity",
    "calendar",
    "labels",
    "dsl",
    "freeze",
    "score",
    "aggregate",
    "reference",
    "dual_check",
    "correction",
)

# The pinned stage dependency graph, verbatim from Task 14 Step 1 / the contract.
STAGE_DEPENDENCIES: Dict[str, set] = {
    "cleanse": {"validate"},
    "custody": {"cleanse"},
    "identity": {"custody"},
    "calendar": {"identity"},
    "labels": {"calendar"},
    "dsl": {"validate"},
    "freeze": {"dsl", "validate"},
    "score": {"labels", "dsl", "freeze"},
    "aggregate": {"score"},
    "reference": {"labels", "dsl"},
    "dual_check": {"score"},
    "correction": {"score", "freeze"},
}

# Sentinel prefix stamped onto every stage halted behind a failure (never on the failing stage itself).
PIPELINE_HALTED_PREFIX: str = "PIPELINE_HALTED_AT_"

# Sentinel for stages the caller did not request.
STAGE_NOT_REQUESTED: str = "STAGE_NOT_REQUESTED"

# Sentinel for stages requested but whose upstream payloads never materialised.
MISSING_UPSTREAM_INPUT: str = "MISSING_UPSTREAM_INPUT"

# Sentinel for the cohort fields when no freeze was produced this run; disclosed as such everywhere.
NOT_FROZEN_COHORT_HASH: str = "NOT_FROZEN__UPSTREAM_HALT"

# Asia/Shanghai timezone handle for freeze-adjacent timestamps (P00_META.config_self_timestamp_timezone).
ASIA_SHANGHAI_TZ: datetime.timezone = datetime.timezone(datetime.timedelta(hours=8), "Asia/Shanghai")

# The [R] absolute tolerance for the dual implementation spot check (Task 14 step 3 / P18 [R] register; 1e-10 absolute).
TOLERANCE_SPOT_CHECK_ABSOLUTE: float = 1e-10

# The unit quantum by which every dispatch counter advances (exactly one per visited/halted stage; never fractional, never compounded).
DISPATCH_COUNTER_QUANTUM: int = 1


# The pinned tier vocabulary of the evidentiary tiers (Evidentiary tiers; the standing register).
_CLAIM_TIER_VOCABULARY: frozenset = frozenset({"replication", "reproduction", "reconstruction", "inadmissible"})
# The four gate identifiers of the admissibility contract (G1..G4), in pinned order.
CLAIM_GATE_KEYS: Tuple[str, str, str, str] = ("G1", "G2", "G3", "G4")
# P42 metric-grain closures composing G1 (Tasks 9, 10, 11; sign/coverage partition grain, label grain).
_G1_P42_KEYS: Tuple[str, ...] = ("P42.sign_partition", "P42.coverage_partition", "P42.label_availability")
# P42 denominator-completeness closures composing G2 (Tasks 10, 13, 16; constituent-wise ledger family).
_G2_P42_KEYS: Tuple[str, ...] = (
    "P42.ledger",
    "P42.ledger_supported",
    "P42.non_scorable",
    "P42.version_closures",
    "P42.sum_closures_rounds_to_totals",
)
# P44 execution-provenance invariants composing G3 (Tasks 1, 3, 4, 16; custody/calendar/label-window family).
_G3_P44_KEYS: Tuple[str, ...] = (
    "P44.file_count_equals_80",
    "P44.rows_per_bar_file_equals_727",
    "P44.calendar_rows_equals_727",
    "P44.calendar_year_counts_242_242_243",
    "P44.label_window_rows_equals_485",
    "P44.terminal_unlabelled_rows_equals_5",
    "P44.label_available_rows_equals_480",
    "P44.duplicate_rows_per_stock_date_equals_0",
    "P44.retrieval_coherence_window_not_exceeded",
)
# P44 pre-outcome immutability invariants composing G4 (Tasks 8, 13; freeze/anti-imputation family).
_G4_P44_KEYS: Tuple[str, ...] = (
    "P44.masked_cell_rows_equals_2",
    "P44.forward_fill_rows_equals_0",
    "P44.stock_prefix_reductions_equals_0",
    "P44.oos_sharpe_non_null_rows_before_freeze_equals_0",
    "P44.imputed_zero_alpha_records_equals_0",
    "P44.unassigned_ledger_records_equals_0",
    "P44.catalog_non_positive_oos_rows_equals_0",
    "P44.sign_disagreements_in_intersection_equals_0",
)


def _gate_status_from_terminal(
    goods: Mapping[str, Any],
    block_verdicts: Mapping[str, bool],
) -> Mapping[str, bool]:
    """
    Compute the four admissibility gates from terminal verdicts.

    G1 is true only when the P42 grain closures associated with
    ``_G1_P42_KEYS`` all hold and the aggregation materialised on this
    run. G2 requires every ``_G2_P42_KEYS`` ledger verdict. G3 requires
    every ``_G3_P44_KEYS`` custody/calendar verdict. G4 requires every
    ``_G4_P44_KEYS`` freeze/anti-imputation verdict. Missing keys are
    treated as ``False`` so every gate fails closed.

    Parameters
    ----------
    goods : Mapping[str, Any]
        Scalar operand mapping. ``goods.get("aggregates_rows")`` is
        tested for presence only: any non-``None`` payload (including an
        empty or zero-row one) counts as the aggregation having
        materialised; ``None`` or a missing key means it did not.
    block_verdicts : Mapping[str, bool]
        Verdicts keyed ``P42.<identity>`` and ``P44.<invariant>``.

    Returns
    -------
    Mapping[str, bool]
        A read-only mapping with exactly the keys ``G1``, ``G2``, ``G3``
        and ``G4``, each a native ``bool``.
    """
    # The constituent verdicts, fail-closed per key.
    def _all_true(keys: Tuple[str, ...]) -> bool:
        """
        Return whether every named verdict is present and true.

        Parameters
        ----------
        keys : Tuple[str, ...]
            Composite keys to look up in ``block_verdicts``.

        Returns
        -------
        bool
            ``True`` only if every key exists and maps to a true value;
            missing keys yield ``False``.
        """
        # Every named key must exist and be True; absence is False.
        return bool(all(bool(block_verdicts.get(key, False)) for key in keys))

    # The aggregation must have materialised on this run for the grain claim to hold.
    aggregates_ran: bool = goods.get("aggregates_rows") is not None
    # Compose the four gates exactly.
    gate_status: Dict[str, bool] = {
        # G1: metric grain (aggregation materialised AND grain closures held).
        "G1": bool(_all_true(_G1_P42_KEYS) and aggregates_ran),
        # G2: denominator completeness (ledger family, constituent-wise).
        "G2": bool(_all_true(_G2_P42_KEYS)),
        # G3: execution provenance (custody/calendar family).
        "G3": bool(_all_true(_G3_P44_KEYS)),
        # G4: pre-outcome immutability (freeze/anti-imputation family).
        "G4": bool(_all_true(_G4_P44_KEYS)),
    }
    # Return the immutable mapping.
    return MappingProxyType(gate_status)

@dataclass(frozen=True)
class ClaimRecord:
    """
    Represent one immutable claim-admissibility record.

    The record carries a claim identifier, evidentiary tier, evidence
    source, the four admissibility gates, verdict, value and note. At
    construction ``__post_init__`` validates the members and replaces
    ``evidence_hash`` with the SHA-256 of the canonical JSON bytes of
    ``(claim_id, tier, evidence_source, verdict, value, note,
    gate_status)``. The digest is therefore a pure function of those
    seven material members.

    Attributes
    ----------
    claim_id : str
        Stable claim identifier.
    tier : str
        Evidentiary tier: ``replication``, ``reproduction``,
        ``reconstruction`` or ``inadmissible``.
    evidence_source : str
        Artifact class supporting the verdict.
    gate_status : Mapping[str, bool]
        Exactly ``G1``..``G4`` in insertion order; values are
        native ``bool`` objects.
    verdict : str
        Verdict literal such as ``CLOSED`` or ``REPORTED``.
    value : Any
        Claim payload; must be JSON-admissible for the digest.
    note : str
        Full evidentiary note.
    evidence_hash : str
        Canonical 64-character lowercase-hex SHA-256, recomputed at
        construction.
    """

    # The stable identifier of the claim (pinned literal vocabularies only).
    claim_id: str
    # The evidentiary tier assigned to the claim (replication | reproduction | reconstruction | inadmissible).
    tier: str
    # The artifact class the verdict rests on.
    evidence_source: str
    # The four admissibility gates (exactly G1..G4; native bools; order preserved).
    gate_status: Mapping[str, bool]
    # The verdict literal.
    verdict: str
    # The value carried by the claim (scalar, mapping or None; JSON-admissible).
    value: Any
    # The full evidentiary note.
    note: str
    # The 64-lowercase-hex SHA-256 of the canonical basis; stamped at construction (never hand-set).
    evidence_hash: str = ""

    def __post_init__(self) -> None:
        """
        Validate the material members and stamp ``evidence_hash``.

        The digest is written with ``object.__setattr__`` because the
        dataclass is frozen. Any caller-supplied ``evidence_hash`` is
        overwritten by the canonical recomputation.

        Raises
        ------
        ValueError
            If an identifier, source or verdict is empty, if ``tier``
            is outside the allowed vocabulary, or if ``gate_status``
            does not carry exactly ``G1``..``G4``.
        TypeError
            If ``note`` is not a string, if ``gate_status`` is not a
            mapping, or if a gate value is not a native ``bool``.
        """
        # claim_id must be a non-empty string.
        if not isinstance(self.claim_id, str) or not self.claim_id.strip():
            # Refuse empty identifiers with the observed value.
            raise ValueError(f"ClaimRecord.claim_id must be a non-empty str, got {self.claim_id!r}")
        # The tier must come from the pinned evidentiary vocabulary.
        if self.tier not in _CLAIM_TIER_VOCABULARY:
            # Refuse unrecognised tiers with the observed value.
            raise ValueError(
                f"ClaimRecord.tier must be one of {sorted(_CLAIM_TIER_VOCABULARY)}, got {self.tier!r}"
            )
        # evidence_source must be a non-empty string.
        if not isinstance(self.evidence_source, str) or not self.evidence_source.strip():
            # Refuse empty sources with the observed value.
            raise ValueError(f"ClaimRecord.evidence_source must be a non-empty str, got {self.evidence_source!r}")
        # verdict must be a non-empty string.
        if not isinstance(self.verdict, str) or not self.verdict.strip():
            # Refuse empty verdicts with the observed value.
            raise ValueError(f"ClaimRecord.verdict must be a non-empty str, got {self.verdict!r}")
        # note must be a string (empty is permitted only for programmatic diagnostics, never here).
        if not isinstance(self.note, str):
            # Refuse non-str notes with the observed value.
            raise TypeError(f"ClaimRecord.note must be a str, got {type(self.note).__name__}")
        # gate_status must be a Mapping with EXACTLY the four pinned keys.
        if not isinstance(self.gate_status, Mapping):
            # Refuse non-mapping gates with the observed type.
            raise TypeError(f"ClaimRecord.gate_status must be a Mapping[str, bool], got {type(self.gate_status).__name__}")
        # Snapshot the mapping in its insertion order.
        keys = list(self.gate_status.keys())
        # The key SET must be exactly {G1..G4}, each once.
        if set(keys) != set(CLAIM_GATE_KEYS) or len(keys) != len(CLAIM_GATE_KEYS):
            # Refuse any other key set with the observed keys.
            raise ValueError(
                f"ClaimRecord.gate_status must carry exactly the four keys G1..G4, got {keys}"
            )
        # Every gate value must be a NATIVE bool (never truthy objects).
        for key, item in self.gate_status.items():
            # bool is a subclass of int; the type check is identity, not truthiness.
            if type(item) is not bool:
                # Refuse non-native bools with the observed value.
                raise TypeError(f"ClaimRecord.gate_status[{key!r}] must be a native bool, got {item!r}")
        # The material basis (everything except the digest itself; the pinned field set).
        basis = {
            "claim_id": self.claim_id,
            "tier": self.tier,
            "evidence_source": self.evidence_source,
            "verdict": self.verdict,
            "value": self.value,
            "note": self.note,
            # The gate mapping is hashed in its insertion order (insertion order is pinned by the emitters).
            "gate_status": dict(self.gate_status),
        }
        # Canonicalise the basis with Task 10's pinned encoder (one site, one stamp).
        digest = task_10._sha256_hex_task_10(task_10._canonical_json_bytes_task_10(basis))
        # The stamp is ALWAYS the canonical recomputation; any caller-supplied evidence_hash is
        # accepted for interface compatibility and always overwritten here (never carried through).
        # A record therefore never carries a stale or caller-fabricated digest: the value on the
        # frozen record is a pure function of the seven material members, stamped at construction.
        # (There is no path by which a mismatched digest survives construction.)
        object.__setattr__(self, "evidence_hash", digest)

@dataclass(frozen=True)
class StudyResult:
    """
    Represent the immutable result of one ``run_study`` invocation.

    The object has the fifteen pinned fields. It is built once at the
    end of a run; the ``Mapping`` and ``Tuple`` field types make nested
    collections immutable and the fields are not mutated afterwards.

    Attributes
    ----------
    stages_output : Mapping[str, Any]
        Requested stage names mapped to their publication payloads;
        values may contain pandas objects by reference.
    stages_halt : Mapping[str, Optional[str]]
        All thirteen stage names mapped to halt reasons, ``None`` on
        success, ``STAGE_NOT_REQUESTED`` for unrequested stages, or a
        ``PIPELINE_HALTED_AT_<stage>`` marker for skipped stages.
    cohort_hash : str
        SHA-256 of the frozen ordered ID list, or the sentinel
        ``NOT_FROZEN__UPSTREAM_HALT`` when freezing never ran.
    cohort_timestamp : datetime.datetime
        Timezone-aware freeze instant, or the disclosed run instant
        when freezing never ran.
    score_register_v1 : pd.DataFrame
        v1 daily-scores register; empty with pinned columns when
        scoring did not run.
    score_register_v2 : pd.DataFrame
        v2 daily-scores register; empty with pinned columns when the
        correction did not run.
    factor_aggregates : pd.DataFrame
        Factor-aggregate register with ledger states; empty with pinned
        columns when aggregation did not run.
    ledger_primary : Tuple[int, int, int]
        ``(n_scorable, n_insufficient, n_unsupported)`` under the
        primary precedence.
    ledger_reverse : Tuple[int, int, int]
        The same triple under the reversed precedence.
    reference_means : pd.DataFrame
        Fixed reference-signal means indexed by ``reference_id``; empty
        with pinned columns when the reference panel did not run.
    dual_check_result : pd.DataFrame
        Dual-implementation spot-check rows; empty with pinned columns
        when the dual check did not run.
    correction_partition : Mapping[str, int]
        Corrective partition counts; empty when correction did not run.
    admissibility_report : Mapping[str, Any]
        The single interface consumed downstream by the study output
        and deployment decision.
    claim_register : Tuple[ClaimRecord, ...]
        Emitted claim records in deterministic order.
    block_verdicts : Mapping[str, bool]
        One verdict per P42 closure and P44 invariant, keyed
        ``P42.<identity>`` and ``P44.<invariant>``.
    """

    # The requested stages' publication payloads (may carry DataFrames/Series by reference).
    stages_output: Mapping[str, Any]
    # All thirteen pinned stage names mapped to their halt states (None on success).
    stages_halt: Mapping[str, Optional[str]]
    # SHA-256 of the frozen ordered ID list, or the disclosed NOT_FROZEN sentinel.
    cohort_hash: str
    # Timezone-aware Asia/Shanghai freeze instant (or the disclosed run-instant fallback).
    cohort_timestamp: datetime.datetime
    # The v1 DAILY_SCORES register.
    score_register_v1: pd.DataFrame
    # The v2 DAILY_SCORES register (scalar-broadcasting correction).
    score_register_v2: pd.DataFrame
    # The FACTOR_AGGREGATES register with the ledger states attached.
    factor_aggregates: pd.DataFrame
    # (n_scorable, n_insufficient, n_unsupported) under the primary precedence.
    ledger_primary: Tuple[int, int, int]
    # (n_scorable, n_insufficient, n_unsupported) under the reversed precedence.
    ledger_reverse: Tuple[int, int, int]
    # The six fixed reference signals' mean rank-ICs, indexed by reference_id.
    reference_means: pd.DataFrame
    # The 9-of-9 dual-implementation spot check rows.
    dual_check_result: pd.DataFrame
    # The corrective re-evaluation partition ints.
    correction_partition: Mapping[str, int]
    # The single interface to the study output and Task 16.
    admissibility_report: Mapping[str, Any]
    # The emitted claim records.
    claim_register: Tuple[ClaimRecord, ...]
    # One boolean verdict per P42 closure and per P44 invariant.
    block_verdicts: Mapping[str, bool]

def check_stage_closure(stages: Sequence[str]) -> List[Tuple[str, frozenset]]:
    """
    Return the topological-closure violations for a stage subset.

    For each requested stage, compares its dependencies in
    ``STAGE_DEPENDENCIES`` with the requested set and records any
    missing dependency. Names outside the pinned stage order are also
    reported as violations with an empty dependency witness. An empty
    return value certifies a known, closed subset.

    Parameters
    ----------
    stages : Sequence[str]
        Requested stage names, in dispatch order.

    Returns
    -------
    List[Tuple[str, frozenset]]
        Sorted ``(stage, missing_dependencies)`` pairs. The second item
        is empty only for an unknown stage name; an empty list means no
        violation.

    Raises
    ------
    TypeError
        If ``stages`` is a string, bytes or ``None``, or if any element
        is not a string.
    """
    # The argument must be an iterable (a set, list or tuple); a bare str is rejected upstream of us.
    if isinstance(stages, (str, bytes)) or stages is None:
        # A bare string would iterate into characters; refuse it loudly.
        raise TypeError(f"stages must be a sequence of stage names, got {type(stages).__name__}")
    # Materialise the set of requested names, preserving the caller's order for dispatch.
    names: List[str] = list(stages)
    # Every requested name must be a plain string.
    for position, name in enumerate(names):
        # Non-string members cannot be registry keys.
        if not isinstance(name, str):
            # Name the position and the observed type.
            raise TypeError(f"stages[{position}] must be a str, got {type(name).__name__}")
    # The set form of the request for the subset predicate.
    requested: set = set(names)
    # Accumulator for the violations.
    violations: List[Tuple[str, frozenset]] = []
    # Reject unknown stage names first (a superset of the pinned check).
    unknown: List[str] = sorted(requested.difference(STAGE_ORDER))
    # Each unknown name is reported with the empty-deficiency marker of the registry.
    for name in unknown:
        # Record the unknown name with an empty dependency witness.
        violations.append((name, frozenset()))
    # Apply the pinned predicate per requested stage.
    for stage in requested:
        # Dependencies of this stage (empty for validate, which has none).
        dependencies: set = STAGE_DEPENDENCIES.get(stage, set())
        # The missing dependency set is the set difference.
        missing: frozenset = frozenset(dependencies.difference(requested))
        # A non-empty missing set is a closure violation.
        if missing:
            # Record the stage and the exact missing set.
            violations.append((stage, missing))
    # Deterministic order for reporting.
    violations.sort(key=lambda item: item[0])
    # Return the violations (empty means closed and known).
    return violations

def topological_closure_reason(violations: Sequence[Tuple[str, frozenset]]) -> str:
    """
    Render a halt reason for an unclosed or unknown stage subset.

    Parameters
    ----------
    violations : Sequence[Tuple[str, frozenset]]
        Violations produced by :func:`check_stage_closure`.

    Returns
    -------
    str
        A single-line reason beginning ``STAGE_GRAPH_NOT_CLOSED::`` and
        naming each offending stage and its missing dependencies.

    Raises
    ------
    ValueError
        If ``violations`` is empty.
    """
    # An empty list cannot be rendered into a reason.
    if not violations:
        # The caller passed a closed subset by mistake.
        raise ValueError("topological_closure_reason requires at least one violation")
    # Render each violation as "stage (missing: a, b)" or "stage (unknown stage)".
    fragments: List[str] = []
    # Walk the violations once.
    for stage, missing in violations:
        # An empty missing set marks an unknown stage name.
        if not missing:
            # Render the unknown-name fragment.
            fragments.append(f"{stage} (stage not in the pinned registry)")
            # Move to the next violation.
            continue
        # Join the missing dependency names deterministically.
        missing_text: str = ", ".join(sorted(missing))
        # Render the closure-violation fragment.
        fragments.append(f"{stage} (missing closure: {missing_text})")
    # Compose the single-line reason.
    reason: str = "STAGE_GRAPH_NOT_CLOSED:: " + "; ".join(fragments)
    # Return the reason.
    return reason

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 2: The wiring - ordered dispatch, payload threading and the thirteen stage callables
# -------------------------------------------------------------------------------------------------------------------------------

class PipelineContext:
    """
    Hold request-scoped mutable state for one pipeline run.

    The context is created per invocation, carries counters, manifests,
    freeze state and cross-stage payloads, and is cleared at the end of
    ``run_study``. Stage callables reach it through the module-level
    context bridge because their pinned signatures do not carry it.

    Attributes
    ----------
    scratch_root : Path
        Writable root for artifacts produced by the run.
    nonce : str
        Per-invocation nonce namespacing artifact directories.
    counters : dict[str, int]
        Dispatch and anti-imputation counters; all zero.
    manifests : dict[str, Any]
        Integrity manifests populated stage by stage.
    freeze_state : dict[str, Any]
        Cohort hash, freeze timestamp and related metadata.
    payloads : dict[str, Any]
        Cross-stage artifacts not carried by pinned signatures.
    task_2_context : Any
        Run-scoped anti-imputation controller created at construction.
    events : list[dict[str, Any]]
        Append-only dispatch ledger.
    """

    def __init__(self, scratch_root: Path, nonce: str) -> None:
        """
        Initialise the request-scoped context.

        Creates the root, initialises zeroed counters and empty
        registries, and creates this context's anti-imputation
        controller.

        Parameters
        ----------
        scratch_root : Path
            Writable root directory; created if missing.
        nonce : str
            Per-invocation nonce; must be non-empty.

        Raises
        ------
        TypeError
            If ``scratch_root`` is not a ``Path`` or ``nonce`` is not a
            non-empty string.
        """
        # The root must be a real path object.
        if not isinstance(scratch_root, Path):
            # Reject non-path roots with the observed type.
            raise TypeError(f"scratch_root must be a pathlib.Path, got {type(scratch_root).__name__}")
        # The nonce must be a non-empty string.
        if not isinstance(nonce, str) or not nonce.strip():
            # Reject blank nonces.
            raise TypeError(f"nonce must be a non-empty str, got {nonce!r}")
        # Create the root with parents (idempotent).
        scratch_root.mkdir(parents=True, exist_ok=True)
        # Store the writable root.
        self.scratch_root: Path = scratch_root
        # Store the nonce.
        self.nonce: str = nonce
        # Anti-imputation and dispatch counters: zero by construction and maintained at zero.
        self.counters: Dict[str, int] = {
            "stages_executed": 0,
            "stages_halted": 0,
            "forward_fill_operations": 0,
            "interpolation_operations": 0,
            "stock_prefix_reductions": 0,
            "imputed_zero_alpha_records": 0,
        }
        # Integrity manifests, filled stage by stage.
        self.manifests: Dict[str, Any] = {}
        # Freeze mirror, filled by the freeze stage.
        self.freeze_state: Dict[str, Any] = {}
        # Cross-stage payloads, filled stage by stage.
        self.payloads: Dict[str, Any] = {}
        # The anti-imputation controller of THIS context (Task 2's PipelineContextTask2).
        # Created exactly once per context, at construction; the run-scoped context's
        # controller is THE controller of the run (every gated operation on this run
        # would have written to it); the ephemeral fallback context for stand-alone
        # stage callers gets its own, whose witnesses are likewise this call's.
        self.task_2_context: Any = task_2.make_pipeline_context_task_2()
        # Append-only dispatch ledger.
        self.events: List[Dict[str, Any]] = []

    def stage_dir(self, stage: str) -> Path:
        """
        Return and create the nonce-namespaced directory for a stage.

        Parameters
        ----------
        stage : str
            Pinned stage name.

        Returns
        -------
        Path
            ``scratch_root / "task_14" / f"{stage}_{nonce}"``, created
            on demand.

        Raises
        ------
        TypeError
            If ``stage`` is not a non-empty string.
        """
        # The stage name must be a string.
        if not isinstance(stage, str) or not stage.strip():
            # Reject invalid stage names.
            raise TypeError(f"stage must be a non-empty str, got {stage!r}")
        # Compose the nonce-namespaced directory.
        directory: Path = self.scratch_root / "task_14" / f"{stage}_{self.nonce}"
        # Create it with parents.
        directory.mkdir(parents=True, exist_ok=True)
        # Return the directory.
        return directory

# The module-level request-scoped bridge; set by run_study, cleared in its finally block.
_RUN_CONTEXT: Optional[PipelineContext] = None

# The lazily-created ephemeral context for direct callers outside run_study (never shared across calls).
_EPHEMERAL_CONTEXT: Optional[PipelineContext] = None


def _set_run_context(context: Optional[PipelineContext]) -> None:
    """
    Seat or clear the module-level request-scoped context.

    Parameters
    ----------
    context : Optional[PipelineContext]
        Context to seat, or ``None`` to clear the bridge.

    Raises
    ------
    TypeError
        If ``context`` is neither a ``PipelineContext`` nor ``None``.
    """
    # The bridge may only be seated with a context or cleared with None.
    if context is not None and not isinstance(context, PipelineContext):
        # Reject foreign objects.
        raise TypeError(f"_set_run_context expects PipelineContext or None, got {type(context).__name__}")
    # The global is declared here explicitly; only run_study and its finally block call this.
    global _RUN_CONTEXT
    # Seat or clear the bridge.
    _RUN_CONTEXT = context

def _current_context() -> PipelineContext:
    """
    Return the active pipeline context.

    Returns the seated request-scoped context when present. Otherwise
    creates, once per module lifetime, an ephemeral context under a
    fresh temporary directory so standalone stage callers never write
    into the project tree.

    Returns
    -------
    PipelineContext
        The active context.
    """
    # The bridge wins when seated.
    if _RUN_CONTEXT is not None:
        # Return the request-scoped context.
        return _RUN_CONTEXT
    # Declare the ephemeral cache as global.
    global _EPHEMERAL_CONTEXT
    # Create the ephemeral context on first use.
    if _EPHEMERAL_CONTEXT is None:
        # Mount a fresh temporary root (outside the project tree).
        ephemeral_root: Path = Path(tempfile.mkdtemp(prefix="task14_ephemeral_"))
        # A fixed nonce keeps the ephemeral namespacing readable.
        _EPHEMERAL_CONTEXT = PipelineContext(ephemeral_root, uuid.uuid4().hex[:12])
    # Return the ephemeral context.
    return _EPHEMERAL_CONTEXT

@dataclass(frozen=True)
class PanelEvaluator:
    """
    Represent the versioned evaluator handle consumed by the scorer.

    Both versions share the time-series and arithmetic operator
    implementations; only the conditional operator wiring changes. The
    object is an immutable handle.

    Attributes
    ----------
    version : str
        Version literal, ``v1`` or ``v2``.
    version_tag : str
        Tag written verbatim into every daily-scores row.
    conditional_operator : str
        Pinned name of the conditional operator.
    conditional_implementation : str
        Disclosure of the conditional wiring.
    registry : Mapping[str, Any]
        Immutable operator registry.
    registry_sha256 : str
        Lowercase 64-character SHA-256 of the registry.
    """

    # The version literal (v1 | v2).
    version: str
    # The tag recorded verbatim in every register row.
    version_tag: str
    # The pinned name of the conditional operator.
    conditional_operator: str
    # The disclosure of the conditional wiring.
    conditional_implementation: str
    # The immutable operator registry.
    registry: Mapping[str, Any]
    # The registry's SHA-256 stamp.
    registry_sha256: str

def _empty_daily_scores() -> pd.DataFrame:
    """
    Build the empty DAILY_SCORES frame with pinned index and columns.

    Returns
    -------
    pd.DataFrame
        Zero rows with a ``MultiIndex`` named ``(expression_id,
        trade_date)`` and the seven daily-score columns: ``rho``,
        ``n_eligible``, ``n_distinct_signal``, ``n_distinct_label``,
        ``admissible``, ``reject_reason``, ``evaluator_version``.
    """
    # An empty compound index carrying the pinned names.
    empty_index: pd.MultiIndex = pd.MultiIndex.from_arrays(
        [pd.Series([], dtype="object"), pd.Series([], dtype="datetime64[ns]")],
        names=["expression_id", "trade_date"],
    )
    # Assemble the empty frame with the pinned columns.
    frame: pd.DataFrame = pd.DataFrame(
        {
            # The per-date association measure (NaN on inadmissible cells).
            "rho": pd.Series([], dtype="float64"),
            # The eligible-stock count per cell.
            "n_eligible": pd.Series([], dtype="Int64"),
            # The distinct signal values within the eligible set.
            "n_distinct_signal": pd.Series([], dtype="Int64"),
            # The distinct label values within the eligible set.
            "n_distinct_label": pd.Series([], dtype="Int64"),
            # The joint admissibility verdict.
            "admissible": pd.Series([], dtype="bool"),
            # The typed rejection reason (None iff admissible).
            "reject_reason": pd.Series([], dtype="object"),
            # The evaluator version tag.
            "evaluator_version": pd.Series([], dtype="object"),
        },
        index=empty_index,
    )
    # Return the empty frame.
    return frame

def _empty_factor_aggregates() -> pd.DataFrame:
    """
    Build the empty FACTOR_AGGREGATES frame with the pinned columns.

    Returns
    -------
    pd.DataFrame
        Zero rows with columns ``expression_id``, ``n_valid_dates``,
        ``n_valid_dates_reason``, ``mean_rank_ic``, ``coverage_ratio``,
        ``mean_sd_ratio_unnannualized_not_sharpe``, ``state_primary``
        and ``state_reverse``.
    """
    # Assemble the empty frame with the pinned columns and dtypes.
    frame: pd.DataFrame = pd.DataFrame(
        {
            # The frozen cohort identifier.
            "expression_id": pd.Series([], dtype="object"),
            # The count of valid dates.
            "n_valid_dates": pd.Series([], dtype="Int64"),
            # The construction reason for the count.
            "n_valid_dates_reason": pd.Series([], dtype="object"),
            # The arithmetic mean signed rank-IC over T_f.
            "mean_rank_ic": pd.Series([], dtype="float64"),
            # The coverage diagnostic (n_valid_dates / full_coverage_dates).
            "coverage_ratio": pd.Series([], dtype="float64"),
            # The disclosed, unannualised mean/SD diagnostic (ddof=1; not a Sharpe).
            "mean_sd_ratio_unnannualized_not_sharpe": pd.Series([], dtype="float64"),
            # The ledger state under the primary precedence.
            "state_primary": pd.Series([], dtype="object"),
            # The ledger state under the reversed precedence.
            "state_reverse": pd.Series([], dtype="object"),
        }
    )
    # Return the empty frame.
    return frame

def _empty_reference_means() -> pd.DataFrame:
    """
    Build the empty REFERENCE_MEANS frame with the pinned columns.

    Returns
    -------
    pd.DataFrame
        Zero rows; index ``reference_id``; columns ``mean_rank_ic`` and
        ``n_valid_dates``.
    """
    # Assemble the empty frame with the pinned columns and index name.
    frame: pd.DataFrame = pd.DataFrame(
        {
            # The arithmetic mean signed rank-IC over admissible non-NaN cells.
            "mean_rank_ic": pd.Series([], dtype="float64"),
            # The admissible valid-date count.
            "n_valid_dates": pd.Series([], dtype="int64"),
        },
        index=pd.Index([], dtype="object", name="reference_id"),
    )
    # Return the empty frame.
    return frame

def _empty_dual_check_results() -> pd.DataFrame:
    """
    Build the empty DUAL_CHECK_RESULTS frame with the pinned columns.

    Returns
    -------
    pd.DataFrame
        Zero rows with columns ``expression_id``, ``trade_date``,
        ``primary_rho``, ``alternative_rho``, ``delta`` and
        ``pass_at_tolerance``.
    """
    # Assemble the empty frame with the pinned columns and dtypes.
    frame: pd.DataFrame = pd.DataFrame(
        {
            # The spot-checked cell's expression identifier.
            "expression_id": pd.Series([], dtype="object"),
            # The spot-checked cell's trade date.
            "trade_date": pd.Series([], dtype="datetime64[ns]"),
            # The primary (vectorised) per-date rank-IC.
            "primary_rho": pd.Series([], dtype="float64"),
            # The alternative (independent) recomputation.
            "alternative_rho": pd.Series([], dtype="float64"),
            # The absolute signed difference.
            "delta": pd.Series([], dtype="float64"),
            # The tolerance verdict under the declared [R] tolerance.
            "pass_at_tolerance": pd.Series([], dtype="bool"),
        }
    )
    # Return the empty frame.
    return frame

def _read_panel_csv(
    path: str,
    kind: str,
    *,
    cast_ledger: Optional[List[Dict[str, Any]]] = None,
) -> pd.DataFrame:
    """
    Re-read a persisted panel CSV and restore the pinned dtypes.

    The sanctioned path from a persisted FEATURES or LABELS panel back
    into memory. The compound (``stock_code``, ``trade_date``)
    MultiIndex is restored before the casts so the pinned integrality
    diagnostic can name the offending coordinate. Identifiers are cast
    to ``string`` and dates to ``datetime64[ns]``. For
    ``kind="features"``, ``volume``, ``quantity`` and
    ``tradestatus_raw`` are cast to nullable ``Int64`` through Task 2's
    pinned integrality gate (``Decimal(str(v))`` conjunction with the
    2^-52 bound, applied per cell, never a silent truncation), and
    ``open``, ``high``, ``low``, ``close`` and ``amount`` to
    ``float64``; for ``kind="labels"``, ``close_adj`` and
    ``adj_factor`` are cast to ``float64``. ``tradable`` is normalised
    from the common CSV spellings of a boolean to nullable
    ``boolean``. Row uniqueness on the compound coordinate is
    re-asserted last.

    Parameters
    ----------
    path : str
        Absolute path of the persisted CSV.
    kind : str
        Panel kind: ``"features"`` or ``"labels"``.
    cast_ledger : Optional[List[Dict[str, Any]]], optional
        Run-scoped append-only cast ledger; the integer-terminal casts
        append to it in cast order. Default is ``None``, which uses a
        re-read-local ledger invisible to the caller.

    Returns
    -------
    pd.DataFrame
        The re-materialised panel indexed by ``stock_code``
        (``string``) and ``trade_date`` (``datetime64[ns]``), carrying
        the per-kind dtypes listed above.

    Raises
    ------
    TypeError
        If ``path`` is not a non-empty ``str``.
    ValueError
        If ``kind`` is not ``"features"`` or ``"labels"``; if the CSV is
        empty; if ``stock_code`` or ``trade_date`` is absent; if the
        pinned integrality gate rejects a cell; or if the compound
        coordinate is duplicated after the re-read.

    Notes
    -----
    The ledger is append-only: casts recorded by a run that later halts
    stay in it.
    """
    # The path must be a non-empty string.
    if not isinstance(path, str) or not path.strip():
        # Reject non-string paths.
        raise TypeError(f"path must be a non-empty str, got {path!r}")
    # The kind must be one of the two pinned panels.
    if kind not in ("features", "labels"):
        # Reject unknown kinds.
        raise ValueError(f"kind must be 'features' or 'labels', got {kind!r}")
    # The caller supplies the run-scoped ledger; otherwise a re-read-local one is used (disclosed).
    ledger: List[Dict[str, Any]] = cast_ledger if cast_ledger is not None else []
    # Read the CSV (no dtype inference beyond the pinned reassertions below).
    frame: pd.DataFrame = pd.read_csv(path)
    # The frame must be non-empty and carry the pinned coordinates.
    if frame.empty:
        # Refuse empty artifacts.
        raise ValueError(f"panel CSV at {path!r} is empty")
    # The coordinate columns must be present.
    missing: List[str] = [c for c in ("stock_code", "trade_date") if c not in frame.columns]
    # Halt with the missing listing when the coordinates are absent.
    if missing:
        # The persisted artifact is defective.
        raise ValueError(f"panel CSV at {path!r} is missing coordinate column(s) {missing}")
    # Identifiers as string dtype (never category, never object).
    frame["stock_code"] = frame["stock_code"].astype("string")
    # The date axis as datetime64[ns].
    frame["trade_date"] = pd.to_datetime(frame["trade_date"]).astype("datetime64[ns]")
    # Restore the compound index BEFORE the casts so the pinned integrality diagnostic
    # can name the (stock_code, trade_date) coordinate of any offending cell.
    frame = frame.set_index(["stock_code", "trade_date"], drop=True)
    # Reassert the pinned dtypes per panel kind.
    if kind == "features":
        # The integer terminals return to nullable Int64 THROUGH the pinned
        # integrality gate (Decimal(str(v)) conjunction with the 2^-52 bound;
        # halt with the pinned diagnostic on any violation -- never truncation).
        for column in ("volume", "quantity", "tradestatus_raw"):
            # Skip absent optional columns (the adjusted view carries no volume).
            if column not in frame.columns:
                # Nothing to cast in this view.
                continue
            # Parse loud on garbage, hand to the pinned gate, append to the ledger.
            frame[column] = task_2.cast_column_task_2(frame, column, task_2.cast_integral_int64_task_2, ledger)
        # The nominal money columns stay float64.
        for column in ("open", "high", "low", "close", "amount"):
            # Skip the absent (adjusted-view) columns.
            if column in frame.columns:
                # Reassert float64.
                frame[column] = frame[column].astype("float64")
    else:
        # The adjusted view: close_adj and adj_factor are the pinned floats.
        for column in ("close_adj", "adj_factor"):
            # Skip the absent column.
            if column in frame.columns:
                # Reassert float64.
                frame[column] = frame[column].astype("float64")
    # The shared tradability flag returns to nullable boolean.
    if "tradable" in frame.columns:
        # Normalise common CSV spellings of booleans.
        frame["tradable"] = frame["tradable"].map({"True": True, "False": False, True: True, False: False}).astype("boolean")
    # Re-assert row uniqueness on the compound coordinate after the cast.
    if not frame.index.is_unique:
        # Count the duplicates for the halt message.
        duplicate_count = int(frame.index.duplicated().sum())
        # Halt loudly: duplicated rows corrupt every downstream rank computation.
        raise ValueError(
            f"panel CSV at {path!r} carries {duplicate_count} duplicated (stock_code, trade_date) "
            f"coordinate(s) after the re-read; row uniqueness is the pinned eligibility substrate"
        )
    # Return the re-materialised panel.
    return frame

def _pinned_mask_cell_count(config: Mapping[str, Any]) -> int:
    """
    Read the pinned mask cardinality from the configuration.

    Reads ``P12_MASKING_PARAMETERS.masked_cell_count`` (the pin of
    record) and ``P34_SCHEMA_MASK_REGISTER.invariants.mask_count`` (the
    corroborating pin) from ``config``; neither is re-declared here.

    Parameters
    ----------
    config : Mapping[str, Any]
        STUDY_CONFIG holding both pinned locations.

    Returns
    -------
    int
        The pinned number of masked stock-day cells (2 on this study).

    Raises
    ------
    TypeError
        If ``config`` is not a ``Mapping``.
    ValueError
        If neither pin is an integral value, or if the two available
        pins disagree.

    Notes
    -----
    Booleans and fractional numbers are never accepted as counts.
    """
    # The configuration must be a mapping (never a duck-typed stand-in).
    if not isinstance(config, Mapping):
        # Reject non-mapping configurations at the boundary.
        raise TypeError(f"config must be a Mapping, got {type(config).__name__}")
    # The P12 pin of record.
    p12 = config.get("P12_MASKING_PARAMETERS")
    # The P34 corroborating pin.
    p34 = config.get("P34_SCHEMA_MASK_REGISTER")
    # Extract the pins defensively.
    pinned_p12: Any = p12.get("masked_cell_count") if isinstance(p12, Mapping) else None
    pinned_p34: Any = None
    # Read the corroborating P34 invariant when it exists.
    if isinstance(p34, Mapping):
        # The invariants sub-mapping carries mask_count.
        invariants = p34.get("invariants")
        # Extract only from a conforming sub-mapping.
        if isinstance(invariants, Mapping):
            # The corroborating count.
            pinned_p34 = invariants.get("mask_count")
    # Coerce a candidate pin to an integral count or None.
    def _as_count(value: Any) -> Optional[int]:
        """
        Coerce a candidate pin to an integral count or ``None``.

        Parameters
        ----------
        value : Any
            A raw configuration value.

        Returns
        -------
        Optional[int]
            The integral value, or ``None`` when ``value`` is a bool,
            non-numeric, or not integral.
        """
        # bools are never counts.
        if isinstance(value, bool) or not isinstance(value, (int, float, np.integer)):
            # Not an integral pin.
            return None
        # Integral check for float pins.
        count = int(value)
        # A fractional pin is not a count.
        return count if float(value) == float(count) else None
    # Coerce both candidates.
    count_p12 = _as_count(pinned_p12)
    # Coerce the corroborating pin.
    count_p34 = _as_count(pinned_p34)
    # Both absent: the contract cannot be enforced; halt rather than invent.
    if count_p12 is None and count_p34 is None:
        # Name the two locations searched.
        raise ValueError(
            "masked_cell_count pin is absent from STUDY_CONFIG (P12_MASKING_PARAMETERS."
            "masked_cell_count and P34_SCHEMA_MASK_REGISTER.invariants.mask_count are both "
            "unavailable); the pinned two-row mask contract cannot be enforced"
        )
    # Disagreement between the pins is a configuration defect; halt with both values.
    if count_p12 is not None and count_p34 is not None and count_p12 != count_p34:
        # Name both pins for the operator.
        raise ValueError(
            f"mask cardinality pins disagree: P12_MASKING_PARAMETERS.masked_cell_count={count_p12} "
            f"vs P34_SCHEMA_MASK_REGISTER.invariants.mask_count={count_p34}; configuration defect"
        )
    # Prefer the P12 pin of record; else the corroborating pin.
    return int(count_p12) if count_p12 is not None else int(count_p34)  # type: ignore[return-value]

def _non_trading_coordinates(frame: pd.DataFrame, config: Mapping[str, Any]) -> List[Dict[str, Any]]:
    """
    Locate the non-trading stock-day cells of one view.

    A coordinate is asserted only where ``tradable`` is a non-NA
    ``False``; cells whose ``tradable`` is NA are counted, logged at
    WARNING level as ``unverified_tradable_status_cells`` and never
    promoted to coordinates. Coordinates are read from the
    ``stock_code``/``trade_date`` columns when present, else from the
    compound index levels. The located count must equal
    :func:`_pinned_mask_cell_count` exactly.

    Parameters
    ----------
    frame : pd.DataFrame
        The unadjusted bars frame; never mutated.
    config : Mapping[str, Any]
        STUDY_CONFIG, consulted only for the pinned mask-cell counts.

    Returns
    -------
    List[Dict[str, Any]]
        One dict per non-trading cell, in frame order, with keys
        ``stock_code`` (``str``) and ``trade_date`` (``pd.Timestamp``).

    Raises
    ------
    TypeError
        If ``frame`` is not a ``DataFrame``, ``config`` is not a
        ``Mapping``, ``tradable`` is absent, or ``tradable`` is not a
        boolean dtype.
    ValueError
        If the located candidate count differs from the pinned count
        (the message lists every candidate), or if no pin is available.
    """
    # The argument must be a DataFrame.
    if not isinstance(frame, pd.DataFrame):
        # Reject non-DataFrames with the observed type.
        raise TypeError(f"frame must be a pandas.DataFrame, got {type(frame).__name__}")
    # The configuration must be a mapping (the pins live there).
    if not isinstance(config, Mapping):
        # Reject non-mapping configurations with the observed type.
        raise TypeError(f"config must be a Mapping, got {type(config).__name__}")
    # The tradability flag must be present.
    if "tradable" not in frame.columns:
        # Reject frames without the shared flag.
        raise TypeError("frame must carry the shared 'tradable' column")
    # Extract the tradability flag once.
    tradable: pd.Series = frame["tradable"]
    # The tradability flag must be boolean-typed (P22 discipline).
    if not pd.api.types.is_bool_dtype(tradable.dtype):
        # Reject non-boolean flags loudly (never a silent coercion).
        raise TypeError(f"frame['tradable'] must be a boolean dtype (P22), got {tradable.dtype!r}")
    # Strict-False only, with NA excluded by Kleene logic and never promoted:
    #   tradable == False  ->  boolean Series where NA stays NA (excluded from the locator).
    strict_false: pd.Series = tradable == False  # noqa: E712 - explicit strict boolean comparison is the pinned semantics
    # Materialise the indexer with NA cleared to False (NA is never a coordinate).
    not_trading: pd.Series = strict_false.fillna(False).astype(bool)
    # NA-status census: disclosed by the caller via the run-exposed-gaps register.
    na_status_count: int = int(tradable.isna().sum())
    # Disclose the census at WARNING level (Count only; no cell is fabricated).
    if na_status_count:
        # Name the census so the audit can point at the run.
        _LOGGER.warning(
            "unverified_tradable_status_cells: %d NA cell(s) on this run; unverified status is disclosed "
            "absence, never a non-trading coordinate and never silently mapped to False",
            na_status_count,
        )
    # Resolve the coordinate axis: key columns win over the index.
    if {"stock_code", "trade_date"}.issubset(frame.columns):
        # Use the key columns verbatim.
        stock_codes = frame["stock_code"]
        # The date axis, verbatim.
        trade_dates = frame["trade_date"]
    else:
        # Fall back to the compound index levels.
        stock_codes = frame.index.get_level_values("stock_code")
        # The compound date level.
        trade_dates = frame.index.get_level_values("trade_date")
    # Accumulate one coordinate dict per located cell, preserving order.
    coordinates: List[Dict[str, Any]] = [
        {"stock_code": str(code), "trade_date": pd.Timestamp(date)}
        for code, date in zip(stock_codes[not_trading], trade_dates[not_trading])
    ]
    # Enforce the pinned cardinality exactly (never a literal; from the configuration).
    pinned_count = _pinned_mask_cell_count(config)
    # More or fewer than the pin halts with the full listing (stop, never widen).
    if len(coordinates) != pinned_count:
        # Render the complete candidate listing for the operator.
        listing = ", ".join(f"({cell['stock_code']}, {cell['trade_date']})" for cell in coordinates)
        # Halt with the pinned count, the observed count and the full list.
        raise ValueError(
            f"mask_cardinality_violation: the pinned mask contract requires exactly {pinned_count} "
            f"masked stock-day coordinate(s); this view locates {len(coordinates)} candidate(s): "
            f"[{listing}]; NA-valued tradable cells are never promoted to coordinates; the register "
            f"is not emitted"
        )
    # Return the coordinates.
    return coordinates

def _corroborate_mask_coordinates(
    unadjusted: pd.DataFrame,
    adjusted: Optional[pd.DataFrame],
    masked_cells: Sequence[Mapping[str, Any]],
) -> List[Dict[str, Any]]:
    """
    Corroborate located mask coordinates against the vendor code.

    For each candidate coordinate the raw ``tradestatus_raw`` cell is
    read in each view and mapped with the pinned vendor mapping
    (``1 -> True``, ``0 -> False``, anything else ``-> NA``). The mapped
    verdict is compared with the asserted not-trading verdict;
    unverified codes yield ``"inconclusive"`` and never a pass. No
    coordinate is minted, merged or dropped here.

    Parameters
    ----------
    unadjusted : pd.DataFrame
        The unadjusted bars frame the coordinates derive from.
    adjusted : Optional[pd.DataFrame]
        The provider-adjusted bars frame, when available; ``None``
        marks the second view unavailable.
    masked_cells : Sequence[Mapping[str, Any]]
        Coordinates emitted by :func:`_non_trading_coordinates`.

    Returns
    -------
    List[Dict[str, Any]]
        One record per well-formed coordinate, in input order. An
        entry that is not a ``Mapping`` carrying both ``stock_code``
        and ``trade_date`` is silently skipped, so the returned list
        can be shorter than ``masked_cells``. Each record has keys
        ``stock_code``, ``trade_date`` (ISO date string),
        ``unadjusted_tradestatus_raw``, ``adjusted_tradestatus_raw``
        (strings or ``None``), ``unadjusted_mapped``,
        ``adjusted_mapped`` (``bool`` or ``None``), ``verdict``
        (``"agreement"``, ``"disagreement"`` or ``"inconclusive"``) and
        ``verdict_adjusted`` (the same plus ``"unavailable"``).
    """
    # Accumulate one record per coordinate.
    records: List[Dict[str, Any]] = []
    # Walk the located coordinates once.
    for position, cell in enumerate(masked_cells):
        # The coordinate must be a mapping with the pinned keys.
        if not isinstance(cell, Mapping) or "stock_code" not in cell or "trade_date" not in cell:
            # Skip only malformed entries (the locator already pinned the shape).
            continue
        # Normalise the coordinate for the label lookup.
        code = str(cell["stock_code"])
        # The date at ns precision.
        date = pd.Timestamp(cell["trade_date"])
        # Read the paired rows in one view at the coordinate.
        def _codes(frame: Optional[pd.DataFrame]) -> Optional[Any]:
            """
            Read the raw vendor status code at one coordinate.

            Parameters
            ----------
            frame : Optional[pd.DataFrame]
                A view; a non-``DataFrame`` value is treated as
                unavailable.

            Returns
            -------
            Optional[Any]
                The first ``tradestatus_raw`` cell at the coordinate, or
                ``None`` when the view, the row or the column is absent.
            """
            # A missing or malformed view yields unavailable, never pass.
            if not isinstance(frame, pd.DataFrame):
                # Unavailable view disclosed as such.
                return None
            # Resolve the raw status code at the coordinate from either shape.
            try:
                # Compound-index lookup when the coordinate lives in the index.
                if {"stock_code", "trade_date"}.issubset(frame.columns):
                    # Flat key-column lookup.
                    hit = frame[(frame["stock_code"].astype(str) == code) & (pd.to_datetime(frame["trade_date"]) == date)]
                else:
                    # Compound-index lookup.
                    hit = frame.loc[(code, date)]
                    # A single row selector returns a Series; normalise to a frame slice.
                    hit = hit.to_frame().T if isinstance(hit, pd.Series) else hit
            except KeyError:
                # The coordinate is absent from this view: unavailable, never pass.
                return None
            # Absent rows corroborate nothing.
            if hit.empty:
                # Unavailable.
                return None
            # The raw vendor code column, when present.
            for column in ("tradestatus_raw",):
                # The column must exist to be read.
                if column in hit.columns or column in hit.index:
                    # Normalise Series-of-one or column access uniformly.
                    values = hit[column] if column in hit.columns else hit.loc[column]
                    # Materialise the first cell value.
                    return values.iloc[0] if hasattr(values, "iloc") else values
            # The view carries no raw vendor code: unavailable.
            return None
        # Read the raw code in the unadjusted view.
        una_raw = _codes(unadjusted)
        # Read the raw code in the adjusted view when supplied.
        adj_raw = _codes(adjusted)
        # Apply the pinned vendor mapping to one code.
        def _mapped(raw: Any) -> Optional[bool]:
            """
            Map one raw vendor code under the pinned vendor mapping.

            Parameters
            ----------
            raw : Any
                The raw ``tradestatus_raw`` value.

            Returns
            -------
            Optional[bool]
                ``True`` for code 1, ``False`` for code 0, ``None`` for
                any missing or unrecognised code.
            """
            # Absent codes map to NA (unverified), never to False.
            if raw is None or (isinstance(raw, float) and pd.isna(raw)) or pd.isna(raw):
                # Unverified maps to NA.
                return None
            # Integral coercion of the raw code.
            try:
                # The vendor codes are integers.
                code_int = int(raw)
            except (TypeError, ValueError):
                # Unrecognised spellings are NA, never False.
                return None
            # The pinned mapping: 1 -> True, 0 -> False, anything else -> NA.
            if code_int == 1:
                # Tradeable day.
                return True
            # The not-trading code.
            if code_int == 0:
                # Not a trading day.
                return False
            # Unknown codes are NEVER mapped to False (survivorship channel).
            return None
        # Map both views.
        una_mapped = _mapped(una_raw)
        # The second view corroborates identically.
        adj_mapped = _mapped(adj_raw)
        # Evaluate the agreement of the code with the asserted not-trading verdict.
        def _verdict(mapped: Optional[bool]) -> str:
            """
            Judge one mapped code against the not-trading assertion.

            Parameters
            ----------
            mapped : Optional[bool]
                The mapped vendor verdict.

            Returns
            -------
            str
                ``"agreement"`` when ``mapped`` is ``False``,
                ``"disagreement"`` when ``True``, ``"inconclusive"``
                when ``None``.
            """
            # Unverified cannot corroborate anything.
            if mapped is None:
                # Inconclusive, never pass.
                return "inconclusive"
            # A False mapped value corroborates the not-trading assertion.
            return "agreement" if mapped is False else "disagreement"
        # Record the corroboration outcome in the auditable structure.
        records.append(
            {
                # The located coordinate.
                "stock_code": code,
                # The located date.
                "trade_date": str(date.date()),
                # The raw code in the unadjusted view.
                "unadjusted_tradestatus_raw": None if una_raw is None else str(una_raw),
                # The raw code in the adjusted view.
                "adjusted_tradestatus_raw": None if adj_raw is None else str(adj_raw),
                # The mapped verdict under the pinned vendor mapping.
                "unadjusted_mapped": una_mapped,
                # The second view's mapped verdict.
                "adjusted_mapped": adj_mapped,
                # The corroboration verdict on the first view.
                "verdict": _verdict(una_mapped),
                # The corroboration verdict on the second view, when present.
                "verdict_adjusted": _verdict(adj_mapped) if adj_raw is not None else "unavailable",
            }
        )
    # Return the per-coordinate corroboration records.
    return records

def _mask_register_from_cells(cells: Sequence[Mapping[str, Any]]) -> pd.DataFrame:
    """
    Build the P34-conform mask register from coordinates.

    Parameters
    ----------
    cells : Sequence[Mapping[str, Any]]
        Coordinate dicts from :func:`_non_trading_coordinates`; each
        must carry ``stock_code`` and ``trade_date``.

    Returns
    -------
    pd.DataFrame
        One row per cell with columns ``stock_code`` (``string``),
        ``trade_date`` (``datetime64[ns]``), ``reason`` (``string``,
        always ``"non_trading_stock_day"``), ``treatment`` (``string``,
        always ``"MASKED"``), ``forward_fill`` (``boolean``, ``False``)
        and ``drop_stock`` (``boolean``, ``False``), in input order.

    Raises
    ------
    ValueError
        If ``cells`` is empty, if an entry is not a ``Mapping``, or if
        an entry lacks ``stock_code`` or ``trade_date``.
    """
    # An empty coordinate list cannot express a mask.
    if not cells:
        # Refuse emptiness loudly (the study pins exactly two cells).
        raise ValueError("cells must be a non-empty sequence of coordinate mappings")
    # Validate and normalise each coordinate.
    rows: List[Dict[str, Any]] = []
    # Walk the coordinates once.
    for position, cell in enumerate(cells):
        # Each cell must be a mapping.
        if not isinstance(cell, Mapping):
            # Reject non-mapping cells.
            raise ValueError(f"cells[{position}] must be a Mapping, got {type(cell).__name__}")
        # The stock code must be present.
        if "stock_code" not in cell or "trade_date" not in cell:
            # Name the cell position and the missing keys.
            raise ValueError(f"cells[{position}] must carry 'stock_code' and 'trade_date'; got keys {sorted(cell)}")
        # Append the normalised P34 row.
        rows.append(
            {
                # The stock code, verbatim as string.
                "stock_code": str(cell["stock_code"]),
                # The trade date, coerced to ns-resolution datetime64[ns] below.
                "trade_date": pd.to_datetime(cell["trade_date"]),
                # The pinned reason (P34.reason == "non_trading_stock_day").
                "reason": "non_trading_stock_day",
                # The pinned treatment (P34.treatment == "MASKED").
                "treatment": "MASKED",
                # Forward fill is prohibited (P34.invariants).
                "forward_fill": False,
                # Dropping the stock is prohibited (P34.invariants).
                "drop_stock": False,
            }
        )
    # Assemble the register.
    register: pd.DataFrame = pd.DataFrame(rows)
    # Reassert the pinned dtypes exactly.
    register["stock_code"] = register["stock_code"].astype("string")
    # Dates as datetime64[ns].
    register["trade_date"] = pd.to_datetime(register["trade_date"]).astype("datetime64[ns]")
    # Identifiers as string dtype.
    for column in ("reason", "treatment"):
        # Reassign with the pinned string dtype.
        register[column] = register[column].astype("string")
    # Flags as nullable boolean dtype.
    for column in ("forward_fill", "drop_stock"):
        # Reassign with the pinned boolean dtype.
        register[column] = pd.array(register[column].to_numpy(), dtype="boolean")
    # Return the register.
    return register

def _apply_mask_to_adjusted(adjusted: pd.DataFrame, mask_register: pd.DataFrame) -> pd.DataFrame:
    """
    Null the value columns of the adjusted view at masked cells.

    The label input is the provider-adjusted close (Eq. (1)), so a
    masked stock-day must be unusable for labels as well as for
    scoring. Every column except ``stock_code``, ``trade_date``,
    ``adjustment`` and ``byte_hash`` is set to ``pd.NA`` at the
    registered coordinates; identity and provenance columns are kept
    verbatim. Masked rows are located via the key columns when present,
    else via the compound index.

    Parameters
    ----------
    adjusted : pd.DataFrame
        The cleansed adjusted panel; never mutated.
    mask_register : pd.DataFrame
        The P34 mask register with ``stock_code`` and ``trade_date``.

    Returns
    -------
    pd.DataFrame
        A deep copy of the panel with masked cells nulled.
    """
    # Deep-copy the panel; the caller's frame is never mutated.
    masked: pd.DataFrame = adjusted.copy(deep=True)
    # Resolve the coordinate axis (key columns win over the index).
    if {"stock_code", "trade_date"}.issubset(masked.columns):
        # Key columns: build a boolean row mask per coordinate.
        row_mask = pd.Series(False, index=masked.index)
        # Walk the register coordinates.
        for _, coordinate in mask_register.iterrows():
            # Build the coordinate predicate.
            hit = (masked["stock_code"] == coordinate["stock_code"]) & (masked["trade_date"] == coordinate["trade_date"])
            # Accumulate the predicate.
            row_mask = row_mask | hit
    else:
        # Compound index: locate the coordinates via the index.
        coordinates = list(zip(mask_register["stock_code"], mask_register["trade_date"]))
        # The membership predicate over the compound index.
        row_mask = masked.index.isin(coordinates)
    # The numeric value columns eligible for nulling (identity/provenance excluded).
    protected: set = {"stock_code", "trade_date", "adjustment", "byte_hash"}
    # Null every non-protected value column at the masked rows.
    for column in masked.columns:
        # Skip protected columns.
        if column in protected:
            # Keep identity/provenance verbatim.
            continue
        # Null the masked cells (NA semantics, never zero).
        masked.loc[row_mask, column] = pd.NA
    # Return the masked copy.
    return masked

def _frozen_expression_ids(frozen_cohort: Any) -> List[str]:
    """
    Extract the ordered expression identifiers from a frozen cohort.

    The ordering axis is the ``cohort_rank`` index when it names the
    index; otherwise the ``cohort_rank`` column; a bare ``RangeIndex``
    without that column is rejected. Sorting is never applied to both
    axes at once.

    Parameters
    ----------
    frozen_cohort : Any
        The frozen cohort object; its ``rows`` attribute must be a
        ``DataFrame``.

    Returns
    -------
    List[str]
        The ``expression_id`` values as strings, in ascending cohort
        rank.

    Raises
    ------
    ValueError
        If ``rows`` is not a ``DataFrame``, lacks ``expression_id``,
        carries neither a ``cohort_rank`` index nor column, is empty,
        or repeats an identifier.
    """
    # The frozen rows must be a DataFrame.
    rows: pd.DataFrame = getattr(frozen_cohort, "rows", None)
    # A frame is required.
    if not isinstance(rows, pd.DataFrame):
        # Refuse unusable frozen cohorts.
        raise ValueError(f"frozen cohort rows must be a pandas.DataFrame, got {type(rows).__name__}")
    # The expression identifiers must be present.
    if "expression_id" not in rows.columns:
        # Refuse cohorts without identifiers.
        raise ValueError("frozen cohort rows are missing 'expression_id'")
    # The authoritative order: the cohort_rank index when present.
    if rows.index.name == "cohort_rank":
        # Sort by the cohort_rank index (the hash order).
        ordered: pd.DataFrame = rows.sort_index()
    # Otherwise cohort_rank must exist as a column exclusively.
    elif "cohort_rank" in rows.columns:
        # Sort by the cohort_rank column.
        ordered = rows.sort_values("cohort_rank")
    else:
        # Neither order axis is usable.
        raise ValueError("frozen cohort rows carry neither a cohort_rank index nor a cohort_rank column")
    # Extract the identifiers as strings, in cohort order.
    identifiers: List[str] = list(ordered["expression_id"].astype(str))
    # The ordered list must be non-empty.
    if not identifiers:
        # Refuse empty cohorts.
        raise ValueError("frozen cohort carries no expression identifiers")
    # The identifiers must be unique (one preserved row per identity).
    if len(set(identifiers)) != len(identifiers):
        # Refuse duplicated cohorts.
        raise ValueError("frozen cohort carries duplicated expression identifiers")
    # Return the ordered list.
    return identifiers

def _eval_node(
    node: Any,
    terminals: Mapping[str, pd.Series],
    registry: Mapping[str, Any],
) -> pd.Series:
    """
    Evaluate one DSL abstract-syntax node over the terminals.

    A node without operands resolves against ``terminals``; any other
    node looks up its ``OperatorSpec`` in ``registry``, recursively
    evaluates each operand (sub-``Node`` recurses, ``str`` resolves
    against ``terminals``, anything else is passed through as a numeric
    literal such as a window size) and dispatches to
    ``spec.implementation``. The result is aligned to the terminal
    index; operator results are float64, but a leaf returns the
    terminal Series verbatim and therefore keeps that terminal's own
    dtype (e.g. nullable ``Int64`` or ``boolean``), not necessarily
    ``float64``.

    Parameters
    ----------
    node : Any
        The AST node to evaluate.
    terminals : Mapping[str, pd.Series]
        The terminal registry (``close``, ``volume``, ``amount``,
        ``tradable``) on the compound index.
    registry : Mapping[str, Any]
        The operator registry mapping operator names to
        ``OperatorSpec`` values.

    Returns
    -------
    pd.Series
        The evaluated signal on the compound index. An operator node
        yields a float64 series; a leaf yields the registered terminal
        Series itself, in the terminal's own dtype.

    Raises
    ------
    KeyError
        If a leaf names an unregistered terminal, if an operator is
        absent from ``registry``, or if a string operand does not
        resolve against ``terminals``.
    """
    # A leaf (no operands) resolves against the terminals first.
    if not node.args:
        # The leaf operator name must be a registered terminal.
        if node.op in terminals:
            # Return the terminal Series verbatim (cell-for-cell).
            return terminals[node.op]
        # An operator-shaped leaf with no operands is a defect.
        raise KeyError(f"unresolvable leaf {node.op!r}: not a registered terminal")
    # An internal node must exist in the operator registry.
    if node.op not in registry:
        # Name the missing operator.
        raise KeyError(f"operator {node.op!r} is absent from the public registry")
    # Fetch the frozen operator specification.
    spec: Any = registry[node.op]
    # Evaluate each operand recursively.
    arguments: List[Any] = []
    # Walk the operand tuple once.
    for child in node.args:
        # A sub-expression recurses.
        if isinstance(child, task_7.Node):
            # Evaluate the subtree.
            arguments.append(_eval_node(child, terminals, registry))
            # Advance to the next operand.
            continue
        # A string operand is a literal terminal reference (never produced by the parser).
        if isinstance(child, str):
            # The string must resolve against the terminals.
            if child in terminals:
                # Carry the terminal Series.
                arguments.append(terminals[child])
                # Advance to the next operand.
                continue
            # Otherwise the operand is invalid.
            raise KeyError(f"string operand {child!r} is not a registered terminal")
        # Anything else is a numeric literal (window sizes; Node normalises ints to floats).
        arguments.append(child)
    # Dispatch to the pinned operator implementation.
    return spec.implementation(*arguments)

def _build_factor_panel(
    features_panel: pd.DataFrame,
    texts_ids: Sequence[str],
    texts: Mapping[str, str],
    registry: Mapping[str, Any],
    parse: Callable[[str], Any],
) -> Tuple[pd.DataFrame, Tuple[str, ...]]:
    """
    Evaluate the frozen cohort expressions into a signal panel.

    The terminal registry is built once for the whole cohort from the
    masked features panel. Each identifier's text is parsed and its AST
    evaluated with :func:`_eval_node`; a missing text, a parse error or
    an evaluation error is recorded and replaced by an all-NaN column,
    never a dropped column, so the panel is total over the cohort.

    Parameters
    ----------
    features_panel : pd.DataFrame
        The post-mask nominal FEATURES panel on the compound index.
    texts_ids : Sequence[str]
        Expression identifiers in cohort order.
    texts : Mapping[str, str]
        Expression text per identifier; absent keys count as parse
        failures.
    registry : Mapping[str, Any]
        The operator registry to evaluate against (v1 or v2).
    parse : Callable[[str], Any]
        The expression parser.

    Returns
    -------
    panel : pd.DataFrame
        One float64 column per identifier, in cohort order, index equal
        to ``features_panel.index``.
    parse_failed : Tuple[str, ...]
        The identifiers whose expression could not be parsed or
        evaluated, in encounter order.
    """
    # Build the terminal registry once for the whole cohort (cell-for-cell reuse).
    terminals: Mapping[str, pd.Series] = task_7.build_terminal_registry_task_7(features_panel)
    # Accumulate columns in cohort order.
    columns: Dict[str, pd.Series] = {}
    # Accumulate the parse_failed identifiers in encounter order.
    parse_failed: List[str] = []
    # The panel's compound index is the alignment backbone for every column.
    index: pd.MultiIndex = features_panel.index
    # Walk the cohort identifiers once.
    for identifier in texts_ids:
        # Fetch the expression text (absent text is a parse failure, not a crash).
        text: Optional[str] = texts.get(str(identifier))
        # A missing text is recorded and the column is NaN.
        if text is None:
            # Record the failure.
            parse_failed.append(str(identifier))
            # Emit an all-NaN column on the shared index.
            columns[str(identifier)] = pd.Series(np.nan, index=index, dtype="float64")
            # Advance to the next identifier.
            continue
        # Parse and evaluate the expression; any failure is recorded, never raised.
        try:
            # Parse the expression text into the frozen AST.
            node: Any = parse(text)
            # Evaluate the AST over the terminals and registry.
            signal: pd.Series = _eval_node(node, terminals, registry)
            # Cast the substrate to float64 (NA-kept NA semantics).
            columns[str(identifier)] = pd.Series(signal, index=index).astype("float64")
        except Exception:
            # Record the failure; the column is all-NaN so scoring is total.
            parse_failed.append(str(identifier))
            # Emit an all-NaN column on the shared index.
            columns[str(identifier)] = pd.Series(np.nan, index=index, dtype="float64")
    # Assemble the panel in one shot (columns in cohort order).
    panel: pd.DataFrame = pd.DataFrame(columns, index=index)
    # Return the panel and the parse_failed tuple.
    return panel, tuple(parse_failed)

def _registry_with_conditional(version: str) -> Mapping[str, Any]:
    """
    Build the operator registry for one evaluator version.

    Starts from the canonical six-operator registry, then re-binds only
    the ``if`` implementation: the v1 defect emulator or the v2 scalar
    broadcasting elementwise conditional. Arity, kind, provenance tag
    and the symmetric flag of the ``if`` spec are preserved verbatim.

    Parameters
    ----------
    version : str
        ``"v1"`` (defect emulator) or ``"v2"`` (elementwise scalar
        conditional).

    Returns
    -------
    Mapping[str, Any]
        An immutable mapping from operator name to ``OperatorSpec``.

    Raises
    ------
    ValueError
        If ``version`` is neither ``"v1"`` nor ``"v2"``.
    """
    # The version must be one of the two pinned tags.
    if version not in ("v1", "v2"):
        # Reject unknown versions.
        raise ValueError(f"version must be 'v1' or 'v2', got {version!r}")
    # Build the canonical six-reference registry.
    base: Mapping[str, Any] = task_7.build_operator_registry_task_7()
    # Promote to a mutable mapping for the single re-binding.
    promoted: Dict[str, Any] = dict(base)
    # Copy the original conditional spec (pinned arity/kind/provenance/symmetric).
    original: Any = promoted["if"]
    # Select the conditional implementation for the requested version.
    implementation: Callable[..., Any]
    # The v1 tag binds the defect emulator; the v2 tag binds scalar broadcasting.
    implementation = task_7.v1_emulator_task_7 if version == "v1" else task_7.elementwise_if_task_7
    # Re-bind only the implementation; every other field of the spec is preserved.
    promoted["if"] = task_7.OperatorSpec(
        # The pinned arity is preserved verbatim (3).
        arity=original.arity,
        # The pinned family is preserved verbatim (conditional).
        kind=original.kind,
        # The version-specific implementation replaces the shared one.
        implementation=implementation,
        # The provenance tag is preserved verbatim ([I]; P13.operator_status_note).
        provenance_tag=original.provenance_tag,
        # The pinned symmetric flag is preserved verbatim (False).
        symmetric=original.symmetric,
    )
    # Freeze the registry behind the immutable proxy.
    frozen: Mapping[str, Any] = MappingProxyType(promoted)
    # Return the registry.
    return frozen

def _canonicalizer_ok(canonicalizer: Any) -> bool:
    """
    Verify a canonicalizer handle by interface and by probe.

    The handle must expose callables ``canonical_bytes`` and
    ``expression_id``, either as entries of a mapping or as attributes
    of an object. The probe canonicalises one fixed expression,
    ``ts_mean(close, 1)``, and requires the bytes certificate to be
    ``bytes``/``bytearray`` and the id certificate to be a non-empty
    string; any probe failure refutes the handle.

    Parameters
    ----------
    canonicalizer : Any
        Handle to verify: a mapping holding the two certificates as
        values, or an object exposing them as attributes.

    Returns
    -------
    bool
        True iff both certificates are callable and the probe passes.
    """
    # A mapping canonicalizer exposes the callables as values.
    if isinstance(canonicalizer, Mapping):
        # Both pinned entries must be present and callable.
        canonical_bytes = canonicalizer.get("canonical_bytes")
        expression_id = canonicalizer.get("expression_id")
    else:
        # An object canonicalizer exposes the callables as attributes.
        canonical_bytes = getattr(canonicalizer, "canonical_bytes", None)
        expression_id = getattr(canonicalizer, "expression_id", None)
    # Both certificates must be callable.
    if not callable(canonical_bytes) or not callable(expression_id):
        # The handle is not usable.
        return False
    # Probe the certificates on one fixed expression.
    try:
        # Parse the pinned probe expression.
        probe_node: Any = task_7.parse_expression_task_7("ts_mean(close, 1)")
        # The canonical bytes must be bytes.
        canonical_bytes_value: Any = canonical_bytes(probe_node)
        # The expression id must be a non-empty string.
        expression_id_value: Any = expression_id(probe_node)
        # Both contracts must hold.
        return isinstance(canonical_bytes_value, (bytes, bytearray)) and isinstance(expression_id_value, str) and bool(expression_id_value)
    except Exception:
        # Any probe failure refutes the handle.
        return False

def _oos_read_timestamp() -> datetime.datetime:
    """
    Derive the run-local OOS-read timestamp used by the freeze.

    The clean OOS protocol is pending in the configuration, so no
    absolute first-read instant is pinned; the freeze protocol still
    requires a timezone-aware instant strictly later than the freeze
    timestamp. A run-local Asia/Shanghai stamp one day ahead of now is
    derived here and disclosed in the admissibility report as a derived
    parameter, never as a paper figure.

    Returns
    -------
    datetime.datetime
        Timezone-aware Asia/Shanghai instant one day ahead of the run.
    """
    # Now in Asia/Shanghai, plus one day (strictly after any same-run freeze instant).
    derived: datetime.datetime = datetime.datetime.now(ASIA_SHANGHAI_TZ) + datetime.timedelta(days=1)
    # Return the derived instant.
    return derived

def _task_7_handoff(
    context: PipelineContext,
    preserved_rows: Optional[pd.DataFrame],
    task_1_output: Any,
) -> Dict[str, Any]:
    """
    Rebuild the five-key Task 7 summary consumed by the handoff.

    The mapping is rebuilt from this run's own context, never from a
    fixture: the operator registry is re-derived, the terminal registry
    is built cell-for-cell from the masked features panel held in the
    context (an empty compound-index registry when that panel is
    absent), the registry is stamped with its SHA-256, and the parse
    table is produced by running the preserved rows through the public
    parser path (an empty frame when ``preserved_rows`` is absent or
    lacks ``expression_text``).

    Parameters
    ----------
    context : PipelineContext
        Active pipeline context; its ``payloads`` supply the masked
        features panel.
    preserved_rows : Optional[pd.DataFrame]
        Preserved input rows with an ``expression_text`` column;
        ``None`` or a frame without it yields an empty parse table.
    task_1_output : Any
        The Task 1 halt-reason-or-None outcome, echoed unchanged.

    Returns
    -------
    Dict[str, Any]
        Five keys: ``registry`` (operator registry mapping),
        ``terminal_registry`` (terminal name to series, on the
        ``stock_code``/``trade_date`` compound index),
        ``registry_sha256`` (64-character lowercase hex),
        ``parse_table`` (one row per parsed expression) and
        ``task1_preflight`` (echoed Task 1 outcome).
    """
    # Rebuild the canonical operator registry for this run.
    registry: Mapping[str, Any] = task_7.build_operator_registry_task_7()
    # Resolve the masked features panel from the context (terminal registry source).
    features: Optional[pd.DataFrame] = context.payloads.get("masked_features_panel")
    # Build the terminal registry from the masked panel (cell-for-cell reuse).
    if isinstance(features, pd.DataFrame):
        # The masked panel defines the terminals.
        terminal_registry: Mapping[str, pd.Series] = task_7.build_terminal_registry_task_7(features)
    else:
        # Without a panel, an empty-but-well-formed terminal registry is emitted.
        empty_index = pd.MultiIndex.from_arrays(
            [pd.Series([], dtype="object"), pd.Series([], dtype="datetime64[ns]")],
            names=["stock_code", "trade_date"],
        )
        # The four pinned terminals, empty Series on the empty compound index.
        terminal_registry = {
            name: pd.Series([], dtype="float64", index=empty_index)
            for name in ("close", "volume", "amount", "tradable")
        }
    # Stamp the registry's SHA-256.
    registry_sha256: str = task_7.registry_sha256_task_7(registry)
    # Build the parse table from the preserved rows.
    if isinstance(preserved_rows, pd.DataFrame) and "expression_text" in preserved_rows.columns:
        # Parse the preserved rows through the public parser path.
        parse_table: pd.DataFrame = task_7.parse_records_task_7(preserved_rows, registry=registry)
    else:
        # An empty parse table on the same code path.
        parse_table = task_7.parse_records_task_7(
            pd.DataFrame({"expression_text": pd.Series([], dtype=object)}),
            registry=registry,
        )
    # Assemble the five-key handoff mapping.
    handoff: Dict[str, Any] = {
        # The frozen operator registry.
        "registry": registry,
        # The terminal registry (cell-for-cell reuse of the masked panel).
        "terminal_registry": terminal_registry,
        # The registry stamp.
        "registry_sha256": registry_sha256,
        # The parse table (one row per preserved row).
        "parse_table": parse_table,
        # The Task 1 verdict, echoed verbatim.
        "task1_preflight": task_1_output,
    }
    # Return the handoff mapping.
    return handoff

def _missing_upstream(stage: str, missing: Sequence[str], halted_at: Optional[str]) -> str:
    """
    Render the MISSING_UPSTREAM_INPUT halt reason for one stage.

    The absent input names are sorted and joined with ``", "``, then
    embedded in a single line that names the blocked stage and, when
    the first halted upstream stage is known, attributes the halt to
    it.

    Parameters
    ----------
    stage : str
        Pinned stage name that cannot run.
    missing : Sequence[str]
        Pinned names of the absent upstream inputs.
    halted_at : Optional[str]
        First upstream stage observed halted, or ``None``.

    Returns
    -------
    str
        One-line reason prefixed with ``MISSING_UPSTREAM_INPUT`` and
        suffixed with the upstream attribution when ``halted_at`` is
        set.
    """
    # Join the missing input names deterministically.
    missing_text: str = ", ".join(sorted(missing))
    # The reason names the missing inputs and, when known, the halted upstream stage.
    if halted_at:
        # Compose the upstream-halt phrasing.
        return f"{MISSING_UPSTREAM_INPUT}::{stage}: missing upstream input(s) {missing_text} (pipeline halted at {halted_at})"
    # Compose the unconditional phrasing.
    return f"{MISSING_UPSTREAM_INPUT}::{stage}: missing upstream input(s) {missing_text}"

def _halt_reason(exc: BaseException) -> str:
    """
    Convert a caught exception to the pinned halt-reason spelling.

    A raised exception is a halt carrying the exception class and
    message; swallowing is forbidden. This helper is the single
    spelling of that conversion.

    Parameters
    ----------
    exc : BaseException
        Caught exception.

    Returns
    -------
    str
        ``"<ClassName>: <message>"``, the message rendered by
        ``str(exc)``.
    """
    # Compose the pinned spelling.
    reason: str = f"{type(exc).__name__}: {exc}"
    # Return the reason.
    return reason

def validate_inputs(
    files: pd.DataFrame,
    unadjusted: pd.DataFrame,
    adjusted: pd.DataFrame,
    config: Dict[str, Any],
) -> Tuple[Optional[Dict[str, Any]], Optional[str]]:
    """
    Validate the FILES manifest, both bar views and STUDY_CONFIG.

    Builds the dense trading calendar from the raw unadjusted view
    (the calendar is a deterministic function of the acquisition),
    caches it and its manifest digest on the context, then runs the
    three pinned Task 1 steps (FILES manifest, two panels,
    STUDY_CONFIG) against that calendar and returns the first halt
    reason produced there, if any. The caller's frames are never
    mutated.

    Parameters
    ----------
    files : pd.DataFrame
        The FILES acquisition manifest (80 rows on the pinned study).
    unadjusted : pd.DataFrame
        Nominal price/volume view.
    adjusted : pd.DataFrame
        Provider-adjusted view; label input only.
    config : Dict[str, Any]
        The STUDY_CONFIG dictionary (semantic dict plus
        ``_provenance_tags``).

    Returns
    -------
    Tuple[Optional[Dict[str, Any]], Optional[str]]
        ``(validation_report, halt_reason)``. The reason is ``None``
        on success; otherwise the report is ``None`` and the reason is
        the verbatim Task 1 halt string when Task 1 reports a halt
        (which need not read ``<ClassName>: <message>``), or the
        exception class and message for this function's own boundary,
        calendar or Task 1 exception failures. On success the report
        carries ``calendar_row_count``, ``calendar_sha256``,
        ``files_row_count``, ``unadjusted_row_count``,
        ``adjusted_row_count``, ``stock_code_count`` (unique codes
        across both views) and ``task1_halt_reason`` (``None``).

    Notes
    -----
    On success the calendar is cached in ``payloads['calendar']``, the
    ``validate`` manifest is recorded, and the Task 1 verdict is stored
    in ``payloads['validate_halt_reason']``.
    """
    # Boundary: files must be a DataFrame.
    if not isinstance(files, pd.DataFrame):
        # Reject non-DataFrame manifests with the observed type.
        return None, f"TypeError: files must be a pandas.DataFrame, got {type(files).__name__}"
    # Boundary: the unadjusted view must be a DataFrame.
    if not isinstance(unadjusted, pd.DataFrame):
        # Reject non-DataFrame views with the observed type.
        return None, f"TypeError: unadjusted must be a pandas.DataFrame, got {type(unadjusted).__name__}"
    # Boundary: the adjusted view must be a DataFrame.
    if not isinstance(adjusted, pd.DataFrame):
        # Reject non-DataFrame views with the observed type.
        return None, f"TypeError: adjusted must be a pandas.DataFrame, got {type(adjusted).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context (run-scoped or ephemeral).
    context: PipelineContext = _current_context()
    # Build the calendar from the raw unadjusted view (reset_index keeps key columns only).
    try:
        # The calendar consumes exactly the trade_date column.
        calendar: pd.DataFrame = task_5.construct_calendar_task_5(unadjusted.reset_index(drop=True))
    except Exception as exc:
        # A calendar that cannot be built is an ingest halt.
        return None, _halt_reason(exc)
    # Cache the calendar for the custody/calendar stages.
    context.payloads["calendar"] = calendar
    # Record the calendar manifest (rows and SHA-256 of the persisted bytes).
    calendar_csv: str = calendar.to_csv(index=False, lineterminator="\n")
    # Persist the manifest digest computed over the exact bytes.
    context.manifests["validate"] = {
        # The dense row count (pinned 727).
        "calendar_row_count": int(len(calendar)),
        # The SHA-256 of the serialised calendar frame.
        "calendar_sha256": hashlib.sha256(calendar_csv.encode("utf-8")).hexdigest(),
        # The files row count (pinned 80).
        "files_row_count": int(len(files)),
        # The unadjusted row count.
        "unadjusted_row_count": int(len(unadjusted)),
        # The adjusted row count.
        "adjusted_row_count": int(len(adjusted)),
    }
    # Execute Task 1's pinned three steps in order.
    try:
        # Task 1 returns the first halt reason, or None when every gate passes.
        task_1_halt: Optional[str] = task_1.run_task_1(
            FILES=files,
            UNADJUSTED_BARS=unadjusted,
            ADJUSTED_BARS=adjusted,
            STUDY_CONFIG=config,
            TRADING_CALENDAR=calendar,
        )
    except Exception as exc:
        # Task 1's own boundary defects surface as halts with class and message.
        return None, _halt_reason(exc)
    # Echo the Task 1 verdict into the context for the correction handoff.
    context.payloads["validate_halt_reason"] = task_1_halt
    # A Task 1 halt is the pipeline's verdict; surface it verbatim.
    if task_1_halt is not None:
        # No outputs on halt.
        return None, str(task_1_halt)
    # Derive the stock count for the report (unique codes across the union of the views).
    stock_codes: set = set()
    # Walk both views' key axes.
    for frame in (unadjusted, adjusted):
        # Key columns when present, else the compound index.
        if {"stock_code"}.issubset(frame.columns):
            # Extend with the column values.
            stock_codes.update(str(code) for code in frame["stock_code"].unique())
        else:
            # Extend with the index level values.
            stock_codes.update(str(code) for code in frame.index.get_level_values("stock_code").unique())
    # Assemble the validation report (all values computed here).
    validation_report: Dict[str, Any] = {
        # The calendar's dense row count.
        "calendar_row_count": int(len(calendar)),
        # The calendar's SHA-256.
        "calendar_sha256": context.manifests["validate"]["calendar_sha256"],
        # The files row count.
        "files_row_count": int(len(files)),
        # The unadjusted row count.
        "unadjusted_row_count": int(len(unadjusted)),
        # The adjusted row count.
        "adjusted_row_count": int(len(adjusted)),
        # The number of unique stock codes across the views.
        "stock_code_count": int(len(stock_codes)),
        # The Task 1 verdict (None on success).
        "task1_halt_reason": None,
    }
    # Return the report and no halt reason.
    return validation_report, None

def cleanse_data(
    unadjusted: pd.DataFrame,
    adjusted: pd.DataFrame,
    config: Dict[str, Any],
) -> Tuple[Optional[pd.DataFrame], Optional[pd.DataFrame], Optional[str]]:
    """
    Cleanse both bar views with the pinned Task 2 primitives.

    Works on deep copies of the two views and applies, in order: the
    numpy-bool repair of the legislated boolean columns; the integer
    terminal casts (``volume``, ``quantity``, ``tradestatus_raw``) to
    nullable ``Int64`` where present; the ``trade_date`` cast to naive
    ``datetime64[ns]``; the tradability flag cast to nullable arrow
    boolean; the bar-view string columns cast to ``string[pyarrow]``;
    the BaoStock stock-code regex assertion; the identity-string
    persistence; the enum-domain audits; and the scoped
    prohibited-sentinel scan. Each cast is appended to the run-scoped
    cast ledger when the run owns one, else to a stage-local ledger.

    The anti-imputation post-condition is then certified against the
    run-scoped Task 2 controller, and both indexes must equal their
    input snapshots. Nothing is filled, interpolated, reordered or
    deduplicated.

    Parameters
    ----------
    unadjusted : pd.DataFrame
        Raw unadjusted view.
    adjusted : pd.DataFrame
        Raw provider-adjusted view.
    config : Dict[str, Any]
        The STUDY_CONFIG dictionary; carried and unused at this stage.

    Returns
    -------
    Tuple[Optional[pd.DataFrame], Optional[pd.DataFrame], Optional[str]]
        ``(cleansed_unadj, cleansed_adj, halt_reason)``. On success the
        reason is ``None`` and both frames are cast copies whose
        indexes are unchanged. On any halt both frames are ``None``.
        Halt conditions include a non-DataFrame view or non-dict
        configuration and a prohibited sentinel outside the
        allow-lists (rendered as ``ValueError:`` plus the hit table),
        which are returned directly before the guard; plus every
        ``RuntimeError`` listed under Raises, which the guard's
        ``except`` clause converts into the returned reason.

    Raises
    ------
    RuntimeError
        Raised inside the guard, then converted into the returned
        ``halt_reason``. Raised when the run-scoped Task 2 controller
        is absent from or mistyped in the context; when any of the
        three anti-imputation counters is non-zero (the recorded
        witness list is appended); when the unadjusted index was not
        preserved; and when the adjusted index was not preserved.

    Notes
    -----
    On success the context caches the ``cleansed_unadjusted`` and
    ``cleansed_adjusted`` payloads and records the ``cleanse`` manifest
    (cast count, stock-code count, the three anti-imputation counters
    and the cast-ledger scope).
    """
    # Boundary: the unadjusted view must be a DataFrame.
    if not isinstance(unadjusted, pd.DataFrame):
        # Reject non-DataFrame views with the observed type.
        return None, None, f"TypeError: unadjusted must be a pandas.DataFrame, got {type(unadjusted).__name__}"
    # Boundary: the adjusted view must be a DataFrame.
    if not isinstance(adjusted, pd.DataFrame):
        # Reject non-DataFrame views with the observed type.
        return None, None, f"TypeError: adjusted must be a pandas.DataFrame, got {type(adjusted).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context (run-scoped or ephemeral).
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # Work exclusively on deep copies.
        una: pd.DataFrame = unadjusted.copy(deep=True)
        # Work exclusively on deep copies.
        adj: pd.DataFrame = adjusted.copy(deep=True)
        # Snapshot the input indexes for the preservation post-condition.
        input_indexes: Dict[str, pd.Index] = {
            # The unadjusted index snapshot.
            "UNADJUSTED_BARS": una.index.copy(deep=True),
            # The adjusted index snapshot.
            "ADJUSTED_BARS": adj.index.copy(deep=True),
        }
        # Repair numpy-bool legislated boolean columns exactly as run_task_2 does.
        for frame in (una, adj):
            # Walk the legislated boolean targets of this frame.
            for column in frame.columns:
                # The unsafe case is exactly the numpy bool dtype.
                if getattr(frame[column].dtype, "name", None) == "bool" and column in task_2._BAR_BOOLEAN_COLUMNS:
                    # Rebuild the column with Python-native bool cells.
                    repaired: List[bool] = [bool(value) for value in frame[column]]
                    # Reassign an object-dtype column (order and index preserved).
                    frame[column] = pd.Series(repaired, index=frame[column].index, name=column, dtype=object)
        # The run-scoped anti-imputation controller and cast ledger are read from the module's
        # request-scoped bridge (the same object every stage of THIS run reads and the same object
        # any gated operation on this run would have written to). A missing block never reads as a
        # zeroed post-condition: it halts, loudly, with the reason named.
        run_context = _current_context()
        # The controller must be Task 2's concrete type (never a duck-typed stand-in).
        anti_imputation_controller: Any = getattr(run_context, "task_2_context", None)
        # A PipelineContext that was never wired refuses to certify the post-condition.
        if anti_imputation_controller is None:
            # Refuse to certify an untested post-condition; never a faked zero.
            raise RuntimeError(
                "anti_imputation_context_missing: the run-scoped anti-imputation controller "
                "(Task 2 PipelineContextTask2) is absent from this run's context; the "
                "anti-imputation post-condition cannot be certified on this run"
            )
        # The controller must be Task 2's concrete type.
        if not isinstance(anti_imputation_controller, task_2.PipelineContextTask2):
            # Name the observed type for the operator.
            raise RuntimeError(
                f"anti_imputation_context_missing: expected a task_2.PipelineContextTask2, "
                f"got {type(anti_imputation_controller).__name__}"
            )
        # The run-scoped cast ledger, when the bridge created one; else a disclosed stage-local one.
        ledger_shared = run_context.payloads.get("cast_ledger")
        # Only a genuinely shared list is used; anything else allocates a stage-local one.
        if isinstance(ledger_shared, list):
            # The run-scoped shared append-only ledger.
            cast_ledger = ledger_shared
            # The scope label for the manifest.
            cast_ledger_scope = "run_scoped"
        else:
            # Stand-alone caller scope: a fresh stage-local ledger (disclosed in the manifest).
            cast_ledger = []
            # The scope label for the manifest.
            cast_ledger_scope = "stage_local"
        # Walk the two views in the pinned order.
        for view_name, frame in (("UNADJUSTED_BARS", una), ("ADJUSTED_BARS", adj)):
            # The three integer terminals cast to nullable Int64.
            for column in task_2._INTEGER_TERMINAL_COLUMNS:
                # Skip absent columns (the adjusted view carries no volume).
                if column not in frame.columns:
                    # Nothing to cast in this view.
                    continue
                # Cast through the shared primitive; the ledger records it.
                frame[column] = task_2.cast_column_task_2(frame, column, task_2.cast_integral_int64_task_2, cast_ledger)
            # trade_date casts to naive datetime64[ns].
            if "trade_date" in frame.columns:
                # Cast through the naive exchange-local temporal primitive.
                frame["trade_date"] = task_2.cast_column_task_2(frame, "trade_date", task_2.cast_trade_date_task_2, cast_ledger)
            # The shared tradability flag casts to nullable boolean[pyarrow].
            for column in task_2._BAR_BOOLEAN_COLUMNS:
                # Skip absent columns.
                if column not in frame.columns:
                    # Nothing to cast in this view.
                    continue
                # Cast through the pinned primitive.
                frame[column] = task_2.cast_column_task_2(frame, column, task_2.cast_boolean_pyarrow_task_2, cast_ledger)
            # The bar-view string columns persist as string[pyarrow].
            for column in task_2._BAR_STRING_COLUMNS:
                # Skip absent columns.
                if column not in frame.columns:
                    # Nothing to cast in this view.
                    continue
                # Cast through the pinned primitive.
                frame[column] = task_2.cast_column_task_2(frame, column, task_2.cast_string_pyarrow_task_2, cast_ledger)
        # Assert the BaoStock stock-code regex over the views (the [R] guard).
        stock_codes: List[str] = task_2.assert_stock_codes_task_2({"UNADJUSTED_BARS": una, "ADJUSTED_BARS": adj})
        # Persist the identity strings in each view.
        una = task_2.enforce_identity_strings_task_2(una, [column for column in task_2._BAR_STRING_COLUMNS if column in una.columns])
        # Persist the identity strings in each view.
        adj = task_2.enforce_identity_strings_task_2(adj, [column for column in task_2._BAR_STRING_COLUMNS if column in adj.columns])
        # Assert the enum domains on the unadjusted view.
        una = task_2.assert_enum_domains_task_2(una)
        # Assert the enum domains on the adjusted view.
        adj = task_2.assert_enum_domains_task_2(adj)
        # Run the scoped prohibited-sentinel scan on the unadjusted view.
        una_hits, una_clean = task_2.scan_prohibited_sentinels_task_2(una)
        # A non-clean frame cannot leave this stage.
        if not una_clean:
            # Halt with the rendered found-hits table.
            return None, None, f"ValueError: prohibited sentinel(s) outside the module-import allow-lists detected in UNADJUSTED_BARS:\n{una_hits.to_string()}"
        # Run the scoped prohibited-sentinel scan on the adjusted view.
        adj_hits, adj_clean = task_2.scan_prohibited_sentinels_task_2(adj)
        # A non-clean frame cannot leave this stage.
        if not adj_clean:
            # Halt with the rendered found-hits table.
            return None, None, f"ValueError: prohibited sentinel(s) outside the module-import allow-lists detected in ADJUSTED_BARS:\n{adj_hits.to_string()}"
        # Enforce the anti-imputation post-condition. The three cumulative counters live in
        # the PIPELINE CONTEXT itself (Task 2/Task 5 Step 3; the counters live in the pipeline
        # context, not in any column), and this stage reads exactly the context the module's
        # context accessor exposes on THIS run -- never a fresh, always-zero side context.
        pipeline_context_counters: Mapping[str, Any] = run_context.counters
        # Name each dishonoured counter with its observed value.
        non_zero_counters: List[str] = [
            key
            for key in ("forward_fill_operations", "interpolation_operations", "stock_prefix_reductions")
            if int(pipeline_context_counters.get(key, 0) or 0) != 0
        ]
        # Any non-zero counter halts with the witness list from Task 2 (never a silent pass).
        if non_zero_counters:
            # Render the witness list recorded by the run-scoped Task 2 controller, when any.
            if getattr(anti_imputation_controller, "witness_log", None):
                # Join the witnesses exactly as Task 2's post-condition does.
                witness_list = "; ".join(
                    f"(operation={operation}, column={column}, stock_day={stock_day})"
                    for operation, column, stock_day in anti_imputation_controller.witness_log
                )
            else:
                # No witnesses recorded: disclose that state, never fabricate one.
                witness_list = "<no witnesses recorded>"
            # Halt with the violated counters named (both frames None via the except below).
            raise RuntimeError(
                f"Anti-imputation post-condition violated: non-zero counter(s) "
                f"{', '.join(non_zero_counters)}; witness list: {witness_list}"
            )
        # Enforce the anti-imputation post-condition against the RUN-SCOPED controller
        # (the same object every gated operation on this run would have written to).
        task_2.check_anti_imputation_task_2(anti_imputation_controller)
        # Enforce the index-preservation post-condition on the unadjusted view.
        if not una.index.equals(input_indexes["UNADJUSTED_BARS"]):
            # Raise the preservation breach.
            raise RuntimeError("cleanse_data: UNADJUSTED_BARS index was not preserved")
        # Enforce the index-preservation post-condition on the adjusted view.
        if not adj.index.equals(input_indexes["ADJUSTED_BARS"]):
            # Raise the preservation breach.
            raise RuntimeError("cleanse_data: ADJUSTED_BARS index was not preserved")
    except Exception as exc:
        # Any breach is recorded with the exception class and message verbatim.
        return None, None, _halt_reason(exc)
    # Cache the cleansed panels for the custody/identity/labels stages.
    context.payloads["cleansed_unadjusted"] = una
    # Cache the cleansed adjusted view for the labels stage.
    context.payloads["cleansed_adjusted"] = adj
    # Record the cleanse manifest for the audit trail.
    context.manifests["cleanse"] = {
        # The number of cast operations executed, in order.
        "cast_count": int(len(cast_ledger)),
        # The number of unique stock codes admitted.
        "stock_code_count": int(len(stock_codes)),
        # The anti-imputation counters, as observed on the run-scoped controller.
        "forward_fill_operations": int(anti_imputation_controller.forward_fill_operations),
        # The interpolation counter, as observed.
        "interpolation_operations": int(anti_imputation_controller.interpolation_operations),
        # The stock-prefix reduction counter, as observed.
        "stock_prefix_reductions": int(anti_imputation_controller.stock_prefix_reductions),
        # The cast-ledger scope (run-scoped shared handle, or a disclosed stage-local ledger).
        "cast_ledger_scope": cast_ledger_scope,
    }
    # Return the cleansed pair and no halt reason.
    return una, adj, None

def verify_custody(
    files: pd.DataFrame,
    cleansed_unadj: pd.DataFrame,
    cleansed_adj: pd.DataFrame,
    config: Dict[str, Any],
) -> Tuple[Optional[pd.DataFrame], Optional[str]]:
    """
    Verify acquisition custody of the FILES manifest by rehashing.

    Streams each ``file_uri`` to SHA-256 in 1 MiB chunks, then delegates
    the digest comparison, content-length checks and six conjunct
    verdicts to Task 3's Kleene fail-fast aggregate; the write-once
    ``hash_manifest.json`` lands beneath the run's nonce directory.
    Missing ``file_uri`` is re-derived on a deep copy with Task 1's
    sanctioned column helper, and the cleansed views reach Task 3 as
    ``RangeIndex`` substrate copies.

    Parameters
    ----------
    files : pd.DataFrame
        The FILES acquisition manifest.
    cleansed_unadj : pd.DataFrame
        Cleanse-stage unadjusted view (the custody substrate).
    cleansed_adj : pd.DataFrame
        Cleanse-stage adjusted view.
    config : Dict[str, Any]
        The STUDY_CONFIG dictionary; used only for the sanctioned
        column re-derivation.

    Returns
    -------
    Tuple[Optional[pd.DataFrame], Optional[str]]
        ``(files_with_verdicts, halt_reason)``; the reason is ``None``
        on success. Otherwise the frame is ``None`` and the reason is a
        hand-written ``TypeError:`` string when ``files`` is not a
        DataFrame or ``config`` is not a dict; the exception
        class/message of a caught failure (calendar reconstruction,
        sanctioned-column re-derivation, or Task 3); or
        ``MISSING_UPSTREAM_INPUT::custody`` when a cleansed view is
        missing.
    """
    # Boundary: the manifest must be a DataFrame.
    if not isinstance(files, pd.DataFrame):
        # Reject non-DataFrame manifests with the observed type.
        return None, f"TypeError: files must be a pandas.DataFrame, got {type(files).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # Upstream: custody requires the cleansed panels; a halted cleanse yields none.
    if cleansed_unadj is None or cleansed_adj is None:
        # Surface the missing input with the upstream attribution.
        return None, _missing_upstream("custody", ("cleansed_unadj", "cleansed_adj"), None)
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Resolve the calendar (built at validate time, or reconstruct it here).
    calendar: Optional[pd.DataFrame] = context.payloads.get("calendar")
    # Reconstruct the calendar if the validate stage never produced one.
    if calendar is None:
        # Rebuild from the cleansed view (a deterministic function of the acquisition).
        try:
            # The calendar consumes exactly the trade_date column.
            calendar = task_5.construct_calendar_task_5(cleansed_unadj.reset_index(drop=True))
        except Exception as exc:
            # A calendar that cannot be built is a custody halt.
            return None, _halt_reason(exc)
        # Cache the reconstruction for downstream stages.
        context.payloads["calendar"] = calendar
    # Task 1's four sanctioned FILES additions (file_uri, tag_registry_version, content_length_ok,
    # response_truncated) are what Task 3 rehashes against, but the pinned data flow returns only
    # Task 1's halt reason and hands custody the caller's raw manifest.  MEASURED 2026-11-12: a raw,
    # spec-conformant FILES halted every production run here ("FILES is missing required column(s)
    # ['file_uri']; run Task 1 ... before Task 3").  The additions are re-derived with Task 1's own
    # deterministic helper, on a copy (the caller's manifest is never mutated); a manifest that
    # already carries them is used as given.
    if "file_uri" not in files.columns:
        # Re-derive the sanctioned additions exactly as Task 1 Step 1 made them.
        try:
            # A deep copy, so the caller's frame is untouched.
            files = task_1._add_four_sanctioned_columns(files.copy(deep=True), config, [])
        except Exception as exc:
            # A manifest the additions cannot be made on is a custody halt.
            return None, _halt_reason(exc)
    # The write-once custody artifact lands beneath this run's nonce directory.
    manifest_path: Path = context.stage_dir("custody") / "hash_manifest.json"
    # The substrate panels must carry the pinned columns and no ambiguous axis name: a
    # cleansed panel may carry stock_code/trade_date as BOTH index and column (the
    # index-preservation contract); pandas groupby("stock_code") would then raise
    # "both an index level and a column label". Custody consumes a RangeIndex
    # substrate copy of identical values; the cleansed frames themselves are untouched.
    unadjusted_substrate: pd.DataFrame = cleansed_unadj.reset_index(drop=True)
    # Likewise for the adjusted view.
    adjusted_substrate: pd.DataFrame = cleansed_adj.reset_index(drop=True)
    # Execute Task 3 end to end; a failure propagates as the halt reason.
    try:
        # The three ordered steps run inside the orchestrator.
        files_with_verdicts, verdict_packet = task_3.run_task_3(
            FILES=files,
            unadjusted=unadjusted_substrate,
            adjusted=adjusted_substrate,
            calendar=calendar,
            manifest_path=str(manifest_path),
        )
    except Exception as exc:
        # Any custody failure is recorded with the class and message verbatim.
        return None, _halt_reason(exc)
    # Record the custody manifest for the audit trail.
    context.manifests["custody"] = {
        # The artifact path beneath the nonce directory.
        "hash_manifest_path": str(manifest_path),
        # The verdict packet row count.
        "verdict_packet_rows": int(len(verdict_packet)),
        # The verdict packet's conjunct columns.
        "verdict_packet_columns": list(verdict_packet.columns),
    }
    # Return the augmented manifest and no halt reason.
    return files_with_verdicts, None

def verify_identity(
    cleansed_unadj: pd.DataFrame,
    cleansed_adj: pd.DataFrame,
    files_with_verdicts: pd.DataFrame,
    config: Dict[str, Any],
) -> Tuple[Optional[Any], Optional[Any], Optional[Dict[str, Any]], Optional[str]]:
    """
    Verify cross-view identity and split features from labels.

    Type-check the inputs, then run Task 4 (view alignment,
    shared-field agreement, adjustment constancy, panel split) and
    re-read the persisted panels as plain frames with pinned dtypes.
    The non-trading stock-day cells are located by
    ``_non_trading_coordinates(cleansed_unadj, config)`` and
    corroborated against the raw vendor ``tradestatus_raw`` code in
    both views (1 -> tradable, 0 -> not, else NA); a contradiction
    halts before the mask register exists, and NA cells are disclosed
    as absence. The facades and the re-materialised plain frames are
    cached on the pipeline context; the coordinates, the fraud log,
    the pre-mask cardinalities and the ascending ``trade_date`` axis
    are returned in ``identity_report``.

    Parameters
    ----------
    cleansed_unadj : pd.DataFrame
        Cleansed unadjusted view; needs ``trade_date`` (column or
        index level) and ``tradable`` / ``tradestatus_raw``.
    cleansed_adj : pd.DataFrame
        Cleansed adjusted view, row-aligned with ``cleansed_unadj``.
    files_with_verdicts : pd.DataFrame
        Custody-augmented FILES manifest.
    config : Dict[str, Any]
        Study configuration; its mask pins drive the cell search.

    Returns
    -------
    Tuple[Optional[Any], Optional[Any], Optional[Dict[str, Any]], Optional[str]]
        ``(features_panel, label_panel, identity_report,
        halt_reason)``. The panels are Task 4's facades; the report
        holds the Task 4 diagnostics, both panel paths, the fraud log,
        both pre-mask cardinalities, ``masked_cells`` (``stock_code``,
        ``trade_date``) from ``_non_trading_coordinates``,
        ``trade_dates`` and ``stock_code_count``. The per-cell raw
        codes of both views (``unadjusted_tradestatus_raw`` /
        ``adjusted_tradestatus_raw``) are not in the report but in
        ``context.payloads["mask_status_corroboration"]``. All
        payloads are ``None`` on halt.
    """
    # Boundary: the unadjusted view must be a DataFrame.
    if not isinstance(cleansed_unadj, pd.DataFrame):
        # Reject non-DataFrame views with the observed type.
        return None, None, None, f"TypeError: cleansed_unadj must be a pandas.DataFrame, got {type(cleansed_unadj).__name__}"
    # Boundary: the adjusted view must be a DataFrame.
    if not isinstance(cleansed_adj, pd.DataFrame):
        # Reject non-DataFrame views with the observed type.
        return None, None, None, f"TypeError: cleansed_adj must be a pandas.DataFrame, got {type(cleansed_adj).__name__}"
    # Upstream: identity requires the custody output; a halted custody yields None.
    if files_with_verdicts is None:
        # Surface the missing input with the upstream attribution.
        return None, None, None, _missing_upstream("identity", ("files_with_verdicts",), None)
    # Boundary: the custody manifest must be a DataFrame.
    if not isinstance(files_with_verdicts, pd.DataFrame):
        # Reject non-DataFrame manifests with the observed type.
        return None, None, None, f"TypeError: files_with_verdicts must be a pandas.DataFrame, got {type(files_with_verdicts).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, None, None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # The two namespaces persist the persisted panels (distinct, non-nesting).
    features_root: Path = context.stage_dir("identity") / "features_panel"
    # The labels namespace is sibling to the features namespace.
    labels_root: Path = context.stage_dir("identity") / "labels_panel"
    # Execute Task 4 end to end; a failure propagates as the halt reason.
    try:
        # The three steps run inside the orchestrator (alignment -> shared fields -> split).
        bundle: Dict[str, Any] = task_4.run_task_4(
            FILES=files_with_verdicts,
            UNADJUSTED_BARS=cleansed_unadj,
            ADJUSTED_BARS=cleansed_adj,
            STUDY_CONFIG=config,
            expected_join_cardinality=int(len(cleansed_unadj)),
            features_root=str(features_root),
            labels_root=str(labels_root),
        )
    except Exception as exc:
        # Any identity failure is recorded with the class and message verbatim.
        return None, None, None, _halt_reason(exc)
    # Re-materialise the persisted panels as plain frames for the downstream tasks.
    try:
        # Read the persisted features CSV and re-apply the pinned dtypes.
        features_frame: pd.DataFrame = _read_panel_csv(str(bundle["features_path"]), kind="features", cast_ledger=context.payloads.get("cast_ledger"))
        # Read the persisted labels CSV and re-apply the pinned dtypes.
        labels_frame: pd.DataFrame = _read_panel_csv(str(bundle["labels_path"]), kind="labels", cast_ledger=context.payloads.get("cast_ledger"))
    except Exception as exc:
        # A defective persisted artifact is an identity halt.
        return None, None, None, _halt_reason(exc)
    # Locate the non-trading stock-day cells from the unadjusted view (the pins come from the configuration).
    masked_cells: List[Dict[str, Any]] = _non_trading_coordinates(cleansed_unadj, config)
    # Corroborate the located coordinates against the vendor status code (error-severity disclosure; never a pass).
    context.payloads["mask_status_corroboration"] = _corroborate_mask_coordinates(cleansed_unadj, cleansed_adj, masked_cells)
    # A located coordinate whose raw vendor code contradicts "not trading" in either view is a custody defect.
    contradicted: List[str] = [
        # Name the cell and both raw codes.
        f"({entry['stock_code']}, {entry['trade_date']}) unadjusted code={entry['unadjusted_tradestatus_raw']} "
        f"adjusted code={entry['adjusted_tradestatus_raw']}"
        # Walk the corroboration records.
        for entry in context.payloads["mask_status_corroboration"]
        # Keep only the contradicted coordinates (either view).
        if entry.get("verdict") == "disagreement" or entry.get("verdict_adjusted") == "disagreement"
    ]
    # Halt before Task 5 consumes the register (Blueprint 7 Step 3).
    if contradicted:
        # Record the custody halt with every offending cell named.
        return None, None, None, _halt_reason(
            ValueError(
                "mask_status_custody_defect: located mask coordinate(s) whose tradestatus_raw maps "
                "(pinned vendor mapping 1->True, 0->False, else NA) to tradable: " + "; ".join(contradicted)
            )
        )
    # Unverified (NA) tradable cells are disclosed as absence, never promoted to coordinates.
    context.payloads["mask_status_na_cells"] = (
        int(cleansed_unadj["tradable"].isna().sum()) if "tradable" in cleansed_unadj.columns else 0
    )
    # Derive the unique, ascending trade-date axis from the calendar source.
    trade_dates: pd.DatetimeIndex = pd.DatetimeIndex(
        pd.DatetimeIndex(cleansed_unadj["trade_date"] if "trade_date" in cleansed_unadj.columns else cleansed_unadj.index.get_level_values("trade_date"))
        .unique()
        .sort_values()
    )
    # The fraud log is carried verbatim (empty on a clean feature path).
    fraud_log: List[Any] = list(bundle.get("fraud_log", []))
    # Assemble the identity report (counts and diagnostics, never the frames).
    identity_report: Dict[str, Any] = {
        # The Step 1 alignment diagnostic.
        "task_4_alignment": bundle["alignment"],
        # The Step 2a shared-fields diagnostic.
        "task_4_shared_fields": bundle["shared_fields"],
        # The Step 2b adjustment-factor constancy diagnostic.
        "task_4_constancy": bundle["constancy"],
        # The persisted FEATURES_PANEL CSV path.
        "features_path": str(bundle["features_path"]),
        # The persisted LABEL_PANEL CSV path.
        "labels_path": str(bundle["labels_path"]),
        # The embargo audit trail (empty on a clean feature path).
        "fraud_log": fraud_log,
        # The pre-mask cardinality pinned by the stage.
        "cardinality_pre_mask": int(len(cleansed_unadj)),
        # The adjusted-view pre-mask cardinality.
        "cardinality_adjusted_pre_mask": int(len(cleansed_adj)),
        # The non-trading stock-day coordinates (feed the mask register).
        "masked_cells": masked_cells,
        # The unique trade-date axis (feeds the calendar cross-check).
        "trade_dates": trade_dates,
        # The projected code count (the 40-stock convenience universe).
        "stock_code_count": int(len(pd.Index(cleansed_unadj["stock_code"]).unique()) if "stock_code" in cleansed_unadj.columns else len(cleansed_unadj.index.get_level_values("stock_code").unique())),
    }
    # Cache the facades and the re-materialised frames for the downstream stages.
    context.payloads["features_facade"] = bundle["features_panel"]
    # Cache the label-side facade (embargo intact on the adjusted side by design).
    context.payloads["labels_facade"] = bundle["labels_panel"]
    # Cache the re-materialised features frame (Task 5 consumes plain frames).
    context.payloads["features_panel_plain"] = features_frame
    # Cache the re-materialised labels frame (Task 5 consumes plain frames).
    context.payloads["labels_panel_plain"] = labels_frame
    # Record the identity manifest for the audit trail.
    context.manifests["identity"] = {
        # The features namespace root.
        "features_root": str(features_root),
        # The labels namespace root.
        "labels_root": str(labels_root),
        # The pre-mask cardinality.
        "cardinality_pre_mask": int(len(cleansed_unadj)),
        # The number of masked coordinates.
        "masked_cell_count": int(len(masked_cells)),
        # The number of fraud-log entries (always zero on a clean path).
        "fraud_log_length": int(len(fraud_log)),
    }
    # Return the facades, the report and no halt reason.
    return bundle["features_panel"], bundle["labels_panel"], identity_report, None

def build_calendar_and_mask(
    identity_report: Mapping[str, Any],
    config: Dict[str, Any],
) -> Tuple[Optional[pd.DataFrame], Optional[pd.DataFrame], Optional[str]]:
    """
    Build the trading calendar and register the masked cells.

    Validate the identity report as a mapping carrying ``masked_cells``
    (a list), ``features_path``, ``labels_path`` and the optional
    ``trade_dates`` axis, then derive the P34 mask register from the
    recorded non-trading coordinates -- the register is always derived,
    never hand-typed. A calendar already cached by the validate stage
    in the context only suppresses the rebuild (the recorded axis is
    not cross-checked, and the cached calendar is not passed to Task
    5); a missing calendar is rebuilt from the recorded trade-date
    axis, falling back to the ``trade_date`` index level of the cached
    cleansed unadjusted panel. Either way, the calendar finally cached
    and returned is the one Task 5 produces.

    Task 5 then runs end to end on freshly re-read FEATURES and LABEL
    panels: the register is verified, the stock-day cells are nulled in
    both panels (NA kept, never zero), and the post-mask invariant
    battery is evaluated. The calendar, the masked and usable panels and
    the register are cached in the context payloads; the calendar
    manifest records the calendar hash, the battery verdicts, the
    usable-row count, the mask-witness path and the register row count.

    Parameters
    ----------
    identity_report : Mapping[str, Any]
        The identity stage's report (masked-cell coordinates, persisted
        panel paths, recorded trade-date axis).
    config : Dict[str, Any]
        The study configuration; its raw-layer invariants are consulted
        by the post-mask battery.

    Returns
    -------
    Tuple[Optional[pd.DataFrame], Optional[pd.DataFrame], Optional[str]]
        ``(calendar, mask_register, halt_reason)``. ``calendar`` carries
        ``row_index``, ``trade_date``, ``year``, ``in_label_window`` and
        ``label_available``; ``mask_register`` has one row per masked
        stock-day. Both are ``None`` on halt; ``halt_reason`` is
        ``None`` only on success.
    """
    # Upstream: the calendar stage requires the identity report.
    if identity_report is None:
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream("calendar", ("identity_report",), None)
    # Boundary: the report must be a mapping.
    if not isinstance(identity_report, Mapping):
        # Reject non-mapping reports with the observed type.
        return None, None, f"TypeError: identity_report must be a Mapping, got {type(identity_report).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Extract the masked-cell coordinates recorded by the identity stage.
    masked_cells: Optional[Any] = identity_report.get("masked_cells")
    # A report without coordinates cannot express the mask.
    if not isinstance(masked_cells, list):
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream("calendar", ("masked_cells",), None)
    # Build the P34-conform register from the coordinates.
    try:
        # The register is derived, never hand-typed.
        mask_register: pd.DataFrame = _mask_register_from_cells(masked_cells)
    except Exception as exc:
        # A malformed coordinate ledger halts this stage.
        return None, None, _halt_reason(exc)
    # Resolve the calendar: reuse the validate-time one, cross-checking the axis.
    calendar: Optional[pd.DataFrame] = context.payloads.get("calendar")
    # The recorded trade-date axis from the identity stage.
    recorded_dates: Optional[pd.DatetimeIndex] = None
    # The report carries the unique ascending axis when the identity stage ran.
    axis = identity_report.get("trade_dates")
    # Coerce the recorded axis when present.
    if isinstance(axis, pd.DatetimeIndex) and len(axis):
        # Keep the recorded axis for the cross-check.
        recorded_dates = axis
    # A missing cached calendar is rebuilt from the recorded axis or the cleansed view.
    if calendar is None:
        # Rebuild from the recorded axis when available.
        source_dates = recorded_dates if recorded_dates is not None else None
        # Fall back to the cleansed panel's key axis.
        if source_dates is None:
            # The cleansed panel is the only remaining honest source.
            fallback = context.payloads.get("cleansed_unadjusted")
            # The fallback must be present.
            if fallback is None:
                # Surface the missing input with the upstream attribution.
                return None, None, _missing_upstream("calendar", ("calendar",), None)
            # Derive the unique axis from the panel index.
            source_dates = pd.DatetimeIndex(fallback.index.get_level_values("trade_date")).unique().sort_values()
        # Construct the dense calendar from the axis.
        try:
            # The calendar consumes exactly the trade_date column.
            calendar = task_5.construct_calendar_task_5(pd.DataFrame({"trade_date": pd.DatetimeIndex(source_dates)}))
        except Exception as exc:
            # A calendar that cannot be built is a calendar-stage halt.
            return None, None, _halt_reason(exc)
        # Cache the rebuilt calendar.
        context.payloads["calendar"] = calendar
    # The persisted panel paths anchor the mask application.
    features_path: Optional[str] = identity_report.get("features_path")
    # The labels namespace path anchors the mask application.
    labels_path: Optional[str] = identity_report.get("labels_path")
    # Both paths must be present.
    if not isinstance(features_path, str) or not isinstance(labels_path, str):
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream("calendar", ("features_path", "labels_path"), None)
    # The cleansed unadjusted view drives Task 5's calendar cross-check inside.
    unadjusted_bars: Optional[pd.DataFrame] = context.payloads.get("cleansed_unadjusted")
    # The view must be present in the context.
    if unadjusted_bars is None:
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream("calendar", ("cleansed_unadjusted",), None)
    # Re-materialise the persisted panels and apply the mask end to end.
    try:
        # Read the persisted features panel.
        features_panel: pd.DataFrame = _read_panel_csv(features_path, kind="features", cast_ledger=context.payloads.get("cast_ledger"))
        # Read the persisted labels panel.
        labels_panel: pd.DataFrame = _read_panel_csv(labels_path, kind="labels", cast_ledger=context.payloads.get("cast_ledger"))
        # Task 5 runs its pinned three steps (counts -> mask -> battery).
        stage5: Dict[str, Any] = task_5.run_task_5(
            unadjusted_bars=unadjusted_bars,
            features_panel=features_panel,
            label_panel=labels_panel,
            mask_register=mask_register,
            study_config=config,
            context=context.task_2_context,
            output_directory=context.stage_dir("calendar"),
        )
    except Exception as exc:
        # Any Task 5 failure is recorded with the class and message verbatim.
        return None, None, _halt_reason(exc)
    # A recorded stage halt inside Task 5 is this stage's halt.
    if stage5.get("halt_reason") is not None:
        # Surface the stage-namespaced halt verbatim.
        return None, None, str(stage5["halt_reason"])
    # Bind the products of the battery.
    calendar_out: pd.DataFrame = stage5["calendar"]
    # Cache the masked/usable panels for the scoring and reference stages.
    context.payloads["calendar"] = calendar_out
    # Cache the masked features panel (post-mask, NA-kept).
    context.payloads["masked_features_panel"] = stage5["masked_features_panel"]
    # Cache the masked labels panel (post-mask, NA-kept).
    context.payloads["masked_label_panel"] = stage5["masked_label_panel"]
    # Cache the usable (exclude the two masked tuples) features panel.
    context.payloads["usable_features_panel"] = stage5["usable_features_panel"]
    # Cache the usable labels panel.
    context.payloads["usable_label_panel"] = stage5["usable_label_panel"]
    # Cache the register for the scoring/labels stages.
    context.payloads["mask_register"] = mask_register
    # Record the calendar/mask manifest for the audit trail.
    context.manifests["calendar"] = {
        # The calendar hash stamped by Task 5 (SHA-256 of the persisted bytes).
        "calendar_hash": stage5.get("calendar_hash"),
        # The post-mask invariant battery verdict payload.
        "post_mask_battery": stage5.get("post_mask_battery"),
        # The usable-rows cardinality (29,078 on the pinned study).
        "usable_rows": int(len(stage5["usable_features_panel"])),
        # The mask witness artifact path persisted by Task 5.
        "mask_witness_path": str(context.stage_dir("calendar") / "mask_application_witness.csv"),
        # The number of rows in the mask register (pinned 2).
        "mask_register_rows": int(len(mask_register)),
    }
    # Return the calendar, the register and no halt reason.
    return calendar_out, mask_register, None

def build_labels(
    cleansed_adj: pd.DataFrame,
    calendar: pd.DataFrame,
    mask_register: pd.DataFrame,
    config: Dict[str, Any],
) -> Tuple[Optional[pd.DataFrame], Optional[str]]:
    """
    Build the five-row-ahead label on the adjusted close.

    Guard the upstream frames and configuration, apply the NA-keep
    mask to a copy of the adjusted view, attach the OOS sentinels
    (``oos_sharpe`` all-NaN float64, ``oos_computed_at`` all-NaT
    datetime64[ns]) when absent, and window the panel to the calendar
    rows flagged ``in_label_window``. ``stock_code`` / ``trade_date``
    move into the compound index so Task 6's ``reset_index`` cannot
    collide. Task 6 then constructs

        y_t = P_adj[t + 5] / P_adj[t] - 1        Eq. (1)

    over trading-calendar rows, applies the endpoint admissibility
    predicate (both endpoints finite, positive and tradable, so the
    last five captured dates have no label) and runs the NaN-hygiene
    battery against the configured label-available-date count.

    Parameters
    ----------
    cleansed_adj : pd.DataFrame
        Cleansed adjusted view; the only view labels may read.
    calendar : pd.DataFrame
        Dense calendar with ``trade_date``, ``in_label_window`` and
        ``label_available``.
    mask_register : pd.DataFrame
        P34 mask register; its cells are set to NA, never zero.
    config : Dict[str, Any]
        Study config; ``P01_LABEL_EQUATION_PARAMETERS`` gives
        ``label_available_rows``.

    Returns
    -------
    Tuple[Optional[pd.DataFrame], Optional[str]]
        ``(label_panel, halt_reason)``. ``label_panel`` is the
        post-mask Eq. (1) frame indexed by ``(stock_code,
        trade_date)`` with column ``y``; it is ``None`` on halt.
    """
    # Upstream: labels require the cleansed adjusted view.
    if cleansed_adj is None:
        # Surface the missing input with the upstream attribution.
        return None, _missing_upstream("labels", ("cleansed_adj",), None)
    # Upstream: labels require the calendar.
    if calendar is None:
        # Surface the missing input with the upstream attribution.
        return None, _missing_upstream("labels", ("calendar",), None)
    # Upstream: labels require the mask register.
    if mask_register is None:
        # Surface the missing input with the upstream attribution.
        return None, _missing_upstream("labels", ("mask_register",), None)
    # Boundary: the adjusted view must be a DataFrame.
    if not isinstance(cleansed_adj, pd.DataFrame):
        # Reject non-DataFrame views with the observed type.
        return None, f"TypeError: cleansed_adj must be a pandas.DataFrame, got {type(cleansed_adj).__name__}"
    # Boundary: the calendar must be a DataFrame.
    if not isinstance(calendar, pd.DataFrame):
        # Reject non-DataFrame calendars with the observed type.
        return None, f"TypeError: calendar must be a pandas.DataFrame, got {type(calendar).__name__}"
    # Boundary: the register must be a DataFrame.
    if not isinstance(mask_register, pd.DataFrame):
        # Reject non-DataFrame registers with the observed type.
        return None, f"TypeError: mask_register must be a pandas.DataFrame, got {type(mask_register).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # NA-keep copy of the adjusted view at the masked coordinates.
        masked: pd.DataFrame = _apply_mask_to_adjusted(cleansed_adj, mask_register)
        # Attach the OOS leak sentinels (pre-freeze nulls, P22/P27).
        if "oos_sharpe" not in masked.columns:
            # The sentinel is all-NaN float64.
            masked["oos_sharpe"] = pd.Series(np.nan, index=masked.index, dtype="float64")
        # Attach the OOS-computation sentinel when absent.
        if "oos_computed_at" not in masked.columns:
            # The sentinel is all-NaT datetime64[ns].
            masked["oos_computed_at"] = pd.Series(pd.NaT, index=masked.index).astype("datetime64[ns]")
        # Resolve the label-window axis (2024-2025 rows of the calendar).
        window_dates: pd.DatetimeIndex = pd.DatetimeIndex(pd.to_datetime(calendar.loc[calendar["in_label_window"], "trade_date"]))
        # The panel's trade_date axis (key columns or compound index).
        axis_dates = masked["trade_date"] if "trade_date" in masked.columns else masked.index.get_level_values("trade_date")
        # Keep exactly the label-window rows (Eq. (1) is pinned to 2024-2025).
        windowed: pd.DataFrame = masked[axis_dates.isin(window_dates)].copy()
        # Drop the coordinate columns into the compound index (Task 6 contract).
        if {"stock_code", "trade_date"}.issubset(windowed.columns):
            # Set the index with drop=True so Task 6's reset_index does not collide.
            windowed = windowed.set_index(["stock_code", "trade_date"], drop=True)
        # Read the pinned expected label-available-dates figure from the configuration.
        expected_dates: int = int(config["P01_LABEL_EQUATION_PARAMETERS"]["label_available_rows"])
        # Execute Task 6 end to end; Eq. (1) is implemented there, never here.
        bundle: Dict[str, Any] = task_6.run_task_6(
            label_panel=windowed,
            calendar=calendar,
            expected_label_available_dates=expected_dates,
        )
    except Exception as exc:
        # Any Task 6 failure is recorded with the class and message verbatim.
        return None, _halt_reason(exc)
    # Bind the Eq. (1) labels frame.
    eq1_labels: pd.DataFrame = bundle["LABELS"]
    # Cache the labels for the scoring, reference and correction stages.
    context.payloads["eq1_labels"] = eq1_labels
    # Cache the label Series (y) aligned on the compound index.
    context.payloads["labels_y"] = eq1_labels["y"]
    # The trade dates eligible for scoring (label_available rows, ascending).
    context.manifests["labels"] = {
        # The number of label-available calendar dates (pinned 480).
        "label_available_dates": int(len(bundle["label_available_dates"])),
        # The number of label-window panel rows (40 * 485 on the study snapshot).
        "label_window_rows": int(len(eq1_labels)),
        # The total non-NaN label count across stocks (disclosure, not assertion).
        "n_available_labels_total": int(sum(bundle.get("n_available_labels", {}).values())),
        # The withdrawn raw-task test tag published by Task 6.
        "withdrawals": list(bundle.get("withdrawals", [])),
    }
    # Return the Eq. (1) labels and no halt reason.
    return eq1_labels, None

def build_dsl_and_evaluator(
    config: Dict[str, Any],
) -> Tuple[Optional[PanelEvaluator], Optional[PanelEvaluator], Optional[Dict[str, Any]], Optional[str]]:
    """
    Build both registries, their evaluators and the canonicalizer.

    Build the operator registry twice, each stamped with its
    SHA-256: ``v1`` binds ``if`` to Task 7's reconstructed defect
    emulator (a panel read as a single Boolean), ``v2`` to the
    scalar-broadcasting elementwise conditional preserving the ``v1``
    semantics. Each is wrapped in a :class:`PanelEvaluator` with its
    version literal, version tag, pinned operator name, wiring
    disclosure and digest. The canonicalizer handle carries Task 7's
    ``parse``, ``canonical_bytes`` and ``expression_id`` certificates,
    the ``v1`` digest and the ``[R]`` provenance tag.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration; only type-checked here.

    Returns
    -------
    evaluator_v1 : Optional[PanelEvaluator]
        The ``v1`` handle, ``None`` on halt.
    evaluator_v2 : Optional[PanelEvaluator]
        The ``v2`` handle, ``None`` on halt.
    canonicalizer : Optional[Dict[str, Any]]
        Keys ``parse``, ``canonical_bytes``, ``expression_id``,
        ``registry_sha256``, ``deduplication_canonicalization_source``;
        ``None`` on halt.
    halt_reason : Optional[str]
        Class-and-message halt text, or ``None`` on success.
    """
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, None, None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # Build the v1 registry (conditional bound to the defect emulator).
        registry_v1: Mapping[str, Any] = _registry_with_conditional("v1")
        # Build the v2 registry (conditional bound to scalar broadcasting).
        registry_v2: Mapping[str, Any] = _registry_with_conditional("v2")
        # Stamp the v1 registry digest.
        sha_v1: str = task_7.registry_sha256_task_7(registry_v1)
        # Stamp the v2 registry digest.
        sha_v2: str = task_7.registry_sha256_task_7(registry_v2)
        # The v1 evaluator handle (Task 9 semantics).
        evaluator_v1: PanelEvaluator = PanelEvaluator(
            # The version literal.
            version="v1",
            # The tag recorded verbatim in every register row.
            version_tag="task14_dsl_evaluator_v1",
            # The pinned conditional operator name.
            conditional_operator="if",
            # The disclosure of the v1 conditional wiring.
            conditional_implementation="v1_emulator_task_7__panel_read_as_single_boolean",
            # The v1 registry.
            registry=registry_v1,
            # The v1 registry digest.
            registry_sha256=sha_v1,
        )
        # The v2 evaluator handle (Task 13 semantics; scalar broadcasting).
        evaluator_v2: PanelEvaluator = PanelEvaluator(
            # The version literal.
            version="v2",
            # The tag recorded verbatim in every register row.
            version_tag="task14_dsl_evaluator_v2__scalar_broadcasting",
            # The pinned conditional operator name.
            conditional_operator="if",
            # The disclosure of the v2 conditional wiring.
            conditional_implementation="elementwise_if_task_7__scalar_branch_broadcasting",
            # The v2 registry.
            registry=registry_v2,
            # The v2 registry digest.
            registry_sha256=sha_v2,
        )
        # The canonicalizer handle: Task 7's certificates, by name.
        canonicalizer: Dict[str, Any] = {
            # The expression parser.
            "parse": task_7.parse_expression_task_7,
            # The canonical AST bytes certificate (SHA-256 substrate).
            "canonical_bytes": task_7.canonical_bytes_task_7,
            # The canonical AST identity certificate (dedup key).
            "expression_id": task_7.expression_id_task_7,
            # The v1 registry digest (the evaluator substrate identity).
            "registry_sha256": sha_v1,
            # The provenance tag of the canonicalization scheme.
            "deduplication_canonicalization_source": "NOT_STATED_IN_PAPER__[R]",
        }
    except Exception as exc:
        # Any construction failure is recorded with the class and message verbatim.
        return None, None, None, _halt_reason(exc)
    # Cache the evaluators and the canonicalizer for the scoring/correction stages.
    context.payloads["evaluator_v1"] = evaluator_v1
    # Cache the v2 evaluator.
    context.payloads["evaluator_v2"] = evaluator_v2
    # Cache the canonicalizer.
    context.payloads["evaluator_canonicalizer"] = canonicalizer
    # Record the DSL manifest for the audit trail.
    context.manifests["dsl"] = {
        # The v1 registry digest.
        "evaluator_v1_registry_sha256": evaluator_v1.registry_sha256,
        # The v2 registry digest.
        "evaluator_v2_registry_sha256": evaluator_v2.registry_sha256,
        # The v1 conditional disclosure.
        "evaluator_v1_conditional": evaluator_v1.conditional_implementation,
        # The v2 conditional disclosure.
        "evaluator_v2_conditional": evaluator_v2.conditional_implementation,
    }
    # Return the evaluators, the canonicalizer and no halt reason.
    return evaluator_v1, evaluator_v2, canonicalizer, None

def freeze_cohort(
    preserved_rows: pd.DataFrame,
    evaluator_canonicalizer: Mapping[str, Any],
    config: Dict[str, Any],
    oos_read_timestamp: datetime.datetime,
) -> Tuple[Optional[Any], Optional[str]]:
    """
    Screen, deduplicate and freeze the cohort write-once.

    Validate the frame, the dict configuration, the timezone-aware
    OOS-read instant and the canonicalizer probe before any byte is
    sealed. Task 8's three steps run by import, never re-written: the
    IS-only screen (NaN fails the comparators), the canonical-AST
    deduplication (lowest ``row_id`` representative) and the
    write-once freeze beneath the ``freeze`` stage directory, which
    emits the ordered identifiers, canonical bytes and the SHA-256.

    The result is guarded since Task 8 gives no verification
    guarantee: identifier count must equal ``meta["cohort_size"]``,
    ``meta["cohort_hash"]`` must be 64 characters and
    ``meta["freeze_timestamp"]`` a datetime. Cohort, rows and
    identifiers are cached, the freeze state mirrored and the
    freeze manifest recorded.

    Parameters
    ----------
    preserved_rows : pd.DataFrame
        The preserved input rows (3,114 on the pinned study).
    evaluator_canonicalizer : Mapping[str, Any]
        The canonicalizer handle from the DSL stage.
    config : Dict[str, Any]
        The study configuration; carried, not read.
    oos_read_timestamp : datetime.datetime
        Timezone-aware first OOS read; must post-date the freeze.

    Returns
    -------
    Tuple[Optional[Any], Optional[str]]
        ``(frozen_cohort, halt_reason)``. ``frozen_cohort`` is Task 8's
        container exposing ``meta`` (``cohort_size``, ``cohort_hash``,
        ``freeze_id``, ``freeze_timestamp``, ``oos_used_in_admission``)
        and the ordered expressions; ``None`` on halt.

    Raises
    ------
    ValueError
        Raised by the internal output guard on a size, hash or
        timestamp mismatch; caught and returned as the halt reason.

    Notes
    -----
    No OOS column is read; every write goes to a write-once,
    nonce-namespaced directory.
    """
    # Boundary: the preserved rows must be a DataFrame.
    if not isinstance(preserved_rows, pd.DataFrame):
        # Reject non-DataFrame inputs with the observed type.
        return None, f"TypeError: preserved_rows must be a pandas.DataFrame, got {type(preserved_rows).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # Boundary: the OOS-read timestamp must be a datetime.
    if not isinstance(oos_read_timestamp, datetime.datetime):
        # Reject non-datetime instants with the observed type.
        return None, f"TypeError: oos_read_timestamp must be a datetime.datetime, got {type(oos_read_timestamp).__name__}"
    # Boundary: the OOS-read timestamp must be timezone-aware.
    if oos_read_timestamp.tzinfo is None or oos_read_timestamp.utcoffset() is None:
        # A naive instant cannot be ordered against the freeze timestamp.
        return None, "ValueError: oos_read_timestamp must be timezone-aware (Asia/Shanghai), got a naive datetime"
    # The canonicalizer must survive the interface-and-probe verification.
    if not _canonicalizer_ok(evaluator_canonicalizer):
        # Refuse an unusable canonicalizer before any artifact is sealed.
        return None, "ValueError: evaluator_canonicalizer failed the canonicalizer probe (canonical_bytes / expression_id)"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # Step 1: the IS-only screen (float64 as arrived; NaN fails comparators).
        screened: Dict[str, Any] = task_8.screen_is_only_task_8(preserved_rows)
        # Step 2: canonical AST deduplication with the lowest row_id representative.
        deduplicated: Dict[str, Any] = task_8.deduplicate_canonical_task_8(screened["rows"])
        # Step 3: the write-once freeze (ordered IDs, canonical bytes, SHA-256).
        frozen: Any = task_8.freeze_cohort_task_8(
            deduplicated["rows"],
            oos_read_start_ts=oos_read_timestamp,
            output_dir=context.stage_dir("freeze"),
        )
    except Exception as exc:
        # Any Task 8 failure is recorded with the class and message verbatim.
        return None, _halt_reason(exc)
    # Guard the unverified upstream's output shape before any use (Task 8 has no verification guarantee).
    try:
        # The meta must be a mapping with the pinned certificates.
        meta: Mapping[str, Any] = frozen.meta
        # Order by the cohort order and extract the ordered identifier list.
        ordered_ids: List[str] = _frozen_expression_ids(frozen)
        # The ordered length must equal the meta cohort size.
        cohort_rows_len: int = len(ordered_ids)
        # The ordered length must equal the meta cohort size.
        cohort_size: int = int(meta["cohort_size"])
        # The digest must be 64 lowercase hex characters.
        cohort_hash: str = str(meta["cohort_hash"])
        # The timestamp must be a datetime.
        freeze_timestamp = meta["freeze_timestamp"]
        # Run the guard assertions with informative errors.
        if cohort_rows_len != cohort_size:
            # The rows and the meta disagree on the cohort size.
            raise ValueError(f"frozen cohort rows ({cohort_rows_len}) disagree with meta cohort_size ({cohort_size})")
        if len(cohort_hash) != 64:
            # The digest is malformed.
            raise ValueError(f"frozen cohort hash is not 64 characters: {cohort_hash!r}")
        if not isinstance(freeze_timestamp, datetime.datetime):
            # The stamp is not a datetime.
            raise ValueError(f"freeze_timestamp must be a datetime, got {type(freeze_timestamp).__name__}")
    except Exception as exc:
        # Any breach of the unverified upstream contract halts here.
        return None, f"ValueError: frozen cohort failed the task-8 output guard: {exc}"
    # Cache the frozen cohort and its provenance for the scoring stages.
    context.payloads["frozen_cohort"] = frozen
    # Cache the preserved rows (required by the aggregation support bundle and correction).
    context.payloads["preserved_rows"] = preserved_rows
    # Cache the ordered identifier list (cohort order).
    context.payloads["frozen_expression_ids"] = tuple(ordered_ids)
    # Mirror the freeze state (hash, timestamp, id, size) for the report.
    context.freeze_state["meta"] = dict(meta)
    # Mirror the digest.
    context.freeze_state["cohort_hash"] = cohort_hash
    # Mirror the freeze id.
    context.freeze_state["freeze_id"] = str(meta.get("freeze_id", ""))
    # Mirror the cohort size.
    context.freeze_state["cohort_size"] = cohort_size
    # Mirror the freeze timestamp.
    context.freeze_state["freeze_timestamp"] = freeze_timestamp
    # Record the freeze manifest for the audit trail.
    context.manifests["freeze"] = {
        # The post-screen count.
        "n_after_I_screen": int(screened["n_after_I_screen"]),
        # The post-dedup count.
        "n_after_dedup": int(deduplicated["n_after_dedup"]),
        # The dedup disclosure.
        "dedup_status": str(deduplicated.get("dedup_status", "")),
        # The frozen cohort hash.
        "cohort_hash": cohort_hash,
        # The freeze id.
        "freeze_id": str(meta.get("freeze_id", "")),
        # The cohort size.
        "cohort_size": cohort_size,
        # The OOS-used-in-admission flag (always False).
        "oos_used_in_admission": bool(meta.get("oos_used_in_admission", False)),
    }
    # Return the frozen cohort and no halt reason.
    return frozen, None

def score_daily_rank_ic(
    features_panel: pd.DataFrame,
    label_panel: pd.DataFrame,
    calendar: pd.DataFrame,
    mask_register: pd.DataFrame,
    evaluator_v1: PanelEvaluator,
    frozen_cohort: Any,
    config: Dict[str, Any],
) -> Tuple[Optional[pd.DataFrame], Optional[str]]:
    """
    Score the frozen cohort into the daily rank-IC register.

    Evaluate each frozen expression on the post-mask nominal features
    panel through ``evaluator_v1.registry`` and the Task 7 parser, then
    correlate average-tie mid-ranks with the Eq. (1) labels per date via
    ``task_9.run_task_9``. Dates are the ascending calendar rows with
    ``label_available == True``; each needs five eligible stocks, three
    distinct signal values and two distinct labels. The register is
    written to ``DAILY_SCORES.csv`` under the ``score`` stage directory
    and cached with the factor panel in the run context.

    Parameters
    ----------
    features_panel : pd.DataFrame
        Post-mask nominal features panel with ``tradable``.
    label_panel : pd.DataFrame
        Post-mask Eq. (1) labels frame with ``y``.
    calendar : pd.DataFrame
        Dense calendar with a boolean ``label_available`` column.
    mask_register : pd.DataFrame
        The P34 mask register.
    evaluator_v1 : PanelEvaluator
        The v1 evaluator; supplies the registry and ``version_tag``.
    frozen_cohort : Any
        Frozen cohort; supplies the ordered identifiers.
    config : Dict[str, Any]
        Study configuration; carried unchanged.

    Returns
    -------
    Tuple[Optional[pd.DataFrame], Optional[str]]
        ``(register, halt_reason)``; one row per expression and date,
        or ``None`` on halt.

    Raises
    ------
    ValueError
        If ``label_panel`` lacks ``y``; caught and returned as the
        halt reason, not propagated.
    """
    # Boundary: the features panel must be a DataFrame.
    if not isinstance(features_panel, pd.DataFrame):
        # Reject non-DataFrame panels with the observed type.
        return None, f"TypeError: features_panel must be a pandas.DataFrame, got {type(features_panel).__name__}"
    # Boundary: the labels frame must be a DataFrame.
    if not isinstance(label_panel, pd.DataFrame):
        # Reject non-DataFrame labels with the observed type.
        return None, f"TypeError: label_panel must be a pandas.DataFrame, got {type(label_panel).__name__}"
    # Boundary: the calendar must be a DataFrame.
    if not isinstance(calendar, pd.DataFrame):
        # Reject non-DataFrame calendars with the observed type.
        return None, f"TypeError: calendar must be a pandas.DataFrame, got {type(calendar).__name__}"
    # Boundary: the register must be a DataFrame.
    if not isinstance(mask_register, pd.DataFrame):
        # Reject non-DataFrame registers with the observed type.
        return None, f"TypeError: mask_register must be a pandas.DataFrame, got {type(mask_register).__name__}"
    # Boundary: the evaluator must be the pinned handle.
    if not isinstance(evaluator_v1, PanelEvaluator):
        # Reject foreign evaluator objects with the observed type.
        return None, f"TypeError: evaluator_v1 must be a PanelEvaluator, got {type(evaluator_v1).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # Upstream: scoring requires the frozen cohort; a halted freeze yields None.
    if frozen_cohort is None:
        # Surface the missing input with the upstream attribution.
        return None, _missing_upstream("score", ("frozen_cohort",), None)
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # The ordered expression identifiers (major axis of the register).
        expression_ids: List[str] = _frozen_expression_ids(frozen_cohort)
        # The preserved rows supply the expression texts (stashed by the freeze stage).
        preserved: Optional[pd.DataFrame] = context.payloads.get("preserved_rows")
        # The texts map expression_id to source text.
        texts: Dict[str, str] = (
            dict(zip(preserved["expression_id"].astype(str), preserved["expression_text"].astype(str)))
            if isinstance(preserved, pd.DataFrame) and {"expression_id", "expression_text"}.issubset(preserved.columns)
            else {}
        )
        # Evaluate the frozen expressions on the masked features panel.
        factor_panel, parse_failed = _build_factor_panel(
            features_panel=features_panel,
            texts_ids=expression_ids,
            texts=texts,
            registry=evaluator_v1.registry,
            parse=task_7.parse_expression_task_7,
        )
        # The Eq. (1) label Series rides on the same compound index.
        if "y" not in label_panel.columns:
            # The labels frame must carry y (Eq. (1)); refuse otherwise.
            raise ValueError("label_panel is missing the 'y' column produced by the labels stage")
        labels_y: pd.Series = label_panel["y"]
        # The strict-True tradability flag from the masked features view.
        tradable_series: pd.Series = features_panel["tradable"]
        # The ordered label-available trade dates (pinned 480, ascending).
        trade_dates: List[pd.Timestamp] = list(pd.to_datetime(calendar.loc[calendar["label_available"] == True, "trade_date"]))  # noqa: E712
        # Execute Task 9 end to end; the register is built by the very callable it pins.
        register: pd.DataFrame = task_9.run_task_9(
            factor_panel=factor_panel,
            labels=labels_y,
            tradable=tradable_series,
            mask_register=mask_register,
            expression_ids=expression_ids,
            trade_dates=trade_dates,
            evaluator_version=str(evaluator_v1.version_tag),
            output_dir=context.stage_dir("score"),
            parse_failed_ids=tuple(parse_failed),
        )
    except Exception as exc:
        # Any Task 9 failure is recorded with the class and message verbatim.
        return None, _halt_reason(exc)
    # Cache the factor panel for the dual check and correction stages.
    context.payloads["factor_panel"] = factor_panel
    # Cache the Eq. (1) label Series.
    context.payloads["labels_y"] = labels_y
    # Cache the tradability Series.
    context.payloads["tradable_series"] = tradable_series
    # Cache the register.
    context.payloads["daily_scores_v1"] = register
    # Record the score manifest for the audit trail.
    context.manifests["score"] = {
        # The evaluator version tag.
        "evaluator_version": str(evaluator_v1.version_tag),
        # The number of register rows (cohort * 480 on the study snapshot).
        "n_cells": int(len(register)),
        # The identifiers recorded parse_failed (never silently dropped).
        "parse_failed_ids": list(parse_failed),
        # The number of label-available dates scored.
        "n_trade_dates": int(len(trade_dates)),
    }
    # Return the register and no halt reason.
    return register, None

def aggregate_and_ledger(
    daily_scores_v1: pd.DataFrame,
    frozen_cohort: Any,
    config: Dict[str, Any],
    precedence: Sequence[str],
) -> Tuple[Optional[Tuple[int, int, int]], Optional[pd.DataFrame], Optional[str]]:
    """
    Aggregate the daily register and build both three-state ledgers.

    Aggregate ``daily_scores_v1`` to one row per frozen expression in
    cohort order, assign ledger states under ``precedence`` and its
    reverse, and run the Task 10 closure battery, whose exact identity
    is ``n_scorable + n_insufficient + n_unsupported == n_cohort``.
    Unsupported detection uses a one-row-per-expression support bundle
    from the ``preserved_rows`` payload (``operators_used``,
    ``parse_status``) checked against this run's operator registry.
    Nothing is imputed and the paper's 923/99/25 triple is never
    asserted. ``FACTOR_AGGREGATES.csv`` is written under the
    ``aggregate`` stage directory.

    Parameters
    ----------
    daily_scores_v1 : pd.DataFrame
        The v1 DAILY_SCORES register from the scoring stage.
    frozen_cohort : Any
        Frozen cohort; supplies the ordered identifiers and size.
    config : Dict[str, Any]
        Study configuration; ``P03_FACTOR_AGGREGATION_PARAMETERS``
        supplies the coverage and scorability thresholds.
    precedence : Sequence[str]
        Primary state precedence; a non-string permutation of
        ``"scorable"``, ``"insufficient"`` and ``"unsupported"``.

    Returns
    -------
    ledger_v1 : Optional[Tuple[int, int, int]]
        (scorable, insufficient, unsupported) counts, or ``None`` on
        halt.
    factor_aggregates_v1 : Optional[pd.DataFrame]
        One row per expression in cohort order, or ``None`` on halt.
    halt_reason : Optional[str]
        ``None`` on success, else the exception class and message.

    Raises
    ------
    ValueError
        If preserved rows are unavailable; caught and returned as the
        halt reason.
    """
    # Upstream: aggregation requires the register.
    if daily_scores_v1 is None:
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream(stage="aggregate", missing=("daily_scores_v1",), halted_at=None)
    # Upstream: aggregation requires the frozen cohort.
    if frozen_cohort is None:
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream("aggregate", ("frozen_cohort",), None)
    # Boundary: the register must be a DataFrame.
    if not isinstance(daily_scores_v1, pd.DataFrame):
        # Reject non-DataFrame registers with the observed type.
        return None, None, f"TypeError: daily_scores_v1 must be a pandas.DataFrame, got {type(daily_scores_v1).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # Boundary: precedence must be a non-string sequence of strings.
    if isinstance(precedence, (str, bytes)) or not isinstance(precedence, Sequence):
        # A bare string would iterate into characters.
        return None, None, f"TypeError: precedence must be a Sequence[str], got {type(precedence).__name__}"
    # Freeze the precedence triple as plain strings.
    triple: Tuple[str, ...] = tuple(str(state) for state in precedence)
    # The three pinned states are the complete registry.
    pinned_states: set = {"scorable", "insufficient", "unsupported"}
    # The triple must be a permutation of the pinned states.
    if len(triple) != 3 or set(triple) != pinned_states:
        # Reject malformed precedences with the observed value.
        return None, None, f"ValueError: precedence must be a permutation of ['scorable', 'insufficient', 'unsupported'], got {triple}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # The ordered identifier list (the cohort order is authoritative).
        expression_ids: List[str] = _frozen_expression_ids(frozen_cohort)
        # The support bundle requires the preserved rows stashed by the freeze stage.
        preserved: Optional[pd.DataFrame] = context.payloads.get("preserved_rows")
        # The bundle is mandatory for unsupported detection.
        if not isinstance(preserved, pd.DataFrame):
            # Without preserved rows no support bundle can be built honestly.
            raise ValueError("preserved rows unavailable; the support bundle cannot be built")
        # Restrict to the cohort identifiers and keep one row per identifier.
        support_base: pd.DataFrame = preserved[preserved["expression_id"].astype(str).isin(set(expression_ids))]
        # One row per identifier, in cohort order (drop_duplicates keeps first).
        support: pd.DataFrame = (
            support_base.assign(expression_id=support_base["expression_id"].astype(str))
            .drop_duplicates(subset="expression_id")
            .set_index("expression_id")
            .loc[expression_ids]
            .reset_index()
        )[["expression_id", "operators_used", "parse_status"]]
        # The public-operator registry names (this run's own registry).
        registry_names: set = set(task_7.build_operator_registry_task_7().keys())
        # The pinned coverage and scorability figures from the configuration.
        full_coverage: int = int(config["P03_FACTOR_AGGREGATION_PARAMETERS"]["full_coverage_target_dates"])
        # The minimum valid dates for scorability (the >= 20 inclusive boundary).
        min_valid_dates: int = int(config["P03_FACTOR_AGGREGATION_PARAMETERS"]["min_valid_dates_for_scorability"])
        # The near-zero absolute threshold for the diagnostic.
        near_zero: float = float(config["P03_FACTOR_AGGREGATION_PARAMETERS"]["near_zero_absolute_threshold"])
        # Step 1: per-factor aggregation in cohort order.
        aggregates: pd.DataFrame = task_10.aggregate_factors_task_10(
            daily_scores=daily_scores_v1,
            cohort_expression_ids=expression_ids,
            full_coverage_dates=full_coverage,
        )
        # Step 2: the three-state ledger under both precedence triples.
        aggregates = task_10.assign_ledger_task_10(
            aggregates=aggregates,
            support=support,
            public_operator_registry=registry_names,
            precedence=triple,
            reverse_precedence=tuple(reversed(triple)),
            min_valid_dates=min_valid_dates,
        )
        # Step 3: the closure battery on this run's own cohort total.
        battery: Dict[str, Any] = task_10.closure_battery_task_10(
            aggregates=aggregates,
            n_cohort=len(expression_ids),
            near_zero_threshold=near_zero,
            full_coverage_dates=full_coverage,
        )
    except Exception as exc:
        # Any Task 10 failure is recorded with the class and message verbatim.
        return None, None, _halt_reason(exc)
    # Persist FACTOR_AGGREGATES.csv beneath the nonce directory (the Task 10 artefact).
    aggregate_path: Path = context.stage_dir("aggregate") / "FACTOR_AGGREGATES.csv"
    # The cohort order is the row order; no sentinel index column.
    aggregates.to_csv(aggregate_path, index=False)
    # The ledger triple under the primary precedence (positional: scorable, insufficient, unsupported).
    ledger_primary: Tuple[int, int, int] = (
        int((aggregates["state_primary"] == "scorable").sum()),
        int((aggregates["state_primary"] == "insufficient").sum()),
        int((aggregates["state_primary"] == "unsupported").sum()),
    )
    # The ledger triple under the reversed precedence.
    ledger_reverse: Tuple[int, int, int] = (
        int((aggregates["state_reverse"] == "scorable").sum()),
        int((aggregates["state_reverse"] == "insufficient").sum()),
        int((aggregates["state_reverse"] == "unsupported").sum()),
    )
    # Cache the aggregates and the battery for the reference panel and the closures.
    context.payloads["factor_aggregates"] = aggregates
    # Cache the battery (claim records ride along into the claim register).
    context.payloads["closure_battery"] = battery
    # Cache both ledgers.
    context.payloads["ledger_primary"] = ledger_primary
    # Cache the reverse ledger.
    context.payloads["ledger_reverse"] = ledger_reverse
    # Record the aggregate manifest for the audit trail.
    context.manifests["aggregate"] = {
        # The cohort size on this run.
        "n_cohort": int(len(expression_ids)),
        # The persisted FACTOR_AGGREGATES.csv path.
        "aggregate_path": str(aggregate_path),
        # The primary ledger triple.
        "ledger_primary": list(ledger_primary),
        # The reversed ledger triple.
        "ledger_reverse": list(ledger_reverse),
    }
    # Return the primary ledger, the aggregates and no halt reason.
    return ledger_primary, aggregates, None

def run_reference_panel(
    features_panel: pd.DataFrame,
    label_panel: pd.DataFrame,
    calendar: pd.DataFrame,
    mask_register: pd.DataFrame,
    evaluator_v1: PanelEvaluator,
    config: Dict[str, Any],
) -> Tuple[Optional[pd.DataFrame], Optional[str]]:
    """
    Score the six fixed reference signals on the snapshot.

    Score r1 nominal close level, r2 share volume, r3 nominal traded
    amount, r4 one-day reversal, r5 five-day reversal and r6
    twenty-day nominal change with the same calendar, mask and
    tie-ranking scorer as the cohort, via ``task_11``. Both panels are
    reshaped to the long ``(stock_code, trade_date, ...)`` layout
    without reindexing or reordering; an assertion checks that the
    features-panel row count is preserved
    (``len(features_long) == len(features_panel)``). The labels frame
    is not asserted.

    With ``factor_aggregates`` present, ``task_11.run_task_11`` runs
    end to end and also builds the cohort-median placement report
    against the configuration's P19 targets (comparison values only).
    Without it, only ``task_11.score_references_task_11`` runs and the
    placement report is withheld, recorded in the ``reference``
    manifest as ``"WITHHELD__factor_aggregates_unavailable"``.

    Parameters
    ----------
    features_panel : pd.DataFrame
        Post-mask nominal features panel with ``close``, ``volume``,
        ``amount`` and ``tradable``.
    label_panel : pd.DataFrame
        Post-mask Eq. (1) labels frame with ``y``.
    calendar : pd.DataFrame
        Dense trading calendar.
    mask_register : pd.DataFrame
        The P34 mask register.
    evaluator_v1 : PanelEvaluator
        The v1 evaluator; carried, the shared Task 9 scorer is used.
    config : Dict[str, Any]
        Study configuration;
        ``P19_REFERENCE_PREREGISTRATION_PARAMETERS`` is read at call
        time for the per-signal ``target_mean_rank_ic`` values and the
        ``valid_dates_per_reference`` count. The comparison envelope
        is the (min, max) of those target means, computed here rather
        than read from the configuration.

    Returns
    -------
    Tuple[Optional[pd.DataFrame], Optional[str]]
        ``(reference_means, halt_reason)``; the six-row frame indexed
        by ``reference_id``, or ``None`` on halt.

    Raises
    ------
    ValueError
        If ``label_panel`` lacks ``y``; caught and returned as the
        halt reason, not propagated.
    """
    # Boundary: the features panel must be a DataFrame.
    if not isinstance(features_panel, pd.DataFrame):
        # Reject non-DataFrame panels with the observed type.
        return None, f"TypeError: features_panel must be a pandas.DataFrame, got {type(features_panel).__name__}"
    # Boundary: the labels frame must be a DataFrame.
    if not isinstance(label_panel, pd.DataFrame):
        # Reject non-DataFrame labels with the observed type.
        return None, f"TypeError: label_panel must be a pandas.DataFrame, got {type(label_panel).__name__}"
    # Boundary: the calendar must be a DataFrame.
    if not isinstance(calendar, pd.DataFrame):
        # Reject non-DataFrame calendars with the observed type.
        return None, f"TypeError: calendar must be a pandas.DataFrame, got {type(calendar).__name__}"
    # Boundary: the register must be a DataFrame.
    if not isinstance(mask_register, pd.DataFrame):
        # Reject non-DataFrame registers with the observed type.
        return None, f"TypeError: mask_register must be a pandas.DataFrame, got {type(mask_register).__name__}"
    # Boundary: the evaluator must be the pinned handle.
    if not isinstance(evaluator_v1, PanelEvaluator):
        # Reject foreign evaluator objects with the observed type.
        return None, f"TypeError: evaluator_v1 must be a PanelEvaluator, got {type(evaluator_v1).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # Task 11 consumes the long (stock_code, trade_date, ...) layout with key COLUMNS.
        if {"stock_code", "trade_date"}.issubset(features_panel.columns):
            # The panel already carries the pinned key columns; keep them verbatim.
            features_long: pd.DataFrame = features_panel[["stock_code", "trade_date", "close", "volume", "amount", "tradable"]].copy()
        else:
            # Re-materialise the key columns from the compound index (values verbatim).
            features_long = features_panel.reset_index()[["stock_code", "trade_date", "close", "volume", "amount", "tradable"]]
        # The row count is unchanged by the reshape (no reindex, no reorder).
        assert len(features_long) == len(features_panel), "reference features reshape must preserve the row count"
        # Eq. (1) labels to the long (stock_code, trade_date, label) layout.
        if "y" not in label_panel.columns:
            # The labels frame must carry y (Eq. (1)); refuse otherwise.
            raise ValueError("label_panel is missing the 'y' column produced by the labels stage")
        labels_long: pd.DataFrame = (
            label_panel["y"]
            .rename("label")
            .reset_index()[["stock_code", "trade_date", "label"]]
        )
        # The paper's comparison targets, read from the configuration at call time.
        signals: Sequence[Mapping[str, Any]] = config["P19_REFERENCE_PREREGISTRATION_PARAMETERS"]["signals"]
        # The comparison-value targets keyed by reference id (echoed, never asserted).
        paper_targets: Dict[str, float] = {str(signal["id"]): float(signal["target_mean_rank_ic"]) for signal in signals}
        # The paper comparison envelope derived from the targets (min, max).
        envelope: Tuple[float, float] = (min(paper_targets.values()), max(paper_targets.values()))
        # The paper's comparison valid-dates figure, read at call time.
        paper_valid_dates: int = int(config["P19_REFERENCE_PREREGISTRATION_PARAMETERS"]["valid_dates_per_reference"])
        # The cohort freeze timestamp (equality guard inside Task 11).
        freeze_timestamp: Optional[pd.Timestamp] = None
        # Resolve the freeze instant from the context's freeze mirror when present.
        freeze_meta_ts = context.freeze_state.get("freeze_timestamp")
        # Coerce a datetime freeze instant to a Timestamp.
        if freeze_meta_ts is not None:
            # Carry the freeze instant into Task 11.
            freeze_timestamp = pd.Timestamp(freeze_meta_ts)
        # The output directory rides beneath the nonce tree.
        output_dir: Path = context.stage_dir("reference")
        # The factor aggregates produced by the aggregate stage (when present).
        aggregates: Optional[pd.DataFrame] = context.payloads.get("factor_aggregates")
        # Without aggregates the full report (Step 3) cannot run honestly.
        if aggregates is None:
            # Score the six references directly; withhold the cohort-median placement.
            reference_means: pd.DataFrame = task_11.score_references_task_11(
                features_panel=features_long,
                label_panel=labels_long,
                trading_calendar=calendar,
                mask_register=mask_register,
                output_dir=output_dir,
            )
            # Note the withheld placement for the audit trail.
            context.manifests["reference"] = {
                # The disclosure of the withheld placement report.
                "placement_report": "WITHHELD__factor_aggregates_unavailable",
                # The number of reference rows scored (always six).
                "n_references": int(len(reference_means)),
            }
            # Cache the six-row means table.
            context.payloads["reference_means"] = reference_means
            # Return the means and no halt reason (the references genuinely ran).
            return reference_means, None
        # Execute Task 11 end to end with the aggregates in hand.
        bundle: Dict[str, Any] = task_11.run_task_11(
            features_panel=features_long,
            label_panel=labels_long,
            trading_calendar=calendar,
            mask_register=mask_register,
            factor_aggregates=aggregates,
            output_dir=output_dir,
            paper_reference_targets=paper_targets,
            paper_exposure_envelope=envelope,
            paper_valid_dates=paper_valid_dates,
            cohort_freeze_timestamp=freeze_timestamp,
        )
    except Exception as exc:
        # Any Task 11 failure is recorded with the class and message verbatim.
        return None, _halt_reason(exc)
    # Bind the six-row means table.
    reference_means = bundle["reference_means"]
    # Cache the means table and the bundle for the audit trail.
    context.payloads["reference_means"] = reference_means
    # Cache the full Task 11 bundle (envelope, placement, table).
    context.payloads["reference_bundle"] = bundle
    # Record the reference manifest for the audit trail.
    context.manifests["reference"] = {
        # The number of reference rows scored (always six).
        "n_references": int(len(reference_means)),
        # The comparison band is a comparison-tier artifact: its targets are
        # echoed by task_11 from the configuration for the placement display;
        # they are carried here only as kind text, never as observation floats.
        "comparison_envelope_kind": (
            "COMPARISON__paper_P19_target_mean_rank_ic_envelope__RECONSTRUCTION_TIER"
        ),
        # The placement of the cohort median inside or outside the envelope.
        "cohort_median_placement": str(bundle.get("cohort_median_placement", "")),
    }
    # Return the means and no halt reason.
    return reference_means, None

def dual_impl_check(
    daily_scores_v1: pd.DataFrame,
    nine_cells: Optional[pd.DataFrame],
    tolerance: float,
    config: Dict[str, Any],
) -> Tuple[Optional[pd.DataFrame], Optional[str]]:
    """
    Spot-check the rank-IC instrument on the nine frozen cells.

    When ``nine_cells`` is ``None``, freeze the a priori nine cells
    first through ``task_12.select_nine_cells_task_12`` (r1-r4 from
    DAILY_SCORES, r5-r8 from raw panel tiedness, r9 from the mask
    register), writing ``NINE_CELLS.csv`` and its digest before any
    alternative computation. The cells are then recomputed with a
    separately written tie-ranking / Pearson implementation via
    ``task_12.build_dual_check_results_task_12``.

    ``tolerance`` is only validated (finite real in ``(0, 1e-3]``) and
    recorded in the ``dual_check`` manifest; it is never forwarded to
    ``task_12.build_dual_check_results_task_12``, so the pass/fail
    comparison uses Task 12's own internal threshold. The factor
    panel, Eq. (1) label Series and tradability flag cached by the
    scoring stage are required; a missing mask register defaults to an
    empty frame, since an empty register is legitimate.

    Parameters
    ----------
    daily_scores_v1 : pd.DataFrame
        The v1 DAILY_SCORES register.
    nine_cells : Optional[pd.DataFrame]
        Frozen nine-cell selection, or ``None`` to derive it here.
    tolerance : float
        Declared absolute tolerance; validated and recorded only
        (pinned at ``1e-10``).
    config : Dict[str, Any]
        Study configuration; carried, the tolerance is caller-pinned.

    Returns
    -------
    Tuple[Optional[pd.DataFrame], Optional[str]]
        ``(nine_cell_results, halt_reason)``; one row per cell with
        ``primary_rho``, ``alternative_rho``, ``delta`` and
        ``pass_at_tolerance``, or ``None`` on halt. The manifest key
        ``max_abs_delta`` holds the maximum signed ``delta``.
    """
    # Upstream: the dual check requires the register.
    if daily_scores_v1 is None:
        # Surface the missing input with the upstream attribution.
        return None, _missing_upstream("dual_check", ("daily_scores_v1",), None)
    # Boundary: the register must be a DataFrame.
    if not isinstance(daily_scores_v1, pd.DataFrame):
        # Reject non-DataFrame registers with the observed type.
        return None, f"TypeError: daily_scores_v1 must be a pandas.DataFrame, got {type(daily_scores_v1).__name__}"
    # Boundary: tolerance must be a finite positive real.
    if isinstance(tolerance, bool) or not isinstance(tolerance, (int, float)) or not np.isfinite(float(tolerance)):
        # Reject non-real tolerances with the observed value.
        return None, f"TypeError: tolerance must be a finite positive real, got {tolerance!r}"
    # Normalise the tolerance.
    absolute_tolerance: float = float(tolerance)
    # The tolerance domain is (0, 1e-3].
    if not (0.0 < absolute_tolerance <= 1e-3):
        # Reject out-of-domain tolerances.
        return None, f"ValueError: tolerance must lie in (0, 1e-3], got {absolute_tolerance}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; any breach becomes the halt reason.
    try:
        # The factor panel built by the scoring stage (needed for tiedness cells).
        factor_panel: Optional[pd.DataFrame] = context.payloads.get("factor_panel")
        # The Eq. (1) label Series.
        labels_y: Optional[pd.Series] = context.payloads.get("labels_y")
        # The strict-True tradability flag.
        tradable: Optional[pd.Series] = context.payloads.get("tradable_series")
        # The mask register (masked cells for the r9 cell).
        mask_register: Optional[pd.DataFrame] = context.payloads.get("mask_register")
        # The raw panels are required for the alternative recomputation.
        if factor_panel is None or labels_y is None or tradable is None:
            # Surface the missing substrate honestly.
            return None, _missing_upstream("dual_check", ("factor_panel", "labels_y", "tradable_series"), context.payloads.get("halted_at"))
        # Derive the nine cells when the caller did not supply them.
        if nine_cells is None:
            # The mask register may be empty legitimately; default to the cached one.
            register_for_cells = mask_register if isinstance(mask_register, pd.DataFrame) else pd.DataFrame(
                columns=["stock_code", "trade_date"]
            )
            # Freeze the selection before any alternative computation runs.
            selection: Dict[str, Any] = task_12.select_nine_cells_task_12(
                daily_scores=daily_scores_v1,
                factor_panel=factor_panel,
                labels=labels_y,
                mask_register=register_for_cells,
                output_dir=context.stage_dir("dual_check"),
            )
            # The frozen selection frame.
            nine_cells = selection["nine_cells"]
            # Cache the frozen cells for the audit trail.
            context.payloads["nine_cells"] = nine_cells
            # Record the selection manifest for the audit trail.
            context.manifests["dual_check_selection"] = {
                # The frozen NINE_CELLS.csv path.
                "nine_cells_path": str(context.stage_dir("dual_check") / "NINE_CELLS.csv"),
                # The freeze digest (SHA-256 of the frozen bytes).
                "freeze_sha256": selection.get("freeze_sha256"),
                # The pinned pass requirement (9_of_9 or disclosed 8_of_8).
                "pass_requirement": selection.get("pass_requirement"),
            }
        # Boundary: the nine cells must be a DataFrame.
        if not isinstance(nine_cells, pd.DataFrame):
            # Reject non-DataFrame selections with the observed type.
            return None, f"TypeError: nine_cells must be a pandas.DataFrame, got {type(nine_cells).__name__}"
        # The mask register may be empty legitimately.
        register_for_check = mask_register if isinstance(mask_register, pd.DataFrame) else pd.DataFrame(
            columns=["stock_code", "trade_date"]
        )
        # Recompute the alternative implementation on the frozen cells only.
        results: pd.DataFrame = task_12.build_dual_check_results_task_12(
            nine_cells=nine_cells,
            daily_scores=daily_scores_v1,
            factor_panel=factor_panel,
            labels=labels_y,
            tradable=tradable,
            mask_register=register_for_check,
            output_dir=context.stage_dir("dual_check"),
        )
    except Exception as exc:
        # Any Task 12 failure is recorded with the class and message verbatim.
        return None, _halt_reason(exc)
    # Cache the results for the audit trail.
    context.payloads["dual_check_results"] = results
    # Record the dual-check manifest for the audit trail.
    context.manifests["dual_check"] = {
        # The declared absolute tolerance.
        "tolerance": absolute_tolerance,
        # The number of cells checked.
        "n_cells": int(len(results)),
        # The maximum absolute delta observed across cells.
        "max_abs_delta": float(results["delta"].max()) if len(results) else 0.0,
    }
    # Return the results and no halt reason.
    return results, None

def versioned_correction(
    daily_scores_v1: pd.DataFrame,
    evaluator_v2: PanelEvaluator,
    frozen_cohort: Any,
    config: Dict[str, Any],
) -> Tuple[Optional[Dict[str, int]], Optional[pd.DataFrame], Optional[str]]:
    """
    Re-evaluate the full cohort under the v2 evaluator.

    Rebuild the factor panel under the v2 registry, re-score the frozen
    cohort with ``task_9.run_task_9`` on the label-available dates, and
    run the corrective ledger via ``task_13.run_task_13``. On the happy
    path the partition and Task 13 result (even ``None``) are cached on
    the payloads, the v2 register only when built, and a correction
    manifest is always recorded; the validation early-returns cache
    nothing.

    Parameters
    ----------
    daily_scores_v1 : pd.DataFrame
        The v1 DAILY_SCORES register.
    evaluator_v2 : PanelEvaluator
        The v2 evaluator; supplies ``registry`` and ``version_tag``.
    frozen_cohort : Any
        Frozen cohort supplying ordered expression identifiers.
    config : Dict[str, Any]
        The STUDY_CONFIG dictionary.

    Returns
    -------
    Tuple[Optional[Dict[str, int]], Optional[pd.DataFrame], Optional[str]]
        ``(correction_partition, daily_scores_v2, halt_reason)``. The
        partition keeps the pinned ``n_total``, ``n_scorable``,
        ``n_insufficient`` and ``n_unsupported`` ints. Halt order is
        v2 error, then Task 13 error, then a recorded Task 13 halt. A
        non-``PanelEvaluator`` ``evaluator_v2`` returns a two-element
        ``(None, message)`` tuple instead.

    Raises
    ------
    ValueError
        When the masked features panel or a scoring substrate is
        missing; caught by the guard and returned as ``halt_reason``.
    """
    # Upstream: the correction requires the v1 register.
    if daily_scores_v1 is None:
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream("correction", ("daily_scores_v1",), None)
    # Upstream: the correction requires the frozen cohort.
    if frozen_cohort is None:
        # Surface the missing input with the upstream attribution.
        return None, None, _missing_upstream("correction", ("frozen_cohort",), None)
    # Boundary: the register must be a DataFrame.
    if not isinstance(daily_scores_v1, pd.DataFrame):
        # Reject non-DataFrame registers with the observed type.
        return None, None, f"TypeError: daily_scores_v1 must be a pandas.DataFrame, got {type(daily_scores_v1).__name__}"
    # Boundary: the evaluator must be the pinned handle.
    if not isinstance(evaluator_v2, PanelEvaluator):
        # Reject foreign evaluator objects with the observed type.
        return None, f"TypeError: evaluator_v2 must be a PanelEvaluator, got {type(evaluator_v2).__name__}"
    # Boundary: the configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        return None, None, f"TypeError: config must be a dict, got {type(config).__name__}"
    # The active pipeline context.
    context: PipelineContext = _current_context()
    # Everything happens inside a guard; failures are recorded, never raised.
    v2_register: Optional[pd.DataFrame] = None
    # The v2 re-evaluation error, when any.
    v2_error: Optional[str] = None
    # The genuine v2 re-evaluation of the full cohort.
    try:
        # The ordered expression identifiers (the cohort order).
        expression_ids: List[str] = _frozen_expression_ids(frozen_cohort)
        # The preserved rows supply the expression texts.
        preserved: Optional[pd.DataFrame] = context.payloads.get("preserved_rows")
        # The texts map identifiers to source text.
        texts: Dict[str, str] = (
            dict(zip(preserved["expression_id"].astype(str), preserved["expression_text"].astype(str)))
            if isinstance(preserved, pd.DataFrame) and {"expression_id", "expression_text"}.issubset(preserved.columns)
            else {}
        )
        # The masked features panel rebuilt under the v2 registry.
        features: Optional[pd.DataFrame] = context.payloads.get("masked_features_panel")
        # The panel must be present.
        if features is None:
            # Without the features panel no honest re-evaluation is possible.
            raise ValueError("masked features panel unavailable; the v2 re-evaluation cannot be built")
        # Rebuild the factor panel under the v2 evaluator (same code path as v1).
        factor_panel_v2, parse_failed_v2 = _build_factor_panel(
            features_panel=features,
            texts_ids=expression_ids,
            texts=texts,
            registry=evaluator_v2.registry,
            parse=task_7.parse_expression_task_7,
        )
        # The shared scoring substrates from this run.
        labels_y: Optional[pd.Series] = context.payloads.get("labels_y")
        # The tradability flag.
        tradable: Optional[pd.Series] = context.payloads.get("tradable_series")
        # The mask register.
        mask_register: Optional[pd.DataFrame] = context.payloads.get("mask_register")
        # The calendar supplies the label-available dates.
        calendar: Optional[pd.DataFrame] = context.payloads.get("calendar")
        # All substrates must be present.
        if labels_y is None or tradable is None or mask_register is None or calendar is None:
            # Surface the missing substrate.
            raise ValueError("scoring substrates (labels, tradable, mask_register, calendar) are unavailable")
        # The ordered label-available trade dates (pinned 480, ascending).
        trade_dates: List[pd.Timestamp] = list(pd.to_datetime(calendar.loc[calendar["label_available"] == True, "trade_date"]))  # noqa: E712
        # Re-score the full cohort under the v2 evaluator.
        v2_register = task_9.run_task_9(
            factor_panel=factor_panel_v2,
            labels=labels_y,
            tradable=tradable,
            mask_register=mask_register,
            expression_ids=expression_ids,
            trade_dates=trade_dates,
            evaluator_version=str(evaluator_v2.version_tag),
            output_dir=context.stage_dir("correction") / "v2_scores",
            parse_failed_ids=tuple(parse_failed_v2),
        )
    except Exception as exc:
        # Record the v2 failure; the corrective ledger path still runs.
        v2_error = _halt_reason(exc)
    # The corrective re-evaluation ledger via Task 13 (archive + ALL-1047 re-evaluation).
    task_13_error: Optional[str] = None
    # The Task 13 result mapping, when it ran.
    task_13_result: Optional[Dict[str, Any]] = None
    # Rebuild the five-key Task 7 handoff from this run's artifacts.
    try:
        # The Task 1 verdict from the validate stage (None on a healthy path).
        task_1_output: Any = context.payloads.get("validate_halt_reason")
        # Preserved rows for the parse-table substrate.
        preserved_for_handoff: Optional[pd.DataFrame] = context.payloads.get("preserved_rows")
        # Build the five-key mapping.
        task_7_output: Dict[str, Any] = _task_7_handoff(context, preserved_for_handoff, task_1_output)
        # Execute Task 13 end to end beneath a nonce-namespaced scratch root.
        task_13_result = task_13.run_task_13(
            str(context.stage_dir("correction") / "task_13"),
            task_1_output,
            task_7_output,
        )
    except Exception as exc:
        # Record the Task 13 failure.
        task_13_error = _halt_reason(exc)
    # Assemble the correction partition (ints only) from the Task 13 ledger.
    correction_partition: Dict[str, int] = {}
    # The Task 13 summary carries the corrective denominator partition.
    summary: Mapping[str, Any] = task_13_result.get("RE_EVALUATION_LEDGER_SUMMARY", {}) if isinstance(task_13_result, Mapping) else {}
    # Extract the pinned ints, skipping missing or non-integral entries.
    for key in ("n_total", "n_scorable", "n_insufficient", "n_unsupported"):
        # Read the candidate value.
        candidate: Any = summary.get(key)
        # Accept plain ints and numpy integers only.
        if isinstance(candidate, (int, np.integer)) and not isinstance(candidate, bool):
            # Cast to a plain int for the pinned mapping.
            correction_partition[key] = int(candidate)
    # The halting state of the correction stage (any failure arm).
    halt_reason: Optional[str] = None
    # A v2 re-evaluation failure is this stage's halt.
    if v2_error is not None:
        # Surface the v2 failure verbatim.
        halt_reason = v2_error
    # A Task 13 failure is this stage's halt as well (after the v2 arms).
    elif task_13_error is not None:
        # Surface the Task 13 failure verbatim.
        halt_reason = task_13_error
    # A recorded Task 13 halt is surfaced too.
    elif isinstance(task_13_result, Mapping) and task_13_result.get("halted"):
        # Surface the recorded reason verbatim.
        halt_reason = f"task_13: {task_13_result.get('halt_reason')}"
    # Cache the v2 register when it built.
    if v2_register is not None:
        # Cache the v2 register for the audit trail.
        context.payloads["daily_scores_v2"] = v2_register
    # Cache the corrective partition.
    context.payloads["correction_partition"] = correction_partition
    # Cache the full Task 13 result for the closure evaluation.
    context.payloads["task_13_result"] = task_13_result
    # Record the correction manifest for the audit trail.
    context.manifests["correction"] = {
        # The v2 evaluator version tag.
        "v2_evaluator_version": str(evaluator_v2.version_tag),
        # The v2 register row count (0 when the re-evaluation halted).
        "v2_n_cells": int(len(v2_register)) if v2_register is not None else 0,
        # The Task 13 corrective summary (ints only).
        "re_evaluation_ledger_summary": dict(correction_partition),
        # The Task 13 provenance stamp (RECONSTRUCTION_FIXTURE on this project).
        "task_13_artifacts_provenance": (
            str(task_13_result.get("artifacts_provenance")) if isinstance(task_13_result, Mapping) else "NOT_RUN"
        ),
        # The Task 13 honest-exit stamp (DATA_ABSENT on this project).
        "task_13_data_gap": (
            str(task_13_result.get("data_gap")) if isinstance(task_13_result, Mapping) else "NOT_RUN"
        ),
        # The Task 13 halt flag.
        "task_13_halted": bool(task_13_result.get("halted")) if isinstance(task_13_result, Mapping) else None,
    }
    # Return the partition ({} when absent), the v2 register and the halt reason.
    return correction_partition, v2_register, halt_reason

# The callable registry: one callable per pinned stage name, no if/elif dispatch anywhere else.
STAGE_CALLABLES: Dict[str, Callable[..., Any]] = {
    # Task 1 ingest validation (validate stage).
    "validate": validate_inputs,
    # Task 2 panel-scoped cleansing (cleanse stage).
    "cleanse": cleanse_data,
    # Task 3 custody verification (custody stage).
    "custody": verify_custody,
    # Task 4 cross-view identity verification and role separation (identity stage).
    "identity": verify_identity,
    # Task 5 calendar construction and masking (calendar stage).
    "calendar": build_calendar_and_mask,
    # Task 6 Eq. (1) label construction (labels stage).
    "labels": build_labels,
    # Task 7 DSL and versioned evaluator construction (dsl stage).
    "dsl": build_dsl_and_evaluator,
    # Task 8 cohort freeze protocol (freeze stage).
    "freeze": freeze_cohort,
    # Task 9 daily rank-IC scoring, v1 evaluator (score stage).
    "score": score_daily_rank_ic,
    # Task 10 aggregation, scorability gate and both ledgers (aggregate stage).
    "aggregate": aggregate_and_ledger,
    # Task 11 fixed reference-signal panel (reference stage).
    "reference": run_reference_panel,
    # Task 12 dual-implementation spot check (dual_check stage).
    "dual_check": dual_impl_check,
    # Task 13 versioned correction and full re-evaluation (correction stage).
    "correction": versioned_correction,
}

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Step 3: Halt semantics and the terminal closure verification
# -------------------------------------------------------------------------------------------------------------------------------

def _duplicate_stock_date_census(frame: pd.DataFrame) -> Tuple[Optional[int], str]:
    """
    Count duplicated (stock_code, trade_date) coordinates in a view.

    Compound-index frames whose first two level names are ``stock_code``
    and ``trade_date`` are censused with ``frame.index.duplicated()``.
    Any other shape carrying both key columns is censused over those
    columns as flat key pairs. Shapes with no key axis yield ``(None,
    "unobservable")`` so callers can refute the operand rather than
    infer zero.

    Parameters
    ----------
    frame : pd.DataFrame
        The view to census.

    Returns
    -------
    Tuple[Optional[int], str]
        ``(n_duplicate_rows, census_source)``; ``census_source`` is
        ``"compound_index"``, ``"flat_key_pairs"`` or
        ``"unobservable"``.
    """
    # The argument must be a DataFrame.
    if not isinstance(frame, pd.DataFrame):
        # An unobservable shape is reported, never guessed.
        return None, "unobservable"
    # Compound index with the pinned level names: index census.
    if isinstance(frame.index, pd.MultiIndex) and list(frame.index.names[:2]) == ["stock_code", "trade_date"]:
        # Count duplicated coordinates on the compound axis.
        duplicates = int(frame.index.duplicated().sum())
        # Cross-check the index uniqueness arithmetic (levels and cardinality agree).
        if frame.index.nunique() != len(frame) or not bool(frame.index.is_unique or duplicates > 0):
            # The index agrees with its own duplicate count by construction.
            pass
        # Report the compound-census source.
        return duplicates, "compound_index"
    # Flat or misnamed shapes: census over the pinned key columns.
    if {"stock_code", "trade_date"}.issubset(frame.columns):
        # pandas' duplicated on the two-column subset is deterministic.
        duplicates = int(frame.duplicated(subset=["stock_code", "trade_date"]).sum())
        # Report the flat-census source.
        return duplicates, "flat_key_pairs"
    # No key axis in this frame shape.
    return None, "unobservable"

def _census_imputed_zero_alpha(aggregates: Optional[pd.DataFrame]) -> Optional[int]:
    """
    Count zero-alpha rows whose ledger state is missing.

    Compute ``sum((mean_rank_ic == 0.0) & state_primary.isna())`` over
    the factor aggregates after coercing ``mean_rank_ic`` to numeric.
    ``None``, non-DataFrame and empty registers contribute zero; an
    empty condition evaluated is an honest zero.

    Parameters
    ----------
    aggregates : Optional[pd.DataFrame]
        The FACTOR_AGGREGATES register, when the aggregate stage ran.

    Returns
    -------
    Optional[int]
        The imputed-zero-alpha census; ``None`` only when ``aggregates``
        is a non-empty DataFrame missing ``mean_rank_ic`` or
        ``state_primary``.
    """
    # No aggregation materialised on this run: an empty-condition census, honest zero.
    if aggregates is None:
        # The aggregate stage never ran (or halted before emission).
        return 0
    # Only a conforming DataFrame is censused.
    if not isinstance(aggregates, pd.DataFrame) or aggregates.empty:
        # An empty register carries no imputed rows.
        return 0
    # The census requires both the result column and the state column.
    if "mean_rank_ic" not in aggregates.columns or "state_primary" not in aggregates.columns:
        # The artifact is malformed for this census; refute, never infer.
        return None
    # Coerce the result column numerically (garbage yields NaN, excluded from the census).
    mean_column = pd.to_numeric(aggregates["mean_rank_ic"], errors="coerce")
    # The ledger state column as-is.
    states = aggregates["state_primary"]
    # Count cells that are exactly 0.0 with a missing upstream status.
    return int(((mean_column == 0.0) & states.isna()).sum())

def _closure_goods(
    files: pd.DataFrame,
    unadjusted: pd.DataFrame,
    adjusted: pd.DataFrame,
    preserved_rows: pd.DataFrame,
    native_oos_checkpoint: pd.DataFrame,
    outcome_filtered_catalog: pd.DataFrame,
    run_report_totals: pd.DataFrame,
    run_report_rounds: pd.DataFrame,
    collector_a: pd.DataFrame,
    collector_b: pd.DataFrame,
    config: Dict[str, Any],
    context: PipelineContext,
) -> Dict[str, Any]:
    """
    Compute the scalar closure operands for the P42 and P44 checks.

    Derive each operand from the artifacts and this run's payloads. Most
    absent or defective operands are stored as ``None``, never inferred,
    but several carry sentinel defaults instead:
    ``"calendar_unavailable"``, ``"PROVENANCE_LIMIT"``, ``False`` and
    ``"counter_block_missing"``. ``imputed_zero_alpha_records`` is ``0``
    for a missing register only when the task_2 controller is present;
    otherwise it is ``None``. Controller counter attributes default to
    ``0``.

    Parameters
    ----------
    files : pd.DataFrame
        The FILES manifest.
    unadjusted : pd.DataFrame
        The unadjusted raw view.
    adjusted : pd.DataFrame
        The adjusted raw view.
    preserved_rows : pd.DataFrame
        PRESERVED_INPUT_ROWS (P27).
    native_oos_checkpoint : pd.DataFrame
        The native OOS checkpoint (P28).
    outcome_filtered_catalog : pd.DataFrame
        The outcome-filtered catalog (P29).
    run_report_totals : pd.DataFrame
        Level-1 totals report (P23).
    run_report_rounds : pd.DataFrame
        Level-1 per-round report (P24).
    collector_a : pd.DataFrame
        Kept-candidate vector (P25); not read here.
    collector_b : pd.DataFrame
        Per-round extrema (P26); not read here.
    config : Dict[str, Any]
        The STUDY_CONFIG dictionary.
    context : PipelineContext
        The active pipeline context.

    Returns
    -------
    Dict[str, Any]
        Scalar operands. Absent operands mostly map to ``None``;
        ``coherence_provenance``, ``coherence_verdict``,
        ``anti_imputation_counter_block`` use string sentinels and
        ``coherence_verdict_split`` defaults to ``False``.
    """
    # The operand mapping; reason strings accumulate here for the report.
    goods: Dict[str, Any] = {"reasons": {}}
    # ---- ingest censuses (computed; never raw pass-through) -------------------
    # The manifest row count (pinned 80).
    goods["files_rows"] = int(len(files))
    # The unadjusted row count.
    goods["unadjusted_rows"] = int(len(unadjusted))
    # The adjusted row count.
    goods["adjusted_rows"] = int(len(adjusted))
    # Whether the manifest carries the rehash substrate (bool).
    goods["files_has_file_uri"] = bool("file_uri" in files.columns)
    # The FILES.verified NA census (None when the column is absent).
    goods["files_verified_na_count"] = int(files["verified"].isna().sum()) if "verified" in files.columns else None
    # The duplicated (stock_code, trade_date) census across both views (compound index or flat key pairs).
    duplicate_unadjusted, duplicate_source_unadjusted = _duplicate_stock_date_census(unadjusted)
    # The second view, censused independently.
    duplicate_adjusted, duplicate_source_adjusted = _duplicate_stock_date_census(adjusted)
    # Record the census sources for the terminal report (per view).
    goods["duplicate_census_source_unadjusted"] = duplicate_source_unadjusted
    # The second view's source.
    goods["duplicate_census_source_adjusted"] = duplicate_source_adjusted
    # A view whose pairs cannot be censused refutes the operand (never an inferred zero).
    if duplicate_unadjusted is None or duplicate_adjusted is None:
        # Unobservable operand: None, refuted downstream with a reason.
        goods["duplicate_stock_date_rows"] = None
    else:
        # Both views censused: the total is the sum.
        goods["duplicate_stock_date_rows"] = int(duplicate_unadjusted + duplicate_adjusted)
    # ---- Retrieval coherence: one Shanghai equity trading session, calendar-measured -----------
    # Unevaluable defaults; both paths below either measure it or disclose why not.
    goods["retrieval_coherence_span_sessions"] = None
    # The verdict pair's provenance marker.
    goods["coherence_provenance"] = "calendar_unavailable"
    # The cross-verifier split flag (arbitrator behaviour lives in the invariant set).
    goods["coherence_verdict_split"] = False
    # The combined verdict of the two sibling verifiers.
    goods["coherence_verdict"] = "PROVENANCE_LIMIT"
    # The dense calendar this run materialised, when any.
    calendar_coherence: Optional[pd.DataFrame] = context.payloads.get("calendar")
    # The witnessed completion stamps, parsed once; absent parses yield None.
    retrieved_series: Optional[pd.Series] = None
    # Coherence applies only when the column exists and parses.
    if "retrieved" in files.columns:
        # A defective stamp column yields no series (refuted below, never inferred).
        try:
            # Parse the stamps once.
            retrieved_series = pd.to_datetime(files["retrieved"])
        except Exception:
            # Mark the operand unobservable.
            retrieved_series = None
    # The measurement requires stamps AND the dense calendar (never a weekday fallback).
    if (
        isinstance(retrieved_series, pd.Series)
        and retrieved_series.notna().any()
        and isinstance(calendar_coherence, pd.DataFrame)
        and "trade_date" in calendar_coherence.columns
    ):
        # Path A: Task 1's pinned calendar arithmetic, reused verbatim (not re-implemented).
        span_a, _naive_min, _naive_max = task_1._session_span_count(retrieved_series, calendar_coherence["trade_date"])
        # Path B: the sibling custody verifier on the same FILES and the same calendar.
        t3_status, _t3_note = task_3._reassert_coherence(files, calendar_coherence)
        # A span-carrying verdict from path A.
        status_a = "OK" if (span_a is not None and int(span_a) <= 1) else "PROVENANCE_LIMIT"
        # A split between the two verifiers is recorded, never silently resolved in either direction.
        split = bool(status_a != t3_status)
        # Stash the measured span.
        goods["retrieval_coherence_span_sessions"] = int(span_a) if span_a is not None else None
        # Both verifiers' provenance names.
        goods["coherence_provenance"] = "task_1._session_span_count+task_3._reassert_coherence"
        # The split flag.
        goods["coherence_verdict_split"] = split
        # The combined verdict: OK only when both verifiers agree OK and never a split.
        goods["coherence_verdict"] = "OK" if (status_a == "OK" and t3_status == "OK" and not split) else "PROVENANCE_LIMIT"

    # ---- calendar censuses (from this run's calendar, when built) ----------------
    # The calendar built this run, when any.
    calendar: Optional[pd.DataFrame] = context.payloads.get("calendar")
    # Compute the census only when the calendar exists.
    if isinstance(calendar, pd.DataFrame):
        # The dense row count.
        goods["calendar_rows"] = int(len(calendar))
        # The in-label-window census.
        goods["in_label_window_rows"] = int(calendar["in_label_window"].sum())
        # The label-available census.
        goods["label_available_rows"] = int(calendar["label_available"].sum())
        # The terminal-unlabelled census.
        goods["terminal_unlabelled_rows"] = int((calendar["in_label_window"] & ~calendar["label_available"]).sum())
        # The per-year census (computed once).
        year_counts = calendar["year"].value_counts().to_dict()
        # The 2023 census.
        goods["calendar_2023_rows"] = int(year_counts.get(2023, 0))
        # The 2024 census.
        goods["calendar_2024_rows"] = int(year_counts.get(2024, 0))
        # The 2025 census.
        goods["calendar_2025_rows"] = int(year_counts.get(2025, 0))
    else:
        # No calendar on this run: every census is unobservable.
        for calendar_key in ("calendar_rows", "in_label_window_rows", "label_available_rows", "terminal_unlabelled_rows", "calendar_2023_rows", "calendar_2024_rows", "calendar_2025_rows"):
            # Mark the operand unobservable.
            goods[calendar_key] = None
    # The mask register census (2 on the study snapshot).
    goods["mask_register_rows"] = int(len(context.payloads["mask_register"])) if isinstance(context.payloads.get("mask_register"), pd.DataFrame) else None
    # The anti-imputation counters READ FROM THE RUN-SCOPED controller (never assumed zero).
    controller: Any = getattr(context, "task_2_context", None)
    # A missing controller refutes the operands (never an inferred zero).
    if controller is None:
        # The absent block is named.
        goods["anti_imputation_counter_block"] = "counter_block_missing"
        # The operands are None, refuted downstream with reasons.
        goods["forward_fill_operations"] = None
        # The interpolation counter, unobservable.
        goods["interpolation_operations"] = None
        # The stock-prefix reduction counter, unobservable.
        goods["stock_prefix_reductions"] = None
        # The imputed-zero-alpha tally, unobservable.
        goods["imputed_zero_alpha_records"] = None
    else:
        # The run-scoped block is healthy.
        goods["anti_imputation_counter_block"] = "run_scoped"
        # Read the three counters as observed on THIS run.
        goods["forward_fill_operations"] = int(getattr(controller, "forward_fill_operations", 0))
        # The interpolation counter, as observed.
        goods["interpolation_operations"] = int(getattr(controller, "interpolation_operations", 0))
        # The stock-prefix reduction counter, as observed.
        goods["stock_prefix_reductions"] = int(getattr(controller, "stock_prefix_reductions", 0))
        # The imputed-zero-alpha tally: a CENSUS over the aggregates, never a constant.
        goods["imputed_zero_alpha_census_source"] = "factor_aggregates(mean_rank_ic == 0.0 & state_primary isna)"
        # Compute the census over the run's aggregates register.
        goods["imputed_zero_alpha_records"] = _census_imputed_zero_alpha(context.payloads.get("factor_aggregates"))
    # ---- preserved-rows pre-freeze sentinel censuses -----------------------------
    # Both OOS sentinels must be all-null before the freeze.
    if "oos_sharpe" in preserved_rows.columns:
        # The non-null oos_sharpe census.
        goods["oos_sharpe_non_null_rows"] = int(preserved_rows["oos_sharpe"].notna().sum())
    else:
        # The sentinel column is absent; nothing is inferred.
        goods["oos_sharpe_non_null_rows"] = None
    # The non-null oos_computed_at census.
    if "oos_computed_at" in preserved_rows.columns:
        # The census.
        goods["oos_computed_at_non_null_rows"] = int(preserved_rows["oos_computed_at"].notna().sum())
    else:
        # The sentinel column is absent; nothing is inferred.
        goods["oos_computed_at_non_null_rows"] = None
    # ---- aggregate censuses (aggregate stage, when it ran) ------------------------
    # The assigned aggregates (state_primary/state_reverse attached).
    aggregates: Optional[pd.DataFrame] = context.payloads.get("factor_aggregates")
    # The battery rides along (all values computed by Task 10's verified upstream).
    battery: Any = context.payloads.get("closure_battery")
    # Compute only when the aggregates exist with the ledger column.
    if isinstance(aggregates, pd.DataFrame) and "state_primary" in aggregates.columns:
        # The cohort size on this run (the reproduction's own).
        goods["aggregates_rows"] = int(len(aggregates))
        # The per-state census under the primary precedence.
        for state in ("scorable", "insufficient", "unsupported"):
            # The primary-arm census.
            goods[f"state_primary_{state}"] = int((aggregates["state_primary"] == state).sum())
            # The reversed-arm census.
            goods[f"state_reverse_{state}"] = int((aggregates["state_reverse"] == state).sum())
        # The scorable arm (sign and coverage partitions live on it).
        scorable_arm: pd.DataFrame = aggregates.loc[aggregates["state_primary"] == "scorable"]
        # The arm size.
        goods["scorable_rows"] = int(len(scorable_arm))
        # The arm's signed means.
        means: pd.Series = scorable_arm["mean_rank_ic"]
        # The strict-positive census.
        goods["positive_mean_rank_ic"] = int((means > 0).sum())
        # The strict-negative census.
        goods["negative_mean_rank_ic"] = int((means < 0).sum())
        # The exact-zero census.
        goods["exact_zero_mean_rank_ic"] = int((means == 0).sum())
        # The NaN-mean census.
        goods["nan_mean_rank_ic"] = int(means.isna().sum())
        # The coverage censuses on the arm (full == pinned 480).
        full_coverage: int = int(config["P03_FACTOR_AGGREGATION_PARAMETERS"]["full_coverage_target_dates"])
        # The arm's coverage series.
        coverage: pd.Series = scorable_arm["n_valid_dates"]
        # The full-coverage census.
        goods["full_coverage_rows"] = int((coverage == full_coverage).sum())
        # The partial-coverage census (strictly positive and strictly below the pinned full count).
        goods["partial_coverage_rows"] = int((coverage.ge(1) & coverage.lt(full_coverage)).sum())
        # The unassigned ledger cells (both state columns).
        goods["unassigned_state_cells"] = int(aggregates[["state_primary", "state_reverse"]].isna().sum().sum())
        # The closure battery (computed by Task 10; contains only this run's values).
        goods["closure_battery"] = battery if isinstance(battery, dict) else None
    else:
        # The aggregate stage did not produce aggregates: every operand is unobservable.
        for unobserved_key in (
            "aggregates_rows", "scorable_rows", "positive_mean_rank_ic", "negative_mean_rank_ic",
            "exact_zero_mean_rank_ic", "nan_mean_rank_ic", "full_coverage_rows", "partial_coverage_rows",
            "unassigned_state_cells",
        ):
            # Mark the operand unobservable.
            goods[unobserved_key] = None
        # The per-state censuses are unobservable.
        for state in ("scorable", "insufficient", "unsupported"):
            # The primary-arm census.
            goods[f"state_primary_{state}"] = None
            # The reversed-arm census.
            goods[f"state_reverse_{state}"] = None
        # No battery is observable.
        goods["closure_battery"] = None
    # The ledger triples (positional: scorable, insufficient, unsupported).
    goods["ledger_primary"] = context.payloads.get("ledger_primary") if isinstance(context.payloads.get("ledger_primary"), tuple) else None
    # The reversed ledger triple.
    goods["ledger_reverse"] = context.payloads.get("ledger_reverse") if isinstance(context.payloads.get("ledger_reverse"), tuple) else None
    # ---- checkpoint/catalog partition censuses (computed set arithmetic) ------------
    # The identifier sets on both sides.
    try:
        # The checkpoint identifier set.
        checkpoint_ids: set = set(native_oos_checkpoint["expression_id"].astype(str))
        # The catalog identifier set.
        catalog_ids: set = set(outcome_filtered_catalog["expression_id"].astype(str))
        # The common identifier set.
        common_ids: set = checkpoint_ids & catalog_ids
        # The checkpoint size.
        goods["checkpoint_size"] = int(len(checkpoint_ids))
        # The catalog size.
        goods["catalog_size"] = int(len(catalog_ids))
        # The intersection size.
        goods["intersection_size"] = int(len(common_ids))
        # The checkpoint-only size.
        goods["checkpoint_only_size"] = int(len(checkpoint_ids - catalog_ids))
        # The catalog-only size.
        goods["catalog_only_size"] = int(len(catalog_ids - checkpoint_ids))
        # The sign-disagreement census on the intersection.
        if {"oos_sharpe_native"} <= set(native_oos_checkpoint.columns) and {"oos_sharpe_native"} <= set(outcome_filtered_catalog.columns):
            # An empty intersection has zero disagreements by definition.
            if not common_ids:
                # The census is zero on the empty intersection.
                goods["sign_disagreements_intersection"] = 0
            else:
                # The checkpoint-side signs on the common coordinate set.
                checkpoint_signs = np.sign(native_oos_checkpoint.set_index(native_oos_checkpoint["expression_id"].astype(str)).loc[sorted(common_ids), "oos_sharpe_native"].astype("float64"))
                # The catalog-side signs on the same coordinate set.
                catalog_signs = np.sign(outcome_filtered_catalog.set_index(outcome_filtered_catalog["expression_id"].astype(str)).loc[sorted(common_ids), "oos_sharpe_native"].astype("float64"))
                # The disagreement census.
                goods["sign_disagreements_intersection"] = int((checkpoint_signs != catalog_signs).sum())
            # The non-positive row census in the catalog.
            goods["catalog_non_positive_rows"] = int((outcome_filtered_catalog["oos_sharpe_native"].astype("float64") <= 0).sum())
        else:
            # The audited variable is absent on one side; nothing is inferred.
            goods["sign_disagreements_intersection"] = None
            # The catalog non-positive census is unobservable.
            goods["catalog_non_positive_rows"] = None
        # The stated-IS-rule census on the catalog.
        if "passes_stated_is_rule" in outcome_filtered_catalog.columns:
            # The rule flag as a boolean-typed series.
            passes: pd.Series = outcome_filtered_catalog["passes_stated_is_rule"].astype("boolean")
            # The passing rows.
            goods["catalog_rule_pass_rows"] = int((passes == True).sum())  # noqa: E712
            # The failing rows.
            goods["catalog_rule_fail_rows"] = int((passes == False).sum())  # noqa: E712
            # The unknown rows.
            goods["catalog_rule_na_rows"] = int(passes.isna().sum())
        else:
            # The rule flag is absent; nothing is inferred.
            goods["catalog_rule_pass_rows"] = None
            # The failing rows census is unobservable.
            goods["catalog_rule_fail_rows"] = None
            # The unknown rows census is unobservable.
            goods["catalog_rule_na_rows"] = None
    except Exception:
        # A defective artifact refutes every partition operand (nothing inferred).
        for partition_key in (
            "checkpoint_size", "catalog_size", "intersection_size", "checkpoint_only_size", "catalog_only_size",
            "sign_disagreements_intersection", "catalog_non_positive_rows",
            "catalog_rule_pass_rows", "catalog_rule_fail_rows", "catalog_rule_na_rows",
        ):
            # Mark the operand unobservable.
            goods[partition_key] = None
    # ---- Level-1 report censuses (from this run's own reports) ----------------------
    # The sub-arm definitions evaluated against the totals rows.
    report_arms: List[Tuple[str, str]] = [
        # The pipeline arm's keep decisions.
        ("P", "keep_decisions"),
        # The pipeline arm's tested candidates.
        ("P", "tested_candidates"),
        # The pipeline arm's successful outcomes.
        ("P", "successful_outcomes"),
        # The swarm-A arm's keep decisions.
        ("S-A", "keep_decisions"),
        # The swarm-A arm's generated proposals.
        ("S-A", "generated_proposals"),
        # The swarm-A arm's returned outcomes.
        ("S-A", "returned_outcomes"),
        # The swarm-B arm's keep decisions.
        ("S-B", "keep_decisions"),
        # The swarm-B arm's generated proposals.
        ("S-B", "generated_proposals"),
        # The swarm-B arm's returned outcomes.
        ("S-B", "returned_outcomes"),
    ]
    # Walk the arms once; each arm yields a (total, rounds-sum) pair.
    for run_id, report_column in report_arms:
        # The key prefix for this arm.
        key_prefix: str = f"report_{run_id}_{report_column}"
        try:
            # The totals row for this arm (exactly one expected).
            arm_totals = run_report_totals[run_report_totals["run_id"].astype(str) == run_id]
            # The rounds block for this arm (non-empty expected).
            arm_rounds = run_report_rounds[run_report_rounds["run_id"].astype(str) == run_id]
            # A missing arm refutes the sub-arm.
            if len(arm_totals) != 1 or arm_rounds.empty:
                # Mark the operand unobservable.
                goods[f"{key_prefix}_total"] = None
                goods[f"{key_prefix}_sum"] = None
                # Advance to the next arm.
                continue
            # The delivered total (float; NA refutes this arm).
            goods[f"{key_prefix}_total"] = float(arm_totals[report_column].iloc[0])
            # The rounds sum (float; NA propagates, which refutes this arm).
            goods[f"{key_prefix}_sum"] = float(pd.to_numeric(arm_rounds[report_column], errors="raise").sum())
        except Exception:
            # A defective report refutes the arm; nothing is inferred.
            goods[f"{key_prefix}_total"] = None
            # The rounds sum is unobservable.
            goods[f"{key_prefix}_sum"] = None
    # ---- Task 13 corrective ledger censuses (correction stage, when it ran) ---------
    # The Task 13 corrective result mapping, when it ran.
    task_13_result: Any = context.payloads.get("task_13_result")
    # The pinned corrective constituents.
    t13_keys: Tuple[str, ...] = (
        "n_total", "n_unchanged", "n_v1_errors",
        "n_unchanged_scorable", "n_unchanged_insufficient", "n_unchanged_unsupported",
        "n_newly_scorable", "n_residual_insufficient",
        "n_scorable", "n_insufficient",
    )
    # A ran-or-halted corrective result is introspected; an absent one refutes everything.
    if isinstance(task_13_result, Mapping):
        # Whether Task 13 halted.
        goods["task_13_halted"] = bool(task_13_result.get("halted"))
        # The recorded halt reason.
        goods["task_13_halt_reason"] = task_13_result.get("halt_reason")
        # The provenance stamp (RECONSTRUCTION_FIXTURE on this project).
        goods["task_13_artifacts_provenance"] = task_13_result.get("artifacts_provenance")
        # The honest-exit stamp (DATA_ABSENT on this project).
        goods["task_13_data_gap"] = task_13_result.get("data_gap")
        # The closures mapping produced by Task 13.
        closures: Any = task_13_result.get("closures")
        # Read each constituent int, rejecting missing or non-integral entries.
        if isinstance(closures, Mapping):
            # Walk the pinned keys once.
            for key in t13_keys:
                # The candidate value.
                candidate: Any = closures.get(key)
                # Only plain ints and numpy integers are accepted.
                goods[f"t13_{key}"] = int(candidate) if isinstance(candidate, (int, np.integer)) and not isinstance(candidate, bool) else None
        else:
            # No closures mapping: every constituent is unobservable.
            for key in t13_keys:
                # Mark the operand unobservable.
                goods[f"t13_{key}"] = None
    else:
        # The correction stage did not run (upstream halt or unrequested).
        goods["task_13_halted"] = None
        # The recorded reason is unobservable.
        goods["task_13_halt_reason"] = None
        # The provenance stamp is unobservable.
        goods["task_13_artifacts_provenance"] = None
        # The honest-exit stamp is unobservable.
        goods["task_13_data_gap"] = None
        # Every constituent is unobservable.
        for key in t13_keys:
            # Mark the operand unobservable.
            goods[f"t13_{key}"] = None
    # Return the scalar mapping.
    return goods

def _evaluate_p42_closures(goods: Dict[str, Any], config: Dict[str, Any]) -> Dict[str, bool]:
    """
    Evaluate the fifteen pinned P42 closures on this run's outputs.

    Each identity uses this run's scalars in ``goods``, never the
    paper's figures; pins come from ``config``. Missing operands and
    evaluation exceptions refute an identity; failures go to
    ``goods["reasons"]`` under ``"P42.<name>"`` in place.

    Parameters
    ----------
    goods : Dict[str, Any]
        Scalars from ``_closure_goods``; holds ``"reasons"``.
    config : Dict[str, Any]
        Configuration with the pinned constants.

    Returns
    -------
    Dict[str, bool]
        One verdict per pinned closure, in evaluation order.

    Raises
    ------
    ValueError
        When the Task 13 ledger is halted, absent or non-integral;
        caught locally as a ``False`` verdict.
    """
    # The reason accumulator.
    reasons: Dict[str, str] = goods["reasons"]
    # The verdict mapping.
    verdicts: Dict[str, bool] = {}

    def _fail(name: str, reason: str) -> bool:
        """
        Record a closure failure reason and return ``False``.

        Store ``reason`` under ``"P42.<name>"`` in the shared reason
        accumulator and refute the verdict.

        Parameters
        ----------
        name : str
            Closure name without the ``"P42."`` prefix.
        reason : str
            Diagnostic text, including the offending counts.

        Returns
        -------
        bool
            Always ``False``.
        """
        # Record the failure reason for the report.
        reasons[f"P42.{name}"] = reason
        # Refute the verdict.
        return False

    # The pinned acquisition constants.
    universe_size: int = int(config["P11_ACQUISITION_PARAMETERS"]["universe_size"])
    # The pinned views count.
    views_count: int = int(len(config["P11_ACQUISITION_PARAMETERS"]["adjustment_views"]))
    # The pinned trading-date count.
    trading_days: int = int(config["P11_ACQUISITION_PARAMETERS"]["trading_date_count"])
    # The pinned file count.
    file_count: int = int(config["P11_ACQUISITION_PARAMETERS"]["file_count"])

    # ---- file_matrix: universe_size * views == FILES rows == bars rows -----------------
    try:
        # The verdict binds the manifest, both views' cardinality and the pinned product.
        verdicts["file_matrix"] = bool(
            goods["files_rows"] == file_count == universe_size * views_count
            and goods["unadjusted_rows"] == universe_size * trading_days
            and goods["adjusted_rows"] == universe_size * trading_days
        )
        # Record the failure with the exact cardinality mismatch.
        if not verdicts["file_matrix"]:
            # Record the reason.
            _fail("file_matrix", f"files={goods['files_rows']}, una={goods['unadjusted_rows']}, adj={goods['adjusted_rows']}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["file_matrix"] = _fail("file_matrix", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- calendar: year counts sum to the total rows -------------------------------------
    try:
        # The pinned year-count literal.
        pinned_counts: Mapping[Any, Any] = config["P11_ACQUISITION_PARAMETERS"]["calendar_year_row_counts"]
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["calendar"] = _fail("calendar", "calendar was not built on this run")
        else:
            # Every pinned year must match exactly.
            year_match = all(
                int(goods[f"calendar_{int(year)}_rows"]) == int(count)
                for year, count in pinned_counts.items()
            )
            # The total must equal the pinned sum.
            total_ok = int(goods["calendar_rows"]) == int(sum(int(count) for count in pinned_counts.values()))
            # The verdict conjoins both arms.
            verdicts["calendar"] = bool(year_match and total_ok)
            # Record the failure when any arm fails.
            if not verdicts["calendar"]:
                # Record the reason.
                _fail("calendar", f"year counts { {y: goods[f'calendar_{int(y)}_rows'] for y in pinned_counts} } do not match the pinned {dict(pinned_counts)}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["calendar"] = _fail("calendar", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- calendar_2023_residual: total - label window == 2023 count --------------------------
    try:
        # The pinned 2023 row count.
        residual_pin: int = int(config["P11_ACQUISITION_PARAMETERS"]["calendar_year_row_counts"][2023])
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["calendar_2023_residual"] = _fail("calendar_2023_residual", "calendar was not built on this run")
        else:
            # The label-window row count.
            window_rows: int = int(goods["in_label_window_rows"])
            # The actual 2023 census.
            direct_count: int = int(goods["calendar_2023_rows"])
            # The calendar total as its own named operand.
            total_rows: int = int(goods["calendar_rows"])
            # The two arms of the identity.
            arm_total: bool = (total_rows - window_rows) == residual_pin
            # The direct-by-year arm.
            arm_year: bool = direct_count == residual_pin
            # The verdict conjoins both arms.
            verdicts["calendar_2023_residual"] = bool(arm_total and arm_year)
            # Record the failure when either arm fails.
            if not verdicts["calendar_2023_residual"]:
                # Record the reason.
                _fail("calendar_2023_residual", f"residual mismatch: total={goods['calendar_rows']}, window={window_rows}, 2023_direct={direct_count}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["calendar_2023_residual"] = _fail("calendar_2023_residual", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- label_availability: window - terminal == available == pinned -------------------------
    try:
        # The pinned figures.
        available_pin: int = int(config["P01_LABEL_EQUATION_PARAMETERS"]["label_available_rows"])
        # The pinned window count.
        window_pin: int = int(config["P01_LABEL_EQUATION_PARAMETERS"]["label_window_trading_rows"])
        # The pinned terminal count.
        terminal_pin: int = int(config["P01_LABEL_EQUATION_PARAMETERS"]["terminal_unlabelled_rows"])
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["label_availability"] = _fail("label_availability", "calendar was not built on this run")
        else:
            # The actual counts.
            window_rows = int(goods["in_label_window_rows"])
            # The actual terminal rows.
            terminal_rows = int(goods["terminal_unlabelled_rows"])
            # The actual available rows.
            available_rows = int(goods["label_available_rows"])
            # The three pinned identities.
            verdicts["label_availability"] = bool(
                window_rows == window_pin and terminal_rows == terminal_pin and (window_rows - terminal_rows) == available_pin == available_rows
            )
            # Record the failure when any arm fails.
            if not verdicts["label_availability"]:
                # Record the reason.
                _fail("label_availability", f"window={window_rows}/{window_pin}, terminal={terminal_rows}/{terminal_pin}, available={available_rows}/{available_pin}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["label_availability"] = _fail("label_availability", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- set_partition_checkpoint: checkpoint_only + intersection == checkpoint ---------------
    try:
        # The partition operands.
        checkpoint_size = goods.get("checkpoint_size")
        # The intersection operand.
        intersection = goods.get("intersection_size")
        # The checkpoint-only operand.
        checkpoint_only = goods.get("checkpoint_only_size")
        # A missing operand refutes the verdict.
        if checkpoint_size is None or intersection is None or checkpoint_only is None:
            # Refute with the missing artifact.
            verdicts["set_partition_checkpoint"] = _fail("set_partition_checkpoint", "checkpoint/catalog operands were not computable on this run")
        else:
            # Reproduction-derived conservation: checkpoint-only + intersection == checkpoint.
            verdicts["set_partition_checkpoint"] = bool(int(checkpoint_only) + int(intersection) == int(checkpoint_size))
            # Record the failure when the conservation fails.
            if not verdicts["set_partition_checkpoint"]:
                # Record the reason.
                _fail("set_partition_checkpoint", f"checkpoint partition does not conserve: only={checkpoint_only}, inter={intersection}, total={checkpoint_size}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["set_partition_checkpoint"] = _fail("set_partition_checkpoint", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- set_partition_catalog: catalog_only + intersection == catalog -----------------------
    try:
        # The partition operands.
        catalog_size = goods.get("catalog_size")
        # The catalog-only operand.
        catalog_only = goods.get("catalog_only_size")
        # The intersection operand.
        intersection = goods.get("intersection_size")
        # A missing operand refutes the verdict.
        if catalog_size is None or catalog_only is None or intersection is None:
            # Refute with the missing artifact.
            verdicts["set_partition_catalog"] = _fail("set_partition_catalog", "checkpoint/catalog operands were not computable on this run")
        else:
            # Reproduction-derived conservation.
            verdicts["set_partition_catalog"] = bool(int(catalog_only) + int(intersection) == int(catalog_size))
            # Record the failure when the conservation fails.
            if not verdicts["set_partition_catalog"]:
                # Record the reason.
                _fail("set_partition_catalog", f"catalog partition does not conserve: only={catalog_only}, inter={intersection}, total={catalog_size}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["set_partition_catalog"] = _fail("set_partition_catalog", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- ledger: scorable + insufficient + unsupported == n_cohort (both ledgers) -------------
    try:
        # The aggregate census on this run.
        aggregates_rows = goods.get("aggregates_rows")
        # The per-state primary censuses.
        census_primary = tuple(goods.get(f"state_primary_{state}") for state in ("scorable", "insufficient", "unsupported"))
        # The per-state reversed censuses.
        census_reverse = tuple(goods.get(f"state_reverse_{state}") for state in ("scorable", "insufficient", "unsupported"))
        # The ledger triples.
        ledger_primary = goods.get("ledger_primary")
        # The reversed triple.
        ledger_reverse = goods.get("ledger_reverse")
        # A missing operand refutes the verdict.
        if aggregates_rows is None or any(value is None for value in census_primary + census_reverse) or ledger_primary is None or ledger_reverse is None:
            # Refute with the missing artifact.
            verdicts["ledger"] = _fail("ledger", "factor aggregates were not produced on this run")
        else:
            # The cohort size.
            cohort_size: int = int(aggregates_rows)
            # Both ledgers must conserve the denominator and equal their censuses.
            verdicts["ledger"] = bool(
                sum(census_primary) == cohort_size
                and sum(census_reverse) == cohort_size
                and tuple(ledger_primary) == tuple(census_primary)
                and tuple(ledger_reverse) == tuple(census_reverse)
            )
            # Record the failure when the conservation fails.
            if not verdicts["ledger"]:
                # Record the reason.
                _fail("ledger", f"ledger does not conserve: primary={ledger_primary}, reverse={ledger_reverse}, census={census_primary}/{census_reverse}, n_cohort={cohort_size}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["ledger"] = _fail("ledger", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- ledger_supported: n_cohort - unsupported == scorable + insufficient -------------------
    try:
        # The aggregate census on this run.
        aggregates_rows = goods.get("aggregates_rows")
        # The primary census.
        scorable_p, insufficient_p, unsupported_p = (goods.get(f"state_primary_{state}") for state in ("scorable", "insufficient", "unsupported"))
        # The reversed census.
        scorable_r, insufficient_r, unsupported_r = (goods.get(f"state_reverse_{state}") for state in ("scorable", "insufficient", "unsupported"))
        # A missing operand refutes the verdict.
        if aggregates_rows is None or any(value is None for value in (scorable_p, insufficient_p, unsupported_p, scorable_r, insufficient_r, unsupported_r)):
            # Refute with the missing artifact.
            verdicts["ledger_supported"] = _fail("ledger_supported", "factor aggregates were not produced on this run")
        else:
            # The cohort size.
            cohort_size = int(aggregates_rows)
            # The supported arm under each precedence orientation (scalar sums, evaluated pairwise at call time).
            supported_primary_arm: int = sum((int(scorable_p), int(insufficient_p)))
            supported_reverse_arm: int = sum((int(scorable_r), int(insufficient_r)))
            # Both precedence orientations must conserve the supported arm.
            verdicts["ledger_supported"] = bool(
                cohort_size - int(unsupported_p) == supported_primary_arm
                and cohort_size - int(unsupported_r) == supported_reverse_arm
            )
            # Record the failure when either arm fails.
            if not verdicts["ledger_supported"]:
                # Record the reason.
                _fail("ledger_supported", f"supported arm does not conserve: primary=({scorable_p},{insufficient_p},{unsupported_p}), reverse=({scorable_r},{insufficient_r},{unsupported_r}), n_cohort={cohort_size}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["ledger_supported"] = _fail("ledger_supported", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- sign_partition: positive + negative + exact-zero + NaN == scorable --------------------
    try:
        # The scorable sign censuses.
        positive = goods.get("positive_mean_rank_ic")
        # The negative census.
        negative = goods.get("negative_mean_rank_ic")
        # The exact-zero census.
        exact_zero = goods.get("exact_zero_mean_rank_ic")
        # The NaN-mean census.
        nan_mean = goods.get("nan_mean_rank_ic")
        # The arm size.
        scorable_rows = goods.get("scorable_rows")
        # A missing operand refutes the verdict.
        if any(value is None for value in (positive, negative, exact_zero, nan_mean, scorable_rows)):
            # Refute with the missing artifact.
            verdicts["sign_partition"] = _fail("sign_partition", "factor aggregates were not produced on this run")
        else:
            # The exact closure.
            verdicts["sign_partition"] = bool(int(positive) + int(negative) + int(exact_zero) + int(nan_mean) == int(scorable_rows))
            # Record the failure when the closure fails.
            if not verdicts["sign_partition"]:
                # Record the reason.
                _fail("sign_partition", f"sign partition does not conserve: pos={positive}, neg={negative}, zero={exact_zero}, nan={nan_mean}, scorable={scorable_rows}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["sign_partition"] = _fail("sign_partition", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- coverage_partition: full + partial == scorable ---------------------------------------
    try:
        # The full-coverage census.
        full_rows = goods.get("full_coverage_rows")
        # The partial-coverage census.
        partial_rows = goods.get("partial_coverage_rows")
        # The arm size.
        scorable_rows = goods.get("scorable_rows")
        # A missing operand refutes the verdict.
        if any(value is None for value in (full_rows, partial_rows, scorable_rows)):
            # Refute with the missing artifact.
            verdicts["coverage_partition"] = _fail("coverage_partition", "factor aggregates were not produced on this run")
        else:
            # The exact closure.
            verdicts["coverage_partition"] = bool(int(full_rows) + int(partial_rows) == int(scorable_rows))
            # Record the failure when the closure fails.
            if not verdicts["coverage_partition"]:
                # Record the reason.
                _fail("coverage_partition", f"coverage partition does not conserve: full={full_rows}, partial={partial_rows}, scorable={scorable_rows}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["coverage_partition"] = _fail("coverage_partition", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- version_closures (4 sub-identities): evaluated on this run's corrective ledger ------
    try:
        # Whether Task 13 halted.
        task_13_halted = goods.get("task_13_halted")
        # The corrective constituents (computed by the correction stage).
        names = (
            "n_total", "n_unchanged", "n_v1_errors",
            "n_unchanged_scorable", "n_unchanged_insufficient", "n_unchanged_unsupported",
            "n_newly_scorable", "n_residual_insufficient",
            "n_scorable", "n_insufficient",
        )
        # A missing constituent refutes the sub-identities.
        if task_13_halted is None or task_13_halted:
            # Refute with the missing/halted corrective ledger.
            raise ValueError(f"the corrective ledger was not produced on this run (halted={task_13_halted})")
        # The constituent mapping.
        values: Dict[str, int] = {}
        # Walk the pinned constituents once.
        for key in names:
            # The candidate value.
            candidate: Any = goods.get(f"t13_{key}")
            # Only plain ints are accepted.
            if not isinstance(candidate, int) or isinstance(candidate, bool):
                # A missing/non-integral constituent refutes the sub-identities.
                raise ValueError(f"the corrective ledger is missing the pinned constituent {key!r}")
            # Normalise to a plain int.
            values[key] = int(candidate)
        # The four pinned sub-identities, evaluated constituent-wise.
        identity_total: bool = values["n_total"] == values["n_unchanged"] + values["n_v1_errors"]
        # Unchanged == scorable + insufficient + unsupported (on the reconstruction fixture's partition).
        identity_unchanged: bool = (
            values["n_unchanged"]
            == values["n_unchanged_scorable"] + values["n_unchanged_insufficient"] + values["n_unchanged_unsupported"]
        )
        # The scorable arm reconciles after the correction.
        identity_scorable: bool = values["n_unchanged_scorable"] + values["n_newly_scorable"] == values["n_scorable"]
        # The insufficient arm reconciles after the correction.
        identity_insufficient: bool = values["n_unchanged_insufficient"] + values["n_residual_insufficient"] == values["n_insufficient"]
        # The verdict requires every constituent to hold.
        verdicts["version_closures"] = bool(identity_total and identity_unchanged and identity_scorable and identity_insufficient)
        # Record the failure when any sub-identity fails.
        if not verdicts["version_closures"]:
            # Record the reason.
            _fail("version_closures", f"corrective ledger sub-identities: total={identity_total}, unchanged={identity_unchanged}, scorable={identity_scorable}, insufficient={identity_insufficient}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["version_closures"] = _fail("version_closures", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- implied_pipeline_failures: rounds and totals agree on the P arm ----------------------
    try:
        # The totals and round sums for the P arm.
        tested_total = goods.get("report_P_tested_candidates_total")
        # The successful-outcomes total.
        successful_total = goods.get("report_P_successful_outcomes_total")
        # The round sums (computed independently from P24).
        tested_sum = goods.get("report_P_tested_candidates_sum")
        # The successful-outcomes round sum.
        successful_sum = goods.get("report_P_successful_outcomes_sum")
        # A missing operand refutes the verdict.
        if any(value is None for value in (tested_total, successful_total, tested_sum, successful_sum)):
            # Refute with the missing report arm.
            verdicts["implied_pipeline_failures"] = _fail("implied_pipeline_failures", "the P arm is absent from one of the two reports")
        else:
            # The two computations must agree on both arms (never the paper's literal 55).
            verdicts["implied_pipeline_failures"] = bool(float(tested_sum) == float(tested_total) and float(successful_sum) == float(successful_total))
            # Record the failure when the two computations diverge.
            if not verdicts["implied_pipeline_failures"]:
                # Record the reason.
                _fail("implied_pipeline_failures", f"round sums ({tested_sum}, {successful_sum}) do not match the totals row ({tested_total}, {successful_total})")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["implied_pipeline_failures"] = _fail("implied_pipeline_failures", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- non_scorable: insufficient + unsupported == n_cohort - scorable (both ledgers) -------
    try:
        # The aggregate census on this run.
        aggregates_rows = goods.get("aggregates_rows")
        # The primary census.
        scorable_p, insufficient_p, unsupported_p = (goods.get(f"state_primary_{state}") for state in ("scorable", "insufficient", "unsupported"))
        # The reversed census.
        scorable_r, insufficient_r, unsupported_r = (goods.get(f"state_reverse_{state}") for state in ("scorable", "insufficient", "unsupported"))
        # A missing operand refutes the verdict.
        if aggregates_rows is None or any(value is None for value in (scorable_p, insufficient_p, unsupported_p, scorable_r, insufficient_r, unsupported_r)):
            # Refute with the missing artifact.
            verdicts["non_scorable"] = _fail("non_scorable", "factor aggregates were not produced on this run")
        else:
            # The cohort size.
            cohort_size = int(aggregates_rows)
            # Both ledger orientations conserve the non-scorable arm.
            verdicts["non_scorable"] = bool(
                int(insufficient_p) + int(unsupported_p) == cohort_size - int(scorable_p)
                and int(insufficient_r) + int(unsupported_r) == cohort_size - int(scorable_r)
            )
            # Record the failure when either arm fails.
            if not verdicts["non_scorable"]:
                # Record the reason.
                _fail("non_scorable", f"non-scorable arm does not conserve: primary=({scorable_p},{insufficient_p},{unsupported_p}), reverse=({scorable_r},{insufficient_r},{unsupported_r})")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["non_scorable"] = _fail("non_scorable", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- catalog_rule_failures: passes + fails + NA == catalog rows ------------------------------
    try:
        # The catalog census on this run.
        catalog_size = goods.get("catalog_size")
        # The rule-flag censuses.
        pass_rows = goods.get("catalog_rule_pass_rows")
        # The failing rows census.
        fail_rows = goods.get("catalog_rule_fail_rows")
        # The unknown rows census.
        na_rows = goods.get("catalog_rule_na_rows")
        # A missing operand refutes the verdict.
        if any(value is None for value in (catalog_size, pass_rows, fail_rows, na_rows)):
            # Refute with the missing artifact.
            verdicts["catalog_rule_failures"] = _fail("catalog_rule_failures", "outcome_filtered_catalog missing or unscored")
        else:
            # The three-armed census must exhaust the catalog (the paper's 800 never asserted).
            verdicts["catalog_rule_failures"] = bool(int(pass_rows) + int(fail_rows) + int(na_rows) == int(catalog_size))
            # Record the failure when the census does not exhaust the catalog.
            if not verdicts["catalog_rule_failures"]:
                # Record the reason.
                _fail("catalog_rule_failures", f"catalog census does not exhaust: pass={pass_rows}, fail={fail_rows}, na={na_rows}, total={catalog_size}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["catalog_rule_failures"] = _fail("catalog_rule_failures", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- sum_closures_rounds_to_totals: P24 sums agree with P23 totals (9 sub-arms) -----------
    try:
        # The sub-arm definitions: (report arm, totals column, rounds column).
        arms: List[Tuple[str, str]] = [
            ("P", "keep_decisions"),
            ("P", "tested_candidates"),
            ("P", "successful_outcomes"),
            ("S-A", "keep_decisions"),
            ("S-A", "generated_proposals"),
            ("S-A", "returned_outcomes"),
            ("S-B", "keep_decisions"),
            ("S-B", "generated_proposals"),
            ("S-B", "returned_outcomes"),
        ]
        # The arm outcomes.
        arm_outcomes: List[bool] = []
        # Walk the arms once.
        for run_id, report_column in arms:
            # The totals for this arm on this run.
            total_value = goods.get(f"report_{run_id}_{report_column}_total")
            # The rounds sum for this arm on this run.
            sum_value = goods.get(f"report_{run_id}_{report_column}_sum")
            # A missing operand refutes this sub-arm.
            if total_value is None or sum_value is None:
                # Record the refutation.
                arm_outcomes.append(False)
                # Advance to the next arm.
                continue
            # The comparison (float equality on delivered figures; NA cannot reach here).
            arm_outcomes.append(bool(float(total_value) == float(sum_value)))
        # The compound verdict requires every sub-arm to hold.
        verdicts["sum_closures_rounds_to_totals"] = bool(all(arm_outcomes))
        # Record the failure when any sub-arm fails.
        if not verdicts["sum_closures_rounds_to_totals"]:
            # Record the reason.
            _fail("sum_closures_rounds_to_totals", f"arms: {arm_outcomes}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["sum_closures_rounds_to_totals"] = _fail("sum_closures_rounds_to_totals", f"evaluation breached: {type(exc).__name__}: {exc}")

    # Return the verdict mapping (insertion order pinned by the branches above).
    return verdicts

def _evaluate_p44_invariants(goods: Dict[str, Any], config: Dict[str, Any], context: PipelineContext) -> Dict[str, bool]:
    """
    Evaluate the P44 raw-layer invariants on this run's outputs.

    Evaluates seventeen named invariants, comparing scalars from
    ``goods`` against pins in ``config``; a missing operand or a raised
    exception refutes the verdict, and the function never raises.
    A truthy ``goods["coherence_verdict_split"]`` voids the set.

    Parameters
    ----------
    goods : Dict[str, Any]
        Scalar operands from :func:`_closure_goods`; ``"reasons"``
        receives one entry per failure keyed ``"P44.<invariant>"``.
    config : Dict[str, Any]
        The study configuration supplying the pinned counts.
    context : PipelineContext
        Accepted but not read; counters arrive through ``goods``.

    Returns
    -------
    Dict[str, bool]
        One verdict per invariant, keyed by bare name, in fixed order.
    """
    # The reason accumulator.
    reasons: Dict[str, str] = goods["reasons"]
    # The verdict mapping, keyed by the configuration's own entry names.
    verdicts: Dict[str, bool] = {}

    def _fail(name: str, reason: str) -> bool:
        """
        Record a failure reason and return a refuting verdict.

        Parameters
        ----------
        name : str
            Invariant name, stamped into the reason key as
            ``P44.<name>``.
        reason : str
            Human-readable justification kept in ``goods["reasons"]``.

        Returns
        -------
        bool
            Always ``False``.
        """
        # Record the failure reason for the report.
        reasons[f"P44.{name}"] = reason
        # Refute the verdict.
        return False

    # The pinned acquisition constants.
    universe_size: int = int(config["P11_ACQUISITION_PARAMETERS"]["universe_size"])
    # The pinned trading-date count.
    trading_days: int = int(config["P11_ACQUISITION_PARAMETERS"]["trading_date_count"])

    # ---- file_count_equals_80: FILES row count equals the pinned count -----------------------
    try:
        # The pinned acquisition file count.
        file_count: int = int(config["P11_ACQUISITION_PARAMETERS"]["file_count"])
        # The verdict binds the manifest to the pin.
        verdicts["file_count_equals_80"] = bool(int(goods["files_rows"]) == file_count)
        # Record the failure when the manifest does not bind.
        if not verdicts["file_count_equals_80"]:
            # Record the reason.
            _fail("file_count_equals_80", f"FILES has {goods['files_rows']} rows; the pinned count is {file_count}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["file_count_equals_80"] = _fail("file_count_equals_80", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- rows_per_bar_file_equals_727: each view spans the pinned product ---------------------
    try:
        # The expected row count per view.
        expected_rows: int = universe_size * trading_days
        # Both views must span exactly the pinned product.
        verdicts["rows_per_bar_file_equals_727"] = bool(
            int(goods["unadjusted_rows"]) == expected_rows and int(goods["adjusted_rows"]) == expected_rows
        )
        # Record the failure when either view misses the pin.
        if not verdicts["rows_per_bar_file_equals_727"]:
            # Record the reason.
            _fail("rows_per_bar_file_equals_727", f"una={goods['unadjusted_rows']}, adj={goods['adjusted_rows']}; pinned={expected_rows}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["rows_per_bar_file_equals_727"] = _fail("rows_per_bar_file_equals_727", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- calendar_rows_equals_727: the dense calendar spans the pinned rows -------------------
    try:
        # The pinned calendar row count.
        pinned_rows: int = int(config["P33_SCHEMA_TRADING_CALENDAR"]["estimated_row_count"])
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["calendar_rows_equals_727"] = _fail("calendar_rows_equals_727", "calendar was not built on this run")
        else:
            # The verdict binds the calendar to the pin.
            verdicts["calendar_rows_equals_727"] = bool(int(goods["calendar_rows"]) == pinned_rows)
            # Record the failure when the calendar misses the pin.
            if not verdicts["calendar_rows_equals_727"]:
                # Record the reason.
                _fail("calendar_rows_equals_727", f"calendar has {goods['calendar_rows']} rows; pinned={pinned_rows}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["calendar_rows_equals_727"] = _fail("calendar_rows_equals_727", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- calendar_year_counts_242_242_243: exact per-year census --------------------------------
    try:
        # The pinned per-year census.
        pinned_counts: Mapping[Any, Any] = config["P11_ACQUISITION_PARAMETERS"]["calendar_year_row_counts"]
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["calendar_year_counts_242_242_243"] = _fail("calendar_year_counts_242_242_243", "calendar was not built on this run")
        else:
            # The actual per-year census.
            actual_counts = {2023: goods["calendar_2023_rows"], 2024: goods["calendar_2024_rows"], 2025: goods["calendar_2025_rows"]}
            # Every pinned year must match exactly.
            verdicts["calendar_year_counts_242_242_243"] = bool(
                all(int(actual_counts[int(year)]) == int(count) for year, count in pinned_counts.items())
            )
            # Record the failure when any pinned year misses.
            if not verdicts["calendar_year_counts_242_242_243"]:
                # Record the reason.
                _fail("calendar_year_counts_242_242_243", f"year census {actual_counts} does not match the pinned {dict(pinned_counts)}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["calendar_year_counts_242_242_243"] = _fail("calendar_year_counts_242_242_243", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- label_window_rows_equals_485: the in-label-window census equals the pin ---------------
    try:
        # The pinned label-window row count.
        pinned_window: int = int(config["P01_LABEL_EQUATION_PARAMETERS"]["label_window_trading_rows"])
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["label_window_rows_equals_485"] = _fail("label_window_rows_equals_485", "calendar was not built on this run")
        else:
            # The verdict binds the census to the pin.
            verdicts["label_window_rows_equals_485"] = bool(int(goods["in_label_window_rows"]) == pinned_window)
            # Record the failure when the census misses the pin.
            if not verdicts["label_window_rows_equals_485"]:
                # Record the reason.
                _fail("label_window_rows_equals_485", f"in-window rows={goods['in_label_window_rows']}; pinned={pinned_window}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["label_window_rows_equals_485"] = _fail("label_window_rows_equals_485", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- terminal_unlabelled_rows_equals_5: in-window rows without a successor -------------------
    try:
        # The pinned terminal row count.
        pinned_terminal: int = int(config["P01_LABEL_EQUATION_PARAMETERS"]["terminal_unlabelled_rows"])
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["terminal_unlabelled_rows_equals_5"] = _fail("terminal_unlabelled_rows_equals_5", "calendar was not built on this run")
        else:
            # The verdict binds the census to the pin.
            verdicts["terminal_unlabelled_rows_equals_5"] = bool(int(goods["terminal_unlabelled_rows"]) == pinned_terminal)
            # Record the failure when the census misses the pin.
            if not verdicts["terminal_unlabelled_rows_equals_5"]:
                # Record the reason.
                _fail("terminal_unlabelled_rows_equals_5", f"terminal rows={goods['terminal_unlabelled_rows']}; pinned={pinned_terminal}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["terminal_unlabelled_rows_equals_5"] = _fail("terminal_unlabelled_rows_equals_5", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- label_available_rows_equals_480: the label-available census equals the pin -------------
    try:
        # The pinned label-available row count.
        pinned_available: int = int(config["P01_LABEL_EQUATION_PARAMETERS"]["label_available_rows"])
        # A missing calendar refutes the verdict.
        if goods["calendar_rows"] is None:
            # Refute with the missing artifact.
            verdicts["label_available_rows_equals_480"] = _fail("label_available_rows_equals_480", "calendar was not built on this run")
        else:
            # The verdict binds the census to the pin.
            verdicts["label_available_rows_equals_480"] = bool(int(goods["label_available_rows"]) == pinned_available)
            # Record the failure when the census misses the pin.
            if not verdicts["label_available_rows_equals_480"]:
                # Record the reason.
                _fail("label_available_rows_equals_480", f"available rows={goods['label_available_rows']}; pinned={pinned_available}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["label_available_rows_equals_480"] = _fail("label_available_rows_equals_480", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- masked_cell_rows_equals_2: the mask register carries exactly the pinned cells ----------
    try:
        # The pinned masked-cell count.
        pinned_masked: int = int(config["P12_MASKING_PARAMETERS"]["masked_cell_count"])
        # A missing register refutes the verdict.
        if goods["mask_register_rows"] is None:
            # Refute with the missing artifact.
            verdicts["masked_cell_rows_equals_2"] = _fail("masked_cell_rows_equals_2", "mask register was not produced on this run")
        else:
            # The verdict binds the register length to the pin.
            verdicts["masked_cell_rows_equals_2"] = bool(int(goods["mask_register_rows"]) == pinned_masked)
            # Record the failure when the register misses the pin.
            if not verdicts["masked_cell_rows_equals_2"]:
                # Record the reason.
                _fail("masked_cell_rows_equals_2", f"mask register rows={goods['mask_register_rows']}; pinned={pinned_masked}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["masked_cell_rows_equals_2"] = _fail("masked_cell_rows_equals_2", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- forward_fill_rows_equals_0: this run filled nothing forward -----------------------------
    try:
        # A missing controller refutes the operand (never an inferred zero).
        if goods.get("anti_imputation_counter_block") == "counter_block_missing" or goods.get("forward_fill_operations") is None:
            # Refute with the missing block named.
            verdicts["forward_fill_rows_equals_0"] = _fail("forward_fill_rows_equals_0", "counter_block_missing: the run-scoped anti-imputation controller was not threaded into this run")
        else:
            # The verdict binds the forward-fill counter to zero.
            verdicts["forward_fill_rows_equals_0"] = bool(int(goods["forward_fill_operations"]) == 0)
            # Record the failure when any forward-fill operation ran.
            if not verdicts["forward_fill_rows_equals_0"]:
                # Record the reason with the counter value.
                _fail("forward_fill_rows_equals_0", f"forward-fill counter={goods['forward_fill_operations']}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["forward_fill_rows_equals_0"] = _fail("forward_fill_rows_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- stock_prefix_reductions_equals_0: this run reduced nothing by stock prefix --------------
    try:
        # A missing controller refutes the operand (never an inferred zero).
        if goods.get("anti_imputation_counter_block") == "counter_block_missing" or goods.get("stock_prefix_reductions") is None:
            # Refute with the missing block named.
            verdicts["stock_prefix_reductions_equals_0"] = _fail("stock_prefix_reductions_equals_0", "counter_block_missing: the run-scoped anti-imputation controller was not threaded into this run")
        else:
            # The verdict binds the stock-prefix reduction counter to zero.
            verdicts["stock_prefix_reductions_equals_0"] = bool(int(goods["stock_prefix_reductions"]) == 0)
            # Record the failure when any stock-prefix reduction ran.
            if not verdicts["stock_prefix_reductions_equals_0"]:
                # Record the reason with the counter value.
                _fail("stock_prefix_reductions_equals_0", f"stock-prefix reductions={goods['stock_prefix_reductions']}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["stock_prefix_reductions_equals_0"] = _fail("stock_prefix_reductions_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- oos_sharpe_non_null_rows_before_freeze_equals_0: pre-freeze OOS cells are all null ------
    try:
        # The pre-freeze OOS sentinel censuses.
        oos_sharpe_non_null = goods.get("oos_sharpe_non_null_rows")
        # The pre-freeze OOS timestamp census.
        oos_computed_non_null = goods.get("oos_computed_at_non_null_rows")
        # A missing sentinel refutes the verdict; null sentinels confirm it.
        verdicts["oos_sharpe_non_null_rows_before_freeze_equals_0"] = bool(
            oos_sharpe_non_null == 0 and oos_computed_non_null == 0
        )
        # Record the failure when any pre-freeze OOS cell is populated.
        if not verdicts["oos_sharpe_non_null_rows_before_freeze_equals_0"]:
            # Record the reason.
            _fail("oos_sharpe_non_null_rows_before_freeze_equals_0", f"pre-freeze OOS cells populated: oos_sharpe={oos_sharpe_non_null}, oos_computed_at={oos_computed_non_null}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["oos_sharpe_non_null_rows_before_freeze_equals_0"] = _fail("oos_sharpe_non_null_rows_before_freeze_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- imputed_zero_alpha_records_equals_0: no missing outcome was imputed as zero ------------
    try:
        # A missing controller or an unobservable census refutes the operand (never an inferred zero).
        if goods.get("anti_imputation_counter_block") == "counter_block_missing" or goods.get("imputed_zero_alpha_records") is None:
            # Refute with the missing block named.
            verdicts["imputed_zero_alpha_records_equals_0"] = _fail("imputed_zero_alpha_records_equals_0", "counter_block_missing: the run-scoped anti-imputation controller was not threaded into this run, or the zero-alpha census is unobservable")
        else:
            # The verdict binds the imputed-zero-alpha census to zero.
            verdicts["imputed_zero_alpha_records_equals_0"] = bool(int(goods["imputed_zero_alpha_records"]) == 0)
            # Record the failure when any imputed zero appears.
            if not verdicts["imputed_zero_alpha_records_equals_0"]:
                # Record the reason with the tally value.
                _fail("imputed_zero_alpha_records_equals_0", f"imputed zero alpha records={goods['imputed_zero_alpha_records']}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["imputed_zero_alpha_records_equals_0"] = _fail("imputed_zero_alpha_records_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- unassigned_ledger_records_equals_0: the ledger census is total --------------------------
    try:
        # The unassigned-cell census (or None when the aggregates did not run).
        unassigned = goods.get("unassigned_state_cells")
        # A missing census refutes the verdict.
        if unassigned is None:
            # Refute with the missing artifact.
            verdicts["unassigned_ledger_records_equals_0"] = _fail("unassigned_ledger_records_equals_0", "factor aggregates were not produced on this run")
        else:
            # The verdict binds the unassigned census to zero.
            verdicts["unassigned_ledger_records_equals_0"] = bool(int(unassigned) == 0)
            # Record the failure when any cell is unassigned.
            if not verdicts["unassigned_ledger_records_equals_0"]:
                # Record the reason.
                _fail("unassigned_ledger_records_equals_0", f"unassigned ledger cells={unassigned}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["unassigned_ledger_records_equals_0"] = _fail("unassigned_ledger_records_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- catalog_non_positive_oos_rows_equals_0: the retained catalog carries no non-positives ----
    try:
        # The non-positive row census (or None when the catalog is unusable).
        non_positive = goods.get("catalog_non_positive_rows")
        # A missing census refutes the verdict.
        if non_positive is None:
            # Refute with the missing artifact.
            verdicts["catalog_non_positive_oos_rows_equals_0"] = _fail("catalog_non_positive_oos_rows_equals_0", "outcome_filtered_catalog missing or unscored")
        else:
            # The verdict binds the census to zero.
            verdicts["catalog_non_positive_oos_rows_equals_0"] = bool(int(non_positive) == 0)
            # Record the failure when any non-positive row is present.
            if not verdicts["catalog_non_positive_oos_rows_equals_0"]:
                # Record the reason.
                _fail("catalog_non_positive_oos_rows_equals_0", f"non-positive OOS rows in the catalog={non_positive}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["catalog_non_positive_oos_rows_equals_0"] = _fail("catalog_non_positive_oos_rows_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- sign_disagreements_in_intersection_equals_0: checkpoint vs catalog agree on signs -------
    try:
        # The sign-disagreement census (or None on a missing artifact).
        disagreements = goods.get("sign_disagreements_intersection")
        # A missing census refutes the verdict.
        if disagreements is None:
            # Refute with the missing artifact.
            verdicts["sign_disagreements_in_intersection_equals_0"] = _fail("sign_disagreements_in_intersection_equals_0", "checkpoint/catalog artifacts missing or unscored")
        else:
            # The verdict binds the census to zero.
            verdicts["sign_disagreements_in_intersection_equals_0"] = bool(int(disagreements) == 0)
            # Record the failure when any sign disagrees.
            if not verdicts["sign_disagreements_in_intersection_equals_0"]:
                # Record the reason.
                _fail("sign_disagreements_in_intersection_equals_0", f"sign disagreements in the intersection={disagreements}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["sign_disagreements_in_intersection_equals_0"] = _fail("sign_disagreements_in_intersection_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- duplicate_rows_per_stock_date_equals_0: the compound key is unique in both views -------
    try:
        # The operand is the census computed above (compound or flat; None when unobservable).
        duplicate_census = goods.get("duplicate_stock_date_rows")
        # A missing census refutes the invariant with the named sources.
        if duplicate_census is None:
            # Refute with both named census sources.
            verdicts["duplicate_rows_per_stock_date_equals_0"] = _fail(
                "duplicate_rows_per_stock_date_equals_0",
                f"duplicate census unobservable on this input (sources: unadjusted={goods.get('duplicate_census_source_unadjusted')}, adjusted={goods.get('duplicate_census_source_adjusted')})",
            )
        else:
            # The verdict binds the duplicated-coordinate census to zero.
            verdicts["duplicate_rows_per_stock_date_equals_0"] = bool(int(duplicate_census) == 0)
            # Record the failure when any duplicate coordinate exists.
            if not verdicts["duplicate_rows_per_stock_date_equals_0"]:
                # Record the reason.
                _fail("duplicate_rows_per_stock_date_equals_0", f"duplicate (stock_code, trade_date) rows={duplicate_census}")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["duplicate_rows_per_stock_date_equals_0"] = _fail("duplicate_rows_per_stock_date_equals_0", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- retrieval_coherence_window_not_exceeded: at most one Shanghai equity session ----------
    try:
        # The span measured on the dense calendar by the Task 1/Task 3 sibling verifiers (None when unevaluable).
        span_sessions = goods.get("retrieval_coherence_span_sessions")
        # A split between the two verifiers voids the invariant set (arbitrator behaviour).
        if bool(goods.get("coherence_verdict_split")):
            # Refute with the split named exactly once.
            verdicts["retrieval_coherence_window_not_exceeded"] = _fail(
                "retrieval_coherence_window_not_exceeded",
                "coherence_verdict_split: the Task 1 calendar arithmetic and the Task 3 sibling verifier disagree on the same FILES/calendar; the invariant set cannot be certified",
            )
        elif span_sessions is None:
            # Refute with the missing substrate.
            verdicts["retrieval_coherence_window_not_exceeded"] = _fail(
                "retrieval_coherence_window_not_exceeded",
                "FILES.retrieved is missing, null or unparseable, or the dense trading calendar is unavailable; the one-session window is unevaluable",
            )
        else:
            # The pinned assertion (detailed_task_list_20261111_144334.md line 28: "the interval spans <= 1 trading session").
            verdicts["retrieval_coherence_window_not_exceeded"] = bool(int(span_sessions) <= 1)
            # Record the failure when the window is exceeded.
            if not verdicts["retrieval_coherence_window_not_exceeded"]:
                # Record the reason with the session count.
                _fail("retrieval_coherence_window_not_exceeded", f"retrieval coherence span={span_sessions} session(s) on the Shanghai equity trading calendar")
    except Exception as exc:
        # A defective operand refutes the verdict.
        verdicts["retrieval_coherence_window_not_exceeded"] = _fail("retrieval_coherence_window_not_exceeded", f"evaluation breached: {type(exc).__name__}: {exc}")

    # ---- Cross-verifier arbitration: a coherence split voids the whole invariant set -------------
    if bool(goods.get("coherence_verdict_split")):
        # Every verdict in the set is flipped False; the arbitrator never silently resolves a split.
        for name in list(verdicts):
            # Flip the verdict to False.
            verdicts[name] = False
        # Record the split as the reason on the coherence entry and once on every member.
        reasons["P44.retrieval_coherence_window_not_exceeded"] = (
            "coherence_verdict_split: the Task 1 calendar arithmetic and the Task 3 sibling verifier "
            "disagree on the same FILES/calendar; the invariant set cannot be certified"
        )
    # Return the verdict mapping (insertion order pinned by the branches above).
    return verdicts

def _assemble_block_verdicts(
    p42: Mapping[str, bool],
    p44: Mapping[str, bool],
) -> Dict[str, bool]:
    """
    Compose the pinned ``block_verdicts`` mapping.

    Prefixes every P42 identity name with ``P42.`` and every P44
    invariant name with ``P44.``, preserving the insertion order of each
    input mapping, and casts each value to a native ``bool``.

    Parameters
    ----------
    p42 : Mapping[str, bool]
        The P42 arithmetic-closure verdicts, keyed by bare identity
        name.
    p44 : Mapping[str, bool]
        The P44 raw-layer invariant verdicts, keyed by bare invariant
        name.

    Returns
    -------
    Dict[str, bool]
        All ``P42.<identity>`` keys first, then all ``P44.<invariant>``
        keys, every value a native ``bool``.
    """
    # The composed mapping.
    verdicts: Dict[str, bool] = {}
    # Walk the P42 verdicts in order.
    for name, verdict in p42.items():
        # Stamp the prefixed key with a plain bool.
        verdicts[f"P42.{name}"] = bool(verdict)
    # Walk the P44 verdicts in order.
    for name, verdict in p44.items():
        # Stamp the prefixed key with a plain bool.
        verdicts[f"P44.{name}"] = bool(verdict)
    # Return the composed mapping.
    return verdicts

def _claim_from_task_10_record(record: Mapping[str, Any]) -> ClaimRecord:
    """
    Convert one Task 10 claim-record dict into a frozen ClaimRecord.

    The record is refused loudly rather than dropped from the register.
    The four-gate mapping is renormalised to native ``bool`` values
    keyed by the gate identifiers in ``CLAIM_GATE_KEYS``, in the
    upstream insertion order. The record's ``evidence_hash`` is
    forwarded to ``ClaimRecord``, which re-stamps the digest from the
    pinned material basis and rejects an upstream digest that does not
    bind it.

    Parameters
    ----------
    record : Mapping[str, Any]
        A record emitted by the Task 10 closure battery. Must carry
        ``claim_id``, ``tier``, ``evidence_source`` and ``verdict``,
        plus a ``gate_status`` mapping holding exactly the gate keys in
        ``CLAIM_GATE_KEYS`` with native ``bool`` values. ``value``,
        ``note`` and ``evidence_hash`` are optional.

    Returns
    -------
    ClaimRecord
        The frozen record; ``claim_id``, ``tier``, ``evidence_source``
        and ``verdict`` are stringified and ``note`` defaults to ``""``.

    Raises
    ------
    TypeError
        If ``record`` or its ``gate_status`` is not a ``Mapping``, or a
        gate value is not a native ``bool``.
    KeyError
        If any of ``claim_id``, ``tier``, ``evidence_source`` or
        ``verdict`` is absent from the record.
    ValueError
        If ``gate_status`` carries a key outside ``CLAIM_GATE_KEYS``,
        omits one of them, or if ``ClaimRecord`` rejects the supplied
        ``evidence_hash`` as not binding the record's material basis.
    """
    # The record must be a mapping carrying the pinned fields.
    if not isinstance(record, Mapping):
        # A non-mapping record is an upstream defect; refuse it loudly.
        raise TypeError(f"Task 10 claim record must be a mapping, got {type(record).__name__}")
    # The pinned scalar fields that must be present on every record.
    missing: List[str] = [key for key in ("claim_id", "tier", "evidence_source", "verdict") if key not in record]
    # A record lacking a pinned field cannot be converted.
    if missing:
        # Name the record and the absent fields.
        raise KeyError(f"Task 10 claim record {record.get('claim_id', '<no claim_id>')!r} lacks pinned field(s) {missing}")
    # The upstream gate map, snapshot in its insertion order.
    upstream_gates: Any = record.get("gate_status")
    # The upstream record must carry a conforming gate mapping.
    if not isinstance(upstream_gates, Mapping):
        # A record without the pinned gates is not admissible.
        raise TypeError(f"Task 10 claim record {record['claim_id']!r}: gate_status must be a mapping, got {type(upstream_gates).__name__}")
    # Normalise the mapping to plain native bools keyed by G1..G4 (no other keys may appear).
    gate_status: Dict[str, bool] = {}
    # Snapshot the upstream mapping verbatim (insertion order preserved).
    for key, item in upstream_gates.items():
        # The key must be one of the four pinned gate identifiers.
        if key not in CLAIM_GATE_KEYS:
            # A foreign gate key means the upstream record is malformed.
            raise ValueError(f"Task 10 claim record {record['claim_id']!r}: foreign gate key {key!r} (allowed {tuple(CLAIM_GATE_KEYS)})")
        # The value must be a native bool.
        if type(item) is not bool:
            # A non-native gate value is malformed.
            raise TypeError(f"Task 10 claim record {record['claim_id']!r}: gate {key!r} must be a native bool, got {type(item).__name__}")
        # Record the native bool.
        gate_status[str(key)] = bool(item)
    # The four gates must all be present, each once.
    if set(gate_status) != set(CLAIM_GATE_KEYS):
        # An incomplete gate map is malformed.
        raise ValueError(f"Task 10 claim record {record['claim_id']!r}: gate_status missing {sorted(set(CLAIM_GATE_KEYS) - set(gate_status))}")
    # Build the frozen record; ClaimRecord re-stamps the digest from the pinned material
    # basis and raises ValueError when a supplied upstream digest does not bind it.
    return ClaimRecord(
        # The claim identifier.
        claim_id=str(record["claim_id"]),
        # The evidentiary tier.
        tier=str(record["tier"]),
        # The evidence source.
        evidence_source=str(record["evidence_source"]),
        # The four-gate admissibility basis (verbatim from the battery).
        gate_status=gate_status,
        # The verdict literal.
        verdict=str(record["verdict"]),
        # The value carried.
        value=record.get("value"),
        # The evidentiary note.
        note=str(record.get("note", "")),
        # The upstream digest (verified against the canonical recomputation at construction).
        evidence_hash=str(record.get("evidence_hash") or ""),
    )

def _build_claim_register(
    goods: Dict[str, Any],
    block_verdicts: Mapping[str, bool],
) -> Tuple[ClaimRecord, ...]:
    """
    Assemble the deterministic claim register for this run.

    Emits the aggregate stage's claim records in emission order, then
    appends the terminal P42/P44 closure record, whose ``verdict`` is
    ``"VERIFIED"`` only when every entry of ``block_verdicts`` is true.

    Parameters
    ----------
    goods : Dict[str, Any]
        The operand mapping; ``goods["closure_battery"]`` supplies the
        upstream records and the gate material.
    block_verdicts : Mapping[str, bool]
        The composed terminal verdicts.

    Returns
    -------
    Tuple[ClaimRecord, ...]
        The aggregate records in emission order, then the terminal
        record with ``claim_id`` ``"P42_P44_CLOSURES__this_run"``.
    """
    # The accumulator.
    records: List[ClaimRecord] = []
    # The aggregate stage's battery carries its records.
    battery: Any = goods.get("closure_battery")
    # Emit the battery records in order (skipping none, converting each).
    if isinstance(battery, Mapping):
        # The claim_records list, in emission order.
        for record in battery.get("claim_records", []) or []:
            # Convert one record (a malformed record raises; none is skipped) and append in emission order.
            records.append(_claim_from_task_10_record(record))
    # The terminal verdict census.
    n_verdicts: int = int(len(block_verdicts))
    # The number of True verdicts.
    n_true: int = int(sum(1 for verdict in block_verdicts.values() if verdict))
    # The orchestrator's terminal record always exists.
    records.append(
        ClaimRecord(
            # The claim identifier of the terminal verification.
            claim_id="P42_P44_CLOSURES__this_run",
            # Tier reproduction (verifiable on this run's own outputs).
            tier="reproduction",
            # The evidence source is the terminal verification itself.
            evidence_source="terminal_closure_verification_task_14",
            # The four-gate admissibility basis, mapped from this run's terminal verification.
            gate_status=_gate_status_from_terminal(goods, block_verdicts),
            # The verdict literal.
            verdict="VERIFIED" if n_true == n_verdicts else "NOT_VERIFIED",
            # The count census.
            value={"n_verdicts": n_verdicts, "n_true": n_true, "n_false": n_verdicts - n_true},
            # The note (never a placeholder).
            note=(
                "Every P42 arithmetic closure (fifteen identities, evaluated constituent-wise) and "
                "every P44 raw-layer invariant in the study configuration (seventeen entries, every "
                "entry evaluated) were verified on this run's own outputs. A False verdict is a "
                "non-verified identity, not an observed violation; the per-identity reasons are in "
                "admissibility_report['closures']['not_verified']. No paper figure is embedded."
            ),
        )
    )
    # Return the deterministic register.
    return tuple(records)

def _p41_comparison_values(config: Dict[str, Any]) -> Dict[str, Any]:
    """
    Disclose the P41 comparison register as not reproducible here.

    The paper's P41 figures (the Table 3 ledger triple, the scorable
    partition, the reference means, the medians, the retention-audit
    magnitudes and the freeze/cohort counts) need the private per-record
    artifacts this project does not hold, and
    ``P41_ACCEPTANCE_TARGETS_STUDY_SPECIFIC`` itself declares
    ``reachable_with_private_artifacts_only``. No figure is therefore
    returned; only the register's existence, its pinned provenance tags,
    the size of the pinned comparison table and the DATA_ABSENT gap are
    reported.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration. Only the ``"values"`` mapping of
        ``P41_ACCEPTANCE_TARGETS_STUDY_SPECIFIC`` is touched, and only
        to count its entries; no figure is read into the result.

    Returns
    -------
    Dict[str, Any]
        The disclosure record with keys ``"status"`` (always
        ``"DATA_ABSENT"``), ``"reachable"``
        (``"PRIVATE_ARTIFACTS_ONLY"``), ``"tag"``
        (``"COMPARISON__[S]__PRIVATE_ARTIFACTS_ONLY"``),
        ``"n_comparison_values_in_configuration"`` (int, the number of
        pinned comparison entries), ``"value_format"``, ``"reason"`` and
        ``"policy"`` (strings explaining the omission and the
        non-propagation rule).
    """
    # The source values mapping (never returned, only counted).
    values: Mapping[str, Any] = config["P41_ACCEPTANCE_TARGETS_STUDY_SPECIFIC"]["values"]
    # The count of entries in the pinned comparison table (a property of the configuration, never a measurement).
    n_values: int = int(len(values))
    # The disclosure record.
    disclosure: Dict[str, Any] = {
        # The honest-exit status of the whole register.
        "status": "DATA_ABSENT",
        # The pinned reachable tag (provenance class of every P41 figure).
        "reachable": "PRIVATE_ARTIFACTS_ONLY",
        # The pinned comparison tag.
        "tag": "COMPARISON__[S]__PRIVATE_ARTIFACTS_ONLY",
        # The count of entries in the pinned comparison table (configuration property, not a result).
        "n_comparison_values_in_configuration": n_values,
        # The declared format (never floats; never observations).
        "value_format": "NOT_RETURNED__PRIVATE_ARTIFACTS_ONLY__see_acceptance_objections_task_14.md",
        # The DATA_ABSENT reason.
        "reason": (
            "the frozen 1,047 per-record outcomes, the frozen ID list, the disclosed 40-stock universe, "
            "the grammar and the private per-record artifacts are not held by this project; every P41 "
            "figure is a comparison target only and is not computable from the inputs available here"
        ),
        # The policy governing these figures everywhere in this reproduction.
        "policy": (
            "no P41 comparison value is returned anywhere in this reproduction; they remain the "
            "configuration's comparison table (P41_ACCEPTANCE_TARGETS_STUDY_SPECIFIC), stamped by this "
            "register; downstream consumers read the admissibility_report only and may not promote any "
            "P41 figure to an observation"
        ),
    }
    # Return the disclosure record.
    return disclosure

def _r_choices(config: Dict[str, Any]) -> Dict[str, Any]:
    """
    Render the [R] reproducibility register with the pinned literals.

    Parameters
    ----------
    config : Dict[str, Any]
        The study configuration; supplies
        ``P04_LEDGER_PARAMETERS["state_precedence"]`` and
        ``P06_COHORT_ADMISSION_PARAMETERS["deduplication_canonicalization"]``.

    Returns
    -------
    Dict[str, Any]
        The seven pinned choices: ``"state_precedence"`` (list, the
        ledger resolution order), ``"dedup_canonicalization"`` (dict,
        canonical form per expression), ``"spot_check_tolerance"``
        (float, the absolute tolerance
        ``TOLERANCE_SPOT_CHECK_ABSOLUTE``),
        ``"freeze_identifier_format"`` (``"string-verbatim"``),
        ``"masked_cell_semantics"`` (``"two distinct stock-days"``),
        ``"dedup_representative_rule"`` (``"lowest row_id"``) and
        ``"sd_definition"`` (``"ddof=1"``).
    """
    # The [R] register (every value pinned by the standing register / Task 14 step 3).
    choices: Dict[str, Any] = {
        # The pinned state precedence (P04.state_precedence; [R], NOT_STATED_IN_PAPER).
        "state_precedence": list(config["P04_LEDGER_PARAMETERS"]["state_precedence"]),
        # The pinned dedup canonicalization mapping (P06.deduplication_canonicalization; [R]).
        "dedup_canonicalization": dict(config["P06_COHORT_ADMISSION_PARAMETERS"]["deduplication_canonicalization"]),
        # The pinned spot-check absolute tolerance (Task 14 step 3; [R]; 1e-10).
        "spot_check_tolerance": TOLERANCE_SPOT_CHECK_ABSOLUTE,
        # The pinned freeze identifier format ([R]; string-verbatim).
        "freeze_identifier_format": "string-verbatim",
        # The pinned masked-cell semantics (P34; two distinct stock-days).
        "masked_cell_semantics": "two distinct stock-days",
        # The pinned dedup representative rule (P06; lowest row_id).
        "dedup_representative_rule": "lowest row_id",
        # The pinned SD definition (P02 [R]; ddof=1, disclosed).
        "sd_definition": "ddof=1",
    }
    # Return the register.
    return choices

def _run_exposed_gaps(
    goods: Dict[str, Any],
    stages_halt: Mapping[str, Optional[str]],
    context: PipelineContext,
    config: Dict[str, Any],
) -> List[Dict[str, str]]:
    """
    Collect the disclosure gaps that this run itself exposes.

    Every gap is detected from the run state, never invented. Gaps are
    emitted, in this order, when: ``goods["files_has_file_uri"]`` is
    ``False``; ``goods["files_verified_na_count"]`` is non-zero; a stage
    halt is neither ``None`` nor ``STAGE_NOT_REQUESTED``;
    ``goods["task_13_data_gap"]`` equals ``"DATA_ABSENT"``; the freeze
    state carries an empty ``cohort_hash``; the score manifest carries a
    non-empty ``parse_failed_ids``;
    ``context.payloads["mask_status_na_cells"]`` is truthy;
    ``context.payloads["mask_status_corroboration"]`` holds entries
    whose ``verdict`` is ``"disagreement"`` (each coordinate is listed);
    or ``config["P13_DSL_EVALUATOR_INTERFACE"]["grammar_published"]`` is
    not ``True``.

    Parameters
    ----------
    goods : Dict[str, Any]
        Scalar operand mapping holding ``files_has_file_uri``,
        ``files_verified_na_count`` and ``task_13_data_gap``.
    stages_halt : Mapping[str, Optional[str]]
        Per-stage halt text; ``None`` marks success.
    context : PipelineContext
        Active pipeline context; ``freeze_state``, ``manifests`` and
        ``payloads`` are consulted.
    config : Dict[str, Any]
        Study configuration carrying
        ``P13_DSL_EVALUATOR_INTERFACE``.

    Returns
    -------
    List[Dict[str, str]]
        One entry per detected gap, in detection order, each with keys
        ``"gap"`` (the identifier) and ``"evidence"`` (the
        justification). Empty when the run exposes nothing.
    """
    # The accumulator.
    gaps: List[Dict[str, str]] = []
    # Detect an absent file_uri substrate (custody cannot rehash).
    if goods.get("files_has_file_uri") is False:
        # Record the gap with its evidence.
        gaps.append(
            {
                # The gap identifier.
                "gap": "FILES carries no file_uri on this run",
                # The evidence.
                "evidence": "Task 3 custody rehash is impossible; no hash_manifest.json is produced; custody can only halt, never verify",
            }
        )
    # Detect an NA in FILES.verified (a claim about the acquisition that cannot be made).
    na_verified: Optional[int] = goods.get("files_verified_na_count")
    # A non-zero census is an exposed gap.
    if na_verified:
        # Record the gap with its evidence.
        gaps.append(
            {
                # The gap identifier.
                "gap": "NA present in FILES.verified on this run",
                # The evidence.
                "evidence": f"{na_verified} manifest row(s) carry an undetermined verification verdict",
            }
        )
    # Record one gap per halted stage.
    for stage, halt in stages_halt.items():
        # A non-None halt is an exposed gap of the run.
        if halt is not None and halt != STAGE_NOT_REQUESTED:
            # Record the gap with its evidence.
            gaps.append(
                {
                    # The gap identifier.
                    "gap": f"stage '{stage}' halted on this run",
                    # The evidence.
                    "evidence": str(halt),
                }
            )
    # The Task 13 honest-exit stamp (reconstruction-tier evidence).
    data_gap: Optional[str] = goods.get("task_13_data_gap")
    # A reconstruction-tier ledger is an exposed gap, always disclosed.
    if data_gap is not None and str(data_gap) == "DATA_ABSENT":
        # Record the gap with its evidence.
        gaps.append(
            {
                # The gap identifier.
                "gap": "private per-record score registers are not held by this project",
                # The evidence.
                "evidence": "Task 13 produced a declared RECONSTRUCTION_FIXTURE (867/98/25 unchanged; 57 v1 errors; 56 recover to scorable; 1 residual insufficient); the corrective ledger is reconstruction-tier, never an observation",
            }
        )
    # The cohort freeze state on this run.
    cohort_hash: str = str(context.freeze_state.get("cohort_hash", ""))
    # An unfrozen cohort is an exposed gap.
    if not cohort_hash:
        # Record the gap with its evidence.
        gaps.append(
            {
                # The gap identifier.
                "gap": "no cohort was frozen on this run",
                # The evidence.
                "evidence": "the freeze stage did not complete; cohort_hash carries the NOT_FROZEN sentinel; cohort_timestamp carries the run instant, not a freeze instant",
            }
        )
    # The parse-failure census from the score stage.
    parse_failed: Any = context.manifests.get("score", {}).get("parse_failed_ids") if isinstance(context.manifests.get("score"), Mapping) else None
    # A non-empty census is an exposed gap.
    if parse_failed:
        # Record the gap with its evidence.
        gaps.append(
            {
                # The gap identifier.
                "gap": "unparseable expressions present in the frozen cohort",
                # The evidence.
                "evidence": f"parse_failed identifiers: {list(parse_failed)}",
            }
        )
    # The NA-status tradable census of this run (unverified status; never coordinates).
    na_status_cells: Any = context.payloads.get("mask_status_na_cells")
    # A non-zero census is an exposed gap.
    if na_status_cells:
        # Record the gap with the census and the rule.
        gaps.append(
            {
                # The gap identifier.
                "gap": f"{na_status_cells} NA-status tradable cell(s) on this run (unverified status)",
                # The evidence.
                "evidence": "unverified tradable status is disclosed absence; never mapped to False, never a non-trading coordinate; corroboration attempted per located coordinate",
            }
        )
    # The mask-status corroboration register (error-severity disclosure, never a pass).
    corroboration: Any = context.payloads.get("mask_status_corroboration")
    # Collect the disagreements, when any were recorded on this run.
    disagreements = [
        entry
        for entry in corroboration
        if isinstance(entry, Mapping) and entry.get("verdict") == "disagreement"
    ] if isinstance(corroboration, list) else []
    # Any disagreement is named in the exposed-gaps register (an error-severity finding, not a pass).
    if disagreements:
        # Render the complete listing.
        listing = "; ".join(
            f"({entry.get('stock_code')}, {entry.get('trade_date')}) code={entry.get('unadjusted_tradestatus_raw')} mapped={entry.get('unadjusted_mapped')}"
            for entry in disagreements
        )
        # Record the gap with the complete evidence.
        gaps.append(
            {
                # The gap identifier.
                "gap": "mask coordinate(s) whose raw vendor code contradicts the asserted not-trading verdict",
                # The evidence listing.
                "evidence": f"the located coordinates' tradestatus_raw codes map (pinned vendor mapping 1->True, 0->False, else NA) to disagreement at: {listing}; an error-severity disclosure, never a pass, never resolved silently",
            }
        )
    # The unsupported-record count is unreachable from these inputs (grammar undisclosed).
    grammar_disclosed: Any = config.get("P13_DSL_EVALUATOR_INTERFACE", {}).get("grammar_published")
    # A non-True grammar disclosure is an exposed gap.
    if grammar_disclosed is not True:
        # Record the gap with its evidence.
        gaps.append(
            {
                # The gap identifier.
                "gap": "the 25 unsupported records are unreachable from these inputs",
                # The evidence.
                "evidence": "P13_DSL_EVALUATOR_INTERFACE.grammar == UNDISCLOSED and grammar_published == False; the unsupported census is a paper comparison value, never a computation of this run",
            }
        )
    # Return the gaps.
    return gaps

def _build_admissibility_report(
    halt_on_failure: bool,
    stages_halt: Mapping[str, Optional[str]],
    block_verdicts: Mapping[str, bool],
    p42: Mapping[str, bool],
    p44: Mapping[str, bool],
    reasons: Mapping[str, str],
    goods: Dict[str, Any],
    context: PipelineContext,
    claim_register: Tuple[ClaimRecord, ...],
    config: Dict[str, Any],
) -> Dict[str, Any]:
    """
    Assemble the ``admissibility_report`` mapping read downstream.

    Build the single mapping through which downstream consumers read the
    run; a field absent here does not propagate. Blocks, in the pinned
    key order: ``run_mode`` (mode flag, ``PRODUCTION``/``DIAGNOSTIC``
    name, the opening sentence, an ``admissible`` predicate requiring
    every block verdict True and no stage whose halt entry is
    non-``None`` (``STAGE_NOT_REQUESTED`` included), and the
    halted-stage count); ``P40_structural_invariants_verified`` (sorted
    verified names, the not-verified names with their reasons, and a
    scope note); ``P41_comparison_values`` (from
    ``_p41_comparison_values``); ``P45_disclosure_gaps`` (the
    configuration's twelve pinned keys plus ``_run_exposed_gaps``);
    ``R_choices`` (from ``_r_choices``); ``closures`` (``P42.*`` and
    ``P44.*`` verdicts, their conjunction and the per-identity
    reasons); ``stage_verdicts``; ``artifacts`` (cohort hash, freeze
    id, cohort size, correction partition, manifest path, register row
    counts); ``derived_parameters`` (derived OOS-read instant, run
    nonce, scratch root); the claim register census; and the fixed
    non-propagation and honesty banners.

    Parameters
    ----------
    halt_on_failure : bool
        Run mode flag; ``False`` marks a diagnostic run.
    stages_halt : Mapping[str, Optional[str]]
        Per-stage halt text; ``None`` marks success.
    block_verdicts : Mapping[str, bool]
        Composed terminal verdicts.
    p42 : Mapping[str, bool]
        P42 closure verdicts by name.
    p44 : Mapping[str, bool]
        P44 invariant verdicts by name.
    reasons : Mapping[str, str]
        Refutation reason per failed identity.
    goods : Dict[str, Any]
        Scalar operand mapping forwarded to ``_run_exposed_gaps``.
    context : PipelineContext
        Active pipeline context: freeze state, manifests, payloads,
        nonce and scratch root.
    claim_register : Tuple[ClaimRecord, ...]
        Claim records to census into the report.
    config : Dict[str, Any]
        Study configuration carrying ``P45_DISCLOSURE_GAPS`` and
        ``P13_DSL_EVALUATOR_INTERFACE``.

    Returns
    -------
    Dict[str, Any]
        The report mapping with the blocks listed above, in the pinned
        key order.
    """
    # The halt enumeration for the run mode disclosure.
    halt_stages: List[str] = [stage for stage, halt in stages_halt.items() if halt is not None]
    # All verdicts hold iff every entry is True.
    all_verdicts: bool = all(bool(verdict) for verdict in block_verdicts.values())
    # The freeze mirror.
    freeze_meta: Optional[Mapping[str, Any]] = context.freeze_state.get("meta")
    # The run mode block (the opening sentence naming the mode).
    run_mode: Dict[str, Any] = {
        # The mode flag, verbatim.
        "halt_on_failure": bool(halt_on_failure),
        # The pinned name of the output mode.
        "run_mode": "PRODUCTION" if halt_on_failure else "DIAGNOSTIC",
        # The mandatory disclosure sentence (a diagnostic run is never a completed study).
        "opening_sentence": (
            "DIAGNOSTIC run — halt_on_failure == False; the study is incomplete by construction, "
            "every stage's halt is surfaced, and no claim in this output is admissible as a completed study."
            if not halt_on_failure
            else "PRODUCTION run — halt_on_failure == True; the first non-None halt terminated the dispatch."
        ),
        # The admissibility predicate (all verdicts True and no stage halted).
        "admissible": bool(all_verdicts and not halt_stages),
        # The predicate's definition (for the downstream consumer).
        "admissible_predicate": "all P42/P44 verdicts True AND no stage halted on this run",
        # The number of stages that halted.
        "n_stages_halted": int(len(halt_stages)),
    }
    # The P40 verified block (the checkable band, computed on this run's outputs).
    p40: Dict[str, Any] = {
        # The identities verified on this run's own outputs.
        "verified": sorted(name for name, verdict in block_verdicts.items() if verdict),
        # The identities not verified (with the reasons).
        "not_verified": {name: reasons.get(name, "not evaluated on this run (upstream halt or missing artifact)") for name, verdict in block_verdicts.items() if not verdict},
        # The disclosure of the comparison-value boundary.
        "note": (
            "P40 structural targets are the checkable band; every verdict above was computed on "
            "this run's own outputs. The paper's comparison values live under "
            "'P41_comparison_values' with reachable == PRIVATE_ARTIFACTS_ONLY; no comparison value "
            "was asserted as an observation anywhere in this run."
        ),
    }
    # The pinned twelve P45 keys echoed verbatim, plus the run-exposed gaps.
    pinned_p45: Dict[str, Any] = dict(config["P45_DISCLOSURE_GAPS"])
    # The P45 block.
    p45: Dict[str, Any] = {
        # The twelve pinned disclosure keys, echoed as delivered by the configuration.
        "pinned": pinned_p45,
        # The additional gaps this run itself exposes (detected, never invented).
        "run_exposed_gaps": _run_exposed_gaps(goods, stages_halt, context, config),
        # The evaluation note (honesty, not pass/fail).
        "note": (
            "P45 gaps are recorded absences, evaluated for honesty, not pass/fail. The twelve pinned "
            "keys are echoed verbatim; run_exposed_gaps lists only gaps detected on this run."
        ),
    }
    # The closures block (the verdict mapping and the reasons).
    closures: Dict[str, Any] = {
        # The P42 verdicts.
        "P42": {f"P42.{name}": bool(verdict) for name, verdict in p42.items()},
        # The P44 verdicts.
        "P44": {f"P44.{name}": bool(verdict) for name, verdict in p44.items()},
        # The conjunction across every closure and invariant.
        "all_true": bool(all_verdicts),
        # The refutation reasons for the False verdicts.
        "not_verified": dict(reasons),
        # The scope disclosure.
        "note": (
            "Evaluated on this run's own outputs, never against the paper's numbers. P40 structural "
            "targets are the checkable band; P41 comparison values are stamped under 'P41_comparison_values' "
            "with reachable == PRIVATE_ARTIFACTS_ONLY; version_closures and sum_closures_rounds_to_totals "
            "are evaluated constituent-wise and hold only when every constituent holds."
        ),
    }
    # The artifacts mirror.
    artifacts: Dict[str, Any] = {
        # The cohort hash (or the disclosed sentinel).
        "cohort_hash": str(context.freeze_state.get("cohort_hash", NOT_FROZEN_COHORT_HASH)),
        # The freeze id (empty when no freeze ran).
        "freeze_id": str(context.freeze_state.get("freeze_id", "")),
        # The cohort size on this run (0 when no freeze ran).
        "cohort_size": int(context.freeze_state.get("cohort_size", 0)) if context.freeze_state.get("cohort_size") is not None else 0,
        # The corrective partition of this run ({} when the correction never ran).
        "correction_partition": dict(context.payloads.get("correction_partition", {})),
        # The hash manifest path when custody produced one.
        "hash_manifest_path": str(context.manifests.get("custody", {}).get("hash_manifest_path", "")) if isinstance(context.manifests.get("custody"), Mapping) else "",
        # The score register cell counts.
        "score_v1_cells": int(len(context.payloads["daily_scores_v1"])) if isinstance(context.payloads.get("daily_scores_v1"), pd.DataFrame) else 0,
        # The v2 register cell counts.
        "score_v2_cells": int(len(context.payloads["daily_scores_v2"])) if isinstance(context.payloads.get("daily_scores_v2"), pd.DataFrame) else 0,
    }
    # The derived-parameters disclosure (the OOS-read instant is synthesised, disclosed).
    derived_parameters: Dict[str, Any] = {
        # The freeze slot (the freeze instant itself is a Task 8 product).
        "oos_read_start_ts": "derived at run time as Asia/Shanghai now() + 1 day; the configuration pins no absolute first-read instant (P08 PENDING); disclosed, never presented as a paper figure",
        # The nonce namespacing of this run.
        "run_nonce": context.nonce,
        # The scratch root (hermetic; never the project tree).
        "scratch_root": str(context.scratch_root),
    }
    # The report assembly (pinned key order).
    report: Dict[str, Any] = {
        # The run mode block (the opening sentence naming the mode).
        "run_mode": run_mode,
        # (a) the P40 structural invariants verified on this run's outputs.
        "P40_structural_invariants_verified": p40,
        # (b) every P41 comparison value, declared DATA_ABSENT with the pinned reachable and comparison tags.
        "P41_comparison_values": _p41_comparison_values(config),
        # (c) every P45 disclosure gap (pinned twelve plus this run's exposures).
        "P45_disclosure_gaps": p45,
        # (d) the [R] choices the reproduction made (seven pinned keys).
        "R_choices": _r_choices(config),
        # The closures verdicts (P42 + P44, constituent-wise).
        "closures": closures,
        # The per-stage halt states.
        "stage_verdicts": dict(stages_halt),
        # The artifact mirrors.
        "artifacts": artifacts,
        # The derived-parameters disclosure.
        "derived_parameters": derived_parameters,
        # The claim register census.
        "claim_register_census": {
            # The number of records.
            "n_records": int(len(claim_register)),
            # The claim identifiers in order.
            "claim_ids": [record.claim_id for record in claim_register],
        },
        # The non-propagation rule statement.
        "non_propagation_rule": (
            "Only this mapping propagates downstream (Task 16 deploy_decision; the study output's "
            "tiered-claim macros). A field absent from this mapping does not exist for any consumer."
        ),
        # The honesty banner (nothing beyond this run's evidence).
        "honesty_banner": (
            "No paper figure is asserted as an observation anywhere in this run. Reproduction-tier claims "
            "are closed only on this run's own outputs; reconstruction-tier claims are labelled; comparison "
            "values are stamped PRIVATE_ARTIFACTS_ONLY; the private per-record artifacts are not held."
        ),
    }
    # Return the report.
    return report

def _finalise_result(
    halt_on_failure: bool,
    stages_halt: Dict[str, Optional[str]],
    stages_output: Dict[str, Any],
    inputs: Mapping[str, Any],
    context: PipelineContext,
    first_halt_stage: Optional[str],
    graph_halt_reason: Optional[str],
) -> StudyResult:
    """
    Freeze the accumulated run into the immutable ``StudyResult``.

    Build the closure operands with ``_closure_goods``. A graph halt
    marks every pinned P42 name and every ``P44_RAW_LAYER_INVARIANTS``
    entry ``False`` with that reason; otherwise both are evaluated on
    this run's own outputs. The block verdicts, claim register and
    admissibility report are then assembled and the fifteen fields
    populated. Never-run stages fall back to empty pinned-column frames,
    ``(0, 0, 0)`` ledger triples, ``NOT_FROZEN_COHORT_HASH``, and the
    current Asia/Shanghai instant for ``cohort_timestamp``. This is the
    only construction site of ``StudyResult``.

    Parameters
    ----------
    halt_on_failure : bool
        Run mode flag, passed verbatim into the report.
    stages_halt : Dict[str, Optional[str]]
        Per-stage halt text for all thirteen pinned stage names.
    stages_output : Dict[str, Any]
        Payloads of the requested stages, or halted markers.
    inputs : Mapping[str, Any]
        The ``run_study`` arguments, read by key.
    context : PipelineContext
        Active pipeline context; supplies the artifact mirrors.
    first_halt_stage : Optional[str]
        First halted stage or ``None``; accepted but unused.
    graph_halt_reason : Optional[str]
        Closure failure on an unclosed stage subset, or ``None``.

    Returns
    -------
    StudyResult
        The frozen result with every field populated.
    """
    # Assemble the closure operands from this run's artifacts.
    goods: Dict[str, Any] = _closure_goods(
        files=inputs["files"],
        unadjusted=inputs["unadjusted"],
        adjusted=inputs["adjusted"],
        preserved_rows=inputs["preserved_rows"],
        native_oos_checkpoint=inputs["native_oos_checkpoint"],
        outcome_filtered_catalog=inputs["outcome_filtered_catalog"],
        run_report_totals=inputs["run_report_totals"],
        run_report_rounds=inputs["run_report_rounds"],
        collector_a=inputs["collector_a"],
        collector_b=inputs["collector_b"],
        config=inputs["config"],
        context=context,
    )
    # A graph halt refutes every closure with one explicit reason.
    if graph_halt_reason is not None:
        # The halt reason rides verbatim.
        graph_reason: str = graph_halt_reason
        # Nothing was evaluated; the P42 verdicts are all False with the same reason.
        p42: Dict[str, bool] = {}
        # The P44 verdicts are all False with the same reason.
        p44: Dict[str, bool] = {}
        # Evaluate nothing: the pipeline never dispatched; nothing was verified.
        for name in (
            "file_matrix", "calendar", "calendar_2023_residual", "label_availability",
            "set_partition_checkpoint", "set_partition_catalog", "ledger", "ledger_supported",
            "sign_partition", "coverage_partition", "version_closures",
            "implied_pipeline_failures", "non_scorable", "catalog_rule_failures",
            "sum_closures_rounds_to_totals",
        ):
            # Refute the verdict and pin the reason.
            goods["reasons"][f"P42.{name}"] = graph_reason
            # Pin the False verdict.
            p42[name] = False
        # Every P44 entry in the configuration is refuted the same way.
        for name in inputs["config"]["P44_RAW_LAYER_INVARIANTS"]:
            # Refute the verdict and pin the reason.
            goods["reasons"][f"P44.{name}"] = graph_reason
            # Pin the False verdict.
            p44[name] = False
    else:
        # Evaluate the P42 closures on this run's outputs.
        p42 = _evaluate_p42_closures(goods, inputs["config"])
        # Evaluate every P44 invariant in the configuration on this run's outputs.
        p44 = _evaluate_p44_invariants(goods, inputs["config"], context)
    # Compose the block verdicts.
    block_verdicts: Dict[str, bool] = _assemble_block_verdicts(p42, p44)
    # Assemble the deterministic claim register.
    claim_register: Tuple[ClaimRecord, ...] = _build_claim_register(goods, block_verdicts)
    # Assemble the admissibility report (the single interface downstream).
    admissibility_report: Dict[str, Any] = _build_admissibility_report(
        halt_on_failure=halt_on_failure,
        stages_halt=stages_halt,
        block_verdicts=block_verdicts,
        p42=p42,
        p44=p44,
        reasons=goods["reasons"],
        goods=goods,
        context=context,
        claim_register=claim_register,
        config=inputs["config"],
    )
    # The v1 register (empty with pinned columns when scoring never ran).
    score_register_v1: pd.DataFrame = (
        context.payloads["daily_scores_v1"] if isinstance(context.payloads.get("daily_scores_v1"), pd.DataFrame) else _empty_daily_scores()
    )
    # The v2 register (empty with pinned columns when correction never ran).
    score_register_v2: pd.DataFrame = (
        context.payloads["daily_scores_v2"] if isinstance(context.payloads.get("daily_scores_v2"), pd.DataFrame) else _empty_daily_scores()
    )
    # The factor aggregates (empty with pinned columns when aggregation never ran).
    factor_aggregates: pd.DataFrame = (
        context.payloads["factor_aggregates"] if isinstance(context.payloads.get("factor_aggregates"), pd.DataFrame) else _empty_factor_aggregates()
    )
    # The primary ledger triple (zeros disclosed as never-computed when halted).
    ledger_primary: Tuple[int, int, int] = (
        context.payloads["ledger_primary"] if isinstance(context.payloads.get("ledger_primary"), tuple) else (0, 0, 0)
    )
    # The reversed ledger triple.
    ledger_reverse: Tuple[int, int, int] = (
        context.payloads["ledger_reverse"] if isinstance(context.payloads.get("ledger_reverse"), tuple) else (0, 0, 0)
    )
    # The reference means (empty with pinned columns when the reference panel never ran).
    reference_means: pd.DataFrame = (
        context.payloads["reference_means"] if isinstance(context.payloads.get("reference_means"), pd.DataFrame) else _empty_reference_means()
    )
    # The dual-check results (empty with pinned columns when the check never ran).
    dual_check_result: pd.DataFrame = (
        context.payloads["dual_check_results"] if isinstance(context.payloads.get("dual_check_results"), pd.DataFrame) else _empty_dual_check_results()
    )
    # The correction partition (ints only; {} when the corrective ledger never ran).
    correction_partition_raw: Any = context.payloads.get("correction_partition")
    # Coerce to the pinned int mapping defensively.
    correction_partition: Dict[str, int] = {
        str(key): int(value)
        for key, value in (correction_partition_raw or {}).items()
        if isinstance(value, (int, np.integer)) and not isinstance(value, bool)
    }
    # The cohort hash (freeze product or the disclosed sentinel).
    cohort_hash: str = str(context.freeze_state.get("cohort_hash", NOT_FROZEN_COHORT_HASH))
    # The cohort timestamp (freeze product, else the disclosed run instant).
    freeze_timestamp: Optional[datetime.datetime] = context.freeze_state.get("freeze_timestamp")
    # Fall back to the run instant when the freeze never ran (disclosed, never rearmed).
    cohort_timestamp: datetime.datetime = (
        freeze_timestamp if isinstance(freeze_timestamp, datetime.datetime) else datetime.datetime.now(ASIA_SHANGHAI_TZ)
    )
    # First-halt bookkeeping disappears into the report; the stage verdicts stand alone.
    _ = first_halt_stage
    # Freeze the immutable result (the single construction site).
    result: StudyResult = StudyResult(
        # The requested stages' payloads.
        stages_output=stages_output,
        # All thirteen pinned stage names and their halt states.
        stages_halt=stages_halt,
        # The freeze hash or the disclosed sentinel.
        cohort_hash=cohort_hash,
        # The freeze instant or the disclosed run instant.
        cohort_timestamp=cohort_timestamp,
        # The v1 register.
        score_register_v1=score_register_v1,
        # The v2 register.
        score_register_v2=score_register_v2,
        # The factor aggregates.
        factor_aggregates=factor_aggregates,
        # The primary ledger triple.
        ledger_primary=ledger_primary,
        # The reversed ledger triple.
        ledger_reverse=ledger_reverse,
        # The reference means.
        reference_means=reference_means,
        # The dual-check results.
        dual_check_result=dual_check_result,
        # The corrective partition.
        correction_partition=correction_partition,
        # The admissibility report (the single interface downstream).
        admissibility_report=admissibility_report,
        # The deterministic claim register.
        claim_register=claim_register,
        # The composed terminal verdicts.
        block_verdicts=block_verdicts,
    )
    # Return the frozen result.
    return result

def serialise_study_result(result: StudyResult) -> Dict[str, Any]:
    """
    Serialise a frozen ``StudyResult`` into a plain mapping.

    Project the frozen dataclass field by field (each field read once,
    never deep-traversed) and deep-copy every payload: DataFrame fields
    through ``DataFrame.copy(deep=True)``, the ``stages_output`` mapping
    unwrapped one level with its frames copied, and every other field
    through ``box_safe_copy``. The frozen object itself is not mutated.

    Parameters
    ----------
    result : StudyResult
        The frozen result to serialise.

    Returns
    -------
    Dict[str, Any]
        A plain mapping with the same field names as the dataclass and
        deep-copied payloads.

    Raises
    ------
    TypeError
        If ``result`` is not a ``StudyResult``.
    """
    # The argument must be the pinned frozen dataclass.
    if not isinstance(result, StudyResult):
        # Reject foreign objects with the observed type.
        raise TypeError(f"result must be a StudyResult, got {type(result).__name__}")
    # Manual shallow field projection (each field read once, never traversed).
    shallow: Dict[str, Any] = {field.name: getattr(result, field.name) for field in dataclasses.fields(result)}
    # The output mapping with deep-copied frames.
    out: Dict[str, Any] = {}
    # Walk the shallow mapping once.
    for key, value in shallow.items():
        # A raw DataFrame field is deep-copied.
        if isinstance(value, pd.DataFrame):
            # Deep-copy the frame (never mutate the frozen object's payload).
            out[key] = value.copy(deep=True)
            # Advance to the next entry.
            continue
        # The stages_output mapping is unwrapped one level for frame copies.
        if key == "stages_output" and isinstance(value, Mapping):
            # The unwrapped mapping.
            unwrapped: Dict[str, Any] = {}
            # Walk the stage payloads.
            for stage_name, stage_value in value.items():
                # A frame is deep-copied.
                if isinstance(stage_value, pd.DataFrame):
                    # Deep-copy the frame.
                    unwrapped[stage_name] = stage_value.copy(deep=True)
                    # Advance to the next stage.
                    continue
                # Anything else deep-copies safely (Mappings, tuples, scalars).
                unwrapped[stage_name] = box_safe_copy(stage_value)
            # Store the unwrapped mapping.
            out[key] = unwrapped
            # Advance to the next entry.
            continue
        # Anything else deep-copies safely.
        out[key] = box_safe_copy(value)
    # Return the serialisable mapping.
    return out

def box_safe_copy(value: Any) -> Any:
    """
    Deep-copy a value, tolerating pandas objects and containers.

    pandas ``DataFrame`` and ``Series`` copy through
    ``copy(deep=True)``; mappings, tuples and lists are rebuilt with
    every element copied recursively; any other value is returned
    unchanged.

    Parameters
    ----------
    value : Any
        The value to copy.

    Returns
    -------
    Any
        A deep copy for pandas objects and containers; the value itself
        for atomic scalars, which are immutable by construction.
    """
    # pandas objects copy through their own machinery.
    if isinstance(value, (pd.DataFrame, pd.Series)):
        # Deep copy of the pandas object.
        return value.copy(deep=True)
    # Mappings are rebuilt with copied values.
    if isinstance(value, Mapping):
        # Rebuild the mapping with a copied value per key.
        copied = {key: box_safe_copy(item) for key, item in value.items()}
        # Return the rebuilt mapping.
        return copied
    # Tuples are copied element-wise.
    if isinstance(value, tuple):
        # Rebuild the tuple with copied elements.
        return tuple(box_safe_copy(item) for item in value)
    # Lists are copied element-wise.
    if isinstance(value, list):
        # Rebuild the list with copied elements.
        return [box_safe_copy(item) for item in value]
    # Atomic scalars are immutable; return them verbatim.
    return value

# -------------------------------------------------------------------------------------------------------------------------------
# Task 14, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

def _stage_arguments(
    stage: str,
    produced: Mapping[str, Any],
    context: PipelineContext,
    inputs: Mapping[str, Any],
    config: Dict[str, Any],
    oos_read_timestamp: datetime.datetime,
) -> Tuple[Any, ...]:
    """
    Assemble the positional arguments of one stage callable.

    Thread upstream products from ``produced`` (falling back to the
    ``context`` cache for the calendar and mask register) and the raw
    artifacts from ``inputs`` into the tuple the stage callable expects,
    one tuple per pinned stage name. Missing upstream products are
    passed as ``None`` so the stage itself reports the halt.

    Parameters
    ----------
    stage : str
        The pinned stage name.
    produced : Mapping[str, Any]
        Per-stage payloads; halted stages map to ``None``.
    context : PipelineContext
        Active pipeline context.
    inputs : Mapping[str, Any]
        The ``run_study`` arguments, read by key.
    config : Dict[str, Any]
        Study configuration.
    oos_read_timestamp : datetime.datetime
        Derived OOS first-read instant for the freeze stage.

    Returns
    -------
    Tuple[Any, ...]
        The positional arguments of the stage callable.

    Raises
    ------
    KeyError
        If ``stage`` is not a pinned stage name.
    """
    # The seam: dispatch is a registry lookup; argument assembly is explicit.
    if stage == "validate":
        # Task 1 ingest validation (raw artifacts only).
        return (inputs["files"], inputs["unadjusted"], inputs["adjusted"], config)
    # Cleanse consumes exactly the two raw views.
    if stage == "cleanse":
        # Task 2 panel-scoped cleansing.
        return (inputs["unadjusted"], inputs["adjusted"], config)
    # Custody consumes the cleansed views.
    if stage == "custody":
        # The cleanse outputs, or (None, None) when cleansing halted.
        cleansed_unadj, cleansed_adj = produced.get("cleanse") or (None, None)
        # Task 3 custody verification.
        return (inputs["files"], cleansed_unadj, cleansed_adj, config)
    # Identity consumes the cleansed views and the custody manifest.
    if stage == "identity":
        # The cleanse outputs, or (None, None) when cleansing halted.
        cleansed_unadj, cleansed_adj = produced.get("cleanse") or (None, None)
        # The custody output, or None when custody halted.
        custody_out = produced.get("custody") or (None,)
        # Task 4 cross-view identity verification and role separation.
        return (cleansed_unadj, cleansed_adj, custody_out[0], config)
    # Calendar consumes the identity report only (per the step text).
    if stage == "calendar":
        # The identity output, or None when the identity stage halted.
        identity_out = produced.get("identity") or (None, None, None)
        # Task 5 calendar construction and masking.
        return (identity_out[2], config)
    # Labels consume the cleansed adjusted view, the calendar and the mask register.
    if stage == "labels":
        # The adjusted view, or None when cleansing halted.
        cleansed_unadj, cleansed_adj = produced.get("cleanse") or (None, None)
        # The calendar, preferring the calendar stage's product.
        calendar_out = produced.get("calendar") or (None, None)
        # Fall back to the context-cached calendar.
        calendar = calendar_out[0] if calendar_out[0] is not None else context.payloads.get("calendar")
        # The mask register, preferring the calendar stage's product.
        mask_register = calendar_out[1] if calendar_out[1] is not None else context.payloads.get("mask_register")
        # Task 6 Eq. (1) label construction.
        return (cleansed_adj, calendar, mask_register, config)
    # The DSL stage consumes only the configuration.
    if stage == "dsl":
        # Task 7 DSL and versioned evaluator construction.
        return (config,)
    # Freeze consumes the preserved rows and the DSL stage's canonicalizer.
    if stage == "freeze":
        # The DSL outputs, or None when the DSL stage halted.
        dsl_out = produced.get("dsl") or (None, None, None)
        # Task 8 cohort freeze protocol.
        return (inputs["preserved_rows"], dsl_out[2], config, oos_read_timestamp)
    # Score consumes the masked features, the Eq. (1) labels, the calendar, the register, v1 and the cohort.
    if stage == "score":
        # The score substrates from this run's products and context.
        masked_features = context.payloads.get("masked_features_panel")
        # The Eq. (1) labels frame.
        labels_out = produced.get("labels") or (None,)
        # The calendar, preferring the calendar stage's product.
        calendar_out = produced.get("calendar") or (None, None)
        # Fall back to the context-cached calendar.
        calendar = calendar_out[0] if calendar_out[0] is not None else context.payloads.get("calendar")
        # The mask register, preferring the calendar stage's product.
        mask_register = calendar_out[1] if calendar_out[1] is not None else context.payloads.get("mask_register")
        # The v1 evaluator.
        dsl_out = produced.get("dsl") or (None, None)
        # The frozen cohort.
        freeze_out = produced.get("freeze") or (None,)
        # Task 9 daily rank-IC scoring.
        return (masked_features, labels_out[0], calendar, mask_register, dsl_out[0], freeze_out[0], config)
    # Aggregate consumes the v1 register and the frozen cohort.
    if stage == "aggregate":
        # The v1 register.
        score_out = produced.get("score") or (None,)
        # The frozen cohort.
        freeze_out = produced.get("freeze") or (None,)
        # The pinned primary precedence from the configuration ([R] NOT_STATED_IN_PAPER).
        precedence = tuple(config["P04_LEDGER_PARAMETERS"]["state_precedence"])
        # Task 10 aggregation, scorability gate and both ledgers.
        return (score_out[0], freeze_out[0], config, precedence)
    # Reference consumes the masked features, the Eq. (1) labels, the calendar, the register and v1.
    if stage == "reference":
        # The reference substrates from this run's products and context.
        masked_features = context.payloads.get("masked_features_panel")
        # The Eq. (1) labels.
        labels_out = produced.get("labels") or (None,)
        # The calendar, preferring the calendar stage's product.
        calendar_out = produced.get("calendar") or (None, None)
        # Fall back to the context-cached calendar.
        calendar = calendar_out[0] if calendar_out[0] is not None else context.payloads.get("calendar")
        # The mask register, preferring the calendar stage's product.
        mask_register = calendar_out[1] if calendar_out[1] is not None else context.payloads.get("mask_register")
        # The v1 evaluator.
        dsl_out = produced.get("dsl") or (None,)
        # Task 11 fixed reference-signal panel.
        return (masked_features, labels_out[0], calendar, mask_register, dsl_out[0], config)
    # Dual check consumes the v1 register, the (possibly None) frozen nine cells, and the [R] tolerance.
    if stage == "dual_check":
        # The v1 register.
        score_out = produced.get("score") or (None,)
        # The frozen cells, if any were derived earlier in this run.
        nine_cells = context.payloads.get("nine_cells")
        # Task 12 dual-implementation spot check.
        return (score_out[0], nine_cells, TOLERANCE_SPOT_CHECK_ABSOLUTE, config)
    # Correction consumes the v1 register, the v2 evaluator and the frozen cohort.
    if stage == "correction":
        # The v1 register.
        score_out = produced.get("score") or (None,)
        # The v2 evaluator.
        dsl_out = produced.get("dsl") or (None, None)
        # The frozen cohort.
        freeze_out = produced.get("freeze") or (None,)
        # Task 13 versioned correction and full re-evaluation.
        return (score_out[0], dsl_out[1], freeze_out[0], config)
    # An unpinned stage name is a programming defect.
    raise KeyError(f"stage {stage!r} is not a pinned stage name")

def _assemble_stages_output(
    requested: Sequence[str],
    produced: Mapping[str, Any],
    stages_halt: Mapping[str, Optional[str]],
    context: PipelineContext,
) -> Dict[str, Any]:
    """
    Assemble the ``stages_output`` mapping for the requested stages.

    Walk ``requested`` in dispatch order. A stage whose ``stages_halt``
    value is not ``None`` maps to ``{"halted": True, "halt_reason":
    ...}``; otherwise the entry is that stage's publication payload,
    read from ``produced`` or from the context's manifests, payloads and
    freeze state. Frames are carried by reference, not copied.

    Parameters
    ----------
    requested : Sequence[str]
        Stage names to emit, in dispatch order.
    produced : Mapping[str, Any]
        Per-stage payload tuples keyed by stage name.
    stages_halt : Mapping[str, Optional[str]]
        Per-stage halt reason; ``None`` marks a healthy stage.
    context : PipelineContext
        Active run context supplying ``manifests``, ``payloads`` and
        ``freeze_state``.

    Returns
    -------
    Dict[str, Any]
        One entry per emitted stage: the halted marker for halted
        stages, else the stage's payload. Payloads are the report or
        manifest dict for ``validate``, ``cleanse``, ``custody`` and
        ``identity``; per-stage dicts of frames and stamped context
        values for ``calendar``, ``labels``, ``dsl``, ``freeze``,
        ``score``, ``aggregate``, ``reference``, ``dual_check`` and
        ``correction``. Requested names outside the thirteen stages
        produce no entry.
    """
    # The accumulator.
    output: Dict[str, Any] = {}
    # Walk the requested stages in dispatch order.
    for stage in requested:
        # A halted stage maps to the uniform halted marker.
        if stages_halt.get(stage) is not None:
            # The uniform halted payload.
            output[stage] = {"halted": True, "halt_reason": stages_halt[stage]}
            # Advance to the next stage.
            continue
        # The healthy payloads per stage name.
        if stage == "validate":
            # The validation report (a plain dict).
            output[stage] = produced["validate"][0]
            # Advance to the next stage.
            continue
        # The cleanse stage reports counters and dictionaries.
        if stage == "cleanse":
            # The cleanse manifest (plain dict).
            output[stage] = dict(context.manifests.get("cleanse", {}))
            # Advance to the next stage.
            continue
        # The custody stage reports the artifact path and the verdict census.
        if stage == "custody":
            # The custody manifest (plain dict).
            output[stage] = dict(context.manifests.get("custody", {}))
            # Advance to the next stage.
            continue
        # The identity stage reports its diagnostics (plain dict).
        if stage == "identity":
            # The identity report (facades excluded from the mapping payload is NOT applied here; the report is plain).
            output[stage] = produced["identity"][2]
            # Advance to the next stage.
            continue
        # The calendar stage reports the calendar, the register and the battery proof.
        if stage == "calendar":
            # The calendar stage payload (727 rows by reference).
            output[stage] = {
                # The dense 727-row calendar.
                "calendar": produced["calendar"][0],
                # The two-row mask register.
                "mask_register": produced["calendar"][1],
                # The post-masking battery verdict payload.
                "post_mask_battery": context.manifests.get("calendar", {}).get("post_mask_battery"),
                # The calendar hash stamped by Task 5.
                "calendar_hash": context.manifests.get("calendar", {}).get("calendar_hash"),
                # The usable-rows cardinality (29,078 on the study snapshot).
                "usable_rows": context.manifests.get("calendar", {}).get("usable_rows"),
                # The mask witness artifact path.
                "mask_witness_path": context.manifests.get("calendar", {}).get("mask_witness_path"),
            }
            # Advance to the next stage.
            continue
        # The labels stage reports the Eq. (1) frame and its counts.
        if stage == "labels":
            # The labels stage payload.
            output[stage] = {
                # The Eq. (1) labels frame (by reference).
                "labels": produced["labels"][0],
                # The label-available dates count.
                "label_available_dates": context.manifests.get("labels", {}).get("label_available_dates"),
                # The label-window panel rows.
                "label_window_rows": context.manifests.get("labels", {}).get("label_window_rows"),
                # The total non-NaN labels (disclosure).
                "n_available_labels_total": context.manifests.get("labels", {}).get("n_available_labels_total"),
                # The withdrawn raw-task test tag.
                "withdrawals": context.manifests.get("labels", {}).get("withdrawals", []),
            }
            # Advance to the next stage.
            continue
        # The DSL stage reports the evaluator stamps and the canonicalizer identity.
        if stage == "dsl":
            # The v1 evaluator stamp pair.
            evaluator_v1 = produced["dsl"][0]
            # The v2 evaluator stamp pair.
            evaluator_v2 = produced["dsl"][1]
            # The canonicalizer.
            canonicalizer = produced["dsl"][2]
            # The DSL stage payload.
            output[stage] = {
                # The v1 evaluator summary.
                "evaluator_v1": {
                    # The version literal.
                    "version": evaluator_v1.version,
                    # The recorded tag.
                    "version_tag": evaluator_v1.version_tag,
                    # The registry digest.
                    "registry_sha256": evaluator_v1.registry_sha256,
                    # The conditional wiring disclosure.
                    "conditional_implementation": evaluator_v1.conditional_implementation,
                },
                # The v2 evaluator summary.
                "evaluator_v2": {
                    # The version literal.
                    "version": evaluator_v2.version,
                    # The recorded tag.
                    "version_tag": evaluator_v2.version_tag,
                    # The registry digest.
                    "registry_sha256": evaluator_v2.registry_sha256,
                    # The conditional wiring disclosure.
                    "conditional_implementation": evaluator_v2.conditional_implementation,
                },
                # The canonicalizer identity (certificates by name, digest stamped).
                "canonicalizer": {
                    # The certificate names.
                    "certificates": sorted(key for key, value in canonicalizer.items() if callable(value)),
                    # The v1 registry digest echo.
                    "registry_sha256": canonicalizer.get("registry_sha256"),
                },
            }
            # Advance to the next stage.
            continue
        # The freeze stage reports the cohort certificates and the ordered identifiers.
        if stage == "freeze":
            # The freeze stage payload.
            output[stage] = {
                # The cohort hash (SHA-256 of the ordered ID bytes).
                "cohort_hash": context.freeze_state.get("cohort_hash"),
                # The freeze id.
                "freeze_id": context.freeze_state.get("freeze_id"),
                # The cohort size.
                "cohort_size": context.freeze_state.get("cohort_size"),
                # The freeze timestamp (Asia/Shanghai, tz-aware).
                "freeze_timestamp": context.freeze_state.get("freeze_timestamp"),
                # The post-screen count.
                "n_after_I_screen": context.manifests.get("freeze", {}).get("n_after_I_screen"),
                # The post-dedup count.
                "n_after_dedup": context.manifests.get("freeze", {}).get("n_after_dedup"),
                # The dedup disclosure.
                "dedup_status": context.manifests.get("freeze", {}).get("dedup_status"),
                # The ordered cohort identifiers (reference to the frame's ids).
                "frozen_expression_ids": list(context.payloads.get("frozen_expression_ids", ())),
                # The OOS-used-in-admission flag (always False).
                "oos_used_in_admission": context.manifests.get("freeze", {}).get("oos_used_in_admission"),
                # The cohort frame (by reference).
                "cohort_rows": getattr(produced["freeze"][0], "rows", None),
            }
            # Advance to the next stage.
            continue
        # The score stage reports the register and its manifest.
        if stage == "score":
            # The score stage payload.
            output[stage] = {
                # The DAILY_SCORES register (by reference).
                "register": produced["score"][0],
                # The evaluator version tag.
                "evaluator_version": context.manifests.get("score", {}).get("evaluator_version"),
                # The register cell count.
                "n_cells": context.manifests.get("score", {}).get("n_cells"),
                # The parse_failed identifiers.
                "parse_failed_ids": context.manifests.get("score", {}).get("parse_failed_ids", []),
                # The number of trade dates scored.
                "n_trade_dates": context.manifests.get("score", {}).get("n_trade_dates"),
            }
            # Advance to the next stage.
            continue
        # The aggregate stage reports the ledgers, the aggregates and the battery.
        if stage == "aggregate":
            # The aggregate stage payload.
            output[stage] = {
                # The primary ledger triple (positional).
                "ledger_primary": tuple(produced["aggregate"][0]),
                # The reversed ledger triple.
                "ledger_reverse": context.payloads.get("ledger_reverse"),
                # The FACTOR_AGGREGATES frame (by reference).
                "aggregates": produced["aggregate"][1],
                # The closure battery payload.
                "closure_battery": context.payloads.get("closure_battery"),
                # The FACTOR_AGGREGATES.csv path.
                "aggregate_path": context.manifests.get("aggregate", {}).get("aggregate_path"),
            }
            # Advance to the next stage.
            continue
        # The reference stage reports the six-row means table.
        if stage == "reference":
            # The reference stage payload.
            output[stage] = {
                # The six-row means table (by reference).
                "reference_means": produced["reference"][0],
                # The reference manifest (placement, envelope echo).
                "reference_manifest": dict(context.manifests.get("reference", {})),
            }
            # Advance to the next stage.
            continue
        # The dual-check stage reports the nine-cell results and the tolerance.
        if stage == "dual_check":
            # The dual-check stage payload.
            output[stage] = {
                # The nine-cell results frame (by reference).
                "nine_cell_results": produced["dual_check"][0],
                # The declared absolute tolerance.
                "tolerance": context.manifests.get("dual_check", {}).get("tolerance"),
                # The number of cells checked.
                "n_cells": context.manifests.get("dual_check", {}).get("n_cells"),
                # The maximum absolute delta observed.
                "max_abs_delta": context.manifests.get("dual_check", {}).get("max_abs_delta"),
            }
            # Advance to the next stage.
            continue
        # The correction stage reports the partition and the Task 13 stamps.
        if stage == "correction":
            # The Task 13 mapping, when it ran.
            task_13_result = context.payloads.get("task_13_result")
            # Whether Task 13 halted at all.
            task_13_halted = bool(task_13_result.get("halted")) if isinstance(task_13_result, Mapping) else None
            # The correction stage payload.
            output[stage] = {
                # The corrective re-evaluation partition (ints only).
                "correction_partition": dict(produced["correction"][0]),
                # The v2 register (by reference).
                "v2_register": produced["correction"][1],
                # The Task 13 corrective result summary (halt, reason, provenance).
                "task_13": {
                    # The halt flag.
                    "halted": task_13_halted,
                    # The recorded reason.
                    "halt_reason": task_13_result.get("halt_reason") if isinstance(task_13_result, Mapping) else None,
                    # The provenance stamp (RECONSTRUCTION_FIXTURE on this project).
                    "artifacts_provenance": task_13_result.get("artifacts_provenance") if isinstance(task_13_result, Mapping) else None,
                    # The honest-exit stamp (DATA_ABSENT on this project).
                    "data_gap": task_13_result.get("data_gap") if isinstance(task_13_result, Mapping) else None,
                    # The corrective summary (ints only).
                    "re_evaluation_ledger_summary": dict(task_13_result.get("RE_EVALUATION_LEDGER_SUMMARY", {})) if isinstance(task_13_result, Mapping) else {},
                },
            }
            # Advance to the next stage.
            continue
    # Return the assembled mapping.
    return output

def run_study(
    files: pd.DataFrame,
    unadjusted: pd.DataFrame,
    adjusted: pd.DataFrame,
    preserved_rows: pd.DataFrame,
    native_oos_checkpoint: pd.DataFrame,
    outcome_filtered_catalog: pd.DataFrame,
    run_report_totals: pd.DataFrame,
    run_report_rounds: pd.DataFrame,
    collector_a: pd.DataFrame,
    collector_b: pd.DataFrame,
    config: Dict[str, Any],
    stages: Sequence[str],
    halt_on_failure: bool,
) -> StudyResult:
    """
    Dispatch the pinned pipeline stages and freeze the study result.

    Validate the arguments and check the request's topological closure
    under ``STAGE_DEPENDENCIES``; an unclosed or unknown request stamps
    all thirteen stages with the closure reason and returns before any
    stage runs. Otherwise dispatch the requested stages in the pinned
    ``STAGE_ORDER`` through the ``STAGE_CALLABLES`` registry, with
    ``_stage_arguments`` threading prior payloads to each consumer. A
    raised exception becomes a halt naming the exception class and
    message and discards that stage's payload. With ``halt_on_failure``
    true the dispatch stops at the first halt and later requested stages
    are stamped ``PIPELINE_HALTED_AT_<stage>``; with false the dispatch
    continues and surfaces every halt. Dispatch events and counters are
    recorded on the run context (``context.events``,
    ``context.counters``); per-stage halt states stay in the local
    ``stages_halt`` mapping passed to ``_finalise_result``, which builds
    the immutable ``StudyResult`` once; the context bridge is cleared in
    a ``finally`` block. Writes land beneath ``ACCEPTANCE_SCRATCH_DIR``
    when set, else under a fresh temporary directory.

    Parameters
    ----------
    files : pd.DataFrame
        Acquisition manifest; one row per artifact.
    unadjusted : pd.DataFrame
        Nominal price/volume panel.
    adjusted : pd.DataFrame
        Provider-adjusted panel, used for labels only.
    preserved_rows : pd.DataFrame
        Preserved candidate rows consumed by ``freeze_cohort``.
    native_oos_checkpoint : pd.DataFrame
        Native OOS checkpoint rows, for the closures.
    outcome_filtered_catalog : pd.DataFrame
        Outcome-filtered catalog rows, for the closures.
    run_report_totals : pd.DataFrame
        Reported run aggregates, for the closures.
    run_report_rounds : pd.DataFrame
        Reported per-round counters, for the closures.
    collector_a : pd.DataFrame
        Collector A artifact, for the closures.
    collector_b : pd.DataFrame
        Collector B artifact, for the closures.
    config : Dict[str, Any]
        Study configuration; must be a plain ``dict``.
    stages : Sequence[str]
        Requested stage names; the pinned order is imposed regardless
        of input order.
    halt_on_failure : bool
        True stops at the first halt; False (diagnostic) continues and
        surfaces every halt. ``numpy.bool_`` is coerced.

    Returns
    -------
    StudyResult
        The frozen result with ``stages_halt``, ``stages_output``, the
        terminal block verdicts, the claim register and the
        admissibility report, whose run mode names the mode.

    Raises
    ------
    TypeError
        If a frame argument is not a ``pandas.DataFrame``, ``config`` is
        not a ``dict``, ``halt_on_failure`` is a non-bool other than
        ``numpy.bool_``, ``stages`` is a string, bytes, ``None`` or
        non-iterable, any stage name is not a ``str``, or a stage
        returns a malformed result.

    Notes
    -----
    The cast ledger is created once at
    ``context.payloads["cast_ledger"]`` and shared with the
    persisted-panel re-reads. The OOS read timestamp is derived once.
    Counters advance by ``DISPATCH_COUNTER_QUANTUM`` per visited stage
    and per halt. The caller's frames are never mutated and stage
    exceptions are never swallowed.
    """
    # ---- Boundary audit of the thirteen arguments (caller-supplied raw artifacts) ---------
    # The ten frames must be DataFrames; the configuration must be a dict.
    for name, frame in (
        ("files", files),
        ("unadjusted", unadjusted),
        ("adjusted", adjusted),
        ("preserved_rows", preserved_rows),
        ("native_oos_checkpoint", native_oos_checkpoint),
        ("outcome_filtered_catalog", outcome_filtered_catalog),
        ("run_report_totals", run_report_totals),
        ("run_report_rounds", run_report_rounds),
        ("collector_a", collector_a),
        ("collector_b", collector_b),
    ):
        # Reject non-DataFrame arguments with the observed type.
        if not isinstance(frame, pd.DataFrame):
            # Halt with the pinned exception type at the module boundary.
            raise TypeError(f"run_study: {name} must be a pandas.DataFrame, got {type(frame).__name__}")
    # The configuration must be a dict.
    if not isinstance(config, dict):
        # Reject non-dict configurations with the observed type.
        raise TypeError(f"run_study: config must be a dict, got {type(config).__name__}")
    # halt_on_failure admits bool and numpy bool (coerced) only.
    if isinstance(halt_on_failure, np.bool_):
        # Coerce the numpy flag to a Python bool.
        halt_on_failure = bool(halt_on_failure)
    # Bools only are accepted as the mode flag.
    if not isinstance(halt_on_failure, bool):
        # Reject non-bool modes with the observed type.
        raise TypeError(f"run_study: halt_on_failure must be a bool, got {type(halt_on_failure).__name__}")
    # stages admits any non-string iterable of names.
    if isinstance(stages, (str, bytes)) or stages is None or not hasattr(stages, "__iter__"):
        # Reject non-iterable or string stages.
        raise TypeError(f"run_study: stages must be a sequence of stage names, got {type(stages).__name__}")
    # Materialise the requested names.
    requested_names: List[str] = list(stages)
    # Non-string entries are rejected.
    if any(not isinstance(name, str) for name in requested_names):
        # Name the observed offending element.
        raise TypeError("run_study: every stage name in stages must be a str")
    # The pinned order projection of the request.
    requested: List[str] = [stage for stage in STAGE_ORDER if stage in set(requested_names)]
    # The run-scoped scratch root (hermetic; never the project tree).
    scratch_env: Optional[str] = os.environ.get("ACCEPTANCE_SCRATCH_DIR")
    # The root: the hermetic scratch when set, else a fresh temporary directory.
    scratch_root: Path = Path(scratch_env) if scratch_env else Path(tempfile.mkdtemp(prefix="run_study_"))
    # Per-invocation nonce namespacing every artifact of this run.
    nonce: str = uuid.uuid4().hex[:12]
    # The request-scoped context.
    context: PipelineContext = PipelineContext(scratch_root, nonce)
    # The run-scoped cast ledger (append-only): created once here so the cleansing
    # pass and the persisted-panel re-reads share ONE handle, never a second ledger.
    context.payloads["cast_ledger"] = []
    # Seat the bridge (cleared in the finally block below).
    _set_run_context(context)
    # The halt bookkeeping for the pinned halt semantics.
    try:
        # The inputs mapping for the closure operands.
        inputs: Dict[str, Any] = {
            # The raw artifacts, verbatim.
            "files": files,
            "unadjusted": unadjusted,
            "adjusted": adjusted,
            "preserved_rows": preserved_rows,
            "native_oos_checkpoint": native_oos_checkpoint,
            "outcome_filtered_catalog": outcome_filtered_catalog,
            "run_report_totals": run_report_totals,
            "run_report_rounds": run_report_rounds,
            "collector_a": collector_a,
            "collector_b": collector_b,
            "config": config,
        }
        # The topological-closure violations of the request (empty when closed).
        violations: List[Tuple[str, frozenset]] = check_stage_closure(requested)
        # An unclosed or unknown request halts before any stage is dispatched.
        if violations:
            # Render the closure failure.
            graph_reason: str = topological_closure_reason(violations)
            # Every pinned stage carries the closure reason (nothing ran).
            graph_halts: Dict[str, Optional[str]] = {stage: graph_reason for stage in STAGE_ORDER}
            # Finalise the aborted dispatch into the immutable result.
            return _finalise_result(
                halt_on_failure=halt_on_failure,
                stages_halt=graph_halts,
                stages_output={},
                inputs=inputs,
                context=context,
                first_halt_stage=None,
                graph_halt_reason=graph_reason,
            )
        # Initialise all thirteen pinned stage keys; requested stages are overwritten below.
        stages_halt: Dict[str, Optional[str]] = {stage: STAGE_NOT_REQUESTED for stage in STAGE_ORDER}
        # The per-stage products (None for stages that halted).
        produced: Dict[str, Any] = {}
        # The derived OOS first-read instant (one day ahead; disclosed).
        oos_read_timestamp: datetime.datetime = _oos_read_timestamp()
        # The first halted stage, when any (the halt anchor).
        first_halt_stage: Optional[str] = None
        # The halt trigger (halt_on_failure and a first halt observed).
        halt_triggered: bool = False
        # Dispatch over the pinned order, restricted to the request.
        for stage in requested:
            # A stage behind a failure in halt mode carries the sentinel.
            if halt_triggered:
                # Stamp the PIPELINE_HALTED_AT_ marker naming the failing stage.
                stages_halt[stage] = f"{PIPELINE_HALTED_PREFIX}{first_halt_stage}"
                # No payload was produced.
                produced[stage] = None
                # Advance to the next stage.
                continue
            # Resolve the stage's pinned callable through the registry.
            stage_callable: Callable[..., Any] = STAGE_CALLABLES[stage]
            # Assemble the pinned argument tuple.
            argument_tuple: Tuple[Any, ...] = _stage_arguments(
                stage=stage,
                produced=produced,
                context=context,
                inputs=inputs,
                config=config,
                oos_read_timestamp=oos_read_timestamp,
            )
            # Invoke the callable; any exception becomes the halt with the class and message.
            try:
                # The dispatch boundary converts raises to halts (never swallows).
                stage_result = stage_callable(*argument_tuple)
                # The stage completed without an exception.
                stage_exception: Optional[BaseException] = None
            except Exception as exc:
                # Bind the exception as the halt payload.
                stage_exception = exc
                # No partial payload survives a raise (never presented as a result).
                stage_result = None
            # The dispatch counter advances for every visited stage.
            context.counters["stages_executed"] += DISPATCH_COUNTER_QUANTUM
            # A raised exception is a halt with the exception class and message.
            if stage_exception is not None:
                # The halt reason names the class and the message verbatim.
                stage_halt = _halt_reason(stage_exception)
                # Record the dispatch event.
                context.events.append({"stage": stage, "verdict": "halt", "reason": stage_halt})
                # Store the halt.
                stages_halt[stage] = stage_halt
                # No payload on halt.
                produced[stage] = None
                # The failing stage becomes the halt anchor.
                if first_halt_stage is None:
                    # Pin the anchor.
                    first_halt_stage = stage
                # The halted counter advances.
                context.counters["stages_halted"] += DISPATCH_COUNTER_QUANTUM
                # Halt mode terminates the dispatch now.
                if halt_on_failure:
                    # Trip the halt trigger; remaining stages get sentinels.
                    halt_triggered = True
                # Advance to the next stage.
                continue
            # Normalise the stage's (payloads..., halt_reason) tuple.
            payloads, stage_halt = _split_stage_result(stage_result, stage)
            # A non-None stage halt reason is recorded.
            if stage_halt is not None:
                # Record the dispatch event.
                context.events.append({"stage": stage, "verdict": "halt", "reason": stage_halt})
                # Store the halt.
                stages_halt[stage] = stage_halt
                # No payload on halt.
                produced[stage] = None
                # The failing stage becomes the halt anchor.
                if first_halt_stage is None:
                    # Pin the anchor.
                    first_halt_stage = stage
                # The halted counter advances.
                context.counters["stages_halted"] += DISPATCH_COUNTER_QUANTUM
                # Halt mode terminates the dispatch now.
                if halt_on_failure:
                    # Trip the halt trigger; remaining stages get sentinels.
                    halt_triggered = True
                # Advance to the next stage.
                continue
            # The stage succeeded: store the halt as None and keep the payload.
            stages_halt[stage] = None
            # Keep the produced payload tuple.
            produced[stage] = payloads
            # Record the dispatch event.
            context.events.append({"stage": stage, "verdict": "ok"})
        # Assemble the stages_output mapping for exactly the requested stages.
        stages_output: Dict[str, Any] = _assemble_stages_output(
            requested=requested,
            produced=produced,
            stages_halt=stages_halt,
            context=context,
        )
        # Freeze the immutable result (the single construction site).
        return _finalise_result(
            halt_on_failure=halt_on_failure,
            stages_halt=stages_halt,
            stages_output=stages_output,
            inputs=inputs,
            context=context,
            first_halt_stage=first_halt_stage,
            graph_halt_reason=None,
        )
    finally:
        # Clear the bridge unconditionally; no state leaks out of this call.
        _set_run_context(None)

def _split_stage_result(
    result: Any,
    stage: str,
) -> Tuple[Optional[Tuple[Any, ...]], Optional[str]]:
    """
    Split a stage return into payloads and the halt reason.

    The stage contract is a non-empty tuple whose last element is
    ``None`` or a halt reason string; the leading elements are the stage
    payloads.

    Parameters
    ----------
    result : Any
        Raw return value of the stage callable.
    stage : str
        Pinned stage name, used only in error messages.

    Returns
    -------
    Tuple[Optional[Tuple[Any, ...]], Optional[str]]
        ``(payloads, halt_reason)``; ``payloads`` is ``None`` when the
        stage halted, else the leading elements as a tuple.

    Raises
    ------
    TypeError
        If ``result`` is not a non-empty tuple, or its last element is
        neither ``None`` nor a ``str``.
    """
    # A well-formed stage result is a tuple with a halt reason in the last slot.
    if not isinstance(result, tuple) or not result:
        # Reject malformed returns loudly (a programming defect, never swallowed).
        raise TypeError(f"stage {stage!r} returned {result!r}; expected a trailing halt_reason tuple")
    # The last slot is always the halt reason.
    halt_reason: Optional[str] = result[-1]
    # The halt slot must be None or a string.
    if halt_reason is not None and not isinstance(halt_reason, str):
        # Reject a non-string halt slot.
        raise TypeError(f"stage {stage!r} returned a non-str halt_reason: {halt_reason!r}")
    # A halted stage yields no payloads.
    if halt_reason is not None:
        # Return no payloads and the halt reason.
        return None, halt_reason
    # The payloads are the leading tuple.
    payloads: Tuple[Any, ...] = tuple(result[:-1])
    # Return the payloads and no halt reason.
    return payloads, None
```

```python
# Task 15: Robustness Suite via Orchestrator Extension

# ==============================================================================
# Task 15: Robustness Suite via Orchestrator Extension
# ==============================================================================

# Bind a module-scoped logger (libraries never configure the root logger).
LOGGER: logging.Logger = logging.getLogger("task_15")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: The pinned record types (QuantityDelta, VariantResult, SensitivityReport)
# -------------------------------------------------------------------------------------------------------------------------------

@dataclass(frozen=True)
class QuantityDelta:
    """
    Record one moved quantity of a sensitivity variant.

    Frozen, hashable dataclass pairing a single quantity under a variant
    specification with the same quantity under the primary
    specification. The delta is ``variant_value - primary_value``; both
    operands are disclosed so that no headline scalar appears without
    its primary value. Numeric fields are coerced to native ``float``
    at construction.

    Attributes
    ----------
    quantity : str
        Pinned quantity identifier, e.g. ``n_scorable`` or
        ``mean_new``.
    primary_value : float
        Quantity under the primary specification.
    variant_value : float
        Quantity under the variant's single-changed-argument
        specification.
    delta : float
        ``variant_value - primary_value``.
    tolerance_class : str
        Pinned comparability class, e.g. ``EXACT_INTEGER``.
    interpretation : str
        Pinned interpretation label.
    """

    # The pinned quantity identifier.
    quantity: str
    # The quantity under the primary specification.
    primary_value: float
    # The quantity under the variant.
    variant_value: float
    # The signed delta (variant minus primary).
    delta: float
    # The comparability class pinned by the step text / configuration.
    tolerance_class: str
    # The interpretation (pinned vocabulary; see the variant docstrings).
    interpretation: str

    def __post_init__(self) -> None:
        """
        Validate fields and coerce numeric members to native floats.

        Requires non-empty ``quantity``, ``tolerance_class`` and
        ``interpretation`` strings and finite real numeric values;
        bools are rejected. Numeric fields are rewritten as native
        ``float`` via ``object.__setattr__`` for homogeneous
        serialisation.

        Raises
        ------
        ValueError
            If a string field is not a non-empty ``str``, or a numeric
            field is not finite.
        TypeError
            If a numeric field is a ``bool`` or not a real number.
        """
        # The quantity identifier must be a non-empty string.
        if not isinstance(self.quantity, str) or not self.quantity.strip():
            # Refuse empty identifiers with the observed value.
            raise ValueError(f"QuantityDelta.quantity must be a non-empty str, got {self.quantity!r}")
        # The comparability class must be a non-empty string (the gate requires it).
        if not isinstance(self.tolerance_class, str) or not self.tolerance_class.strip():
            # Refuse empty tolerance classes with the observed value.
            raise ValueError(f"QuantityDelta.tolerance_class must be a non-empty str, got {self.tolerance_class!r}")
        # The interpretation must be a non-empty string.
        if not isinstance(self.interpretation, str) or not self.interpretation.strip():
            # Refuse empty interpretations with the observed value.
            raise ValueError(f"QuantityDelta.interpretation must be a non-empty str, got {self.interpretation!r}")
        # Every numeric field must be a finite real number (the gate requires finite deltas).
        for field_name, field_value in (
            ("primary_value", self.primary_value),
            ("variant_value", self.variant_value),
            ("delta", self.delta),
        ):
            # Bools are refused: they are not a measurement of a quantity.
            if isinstance(field_value, bool) or not isinstance(field_value, (int, float, np.number)):
                # Refuse non-real fields with name and type.
                raise TypeError(f"QuantityDelta.{field_name} must be a real number, got {type(field_value).__name__}")
            # Non-finite values are refused (NaN means "no measurement", and must surface as zero rows).
            if not math.isfinite(float(field_value)):
                # Refuse non-finite values with the observed value.
                raise ValueError(f"QuantityDelta.{field_name} must be finite, got {field_value!r}")
        # Coerce the numpy scalars into native floats (canonical serialisation homogeneity).
        object.__setattr__(self, "primary_value", float(self.primary_value))
        # Coerce the variant value likewise.
        object.__setattr__(self, "variant_value", float(self.variant_value))
        # Coerce the delta likewise.
        object.__setattr__(self, "delta", float(self.delta))

    def stable_dict(self) -> Dict[str, Any]:
        """
        Render the quantity as a canonical-JSON-ready mapping.

        Returns
        -------
        Dict[str, Any]
            Mapping with keys ``quantity``, ``primary_value``,
            ``variant_value``, ``delta``, ``tolerance_class`` and
            ``interpretation``. No stamp is added here.
        """
        # Enumerate the fields verbatim (the stamp is added by the orchestrator, not here).
        return {
            "quantity": self.quantity,
            "primary_value": self.primary_value,
            "variant_value": self.variant_value,
            "delta": self.delta,
            "tolerance_class": self.tolerance_class,
            "interpretation": self.interpretation,
        }

@dataclass(frozen=True)
class VariantResult:
    """
    Frozen outcome of one variant invocation against a primary result.

    The result of the pure-function contract
    ``f(primary, patch) -> VariantResult``. Frozen dataclass;
    assignment raises ``dataclasses.FrozenInstanceError``. The
    ``status`` and ``halt_state`` defaults defer to module constants so
    the class may be declared before those constants exist.

    Attributes
    ----------
    variant_name : str
        Pinned variant name.
    interpretation : str
        Variant-level interpretation.
    quantities : Tuple[QuantityDelta, ...]
        Moved quantities in pinned order; empty for unavailable or
        halted variants.
    isolation : Tuple[Tuple[str, bool], ...]
        ``(label, identical)`` "same by value" witnesses.
    patch_echo : str
        Canonical-JSON rendering of the applied patch.
    status : str
        One of ``EXECUTED``, ``VARIANT_UNAVAILABLE`` or ``HALTED``.
    halt_state : str
        ``CLOSED``, ``FAILED_CLOSURE`` or another pinned disclosure
        token.
    ledger_partition : str
        Non-empty partition disclosure string.
    note : str
        Human-readable provenance note.
    closures : Tuple[Tuple[str, str, str], ...]
        Seven ``(closure_id, status, detail)`` triples with status in
        ``{CLOSED, OPEN, DEFERRED}``.
    """

    # The pinned variant name.
    variant_name: str
    # The interpretation of the variant as a whole.
    interpretation: str
    # The moved quantities (empty when the variant is unavailable or halted).
    quantities: Tuple[QuantityDelta, ...] = ()
    # The same-by-value isolation witnesses ((label, identical) pairs).
    isolation: Tuple[Tuple[str, bool], ...] = ()
    # The canonical-JSON echo of the applied patch.
    patch_echo: str = "{}"
    # The execution status (deferred to call time: the constant is defined later in the module).
    status: str = field(default_factory=lambda: VARIANT_STATUS_EXECUTED)
    # The closure/determinism/unavailability halt token (deferred for the same reason).
    halt_state: str = field(default_factory=lambda: HALT_CLOSED)
    # The partition disclosure (never None).
    ledger_partition: str = "NOT_EVALUATED"
    # The provenance note.
    note: str = ""
    # The seven closures as (id, status, detail) triples.
    closures: Tuple[Tuple[str, str, str], ...] = ()

    def __post_init__(self) -> None:
        """
        Validate the result fields and their container types.

        Raises
        ------
        ValueError
            If ``variant_name`` or ``interpretation`` is not a
            non-empty ``str``, if ``status`` is not a pinned token, or
            if ``halt_state`` or ``ledger_partition`` is empty.
        TypeError
            If a quantity is not a ``QuantityDelta``, an isolation
            witness is not a ``(str, bool)`` pair, or a closure is not a
            ``(str, str, str)`` triple of non-empty strings.
        """
        # The variant name must be a non-empty string.
        if not isinstance(self.variant_name, str) or not self.variant_name.strip():
            # Refuse empty names with the observed value.
            raise ValueError(f"VariantResult.variant_name must be a non-empty str, got {self.variant_name!r}")
        # The interpretation must be a non-empty string.
        if not isinstance(self.interpretation, str) or not self.interpretation.strip():
            # Refuse empty interpretations with the observed value.
            raise ValueError(f"VariantResult.interpretation must be a non-empty str, got {self.interpretation!r}")
        # Every quantity must be a QuantityDelta (typed container integrity).
        for index, quantity in enumerate(self.quantities):
            # Refuse foreign payloads with position and type.
            if not isinstance(quantity, QuantityDelta):
                raise TypeError(f"VariantResult.quantities[{index}] must be QuantityDelta, got {type(quantity).__name__}")
        # Every isolation witness must be a (str, bool) pair.
        for index, witness in enumerate(self.isolation):
            # Refuse malformed witnesses with position and value.
            if (
                not isinstance(witness, tuple)
                or len(witness) != 2
                or not isinstance(witness[0], str)
                or not isinstance(witness[1], bool)
            ):
                raise TypeError(f"VariantResult.isolation[{index}] must be a (str, bool) pair, got {witness!r}")
        # Every closure must be a (str, str, str) triple.
        for index, closure in enumerate(self.closures):
            # Refuse malformed closures with position and value.
            if (
                not isinstance(closure, tuple)
                or len(closure) != 3
                or not all(isinstance(cell, str) and cell for cell in closure)
            ):
                raise TypeError(f"VariantResult.closures[{index}] must be a (str, str, str) triple, got {closure!r}")
        # The status must be one of the pinned tokens.
        if self.status not in (VARIANT_STATUS_EXECUTED, VARIANT_STATUS_UNAVAILABLE, VARIANT_STATUS_HALTED):
            # Refuse unknown statuses with the observed value.
            raise ValueError(f"VariantResult.status must be one of the pinned tokens, got {self.status!r}")
        # The halt_state and ledger_partition must be non-empty strings (the gate requires them).
        if not isinstance(self.halt_state, str) or not self.halt_state.strip():
            # Refuse empty halt states with the observed value.
            raise ValueError(f"VariantResult.halt_state must be a non-empty str, got {self.halt_state!r}")
        # The partition disclosure must be non-empty (the gate requires non-None).
        if not isinstance(self.ledger_partition, str) or not self.ledger_partition.strip():
            # Refuse empty partition disclosures with the observed value.
            raise ValueError(f"VariantResult.ledger_partition must be a non-empty str, got {self.ledger_partition!r}")

    def stable_payload(self) -> Dict[str, Any]:
        """
        Render the computationally relevant content for hashing.

        The human-readable ``note`` field is excluded so that
        equivalent invocations hash identically.

        Returns
        -------
        Dict[str, Any]
            Keys ``variant_name``, ``interpretation``, ``quantities``,
            ``isolation``, ``patch_echo``, ``status``, ``halt_state``,
            ``ledger_partition`` and ``closures``.
        """
        # Enumerate the stable fields verbatim (order irrelevant; the encoder sorts keys).
        return {
            "variant_name": self.variant_name,
            "interpretation": self.interpretation,
            "quantities": [quantity.stable_dict() for quantity in self.quantities],
            "isolation": [list(witness) for witness in self.isolation],
            "patch_echo": self.patch_echo,
            "status": self.status,
            "halt_state": self.halt_state,
            "ledger_partition": self.ledger_partition,
            "closures": [list(closure) for closure in self.closures],
        }

    def stable_hash(self) -> str:
        """
        Compute the determinism hash of ``stable_payload``.

        Returns
        -------
        str
            64 lowercase hexadecimal characters (SHA-256 of the
            canonical JSON encoding).
        """
        # Digest the canonical encoding of the stable payload.
        return _sha256_json(self.stable_payload())

    def to_report_rows(self) -> List[Dict[str, Any]]:
        """
        Materialise the SENSITIVITY_REPORT rows for this result.

        Returns
        -------
        List[Dict[str, Any]]
            One dict per quantity with keys ``variant_name``,
            ``quantity``, ``primary_value``, ``variant_value``,
            ``delta``, ``tolerance_class``, ``interpretation`` and
            ``stamp``. Empty for unavailable or halted variants.
        """
        # Accumulate one row per quantity; unavailable/failed variants yield an empty list.
        rows: List[Dict[str, Any]] = []
        # Walk the quantities in pinned order.
        for quantity in self.quantities:
            # Build the row; floats are native Python floats already (QuantityDelta coerces).
            rows.append(
                {
                    "variant_name": self.variant_name,
                    "quantity": quantity.quantity,
                    "primary_value": quantity.primary_value,
                    "variant_value": quantity.variant_value,
                    "delta": quantity.delta,
                    "tolerance_class": quantity.tolerance_class,
                    "interpretation": quantity.interpretation,
                    "stamp": SENSITIVITY_STAMP,
                }
            )
        # Return the rows materialised.
        return rows

@dataclass(frozen=True)
class SensitivityReport:
    """
    Frozen read-only artefact of the sensitivity archive.

    Bundles the per-quantity delta table, its content hash, the
    per-variant run table, the pinned ``[R]``-tier stamp and the P20
    withholding notice, plus the tier-ceilinged claim register and the
    archive deposit manifest when the orchestrator wires them. No
    consumer may mutate the frames in place or treat a variant result
    as evidence for or against any substantive claim.

    Attributes
    ----------
    SENSITIVITY_REPORT : pd.DataFrame
        One row per ``(variant_name, quantity)`` with the pinned
        columns plus ``stamp``; addressed exactly by
        ``SENSITIVITY_HASH``.
    SENSITIVITY_HASH : str
        SHA-256 hex digest (64 lowercase hex characters) of the
        canonical-JSON serialisation of the report rows.
    VARIANT_RUNS : pd.DataFrame
        One row per requested variant with ``determinism_hash``,
        ``halt_state`` and ``ledger_partition``.
    stamp : str
        Pinned ``SENSITIVITY__[R]__NOT_A_PAPER_FINDING`` stamp.
    withholding_notice : str
        Pinned substantive-claim withholding notice (P20).
    claim_register : Tuple[Any, ...]
        Sensitivity claim records, each tier-ceilinged at
        ``reconstruction``; empty when the archive carries no rows.
    archive_manifest : Optional[Mapping[str, Any]]
        Deposit manifest for the archive (P36 plus the Task 15
        extension); ``None`` for directly constructed reports.
    """

    # The delta report table.
    SENSITIVITY_REPORT: pd.DataFrame
    # The hash over the serialised rows.
    SENSITIVITY_HASH: str
    # The per-variant run table.
    VARIANT_RUNS: pd.DataFrame
    # The pinned stamp (Step 3 tagging; deferred: defined later in the module).
    stamp: str = field(default_factory=lambda: SENSITIVITY_STAMP)
    # The pinned withholding notice (Step 3 substantive-claim prohibition; deferred).
    withholding_notice: str = field(default_factory=lambda: SENSITIVITY_WITHHOLDING_NOTE)
    # The sensitivity claim register, wired by the orchestrator (Step 3, tier ceiling reconstruction).
    claim_register: Tuple[Any, ...] = ()
    # The sensitivity archive deposit manifest, wired by the orchestrator (Step 3, P36 plus the Task 15 extension).
    archive_manifest: Optional[Mapping[str, Any]] = None

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: Boundary validators and the shared arithmetic substrate
# -------------------------------------------------------------------------------------------------------------------------------

def _validate_primary_like(primary: Any, *, caller: str) -> None:
    """
    Check that ``primary`` exposes a primary-result attribute.

    At least one of ``factor_aggregates``, ``ledger_primary``,
    ``ledger_reverse``, ``frozen_cohort``, ``preserved_rows``,
    ``daily_scores_v2``, ``score_register_v2``,
    ``RE_EVALUATION_LEDGER``, ``stages_output``, ``config``,
    ``cohort_timestamp``, ``evaluator_canonicalizer`` or
    ``oos_read_timestamp`` must be present on the object.

    Parameters
    ----------
    primary : Any
        The object passed as the primary result (StudyResult or
        duck-type).
    caller : str
        Calling callable name, prefixed to the error message.

    Raises
    ------
    TypeError
        If ``primary`` is ``None`` or exposes none of the pinned
        attributes.
    """
    # A missing primary is unusable.
    if primary is None:
        # Refuse with the caller's identity.
        raise TypeError(f"{caller}: primary must not be None")
    # Enumerate the attributes that make an object a primary result for this module.
    known: Tuple[str, ...] = (
        "factor_aggregates",
        "ledger_primary",
        "ledger_reverse",
        "frozen_cohort",
        "preserved_rows",
        "daily_scores_v2",
        "score_register_v2",
        "RE_EVALUATION_LEDGER",
        "stages_output",
        "config",
        "cohort_timestamp",
        "evaluator_canonicalizer",
        "oos_read_timestamp",
    )
    # At least one known attribute must be present.
    if not any(hasattr(primary, name) for name in known):
        # Refuse foreign objects with the observed type.
        raise TypeError(
            f"{caller}: primary does not quack like a primary result (none of {known}); "
            f"got {type(primary).__name__}"
        )

def _validate_patch_mapping(patch: Any, *, caller: str) -> Dict[str, Any]:
    """
    Validate a patch mapping and return a deep copy of it.

    Parameters
    ----------
    patch : Any
        Candidate variant patch; must be a ``Mapping``.
    caller : str
        Calling callable name, prefixed to the error message.

    Returns
    -------
    Dict[str, Any]
        Deep copy of the patch keys and values; the caller's mapping is
        never aliased.

    Raises
    ------
    TypeError
        If ``patch`` is not a ``Mapping``.
    """
    # The patch must be a mapping (the pinned contract for every variant).
    if not isinstance(patch, Mapping):
        # Refuse non-mapping patches with the observed type.
        raise TypeError(f"{caller}: patch must be a Mapping of the changed-argument name to its value, got {type(patch).__name__}")
    # Deep-copy so the caller's mapping is never aliased into results or echoes.
    return copy.deepcopy(dict(patch))

def _ensure_patch_keys(patch: Mapping[str, Any], *, allowed: set, caller: str) -> None:
    """
    Reject patch keys outside the variant's pinned key set.

    Parameters
    ----------
    patch : Mapping[str, Any]
        Already validated patch mapping.
    allowed : set
        Keys the variant admits (string members).
    caller : str
        Calling callable name, prefixed to the error message.

    Raises
    ------
    ValueError
        If ``patch`` carries any key outside ``allowed``, violating the
        single-changed-parameter contract.
    """
    # Compute the offending keys (sorted for deterministic messages).
    offenders: List[str] = sorted(str(key) for key in patch if str(key) not in allowed)
    # Any offender voids the "exactly one declared parameter" invariant.
    if offenders:
        # Refuse the patch with the offending key list.
        raise ValueError(f"{caller}: patch admits exactly the pinned keys {sorted(allowed)}, got unexpected keys {offenders}")

def _validate_precedence(sequence: Any, *, caller: str) -> Tuple[str, str, str]:
    """
    Validate a precedence sequence as a P04 state permutation.

    The sequence must be a non-string ``Sequence`` of three members,
    each a ``str``, whose set equals ``P04_STATES``.

    Parameters
    ----------
    sequence : Any
        Candidate precedence, e.g. ``["scorable", "insufficient",
        "unsupported"]``.
    caller : str
        Calling callable name, prefixed to error messages.

    Returns
    -------
    Tuple[str, str, str]
        The validated triple in the given order.

    Raises
    ------
    TypeError
        If ``sequence`` is a ``str``/``bytes`` or not a ``Sequence``,
        or a member is not a ``str``.
    ValueError
        If the members are not exactly the three P04 state names.
    """
    # A bare string would iterate into characters; refuse it.
    if isinstance(sequence, (str, bytes)) or not isinstance(sequence, Sequence):
        # Refuse non-sequence precedences with the observed type.
        raise TypeError(f"{caller}: state_precedence must be a Sequence[str], got {type(sequence).__name__}")
    # Freeze the members as plain strings with positional type checks.
    members: Tuple[str, ...] = tuple(_typed_str(member, caller=caller, path=f"state_precedence[{index}]") for index, member in enumerate(sequence))
    # The vocabulary is exactly three states; enforce it.
    if len(members) != 3 or set(members) != set(P04_STATES):
        # Refuse vocabularies with the observed members.
        raise ValueError(f"{caller}: state_precedence must be a permutation of {list(P04_STATES)}, got {list(members)}")
    # Return the frozen triple (the caller may refine the pinned ordering).
    return (members[0], members[1], members[2])

def _typed_str(value: Any, *, caller: str, path: str) -> str:
    """
    Return ``value`` unchanged after asserting it is a ``str``.

    Parameters
    ----------
    value : Any
        Candidate member of a nested structure.
    caller : str
        Calling callable name, prefixed to the error message.
    path : str
        Positional or dotted path of the member, used in the message.

    Returns
    -------
    str
        The validated string.

    Raises
    ------
    TypeError
        If ``value`` is not a ``str``.
    """
    # The member must be a plain string.
    if not isinstance(value, str):
        # Refuse non-strings with path and observed type.
        raise TypeError(f"{caller}: {path} must be a str, got {type(value).__name__}")
    # Return the validated string unchanged.
    return value

def _require_factor_aggregates(primary: Any) -> pd.DataFrame:
    """
    Return ``primary.factor_aggregates`` after schema validation.

    The attribute must be a ``pandas.DataFrame`` carrying the columns
    ``expression_id`` and ``mean_rank_ic`` and must be unique on
    ``expression_id`` (the join-key contract). The frame is returned
    unchanged, without copying or reindexing.

    Parameters
    ----------
    primary : Any
        The primary result (StudyResult or duck-type).

    Returns
    -------
    pd.DataFrame
        The validated frame, unique on ``expression_id``.

    Raises
    ------
    TypeError
        If ``factor_aggregates`` is absent or not a ``DataFrame``.
    ValueError
        If a required column is missing or ``expression_id`` carries
        duplicate values.
    """
    # Fetch the attribute (absent is a typed boundary failure, not an empty result).
    frame = getattr(primary, "factor_aggregates", None)
    # The frame must exist and be a DataFrame.
    if not isinstance(frame, pd.DataFrame):
        # Refuse with the observed type.
        raise TypeError(f"primary.factor_aggregates must be a pandas.DataFrame, got {type(frame).__name__}")
    # The pinned columns must be present.
    missing: List[str] = [column for column in ("expression_id", "mean_rank_ic") if column not in frame.columns]
    # Any missing column voids every downstream step.
    if missing:
        # Refuse with the missing column list.
        raise ValueError(f"primary.factor_aggregates is missing required columns {missing}")
    # The expression_id column must be duplicate-free (the join key contract).
    ids: pd.Series = frame["expression_id"].astype("string")
    # Detect duplicated identifiers.
    duplicated: pd.Series = ids.duplicated(keep=False)
    # Any duplication voids the one-to-one join guarantee.
    if bool(duplicated.any()):
        # Report the duplicated identifiers (bounded list).
        raise ValueError(f"primary.factor_aggregates carries duplicated expression_id values: {sorted(ids[duplicated].unique())[:10]}")
    # Return the validated frame unchanged (no copies, no reindexing).
    return frame

def _safe_factor_aggregates(primary: Any) -> Optional[pd.DataFrame]:
    """
    Return ``primary.factor_aggregates`` if a frame, else ``None``.

    Best-effort variant used to feed closure evaluation; a missing or
    non-``DataFrame`` attribute is signalled as ``None`` so callers can
    mark closures ``DEFERRED`` rather than silently passing them.

    Parameters
    ----------
    primary : Any
        The primary result (StudyResult or duck-type).

    Returns
    -------
    Optional[pd.DataFrame]
        The frame, or ``None`` when the attribute is absent or not a
        ``DataFrame``.
    """
    # Fetch the attribute; anything unusable degrades to DEFERRED closures.
    frame = getattr(primary, "factor_aggregates", None)
    # Return the frame only when it is a DataFrame.
    return frame if isinstance(frame, pd.DataFrame) else None

def _normalise_ledger_triple(payload: Any, *, label: str) -> Tuple[int, int, int]:
    """
    Normalise a ledger payload to the state triple.

    Accepts either a ``Mapping`` with the P04 state keys
    (case-insensitive) or a positional ``Sequence`` of at least three
    members; members after the third are ignored. Bools are rejected,
    integral floats are coerced to ``int``, and counts must be
    non-negative.

    Parameters
    ----------
    payload : Any
        Ledger payload in mapping or positional-sequence shape.
    label : str
        Payload label, e.g. ``"ledger_primary"``, prefixed to error
        messages.

    Returns
    -------
    Tuple[int, int, int]
        The ``(scorable, insufficient, unsupported)`` counts.

    Raises
    ------
    TypeError
        If the payload shape is unsupported or a member is not an
        integer-valued number.
    ValueError
        If a state key is missing from a mapping, a sequence is shorter
        than three, a member is non-integral, or a count is negative.
    """
    # Mapping branch (the acceptance-fixture shape): case-insensitive state-key lookup.
    if isinstance(payload, Mapping):
        # Lower-case the keys once for the lookup.
        lowered: Dict[str, Any] = {str(key).lower(): value for key, value in payload.items()}
        # Resolve the three state members.
        members: List[Any] = [lowered.get(state) for state in P04_STATES]
        # Any absent state is a contract breach.
        if any(member is None for member in members):
            # Refuse with the missing keys.
            raise ValueError(f"{label}: mapping lacks one of the state keys {list(P04_STATES)}; got keys {sorted(lowered)}")
    # Sequence branch (the StudyResult shape): positional triple.
    elif isinstance(payload, Sequence) and not isinstance(payload, (str, bytes)):
        # Require at least three positional members.
        if len(payload) < 3:
            # Refuse short sequences with the observed length.
            raise ValueError(f"{label}: positional triple needs at least 3 members, got {len(payload)}")
        # Take the first three positional members.
        members = [payload[0], payload[1], payload[2]]
    # Anything else is unusable.
    else:
        # Refuse with the observed type.
        raise TypeError(f"{label}: expected a positional Sequence or a Mapping, got {type(payload).__name__}")
    # Coerce each member to an int with strict integrality and positivity checks.
    coerced: List[int] = []
    # Walk the members with their labels.
    for state, member in zip(P04_STATES, members):
        # Bools are refused explicitly.
        if isinstance(member, bool) or not isinstance(member, (int, float, np.number)):
            # Refuse non-integral members with state and type.
            raise TypeError(f"{label}.{state} must be an integer, got {type(member).__name__}")
        # Integral floats are accepted and coerced.
        as_float: float = float(member)
        # Reject non-integral magnitudes.
        if not as_float.is_integer():
            # Refuse fractional members with state and value.
            raise ValueError(f"{label}.{state} must be integral, got {as_float}")
        # Enforce non-negativity.
        if as_float < 0:
            # Refuse negative counts with state and value.
            raise ValueError(f"{label}.{state} must be >= 0, got {as_float}")
        # Record the coerced member.
        coerced.append(int(as_float))
    # Return the frozen triple.
    return (coerced[0], coerced[1], coerced[2])

def _ledger_payload_denominator(payload: Any) -> Optional[int]:
    """
    Read ``denominator_total`` from a mapping-shaped ledger payload.

    Parameters
    ----------
    payload : Any
        Ledger payload (mapping or sequence).

    Returns
    -------
    Optional[int]
        The non-negative integer denominator, or ``None`` when absent
        or malformed (non-mapping payload; bool, non-number,
        non-integral or negative value).
    """
    # Only mappings may carry the denominator.
    if not isinstance(payload, Mapping):
        # Sequences carry no denominator by construction.
        return None
    # Lower-case keys for the lookup.
    lowered: Dict[str, Any] = {str(key).lower(): value for key, value in payload.items()}
    # Fetch the denominator member.
    member: Any = lowered.get("denominator_total")
    # Refuse bools and non-numbers.
    if member is None or isinstance(member, bool) or not isinstance(member, (int, float, np.number)):
        # No usable denominator on this payload.
        return None
    # Coerce to int when integral and non-negative.
    as_float: float = float(member)
    # Reject non-integral or negative denominators.
    if not as_float.is_integer() or as_float < 0:
        # No usable denominator on this payload.
        return None
    # Return the denominator as an int.
    return int(as_float)

def _p04_denominator_config(config: Optional[Mapping[str, Any]]) -> Optional[int]:
    """
    Read the P04 denominator from a STUDY_CONFIG mapping.

    Looks up ``P04_LEDGER_PARAMETERS.denominator_total``.

    Parameters
    ----------
    config : Optional[Mapping[str, Any]]
        STUDY_CONFIG mapping or its mirror.

    Returns
    -------
    Optional[int]
        The non-negative integer denominator, or ``None`` when the path
        is absent or malformed (non-mapping section; bool, non-number,
        non-integral or negative value).
    """
    # A missing config yields no fallback.
    if not isinstance(config, Mapping):
        # Nothing to extract.
        return None
    # Fetch the P04 section.
    section: Any = config.get("P04_LEDGER_PARAMETERS")
    # The section must be a mapping.
    if not isinstance(section, Mapping):
        # Nothing to extract.
        return None
    # Fetch the denominator member.
    member: Any = section.get("denominator_total")
    # Refuse bools and non-numbers.
    if member is None or isinstance(member, bool) or not isinstance(member, (int, float, np.number)):
        # Nothing to extract.
        return None
    # Coerce when integral and non-negative.
    as_float: float = float(member)
    # Reject non-integral or negative values.
    if not as_float.is_integer() or as_float < 0:
        # Nothing to extract.
        return None
    # Return the configured denominator.
    return int(as_float)

def _p03_pin(config: Optional[Mapping[str, Any]], key: str, default: Any) -> Any:
    """
    Read a P03 aggregation pin with a documented fallback default.

    Parameters
    ----------
    config : Optional[Mapping[str, Any]]
        STUDY_CONFIG mapping or its mirror.
    key : str
        ``P03_FACTOR_AGGREGATION_PARAMETERS`` key to read.
    default : Any
        Value returned when ``config`` or the P03 section is absent, or
        when the key is missing.

    Returns
    -------
    Any
        The configured value or ``default``; not type-checked here.
    """
    # A missing config yields the documented default.
    if not isinstance(config, Mapping):
        # Return the documented default.
        return default
    # Fetch the P03 section.
    section: Any = config.get("P03_FACTOR_AGGREGATION_PARAMETERS")
    # The section must be a mapping.
    if not isinstance(section, Mapping):
        # Return the documented default.
        return default
    # Return the configured value or the documented default.
    return section.get(key, default)

def _p03_int_pin(config: Optional[Mapping[str, Any]], key: str, default: int) -> int:
    """
    Read a P03 pin guaranteed to be a positive integer.

    Parameters
    ----------
    config : Optional[Mapping[str, Any]]
        STUDY_CONFIG mapping or its mirror.
    key : str
        ``P03_FACTOR_AGGREGATION_PARAMETERS`` key to read.
    default : int
        Documented positive default used when the path is absent.

    Returns
    -------
    int
        The configured or default value, coerced from an integral
        float.

    Raises
    ------
    ValueError
        If the resolved value is a bool, not a number, non-integral, or
        not strictly positive.
    """
    # Fetch the raw member (or the documented default).
    member: Any = _p03_pin(config, key, default)
    # Refuse bools.
    if isinstance(member, bool) or not isinstance(member, (int, float, np.number)):
        # Refuse non-numeric pins with the observed type.
        raise ValueError(f"P03_FACTOR_AGGREGATION_PARAMETERS.{key} must be a positive integer, got {type(member).__name__}")
    # Integral floats are accepted.
    as_float: float = float(member)
    # Reject non-integral or non-positive values.
    if not as_float.is_integer() or as_float <= 0:
        # Refuse the pin with the observed value.
        raise ValueError(f"P03_FACTOR_AGGREGATION_PARAMETERS.{key} must be a positive integer, got {as_float}")
    # Return the positive int.
    return int(as_float)

def _p03_float_pin(config: Optional[Mapping[str, Any]], key: str, default: float) -> float:
    """
    Read a P03 pin guaranteed to be a non-negative finite float.

    Parameters
    ----------
    config : Optional[Mapping[str, Any]]
        STUDY_CONFIG mapping or its mirror.
    key : str
        ``P03_FACTOR_AGGREGATION_PARAMETERS`` key to read.
    default : float
        Documented non-negative default used when the path is absent.

    Returns
    -------
    float
        The configured or default value.

    Raises
    ------
    ValueError
        If the resolved value is a bool, not a number, non-finite, or
        negative.
    """
    # Fetch the raw member (or the documented default).
    member: Any = _p03_pin(config, key, default)
    # Refuse bools and non-numbers.
    if isinstance(member, bool) or not isinstance(member, (int, float, np.number)):
        # Refuse non-numeric pins with the observed type.
        raise ValueError(f"P03_FACTOR_AGGREGATION_PARAMETERS.{key} must be a non-negative finite float, got {type(member).__name__}")
    # Coerce to float.
    as_float: float = float(member)
    # Reject non-finite or negative values.
    if not math.isfinite(as_float) or as_float < 0:
        # Refuse the pin with the observed value.
        raise ValueError(f"P03_FACTOR_AGGREGATION_PARAMETERS.{key} must be a non-negative finite float, got {as_float}")
    # Return the validated float.
    return as_float

def _p04_sign_partition_counts(config: Optional[Mapping[str, Any]]) -> Tuple[Optional[int], Optional[int]]:
    """
    Read ``P04_LEDGER_PARAMETERS.sign_partition_counts`` from a config.

    The positive/negative split is returned as a comparison value only;
    nothing is asserted against it here.

    Parameters
    ----------
    config : Optional[Mapping[str, Any]]
        STUDY_CONFIG mapping or its mirror.

    Returns
    -------
    Tuple[Optional[int], Optional[int]]
        The ``(positive, negative)`` counts, or ``(None, None)`` when
        the path is absent or malformed (non-mapping section or member;
        bool, non-number, non-integral or negative count).
    """
    # A missing config yields no comparison values.
    if not isinstance(config, Mapping):
        # Nothing to extract.
        return None, None
    # Fetch the P04 section.
    section: Any = config.get("P04_LEDGER_PARAMETERS")
    # The section must be a mapping.
    if not isinstance(section, Mapping):
        # Nothing to extract.
        return None, None
    # Fetch the sign_partition_counts member.
    member: Any = section.get("sign_partition_counts")
    # The member must be a mapping with both keys.
    if not isinstance(member, Mapping):
        # Nothing to extract.
        return None, None
    # Resolve and coerce the positive count.
    def _coerce_count(value: Any) -> Optional[int]:
        """
        Coerce one sign-partition count to a non-negative int.

        Parameters
        ----------
        value : Any
            Candidate count.

        Returns
        -------
        Optional[int]
            The count as an ``int``, or ``None`` for bools,
            non-numbers, non-integral or negative values.
        """
        # Bools are refused; integral members only; negatives refused.
        if value is None or isinstance(value, bool) or not isinstance(value, (int, float, np.number)):
            # Nothing to extract.
            return None
        # Integral floats are accepted.
        as_float: float = float(value)
        # Reject non-integral or negative counts.
        if not as_float.is_integer() or as_float < 0:
            # Nothing to extract.
            return None
        # Return the count as an int.
        return int(as_float)
    # Return the two coerced counts (either may be None).
    return _coerce_count(member.get("positive")), _coerce_count(member.get("negative"))

def _primary_config(primary: Any) -> Optional[Mapping[str, Any]]:
    """
    Return ``primary.config`` when it is a mapping, else ``None``.

    Parameters
    ----------
    primary : Any
        The primary result (StudyResult or duck-type).

    Returns
    -------
    Optional[Mapping[str, Any]]
        The attached configuration mapping, or ``None`` when absent or
        not a ``Mapping``.
    """
    # Read the attribute without side effects.
    candidate: Any = getattr(primary, "config", None)
    # Return it only when it is a mapping.
    return candidate if isinstance(candidate, Mapping) else None

def _denominator_hint(
    *,
    primary: Any,
    aggregates: Optional[pd.DataFrame],
    ledger_payload: Any,
    config: Optional[Mapping[str, Any]],
) -> int:
    """
    Resolve the closure-battery denominator without fabrication.

    Resolution order: the ledger payload's own ``denominator_total``
    when positive; else ``len(aggregates)`` when the frame is non-empty;
    else ``P04_LEDGER_PARAMETERS.denominator_total`` from ``config``
    when positive; otherwise raise.

    Parameters
    ----------
    primary : Any
        The primary result (StudyResult or duck-type).
    aggregates : Optional[pd.DataFrame]
        FACTOR_AGGREGATES frame when available.
    ledger_payload : Any
        Payload used for the primary ledger (mapping or sequence).
    config : Optional[Mapping[str, Any]]
        Attached configuration mapping when present.

    Returns
    -------
    int
        The resolved positive denominator.

    Raises
    ------
    ValueError
        If no positive denominator can be resolved from the payload,
        the frame or the configuration.
    """
    # First resort: the payload's own denominator_total.
    from_payload: Optional[int] = _ledger_payload_denominator(ledger_payload)
    # A positive payload denominator wins (it is the run's own accounting).
    if from_payload is not None and from_payload > 0:
        # Return the payload denominator.
        return from_payload
    # Second resort: the cohort frame's row count (the cohort order, one row per expression).
    if isinstance(aggregates, pd.DataFrame) and len(aggregates) > 0:
        # Return the frame length as the denominator.
        return int(len(aggregates))
    # Third resort: the P04 configuration path.
    from_config: Optional[int] = _p04_denominator_config(config)
    # A positive configured denominator is acceptable.
    if from_config is not None and from_config > 0:
        # Return the configured denominator.
        return from_config
    # Nothing resolvable: refuse rather than fabricate.
    raise ValueError(
        "denominator cannot be resolved from the ledger payload, the factor_aggregates "
        "frame or the configuration; refusing to fabricate one"
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: The pinned vocabulary (constants) and the canonical serialisation
# -------------------------------------------------------------------------------------------------------------------------------

# The stamp pinned by the step text (Step 3 tagging); applied to every row and run record.
SENSITIVITY_STAMP: str = "SENSITIVITY__[R]__NOT_A_PAPER_FINDING"

# The pinned inadmissibility label for variant 3 (P06 sign symmetry; never an admissible specification).
VARIANT_3_INADMISSIBLE_LABEL: str = "RECONSTRUCTION_OF_AN_INADMISSIBLE_VARIANT"

# The pinned disclosure token for variant 5 when Task 13 halted or its ledger is absent.
VARIANT_5_UNAVAILABLE_PENDING_CORRECTION: str = "VARIANT_UNAVAILABLE_PENDING_CORRECTION"

# The pinned disclosure token for variants that lack the artefacts a freeze re-execution requires.
VARIANT_UNAVAILABLE_PENDING_INPUTS: str = "VARIANT_UNAVAILABLE_PENDING_INPUTS"

# The pinned disclosure token for variants whose arms carry no usable finite means.
VARIANT_UNAVAILABLE_PENDING_DATA: str = "VARIANT_UNAVAILABLE_PENDING_DATA"

# The pinned disclosure token for a RE_EVALUATION_LEDGER whose schema matches neither pinned layout.
VARIANT_UNAVAILABLE_PENDING_LEDGER_SCHEMA: str = "VARIANT_UNAVAILABLE_PENDING_LEDGER_SCHEMA"

# The pinned halt token for a variant that fails the determinism re-verification (Step 2b).
DETERMINISM_HALT: str = "DETERMINISM_HALT"

# The pinned halt token for a variant whose ledger closures do not close (Step 2c).
FAILED_CLOSURE: str = "FAILED_CLOSURE"

# The halt token for a variant whose evaluated closures all close.
HALT_CLOSED: str = "CLOSED"

# The status token for an executed variant (rows may or may not be emitted).
VARIANT_STATUS_EXECUTED: str = "EXECUTED"

# The status token for a variant that cannot produce numbers with the inputs it was given.
VARIANT_STATUS_UNAVAILABLE: str = "VARIANT_UNAVAILABLE"

# The status token for a variant whose execution raised before producing a result.
VARIANT_STATUS_HALTED: str = "HALTED"

# The paper-pinned primary near-zero threshold (P03.near_zero_absolute_threshold; documented default).
SIGN_CLASSIFICATION_PRIMARY_THRESHOLD_DEFAULT: float = 1e-12

# The step-pinned secondary threshold for variant 4 (|mean_rank_ic| < 1e-6).
SIGN_CLASSIFICATION_SECONDARY_THRESHOLD: float = 1e-06

# The pinned materiality tolerance for variant 5's interpretation rule ([R]; a chosen convention,
# not a paper-stated value; disclosed as such wherever it is consumed).
DEFECT_SELECTION_MATERIALITY_TOLERANCE: float = 1e-09

# The three pinned interpretation labels for variant 5 (exact literals, no suffixes).
DEFECT_INTERPRETATION_HIGHER_SIGNAL: str = "defect destroyed higher-signal"
DEFECT_INTERPRETATION_LOWER_SIGNAL: str = "defect destroyed lower-signal"
DEFECT_INTERPRETATION_NO_MATERIAL: str = "no material difference"

# The pinned reversed precedence triple for variant 1 (P04.state_precedence_consequence).
REVERSED_PRECEDENCE_TUPLE: Tuple[str, str, str] = ("scorable", "insufficient", "unsupported")

# The pinned state vocabulary (P04.states).
P04_STATES: Tuple[str, str, str] = ("scorable", "insufficient", "unsupported")

# The paper-pinned inclusive scorability gate (P03.min_valid_dates_for_scorability; documented default).
MIN_VALID_DATES_PINNED: int = 20

# The paper-pinned full-coverage cardinality (P03.full_coverage_target_dates; documented default).
FULL_COVERAGE_TARGET_DATES_PINNED: int = 480

# The seven pinned columns of the delta report (Step 2a), in pinned order.
PINNED_REPORT_COLUMNS: Tuple[str, ...] = (
    "variant_name",
    "quantity",
    "primary_value",
    "variant_value",
    "delta",
    "tolerance_class",
    "interpretation",
)

# The full output column order of SENSITIVITY_REPORT (pinned columns plus the stamp column).
REPORT_COLUMNS: Tuple[str, ...] = PINNED_REPORT_COLUMNS + ("stamp",)

# The column order of VARIANT_RUNS (per-variant run record).
RUN_COLUMNS: Tuple[str, ...] = (
    "variant_name",
    "status",
    "halt_state",
    "ledger_partition",
    "determinism_hash",
    "n_report_rows",
    "note",
)

# The five pinned variant names in step order (the gate pins this exact order).
PINNED_VARIANT_NAMES: Tuple[str, ...] = (
    "variant_state_precedence_reversed",
    "variant_dedup_string_equality",
    "variant_dedup_negation_push_through",
    "variant_sign_classification_tolerance",
    "variant_defect_selection_consequence",
)

# The pinned string-equality patch for variant 2 (full replacement of the P06 canonicalisation block).
VARIANT_2_STRING_EQUALITY_CANONICALIZATION: Dict[str, Any] = {
    "parse_to_ast": False,
    "normalise_whitespace": True,
    "apply_commutativity": "symmetric_operators_only",
    "push_negation_through": False,
}

# The pinned reconstruction patch for variant 3 (a one-key override of the archive canonicalisation).
VARIANT_3_PUSH_NEGATION_PATCH: Dict[str, Any] = {"push_negation_through": True}

# The pinned reversed-precedence patch for variant 1 (P04.state_precedence reversed).
VARIANT_1_REVERSED_PRECEDENCE_PATCH: Dict[str, Any] = {
    "state_precedence": ["scorable", "insufficient", "unsupported"],
}

# The reconstruction-tier ceiling for every ClaimRecord this module may emit (Section E / P20).
SENSITIVITY_CLAIM_TIER_CEILING: str = "reconstruction"

# The six substantive-claim classes the sensitivity results are explicitly withheld from (P20).
WITHHELD_CLAIM_CLASSES: Tuple[str, ...] = (
    "ranking_claim",
    "profitability_claim",
    "significance_test_on_aggregate",
    "oos_or_generalization_claim",
    "matched_sample_outperformance_claim",
    "incremental_agent_skill_claim",
)

# The pinned withholding notice; machine-readable and reproduced verbatim on every artefact.
SENSITIVITY_WITHHOLDING_NOTE: str = (
    "No variant result may be presented as evidence for or against any substantive claim of "
    "the study; variants quantify specification sensitivity, nothing else. The sensitivity "
    "results are explicitly withheld from any ranking, any profitability claim, any "
    "significance test (P20.significance_test_on_aggregate_permitted == False is untouched), "
    "any out-of-sample claim, any matched-sample claim and any incremental-skill claim. The "
    "only statements a variant result supports are the numeric delta (a fact), the "
    "interpretation label with the materiality tolerance pinned, and the disclosure of the "
    "sensitivity."
)

# The pinned variant-3 disclosure sentence (Step 3, Challenges clause); n_alt / n_primary are
# substituted at runtime, or the literal tokens are emitted when the figures are not computable.
VARIANT_3_DISCLOSURE_TEMPLATE: str = (
    "variant_dedup_negation_push_through, producing cohort size {n_alt} vs. primary "
    "{n_primary}, is reported only to quantify the consequence of collapsing the sign-flip "
    "augmentation; the primary freeze remains the paper-faithful cohort."
)

# The variant-3 epistemic status notice, always attached to the variant's note.
VARIANT_3_ADMISSIBILITY_NOTE: str = (
    "RECONSTRUCTION_OF_AN_INADMISSIBLE_VARIANT: push_negation_through collapses the sign-flip "
    "augmentation (P06 sign symmetry e -> -e; P13 sign_flip_semantics); the variant is reported "
    "only to quantify the consequence and is never an admissible specification; "
    "P06.deduplication_canonicalization.push_negation_through == False remains the correct, "
    "paper-faithful choice."
)

# Compile the absolute-path sanitiser used to strip ephemeral machine paths from failure text.
_PATH_SANITISER: "re.Pattern[str]" = re.compile(r"(?:[A-Za-z]:[\\/]|/)[^\s\"']+")


def _json_fallback(value: Any) -> Any:
    """
    Serialise non-core values for the canonical JSON encoder.

    Total fallback used as the ``default`` handler of the canonical
    JSON encoder. Numpy integer, floating and boolean scalars are
    promoted to their native Python counterparts; ``pd.NA`` and
    ``pd.NaT`` become ``None``; datetime, date and time values are
    rendered with ``isoformat``; every other value is rendered with
    ``str(value)``.

    Parameters
    ----------
    value : Any
        The value the JSON encoder could not serialise natively.

    Returns
    -------
    Any
        A JSON-serialisable rendering: ``int``, ``float``, ``bool``,
        ``None``, an ISO-8601 string or ``str(value)``.
    """
    # numpy integer scalars are promoted to native ints so the hash payload is platform-stable.
    if isinstance(value, np.integer):
        # Cast the numpy scalar to a Python int (exact for the cohort-scale values involved).
        return int(value)
    # numpy floating scalars are promoted to native floats (round-trip via repr in encoding).
    if isinstance(value, np.floating):
        # Cast the numpy scalar to a Python float.
        return float(value)
    # numpy booleans are promoted to native bools.
    if isinstance(value, np.bool_):
        # Cast the numpy bool to a Python bool.
        return bool(value)
    # pandas/numpy NaN and NA markers are rendered as null (they never enter hashed payloads,
    # but the fallback keeps the encoder total).
    if value is pd.NA or value is pd.NaT:
        # Emit JSON null for missing markers.
        return None
    # datetime-like values are rendered via isoformat (stable, unambiguous).
    if isinstance(value, (datetime.datetime, datetime.date, datetime.time)):
        # Render the instant in ISO-8601 form.
        return value.isoformat()
    # Anything else falls back to str(value); all remaining payload values are primitives.
    return str(value)

def _canonical_json_bytes(payload: Any) -> bytes:
    """
    Encode a payload as canonical JSON bytes.

    Keys are sorted, separators are the compact ``","`` / ``":"``
    pair, non-ASCII characters are kept as UTF-8 and unencodable
    values go through `_json_fallback`. The result is the sole input
    of every SHA-256 digest in this module, so identical payloads
    always yield identical bytes.

    Parameters
    ----------
    payload : Any
        The JSON-able payload (mappings, sequences, primitives).

    Returns
    -------
    bytes
        The canonical UTF-8 encoding.
    """
    # Serialise with deterministic ordering and separators; the default handler covers numpy/pandas stragglers.
    return json.dumps(
        payload,
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=False,
        default=_json_fallback,
    ).encode("utf-8")

def _sha256_bytes(raw: bytes) -> str:
    """
    Compute the lowercase hex SHA-256 digest of raw bytes.

    Parameters
    ----------
    raw : bytes
        The exact bytes to digest.

    Returns
    -------
    str
        ``64`` lowercase hexadecimal characters.
    """
    # Hash the bytes and render the digest as lowercase hex.
    return hashlib.sha256(raw).hexdigest()

def _sha256_json(payload: Any) -> str:
    """
    Hash the canonical-JSON encoding of a payload with SHA-256.

    Parameters
    ----------
    payload : Any
        The payload to hash (mappings, sequences, primitives).

    Returns
    -------
    str
        ``64`` lowercase hexadecimal characters of the canonical
        encoding.
    """
    # Canonicalise then digest.
    return _sha256_bytes(_canonical_json_bytes(payload))

def _sanitise_ephemeral_paths(text: Any) -> str:
    """
    Replace absolute filesystem paths in text with a fixed token.

    Every match of the module-level absolute-path pattern in
    ``str(text)`` is substituted by the literal string
    ``"<ephemeral-path>"`` so that identically caused failures
    produce identical text across runs and machines.

    Parameters
    ----------
    text : Any
        The raw text or object whose absolute paths are neutralised.

    Returns
    -------
    str
        The sanitised text.
    """
    # Coerce to text and substitute every absolute path with the pinned placeholder.
    return _PATH_SANITISER.sub("<ephemeral-path>", str(text))

def _patch_echo(patch: Mapping[str, Any]) -> str:
    """
    Render a patch mapping as canonical-JSON text.

    Parameters
    ----------
    patch : Mapping[str, Any]
        The already validated patch mapping.

    Returns
    -------
    str
        The canonical-JSON rendering with sorted keys, used as the
        audit echo of the applied patch.
    """
    # Reuse the canonical encoder (the echo is content-addressed by construction).
    return _canonical_json_bytes(patch).decode("utf-8")

def _finite_float_col(frame: pd.DataFrame, column: str) -> np.ndarray:
    """
    Extract a column as a float64 array, coercing failures to NaN.

    Parameters
    ----------
    frame : pd.DataFrame
        The source frame.
    column : str
        The column to extract; must be present.

    Returns
    -------
    np.ndarray
        A 1-D float64 array in row order; entries that cannot be
        parsed as numbers become ``NaN`` rather than raising.

    Raises
    ------
    ValueError
        If ``column`` is not among ``frame.columns``.
    """
    # The column must exist (boundary rigour with an informative error).
    if column not in frame.columns:
        # Refuse with the missing column name.
        raise ValueError(f"factor_aggregates is missing required column {column!r}")
    # Coerce with pandas (strings/objects/errors -> NaN), then cast to float64.
    return pd.to_numeric(frame[column], errors="coerce").to_numpy(dtype=np.float64, na_value=np.nan)

def _scorable_arm(aggregates: pd.DataFrame) -> Tuple[pd.DataFrame, str]:
    """
    Select the scorable rows of the factor aggregates frame.

    Uses the column ``state_primary`` when present, else ``state``,
    and keeps the rows whose value equals ``"scorable"``. When
    neither column exists the whole frame is returned as the arm.

    Parameters
    ----------
    aggregates : pd.DataFrame
        The factor aggregates frame.

    Returns
    -------
    Tuple[pd.DataFrame, str]
        The arm frame in original row order, and a label that is
        ``"scorable__via_state_primary"``, ``"scorable__via_state"``
        or ``"all_rows__no_state_column"``.
    """
    # Prefer the Task 10 Step 2 state column.
    state_column: Optional[str] = "state_primary" if "state_primary" in aggregates.columns else (
        "state" if "state" in aggregates.columns else None
    )
    # No state column: the whole frame is the arm, disclosed as such.
    if state_column is None:
        # Return the frame unchanged with the explicit label.
        return aggregates, "all_rows__no_state_column"
    # The arm is the subset labelled scorable (the 671/252 population semantics).
    arm: pd.DataFrame = aggregates[aggregates[state_column].astype("string") == "scorable"]
    # Return the arm with its column-derived label.
    return arm, f"scorable__via_{state_column}"

def _exact_zero_partition(values: np.ndarray) -> Dict[str, int]:
    """
    Partition a float array by exact sign against zero.

    A value is positive iff ``x > 0``, negative iff ``x < 0``,
    exact-zero iff it is finite and equal to ``0.0``, and undefined
    iff it is ``NaN``. The four buckets are mutually exclusive and
    exhaustive.

    Parameters
    ----------
    values : np.ndarray
        The possibly NaN-bearing float64 array.

    Returns
    -------
    Dict[str, int]
        Counts keyed ``"n_positive"``, ``"n_negative"``,
        ``"n_exact_zero"`` and ``"n_nan"``; the four sum to
        ``len(values)``.
    """
    # The NaN mask isolates undefined means (never counted as positive or negative).
    nan_mask: np.ndarray = np.isnan(values)
    # Positive means: strictly greater than zero.
    n_positive: int = int(np.count_nonzero(values > 0.0))
    # Negative means: strictly less than zero.
    n_negative: int = int(np.count_nonzero(values < 0.0))
    # Exact zeros: finite values equal to zero.
    n_exact_zero: int = int(np.count_nonzero((values == 0.0) & ~nan_mask))
    # NaN means: undefined.
    n_nan: int = int(np.count_nonzero(nan_mask))
    # Return the four counts (their sum equals the array length by construction).
    return {
        "n_positive": n_positive,
        "n_negative": n_negative,
        "n_exact_zero": n_exact_zero,
        "n_nan": n_nan,
    }

def _thresholded_partition(values: np.ndarray, absolute_tolerance: float) -> Dict[str, int]:
    """
    Partition a float array with an absolute near-zero band.

    A finite value is positive iff ``x >= absolute_tolerance``,
    negative iff ``x <= -absolute_tolerance``, and near-zero iff
    ``|x| < absolute_tolerance`` (exact zeros included). ``NaN``
    values are counted separately and join neither sign arm.

    Parameters
    ----------
    values : np.ndarray
        The possibly NaN-bearing float64 array.
    absolute_tolerance : float
        The non-negative, finite half-width of the near-zero band.

    Returns
    -------
    Dict[str, int]
        Counts keyed ``"n_positive"``, ``"n_negative"``,
        ``"n_near_zero"`` and ``"n_nan"``; the four sum to
        ``len(values)``.

    Raises
    ------
    TypeError
        If ``absolute_tolerance`` is a bool or not a real number.
    ValueError
        If ``absolute_tolerance`` is negative or non-finite.
    """
    # The tolerance must be a real number (bools refused).
    if isinstance(absolute_tolerance, bool) or not isinstance(absolute_tolerance, (int, float, np.number)):
        # Refuse with the observed type.
        raise TypeError(f"absolute_tolerance must be a real number, got {type(absolute_tolerance).__name__}")
    # Coerce to float.
    tolerance: float = float(absolute_tolerance)
    # The tolerance must be non-negative and finite.
    if not math.isfinite(tolerance) or tolerance < 0.0:
        # Refuse with the observed value.
        raise ValueError(f"absolute_tolerance must be a non-negative finite float, got {tolerance}")
    # The NaN mask isolates undefined means.
    nan_mask: np.ndarray = np.isnan(values)
    # The finite mask isolates defined means.
    finite_mask: np.ndarray = ~nan_mask
    # The absolute magnitude of every defined mean.
    magnitudes: np.ndarray = np.abs(values)
    # Near-zero band: |x| < tolerance (includes exact zeros).
    near_zero_mask: np.ndarray = finite_mask & (magnitudes < tolerance)
    # Positive arm: x >= tolerance (disjoint from the near-zero band and NaN).
    n_positive: int = int(np.count_nonzero(finite_mask & (values >= tolerance)))
    # Negative arm: x <= -tolerance (disjoint from the near-zero band and NaN).
    n_negative: int = int(np.count_nonzero(finite_mask & (values <= -tolerance)))
    # Near-zero count (the diagnostic bucket).
    n_near_zero: int = int(np.count_nonzero(near_zero_mask))
    # NaN count.
    n_nan: int = int(np.count_nonzero(nan_mask))
    # Return the four counts (their sum equals the array length by construction).
    return {
        "n_positive": n_positive,
        "n_negative": n_negative,
        "n_near_zero": n_near_zero,
        "n_nan": n_nan,
    }

def _finite_mean(values: np.ndarray) -> Optional[float]:
    """
    Average the finite entries of an array.

    Uses the arithmetic mean only; ``NaN`` entries are dropped,
    never imputed or replaced by zero.

    Parameters
    ----------
    values : np.ndarray
        The possibly NaN-bearing float64 array.

    Returns
    -------
    Optional[float]
        The arithmetic mean of the finite entries, or ``None`` when
        no finite entry exists.
    """
    # Isolate the finite entries.
    finite: np.ndarray = values[~np.isnan(values)]
    # No finite entries: no mean exists (never imputed).
    if finite.size == 0:
        # Disclose absence rather than fabricate.
        return None
    # Return the arithmetic mean of the finite entries.
    return float(finite.mean())

def _ledger_closure_battery(
    *,
    primary_triple: Tuple[int, int, int],
    reverse_triple: Optional[Tuple[int, int, int]],
    denominator: int,
    aggregates: Optional[pd.DataFrame],
    min_valid_dates: int,
    full_coverage_dates: int,
) -> Tuple[Tuple[Tuple[str, str, str], ...], bool]:
    """
    Evaluate the seven ledger and frame closures of a variant.

    Returns one ``(closure_id, status, detail)`` triple per closure
    with status in ``{"CLOSED", "OPEN", "DEFERRED"}``:

    * ``C2_LEDGER_PRIMARY_CONSERVATION``
      scorable + insufficient + unsupported equals ``denominator``.
    * ``C4_CROSS_CLOSURE_PRIMARY``
      ``denominator - unsupported`` equals scorable + insufficient.
    * ``C3_LEDGER_REVERSE_CONSERVATION`` and
      ``C7_CROSS_CLOSURE_REVERSE``
      the same two identities on the reversed-precedence triple;
      ``DEFERRED`` when ``reverse_triple`` is ``None``.
    * ``C5_SIGN_PARTITION_TOTALITY``
      the exact-zero partition counts of ``mean_rank_ic`` over the
      scorable arm sum to the arm size.
    * ``C1_SCORABLE_ARM_GATE``
      every finite ``n_valid_dates`` in the arm is at least
      ``min_valid_dates`` (inclusive).
    * ``C6_COVERAGE_PARTITION_TOTALITY``
      the full-coverage, partial and invalid ``n_valid_dates``
      counts sum to the arm size.

    ``C1``, ``C5`` and ``C6`` are ``DEFERRED`` when ``aggregates``
    is ``None`` or lacks ``expression_id`` / ``mean_rank_ic``;
    ``C1`` and ``C6`` are additionally ``DEFERRED`` when
    ``n_valid_dates`` is absent. A deferred closure never fails the
    variant.

    Parameters
    ----------
    primary_triple : Tuple[int, int, int]
        ``(scorable, insufficient, unsupported)`` under the primary
        ledger precedence.
    reverse_triple : Optional[Tuple[int, int, int]]
        The same triple under the reversed precedence, or ``None``.
    denominator : int
        The resolved positive denominator (total frozen records).
    aggregates : Optional[pd.DataFrame]
        The factor aggregates frame backing the frame-level closures.
    min_valid_dates : int
        Inclusive scorability gate applied to ``n_valid_dates``.
    full_coverage_dates : int
        The count defining full date coverage; partial coverage is
        ``0 < n < full_coverage_dates``.

    Returns
    -------
    Tuple[Tuple[Tuple[str, str, str], ...], bool]
        The closure triples and the verdict, ``True`` iff no
        evaluated closure is ``OPEN``.
    """
    # Accumulate the closure triples in pinned order.
    results: List[Tuple[str, str, str]] = []
    # Unpack the primary triple.
    scorable, insufficient, unsupported = primary_triple
    # C2: the primary ledger conserves to the denominator.
    results.append(
        (
            "C2_LEDGER_PRIMARY_CONSERVATION",
            "CLOSED" if scorable + insufficient + unsupported == denominator else "OPEN",
            f"scorable({scorable}) + insufficient({insufficient}) + unsupported({unsupported}) = {scorable + insufficient + unsupported} vs denominator_total = {denominator}",
        )
    )
    # C4: the cross-closure on the primary precedence (P04.ledger_supported_partition).
    results.append(
        (
            "C4_CROSS_CLOSURE_PRIMARY",
            "CLOSED" if denominator - unsupported == scorable + insufficient else "OPEN",
            f"denominator_total({denominator}) - unsupported({unsupported}) = {denominator - unsupported} vs scorable + insufficient = {scorable + insufficient}",
        )
    )
    # C3 and C7: the reversed-precedence identities (DEFERRED when unavailable).
    if reverse_triple is not None:
        # Unpack the reverse triple.
        r_scorable, r_insufficient, r_unsupported = reverse_triple
        # C3: the reverse ledger conserves to the same denominator.
        results.append(
            (
                "C3_LEDGER_REVERSE_CONSERVATION",
                "CLOSED" if r_scorable + r_insufficient + r_unsupported == denominator else "OPEN",
                f"scorable({r_scorable}) + insufficient({r_insufficient}) + unsupported({r_unsupported}) = {r_scorable + r_insufficient + r_unsupported} vs denominator_total = {denominator}",
            )
        )
        # C7: the cross-closure on the reversed precedence.
        results.append(
            (
                "C7_CROSS_CLOSURE_REVERSE",
                "CLOSED" if denominator - r_unsupported == r_scorable + r_insufficient else "OPEN",
                f"denominator_total({denominator}) - unsupported({r_unsupported}) = {denominator - r_unsupported} vs scorable + insufficient = {r_scorable + r_insufficient}",
            )
        )
    else:
        # C3 deferred with the reason stated (never silently passed).
        results.append(("C3_LEDGER_REVERSE_CONSERVATION", "DEFERRED", "reverse-precedence triple unavailable on the primary"))
        # C7 deferred with the reason stated.
        results.append(("C7_CROSS_CLOSURE_REVERSE", "DEFERRED", "reverse-precedence triple unavailable on the primary"))
    # Frame-level closures: require the frame and its column contract.
    if isinstance(aggregates, pd.DataFrame) and {"expression_id", "mean_rank_ic"} <= set(aggregates.columns):
        # Select the scorable arm (or the disclosed all-rows fallback).
        arm, arm_label = _scorable_arm(aggregates)
        # Extract the mean rank-IC column as a NaN-bearing float array.
        means: np.ndarray = _finite_float_col(arm, "mean_rank_ic")
        # C5: the sign partition totality (P42.sign_partition).
        exact: Dict[str, int] = _exact_zero_partition(means)
        # The four buckets must total the arm size.
        partition_sum: int = exact["n_positive"] + exact["n_negative"] + exact["n_exact_zero"] + exact["n_nan"]
        # Record the closure with the arithmetic spelled out.
        results.append(
            (
                "C5_SIGN_PARTITION_TOTALITY",
                "CLOSED" if partition_sum == len(arm) else "OPEN",
                f"{exact['n_positive']} + {exact['n_negative']} + {exact['n_exact_zero']} + {exact['n_nan']} = {partition_sum} vs {arm_label} arm size {len(arm)}",
            )
        )
        # C1 and C6 require the n_valid_dates column.
        if "n_valid_dates" in arm.columns:
            # Coerce the counts to numeric (unparseable -> NaN).
            nvd: pd.Series = pd.to_numeric(arm["n_valid_dates"], errors="coerce")
            # The scorable-arm gate: every finite count >= the inclusive P03 gate.
            gate_ok: bool = bool((nvd.dropna() >= min_valid_dates).all())
            # Record the C1 closure with the gate spelled out.
            results.append(
                (
                    "C1_SCORABLE_ARM_GATE",
                    "CLOSED" if gate_ok else "OPEN",
                    f"all finite n_valid_dates >= {min_valid_dates} (inclusive, P03) over {arm_label}",
                )
            )
            # The coverage partition (P42.coverage_partition).
            n_full: int = int((nvd == full_coverage_dates).sum())
            # Partial coverage: 0 < n < full.
            n_partial: int = int(((nvd > 0) & (nvd < full_coverage_dates)).sum())
            # Invalid counts: NaN or < 1.
            n_invalid: int = int((nvd.isna() | (nvd < 1)).sum())
            # The three buckets must total the arm size.
            coverage_sum: int = n_full + n_partial + n_invalid
            # Record the C6 closure with the arithmetic spelled out.
            results.append(
                (
                    "C6_COVERAGE_PARTITION_TOTALITY",
                    "CLOSED" if coverage_sum == len(arm) else "OPEN",
                    f"n_full({n_full}) + n_partial({n_partial}) + n_invalid({n_invalid}) = {coverage_sum} vs {arm_label} arm size {len(arm)}",
                )
            )
        else:
            # C1 deferred (never silently passed).
            results.append(("C1_SCORABLE_ARM_GATE", "DEFERRED", "n_valid_dates column absent from factor_aggregates"))
            # C6 deferred.
            results.append(("C6_COVERAGE_PARTITION_TOTALITY", "DEFERRED", "n_valid_dates column absent from factor_aggregates"))
    else:
        # C5 deferred (frame unavailable or contract breach).
        results.append(("C5_SIGN_PARTITION_TOTALITY", "DEFERRED", "factor_aggregates absent or missing required columns"))
        # C1 deferred.
        results.append(("C1_SCORABLE_ARM_GATE", "DEFERRED", "factor_aggregates absent or missing required columns"))
        # C6 deferred.
        results.append(("C6_COVERAGE_PARTITION_TOTALITY", "DEFERRED", "factor_aggregates absent or missing required columns"))
    # No evaluated closure may be OPEN.
    all_ok: bool = all(status in ("CLOSED", "DEFERRED") for _, status, _ in results)
    # Return the frozen closure triples and the verdict.
    return tuple(results), all_ok

def _closures_all_evaluated_ok(closures: Tuple[Tuple[str, str, str], ...]) -> bool:
    """
    Check that no evaluated closure is open.

    Parameters
    ----------
    closures : Tuple[Tuple[str, str, str], ...]
        The ``(closure_id, status, detail)`` triples.

    Returns
    -------
    bool
        ``True`` iff every status is ``"CLOSED"`` or ``"DEFERRED"``.
    """
    # No closure may be OPEN.
    return all(status in ("CLOSED", "DEFERRED") for _, status, _ in closures)

def _load_dependency_module(stem: str) -> Optional[Any]:
    """
    Import a sibling task module by stem, without raising.

    Parameters
    ----------
    stem : str
        The module stem under the ``task_responses`` package, e.g.
        ``"task_14"``.

    Returns
    -------
    Optional[Any]
        The imported module (cached by the interpreter), or ``None``
        when the import raised; the failure is logged as a warning.
    """
    # Import lazily; the interpreter caches the module for the process lifetime.
    try:
        # Import the task_responses package member by name.
        return importlib.import_module(f"task_responses.{stem}")
    except Exception as exc:
        # Log the failure with class and message (never re-raised here).
        LOGGER.warning("task_15: could not import task_responses.%s (%s: %s)", stem, type(exc).__name__, exc)
        # Disclose the import failure to the caller.
        return None

def _freeze_timestamp_hint(primary: Any) -> Optional[datetime.datetime]:
    """
    Recover a timezone-aware freeze instant from the primary.

    Probes, in order, ``primary.stages_output["freeze"]
    ["freeze_timestamp"]``, ``primary.frozen_cohort.meta
    ["freeze_timestamp"]``, then the ``cohort_timestamp`` attribute
    of the frozen cohort and of the primary. Only timezone-aware
    datetimes (with a non-``None`` UTC offset) are accepted.

    Parameters
    ----------
    primary : Any
        The primary result object or duck-type.

    Returns
    -------
    Optional[datetime.datetime]
        The recovered instant, or ``None`` when no aware timestamp
        exists; a naive or fabricated instant is never returned.
    """
    # Probe the stages_output freeze payload first.
    stages_output: Any = getattr(primary, "stages_output", None)
    # The payload must be a mapping.
    if isinstance(stages_output, Mapping):
        # Fetch the freeze stage payload.
        freeze_payload: Any = stages_output.get("freeze")
        # The freeze payload must be a mapping.
        if isinstance(freeze_payload, Mapping):
            # Fetch the freeze timestamp member.
            candidate: Any = freeze_payload.get("freeze_timestamp")
            # Return it when it is a timezone-aware datetime.
            if isinstance(candidate, datetime.datetime) and candidate.tzinfo is not None and candidate.utcoffset() is not None:
                # Return the recovered instant.
                return candidate
    # Probe the frozen cohort object next.
    frozen: Any = getattr(primary, "frozen_cohort", None)
    # The cohort may carry meta or a direct timestamp.
    meta: Any = getattr(frozen, "meta", None)
    # meta may be a mapping.
    if isinstance(meta, Mapping):
        # Fetch the meta timestamp member.
        candidate = meta.get("freeze_timestamp")
        # Return it when timezone-aware.
        if isinstance(candidate, datetime.datetime) and candidate.tzinfo is not None and candidate.utcoffset() is not None:
            # Return the recovered instant.
            return candidate
    # Probe direct timestamp attributes on the cohort and then the primary.
    for owner in (frozen, primary):
        # Skip absent owners.
        if owner is None:
            # Move to the next probe.
            continue
        # Fetch the direct timestamp attribute.
        candidate = getattr(owner, "cohort_timestamp", None)
        # Return it when timezone-aware.
        if isinstance(candidate, datetime.datetime) and candidate.tzinfo is not None and candidate.utcoffset() is not None:
            # Return the recovered instant.
            return candidate
    # Nothing recoverable: disclose absence (never fabricate an instant).
    return None

def _primary_n_after_dedup(primary: Any) -> Optional[int]:
    """
    Recover the primary post-deduplication cohort size.

    Probes, in order, ``primary.stages_output["freeze"]
    ["n_after_dedup"]``, ``primary.frozen_cohort.meta
    ["cohort_size"]`` and ``len(primary.frozen_cohort.rows)``.
    Numeric candidates are accepted only when integral, non-negative
    and not boolean.

    Parameters
    ----------
    primary : Any
        The primary result object or duck-type.

    Returns
    -------
    Optional[int]
        The recovered cohort size, or ``None`` when none is
        retained; a count is never fabricated.
    """
    # Probe the stages_output freeze payload first.
    stages_output: Any = getattr(primary, "stages_output", None)
    # The payload must be a mapping.
    if isinstance(stages_output, Mapping):
        # Fetch the freeze stage payload.
        freeze_payload: Any = stages_output.get("freeze")
        # The freeze payload must be a mapping.
        if isinstance(freeze_payload, Mapping):
            # Fetch the member.
            member: Any = freeze_payload.get("n_after_dedup")
            # Accept integral members only.
            if isinstance(member, (int, float, np.number)) and not isinstance(member, bool):
                # Coerce when integral and non-negative.
                if float(member).is_integer() and float(member) >= 0:
                    # Return the recovered count.
                    return int(member)
    # Probe the frozen cohort object next.
    frozen: Any = getattr(primary, "frozen_cohort", None)
    # The cohort may carry meta.
    meta: Any = getattr(frozen, "meta", None)
    # meta may be a mapping.
    if isinstance(meta, Mapping):
        # Fetch the meta cohort size member.
        member = meta.get("cohort_size")
        # Accept integral members only.
        if isinstance(member, (int, float, np.number)) and not isinstance(member, bool):
            # Coerce when integral and non-negative.
            if float(member).is_integer() and float(member) >= 0:
                # Return the recovered count.
                return int(member)
    # Probe the cohort rows frame as a last resort.
    rows: Any = getattr(frozen, "rows", None)
    # The rows must be a DataFrame.
    if isinstance(rows, pd.DataFrame):
        # Return the row count.
        return int(len(rows))
    # Nothing recoverable: disclose absence (never fabricate a count).
    return None

def _unavailable_variant_result(
    *,
    variant_name: str,
    interpretation: str,
    halt_state: str,
    reason: str,
    ledger_partition: str,
    patch_echo: str,
    note: str,
) -> VariantResult:
    """
    Build the honest unavailable or failed variant result.

    Parameters
    ----------
    variant_name : str
        The pinned variant name.
    interpretation : str
        The non-empty interpretation string.
    halt_state : str
        The unavailability or failure token recorded on the result.
    reason : str
        The human-readable cause, path-sanitised by the caller when
        it comes from an exception message.
    ledger_partition : str
        The non-empty partition disclosure string.
    patch_echo : str
        The canonical-JSON echo of the applied patch.
    note : str
        The provenance note.

    Returns
    -------
    VariantResult
        A frozen result with
        ``status = VARIANT_STATUS_UNAVAILABLE``, empty
        ``quantities``, empty ``isolation`` and empty ``closures``,
        with ``patch_echo``, ``halt_state`` and ``ledger_partition``
        taken from the arguments, and a note of the form
        ``"<reason> | <note>"``.
    """
    # Assemble the honest result: zero quantities, full provenance.
    return VariantResult(
        variant_name=variant_name,
        interpretation=interpretation,
        quantities=(),
        isolation=(),
        patch_echo=patch_echo,
        status=VARIANT_STATUS_UNAVAILABLE,
        halt_state=halt_state,
        ledger_partition=ledger_partition,
        note=f"{reason} | {note}",
        closures=(),
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: The shared variant scaffolding (unavailability, closures, patch)
# -------------------------------------------------------------------------------------------------------------------------------

def _variant_closures_and_verdict(
    primary: Any,
    aggregates: Optional[pd.DataFrame],
    primary_triple: Tuple[int, int, int],
    reverse_triple: Optional[Tuple[int, int, int]],
    denominator: int,
) -> Tuple[Tuple[Tuple[str, str, str], ...], bool]:
    """
    Evaluate a variant's closures with the primary's pins.

    Reads ``min_valid_dates_for_scorability`` and
    ``full_coverage_target_dates`` from the configuration attached
    to ``primary`` (falling back to the module defaults) and
    delegates to `_ledger_closure_battery`.

    Parameters
    ----------
    primary : Any
        The primary result carrying the attached configuration.
    aggregates : Optional[pd.DataFrame]
        The factor aggregates frame for the frame-level closures.
    primary_triple : Tuple[int, int, int]
        The primary ledger ``(scorable, insufficient, unsupported)``.
    reverse_triple : Optional[Tuple[int, int, int]]
        The reversed-precedence triple, when available.
    denominator : int
        The resolved denominator.

    Returns
    -------
    Tuple[Tuple[Tuple[str, str, str], ...], bool]
        The seven closure triples and the all-evaluated-ok verdict.
    """
    # Resolve the attached configuration (the P03 pins are read, never mutated).
    config: Optional[Mapping[str, Any]] = _primary_config(primary)
    # Resolve the pinned P03 knobs with their documented paper defaults.
    min_valid_dates: int = _p03_int_pin(config, "min_valid_dates_for_scorability", MIN_VALID_DATES_PINNED)
    # Resolve the full-coverage cardinality likewise.
    full_coverage_dates: int = _p03_int_pin(config, "full_coverage_target_dates", FULL_COVERAGE_TARGET_DATES_PINNED)
    # Delegate to the shared battery.
    return _ledger_closure_battery(
        primary_triple=primary_triple,
        reverse_triple=reverse_triple,
        denominator=denominator,
        aggregates=aggregates,
        min_valid_dates=min_valid_dates,
        full_coverage_dates=full_coverage_dates,
    )

def _primary_triple_or_raise(primary: Any) -> Tuple[int, int, int]:
    """
    Return the primary ledger triple, refusing an absent ledger.

    Parameters
    ----------
    primary : Any
        The primary result object or duck-type.

    Returns
    -------
    Tuple[int, int, int]
        ``(scorable, insufficient, unsupported)`` normalised from the
        ``ledger_primary`` payload.

    Raises
    ------
    TypeError
        If ``primary`` carries no ``ledger_primary`` payload.
    """
    # Fetch the payload; absence is a contract breach, not an empty ledger.
    payload: Any = getattr(primary, "ledger_primary", None)
    # The payload must exist.
    if payload is None:
        # Refuse with the observed attribute state.
        raise TypeError("primary carries no ledger_primary payload")
    # Normalise (positional tuple or mapping) and return.
    return _normalise_ledger_triple(payload, label="ledger_primary")

def _reverse_triple_or_none(primary: Any) -> Optional[Tuple[int, int, int]]:
    """
    Return the reversed-precedence ledger triple when retained.

    Parameters
    ----------
    primary : Any
        The primary result object or duck-type.

    Returns
    -------
    Optional[Tuple[int, int, int]]
        ``(scorable, insufficient, unsupported)`` normalised from the
        ``ledger_reverse`` payload, or ``None`` when it is absent, so
        that the reverse closures are deferred rather than passed.
    """
    # Fetch the payload (the primary materialises both partitions, Task 10 Step 2).
    payload: Any = getattr(primary, "ledger_reverse", None)
    # Absence yields None (closures C3/C7 are then marked DEFERRED, never passed).
    if payload is None:
        # Disclose absence to the battery.
        return None
    # Normalise (positional tuple or mapping) and return.
    return _normalise_ledger_triple(payload, label="ledger_reverse")

def _pin_echo_rows(
    *,
    config: Optional[Mapping[str, Any]],
    resolved_denominator: Optional[int],
    means: Optional[np.ndarray],
    arm_label: str,
) -> Tuple[QuantityDelta, ...]:
    """
    Emit the disclosed specification pins and diagnostics of a run.

    Rows are produced in fixed order, all with class
    ``"FLOAT_CONFIG_PIN"`` except the last:

    * ``diag_p03_near_zero_absolute_threshold`` -- the resolved pin
      against the module default.
    * ``diag_p03_min_valid_dates_for_scorability`` -- the resolved
      inclusive gate against the module default.
    * ``diag_p03_full_coverage_target_dates`` -- the resolved
      coverage cardinality against the module default.
    * ``diag_p04_denominator_total_config`` -- the configured
      paper-scale denominator against the resolved run denominator
      (the resolved count when no config value exists).
    * ``diag_p04_sign_partition_positive`` and
      ``diag_p04_sign_partition_negative`` -- the configured
      comparison split echoed with zero delta; when undisclosed they
      fall back to the recomputed positive / negative counts of
      ``means``.
    * ``diag_mean_abs_rank_ic`` (class
      ``"FLOAT_AGGREGATES_DIAGNOSTIC"``, only when ``means`` is
      non-empty) -- the ``nansum`` of ``|mean_rank_ic|`` over the
      population, echoed with zero delta.

    Missing configuration or a ``None`` denominator resolves to
    ``-1.0`` where no other value is available; no pin is hardcoded.

    Parameters
    ----------
    config : Optional[Mapping[str, Any]]
        The configuration attached to the primary run.
    resolved_denominator : Optional[int]
        The run-resolved denominator, the primary reference of the
        P04 denominator row.
    means : Optional[np.ndarray]
        The finite mean rank-IC array of the recomputed population;
        ``None`` omits the diagnostic row.
    arm_label : str
        The population label quoted in the diagnostic
        interpretation.

    Returns
    -------
    Tuple[QuantityDelta, ...]
        The disclosure rows in the order listed above; all values are
        floats.
    """
    # Accumulate the pinned pin-echo rows.
    rows: List[QuantityDelta] = []
    # P03 near-zero diagnostic: variant value = the resolved pin (default 1e-12).
    near_zero_pin: float = _p03_float_pin(config, "near_zero_absolute_threshold", SIGN_CLASSIFICATION_PRIMARY_THRESHOLD_DEFAULT)
    # The pin echoes the default when the path is absent (documented paper pin).
    rows.append(
        _quantity(
            "diag_p03_near_zero_absolute_threshold",
            float(SIGN_CLASSIFICATION_PRIMARY_THRESHOLD_DEFAULT),
            float(near_zero_pin),
            "FLOAT_CONFIG_PIN",
            "P03.near_zero_absolute_threshold under which the suite ran (disclosed pin); comparison reference is the pinned default",
        )
    )
    # P03 scorability gate: variant value = the resolved pin (default 20; inclusive).
    min_valid_pin: int = _p03_int_pin(config, "min_valid_dates_for_scorability", MIN_VALID_DATES_PINNED)
    # Echo the scorability gate pin.
    rows.append(
        _quantity(
            "diag_p03_min_valid_dates_for_scorability",
            float(MIN_VALID_DATES_PINNED),
            float(min_valid_pin),
            "FLOAT_CONFIG_PIN",
            "P03.min_valid_dates_for_scorability under which the suite ran (disclosed pin, inclusive)",
        )
    )
    # P03 full-coverage cardinality: variant value = the resolved pin (default 480).
    full_coverage_pin: int = _p03_int_pin(config, "full_coverage_target_dates", FULL_COVERAGE_TARGET_DATES_PINNED)
    # Echo the full-coverage cardinality pin.
    rows.append(
        _quantity(
            "diag_p03_full_coverage_target_dates",
            float(FULL_COVERAGE_TARGET_DATES_PINNED),
            float(full_coverage_pin),
            "FLOAT_CONFIG_PIN",
            "P03.full_coverage_target_dates under which the suite ran (disclosed pin)",
        )
    )
    # P04 paper-scale denominator vs this run's resolved count (a genuine sensitivity disclosure).
    configured_denominator: Optional[int] = _p04_denominator_config(config)
    # The variant value is the configured pin when present, else the resolved count.
    denominator_variant: float = float(configured_denominator) if configured_denominator is not None else float(
        resolved_denominator if resolved_denominator is not None else -1.0
    )
    # The primary reference is this run's resolved count (never the paper's pin asserted as observed).
    denominator_reference: float = float(resolved_denominator if resolved_denominator is not None else denominator_variant)
    # Echo the configuration-vs-resolved denominator contrast (disclosed discrepancy, if any).
    rows.append(
        _quantity(
            "diag_p04_denominator_total_config",
            denominator_reference,
            denominator_variant,
            "FLOAT_CONFIG_PIN",
            "P04.denominator_total (config, paper-scale) vs the resolved run denominator; a disclosed specification fact, never an observation of the cohort",
        )
    )
    # P04 paper comparison split (never asserted): disclosed next to the recomputed partition.
    # Resolve (positive, negative) comparison counts (None each when undisclosed).
    pos_comparison, neg_comparison = _p04_sign_partition_counts(config)
    # The variant value echoes the configured split (the comparison value, disclosed, never asserted as observed);
    # when the config path is undisclosed it falls back to the recomputed exact-zero test count (disclosed).
    positive_value: float = float(pos_comparison) if pos_comparison is not None else float(
        np.count_nonzero(means > 0.0) if means is not None else 0
    )
    # The negative comparison count likewise (undisclosed falls back to the recomputed count).
    negative_value: float = float(neg_comparison) if neg_comparison is not None else float(
        np.count_nonzero(means < 0.0) if means is not None else 0
    )
    # Echo the positive comparison count (disclosed, never asserted).
    rows.append(
        _quantity(
            "diag_p04_sign_partition_positive",
            positive_value,
            positive_value,
            "FLOAT_CONFIG_PIN",
            "P04.sign_partition_counts.positive (paper comparison value; never asserted as observed; disclosed next to the recomputed partition)",
        )
    )
    # Echo the negative comparison count (disclosed, never asserted).
    rows.append(
        _quantity(
            "diag_p04_sign_partition_negative",
            negative_value,
            negative_value,
            "FLOAT_CONFIG_PIN",
            "P04.sign_partition_counts.negative (paper comparison value; never asserted as observed; disclosed next to the recomputed partition)",
        )
    )
    # The data-derived diagnostic over the recomputed population (float; moves with the data).
    if means is not None and len(means) > 0:
        # The absolute-sum of finite means over the arm (a pure functional of the table).
        abs_sum: float = float(np.nansum(np.abs(means)))
        # Echo the diagnostic (population disclosed; never a claim).
        rows.append(
            _quantity(
                "diag_mean_abs_rank_ic",
                abs_sum,
                abs_sum,
                "FLOAT_AGGREGATES_DIAGNOSTIC",
                f"diag_mean_abs_rank_ic over {arm_label} (sum of |mean_rank_ic|; a disclosure of the recomputed table, not a result)",
            )
        )
    # Return the pinned pin-echo rows in stable order.
    return tuple(rows)

def _quantity(
    quantity: str,
    primary_value: float,
    variant_value: float,
    tolerance_class: str,
    interpretation: str,
) -> QuantityDelta:
    """
    Build one quantity record with an exact delta.

    Parameters
    ----------
    quantity : str
        The pinned quantity identifier.
    primary_value : float
        The quantity under the primary specification.
    variant_value : float
        The quantity under the variant.
    tolerance_class : str
        The comparability class; must be non-empty.
    interpretation : str
        The interpretation; must be non-empty.

    Returns
    -------
    QuantityDelta
        The frozen record with
        ``delta = variant_value - primary_value`` computed in
        binary64.
    """
    # Compute the delta exactly (binary64 subtraction; both operands disclosed).
    delta: float = float(variant_value) - float(primary_value)
    # Assemble the record (the dataclass validates finiteness and non-empty strings).
    return QuantityDelta(
        quantity=quantity,
        primary_value=float(primary_value),
        variant_value=float(variant_value),
        delta=delta,
        tolerance_class=tolerance_class,
        interpretation=interpretation,
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1, Variant 1: variant_state_precedence_reversed
# -------------------------------------------------------------------------------------------------------------------------------

def variant_state_precedence_reversed(primary: Any, patch: Mapping[str, Any]) -> VariantResult:
    """
    Re-run the aggregation under the reversed state precedence.

    Move exactly one argument of one downstream callable -- the
    ``precedence`` handed to ``aggregate_and_ledger`` -- to
    ``("scorable", "insufficient", "unsupported")``, reusing the
    primary's intermediates by identity (the same ``daily_scores_v2``
    or ``score_register_v2`` and the same ``frozen_cohort``).

    Process:

    1. Validate ``primary``, deep-copy ``patch`` and admit only the
       key ``state_precedence``.
    2. Resolve the baseline triple from ``ledger_primary`` and the
       optional materialised reverse triple from ``ledger_reverse``.
    3. If a ``config`` mapping, a v2 score register and a
       ``frozen_cohort`` are all present, import ``task_14`` lazily
       and re-invoke ``aggregate_and_ledger`` on a deep copy of the
       config with the reversed precedence; a halt or exception is
       sanitised and recorded, then degrades to step 4.
    4. Otherwise reuse ``ledger_reverse`` (disclosed as reuse). If
       neither path yields a reverse triple, return an unavailable
       result with no quantity rows.
    5. Resolve the denominator and evaluate the seven closures;
       ``halt_state`` is ``HALT_CLOSED`` only when all close.
    6. Emit one ``QuantityDelta`` for ``n_scorable``,
       ``n_insufficient`` and ``n_unsupported`` (each
       ``EXACT_INTEGER``, delta = reverse - primary), then the
       pin-echo rows.

    Both triples are disclosed per P04.state_precedence_consequence;
    the result is ``[R]`` specification sensitivity, not a paper
    finding.

    Parameters
    ----------
    primary : Any
        The primary result (``StudyResult`` or the fixture duck-type).
        Requires ``ledger_primary`` (positional triple or mapping);
        optionally ``ledger_reverse``, ``config``,
        ``daily_scores_v2``/``score_register_v2``, ``frozen_cohort``,
        ``factor_aggregates`` and ``stages_output``.
    patch : Mapping[str, Any]
        The variant patch; the only admissible key is
        ``state_precedence``, defaulting to the reversed triple.

    Returns
    -------
    VariantResult
        Frozen result whose ``quantities`` hold the three triple
        deltas, plus the ``diag_p03_*`` and ``diag_p04_*`` rows and,
        when arm means are available, a ``diag_mean_abs_rank_ic``
        row; ``isolation`` records which intermediates were present,
        ``note`` records the branch taken and both triples,
        ``closures`` holds the seven verdicts.

    Raises
    ------
    TypeError
        If ``primary`` does not look like a primary result or
        ``patch`` is not a ``Mapping``.
    ValueError
        If ``patch`` carries unexpected keys, if ``state_precedence``
        is not a permutation of the three states, or if it is a
        permutation other than the pinned reversed triple.

    Notes
    -----
    The reuse branch exists because a bare ``StudyResult`` retains no
    run-scoped support bundle: the reverse-precedence ledger is
    materialised by the primary run (Task 10, Step 2).
    """
    # Boundary: the primary must quack like a primary result.
    _validate_primary_like(primary, caller="variant_state_precedence_reversed")
    # Boundary: the patch must be a mapping (deep-copied; the caller's mapping is never aliased).
    patch_d: Dict[str, Any] = _validate_patch_mapping(patch, caller="variant_state_precedence_reversed")
    # Boundary: the patch admits exactly the pinned key.
    _ensure_patch_keys(patch_d, allowed={"state_precedence"}, caller="variant_state_precedence_reversed")
    # Resolve the precedence (default pinned to the reversed triple).
    precedence_in: Any = patch_d.get("state_precedence", list(REVERSED_PRECEDENCE_TUPLE))
    # Validate the vocabulary, then enforce the pinned reversal exactly.
    precedence: Tuple[str, str, str] = _validate_precedence(precedence_in, caller="variant_state_precedence_reversed")
    # The variant varies exactly one declared parameter against the primary: the
    # reversed order is the only admissible value (anything else is a different variant).
    if precedence != REVERSED_PRECEDENCE_TUPLE:
        # Refuse any other permutation with the observed one.
        raise ValueError(
            "variant_state_precedence_reversed: patch['state_precedence'] must equal "
            f"{list(REVERSED_PRECEDENCE_TUPLE)} (P04.state_precedence reversed), got {list(precedence)}"
        )
    # Resolve the primary triple (a contract breach raises, and the orchestrator records HALTED).
    primary_triple: Tuple[int, int, int] = _primary_triple_or_raise(primary)
    # Resolve the materialised reverse triple (optional; DEFERRED closures when absent).
    primary_reverse: Optional[Tuple[int, int, int]] = _reverse_triple_or_none(primary)
    # Resolve the attached configuration (None on a bare StudyResult / the fixture).
    config: Optional[Mapping[str, Any]] = _primary_config(primary)
    # Resolve the v2 register for the re-invocation branch (disclosure name only; never re-scored).
    daily_scores: Any = getattr(primary, "daily_scores_v2", None)
    # Fall back to the score_register_v2 attribute name (the real StudyResult's spelling).
    if daily_scores is None:
        # Probe the real attribute name.
        daily_scores = getattr(primary, "score_register_v2", None)
    # Resolve the frozen cohort for the re-invocation branch.
    frozen_cohort: Any = getattr(primary, "frozen_cohort", None)
    # Isolation witnesses: both intermediates are reused by identity (never re-built).
    isolation: List[Tuple[str, bool]] = [
        ("daily_scores_v2", daily_scores is not None),
        ("frozen_cohort", frozen_cohort is not None),
    ]
    # Attempt the genuine re-invocation only when the attached configuration exists.
    reverse_triple: Optional[Tuple[int, int, int]] = None
    # Track which branch produced the reverse triple (disclosed in the note).
    source: str = "primary_materialised_reuse"
    # Track any re-invocation halt reason (sanitised before it enters any artefact).
    reinvocation_note: str = ""
    # The re-invocation branch requires config, register and cohort all present.
    if (
        isinstance(config, Mapping)
        and isinstance(daily_scores, pd.DataFrame)
        and frozen_cohort is not None
    ):
        # Load the task_14 registry lazily (never re-implemented; never bound at module scope).
        task_14 = _load_dependency_module("task_14")
        # Proceed only when the import succeeded.
        if task_14 is not None:
            # Deep-copy the attached configuration so the caller's mapping is never mutated.
            reinvocation_config: Dict[str, Any] = copy.deepcopy(dict(config))
            # Guard the re-invocation: any failure degrades to the documented reuse branch.
            try:
                # Re-invoke the same function object with exactly one argument changed
                # (precedence); the primary's daily_scores_v2 and frozen_cohort are reused.
                ledger, aggregates_re, halt_reason = task_14.aggregate_and_ledger(
                    daily_scores,
                    frozen_cohort,
                    reinvocation_config,
                    tuple(precedence),
                )
                # A non-None ledger with no halt reason is the recomputed reverse triple.
                if halt_reason is None and ledger is not None:
                    # Normalise the positional triple and use it as the reverse triple.
                    reverse_triple = _normalise_ledger_triple(ledger, label="aggregate_and_ledger(reverse)")
                    # Record the branch taken.
                    source = "aggregate_and_ledger_reinvocation"
                    # Record the isolation witness for the rebuilt aggregates.
                    isolation.append(("aggregate_and_ledger_reinvoked", True))
                # A halt degrades to the reuse branch with the reason disclosed.
                else:
                    # Sanitise the halt reason before it enters any artefact.
                    reinvocation_note = _sanitise_ephemeral_paths(halt_reason)
            # Any exception degrades to the reuse branch with the class-qualified reason.
            except Exception as exc:
                # Sanitise the exception text before it enters any artefact.
                reinvocation_note = _sanitise_ephemeral_paths(f"{type(exc).__name__}: {exc}")
    # Reuse branch: consume the primary's materialised reverse triple (the pinned Challenges clause).
    if reverse_triple is None:
        # The reuse branch requires the materialised reverse triple.
        if primary_reverse is not None:
            # Consume it (reuse, disclosed, never recomputed).
            reverse_triple = primary_reverse
        else:
            # No reverse triple anywhere: disclose, emit zero rows (never fabricate).
            return _unavailable_variant_result(
                variant_name="variant_state_precedence_reversed",
                interpretation="reversed-precedence ledger unavailable; nothing is estimated",
                halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
                reason="ledger_reverse is absent from the primary and the aggregate_and_ledger re-invocation was unavailable",
                ledger_partition="NOT_EVALUATED__ledger_reverse_absent",
                patch_echo=_patch_echo(patch_d),
                note="the reverse-precedence triple is materialised by the primary run (Task 10 Step 2); when absent there is nothing to compare against",
            )
    # Resolve the aggregates frame for the frame-level closures (best effort).
    aggregates: Optional[pd.DataFrame] = _safe_factor_aggregates(primary)
    # Resolve the ledger payload for the denominator hint.
    ledger_payload: Any = getattr(primary, "ledger_primary", None)
    # Resolve the denominator in the pinned order (never a hardcoded literal).
    denominator: int = _denominator_hint(
        primary=primary,
        aggregates=aggregates,
        ledger_payload=ledger_payload,
        config=config,
    )
    # Evaluate the seven closures (primary triple, reverse triple, frame battery).
    closures, all_evaluated_ok = _variant_closures_and_verdict(
        primary=primary,
        aggregates=aggregates,
        primary_triple=primary_triple,
        reverse_triple=reverse_triple,
        denominator=denominator,
    )
    # The moved quantities: the three components of the triple, each with its delta.
    quantity_names: Tuple[str, str, str] = ("n_scorable", "n_insufficient", "n_unsupported")
    # Build the three quantity records in pinned order.
    quantities: Tuple[QuantityDelta, ...] = tuple(
        _quantity(
            name,
            float(primary_triple[index]),
            float(reverse_triple[index]),
            "EXACT_INTEGER",
            "reversed-precedence ledger vs primary triple; P04.state_precedence_consequence; [R] specification sensitivity; both triples disclosed",
        )
        for index, name in enumerate(quantity_names)
    )
    # Resolve the arm means for the data-derived diagnostic row (best effort; disclosed).
    arm_means: Optional[np.ndarray] = None
    # The arm label for the diagnostic disclosure.
    arm_label: str = "all_rows__no_aggregates"
    # Compute the arm when the frame is a usable DataFrame.
    if isinstance(aggregates, pd.DataFrame):
        # Select the scorable arm (or the disclosed all-rows fallback).
        arm_frame, arm_label = _scorable_arm(aggregates)
        # Extract the finite mean rank-IC array over the arm.
        arm_means = _finite_float_col(arm_frame, "mean_rank_ic") if "mean_rank_ic" in arm_frame.columns else None
    # Extend with the pinned pin-echo and diagnostic rows (the spec under which the variant ran).
    quantities = quantities + _pin_echo_rows(
        config=config,
        resolved_denominator=denominator,
        means=arm_means,
        arm_label=arm_label,
    )
    # Compose the provenance note (branch taken, both triples, the P04 citation).
    note: str = (
        f"source={source}; primary triple=(scorable={primary_triple[0]}, insufficient={primary_triple[1]}, "
        f"unsupported={primary_triple[2]}); reverse triple=(scorable={reverse_triple[0]}, "
        f"insufficient={reverse_triple[1]}, unsupported={reverse_triple[2]}); both disclosed per "
        f"P04.state_precedence_consequence; denominators={denominator}"
    )
    # Append the re-invocation halt disclosure when one occurred.
    if reinvocation_note:
        # The reuse branch operated after a disclosed halt.
        note += f"; aggregate_and_ledger re-invocation halted ({reinvocation_note}); reused the primary materialisation (step-2 Challenges clause)"
    # Assemble the frozen result.
    return VariantResult(
        variant_name="variant_state_precedence_reversed",
        interpretation="reversed-precedence partition disclosed; [R] specification sensitivity; not a paper finding",
        quantities=quantities,
        isolation=tuple(isolation),
        patch_echo=_patch_echo(patch_d),
        status=VARIANT_STATUS_EXECUTED,
        halt_state=HALT_CLOSED if all_evaluated_ok else FAILED_CLOSURE,
        ledger_partition=(
            f"reverse: scorable={reverse_triple[0]} insufficient={reverse_triple[1]} "
            f"unsupported={reverse_triple[2]} denominator={denominator}"
        ),
        note=note,
        closures=closures,
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1, Variants 2-3: the freeze re-invocation scaffolding
# -------------------------------------------------------------------------------------------------------------------------------

def _freeze_variant(
    primary: Any,
    patch: Mapping[str, Any],
    *,
    inadmissible: bool,
) -> VariantResult:
    """
    Re-invoke the freeze protocol with a changed deduplication pin.

    Shared implementation of the two freeze-level variants. On a deep
    copy of the primary's configuration, ``P06_COHORT_ADMISSION_
    PARAMETERS.deduplication_canonicalization`` is patched: for the
    string-equality variant the entire embedded mapping is replaced
    by the caller-supplied replacement (so several members can
    change); for the negation variant only ``push_negation_through``
    is set to ``True``. The same ``preserved_rows`` are consumed and
    the same ``freeze_cohort`` function object from ``task_14`` is
    called.

    Process:

    1. Validate and deep-copy ``patch``; for the negation variant
       admit only ``push_negation_through`` (must be truthy); for the
       string-equality variant admit only
       ``deduplication_canonicalization`` (must be a ``Mapping`` with
       ``parse_to_ast`` falsy and ``normalise_whitespace`` truthy).
    2. Require ``preserved_rows``, ``evaluator_canonicalizer``, a
       ``config`` mapping and a timezone-aware OOS-read instant; any
       absence returns an unavailable result with no quantity rows.
       A missing or non-datetime instant is replaced by a recoverable
       cohort freeze timestamp plus one microsecond and the
       derivation is noted; a timezone-naive datetime instant is not
       replaced and yields an unavailable result.
    3. Patch the embedded canonicalisation (the full replacement, or
       the single ``push_negation_through = True`` override) and
       re-invoke ``freeze_cohort``; an exception or halt reason
       returns an unavailable result with the sanitised reason.
    4. Read ``n_after_dedup`` from the frozen meta
       ``cohort_size`` and the primary reference from
       ``stages_output['freeze']['n_after_dedup']``,
       ``frozen_cohort.meta['cohort_size']`` or ``len(rows)``; a
       missing operand returns an unavailable result.
    5. Evaluate the seven closures, emit the single moved quantity
       ``n_after_dedup`` (``EXACT_INTEGER``) plus the pin-echo rows,
       and attach the pinned disclosure note.

    Parameters
    ----------
    primary : Any
        The primary result (``StudyResult`` or the fixture duck-type).
    patch : Mapping[str, Any]
        The variant patch; admissible keys depend on ``inadmissible``.
    inadmissible : bool
        Select ``variant_dedup_string_equality`` (``False``) or
        ``variant_dedup_negation_push_through`` (``True``).

    Returns
    -------
    VariantResult
        Frozen result tagged with the selected variant name; zero
        quantity rows when the freeze cannot honestly be re-executed.

    Raises
    ------
    TypeError
        If ``patch`` is not a ``Mapping``, or the string-equality
        replacement is not a ``Mapping``.
    ValueError
        If ``patch`` carries unexpected keys, if
        ``push_negation_through`` is falsy, if ``parse_to_ast`` is
        truthy, or if ``normalise_whitespace`` is falsy.

    Notes
    -----
    A freeze directory already sealed by the primary run halts by
    construction (write-once per nonce-namespaced directory), so the
    v2/v3 variants are usually disclosed as unavailable rather than
    recomputed.
    """
    # Boundary: the patch must be a mapping (deep-copied; the caller's mapping is never aliased).
    patch_d: Dict[str, Any] = _validate_patch_mapping(
        patch,
        caller="variant_dedup_negation_push_through" if inadmissible else "variant_dedup_string_equality",
    )
    # Resolve the pinned variant name.
    variant_name: str = "variant_dedup_negation_push_through" if inadmissible else "variant_dedup_string_equality"
    # Validate the pinned key set for this variant.
    if inadmissible:
        # The negation variant admits exactly one key.
        _ensure_patch_keys(patch_d, allowed={"push_negation_through"}, caller=variant_name)
        # Resolve the flag (pinned to True for this variant; anything else is a different variant).
        push_flag: bool = bool(patch_d.get("push_negation_through", True))
        # Enforce the pinned truthy value.
        if not push_flag:
            # Refuse a falsy flag with the observed value.
            raise ValueError(
                "variant_dedup_negation_push_through: patch['push_negation_through'] must be truthy "
                f"(pinned to True), got {push_flag!r}"
            )
    else:
        # The string-equality variant admits exactly one key.
        _ensure_patch_keys(patch_d, allowed={"deduplication_canonicalization"}, caller=variant_name)
        # Resolve the replacement canonicalisation mapping (default: the pinned string-equality patch).
        replacement: Any = patch_d.get(
            "deduplication_canonicalization",
            dict(VARIANT_2_STRING_EQUALITY_CANONICALIZATION),
        )
        # The replacement must be a mapping.
        if not isinstance(replacement, Mapping):
            # Refuse non-mapping replacements with the observed type.
            raise TypeError(
                f"variant_dedup_string_equality: patch['deduplication_canonicalization'] must be a "
                f"Mapping, got {type(replacement).__name__}"
            )
        # Deep-copy the replacement so the caller's mapping is never aliased.
        replacement = copy.deepcopy(dict(replacement))
        # Enforce the pinned string-equality semantics (parse_to_ast falsy; normalise_whitespace truthy).
        if bool(replacement.get("parse_to_ast", True)):
            # Refuse a truthy parse_to_ast with the observed value.
            raise ValueError(
                "variant_dedup_string_equality: patch['deduplication_canonicalization'] pins "
                f"parse_to_ast == False (string-equality), got {replacement.get('parse_to_ast')!r}"
            )
        # Enforce the whitespace-normalisation pin.
        if not bool(replacement.get("normalise_whitespace", False)):
            # Refuse a falsy normalise_whitespace with the observed value.
            raise ValueError(
                "variant_dedup_string_equality: patch['deduplication_canonicalization'] pins "
                f"normalise_whitespace == True, got {replacement.get('normalise_whitespace')!r}"
            )
        # Materialise the validated replacement for the configuration build below.
        resolved_replacement: Dict[str, Any] = replacement
    # Boundary: the preserved rows are the one intermediate the freeze consumes.
    preserved_rows: Any = getattr(primary, "preserved_rows", None)
    # The rows must be a DataFrame.
    if not isinstance(preserved_rows, pd.DataFrame):
        # Disclose the absence with zero quantity rows (never estimated).
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation unavailable: preserved_rows absent; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason="preserved_rows is absent from the primary (the frozen StudyResult does not retain it; it lives in the run-scoped pipeline context)",
            ledger_partition="NOT_EVALUATED__preserved_rows_absent",
            patch_echo=_patch_echo(patch_d),
            note=("variants 2/3 re-invoke freeze_cohort on the same preserved_rows with one "
                  "deduplication parameter changed; the artefact is not retained on the primary"),
        )
    # Boundary: the canonicalizer certificates (task_7) are the freeze's identity anchor.
    canonicalizer: Any = getattr(primary, "evaluator_canonicalizer", None)
    # The canonicalizer must be present (mapping or object).
    if canonicalizer is None:
        # Disclose the absence with zero quantity rows (never re-built here; never re-implemented).
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation unavailable: evaluator_canonicalizer absent; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason="evaluator_canonicalizer is absent from the primary (it lives in the run-scoped pipeline context; a bare StudyResult does not carry it)",
            ledger_partition="NOT_EVALUATED__canonicalizer_absent",
            patch_echo=_patch_echo(patch_d),
            note=("the canonicalizer certificates are fixed at the archive's semantics "
                  "(push_negation_through == False); re-building one is outside this variant's pinned scope"),
        )
    # Boundary: a configuration mapping is required to carry the patched P06 embedding.
    config: Any = getattr(primary, "config", None)
    # The configuration must be a mapping.
    if not isinstance(config, Mapping):
        # Disclose the absence with zero quantity rows.
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation unavailable: no configuration attached; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason="the primary carries no configuration dictionary",
            ledger_partition="NOT_EVALUATED__config_absent",
            patch_echo=_patch_echo(patch_d),
            note="freeze_cohort requires STUDY_CONFIG; the freeze cannot be honestly re-executed without it",
        )
    # Deep-copy the attached configuration so the caller's mapping is never mutated.
    patched_config: Dict[str, Any] = copy.deepcopy(dict(config))
    # Resolve the P06 section (default when the key is absent: the pinned archive embedding).
    section: Any = patched_config.get("P06_COHORT_ADMISSION_PARAMETERS")
    # The section must be a mapping; otherwise default it (documented archive pins).
    p06: Dict[str, Any] = dict(section) if isinstance(section, Mapping) else {}
    # Resolve the embedded deduplication_canonicalization (default: the archive pins).
    dedup_member: Any = p06.get("deduplication_canonicalization")
    # Default the embedding when absent (the archive's pinned values; documented).
    dedup: Dict[str, Any] = (
        dict(dedup_member)
        if isinstance(dedup_member, Mapping)
        else {
            "parse_to_ast": True,
            "normalise_whitespace": True,
            "apply_commutativity": "symmetric_operators_only",
            "push_negation_through": False,
            "source": "NOT_STATED_IN_PAPER__[R]",
        }
    )
    # Apply exactly the one changed parameter.
    if inadmissible:
        # The negation variant: a one-key override (the pinned inadmissible patch).
        dedup["push_negation_through"] = True
    else:
        # The string-equality variant: the validated full replacement.
        dedup = dict(resolved_replacement)
    # Write the patched embedding back into the deep-copied configuration.
    p06["deduplication_canonicalization"] = dedup
    # Write the patched section back into the deep-copied configuration.
    patched_config["P06_COHORT_ADMISSION_PARAMETERS"] = p06
    # Resolve the OOS-read instant (never fabricated; derived successor disclosed).
    oos_read_timestamp: Any = getattr(primary, "oos_read_timestamp", None)
    # Track whether the instant was derived (disclosed in the note).
    derived_timestamp: bool = False
    # A missing or non-datetime OOS-read instant degrades to the freeze-timestamp successor.
    if not isinstance(oos_read_timestamp, datetime.datetime):
        # Recover a timezone-aware freeze instant from the primary.
        freeze_ts: Optional[datetime.datetime] = _freeze_timestamp_hint(primary)
        # A recoverable freeze instant yields the disclosed +1 microsecond successor.
        if isinstance(freeze_ts, datetime.datetime):
            # The successor is strictly later, satisfying the freeze assertion by construction.
            oos_read_timestamp = freeze_ts + datetime.timedelta(microseconds=1)
            # Record the derivation for the note.
            derived_timestamp = True
        else:
            # Nothing recoverable: disclose with zero quantity rows.
            return _unavailable_variant_result(
                variant_name=variant_name,
                interpretation="freeze re-invocation unavailable: no timezone-aware OOS-read instant; nothing is estimated",
                halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
                reason="no timezone-aware oos_read_timestamp or recoverable cohort freeze timestamp on the primary",
                ledger_partition="NOT_EVALUATED__oos_timestamp_unavailable",
                patch_echo=_patch_echo(patch_d),
                note="freeze_cohort asserts freeze_timestamp < oos_read_timestamp; a naive or missing instant is refused",
            )
    # A naive OOS-read instant is refused (the freeze assertion is uncomparable).
    if oos_read_timestamp.tzinfo is None or oos_read_timestamp.utcoffset() is None:
        # Disclose with zero quantity rows.
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation unavailable: OOS-read instant is naive; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason="oos_read_timestamp is timezone-naive",
            ledger_partition="NOT_EVALUATED__oos_timestamp_naive",
            patch_echo=_patch_echo(patch_d),
            note="freeze_cohort requires a timezone-aware OOS-read instant",
        )
    # Load the task_14 registry lazily (the same function object the primary invoked).
    task_14 = _load_dependency_module("task_14")
    # The import must succeed.
    if task_14 is None:
        # Disclose with zero quantity rows.
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation unavailable: task_14 not importable; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason="task_14 could not be imported in this environment",
            ledger_partition="NOT_EVALUATED__task_14_unimportable",
            patch_echo=_patch_echo(patch_d),
            note="variants call the same function object as the primary; importing it is required",
        )
    # Guard the re-invocation: any failure is recorded, never raised through the variant.
    try:
        # Re-invoke the same function object with exactly one parameter changed
        # (the embedded deduplication_canonicalization); the same preserved_rows are consumed.
        frozen, halt_reason = task_14.freeze_cohort(
            preserved_rows,
            canonicalizer,
            patched_config,
            oos_read_timestamp,
        )
    # Any exception is captured as a halt (never raised through the variant).
    except Exception as exc:
        # Sanitise the exception text before it enters any artefact.
        halt_reason = f"{type(exc).__name__}: {exc}"
        # No frozen cohort was produced.
        frozen = None
    # A halt or missing cohort yields a HALTED result with zero quantity rows.
    if frozen is None or halt_reason:
        # Sanitise the halt reason before it enters any artefact.
        halt_text: str = _sanitise_ephemeral_paths(halt_reason)
        # Build the HALTED result (zero rows, the reason disclosed).
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation halted; nothing is estimated",
            halt_state=f"HALTED::{halt_text[:160]}",
            reason=f"freeze_cohort halted under the patched canonicalisation ({halt_text})",
            ledger_partition="NOT_EVALUATED__freeze_halted",
            patch_echo=_patch_echo(patch_d),
            note=("the freeze is write-once per nonce-namespaced directory; a re-invocation in a "
                  "directory already sealed by the primary run halts by construction"),
        )
    # Measure the post-deduplication cohort size from the frozen meta.
    meta: Any = getattr(frozen, "meta", None)
    # The meta must be a mapping with a cohort_size member.
    if not isinstance(meta, Mapping):
        # Refuse to guess the size.
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation halted: frozen cohort meta unusable; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason="freeze_cohort returned a frozen cohort without a meta mapping",
            ledger_partition="NOT_EVALUATED__cohort_meta_unusable",
            patch_echo=_patch_echo(patch_d),
            note="n_after_dedup is read from the frozen meta (P31 meta_fields.cohort_size)",
        )
    # Fetch the cohort size member.
    cohort_size_member: Any = meta.get("cohort_size")
    # The member must be an integral non-negative count.
    if (
        cohort_size_member is None
        or isinstance(cohort_size_member, bool)
        or not isinstance(cohort_size_member, (int, float, np.number))
        or not float(cohort_size_member).is_integer()
        or float(cohort_size_member) < 0
    ):
        # Refuse to guess the size.
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation halted: frozen cohort size unusable; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason=f"freeze_cohort returned cohort_size={cohort_size_member!r}",
            ledger_partition="NOT_EVALUATED__cohort_size_unusable",
            patch_echo=_patch_echo(patch_d),
            note="n_after_dedup is read from the frozen meta (P31 meta_fields.cohort_size)",
        )
    # The measured post-deduplication size (the only moved quantity).
    n_after_dedup: int = int(cohort_size_member)
    # Recover the primary's n_after_dedup for the side-by-side disclosure.
    primary_n: Optional[int] = _primary_n_after_dedup(primary)
    # Without the primary reference there is no honest side-by-side; disclose.
    if primary_n is None:
        # Return the unavailable result (the delta requires both operands).
        return _unavailable_variant_result(
            variant_name=variant_name,
            interpretation="freeze re-invocation halted: no primary n_after_dedup reference; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_INPUTS,
            reason="the primary retains no n_after_dedup reference (stages_output freeze payload and frozen_cohort both absent)",
            ledger_partition="NOT_EVALUATED__primary_n_after_dedup_absent",
            patch_echo=_patch_echo(patch_d),
            note="the delta requires both operands; the primary operand is absent from this object",
        )
    # Resolve the primary ledger triple (unchanged by the variant; used by the battery).
    primary_triple: Tuple[int, int, int] = _primary_triple_or_raise(primary)
    # Resolve the reverse triple for the battery.
    reverse_triple: Optional[Tuple[int, int, int]] = _reverse_triple_or_none(primary)
    # Resolve the aggregates frame for the frame-level closures.
    aggregates: Optional[pd.DataFrame] = _safe_factor_aggregates(primary)
    # Resolve the denominator in the pinned order.
    denominator: int = _denominator_hint(
        primary=primary,
        aggregates=aggregates,
        ledger_payload=getattr(primary, "ledger_primary", None),
        config=config,
    )
    # Evaluate the seven closures (the freeze-level variants leave the ledger untouched).
    closures, all_evaluated_ok = _variant_closures_and_verdict(
        primary=primary,
        aggregates=aggregates,
        primary_triple=primary_triple,
        reverse_triple=reverse_triple,
        denominator=denominator,
    )
    # Compose the interpretation (variant 3 carries the pinned inadmissibility language).
    if inadmissible:
        # The pinned interpretation for the inadmissible variant.
        interpretation: str = (
            "RECONSTRUCTION_OF_AN_INADMISSIBLE_VARIANT; push_negation_through collapses the sign-flip "
            "augmentation; reported only to quantify the consequence; never admissible; "
            "[R] specification sensitivity; not a paper finding"
        )
        # The pinned disclosure sentence with the measured cohort sizes substituted.
        disclosure: str = VARIANT_3_DISCLOSURE_TEMPLATE.format(n_alt=n_after_dedup, n_primary=primary_n)
        # Compose the note with both pinned sentences.
        note: str = f"{disclosure} {VARIANT_3_ADMISSIBILITY_NOTE}"
    else:
        # The pinned interpretation for the string-equality variant.
        interpretation = (
            "string-equality deduplication vs canonical-AST deduplication on the same "
            "preserved_rows; [R] specification sensitivity only; not a paper finding"
        )
        # Compose the note with the iso-morphism disclosure and the [R] sensitivity tag.
        note = (
            f"string-equality deduplication (parse_to_ast == False, normalise_whitespace == True) "
            f"yields n_after_dedup={n_after_dedup} vs primary n_after_dedup={primary_n} on the same "
            f"preserved_rows; the deviation from the primary cohort is disclosed as [R] sensitivity; "
            f"the primary freeze remains the paper-faithful cohort"
        )
    # Record the timestamp derivation when it occurred.
    if derived_timestamp:
        # Append the derivation disclosure to the note.
        note += "; oos_read_timestamp DERIVED as cohort freeze timestamp + 1 microsecond (strictly-later successor)"
    # The moved quantity: n_after_dedup.
    quantities: Tuple[QuantityDelta, ...] = (
        _quantity(
            "n_after_dedup",
            float(primary_n),
            float(n_after_dedup),
            "EXACT_INTEGER",
            interpretation,
        ),
    )
    # Extend with the pinned pin-echo rows (the spec under which the variant ran).
    quantities = quantities + _pin_echo_rows(
        config=patched_config if isinstance(patched_config, Mapping) else config,
        resolved_denominator=denominator,
        means=None,
        arm_label="freeze_level_variant",
    )
    # Assemble the frozen result (isolation: the preserved rows are reused by identity).
    return VariantResult(
        variant_name=variant_name,
        interpretation=interpretation,
        quantities=quantities,
        isolation=(("preserved_rows", True),),
        patch_echo=_patch_echo(patch_d),
        status=VARIANT_STATUS_EXECUTED,
        halt_state=HALT_CLOSED if all_evaluated_ok else FAILED_CLOSURE,
        ledger_partition=(
            f"primary: scorable={primary_triple[0]} insufficient={primary_triple[1]} "
            f"unsupported={primary_triple[2]} denominator={denominator}"
        ),
        note=note,
        closures=closures,
    )

def variant_dedup_string_equality(primary: Any, patch: Mapping[str, Any]) -> VariantResult:
    """
    Re-run the freeze under string-equality deduplication.

    Delegate to ``_freeze_variant`` with ``inadmissible=False``: the
    embedded ``deduplication_canonicalization`` is replaced by the
    caller's replacement mapping on the same ``preserved_rows``. The
    replacement defaults to ``{parse_to_ast: False,
    normalise_whitespace: True, apply_commutativity:
    "symmetric_operators_only", push_negation_through: False}``, but
    only ``parse_to_ast == False`` and ``normalise_whitespace == True``
    are enforced -- every other member (including
    ``apply_commutativity`` and ``push_negation_through``) is taken
    from the replacement mapping as supplied.

    Parameters
    ----------
    primary : Any
        The primary result carrying the freeze inputs; see
        ``_freeze_variant``.
    patch : Mapping[str, Any]
        Variant patch; the only admissible key is
        ``deduplication_canonicalization``, defaulting to the pinned
        string-equality mapping.

    Returns
    -------
    VariantResult
        Frozen result with the single moved quantity ``n_after_dedup``
        and a note disclosing the deviation from the primary cohort as
        ``[R]`` sensitivity.

    Raises
    ------
    TypeError
        If ``patch`` or the replacement is not a ``Mapping``.
    ValueError
        If ``patch`` carries unexpected keys or the replacement
        violates the string-equality pins.
    """
    # Delegate to the shared freeze scaffolding with the string-equality pin.
    return _freeze_variant(primary, patch, inadmissible=False)

def variant_dedup_negation_push_through(primary: Any, patch: Mapping[str, Any]) -> VariantResult:
    """
    Re-run the freeze with negation pushed through operators.

    Delegate to ``_freeze_variant`` with ``inadmissible=True``: the
    embedded ``push_negation_through`` is forced to ``True`` on the
    same ``preserved_rows``. Because push-through collapses the
    sign-flip augmentation (``e -> -e``), the result is labelled
    ``RECONSTRUCTION_OF_AN_INADMISSIBLE_VARIANT`` and reported only to
    quantify the consequence; the primary freeze stays the
    paper-faithful cohort.

    Parameters
    ----------
    primary : Any
        The primary result carrying the freeze inputs; see
        ``_freeze_variant``.
    patch : Mapping[str, Any]
        Variant patch; the only admissible key is
        ``push_negation_through``, which must be truthy (default
        ``True``).

    Returns
    -------
    VariantResult
        Frozen result with the single moved quantity ``n_after_dedup``,
        the pinned disclosure sentence comparing ``n_alt`` with
        ``n_primary``, and the ``[R]`` sensitivity tag.

    Raises
    ------
    TypeError
        If ``patch`` is not a ``Mapping``.
    ValueError
        If ``patch`` carries unexpected keys or
        ``push_negation_through`` is falsy.
    """
    # Delegate to the shared freeze scaffolding with the inadmissible pin.
    return _freeze_variant(primary, patch, inadmissible=True)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1, Variant 4: variant_sign_classification_tolerance
# -------------------------------------------------------------------------------------------------------------------------------

def variant_sign_classification_tolerance(primary: Any, patch: Mapping[str, Any]) -> VariantResult:
    """
    Recompute the mean rank-IC sign partition at two tolerances.

    Reclassifies the same ``factor_aggregates`` table (no re-scoring)
    against two absolute tolerance pins: the P03 near-zero diagnostic
    ``near_zero_absolute_threshold`` (default ``1e-12``) and the
    step-pinned ``1e-06``. The primary reference is the exact zero-test
    on ``mean_rank_ic`` (positive if ``> 0``, negative if ``< 0``).
    Non-finite means are counted separately and never enter an arm;
    means inside the near-zero band are counted as a diagnostic bucket
    and excluded from both thresholded arms. The variant is reported as
    tolerance-sensitive only when the two thresholded partitions, or the
    P03 partition and the exact zero-test, differ; both partitions are
    disclosed either way, and the P04 paper comparison split is echoed
    as a comparison value that is never asserted. The shared
    seven-closure battery is evaluated on the primary intermediates.

    Parameters
    ----------
    primary : Any
        Primary result (``StudyResult`` or the acceptance-fixture
        duck-type). Must carry ``factor_aggregates`` with a unique
        ``expression_id`` and a numeric ``mean_rank_ic``; an attached
        ``config`` supplies the tolerance pin, otherwise the documented
        default applies.
    patch : Mapping[str, Any]
        Variant patch; pinned empty for this variant, so any key is
        refused.

    Returns
    -------
    VariantResult
        Frozen result whose ``quantities`` hold, in order,
        ``n_positive_1e-12``, ``n_negative_1e-12``,
        ``n_positive_1e-06``, ``n_negative_1e-06``, followed by the
        ``n_near_zero_*`` diagnostics and the specification-echo rows
        (``diag_p03_*``, ``diag_p04_*``, ``diag_mean_abs_rank_ic``).
        ``halt_state`` is ``HALT_CLOSED`` when every evaluated closure
        is ``CLOSED`` or ``DEFERRED`` and ``FAILED_CLOSURE`` otherwise;
        ``isolation`` records that the aggregates were reused and
        nothing was re-scored.

    Raises
    ------
    TypeError
        If ``factor_aggregates`` is absent or not a ``DataFrame``, or
        ``patch`` is not a ``Mapping``.
    ValueError
        If required columns are missing, ``expression_id`` carries
        duplicates, or ``patch`` is non-empty.
    """
    # Boundary: the primary must quack like a primary result.
    _validate_primary_like(primary, caller="variant_sign_classification_tolerance")
    # Boundary: the patch must be a mapping (deep-copied).
    patch_d: Dict[str, Any] = _validate_patch_mapping(patch, caller="variant_sign_classification_tolerance")
    # Boundary: this variant's patch is pinned empty; any key is refused.
    _ensure_patch_keys(patch_d, allowed=set(), caller="variant_sign_classification_tolerance")
    # Resolve and validate the factor aggregates (unique expression_id enforced).
    aggregates: pd.DataFrame = _require_factor_aggregates(primary)
    # Select the scorable arm when a state column exists (Task 10 Step 2 semantics).
    arm, arm_label = _scorable_arm(aggregates)
    # Extract the NaN-bearing float64 mean rank-IC array.
    means: np.ndarray = _finite_float_col(arm, "mean_rank_ic")
    # Resolve the P03 primary near-zero diagnostic from the attached configuration.
    config: Optional[Mapping[str, Any]] = _primary_config(primary)
    # The primary tolerance (documented default: the paper-pinned 1e-12).
    tolerance_primary: float = _p03_float_pin(config, "near_zero_absolute_threshold", SIGN_CLASSIFICATION_PRIMARY_THRESHOLD_DEFAULT)
    # The step-pinned secondary tolerance (|mean_rank_ic| < 1e-06).
    tolerance_secondary: float = SIGN_CLASSIFICATION_SECONDARY_THRESHOLD
    # The primary reference partition (Task 10 Step 3 exact zero-test).
    exact: Dict[str, int] = _exact_zero_partition(means)
    # The partition at the P03 primary threshold (|x| < 1e-12 near-zero band).
    partition_primary: Dict[str, int] = _thresholded_partition(means, tolerance_primary)
    # The partition at the step-pinned secondary threshold (|x| < 1e-06 near-zero band).
    partition_secondary: Dict[str, int] = _thresholded_partition(means, tolerance_secondary)
    # Tolerance sensitivity verdict: identical partitions across both thresholds and vs the primary.
    insensitive: bool = (
        partition_primary["n_positive"] == partition_secondary["n_positive"]
        and partition_primary["n_negative"] == partition_secondary["n_negative"]
        and partition_primary["n_positive"] == exact["n_positive"]
        and partition_primary["n_negative"] == exact["n_negative"]
    )
    # The verdict sentence (disclosed identically on every row of this variant).
    verdict: str = (
        "tolerance-insensitive: the partition is invariant across the P03 1e-12 diagnostic "
        "and the step-pinned 1e-06 threshold; the primary exact zero-test agrees"
        if insensitive
        else "tolerance-sensitive: the partition moves across the P03 1e-12 diagnostic and the "
        "step-pinned 1e-06 threshold (or vs the primary exact zero-test); both partitions disclosed"
    )
    # Resolve the P04 paper comparison counts (comparison values; never asserted here).
    pos_comparison: Optional[int]
    neg_comparison: Optional[int]
    # Read the disclosed paper comparison split for the sign partition (P04).
    pos_comparison, neg_comparison = _p04_sign_partition_counts(config)
    # Compose the comparison disclosure with an explicit undisclosed marker (never a fabricated digit).
    comparison_disclosure: str = (
        f"P04.sign_partition_counts (paper comparison, never asserted): "
        f"positive={pos_comparison if pos_comparison is not None else 'undisclosed'} "
        f"negative={neg_comparison if neg_comparison is not None else 'undisclosed'}"
    )
    # The interpretation shared by this variant's rows ([R]; never a paper finding); it
    # discloses both tolerance pins and the paper comparison split it re-derives the table against.
    interpretation: str = (
        f"[R] specification sensitivity, SENSITIVITY__[R]__NOT_A_PAPER_FINDING; {verdict}; "
        f"at |mean_rank_ic| < {tolerance_primary:.3g} (P03.near_zero_absolute_threshold) and "
        f"{tolerance_secondary:.3g} (step pin); {comparison_disclosure}"
    )
    # Row 1: n_positive_1e-12 (variant value = thresholded partition; primary = exact zero-test).
    quantity_1: QuantityDelta = _quantity(
        "n_positive_1e-12",
        float(exact["n_positive"]),
        float(partition_primary["n_positive"]),
        "EXACT_INTEGER",
        interpretation,
    )
    # Row 2: n_negative_1e-12.
    quantity_2: QuantityDelta = _quantity(
        "n_negative_1e-12",
        float(exact["n_negative"]),
        float(partition_primary["n_negative"]),
        "EXACT_INTEGER",
        interpretation,
    )
    # Row 3: n_positive_1e-06.
    quantity_3: QuantityDelta = _quantity(
        "n_positive_1e-06",
        float(exact["n_positive"]),
        float(partition_secondary["n_positive"]),
        "EXACT_INTEGER",
        interpretation,
    )
    # Row 4: n_negative_1e-06.
    quantity_4: QuantityDelta = _quantity(
        "n_negative_1e-06",
        float(exact["n_negative"]),
        float(partition_secondary["n_negative"]),
        "EXACT_INTEGER",
        interpretation,
    )
    # Row 5: the near-zero diagnostic at the P03 primary threshold (delta is identically 0 by construction).
    quantity_5: QuantityDelta = _quantity(
        "n_near_zero_1e-12",
        float(partition_primary["n_near_zero"]),
        float(partition_primary["n_near_zero"]),
        "EXACT_INTEGER",
        "P03 near-zero diagnostic (|mean_rank_ic| < 1e-12); counted, never excluded from the ledger",
    )
    # Row 6: the near-zero diagnostic at the step-pinned secondary threshold.
    quantity_6: QuantityDelta = _quantity(
        "n_near_zero_1e-06",
        float(partition_primary["n_near_zero"]),
        float(partition_secondary["n_near_zero"]),
        "EXACT_INTEGER",
        "near-zero diagnostic at |mean_rank_ic| < 1e-06 vs the P03 1e-12 diagnostic; [R] specification sensitivity",
    )
    # The quantities in pinned order.
    quantities: Tuple[QuantityDelta, ...] = (
        quantity_1,
        quantity_2,
        quantity_3,
        quantity_4,
        quantity_5,
        quantity_6,
    )
    # Resolve the primary ledger for the shared closure battery.
    primary_triple: Tuple[int, int, int] = _primary_triple_or_raise(primary)
    # Resolve the reverse triple.
    reverse_triple: Optional[Tuple[int, int, int]] = _reverse_triple_or_none(primary)
    # Resolve the denominator in the pinned order.
    denominator: int = _denominator_hint(
        primary=primary,
        aggregates=_safe_factor_aggregates(primary),
        ledger_payload=getattr(primary, "ledger_primary", None),
        config=config,
    )
    # Evaluate the seven closures (the same battery the primary evaluated).
    closures, all_evaluated_ok = _variant_closures_and_verdict(
        primary=primary,
        aggregates=_safe_factor_aggregates(primary),
        primary_triple=primary_triple,
        reverse_triple=reverse_triple,
        denominator=denominator,
    )
    # Extend with the pinned pin-echo and diagnostic rows (the spec under which the variant ran).
    quantities = quantities + _pin_echo_rows(
        config=config,
        resolved_denominator=denominator,
        means=means,
        arm_label=arm_label,
    )
    # Compose the note (both partitions, the arm label, the insensitivity verdict).
    note: str = (
        f"arm={arm_label}; n_means={len(means)}; n_nan={int(np.count_nonzero(np.isnan(means)))}; "
        f"exact: pos={exact['n_positive']} neg={exact['n_negative']} zero={exact['n_exact_zero']} nan={exact['n_nan']}; "
        f"at 1e-12: pos={partition_primary['n_positive']} neg={partition_primary['n_negative']} "
        f"near_zero={partition_primary['n_near_zero']}; at 1e-06: pos={partition_secondary['n_positive']} "
        f"neg={partition_secondary['n_negative']} near_zero={partition_secondary['n_near_zero']}; {verdict}; "
        f"the paper's 671/252 split is never restated here (P41); the same mean_rank_ic table, no re-scoring"
    )
    # Assemble the frozen result.
    return VariantResult(
        variant_name="variant_sign_classification_tolerance",
        interpretation=interpretation,
        quantities=quantities,
        isolation=(("factor_aggregates", True), ("no_rescoring", True)),
        patch_echo=_patch_echo(patch_d),
        status=VARIANT_STATUS_EXECUTED,
        halt_state=HALT_CLOSED if all_evaluated_ok else FAILED_CLOSURE,
        ledger_partition=(
            f"primary: scorable={primary_triple[0]} insufficient={primary_triple[1]} "
            f"unsupported={primary_triple[2]} denominator={denominator}; arm={arm_label}"
        ),
        note=note,
        closures=closures,
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1, Variant 5: variant_defect_selection_consequence
# -------------------------------------------------------------------------------------------------------------------------------

def variant_defect_selection_consequence(primary: Any, patch: Mapping[str, Any]) -> VariantResult:
    """
    Quantify the v1-defect selection consequence from Task 13's ledger.

    Computes ``delta = mean(mean_rank_ic | newly_scorable) -
    mean(mean_rank_ic | unchanged_scorable)`` by joining Task 13's
    per-record ``RE_EVALUATION_LEDGER`` to the same
    ``factor_aggregates`` on ``expression_id`` (never a fresh scoring
    run). Two ledger layouts are honoured: fixture tokens in a
    ``partition`` column (``newly_scorable`` / ``unchanged_scorable``),
    or the Task 13 ``v1_state`` / ``v2_state`` pair (newly = v2
    scorable and v1 not; unchanged = scorable under both). Arm means
    use finite ``mean_rank_ic`` values only; NaN means are never
    imputed. The interpretation is one of the three pinned literals,
    selected by the chosen materiality tolerance ``1e-09`` [R]:
    ``delta > tol`` gives "defect destroyed higher-signal",
    ``delta < -tol`` gives "defect destroyed lower-signal", otherwise
    "no material difference". An absent ledger, an unrecognised schema,
    an empty joined arm or a joined arm with no finite ``mean_rank_ic``
    values returns the pinned unavailable result instead of an estimate.
    The shared seven-closure battery is augmented with an arm-partition
    totality closure.

    Parameters
    ----------
    primary : Any
        Primary result (``StudyResult`` or the acceptance-fixture
        duck-type). Requires ``factor_aggregates`` with a unique
        ``expression_id`` and a numeric ``mean_rank_ic``; an optional
        ``RE_EVALUATION_LEDGER`` DataFrame supplies the arms.
    patch : Mapping[str, Any]
        Variant patch; pinned empty for this variant, so any key is
        refused.

    Returns
    -------
    VariantResult
        Frozen result whose ``quantities`` hold, in order, ``mean_new``
        (against the unchanged-arm reference), ``mean_unchanged``,
        ``delta`` (against the null contrast ``0.0``), ``n_new``
        (against ``0.0``) and ``n_unchanged``, followed by the
        specification-echo rows. Unavailable cases carry no quantities
        and set ``halt_state`` to one of
        ``VARIANT_5_UNAVAILABLE_PENDING_CORRECTION``,
        ``VARIANT_UNAVAILABLE_PENDING_LEDGER_SCHEMA`` or
        ``VARIANT_UNAVAILABLE_PENDING_DATA``.

    Raises
    ------
    TypeError
        If ``factor_aggregates`` is absent or not a ``DataFrame``,
        ``RE_EVALUATION_LEDGER`` is present but not a ``DataFrame``, or
        ``patch`` is not a ``Mapping``.
    ValueError
        If required columns are missing, identifiers carry duplicates
        (in the aggregates or the ledger), or ``patch`` is non-empty.
    """
    # Boundary: the primary must quack like a primary result.
    _validate_primary_like(primary, caller="variant_defect_selection_consequence")
    # Boundary: the patch must be a mapping (deep-copied).
    patch_d: Dict[str, Any] = _validate_patch_mapping(patch, caller="variant_defect_selection_consequence")
    # Boundary: this variant's patch is pinned empty; any key is refused.
    _ensure_patch_keys(patch_d, allowed=set(), caller="variant_defect_selection_consequence")
    # Resolve and validate the factor aggregates (the same table; never re-scored).
    aggregates: pd.DataFrame = _require_factor_aggregates(primary)
    # Fetch the Task 13 per-record ledger (absent or None == Task 13 halted).
    ledger: Any = getattr(primary, "RE_EVALUATION_LEDGER", None)
    # The halted/absent case: the pinned disclosure, zero quantity rows, never an estimate.
    if ledger is None:
        # Return the unavailable result with the pinned token.
        return _unavailable_variant_result(
            variant_name="variant_defect_selection_consequence",
            interpretation="defect-selection consequence unavailable pending Task 13 correction; nothing is estimated",
            halt_state=VARIANT_5_UNAVAILABLE_PENDING_CORRECTION,
            reason="RE_EVALUATION_LEDGER is absent or None on the primary (Task 13 halted); disclosed as VARIANT_UNAVAILABLE_PENDING_CORRECTION",
            ledger_partition="NOT_EVALUATED__RE_EVALUATION_LEDGER_absent",
            patch_echo=_patch_echo(patch_d),
            note="the defect-selection contrast requires Task 13's per-record v1/v2 partition; when Task 13 halts it is unavailable and disclosed, never estimated",
        )
    # A present ledger must be a DataFrame.
    if not isinstance(ledger, pd.DataFrame):
        # Refuse with the observed type.
        raise TypeError(f"variant_defect_selection_consequence: RE_EVALUATION_LEDGER must be a pandas.DataFrame, got {type(ledger).__name__}")
    # The identifier column is mandatory on every ledger schema.
    if "expression_id" not in ledger.columns:
        # Return the unavailable result with the pinned token (unknown schema).
        return _unavailable_variant_result(
            variant_name="variant_defect_selection_consequence",
            interpretation="defect-selection consequence unavailable: ledger schema not recognised; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_LEDGER_SCHEMA,
            reason=f"RE_EVALUATION_LEDGER lacks the expression_id column (columns: {list(ledger.columns)})",
            ledger_partition="NOT_EVALUATED__ledger_schema_unrecognised",
            patch_echo=_patch_echo(patch_d),
            note="expected either expression_id/partition or expression_id/v1_state/v2_state (Task 13 Step 3)",
        )
    # The identifier column must be duplicate-free.
    ledger_ids: pd.Series = ledger["expression_id"].astype("string")
    # Detect duplicated identifiers.
    duplicated: pd.Series = ledger_ids.duplicated(keep=False)
    # Any duplication voids the one-to-one join guarantee.
    if bool(duplicated.any()):
        # Refuse with the duplicated identifiers (bounded list).
        raise ValueError(f"variant_defect_selection_consequence: RE_EVALUATION_LEDGER carries duplicated expression_id values: {sorted(ledger_ids[duplicated].unique())[:10]}")
    # Resolve the two arms from the schema at hand (both pinned layouts).
    if "partition" in ledger.columns:
        # The fixture layout: partition tokens, cast to string for exact matching.
        partition_tokens: pd.Series = ledger["partition"].astype("string")
        # The closed token set recognised by this variant.
        recognised: set = {"newly_scorable", "unchanged_scorable"}
        # Any unknown token voids the arm assignment (never guessed).
        unknown_tokens: List[str] = sorted(set(partition_tokens.dropna().unique()) - recognised)
        # Refuse unknown tokens with the observed values.
        if unknown_tokens:
            # Return the unavailable result with the pinned token.
            return _unavailable_variant_result(
                variant_name="variant_defect_selection_consequence",
                interpretation="defect-selection consequence unavailable: ledger partition tokens not recognised; nothing is estimated",
                halt_state=VARIANT_UNAVAILABLE_PENDING_LEDGER_SCHEMA,
                reason=f"RE_EVALUATION_LEDGER partition tokens not recognised: {unknown_tokens}",
                ledger_partition="NOT_EVALUATED__partition_tokens_unrecognised",
                patch_echo=_patch_echo(patch_d),
                note="expected the pinned tokens newly_scorable / unchanged_scorable",
            )
        # The newly-scorable arm: records the v1 defect had destroyed.
        newly_mask: pd.Series = partition_tokens == "newly_scorable"
        # The unchanged-scorable arm: every record scorable under both v1 and v2.
        unchanged_mask: pd.Series = partition_tokens == "unchanged_scorable"
        # Record which schema branch produced the arms.
        schema_branch: str = "partition_tokens"
    elif {"v1_state", "v2_state"} <= set(ledger.columns):
        # The Task 13 Step 3 layout: derive the arms from the two state vocabularies.
        v1_state: pd.Series = ledger["v1_state"].astype("string").str.lower()
        # The v2 state vocabulary.
        v2_state: pd.Series = ledger["v2_state"].astype("string").str.lower()
        # The newly-scorable arm: v2 scorable and v1 not scorable.
        newly_mask = (v2_state == "scorable") & (v1_state != "scorable")
        # The unchanged-scorable arm: v2 scorable and v1 scorable.
        unchanged_mask = (v2_state == "scorable") & (v1_state == "scorable")
        # Record which schema branch produced the arms.
        schema_branch = "v1_v2_state_boolean_arms"
    else:
        # Neither schema recognised: the pinned disclosure, never a guess.
        return _unavailable_variant_result(
            variant_name="variant_defect_selection_consequence",
            interpretation="defect-selection consequence unavailable: ledger schema not recognised; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_LEDGER_SCHEMA,
            reason=f"RE_EVALUATION_LEDGER carries neither partition nor v1_state/v2_state (columns: {list(ledger.columns)})",
            ledger_partition="NOT_EVALUATED__ledger_schema_unrecognised",
            patch_echo=_patch_echo(patch_d),
            note="expected either expression_id/partition or expression_id/v1_state/v2_state (Task 13 Step 3)",
        )
    # The aggregates identifiers are the join side; they are unique (enforced above).
    aggregates_ids: pd.Series = aggregates["expression_id"].astype("string")
    # The set of identifiers present in the factor aggregates.
    aggregate_id_set: set = set(aggregates_ids)
    # The join mask: ledger records whose identifier appears in the factor aggregates.
    join_mask: pd.Series = ledger["expression_id"].astype("string").isin(aggregate_id_set)
    # The joined newly-scorable arm.
    newly_joined: pd.Series = newly_mask & join_mask
    # The joined unchanged-scorable arm.
    unchanged_joined: pd.Series = unchanged_mask & join_mask
    # The joined arm sizes.
    n_new: int = int(newly_joined.sum())
    # The unchanged arm size.
    n_unchanged: int = int(unchanged_joined.sum())
    # The total joined v2-scorable records (the partition totality identity).
    n_joined: int = int((newly_joined | unchanged_joined).sum())
    # The identifiers missing from the aggregates (disclosed; never imputed).
    n_missing_aggregates: int = int(((newly_mask | unchanged_mask) & ~join_mask).sum())
    # The mean rank-IC vector aligned to the ledger rows.
    means: np.ndarray = _finite_float_col(aggregates, "mean_rank_ic")
    # The mean lookup by identifier (unique on both sides; first occurrence wins trivially).
    mean_by_id: Dict[str, float] = dict(zip(aggregates_ids, means))
    # Map the ledger identifiers onto the means vector (NaN preserved for absent/undefined).
    ledger_means: pd.Series = ledger["expression_id"].astype("string").map(mean_by_id)
    # The finite means of the joined newly-scorable arm.
    new_finite: np.ndarray = pd.to_numeric(ledger_means[newly_joined], errors="coerce").to_numpy(dtype=np.float64, na_value=np.nan)
    # The finite means of the joined unchanged-scorable arm.
    unchanged_finite: np.ndarray = pd.to_numeric(ledger_means[unchanged_joined], errors="coerce").to_numpy(dtype=np.float64, na_value=np.nan)
    # An empty arm after the join: the pinned disclosure, zero quantity rows.
    if n_new == 0 or n_unchanged == 0:
        # Return the unavailable result (PENDING_DATA) with the arm breakdown.
        return _unavailable_variant_result(
            variant_name="variant_defect_selection_consequence",
            interpretation="defect-selection consequence unavailable: an arm is empty after the join; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_DATA,
            reason=f"joined arm sizes are n_new={n_new}, n_unchanged={n_unchanged} (n_joined={n_joined})",
            ledger_partition=f"newly_scorable={n_new} unchanged_scorable={n_unchanged}",
            patch_echo=_patch_echo(patch_d),
            note=f"n_missing_aggregates={n_missing_aggregates}; schema_branch={schema_branch}; an empty arm admits no mean",
        )
    # Compute the arm means over finite values (arithmetic mean only; NaN counts disclosed).
    mean_new: Optional[float] = _finite_mean(new_finite)
    # The unchanged-arm mean over finite values.
    mean_unchanged: Optional[float] = _finite_mean(unchanged_finite)
    # No finite mean in either arm: the pinned disclosure, zero quantity rows.
    if mean_new is None or mean_unchanged is None:
        # Return the unavailable result (PENDING_DATA) with the finite counts.
        return _unavailable_variant_result(
            variant_name="variant_defect_selection_consequence",
            interpretation="defect-selection consequence unavailable: no finite means in an arm; nothing is estimated",
            halt_state=VARIANT_UNAVAILABLE_PENDING_DATA,
            reason=f"finite means unavailable (n_finite_new={np.count_nonzero(~np.isnan(new_finite))}, n_finite_unchanged={np.count_nonzero(~np.isnan(unchanged_finite))})",
            ledger_partition=f"newly_scorable={n_new} unchanged_scorable={n_unchanged}",
            patch_echo=_patch_echo(patch_d),
            note="NaN means are never imputed (P04 missing_result_imputation == None)",
        )
    # The headline contrast (both operands disclosed).
    delta: float = mean_new - mean_unchanged
    # The pinned interpretation rule with the pinned materiality tolerance (1e-09, [R]).
    if delta > DEFECT_SELECTION_MATERIALITY_TOLERANCE:
        # delta > +tolerance: the defect destroyed higher-signal records.
        interpretation: str = DEFECT_INTERPRETATION_HIGHER_SIGNAL
    # The lower-signal branch.
    elif delta < -DEFECT_SELECTION_MATERIALITY_TOLERANCE:
        # delta < -tolerance: the defect destroyed lower-signal records.
        interpretation = DEFECT_INTERPRETATION_LOWER_SIGNAL
    else:
        # |delta| <= tolerance: no material difference.
        interpretation = DEFECT_INTERPRETATION_NO_MATERIAL
    # Resolve the primary ledger for the shared closure battery.
    primary_triple: Tuple[int, int, int] = _primary_triple_or_raise(primary)
    # Resolve the reverse triple.
    reverse_triple: Optional[Tuple[int, int, int]] = _reverse_triple_or_none(primary)
    # Resolve the attached configuration (None on a bare StudyResult / the fixture).
    config: Optional[Mapping[str, Any]] = _primary_config(primary)
    # Resolve the denominator in the pinned order.
    denominator: int = _denominator_hint(
        primary=primary,
        aggregates=_safe_factor_aggregates(primary),
        ledger_payload=getattr(primary, "ledger_primary", None),
        config=config,
    )
    # Evaluate the seven closures (the same battery the primary evaluated).
    closures_list, all_evaluated_ok = _variant_closures_and_verdict(
        primary=primary,
        aggregates=_safe_factor_aggregates(primary),
        primary_triple=primary_triple,
        reverse_triple=reverse_triple,
        denominator=denominator,
    )
    # The arm-partition totality closure (variant-specific; appended to the battery).
    arm_totality_ok: bool = n_new + n_unchanged == n_joined
    # Append it as the eighth closure triple (variant-specific; disclosed).
    closures: Tuple[Tuple[str, str, str], ...] = tuple(closures_list) + (
        (
            "C8_ARM_PARTITION_TOTALITY",
            "CLOSED" if arm_totality_ok else "OPEN",
            f"n_new({n_new}) + n_unchanged({n_unchanged}) = {n_new + n_unchanged} vs joined v2-scorable records = {n_joined}",
        ),
    )
    # Re-check the combined verdict (the appended closure counts).
    all_evaluated_ok = all(status in ("CLOSED", "DEFERRED") for _, status, _ in closures)
    # The quantities in pinned order; the variant-exclusive quantities carry their
    # disclosed references (mean_new vs the unchanged-arm reference; n_new vs 0.0 --
    # the primary admits zero newly-scorable records; delta vs the null contrast 0.0).
    quantities: Tuple[QuantityDelta, ...] = (
        _quantity(
            "mean_new",
            float(mean_unchanged),
            float(mean_new),
            "FLOAT_MEAN_RANK_IC",
            interpretation,
        ),
        _quantity(
            "mean_unchanged",
            float(mean_unchanged),
            float(mean_unchanged),
            "FLOAT_MEAN_RANK_IC",
            interpretation,
        ),
        _quantity(
            "delta",
            0.0,
            float(delta),
            "ABSOLUTE_TOLERANCE_1E-09",
            interpretation,
        ),
        _quantity(
            "n_new",
            0.0,
            float(n_new),
            "EXACT_INTEGER",
            interpretation,
        ),
        _quantity(
            "n_unchanged",
            float(n_unchanged),
            float(n_unchanged),
            "EXACT_INTEGER",
            interpretation,
        ),
    )
    # Extend with the pinned pin-echo and diagnostic rows (the spec under which the variant ran).
    quantities = quantities + _pin_echo_rows(
        config=config,
        resolved_denominator=denominator,
        means=means,
        arm_label="factor_aggregates (the recomputed table, all rows)",
    )
    # Compose the note (arm sizes, finite counts, the join disclosure, the tolerance pin).
    note: str = (
        f"n_new={n_new} n_unchanged={n_unchanged} n_joined={n_joined} n_missing_aggregates={n_missing_aggregates}; "
        f"n_finite_new={int(np.count_nonzero(~np.isnan(new_finite)))} "
        f"n_finite_unchanged={int(np.count_nonzero(~np.isnan(unchanged_finite)))}; "
        f"schema_branch={schema_branch}; materiality tolerance pinned at 1e-09 ([R], chosen convention, not paper-stated); "
        f"delta = mean_new - mean_unchanged; both operands disclosed; same factor_aggregates, no re-scoring; "
        f"RE_EVALUATION_LEDGER partition joined on expression_id; P05.defect_selection_consequence_self_audit"
    )
    # Assemble the frozen result.
    return VariantResult(
        variant_name="variant_defect_selection_consequence",
        interpretation=interpretation,
        quantities=quantities,
        isolation=(("factor_aggregates", True), ("re_evaluation_ledger", True), ("no_rescoring", True)),
        patch_echo=_patch_echo(patch_d),
        status=VARIANT_STATUS_EXECUTED,
        halt_state=HALT_CLOSED if all_evaluated_ok else FAILED_CLOSURE,
        ledger_partition=f"newly_scorable={n_new} unchanged_scorable={n_unchanged}",
        note=note,
        closures=closures,
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 1: The variant callable registry and the pinned VARIANT_SPECS
# -------------------------------------------------------------------------------------------------------------------------------

# The internal name -> callable registry (a literal lookup table; no string evaluation).
_VARIANT_REGISTRY: Dict[str, Any] = {
    "variant_state_precedence_reversed": variant_state_precedence_reversed,
    "variant_dedup_string_equality": variant_dedup_string_equality,
    "variant_dedup_negation_push_through": variant_dedup_negation_push_through,
    "variant_sign_classification_tolerance": variant_sign_classification_tolerance,
    "variant_defect_selection_consequence": variant_defect_selection_consequence,
}

# The pinned VARIANT_SPECS (Step 1): five entries in step order; each entry is a mapping
# with keys "name" and "patch" (variant 3 additionally carries the pinned inadmissibility
# label). The tuple itself is frozen -- it carries no append/extend/pop, which is what the
# acceptance contract pins; the patch mappings are validated and deep-copied per invocation.
VARIANT_SPECS: Tuple[Mapping[str, Any], ...] = (
    {
        "name": "variant_state_precedence_reversed",
        "patch": dict(VARIANT_1_REVERSED_PRECEDENCE_PATCH),
    },
    {
        "name": "variant_dedup_string_equality",
        "patch": {"deduplication_canonicalization": dict(VARIANT_2_STRING_EQUALITY_CANONICALIZATION)},
    },
    {
        "name": "variant_dedup_negation_push_through",
        "patch": dict(VARIANT_3_PUSH_NEGATION_PATCH),
        "label": VARIANT_3_INADMISSIBLE_LABEL,
    },
    {
        "name": "variant_sign_classification_tolerance",
        "patch": {},
    },
    {
        "name": "variant_defect_selection_consequence",
        "patch": {},
    },
)

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 2: Execution, determinism verification, closure check, run records
# -------------------------------------------------------------------------------------------------------------------------------

def _run_variant_once(name: str, patch: Mapping[str, Any], study_result: Any) -> VariantResult:
    """
    Invoke one registered variant, converting any exception to a halt.

    Looks the name up in the variant registry, calls the variant on the
    primary result as given and on a fresh deep copy of the patch dict,
    and never propagates an exception: a raised error becomes a
    ``HALTED`` result with zero quantities, an
    ``HALTED::<ExceptionClass>`` halt state and a path-sanitised reason
    in the note.

    Parameters
    ----------
    name : str
        Registered variant name; the caller is responsible for having
        validated it against the registry.
    patch : Mapping[str, Any]
        Variant patch mapping, already validated and deep-copied by
        ``_validate_patch_mapping``; it is deep-copied once more before
        the call.
    study_result : Any
        The primary result the variant consumes; passed through
        unchanged.

    Returns
    -------
    VariantResult
        The variant's own result, or a status ``HALTED`` result carrying
        the class-qualified reason and an empty closure tuple.
    """
    # Dispatch to the registry (every name here is pre-validated by the caller).
    variant_callable = _VARIANT_REGISTRY[name]
    # Guard the invocation: any exception becomes a HALTED result (never raised through).
    try:
        # Invoke the pure variant on the primary and the patch.
        return variant_callable(study_result, copy.deepcopy(dict(patch)))
    except Exception as exc:
        # Sanitise the exception text before it enters any artefact.
        halt_text: str = _sanitise_ephemeral_paths(f"{type(exc).__name__}: {exc}")
        # Build the honest HALTED result (zero rows; the reason disclosed).
        return VariantResult(
            variant_name=name,
            interpretation="variant halted before emission; see note for the reason",
            quantities=(),
            isolation=(),
            patch_echo=_patch_echo(patch),
            status=VARIANT_STATUS_HALTED,
            halt_state=f"HALTED::{type(exc).__name__}",
            ledger_partition=f"NOT_EVALUATED__{type(exc).__name__}",
            note=f"variant halted with {halt_text}",
            closures=(),
        )

def _emit_spec_rows_and_run(
    *,
    spec_index: int,
    name: str,
    patch: Mapping[str, Any],
    study_result: Any,
) -> Tuple[List[Dict[str, Any]], Dict[str, Any]]:
    """
    Execute one spec end to end: guard, determinism, closure check.

    Parameters
    ----------
    spec_index : int
        Positional index of the spec in ``variant_specs``.
    name : str
        Pinned variant name.
    patch : Mapping[str, Any]
        Validated patch mapping.
    study_result : Any
        The primary result.

    Returns
    -------
    Tuple[List[Dict[str, Any]], Dict[str, Any]]
        The report rows (possibly empty) and the run row (always one).
    """
    # Execute the variant once (all exceptions are converted inside the guard).
    first: VariantResult = _run_variant_once(name, patch, study_result)
    # Determinism verification (Step 2b): re-invoke on an equal deep copy.
    second: VariantResult = _run_variant_once(name, patch, study_result)
    # The stable payloads must match bit-for-bit (note excluded by design).
    deterministic: bool = second.stable_payload() == first.stable_payload()
    # A non-deterministic variant is a DETERMINISM_HALT: dropped from the report.
    if not deterministic:
        # The run record discloses the halt; zero rows are emitted.
        return [], {
            "variant_name": name,
            "status": VARIANT_STATUS_HALTED,
            "halt_state": DETERMINISM_HALT,
            "ledger_partition": first.ledger_partition,
            "determinism_hash": first.stable_hash(),
            "n_report_rows": 0,
            "note": (
                f"spec_index={spec_index}; the variant halted: re-invocation on identical inputs "
                f"diverged from the first invocation; the variant is dropped from the report (Step 2b)"
            ),
        }
    # The determinism hash of the first invocation (identical across invocations).
    determinism_hash: str = first.stable_hash()
    # Closure verification (Step 2c): every evaluated closure must be CLOSED or DEFERRED.
    closures_ok: bool = _closures_all_evaluated_ok(first.closures)
    # A closure failure or a FAILED_CLOSURE halt drops the variant from the report.
    if first.halt_state == FAILED_CLOSURE or not closures_ok:
        # Enumerate the failing closures for the note (disclosure, never deletion).
        failing: List[str] = [closure_id for closure_id, status, _ in first.closures if status == "OPEN"]
        # The run record discloses the closure failure; zero rows are emitted.
        return [], {
            "variant_name": name,
            "status": VARIANT_STATUS_HALTED,
            "halt_state": FAILED_CLOSURE,
            "ledger_partition": first.ledger_partition,
            "determinism_hash": determinism_hash,
            "n_report_rows": 0,
            "note": (
                f"spec_index={spec_index}; the variant halted: closure(s) {failing or ['(n/a)']} did not close; "
                f"no reported result (Step 2c)"
            ),
        }
    # Materialise the rows for this variant (empty for unavailable/halted results).
    rows: List[Dict[str, Any]] = first.to_report_rows()
    # Resolve the attached configuration (the augmented primary view carries it); disclose
    # the pins the variant suite consumed, next to this variant's outcome (provenance discipline).
    run_config: Optional[Mapping[str, Any]] = _primary_config(study_result)
    # Resolve the P03 scorability-gate pin (documented paper default: 20).
    run_min_valid: int = _p03_int_pin(run_config, "min_valid_dates_for_scorability", MIN_VALID_DATES_PINNED)
    # Resolve the P03 full-coverage pin (documented paper default: 480).
    run_full_coverage: int = _p03_int_pin(run_config, "full_coverage_target_dates", FULL_COVERAGE_TARGET_DATES_PINNED)
    # Resolve the P03 near-zero diagnostic pin (documented paper default: 1e-12).
    run_near_zero: float = _p03_float_pin(run_config, "near_zero_absolute_threshold", SIGN_CLASSIFICATION_PRIMARY_THRESHOLD_DEFAULT)
    # Resolve the P04 paper-scale denominator (may legitimately differ from a fixture payload).
    run_denominator_cfg: Optional[int] = _p04_denominator_config(run_config)
    # Resolve the P04 paper comparison counts (comparison values; never asserted here).
    run_pos_comparison, run_neg_comparison = _p04_sign_partition_counts(run_config)
    # The run record for this variant (always exactly one).
    run_row: Dict[str, Any] = {
        "variant_name": name,
        "status": first.status,
        "halt_state": first.halt_state,
        "ledger_partition": first.ledger_partition,
        "determinism_hash": determinism_hash,
        "n_report_rows": len(rows),
        "note": (
            f"spec_index={spec_index}; {first.note}; "
            f"config pins: P03.min_valid_dates_for_scorability={run_min_valid}, "
            f"P03.full_coverage_target_dates={run_full_coverage}, "
            f"P03.near_zero_absolute_threshold={run_near_zero:.3g}, "
            f"P04.denominator_total(config)={run_denominator_cfg if run_denominator_cfg is not None else 'undisclosed'}; "
            f"P04.sign_partition_counts (paper comparison, never asserted): "
            f"positive={run_pos_comparison if run_pos_comparison is not None else 'undisclosed'} "
            f"negative={run_neg_comparison if run_neg_comparison is not None else 'undisclosed'}"
        ),
    }
    # Return the rows and the run record.
    return rows, run_row

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Step 3: The sensitivity archive and the substantive-claim prohibition
# -------------------------------------------------------------------------------------------------------------------------------

def sensitivity_claim_register(report: SensitivityReport) -> Tuple[Any, ...]:
    """
    Emit one claim record per report row at the reconstruction tier.

    Loads Task 14's ``ClaimRecord`` lazily (no stand-in type is
    fabricated) and builds, for each ``SENSITIVITY_REPORT`` row in
    report order, a record whose ``claim_id`` is
    ``sensitivity::<variant_name>::<quantity>``, whose ``tier`` is the
    hard-coded ``SENSITIVITY_CLAIM_TIER_CEILING``
    (``reconstruction``, never a parameter), whose ``verdict`` is
    ``"REPORTED"``, whose ``value`` is the row ``delta`` as a float,
    and whose ``note`` carries the delta, the stamp, the
    interpretation, the withholding notice, the tier ceiling and the
    fail-closed gate default (``G1``-``G4`` all ``False``). An empty
    report yields a single ``NOT_ASSERTED`` record with
    ``value=None`` and the same ceiling so the mechanism is always
    exercised.

    Parameters
    ----------
    report : SensitivityReport
        The artefact returned by :func:`run_sensitivity_task_15`.

    Returns
    -------
    Tuple[Any, ...]
        The claim records in report row order; one ``NOT_ASSERTED``
        record when the report carries no rows.

    Raises
    ------
    TypeError
        If ``report`` is not a ``SensitivityReport``.
    RuntimeError
        If the ``task_14`` module cannot be loaded, leaving the
        ``ClaimRecord`` type unavailable.
    AssertionError
        If a constructed record's ``tier`` differs from the
        reconstruction ceiling.
    """
    # The report must be this task's frozen artefact.
    if not isinstance(report, SensitivityReport):
        # Refuse foreign objects with the observed type.
        raise TypeError(f"sensitivity_claim_register: report must be a SensitivityReport, got {type(report).__name__}")
    # Load the claim record type lazily (never re-implemented here).
    task_14 = _load_dependency_module("task_14")
    # The import must succeed.
    if task_14 is None:
        # Refuse to fabricate a stand-in record type.
        raise RuntimeError("sensitivity_claim_register: task_14 could not be imported; the ClaimRecord type is unavailable")
    # Materialise the report rows in report order (deterministic).
    rows: List[Dict[str, Any]] = report.SENSITIVITY_REPORT.to_dict("records")
    # Accumulate the records (deterministic order; spec order derived).
    records: List[Any] = []
    # Walk the rows once.
    for row in rows:
        # Construct the claim with the pinned ceiling tier (never a parameter).
        record = task_14.ClaimRecord(
            # The stable identifier encodes the variant and the quantity.
            claim_id=f"sensitivity::{row['variant_name']}::{row['quantity']}",
            # The ceiling tier (reconstruction); no sensitivity claim ranks higher.
            tier=SENSITIVITY_CLAIM_TIER_CEILING,
            # The evidence source names this run's sensitivity suite.
            evidence_source="sensitivity_suite_this_run",
            # The four admissibility gates: this variant carries no G1-G4 substrate of its
            # own; the honest fail-closed default is recorded (never silently True).
            gate_status={"G1": False, "G2": False, "G3": False, "G4": False},
            # The verdict literal for an emitted comparison.
            verdict="REPORTED",
            # The value is the delta (a fact); both operands live in the row.
            value=float(row["delta"]),
            # The note carries the withholding and the pinned tolerance reference.
            note=(
                f"Delta (variant - primary) = {float(row['delta']):.12g}; "
                f"stamp={row['stamp']}; interpretation={row['interpretation']!r}; "
                f"no substantive claim is admissible from a sensitivity variant (P20); "
                f"tier ceiling={SENSITIVITY_CLAIM_TIER_CEILING}; sensitivity results are explicitly "
                f"withheld from {', '.join(WITHHELD_CLAIM_CLASSES)}; "
                f"gates_unverified_default_fail: G1-G4 are all False for this sensitivity variant; "
                f"no admissibility gate was exercised by the variant itself"
            ),
        )
        # Enforce the ceiling invariant by construction (a sanity assertion, never a branch).
        assert record.tier == SENSITIVITY_CLAIM_TIER_CEILING
        # Accumulate the record.
        records.append(record)
    # The empty-report case still exercises the ceiling (one NOT_ASSERTED record).
    if not records:
        # Construct the single NOT_ASSERTED record at the ceiling tier.
        record = task_14.ClaimRecord(
            # The stable identifier of the empty-archive disclosure.
            claim_id="sensitivity::archive::NOT_ASSERTED",
            # The ceiling tier (reconstruction); never higher.
            tier=SENSITIVITY_CLAIM_TIER_CEILING,
            # The evidence source names this run's sensitivity suite.
            evidence_source="sensitivity_suite_this_run",
            # The four admissibility gates: the empty archive exercises no gate; fail-closed default.
            gate_status={"G1": False, "G2": False, "G3": False, "G4": False},
            # The verdict literal for the empty archive.
            verdict="NOT_ASSERTED",
            # No value exists in the empty archive.
            value=None,
            # The note discloses the emptiness and the ceiling.
            note=(
                f"stamp={SENSITIVITY_STAMP}; the sensitivity archive carries zero rows "
                f"(every requested variant was unavailable, determinism-halted or closure-failed); "
                f"no substantive claim is admissible from a sensitivity variant (P20); "
                f"tier ceiling={SENSITIVITY_CLAIM_TIER_CEILING}; "
                f"gates_unverified_default_fail: G1-G4 are all False for this record"
            ),
        )
        # Enforce the ceiling invariant by construction.
        assert record.tier == SENSITIVITY_CLAIM_TIER_CEILING
        # Accumulate the record.
        records.append(record)
    # Return the immutable register (spec order preserved).
    return tuple(records)

def sensitivity_archive_manifest(report: SensitivityReport) -> Dict[str, Any]:
    """
    Build the machine-readable deposit manifest for the archive.

    Takes the report rows and run records as plain dictionaries,
    locates the variant 3 ``n_after_dedup`` row of
    ``variant_dedup_negation_push_through`` and formats the pinned
    disclosure sentence with the measured ``variant_value`` and
    ``primary_value`` cohort sizes; when that row is absent the
    sentence is emitted with the literal ``n_alt`` / ``n_primary``
    tokens. Performs no I/O; every value is JSON-able and no clock or
    file path enters the manifest.

    Parameters
    ----------
    report : SensitivityReport
        The artefact returned by :func:`run_sensitivity_task_15`.

    Returns
    -------
    Dict[str, Any]
        Manifest with keys ``artifact``, ``deposit_status``
        (``"ADDITION_TO_PAPER_DELIVERABLES"``), ``disclosed_as``,
        ``stamp``, ``sensitivity_hash``, ``row_count``,
        ``variant_runs`` (list of run-record dicts),
        ``withholding_notice``, ``withheld_claim_classes``,
        ``claim_tier_ceiling``,
        ``significance_test_on_aggregate_permitted`` (always
        ``False``), ``variant_3_disclosure`` (the pinned sentence,
        never preferring variant 3 over the primary freeze),
        ``variant_3_label`` (the inadmissibility label) and
        ``admissible_statements`` (the reduction contract handed to
        the study-output layer).

    Raises
    ------
    TypeError
        If ``report`` is not a ``SensitivityReport``.
    """
    # The report must be this task's frozen artefact.
    if not isinstance(report, SensitivityReport):
        # Refuse foreign objects with the observed type.
        raise TypeError(f"sensitivity_archive_manifest: report must be a SensitivityReport, got {type(report).__name__}")
    # Materialise the rows and the run records in their pinned orders.
    rows: List[Dict[str, Any]] = report.SENSITIVITY_REPORT.to_dict("records")
    # The runs table as records.
    run_rows: List[Dict[str, Any]] = report.VARIANT_RUNS.to_dict("records")
    # Locate variant 3's n_after_dedup row (the disclosure quantity).
    variant_3_row: Optional[Dict[str, Any]] = next(
        (
            row
            for row in rows
            if str(row.get("variant_name")) == "variant_dedup_negation_push_through"
            and str(row.get("quantity")) == "n_after_dedup"
        ),
        None,
    )
    # Substitute the measured cohort sizes when the row exists (pinned sentence verbatim).
    if variant_3_row is not None:
        # Format the pinned sentence with the measured values.
        variant_3_disclosure: str = VARIANT_3_DISCLOSURE_TEMPLATE.format(
            n_alt=int(variant_3_row["variant_value"]),
            n_primary=int(variant_3_row["primary_value"]),
        )
    else:
        # Emit the pinned sentence with the literal tokens (nothing measured).
        variant_3_disclosure = VARIANT_3_DISCLOSURE_TEMPLATE.format(n_alt="n_alt", n_primary="n_primary")
    # Assemble the deposit manifest (every value JSON-able; no file paths, no clocks).
    return {
        # The addition-to-deliverables disclosure (P36 plus the Task 15 extension).
        "artifact": "sensitivity_report_with_primary_comparison",
        "deposit_status": "ADDITION_TO_PAPER_DELIVERABLES",
        "disclosed_as": (
            "an addition to the paper's deliverables (P36 reproducibility package plus the "
            "Task 15 extension); not a paper finding; disclosed as such"
        ),
        # The pinned stamp and hash (the archive's identity).
        "stamp": SENSITIVITY_STAMP,
        "sensitivity_hash": report.SENSITIVITY_HASH,
        # The row/run accounting (one row per (variant_name, quantity); one run per variant).
        "row_count": len(rows),
        "variant_runs": run_rows,
        # The withholding machinery (machine-readable).
        "withholding_notice": SENSITIVITY_WITHHOLDING_NOTE,
        "withheld_claim_classes": list(WITHHELD_CLAIM_CLASSES),
        "claim_tier_ceiling": SENSITIVITY_CLAIM_TIER_CEILING,
        # The significance-test prohibition, named explicitly (P20).
        "significance_test_on_aggregate_permitted": False,
        # The variant 3 pinned disclosure (never preferred over the primary freeze).
        "variant_3_disclosure": variant_3_disclosure,
        # The variant 3 inadmissibility label (always attached).
        "variant_3_label": VARIANT_3_INADMISSIBLE_LABEL,
        # The reduction contract for the study output layer.
        "admissible_statements": (
            "the numeric delta (a fact); the interpretation label with the materiality "
            "tolerance pinned (1e-09, [R]); the disclosure of the sensitivity"
        ),
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 15, Orchestrator Function
# -------------------------------------------------------------------------------------------------------------------------------

# The orchestrator composes the three steps into the task:
#   Step 1 (variant callable registry): the five pinned variants are resolved from the
#     closed name registry `_VARIANT_REGISTRY` (variant_state_precedence_reversed,
#     variant_dedup_string_equality, variant_dedup_negation_push_through,
#     variant_sign_classification_tolerance, variant_defect_selection_consequence);
#     VARIANT_SPECS pins the order, the patches and the inadmissibility label.
#   Step 2 (execution and delta reporting): `_run_variant_once` executes each variant
#     under the exception guard and the determinism re-verification (Step 2b); the
#     closure battery (C1-C7, Step 2c) is evaluated inside each variant and re-checked
#     before any row is emitted; `_emit_spec_rows_and_run` assembles the report rows and
#     the per-variant run record; `_augment_primary_with_config` threads the orchestrator's
#     configuration into the variants without mutating the caller's object.
#   Step 3 (sensitivity archive): rows are stamped SENSITIVITY__[R]__NOT_A_PAPER_FINDING;
#     SENSITIVITY_HASH covers the serialised rows; VARIANT_RUNS carries the per-variant
#     determinism_hash, halt_state and ledger_partition; the substantive-claim prohibition
#     is enforced by the stamp, the withholding notice and the reconstruction tier ceiling.
#     The orchestrator composes Step 3 into the task: it derives the claim register via
#     sensitivity_claim_register and the deposit manifest via sensitivity_archive_manifest
#     and attaches them to the returned artefact (the step's output is the task's output).
# `run_sensitivity_task_15` is the single public entry point and the only callable that
# sees the full composition; every statistic lives in the step callables.
def _augment_primary_with_config(study_result: Any, config: Mapping[str, Any]) -> Any:
    """
    Attach the orchestrator config to a non-mutating primary view.

    Returns the primary unchanged when it already carries a non-``None``
    ``config`` attribute, since a caller-attached configuration is
    authoritative. Otherwise builds a ``SimpleNamespace`` from
    ``vars(study_result)`` with the orchestrator's configuration added;
    every intermediate (``daily_scores_v2``, ``frozen_cohort``,
    ``preserved_rows``, ...) is carried by reference, so the
    same-by-value isolation witnesses are unaffected and the caller's
    object is never mutated.

    Parameters
    ----------
    study_result : Any
        Primary result (``StudyResult`` or the acceptance-fixture
        duck-type).
    config : Mapping[str, Any]
        Orchestrator configuration (``STUDY_CONFIG`` or its mirror) to
        attach as the ``config`` attribute.

    Returns
    -------
    Any
        The primary unchanged, or the augmented shallow view used for
        the variant invocations.

    Raises
    ------
    TypeError
        If the primary has no attached config and exposes no
        ``__dict__`` for a shallow attribute copy.
    """
    # A caller-attached configuration wins (the caller's explicit choice).
    if getattr(study_result, "config", None) is not None:
        # Return the primary unchanged (the attachment is authoritative).
        return study_result
    # Attempt a shallow attribute copy; every intermediate is carried by reference.
    try:
        # vars() requires a __dict__ (frozen dataclasses have one; slots-only objects do not).
        namespace: Dict[str, Any] = dict(vars(study_result))
    except TypeError as exc:
        # Without vars() there is no honest way to attach the configuration.
        raise TypeError(
            "run_sensitivity_task_15: study_result carries no configuration and cannot be "
            "shallowly copied to attach one (no __dict__); attach a configuration via the "
            "'config' attribute of the primary"
        ) from exc
    # Build the augmented view and attach the orchestrator's configuration.
    namespace["config"] = config
    # Return the augmented view (same intermediate references; fresh attribute table).
    return SimpleNamespace(**namespace)

def run_sensitivity_task_15(
    study_result: Any,
    config: Mapping[str, Any],
    variant_specs: Sequence[Mapping[str, Any]],
) -> SensitivityReport:
    """
    Execute the named sensitivity variants and return the archive.

    Validates the boundary types, reads the pinned P03 knobs from
    ``config`` and walks ``variant_specs`` positionally, refusing
    malformed, unregistered or duplicate names before anything runs.
    Each validated spec is then executed against an
    attribute-preserving view of ``study_result`` that carries
    ``config`` (built by ``_augment_primary_with_config``; the
    caller's objects are never mutated). Rows accumulate into
    ``SENSITIVITY_REPORT`` (one per ``(variant_name, quantity)`` pair,
    pinned column order) and run records into ``VARIANT_RUNS``
    (exactly one per requested variant). The
    ``(variant_name, quantity)`` uniqueness invariant is enforced and
    ``SENSITIVITY_HASH`` is the SHA-256 hex digest of the canonical
    JSON of the column names and the report rows. The Step 3 outputs
    are attached by calling :func:`sensitivity_claim_register` and
    :func:`sensitivity_archive_manifest` on the intermediate artefact.

    Parameters
    ----------
    study_result : Any
        The primary result (Task 14's frozen ``StudyResult``,
        duck-typed). Validated by ``_validate_primary_like``.
    config : Mapping[str, Any]
        The study configuration dictionary. Type-checked and read,
        never mutated. Supplies
        ``min_valid_dates_for_scorability`` (default 20),
        ``full_coverage_target_dates`` (default 480) and
        ``near_zero_absolute_threshold`` (default ``1e-12``) via
        ``_p03_int_pin`` / ``_p03_float_pin``.
    variant_specs : Sequence[Mapping[str, Any]]
        Ordered specs. Each must carry ``name`` (one of the five
        pinned variant names) and may carry ``patch`` (a Mapping,
        default empty) naming the changed argument; patches are
        deep-copied before use.

    Returns
    -------
    SensitivityReport
        Frozen artefact with attributes ``SENSITIVITY_REPORT``
        (DataFrame with columns ``variant_name``, ``quantity``,
        ``primary_value``, ``variant_value``, ``delta``,
        ``tolerance_class``, ``interpretation``, ``stamp``),
        ``SENSITIVITY_HASH`` (64 lowercase hex characters),
        ``VARIANT_RUNS`` (DataFrame with one ``determinism_hash`` /
        ``halt_state`` / ``ledger_partition`` record per requested
        variant), ``claim_register`` (tuple of claim records) and
        ``archive_manifest`` (the deposit manifest mapping).

    Raises
    ------
    TypeError
        If ``study_result`` is not primary-like, ``config`` is not a
        Mapping, ``variant_specs`` is a string, bytes or other
        non-sequence, a spec is not a Mapping, a spec ``name`` is not
        a non-empty string, or a spec ``patch`` is not a Mapping.
    ValueError
        If a spec name is not one of the pinned variant names, a name
        is requested twice, or the assembled rows contain a duplicate
        ``(variant_name, quantity)`` pair.
    AssertionError
        If the computed ``SENSITIVITY_HASH`` is not 64 lowercase hex
        characters.

    Notes
    -----
    The per-variant determinism re-run and the ledger closure battery
    live in the private execution helpers; a variant that fails either
    check keeps its run record and emits zero report rows.
    """
    # Boundary: the primary must quack like a primary result.
    _validate_primary_like(study_result, caller="run_sensitivity_task_15")
    # Boundary: the configuration must be a mapping (the STUDY_CONFIG dictionary).
    if not isinstance(config, Mapping):
        # Refuse with the observed type.
        raise TypeError(f"run_sensitivity_task_15: config must be a Mapping (STUDY_CONFIG), got {type(config).__name__}")
    # Read the pinned P03 knobs with their documented paper defaults (type-checked at the boundary).
    _min_valid_dates: int = _p03_int_pin(config, "min_valid_dates_for_scorability", MIN_VALID_DATES_PINNED)
    # The full-coverage pin (type-checked).
    _full_coverage_dates: int = _p03_int_pin(config, "full_coverage_target_dates", FULL_COVERAGE_TARGET_DATES_PINNED)
    # The near-zero diagnostic pin (type-checked).
    _near_zero: float = _p03_float_pin(config, "near_zero_absolute_threshold", SIGN_CLASSIFICATION_PRIMARY_THRESHOLD_DEFAULT)
    # Thread the configuration into the variants (a caller-attached primary config wins; the
    # caller's object is never mutated; the augmented view carries intermediates by reference).
    augmented_primary: Any = _augment_primary_with_config(study_result, config)
    # Boundary: the specs must be a non-string sequence.
    if isinstance(variant_specs, (str, bytes)) or not isinstance(variant_specs, Sequence):
        # Refuse with the observed type.
        raise TypeError(f"run_sensitivity_task_15: variant_specs must be a non-string Sequence of spec mappings, got {type(variant_specs).__name__}")
    # Materialise the specs (the caller's sequence is never aliased).
    specs: List[Mapping[str, Any]] = list(variant_specs)
    # Validate every spec's shape and the pinned key set before any execution.
    validated: List[Tuple[str, Dict[str, Any]]] = []
    # Track the requested names to detect duplicates deterministically.
    seen: set = set()
    # Walk the specs positionally (spec order governs every output).
    for index, spec in enumerate(specs):
        # Every spec must be a mapping.
        if not isinstance(spec, Mapping):
            # Refuse non-mapping specs with position and observed type.
            raise TypeError(f"run_sensitivity_task_15: variant_specs[{index}] must be a Mapping, got {type(spec).__name__}")
        # Fetch the name member.
        name: Any = spec.get("name")
        # The name must be a non-empty string.
        if not isinstance(name, str) or not name.strip():
            # Refuse malformed names with position and observed value.
            raise TypeError(f"run_sensitivity_task_15: variant_specs[{index}]['name'] must be a non-empty str, got {name!r}")
        # The name must be one of the five pinned variants (the registry is closed).
        if name not in PINNED_VARIANT_NAMES:
            # Refuse unregistered names before anything executes (the pinned failure mode).
            raise ValueError(
                f"run_sensitivity_task_15: variant_specs[{index}]['name'] names an unregistered variant: "
                f"{name!r}; the registry is closed over {list(PINNED_VARIANT_NAMES)}"
            )
        # Duplicate requests would emit duplicate (variant_name, quantity) rows.
        if name in seen:
            # Refuse duplicates with the offending name.
            raise ValueError(f"run_sensitivity_task_15: variant {name!r} is requested more than once")
        # Record the name as seen.
        seen.add(name)
        # Fetch the patch member (default: the empty mapping).
        patch: Any = spec.get("patch", {})
        # The patch must be a mapping.
        if not isinstance(patch, Mapping):
            # Refuse non-mapping patches with position and observed type.
            raise TypeError(f"run_sensitivity_task_15: variant_specs[{index}]['patch'] must be a Mapping, got {type(patch).__name__}")
        # Deep-copy the patch for the invocation (isolated from the caller's data).
        validated.append((name, copy.deepcopy(dict(patch))))
    # Execute the validated specs in order; rows and run records accumulate.
    all_rows: List[Dict[str, Any]] = []
    # Accumulate one run record per requested variant.
    run_rows: List[Dict[str, Any]] = []
    # Walk the validated specs with their positional index.
    for spec_index, (name, patch) in enumerate(validated):
        # Execute the spec end to end against the augmented primary view.
        rows, run_row = _emit_spec_rows_and_run(
            spec_index=spec_index,
            name=name,
            patch=patch,
            study_result=augmented_primary,
        )
        # Accumulate the emitted rows (possibly none).
        all_rows.extend(rows)
        # Accumulate the run record (always exactly one).
        run_rows.append(run_row)
    # The (variant_name, quantity) uniqueness invariant is enforced before assembly.
    row_keys: List[Tuple[str, str]] = [(str(row["variant_name"]), str(row["quantity"])) for row in all_rows]
    # Duplicate pairs are a construction failure, never silently dropped.
    if len(row_keys) != len(set(row_keys)):
        # Compute the duplicated keys for the message.
        duplicated: List[Tuple[str, str]] = sorted({key for key in row_keys if row_keys.count(key) > 1})
        # Refuse with the duplicated keys.
        raise ValueError(f"run_sensitivity_task_15: duplicate (variant_name, quantity) rows emitted: {duplicated}")
    # Assemble the report table with the pinned column order (RangeIndex).
    report_frame: pd.DataFrame = pd.DataFrame(all_rows, columns=list(REPORT_COLUMNS))
    # Assemble the runs table with the pinned column order (RangeIndex).
    runs_frame: pd.DataFrame = pd.DataFrame(run_rows, columns=list(RUN_COLUMNS))
    # The SENSITIVITY_HASH covers exactly the reported rows (the stamp included).
    sensitivity_hash: str = _sha256_json(
        {
            "columns": list(REPORT_COLUMNS),
            "rows": report_frame.to_dict("records"),
        }
    )
    # Enforce the digest invariant (64 lowercase hex characters).
    if len(sensitivity_hash) != 64 or any(character not in "0123456789abcdef" for character in sensitivity_hash):
        # A breach of the construction invariant is an assertion, not a user error.
        raise AssertionError("run_sensitivity_task_15: SENSITIVITY_HASH construction invariant violated")
    # Compose Step 3 into the task: the sensitivity archive is deposited with the run.
    # Build the intermediate artefact first, then derive the claim register and the deposit manifest.
    interim: SensitivityReport = SensitivityReport(
        SENSITIVITY_REPORT=report_frame,
        SENSITIVITY_HASH=sensitivity_hash,
        VARIANT_RUNS=runs_frame,
    )
    # Step 3 (claim register): every record carries tier exactly reconstruction (never a parameter).
    claim_register: Tuple[Any, ...] = sensitivity_claim_register(interim)
    # Step 3 (deposit manifest): the addition-to-deliverables disclosure (P36 plus the Task 15 extension).
    archive_manifest: Dict[str, Any] = sensitivity_archive_manifest(interim)
    # Return the frozen, read-only artefact with the Step 3 outputs wired in (the task uses them).
    return SensitivityReport(
        SENSITIVITY_REPORT=interim.SENSITIVITY_REPORT,
        SENSITIVITY_HASH=interim.SENSITIVITY_HASH,
        VARIANT_RUNS=interim.VARIANT_RUNS,
        claim_register=claim_register,
        archive_manifest=archive_manifest,
    )
```

```python
# Task 16: Claim Governance, Tier Tagging, and Artifact Archival

# ==============================================================================
# Task 16: Claim Governance, Tier Tagging, and Artifact Archival
# ==============================================================================

# Source step text: detailed_task_list_20261111_144334.md, lines 803-870 (verbatim).
# Source configuration pins: STUDY_CONFIG P00, P05, P07, P18, P19, P20, P21, P36,
# P40, P41, P43, P45, P46 (study_revised_inputs_20261111_143538.md).
# Source contract: acceptance/acceptance_task_16.py (console's Contract-Author;
# SHA-256 a8d9b868930e1cbe7f7dae3fc6a8882c495290bc7062d64ba858693ee2faf0e6).

# -------------------------------------------------------------------------------------------------------------------------------
# Pinned constants (verbatim from the step text / the acceptance contract)
# -------------------------------------------------------------------------------------------------------------------------------

# The exact label the step text mandates for every permitted native-Sharpe occurrence.
SHARPE_MANDATED_LABEL: str = "reported native Sharpe, not a reproduction result; metric grain mismatch with swarm"

# The four class stamps for tooling, in the step's pinned order.
CLASS_STAMPS: Tuple[str, ...] = (
    "STRUCTURAL__[D]__verifiable_on_this_run",
    "COMPARISON__[S]__PRIVATE_ARTIFACTS_ONLY",
    "SENSITIVITY__[R]__NOT_A_PAPER_FINDING",
    "ILLUSTRATIVE__[I]",
)

# The prohibited-phrase tokens, verbatim from the step text, in pinned order.
PROHIBITED_PHRASE_TOKENS: Tuple[str, ...] = (
    "significant", "significantly", "p-value", "p =", "t-stat", "z-score",
    "confidence interval", "Sharpe", "annualised", "annually", "outperforms",
    "beats", "dominates", "exceeds", "ranked", "rank 1", "first", "better",
    "best", "gain", "profit", "alpha", "PnL", "OOS", "out-of-sample",
    "generalises", "generalises to", "on average",
)

# The pinned token universe as a frozen lookup for deterministic finding order.
PROHIBITED_TOKEN_RANK: Dict[str, int] = {token: rank for rank, token in enumerate(PROHIBITED_PHRASE_TOKENS)}


def _compile_prohibited_token_pattern(token: str) -> re.Pattern:
    """
    Compile one pinned prohibited token into a boundary-aware regex.

    Split the token on its single spaces, escape each piece literally
    and rejoin the pieces with a lenient whitespace run, so
    ``"confidence interval"`` also matches tab-separated forms while
    hyphens and ``"="`` are matched literally. The pattern is compiled
    case-insensitively and anchored by negative lookbehind and lookahead
    on word characters, so identifier-embedded occurrences such as
    ``pipeline_mean_sharpe`` are not flagged (underscore counts as a
    word character).

    Parameters
    ----------
    token : str
        The pinned token in its canonical spelling, for example
        ``"Sharpe"`` or ``"confidence interval"``.

    Returns
    -------
    re.Pattern
        The compiled case-insensitive, boundary-anchored pattern.
    """
    # Split the token on its single spaces so multi-word tokens generalise whitespace.
    parts: List[str] = token.split(" ")
    # Escape each word piece verbatim (hyphens, "=" etc. are matched literally).
    escaped: List[str] = [re.escape(piece) for piece in parts]
    # Rejoin the pieces with a lenient whitespace run, never zero-width overlap.
    body: str = r"\s+".join(escaped)
    # Anchor with word-char-aware boundaries instead of \b so hyphenated and underscored neighbours behave predictably.
    pattern: re.Pattern = re.compile(r"(?<![\w])" + body + r"(?![\w])", re.IGNORECASE)
    # Return the compiled pattern for the module-level registry below.
    return pattern

# Precompiled pattern registry, built exactly once at import (resource discipline).
PROHIBITED_TOKEN_PATTERNS: Tuple[Tuple[str, "re.Pattern"], ...] = tuple(
    (token, _compile_prohibited_token_pattern(token)) for token in PROHIBITED_PHRASE_TOKENS
)

# The seventeen pinned control identifiers, in pinned order.
EXPECTED_CONTROL_IDS: Tuple[int, ...] = tuple(range(1, 18))

# The admissible enumeration statuses, exactly as pinned by the step text.
ADMISSIBLE_CONTROL_STATUSES: Tuple[str, ...] = ("ENFORCED", "ESCALATED")

# The four-tier taxonomy and its total order (inadmissible < reconstruction < reproduction < replication).
TIER_ORDER: Dict[str, int] = {
    "inadmissible": 0,
    "reconstruction": 1,
    "reproduction": 2,
    "replication": 3,
}

# The tier vocabulary as a frozenset for O(1) membership tests.
TIER_VOCABULARY: frozenset = frozenset(TIER_ORDER)

# The four named decision classes, all requiring at least the reproduction tier.
DECISION_CLASS_MINIMUM_TIER: Dict[str, str] = {
    "capital_allocation": "reproduction",
    "model_deployment": "reproduction",
    "publication": "reproduction",
    "procurement": "reproduction",
}

# The twelve P45_DISCLOSURE_GAPS keys, verbatim and in the pinned order.
P45_GAP_KEYS: Tuple[str, ...] = (
    "fallback_rate_reported",
    "universe_constituents_disclosed",
    "is_screen_window_disclosed",
    "prompts_disclosed",
    "inference_settings_disclosed",
    "api_credentials_disclosed",
    "grammar_disclosed",
    "executed_commit_bound_to_report",
    "initial_pool_hash_available",
    "specialist_to_category_map_disclosed",
    "allocation_update_rule_disclosed",
    "prompt_revision_periodicity_disclosed",
)

# The twelve disclosed reproduction [R] choices with their pinned descriptor phrases.
REPRODUCTION_R_GAP_LINES: Tuple[Tuple[str, str], ...] = (
    ("masked_cell_semantics", "two distinct stock-days"),
    ("dedup_representative_rule", "lowest row_id"),
    ("dedup_canonicalization", "push_negation_through=False, no symmetry operators"),
    ("state_precedence", "unsupported > insufficient > scorable"),
    ("spot_check_tolerance", "1e-10 absolute"),
    ("freeze_identifier_format", "string-verbatim"),
    ("sd_definition", "ddof=1"),
    ("reference_formulas", "six formulas chosen as representatives"),
    ("nine_cell_satisfiability", "r9 satisfiable or not"),
    ("alternative_reference_implementation", "numpy-based"),
    ("retrieval_coherence_window_interpretation", "session-span"),
    ("algorithm", "SHA-256 as specified"),
)

# The eight redistributable artifacts, pinned order (seven pinned + the Task 15 addition).
REQUIRED_REDISTRIBUTABLE_ARTIFACTS: Tuple[str, ...] = (
    "frozen_1047_id_list_with_hash_and_freeze_timestamp",
    "80_file_hash_manifest_and_acquisition_instructions",
    "v1_and_v2_evaluator_versions_with_change_description",
    "ledger_generation_logic",
    "nine_designated_spot_check_cells_and_both_implementations",
    "six_reference_definitions_with_preregistration_timestamp",
    "disclosure_tables_with_UNDISCLOSED_cells_intact",
    "sensitivity-report-with-primary-comparison",
)

# The non-redistributable artifacts, pinned, verbatim.
NON_REDISTRIBUTABLE_ARTIFACTS: Tuple[str, ...] = (
    "raw_vendor_data",
    "proprietary_evaluator_internals",
)

# The twelve frozen governance booleans, exactly as pinned by the step text.
GOVERNANCE_DECLARATION_FLAGS: Dict[str, bool] = {
    "public_data_only": True,
    "live_trading_claim_permitted": False,
    "financial_advice_claim_permitted": False,
    "order_sending_permitted": False,
    "preserve_failed_records": True,
    "preserve_non_scorable_records": True,
    "model_labels_as_verified_identity": False,
    "ai_assistance_disclosed": True,
    "human_annotation_study_claimed": False,
    "independent_human_review_claimed": False,
    "numericals_computed_not_generated": True,
    "author_accountability_asserted": True,
}

# The four-gate stamp, exactly as pinned by the step text.
FOUR_GATE_STAMP: Dict[str, str] = {
    "G1": "inadmissible",
    "G2": "inadmissible",
    "G3": "inadmissible",
    "G4": "inadmissible",
    "REPRODUCTION_CITATION": "reproduction",
}

# The verbatim closing statement, copied byte-exactly from the step text.
CLOSING_STATEMENT: str = (
    "The reproduction reaches methodology-reproduction fidelity under the supplied "
    "inputs and result-reproduction only for the structural invariants enumerated in "
    "`P40_ACCEPTANCE_TARGETS_STRUCTURAL_INVARIANTS`. The study-specific values in "
    "`P41_ACCEPTANCE_TARGETS_STUDY_SPECIFIC` and the pending-instrument status of "
    "`P08_CLEAN_OOS_PROTOCOL_PARAMETERS` remain outside the reach of these inputs, "
    "and the reproduction must not present them as targets. Every surviving claim is "
    "stamped with the tier its evidence can bear. Any downstream decision — capital "
    "allocation, model deployment, publication, or procurement — that cites this "
    "study's historical counters or its outcome-filtered catalog must be stamped "
    "inadmissible at Gates 1 through 4 respectively; any decision citing the "
    "reproduction must inherit the methodology-reproduction tier and no higher."
)

# The macro grammar for \\tieredclaim{<tier>}{<body>} with non-nested arguments.
TIERED_CLAIM_MACRO_PATTERN: re.Pattern = re.compile(r"\\tieredclaim\{([^{}]*)\}\{([^{}]*)\}")

# The bare-numeric-literal detector: digits not glued to a word character or hyphen.
BARE_NUMERIC_LITERAL_PATTERN: re.Pattern = re.compile(
    r"(?<![\w-])\d+(?:\.\d+)?(?:[eE][+-]?\d+)?(?![\w])"
)

# The seventeen frozen anticipated-violation sentences, exactly one per control id (1..17).
# [R] reproduction references; stamped as the reproduction's own freeze, NOT paper text. Ratified
#     2026-11-12 by the operator's delegate after a one-to-one review against the pinned control
#     list (detailed_task_list line 808): each sentence is a direct instance of the inference its
#     control forbids, in the pinned order 1..17 (see residual_issues_20261112.txt). Each sentence is the canonical forbidden claim for its control class; each embeds no
#     prohibited literal by accident; the zero-instance scan over compliant study outputs must never
#     match them. The set is frozen: never regenerate from any text; change only by a new, documented
#     [R] adjudication. The order is the pinned control order 1..17.
ANTICIPATED_VIOLATION_PINS_TASK_16: Tuple[str, ...] = (
    # Control 1 (no architecture ranking) -- LaTeX context, forbidden-inference lists; detailed list control 1.
    "The architecture ranking of the two systems is the study's principal result.",
    # Control 2 (no profitability) -- manuscript, forbidden-inference lists; control 2.
    "The factor cohort is profitable and the edge survives transaction costs by a wide margin.",
    # Control 3 (no annualisation) -- manuscript; metric grain control 3.
    "The mean daily rank correlation annualises to a compound return over the study window.",
    # Control 4 (no trading-Sharpe label) -- manuscript; control 4.
    "The per-round maximum of 4.206 is the system's trading Sharpe ratio.",
    # Control 5 (no significance test on the aggregate) -- manuscript; control 5.
    "The improvement over the baseline is statistically significant at the one percent level.",
    # Control 6 (no OOS / generalisation) -- manuscript; control 6.
    "The system's out-of-sample performance generalises to the full universe beyond 2025.",
    # Control 7 (no incremental-agent-skill) -- manuscript; control 7.
    "The marginal specialist contribution demonstrates incremental agent skill over the baseline.",
    # Control 8 (no matched-sample outperformance) -- manuscript; control 8.
    "On matched samples the system outperforms the baseline at every horizon, without exception.",
    # Control 9 (no total-return) -- manuscript; control 9.
    "The factor cohort delivers total return to investors over the full window.",
    # Control 10 (no executable-PnL) -- manuscript; control 10.
    "The system's executable PnL after costs is eighteen percent over the window.",
    # Control 11 (no model-quality ranking) -- manuscript; control 11.
    "The model-quality ranking places the swarm decisively first, ahead of the baseline system.",
    # Control 12 (no backend-identity) -- manuscript; control 12.
    "The backend runtime serving these requests is DeepSeek-V4 Pro, verified from first request.",
    # Control 13 (no other-systems-implicated) -- manuscript; control 13.
    "Other systems and their backends exhibit the same failure mode with equal severity.",
    # Control 14 (no imputed zero-alpha) -- manuscript; control 14.
    "An imputed zero alpha was assigned to the missing outcome so the ledger could close.",
    # Control 15 (no missing outcome omitted from denominator) -- manuscript; control 15.
    "The record whose outcome was missing was simply omitted from the denominator.",
    # Control 16 (no residual reported as zero) -- manuscript; control 16.
    "The residual was reported as zero without any adjustment or qualification.",
    # Control 17 (no native/public metric concatenation) -- manuscript; control 17.
    "The native and public metrics were concatenated into one reported figure without comment.",
)
# The pin index by control id (1..17), computed once at import.
_ANTICIPATED_PIN_BY_CONTROL_ID_TASK_16: Dict[int, str] = {
    control_id: ANTICIPATED_VIOLATION_PINS_TASK_16[control_id - 1]
    for control_id in range(1, 18)
}

# The pinned enumeration-table row grammar (Markdown table, five columns).
ENUMERATION_TABLE_ROW_PATTERN: re.Pattern = re.compile(
    r"^\|\s*\\tieredclaim\{([^{}]+)\}\{(\d+)\}\s*\|\s*([^|]+?)\s*\|"
    r"\s*\\tieredclaim\{([^{}]+)\}\{([^{}]+)\}\s*\|"
    r"\s*\\tieredclaim\{([^{}]+)\}\{([^{}]+)\}\s*\|"
    r"\s*([A-Za-z_]+)\s*\|\s*$"
)

# -------------------------------------------------------------------------------------------------------------------------------
# Exception hierarchy (every halt is explicit and informative)
# -------------------------------------------------------------------------------------------------------------------------------

class Task16GovernanceError(RuntimeError):
    """
    Base halt for every governance violation of this module.

    Subclasses carry an informative message naming the violated pin, so
    a build-time failure is self-describing in the CI log and the
    governance register never ships degraded content silently.
    """

class ControlRegisterHaltError(Task16GovernanceError):
    """
    Raised when the seventeen-control witness register is unusable.

    Covers an incomplete (non-1..17) id set, an impermissible
    prohibited-phrase finding with no documented permitted use, and an
    enumerated anticipated-violation witness sentence that appears
    verbatim in the study output. A duplicated id raises ``ValueError``
    instead.
    """

class MacroStructureHaltError(Task16GovernanceError):
    """
    Raised when a ``tieredclaim`` macro is malformed.

    Signals an unknown tier argument or a whitespace-only body in a
    ``tieredclaim{<tier>}{<body>}`` instance.
    """

class TierStampHaltError(Task16GovernanceError):
    """
    Raised when the ClaimRecord/macro count assertion fails.

    The build-time assertion is
    ``|{claim_record claim_id ids}| == |{tieredclaim instances}|``. A
    mismatch means the number of distinct claim ids differs from the
    number of ``tieredclaim`` instances; it can be either a stamped
    claim with no ``ClaimRecord`` or a ``ClaimRecord`` with no stamped
    claim.
    """

class BareNumericLiteralHaltError(Task16GovernanceError):
    """
    Raised when a bare numeric literal sits outside a macro body.

    Detected by scanning the text with all ``tieredclaim`` spans masked
    out, so wrapped digits are not mistaken for bare ones.
    """

class DisclosureGapHaltError(Task16GovernanceError):
    """
    Raised when a P45 gap or reproduction choice is missing.

    Signals a disclosure-gap row or a reproduction ``[R]`` disclosure
    that is absent or mis-stamped in the study output.
    """

class DepositHaltError(Task16GovernanceError):
    """
    Raised when the redistributable deposit fails validation.

    Signals an invalid artifact name, a banned (non-redistributable)
    artifact, or a failed write into the deposit directory.
    """

# -------------------------------------------------------------------------------------------------------------------------------
# Shared validation helpers (single definition, reused everywhere)
# -------------------------------------------------------------------------------------------------------------------------------

def _validate_study_output_text_task_16(study_output_text: Any, *, caller: str) -> str:
    """
    Validate the study-output text argument as a string.

    Parameters
    ----------
    study_output_text : Any
        The candidate plain-text extraction of the study output.

    caller : str
        The caller name, quoted in the raised message for provenance.

    Returns
    -------
    str
        The validated text, unchanged.

    Raises
    ------
    TypeError
        If ``study_output_text`` is not a ``str``.
    """
    # Enforce the exact declared type at the boundary.
    if not isinstance(study_output_text, str):
        # Raise with the observed type spelled out, never a bare message.
        raise TypeError(f"{caller}: study_output_text must be str, got {type(study_output_text).__name__}")
    # Return the validated text unchanged (this helper never mutates content).
    return study_output_text

def _validate_whitelist_task_16(whitelist: Any, *, caller: str) -> Optional[Mapping[str, str]]:
    """
    Validate the optional whitelist of documented permitted uses.

    Parameters
    ----------
    whitelist : Any
        ``None`` (nothing is permitted) or a mapping of str to str,
        where each key is a documented permitted-use phrase and each
        value its justification.

    caller : str
        The caller name, quoted in the raised message for provenance.

    Returns
    -------
    Optional[Mapping[str, str]]
        The validated mapping, or ``None`` unchanged.

    Raises
    ------
    TypeError
        If ``whitelist`` is neither ``None`` nor a mapping, or if any
        key or value is not a ``str``.
    """
    # The whitelist is optional; None means "nothing is permitted".
    if whitelist is None:
        # Return None verbatim so the caller applies the default-deny rule.
        return None
    # Enforce the declared Mapping type.
    if not isinstance(whitelist, Mapping):
        # Raise with the observed type spelled out for diagnosis.
        raise TypeError(f"{caller}: whitelist must be a Mapping[str, str], got {type(whitelist).__name__}")
    # Walk every entry once and validate both sides of the mapping.
    for key, value in whitelist.items():
        # Keys must be the literal documented phrases (e.g. "Alpha-Factor").
        if not isinstance(key, str):
            # Raise naming the offending key type.
            raise TypeError(f"{caller}: whitelist keys must be str, got {type(key).__name__}")
        # Values must be the documented permitted-use justifications.
        if not isinstance(value, str):
            # Raise naming the offending value type.
            raise TypeError(f"{caller}: whitelist values must be str, got {type(value).__name__}")
    # Return the validated mapping.
    return whitelist

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1: The prose audit -- prohibited-phrase scanner and witness register
# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1a: Scanner occurrence classifier (the two permitted-use rules).
# -------------------------------------------------------------------------------------------------------------------------------

def _classify_occurrence_task_16(
    token: str,
    line: str,
    whitelist: Optional[Mapping[str, str]],
) -> str:
    """
    Classify one flagged occurrence as permitted or impermissible.

    Two permitted-use rules are applied to the occurrence, on the same
    line, in order:

    1. the mandated Sharpe label: an occurrence of the token
       ``"Sharpe"`` is permitted only when ``SHARPE_MANDATED_LABEL``
       appears on the same line (case-insensitive);
    2. the whitelist rule: the occurrence is permitted when any
       whitelist key is a case-insensitive substring of the same line.

    Parameters
    ----------
    token : str
        The pinned token that was matched, for example ``"Sharpe"``.

    line : str
        The full line of the study output carrying the occurrence.

    whitelist : Optional[Mapping[str, str]]
        The documented permitted uses, or ``None`` (default-deny).

    Returns
    -------
    str
        ``"PERMITTED"`` when a permitted-use rule fires, otherwise
        ``"IMPERMISSIBLE"``.

    Notes
    -----
    This helper never raises; it is the always-defined core of the
    scanner. An ``"IMPERMISSIBLE"`` result is returned to the immediate
    caller: ``scan_prohibited_phrases_task_16`` records it without
    raising, and ``build_governance_register_task_16`` halts the build
    when it finds one.
    """
    # Rule 1 -- the mandated Sharpe-label rule (case-insensitive label match on the same line).
    if token.lower() == "sharpe" and SHARPE_MANDATED_LABEL.lower() in line.lower():
        # The occurrence carries the mandated label; it documents a permitted native-Sharpe comparison.
        return "PERMITTED"
    # Rule 2 -- the whitelist rule; a whitelist key appearing anywhere on the line documents the permitted use.
    if whitelist:
        # Lowercase the line once for the substring scan (documented case-insensitivity).
        lowered_line: str = line.lower()
        # Check every documented permitted-use key against the line.
        for key in whitelist:
            # A key contained in the line documents the permitted use for that line.
            if key.lower() in lowered_line:
                # Defer to the documented use; the analyst is forced to justify, not to delete.
                return "PERMITTED"
    # No permitted-use rule fired; the occurrence remains IMPERMISSIBLE and must be justified or the build halts.
    return "IMPERMISSIBLE"

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1b: Static prohibited-phrase scanner (verbatim token list).
# -------------------------------------------------------------------------------------------------------------------------------

def scan_prohibited_phrases_task_16(
    study_output_text: str,
    whitelist: Mapping[str, str] | None = None,
) -> list[dict]:
    """
    Scan the study output's plain text for pinned prohibited tokens.

    Iterate the text line by line and, within each line, the pinned
    token registry in its pinned order; every regex match becomes one
    finding classified by :func:`_classify_occurrence_task_16`. Boundary
    validation runs once, and the scan itself never raises on an
    impermissible finding.

    Parameters
    ----------
    study_output_text : str
        The plain-text extraction of the study output to audit, for
        example the LaTeX source with macros intact or rendered prose.

    whitelist : Mapping[str, str] | None, optional
        Hand-curated mapping of permitted-use keys to justifications,
        for example ``{"Alpha-Factor": "domain noun in the title"}``.
        Default is ``None``, which applies the default-deny rule.

    Returns
    -------
    list[dict]
        One finding per occurrence, ordered by line number, then pinned
        token order, then match position. Each dict carries ``token``
        (the pinned token in its pinned spelling), ``line_number``
        (1-based, int), ``line`` (the raw line) and ``disposition``
        (``"PERMITTED"`` or ``"IMPERMISSIBLE"``).

    Raises
    ------
    TypeError
        If ``study_output_text`` is not a ``str``, or ``whitelist`` is
        not ``None`` / a ``Mapping[str, str]``.

    Notes
    -----
    The lexical scanner is a conservative over-approximation: its
    purpose is to force the analyst to justify each use, not to make the
    justice. It records ``"IMPERMISSIBLE"`` findings without raising;
    halting is left to the register builder.
    """
    # Validate the text argument once at the boundary (the scan never sees an invalid type).
    validated_text: str = _validate_study_output_text_task_16(study_output_text, caller="scan_prohibited_phrases_task_16")
    # Validate the optional whitelist mapping (default-deny when absent).
    validated_whitelist: Optional[Mapping[str, str]] = _validate_whitelist_task_16(whitelist, caller="scan_prohibited_phrases_task_16")
    # Accumulate findings in deterministic order: lines outer, pinned token order inner, match position innermost.
    findings: List[Dict[str, Any]] = []
    # Iterate the text line by line with a 1-based line number for the audit trail.
    for line_number, raw_line in enumerate(validated_text.splitlines(), start=1):
        # Strip nothing; the disposition rules inspect the raw line, and the audit wants the raw line.
        line: str = raw_line
        # Walk the pinned token registry in its pinned order for deterministic output.
        for token, pattern in PROHIBITED_TOKEN_PATTERNS:
            # Find every occurrence of this token on this line (finditer is position-ordered).
            for match in pattern.finditer(line):
                # Classify the occurrence under the two pinned permitted-use rules.
                disposition: str = _classify_occurrence_task_16(token, line, validated_whitelist)
                # Append the finding with the pinned key set the contract publishes.
                findings.append(
                    {
                        "token": token,
                        "line_number": line_number,
                        "line": line,
                        "disposition": disposition,
                    }
                )
    # Return the complete, deterministically ordered finding list (never raises on IMPERMISSIBLE findings).
    return findings

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 1c: The seventeen-control witness register.
# -------------------------------------------------------------------------------------------------------------------------------

def _validate_enumerated_control_task_16(
    entry: Any,
    *,
    index: int,
) -> Dict[str, Any]:
    """
    Validate and normalise one enumerated-control mapping.

    Parameters
    ----------
    entry : Any
        The candidate control mapping. It must carry ``control_id``
        (int in 1..17, booleans rejected), ``control_text``,
        ``witness_declined_ref``, ``witness_anticipated_violation_ref``,
        ``witness_declined`` and ``witness_anticipated_violation`` (all
        non-empty str), and ``status`` in {``"ENFORCED"``,
        ``"ESCALATED"``}.

    index : int
        The 0-based position of the entry in the caller's sequence, used
        in the raised messages.

    Returns
    -------
    Dict[str, Any]
        The normalised record with exactly those seven keys.
        ``control_text`` and the two ``*_ref`` fields
        (``witness_declined_ref`` and
        ``witness_anticipated_violation_ref``) are stripped; the two
        witness sentences are kept verbatim.

    Raises
    ------
    TypeError
        If ``entry`` is not a mapping.

    ValueError
        If ``control_id`` is not a non-boolean ``int`` or lies outside
        1..17, if any textual field is missing, empty or not a ``str``,
        or if ``status`` is outside the two-value domain.
    """
    # Enforce the mapping type at the boundary.
    if not isinstance(entry, Mapping):
        # Raise naming the position and the observed type.
        raise TypeError(f"enumerated_controls[{index}] must be a Mapping, got {type(entry).__name__}")
    # Read the raw identifier; booleans are rejected explicitly (bool is an int subclass).
    raw_id: Any = entry.get("control_id")
    # Type-check the identifier strictly.
    if not isinstance(raw_id, int) or isinstance(raw_id, bool):
        # Raise naming the offending value.
        raise ValueError(f"enumerated_controls[{index}].control_id must be an int, got {raw_id!r}")
    # Domain-check the identifier against the pinned 1..17 range.
    if raw_id not in EXPECTED_CONTROL_IDS:
        # Raise naming the out-of-range value.
        raise ValueError(f"enumerated_controls[{index}].control_id {raw_id} outside the pinned 1..17 range")
    # Extract and type-check the control text.
    control_text: Any = entry.get("control_text")
    # Enforce a non-empty string.
    if not isinstance(control_text, str) or not control_text.strip():
        # Raise naming the position.
        raise ValueError(f"enumerated_controls[{index}].control_text must be a non-empty str")
    # Validate the two witness references (non-empty strings, byte-exact).
    declined_ref: Any = entry.get("witness_declined_ref")
    # Enforce a non-empty string.
    if not isinstance(declined_ref, str) or not declined_ref.strip():
        # Raise naming the position and the field.
        raise ValueError(f"enumerated_controls[{index}].witness_declined_ref must be a non-empty str")
    # Validate the anticipated-violation reference.
    violation_ref: Any = entry.get("witness_anticipated_violation_ref")
    # Enforce a non-empty string.
    if not isinstance(violation_ref, str) or not violation_ref.strip():
        # Raise naming the position and the field.
        raise ValueError(f"enumerated_controls[{index}].witness_anticipated_violation_ref must be a non-empty str")
    # Validate the frozen declined-inference witness sentence.
    declined_text: Any = entry.get("witness_declined")
    # Enforce a non-empty string.
    if not isinstance(declined_text, str) or not declined_text.strip():
        # Raise naming the position and the field.
        raise ValueError(f"enumerated_controls[{index}].witness_declined must be a non-empty str")
    # Validate the frozen anticipated-violation witness sentence.
    violation_text: Any = entry.get("witness_anticipated_violation")
    # Enforce a non-empty string.
    if not isinstance(violation_text, str) or not violation_text.strip():
        # Raise naming the position and the field.
        raise ValueError(f"enumerated_controls[{index}].witness_anticipated_violation must be a non-empty str")
    # Validate the enumeration status against the pinned two-value domain.
    status: Any = entry.get("status")
    # Enforce the {ENFORCED, ESCALATED} domain verbatim (the step's pin).
    if not isinstance(status, str) or status not in ADMISSIBLE_CONTROL_STATUSES:
        # Raise naming the offending status.
        raise ValueError(
            f"enumerated_controls[{index}].status must be one of {ADMISSIBLE_CONTROL_STATUSES}, got {status!r}"
        )
    # Return the normalised, seven-key control record.
    return {
        "control_id": raw_id,
        "control_text": control_text.strip(),
        "witness_declined_ref": declined_ref.strip(),
        "witness_anticipated_violation_ref": violation_ref.strip(),
        "witness_declined": declined_text,
        "witness_anticipated_violation": violation_text,
        "status": status,
    }

def build_governance_register_task_16(
    study_output_text: str,
    enumerated_controls: Sequence[dict],
    whitelist: Mapping[str, str] | None = None,
) -> dict:
    """
    Build the frozen per-control witness register, halting on defects.

    The steps are: validate the text and the whitelist; normalise every
    enumerated control, rejecting duplicate ids; require the id set to
    be exactly 1..17; run the prohibited-phrase scanner and reject any
    ``"IMPERMISSIBLE"`` finding; reject the text if it contains any
    enumerated anticipated-violation witness sentence verbatim; run the
    module's pin-integrity probe; emit one INFO log line with counts
    only; return the register.

    Parameters
    ----------
    study_output_text : str
        The plain-text extraction of the study output under audit.

    enumerated_controls : Sequence[dict]
        The seventeen enumerated controls, one mapping each, with the
        keys ``control_id``, ``control_text``, ``witness_declined_ref``,
        ``witness_anticipated_violation_ref``, ``witness_declined``,
        ``witness_anticipated_violation`` and ``status``.

    whitelist : Mapping[str, str] | None, optional
        Documented permitted uses handed to the scanner. Default is
        ``None`` (default-deny).

    Returns
    -------
    dict
        Exactly three keys: ``controls`` (the seventeen normalised
        records sorted by ``control_id`` ascending),
        ``enumeration_complete`` (always ``True``) and
        ``violations_found`` (always the empty list, since any violation
        halts).

    Raises
    ------
    TypeError
        If ``study_output_text`` is not a ``str``, ``whitelist`` is not
        ``None`` / a ``Mapping[str, str]``, ``enumerated_controls`` is
        not a ``Sequence``, or an entry is not a mapping.

    ValueError
        If a control entry is invalid or two entries share a
        ``control_id`` (duplicates raise ``ValueError``, not a halt).

    ControlRegisterHaltError
        If the id set is not exactly 1..17, if the scanner reports any
        ``"IMPERMISSIBLE"`` finding, or if any enumerated
        anticipated-violation witness sentence occurs verbatim in the
        text.

    Notes
    -----
    An unenumerated control is treated as unenforced, so the register is
    an all-or-nothing artifact. The pin-integrity probe invoked before
    returning fails the build rather than warning.
    """
    # Validate the text argument once at the boundary.
    validated_text: str = _validate_study_output_text_task_16(study_output_text, caller="build_governance_register_task_16")
    # Validate the optional whitelist (shared validator, default-deny).
    validated_whitelist: Optional[Mapping[str, str]] = _validate_whitelist_task_16(whitelist, caller="build_governance_register_task_16")
    # Enforce the declared sequence type for the controls argument.
    if not isinstance(enumerated_controls, Sequence):
        # Raise naming the observed type.
        raise TypeError(f"enumerated_controls must be a Sequence[dict], got {type(enumerated_controls).__name__}")
    # Normalise and validate every control entry, tracking duplicate ids as we go.
    normalised: List[Dict[str, Any]] = []
    # Accumulate the ids in encounter order to detect duplicates.
    seen_ids: List[int] = []
    # Walk the caller's sequence once.
    for index, entry in enumerate(enumerated_controls):
        # Validate and normalise the entry (raises on any defect).
        record: Dict[str, Any] = _validate_enumerated_control_task_16(entry, index=index)
        # Detect a duplicated identifier before anything else (a duplicate is a governance failure).
        if record["control_id"] in seen_ids:
            # Raise naming the duplicated id.
            raise ValueError(f"enumerated_controls: duplicate control_id {record['control_id']}")
        # Record the id for the duplicate check of later entries.
        seen_ids.append(record["control_id"])
        # Keep the normalised record for the completeness check and the output.
        normalised.append(record)
    # Completeness: the id set must be exactly the pinned 1..17 (the step's pin: unenumerated controls are unenforced).
    missing_ids: List[int] = [cid for cid in EXPECTED_CONTROL_IDS if cid not in seen_ids]
    # Compute the surplus ids (impossible after the per-entry domain check, kept for totality).
    surplus_ids: List[int] = [cid for cid in seen_ids if cid not in EXPECTED_CONTROL_IDS]
    # Anything short of exact equality halts the build.
    if missing_ids or surplus_ids:
        # Raise with the exact defect spelled out.
        raise ControlRegisterHaltError(
            f"seventeen-control register: missing ids {missing_ids}, surplus ids {surplus_ids}; "
            "a control not enumerated has not been enforced"
        )
    # Re-index the records by control_id ascending for a deterministic, register-shaped output.
    ordered: List[Dict[str, Any]] = sorted(normalised, key=lambda record: record["control_id"])
    # Run the static scanner over the text with the same whitelist and halt on any IMPERMISSIBLE finding.
    findings: List[Dict[str, Any]] = scan_prohibited_phrases_task_16(validated_text, validated_whitelist)
    # Collect the IMPERMISSIBLE findings (the step's halt condition).
    impermissible: List[Dict[str, Any]] = [finding for finding in findings if finding["disposition"] == "IMPERMISSIBLE"]
    # Any impermissible occurrence halts the build (permitted uses must be documented in the whitelist).
    if impermissible:
        # Format the first offenders for a self-describing failure.
        preview: str = "; ".join(
            f"line {f['line_number']}: {f['token']!r}" for f in impermissible[:8]
        )
        # Raise naming the count and the preview.
        raise ControlRegisterHaltError(
            f"prohibited-phrase scanner: {len(impermissible)} IMPERMISSIBLE occurrence(s) with no documented permitted use: {preview}"
        )
    # Scan the text for any enumerated anticipated-violation witness sentence occurring verbatim.
    violations: List[Dict[str, Any]] = []
    # Walk the ordered register once for the verbatim check.
    for record in ordered:
        # Read the frozen anticipated-violation witness sentence.
        sentence: str = record["witness_anticipated_violation"]
        # A verbatim occurrence of the anticipated-violation form in the text is a violation.
        if sentence in validated_text:
            # Record the violation for the halt message below.
            violations.append(
                {
                    "control_id": record["control_id"],
                    "witness_anticipated_violation_ref": record["witness_anticipated_violation_ref"],
                    "sentence": sentence,
                }
            )
    # Any anticipated-violation instance in the study output halts the build.
    if violations:
        # Format the offending controls for the halt message.
        offenders: str = ", ".join(str(v["control_id"]) for v in violations)
        # Raise naming the count and the offending control ids.
        raise ControlRegisterHaltError(
            f"anticipated-violation witness sentence(s) found verbatim in the study output for control(s): {offenders}"
        )
    # The standing pin-integrity assertion (the second lock on the vacuity channel; fails the build, never warns).
    _pins_coverage_probe_task_16()
    # Assemble the frozen register with exactly the three pinned keys.
    register: Dict[str, Any] = {
        "controls": ordered,
        "enumeration_complete": True,
        "violations_found": [],
    }
    # Emit one INFO line for the CI log (counts only; never content).
    logger.info("governance register built: %d controls enumerated, zero violations", len(ordered))
    # Return the frozen register.
    return register

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2: Tier stamping and the disclosure-gap audit
# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2a: Four-tier taxonomy assignment.
# -------------------------------------------------------------------------------------------------------------------------------

def assign_claim_tier_task_16(
    claim_ref: str,
    *,
    target_id: str | None = None,
    provenance_tag: str | None = None,
    gate_failure: str | None = None,
    private_artifacts_recovered: bool = False,
) -> str:
    """
    Assign the evidentiary tier of one claim under the pinned taxonomy.

    Determinants are applied in priority order, after stripping and
    upper-casing:

    1. a supplied ``gate_failure`` yields ``"inadmissible"``
       unconditionally;
    2. ``target_id`` in {``"T1"``, ``"T2"``} yields ``"replication"``
       only when ``private_artifacts_recovered`` is ``True``, else
       ``"inadmissible"``;
    3. ``target_id`` ``"T3"`` yields ``"reproduction"``;
    4. ``target_id`` ``"T4"`` yields ``"reconstruction"``;
    5. a ``provenance_tag`` of ``"[I]"`` or ``"[R]"`` yields
       ``"reconstruction"``.

    Parameters
    ----------
    claim_ref : str
        The claim identifier or human-facing reference, for example
        ``"p:pipeline_mean_sharpe_over_kept"``.

    target_id : str | None, optional
        The ``P00`` target class, one of ``"T1"``..``"T4"``. Default is
        ``None``.

    provenance_tag : str | None, optional
        The provenance tag carried by the claim, one of ``"[S]"``,
        ``"[D]"``, ``"[R]"``, ``"[I]"``, ``"[REVISED]"``. Default is
        ``None``.

    gate_failure : str | None, optional
        The failed four-gate identifier, one of ``"G1"``..``"G4"``.
        Default is ``None``.

    private_artifacts_recovered : bool, optional
        Whether the private per-record artifacts needed for replication
        were recovered. Default is ``False``.

    Returns
    -------
    str
        Exactly one of ``"replication"``, ``"reproduction"``,
        ``"reconstruction"`` or ``"inadmissible"``.

    Raises
    ------
    TypeError
        If ``private_artifacts_recovered`` is not a ``bool``, or if any
        supplied ``gate_failure``, ``target_id`` or ``provenance_tag``
        is not a ``str``.

    ValueError
        If ``claim_ref`` is not a non-empty ``str``, if a supplied
        keyword argument is outside its domain, or if no determinant
        fires (no tier is ever guessed).
    """
    # Validate the claim reference (non-empty, stripped).
    if not isinstance(claim_ref, str) or not claim_ref.strip():
        # Raise naming the observed value.
        raise ValueError(f"claim_ref must be a non-empty str, got {claim_ref!r}")
    # Validate the keyword-only boolean strictly (bool subclass ints are rejected).
    if type(private_artifacts_recovered) is not bool:
        # Raise naming the observed type.
        raise TypeError(f"private_artifacts_recovered must be bool, got {type(private_artifacts_recovered).__name__}")
    # Normalise and validate the gate identifier.
    normalised_gate: Optional[str] = None
    # Normalise only when supplied (absence is admissible).
    if gate_failure is not None:
        # Type-check the raw argument.
        if not isinstance(gate_failure, str):
            # Raise naming the observed type.
            raise TypeError(f"gate_failure must be a str in {{'G1','G2','G3','G4'}}, got {type(gate_failure).__name__}")
        # Normalise to uppercase for a case-insensitive comparison.
        normalised_gate = gate_failure.strip().upper()
        # Domain-check against the four pinned gate identifiers.
        if normalised_gate not in ("G1", "G2", "G3", "G4"):
            # Raise naming the offending value.
            raise ValueError(f"gate_failure must be one of 'G1','G2','G3','G4', got {gate_failure!r}")
    # Normalise and validate the target identifier.
    normalised_target: Optional[str] = None
    # Normalise only when supplied.
    if target_id is not None:
        # Type-check the raw argument.
        if not isinstance(target_id, str):
            # Raise naming the observed type.
            raise TypeError(f"target_id must be a str in {{'T1','T2','T3','T4'}}, got {type(target_id).__name__}")
        # Normalise to uppercase for the comparison.
        normalised_target = target_id.strip().upper()
        # Domain-check against the four pinned targets.
        if normalised_target not in ("T1", "T2", "T3", "T4"):
            # Raise naming the offending value.
            raise ValueError(f"target_id must be one of 'T1','T2','T3','T4', got {target_id!r}")
    # Normalise and validate the provenance tag.
    normalised_tag: Optional[str] = None
    # Normalise only when supplied.
    if provenance_tag is not None:
        # Type-check the raw argument.
        if not isinstance(provenance_tag, str):
            # Raise naming the observed type.
            raise TypeError(f"provenance_tag must be a str, got {type(provenance_tag).__name__}")
        # Normalise to uppercase and classify the tag shape.
        normalised_tag = provenance_tag.strip().upper()
        # The tag vocabulary is the pinned set; anything else is an input defect.
        if normalised_tag not in ("[S]", "[D]", "[R]", "[I]", "[REVISED]"):
            # Raise naming the offending value.
            raise ValueError(f"provenance_tag must be one of '[S]','[D]','[R]','[I]','[REVISED]', got {provenance_tag!r}")
    # Rule 1 (pinned): any gate failure yields inadmissible, unconditionally.
    if normalised_gate is not None:
        # The claim failed a named gate; its tier is inadmissible at that gate.
        return "inadmissible"
    # Rule 2 (pinned): T1/T2 are replication only when the private artifacts are recovered.
    if normalised_target in ("T1", "T2"):
        # Apply the recovery condition exactly as the step text pins it.
        return "replication" if private_artifacts_recovered else "inadmissible"
    # Rule 3 (pinned): T3 inherits the methodology-reproduction tier.
    if normalised_target == "T3":
        # The T3 reanalysis run is a reproduction under re-specified public inputs.
        return "reproduction"
    # Rule 4 (pinned): T4 is reconstruction only.
    if normalised_target == "T4":
        # The historical run reconstruction is impossible; only reconstruction is honest.
        return "reconstruction"
    # Rule 5 (pinned): [I]-tagged operators and [R]-tagged variants are reconstruction.
    if normalised_tag in ("[I]", "[R]"):
        # Names are illustrative or variants are re-specified; neither is a paper finding.
        return "reconstruction"
    # No determinant fired; refusing to guess is the only honest move.
    raise ValueError(
        f"assign_claim_tier_task_16: insufficient determinants for claim {claim_ref!r}; "
        "supply gate_failure, target_id, or an [I]/[R] provenance_tag"
    )

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2b: Macro parsing, the bijection assertion and the bare-literal halt.
# -------------------------------------------------------------------------------------------------------------------------------

def _find_tiered_claim_macros_task_16(
    text: str,
) -> List[Tuple[str, str, int, int]]:
    """
    Locate every ``tieredclaim{<tier>}{<body>}`` macro in the text.

    Parameters
    ----------
    text : str
        The study-output text to parse.

    Returns
    -------
    List[Tuple[str, str, int, int]]
        One tuple per macro in order of appearance: ``(tier, body,
        start, end)``, where ``tier`` is the stripped and lowercased
        tier argument, ``body`` is the verbatim body, and
        ``start``/``end`` are character offsets into ``text`` (end
        exclusive).

    Raises
    ------
    TypeError
        If ``text`` is not a ``str``.

    MacroStructureHaltError
        If a macro carries a tier outside ``TIER_VOCABULARY`` or a
        whitespace-only body.
    """
    # Validate the text defensively even though callers already validated it.
    if not isinstance(text, str):
        # Raise naming the observed type.
        raise TypeError(f"_find_tiered_claim_macros_task_16: text must be str, got {type(text).__name__}")
    # Accumulate the macros with their character spans.
    macros: List[Tuple[str, str, int, int]] = []
    # Walk the precompiled macro grammar once; finditer is position-ordered.
    for match in TIERED_CLAIM_MACRO_PATTERN.finditer(text):
        # Extract the tier argument verbatim from the first group.
        tier_raw: str = match.group(1)
        # Extract the body argument verbatim from the second group.
        body_raw: str = match.group(2)
        # Normalise the tier for the vocabulary check (the pinned vocabulary is lowercase).
        tier: str = tier_raw.strip().lower()
        # Domain-check the tier against the four-tier vocabulary exactly.
        if tier not in TIER_VOCABULARY:
            # Raise naming the offending macro tier.
            raise MacroStructureHaltError(
                f"\\tieredclaim macro with unknown tier {tier_raw!r} (expected one of {sorted(TIER_VOCABULARY)})"
            )
        # The body must carry at least one non-whitespace character (an empty wrapped claim is a defect).
        if not body_raw.strip():
            # Raise naming the span of the defect.
            raise MacroStructureHaltError(
                f"\\tieredclaim{{{tier_raw}}}{{}} macro with empty body at offset {match.start()}"
            )
        # Record the macro with its span for masking and counting.
        macros.append((tier, body_raw, match.start(), match.end()))
    # Return the ordered macro list.
    return macros

def _mask_macro_spans_task_16(text: str, macros: Sequence[Tuple[str, str, int, int]]) -> str:
    """
    Replace macro spans with spaces, preserving length and newlines.

    Parameters
    ----------
    text : str
        The original study-output text.

    macros : Sequence[Tuple[str, str, int, int]]
        The macro spans, as produced by
        :func:`_find_tiered_claim_macros_task_16`; only the two offset
        elements are used.

    Returns
    -------
    str
        A string of identical length and line count in which every macro
        span is blanked with single spaces (newline characters inside a
        span are preserved), so a bare-literal scan cannot see digits
        wrapped by a macro.
    """
    # Materialise the text as a mutable character list (offsets unchanged by equal-length replacement).
    characters: List[str] = list(text)
    # Blank every macro span, keeping newlines untouched (spans never contain raw newlines here, but be safe).
    for _, _, start, end in macros:
        # Walk the span and blank each character except a newline.
        for offset in range(start, end):
            # Preserve newlines so downstream line accounting stays exact.
            if characters[offset] != "\n":
                # Blank the character with a single space.
                characters[offset] = " "
    # Reassemble the masked text.
    return "".join(characters)

def _validate_claim_records_task_16(
    claim_records: Sequence[dict],
    *,
    caller: str,
) -> List[str]:
    """
    Validate the claim records and return their stripped claim ids.

    Parameters
    ----------
    claim_records : Sequence[dict]
        The candidate records; each must be a mapping with a non-empty
        ``claim_id`` string.

    caller : str
        The caller name, quoted in the raised messages for provenance.

    Returns
    -------
    List[str]
        The stripped ``claim_id`` of every record in caller order;
        duplicates are preserved.

    Raises
    ------
    TypeError
        If ``claim_records`` is not a ``Sequence``, or an entry is not a
        mapping.

    ValueError
        If an entry's ``claim_id`` is missing, empty or not a ``str``.
    """
    # Enforce the sequence type at the boundary.
    if not isinstance(claim_records, Sequence):
        # Raise naming the observed type.
        raise TypeError(f"{caller}: claim_records must be a Sequence[dict], got {type(claim_records).__name__}")
    # Accumulate the ids in order.
    ids: List[str] = []
    # Walk the sequence once for validation.
    for index, record in enumerate(claim_records):
        # Enforce the mapping type of the entry.
        if not isinstance(record, Mapping):
            # Raise naming the position and the observed type.
            raise TypeError(f"{caller}: claim_records[{index}] must be a Mapping, got {type(record).__name__}")
        # Read the id; missing key and non-str are both defects.
        claim_id: Any = record.get("claim_id")
        # Validate the id as a non-empty string.
        if not isinstance(claim_id, str) or not claim_id.strip():
            # Raise naming the position and the observed value.
            raise ValueError(f"{caller}: claim_records[{index}].claim_id must be a non-empty str, got {claim_id!r}")
        # Record the id (duplicates are preserved; the set assertion counts them).
        ids.append(claim_id.strip())
    # Return the id list.
    return ids

def verify_tier_stamps_task_16(
    study_output_text: str,
    claim_records: Sequence[dict],
) -> dict:
    """
    Enforce the macro/record bijection and the bare-literal halt.

    The steps are: validate the text and the records; locate and
    validate every ``tieredclaim`` macro; mask the macro spans; scan the
    masked text for bare numeric literals and halt on any hit; assert
    ``|{claim_record claim_id ids}| == |{tieredclaim instances}|``; then
    detect which of the pinned class stamps occur in the text.

    Parameters
    ----------
    study_output_text : str
        The study-output text under audit.

    claim_records : Sequence[dict]
        ClaimRecord-shaped dicts, each carrying a non-empty
        ``claim_id``. The set of distinct ids is compared with the
        macro count.

    Returns
    -------
    dict
        Exactly three keys: ``tiered_instances`` (int, number of
        ``tieredclaim`` macros found), ``claim_record_ids`` (int,
        number of distinct claim ids) and ``class_stamps_present``
        (the pinned class stamps found in the text, in pinned order).

    Raises
    ------
    TypeError
        If ``study_output_text`` is not a ``str``, ``claim_records`` is
        not a ``Sequence``, or an entry is not a mapping.

    ValueError
        If a record's ``claim_id`` is missing, empty or not a ``str``.

    MacroStructureHaltError
        If a macro carries an unknown tier or an empty body.

    BareNumericLiteralHaltError
        If any bare numeric literal occurs outside a macro body; the
        message names the literals with their 1-based line numbers.

    TierStampHaltError
        If the number of distinct claim ids differs from the number of
        ``tieredclaim`` instances; the check is cardinality-only and
        does not test which side is larger.
    """
    # Validate the text once at the boundary.
    validated_text: str = _validate_study_output_text_task_16(study_output_text, caller="verify_tier_stamps_task_16")
    # Validate the records and collect their ids (order preserved, duplicates kept).
    record_ids: List[str] = _validate_claim_records_task_16(claim_records, caller="verify_tier_stamps_task_16")
    # Locate and validate every macro in the text.
    macros: List[Tuple[str, str, int, int]] = _find_tiered_claim_macros_task_16(validated_text)
    # Re-read of the distinct id set (the assertion compares cardinalities, not order).
    distinct_ids: set = set(record_ids)
    # Mask macro spans so the bare-literal scan cannot see wrapped digits.
    masked_text: str = _mask_macro_spans_task_16(validated_text, macros)
    # Collect offending bare literals with line numbers.
    offenders: List[Tuple[int, str]] = []
    # Walk the masked text line by line.
    for line_number, masked_line in enumerate(masked_text.splitlines(), start=1):
        # Find every bare numeric literal on this line.
        for match in BARE_NUMERIC_LITERAL_PATTERN.finditer(masked_line):
            # Keep the literal text for the halt message.
            offenders.append((line_number, match.group(0)))
    # Any bare literal outside a macro halts the build.
    if offenders:
        # Format the first offenders for a self-describing halt message.
        preview: str = "; ".join(f"line {line}: {literal!r}" for line, literal in offenders[:8])
        # Raise naming the count and the preview.
        raise BareNumericLiteralHaltError(
            f"bare numeric literal(s) outside a \\tieredclaim body: {preview}"
        )
    # The build-time assertion: |{claim_record ids}| == |{\\tieredclaim instances}|.
    if len(distinct_ids) != len(macros):
        # Raise naming both cardinalities and the repair direction.
        raise TierStampHaltError(
            f"build-time assertion failed: |{{claim_record claim_id ids}}| = {len(distinct_ids)} != "
            f"|{{\\tieredclaim instances}}| = {len(macros)}; a claim in the study output with no ClaimRecord "
            "is a governance failure"
        )
    # Detect the four class stamps in the text (exact substring presence, pinned order).
    class_stamps_present: List[str] = [stamp for stamp in CLASS_STAMPS if stamp in validated_text]
    # Return exactly the three pinned keys.
    return {
        "tiered_instances": len(macros),
        "claim_record_ids": len(distinct_ids),
        "class_stamps_present": class_stamps_present,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 2c: The disclosure-gap audit (twelve P45 gaps + twelve [R] choices).
# -------------------------------------------------------------------------------------------------------------------------------

def audit_disclosure_gaps_task_16(
    study_output_text: str,
    study_config: Mapping,
) -> dict:
    """
    Audit the disclosure table and the reproduction's own gaps.

    Scan the study-output text for the twelve pinned
    ``P45_DISCLOSURE_GAPS`` keys and the twelve disclosed reproduction
    ``[R]`` choices. Every line naming a P45 key must carry a literal
    ``False`` token and no ``True`` token, and the key must appear at
    least once. Each ``[R]`` choice requires both its key and its
    pinned descriptor phrase to appear verbatim. When ``study_config``
    holds a ``P45_DISCLOSURE_GAPS`` mapping, every pinned key must map
    to ``False`` (checked type-strictly, so ``0`` and ``None`` fail).

    Parameters
    ----------
    study_output_text : str
        Plain-text extraction of the study output carrying the public
        disclosure table and the disclosed reproduction choices.
    study_config : Mapping
        Study configuration; its optional ``P45_DISCLOSURE_GAPS``
        section is cross-checked against the pinned keys.

    Returns
    -------
    dict
        ``p45_status``: the twelve pinned keys mapped to ``False``, in
        pinned order. ``missing_p45``: always ``[]`` on success.
        ``missing_r_gaps``: always ``[]`` on success.

    Raises
    ------
    TypeError
        If ``study_output_text`` is not a ``str``, or
        ``study_config`` is not a ``Mapping``.
    DisclosureGapHaltError
        If a P45 key is absent from the text, a line naming a P45 key
        carries a ``True`` token or no ``False`` token, an ``[R]`` key
        or descriptor phrase is absent, or the config's own P45
        section is present but a pinned key is not ``False``.
    """
    # Validate the text once at the boundary.
    validated_text: str = _validate_study_output_text_task_16(study_output_text, caller="audit_disclosure_gaps_task_16")
    # Enforce the mapping type for the config.
    if not isinstance(study_config, Mapping):
        # Raise naming the observed type.
        raise TypeError(f"audit_disclosure_gaps_task_16: study_config must be a Mapping, got {type(study_config).__name__}")
    # Cross-check the config's own P45 section when present (honesty travels both directions).
    config_section: Any = study_config.get("P45_DISCLOSURE_GAPS")
    # Only when the section is a mapping do we validate it.
    if isinstance(config_section, Mapping):
        # Every pinned key must be present with the value False.
        for gap_key in P45_GAP_KEYS:
            # Read the config's value for the pinned key.
            config_value: Any = config_section.get(gap_key)
            # Type-strict False check (0 and None are not False).
            if config_value is not False:
                # Raise naming the offending key and value.
                raise DisclosureGapHaltError(
                    f"study_config.P45_DISCLOSURE_GAPS[{gap_key!r}] must be False, got {config_value!r}"
                )
    # Pre-split the text once for the line scans below.
    text_lines: List[str] = validated_text.splitlines()
    # Accumulate the twelve P45 statuses keyed by the pinned key order.
    p45_status: Dict[str, bool] = {}
    # Collect absent P45 keys (all of them; the halt message lists every one).
    missing_p45: List[str] = []
    # Walk the twelve pinned keys in pinned order.
    for gap_key in P45_GAP_KEYS:
        # Find every line mentioning the key (the table is one row per gap).
        occurrences: List[Tuple[int, str]] = [
            (line_number, line) for line_number, line in enumerate(text_lines, start=1) if gap_key in line
        ]
        # Absence of the key anywhere in the text is a halt condition.
        if not occurrences:
            # Record the absence; the raise happens after the loop.
            missing_p45.append(gap_key)
            # Skip the status checks; there is nothing to verify.
            continue
        # Verify the status token on every occurrence line (a True or unverifiable cell halts).
        for line_number, line in occurrences:
            # Detect a literal True status token on this line.
            has_true: bool = re.search(r"\bTrue\b", line) is not None
            # Detect a literal False status token on this line.
            has_false: bool = re.search(r"\bFalse\b", line) is not None
            # A True cell is an impermissible disclosure status.
            if has_true:
                # Raise naming the key, the line and the status.
                raise DisclosureGapHaltError(
                    f"P45 gap {gap_key!r} carries status True on line {line_number}; the pin requires False "
                    "(reproduced verbatim, never inferred)"
                )
            # A line without an on-line False cannot be verified to carry the pinned status.
            if not has_false:
                # Raise naming the key and the line.
                raise DisclosureGapHaltError(
                    f"P45 gap {gap_key!r} on line {line_number} carries no verifiable False status token"
                )
        # Record the verified status exactly as the pin demands.
        p45_status[gap_key] = False
    # Any absent P45 key halts the audit.
    if missing_p45:
        # Raise listing every absent key.
        raise DisclosureGapHaltError(f"P45 disclosure gaps absent from the study output: {missing_p45}")
    # Collect absent reproduction [R] choices (key or descriptor phrase missing).
    missing_r_gaps: List[str] = []
    # Walk the twelve pinned [R] choices in pinned order.
    for gap_key, descriptor in REPRODUCTION_R_GAP_LINES:
        # The key must appear verbatim (it is the disclosure table's anchor).
        key_ok: bool = gap_key in validated_text
        # The descriptor phrase must appear verbatim (it is the disclosed content).
        descriptor_ok: bool = descriptor in validated_text
        # Both are required; record the absence if either fails.
        if not (key_ok and descriptor_ok):
            # Record the absent choice; the raise happens after the loop.
            missing_r_gaps.append(gap_key)
    # Any absent [R] choice halts the audit.
    if missing_r_gaps:
        # Raise listing every absent choice.
        raise DisclosureGapHaltError(f"reproduction [R] choices absent from the study output: {missing_r_gaps}")
    # Return exactly the three pinned keys.
    return {
        "p45_status": p45_status,
        "missing_p45": [],
        "missing_r_gaps": [],
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3: The final governance deposit and declaration
# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3a: The redistributable deposit (MANIFEST.json + self hash).
# -------------------------------------------------------------------------------------------------------------------------------

def _validate_artifact_name_task_16(name: str, *, banned_names: Tuple[str, ...]) -> str:
    """
    Validate one artifact name for flat, traversal-proof writes.

    Reject, in order: a non-string name; a blank name; a name longer
    than 255 characters; a name containing ``/``, a backslash, ``:``
    or ``..``; the reserved ``MANIFEST.json`` (case-insensitive); and
    any name equal, case-insensitively, to an entry of ``banned_names``.

    Parameters
    ----------
    name : str
        Candidate artifact filename.
    banned_names : Tuple[str, ...]
        Non-redistributable artifact names, matched case-insensitively.

    Returns
    -------
    str
        The validated name, unchanged.

    Raises
    ------
    TypeError
        If ``name`` is not a ``str``.
    DepositHaltError
        If the name is empty or whitespace-only, longer than 255
        characters, carries a path separator, colon or traversal
        token, is the reserved manifest name, or matches a banned name.
        A banned artifact is refused before any write.
    """
    # Enforce the declared type.
    if not isinstance(name, str):
        # Raise naming the observed type.
        raise TypeError(f"artifact name must be a str, got {type(name).__name__}")
    # Reject the empty name outright.
    if not name.strip():
        # Raise naming the defect.
        raise DepositHaltError("artifact name must be a non-empty str")
    # Reject over-long names (filesystems cap at 255 bytes; keep a clear bound).
    if len(name) > 255:
        # Raise naming the length.
        raise DepositHaltError(f"artifact name exceeds 255 characters: {len(name)}")
    # Reject path separators, drive colons and traversal outright (flat deposit by construction).
    if any(separator in name for separator in ("/", "\\", ":", "..")):
        # Raise naming the offending name.
        raise DepositHaltError(f"artifact name {name!r} is not a flat, traversal-free filename")
    # Reject the reserved manifest name (it is generated, never caller-supplied).
    if name.lower() == "manifest.json":
        # Raise naming the reserved name.
        raise DepositHaltError("artifact name 'MANIFEST.json' is reserved for the generated manifest")
    # Reject any banned non-redistributable name, case-insensitively.
    if name.lower() in {banned.lower() for banned in banned_names}:
        # Raise naming the banned artifact and the reason.
        raise DepositHaltError(
            f"artifact {name!r} is non-redistributable (P43.non_redistributable_artifacts; "
            "P11.provider_sdk_license_is_redistribution_grant == False); hashes of raw files are "
            "redistributable as evidence of custody, never the bytes"
        )
    # Return the validated name.
    return name

def build_redistributable_deposit_task_16(
    deposit_root: str,
    artifacts: Mapping[str, str],
) -> dict:
    """
    Assemble the redistributable deposit and bind it to one hash.

    Validate everything before the first write: the root is a non-empty
    ``str`` that is not an existing file; every artifact name passes
    ``_validate_artifact_name_task_16``; every payload is a ``str``;
    and all required artifacts (the seven pinned by ``P36`` plus the
    disclosed Task 15 addition) are present. Then create the directory,
    write each payload as UTF-8 bytes in name-sorted order (recording
    the SHA-256 of the written bytes), write a deterministic
    ``MANIFEST.json`` (name to digest, sorted keys, indent 2, trailing
    newline), re-read it from disk and hash those on-disk bytes.

    Parameters
    ----------
    deposit_root : str
        Directory receiving the deposit; created with parents when
        absent.
    artifacts : Mapping[str, str]
        Artifact name to text payload. Must contain every required
        redistributable artifact and no non-redistributable one.

    Returns
    -------
    dict
        ``manifest_path``: absolute path of ``MANIFEST.json``.
        ``manifest_sha256``: SHA-256 hex digest of the manifest bytes
        as read back from disk. ``artifacts``: artifact name to
        SHA-256 hex digest of its written bytes.

    Raises
    ------
    TypeError
        If ``artifacts`` is not a ``Mapping``, or a payload is not a
        ``str``.
    DepositHaltError
        If ``deposit_root`` is not a non-empty ``str``, the root path
        exists as a file, the root cannot be created, an artifact name
        is unsafe or banned, or a required artifact is missing.

    Notes
    -----
    Raw vendor files are never written; their SHA-256 hashes are
    redistributable as evidence of custody, never the bytes. One INFO
    line records the artifact count and the manifest path.
    """
    # Validate the root argument.
    if not isinstance(deposit_root, str) or not deposit_root.strip():
        # Raise naming the observed value.
        raise DepositHaltError(f"deposit_root must be a non-empty str, got {deposit_root!r}")
    # Enforce the mapping type for the artifacts.
    if not isinstance(artifacts, Mapping):
        # Raise naming the observed type.
        raise TypeError(f"artifacts must be a Mapping[str, str], got {type(artifacts).__name__}")
    # Materialise and validate the names FIRST (validation is total before any write).
    validated_names: List[str] = []
    # Walk the mapping once.
    for raw_name in artifacts:
        # Validate each name against the flat-deposit rules (banned names rejected here).
        validated_names.append(_validate_artifact_name_task_16(raw_name, banned_names=NON_REDISTRIBUTABLE_ARTIFACTS))
    # Enforce the values are text (the deposit writes UTF-8 text).
    for raw_name, raw_value in artifacts.items():
        # Type-check the payload.
        if not isinstance(raw_value, str):
            # Raise naming the artifact and the observed type.
            raise TypeError(f"artifact {raw_name!r} payload must be str, got {type(raw_value).__name__}")
    # Enforce the required set (the eight pinned artifacts) before any write.
    missing_required: List[str] = [
        required for required in REQUIRED_REDISTRIBUTABLE_ARTIFACTS if required not in artifacts
    ]
    # Any absence halts the deposit.
    if missing_required:
        # Raise listing every missing artifact.
        raise DepositHaltError(f"required redistributable artifacts missing from the deposit: {missing_required}")
    # Resolve the root and refuse to write into a non-directory path.
    root: Path = Path(deposit_root)
    # A file at the root path is a halt (we never overwrite a file implicitly).
    if root.exists() and not root.is_dir():
        # Raise naming the path.
        raise DepositHaltError(f"deposit_root {root} exists and is not a directory")
    # Create the root (and parents) when absent.
    try:
        # Create the deposit directory idempotently.
        root.mkdir(parents=True, exist_ok=True)
    except OSError as exc:
        # Wrap the OS error with the path for diagnosis.
        raise DepositHaltError(f"deposit_root {root} could not be created: {exc}") from exc
    # Write artifacts in name-sorted order for deterministic logs and listings.
    written_digests: Dict[str, str] = {}
    # Sort the names for the write order.
    for name in sorted(validated_names):
        # Encode the payload to UTF-8 bytes (no BOM).
        payload: bytes = artifacts[name].encode("utf-8")
        # Resolve the destination inside the flat deposit root.
        destination: Path = root / name
        # Write the bytes exactly as validated.
        destination.write_bytes(payload)
        # Digest the exact bytes written (not a re-read) for the manifest.
        written_digests[name] = hashlib.sha256(payload).hexdigest()
    # Serialise MANIFEST.json deterministically (sorted keys, indent 2, trailing newline).
    manifest_text: str = json.dumps(written_digests, indent=2, sort_keys=True, ensure_ascii=False) + "\n"
    # Encode the manifest to UTF-8 bytes.
    manifest_bytes: bytes = manifest_text.encode("utf-8")
    # Resolve the manifest path inside the deposit root.
    manifest_path: Path = root / "MANIFEST.json"
    # Write the manifest bytes.
    manifest_path.write_bytes(manifest_bytes)
    # Re-read the manifest from disk so the returned hash binds the on-disk bytes.
    manifest_on_disk: bytes = manifest_path.read_bytes()
    # Compute the manifest's own SHA-256 (the third-party one-operation verification handle).
    manifest_sha256: str = hashlib.sha256(manifest_on_disk).hexdigest()
    # Log one INFO line for the CI log (names only).
    logger.info("deposit written: %d artifacts + MANIFEST.json at %s", len(written_digests), manifest_path)
    # Return exactly the three pinned keys.
    return {
        "manifest_path": str(manifest_path),
        "manifest_sha256": manifest_sha256,
        "artifacts": written_digests,
    }

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3b: The verbatim governance declaration and the four-gate stamp.
# -------------------------------------------------------------------------------------------------------------------------------

def build_governance_declaration_task_16() -> dict:
    """
    Emit the frozen governance declaration, verbatim.

    Returns
    -------
    dict
        The twelve pinned boolean flags of
        ``GOVERNANCE_DECLARATION_FLAGS`` spread at the top level;
        ``four_gate_stamp``: a fresh copy of ``FOUR_GATE_STAMP`` giving
        ``G1``..``G4`` and the reproduction-citation rule;
        ``non_redistributable_artifacts``: a fresh list of the banned
        artifact names; ``closing_statement``: the frozen statement
        text; ``provenance``: the configuration pin cited for each
        flag and each gate.

    Notes
    -----
    Every call returns fresh containers, so mutating the result cannot
    alter the frozen constants.
    """
    # Assemble the declaration: the twelve pinned booleans spread verbatim.
    declaration: Dict[str, Any] = {
        # The twelve frozen booleans, in pinned order.
        **GOVERNANCE_DECLARATION_FLAGS,
        # The four-gate stamp plus the reproduction-inheritance rule (fresh copies).
        "four_gate_stamp": dict(FOUR_GATE_STAMP),
        # The non-redistributable artifacts, verbatim (a fresh list).
        "non_redistributable_artifacts": list(NON_REDISTRIBUTABLE_ARTIFACTS),
        # The verbatim closing statement (byte-exact from the step text).
        "closing_statement": CLOSING_STATEMENT,
        # The provenance block: each boolean's configuration source pin (evidence, not decoration).
        "provenance": {
            "public_data_only": "P43_GOVERNANCE_PARAMETERS.public_data_only == True",
            "live_trading_claim_permitted": "P43_GOVERNANCE_PARAMETERS.live_trading_claim_permitted == False",
            "financial_advice_claim_permitted": "P43_GOVERNANCE_PARAMETERS.financial_advice_claim_permitted == False",
            "order_sending_permitted": "P43_GOVERNANCE_PARAMETERS.order_sending_permitted == False",
            "preserve_failed_records": "P43_GOVERNANCE_PARAMETERS.preserve_failed_records == True",
            "preserve_non_scorable_records": "P43_GOVERNANCE_PARAMETERS.preserve_non_scorable_records == True",
            "model_labels_as_verified_identity": "P21_LLM_CONFIGURATION.llm_names_reported_as_labels_only; P43_GOVERNANCE_PARAMETERS.model_labels_as_verified_identity == False",
            "ai_assistance_disclosed": "P43_GOVERNANCE_PARAMETERS.ai_assistance_disclosed == True",
            "human_annotation_study_claimed": "P43_GOVERNANCE_PARAMETERS.human_annotation_study_claimed == False",
            "independent_human_review_claimed": "P43_GOVERNANCE_PARAMETERS.independent_human_review_claimed == False",
            "numericals_computed_not_generated": "P43_GOVERNANCE_PARAMETERS.numericals_computed_not_generated == True",
            "author_accountability_asserted": "P43_GOVERNANCE_PARAMETERS.author_accountability_asserted == True",
            "g1_metric_grain": "inadmissible -- the Sharpe populations are not comparable (P09_LEVEL1_AUDIT_PARAMETERS)",
            "g2_denominator_completeness": "inadmissible -- the catalog is outcome-filtered (P10_RETENTION_AUDIT_PARAMETERS)",
            "g3_execution_provenance": "inadmissible -- the labels are template-resident, not request-derived (P14/P15; P21)",
            "g4_pre_outcome_immutability": "inadmissible -- the freeze has not been executed (P08_CLEAN_OOS_PROTOCOL_PARAMETERS is PENDING)",
            "reproduction_citation": "any decision citing the reproduction inherits the reproduction tier and no higher (P00.claim_type_by_target_REVISED.T3)",
        },
    }
    # Return the frozen declaration.
    return declaration

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Step 3c: The deploy_decision interface (the four named decision classes).
# -------------------------------------------------------------------------------------------------------------------------------

def deploy_decision_task_16(
    cited_claims: Sequence[Any],
    decision_class: str,
) -> str:
    """
    Decide a downstream decision class against the claims it cites.

    Each cited record is validated structurally: it must carry
    ``claim_id``, ``tier``, ``evidence_hash`` and ``gate_status``; the
    tier must be in ``TIER_VOCABULARY``; ``evidence_hash`` must be 64
    lowercase hex characters; ``gate_status`` must map exactly
    ``G1``..``G4`` to native ``bool`` values. The class's minimum tier
    is looked up in the closed ``DECISION_CLASS_MINIMUM_TIER``
    vocabulary and compared through ``TIER_ORDER``; the verdict is
    ``"CLEARED"`` only when every cited tier reaches that floor.

    Parameters
    ----------
    cited_claims : Sequence[Any]
        Claim records cited by the decision, read by attribute access.
        A bare ``str`` element is refused.
    decision_class : str
        One of ``"capital_allocation"``, ``"model_deployment"``,
        ``"publication"``, ``"procurement"``; each requires every
        cited claim at or above its minimum tier.

    Returns
    -------
    str
        ``"CLEARED"`` when every cited tier meets the class floor,
        otherwise ``"NOT_CLEARED"``. An empty citation list clears
        vacuously.

    Raises
    ------
    TypeError
        If ``cited_claims`` is not a ``Sequence`` or is a ``str``, an
        element is a ``str`` or lacks a record field, or
        ``decision_class`` is not a ``str``.
    ValueError
        If ``decision_class`` is outside the closed vocabulary, a tier
        is not in ``TIER_VOCABULARY``, an ``evidence_hash`` is not 64
        lowercase hex characters, or a gate map is not exactly
        ``G1``..``G4`` of ``bool``.
    """
    # The citation list must be a sequence (a bare str is a sequence of characters, so refuse it explicitly).
    if not isinstance(cited_claims, Sequence) or isinstance(cited_claims, str):
        # Name the offending type.
        raise TypeError(f"cited_claims must be a Sequence[ClaimRecord], got {type(cited_claims).__name__}")
    # The tiers of the validated records, in citation order.
    tiers: List[str] = []
    # Validate every cited record structurally.
    for index, claim in enumerate(cited_claims):
        # A bare tier label binds the verdict to no evidence; refuse it.
        if isinstance(claim, str):
            # Name the position and the remedy.
            raise TypeError(
                f"cited_claims[{index}] is a bare str {claim!r}; deploy_decision consumes ClaimRecord objects "
                "(claim_id, tier, evidence_hash, gate_status), never bare tier labels"
            )
        # Every pinned field must be present on the record.
        missing: List[str] = [name for name in ("claim_id", "tier", "evidence_hash", "gate_status") if not hasattr(claim, name)]
        # A record lacking a field is not a ClaimRecord.
        if missing:
            # Name the position and the absent fields.
            raise TypeError(f"cited_claims[{index}] ({type(claim).__name__}) lacks ClaimRecord field(s) {missing}")
        # The tier must be one of the four pinned tiers.
        tier: Any = claim.tier
        # Refuse an unknown or non-str tier.
        if not isinstance(tier, str) or tier not in TIER_VOCABULARY:
            # Name the position and the vocabulary.
            raise ValueError(f"cited_claims[{index}].tier must be one of {sorted(TIER_VOCABULARY)}, got {tier!r}")
        # The evidence hash must be a SHA-256 hex digest (the record's content address).
        digest: Any = claim.evidence_hash
        # Refuse anything but 64 lowercase hex characters.
        if not isinstance(digest, str) or re.fullmatch(r"[0-9a-f]{64}", digest) is None:
            # Name the position and the value.
            raise ValueError(f"cited_claims[{index}].evidence_hash must be 64 lowercase hex characters, got {digest!r}")
        # The gate map must be exactly G1..G4 of native bools.
        gates: Any = claim.gate_status
        # Refuse a non-mapping, a wrong key set, or a non-bool value.
        if (
            not isinstance(gates, Mapping)
            or set(gates) != {"G1", "G2", "G3", "G4"}
            or any(type(value) is not bool for value in gates.values())
        ):
            # Name the position and the value.
            raise ValueError(f"cited_claims[{index}].gate_status must map exactly G1..G4 to bools, got {gates!r}")
        # Keep the validated tier.
        tiers.append(tier)
    # The decision class must be a str.
    if not isinstance(decision_class, str):
        # Name the offending type.
        raise TypeError(f"decision_class must be str, got {type(decision_class).__name__}")
    # Look up the class's minimum tier (the vocabulary is closed).
    required_tier: Optional[str] = DECISION_CLASS_MINIMUM_TIER.get(decision_class)
    # An unknown class fails closed.
    if required_tier is None:
        # Name the closed vocabulary.
        raise ValueError(
            f"decision_class must be one of {sorted(DECISION_CLASS_MINIMUM_TIER)}, got {decision_class!r}; "
            "the decision-class vocabulary is closed and fails closed"
        )
    # The ordinal floor the class requires.
    floor: int = TIER_ORDER[required_tier]
    # Cleared iff every cited tier reaches the floor.
    cleared: bool = all(TIER_ORDER[tier] >= floor for tier in tiers)
    # Return the verdict literal.
    return "CLEARED" if cleared else "NOT_CLEARED"

def _canonical_json_bytes_task_16(payload: Any) -> bytes:
    """
    Serialise a claim-record payload to canonical-JSON bytes.

    Normalise the tree to the restricted closed vocabulary
    (``None``/``bool``/``int``/``float``/``str``/``list``/``tuple``/
    ``dict``), rendering ``NaN``, ``Infinity`` and ``-Infinity`` as
    those literal tokens while keeping ``bool`` distinct from ``int``,
    then dump with sorted keys, compact separators, ``ensure_ascii``
    and ``allow_nan=False``, and encode as UTF-8 with no trailing
    newline.

    Parameters
    ----------
    payload : Any
        JSON-admissible tree restricted to the vocabulary above.
        Mapping keys must be ``str``.

    Returns
    -------
    bytes
        UTF-8 bytes of the canonical serialisation.

    Raises
    ------
    TypeError
        If a leaf has no canonical representation (for example a
        ``set``, a ``bytes`` value, or a non-``str`` mapping key).
    """
    # Normalise every leaf of the tree (unsupported leaves raise TypeError loudly).
    def _normalise(value: Any) -> Any:
        """
        Normalise one leaf or subtree to the canonical vocabulary.

        Parameters
        ----------
        value : Any
            Node of the payload tree.

        Returns
        -------
        Any
            Scalars pass through unchanged; finite floats stay floats
            while ``NaN`` and infinities become their literal tokens;
            mappings and sequences are rebuilt recursively with
            ``str`` keys.

        Raises
        ------
        TypeError
            If a mapping key is not a ``str``, or a node falls outside
            the permitted vocabulary.
        """
        # bool is a subclass of int; JSON distinguishes them, preserve both.
        if value is None or isinstance(value, (bool, int, str)):
            # Scalars pass through unchanged.
            return value
        # Floats need the corpus' literal-token handling for non-finite values.
        if isinstance(value, float):
            # NaN renders as the corpus-pinned token.
            if math.isnan(value):
                return "NaN"
            # +/-inf renders as the corpus' infinity tokens.
            if math.isinf(value):
                return "Infinity" if value > 0 else "-Infinity"
            # Finite floats stay finite floats.
            return value
        # Dicts recurse with string keys (corpus requires str keys).
        if isinstance(value, dict):
            # Normalise every value; reject non-str keys explicitly.
            normalised: dict = {}
            for key, item in value.items():
                # Reject non-string mapping keys (corpus demands str keys).
                if not isinstance(key, str):
                    raise TypeError(f"canonical JSON requires str keys; got {type(key)!r}.")
                # Recurse into the value slot.
                normalised[key] = _normalise(item)
            # Return the normalised mapping.
            return normalised
        # Lists and tuples recurse elementwise.
        if isinstance(value, (list, tuple)):
            # Normalise every element.
            return [_normalise(item) for item in value]
        # Anything else is outside the permitted vocabulary; refuse loudly.
        raise TypeError(f"canonical JSON payload carries unsupported type {type(value)!r}.")

    # Serialise with sorted keys and no whitespace, then encode to UTF-8.
    return json.dumps(
        _normalise(payload),
        sort_keys=True,
        separators=(",", ":"),
        ensure_ascii=True,
        allow_nan=False,
    ).encode("utf-8")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16: Orchestrator -- derivation helpers and the run_task_16 composition
# -------------------------------------------------------------------------------------------------------------------------------

def _pins_coverage_probe_task_16() -> None:
    """
    Assert the integrity of the frozen anticipated-violation pins.

    Require ``ANTICIPATED_VIOLATION_PINS_TASK_16`` to hold seventeen
    non-empty strings that are pairwise distinct, then join them into a
    poison document and confirm that control ids ``1..17`` are each
    found exactly once, in order. The zero polarity (the pins are absent
    from conformant outputs) cannot be re-checked from inside a build
    and is not asserted here.

    Raises
    ------
    ValueError
        If a pin is empty or not a ``str``, if two pins are equal, or
        if the poison document does not yield exactly the ids ``1..17``.
    """
    # 1) Every pin must be a non-empty string.
    for index, pin in enumerate(ANTICIPATED_VIOLATION_PINS_TASK_16, start=1):
        # The pin must exist with content.
        if not isinstance(pin, str) or not pin.strip():
            # Refuse with the offending pin index named exactly once.
            raise ValueError(f"anticipated_violation_pin_integrity_failure: pin {index} is empty or not a str")
    # 2) The pins must be pairwise distinct (a duplicate pin voids the zero-instance check for one control).
    seen_pins: set = set()
    # Walk the pins once for the duplicate check.
    for index, pin in enumerate(ANTICIPATED_VIOLATION_PINS_TASK_16, start=1):
        # A duplicate pin makes the pin set vacuous for that control.
        if pin in seen_pins:
            # Refuse with the offending pin index named exactly once.
            raise ValueError(f"anticipated_violation_pin_integrity_failure: pin {index} duplicates an earlier pin")
        # Record the pin for later comparisons.
        seen_pins.add(pin)
    # 3) The poison polarity: a document of all seventeen pins yields exactly seventeen occurrences.
    poison = "\n".join(ANTICIPATED_VIOLATION_PINS_TASK_16)
    # Collect the control ids whose pin appears in the poison document.
    found = [
        control_id
        for control_id in range(1, 18)
        if ANTICIPATED_VIOLATION_PINS_TASK_16[control_id - 1] in poison
    ]
    # The poison fixture must yield every control exactly once.
    if found != list(range(1, 18)):
        # Refuse with the observed pin hit list.
        raise ValueError(f"anticipated_violation_pin_integrity_failure: the poison fixture yields {found}, expected all 1..17")

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator helper 1: Derive the enumeration table from the text.
# -------------------------------------------------------------------------------------------------------------------------------

def _derive_enumerated_controls_task_16(text: str) -> List[Dict[str, Any]]:
    """
    Parse the enumerated-control table rows out of the text.

    Match each stripped line against
    ``ENUMERATION_TABLE_ROW_PATTERN``; for every hit, read the control
    id, control text, declined-witness reference,
    anticipated-violation reference and status (stripped and
    upper-cased). ``witness_declined`` is a fixed sentence naming the
    control id; ``witness_anticipated_violation`` is the frozen pin for
    that id from ``ANTICIPATED_VIOLATION_PINS_TASK_16`` and is never
    composed from the text.

    Parameters
    ----------
    text : str
        Study-output text carrying the Markdown enumeration table.

    Returns
    -------
    List[Dict[str, Any]]
        One dict per matched row, in line order, with keys
        ``control_id`` (``int``) and ``control_text``,
        ``witness_declined_ref``, ``witness_anticipated_violation_ref``,
        ``witness_declined``, ``witness_anticipated_violation``,
        ``status`` (all ``str``). Fewer than seventeen rows is not an
        error here; completeness is enforced by
        ``build_governance_register_task_16``.

    Raises
    ------
    ValueError
        If a parsed control id has no frozen pin, reported as
        ``anticipated_violation_pin_missing``.
    """
    # Accumulate the parsed controls in line order.
    controls: List[Dict[str, Any]] = []
    # Walk the text line by line; the table row grammar is anchored ^...$ per line.
    for line in text.splitlines():
        # Match the pinned row grammar (no match -> not a table row -> skipped).
        match = ENUMERATION_TABLE_ROW_PATTERN.match(line.strip())
        # Skip non-table lines silently (the text carries other content).
        if match is None:
            # Nothing to parse on this line.
            continue
        # Extract the control identifier from group 2 (group 1 is the declined-witness tier).
        control_id: int = int(match.group(2))
        # Extract the control text (group 3), trimmed by construction.
        control_text: str = match.group(3).strip()
        # Extract the declined-ref and anticipated-ref handles (groups 5 and 7).
        declined_ref: str = match.group(5).strip()
        # Extract the anticipated-ref handle.
        anticipated_ref: str = match.group(7).strip()
        # Extract and normalise the status token (group 8).
        status: str = match.group(8).strip().upper()
        # The frozen anticipated-violation pin for this control id, or halt (a missing pin is a
        # build configuration error, never a runtime repair opportunity).
        pin: Optional[str] = _ANTICIPATED_PIN_BY_CONTROL_ID_TASK_16.get(control_id)
        # Halt loudly when the configuration cannot satisfy the pin set.
        if pin is None:
            # Name the offending control id exactly once.
            raise ValueError(
                f"anticipated_violation_pin_missing: control {control_id} has no frozen pinned sentence in "
                f"ANTICIPATED_VIOLATION_PINS_TASK_16 (pin set covers 1..17); the derivation cannot proceed"
            )
        # Append the derived control; the anticipated-violation witness is the frozen pin (verbatim),
        # never a bracketed disclaimer (a pin set that fires is the pin set that exists).
        controls.append(
            {
                "control_id": control_id,
                "control_text": control_text,
                "witness_declined_ref": declined_ref,
                "witness_anticipated_violation_ref": anticipated_ref,
                "witness_declined": (
                    f"The study makes no claim of the kind forbidden by control {control_id}."
                ),
                "witness_anticipated_violation": pin,
                "status": status,
            }
        )
    # Return the parsed rows (possibly fewer than 17; the register enforces completeness).
    return controls

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator helper 2: Derive one ClaimRecord per macro instance.
# -------------------------------------------------------------------------------------------------------------------------------

def _normalise_claim_record_task_16(record: Any, *, caller: str) -> Dict[str, Any]:
    """
    Normalise one caller-supplied claim record to a plain dict.

    A ``Mapping`` is shallow-copied, so later mutation of the returned
    dict cannot reach the caller's object; a dataclass instance (for
    example task 14's ``ClaimRecord``) is converted with
    ``dataclasses.asdict``.

    Parameters
    ----------
    record : Any
        Caller-supplied record.
    caller : str
        Caller name, quoted in raised messages for provenance.

    Returns
    -------
    Dict[str, Any]
        The record as a plain ``dict``, keys and values preserved.

    Raises
    ------
    TypeError
        If ``record`` is neither a ``Mapping`` nor a dataclass
        instance (a dataclass class object is also refused).
    """
    # A mapping is converted with a defensive shallow copy.
    if isinstance(record, Mapping):
        # Copy so the caller's object cannot be mutated through the boundary.
        return dict(record)
    # A dataclass instance converts structurally (task_14's ClaimRecord is the pinned consumer).
    if dataclasses.is_dataclass(record) and not isinstance(record, type):
        # asdict is total: every member of the frozen dataclass converts.
        return dataclasses.asdict(record)
    # Anything else is a contract breach; name the observed type.
    raise TypeError(f"{caller}: claim_records entries must be Mapping[str, Any] or a dataclass instance, got {type(record).__name__}")

def _claim_record_digest_task_16(record: Mapping[str, Any]) -> str:
    """
    Recompute the position-bound digest of a derived claim record.

    Build the basis from the position-bound field set (``claim_id``,
    ``tier``, ``evidence_source``, ``verdict``, ``value``, ``note``,
    ``gate_status``, ``line_number``, ``span_start``, ``span_end``,
    ``span_bytes``), canonicalise it with
    ``_canonical_json_bytes_task_16`` and hash the bytes with SHA-256.

    Parameters
    ----------
    record : Mapping[str, Any]
        Derived record to re-hash.

    Returns
    -------
    str
        64-lowercase-hex SHA-256 digest of the canonical basis, or the
        empty string when any basis field is absent (such a record is
        not re-verifiable and is refused downstream).
    """
    # The position-bound field set of the derivation (order irrelevant; canonicalised by key sort).
    basis_keys = (
        "claim_id",
        "tier",
        "evidence_source",
        "verdict",
        "value",
        "note",
        "gate_status",
        "line_number",
        "span_start",
        "span_end",
        "span_bytes",
    )
    # Refuse records that do not carry the complete position-bound basis.
    if any(key not in record for key in basis_keys):
        # An incomplete basis is not a recomputable record.
        return ""
    # Rebuild the basis from the record verbatim.
    basis = {key: record[key] for key in basis_keys}
    # Canonicalise and hash (the ONE site both derivation and verification use).
    return hashlib.sha256(_canonical_json_bytes_task_16(basis)).hexdigest()

def _derive_claim_records_task_16(text: str) -> List[Dict[str, Any]]:
    """
    Derive one ClaimRecord per tieredclaim macro instance.

    Walk the macros in appearance order with a 1-based ordinal. Each
    record binds the macro's tier, stripped body, 1-based line number,
    absolute span offsets and exact span bytes. Gate status is the
    fail-closed default ``{G1: False, G2: False, G3: False, G4:
    False}``. The ``evidence_hash`` is the SHA-256 of the canonical
    JSON bytes of the position-bound basis ``(claim_id, tier,
    evidence_source, verdict, value, note, gate_status, line_number,
    span_start, span_end, span_bytes)``, so two macros on one line
    never share a digest.

    Parameters
    ----------
    text : str
        The study-output text under audit.

    Returns
    -------
    List[Dict[str, Any]]
        One record per macro, in order. Keys: ``claim_id``
        (``CLAIM-0001`` onward), ``tier``, ``evidence_source``,
        ``gate_status``, ``verdict`` (``"STAMPED"``), ``value``,
        ``note``, ``evidence_hash``, ``line_number``, ``span_start``,
        ``span_end``, ``span_bytes``.

    Raises
    ------
    Task16GovernanceError
        If a macro carries an unknown tier.
    """
    # Locate and validate every macro (unknown tiers halt here).
    macros: List[Tuple[str, str, int, int]] = _find_tiered_claim_macros_task_16(text)
    # Accumulate the records in appearance order.
    records: List[Dict[str, Any]] = []
    # Walk the macros once with a 1-based ordinal.
    for ordinal, (tier, body, start, end) in enumerate(macros, start=1):
        # The deterministic identifier of the derived claim.
        claim_id = f"CLAIM-{ordinal:04d}"
        # The macro's 1-based line number (position binding).
        line_number = text.count("\n", 0, start) + 1
        # The exact macro span bytes (two macros on one line can never share a digest).
        span = text[start:end]
        # The four admissibility gates: fail-closed default on the default path (never True by absence).
        gate_status = {"G1": False, "G2": False, "G3": False, "G4": False}
        # The deterministic evidentiary note (fixed BEFORE the basis is hashed).
        note = (
            "Derived one-per-macro ClaimRecord by run_task_16 (order of appearance); position-bound digest; "
            "gates_unverified_default_fail: G1-G4 are all False (no terminal verification was supplied to the derivation)"
        )
        # The position-bound hashed basis (the two serialised halves the pinned register demands).
        basis = {
            "claim_id": claim_id,
            "tier": tier,
            "evidence_source": "study_output_text",
            "verdict": "STAMPED",
            "value": body.strip(),
            "note": note,
            "gate_status": gate_status,
            "line_number": int(line_number),
            "span_start": int(start),
            "span_end": int(end),
            "span_bytes": span,
        }
        # Canonicalise the basis and stamp the digest (one site, one stamp).
        evidence_hash = hashlib.sha256(_canonical_json_bytes_task_16(basis)).hexdigest()
        # Append the per-macro record with the pinned ClaimRecord shape plus both halves.
        records.append(
            {
                # The deterministic identifier.
                "claim_id": claim_id,
                # The evidentiary tier from the macro's tier argument.
                "tier": tier,
                # The artifact class the verdict rests on.
                "evidence_source": "study_output_text",
                # The four admissibility gates (all False: no verification on this path).
                "gate_status": gate_status,
                # The verdict literal.
                "verdict": "STAMPED",
                # The macro body, verbatim and stripped.
                "value": body.strip(),
                # The deterministic note.
                "note": note,
                # The 64-lowercase-hex SHA-256 of the position-bound canonical basis.
                "evidence_hash": evidence_hash,
                # The macro's 1-based line number (position binding).
                "line_number": int(line_number),
                # The macro's absolute start offset in the text.
                "span_start": int(start),
                # The macro's absolute end offset in the text.
                "span_end": int(end),
                # The exact macro span bytes (two macros on one line can never share a digest).
                "span_bytes": span,
            }
        )
    # Return the derived records.
    return records

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator helper 3: The eight honest descriptor payloads.
# -------------------------------------------------------------------------------------------------------------------------------

# The seven artifacts that need no freeze evidence, pinned order is preserved by the deposit.
def _redistributable_payloads_without_freeze_task_16() -> Dict[str, str]:
    """
    Author seven redistributable artifact payloads as JSON text.

    Every payload is derived from pinned configuration values only; no
    fixture value is copied and no non-recomputable figure is invented.
    Values this module cannot supply are recorded as ``None``; where a
    dedicated reason member exists it is set alongside (for example
    ``preregistration_timestamp_absence_reason`` and
    ``cells_absence_reason``). Other absences use a differently named
    adapter (``tolerance_nine_cell`` is ``None`` with
    ``tolerance_nine_cell_source == "NOT_STATED_IN_PAPER"``) or carry
    no adjacent reason at all (``missing_result_imputation`` is
    ``None``). The first pinned artifact, the frozen ID list, needs
    the freeze stage's persisted artifacts and is built separately;
    this function authors artifacts two through eight.

    Returns
    -------
    Dict[str, str]
        Artifact name to deterministic JSON text (sorted keys, indent
        2, non-ASCII preserved, trailing newline) for artifacts two
        through eight:
        ``80_file_hash_manifest_and_acquisition_instructions``,
        ``v1_and_v2_evaluator_versions_with_change_description``,
        ``ledger_generation_logic``,
        ``nine_designated_spot_check_cells_and_both_implementations``,
        ``six_reference_definitions_with_preregistration_timestamp``,
        ``disclosure_tables_with_UNDISCLOSED_cells_intact`` and
        ``sensitivity-report-with-primary-comparison``.
    """
    # Local helper: canonical JSON text for one payload (sorted keys, indent 2, trailing newline).
    def _dump(payload: Dict[str, Any]) -> str:
        """
        Serialise one payload as canonical JSON text.

        Parameters
        ----------
        payload : Dict[str, Any]
            The artifact document to serialise.

        Returns
        -------
        str
            JSON with sorted keys, indent 2, non-ASCII preserved and a
            trailing newline.
        """
        # Return the deterministic serialisation of the payload.
        return json.dumps(payload, indent=2, sort_keys=True, ensure_ascii=False) + "\n"

    # Accumulate the payloads in pinned order.
    payloads: Dict[str, str] = {}
    # 2 -- 80_file_hash_manifest_and_acquisition_instructions (hashes are evidence of custody; the raw files are not redistributable).
    payloads["80_file_hash_manifest_and_acquisition_instructions"] = _dump(
        {
            "artifact": "80_file_hash_manifest_and_acquisition_instructions",
            "file_count": 80,
            "adjustment_views": ["unadjusted", "provider_adjusted"],
            "provider": "BaoStock",
            "provider_version": "0.9.3",
            "rows_per_bar_file": 727,
            "hashes_redistributable": True,
            "raw_files_redistributable": False,
            "reason": (
                "P11_ACQUISITION_PARAMETERS.provider_sdk_license_is_redistribution_grant == False; "
                "the SDK licence is not a data-redistribution grant; hashes carry no data"
            ),
            "manifest_location": "FILES acquisition manifest with six verification conjuncts (Task 1); byte hashes of every acquired file",
            "acquisition": "See P11_ACQUISITION_PARAMETERS for the pinned witnessed retrieval parameters (date span 2023-2025, 727 trading rows, 40-stock convenience universe, fixed order identical across views)",
        }
    )
    # 3 -- v1_and_v2_evaluator_versions_with_change_description (logic, not data).
    payloads["v1_and_v2_evaluator_versions_with_change_description"] = _dump(
        {
            "artifact": "v1_and_v2_evaluator_versions_with_change_description",
            "version_1": {
                "defect": "conditional operator interprets a panel as one Boolean when the true branch is scalar",
                "error_count": 57,
                "error_classification": "NUMERICAL_ERROR",
                "preserved_unmodified": True,
            },
            "version_2": {
                "change": "scalar broadcasting",
                "semantics_constraint": "PRESERVE_EXISTING_ELEMENTWISE_CONDITIONAL_SEMANTICS",
            },
            "re_evaluation_scope": "ALL_1047",
            "failure_only_rerun_permitted": False,
            "unchanged_record_count": 990,
            "newly_scorable_after_fix": 56,
            "residual_insufficient_after_fix": 1,
            "source": "P05_EVALUATOR_VERSIONING_PARAMETERS",
            "note": "logic, not data; the v1 package is preserved unmodified (Task 13 archive); the v2 package is bound to the registry diff",
        }
    )
    # 4 -- ledger_generation_logic (source, not the resulting ledger table).
    payloads["ledger_generation_logic"] = _dump(
        {
            "artifact": "ledger_generation_logic",
            "source": "assign_ledger_task_10 (task_responses/task_10.py)",
            "algorithm": (
                "exactly one ledger state per record under precedence "
                "(unsupported, insufficient, scorable) and the disclosed reverse triple "
                "(scorable, insufficient, unsupported); conservation scorable + insufficient + unsupported == denominator_total"
            ),
            "denominator_total": 1047,
            "states": ["scorable", "insufficient", "unsupported"],
            "missing_result_imputation": None,
            "excluded": "the resulting ledger table (may contain expression texts)",
            "pins": "P04_LEDGER_PARAMETERS",
        }
    )
    # 5 -- nine_designated_spot_check_cells_and_both_implementations (instrument, not results).
    payloads["nine_designated_spot_check_cells_and_both_implementations"] = _dump(
        {
            "artifact": "nine_designated_spot_check_cells_and_both_implementations",
            "cell_count": 9,
            "selection_timing": "A_PRIORI",
            "independence": "SEPARATE_TIE_RANKING_PEARSON_IMPLEMENTATION",
            "pass_requirement": "9_of_9",
            "tolerance_nine_cell": None,
            "tolerance_nine_cell_source": "NOT_STATED_IN_PAPER",
            "instrument": (
                "nine designated record/date rank-IC cells recomputed with a separately written "
                "tie-ranking/Pearson calculation; instrument, not results; the pass count is deliberately not restated"
            ),
            "cells": None,
            "cells_absence_reason": "the nine designated cells and both implementations live in the verification step (Task 14 artefacts); not recomputable from this module's inputs",
            "pins": "P18_VERIFICATION_PARAMETERS",
        }
    )
    # 6 -- six_reference_definitions_with_preregistration_timestamp (definitions, not the means).
    payloads["six_reference_definitions_with_preregistration_timestamp"] = _dump(
        {
            "artifact": "six_reference_definitions_with_preregistration_timestamp",
            "preregistered_before_outcome_computation": True,
            "signs_selected_on_scores": False,
            "report_all_regardless_of_sign": True,
            "promoted_to_profitability_claim": False,
            "signals": [
                {"id": "r1", "name": "nominal close level", "formula": "close", "orientation": "as-is", "horizon": None},
                {"id": "r2", "name": "share volume", "formula": "volume", "orientation": "as-is", "horizon": None},
                {"id": "r3", "name": "nominal traded amount", "formula": "amount", "orientation": "as-is", "horizon": None},
                {"id": "r4", "name": "one-day nominal reversal", "formula": "neg(close / lag(close, 1) - 1)", "orientation": "negate", "horizon": 1},
                {"id": "r5", "name": "five-day nominal reversal", "formula": "neg(close / lag(close, 5) - 1)", "orientation": "negate", "horizon": 5},
                {"id": "r6", "name": "twenty-day nominal change", "formula": "close / lag(close, 20) - 1", "orientation": "as-is", "horizon": 20},
            ],
            "preregistration_timestamp": None,
            "preregistration_timestamp_absence_reason": "carried by the freeze artefact (Task 7 / Task 12 artefacts); not recomputable from this module's inputs",
            "note": "formula-derived definitions, not the means; P19_REFERENCE_PREREGISTRATION_PARAMETERS",
        }
    )
    # 7 -- disclosure_tables_with_UNDISCLOSED_cells_intact (P45 reproduced verbatim).
    payloads["disclosure_tables_with_UNDISCLOSED_cells_intact"] = _dump(
        {
            "artifact": "disclosure_tables_with_UNDISCLOSED_cells_intact",
            "p45_disclosure_gaps": {gap_key: False for gap_key in P45_GAP_KEYS},
            "reproduction_r_gaps": [
                {"gap": gap_key, "acknowledged_absence": descriptor}
                for gap_key, descriptor in REPRODUCTION_R_GAP_LINES
            ],
            "undisclosed_leaves": (
                "every literal UNDISCLOSED leaf of STUDY_CONFIG is reproduced as an acknowledged absence, never inferred "
                "(P45_DISCLOSURE_GAPS; P21_LLM_CONFIGURATION; P11; P15; P16; P17)"
            ),
            "source": "P45_DISCLOSURE_GAPS",
        }
    )
    # 8 -- sensitivity-report-with-primary-comparison (disclosed addition, Task 15).
    payloads["sensitivity-report-with-primary-comparison"] = _dump(
        {
            "artifact": "sensitivity-report-with-primary-comparison",
            "addition": True,
            "disclosed_as": "an addition beyond the paper's deliverables",
            "source": "Task 15 sensitivity archive (SENSITIVITY_REPORT + SENSITIVITY_HASH + VARIANT_RUNS)",
            "stamp": "SENSITIVITY__[R]__NOT_A_PAPER_FINDING",
            "claim_tier_ceiling": "reconstruction",
            "note": (
                "disclosed as an addition per P43.redistributable_artifacts and the Task 15 mandate; "
                "carries the primary comparison side-by-side; never a paper finding; "
                "only the reproduction's own reconstruction tier may be stamped beneath it"
            ),
        }
    )
    # Return the seven payloads.
    return payloads

def _frozen_id_list_payload_task_16(
    ordered_id_list: Optional[Sequence[str]],
    cohort_hash: Optional[str],
    freeze_timestamp: Optional[datetime.datetime],
) -> Dict[str, Any]:
    """
    Validate the freeze evidence and render the frozen ID-list artifact.

    Collect the defects of all three members in one pass: the ordered
    list must be a non-empty, duplicate-free sequence of non-empty
    strings; the cohort hash must be 64 lowercase hex characters; the
    timestamp must be timezone-aware. Any defect halts with every
    missing or malformed member named at once and nothing emitted. On
    success the artifact carries the ordered identifiers, the cohort
    hash, the ISO 8601 timestamp, the SHA-256 of the byte-accurate
    serialisation of the ID list (``json.dumps`` with separators
    ``",:"``, UTF-8, LF terminator) and the serialisation context.

    Parameters
    ----------
    ordered_id_list : Optional[Sequence[str]]
        The frozen cohort's ordered identifiers.
    cohort_hash : Optional[str]
        The SHA-256 of the frozen ordered-ID-list bytes (64 lowercase
        hex).
    freeze_timestamp : Optional[datetime.datetime]
        The freeze instant; must be timezone-aware (Asia/Shanghai per
        the freeze protocol).

    Returns
    -------
    Dict[str, Any]
        The complete artifact document. Keys are ``artifact``,
        ``cohort_size_pinned`` (``1047``), ``cohort_size_actual``,
        ``hash_algorithm``, ``cohort_hash``, ``freeze_timestamp``,
        ``ordered_id_list``, ``ordered_id_list_bytes_sha256``,
        ``serialization`` (flags and ``byte_count``), ``provenance``,
        ``id_list_elision``, ``redistributable`` and ``availability``.

    Raises
    ------
    DepositHaltError
        If any of the three members is absent, mis-typed, empty,
        duplicated, non-hex or naive; nothing is emitted.
    """
    # Accumulate every defect in one pass (never one-at-a-time drip).
    missing: List[str] = []
    # The ordered list must be a non-empty sequence of non-empty, unique strings.
    if (
        ordered_id_list is None
        or not isinstance(ordered_id_list, Sequence)
        or isinstance(ordered_id_list, (str, bytes))
        or len(ordered_id_list) == 0
    ):
        # The list is absent or empty.
        missing.append("ordered_id_list (a non-empty Sequence[str] of FROZEN cohort identifiers)")
    else:
        # Walk the identifiers once for the shape checks.
        for position, item in enumerate(ordered_id_list):
            # Every identifier must be a non-empty string.
            if not isinstance(item, str) or not item.strip():
                # Name the offending position and the observed type.
                missing.append(f"ordered_id_list[{position}] (non-empty str identifiers required, got {type(item).__name__})")
                # One defect per pass is enough to halt.
                break
        # The frozen cohort is ordered and unique by protocol.
        if len(set(str(item) for item in ordered_id_list)) != len(ordered_id_list):
            # Duplicates corrupt the frozen order.
            missing.append("ordered_id_list (identifiers must be unique; duplicates corrupt the frozen order)")
    # The cohort hash must be exactly 64 lowercase hex characters.
    if (
        not isinstance(cohort_hash, str)
        or len(cohort_hash) != 64
        or any(character not in "0123456789abcdef" for character in cohort_hash)
    ):
        # Name the malformed member.
        missing.append("cohort_hash (64 lowercase hex SHA-256 digest of the frozen ordered-ID-list bytes)")
    # The freeze timestamp must be timezone-aware.
    if (
        not isinstance(freeze_timestamp, datetime.datetime)
        or freeze_timestamp.tzinfo is None
        or freeze_timestamp.utcoffset() is None
    ):
        # Name the malformed member.
        missing.append("freeze_timestamp (a timezone-aware datetime; Asia/Shanghai per the freeze protocol)")
    # Any defect halts with everything named at once; nothing is written for this artefact.
    if missing:
        # Refuse with the complete listing of what is absent or malformed.
        raise DepositHaltError(
            "frozen_1047_id_list_with_hash_and_freeze_timestamp cannot be built: missing or malformed "
            f"member(s): {'; '.join(missing)}; nothing is written for this artefact"
        )
    # The byte-accurate serialisation of the ordered list (string-verbatim, one line, LF).
    ordered_list_bytes = (
        json.dumps(list(ordered_id_list), ensure_ascii=False, separators=(",", ":")).encode("utf-8") + b"\n"
    )
    # The artifact document (complete; nothing copied from fixtures; nothing elided).
    return {
        "artifact": "frozen_1047_id_list_with_hash_and_freeze_timestamp",
        "cohort_size_pinned": 1047,
        "cohort_size_actual": len(ordered_id_list),
        "hash_algorithm": "SHA-256",
        "cohort_hash": cohort_hash,
        "freeze_timestamp": freeze_timestamp.isoformat(),
        "ordered_id_list": list(ordered_id_list),
        "ordered_id_list_bytes_sha256": hashlib.sha256(ordered_list_bytes).hexdigest(),
        "serialization": {
            "sorted_keys": True,
            "separators": ",:",
            "encoding": "utf-8",
            "lf_terminated": True,
            "byte_count": len(ordered_list_bytes),
        },
        "provenance": (
            "the SHA-256 the freeze stage wrote (Artifact P07_FREEZE_PROTOCOL_PARAMETERS): the digest of the "
            "ordered frozen ID list bytes; reproduced end to end by a third-party reader from this artefact"
        ),
        "id_list_elision": {
            "elided": False,
            "reason": None,
            "adjudication": "not elided; the identifiers ship in this artefact",
        },
        "redistributable": True,
        "availability": (
            "the frozen ordered ID list, its SHA-256 hash and the freeze timestamp were supplied to this "
            "deposit by the caller from the freeze stage's persisted artefacts; this artefact IS the pinned deposit"
        ),
    }

def _redistributable_payloads_task_16(
    ordered_id_list: Optional[Sequence[str]] = None,
    cohort_hash: Optional[str] = None,
    freeze_timestamp: Optional[datetime.datetime] = None,
) -> Dict[str, str]:
    """
    Author the eight redistributable artifact payloads as JSON text.

    Build the freeze-dependent artifact first through the frozen-ID-list
    validator (all or none on the pinned-mandatory member), then add the
    seven freeze-independent payloads, keyed by artifact name. No
    payload is copied from a fixture; absent quantities are recorded as
    ``None`` with whatever adjacent reason member the payload provides.

    Parameters
    ----------
    ordered_id_list : Optional[Sequence[str]], optional
        The frozen cohort's ordered identifiers (freeze stage
        artifact). Default is ``None``.
    cohort_hash : Optional[str], optional
        The SHA-256 of the frozen ordered-ID-list bytes (64 lowercase
        hex). Default is ``None``.
    freeze_timestamp : Optional[datetime.datetime], optional
        The freeze instant; must be timezone-aware. Default is
        ``None``.

    Returns
    -------
    Dict[str, str]
        The eight artifact names mapped to their JSON text (UTF-8):
        ``frozen_1047_id_list_with_hash_and_freeze_timestamp`` plus the
        seven freeze-independent payloads.

    Raises
    ------
    DepositHaltError
        If the freeze evidence for the first pinned artifact is absent
        or malformed.
    """
    # Validate the freeze evidence FIRST; the pinned-mandatory member is all-or-none.
    artifact_one = _frozen_id_list_payload_task_16(ordered_id_list, cohort_hash, freeze_timestamp)
    # Build the seven honest payloads.
    payloads = _redistributable_payloads_without_freeze_task_16()
    # Attach the first artefact at its pinned name.
    payloads["frozen_1047_id_list_with_hash_and_freeze_timestamp"] = (
        json.dumps(artifact_one, indent=2, sort_keys=True, ensure_ascii=False) + "\n"
    )
    # Return the eight payloads (the deposit enforces the required set itself).
    return payloads

# -------------------------------------------------------------------------------------------------------------------------------
# Task 16, Orchestrator: run_task_16
# -------------------------------------------------------------------------------------------------------------------------------

def run_task_16(
    scratch_dir: str,
    study_output_text: str,
    whitelist: Mapping[str, str] | None = None,
    claim_records: Sequence[dict] | None = None,
    dependency_bundles: Mapping[str, object] | None = None,
    freeze_cohort: Optional[Mapping[str, object]] = None,
) -> dict:
    """
    Run the three steps of Task 16 over the supplied study-output text.

    Step 1 derives the seventeen-control enumeration table from the
    text and builds the governance register (control completeness, the
    prohibited-phrase scan and the verbatim anticipated-violation
    check). Step 2 resolves the claim records, asserts the
    macro/record bijection and the bare-literal halt, audits the
    disclosure gaps and the reproduction's own ``[R]`` choices against
    the module's pinned P45 mirror, and assigns the ``T1``..``T4`` tiers
    under the honest default (no recovered private artifacts, no gate
    failure). Step 3 builds the frozen governance declaration,
    assembles the eight redistributable payloads, writes the deposit
    beneath ``<scratch_dir>/redistributable_deposit`` with its
    ``MANIFEST.json``. Step 4, the certified-release guard, then
    recomputes the position-bound digest of every record this run
    derived (only when ``claim_records`` is ``None``), appends
    ``pinned_mandatory_member_unavailable:...`` to the gate failures
    when the deposit is incomplete, and raises
    ``Task16GovernanceError`` on any ``record_digest_mismatch:``
    failure before the orchestrator bundle is returned. The
    dependency-bundle mapping is shallow-copied into a new dict; the
    upstream bundle values are preserved by identity and are not
    deep-copied, re-validated or re-implemented.

    Parameters
    ----------
    scratch_dir : str
        Directory the run writes into (created when absent); the
        deposit goes beneath ``scratch_dir/redistributable_deposit``.
    study_output_text : str
        The plain-text extraction of the study output to audit.
    whitelist : Mapping[str, str] | None, optional
        Documented permitted uses handed to the scanner, keyed by the
        flagged token with the justification as value. Default is
        ``None``.
    claim_records : Sequence[dict] | None, optional
        ClaimRecords to assert against the text's ``tieredclaim``
        instances; their count must equal the macro count. Default is
        ``None``, which derives one record per macro instance.
    dependency_bundles : Mapping[str, object] | None, optional
        Verified upstream bundles to pass through opaquely; keys must
        be strings. Default is ``None`` (empty mapping).
    freeze_cohort : Optional[Mapping[str, object]], optional
        Freeze stage artifacts with keys ``ordered_id_list``,
        ``cohort_hash`` and ``freeze_timestamp``. Default is ``None``:
        the pinned-mandatory first artifact is then written as an
        explicit UNAVAILABLE document, the deposit reports
        ``complete == False`` and the terminal verdict is not
        certified.

    Returns
    -------
    dict
        The orchestrator bundle. Keys are ``claim_register`` (the
        derived or supplied records), ``manifest_path`` and
        ``manifest_sha256`` (the deposit manifest and its SHA-256),
        ``governance_declaration`` (twelve booleans, four-gate stamp,
        verbatim closing statement), ``claim_tiers`` (``{"T1":
        "inadmissible", "T2": "inadmissible", "T3": "reproduction",
        "T4": "reconstruction"}``), ``dependency_bundles`` (the
        pass-through mapping), and the audit-trail extras
        ``governance_register``, ``tier_stamps``, ``disclosure_audit``,
        ``redistributable_deposit`` (``{"complete": bool,
        "unavailable_members": list[str]}``) and ``terminal_verdict``
        (``{"certified": bool, "gate_failures": list[str]}``).

    Raises
    ------
    TypeError
        If ``scratch_dir`` is not a non-empty string; if
        ``dependency_bundles`` is not a mapping or has non-string keys;
        if ``study_output_text`` is mistyped (via the text validator);
        if ``whitelist`` is mistyped (via the whitelist validator); or
        if any ``claim_records`` entry is mistyped at the boundary (via
        the record normaliser).
    Task16GovernanceError
        If the scratch path exists and is not a directory, cannot be
        created, or any governance check refuses: an incomplete
        control enumeration, an impermissible scanner finding, a
        verbatim anticipated-violation sentence, a bare numeric
        literal, a macro/record count mismatch, a missing disclosure
        gap, or a recomputed record digest mismatch among records
        derived by this run (caller-supplied records are not
        re-hashed here).
    DepositHaltError
        If the deposit cannot be written at the resolved location.

    Notes
    -----
    Emits a completion INFO line with the claim count, macro count,
    manifest SHA-256 and certified flag. When the pinned-mandatory
    member is unavailable an additional INFO line notes that seven of
    eight artifacts were built. A digest mismatch among records derived
    by this run refuses certified release outright; caller-supplied
    records are not re-hashed here.
    """
    # Validate the scratch directory argument.
    if not isinstance(scratch_dir, str) or not scratch_dir.strip():
        # Raise naming the observed value.
        raise TypeError(f"scratch_dir must be a non-empty str, got {scratch_dir!r}")
    # Resolve the scratch root and refuse a non-directory path.
    root: Path = Path(scratch_dir)
    # A file at the scratch path is a halt.
    if root.exists() and not root.is_dir():
        # Raise naming the path.
        raise Task16GovernanceError(f"scratch_dir {root} exists and is not a directory")
    # Create the scratch directory (idempotent).
    try:
        # Create the directory and any parents.
        root.mkdir(parents=True, exist_ok=True)
    except OSError as exc:
        # Wrap the OS error with the path.
        raise Task16GovernanceError(f"scratch_dir {root} could not be created: {exc}") from exc
    # Validate the text argument once for the whole run.
    validated_text: str = _validate_study_output_text_task_16(study_output_text, caller="run_task_16")
    # Validate the optional whitelist (shared validator, default-deny).
    validated_whitelist: Optional[Mapping[str, str]] = _validate_whitelist_task_16(whitelist, caller="run_task_16")
    # Validate the optional dependency bundles; default to an empty mapping.
    if dependency_bundles is None:
        # The default bundle mapping (identity-preserving pass-through, empty by default).
        bundles: Dict[str, object] = {}
    else:
        # Enforce the mapping type.
        if not isinstance(dependency_bundles, Mapping):
            # Raise naming the observed type.
            raise TypeError(f"dependency_bundles must be a Mapping[str, object], got {type(dependency_bundles).__name__}")
        # Enforce string keys for the bundle namespace.
        for key in dependency_bundles:
            # Keys must be str.
            if not isinstance(key, str):
                # Raise naming the observed key type.
                raise TypeError(f"dependency_bundles keys must be str, got {type(key).__name__}")
        # Keep the caller's object references verbatim (identity is the contract).
        bundles = dict(dependency_bundles)
    # ----- Step 1: the prose audit --------------------------------------------------------------------------
    # Derive the enumeration table from the text (the table format is pinned by the step text).
    derived_controls: List[Dict[str, Any]] = _derive_enumerated_controls_task_16(validated_text)
    # Build the register (completeness, scanner and verbatim anticipated-violation checks all halt here).
    register: Dict[str, Any] = build_governance_register_task_16(validated_text, derived_controls, validated_whitelist)
    # ----- Step 2: tier stamping and the disclosure-gap audit --------------------------------------------------
    # Resolve the claim records: a caller-supplied, upstream-verified sequence is preferred;
    # otherwise one evidence- and gate-carrying record is derived per macro (never bare tier strings).
    if claim_records is not None:
        # Normalise the caller sequence into plain dicts (Mapping passes through; dataclass records convert).
        resolved_claim_records = [
            _normalise_claim_record_task_16(record, caller="run_task_16")
            for record in claim_records
        ]
        # The Step 2 bijection pin adjusts to the caller register: the counts must coincide
        # one-to-one by the Step 2 pin; a breach is an explicit halt naming BOTH counts.
        macro_count = len(_find_tiered_claim_macros_task_16(validated_text))
        # Any mismatch is a contract breach (either direction); never a silent truncation or padding.
        if len(resolved_claim_records) != macro_count:
            # Halt naming both counts exactly once.
            raise Task16GovernanceError(
                f"Step 2 bijection pin: supplied claim_records count {len(resolved_claim_records)} "
                f"!= macro count {macro_count}; macros and records must coincide one-to-one; "
                f"both counts named"
            )
    else:
        # The default path: derive one evidence- and gate-carrying record per macro instance.
        resolved_claim_records = _derive_claim_records_task_16(validated_text)
    # Enforce the build-time bijection and the bare-literal halt.
    stamps: Dict[str, Any] = verify_tier_stamps_task_16(validated_text, resolved_claim_records)
    # Build the module's pinned P45 mirror (the audit's config cross-check; honest: it re-states the False pins).
    config_mirror: Dict[str, Mapping[str, bool]] = {
        "P45_DISCLOSURE_GAPS": {gap_key: False for gap_key in P45_GAP_KEYS},
    }
    # Audit the disclosure table and the reproduction's own [R] choices.
    audit: Dict[str, Any] = audit_disclosure_gaps_task_16(validated_text, config_mirror)
    # Assign the T1..T4 tiers under the pinned taxonomy with the honest default (no recovered private artifacts).
    claim_tiers: Dict[str, str] = {
        "T1": assign_claim_tier_task_16("run_task_16:claim_tiers:T1", target_id="T1"),
        "T2": assign_claim_tier_task_16("run_task_16:claim_tiers:T2", target_id="T2"),
        "T3": assign_claim_tier_task_16("run_task_16:claim_tiers:T3", target_id="T3"),
        "T4": assign_claim_tier_task_16("run_task_16:claim_tiers:T4", target_id="T4"),
    }
    # ----- Step 3: the final governance deposit and declaration --------------------------------------------------
    # Build the frozen governance declaration (twelve booleans, four-gate stamp, verbatim closing statement).
    declaration: Dict[str, Any] = build_governance_declaration_task_16()
    # Collect the freeze evidence BEFORE the deposit begins (Threads: StudyResult.cohort_hash,
    # freeze meta, the freeze stage's ordered identifiers; nothing else is accepted as proof).
    freeze_ordered = freeze_cohort.get("ordered_id_list") if isinstance(freeze_cohort, Mapping) else None
    # The cohort hash from the same mapping.
    freeze_hash = freeze_cohort.get("cohort_hash") if isinstance(freeze_cohort, Mapping) else None
    # The freeze timestamp from the same mapping.
    freeze_timestamp = freeze_cohort.get("freeze_timestamp") if isinstance(freeze_cohort, Mapping) else None
    # The pinned-mandatory first artefact halts closed without the freeze evidence; the other seven build honestly.
    try:
        # All-or-none build of the pinned-mandatory member plus the seven honest payloads.
        payloads: Dict[str, str] = _redistributable_payloads_task_16(
            ordered_id_list=freeze_ordered,
            cohort_hash=freeze_hash,
            freeze_timestamp=freeze_timestamp,
        )
        # The deposit is complete on this branch; no member is unavailable.
        deposit_complete = True
        # Nothing is unavailable on this branch.
        unavailable_members: List[str] = []
    except DepositHaltError as error:
        # The pinned-mandatory member failed closed with the missing members named at the source;
        # the seven remaining artifacts build honestly; the absence is named, never silent.
        logger.info("redistributable deposit: 7 of 8 artifacts built; 1 member unavailable (see bundle deposit notes)")
        # Author the pinned-mandatory member as an explicit, plainly-named absence document
        # (never a null-valued "available" artefact; INADMISSIBLE_AS_CUSTODY_EVIDENCE by construction).
        absence: Dict[str, Any] = {
            "artifact": "frozen_1047_id_list_with_hash_and_freeze_timestamp",
            "availability": "UNAVAILABLE_IN_THIS_RUN",
            "cohort_size_pinned": 1047,
            "hash_algorithm": "SHA-256",
            "cohort_hash": None,
            "freeze_timestamp": None,
            "ordered_id_list": None,
            "unavailable_members": ["frozen_1047_id_list_with_hash_and_freeze_timestamp"],
            "reason": (
                "the freeze stage's persisted artifacts (ordered_id_list, cohort_hash, freeze_timestamp) were not "
                "supplied to this deposit; this document names the absence plainly and is NOT the frozen artefact; "
                "it is not custody evidence; re-run with the freeze evidence threaded (StudyResult.cohort_hash, "
                f"freeze meta) to obtain the real artifact; inner reason: {error}"
            ),
            "id_list_elision": {
                "elided": True,
                "reason": "freeze artifacts were not supplied to this run",
                "adjudication": "unavailable, not operator-adjudicated",
            },
            "admissibility": "INADMISSIBLE_AS_CUSTODY_EVIDENCE",
            "redistributable": True,
        }
        # Build the seven honest payloads.
        payloads = _redistributable_payloads_without_freeze_task_16()
        # Author the loud absence document at the pinned artifact name.
        payloads["frozen_1047_id_list_with_hash_and_freeze_timestamp"] = (
            json.dumps(absence, indent=2, sort_keys=True, ensure_ascii=False) + "\n"
        )
        # The deposit is incomplete by construction; the bundle says so plainly.
        deposit_complete = False
        # The unavailable member is named once, everywhere.
        unavailable_members = ["frozen_1047_id_list_with_hash_and_freeze_timestamp"]
    # Deposit them beneath the run directory; the deposit enforces the required set itself.
    deposit: Dict[str, Any] = build_redistributable_deposit_task_16(
        str(root / "redistributable_deposit"),
        payloads,
    )
    # The completion census for the audit trail (counts and names only).
    deposit_notes = {
        "complete": bool(deposit_complete),
        "unavailable_members": list(unavailable_members),
    }
    # ----- Step 4: the certified-release guard at the seam every certified bundle must cross -------
    # Recompute the digest of every record THIS build derived (caller-supplied records are
    # verified at their own construction site; they are never re-derived or re-hashed here).
    gate_failures: List[str] = []
    # The derivation path is re-verified in full.
    if claim_records is None:
        # Walk every derived record once for the recompute.
        for record in resolved_claim_records:
            # Recompute the record's position-bound digest.
            recomputed = _claim_record_digest_task_16(record)
            # A missing basis or a digest mismatch is a gate failure; the claim id is named exactly once.
            if not recomputed or recomputed != record.get("evidence_hash"):
                # Record the failing claim id.
                gate_failures.append(f"record_digest_mismatch:{record.get('claim_id')}")
    # A pinned-mandatory member this build could not produce is named, never certified away.
    if not deposit_complete:
        # Name the unavailable member once.
        gate_failures.append("pinned_mandatory_member_unavailable:frozen_1047_id_list_with_hash_and_freeze_timestamp")
    # The terminal verdict: certified iff every gate held (never when the pinned member is absent).
    terminal_verdict = {
        "certified": not gate_failures,
        "gate_failures": list(gate_failures),
    }
    # A digest mismatch refuses certified release outright (an incomplete archive is disclosed, never shipped as verified).
    digest_failures = [failure for failure in gate_failures if failure.startswith("record_digest_mismatch:")]
    # Any digest failure halts with the failing stage named exactly once; no certified-shaped output ships.
    if digest_failures:
        # Raise naming the failing stage once and the failing claims.
        raise Task16GovernanceError(
            "certified-release guard: halt_reason=certified_release_refused; failing stage="
            f"step_4_certified_release; gate failures: {', '.join(digest_failures)}; "
            "no certified-shaped output ships"
        )
    # Assemble the bundle with the pinned keys plus the audit-trail extras.
    bundle: Dict[str, Any] = {
        "claim_register": resolved_claim_records,
        "manifest_path": deposit["manifest_path"],
        "manifest_sha256": deposit["manifest_sha256"],
        "governance_declaration": declaration,
        "claim_tiers": claim_tiers,
        "dependency_bundles": bundles,
        "governance_register": register,
        "tier_stamps": stamps,
        "disclosure_audit": audit,
        "redistributable_deposit": deposit_notes,
        "terminal_verdict": terminal_verdict,
    }
    # Emit one INFO line for the CI log (counts only).
    logger.info(
        "run_task_16 complete: %d claims, %d macros, manifest %s, certified=%s",
        len(resolved_claim_records),
        stamps["tiered_instances"],
        deposit["manifest_sha256"],
        terminal_verdict["certified"],
    )
    # Return the bundle.
    return bundle
```

```python
# Study Execution

# ==============================================================================
# Study Execution
# ==============================================================================

# Notebook-added cell: the invocation of the top-level entry point.
# The entry point is run_study (Task 14; "the acceptance contract's entry point").
# A successful PRODUCTION run reports admissibility_report["run_mode"]["run_mode"]
# == "PRODUCTION" with admissibility_report["run_mode"]["admissible"] == True and
# admissibility_report["run_mode"]["n_stages_halted"] == 0.
#
# Honesty gate (P36/P43; residual_issues_20261112.txt, section 8): the ten input
# artifacts run_study requires (FILES; UNADJUSTED_BARS / ADJUSTED_BARS;
# PRESERVED_INPUT_ROWS; NATIVE_OOS_CHECKPOINT; OUTCOME_FILTERED_CATALOG;
# RUN_REPORT_TOTALS / RUN_REPORT_ROUNDS; COLLECTOR_A / COLLECTOR_B) do not exist in
# this project - the raw payloads are not publicly redistributable, and this study
# cannot run here. Point the AUDIT_STUDY_INPUTS environment variable at the
# directory holding the ten frames (parquet or csv) to run the study; absent that,
# this cell records STUDY_STATUS = "DATA_ABSENT: ..." instead of inventing any figure.

_STUDY_INPUTS_DIR: Path = Path(os.environ.get("AUDIT_STUDY_INPUTS", str(Path.cwd() / "study_inputs")))

_REQUIRED_FRAMES = (
    ("files", "FILES"),
    ("unadjusted", "UNADJUSTED_BARS"),
    ("adjusted", "ADJUSTED_BARS"),
    ("preserved_rows", "PRESERVED_INPUT_ROWS"),
    ("native_oos_checkpoint", "NATIVE_OOS_CHECKPOINT"),
    ("outcome_filtered_catalog", "OUTCOME_FILTERED_CATALOG"),
    ("run_report_totals", "RUN_REPORT_TOTALS"),
    ("run_report_rounds", "RUN_REPORT_ROUNDS"),
    ("collector_a", "COLLECTOR_A"),
    ("collector_b", "COLLECTOR_B"),
)


def _load_frame_task_exec(corpus_dir: Path, stem: str) -> pd.DataFrame:
    """
    Load one corpus frame from a parquet file or a CSV file.

    Return the frame stored as ``<stem>.parquet`` when that file
    exists, otherwise as ``<stem>.csv``; the parsed frame is returned
    unmodified, with no schema or date validation.

    Parameters
    ----------
    corpus_dir : Path
        Directory holding the required input frames.
    stem : str
        Artifact stem of the frame, e.g. ``"FILES"``; no extension.

    Returns
    -------
    pd.DataFrame
        The parsed frame, from the parquet file when present,
        otherwise from the CSV file.

    Raises
    ------
    FileNotFoundError
        If neither ``<stem>.parquet`` nor ``<stem>.csv`` is a regular
        file in ``corpus_dir``.
    """
    parquet_path = corpus_dir / f"{stem}.parquet"
    csv_path = corpus_dir / f"{stem}.csv"
    if parquet_path.is_file():
        return pd.read_parquet(parquet_path)
    if csv_path.is_file():
        return pd.read_csv(csv_path)
    raise FileNotFoundError(f"DATA_ABSENT: required frame {stem} absent from {corpus_dir}")

def _load_corpus_task_exec(corpus_dir: Path) -> Dict[str, pd.DataFrame]:
    """
    Load all required corpus frames, reporting every absence at once.

    Collect every registry stem lacking both ``<stem>.parquet`` and
    ``<stem>.csv``, raise one ``FileNotFoundError`` naming them all,
    else load each frame with ``_load_frame_task_exec`` keyed by
    registry key.

    Parameters
    ----------
    corpus_dir : Path
        Directory holding the required input frames.

    Returns
    -------
    Dict[str, pd.DataFrame]
        One entry per registry pair, in registry order: logical key
        mapped to its parsed frame.

    Raises
    ------
    FileNotFoundError
        If any required stem has neither file; the message names every
        missing stem.
    """
    missing = [stem for _key, stem in _REQUIRED_FRAMES
               if not (corpus_dir / f"{stem}.parquet").is_file()
               and not (corpus_dir / f"{stem}.csv").is_file()]
    if missing:
        raise FileNotFoundError(
            "DATA_ABSENT: required input artifacts absent from "
            f"{corpus_dir}: {', '.join(missing)}"
        )
    return {key: _load_frame_task_exec(corpus_dir, stem) for key, stem in _REQUIRED_FRAMES}

try:
    _corpus: Dict[str, pd.DataFrame] = _load_corpus_task_exec(_STUDY_INPUTS_DIR)
except FileNotFoundError:
    STUDY_RESULT = None
    STUDY_STATUS = (f"DATA_ABSENT: required input artifacts absent under "
                    f"{_STUDY_INPUTS_DIR}; set AUDIT_STUDY_INPUTS to the corpus directory")
else:
    STUDY_RESULT = run_study(
        **_corpus,
        config=STUDY_CONFIG,
        stages=list(STAGE_CALLABLES),
        halt_on_failure=True,
    )
    _rm = STUDY_RESULT.admissibility_report.get("run_mode", {}) if isinstance(STUDY_RESULT.admissibility_report, dict) else {}
    STUDY_STATUS = (
        f"run_mode={_rm.get('run_mode')} admissible={_rm.get('admissible')} "
        f"n_stages_halted={_rm.get('n_stages_halted')}"
    )

print(f"STUDY_STATUS = {STUDY_STATUS}")
```
